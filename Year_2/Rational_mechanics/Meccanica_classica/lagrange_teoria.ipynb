### Roba utile

**Nomenclatura**

$q_a$: coordinata generalizzata  
$p_a$. il momento coniugato alla variabile $q_a$  

utile così Euler-Lagrange diventa:  
$\dot{p}_a= \frac{\partial L}{\partial q_a}$  

simile a: $\dot{p}=F=\frac{\partial U}{\partial x}$  

**Calcolare altre coordinate**

Usando Lagrangiana  

$\begin{cases}
p_a = \frac{\partial L}{\partial \dot{q}_a} \\ 
\dot{p}_a = \frac{\partial L}{\partial q_a} 
\end{cases}$

Usando Hamiltoniana  

$\begin{cases}
\dot{q}_a = \frac{\partial H}{\partial p_a} \\
\dot{p}_a = - \frac{\partial H}{\partial q_a} \\ 
\end{cases}$

rapporto:  
$H(q,p,t) = p\dot{q} -L(q, \dot{q}, t)$  

**Note per ricordare**

la prima di ogni funzione serve per passare da L a H  
quindi deriva la funzione che ho per la var  che appare solo lì, ottengo la var che appare solo nell'atra funzione  

poi entrambe creano $\dot{p}$ derivando per q, la var che hanno entrambe  
attena, con la H ci va il -  

**Cosa fa H**

1. $H=E$, cioè è l'energia meccanica del sistema, se:  
T quadratica omogenea nelle velocità  
U indipendente da dalle velocità  

2. H si conserva se:  
$\frac{\partial L}{\partial t}=0$  
perchè c'è un teorema secondo cui $\frac{dH}{dt}= -\frac{\partial L}{\partial t}$

### Scegliere L

data una L, posso sostituirla con una qualunque L' che differisca da questa per una derivata totale rispetto al tempo di una trasformazione continua, funzione solo di coordinate e tempo  

$L \rightarrow L'=L+\frac{dF(q,t)}{dt}$  

è quello che succede quando faccio una trasformazione di Galileo per andare ad un altro sistema inerziale, che si muove a velocità V rispetto a quello di prima  

### Trasformazioni

**Nomi**

variabile ciclica: che non appare direttamente in L, cioè L è invariante rispetto a q    
L invariante rispetto a q vuol dire: $\frac{\partial L}{\partial q}=0$  

se $q_a$ è ciclica, $p_a$ è detto integrale del moto, o costante del moto 

**Scrivere una trasformazione**

in generale, si scrivono:  
$x \rightarrow x + \epsilon \, A$  
vedi a cosa corrisponde A ogni volta  

- traslazione in direzione $\hat{n}$  
$x \rightarrow x+\epsilon \, \hat{n}$  

- rotazione attorno ad asse $\hat{n}$  
$x \rightarrow x +\epsilon \, (\hat{n} \land x)$  
qui $\epsilon = \delta \theta$  


**Attiva vs passiva**

sono due modi diversi di vedere la stessa trasformazione  

- attiva: sposto il sistema meccanico nella vita vera  
eg. traslo in avanti la posizione iniziale di una pallina  

- passiva: modifico il sistema di coord scelto  
eg. traslo indietro il sistema di riferimento, così la pallina sembra più avanti  

sono solo modi di vederla  
di solito, pensiamo a quelle attivo  

**Trasformazione**

insieme di tutte le trasformazioni operabili su un sistema fisico ha la struttura di un gruppo:  
chiuso rispetto all'operazione di composizione: $g_1 \; o ; g_2=g_3$  
operazione ha proprietà associativa  
esiste inverso  
esiste identità  

le trasformazioni simmetriche sono un sottogruppo, il 'gruppo di simmetria del sistema'

**Tipi di gruppi**

gruppi continui: trasformazioni sono parametrizzabili con parametri continui  
cioè contiene trasformazioni continue, che posso fare piccole quanto voglio  
eg. traslazione, rotazione  

gruppi discreti: eg. riflessione, traslazione discreta  

gruppi di Lie: dipendenza continua e differenziale dai parametri  
cioè posso fare le derivate parziali di tutti i parametri  

**Matrici ortogonali**

$O(n)$: gruppo delle matrici ortogonali nxn    
$SO(n)$: gruppo ortogonale delle rotazioni nxn (cioè hanno det=+1)  

$x_i \rightarrow x_i' = R_{ij}x_j$  
dove $R^tR=I$  
per definizione, R preserva il prodotto scalare, quindi la norma dei vettori  

$R^tR=I$ non è lineare  
ma per trasformazioni infinitesime vicino all'identità posso approssimare con:  
$R=I+\epsilon \Omega$  
$\epsilon$: parametro infinitesimo  
$\Omega$: generatore infinitesimo del gruppo -> l'insieme dei generatori infinitesimi è una base per uno spazio vettoriale, che è oiù facile da stuiare del gruppo

**Forma lineare e generatore**

nella formula non lineare pluggo l'approssimazione vicino all'identità, poi trascuro gli infinitesimi di ordine superiore ($\epsilon^2$):  
$(I+\epsilon \Omega)^t (I+\epsilon \Omega)=I$  
$\Omega + \Omega^t = 0$  
cioè il generatore $\Omega$ è una matrice antisimmetrica  
ad esempio, una base è:  
$\Omega = \begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}$  

**Da forma lineare a completa**

la forma lineare mostra come cosa succede se faccio una trsformazione infinitesima: $I+\delta \theta \, \Omega$  
quindi se metto insieme tante trasformazioni infinitesime, ho una trasformazione più grande  
in pratica, dall'approssimazione di taylor di una funzione, voglio ricavare quale era la funzione completa  

una rotazione di angolo $\theta$ è fatta da n rotazioni di angolo $\delta \theta  = \frac{\theta}{n}$, che posso rendere infinitesime:  
$\lim _{n \rightarrow \infty} (I+ \frac{\theta}{n} \, \Omega)^n = e^{\theta \Omega}$   

lo approssimo con Taylor, notando che la matrice del generatore ha: $\Omega ^2 = -1, \Omega^3=-\Omega$  
quindi viene:  
$= (1+\frac{\theta^2}{2}...)I + (\theta - \frac{\theta^3}{3!}...)\Omega = \cos \theta \, I + \sin \theta \, \Omega = \begin{bmatrix} \cos \theta & - \sin \theta \\ \sin \theta & \cos \theta \end{bmatrix}$

SO(n) è la parte di S(n) connessa con l'identità  
perchè le rotazioni sono ottenibile con una trasformazione continua a partire dall'identità, le riflessioni no, visto che deve cambiare il determinante a -1 


**In 3D**

la mia base è composta da 3 matrici: $\Omega_1, \Omega_2, \Omega_3$  
rappresentano le rotazioni attorno agli assi x,y,z  

le posso indicare con:  
$(\Omega_i)_{jk}=-\epsilon_{ijk}$  
i: quale matrice sto considerando  
jk: quale posto della matrice  

esempio:  
$\Omega_3 = \begin{bmatrix} 0 & 1 & 0 \\
-1 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix}$  
infatti: $(\Omega_{3})_{12}= -\epsilon_{213} = - (-1) = 1$  

**Relazione di commutazione**

$[\Omega_i, \Omega_j] = \epsilon_{ijk} \Omega_k$  

come si scrive:  
$[\Omega_i, \Omega_j] = \Omega_i \Omega_j - \Omega_j \Omega_i$  
dove in realtà si somma sull'indice ripetuto p:  
$[\Omega_i, \Omega_j]_{mn} = (\Omega_i)_{mp} (\Omega_j)_{pn} - (\Omega_j )_{mp}(\Omega_i)_{pn}$  
sostituendo la definizione: $(\Omega_i)_{jk}=-\epsilon_{jk}$  
poi uso el proprietà del simbolo di Levi-Civita  



**Teorema di Eulero**

un qualsiasi moto rigido con un punto fisso è una rotazione intorno ad un qualche asse passante per quel punto  

cioè:  
Se prendi un oggetto rigido (ad esempio un libro) e lo muovi nello spazio senza traslarlo, ma solo ruotandolo  
allora esiste sempre un asse (una retta passante per quel punto fisso) intorno al quale quella rotazione è avvenuta.

perchè:  
Tutte le rotazioni in 3D si possono scrivere come:  
$R=e^{θ Ω}$  
dove Ω è una matrice antisimmetrica (il generatore infinitesimo della rotazione).  
il vettore $\hat{n}$ tale che $\Omega \theta{n}=0$ è l'asse di rotazione  


**Moto rigido**

Un moto rigido con un punto fisso può essere rappresentato da una matrice di rotazione R del gruppo SO(3)   

perchè:  
$x'=Mx$  
per mantenere la distanza uguale (condizione di un corpo rigido)  
$|x'|^2=|x^2|$  
$|x'|^2 = (Mx)\cdot (Mx) = x^rM^t Mx$ 
$ |x^2 | = x^t x$  
quindi $M^tM=I$, cioè M è una matrice ortogonale  
voglio solo i moti rigidi continui, quindi $det =+1$  

**In rotazione con punto fisso**

il moto si scrive:  
$x(t+dt)= R(d\theta)x(t)$  
$x+dx = x+ d\theta(n \land x)$  
cioè fa uno spostamento tangenziale a se stesso, per mantenere la distanza con l'origine (il punto fisso) costante

velocità istantanee:  
$v = \frac{dx}{dt} = \frac{1}{dt} d\theta \hat{n} \land x = \omega \land x$  


**Per calcolare**

identità:  
$|a \land b |^2 =  |a^2||b^2| - (a \cdot b)^2$

espandere in componenti:  
$|x|^2 = (x \cdot x)= x_ix_i = \delta_{ij}x_ix_j$    
$(\omega \cdot x) = \omega_i x_i$  
$(\omega \cdot x)^2 = (\omega_i x_i) (\omega_j x_j)$ (usa due indici diversi, perchè sono due indici muti)     


**Tensore**

tensore: generalizzazione di un vettore in più dimensioni  
matrice: tensore di ordine 2  

come scrivere un tensore di ordine 2 (matrice):  
$M_{ij}=x_ix_j$  
cioè moltiplicando du vettori 3x1 e 1x3 ottengo matrice 3x3  

rotazione applicata ad un vettore:  
$x_i'=R_{ij}x_j$  

rotazione applicata ad un tensore:  
$M'_{ij}=x'_ixì_j = (R_{ik}x_k)(R_{jl}x_l)=R_{ik}R_{jl}(x_kx_l) = R_{ik}R_{jl}M_{kl}$  

**Indici**

$C_{ij}=A_{ik}B_{kj}$

vuol dire, per prendere l'elemento nel posto ij della matrice C, devo moltiplicare la riga i di A e la colonna j di B, e sommare i prodotti

### Teorema di Eulero per funzioni omogenee

**Teorema**

ipotesi:  
f quadratica omogenenea  
di grado k  

th:  
$\Sigma x \; \frac{df}{dx}=kf$  

**Funzione omogenea di grado k**

$f(x_1, ... x_n)$ è quadratica di rgado k se:    
$f(\lambda x_1, ... \lambda x_n) = \lambda ^k f(x_1, ... x_n)$  

quadratica omogenea significa k=2  

**Funzione omogenea quadratica**

significa che ogni termine ha il prodotto di due velocità, anche diverse  
cioè posso scrivere:  
$T=\frac{1}{2}\Sigma A_{i,j} \; \dot{q}_i \; \dot{q}_j$  

A è una amtrice di cefficienti  
1/2 è per evitare di contare due volte la stessa coppia i,j e j,i

**T è una funzione omogenea?**

in coord cartesiane sì, è omogenea quadratica, perchè $T=\Sigma \frac{1}{2}m\dot{x}^2$  
ma quando passo alle coordinata generalizzate potrebbe non esserlo più

$\frac{d}{dt}x(q,t) = \Sigma \frac{\partial x}{\partial q} \dot{q} +\frac{\partial x}{\partial t} $   
se x non dipende esplicitamente dal tempo, cioè $\frac{\partial x}{\partial t} =0$, allora le $\dot{x}$ sono funzioni lineari delle $\dot{q}$, quindi T è quadratica omogenea anche nelle coord genralizzate  
ma se x ha una dipendenza esplicita dal tempo, allora x è una funzione affine di q, non lineare, quindi in T possono apparire delle costanti, non sarà omogenea

### Modi normali

**Espando intorno a punto di equilibrio stabile**

in un punto di equilibrio stabile:  
$V'(r)=0$  
$V''(r)>0$  

V è definito a meno di una costante  

quindi quando espando V attorno ad un punto di equillibrio stabile:  
$V(x+\epsilon)=V(x)+V'(x) \epsilon + \frac{1}{2}V''(x)\epsilon^2$  
V(x) è un costante, la tolgo ( in pratica cambio il punto dove V=0)  
la derivata prima è nulla 
quindi:  
$V(x+\epsilon)= \frac{1}{2}V''(x)\epsilon^2$  

**Ottenere equazione differenziale per piccole oscillazioni**

$L=\frac{1}{2}T_{ij}(q)\dot{q}_i \dot{q}_j-V_{ij}(q)$  
o in 1D, più semplice: $L=\frac{1}{2}T(q) \dot{q}^2-V(q)$  
ij in absso vuol dire che è una amtrice   

dove $\overline{q}_i$ è il punto di equilibrio  

piccole oscillazioni:

1- faccio cambio di coordinate, per avere lo 0 nel punto di equilibrio stabile:  
$\eta_i = q_i-\overline{q}_i$

2- espando V intorno al punto di equilibrio stabile  

ottengo:  
$L=\frac{1}{2}T_{ij}(\overline{q}) \dot{\eta}_i \dot{\eta}_j-V(\overline{q})$  
uso eulero-lagrange, ottengo un sistema di equazioni differenziali (una per ogni coord):      
$T_{ij}\ddot{\eta}_j = -V_{ij} \eta_j$  

**Modi normali di oscillazioni**

le soluzioni di un'equazione differenziale formano uno spazio vettoriale  
posso trovarne una base formata da soluzioni periodiche  

nota ch il moto non è per forza periodico  
in 1D limitato = periodico, ma in dimensioni maggiori no  
però posso avere una base di soluzioni periodiche, da combianre linearmente per avere una soluzione generale non periodica

#### Modi degeneri

**Cosa sono**

trova le frequenze di ogni autovettore:  
frequenza non degenere: è unica, cioè non ci sono altri modi con la stessa frequenza  
frequenza degenere: esiste più di un modo (indipendente) con la stessa frequenza  

modo degenere: se la sua frequenza è degenere  

**Permutazioni: proof che si sono 2 frequenze degeneri**

se ho come simmetria le permutazioni tra 3 oggetti, cioè $S_3$  

frequenze non possono essere tutte uguali (no tutte degeneri):    
sennò sistema sarebbe simmetrico come una sfera (avrebbe simmetria continua $O_3$,), mentre ce l'ha discreta  
cioè le matrici T e V sarebbero proporzionali all'identità, posso vedere a occhio che V non lo è  

frequenze non possono essere tutte diverse:  
Se tutte le frequenze fossero diverse → il gruppo 
$S_3$ agirebbe in modo diagonale → rappresentazione abeliana → impossibile perché $S_3$ non è abeliano.  
se frequenze tutte diverse, tutti i modi normali sono indipendenti, perchè autovettori di autovalori diversi sono indipendenti  
allora ognuno è mappato in sè, L sarebbe diagonale (L è la matrice della simmetria lienare, non la lagrangiana)  
una matrice diagonale commuta sempre   
ma non abeliano significa non commutativo, non è possibile    
cioè: gruppo di scambi non abeliano -> esistono modi degeneri

#### Teoria soluzioni equazioni moto

**Soluzione**

voglio una soluzione oscillante:  
$\eta_i(t) = Re(v_i e^{-i \omega t})$  
$v_i$ è un vettore di N coefficienti complessi, voglio solo la parte reale, quale complessa non ha senso fisico  
la parte reale è il 'modo normale di oscillazione'  

**Problema degli autovalori**

utilizzando l'ansatz (guess per la forma della soluzione):  
$\eta_i(t) = v_i e^{-i \omega t}$  
$\ddot{q} = (i\omega)^2 ve^{-i\omega t} = -\omega^2 ve^{-i\omega t}$  

quindi sostituendo, $T\ddot{q} =-Uq$ diventa:  
$-\omega ^2 Tv +Uv = 0$  
$(U-\omega^2T)v=0$

**Piccole oscillazioni in 2D**

moto non periodico: traiettoria è curva di Lissajous  
oscillatore armonico anisotropo: la forza in x e y è diversa, $k_x \neq k_y$  

ma posso trovare delle soluzioni particolari periodiche, che formano una base per lo spazio delle soluzioni  
detta 'base dei modi normali di oscillazione'

allora nella formula di H cerca il potenziale efficace, cioè: $V_{eff}(r)=V(r)+\frac{L^2}{2mr^2}$
in questo caso, $L=p_\theta$, perchè è conservato  

punti di equilibrio: $\partial_r V_{eff} = 0$  
è stabile se: $\partial_{rr} V_{eff} > 0$ (derivata seconda positiva)  

**Punto di equilibrio se ho solo 1 coord generalizzata**

quando scrivo la formula di Eulero-Lagrange, dopo aver derivato dovrei avere la legge del moto $m\ddot{q}=...$  
metti $\ddot{q}=0$, vedi che q serve per annullare dall'altro lato  
metodo solito per trovare soluzione particolare di eq differenziali  

**Studia variazione energia meccanica**

prima di iniziare, controlla se H si conserva e se è l'energia  
se H si conserva, ma non è l'energia, vedi a occhio quale è la differenza e calcola il rate di perdita dell'energia:  

$E=H+...$  
$\frac{dE}{dt} = \frac{dH}{dt} + \frac{d(...)}{dt} =  \frac{d(...)}{dt}$  

ricorda che H si conserva  
alla fine, falla la derivata della ..., mi dà il rate della perdita di energia
