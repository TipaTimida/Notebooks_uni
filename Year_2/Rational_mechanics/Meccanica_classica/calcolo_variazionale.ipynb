### Principio variazionale

**Problema variazionale**

problema di massimo/minimo su SPAZIO DI FUNZIONE  
trovare q(t) che minimizzi il funzionale S (qui, la traiettoria e l'azione)  

di solito, in analisi faccio un problema di massimo/minimo su spazio di variabili  
cioè devo scegliere il valore di una var che estremizza una funzione  
qui invece devo scegliere la funzione che ne estremizza un'altra

**Esempi di problemi variazionali**

- problema della brachiostomia:  
ho un campo gravitazionale e una guida di qualche profilo, parametrizzato y(x)  
oggetto deve andare da A a B, due punti fissi  
quale è la guida $y(x)$ che riduce il tempo per fare il tragitto?  
cioè voglio trovare $y(x)$, la parametrizzazione della curva che minimizza il tempo di caduta  
risposta: cicloide (tipo il moto della valvola della ruota della bici, che ruota mentre il cm della ruota trasla avanti)  
se non ci fosse campo grav, risposta sarebbe una retta  

- problema della catenaria  
che forma ha una catena attaccata ai punti A e B e lasciata appesa?  
devo trovare la sua parametrizzazione $y(x)$, è quella che minimizza l'energia potenziale  

nel primo, ho:   
$t=\int_{x_A}^{x_B} dx \; f(y(x), y'(x))$  
t è la cosa da minimizzare  
f di è la formula di ciò che voglio minimizzare  
y(x) è cio che voglio trovare  

### Problema meccanico

**Azione**

$S[q(t), t_A, t_B] = \int_{t_A}^{t_B}L(q, \dot{q}, t) \; dt$   

è l'azione associata ad una traiettoria  
in pratica l'integrale nel tempo della lagrangiana  
è un funzionale: in inpt un vettore (tutti i punti della traiettoria), in output uno scalare (il valore dell'azione, una specie di 'bontà' della traiettoria)    

**Modi di scrivere l'azione**

con lagrangiana:  
$S[q(t)] = \int_{t_A}^{t_B}L(q, \dot{q}, t) \; dt$   

con hamiltoniana:  
$S[q(t),p(t)]= \int_{t_A}^{t_B}p\dot{q} - H(q,p,t) \; dt$   

specificando in modo diverso gli estremi:  
$S[A,B,q(t),p(t)] = \int_{A}^{B}p\, dq \; - H(q,p,t) \, dt$   


**Principle of least action**

considera tutte le funzioni q(t) con estremi fissati ad A, B  
l'estremo tra queste funzioni è quella la cui azione è un estremo locale   
la traiettoria estremale è quella effettivamente percorsa  

non è per forza un minimo assoluto, potrebbe essere un minimo locale o un punto di sella  

**Cosa fare**

$q(t)=\overline{q}(t) +\epsilon \, h(t)$  
$h(t)$: controlla la direzione della variazione  
$\epsilon$: controlla l'ampiezza della variazione  
nota che in $t_A$ e $t_B$, h deve essere 0, perchè i punti estremi sono fissi  

voglio provare che $S[q(t)]-S[\overline{q}(t)] \ge 0$  

### Osservazioni

**Lemma fondalmentale del calcolo delle variazioni**

se $\int f(x) \, h(x) =0$  per ogni possibile h(x), allora f(x) deve esere identicamente nulla  

come usare:  
$\frac{dS}{d\epsilon} =  \int_{t_A}^{t_B} \frac{dL}{d \epsilon} \, dt = \int_{t_A}^{t_B} (\frac{\partial L}{\partial q}-\frac{d}{dt}\frac{\partial L}{\partial \dot{q}}) h(t) \, dt = 0$  
se deve essere vero per ogni h(t), la cosa tra parentesi deve essere nulla  

**Derivata funzionale**

genralizzazione della derivata direzionale, differenzia nella direzione di na funzione invece che di un vettore  
esempio: $\frac{\partial S}{\partial q(t)}$   

come usare:  
$\frac{\partial S}{\partial q(t)} = \frac{\partial L}{\partial q}-\frac{d}{dt}\frac{\partial L}{\partial \dot{q}}$  
altro modo di scrivere Euler-Lagrange  

**Come provare che qualcosa è la soluzione**

$q(t)=\overline{q}(t) +\epsilon \, h(t)$  
$h(t)$: controlla la direzione della variazione  
$\epsilon$: controlla l'ampiezza della variazione  
nota che in $t_A$ e $t_B$, h deve essere 0, perchè i punti estremi sono fissi  

voglio provare che $S[q(t)]-S[\overline{q}(t)] \ge 0$  

**Individua i minimi**

ho una massa m che deve muoversi lungo una guida circolare, dal punto A al punto B  
i minimi locali sono TUTTI i percorsi da A a B, in entrambe le direzioni, dopo aver fatto un qualunque numero di giri completi del cerchio  
quindi sono infiniti  
perchè sono minimi locali: perchè se aggiungo PICCOLE variazioni non posso trovare una strada più veloce, e un giro intero in meno non è una piccola variazione   
minimo assoluto: quello che fa la strada più breve  

altro esempio: a biliardo, mandare la palla dal punto A a B  
minimo assoluto: linea retta da A a B  
minimo locale: colpo di sponda per andare da A a B  
è un minimo perchè, se cambio la traiettoria DI POCO, l'azione aumenta perchè aumenta di molto il percorso che compie prima di arrivare finalmente in B, quindi localmente è un minimo  

**Condizioni al contorno**

condizioni al contorno non cambiano l’equazione di Eulero-Lagrange, ma determinano quali soluzioni sono ammissibili  
non confondere condizioni al contorno con condizioni iniziali  

- condizioni al contorno:  
per problema variazionale  
setta un valore in ogni punto A e B, di inizio e fine (la posizione per le condizioni di DIrichlet, il momento per quelle di Neumann)  

- condizioni iniziali:  
per problema di Cauchy  
setta posizione e momento solo all'inizio  

unicità della soluzione:  
se fisso $q$ (pos) iniziale e finale: soluzione non è unica  
potrei avere più $q(t)$ che sono minimi locali  
se fisso le condizioni iniziali invece ho un problema di Cauchy, che ha soluzione unica  

**Geodetica**

cos'è: su una superficie curva, è la curva con percorso minore  
su una sfera, sono gli archi di cerchio massimo  

per una particella libera che si muove su una superficie curva (o in uno spazio con metrica data), SENZA potenziale:  
Equivalenza tra principio d’azione e principio di minima lunghezza  
cioè la la strada con minima azione è anche la più corta  
poi il modo migliore di percorrerla è a velocità uniforme

**Meccanica quantistica e minima azione**

il principio di minima azione è il limite classico della meccanica quantistica  
limite classico: ℏ→0, cioè le masse m in gioco sono così grandi che la costante di Plank è trascurabilmente piccola  

esempio: ho osservato che  $q(t_a)=q_a$   
quale è la probabilità che $q(t_b)=q_B$?   
posso calcolarla con l'INTEGRALE SUI CAMMINI  
cioè considero tutti i cammini tra A e B  
ad ognuno associo una fase complessa, cioè un punto in unc erchio di raggio 1  
normalmente i punti sono equalmente distribuiti sul cerchio  
ma se c'è un punto di minimo, il gradiente lì è nullo, quindi per piccole variazioni le traiettorie sono sempre lì vicino  
si rompe la simmetria, la soluzione è non-0, mi dà una risposta  

$p_{A->B}  \propto |\Sigma e^{i\frac{S[q(t)]}{h}}|^2$  


Nel limite classico, i cammini non estremali si cancellano per interferenza distruttiva → rimane solo il cammino stazionario (classico).

### Teorema di Noether

**Carica di Noether**

$Q = \Sigma \frac{\partial L}{\partial \dot{q}} A = \Sigma p \,A$  

Q: quantità conservata  
A: trasformazione continua  

Q si conserva se il sistema per trasformazione A applicata alle cooed $q$  

**Forma completa**

se cambia anche il tempo:  

$Q=∑_i \frac{​\partial L}{\partial \dot q_i} \delta_i -L \tau$  

per le trasformazioni infinitesime:  
$\begin{cases}
t' = t+ \epsilon \, \tau(q,t) \\
q'_i(t') = q_i(t)+\epsilon \, \delta (q,t)
\end{cases}$

se il tempo non cambia, $\tau=0$  

**Esempio: traslazione spaziale**

ho sistema di particelle con coord: $q_{a,i}$
a: quale particella  
i: quale coord, x,y,z  

trasformazione: traslazione globale lungo l'asse i  
$q_{ai} \rightarrow q_{ai}+\epsilon$  
quindi $\delta_{ai}=1$  

applico Noether:  
$Q = \Sigma \frac{\partial L}{\partial \dot{q}_{ai}} A_{ai} = \Sigma \frac{\partial L}{\partial \dot{q}_{ai}}  = \Sigma p_{ai}$  
cioè Q, la cosa che si conserva, è la somma dei momenti di ogni particella in direzione i, cioè il momento totale in direzione i  

**Esempio: traslazione temporale**

ho lagrangiana che non dipende dal tempo: $\frac {\partial L}{\partial t}=0$  

trasformazione: $t \rightarrow t+ \epsilon$  
quindi, per le coord, con Taylor: $q_{ai}(t +\epsilon)=q_{ai}(t)+\epsilon \dot{q_{ai}}(t)$  
variazione infinitesima: $\delta q_{ai} = q'_{ai}(t)-q_{ai}(t) = \epsilon \dot q_{ai}$  
confronto con forma di Noether: $\delta q_{ai} = \epsilon A_{ai}$  
allora $A_{ai} = \dot q_{ai}$  
recap: $\tau=1, \delta_i = \dot{q_i}$

applico Noether:  
$Q = \Sigma \frac{\partial L}{\partial \dot{q}_{ai}} \dot q_{ai} -L = \Sigma p_i \dot q_i -L = H$  
quindi la cosa conservata è l'hamiltoniana, cioè l'energia  

**A cosa mi serve**  

ho trasformazione:  
$q \rightarrow q'=q+\epsilon A$  
dove $A(q)$ è una trasformazione CONTINUA (perchè dopo devo differenziarla)  

se L è invariante sotto tale trasformazione, cioè $\delta L=0$, allora Q è costante lungo la traiettoria  

**Proof**

1- scrivo la variazione di L (espando L al primo ordine):   
$\delta L = L(q', \dot{q}', t)- L(q,\dot{q},t) = \Sigma  \frac{\partial L}{\partial q}\delta q + \Sigma  \frac{\partial L}{\partial \dot{q}}\delta \dot{q} =$  

2- sostituisco $\delta q_a = \epsilon A_a$  
$ = \epsilon \Sigma (\frac{\partial L}{\partial q} A)+ \epsilon \Sigma (\frac{\partial L}{\partial \dot{q}} \dot{A}) =  \epsilon \Sigma (\dot{p} \; A)+ \epsilon \Sigma (p \; \dot{A}) =$

3- uso formula della derivata di un prodotto  
$ = \epsilon \Sigma (\frac{d}{dt} (p \; A)) = \epsilon \frac{d}{dt}Q $  

quindi, se $\delta L=0$, la mia ipotesi del torema, allora Q non cambia lungo la traiettoria, ha derivata nulla
