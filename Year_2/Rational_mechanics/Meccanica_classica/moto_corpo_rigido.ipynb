### Teoria

**Formula generale tensore di inerzia**

$I_{ij} = \Sigma m_a(|x_a|^2 \delta_{ij}-x_{ai}x_{aj})$  
questa è la formula per ogni elemento della mtrice  
sostituisci in i,j gli indici dell'elemento che voglio per calcolarne uno di preciso  

oppure $\Sigma_a m_a = \int \rho(x) \; d^3 x$

nota che $|x_a|^2 = x_a^2+y^2+z^2$ è un vettore in 3D  
mentre $x_{ai}$ è semplicemente $x_a$  

$(|x_a|^2 \delta_{ij}-x_{ai}x_{aj})$ significa la distanza dall'asse  
tipo, se  considero l'asse z, sarebbe: $|x_a|^2 -z_a^2 = x_a^2+y_a^2$

quindi per un asse principale:  
$I_i = \Sigma _a m_a (x_{aj}^2+x_{ak}^2)$  
cioè se volgio $I_x$, considero solo le coordinate y e z, che rappresentano la distanza del punto di massa m dall'asse x  


**Tensore di inerzia**

$I = \begin{bmatrix} I_{xx} & I_{xy} & I_{xz} \\
I_{xy} & I_{yy} & I_{yz} \\  
I_{xz} & I_{yz} & I_{zz} \\  
\end{bmatrix} =  \begin{bmatrix} I_1 & 0 & 0 \\
0 & I_2 & 0 \\  
0 & 0 & I_3 \\  
\end{bmatrix}  
$  

definisce le proprietà geometriche e meccaniche di un corpo rigido  

**Observations**

the inertia tensor is a matrix:  
- symmetric  
- real  
- defined positive  

therefore:  
- it's diagonalizable  
- has 3 positive real eigenvalues: $I_{xx}^P, I_{yy}^P, I_{zz}^P$  
which are the momentums of inertia referred to the principal axes  
- their eigenvectors are the principal axes themselves  

this is proof that every rigid body has at least 3 principal axes ortogonal to each other  

**Proprietà del tensore**

traccia:  
$tr(I)=2\Sigma _a m_a |x_a|^2$  
e ovviamente, $tr(I)=I_1+I_2+I_3$  

se la distribuzione è su un piano, allora:  
$I_1+I_2=I_3$  
se non è su un piano, diventa:  
$I_1+I_2= \Sigma _a m_a(x_{a1}^2+x_{a2}^2+2x_{a3}^2) = I_3 + 2 \Sigma_a m_a a_{a3}^2 \ge I_3$  
è ovvio, basta sotituire a sinistra per i due I: $I_i = \Sigma _a m_a (x_{aj}^2+x_{ak}^2)$  
dove c'è uguaglianza se tutit i punti hanno $x_{a3}=0$, cioè tutta la massa è concentrata su un piano  

**Come trovare un asse principale**

in 3D:  
la distribuzione di massa ha una simmetria per riflessione rispetto ad un piano: il vettore normale al piano di simmetria è un asse principale   
e gli altri due si trovano su quel piano  

in 2D:  
la distribuzione di massa ha una simmetria per riflessione rispetto ad un un asse: quell'asse è un asse principale  

**Un momento nullo**

se ad esempio $I_3$ è nullo:  
$I_3= \Sigma _a m_a (x_{a1}^2+x_{a2}^2)=0$  
cioè  per ogni punto, $x_{a1}=x_{a2}=0$  
cioè tutta la massa è concentrata su una linea, l'asse $x_3$  

**Riassunto**

corpo 3D: $I_1 +I_2 \ge I_3$  
corpo planare: $I_1+I_2=I_3$  
corpo lineare: un momento è 0, gli altri opposti    
corpo puntiforme: tutti i momenti sono 0  

### Forme della matrice del tensore

**Come studiarla**

1. la diagonalizzo  
concretamente, significa applico una rotazione al sistema di assi, finchè ne ottengo uno composto da 3 assi principali  

2. studio gli autovalori (termini sulla diagonale)

**Tipi**

trottola asimmetrica: tutti gli I diversi  
trottola simmetrica: due I uguali  
trottola sferica: tutti gli I uguali  

**Tutti autovalori uguali**

cioè $I=I_1 \; Id$  
qualunque scelta di assi è uguale, tutte le terne hanno 3 assi principali  

perchè:  
$I'=RIR^t = I_1(R\; Id \; R^t) = I_1 \; Id$  

**Trottola simmetrica**

$I = \begin{bmatrix} I_1 & 0 & 0 \\
0 & I_2 & 0 \\  
0 & 0 & I_2 \\  
\end{bmatrix}$  

una coppia di autovalori uguali, uno diverso  
cioè un asse è diverso, gli altri due interscambiabili  

### Esempi

#### Guscio sferico

**Trova assi principali**

c'è simemtria per riflessione rispetto ad ogni piano equatoriale (passante per il centro)  
quindi qualunque asse è principale (perchè ogni asse è il vettore normale ad un piano)  
quindi qualunque terna di assi ha tutti assi principali

**Scrivi tensore**

$I = I_1 \; Id$  

quindi posso calcolare il momento di inerzia rispetto ad un'asse qualunque (visto che sono tutte equivalenti)  

$tr(I)=3I_1$ (dalla matrice)  
$tr(I)=2\Sigma _a m_a |x_a|^2 = 2MR^2$    
$I_1 = \frac{2}{3}MR^2$  

**Sfera piena**

come nel guscio, tutti gli assi passanti per il centro sono assi principali  
ma calcolare I è difficile, dovrei fare un integrale  

#### Cubo omogeneo

**Trova assi principali**

considero i 3 piani, ognuno parallelo ad una coppia di facce  
sono piani di simemtria, quindi i vettori normali ad essi sono assi principali  
sono i vettori perpendicolari alle facce del cubo  

**Scrivi tensore**

a occhio, i tre assi sono uguali, nessuno è più importante o diverso  
quindi, come nel guscio e nella sfera:      
$I=I_1 \; Id$  

conseguenze:  
1- nella rotazione, un cubo si comporta come una sfera  
2- tutte le terne sono fatte di assi principali, anche quelle coi vettori non ortognali ai piani di simemtria, e che quindi a occhio non lo sembrano  

#### Anello omogeneo

**Trova assi principali**

è una distribuzione planare  
ovviamente il piano su cui c'è la distribuzione è un piano di simmetria, quindi l'asse ortogonale è principale  
per fare una base ortogonale, mi mancano 2 vettori da scegliere sul piano della distribuzione  
ogni diametro è un asse di simemtria del piano con la distribuzion, quindi ogni diametro è un asse principale   

**Tensore**

trottola simmtrica  
perchè l'asse perpendicolare al piano è di un tipo, i due sul piano sono uguali, interscambiabili  

**Calcola**

uso la traccia:  
$tr(I)=I_1+2I_2$  
$tr(I)=2\Sigma _a m_a |x_a|^2 = 2MR^2$   

ne calcolo uno:  
$I_1 = \Sigma m_a|x_a|^2 = MR^2$  

da cui ottengo:  
$I_2 = \frac{1}{2}MR^2$  

#### Poligono regolare inscritto

**Assi principali**

è una distribuzione piana  
l'asse normale al piano è principale  

sul piano, scelgo un asse di simmetria (diametro passante per un vertice), è un asse principale  
per esclusione, l'asse perpendicoalre a questi due deve esserlo pure???  

scelgo due assi di simmetria sul piano  
sono vettori indipendenti tra loro  
ovviamente, per simemtria del disegno, nessuno è più importante dell'altro, quindi hanno stesos autovalore  
cioè posso scrivere il piano come combinazione di due autovettori con stesso autovalore, quindi il piano è un autospazion di quell'autovalore  
cioè tutti i vettori del piano hanno quell'autovalore, perchè sono combinazione di quei due  
$A(a_1v_1+a_2v_2)=\lambda (a_1v_1+a_2v_2)$  
quindi anche questa è una trottola simemtrica, come l'anello  
durante le rotazioni, assume una simmetria cilindrica  
anche se normalmente ha simmetria solo per rotazioni discrete (dello stesso l'angolo dell'angolo interno)  



### Moto roto-traslatorio

**Teorema di Koenig**

ogni moto può essere scomposto in:  
traslazione del cm  
rotazione del corpo attorno al cm  

$\vec{l}_i = M(\vec{x}_{cm} \land \vec{v}_{cm})_i + I_{ij}^{cm} \omega_j$  
primo addendo: corpo è punto rigido che ruota attorno al polo O, attorno a cui voglio sapere l    
secondo addendo: corpo ruota attorno al suo cm  

$T=\frac{1}{2}M|\vec{v}_{cm}|^2 +\frac{1}{2}I_{ij}^{cm}\omega_i \omega_j$  
primo addendo: corpo è punto concentrato nel cm che trasla solo, energia data dalla sua traslazione   
secondo addendo: energia data dalla rotazione del corpo attorno al suo cm  

**Cambiare polo**

$I_{ij}^{O}=I_{ij}^{cm} +M(|x_{cm}|^2 \delta_{ij}-x_{cm,i}x_{cm,j})$  

che vuol dire:  
il tensore di inerzia nel nuovo sistema è quello nel centro di massa più quello che si avrebbe mettendo un punto materiale di massa M nella posiziondel centro di massa  

è il caso generale del th di Huygens-Steiner, che si ottiene considerando gli elementi diagonali, cioè i=j  

#### Precessione in trottola simmetrica

conservazione del momento angolare:    
$\dot{l}_i = \dot{I}_{ij} \omega_j + I_{ij} \dot{\omega}_j = 0$  

$\dot{I}_{ij}$ è il cambiamento di I nel tempo  
perchè l'oggetto ruota, mentre il sistema di assi resta fermo, quindi anche se le proprietà dell'oggetto non cambiano, i valori di I rispetto a quegli assi sì  
tipo, se prima un'asse era nella direzione di un asse principale, dopo che l'oggetto ha ruotato potrebbe non esserlo più    


come scrivere queste derivate:  
$\dot{I}_{ji} \omega_j = (\omega \land l)_i = \epsilon_{ijk} \omega_j I_{kl} \omega_{l}$  
perchè? che vuol dire?

$I\dot{\omega} = -(\omega \land l)$  
$\dot{\omega}_i = I_{ij}^{-1} (l \land \omega)_j$  


**Solo per trottola simmetrica**

$\omega = \omega_\perp + \omega _\parallel$  
scompongo in rotazione perpendicoalre all'asse e attorno all'asse di simmetria  

limitato al piano ortogonale all'asse di simmetria, ho $I = I_\perp \, Id$  
quindi:  
$\dot{\omega}_i = \frac{1}{I_\perp} \delta_{ij} (l \land \omega)_j$  
$\dot{\omega} = \frac{l}{I_\perp} \land \omega$   
l’equazione differenziale per un vettore che ruoti con velocità angolare $\Omega$  

dove $\Omega = \frac{l}{I_\perp}$ è la velocità angolare di precessione

Il moto della trottola è quindi dato da una
rotazione istantanea intorno all’asse individuato da ω , con il modulo che si mantiene costante
nel tempo, mentre l’asse precede ruotando intorno al vettore momento angolare (che rimane fisso)
con velocità di precessione Ω.  

Tale precessione corrisponde proprio alle piccole oscillazioni che osserviamo quando lanciamo
un piatto o oggetti simili con velocità di rotazione iniziale non esattamente lungo l’asse principale.
