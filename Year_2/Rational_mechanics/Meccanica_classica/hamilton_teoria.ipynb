#### DOF, vincoli

**Gradi di libertà**

N: variabili lagrangiane, $q_i$ (DOF)  
2N: coordinate e momenti generalizzati    

perchè:  
le equazioni del moto sono differenziali al secondo ordine  
quindi specificare il moto devo assgnare 2N condizioni iniziali indipendenti, le $q$ e $\dot{q}$ al tempo iniziale (oppure p e q)   

nota che c'è una corrispondenza biunivoca tra p e $\dot{q}$, momento e velocità  

**Vincolo nello spazio delle fasi**

cos'è:  
è una quantità conservata (cioè indipendente dal tempo, $\frac{dQ}{dt}=0$)  

cosa fa:  
riduce di 1 la dimensione della superficie su cui è confinata la dinamica (cioè la dimensione della varietà accessibile allo stato del sistema)  
NON chiamarla DOF, che è la quantità di coord generalizzate necessarie, qualla iniziale prima di sottrarre i vincoli  

quanti:  
al massimo: 2N-1 vincoli  
per essere contato come vincolo, la quantità conservata deve essere indipendente a quelle già considerate prima, cioè:  $\{Q_i, Q_j\}$  

**Sistema integrabile**

def: posso trovare esplicitamente le soluzioni del moto usando integrali   
cioè esiste una scelta di N coordinate in cui posso DISACCOPPIARE le equazioni del moto, avere tutti moti unidimensionali indipendenti  

in pratica, integrale = disaccoppiabile  

esempi:  
oscillatore anisotropo: sì, posso disaccoppiare $H_x$ e $H_y$   
orscillatore in regime di piccole oscillazioni: sì, posso dissaccopiare usando i modi normali  
pendolo doppio: no, non disaccoppiabile  


**Th di Liouville-Arnold**

sistema integrabile <-> esistono almeno N costanti del moto (indipendenti)  

se ne trovo di più, si chiama super-accoppiabile  
cioè ha anche simmetrie aggiuntive                           

**Dimensioni della dinamica la dinamica**

N: DOF  
2N: coordinate e momenti generalizzati    
v: numero di vincoli  
2N-v: dimensione della superficie su cui è confinata la dinamica  

valori di 2N-v:

**Esempio: vincoli**

#### Struttura simplettica  

**Struttura simplettica**

La struttura simplettica è la legge geometrica che determina come funzioni e variabili dinamiche “si accoppiano” nello spazio delle fasi  

Γ, la matrice simplettica standard, è la sua rappresentazione in coordinate canoniche: $\{z_i, z_j\}=\Gamma_{ij}$    

cioè $\Gamma$ implementa la struttura simplettica dello spazio delle fasi, cioè la forma canonica della parentesi di Poisson

**Riscrivere coords con matrice simplettica**

invece di avere due vettori, $q=(q_1, ... q_N)$ e $p=(p_1, ., p_N)$, ne uso uno solo:  
$z=(q_1, ... q_N, p_i, ... p_N)$  

intuitivamente, la matrice simplettica serve per passare da q a p  
se ho $\Gamma_{ij}$, allora $z_i$ e $z_j$ sono rispettivamente $q_i$ e $p_i$  
perchè: $\Gamma = \begin{bmatrix} 0 & I \\ -I & 0 \end{bmatrix}$  

così posso scrivere le cose più facilmente:  
equzioni di hamilton: $\dot{z}_i=\Gamma_{ij} \frac{\partial H(z,t)}{\partial z_j}$  
parentesi di Poisson: $\{f,g\}=\Gamma _{ij}  \frac{\partial f}{\partial z_i} \frac{\partial g}{\partial z_j}$

### Teorema di Liouville

**Teorema di Liouville**

diversi modi per dirlo:  
1- la densità lungo le traiettorie rimane costante (quello della dimostrazione)  
2- l’evoluzione temporale (la mappa) conserva i volumi dello spazio delle fasi    
3- campo hamiltoniano ha div=0  

validità: ogni evoluzione temporale hamiltoniana (anche campi non stazionari)  

**Analogia con fluidodinamica**

campo di velocità hamiltoniano è equiparabile a a quello di un fluido  
così $\rho(q,p,t)$ sarebbe:  
fluidodinamica: densità del fluido  
hamilton: densità dei possibili sistemi nello spazio delle fasi (cioè numero dei sustemi nel volumetto (dq, dp) al tempo t)   

$\vec{j}=\rho\dot{z}$: densità di corrente, densità per velocità del fluido  

**Proof: th Liuoville**

ipotesi:  
1- il moto è hamiltoniano (spazio delle fasi evolve secondo le equazioni di Hamilton)  
2- Nei sistemi hamiltoniani le traiettorie non si interrompono e non si creano o distruggono sistemi  

procedimento:  

1- uso l'equazione di continuità (come per la corrente in elettrostatica):    
$\frac{\partial \rho}{\partial t} +\frac{\partial (\rho\dot{z})}{\partial z_i} = 0$  

2- sviluppo la divergenza:  
$\frac{\partial \rho}{\partial t} +\frac{\partial}{\partial z_i} (\rho\dot{z}) = \frac{\partial \rho}{\partial t} +\dot{z} \frac{\partial \rho}{\partial z_i}+ \rho \frac{\partial \dot{z}}{\partial z_i} = \frac{\partial \rho}{\partial t} +\dot{z} \frac{\partial \rho}{\partial z_i} = \frac{d\rho}{dt}=0$  

ho cancellato un addendo perchè:  
$\frac{\partial \dot{z}_i}{\partial z_i} = \Gamma_{ij} \frac{\partial^2 H}{\partial_i \partial _j}$  
ma $\Gamma$ è antisimmetrica, le derivate seconde di H simemtriche, e simmetrico per antisimmetrico = 0  

$\frac{d\rho}{dt}$ è ciò che volevo dimostrare  

**Scriverlo con la divergenza**

la divergenza:  
si applica ad un campo vettoriale, tipo quella di velocità    
serve a fare la derivata rispetto a tutte le componenti di un vettore, poi a somamrle  
$\frac{\partial \dot{z}_i}{\partial z_i} = \nabla_z \cdot \dot{z}$

equazione di continuità:  
$\frac{\partial \rho}{\partial t} +\frac{\partial (\rho\dot{z})}{\partial z_i} = 0$  
$\frac{\partial \rho}{\partial t} +\nabla _z  \cdot (\rho\dot{z}) = 0$  

regola di Leibniz:  
$\frac{\partial}{\partial z_i} (\rho\dot{z}) = \dot{z} \frac{\partial \rho}{\partial z_i}+ \rho \frac{\partial \dot{z}}{\partial z_i}$  $\nabla_z \cdot (\rho \dot{z}) = (\nabla _z \rho) \cdot \dot{z} + \rho (\nabla _x \cdot \dot{z})$    
$\dot{z}$: campo vettoriale, è sempre preceduto dal prodottos calare, se con nabla fa la divergenza  
$\rho$: campo scalare, può fare il gradiente o essere un coefficiente  

$\nabla \cdot v=0$: è la condizione di divergenza nulla  


**Note**

differenza:  
$\frac{\partial \rho}{\partial t}$: come cambia la densità in un punto fisso dello spazio delle fasi   
$\frac{d\rho}{dt}$: come cambia la densità vista da un punto che si muove insieme al fluido  


la divergenza è una proprietà del campo di velocità, non del flusso  
divergenza nulla significa consevazione del volume, come in liquido incomprimibile


### Flusso del pendolo

**Moto limitato vs periodico**

moto limitato: ristretto ad una regione di spazio  

in 1D: 
moto limitato -> periodico  

in 2D (o più):    
non necessariamente, può essere limitato ma non periodico  
eg. oscillatore armonico anisotropo con le frequenze in rapporto irrazionale  
anisotropo: $\omega_x \neq \omega_y$, cioè $k_x \neq k_y$, nelle due direzioni ha forze diverse  
se $\frac{\omega_x}{\omega_y}$ sono in rapporto irrazionale, non c'è alcun periodo comune, le oscillazioni non si sincronizzano mai  

![image.png](https://www.researchgate.net/publication/225622678/figure/fig5/AS:668281627234319@1536342164986/The-phase-space-of-the-simple-pendulum.png)

**Linee di flusso nel pendolo**

attorno a punto di equilibrio stabile: ellissi (perchè per piccoli angoli è un oscillatore armonico, che ellissi come linee di flusso)  
se aumento energia: le ellissi si deformano, finchè ho abbastanza E per fare il giro della morte  

**Intersezioni linee**

le linee nello spazio delle fasi non possono mai intersecarsi  
in alcuni punti, ad esempio in quello del pendolo semplice, sembra che si incrocino (intersezione linee rosse)  
ma in realtà quelli sono punti di equilibrio instabile, non appartengono a nessuna traiettoria  

cioè:  
i punti sono $(q=\pi, p=0)$, cioè il pendolo è fermo e in verticale verso l'alto  
se poggio lì il pendolo fermo ci rimane, ma durante il movimento non si potrà mai fermare lì  
servirebbe un tempo infinito perchè la velocità si annullasse proprio in quel punto, senza farlo ricadere  
quindi le traiettorie tendono a quel punto asintoticamente, ma mai in tempo finito    


**Tipi di moto: spazio delle fasi del pendolo semplice**

- moto librativo: i cerchi chiusi all'interno della linea rossa  
il pendolo non compie una rotazione completa (giro della morte)  
traiettoria limitata  

- moto di circolazione: le curve fuori dalla linea rossa  
il pendolo continua a fare giri della morte  

- moto di separazione: la linea rossa  
separa i due tipi  

**Condizioni sui tipi di moto**  

moto librativo: non fa giro della morte  
quindi ci deve essere un punto di inversione della velocità e della direzione del moto  
cioè un punto dove $T=0$, perchè la velocità deve annullarsi pe rpotersi invertire  

1- scrivendo T, vedo che è funzione quadratica della velocità  
quindi H=T+U=E  

2- ovviamente, $T \ge 0$  

3- calcolo che valori può avere U:  
$U(\theta) = mgl\cos \theta$  
$-mgl \le U \le mgl$  

4- vedo quando T si può annullare:  
$E-U=T\ge 0$  
E si annulla per qualche angolo se $E<mgl$ -> moto librativo  
E non si annulla per nessun angolo se $E>mgl$ -> moto di circolazione  
$E=mgl$: moto di separazione  
notqa che mgl è il massimo valore di U  
se E è maggiore del massimo valore di U, T non potrà mai annullarsi, deve essere positiva per avere E=T+U

### Trasformata di Legendre

**Cosa è**

per cambiare punto di vista su una funzione  
passo da una descrizione in termini di una variabile ad una in termini della sua derivata (pendenza)  

cioè se non ho a dispozione una variabile, ma solo la sua derivata, quindi voglio riscrivere la funzione in termini di quella  

$g(y)=sup_x (xy-f(x))$  

nota che x è in realtà $x(y)$, cioè il valore di x che corrisponde alla y ricevuta in input, cioè tale che $y=f'(x)$  
quindi anche f(x) è in realtà f(x(y))  
quindi g(y) è solo in funzione di y, non di x  

**Esempio di calcolo**

$f(x)=\frac{1}{3}x^3$  

derivo: $y=f'(x)=x^2$  
trovo l'inversa: $x= \sqrt{y}$  
la pluggo nella formula: $g(y) = xy-f(x) = \sqrt{y} y - f(\sqrt{y}) = y^{3/2}-\frac{1}{3}y^{3/2} = \frac{2}{3}y^{3/2}$

recap:  
$f(x)=\frac{1}{3}x^3$  
$g(y)=  \frac{2}{3}y^{3/2}$

**Nota**

f e g sono funzioni diverse, danno valori diversi   

esempio: se prendo x=2, e quindi y=f'(2)= 4, ottengo:  
f(x=2)=2.6  
g(y=4)=5.3

**Usarla per cambiare variabili**

ho: $f(x,y,z,t)$  
conosco alcune derivate: $u = \frac{\partial f}{\partial x}, \quad v = \frac{\partial f}{\partial y}$  
voglio sostituire x,y con u,v, cioè la cosa per cui derivo con il valore della derivata    

1- devo sommare adddendi con le coppie da scambiare  
2- sottraggo l'oiginale  
3- in x,y metto le loro espressioni in termini di u,v,z,t (le funzioni inverse)

$g(u,v,z,t)=ux + vy -f(x,y,z,t)$  

**Esempio di applicazione**

$F=U-TS$  
F: energia libera  
U: energia totale  
T: temperatura  
S: entropia  

so che $dS = \frac{dQ}{dT}$  
cioè F è la trasformata di Legendre di U, quando scambio T con S (al posto di T, ci metto la sua pendenza, S)  

**Usata con Hamilton**

ho: $L(q, \dot{q}, t)$   
so: $p = \frac{\partial L}{\partial \dot{q}}$  

quindi posso scambiare $\dot{q}$ con la sua derivata, $p$, e ottengo:  
$H(q,p,t) = p\dot{q} -L(q, \dot{q}, t)$  

nel primo addendo, devo sommare tutte le coppie di coordinate che sto cambiando, sarebbe $\Sigma_a p_a \dot{q}_a$

le nuove equazioni del moto sono:  
$\begin{cases}
\dot{p}=\frac{\partial L}{\partial q}= - \frac{\partial H}{\partial q} \\
\dot{q} = \frac{\partial H}{\partial p}
\end{cases}$

**Nota con Hamilton**

$p=\frac{\partial L}{\partial \dot{q}}$  
$\dot{q} = \frac{\partial H}{\partial p}$  

per scambiare p e $\dot{q}$, la variabile per cui derivo e il valore della derivata, devo scambiare la funzione con la sua trasformata di Legendre (H è la trasformata di Legendre di L)  
ovviamente, scambiarle due volte dà l'identità, quindi la trasformata è un'involuzione (f^2 = I)  

**Perchè serve**

se sostituisssi semplicemente le derivate che voglio, il differenziale non verrebbe bene  
perchè mentre faccio la derivata parziale di una variabile, nonmmi verrebbero fisse le altre  

### Parentesi di Poisson

**Parentesi di Poisson**

$\{f,g\} = \Sigma (\frac{\partial f}{\partial q}\frac{\partial g}{\partial p} - \frac{\partial f}{\partial p}\frac{\partial g}{\partial q})$  
in pratica dice se f e g sono correlate  

**Derivata totale rispetto al tempo**

è comodo usare le parentesi di Poissson per scriverle, invece della chain rule:  
$\frac{df(q,p,t)}{dt} = \frac{\partial f}{\partial q} \frac{\partial q}{\partial t} + \frac{\partial f}{\partial p}\frac{\partial p}{ \partial t} + \frac{\partial f}{\partial t} =  \frac{\partial f}{\partial q} \dot{p} - \frac{\partial f}{\partial p} \dot{q} + \frac{\partial f}{\partial t}$   
$\frac{df(q,p,t)}{dt} = \{f, H\} + \frac{\partial f}{\partial t}$  

così posso scrivere più facilmente la definizione di quantità costante  

**Relazioni canoniche di Poisson**

$\{q_a, q_b\}=\{p_a, p_b\}=0$  
perchè per particelle diverse sono indipendenti, e per particelle uguali è una sottrazioni di due addendi uguali  

$\{q_a, p_b\}=\delta_{ab}$   
cioè 1 solo se a=b, se si riferiscono alla stessa particella  
0 else 

**Proprietà**

1- bilineare e antisimmetrica  

2- identità di Jacobi: $\{A, \{B,C\}\} + \{B, \{C,A\}\} +  + \{C, \{A,B\}\} =0$ 

3- teorema di Poisson:  
se $\frac{d}{dt} A=0$ e $\frac{d}{dt} B=0$, allora $\frac{d}{dt} C=0$ dove $C=\{A,B\}$  
cioè se ho 2 quantità conservate A e B, posso trovarne una terza C  
nota che C potrebbe esere identicamente nulla, quindi potrebbe non servire a niente  

#### Commutatori

**Commutatori tra matrici**

A, B matrici  
commutatore: [A,B]=AB-BA  
se =0 commutano, sennò no  

in pratica, anche qui dice se A e B sono legate in qualche modo  

**Commutatori e parentesi di poisson**

il passaggio fra la meccanica classica e quella quantistica viene effettuato sostituendo la parentesi di Poisson fra quantità fisiche il commutatore dei corrispondenti operatori diviso per ih̄:  
$\{f,g\} \rightarrow \frac{1}{ih̄}[F,G]$  

matematicamente:  
la parentesi di Poisson realizza l’algebra di Lie delle trasformazioni canoniche nella meccanica classica;  
il commutatore di matrici realizza l’algebra di Lie degli operatori lineari (come in meccanica quantistica).  

**Algebra di Lie**  

spazio vettoriale con un’operazione bilineare [⋅,⋅] che soddisfa:  
antisimmetria  
identità di Jacobi
linearità  
regola di Leibniz  

che sono le proprietà condivise da parentesi di Poisson e commutatori   

Una algebra di Lie è il linguaggio matematico che descrive:  
generatori di trasformazioni  
simmetrie  
evoluzioni infinitesime  

quindi le trasformazioni canoniche formano un gruppo di simmetria, e la loro algebra infinitesima è la parentesi di Poisson.  

### Trasformazioni canoniche

**Trasformazione**

1- trasformazione dallo spazio delle fasi allo spazio delle fasi  
2- invertibile  

$q_i,p_i \rightarrow Q_i(q_i, p_i, t), \, P_i(q_i, p_i, t)$    
cioè ogni nuova Q e P può dipendere da tutte le q e p di prima  

**Trasformazione canonica**

per ogni $H(p,q,t)$, esiste $K(Q,P, t)$, tali che le equazioni di hamilton danno un sistema di equazioni del moto equivalenti  
in teoria, questa è def di geenralizzata, ma poi a me interesserà solo canonica in senso stretto  

è canonica in senso stretto? 3 criteri equivalenti:  
1- matrice jacobiana simplettica  
2- equivalenza delle parentesi di Poisson  
3- esistenza di una funzione generatrice  

**1. Simplettica**

in senso stretto (quello che interessa a me):  
jacobiana simplettica: se J rispetta  
$J \Gamma J^t = \Gamma$    

in sesnso generalizzato:  
$J \Gamma J^t = \lambda \Gamma$    
$\lambda$ può essere un qualunque valore  
se c'è dipendenza dal tempo, J deve essere simplettica ad ogni t, ma può esserlo per valori diversi di $\lambda$  

**2. Metodo della matrici canoniche**

è canonica se, prese due quantità $G_1, G_2$, le loro parentesi di Poisson rispetto alle vecchie coordinate e rispetto alle nuove coordinate coincidono:  
$\{G_1, G_2\}_{q,p} = \{G_1, G_2\}_{Q,P}$   

cioè, in modo più facile, se Q e P soddsfano le relazioni canoniche di Poisson, come lo facevano q e p  

### Funzione generatrice

#### Teoria

**Problema**

cosa voglio fare:  
voglio fare il passaggio di coordinate (q,p)->(Q,P)  
e voglio sapere come è la nuova hamiltoniana (o lagrangiana)  

a cosa devo stare attenta:  
il sistema di equazioni di hamilton deve dare una soluzione uguale a quella di prima, cioè produrre le stesse equazioni del moto  
cioè, visto che la soluzione è la traiettoria dove l'azione è un estremo, la nuova H deve mantenere la stessa traiettoria estremale

**Come si fa: introduco funzione generatrice**

due funzionali hanno stessa traiettoria estremale se:  
differiscono per un differenziale ESATTO dF  
dove F(q,p,t) è detta funzione generatrice  

cioè, fissati A e B, ogni traiettoria ha valori dell'azione diverse nei due funzionali, ma quella estrema è la stessa in entrambi  
cambia il valore dell'azione su di essa, ma non il fatto che sia estremale   

**Note**

F è indipendente da H  
deve essere valida per qualunque hamiltoniana, non solo per un H specifico  
deve dipendere solo dalla trasformazione delle coordinate, non dalla dinamica

inversione temporale e trasformazione di scala non sono simplettiche in senso stretto, le escludo  


**In formule**

passaggio di coordinate:  
$p \, dq -H(q,p,t) \, dt = P \, dQ - K(P,Q,t) \, dt + dF$  

riordinando:

#### H si conserva (Derivata di L e H)

**Raporto tra L e H**

$\frac{dH}{dt}= -\frac{\partial L}{\partial t}$   

significa che:  
se la lagrangiana è indipendente dal tempo (derivata parziale)  
allora l'Hamiltoniana è una costante del moto (derivata totale)  

legame fra simmetria temporale e conservazione dell’energia

**Proof**

1- scrivo la definizione di L  
2- trasformo le cose in momenti generalizzati, per scrivere più compatto  
5- uso Legendre per ottenere H   

$\frac{dL}{dt} = \Sigma \frac{\partial L}{\partial q} \dot{q} + \Sigma \frac{\partial L}{\partial \dot{q}} \ddot{q} + \frac{\partial L}{\partial t} =$    
$= \Sigma \dot{p}\dot{q} +\Sigma p\ddot{q} + \frac{\partial L}{\partial t} =$    
$ = \frac{d}{dt}(\Sigma p\dot{q}) +\frac{\partial L}{\partial t} = $   
$ = \frac{d}{dt}(\Sigma p\dot{q}-L+L) +\frac{\partial L}{\partial t} = $   
$ = \frac{d}{dt}(H+L) +\frac{\partial L}{\partial t}$ 

**Rapporto tra derivata parziale e totale**

SOLO lungo le soluzioni delle equazioni di hamilton:  
$\frac{\partial H}{\partial t}= \frac{dH}{dt}$   
cioè se H non dipende ESPLICITAMENTE dal tempo, allora è conservata  

**Proof**

metodo 1:  
$\frac{d H}{d t}= \dot{q} \frac{\partial H}{\partial q} + \dot{p} \frac{\partial H}{\partial p}+ \frac{\partial H}{\partial t} = -\dot{q} \dot{p}  +  \dot{p} \dot{q} +\frac{\partial H}{\partial t} = \frac{\partial H}{\partial t}$  
1- usa chain rule per scrivere differenziale di H  
2- quegli addendi sono 0, per le equazioni di hamilton

metodo 2: per sistema hamiltoniano, posso scrivere:  
$\frac{dH}{dt}= \{ H, H \} + \frac{\partial H}{\partial t}$  
dove la parentesi di poisson =0
