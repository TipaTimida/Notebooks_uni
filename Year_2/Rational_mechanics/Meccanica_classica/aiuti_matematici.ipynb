### Simboli

#### Delta di Kronecker

$\delta_{ij} = \begin{cases} 1 & se \; i=j  \\ 0 &else \end{cases}$  

**Uso**  

usata per scrivere il prodotto scalare:  
$a \cdot b = \delta_{ij} a_ib_j$  

significa: rendi uguali i due simboli

#### Simbolo di Levi-Civita / tensore completamente antisimmetrico

$\epsilon_{ijk} = \begin{cases} \pm 1 & se \; ijk \; permutazione \; di \; 123 \\
0 & else \end{cases}$  

+1 se leggendo da sinistra a destra ijk, ritrovo l'ordine 123, sennò -1  


**Uso**

usata per scrivere prodotto vettoriale:  
$(a \land b)_i = \epsilon _{ijk}a_jb_k$  


#### Delta di Dirac

**Definition**

cosa:  
funzione generalizzata (essenzialmente una distribuzione)

proprietà:  
1- è zero ovunque, tranne nel punto dove il suo argomento vale 0  
cioè se $\delta (\vec{r}-\vec{r}_0)$, è non-0 solo in $r_0$  
2- il suo integrale totale vale 1  
cioè, moltiplicare sotto integrale una funzione f(x) per $\delta(x-x_0)$ vuol dire calcolare il valore di f(x) nel punto $x_0$, dove c'è lo spike della delta di Dirac  

utilizzi:  
per esprimere cose puntiformi (eg. densità di una carica puntiforme)   

**Come si calcola**

$\int g(t) \; \delta(f(t)) \; dt = \frac{g(t_0)}{|f'(t_0)|}$  
where $t_0$ is the only zero of f, so $f(t_0)=0$  

other ways to write it, for 1D and 3D space:  
$\int f(r') \delta^{(3)}(r'-a) \, d^3r' = f(a)$  
$\int f(x) \delta(x-a) \, dx = f(a)$  

**Esempio: densità di carica puntiforme**

$\rho(\vec{r})=q \, \delta (\vec{r}-\vec{r}_0)$   
$\int _V \rho(\vec{r}) \, d^3r = \int _V q \, \delta (\vec{r}-\vec{r}_0) d^3r = \begin{cases} 
q  & se \, \vec{r}_0 \in V \\
0 & else \end{cases}$  

posso aggiugnere anche il tempo:  
normal: $\rho(r',t) d^3r' = \int q \delta^{(3)}(r-r_q(t))d^2r' = q$  

**Graficamante**

è un rettangolo di area 1 con base x e altezza y, dove dx -> 0  
così diventa un linea verticale infinita, ma sermpre di area 1  
nei segnali, sarebbe chiamato un impulso  

### Indici

**Leggere indici**

indice muto / ripetuto: sottintende una sommatoria  
scope locale   
si chiama 'convenzione di Einstein' il fatto per cui implica la sommatoria  

indice libero / non ripetuto: fa scegliere la componente del vettore  
scope globale  

**Scope**

locale: ogni volta che metto il segno di = potrei cambiargli nome  
cioè potrei fare $a \cdot b = a_ib_i = a_jb_j$   

globale: non posso cambiargli nome, sennò non corrisponde più con il nome della componente del risultato che ho già scritto in precedenza, prima dell'=  


**Esempio**

con indici: $v_i = M_{ij} w_j$  
senza indici: $\vec v = M \vec w$  

i: appare una volta da ogni lato, non ripetuto   
stabilisce componente del risultato, il vettore v  

j: appare due volte dallo stesso lato, ripetuto  
dice: somma implicita (fare una summatoria su tutte le componenti di w)    

se togliessi gli indici i e j, dovrei mettere i segni di vettore  

#### Usare indici per scrivere cose

**Matrici**

matrice: $v_i = M_{ij} w_j$

**Nabla**

$\nabla _i = \partial_i = \frac{\partial}{\partial x_i}$  

gradiente: $(\nabla f)_i = \partial_i f$  
dà in output un vettore, quindi devo precisare quale componente voglio  
qui i non è un indice ripetuto, è indice libero, appare anche a sinistra, serve a scegliere la componente, non a fare sommatoria

dà in output uno scalare, quindi non posso precisare la componente  
l'indice i ripetuto sottintende una somma: $\partial_iE_i = \frac{\partial E_1}{\partial x_1} + \frac{\partial E_2}{\partial x_2} + \frac{\partial E_3}{\partial x_3}$

rotore: $(\nabla \times A)_i =  \epsilon_{ijk} \nabla _j A_k = \epsilon_{ijk} \frac{\partial A_k}{\partial x_j}$  

nota: $\nabla_i = \partial _i$ solo in spazio euclideo cartesiano  
in realtività o in coordinate polari non vale, ci sono termini extra (connessioni)  
infatti le formule con nabla in coordinate polari sono diverse

**Prodotti**

scalare: $a \cdot b = a_ib_i$  
vettoriale: $(a \land b)_i = \epsilon _{ijk}a_jb_k$   

nota che prodotto scalare dà in output uno scalare, quindi non posso selezionare una componente  

**Pluggare cose**

so: $B_i = (\nabla \times A)_i = \epsilon_{ijk} \nabla_j A_k$  
voglio pluggarlo in: $(v \times B)_i = \epsilon_{ijk}v_j B_k$  

$(v \times B)_i = \epsilon_{ijk}v_j \, (\epsilon_{klm} \nabla_l A_m)$

### Derivare

#### Tipi di derivate

**Tipi**

1- derivata totale (derivata lungo il moto)
$\frac{d}{dt}L$  
come cambia L lungo il moto, quindi cambiano sia tempo che posizione  
evolution of L along the actual motion in time  

2- derivata parziale (rispetto al tempo)
$\frac{\partial }{\partial t}L$  
come cambia L nel tempo, mentre le coordinate spaziali restano fisse  
cioè cosa succede, in un punto fissato, allo scorrere il tempo   

3- variazione (variazione funzionale)  
$\delta L$  
come cambia L lungo la traiettoria, ma a tempo fermo  
quindi cambiano le coordinate spaziali, ma non quelle temporali  
linear response of L under a virtual change of the path  
virtuale vuol dire considera nulla la derivata col tempo, perchè se il tempo non passa il vincolo non si sposta     

**Esempio**

$\frac{\partial \rho}{\partial t}$: come cambia la densità in un punto fisso, mentre passa il tempo  

$\frac{d\rho}{dt}$: come cambia la densità vista da un punto che si muove insieme al fluido  

#### Modi di scrivere

**Come si scrive ogni tipo**

dalla definizione vedi se qualcosa non cambia, in quel caso quella derivata è nulla  
ricorda che la sommatoria la potrei togliere se ci sono gli indici ripetuti  

$\frac{d}{dt} L(q_a, \dot{q}_a, t)= \Sigma \frac{\partial L}{\partial q_a} \dot{q}_a + \Sigma \frac{\partial L}{\partial \dot{q}_a} \ddot{q}_a + \frac{\partial L}{\partial t}$  
uso la chain rule  

$\frac{\partial }{\partial t} L(q_a, \dot{q}_a, t)$  
non c'è un altro modo di scriverlo  


$\delta L(q_a, \dot{q}_a, t) = \Sigma  \frac{\partial L}{\partial q_a}\delta q_a + \Sigma  \frac{\partial L}{\partial \dot{q}_a}\delta \dot{q}_a$  

**Come fare derivata totale**

fai la derivata rispetto a tutte le var della funzione  
$dU= \nabla U \cdot dr = \Sigma \frac{\partial U}{\partial x_i} dx_i + \frac{\partial U}{\partial t} dt$  

esempi:  
$dU(x,y) =  \frac{\partial U}{\partial x} dx  + \frac{\partial U}{\partial y} dy$  
$dU(x,y,t) =  \frac{\partial U}{\partial x} dx  + \frac{\partial U}{\partial y} dy +  \frac{\partial U}{\partial t} dt $  

Modi di scrivere alternativi  
$dU(x,y) = U(x+dx, y+dy)-U(x,y) = \frac{\partial U}{\partial x} dx  + \frac{\partial U}{\partial y} dy = \vec{\nabla} U \cdot d\vec{r} $  
$\frac{dU}{dt}(x(t),y(t),t)= \vec{\nabla} U \cdot \dot{r} + \frac{\partial U}{\partial t}$   
le derivate parziali permettono uno sviluppo di Taylor al primo ordine  


**Come scrivere derivara totale**

due modi per scrivere: con $\dot{x}$ o $dx$  
essenzialmente dipende se è tutto diviso per dt o no  

esempio: $x(q)=\sin(q)$, dove $q=q(t)$  
$\frac{dx}{dt} = \dot{x} = \cos (q) \; \dot{q}$  
$dx = \cos (q) \; dq$  

**Come fare derivata paziale**  

usa chain rule:  

$ L(q_\epsilon \, \dot{q}_\epsilon, t)$  
$q_\epsilon(t)=q(t)+\epsilon h(t)$  

$\frac{\partial L}{\partial \epsilon} = \frac{\partial L}{\partial q} \frac{\partial q}{\partial t}+ \frac{\partial L}{\partial \dot{q}} \frac{\partial \dot{q}}{\partial t}  = \frac{\partial L}{\partial q} h(t)+ \frac{\partial L}{\partial \dot{q}} \dot{h}(t)$  

**Chain rule o no?**

se derivo per t: usa chain rule per le coordinate, perchè di solito sono funzioni del tempo  
se derivo per una coordinata: è una derivata parziale, no chain rule  

$x(q,t)=q(t)^2$   
$\frac{d x}{dt} = 2q \; \dot{q}$  
$\frac{\partial x}{\partial q} = 2q$  

**Nota**

$df = (\frac{\partial f}{\partial x})_y \, dy + (\frac{\partial f}{\partial _y})_x dy$  
la lettera accanto alla paretnesi vuol dire cosa resta costante  s 

**Derivare una sommatoria**

togli la sommatoria, deriva solo dentro  
perchè se ho sommatoria $\Sigma_j$, e derivo rispetto a $\partial_i$, della sommatoria mi interessa solo l'addendo dove i=j, non gli altri  
quindi la sommatoria collassa a un solo termine

$\frac{\partial a}{\partial b} \frac{\partial b}{\partial c} = \frac{\partial a}{\partial c}$

$\frac{\partial L}{\partial q} = \frac{d}{dt} \frac{\partial L}{\partial \dot{q}}$

#### Derivate parziali e trasformata di Legendre

**Quale var lascio costante**

con le derivate parziali è fondamentale quale var lascio costante  
se è specificato quali sono le var della funzione, assumo che tutte quelle per cui non sto derivando sono lasciate costanti  
o posso indicarlo con un pedice ad una parentesi tonda  

esempio fisico:  
$\partial_V U(S,V)=T$  
$\partial_V U(T,V)=0$  
cioè mi sto chiedendo se l'energia cambia quando cambia il volume  
mi dà risposte diverse per una trasformazione ad entropia o temperatura costante (adiabatica o isoterma) 

**Leggere le derivate parziali**

forma differenziale:  
$dU = T dS -P dV$  

quindi:  
$(\frac{\partial U}{\partial S})_V = T$  
$(\frac{\partial U}{\partial V})_S = -P$  

#### Trasformata di Legendre

**A che serve fisicamente**

se nella realtà posso controllare la var p, e non la x  

**Esempio: lasciare var costanti**  

ho: $f(x,y)=x^2+xy$  
definisco: $p=(\frac{\partial f}{\partial x})_y$  
cioè $p = 2x+y$ e $x= \frac{p-y}2$  

posso fare 2 cose:  
1- plug normale  
$f(p,y)=\frac{(p-y)^2}{4}+\frac{y(p-y)}2$  
2- uso Legendre  
$g(p,y)=px-f(x,y) = px-(x^2+xy) = \frac{(p-y)^2}4$  

SOLO se ho usato Legendre, posso fare:  
$(\frac{\partial g}{\partial p})_y = x$  


#### Differenziali esatti e non

**Note: differenziali esatti**

quando faccio il differenziale di una qualche grandezza f:  

1- se f è una funzione di stato: ho un differenziale esatto df    
cioè: il valore scambiato tra due stati dipende solo dallo stato iniziale e finale  
so che esiste una funzione f(x,y...)  

2- se f non è una funzione di stato: ho un differenziale non esatto $\partial f$    
cioè: il valore dipende dal percorso  
non esiste una funzione f(x,y...)  

esempio fisico:  
$dU$: vuol dire che esiste una funzione di stato $U(S,V)$, e la differenza dipende solo dai punti di inizio e fine  
$\partial L$: vuol dire che non esiste una funzione di stato $L(S,V)$, e il lavoro dipende dal percorso, quindi dal tipo di trasformazione compiuta (isocora, isobara, isoterma...)  

### Altro: derivare

**Chain rule**

ho $f(x)$  

$\nabla' f= \frac{\partial f}{\partial x_i'} = \frac{\partial f }{\partial x_j} \frac{\partial x_j }{\partial x'_i}$  

come procedere: calcolae la seconda derivata  
ho $x'_i = L_{ij}x_j$  
$x_j = (L^{-1})_{jk}x_k'$  
$\frac{\partial x_j }{\partial x'_i} = (L^{-1}_{ji})$  
lo pluggo:  
$\nabla' f= \frac{\partial f}{\partial x_j} (L^{-1})_{ji}=\nabla f \, (L^{-1})_{ij}$

**Altro**

$\frac {\partial x_k}{\partial x_i}=\delta _{ki}$  
cioè, tutte le var sono indipendenti tra loro  
quindi due diverse danno 0 perchè indipendenti, mentre la stessa dà 1 (derivata di sè stessa)  

### Altre info - non servono per dimostrazioni

**Campo conservativo**

2 definizioni:  
- posso scrivere $F=-\vec{\nabla} U$  
- $W = \int F \, dx$ è indipendente dal percorso fatto, solo dagli estremi della traiettoria

#### Spostamento virtuale

**Principio dei lavori virtuali**

Un sistema meccanico/oggetto è in equilibrio se e solo se: il lavoro totale delle forze applicate, corrispondente a qualsiasi spostamento virtuale compatibile con i vincoli, è nullo  
cioè se il lavoro calcolato sullo spostamento VIRTUALE è nullo  

**Spostamento virtuale**

Sono piccole variazioni matematiche delle coordinate, istantanee e compatibili con i vincoli  
cioè, se f(q)=0 è il vincolo, non deve violare f=0  

la differenziale totale del vincolo:  
$df = \Sigma \frac{\partial f}{\partial x_i} dx_i + \frac{\partial f}{\partial t} dt$  
uno spostamento virtuale avviene a TEMPO FISSATO, quindi dt=0, quindi  
$\delta f = \Sigma \frac{\partial f}{\partial x_i} dx_i = 0$  

nota che, in uno spostamento VIRTUALE, cioè a tempo fissato, lo scrivo senza l'ultima parte, perchè dt=0:  
$\delta {x}_i = \Sigma \frac{\partial x_i}{\partial q_a} \dot{q}_a$  


**Esempio di come usarlo in meccanica**  

ho un blocco su un piano inclinato, senza attrito  
metto asse x lungo il piano   

spostamenti virtuali: il piano, che è il vincolo, ammette solo spostamenti lungo il piano  
quindi scelgo come piccolo spostamento virtuale: $\delta \vec{r} = \hat{x} \delta s$  

principio dei lavori virtuali: il sistema è in equilibrio se:  
$L=\Sigma \vec{F} \cdot \delta \vec{r} =0$  
la reazione normale del piano è perpendicolare al piano, quindi fa lavoro nullo, quindi:  
$L=mg \sin \theta \, \delta s$  
l'unico modo per avere equilibrio è $\theta=0$    

**A cosa serve in meccanica newtoniana**

problema:  
in meccanica newtoniana, ho scritto tutte le leggi del moto: $ \quad m_i\ddot{x}_i=\Sigma F$  
adesso devo combinarle linearmente per far cancellare i vincoli, tipo T  
potrei fare a tentativi, con le sostituzioni  
ma c'è un modo sicuro usando lo spostamento virtuale  

soluzione:  
moltiplico tutte le equazioni per lo spostamento virtuale $\delta x$   
le forze vincolari, a meno che non sia presente attrito, saranno sempre
ortogonali a tali direzioni e quindi scompariranno
