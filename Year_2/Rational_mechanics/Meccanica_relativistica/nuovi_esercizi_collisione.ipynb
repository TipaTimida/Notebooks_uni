### Notazione covariante

#### Quadrimpulso

**Cos'è**

$p^\mu=(P^0, \vec p)$  
è quello che prima scrivevo come $w$   

come distinguere le cose:  
quadrivettore: metti $\mu$ in alto (o in basso)  
energia: metti $0$ in alto (se è ciò che voglio calcolare, potrei chiamarla E)  
modulo: metti freccia di vettore  

**Componenti**

componente controvariante:  
$p^\mu = (E, \vec p)$  

componente covariante:  
$p_\mu = (E, -\vec p)$  

stesso oggetto fisico   
cambia solo il segno della parte spaziale   
serve a scivere più facilmente il prodotto vettore  
succede a causa di come è definita la matrice della metrica di Minkowski  

**Scrivere i nomi**

1- che lettera usare  
dipende dal tipo di particella, di solito uso: protone=P, fotone=k, pione=q  

2- il $\mu$ va sotto o sopra?  
di default, sopra  
lo metto sotto solo se c'è un prodotto scalare (moltiplicazione tra due quadrimpulsi), allora invece di mettere il puntino del prodotto scalare porto $\mu$ sotto al primo quadrimpulso  
cioè: $P^\mu \cdot k^\mu = P_\mu k^\mu$  

3- aggiugnere pedici:  
se la particella esiste sia prima che dopo l'urto, per distinguere i quadrimpulsi aggiungo il pedice i=inizio o f=fine, per prima e dopo  
cioè: $P_{i}^\mu \; P_f^\mu$

#### Cose

**Quadrimpulso totale**

somma dei quadrimpulsi di tutte le particelle in uno specifico frame:  
$p_{tot}^\mu = \Sigma_i \, p^\mu _i$  

**Massa invarianti**

$M_{inv}=\Sigma \sqrt{m_i^2+|\vec p_i|^2|} $

**Energia di soglia**

è l'energia  minima per far avvenire la reazione   
quindi, nel frame dove avviene, la massa invariante è minima ed è la somma delle masse, perchè tutti i momenti sono nulli  

cioè, massa invariante in questo caso ha l'uguaglianza:      
$M_{inv}=\Sigma \sqrt{m_i^2+|\vec p_i|^2|} \ge \Sigma m_i$  

### Prodotto scalare relativistico

**Come calcolarlo**

formula generale: $P_\mu k^\mu = P^0 k^0 - \vec P \cdot \vec k$     

guarda se la parte covariante e controvariante sono:  

1- della STESSA particella: massa della particella  
$P_\mu P^\mu = M^2$   

2- del quadrimpulso TOTALE: massa invariante  
$p_{\mu, tot}p^{\mu}_{tot} = M_{inv}^2$  

3- di particelle DIVERSE:  
senza angolo: $P_\mu k^\mu = P^0k^0 - P_\mu k^\mu$     
con angolo: $P_\mu k^\mu = P^0k^0 - |\vec P| \, |\vec k| \cos \phi$ 

**Spiegazione**

1- perchè $P_\mu P^\mu = m^2$ ?  
$p^\mu p_\mu = (P^0)^2+|\vec p^2| = m^2$  
nota: fotone ha massa 0, quindi $k_\mu k^\mu = 0$  
cosa sono i moduli:   
per un fotone, $|\vec k| = k^0$  
per una generica particella che si muove, lascio scritto $|\vec q|$, ma sarebbe: $|\vec q |^2 =(q^0)^2 -m^2$   

2- modi di scrivere prodotto scalare  
posso esplicitare l'angolo e il modulo: $P_\mu k^\mu = P^0k^0 - |\vec P| \, |\vec k| \cos \phi$  

**Simmetria**

questo prodotto è simmetrico:  
$P_{\mu}k^\mu = k_\mu P^\mu$  

utile per i quadrati:  
$(P_\mu+k_\mu)(P^\mu +k^\mu) = P_\mu P^\mu + k_\mu k^\mu + 2P_\mu k^\mu$  

**Nota sugli angoli**

$\phi$: angolo tra $\vec k$ e $-\vec p$  
$\theta$ : angolo tra $\vec{k}$ e $\vec p$  
allora: $\theta = \pi -\phi$, quindi $\cos \theta = -\cos \phi$  

nel testo dice solo che $\phi$ deve essere l'angolo di incidemza   
non dice se con +p o -p, sono entrambi chiamati angoli di incidenza???  


**Scrivere quadrati**

1- di un binomio:  $(A+k)^2 = A\cdot A + k\cdot k +2 A \cdot k$  
perchè: usando il fatto che il prodotto è simmetrico  
$(P_{\mu}+q_\mu)(P^\mu +q^\mu) = P_{\mu}P^\mu + P_{\mu}q^\mu + q_\mu P^\mu + q_\mu q^\mu = M^2 + 2P_{\mu}q^\mu +m^2$  

2- di un trinomio:  
considero una var A=(P-q), così viene:  
$(A+k)^2 = A\cdot A + k\cdot k +2 A \cdot k = (P_\mu-q_\mu)(P^\mu-q^\mu) + k_\mu k^\mu +2 (P_\mu-q_\mu)k^\mu$  

### Risolvere esercizi

**Cosa posso fare**

1- conservazione massa invariante (quadrimpulso totale) (norma)    
$p_{tot, \mu}p_{tot}^\mu = M_{inv}^2$  
quando vale: sempre  
quando usare: energia di soglia  

2- conservazione quadrimpulso (tutte le componenti)  
$p_{tot,i}^\mu = p_{tot, f}^\mu$  
quando vale: solo se cambia tempo, non se cambia frame  
quando usare: calcolare una var del dopo l'urto in lab frame, tipo energia di una particella dopo l'urto in lab frame  

**Decadimento**

$L=\beta \, \Delta T$    
$L$: spazio percorso, nel lab  
$\Delta T$: tempo per decadere, nel lab  
$\beta$: la trovo sapendo $q^0 = m \gamma = \frac{m}{\sqrt{1-v^2}}$   

passare a proper frame:  
$\tau = \frac{\Delta T}{\gamma}$  

#### Calcola energia di soglia

**Siuazione**

prima: fotone + protone   
dopo: pione (m)+ protone (M)

caso 1: prima, fotone a riposo  
caso 2: prima, fotone in moto (scrivi l'energia in funzione dell'angolo di incidenza)  

la prima parte uguale per i due casi, solo dopo uso l'informazione sul protone  

**Steps**

1- conservazione di quadrimpulso totale:  
$p_{tot, \mu}p_{tot}^\mu = M_{inv}^2$  

2- espando il lato sinitro:  
$p_{tot}^\mu = P_i^\mu+k^\mu$  
$p_{tot, \mu}p_{tot}^\mu = (P_{i,\mu}+k_\mu)(P_i^\mu +k^\mu) = P_{i,\mu}P_i^\mu + P_{i,\mu}k^\mu + k_\mu P_i^\mu + k_\mu k^\mu = M^2 + 2P_{i,\mu}k^\mu $  

3- trovo disuguaglianza nel lato destro:  
$M_{inv} = \Sigma \sqrt{m_i^2}+|\vec p_i|^2 \ge \Sigma m_i = (M+m)$    

4- recap:  
$M^2 + 2P_{i,\mu}k^\mu \ge (M+m)^2$  

**Versione 1: protone fermo**

1- uso l'info: $\vec P =0$  
$P_{i,\mu}k^\mu  = P^0k^0-|\vec P|  \cdot| \vec k|  = P^0k^0 = Mk^0$  

2- metto tutto insieme:  
$ M^2 +2Mk^0 \ge (M+m)^2$  
quindi: $k^0 = m + \frac{m^2}{2m}$  



**Versione 2: protone in moto**

1- uso formula generica:  
$P_{i,\mu}k^\mu  = P^0k^0-|\vec P|\, | \vec k|  = P^0k^0 -|\vec P| k^0 \cos \phi$  
diventa +, ma perchè???   


DA QUI, NON CAPISCO:  
2- l'angolo di incidenta $\phi$ è quello tra i vettori $\vec k$ e $\vec P_i$, così:  
$P_{i,\mu}k^\mu = P^0k^0 + |\vec P| k^0 \cos \phi$    

3- nel limite $k^0 \ll m$, allora $P^0+|\vec p| \gg M$ (perchè???)   
quindi $\vec|p| \sim P^0$  
quindi: $P^0 \ge \frac{Mm}{(1+\cos \phi)k^0}$

#### Conservazione del quadri-impulso (tutte le componenti)

**Situazione**

prima: fotone + protone a riposo  
dopo: pione (m)+ protone (M)

1- conservazione del quadrimpulso:  
$P_i^\mu +k^\mu = P_f^\mu +q^\mu$  
cioè: $P_i^\mu +k^\mu -q^\mu= P_f^\mu$  

2- faccio il quadrato (cioè prodotto scalare)  
$(P_{i\mu} +k_\mu -q_\mu)(P_i^\mu +k^\mu -q^\mu)= P_f^\mu P_{f\mu}$  

3- calcolo lato destro:  
$P_f^\mu P_{f\mu} = M^2$  

4- calcolo lato sinistro:  
formula generale: $(A+k)^2 = A\cdot A + k\cdot k +2 A \cdot k = (P-q)\cdot (P-q) + k \cdot k +2 (P-q) \cdot k$  
$=  M^2+m^2 -2P_{i\mu}q^\mu +0 +  2 (P_\mu-q_\mu)k^\mu$   

5- metto insieme:  
$M^2+m^2 -2P_{i\mu}q^\mu +2 (P_\mu-q_\mu)k^\mu = M^2$    
$2 (P_\mu-q_\mu)k^\mu = 2P_{i\mu}q^\mu
-m^2$  

6- nel frame del LAB, il protone è a riposo  
lato sinistro:  
$2k_\mu (P^\mu-q^\mu) = 2(k_\mu P_i^\mu-k_\mu q^\mu) = 2(k^0M-\vec k \cdot \vec 0 \; - k^0 q^0+\vec k \cdot \vec q) = 2(k^0M -k^0 q^0 + \vec k \cdot \vec q) = 2k^0(M-q^0) + 2k^0 \vec q \cos \theta$  
lato destro:  
$P_{i\mu}q^\mu = 2Mq^0$, perchè il protone è è fermo  

7- ora, con l'algebra, mi basta isolare $k^0$:  
$2k^0(M-q^0)+2k^0|\vec q|\cos \theta = 2Mq^0-m^2$  
$k^0 = \frac{q^0-m^2/(2M)}{1-(q^0-|\vec q|\cos \theta)/M}$  

8- limite in cui $m, k^0 \ll M$  
riscrivo lo step 7, lasciando solo gli addendi dove appare M:  
$2Mk^0 \sim 2Mq^0$
