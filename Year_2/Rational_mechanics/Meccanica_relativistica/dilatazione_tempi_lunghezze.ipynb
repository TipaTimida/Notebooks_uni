### Dilatazione tempi

**Definizione**

se un osservatore (O) guarda un oggetto che gli appare in moto (O')    
il tempo percepito da O è maggiore di quello percepito da O'  

**Proof**

situazione:  
O vede O' muoversi con velocità v lungo asse x  
c'è un orologio nell'origine di O'  
assumo che batta a t=t'=0    
quando è il prossimo battito/evento?  

battito nei due frame:  
frame O': $(\tau, 0)$ (l'orologio è sempre nella sua origine)   
frame O: facendo trasformazione Lorentz, avviene in:  
$ct = \gamma (ct'-\beta x') = \gamma c\tau$

**Decadimento particelle**

$\tau$: tempo di decadimento, tra eventi A (creazione) e B (decadimento)  
è definito nel frame proprio della particella  

nel frame di un osservatore, che vede il muone in movimento:  
il tempo di decadimento è più lungo di $\tau$  

conseguenza:  
muoni sono prodotti nell'alta atmosfera  
arrivano sulla superficie terrestre, ma non ce la farebbero se avessero tempo di decadimento $\tau$ in tutti i frame  
il fatto che li vediamo sulla superficie terrestre è la prova dell'esistenza della dilatazione dei tempi  

**Paradosso**

domanda:  
metto 2 orologi uguali, li chiamo 1 e 2, in origini di O e O'  
O vede orologio 2 muoversi, quindi batte più lento del suo  
O' vede orologio 1 muoversi, quindi batte più lento del suo  
chi ha ragione?  

soluzione:  
la domanza è mal posta  
i due osservatori non avranno mai modo di verificare veramente quale dei due orologi vada più piano  
Infatti O ed O′ possono confrontare e sincronizzare i due orologi solo quando questi si incrociano (cioè si trovano nello stesso punto e allo stesso istante), ma poiché i due sistemi sono inerziai può avvenier una sola volta  

**Due orologi**

voglio sapere quale è più lento:  

se gli orogologi si separano e poi si reincontrano:  
posso considerare il tempo misurato da ognuno tra evento A (separazione) e evento B (reincontro)  
quindi ok, posso trovare il più lento   

se due orologi si separano e non si incotrano più:  
non ha senso chiedersi quale sia in ritardo  
perchè non ci saranno due eventi simultanei che posso considerare come 'fine', punto B  
visto che la simultaneità non esiste  

### Contrazione lughezze

**Definizione**

la distanza fra due punti materiali a riposo in un dato riferimento (ad esempio gli estremi di un’asta rigida), risulta contratta per un osservatore in moto relativo in direzione parallela alla distanza stessa  

**Proof**

situazione:  
ho due orologi fermi in O, ai punti x=0 e x=L  
secondo O, O' si muove a velocità v  

il tempo misurato da O' tra quando incrocia il primo e quando incrocia il secondo è: $t_2'-t_1'=\frac L {\gamma v}$  
O' vede gli orologi muoversi a velocità -v, quindi la distanza tra di loro gli sembra $\frac L \gamma$

### Proper speed and time

**Definizioni**

tempo proprio: tempo che intercorre fra due eventi che accadono nello stesso punto  
cioè misurato da un frame solidale all'oggetto che si muove  

lunghezza propria: lo spazio che separa due eventi che accadono allo stesso istante  
cioèmisurato in un frame in cui i due eventi appaiono contemporanei  

**Esempio**

a ship travels from Earth to Mars    
define 2 events:  
A: ship leaves Earth    
B: ship reach Mars  

1- frame of ship  
the ship is always in its origin, so events A and B happen in the same place  
so i'm measuring proper time (but not proper length???)    
travel takes time $\tau$ and is long $L$  

2- frame of person on Earth  
i see the ship moving, and events A and B happening in different places  
travel time takes $\gamma \tau$ and is long $\frac L \gamma$ (dilatazione tempi e contrazione lunghezze)
