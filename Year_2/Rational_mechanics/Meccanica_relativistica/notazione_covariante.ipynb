### Metric spaces

**Covariant formulation**

a way of writing physical laws so that they keep the same mathematical form in every INERTIAL??? coordinate system  
so, adapted to the metric of the spacetime i'm working in  
for example, i could adapt it to the Minkowski metric  

**Euclidean space vs Minkoski space**

how to describe space-time:  
by giving it a scalar product, which is decribed by a matrix $g_{\mu v}$, called 'metric tensor'   

examples:  
euclidean, classical mechanics: $g_{ij} = \delta_{ij}$, cioè l'identità I     
minkowki, relativity: $g_{\mu v} = \eta_{\mu v}=diag(1, -1, -1, -1)$  

**When appears covariant formulation**  

in euclidean, covariant and contravariant are the same  
$v_i = \delta_{ij}v^j = v^i$  

but in minkowski they're different  
it changes the sign of the spatial vector  
$k_\mu = \eta_{\mu v} k^v = \eta_{\mu v}(k^0, k) = (k^0, -k)$  

**How to incorporate the metric**

the metric appears when doing a scalar prduct  
which is describes by the metric tensor (matrix describing the scalar product)  
$a^i \cdot b^j = g_{ij} a^ib^j$

covariant formulation is a differet way to say it:  
1- do scalar product between a and b: $g_{ij} a^ib^j$  
2- pass vector b as input, to the functional that does scalar product with vector a: $a_jb^i$  
where $a_i = g_{ij} a^i$  

note the indexes:   
in version 1, a and b are inputs of g, so so a and b have different indexes, and are copied in g  
in version 2, b is the input of a, so a and b have the same index  

**Tipi di trasformzioni**

in ogni spazio, sonofisicamente ammissibili solo le trasformazioni che lasciano invariato il tensore metrico di quello spazio  
cioè per cui: $\Lambda^t g\Lambda = g$  

1- spazio euclideo: $Λ^t I Λ = 0$  
cioè $Λ^tΛ=0$, $Λ$ è una trasformazione ortogonale, cioè una rotazione  

2- spazio di Minkowski:  $\Lambda^t \eta_{\mu v} \Lambda = \eta_{\mu v}$  
le trasformazioni con questa proprietà si chiamano 'trasformazioni di Lorentz'  

**Invarianza**

un oggetto che ha stesso numero di indici covarianti e controvarianti  è invariante per cambio di coordinate  

perchè:  
durante un cambio base, controvariante significa moltiplicare per $\Lambda$, covariante significa moltiplicare per $\Lambda^{-1}$  
quindi se sono di uguale numero, in totale non cambia niente durante il cambio base  

esempio:  
prodotto scalare tra due vettori si scrive: $x_\mu y^\mu$  
se faccio cambio base, y è controvariante e si moltiplica per $\Lambda$, x è covariante e si moltiplica per $\Lambda^{-1}$  
in totale non cambia niente  
che è ciò che voglio, le trasformazioni devono lasciare invariata la metrica dello spazio, che è descritta dal prodotto scalare    

### Dual space

**Dual spaec**

1- $V$: vector space  
made of contravariant vectors = vectors  
must be defined on a scalar field, generally R (cioè chiuso per moltiplicazione rispetto a R)  

2- $V^*$: dual space  
made of covariant vectors = functionals  
composto dalle applicazioni lineari da V sul campo di definizione di V (R)  

**Isomorprhism**

how it's done:  
the metric (so the scalar product, $g_{ij}$) creates an isomorphism between $V$ and $V^*$  
so to each vector, it univocally maps a dual vector  
bc for each vector v, can UNIVOCALLY define the functional that makes the scalar product with that vector  

what it looks like:  
l’applicazione lineare associata ad un qualsiasi vettore v di V è quella che fornisce il prodotto scalare con v  
so v is mapped to the functional $f(x)=v \cdot x$  

E’ proprio questa la corrispondenza biunivoca che realizziamo quando abbassiamo o alziamo gli indici: stiamo passando da un vettore al suo corrispondente nel duale (secondo la metrica g assegnata) o viceversa  
$v_\mu = g_{\mu v}v^v$  
$v^v$: vettore  
$v_\mu$: covettore associato

### Indexes

#### Check something is well written

**Nomenclature**

$v^\mu$: controvariant index, put higher  
means that object serve da input  

$v_\mu$: ovariant index, put lower  
means that object needs an input  

**Mute and free index**

1- free index:   
what: tells me the component i want (i'm in 4D space-time)    
when: in between each pair there must be an = sign  
types: on both sides of the equal sign, must be of same type, both covariant or both controvariant   
intuitive: immagina che si copia oltre l'uguale    

2- mute index:   
what: tells me relationship between a functional and the vector it takes in input  
when: appears twice without an = sign in between  
types: in the 2 objects where it appear (between who there must not be an = sign), it must be of opposite type  
intuitive: immagina che si cancellano, tipo numeratore e denominatore di una frazione  

**Example: mute and free index**

$k_\mu = \eta_{\mu v} k^v$  

$\mu$: free index   
in both sides appear as covariant, tells me i want a covariant component as a result  

$v$: mute index  
in $\eta$ is covariant, in $k$ controvariant, means $k$ is given as input to k  

**Check a formula is correct**

it must be still true for base changes  
so, when do a base change, the number of $\Lambda$ multiplied to each side must be the same  

how to check:  
to each part in between two = signs, give it a rank  
rank=(a,b), dove a è il numero di indici controvarianti, b covarianti  
eg. $g_{ij}$ is a (0,2) tensor   
guarda i due lati dell'uguaglianza, a e b possono essere diversi, ma la differenza a-b deve essere uguale  

why:  
se tipo ho a-b=1, vuol dire che quell'espressione, in un cambio di base, viene moltiplicata per M   
quindi anche l'altro lato dell'uguaglianza, durante il cambio coord, deve venire moltiplciato per M, sennò l'uguaglianza è rotta  

**Eccezione**

se ho una una derivata, la cosa al denominatore ha il segno opposto  
esempio: $\nabla_i = \frac{\partial}{\partial x^i}$  
anche se $x^i$ sembra essere controvariante, in realtà la frazione è un oggetto covariante
