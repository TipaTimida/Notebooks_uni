**Steps: per moto in 1D**

**1. Scrivi l'hamiltoniana**

in coord cartesiane: $H=\frac{p^2_x}{2m}+\frac{p^2_y}{2m}+V(x,y)$    
in coord polari: $H=\frac{p^2_r}{2m}+\frac{p^2_\varphi}{2mr^2}+V(r, \varphi)$

cosa è q:

pendolo in piccole oscillazioni: spostamento orizzontale rispetto alla verticale (sarebbe angolo, ma in piccole oscillazioni sin x = x, quindi non è un angolo e non serve r^2 al denominatore)

**Oscillatore armonico con massa efficace**

è un oscillatore armonico la cui massa non è costante, ma dipende tempo o coordinate    
cioè lagrangiana si scrive: $L=\frac{1}{2}m(q,t) \dot{q}^2-\frac{1}{2}m(q,t)\omega q^2$  

**5. Se potenziale V(x) definito a tratti**

alla separazione tra due V diversi, cioè la discontinuità, devo vedere cosa succede   

se una regione di spazio ha $V=\infty$, vuol dire che non si può andare in quella zona  
quindi lì non ci sarà diesgnata nessuna linea di flusso  
esempio:  
$x>0:\quad V=A|x|^p$ (ellisse)  
$x<0 \quad V=\infty$ (inagibile)

**5. Controllo discontinuità**

esempio: rimbalzo su parete

### Periodo della traiettoria

**Periodo della traiettoria**

$T(E)=\frac{dI(E)}{dE}$  
validità: tutti i sistemi HAMILTONIANI con moto periodico (cioè per cui T ha un senso)  

$I=T \, E$  
validità: solo per un oscillatore armonico  

queste sono per 1 grado di libertà, se sono di più diventa più complicato, con le 'action variables'  

**Azione: I**

è l'area sottesa dalla traiettoria (curva nello spazio delle fasi)     
se traiettoria chiusa: è l'area racchiusa  
sennò: area sottesa dalla traiettoria (quindi tra traiettoria e asse x)  

**Calcola I: metodi**

1. formule di geometria  
cioè disegna la traiettoria, vedi che forma ha e se conosco la formula dell'area usala (per forme facili, tipo ellisse o rettangolo)  
area ellisse: $I=\pi ab$, con a e b semiassi (cioè interesezioni di ellisse e assi x,y)    

2.  SOLO per oscillatore armonico: $I=T \, E = \frac{2 \pi}{\omega} E$  
quindi devo trovare $E$ ed $\omega$ per quel problema  

3. generale: $I=\int p \, dx$ oppure $I=2\int p \, dx$  
se è chiusa: metti coeff 2, perchè voglio anche la parte sotto  
sennò: senza coeff 2, solo parte tra curva e asse x  

**Esempio metodo generale - CHECK!!!**

1. trova espressione di p(x)  
basta scrivere l'hamiltoniana, uguagliare a E, e invertire per avere p  

2. trova punti estremi (per estremi di integrazione)  +sono quando E=V, cioè K=0   

$H=\frac{p^2}{2m}+A|x|^p$  
1- $E=\frac{p^2}{2m}+A|x|^p \rightarrow p(x) =\sqrt{2m(E-Ax^p)}$  
2- $E = A|x|^p \rightarrow x_0 = \pm (\frac{E}{A})^{1/p}$  

hint: se i punti di inversione sono $\pm$ qualcosa, se p è simmetrico posso dimezzare l'intervallo di integrazione e moltiplicare per 2  
questo tipo di integrale, già visto negli esercizi vari, si risolve come $y==\frac{x}{x_0} = \frac{x}{(E/A)^{1/p}}$, così resta $\int_0^1 \sqrt{1-y^p} \, dy=I_p$, che è semplicemente un numero, gli dò un nome a caso  
lo faccio per estrarre la dipendenza dall'energia, così il valore dell'integrale dipende solo da p, lo chiamo $I_p$  
quindi se sto studiando qualcosa relativo all'energia, poi $I_p$ si cancellerà  

$2 \int_0^{x_0} \sqrt{2m(E-Ax^p)} dx =  \sqrt{2mE} (\frac{E}{A})^{1/p} \int_0^1 \sqrt{1-y^p} \, dy =  \sqrt{2mE} (\frac{E}{A})^{1/p} I_p$  

hint: potrei accorgermi che visto che l'integranda non cambia ma i limiti passano da [0,x] a [-x,x], allora $J'=2J$     

### Regime adiabatico

#### Teoria

**Regime adiabatico**

avviene una variazione di un parametro molto lenta    
cioè il parametro è LENTAMENTE VARIABILE  

esempi:  
1- pendolo semplice con lunghezza l che cambia piano (lo tiro sù o giù)  
2- orbita terrestre attorno al Sole, con massa del sole che diminusice poco a poco  

**Parametro lentamente variabile**

$\frac{d\lambda}{dt} \cdot T(t) \ll \lambda$  
$\lambda$ è il parametro  
vuol dire che la variazione del parametro in un periodo è tanto piccola da essere trascurabile  

**Note**

se il parametro è LENTAMENTE VARIABILE:  
1- le orbite sono ordinate in rodine di E crescente  
2- le orbite non si intersecano mai  
3- se $E_1 > E_2$, allora $E_1' > E_2'$  

se il parametro varia velocemente, queste cose non sono vere  
ad esempio, se ho pallina m in una scatola, e sposto parete della scatola  
se sposto la parete veloce, cosa succede dipende da in che direzione sta viaggiando la pallina in quel momento???

**Invariante adiabatico**

qualcosa che, in regime adiabatico (durante una variazione lenta di un altro parametro), rimane costante   

I, l'azione, è un invariante adiabatico  
th di Liuoville: la mappa (di evoluzione temporale?) conserva l'area  
cioè $I(E,\lambda)=I(E', \lambda')$   

**Validità soluzione**

questa non à la soluzione perfetta, ma è un'approssimazione, tanto più giusta quanto il parametro varia lentamente  
per la soluzione perfetta dovrei risolvere con la conservazione dell'energia  

se il parametro varia velocemente, il th di Liuoville vale comunque  
però non so come è la mappa, poterbbe frantumarsi invece di avere come immagine un insieme connesso  
quindi il metodo non funziona  

#### Esercizi

**Esercizio con parametro lentamente variabile**

$\lambda$: parametro lentamente variabile  
$A$: variabile di cui voglio sapere la variazione  

ci sono due modi di rispondere:    

1. legge di evoluzione di A  
$A' = f(A, \lambda, \lambda')$  

2. dipendenza istantanea di equilibrio di A da $\lambda$  
$A \propto g(\lambda)$  

in entrambi uso l'invariante adiabatico I,  
ma:  
nel primo uguaglio due momenti definiti, $I=I'$   
nel secondo ne metto uno generico costante, $I=const$  


**1. Legge di evoluzione di A**

steps:  
1. scrivi I per un $\lambda$ e E generici    
2. fai uguaglianza $I(E,\lambda)=I(E', \lambda')$  
basta cambiare i nomi delle var mettendo un '  
3. riordina per avere $E'(E, \lambda, \lambda')$  

tip: utile se vedo graficamente il rapporto tra le aree  
ad esempio, per trovare Relazione tra energie a inizio e fine di trasformazione adiabatica  
ad esempio, $I'=2I$, magari perchè ho ampliato le zone ammissibili, definendo V su tutto lo spazio, mentre prima era solo su x>0  
quindi guarda la formula generica di I, sostituisci da entrambi i lati co E e E', vedi rapporto tra E e E'  

**2. Dipendenza istantanea di A**

steps: 

1. $I(A, \lambda)$   
cioè scrivi I per un $\lambda$ e A generici  

2. $I(A, \lambda) = cosnt$     
cancella da I tutti i paramtri che rimangono costanti, cioè lascia solo A e $\lambda$  
e uguaglialo a const

3. riordina per trovare la forma che voglio  

aggiunta: se poi voglio anche B in funzione di $\lambda$, e so $B(A)$, basta pluggare:  
$B = B(A) \propto B( A(\lambda))$    

**Esempio: dipendenza istantanea**

parametro variabile: $l$  
var che voglio sapere: $\theta_0$  
è un pendolo semplice col filo che si acorcia, voglio sapere ampiezza di oscillazione  

preparazione:  
$E=mgl(1- \cos \theta)=\frac{1}{2}mgl\theta_0^2$  
in oscillatore armonico, $I=TE$  

1- $I(E,l) = \frac{2 \pi}{\omega} E = \frac{2 \pi}{\sqrt{\frac{g}{l}}} \frac{1}{2}mgl \theta_0^2$  
2- $I \propto l^{3/2}\theta_0^2 = const$  
3- $\theta_0 \propto l^{-3/4}$


**Forza media**

formule:  
1- $F=\frac{\Delta p}{\Delta t}$   
2- $F = - \frac{dE'}{d\lambda'}|_\lambda$  

1- se voglio forza media su un periodo, $\Delta T$ è il periodo, calcolato con dI/dE  
$\Delta p$ lo vedo dalla traiettoria del flusso  

2- è la proiezione della forza lungo una coordinata generalizzata  
$\lambda$ deve essere la coordinata generalizzata, di solito in 1D chiamata $x$   

modi di usare metodo 2:  
se ho legge di evoluzione $E'(E, \lambda, \lambda')$   
calcola $F = -\frac{dE'}{d\lambda'}|_{\lambda}$  
cioè alla fine, sostituisci le $\lambda'$ rimaste con $\lambda$        
se ho dipendenza istantanea: $E(\lambda, c)$ (c =const)  
calcola $F=-\frac{dE}{d\lambda}$  
semplicemente, non devo preoccuparmi di sostiturire cose dopo  

**H dipende esplicitamente dal tempo, ma è costante in media su un periodo?**

moto durante cui: $\gamma \ll \omega$    

metodo 1: conservazione di I   
scrivi I ad un tempo generico, so che è conservato  
quindi se è tipo $I=\frac{2 \pi H}{\omega}$, e so che $\omega$ è costante, allora anche H è costante  

metodo 2: calcolo diretto di $\dot{H}$  
calcola $\dot{H}=\frac{\partial H}{\partial t}$  
(ricorda che in sistema hamiltoniano, differenziale e derviata parziale di H per il tempo sono uguali)  
nel risultato resterà comunque il tempo t, ma calcola la media su un periodo, se dà 0 allora H è costante  

attenzione!    
H(t) non è esattamente costante istante per istante, ma è costante in media su ogni oscillazione  
vale per entrambi i metodi  

**Moto con piccole variazioni**

so che $\gamma \ll \omega$  
e ho: $H=e^{-\gamma t} \frac{p^2}{2m} + e^{\gamma t} \frac{m\omega^2}{2}q^2$  
$\gamma$ varia lentamente, come varia E  

come fare:  
posso approssimare il moto come quello di un oscilaltore armonico, come detto prima, quindi $q(t)\sim A(t) \cos (\omega t) \quad p(t)=B(t) \sin (\omega t)$   
dove A(t) e B(t) sono i semiassi in ogni momento  

spiegazione:  
studio all'interno di una sola oscillazione, cioè per $t < T$  
il termine problematico è: $e^{\gamma t} < e^{\gamma T} =e^{\gamma 2 \pi / \omega}$  
dove $\gamma/ \omega \ll 1$, quindi ho circa $e^0 \sim 1$, cioè posso ignorare questo termine con la dipendenza dal tempo  
quindi: sulla scala dei tempi di una singola oscillazione posso considerare il sistema un oscillatore armonico
