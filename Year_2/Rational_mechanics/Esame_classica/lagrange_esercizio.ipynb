### Scegliere coordinate generalizzate

**Guida**

quante: abbastanza per cui, settandole tutte, è definito UNIVOCAMENTE uno stato del sistema    

come:   
1- in modo che (0,0) sia il punto di equilibrio stabile    
2- se l'apparato disegnato ha un asse di simmetria, q=0 sull'asse  

**Per disco di massa M**

se ruota intorno al suo centro, ma il cm non si sposta:   
mi serve solo $\theta$, l'angolo di un suo punto qualunque, così $\dot{\theta}$ è la sua velocità angolare  

se rotola senza strisciare (pure rolling):  
mi basta x del suo centro, non mi serve un'altra coordinata per l'angolo, perchè $R\omega=\dot{x}$  
o potrei prendere solo $\theta$ invece di $x$, se mi è più comodo per qualcos'altro  

**Esempi: come scegliere**

esempio 1: ho tre masse m, ognuna sui lati di una triangolo equilatero  
intuitvamente, l'equilibrio è quando tutte sono nel centro del proprio lato  
quindi scegli come coord la distanza dal centro del proprio lato per ogni particella   
oppure, l'asse di un segmento è un asse di simmetria dell'apparato  
quindi setta q=0 quando è nel centro del lato  
e setta la direzione è tutte in verso antiorario, così ho il moto 1,1,1  

esempio 2: pendolo  
il punto di equilibrio è quando è in verticale  
quindi metti come $\theta$ l'angolo rispetto alla verticale in basso  
oppure: la linea verticale è l'asse di simmetria dell'apparato  

**CM e relative**

può essere utile scrivere come coord:  
$x_{cm} =\frac{m_1x_1+m_2x_2}{m_1+m_2}$  
$x_{rel} = x_2-x_1$  

**Direzione**

su una riga 1D, due coord $x_1$ e $x_2$ potrebbero avere il segno positivo in direzione opposta  
utile per vedere simemtria rispetto al punto centrale  
se lo faccio, comunque non ci va il - nella velocità in T  

#### Polari

**Coordiante polari**

per ognuna, scrivo le componenti $\vec{r}=(x,y,z)$    

polari in 2D:  
$\vec{r} = (r \cos \theta, r \sin \theta)$  

cilindriche:  
$\vec{r} = (r \cos \theta, r \sin \theta, z)$

sferiche:  
$\vec{r} = (r \, \sin \theta \, \cos \varphi,  r \, \sin \theta \, \sin \varphi,  -r \, \cos \theta) $  
$\varphi$: come l'angolo in coordinate polari, parte dall'asse x e poi in senso antiorario verso l'asse y  
$\theta$: l'angolo verticale, l'altezza del punto 
il - sull'ultima componente di solito non c'è  si aggiunge perchè così quando $\theta=0$, il pendolo è sul fondo (sennò sarebbe in cima)  
cioè $\theta$ è l'angolo rispetto alla verticale in basso, non in alto  
cioè sto facendo la sostituzione $\theta \rightarrow \pi-\theta$  
le altre componenti non cambiano, perchè c'è il seno in entrambi, e il seno di queste due è uguale  

**Come usare**

usare se ci sono:  
pendoli (se il punto a cui è attaccato si muove, definisci le coord anche di quello, anche se non ha massa)    
disco che ruota attorno al suo asse (definisci le coord di un punto qualunque sul suo bordo)  
particelle puntiformi, ognuna su guida circolare, senza gravità, solo forza di Coulomb (scelgo due angoli,  $\theta_1$ e $\theta_2$, ognuno che indica una diversa particella)  

attenzione:  
l'angolo è sempre rispetto alla verticale in basso, così $\theta=0$ è la posizione di equilibrio stabile, sarà più facile fare le piccole oscillazioni, on servità il var change     
se la lunghezza l è fissata non mi serve la coord r, basta l'angolo  

attenzione 2:  
se ho più di una coordinata angolare, scegli l'angolo positivo NELLA STESSA DIREZIONE    
così mi dà meno problemi di + e -  
cioè quando faccio il disegno, disegna $\theta_1$ e $\theta_2$ entrambi verso destra, in senso antiorario, che è la direzione che ho scelto positiva

### Scrivere T

**Steps**

$\vec{q}$: generalizzate, $\vec{x}$: cartesiane  

1. scrivi $x(q)$  
2. deriva per dt, così ho $\dot{x}(q, \dot{q})$  
3. pluggalo nella formula di $T(x)$  

**Esempio: derivare rispetto al tempo**

derivo rispetto al tempo, quindi ricorda la chain rule sulle coord  

$\begin{cases}
x_1 = l_1 \sin \theta_1 & \dot{x}_1 = l_1 \dot{\theta}_1 \cos \theta_1 \\
y_1 = l_1 \cos \theta_1 & \dot{y}_1  = - l_1 \dot{\theta}_1 \sin \theta_1\\
x_2 = x_1 + l_2 \sin \theta_2 = l_1 \sin \theta_1 + l_2 \sin \theta_2 & \dot{x}_2 = l_1 \dot{\theta}_1 \cos \theta_1  + l_2 \dot{\theta}_2 \cos \theta_2 \\
y_2 = y_1 + l_2 \cos \theta_2 = l_1 \cos \theta_1 + l_2 \cos \theta_2 & \dot{y}_2 = - l_1 \dot{\theta}_1 \sin \theta_1  - l_2 \dot{\theta}_2 \sin \theta_2 
\end{cases}$

**Formula di T**

formula in coordinate cartesiane:    
$T = \Sigma \frac{1}{2}m \dot{x}_{cm}^2 + \Sigma \frac{1}{2}I_{cm}\omega^2$     

con la m: per oggetti che traslano    
con la I: per oggetti che ruotano, tipo un'asta o un disco  

**Note: formula di T**

oggetto con moto circolare:  
con I è per oggetti che ruotano intornoal proprio asse  
per oggetto puntiforme che fa traiettoria circolare, metti quella con m, e poi $v=R \dot{\theta}$

se c'è qualcosa che ruota e quindi mi serve $I\omega^2$:  
per quell'oggetto scegli una coord $\theta$, l'angolo rispetto alla verticale e al centro dell'oggetto, così che $\dot{\theta}=\omega$  

$v_{cm}^2 = \dot{x}^2+\dot{y}^2$

### Scrivere V

**Steps**

1. individua tutte le energie potenziali presenti  

2. scrivi $V(x)$  
per ora la distanza tra particelle può essere un generico d      

3. plugga per avere $V(q)$  
se richiesto, trova d      

4. cancella addendi costanti  

5. vedi se puoi raccogliere qualche angolo (se ho usato coordinate polari)  

U elettrostatica tra due cariche la considero solo se lo dice espressamente, sennò no  

**3. Distanza tra particelle: coord polari**  

2 metodi:  

1. $d^2=\Delta x^2+\Delta y^2$   
cosa: uso formula in cartesiane, poi plugga $x(q)$  
ricorda che spesso qui posso raccogliere gli angoli (punto 5)

2. $l^2=x^2+y^2-2xy \cos \theta$   
cosa: th del coseno, dove ho un triangolo di lati x,y,l, e $\theta$ è l'angolo opposto a l, il lato che voglio sapere  

il metodo 2 è meglio, evito anche di scrivere le coord cartesiane in funzione di quelle generalizzate  

**1. Tipi di energia potenziale**

gravitazionale$: $V=mgz$    
elastica: $V=\frac{1}{2}k(r^2)$    
elettrostatica  

**4. Cancellare addendi costanti**

se mi serve per scrivere la lagrangiana, posso cancellare gli addendi costanti  
perchè V è definito a meno di una costante  

esempio: $V=\frac{1}{2}k(x^2+d^2)$, con d costante  
posso scrivere invece: $V=\frac{1}{2}kx^2$   
particolarmente utile con le molle  

**5.Riscrivere angoli**

se ho usato coord polari con più di un angolo:   
spesso devo raccogliere come cos o sin di una differenza, tipo:  
$\cos \theta \cos \varphi - \sin \theta \sin \varphi = \cos (\theta+\varphi)$  
accade spesso in d, la distanza tra due particelle  

esempio:  
$d^2 = (\Delta x) ^2 - (\Delta y)^2 =  l^2(\cos \theta - \cos \varphi)^2 + l^2 (\sin \theta - \sin \varphi)^2 = 2l^2(1-\cos \theta \cos \varphi - \sin \theta \sin \varphi) = 2l^2 (1- \cos (\varphi-\theta))$
