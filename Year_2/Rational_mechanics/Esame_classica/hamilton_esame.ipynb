#### 1. da L ad H (e viceversa)

**Steps**

1. tra $L(q, \dot{q})$ e $H(q,p)$, determino quale var ho e quale voglio      

2. calcolo la variabile che voglio, in funzione della variabile che ho  
per farlo, deriva la funzione che ho che ho per la var che ho (non q, l'altra)  
$p(q, \dot{q}) = \frac{\partial L}{\partial \dot{q}}$    
$\dot{q}(q,p) =\frac{\partial H}{\partial p}$  

3. invertile per ottenere la variabile che ho in funzione della variabile che voglio  

3. usa la trasformazione di Legendre  
$H(p,q)=p\dot{q}-L(q, \dot{q})$  
var scambiate - originale  
in una delle due var moltiplicate (quella che non deve apparire in questa funzione), pluggaci la formula del punto 3    
poi è tutta algebra  


**Esempio pendolo semplice: lagrangiana -> hamiltoniana**

lagrangiana:  
$L(\theta, \dot{\theta}) = \frac{1}{2}ml^2\dot{\theta}^2+mgl \cos \theta$  

momento generalizzato:  
$p_\theta = ml^2 \dot{\theta}$  

invertilo:  
$\dot{\theta}=\frac{p_\theta}{ml^2}$  

hamiltoniana:  
$H(\theta, p_\theta) =  p_\theta (\frac{p_\theta}{ml^2}) - L(\theta, \frac{p_\theta}{ml^2}) = p_\theta (\frac{p_\theta}{ml^2}) - \frac{1}{2}ml^2(\frac{p_\theta}{ml^2})^2+mgl \cos \theta$

**Quando si può fare il passaggio?**

in teoria:  
quando la trasformazione è NON degenere   

in pratica:  
prova a farlo, se al punto 3 non riesco ad invertire allora non si può fare  

**Esempio di fallimento**

$H(q,p)=p$   
al punto 2 ottengo: $\dot{q}=1$  
quindi non riesco ad invertire per ottenere $p(\dot{q})$    
quindi non posso trovare L   

**Trasformazione degenere -DELETE???**

quando è NON-degenere:  
quando la hessiana (di H rispetto a p o di L rispetto a $\dot{q}$) sia non-nulla, cioè:  
per fare lagrangiana -> hamiltoniana:  $det (\frac{\partial^2 L}{\partial \dot{q}^2}) \neq 0$  
per fare hamiltoniana -> lagrangiana: $det (\frac{\partial^2 H}{\partial p^2}) \neq 0$  

perchè:  
se viene non-0, vuol dire che alla prima derivata ancora apparivano la p e la $\dot{q}$, cioè che la posso invertire, che è ciò che voglio fare  
infatti viene 0 solo se alla derivata prima c'era già un termine noto, non più variabili  

ma quindi è una proprietà solo di L o di H, non della coppia??? come è possibile???

#### Parentesi di Poisson

**Trasformazione canonica**

ho: Q(q,p) e P(q,p)   
la trasformazione è canonica se:  
$\{Q,Q\}=\{P,P\}=0$  
$\{Q,P\}=1$  

non appare il $\delta$ perchè c'è una sola particella  
infatti le coordinate originali sono solo 2, p e q, non ci sono $p_1$ e $p_2$  
quindi non devo preoccuparmi di quando parlo della stessa particella e quando no  

se ho $q(Q,P)$, posso anche applciarla al contrario, cioè calcolare $\{q,p\}_{Q,P}$  

**Steps**

1. calcola le derivate parziali delle nuove variabili in termini delle vecchie:  
$\frac{\partial P}{\partial p}, \frac{\partial P}{\partial q}, \frac{\partial Q}{\partial p}, \frac{\partial Q}{\partial q}$  

2. se c'è una sola particella (cioè se sono in 1D) $\{Q,Q\}=\{P,P\}=0$ è ovvio  
perchè se la scrivo per esteso è sottrazione di 2 addendi uguali  

3. scrivo per esteso $\{Q,P\}$, sostituisco le cose del punto 1 e vedo se mi ridà =1  
$\{f,g\} = \Sigma (\frac{\partial f}{\partial q}\frac{\partial g}{\partial p} - \frac{\partial f}{\partial p}\frac{\partial g}{\partial q})$  


**Con più var**

$\{Q_i,Q_j\}=0$  
$\{P_i, P_j\}=0$   
qui quelli con $i=j$ sono ovvi, controlla solo i casi tra particelle diverse   

$\{Q_i,P_i\}=\delta_{ij}$
