### Punti di equilibrio stabile

**Steps**

1. punti di equilibrio: $\nabla V (x,y)=(0,0)$   

2. trova quello stabile, 2 metodi:  
    - cosa: $V(x,y)$ è un minimo, vedilo a occhio  
    quando: utile con le molle, quando ho cose al quadrato, quindi il minimo è quando tutto è 0  

    - cosa: la hessiana in quel punto $H_{(x,y)}$ è definita positiva  
    quando: tutti gli altri casi  

#### Punti di equilbrio

**Minimo di V a occhio**

graficamente: sono i punti che se li metti con velocità nulla, restano lì   
se c'e solo una molla che dà il potenziale, i punti dove o la molla ha la lunghezza di riposo  
o dove la forza è ortogonale alla guida in quel punto, quindi non muove le masse  

analiticamente: guarda la formula di V, come si minimizza?  

**Esempio: Trova minimo di V a occhio**

$V = k\frac{q^2}{R\sqrt{5-4\cos(\theta_1-\theta_2)}}$    

minimo è per $\cos = 1$, quindi $\theta_1=\theta_2$  
massimo è per $\cos = -1$, quindi $\theta_2 = \theta_1 + \pi$  

trova invarianze:  
V dipende solo dalla differenza $\theta_1-\theta_2$, cioè c'è invarianza sotto rotazioni globali (se fanno entrambi un giro completo, posos sommare $2 \pi$ a entrambe e V non cambia) (ma NON è per forza un modo zero)      
quindi esistono una serie di punti di equilibrio stabili per $\theta_1 = \theta_2$, legati fra loro da rotazioni globali  

**Esempio: Trova punti di equilibrio con hessiana**

$V=-mgl(\cos \theta +\cos \varphi)$

tutti i punti di equilibrio:  
$\nabla V=(mgl \sin \theta, mgl \sin \varphi)=(0,0)$  
quindi i punti di equilibrio sono con gli angoli a $0$ o $\pi$  

punti di equilibrio stabili:  
calcolo le derivate e scrivo la hessiana:  
$H=\begin{bmatrix} mgl \cos \theta & 0 \\ 0 & mgl\cos\varphi \end{bmatrix} = mgl \begin{bmatrix} \cos \theta & 0 \\ 0 & \cos\varphi \end{bmatrix} $  

deve essere definita positiva, quindi avere entrambi gli autovalori positivi  
tra le opzioni di prima, solo $\theta=\varphi=0$

**Teoria: punto di equilibrio**

condizione di equilibrio: equazione di stazionarietà  
$\nabla V (x,y)=(0,0)$   

definizione:  
un punto (x,y) dove le derivate parziali di V (non di L) rispetto ad ogni coordinata sono nulle  
cioè se lo metto lì con velocità nulla, rimane lì  

#### Hessiana

**Come si scrive**

calcola le derivate parziali seconde di V, sia uniche che miste  
nota che è simmetrica  

$H = \nabla^2 V = \begin{bmatrix} \partial^2_x V & \partial _{xy} V  \\ \partial _{xy} V & \partial ^2_y V \end{bmatrix}$  

**È definita positiva?**

metodo 1: tutti gli autovalori sono positivi  
utile se è già diagonale, metti che tutti gli elementi sono positii  

metodo 2: det>0 e almeno un elemento positio sulla diagonale principale  
utile se matrice 2x2, facile calcolare det  

**Test dei punti di equilibrio**

definita positiva: V minimo locale, stabile (come quando a>0)   
definita negativa: V massimo locale, instabile    
indefinita: punto di sella, instabile (sia autovalori + che -)  

se ci sono autovalori 0: il suo autovettore è un modo zero???  
lungo quella direzione la curvatura del potenziale è nulla: il potenziale è piatto al secondo ordine  

### Scrivere matrici V e T

**Scrivere T**

le voglio calcolate nel punto di equilibrio stabile, quindi:  

1. riprendi $T(q)$    

2. se non è una forma quadratica, ma ho trig, espandi attorno al punto di equilibrio stabile  

3. le caselle sulla diag principale contengono il quadrato di una velocità, le altre i prodotti misti  
poichè devono essere simmetriche, per i prodotti misti metti metà del coeff in una casella e metà nell'altra  

**Scrivere V**

se è una forma quadratica:   
come faccio per la T, metti i coeff al posto giusto  
in questo caso, come per la T ignora cancella un fattore 1/2 (viene dal fatto che in teoriafarei la derivata) (se non c'è aggiungi 2/2 davanti, così cancello 1/2, in pratica ho moltiplciato per 2)  

se non lo è:  
$V=V''=H$ (V in punto di equilibrio STABILE è la hessiana in quel punto)  

**Attenzione quando scrivo**

1. porta fuori più costanti che puoi, come coeff di tutta la matrice  

2. in T, devo ignorare l'1/2 davanti all'intera matrice  
ma SOLO QUELLO, se c'è un altro coefficiente che viene tipo dalla I, QUELLO CI DEVE RIMANERE, sennò mi vengono le frequenze sbagliate  

3. entrambe devono essere simmetriche  
quindi sulla diagonale non principale metti metà del termine misto in ogni casella  

4. l'ordine delle coordinate, $q_1$ e $q_2$; è importante, ricordatelo  
dovrà essere sempre uguale  

**Suggerimento**

se le coordinate sono di tipo diverso, tipo una lineare e una angolare, $(x, \theta)$:  
non riesco ad ottenere $T=I$ perchè mi viene la R in corrispondenza della coordinata angolare   

invece, posso scgliere come coord generalizzate $(\mu_1, \mu_2)=(x, R\theta)$  
così ho $T=I$  

**Nomenclatura**

T, M: matrice cinetica, matrice dele masee generalizzate    
U, K: matrice potenziale, hessiana, matrice di rigidità, matrice di forza  

**Tip**

so per piccoli spostamenti $\sin \theta \sim \theta$  
ma se invece voglio usarlo nel senso inverso, so anche $\theta \sim \tan \theta$  
utile se ho asta che ruota, quindi so $\theta$ angolo di rotazione dell'asta, ma ho come coord gli spostamenti ortogonali $y_1$ e $y_2$ delle sue estremità   
$\theta = \tan \theta = \frac{\Delta y}{\Delta x} = \frac{y_1-y_2}{2}$  


#### Esempi

**Scrivo V: con hessiana**

$V=-mgl(\cos \theta +\cos \varphi)$  
punto di equilibrio stabile: $(0,0)$  

calcolo la hessiana nel punto di equilibrio stabile (in generale, l'avevo già calcolata prima):     
$H= mgl \begin{bmatrix} \cos \theta & 0 \\ 0 & \cos\varphi \end{bmatrix} $  
$V=mgl \begin{bmatrix} 1& 0 \\ 0 & 1 \end{bmatrix}$  

**Scrivo V: forma quadratica**

$V= \frac{1}{2}\frac{2kq^2}{R}(\theta_2-\theta_1)^2 = \frac{1}{2}\frac{2kq^2}{R} (\theta_1^2+\theta_2^2-2\theta_1\theta_2)$  

come faccio per la T:  
$V =\frac{2kq^2}{R}\begin{bmatrix} 1 & -1 \\ -1 & 1 \end{bmatrix}$  

**Scrivo T**

$T=\frac{1}{2}ml^2 (3\dot{\theta}^2+\dot{\varphi}^2+2\dot{\theta}\dot{\varphi})$  
$T=ml^2 \begin{bmatrix} 3 & 1 \\ 1 & 1\end{bmatrix}$  

ricorda di dividere a metà il prodotto misto  
con coord angolari, l potrebbe essere R  

### Piccole oscillazioni su eq stabile

**Steps**

1. trova modi normali  

2. trova frequenza per ogni modo normale  

**Scegliere metodo**

1. modo zero:    
    - se c'è simmetria continua (rotazioni/traslazioni globali) (ad esempio se c'è equilibrio ogni volta che due angoli sono uguali)     
    - se in L c'è una variabile ciclica  (L(x+c)=L(x))
    - se $det \; V = 0$  
2. metodo delle simmetrie:   
    - se c'è simmetria discreta (permutazione tra particelle, scambio ciclico)     
    - se T=I      
3. metodo lungo: negli altri casi  

**Nota**

se c'è un modo zero + permutazione tra TUTTE le var:  
se uso metodo delle simmetrie, apparirà naturalmente il modo zero, non devo cercarlo a parte  

se c'è modo zero + permutazione tra QUALCHE var:  
devo trovare esplicitamente il modo zero, e per primo  

**Steps: metodo delle simmetrie**

1. cerca simmetria continua -> modo zero  
scrivi il suo autovettore  

2. cerca simmetrie discrete   
frequenze NON degeneri: trova autovettori, sono modi normali  
frequenze degeneri: prendi una base qualunque, purchè sia T-ortogonale a tutti i modi trovati finora  

RICORDA DI T-NORMALIZZARE I MODI  
tipo (1,1) e (1,-1), se T è multiplo di I, vanno divisi per sqrt(2)  

#### Metodo lungo  

**Steps**

$det(V-\omega^2T)v=0$  

1. porta tutte le costanti davanti alla matrice T, e riuniscile in una sola, chiamata $\lambda$  

2. sottrai le due matrici  
così ho una matrice con $\lambda$ all'interno, simile a quando calcolo il char poly  

3. trova gli autovalori  
calcola il det della matrice trovata, imponilo uguale a 0, trova valori di $\lambda$ che lo permettono

4. per ogni $\lambda$, calcola:  
$\omega$, usando formula dello step 1  
$v$, col metodo sotto  

**Modi normali**

nota che: $v = \begin{bmatrix} v_1 \\ v_2 \end{bmatrix}$  

1. espando un'equazione a scelta di $[...]v=0$  
[...] è la matrice con dentro $\lambda$ ottenuta dopo lo step 2.2    
ottengo tipo $(...)v_1 - (...)v_2=0$  

2. scrivo la seconda in funzione della prima: $v_2 (v_1)=...$  

3. scelgo una componente a caso, tipo $v_1=1$, e trovo la seconda dal rapporto di prima  
così ho l'intero autovettore riferito a quella frequenza  



**Se T proporzionale all'identità: $T \sim I$**

$v$ sono gli autoettori di V (con autovalore $\lambda$)  
$\omega^2 =  \frac{\lambda}{m}$ (quindi $\omega$ non è l'autovalore in V)

perchè:  
voglio risolvere: $(V-\omega^2T)v=0$  
tutti i vettori sono autovettori di I, quindi per essere vera la condizione è che:  
v deve essere autovettore di V  

#### Esempio: metodo lungo

**Risolvo**

1- scrivo l'equazione:  
$U-\omega^2 T =  ml^2 \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix} -\omega^2  \; mgl\begin{bmatrix} 3 & 1 \\ 1 & 1\end{bmatrix}  = \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix} -\lambda \begin{bmatrix} 3 & 1 \\ 1 & 1\end{bmatrix}  = \begin{bmatrix} 1-3\lambda & -\lambda \\ -\lambda & 1-\lambda \end{bmatrix}$

dove: $\lambda = \omega^2 \frac{ml^2}{mgl} = \omega^2 \frac{l}{g}$   

2- cerco $\lambda$ che mi dia det=0  
sono: $\lambda=\frac{2\pm \sqrt{2}}{2}$

**Trova modi normali**

$\begin{bmatrix} 1-3\lambda & -\lambda \\ -\lambda & 1-\lambda \end{bmatrix} \begin{bmatrix} v_1 \\ v_2 \end{bmatrix}=0$  

ne espando una a scelta, scelgo la seconda riga:  
$-\lambda v_1 +(1-\lambda)v_2 =0$  
$v_2 = \frac{\lambda}{1+\lambda}v_1$  

setto $v_1=1$ e pluggo $\lambda$:  
$(1, \frac{\lambda_\pm}{1+\lambda_\pm}) = (1, -1\mp \sqrt{2})$


#### Metodo delle simmetrie

**Steps**

1. cerco una trasformazione che descrive una simmetria    

2. cerco quante volte va usata per riavere l'identità    

3. trovo le $\alpha$  

4. trovo gli autovettori  
se sono non degeneri sono modi normali, sennò non so  

5. trovo le frequenze  
pluggo v in $(V-\omega^2T)v=0$, trovo $\omega$ che lo rispetti  

nota che qui L significa una trasformazione lineare, NON la lagrangiana  

'**1. Esempi di trasformazioni che descrivono una simmetria**

scambi di variabili:  $L(q_1, q_2) = (q_2, q_1)$  
rotazione ciclica / permutazioni : $L(v_1, v_2, v_3) = (v_3, v_2, v_1)$  

come:  
a occhio  
o guarda se scambiando due var nella lagrangiana, se quella cambia  

**2. Trova identità**

in questi esempi, quante volte va applicata L (lo scambio) per tornare al punto di partenza  

$L(q_1, q_2) = (q_2, q_1) \rightarrow L^2(q)=q$   
$L(v_1, v_2, v_3) = (v_3, v_1, v_2) \rightarrow L^3(v)=v$    

**3. Trovo le $\alpha$**

1. $L^n(v)=\alpha^n v = v$    
l'identità: $L^n(v)=v$   
v è un autovettore di L: $Lv=\alpha v$   

2. risolvi per $\alpha$: è l'autovalore di L, NON la frequenza dei modi  

**4. Modi normali**

1. per ogni autovalore di L trovato, a occhio trovo il suo autovettore   
è facile, sono tipo (1,1) o (1,-1)  

2. se mi mancano autovettori per fare una base:  
vuol dire che ci sono dei modi degeneri  
scelgo un altro vettore, basta che sia ortogonale agli altri  

nota: il metodo delle simmetrie lo uso se T=I  
quindi T-ortogonale = ortogonale  

**Simmetria per permutazione**

considera una delle simmetrie del gruppo  

metodo 1: considera 2 scambi  
per trovare tutti gli autovalori, potrei considerare come simmetrie separate gli scambi di 2 masse  
cioè se ho permutazioni di 4 masse, considero scambio 1-3 e scambio 2-4  
ognuno ha autovalore $\pm \alpha$, quindi trovo 2 autovettori per ognuno, in totale ho i 4 autovettori richiesti  

metodo 2: considera la rotazione ciclica  
ha identità, $L^nv=\alpha^n v=v$, e la radice di 1 ha sempre tutte radici diverse, quindi tutti modi normali non degeneri   
trova tutti gli autovalori: $\alpha=e^{i2\pi k/n}$, formano un poligono regolare  
trova tutti gli autovettori: $v=(1, \alpha, \alpha^2, .., \alpha^{n-1})$, perchè $v_2=\alpha v_1, v_3=\alpha^2 v_1...$ [visto che $Lv = \alpha v$, e $L(x_1, x_2)=(x_2, x_1)$]   
tip: se ho due auovettori complessi coniugati, con stessa frequenza, sostituiscili con Re(v)e Im(v), così sono reali (mantengono stessa frequenza di prima)  
perchè combinazione lineare di modi normali con stessa frequenza dà un modo normale con stessa frequenza  

#### Esempio: metodo delle simmetrie

**Permutazione/scambio di 2 variabili**

trasformazione: $L(q_1, q_2) = (q_2, q_1)$    
identità: $L^2(q)=q$    
$L^2(v)  = \alpha^2 v = v$   
autovalori: $\alpha = \pm 1$   
modi normali: (1,1),(1,-1)  

**Permutazione di 3**

rotazione ciclica: $L^3v=\alpha^3v=v$  
autovalori: radici terze di 1, $\alpha = 1, e^{i2\pi/3}, e^{i4\pi/3}$  
autovettori: $v=(1,\alpha,\alpha^2)$   
i due complessi coniugari, li sostituisco con parte reale e immaginaria:(1,-1/2, -1/2), $(0, \sqrt{3}/2, -\sqrt{3}/2)$

altro metodo:  
prendo solo $\alpha=1$, reale  
autovettore:(1,1,1)  
trova modi degeneri: prendo una qualunque coppia ortogonale a quello, (1,-1,0) e (0,1,-1) ???VALE???

**Permutazione di 4**

rotazione ciclica: $L^4v=\alpha^4v=v$  
autovalori: le radici quarte di 1, cioè $\alpha = 1,i,-1,-i$ (i vertici di un poligono)   
autovettori: $v=(1, \alpha, \alpha^2, \alpha^3)$, quindi (1,1,1,1), (1,i,-1,-i), (1,-1,1,-1), (1,-1,-1,i)  
i due complessi coniugari, li sostituisco con parte reale e immaginaria: (1,0,-1,0), (0,1,0,-1), con la stessa frequenza di quelli complessi (oppure li sommo e sottraggo)  

**Intuire la degenerazione**

motivo matematico:  
Quando compaiono autovettori complessi coniugati, indicano due soluzioni indipendenti con la stessa frequenza (quindi degeneri)  
perchè quando vado a calcolare la frequenza, viene uguale???  

motivo fisico:  
il gruppo di simmetria non è commutativo  

**Nota**

un autovettore è un modo normale solo se è l'unico possibile  
ad esempio, se ho la funzione: (1,2,3,4)->(3,2,1,4), che ha identità $L^2$. allora ho $\alpha = \pm 1$   
però a $\alpha=+1$ corrispondono più autovettori, perchè vanno bene (1,0,1,0) o (1,1,1,1) ...  
quindi nessuno di questi è sicuramente un modo normale 

nota che però è unico per $\alpha = -1$, quindi intanto trova quello  
per il restante, o uso un modo zero, o uso il fatto che deve essere T-ortogonale a tutti gli altri

#### Metodo del modo zero

**Steps**

1. trova la simmetria e scrivine il vettore  

3. trova gli altri vettori per fare una base  
i vettori devono essere T-ortogonali (normalizzati ci penserò dopo)  

4. trova le frequenze:  
come nel metodo delle simmetrie, ho gli autovettori, pluggali  

**Step 1: trova la simmetria**

simmetria globale per traslazioni   
esempio: due masse su due guide parallele, collegate da una molla  

simmetria per rotazione continua  
esempio: due traiettorie circolari concentriche  

l'autovettore è (1,1) per entrambi  
per la traslazione, trasla di uno stesso dx  
per la rotazione, vuol dire ruota di uno stesso angolo  

nota: per questo non serve che le masse siano uguali  
perchè voglio solo una simmetria di V, non anche di T  

**Step 3: altri vettori**

T-ortogonale: $v_1^t T v_2=0$  

1- vettore con param generici,  $v_2=(a,b)$   
2- calcola, come in algebra lineare $am_1 + bm_2=0$  
3- setta a=1, poi trova il valore di b

### Sol con specifica configurazione iniziale

#### Metodo 1: generale e lungo

**Trova sol generale**  

1. scrivi condizioni su posizione e velocità iniziale  

1. scrivi $q(t)$ generale con incognite      

2. deriva $q(t)$ per trovare $\dot{q}(t)$    

3. plugga le condizioni iniziali per trovare A,B,C,D  

**Scrivere la soluzione generale**

$q,v$ sono vettori colonna, lunghi il numero di coordinate generalizzate  
per avere $q_i$, prendi il valore $v_i$  
$\begin{bmatrix} q_1(t) \\ q_2(t)\end{bmatrix} = q_{eq} + v_1 \eta_1 + v_2 \eta_2$   

per ogni v, il suo $\eta$ è:  
- se $\omega \neq 0$: $\eta =  A \cos(\omega t) + B\sin(\omega t) = A \cos(\omega t + \phi) $  
- se $\omega=0$: $\eta(t) = (A+Bt)$      
per modo zero: usa questo, non pluggare $\omega=0$ nell'altro 

$q_{eq}$: la posizione di equilibrio. può essere diversa nelle due righe  
di solito equilibrio è a (0,0), quindi lo ignoro  
ma se uso la simmetria, potrei trovare equilibrio in altri pisti, allora lo metto  

#### Metodo alternativo

**Steps**

1. normalizzo i modi normali  

2. trovo i coefficienti $c$  
$c_i = v_i^t \, T \, q(0)$  

3. scrivo la formula di $q_i(t)$ pluggandoci c e v, ed estraggo la parte reale   
pluggare $\omega$ non serve, tanto non si semplifica con niente  


**Come estrarre la parte reale**

$e^{-i \omega t} = \cos (\omega t) + i \sin (\omega t)$   
$c \; e^{-i \omega t} = c [\cos (\omega t) + i \sin (\omega t)]$ 

quindi di solito:  
$Re(c \; e^{-i \omega t}) = c \; \cos (\omega t)$   

se $c$ o $v$ sono COMPLESSI:    
in tal caso avrei una $i$ moltiplicata alle funzioni trigonometriche  
che quindi cancella la $i$ al sin e la fa apparire al cos     
quindi la parte reale diventerebbe il seno  

**Come normalizzare un vettore**

ho:  
$v$ che voglio normalizzare  
la matrice T (senza le costanti davanti)  

1. calcola la norma di v, rispetto al prodotto scalare descritto da T  
cioè fai la moltiplciazione matriciale: $|v|^2= v^t T v$  

2. dividi v per la RADICE della sua norma:  
$v_{norm}= \frac{v}{|v|}$  


**Esempio: normalizzare vettori**

$T=ml^2 \begin{bmatrix} 2 & 1 \\ 1 & 1 \end{bmatrix}$  
$v= (1, + \sqrt{2})$   

ortonalizzo rispetto a $\frac{T}{ml^2}$:  
$|v|^2 = v^t T v = 2(2+\sqrt{2})$  
$v_{n} = \frac{1}{\sqrt{2(2+\sqrt{2})}}(1, \sqrt{2})$  

**Trova i coeff $c_1, c_2$**

dopo aver ortonormalizzato i modi, so che:  
$c_i =v_i^t \; T \;  q(0)$  

dove:  
$q(0)$: il vettore colonna con le condizioni iniziali sulla posizione  
$v_i$: un vettore dei modi normali T-ORTONORMALIZZATO  
T: la matrice  

è una moltiplicazione di cose, tipo altra lineare  
falla, trova $c_+$ e $c_-$ (usando $v_+$ e $v_-$)  

nota:   
$q(0)$ è il vettore delle posizioni perchè voglio sapere i coefficienti per scrivere $q_i(t)$  
se metto $\dot{q}(0)$ ottengo i coefficienti per scrivere $\dot{q}_i(t)$

#### Condizione di piccole oscillazioni

**Validità del regime di piccole oscillazioni**

metodo 1:  
$x_1(t)-x_2(t) \ll 1$  
se 1 e 2 sono masse m collegate da una forza interna, tipo di Coulomb o una molla  
o se fare le derivate di V è troppo difficile  

metodo 2:  
piccole oscillazioni: $V=V''$, come un oscillatore armonico  
quindi condizione è che il terzo ordine dello sviluppo di taylor (calcolato nel punto di massima ampiezza, così è vero in ogni caso) sia molto più piccolo del secondo  
$\frac{1}{3!}V^{(3)}_{eff}(R) \, \Delta R^3 \ll \frac{1}{2!} V^{(2)}_{eff}(R) \, \Delta R^2$  
ottengo tipo: $\frac{\Delta R}{R} \ll 3$

#### Esamina casi quando mg >> kl

**Steps**

1. trova variabile di supporto per descrivere i casi  
2. spandila con Taylor, un'espansione per ogni caso  

**1. Variabile di upporto**

se devo esaminare casi dove:  $mg \gg kl$  
probabilmente, mentre scrivevo la hessiana, ho usato una variabile di supporto, tipo $\alpha = \frac{mg}{kl}$   
ora devo distinguere i casi $\alpha \gg 1$ e $\alpha \ll 1$

tip:  
questo è un indizio  
se l'ultima domanda mi chiede di studiare quando $mg \gg kl$, probabilmente prima è comodo far apparire $\alpha = \frac{mg}{kl}$ da qualche parte, tipo nella hessiana    

**2. Espandere**

probabilmente $\alpha$ appare nella formula degli $\omega^2$   
devo espandere $\alpha$ con Taylor, a seconda del caso che sto considerando  

se nella formula di $\alpha$ c'è un radice:  
1- cerca di trovare $\sqrt{1+x}$ con $x \sim 0$  
2- espandi come  $\sqrt{1+x} \sim 1+ \frac{x}{2}$   
così viene diverso da semplicemente cancellare a o i termini noti, fai in questo modo  


se $\alpha \ll 1$:  
$\sqrt{(\alpha +1)^2+48} = \sqrt{49+2\alpha+\alpha^2}= 7\sqrt{1+\frac{2\alpha + \alpha^2}{49}}$  
in pratica: $\alpha \sim 0$, sparisce  

se $\alpha \gg 1$:  
$\sqrt{(\alpha +1)^2+48}  =(\alpha +1) \sqrt{1+\frac{48}{(\alpha +1)^2}}$  
in pratica: cancello i termini noti 

**Nota**

la cosa richiesta potrebbe essere i casi: $d>R, \; d<R, \; d=R$  
usa tutti e 3 i segni  
fai apparire e sostituisci $\alpha=\frac{d}{R}$ in tutte le formule 

#### Disaccoppia i moti

**Cosa**

trova coordinate q' per cui la legge del moto è disaccoppiata  
quindi oplugga $q(q')$ in tutte le formule  
alla fine posso pluggare la cosa inversa, e tornare alle coord iniziali  

**Trova coord che disaccoppiano**

mi servono:  
1- coordinate generalizzate: $x_1, x_2$  
2- modi normali: $v_+ = \frac{1}{\sqrt{2}}(1,1), \quad v_- =\frac{1}{\sqrt{2}}(1,-1)$  
posso anche ignorare il fattore di normalizzazione al denominatore, va bene comunque  

come si fa:  
$\begin{bmatrix} \varphi \\ \theta \end{bmatrix} = q_1 v_1 + q_2 v_2$  
dove   
$\varphi, \theta$: sono le vecchie coord  
$q_1,q_2$: sono le nuove coord  
e $v_1,v_2$: sono vettori, i modi normali trovarti (anche non normalizzati)  
così ho già le cose da pluggare  

2 modi di disaccoppiare:  
1- plugga nella lagrangiana, ricalcola le eq. di eulero-lagrange  
2- se ho già calcolato le eq. di eulero-lagrange, plugga in quelle  
poi **una volta sommale e una volta sottraile, e ottengo le equazioni disaccoppiate**   

per avere trormazione inversa, $q'(q)$:  
definisco delle nuove coordinate, pluggando quelle vecchie nelle componenti dei modi:  
$q_1 = \frac{x_1+x_2}{\sqrt{2}}, \quad q_2 = \frac{x_1-x_2}{\sqrt{2}}$   

**Altre coord che disaccoppiano**

se ho 2 oggetti, è comodo considerare centro di massa e coordinata relativa:  
$q_1 = \theta-\varphi$  
$q_2 = \frac{m_1 \theta + m_2 \varphi}{m_1+m_2}$  

conseguenza:  
se trovo $\ddot{q}_1=0 \quad \mu \ddot{q}_2 = -2k \sin q_2$  
il sistema è composto da un oggetto libero (che si muove di moto rettilineo uniforme) e da un pendolo semplice (con la massa ridotta $\mu$)  

altre coord che disaccoppiano:  
$q_2 = \frac{I_1 \varphi + I_2 \theta}{I_1+I_2}$  
dove tipo $I_1 = mR^2$, quindi le R potrebbero essere diverse  

**Quale coord usare**

prendi quelle più facili da pluggare in L  
tipo, se in L c'è $\cos (\theta-\varphi)$, prendi cm e realtiva, così diventa $\cos q_1$
