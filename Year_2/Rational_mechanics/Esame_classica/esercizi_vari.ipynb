### Variabile ciclica / momento coniugato è costante

**Quando un momento coniugato è costante**

q variabile ciclica -> $\dot{p}_q=0$ -> $p_q$ momento costante  
cioè se $q$ non appare mai in L o H (le sue derivate invece possono apparire)  
perchè se non appare, il lato destro della sua equazione di lagrange è 0 (langrangiana) o l'equazione di hamilton dà 0  

**Cosa fare se ho variabile ciclica**

1. calcolo $p$ (eq di Hamilton) e gli dò un nome diverso, per ricordarmi che è una var fissa  
tipo se mi sembra un momento angolare, lo chiamo $l$  

2. lo sostituisco nel posto giusto:  
    H: lo sostituisco direttamente in H, riduco la dimensionalità    
    L: NON posso metterlo in L, ma solo nell'eq di Eulero-lagrange (sennò dà problemi coi segni)  

### Cose conservate

**Riconoscere conservate**

$H$ è conservata: se H non dipende esplicitamente dal tempo  
$p_y$ è conservata: se y è ciclica  
$H_1$ e $H_2$ sono conservati: se a occhio, H è disaccoppiata  

più difficile:

allora noto che $\dot{p}_x-m\omega_c \dot{y}=0$, cioè $p_x-\omega_c y$ è conservato (perchè la sua deruvata è nulla)  
come accorgermene???  

**Sistema disaccoppiato**

se posso separare H in addendi che contengono solo $q_1, p_1$ e quelli che contengono solo $q_2, p_2$ il sistema è disaccoppiato (cioè non devo avere tipo $\cos(q_1-q_2)$)  
se è disaccoppiato posso fare $H=H_1+H_2$  

se H si conserva (perchè non dipende esplicitamente dal tempo):  
se è disaccoppiato si conservano anche $H_1$ e $H_2$, perchè non possono interagire tra loro  
mentre se non è disaccoppiato non posso dire che nient'altro si conserva  

#### Momento angolare

**Formula del momento angolare**

se ho una coord: $L=qp$   
se ho 2 coord (quindi L su un piano): $L=q_1p_2-q_2p_1$  
attenta a non accoppiare gli indici negli addendi    
per vedere se è conservato, fai la derivata dL/dt e plagga le eq di Hamilton nelle cose con la derivata  

**Conseguenze: momento angolare costante**

$\vec{l}=\vec{r} \land \vec{p}$  
quindi l è ortognale al piano formato da r e p, quello su cui avviene il moto    

se l è costante, lo è anhe la sua direzione  
quindi lo è anche il piano a essa ortogonale, cioè quello su cui avviene il moto  
quindi se l è costante, il moto è planare   

nota che si conserva se $\dot{l}=0$

#### Teoria

**Definizione**

l’oscillatore armonico rappresenta il comportamento locale di qualsiasi sistema vicino a un equilibrio stabile (Piccole oscillazioni di un oggetto in un potenziale qualunque)   
cioè è qualunque cosa con equazione del moto $\ddot{x}=-\omega^2 x$

cosa può essere:  
molla  
pendolo, ma SOLO in regime di piccole oscillazioni  

**Teoria: espandere potenziale intorno a punto di equilibrio stabile**

definizione punto di equilibrio stabile $x_0$:  
$\partial_r V(r) |_{x_0}= 0$  
$\partial_r^2 V(r)|_{x_0} > 0$  

allora:  
$V(x)=V(x_0) + \partial_r V(r) (x-x_0)+ \frac{1}{2} \partial_r^2 V(r) (x-x_0)^2$  
potenziale è definito a meno di una costante -> posso cancellare la costante $V(x_0)$  
per definizione di punto di equilibrio, la correzione al primo ordine è nulla  
quindi serve la correzione al secondo ordine  
ignoro i termini dal terzo grado in su, visto che sono trascurabili  
$V(x)= \frac{1}{2} \partial_r^2 V(r) (x-x_0)^2$  

se invece di V metto $V_{eff}$ è uguale  

quindi $V= \frac{1}{2}V''x^2 $

### Orbite circolari

#### Come risolvere

**Steps: studia orbita circolare**

1. esistenza: $V_{eff}'=0$       
2. stabilità:  $V_{eff}''>0$  
3. periodo: $T= \frac{ 2 \pi}{\dot{\varphi}}$  
4. chiusa (spesso con piccola perturbazione radiale)  

**1. Esistenza e stabilità**

come per i normali punti di equilibrio stabile:  

punto di equilibrio (esistenza): gli $r_0$ tali che $\partial_r V_{eff}(r)|_{r_0}=0 \quad$ ($\nabla V = 0$)  
stabilità: i punti di equilibrio che sono punti di minimo, cioè $\partial^2_{r}V_{eff}(r)>0$  

mentre calcolo se è un punto di equilibrio segnati:  
$\dot{\theta}(r)$ -> mi servirà per calcolare il periodo  
$l^2(r)$ -> mi servirà per calcolare la derivata seonda (tipo $l^2=4m\lambda r^6$)

**3. Periodo**

$T= \frac{ 2 \pi}{\dot{\varphi}}$  

$\dot{\varphi}$ lo prendo dall'eq di Hamilton, è tipo:  
$\varphi^2 = \frac{l^2}{mr^2}$   

se orbite circolari:  
per esprimere T solo in funzione di $r$, posso trovare $l(r)$ dalla condizione di orbite circolari e pluggarlo dentro  


**4. Chiusa**

è chiusa se il rapporto tra frequenze angolare e radiale è razionale  
quindi il rapporto deve essere costante e indipendente dall’ampiezza, ciò accade solo per alcuni potenziali particolari    
se non c'è perturbazione radiale, $\omega_r=0$, quindi è chiusa?  

**Piccola perturbazione radiale**

cosa succede:  

- $L$ è conservato, è uguale a prima   
perchè l'impulso dato è in direzione radiale, non cambia il momento angolare  

- il raggio non è più $r_0$, quello di equilibrio stabile senza perturbazione, ma oscilla intorno a quel punto   

cosa fare:  
il comportamento locale di qualsiasi sistema vicino a un equilibrio stabile è un oscillatore armonico    
quindi studia l'espansione del potenziale per sapere quanto vale $\omega$, dove rinomino: $V(r)=\frac{1}{2}k(\Delta x)^2$

angolare: non cambia, viene dalla T di prima ($\omega_\theta=\frac{2 \pi}{\dot{\theta}}$)

#### Trovare la formula

**Trovare la formula**

1. scrivi $E=T+V$  

2. isola $\dot{x}=\frac{dx}{dt}$, presa da T

4. sostituisci:  
$V=V(x)$, la sua formula con la var x  
$E=V(x_0)$, è un valore costante, V calcolato nel punto di inversione

**Esempio: pendolo**


1- scrivo le energie  
$T=\frac{1}{2}m(l\dot{\theta})^2$  
$U=-mgl \cos \theta$   
$E=\frac{1}{2}m(l\dot{\theta})^2 -U$  

2- trovo la derivata  
$\dot{\theta}^2 = \frac{2}{ml^2}(E-U)$  

3- integra  
$T = \int_{t1}^{t2} dt = \int_{\theta 1}^{\theta 2} \frac{d\theta}{\dot{\theta }} =   \int_{\theta 1}^{\theta 2} \frac{d\theta}{\sqrt{ \frac{2}{ml^2}(E-U)}} = 2 \int_{-\theta_0 }^{\theta _0} \frac{d\theta}{\sqrt{ \frac{2}{ml^2}(E-U)}}$  

4- sostituisci  
$T=2 \int_{-\theta_0 }^{\theta _0}\frac{d\theta}{\sqrt{\frac{2}{ml^2} (-mgl \cos \theta_0 + mgl \cos \theta)}} =4 \int_{0}^{\theta _0} \frac{d\theta}{ \sqrt{2\frac{g}{l} (\cos \theta - \cos \theta_0)}}$  
se voglio, posso chiamare $\frac{g}{l}=\omega^2$   
dove $\theta_0$ è l'ampiezza di oscillazione

$T = 4 \int_0^{x_0} \frac{dx}{\sqrt{\frac{2}{m}\lambda x_0^p -\lambda x^p}}=$   
$= \frac{4}{\sqrt{\frac{2}{m}(\lambda x_0^p)}} \int_0^{x_0} \frac{dx}{\sqrt{1-(\frac{x}{x_0})^p}} = $  
$= \frac{4}{\sqrt{\frac{2}{m}(\lambda x_0^p)}} \; x_0 \int_0^1 \frac{dy}{\sqrt{1-y^p}} =$  
$= \frac{4}{\sqrt{\frac{2}{m}(\lambda x_0^p)}} \; x_0 \; I_p$    


**Spiegazione**

nota: ricorda che basta fare la sotituizoine per ottenre 0 e 1 come estremi di integrazione, il mio oviettivo è quello  

$I_p$ è semplicemente un numero, gli dò un nome a caso  
lo faccio per estrarre la dipendenza dall'energia, così il valore dell'integrale dipende solo da p, lo chiamo $I_p$  
tanto se sto studiando qualcosa relativo all'energia, poi $I_p$ si cancellerà  

**Altro esempio di potenziale generico**

$E=\frac{p^2}{2m}+A|x|^p \rightarrow p(x) =\sqrt{2m(E-Ax^p)}$  
$\int p(x) \, dx$  

$2 \int_0^{x_0} \sqrt{2m(E-Ax^p)} dx =  \sqrt{2mE} (\frac{E}{A})^{1/p} \int_0^1 \sqrt{1-y^p} \, dy =  \sqrt{2mE} (\frac{E}{A})^{1/p} I_p$  

**Scrivere il risultato**

T in funzione dell'ampiezza:  
ricorda che l'ampiezza è $x_0$, oppure $2x_0$, dipende da cosa intende il testo  
semplicemente raccoglila e mettila in risalto  

T in funzione dell'energia:  
poichè $E=\lambda x_0^p$, basta pluggare $x_0 =(\frac{E}{\lambda})^{1/p}$ nella formula di T in dunzione dell'ampiezza   
così da $T(x_0)$ passo a $T(E)$   
nota che entrambe le versioni devono essere in funzione di p  

**Deviazione da un risultato**

dopo aver espanso in serie, ho $T(\theta_0)=T_0(1+\frac{\theta_0^2}{16})$   
cioè T è il valore che avrei per $\theta_0=0$   
voglio vedere per quale angolo $\theta_0$ devia dell'1%  

$\frac{\Delta T}{T_0}=\frac{T(\theta_0-T_0)}{T_0} \sim \frac{\theta_0^2}{16} > 0.01$  

**Quando il pendolo può raggiungere la cima**

domanda: può raggiungere la cima?  
cioè, per $\theta_0 = \pi$ (la cima), T è finito?  

cosa faccio:  
riprendi la formula: $T(\theta_0)= 4 \int_{0}^{\theta _0} \frac{d\theta}{ \sqrt{2\frac{g}{l} (\cos \theta - \cos \theta_0)}}$   
l'unico punto dove potrebbe divergere è il punto critico $\theta \sim \pi$, lo studio:  
metto $\cos \theta_0 =-1$ ed espando lì intorno $\cos \theta = -1 + \frac{(\theta-\pi)^2}{2}$  
quindi l'integrale diventa: $\frac{1}{\sqrt{-1 +\frac{(\theta-\pi)^2}{2}} +1} \sim \frac{1}{\theta - \pi}$  
è una p-serie divergente, T è infinito   
cioè quando $\theta_0 \rightarrow \pi$, allora $T(\theta_0) \rightarrow \infty$   

che vuol dire periodo infinito:  
Quando l’energia è appena sufficiente per arrivare alla cima, la velocità in quel punto sarà esattamente zero.  
Questo significa che il pendolo impiega un tempo sempre più lungo per avvicinarsi alla cima, rallentando continuamente.  
rallenta sempre di più e impiega un tempo infinito per arrivarci esattamente

### Sistema rotante

**Passare a sistema rotante**

S: sistema inerziale  
usa coordinata $r$  

S': sistema non inerziale, ruota con $\omega$  
uso coordinata $r'$  

rapporto tra velocità:
$\dot{r}=\dot{r}'+\omega \land r'$  
per passare da coordinate di S a S'  
basta pluggarla in T  

**Cosa voglio fare**

scrivere la lagrangiana nel sistema inerziale (non rotante) o inerziale (rotante)  
darà la stessa soluzione, ma espressa in coordinate diverse  
quindi in r', non r  

**Trasforma da S' a S**

trasforma T:  
in S, ho: $T=\frac{1}{2}m\dot{|r|}^2$   
in $|\dot{r}|^2$, pluggaci il rapporto tra velocità $\dot{r}$ e $\dot{r}'$  

trasforma V:  
$V(r)=V(r')$  
perchè r e r' indicano lo stesso punto, cambia solo l'origine  
e V dipende solo dal punto in cui si trova l'oggetto (assumo che V sia dovuto alla gravità??)  

**Esempio: lagrangiana di massa puntiforme (senza potenziale, V=0)**

in S:  
scrivo lagrangiana: $L=T=\frac{1}{2}m\dot{r}^2$  
uso Eulero-Lagrange: $m\ddot{r}=0$   
cioè moto rettilineo uniforme  

in S':  
scrivo lagrangiana: $L= T=\frac{1}{2}m(\dot{r}'+\omega \land r')^2= \frac{1}{2}m(\dot{r}'^2 + 2r'\cdot(\omega \land r') + (\omega \land r')^2)$  
trovala come dice sopra, pluggando la relazione tra velocità  
uso Eulero-Lagrange: $m\ddot{r}=-2m(\omega \land \dot{r}')-m\omega \land (\omega \land r)$    
i due termini a destra sono la forza di Coriolis e la forza centrifuga, cioè le forze apparenti

### Campo magnetico

**Nota lagrangiana**

la lagrangiana non ha sempre questa forma $L=T-V$    
ce l'ha solo se U NON dipende dalla velocità delle particelle (quindi no attrito o forze magnetiche di Lorentz)  

**Steps**

ho un campo B, di cui so le componenti, e una particella carica   

1. scegli un $\vec{A}$ adatto (potenziale vettore)

2. scrivi L e H di una particella carica  

**Esempio**

$B(x,y)=2C x \hat{z}$  

1- scelgo A  
$A=(0, Cx^2, 0)$  

2- scrivo L   
$v \cdot A = A_x\dot{x}+A_y \dot{y} = Cx^2 \dot{y}$   
$L=\frac{1}{2}m(\dot{x}^2+\dot{y}^2) + qA \cdot v$   

3- scrivo H    
$H=\frac{p_x^2}{2m} +\frac{(p_y-qCx^2)^2}{2m}$  

**1. Scegliere A**

$B = \nabla \times A$  
cioè fai il rotore, quello che faccio il det di una matrice  

esempio: se il campo è lungo z, cioè $B=(0,0,B_z)$, si calcola con:  
$B_z = \partial_x A_y -\partial_y A_x$  
quindi per avere B costante, posso scegliere tipo:  
$A= (\beta By, \, \alpha B x, \, 0 )$  
con $\alpha +\beta = 1$  

le diverse scelte che posso fare per A mi evidenziano simmetrie diverse  
$A=(0,Bx,0)$: simmetria di traslazione lungo l'asse y  
$A=(By, 0, 0)$: '' lungo asse x  
$A=(1/2By, 1/2 Bx, 0)$: invarianza sotto rotazione al

**2. Lagrangiana e hamiltoniana di una particella carica**

$L=\frac{1}{2}mv^2 + qv \cdot A- q\phi$  
$H = \frac{1}{2m} |\vec{p}-q\vec{A}|^2+q \phi$  

per L aggiungo una parte simile alla forza di Lagrange, ma con A invece di B  
la parte con il modulo deve essere un addendo per ognuna delle componenti di A, che ho stabilito prima  

quando faccio le eq di hamilton, in generale: $\vec{p}=m\vec{v}+q\vec{A}$  

**Note su H e L**

#### Campo varia nel tempo

**Come si scrive la condizione**

campo uniforme lungo z, e varia lentamente nel tempo  
cioè $\frac{1}{B}\frac{\partial B}{\partial t}\ll \omega_c$  
che vuol dire  $\frac{\partial B}{\partial t} (\frac{2 \pi}{\omega_c})\ll B$  
cioè la variazione di B in un periodo (una rotazione della pallina) è trascurabile    

come prima, riduco il problema in 1D  
è un oscillatore, quindi $I=T \, E  = \frac{2 \pi}{\omega_c} E = const$  

cosa posso trovare:  
$I \propto \frac{E}{\omega_c} \propto \frac{v^2_\perp}{B} \propto \omega_c R^2 \propto BR^2$

che vuol dire:  
$v^2/B$: se aumento B, aumenta la velocità della particella (ovvio della def di $\omega_c$)  
$BR^2$: è il flusso del campo magnetico, il raggio dell'orbita cambia il modo che il flusso di B attraverso essa sia costante  

nota che: $\mu = \frac{m_v \perp^2}{2B}$ è il momento magnetico  
che quindi è conservato, è un invariante adiabatico  

**Conseguenza**

$\mu$ è un ivariante adiabatico  
visto che si conserva, posso riscrivere il raggio ciclotronico (dell'orbita) come:  
$R = \frac{v_\perp}{\omega_c} =\sqrt{\frac{2 \mu m}{q^2 B}}$  

**Nota**

ho visto che E della particella varia, ma cosa la fa variare?  
non può essere il campo B e la forza di lorentz, perchè non fa lavoro  
ma visto che B varia nel tempo, si crea un E indotto  
è questo E a fare lavoro  

**Cambio varia lentamente nello spazio**

anche qui si conserva il momento magnetico 


**Magnetic mirror**

l'energia $E=\frac{1}{2}m(v_\perp^2+v_\parallel^2)$ è conservata  
$\mu$ è conservato perchè è un invariante adiabatico, quindi se B aumenta anche $v_\perp$ deve aumentare  
per conservare E, $v_\parallel$ deve diminuire  
può arrivare a 0, in tal caso la particella si riflette  


quindi: l'orbita si restringe quando B aumenta  
è il fenomeno dello specchio magnetico, la particella può addirittura rimbalzare indietro  
lo specchio magnetico avviene se B(z) è monotona  
se invece B(z) ha un minimo in mezzo, e aumenta ai lati, è come mettere di fronte due specchi magnetici, così la particella ci rimbalza contro e ci rimane intrappolata (tipo ping pong)  
si chiama fenomeno della bottiglia magnetica  



**Recap**

$\mu$, il momento magnetico è costante sia se B varia lungo z, sia se avria lungo t  
gli effetti cambiano:  
varia lungo t: raggio ciclotronico varia  
varia lungo z: crea specchio magentico  

se B varia lungo t: non c'è nessuna direzione preferita nello spazio, niente specchio
