### Doppler relativistico

**Steps**

1- disegna nel piano xy: traiettorie e vettori  
ATTENTA in che frame sei  

2- usa la formula di f'  
quindi devo calcolare i valori da pluggarci  

**1. Formula generale**

$f'(t)=\gamma f (1-\vec{\beta} \cdot \hat{n})  =\gamma f (1-\beta \cos \theta)$  

sono 2 modi uguali, meglio il primo  
il prodotto scalare fa apparire il coseno tra i due e il prodotto dei vettori, ma il versore ha modulo 1    

**2. Trovare vettori**

1- $\vec \beta$  
è tangente alla traiettoria dell'astronave, disegnala  

2- $\hat n$  
segna due punti, che potrebbero dipendere dal tempo:  
1- dove si trovava la sorgente quando ha lanciato il raggio      
2- dove si trova il ricevitore quando riceve il raggio   
$\hat{n}$ è il vettore che va dal primo punto al secondo    

**3. Angolo $\theta$ in funzione del tempo**

situazione:  
sorgente fermo nell'origine  
ricevitore si muove con legge del moto $R(t)=(vt, d)$ nel frame della sorgente (cioè lungo asse x)  

disegno nel frame della sorgente  
assumo che il ricevitore riceva il segnale ad un generico momento $t$, fisso il putno dove si trova (un punto qualunque sull'asse x)  
quel punto mi definisce un triangolo rettangolo, so i valori dei lati dalle componenti x e y di R(t)   
$\cos \theta = \frac{adiacente}{ipotenusa} =\frac{vt}{\sqrt{(vt)^2+d^2}}$  

#### Attenzione frame

cosa | O: frame di partenza | A: frame di arrivo  
---|---|---
quale è | quello in cui so disegnare | quello dell'oggetto A che si muove con velocità v nel disegno  
quela è la sua frequenza | f | f'  
di cosa è | dell'oggetto fermo nel disegno <br> se nulla è fermo, è di un osservatore esterno | dell'oggetto A  

**Spiegazione**

1- risolvi il problema senza dati  
per semplicità, fai sempre trasformazione: O->A   
vai dal frame che so disegnare a quello che no  
quindi risolvi semplicemente il problema di trovare f'(f)  

2- capisci a cosa corrispondono i dati  
devo vedere se l'oggetto proprio dei fue frame (rispettivamente, un oggetto fermo e A) sono la sorgente, il ricevitore o un osservatore esterno  
così posso associare la frequenza in quel frame ad uno dei dati  
eg. se A è la sorgente, allora $f'=f_A$  

**Esempio: facile**

nel frame di partenza:    
ricevitore fermo nell'origine  
ho astronave A che fa un cerchio di raggio R a velocitò v, intorno all'origine  
ed emette segnale $f_s$, misurato nel suo frame, verso il centro  

1- passaggio O->A (da disegno ad astronave)  
disegno v e n nel frame O  
poichè sono perpendicolari (tangenziale e radiale), ho doppler trasverso: $f'=\gamma f$  

2- associa i dati  
A è la sorgente, quindi A è il frame della sorgente, quindi $f'=f_s$  
in O c'è il ricevitore fermo, quindi è il frame del ricevitore, quindi $f=f_r$  
se io so $f_s$ e voglio trovare $f_r$, la soluzione si scrive: $f_r = \frac{f_s}{\gamma}$  

**Esempio: difficile**

so disegnare nel frame O, di un'osservatore esterno  
sia sorgente che ricevitore sono in moto, sono i frame A e B  

soluzione:  
nel frame di O, disegna $\hat{n}, \beta_A, \beta_B$  
poi fai 2 trasformazioni, O->A e O->B  

nota: non fare A->O, ti confondi solo a invertire i vettori  
semplicemente dai O->A, poi come nell'esempio sopra fai la formula inversa   
quindi ho $f_B(f_0)$ e $f_A(f_0)$, inverti la seconda per avere $f_0(f_A)$, così ottengo $f_B(f_A)$  

**Nota: disegnare un altro frame**

se due frame sono inerziali, cioè la loro velocità relativa è costante in modulo e verso  
esempio: disegno il frame O, dove c'è l'oggetto A che si muove a velocità $\beta$ lungo l'asse x  
allora posso disegnare anche il frame A, e le velocità di tutti gli altri oggetti le ottengo facendo la somma delle velocità con la formula di Lorentz (somma tra velocità dell'oggetto in O e boost tra O e A)  

ma cerca di non farlo, meglio fare come nell'esempio sopra, disegna il più facile e basta  

#### Casi particolari

**Casi particolari**

raggio e $\beta$ paralleli: effetto Doppler longitudinale  
$f'=f \sqrt{\frac{1 \pm \beta}{1 \mp \beta}}$  
sopra ci va il + se si avvicinano (perchè schiacciano l'onda), - se si allontatano  
e sotto il segno opposto  

raggio e $\beta$ ortogonali: effetto Doppler trasverso  
$f'=\gamma f$  

semplicemente plugga il coseno dell'angolo giusto  
per longitudinale, quella forma è un modo più bello di riscrivere  

**ALtro metodo per caso difficile??? Distanza minima tra due navi**

ho due navi, di frame A e B  
e frame generico O, dove so diseganre le rotte delle due navi  

1- scegli uno dei due frame delle navi, tipo A  
così la nave A è sempre nell'origine  

2- con trasf di Lorentz, passa dalle coord di B in O alle coord di B in A  

3- calcola distanza di B in coord di A  
ricordando che A è nell'origine, è solo la norma del vettore spaziale, con pitagora  

4- per avere il minimo, fai lad erivata rispetto al tempo  

**Nota**

effetto longitudinale: quando $\vec k$ di onda è parallelo a $\vec v$ del frame     
esiste anche in caso classico  

effetto trasversale: uando $\vec k$ di onda è ortogonale a $\vec v$ del frame  
esiste solo in caso realtivistico  

**Nota: red shift**

luce che arriva dalle galassie lontane subisce effetto Doppler che gli abbassa la frequenza, quindi cambia colore, appare rossa  
prova dell'espansione dell'universo  

#### Cambia tempo

**Domanda**

t: tempo nel frame O     
t': tempo nel frame A     
ho f'(t), ma voglio f'(t')  
quindi cambiare solo il frame del tempo, non della frequenza  

come fare:  
1-calcola f'(t) con Doppler, che non cambia il tempo  
2- poi calcola con trasformazioni t(t') e pluggalo dentro, così ho f'(t(t'))  

**2. Trasformazione di tempo**

legge del di frame A in frame O: $R(t)=(vt, d)$     
voglio passare da tempo di frame O a tempo di frame A  

uso la generica trasformazione di Lorentz:  
$t'=\gamma(t-v \, x(t))= \gamma t (1- v^2) = \frac{t}{\gamma}$  
sto pluggando solo la componente x di R; quella lungo cui avviene il boost  
ricorda, $v=\beta$  

#### Teoria

**Trasformazione di Lorentz**

è la trasformazione della componente TEMPORALE del quadrivettore d'onda:   
$\frac {\omega'}c = \frac \omega c \gamma (1- \vec \beta \cdot \hat n)$  
$f' = f \gamma (1- \vec \beta \cdot \hat n)$  


quadrivettore d'onda:  
$k^\mu = (\frac \omega c, \vec k) = \frac \omega c (1, \hat n)$   
nota che $\vec {|k|} = \frac  \omega c$ solo per onda EM, cioè luce  
n: direzione di propagazione  

per ottenere la frequenza:  
$\omega = 2 \pi f$  
quindi trasformo $\omega$ in f, e semplifico $2 \pi$ a destra e sinistra  

**Dipendenza dal tempo**

f' può dipendere dal tempo anche se f no, perchè il versore $\hat{n}$ dipende dal tempo   
cioè la sorgente lancia raggi in tutte le direzioni, ma a seconda di che tempo t mi interessa, in quel momento riceve un raggio diverso, che quindi era stato lanciato a diversa $\hat{n}$  

oppure pensa che l'angolo $\theta$ può dipendere dal tempo  

**Frame e oggetti** 

ogni oggetto crea un suo proprio frame  
in cui l'oggetto è sempre fermo nell'origine  
e il frame si muove con la velocità dell'oggetto  
in pratica oggetto = origine del frame  
pensa sempre all'origine del frame, mai all'oggetto  

### Distribuzione e valore medio

**Formula media di una distribuzione**

$⟨f(x)⟩=\frac{∫f(x)w(x)dx​}{∫w(x)dx}$  
se w(x) è una probabilità, non serve normalizzare, cioè il denominatore è 1  

**Esempio**

$f(\theta)=\cos \theta$, che è in $f(\theta) \in [-1, 1]$  
$w(\theta) = 1- \cos \theta$  

1- faccio cambio var, $x = \cos \theta$, $x \in [-1,1]$    
$f(x)=x$  
$w(x)=1-x$  

2- calcolo:  
$⟨\cos \theta⟩ = ⟨x⟩=\frac{∫_{-1}^1 x \, (1-x) \, dx​}{∫_{-1}^1 1- x \; dx} =- \frac 13$  


#### Esempio

**Situazione**

ho un particella emessa con stessa prob ad ogni angolo (distribuzione angolare isotropa)  
se è messa con angolo $\cos \theta \in [-1, v]$ ha energia $E_1$, se emessa con angolo $\cos \theta \in [v, 1]$ ha energia $E_2$  
per comodità, chiamo $x=\cos \theta$  

**Esempio spiegato bene**

devo trovare $f(\theta)$ e $w(\theta)$  

1- valore: è il valore della cosa di cui voglio la media, in funzione della cosa che varia  
$f(x) = E(x) = \begin{cases} E_1 & \cos \theta \in [-1, v] \\ E_2 & \cos \theta \in [v, 1] \end{cases}$

2- pesi: quanto è probabile che la variabile $\theta$ abbia un certo valore  
se emissione è isotropa, $\omega(x)=1$  

3- calcola:  
$⟨E(x)⟩=\frac{∫_{-1}^1E(x)w(x)dx​}{∫_{-1}^1 w(x)dx} = \frac 12 (\int_{-1}^v E_1 \; dx + \int_v^1 E_2 \; dx)$  


**Prob per emissione isotropa**

emissione isotropa: la prob è uniforme nell'angolo solido $\Omega$  
(NON in $\theta$, sennò ogni meridiano parallelo sarebbe colpito dallo stesso numero di fotoni emessi, ma non può essere, l'equatore è più grande e dovrebbe riceverne di più)      

1- in funzione di $\theta$  
$dP ∝ d\Omega ∝ \sin \theta  \; d\theta$  
cioè: $\frac{dP}{d\theta} = \omega(\theta) \sim \sin \theta$  

2- cambio var  
perchè i valori di $f(\theta)$ sono espressi in funzione del cos, mi farebbe più comodo  
$x = \cos \theta \quad \rightarrow \quad dx = -\sin \theta d\theta$  
$dP \sim d\Omega  \sin \theta  \; d\theta  = dx$  
così $\frac{dP}{dx} = w(x) = 1$  

3- volendo potrei normalizzarla, cioè:  
$p(x) = w(x) \; \int_{-1}^1 w(x) \, dx = 1 \, \frac 12  = \frac 12$   
è il denominatore della formula grande  

**Recap statistica**

1- prob che x sia un valore:  
$p(x)$  

2- prob che x sia sopra un valore:  
$P(X>x) = \int_x^\infty \, p(x) \;dx $  
X: variabile aleatoria che segue la distribuzione $p(x)$, voglio che il valore estratto sia maggiore di un valore x  
semplicemente, prendi la prob che x sia un certo valore, e intrgrala su tutti i valori di x accettabili  

3- prob che una var aleatoria sia maggiore di un'altra   
$p(A>B) = \int_{- \infty}^{\infty} p_B(x) \; P(A>x) \; dx$     
ho due var aleatorie A e B, che seguono le distrib $p_A(x)$ e $p_B(x)$  
per ogni valore estrabile, calcolo prob che B sia quel valore, e A più grande   
il dominio dell'integrale sono i valori che A e B possono assumere, cioè definisco p(x) a tratti, e la faccio valere 0 per i valori impossibili, poi separo in somma d integrali 

4- valore medio:  
$⟨f(x)⟩= \int_{-\infty}^{\infty} f(x) \, p(x) \; dx = \frac{∫f(x)w(x)dx​}{∫w(x)dx}$  
per ogni valore possibile, moltiplico il suo valore e la prob di averlo  
se fosse in un dato, farei la media aritmetica dei valori possibili  

**Come normalizzare**

dividi per l'area totale:  
$p(x) =\frac{w(x)}{\int_{-\infty}^{\infty} w(x) \, dx}$
