### Limiti

**Limite relativistico e non**

1- NON relativistico: $p \ll m$  
2- relativistico: $ m \ll p$  

raccogli in modo di avere la cosa grande al denominatore, e quella frazione diventa 0  
metti a 0 sono gli infinitesimi di SECONDO ordine, cioè $(p/m)^2 = 0$  
cioè, se $m<<p$, allora $m^2+p^2 \sim p^2$  


**Altro limite**

il metodo sopra va bene solo se qualcosa tende a 0 (cioè se qualcosa è molto più grande di un'altro)     
invece, se qualcosa tende uguale a un'altra, tipo m->M, allora posso solo raccogliere tutto in una sola frazione e poi razionalizzare  

**Esempio: limite relativistico**

$v = \frac{p}{m+\sqrt{m^2+p^2}}$  

1- NON relativistico  
$v= \frac{p}{m + m\sqrt{1+(p/m)^2}} \sim \frac{p}{2m}$  

2- relativistico  
$v= \frac{p}{m + p\sqrt{(m/p)^2+1}} \sim \frac{p}{m+p} =  \frac{1}{m/p+1} = (\frac mp  +1)^ {-1} \sim 1- \frac mp$   

**Esempio: M->m**

$v = \sqrt{1-\frac{4M^2}{\mu^2}}$  

faccio M->m, ed esplicito $\mu$  
$= \sqrt{1-\frac{4m^2}{2m^2+2m\sqrt{m^2+p^2}}} = \sqrt{1-\frac{2m}{m+\sqrt{m^2+p^2}}} = $  

metto tutto a denominatore comune:  
$ = \sqrt{\frac{m+\sqrt{m^2+p^2}- \; 2m}{m+\sqrt{m^2+p^2}}} = \sqrt{\frac{\sqrt{m^2+p^2} -m}{\sqrt{m^2+p^2} + m}}$ 

razionalizzo (in realtà non sto razionalizzando, sto togliendo la radice sopra invece di sotto, ma vabbè):  
$= \sqrt{\frac{p^2}{[\sqrt{m^2+p^2} + m]^2}} = \frac{p}{\sqrt{m^2+p^2} + m}$  

### Energia di soglia

**Come fare**

calcola $\mu$ in due momenti e uguagliali:  
1- prima, lab frame  
2- dopo, cm frame (e metti tutto fermo)   

**Esempio**

situazione:   
m ferma + m in moto -> M + M

$k_{lab}=(m+\sqrt{E^2-m^2}, p)$  
$k'_{cm}=(2M,0)$   


prima: $\mu^2 = (m+\sqrt{E^2-m^2})^2 -p^2 = 2m^2-2m\sqrt{E^2-m^2}$       
dopo: $\mu^2 = 4M^2$  

uguagliali, trova espressione di p (o di E, se ho scritto in funzione di quella)  

### Velocità

**Come si fa**

conoscere la velocità in un frame = conoscere l'energia in un frame  
se so l'energia, posso trovare:  

$\gamma = \frac E m$   
$v = \sqrt{1-\frac 1{\gamma^2}} = \sqrt{1- \frac{m^2}{E^2}}$  

nota che trovo la velocità vista nel frame dove è calcolata E  

**Velocità del CM**

formula generale: $v_{cm} = \frac{p_{tot}}{E_{tot}}$   
calcolati nel lab 

formula particolare: $v_{cm} =  \sqrt{1- \frac 1 {\gamma^2}} = \sqrt{1-4 \frac{m^2}{\mu^2}}$  
quando: 2 particelle di massa uguale m, di cui una ferma  
perchè: nel CM le due particelle hanno uguale velocità, così $p_{tot, cm}=0$, ed energia $E=\frac{\mu}{2}$     
La particella che nel LAB era ferma, nel CM viene vista muoversi all'indietro esattamente con la velocità del CM  
quindi ho due particelle in moto con $v_{cm}$  
che posso calcoalre, visto che so l'energia  

**Velocità relativa**

devo fare la somma delle velocità con la formula di Lorentz  

perpendicular component (wrt boost direction):  
$v_y' = \frac{v_y}{\gamma (1+\beta v_x)}$  

parallel component:  
$v_x' = \frac{v_x+\beta}{1+\beta v_x}$ 

$\beta$: relative speed of frames (speed of final frame seen by initial frame, so boost)   
$v$: speeds in initial frame  
$v'$: speeds in final frame  

**Velocità relativa: shortcut**

due masse che si muovono in +v e -v (stessa velocità, direzione opposta)  
$v_{rel} =\frac{2v}{1+v^2}$  

nota: per sommare devo sempr fare la somma delle velcità  
tranne se una delle due è 0  
tipo se so v_{cm}, ho una particella ferma in cm frame, allora quella particella vede il lab frame muoversi a -v_{cm}

**Prima dell'urto: modellizza**

sistema con massa invariante $\mu$    
quindi può essere qualunque cosa, una particella o più, in moto o non, mi interessa solo la massa invariante   

come calcolo $\mu$:  
si può calcolare anche nel LAB frame  
$\mu^2=(E_e+E_\gamma)-(\vec p_e+ \vec p_\gamma)^2$ 

#### Calcola energie

**Nota**

avere E significa avere v, usa le formule scritte più in alto   

se la massa invariante del 'prima' è un multiplo di m, sostituisci subito, mi aiuta coi calcoli   

**Energia dei prodotti IN CM FRAME**

1- emetto 2 particelle diverse:  
$E_1 = \frac {\mu^2+m_1^2-m_2^2}{2 \mu}$  
se voglio $E_2$, scambia $m_1$ e $m_2$ al numeratore  

2- emetto 2 particelle uguali:

**Passa a lab frame**

$\theta$: angolo tra $\vec p$ e $\vec v$ (momento di m e elocità di $\mu$, per passare a cm frame)   

$\vec v = -v_{cm}$, quindi appare segno +   
perchè qui mi serve la velocità con cui il CM vede il LAB, ma $v_{cm}$ p la velocità con cui il LAB vede il CM

$m_{23}^2=(E_2+E_3)-(\vec p_2+ \vec p_3)^2$  
nota che $m_{23} \ge m_2 + m_3$

**Energia delle cose finali (m3) in CM frame**

1- in frame di $m_2$: $E_3 = \frac {m_2} 2$    
perchè: ho sistema di $\mu=m_2$ che emette due particelle uguali m3   

2- in frame di m1 / CM: $E_{3, cm} = \gamma (E_3-v_2 p_3 \cos \theta)$    
come: $p= \sqrt{E_3^2-m_3^2}$   
ho già trovato $E_2 = \mu /2$, quindi posso trovare $v_2$ e $\gamma$ (è la velocità di m2, perchè il boost è tra cm e frame di m2)  

3- max e min:  
min: cos = 1  
max: cos = -1  
calcola i valori  

4- distribuzione  
$E_{cm} \sim \cos \theta$  
quindi le particelle sono emesse in modo isotropo nel sistema di $m_2$, allora la distriuzione delle energia è unifore nell'intervallo $E_{cm} \in [E_{min}, E_{max}]$  
quindi: $\frac{dP}{dE}=\frac{1}{E_{min}-E_{max}}$  

**Velocità relative tra una coppia di m3**

2- massima

**Massa invariante di una coppia di m**

chiamo i due componenti della coppia A,B 

1- se vengono dalla stessa m2:   
$\mu$ non cambia, quindi $\mu = m_2$ (calcolata prima dell'urto, nel frame di m2)  

2- se vengono da m2 diverse:  
$\mu^2 = (E_a^2+E_b^2) - (\vec p_a^2 + \vec p_b ^2) = 2m^2 +E_aE_b-p_ap_b \cos \theta$

### Esempio esercizio decadimenti

$m_1 \rightarrow 2m_2 \rightarrow 4m_3$  

$m_1 = 16 m$  
$m_2 = 4m$  
$m_3 = m$  
$m_1$ è ferma  

le m3 sono emesse in modo isotropo nel sistema di m2  

**1. Velocità di m2**  

decade in due oggetti uguali:  
$E_2 = \frac{\mu}2 = \frac{m_1}{2} = 8m$  
$v_2 = \sqrt{1-\frac{m^2}{E^2}} = \frac {\sqrt 3}2$ 

**2. Energia max e min di m**

1- in frame di m_2:  
$E_3 = \frac{m_2}2 = 2m$  
$p_3 = \sqrt{E_3^2-m_3^2} = \sqrt 3 m$  
$\gamma =\frac{E_3}m = 2$  

2- in frame di m_1 / CM:  
$E_{3,CM} = \gamma (E_m -\, \vec v_2 \, p_3 \, \cos \theta) = 2(2m+ \frac {\sqrt 3}2 m \cos \theta) = m(4+3 \cos \theta)$  

3- max e min:  
max: cos = 1, $\theta =  \pi \quad \rightarrow \quad E_{cm}=m$  
max: cos = -1, $\theta = 0 \quad \rightarrow \quad E_{cm}=7m$  

**3. Distribuzione di di E'_m$**  

le m sono emesse in modo isotropo nel sistema di m2, quindi tutti gli angoli sono probabili uguali  
poichè $E'_m \sim \cos \theta$, allora la distribuzione è uguale all'interno dell'intervallo $E'_m \in [m, 7m]$  

cioè $\frac{dE}{6m}$  

**4. Velocità relative tra una coppia di m**

1- min:

2- max:

faccio la somma delle velocità: $v_{rel} =\frac{2v}{1+v^2}$  
dove v la trovo da $E_{max} = 7m = \gamma m $, quindi $\gamma = 7$   

### Angoli di emissione

**Situazione**

prima di decadimento: sistema con v_{cm} verso destra  
cioè ho almeno una particella in moto verso destra  
ricorda $v_{cm} = \frac{p_{tot}}{E_{tot}}$  

**Domanda**

voglio che una particella sia emessa verso destra (anche in diagonale a destra va bene), non verso sinistra  

condizione: $v<v_{cm}$  
v: velocità di particella emessa, in CM frame  
$v_{cm}$: velocità del cm frame (calcolato nel lab frame, prima)  
se è rispettata, esiste un angolo massimo di emissione della particella (considerando theta=0 verso destra  )

perchè:  
immagina la particella come una persona su un treno  
se è più lenta del treno, anche se corre nella direzione opposta al moto del treno, in totale si muoverà sempre avanti  

**Angolo massimo di emissione (nel lab)**

1- angolo di emissione in CM  
$\tan \theta = \frac{p_y}{p_x} = \frac{p \sin \theta}{p \cos \theta}$  

2- angolo di emissione in LAB  
$\tan \theta' = \frac{p_y}{\gamma(p_x-\vec v_{cm} E)} = \frac{p \sin \theta}{\gamma(p \cos \theta +v_{cm} E)} = \frac{\sin \theta}{\gamma(\cos \theta+v_{cm} E/p)} = \frac{\sin \theta}{\gamma(\cos \theta+v_{cm} v)}$    
ora deriva rispetto a $\theta$, trova il massimo  

3- note:  
$\vec v_{cm} = -v_{cm}$, perchè v è la velocità di LAB visto da CM, dovrei già saperla (da cui calcolo $\gamma$)  
$p/E = v$ è la velocità di m nel cm frame, si trova sapendo la sua energia nel cm, tipo $E=\mu/2$   
il boost è lungo x, quindi p_y rimane uguale, p_x cambia  

**Trova condizione**

1- frame CM:  
$k_{cm}=(E, p \cos \theta, p \sin \theta)$  
quindi $\theta$ è l'angolo misurato nel CM ($\theta=0$ è emissione verso destra)  
$\vec v_{lab}=-v_{cm}$, perchè $\vec v_{cm}=v_{cm}$ (cioè il cm si muove verso destra)  

2- frame lab:   
$\vec p_{x, lab} = \gamma (\vec p_{x,cm}- \vec v_{lab} E_{cm}) = \gamma (p \cos \theta +v_{cm} E)$  

3- condizione:  
nel lab, si muove verso destra se $\vec p_{x,lab} >0$, cioè:  
$p \cos \theta +v_{cm} E >0$   
$v \cos \theta +v_{cm} > 0$  
v=p/E è la massa dell'oggetto  

4- esempio:  
se una massa è emessa verso sinistra nel cm, cioè cos=-1, allora nel lab ba verso avanti se:  
$-v +v_{cm} >0$  
$v_{cm}>v$  
cioè se il modula della velocità della massa è minore di quella del cm 


### Esempio: molte trasformazioni, con specchio

**Situazione**

frame O:  
fotone emesso con $E_0$ e con angolo $\theta$ (rispetto a +x)  
specchio si muove verso sinistra, quindi $\vec v_s = -v$  

calcola energia del fotone in frame O dopo la riflessione  

**Note**

1- chiamo $x=\cos \theta$, per scrivere veloce   
2- scrivi il quadrivettore in tutti i casi  
ma posso ignorare $v_y$, tanto è ortogonale al boost e alla specchio, non cambia mai    
3- è un fotone, così $|p|=E$  
4- scrivi p e v come vettori, e nei dati metti il segno che indica la loro direzione, tipo $\vec v_s=-v$  
o per essere sicura, posso diseganre $v$ e $p$ e vedere cosa dà il prodotto scalare  
v ha sempre il segno di vettore, ma quando è accanto a p c'è anche il prodotto scalare, accanto a E no  
$\vec v_s$ è la vellocità di S misurata in O  
$\vec v_O = -\vec v_S$ è la velocità d9 O misurata in S (quindi a destra devo mettere la velocità di ciò a cui voglio arrivare, così ho un valore che so misurare)  
5- $E_0$ lo uso come nome della componente di $k_0$, quindi non può essere anche il suo coeff, o mi confondo  

**Soluzione**

1- frame O, prima: $k_0=E(1, x)$  
$\vec v_s =-v$  
quindi $\vec v_0 = +v$
 
2- frame S, prima: $k_S = \gamma E (1+vx, x+v)$  
$E_s = \gamma (E_0 - \vec v_s \cdot \vec p_{0x}) = \gamma E(1 + v \, x)$  
$p_s = \gamma (\vec p_{0x}- \vec v_s E_0) = \gamma (Ex \; +v \, E) = \gamma E (x \, + \, v)$

3- frame S, dopo riflessione: $k'_S = \gamma E (1+vx, -x-v)$  
la riflessoine cambia il segno del momento ortogonale alla superficie dello specchio, cioè qui x  

4- frame O, dopo riflessione:  
$E_O' = \gamma ( E_s' - \vec v_O \cdot p_{sx}') = \gamma^2 E [(1+vx) \, + v \, (x+v)] = \gamma^2 E (v^2+2vx +1)$  

#### Tempo di decadimento

**Situazione**

pione percorre distanza L nel lab  
quale è il suo tempo di decadimento, misurato nel sistema solidale al pione   

1- calcola velocità del pione nel lab frame  
$v=\sqrt{1-\frac{1}{\gamma^2}} =\sqrt{1-\frac{m^2}{E^2}}$

2- tempo di decadimento del lab frame:  
$\Delta T = L/v$  

3- passo al sistema solidale al pione:  
$\tau = \frac{\Delta T}{\gamma} = (1-v^2) \Delta T = \frac{m^2}{E^2} \Delta T$
