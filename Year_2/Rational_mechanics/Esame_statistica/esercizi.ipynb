### Funzione di partizione

**Steps**

1- scrivi formula di $Z_n(Z_1)$  

2- scrivi formula di $Z_1$  

3- calcola gli integrali dp e dr  

#### 1. Funzione partizione sistema

**Come si scrive**

per particelle non interagenti tra loro:  

con $Z_1$ uguale:  
1- distinguibili: $Z_n = (Z_1)^N$  
2- indistinguibili: $Z_n = \frac{(Z_1)^N}{N!}$ 

con $Z_2$ diverse (ad esempio se l'energia si scrive diversa, perchè hanno massa diversa):    
$Z=Z_mZ_M$  

**Esempio**

ho N particelle di massa m, e N di massa M, entrambi i gruppi indistinguibili:  
$Z_{tot} = \frac{Z^N(m)}{N!} \frac{Z^N(M)}{N!} $  

tip: scrivi una generica $Z(\mu)$, per una singola particella di massa $\mu$  
così poi avrò anche: $U_{tot} = U_{tot}(m)+U_{tot}(M)$, stessa cosa per C   

#### 2. Funzione di partizione particella $Z_1$ 

Step | formula | what  
---|---|---
1- energia di UNA particella | $E= E_{trasl}+E_{int}$ <br>  $E_{trasl}=\frac{p^2}{2m} $  <br> livelli continui: $E_{int}=V(r)$   <br> livelli discreti:  $E_{int} \in [0, \epsilon]$  | separa in parte traslazionale e interna
2- peso di Boltzmann | $e^{-\beta E}= e^{-\beta E_{trasl}} e^{-\beta E_{int}}$ | separa in prodotto di traslazionale e interna
3- formula di Z | livelli continui $Z_1 = \frac{1}{h^3} \int d\Gamma \; e^{-\beta E} = \frac{1}{h^3} \int d^3p \; \int d^3q \; e^{-\beta E}$  <br> livelli discreti: $Z_1=\Sigma_i e^{-\beta E_i}$ | ogni fattore del peso, vedi se integrale (continuo) o sommatoria (discreto)

$Z_1 = \frac 1{h^3} \quad \int_{-\infty}^{\infty} \, dp^3 \, e^{-\beta p^2/(2m)} \quad  \int_V \, d^3q \, e^{-\beta V(r)} \quad \Sigma e^{-\beta ks} $


##### Note

**Attenzione**

1- RICORDA IL - ALL'ESPONENTE del peso di Boltzmann  
2- se sono in spazio 2D: $\frac{1}{h^2}$ e $\sqrt{\frac{\pi}{a}}$ e $d^2p \, d^2q$ hanno esponente 2, non 3  


**Note: energia**

1- per energia interna con livelli discreti, devo metterci un parametro e segnarmi che valori può avere  
non sommare tutti i possibili livelli, perchè una particella può essere in un solo livello alla volta   
è nel peso che sommo tutti i livelli possibili  

2- se ho livelli discreti e lui mi dà una formula $E=...$ in cui NON appare p, NON la devo aggiungere, è già contata  
questo caso, in cui manca p, è l'unico in cui per Z userò la formula con la sommatoria, e non con l'integrale  

**Scegli coord: cartesiane o polari**

quando uso le polari?:  
SOLO SE l'oggetto è vincolato a muoversi SU una circonferenza di raggio fisso   
cioè se in E appare $\theta$, ma NON r, perchè r è fissato    
NON se può muoversi ALL'INTERNO di qualcosa tondo  

differenza: come si scrive energia cinetica  
carth: $T = \frac{p^2}{2m}$    
pol: $T=\frac{p^2}{2mR^2}$  

nota: anche se scrivo in carth, posso usare $r$  
probabilmente poi, quando integro dq, lo trasformerò in dr 

**Teoria energia**

traslazionale = cinetica (integrale in p)  
interna = potenziale (integrale in q)  

dove si vede:  
in E sono sommate  
in Z sono moltiplicate  
in U e C ritornano ad essere sommte (perchè il log(Z) trasforma moltiplicazione in somma)  

##### Esempi

**Esempio: normale**

$E=\frac{p^2}{2m} + ks $ dove $s \in [-1, 0, +1]$   
$e^{-\beta E} = e^{-\beta p^2/(2m)}e^{-\beta ks}$    
$Z_1 = \frac 1{h^3} \quad \int d\Gamma \; e^{-\beta p^2/(2m)} \quad \Sigma e^{-\beta ks}  =  \frac 1{h^3} (e^{\beta k}+1+e^{-\beta k})  \; \int d^3p \int d^3q \; \; e^{-\beta p^2/(2m)} $   


**Esempio: senza energia traslazionale**

data del testo:  
$E_{i,j} = i\epsilon + j\epsilon + ∆δ_{ij}$  
quindi mi specifica di non considerare la parte di energia cinetica, l'ha già inclusa lui  

1- individua i possibili stati e la loro energia  
(i,j)=(0,0) -> $\Delta$  
(1,0)=(0,1) -> $\epsilon$  
(1,1) -> $2 \epsilon + \Delta$  

2- scrivi peso:  
$e^{-\beta E} = e^{-\beta E_{int}}$ con $E_{int} \in [\epsilon, \epsilon, \Delta, 2\epsilon+\Delta]$    

3- scrivi Z:  
$Z=\Sigma e^{-\beta E_i}=e^{-\beta \Delta}+2e^{-\beta \epsilon}+e^{-\beta(2 \epsilon + \Delta)}$  


#### 3. Calcola integrali

**Steps**

1- integrale dp: sempre uguale, controlla solo numero di coord  
2- integrale dq: controlla la dipendenza dalle coord di $E_{int}$ e il dominio di integrazione   

**1. Integrale dp**

come si fa:  
$\int d^3 p \; e^{-\beta \frac{p^2}{2m}} = (\frac{2 \pi m}{\beta})^{3/2}$  
elevo al cubo perchè p è un vettore a 3 componenti (particella si muove in 3D)   
ma se mi muovo in 2D o 1D elevo per 2 o 1  

perchè:  
è l'integrale gaussiano:    
$\int_{-\infty}^{\infty} e^{-a x^2} dx = \sqrt{\frac{\pi}{a}}$  

**2. Integrare dr**

dipende da:  
1- dipendenza di $E_{int}$ da var spaziali r,x (quindi formula di V(r))  
2- forma del dominio di integrazione V  

dipendenza da coord di $E_{int}$ | dominio di integr | formula
---|---|---
no dipendenza | V me lo dice il testo, so quanto vale | $\int d^3q = V$ 
V(r) o V(z) | qualche simmetria in coord polari | a seconda della simmetria di V, fai cambio coord: <br> sferica: $\int _V d^3 q = \int _{R_1}^{R_2} 4 \pi r^2 \, dr \quad$, poi   $y = r^3$  <br>   cilindrica assiale: $\int d^3 q = \int_A dA \; \int_0^L dz = A \int_0^L dz$ <br> cilindrica radiale: $\int d^3q = \int_{R_1}^{R_2} \, 2 \pi h \; r \, dr$
dipendenza diversa da x e y | qualche simmetria in cord cartesiane | separala il diff: $dq^2 = dx \; dy$  <br> $\int_A d^2q  = \int_0^L dy \, \int_0^L dx  = L \, \int_0^L dx$ <br> poi, comodo fare sostituzione per avere $e^t$, facile da integrare 
qualunque delle precedenti | due separati | separali: $\int_V d^3q  = \int_{V1} d^3q + \int_{V2} d^3q$  

#### Note dr

**Nota**

se E dipende in modo diverso da coord diverse, allora separa dq = dx dy  
e se qualcosa dipende solo da $\theta_1-\theta_2$, può essere utile fare la sostituzione: $\int _0^{2\pi} d\theta_1 \, \int _0^{2\pi} d\theta_2 \; (\theta_1, \theta_2)= 2 \pi \int _0^{2\pi} d\phi \; f(\phi)$  
dove $\phi = \theta_1-\theta_2$, e faccio questa sostituzione anche nella formula di E, così risolvo più facile  
cioè: $d\theta_1 \, d\theta_2 = 2\pi \, d\phi$  
dove $\phi=\theta_1-\theta_2$  

esempio:  
$T = \frac{p^2_1}{2m_1}+\frac{p^2_2}{2m_2}$  
$\int dp \; e^{-\beta T} = \int dp_1 \int dp_2 \; e^{-\beta T} = \sqrt{2 \pi m_1}{\beta} \, \sqrt{2 \pi m_1}\beta$

**Note**

toroide: uso la simmetria sferica, è come una sfera con $R_1 \neq 0$ 

**Esempio: due domini separati**

dominio: due cilindri di base A, altri L, uno sopra l'altro  
uno nelle cord [0, L], l'altro in [D, D+L]  

la V(r) da inserire nell'integrale resta uguale  
ma cambia il volume di integrazione V, quindi diventa:  
$\int_V d^3q  = \int_{V1} d^3q + \int_{V2} d^3q = A\int_{0}^L dz \; + \; A\int_{D}^{D+L} dz$  

trucco per calcolare:  
fai apparire nel secondo integrale il primo  
oppure risolvili entrambi e poi raccogli  
$e^{−βmgz}=e^{−βmgD}e^{−βmg(z−D)}$  
$\int_{D}^{D+L} dz e^{−βmgz} = e^{−βmgD}\int_0^L ​e^{−βmgz}dz$  
ho ritrovato il primo integrale, ora posso risolverlo solo una volta  

### Trig iperboliche

**A cosa serve**

a riscrivere in modo più facile le somme dei pesi quando ho energia interna discreta  
ma rende più difficile la parte dopo, meglio non farlo  

**Trig iperboliche**

se ho come stati $e^{x}+e^{-x}$, usa le trig iperboliche per rappresentarli  

$\begin{cases}
2\cosh x = e^x + e^{-x} \\
2\sinh x = e^x - e^{-x} \\
\cosh ^2 - \sinh ^2 = 1
\end{cases}$

**Nota: quando usare trig iperboliche**

se esponente è reale:  
$e^x + e^{-x} = 2 \cosh x$  

se esponente è immaginario:    
$e^{ix} +e^{-ix} = 2 \cos x$  
nota che non ci va la i nell'argomento  
la i va solo accanto al 2 per distinguere cos e sin, in entrambi i casi (iperbolico e non)  

**Derivate**

$\frac{d}{dx} \cosh x = \sinh x$  
$\frac{d}{dx} \sinh x = \cosh x$   

there is not - sign in neither, unlike with normal sin and cos  

**Taylor approx**

$\sinh(x)=x$  
$\cosh(x) =1$  
come per trig normali

### Cose macroscopiche da calcolare

#### Energia interna e capacità termica

**Formula**

$U=-\frac{\partial}{\partial \beta} \ln (Z_n) = -N\frac{\partial}{\partial \beta} \ln (Z_1)$  
   
$C=\frac{dU}{dT} = \frac{dU}{d\beta} \frac{d\beta}{dT}$  

vale sia per particelle distinguibili che non  
usa sempre la seconda versione  

**Esempio**

1- trova dipendenza di $Z_1$ da $\beta$:  
$Z_1 = A \; \beta ^{-5/2} \; (1-e^{-\beta V})$  

2- calcolo U  
$U = -N \frac{ \partial}{\partial \beta} [ \ln (\beta ^{-5/2})+ \ln (1-e^{-\beta V}) ]= -N (-\frac 52 \frac 1 \beta + \frac{-Ve^{-\beta V}}{1-e^{-\beta V}}) = \frac{5N}{2 \beta} + NV \frac 1{e^{\beta V}-1}$  

3- calcolo C  
$C = \frac{dU}{d\beta} \frac{d\beta}{dT} = \frac 52 N k_B + (-\frac 1{k_B T^2}) \, NV \, \frac{V e^{\beta V}}{(e^{\beta V}-1)^2 }  =N k_B [\frac 52 - (\beta V)^2\frac{e^{\beta V}}{(e^{\beta V}-1)^2 }]$

tip: se l'esponente di e è complicato, fai una sostituzione  
tipo: $e^{-mghz} = e^{-\eta z}$  


**Tips calcolo**

1- calcolare U  
considera solo i fattori di Z dove appare $\beta$  
il log trasforma un prodotto (Z) in una somma (U), non serve la derivata del prodotto   
ricorda: $[\log(f)]'= \frac{f'}{f}$  
ma prima posso portare fuori l'esponente  

2- calcolare C   
so che: $ \frac{d\beta}{dT} = -\frac 1 {k_B T^2}$  
e: $\frac{1}{\beta} = k_B T$, utile per fare veloce la parte traslazionale  

3- se ho molti esponenziali (magari per stati discreti):  
scrivili in trig iperboliche   
o moltiplica per $e^x$, faccio $\frac{e^{-x}}{1-e^{-x}} = \frac {1}{e^x-1}$  

#### Energia libera di Helmholtz ed entropia

**Formule**

$F=-k_B T \log Z$  

$S=-\frac{\partial F}{\partial T}$  

**Cosa scrivere**

1- per calcolare F:   
$F= -k_BT \log (\frac{Z_1^N}{N!}) = -k_BT \log [(Z_1 \frac{e}{N})^N] = -k_B T N \log (Z_1 \frac eN)$  
per N grandi: $N! \sim N^N e^{-N}$, così porto tutto dentro l'esponente   
tutta la rob dentro il log lasciala così, non toccarla  
poi la manipolerò per calcolare l'entropia   
IN PRATICA, non faccio niente  

2- per calcolare S    
$S = k_B \log (Z_n) + k_BT \frac{\partial}{\partial \beta} (\log Z_n ) \frac{\partial \beta}{\partial T} =  k_B \log (Z_n) + k_BT U \, (-\frac{1}{k_B T^2}) = k_B N \log (Z_1 \frac eN) + k_B\beta U$  
il primo addendo lo trasformo come per F  
nel secondo ritrovo la formula di U  

3- per trovare limiti di S ad alte e basse T  
i limiti di U li ho già calcolati, plugga quelli  
poi calcola i limiti $Z_1$, e plugga anche quello (è dentro il log, quindi dopo devo farci il log)  

**Esempio**

$Z_1=\frac 1{h^3} (\frac \pi \beta)^{5/2} (2m)^{3/2} \frac{4e}{3AN}[1+7\beta AR^3-1]e^{-\beta AR^3}$  

1- separo in Z le parti che contengono $\beta$  
$Z= C Z_\beta = C (\beta^{-5/2})(1+[7\beta AR^3-1]e^{-\beta AR^3})$  

2- scrivo F:    
$ F = -k_B TN \log (Z_1 \frac eN)$  

3- scrivo i due addendi di S:  
$S = k_B N \log (Z_1 \frac eN) +k_B TN (\frac{\partial}{\partial \beta} \log (Z_1 \frac eN) ) \, (-\frac{1}{k_B T^2}) = k_B N \log (Z_1 \frac eN) + k_B\beta U$  

### Regimi di alta e bassa T

#### Come si fa

**Esprimila in funzione di T**

vuol dire scambia le $\beta$ con: $\frac{1}{\beta} = k_B T$  

**Regimi con alta e bassa T: calcola analiticamente**

bassa T: $T \sim 0, \quad \beta \sim \infty$  
alta T: $T \sim \infty, \quad \beta \sim 0$   

steps:   
1- voglio avere frazione polinomiale, quindi togli esponenziali o trig, espandi con taylor   
se arg tende a 0 usa esapensione di Taylor $e^{x}=1+x$, se arg tende a $-\infty$ vale 0, se tende a $\infty$ vince contro il resto  
2- nella somma vince il più forte se il limite tende a infinito e il più debole se tende a 0  
3- nella moltiplicazione vince sempre il più forte   
4- plugga il valore di $\beta$, cioè 0 oppure $\infty$  

**Esempio**

$U= NAR^3[\frac 52 \frac{1}{\beta AR^3} +\frac{7\beta AR^3 -8}{7\beta AR^3-1+e^{\beta AR^3}}]$  

alta T: $\beta \rightarrow 0$   
$U \sim NAR^3 [\frac 52 \frac{1}{\beta AR^3} + \frac{7\beta AR^3 -8}{8\beta AR^3}] = \frac N \beta [\frac 52 + \frac 78 \beta AR^3 -1] \sim \frac N \beta \frac 32$

bassa T: $\beta \rightarrow \infty$   
$U \sim NAR^3 [\frac 52 \frac 1{\beta AR^3} + \frac 1{e^{\beta AR^2}}] = \frac 52 \frac N \beta$  
perchè quello di destra va a 0 prima  

**Esempio**

$C=N k_B [\frac 52 - (\beta V)^2\frac{e^{\beta V}}{(e^{\beta V}-1)^2 }]$

bassa T: $\beta \rightarrow \infty$  
$C \sim Nk_B \frac 52$  
perchè l'esponenziale al quadrato va a 0 per primo    

alta T: $\beta \rightarrow 0$  
$C \sim Nk_B [\frac 52 - (\beta V)^2 \frac{1}{(\beta V)^2}] = Nk_B \frac 32$  
perchè uso taylor per trasformare gli esponenziali in polinomi  

#### Spiegazione

cosa | T bassa | T alta
---|---|---
prob microstati | livello di energia più bassa ha prob=1, gli altri p=0 | tutti stati equiprobabili
perchè | $\beta = \infty$, anche piccole differenze di E diventano enormi nel peso | $\beta = 0$, quindi il peso è uguale per tutti gli stati (si perde dipendenza da E)  
conseguenza | il sistema si congela nello stato più stabile ('grado di libertà interno è congelato al livello più stabile') | come microcanonico
valori comuni | $U \sim \frac 52  \frac N \beta \quad C \sim \frac 52 N k_B$ | $U \sim \frac 32  \frac N \beta \quad C \sim \frac 32 N k_B$  




**Note** 

$\beta$ è il coefficiente dell'energia nel peso, $e^{-\beta E}$    
$\beta =0$ vuol dire che l'energia non influsice affatto sull'estrazione (tutti equivalenti)  
$\beta=\infty$ vuol dire che l'energia è importantissima, quindi riesco ad estrarre solo lo stato migliore in assoluto  

in entrambi i regimi, U e C non cambiano per piccole variazioni di T   
tanto le prob rimangono più o meno sempre 100% sullo stato con energia minore, o tutti equiprobabili  

T intermedia:  
C ha un punto di massimo e minimo  
quindi C non è monotona in T  

**Significato fisico: esempi**

1- ho microstati discreti, ognuno con una sua energia  
T alta: tutti microstati equiprobabili,  $<s>=\Sigma s \, p(s) = \Sigma s$ (media aritmetica)  
T bassa: tutte particelle nel microstato con E più bassa, $<s> = s_{min} \, p(s_{min}) = s_{min} \, 1 = s_{min}$  

2- ho contenitore finito con dentro gravità   
T alta: tutte posizioni equiprobabili, cioè le m sono in tutto il contenitore    
T bassa: tutte m nel punto con E minore, cioè il pavimento (ha energia grav minore)  

**Temperatura caratteristica**

se in Z o altre cose appare un esponenziale, tipo $e^{-\beta mgH}$    
i regimi sono:  
$\beta mgH \ll 1 $, oppure $mgH \ll k_B T$, cioò bassa T  
$\beta mgH \gg 1 $, oppure $mgH \gg k_B T$, cioò alta T  

poi, per fare i limiti, uso $T \sim 0$ e $T \sim \infty$  

### Altre domande

#### Regime di R

**Riflessioni: regime di R**

usa il teorema di equipartizione:  
$U = \frac {n_{dof}}2 \; N k_B T$  

**Contare i dof**

devo scrivere la H, e contare quanti $p^2$ (dof cinetici) e $x^2$ (dof potenziali) ci sono  

attenzione: usa coord cartesiane per scrivere H, NON polari, perchèdevo contare i termini quadratici, e in coord polari è strano  

**Regime di R**

situazione: gas in un cilindro, con asse lungo z  
c'è potenziale armonico $V(r) \sim r^2$   
energia in coord cart: $H=\frac{p_x^2}{2m}+\frac{p_x^2}{2m}+\frac{p_x^2}{2m} + \frac k2 (x^2+y^2)$  

1- caso $R \rightarrow \infty$  
gradi di libertà:  
3 cinetici (x,y,z)  
2 potenziali (perchè V dipende solo da x e y, NON da z, visto che V è radiale non ha componente assiale)   
quindi $U = \frac 52 Nk_BT$  

2- caso $R \rightarrow 0$  
gradi di libertà:  
3 cinetici (perchè non può muoversi lungo x, ma può acceleare, quindi $p_x \neq 0$)  
0 potenziali (perchè $x=y=0$, non ci sono termini quadratici)  
quindi: $U = \frac 32 Nk_BT$

$E=E_{cin}+E_{int}$

risultato noto per microcanonico: $\Omega(E_{cin}) ∝ (E_{cin})^{3N/2}$

$(E-n \epsilon)^{3N/2} \sim (1- (n\epsilon)/E)^{3N/2} \sim e^{-\overline \beta n \epsilon}$  
con $\overline \beta = 3N/2E$
