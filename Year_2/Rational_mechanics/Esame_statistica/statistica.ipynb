**Formule**

Cosa | formula 
---|---  
peso di Boltzmann | $e^{-\beta E}$  
funzione di partizione | $Z = \frac 1 {h^3} \int d^3p \; d^3q \; e^{-\beta E}$  
densità sullo spazio delle fasi (peso di Boltzmann) | $\pi(q,p) = \frac {e^{-\beta E}}{h^3 \; Z} = \frac {e^{-\beta E}}{\int d^3p \; d^3q \; e^{-\beta E}}$ 
densità spaziale (prob marginale) | $\pi(q) = \int d^3p \; \pi(q,p) = \frac {e^{-\beta E}}{\int  d^3q \; e^{-\beta E}}$  
valore medio | $<A> = \int d^3p \; d^3q \; A(q,p) \; \pi(q,p)$    


#### Teoria

**Teoria**

densità sullo spazio delle fasi: normalizzando il peso di Boltzmann  
densità spaziale: prob marginale, integro su tutti i valori di p perchè non mi interessa  
valore medio: tutti i valori possibli, pesati con la densità sullo spazio delle fasi  

**Definizioni**

1- $\pi(q,p)$: densità di probabilità di un microstato (distrbuzione completa, canonica)   
cos'è: prob che una particella abbia questa posizione e questo impulso (cioè tutte le var di E fissate)  

2- $\pi(r)$: probabilità marginale  
cioè: prob che una particella abbia posizione r, mentre il momento non importa, ne può avere uno qualunque (quindi integro su tutti quelli possibili)  

nota:  
se una sola particella, metti $Z_1$  
se mettessi $Z$, sarebbero le prob che l'intero sistema avesse qualche caratteristica, quindi che fosse preso da un certa regione dello spazio delle fasi  

#### Prob marginale

**Shortcut**
 
prendo il peso di Boltzmann in cui compare ciò che mi interessa (ad esempio, la posizione), e lo integro/sommo su:  
denominatore: valori totali assumibili  
numeratore: valori che mi vanno bene, per cui voglio la prob (se mi va bene solo una specifica coord r, non devo integrare)    

**Steps: trova prob marginale di qualcosa**

1- scrivi l'energia E, vedi che variabili ha  

2- individua che valore voglio che abbia ogni var  

3- vedi se tra gli integrali sopra e otto posso semplificarne qualcuno  

**1. Variabili di E**

dipende da:  

1- SEMPRE da q e p, anche se non appaiono  
se appaiono le coord, invece di un generico q puoi chiamarle r oppure x  

2- SE appare un parametro, anche da quello   
ad esempio, s per lo stato interno  

**2. Che valori voglio abia ogni var**

valore | cosa fare 
---|---
qualunque valore (non mi interessa) | integra per quella var, su tutto il dominio <br> non appare più in argomento della prob (perchè si semplifica)   
un valore specifico (voglio prob se quella var ha quel valore preciso) | non integrare nulla <br> lascia la var generica come argomento della prob
un insieme di valori (tipo va bene qualunque punto dentro una certa regione D)| integra mettendo come estremi di integrazione i valori accettabili, tipo D  <br> come argomento della prob metti D

**3. Semplifica tra numeratore e denominatore**

come semplificare correttamente:  
posso semplificare due integrali se hano stessa integranda (solo la parte con dipende da quel differenziale mi interessa) e gli stessi estremi di integrazione    
nota che Z integra su tutti i valori possibili di un parametro (o somma, se livelli discreti)  
quindi se per una var mi vanno bene tutti i valori, allora il numeratore e denominatore si semplificano a vicenda  

recap:  
alla fine ho una frazione con solo il peso di Boltzman con la var che mi interessa (quella tra le parentesi di $\pi$)  
sopra: peso con quella var, su tutti gli stati accettabili  
sotto: peso con quella var, su tutti gli stati possibili (normalizzazione)  

**Densità spaziale: forme utili**

$\pi(q) = \int d^3p \; \pi(q,p)$  
cosa: prob che si trovi nella coordinata q  

$\pi(r)=\int d^3p \; \pi(q,p)$  
cosa: prob che sia a distanza r  
come: fai cambio differenziale, tipo $d^3q =  2 \pi r \, dr$, mentre $e^{-\beta E}$ dovrei averlo già in funzione di r  

$\pi(D)=\int d^3p \; \int_D  d^3q \; \pi(q,p)$    
cosa: prob che si trovi nella regione D   
come: integra q su tutti i valori accettabili, quindi sullo spazio D  

nota: se V(q) è definita a tratti, anche $\pi(q)$ sarà definita a tratti  
in ogni tratto, pluggo nel denominatore V(q) come è definita in quel tratto  
mentre al denominatore integro sempre su tutte le q possibili, anche negli altri tratti, perchè è per normalizzare  

#### Valore medio

**Semplificare integrali**

gli estremi di integrazione degli integrali sono uguali sopra e sotto   
quindi se l'integranda è uguale, posso cancellarli sopra o sotto  
cioè, se $A(q,p)$ è in realtà $A(q)$, posso cancellare gli integrali in p  

**Formula**

$⟨s⟩ = \Sigma_D s \, p(s) = \int_D s \, p(s) \, ds$   
a seconda che s possa avere valori discreti o continui   
D: insieme dei valori di s accettabili (intervallo di integrazione o tutti i valori su cui sommo)  

**Media di una distrbuzione**

$⟨f(x)⟩=\frac{∫f(x)w(x)dx​}{∫w(x)dx}$  
il denominatore è per normalizzare la prob, se è un peso relativo e non una prob   
qui il peso è quello di Boltzmann, $w(x)=e^{-\beta E}$  

#### Esempi: prob marginale

**Prob di stato interno**

voglio prob che una particella abbia uno specifico stato interno  

1- $E(q,p,s)=\frac{p^2}{2m}+ks \quad $ dove $s \in [-1, 0, +1]$  

2- che valori possono avere:    
$q,p$: qualunque, non mi interessa   
$s$: uno specifico  
$\pi(s) = \int d^3q \; d^3p \; \pi(q,p,s)$  

3- esplicito la densità di probabilità  
$\pi(s) = \frac{1}{h^3} \frac{\int d^3q \; d^3p \; e^{-\beta E(r,p,s)}}{Z_1} =  \frac{\int d^3q \; d^3p \; e^{-\beta p^2/(2m)}e^{-\beta ks}}{\int d^3q \, d^3p \; e^{-\beta p^2/(2m)}\Sigma e^{-\beta ks}} = \frac{e^{-\beta ks}}{\Sigma e^{-\beta ks}}$  
è rimasta solo la var s, dscreta   

**Esempio: prob di posizione in una certa regione**

1- $E(q,p)$  

2- che valori possono avere  
$p$: uno qualunque  
$q$: all'interno della regione D  
$\pi(D) = \int_D d^3q \; \int d^3p \; \pi(q,p)$  

3- pluggo:
$\pi(D) = \frac{\int_D e^{-\beta E} d^3q}{\int_V e^{-\beta E} d^3q}$  
è rimasto solo l'integrale su q, continua 

**Esempio: prob di una specifia posizione**

1- $E(q,p) = \frac{p^2}{2m} + V(r)$  

2- che valori possono avere:  
$p$: uno qualunque  
$q$: uno preciso  
$\pi(q) =  \int d^3p \; \pi(q,p)$  

3- pluggo:  
$\pi(q) =  \frac{e^{-\beta V(r)}}{\int d^3q \; \; e^{-\beta V(r)}}$  

**Esempio: simmetria sferica**

ho trovato $\pi(q)$  
ma voglio $\pi(r)$, e ho simmetria sferica  

come: cambio var $dq^3 = 4 \pi^2 dr$

$E=\frac {p^2}{2m}+ar^3$   
$\pi(q)=\frac{e^{-\beta V(r)}}{\int e^{-\beta V(r)} \, d^3q}$  
$\pi(r) = \frac{e^{-\beta  ar^3}}{\int_{R_1}^{R_2} e^{-\beta  ar^3} \, 4 \pi^2 \, dr} = \frac{e^{-\beta ar^3}}{e^{-\beta aR_2^3 }-e^{-\beta a R_1^3}} \frac{a}{\frac 43 \pi}$ 

#### Esempi: valore medio

**Esempio: valore medio dello stato interno**

ho 3 stati discreti: s=-1, 0, +1  
calcola la media di s  

1- calcolo la prob marginale (di uno specifico stato):  
$p(s) = \frac{e^{-\beta \, ks}}{\Sigma e^{-\beta ks}}= \frac{e^{-\beta ks}}{e^{\beta k} + e^{-\beta k} + 1}$  

2- uso la formula di media:  
$⟨s⟩=\Sigma_D s \, p(s) = (-1)p(-1)+(1)p(1)+(0)p(0)$  
perchè $s \in D = [-1, 0, +1]$  

**Esempio: valore medio del raggio**

1- prob marginale:   
$\pi(q) =  \frac{e^{-\beta V(r)}}{\int d^3q \; \; e^{-\beta V(r)}}$

2- uso la formula di media:  
$⟨r^3⟩= \int_D r^3 \, \pi(r^3) = \frac{\int_D d^3q \, r^3 e^{-\beta V(r^3)}}{\int_D d^3q \,  e^{-\beta V(r^3)}}$  
anche qui, per risolvere devo fare cambio coord, per simmetria sferica

dipened dalla forma della forza

$<F_r> = k <r>$  

invece in coord cartesiane:  
$F_z=0$, perchè la forza è radiale, non assiale  
$F_x = F_y$, perchè non c'è ragione per cui una dovrebbe essere maggiore dell'altra  


**Valore più probabile di r ???**

1- prob di un punto: $\pi(r)$  
$\pi(q) = \int d^3p \; \pi(q,p) = \frac {e^{-\beta E}}{\int  d^3q \; e^{-\beta E}} = \frac {e^{-\beta V}}{\int  d^3q \; e^{-\beta V}}$  

2- prob di una distanza:  
$\pi(r) \, dr = \pi(q) \, d^3q = \pi(q) \, 2 \pi r \, dr$  
$\pi(r) = 2 \pi h \, r \pi(q)$  

3- voglio trovare r per cui la prob è massima  
cioè r che massimizza:  
$\pi(r) \sim r \; e^{-\beta V}$  
fai la derivata per r del lato destro, mettila a 0, vedi che r serve   

esempio: forza lungo r  
$<F_z> = \frac{ \int_0^\infty kr \; e^{-\beta \frac 12 kr^2} \, r \, dr}{ \int_0^\infty \; e^{-\beta \frac 12 kr^2} \, r \, dr}$  
$<F_x> = \frac{ \int_{-\infty}^\infty kx \; e^{-\beta \frac 12 kx^2}  \, dx}{ \int_{-\infty}^\infty \; e^{-\beta \frac 12 kr^2} \, dx}$  
nota che: $d^3q = 2 \pi h \; r \, dr = dx \, dy \, dz$   

#### Prob che una massa sia più in alto di un'altra

**Situazione**

ho due masse, m e M  
in un contenitore alto H, con dentro la gravità  

voglio prob che M sia sopra la massa m  
(la domanda potrebbe essere: estrendo a caso una m e una M, M deve stare sopra - l'estrazione non vuol dire niente qui, non va conteggiata col calcolo combinatorio)

**Come si fa**

1- prob che una massa $\mu$ sia ad altezza z  
$p_\mu(z) = \frac{e^{-\beta \mu gz}}{\int_0^H e^{-\beta \mu gz}} = \beta \mu g \; e^{-\beta mu gz}$  
spiegato sopra, il peso di Boltzmann integrato su valori accettabili (solo quella z) fratto totali    

2- prob che una mssa $\mu$ sia sopra una certa altezza $z$:  
$P (Z_\mu >z) = \int_z^H \, p_\mu(z) \, dz \sim e^{-\beta \mu gz}$   
P scritta in maiuscolo perchè è una densità cumulativa, integro la prob della presenza della massa in un punto z su tutti i punti che mi vanno bene (cioè quelli sopra z)   
qui ho usato $H = \infty$, per il regime che mi dava il testo  
Z è la variabile aleatoria da estrarre, z è la quota non precisata che voglio superare  

3- prob che m sia sopra M  
$p(M>m) = \int_0^H p_m(z) \, P(Z_M > z) \; dz = \int_0^H \beta m g \, e^{-\beta m gz} \; \; e^{-\beta Mgz} \; dz= \beta m g \int_0^H e^{-\beta gH (m+M)} \; dz \sim \frac{m}{m+M}$  
per ogni altezza possibile, probabilità che m sia a quell'altezza, e M sia sopra  
anche qui, riuso $H = \infty$
