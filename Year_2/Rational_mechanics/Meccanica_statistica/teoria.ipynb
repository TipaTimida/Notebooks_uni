### Non-distinguibilità delle particelle

**Cosa vuol dire 'indistinguibili'**

facile:  
c'è simmetria sotto scambio di particelle  
se facessi una foto, non potrei distinguere una dall'altra 

lungo:  
prendo due particelle specifiche, le chiamo x1 e x2  
queste due sono indistinguibili se:  
in due sistemi in cui x1 e x2 sono scambiate, i due sistemi evolvono esattamente nello stesso modo  

**Conseguenze**

problema: i due microstati con x1 e x2 scambiati vanno considerati un solo microstato  

soluzione: devo dividere il volume dello spazio delle fasi per $N!$, il numero di permutazioni delle particelle    
questo SOLO per particelle indistinguibili  

nota: ignoriamo i casi di scambio di 3 o più particelle, tanto sono molti meno  
1/N! si chiama 'fattore di Gibbs'  

**Spiegazione**

1- perchè i due sistemi con scambio di x1 e x2 sono un unico microstato?  
abbiamo definito il microstato come la più dettagliata descrizione microscopica possibile per lo stato meccanico del sistema, cioè quella che contiene tutta l’informazione fisicamente accessibile  
se le leggi della Natura ci dicono che non è possibile distinguere le due particelle, contare due microstati differenti vorrebbe dire assegnare dell’informazione che in realtà non esiste  

2- perchè divido per N! ?  
nello spazio delle fasi, quando conto i microstati in ogni zona, se conto tutti i punti ne sto contando più di quelli che ci sono davvero  

**Entropia e indistinguibilità**

l’indistinguibilità riguarda l’informazione accessibile da un punto di vista microscopico  
causerà problemi nell'entropia, che è in qualche modo una misura dell’informazione contenuta nella nostra conoscenza macroscopica  

N! serve a:  
1- garantire l’estensività dell’entropia   
2- risolvere il paradosso di Gibbs:   
il mescolamento di due gas identici genera un aumento di entropia, contrariamente all'evidenza macroscopica che non mostra cambiamenti (nelle var de gas, tipo T)  

quindi il fatto che serve per l'entropia è un'altra dimostrazione del fatto che ci deve stare 1/N!  

**Estensiva**

estensiva: raddoppia se considero due sistemi identici, eg: V, S    
intensiva: non raddoppiano, eg. T

### Energia termica

**Definizione**

$\frac{\partial U}{\partial T} =C_V$  

l'avevo già scoperto nella termodinamica normale 

**Cambiamenti di energia**

definizioni:  
$<\delta E^2>$: fluttuazioni spontanee dell'energia all'equilibrio (a temperatura fissa)  
$k_BT^2 C_V$: quanto l’energia interna cambia se io vario la temperatura   

formula:  
$<\delta E^2> = ... = \frac{\partial^2}{\partial \beta^2} \log Z = \frac{\partial}{\partial \beta} U = -\frac{\partial T}{\partial \beta} \frac{\partial U}{\partial T} = k_B T^2 C_V$  
dove: $<\delta E^2> = <(E-U)^2> = <E^2>+U^2 -U<E>=<E^2> -<E>^2$  

cosa dice:   
lega le fluttuazioni dell’energia a temperatura fissa alla variazione dell’energia interna sotto variazioni di temperatura   
È un caso particolare di una classe di relazioni generali che legano le fluttuazioni del sistema alla sua risposta alle variazioni dei parametri esterni  

**Fluttuazioni dell'energia**

in un gas ideale: $U= 3/2 \; Nk_B T$  
quindi: $\frac{\sqrt{\delta E^2}}{U} \sim \frac{1}{\sqrt{N}}$  

che vuol dire:  
aumentando il numero di particelle, le fluttuazioni diminuiscono  
quindi al limite per $N \sim \infty$ le fluttuazioni sono sparite, l'energia di ogni microstato è fissa (come nel microcanonico, anche se qui siamo in canonico)     

**Limite termodinamico**

cos'è:  
limite in cui la dimensione del sistema diventa molto grande (sia nel numero di particelle N che nel volume V, a quindi densità N/V costante)  
nota: per un gas interagente la densità importa, quindi quando faccio il limite è importante mantenerla  
per uno non interagente, devo comunque mantenerla, perchè se diminuisse avrei le particelle più vicine, e non sarebbe più vero che è non interagente   

cosa succede:  
in tale limite la distinzione fra ensemble microcanonico ed ensemble canonico diventa del tutto irrilevante da entrambi i punti di vista:  
l’energia è di fatto fissata anche nel caso dell’ensemble canonico (perchè ho calcolato che le fluttuazioni sono bassissime)   
nell’ensemble microcanonico la distribuzione per la singola particella diventa
quella di Boltzmann-Gibbs, come per il canonico (perchè ogni singola particella “vede” il resto come un termostato)

Esistono tuttavia situazioni in cui tale equivalenza non è più valida: sono quelle in cui il raggiungimento del limite termodinamico presenta delle criticità, ad esempio in presenza di transizioni di fase.

#### Primo principio della termodinamica

**Formule**

forma differenziale:  
$dU = \partial Q -\partial L = T dS -P dV$  

da questa, posso vedere le derivate parziali di U:  
$(\frac{\partial U}{\partial S})_V = T$  
$(\frac{\partial U}{\partial V})_S = -P$  

**Note: derivate parziali**

con le derivate parziali è fondamentale quale var lascio costante  

esempio:  
sopra vedo: $\partial_V U(S,V)=T$  
ma so anche $U=\frac{3}{2}Nk_BT$, da cui direi: $\partial_V U(V,T)=0$  
sono entrambe giuste, perchè nella seconda tengo costante la temperatura, nella prima l'entropia  
quindi mi sto chiedendo se l'energia interna cambia durante un'espansione isoterma o a entropia costante, sono trasformazioni diverse  

**Note: differenziali esatti**

quando faccio il differenziale di una qualche grandezza f:  

1- se f è una funzione di stato: ho un differenziale esatto df  
cioè il valore scambiato tra due stati dipende solo dallo stato iniziale e finale  
inoltre so che esiste una funzione f(x,y...)  

2- se f non è una funzione di stato: ho un differenziale non esatto $\partial f$  
quindi il valore dipende dal percorso  
e non esiste una funzione f(x,y...)  

sempio:  
$dU$: vuol dire che esiste una funzione di stato $U(S,V)$, e la differenza dipende solo dai punti di inizio e fine  
$\partial L$: vuol dire che non esiste una funzione di stato $L(S,V)$, e il lavoro dipende dal percorso, quindi dal tipo di trasformazione compiuta (isocora, isobara, isoterma...)  

#### Entropia

**Definizione**

$S = -k_B \Sigma p_i \log p_i$  

**Formule**

microcanonico: $S= k_B \log W(E)$  
perchè uguale prob per ogni microstato, quindi $p_i = 1/W(E)$, con W(E) il numero di microstati permessi con una data energia  
la sommatoria diventa solo moltiplicare per W(E), quindi cancella p_i  
fu scritta da Boltzmann, a meno di costanti moltiplicative, in quanto la definizione di costante di Boltzmann fu in realtà introdotta da Planck  

canonico: $S = \frac{U}{T}+k_B \log Z$    
basta sostituire: $U= \Sigma p_iE_i$ e $p_i = e^{-\beta E/Z}$  

$U=\Sigma p_iE_i$  
$dU = \Sigma E_i \; dp_i + \Sigma  p_i \; dE_i$  
primo addendo: variazione dovuta al fatto che il sistema si distribuisce in modo diverso fra i vari microstati accessibili, mantenendo l’e-
nergia meccanica dei vari microstati invariata;  
cioè il calore scambiato  
seconda parte: associata
alla variazione dell’energia meccanica dei singoli microstati, mantenendo la
distribuzione invariata, cioè alla sola variazione da parte di agenti esterni dei
parametri che fissano l’Hamiltoniana del sistema.  
cioè il lavoro scambiato  

$\partial Q = \Sigma E_i dp_i$  
$\partial L = -\Sigma p_i dE_i$

posso definirla anche senza usare i potenziali termodinamici, ma con la definizione microscopica di lavoro: $\partial L = -\Sigma p_i dE_i = -\Sigma p_i \frac{\partial E_i}{\partial V}dV$    
def macroscopica: $\partial L = P dV$  
confrontandoli, ottengo: $P=-\Sigma p_i  \frac{\partial E_i}{\partial V}$  
la pressione è la media statistica di come i livelli energetici cambiano quando cambi il volume  

questa è P per una trasformazione adiabatica (cioè non scambia calore con esterno, quindi S dell'ambiente non aumenta, ma potrebbe aumentare S del sistema)  

la derivata parziale vuol dire: “come cambia l’energia meccanica di un microstato quando cambio un parametro esterno dell’Hamiltoniana”  
se faccio questa derivata, ottengo P=2U/(3V), che è l'equazione dei as perfetti  

usare l'energia libera è comunque meglio  

### Probabilità

**Peso di Boltzmann**

peso di Boltzmann: $e^{-\beta E}$  
è un peso relativo, non una probabilità  
quanto è probabile occupare uno stato di energia E, quando il sistema è a temperatura T  

nota: ha senso soltanto in ensemble canonico  
perchè in microcanonico, tutti i microstati hanno peso uguale  

**Probabilità di estrarre un microstato**

1- canonico:   
discreta: $p_i = \frac{e^{-\beta E_i}}{Z}$  
continua: $p(E) =  \int \pi(r,p) \; d\Gamma =  \frac{\frac{1}{h^3} \int d\Gamma \; e^{-\beta E}}{Z} $

2- microcanonico: $p_i= const$   

nota: la Z serve a normalizzare, a farla diventare una probabilità  
anche quella si può scrivere in forma continua o discreta  

**Più avanzato: probabilità di trovare particella in una specifica regione**

come: uso la probabilità continua, e l'integrale sulle coordinate lo faccio sulla regione richiesta   
$p(D) = \frac{\frac{1}{h^3}\int_D e^{-\beta E} d\Gamma}{Z_1}$  

**Definizioni**

1- densità di probabilità nello spazio delle fasi   
$\pi(r, p) = \frac{1}{Z_1 h^3} e^{-\beta E}$  

2- densità spaziale (densità di probabilità per unità di volume)  
$\pi(r) = \int d^3p \; \pi(r,p)$  
nota che integro solo per dp, quindi l'esopnenziale dell'energia potenziale non viene integrato, è costante, esce da lì 

3- probabilità che una particella sia in una certa regione spaziale D  
$p(D) =  \int_D \pi(r) d^3r = \int_D \pi(r,p) \; d\Gamma$  

4- densità di particelle  
$\rho(r) = N \pi(r)$  

Per un gas perfetto monoatomico classico di N particelle, sappiamo che il numero di microstati (o meglio il volume di spazio delle fasi) con energia cinetica K scala come:
$Ω_{tr}(K)∝K^{3N/2}$  

**Esercizio: In microcanonico**

dati:  
$E$: energia totale sistema fissata (perchè microcanonico)  
n particelle in microstato con energia $\epsilon$  
N-n particelle in microstato con energia 0  
voglio la prob di questa configurazione  

risolvi:  
ho gas perfetto monoatomico di energia traslazione (cinetica): $E_t = (E-n \epsilon)^{3N/2}$  
manipolo:  
$E_t = (E-n \epsilon)^{3N/2} = E^{3N/2} \frac(1-\frac{n\epsilon}{E})^{3N/2}  \propto \frac(1-\frac{n\epsilon}{E})^{3N/2}  \sim e^{\beta n\epsilon}$  
dove $\beta = \frac{3N}{2E}$  

prob di una certa configurazione:  
$P_n = (\frac{N}{n}) e^{-\beta n \epsilon} = (\frac{N}{n}) p_1^m p_0^{N-n}$  
dove: $p_1 = \frac{e^{-\beta \epsilon}}{1+e^{-\beta \epsilon}}$  
$p_0 = \frac{1}{1+ e^{-\beta \epsilon}}$
