**Moltiplicatori di Lagrange: significato**

condizione per cui la funzione f risulti stazionaria lungo il vincolo:  
gli spostamenti lungo il vincolo (g) devono essere tutti ortogonali al gradiente di f   
che non deve quindi necessariamente annullarsi  
$\nabla f = \lambda \nabla g$  

spiegazione:  
disegno la funzione f sul piano, quindi tipo una montagna che va su e giù, l'altitudine è il valore di f  
disegno il vincolo g, che sarebbe una strada  
che percorso (insieme di punti) voglio fare io:  
deve essere ortogonale al gradiente di f: csì il valore di f non cambia, f è stazionaria lungo il vincolo  
deve essere ortogoanle al gradiente di g: così nel vincolo $\nabla g = c$, non cambia il valore di c, il vincolo rimane soddisfatto  
quindi, i gradienti di f e g devono essere paralleli, come si vede dalla formula  

questo vuol dire trovare i punti stazionari della funzione: $f-\lambda g$

nota: se ci sono più vincoli, lo spostamento deve essere ortogonale a tutti  
quindi riscrivo tante volte la formula, con tutti i $g_i$ diversi  

**Teorema del limite centrale**

Se prendi molte variabili casuali indipendenti, tutte distribuite allo stesso modo, e fai la loro media, allora:
1- la media ha una distribuzione gaussiana
2- la sua varianza è la varianza della singola variabile divisa per N  

esempio: lanci di un moneta  
ogni lancio è una var casuale: mi esce un numero compreso da 1 a 6, che segue una distribuzione uniforme  
se faccio la media di N lanci, tale valore non seguirà una distribuzione uniforme, ma una gaussiana  
$q = \frac{1}{N}\Sigma^N X_i$  


#### Calcolo combinatorio

**Modi di fare una parola**

$M=\frac{n!}{n_a!n_b!...}$  
$M$: molteplicità della distribuzione  
$n$: lunghezza della parola  
$n_a$: quante volte appare la lettera a  

in meccanica statistica non uso le lettere a,b,..., ma $N_1, N_2$... cioè quanto volte durante l'estrazione sono usciti il microstato/celletta 1, il 2..., e N il numero di estrazioni  

facendo il calcolo, trovo:  
$\log M = -\Sigma_i p_i \log p_i$  

**Approssimazione di Stirling**

$n! = (\frac{n}{e})^n \sqrt{2 \pi n}$  
$\log(n!) \sim (n \log(n)-1)+\frac{1}{2}\log(2 \pi n)$

### Ipotesi ergodica

**Media temporale**

$\overline{O}(t) = \frac{1}{\Delta t}\int_t^{t+\Delta t} O(t) \, dt$  

cos'è:  
integrare qualcosa su un periodo e dividere per il tempo di un periodo  
cioè osservare la distribuzione al variare del tempo  

**Media di ensemble - Distribuzione di probabilità sullo spazio delle fasi**

$<O> = \int O(q,p) \, \rho(q,p,t) \, d\Gamma$  
$\rho$: distribuzione di probabilità sullo spazio delle fasi   

cos'è:  
ho costruito $\rho$, una distribuzione di probabilità sullo spazio delle fasi   
studio la distribuzione campionandola tante volte  
campionare, qui, significa creare un sistema a caso, e vedere con che parametri è stato creato  

**Usare le medie nelle dimostrazioni???**

1- media sperimentale / temporale: $\frac{1}{N}\Sigma E_i$  
media delle energie dei microstati realmente osservati nella sequenza  

2- media probabilistica / di ensemble: $\Sigma p_iE_i$  
scritta con la distribuzione di probabilità

**Ipotesi ergodica**

il sistema, evolvendo nel tempo, visita uniformemente tutta la superficie di energia costante nello spazio delle fasi  

spiegazione:  
uniformemente: vuol dire che una celletta è più visitata (vengono estratti più microstati da essa) quanto più il suo volume è grande  
energia costante: perchè è il mio vincolo, ciò che mi dà un sistema canonico    

conseguenza:   
Principio di equivalenza tra media temporale e di ensemble, $<O>=\overline{O}$   
media temporale: sistema reale  
media di ensemble: meccanica statistica  

### Ensemble e stati

#### Definizioni: ensemble

**Cos'è un ensemble**

voglio studiare un sistema  
invece di prendere tante foto dello stesso sistema a t diversi, prendo foto di copie diverse dello stesso sistema, allo stesso t  
questi sistemi (copie) sono distribuiti a caso nello spazio delle fasi, proprio secondo la distribuzione $\pi(q, p, t)$  

**Tipi di ensemble**

a seconda di come è il sistema reale, l'ensemble che creo per descriverlo avrà caratteristiche diverse:  

1- microcanonico  
rappresenta: sistema isolato (con energia nota)  
vincolo: $E$ fissa  

2- canonico:   
rappresenta: sistema non isolato, scambia calore con ambiente/termostato (con energia interna nota)     
vincolo: T fissa  

**Rapporto canonico e microcanonico**

canonico e microcanonico diventano uguali se:  
1- le particelle non interagiscono tra loro   
2- $N \rightarrow \infty$  

spiegazione:  
se $N \rightarrow \infty$, per la regola 3, considero solo l'estrazione più frequente  
che per non interazioni, è quella in cui tutti i microstati hanno stessa prob???  
quindi i due ensemble danno le stesse previsioni macroscopiche  

#### Spiegazione vincoli

**Da dove vengono i vincoli**

1- microcanonico:  
poichè è isolato, E (oppure H) si conserva  
quindi posso estrarre solo microstati con questa E   
cioè  $E_i=E$, cioè ogni microstato deve avere una specifica energia  

2- canonico:  
E non è conservata (sistema non isolato)  
ma il sistema è in equilibrio con un termostato, quindi T resta costante  
quindi resta costante anche U, l'energia l'interna, che è il valore medio dell'energia  
posso estrarre solo sequenze in cui la media delle energie delle estrazioni è U   
cioè $<E> = \frac{1}{N} \Sigma N_iE_i = U$,

**Spiegazione: vincoli e probabilità**

a priori, tutti i microstati hanno uguale prob    
ma se impongono vincoli, questo può cambiare  

1- microcanonico  
vincolo: solo sequenze composte solo dai microstati con quella E (quindi è un vincolo più forte)  
prob: il vincolo mi riduce che stati posso estrarre, ma tra questi ammessi nessuno è favorito, sono tutti equiprobabili  

2- canonico  
vincolo: non limita i microstati ammissibili, ma limita i modi di combinarli, perchè l'energia media alla fine deve essere U  
prob: ogni microstato ha prob diversa  

**Nota: probabilità microstati in canonico**

1- perchè microstati con energia vicina ad U sono più probabili    
$E_i$ in teoria possono avere qualunque valore  
ma è improbabile che siano molto distanti da U, perchè poi è difficile avere U come media  
quindi i microstati con energia più vicina ad U sono più probabili  

2- perchè non scelgo solo microstati con energia U   
ma è anche improbabile che abbiano tutti esattamente $E_i=U$, perchè come visto a lab 1, un po' di varianza è sempre attesa  

#### Definizioni: stati e molteplicità

**Tipi di stati**

sono diversi modi di descrivere il comportamento di un sistema, a livello micro o mascoscopico  

1- microstato: la più dettagliata descrizione microscopica che è possibile fare dello stato meccanico del sistema  
dico: (q,p) di ogni particella del sistema  

2- macrostato: la specifica di varie variabili macroscopiche che permettono di definire la distribuzione dell’ensemble sui microstati   
cioè una descrizione grossolana del sistema, che specifica solo alcune proprietà globali   
dico: solo alcune var macroscopiche del sistema, tipo (P,V,T) 

**Molteplicità**

il macrostato è una descrizione più grossolana  
quindi ad uno specifico set di valori per le var macroscopiche possono corrispondere più descrizioni microscopiche che danno quel risultato  

questa è la molteplicità di un macrostato: quanti microstati possono corrispondere ad un tale set di valori per le var macroscopiche  
quanti microstati sono compatibili con un tale macrostato  

**Descrivere gli stati graficamente**

2 livelli di descrizione:  

1- livello fisico   
micro: un punto nello spazio delle fasi $(q_1, ...q_N, p_1, ...p_N)$    
macro: una regione nello spazio delle fasi (insieme di microstati)

2- livello simbolico  
discretizzo lo spazio delle fasi, quindi lo divido in cellette chiamate A,B,C...
perchè: per il principio di indeterminazione di Heisenberg, non posso sapere con certezza coordinata e velocità di ogni particella  
micro: una lettera, tipo A (vuol dire che il punto che avrei disegnato nel precedente modello cade nella cella con questa lettera)  
macro: una classe di sequenze, tipo '3A e 1B' (vuol dire un insieme di punti che rispetti questa condizione)  

#### Molteplicità

**Formula**

$\log M = -\Sigma_i p_i \log p_i$

**Definizione intuitiva**

M è una misura della nostra ignoranza  
più è alta, più, anche se conosciamo il macrostato, siamo indecisi su da che microstato sia partito il sistema  
quindi, rappresenta il disordine associato al macrostato  

**Molteplicità ed entropia**

formalmente:  
M è la misura delle diverse combinazioni di moti microscopici corrispondenti alla stessa descrizione macroscopica
è legata alla nostra mancanza di informazione sullo stato microscopico del sistema  
quindi al disordine associato al macrostato stesso;  

entropia:   
l'entropia è una misura del disordine   
quindi M è la def di entropia (entropia di Gibbs–Shannon)  
si scrive S  

**Ricavare la formula**

N: quante estrazioni ho fatto  
$N_i$: quante estrazioni sono risultate nella celletta/nel microstato i-esimo  
$p_i$: probabilità (normalizzata) che un'estrazione risulti nella celletta i-esima  
$M$: molteplicità, numero di microstati che corrispondono a un macrostato     

condizioni:  
$\Sigma N_i = N$ (ogni estrazione è per forza in una cella)  
$N_i = Np_i$ (definizione di prob)  
$\Sigma p_i = 1$ (prob è normalizzata)  

cosa viene:  
$\log M = -\Sigma_i p_i \log p_i$

**Costruzione dell'ensemble: graficamente**

1- separo lo spazio delle fasi (piano qp) in cellette  
ogni celletta contiene più punti del piano qp, così ho un numero di celle numerabili, non infinito  
così la probabilità p è discreta, non contninua (perchè i microstati sono numerabili)  
questo anche per il prinicpio di indeterminazione della meccanica quantistica, quindi non posso sapere q e p con certezza  

2- guardando le informazioni macroscopiche che ho, cancello le cellette impossibili  
è la regola 2  

3- estraggo tanti stati    
estrarre uno stato significa fare un puntino a caso nel piano qp (nelle cellette ammissibili), e vedere in che celletta finisce  
è la regola 1  

4- vedo come sono distribuiti gli stati estratti  

5- immaginando di arrivare al limite di infinite estrazioni, mi interessa solo quale esce più frequentemente  
è la regola 3  
nota: NON significa scegli la celletta uscita più spesso  
significa scegli l'estrazione (cioè la sequenza di celle estratte, SENZA considerare l'ordine) più probabile   

**Costruzione dell'ensemble: analiticamente**

1- costruisco la funzione di probabilià $\rho$

2- cerco il massimo della distribuzione  
è la regola 3  

3- facendo attenzione che sia un massimo vincolato  
è la regola 2  

**Regole sull'estrazione**

1- Principio di equiprobabilità a priori:   
cellette di uguale volume nello spazio delle fasi hanno a priori la stessa probabilità di essere estratte   
cioè: l'area di una celletta è proporzionale alla sua probabilità  
serve per verificare il teorema di Liuoville  

2- Non tutte le estrazioni sono ammissibili:  
ma solo quelle compatibili con le informazioni macroscopiche a disposizione    
questo definisce il tipo di ensemble considerato  
cioè: se so E del sistema, posso estrarre solo microstati che portano a quella E  

3- Nel limite N → ∞, l’unica distribuzione di popolazioni Ni da tenere in considerazione è quella che ha la maggior probabilità di essere estratta   
perchè le altre escono infinitamente meno frequentemente  
è un'estremizzazione della legge dei grandi numeri  
altro modo di dirla: tra tutte le distribuzioni ammissibili, scegli quella che rende massima l’entropia (principio di massima entropia)  

#### Ricavare Z

**Trovare il massimo della molteplicità**

$f- \Sigma \lambda g=0$  
se ho piò vincoli, li metto tutti  
nota che la sommatoria se ne va via  
esempio: $-\log p_{a,\alpha} -1-\lambda \; -\beta _a e_a -\beta _\alpha \epsilon _\alpha=0$  

isolo $p$ e nel lato destro trovo cosa è Z  
nota: $\lambda$ è fissato dalla condizione di normalizzazione, cioè $e^\lambda$ diventerà parte di Z  

variante: questo era se so $U_A$ e $U_B$ separatamente, quindi mi dà due vincoli  
se invece so solo la loro somma, $U=U_A+U_B$, e so che A e B sono due sistemi con energia fissata, ma non so quale (come passare da microcanonico a canonico)  
I sistemi A e B sono quindi in contatto termico fra di loro, la distribuzione di
equilibrio corrisponderà a quella di equilibrio termico fra i due sistemi  
Il parametro che è funzione dell’energia interna e che è uguale, all’equilibrio,
per due sistemi in contatto termico, è la temperatura. Quindi il parametro β deve
essere una qualche funzione della temperatura T e tale funzione deve essere universale,
cioè uguale per tutti i sistemi fisici, in quanto possiamo pensare di mettere a contatto
termico una qualsiasi coppia di sistemi, anche diversi fra loro.  



**Esemble microcanonico**

il vincolo che l'ernegia sia E l'ho giù usato per cancellare delle cellette  
mi rimagono: 
funzione: $S$, l'entropia    
vincoli: $\Sigma p_i=1$  

sottraggo f e g:  
$-\Sigma p_i \log p_i \; -\lambda \Sigma p_i$  

$p_i = e^{-(1+\lambda)}$  
$\lambda$: fissato dalla condizione di normalizzazione, quindi dipende dal numero totale di microstati accessibili dato il valore dell’energia (quante cellette non ho cancellato)   
l'estrazioe più probabile è quella in cui ogni microstato appare con uguale frequenza, cioè $p_i = cost$  

**Esempio canonico**

tutti i microstati sono accessibili  
ma la condizione su U impone: $<E>=\Sigma p_i E_i = U$  
quindi oltre alla normalizzazione di p, ho un altro vincolo  

sottraggo f e vincoli:  
$-\Sigma p_i \log p_i \; -\lambda \Sigma p_i \; -\beta \Sigma p_iE_i$  
$\beta$ è un altro moltiplicatore di Lgrange, come $\lambda$    
$p_i=\frac{e^{-\beta E_i}}{Z(\beta)}$  
dove: $Z(\beta) =\Sigma e^{-\beta E_i}$, è tipo un fattore di normalizzazione  
Z si chiama 'funzione di partizione' del sistema  
racchiude buona parte delle informazioni sulla termodinamica del sistema stesso  
appare quando normalizzo la distribuzione canonica  

la funzione di partizione esiste solo nell'ensemble canonico, non nel microcanonico  

**Beta**

che valori può avere: 
se ho infiniti microstati (cioè se $E_i$ possibile per un microstato non ha un limite superiore, come nella realtà), allora deve essere $\beta >0$, sennò $<E>$ sarebbe divergente  
perchè i microstati con energia grange peserebbero troppo, mentre il loro contributo delle andare a 0, visto che sono poco probabili  

### Canonico a temperature diverse

**Ricorda**

peso di Boltzmann: $e^{-\beta E}$  
prob: $\frac{e^{-\beta E}}{Z}$  
$\frac{1}{\beta} = k_B T$  

**T = infinita**

1- pesi di Boltzman:  
poichè $\beta=0$, il peso è uguale per tutti gli stati di stati  
quindi anche la prob è uguale per tutti gli stati  

2- che succede:  
tutti gli stati sono equiprobabili  
come microcanonico  

**T = 0**

1- pesi di Boltzmann  
poichè $\beta = \infty$, anche piccole differenze di E diventano enormi  

2- che succede: il sistema si congela nello stato più stabile   
cioè 'grado di libertà interno è congelato al livello più stabile'     
stato di minima energia ha prob=1  
gli altri hanno prob=0  


**T e disordine**

devo minimizzare: $F=E-TS$  

T bassa: $F \sim E$  
quindi devo minimizzare E, ho sistema ordinato  

T alta: $F \sim -TS$  
quindis evo massimizzare S, ho sistema disordinato
