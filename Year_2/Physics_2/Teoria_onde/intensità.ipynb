### Vettore di Poynting

**Formula**

$\vec S=\frac{1}{\mu_0} \vec E \times \vec B$  

**Definizione**

potenza trasportata dall’onda elettromagnetica per unità di superficie incidente A  
energia che fluisce nel tempo unitario (potenza) attraverso una superficie unitaria  

$|\vec{S}| = \frac{1}{A}\frac{dE}{dt}$  

**Nota**

se calcolo S con E di radiazione, è la potenza irraggiata  
E di radiazione è quello che scala come $E \sim \frac{1}{r}$  
quindi un dipolo ha tutto il campo di radiazione, non c'è parte di velocità (non avrebbe senso,perchè l'approssimazione di dipolo vale solo a grandi distanze)  

#### Esercizi

**Esercizi: come calcolarlo**

1- calcolo E e B  
2- uso la formula  

tips:  
1- di solito E e B sono perpendicolari, quendi semplicemente moltiplica i valori, e poi S punta nella direzione della mano destra  
tipo, se sono uno assiale e uno azimutale, S punta in direzione radiale, metti $\hat{r}$ nella risposta   
2- quando devo calcolare il flusso, controlla in che direzione punta S, così so che area devo moltiplicare  
3- se ho un'onda piana, S punta lungo $\vec k$ (perchè punta lungo $ E \times B$)  

**Esempio calcolo**

ho trovato: $\vec S = S_0 \frac{1}{r^2} \hat z$   
dove z è l'asse di un cilindro, con apreti a r=a e r=b    

$\Phi(S) = \int_a^b S(r) \, 2 \pi r \, dr$  
non posso fare $4 \pi r^2$, perchè S non è uguale su tutta la sezione  

#### Teorema di Poynting

**Teorema di Poynting**

forma locale: $\frac{\partial}{\partial t} u + \nabla \cdot S = - E \cdot J$  
forma integrale (e nel vuoto, cioè J=0): $\frac{dU}{dt}-\int S \cdot dA = 0$ 

$u=\frac{1}{2}\epsilon_0E^2 + \frac{B^2}{2\mu}$: densità di energia elettromagnetica  
$E \cdot J$: potenza trasferita alla materia (di solito dissipata, ma non per forza)    

**Spiegazione**
 
S è u vettore, quindi posso calcolare il suo flusso attraverso una superficie A  
il flusso di S è uguale alla derivata parziale temporale dell'energia magnetica 

**Univoco**

dal teorema di Poynting, vedo che di S mi interessa solo la divergenza    
quindi S non ha una forma univoca, perchè se $S'=S+\nabla \times \vec v$, allora S e S' hanno la stessa divergenza  
dove v è un vettore qualunque  
questa NON è una scelta di Gauge, quella si avrebbe se S e S' devono avere lo stesso rotore, non la stessa divergenza  

**Esercizio: prova che è uguale a energia**

situazione:  
ho condensatore cilindrico con E(t)  
voglio flusso di S su superficie cilindrica   

1- calcola B  
$2 \pi r B = \pi r^2 \frac 1{c^2} \partial_t E$  
$B= r\frac{1}{2c^2}\partial_t E$  

2- calcola S  
$S= \epsilon_0 c^2 E \times B = \frac r2 \partial_t (\frac{\epsilon_0}2 E^2)$  
$\Phi(S) = 2 \pi r h \, S = \pi r^2 h \, \partial_t(\frac{\epsilon_0}2 E^2)   = \frac d{dt} U_{es}$  

nota che quando scrivo S appare un 1/2 in più, perchè $E \, \partial_t E = \frac 12 \partial_t E^2$  

#### Leggi di conservazione

**Forma generale**

$\frac{\partial A}{\partial t} + \vec \nabla \cdot \vec B = C$  
variazione interna + entrata/uscita = produzione interna  
accumulo + flusso = sorgenti  

A: la cosa che viene conservata, una densità della quantità nel volume    
B: la forma sotto cui può spostarsi fuori da quel volume, flusso di quella quantità   
convenzione: flusso uscnte = positivo  
C: se può trasformarsi in una'altra forma (sorgenti o perdite interne) (produzione o distruzione locale)


**Esempi** 

conservazione della carica:  
$\frac{\partial \rho}{\partial t} +\vec \nabla \cdot \vec J = 0$  
$\rho$: conservazione della carica  
$\vec J$: la carica può uscire da un volume sotto forma di corrente  
$0$: la carica non può trasformarsi in altre cose  

conservazione dell'energia elettromagnetica  
$\frac{\partial u_{em}}{\partial t} +\vec \nabla \cdot \vec S = -\vec J \cdot \vec E$  
$u_{em}=\frac 12 \epsilon E^2 + \frac{B^2}{2\mu}$: conservazione dell'energia  
$S$: vettore di Poynting, indica il flusso di energia  
$J \cdot E$: energia ceduta alla materia, perchè l'energia elettromagnetica può passare ad altre forme, tipo il calore  

**Dissipazione Joule**

calore dissipato quando il materiale è ohmico  
cioè quando vale la legge di Ohm locale: $J = \sigma E$

### Intensità

**Formula**

$I = <S>$  
cioè la media temporale  
ma la cosa importante è $I \sim |E|^2$

ricorda che:  
$<AB>=\frac 12 Re[A_0B_0^*]$  
$<A^2>=\frac 12 A_0^2$  

**Calcola intensità: onda piana**

in onda piana:  
nel vuoto: $E=cB$  
in mezzo: $B = \frac Ev = \frac{E \, n}c$   
cioè, in un mezzo moltiplico per n   

nel vuoto:  
$<S>=   <\frac{1}{\mu_0}E \times B>  =\frac{1}{\mu_0 \, c} <E^2>= \epsilon_0 c \frac 12 E_0^2$   
in un mezzo:  
$<S> = \frac 12 n  \epsilon_0 c  E_0^2$  

**Calcola intensità: superposizione di onde**

$I = <S> = <\frac{1}{\mu_0 c}(E_1+E_2)^2> =  \frac{1}{\mu_0 c} <\vec E_1^2+\vec E_2^2 +2\vec  E_1 \cdot\vec E_2> =I_1+I_2 + 2\frac{1}{\mu_0 c} |E_1|\,  |E_2| \, (\hat e_1 \cdot \hat e_2)\cos (k \delta + \Phi)  =  I_1+I_2 +2 \sqrt{I_1I_2} (\hat e_1 \cdot \hat e_2)\cos (k \delta + \Phi)$ $  

$\hat e $ appaiono per il prodotto scalare, lo sfasamento appare per la formula della media temporale tra fasori  
e poi $I = \frac{E^2}{\mu_0 c}$  

### Potenza

**Potenza dell'onda**

$P= I \cdot A$  
prodotto scalare tra intensità e superficie  

non confondere:  
P: potenza, energia per unità di tempo  
S: energia per unità di tempo e spazio, potenza per unità di spazio
