### D'Alembert vs Helmholtz

cosa | Alambert | Helmholtz
---|---|---
equazione | $\nabla^2 u -\frac{1}{c^2} \partial_{tt}u=0$ | $\nabla^2 u + k^2 u =0$  
variabilie | spazio e tempo | spazio
dominio | tempo | frequenza
significato | propagazione | scattering, risonanza

### Trova equazione di d'Alembert

**Recap**

continuo:  
$\frac{\partial^2 z}{\partial t^2} = v^2 \frac{\partial^2 z}{\partial x^2}$  

discreto:  
$\ddot z = \omega_0^2 (z_{n-1} + z_{n+1} -2z_{n})$  

**Nota**

possono descrivere sia onda longitudinale che trasversale  
dipende solo da cosa rappresenta z  

#### Caso continuo: generale

**Come ottenerla**

1- EM waves devono rispettare le equazioni di Maxwell, parto da quelle  
assumo il caso nel vuoto:  
$\begin{cases}
\nabla \cdot E =0 & \nabla \cdot B = 0 \\
\nabla \times E = -\frac{\partial}{\partial t}B & \nabla \times B = \mu_0 \epsilon_0 \frac{\partial}{\partial t}E
\end{cases}$  

2- nelle equazioni del rotore, applico di nuovo il rotore da entrambe le parti  
così uso la proprietà: $\nabla \times \nabla \times w = \nabla(\nabla \cdot w)-\nabla^2 w$   
ottengo:   
$\begin{cases}
-\nabla^2 E = -\frac{\partial}{\partial t}(\nabla \times B) \\
-\nabla^2 B = \mu_0 \epsilon_o \frac{\partial}{\partial t}(\nabla \times E) \end{cases}$  

3- pluggo le originali equazioni del rotore nei rotori del punto prima, ottengo le equazioni d'onda:  
$\begin{cases}
\nabla ^2 E = \mu_0 \epsilon_0 \frac{\partial^2}{\partial t^2} E \\
\nabla ^2 B = \mu_0 \epsilon_0 \frac{\partial^2}{\partial t^2} B \end{cases}$   

**Altro modo**

ricorda che: $\epsilon_0 \mu_0 = \frac{1}{c^2}$

#### Caso discreto

**Come trovarla**

scrivi la legge di Newton  
cerca la forma generale dell'equazione del moto per un onda nel discreto  

![](https://www.researchgate.net/publication/304499230/figure/fig4/AS:377767855181827@1467078283367/A-chain-of-springs-and-masses.png)

**Esempio: corda con molla e massa intervallate**

situazione:  
masse numerate da 0 a n, con n infinito  
quindi posizione (coord lungo asse x) di una massa n è: $z_n$  
distanza tra punto di equilibrio di due masse è a  

scrivo legge di Newton per una massa n  
la massa risenta delle forze elastiche dalle due masse vicine:  
$m\ddot{z}_n=-k(z_n-z_{n-1})-k(z_n-z_{n+1}) = k(z_{n-1} + z_{n+1} -2z_{n})$  

quindi:  
$\ddot z = \frac km (z_{n-1} + z_{n+1} -2z_{n})$  
posso chiamare $\omega^2_0 = \frac km$   

**Nota: esempio corda**

la stessa equazione descrive sia onda longitudinale che trasversale, quale è dipende da cosa rappresenta z:  

1- longitudinale  
ho catena orizzontale di m e k intervallate, come in immaigne  
m può fare solo spostamento orizzontale  

2- trasversale  
ho una serie di guide parallele, verticali, disposte come le tacchette su un righello  
su ognuna c'è una massa m, legata alle due vicine con una molla k  
m può fare solo spostamento verticale

**Esempio: circuito** 

situazione: chiamo un'induttanza $L_n$, $I_n$ la corrente che la attraversa, $V_n$ il nodo subito a sinistra, e $C_n$ il condensatore sotto quel nodo, quindi a sinistra di $L_n$  

scrivo legge dei nodi sul nodo n e voltage drop sull'induttanza n  
$\begin{cases}
I_{n-1} = I_n + \frac{dQ_n}{dt} \\ 
V_n-V_{n-1} =L \frac{dI_n}{dt} \end{cases}$   
derivo la seconda per il tempo, e pluggo la seconda nel lato destro  
$L \frac{d^2I}{dt^2} = \frac{1}{C} (\frac{dQ_n}{dt}-\frac{dQ_{n+1}}{dt}) = \frac 1C[(I_{n-1}-I_n)-(I_n-I_{n-1})] = \frac 1C (I-{n-1}-2_n+I_{n+1})$  

quindi:  
$\frac{d^2I}{dt^2} = \frac1 {LC}(I_{n-1}-2I_n+I_{n+1})$   
posso chiamare $\omega_0^2 = \frac 1{LC}$  


#### Passaggio da discreto a continuo

**Passaggio al continuo**

$a$: distanza tra due guide/ punti vicini, quindi nd dice a che punto della catena sono  
$\lambda$: ampiezza delle fluttuazioni  

contiuo: $z(x,t) = z(na, t)$  
discreto: $z_n(t)$  

condizione per passare al continuo: $a/\lambda \ll 1$:  

1- equazione del moto in continuo:  
$m\ddot{z}(na)=k(z_{n-1} + z_{n+1} -2z_{n})$  
2- approssimo con Taylor:  
$z_{n+1} = z(x+a) = z(x)+az'(x)+\frac{a^2}{2}z''(x) +...$  
$z_{n-1} = z(x-a) = z(x)-az'(x)+\frac{a^2}{2}z''(x) +...$  
3- ottengo:  
$m\ddot{z}=k(a^2z'' + ...)$    
$\ddot{z}=-\frac{T}{\rho}z''=0$  
4- è come equazione in continua:  
$\frac{\partial^2 \phi}{\partial t^2}-v^2\frac{\partial^2 \phi}{\partial x^2} =0$  


nota:  
$\ddot{z}$: derivata seconda rispetto al tempo (accelerazione)  
$z''$: derivata seconda rispetto allo spazio  
$T=ka$: tensione  
$\rho=m/d$: densità di massa/lineare  
$\phi(x,t)$: campo  
$v=\sqrt{T/\rho}$: un parametro con le dimensioni di una velocità

**Caso discreto vs continuo**

equazione:  
continua: $u_{tt}-v^2u_{xx}=0$  
discreta: $\ddot x_n =-\omega_0^2 (-2x_n + x_{n+1}+x_{n-1})$  

soluzione:  
continua: $u(x,t)$  
discreta: $x_n(t)=u(nd, t)$

#### 1. Problema senza bordi, studia propagazione

**Steps**

1- ho l'equazione differenziale  
2- plugga come ansatz un'onda piana  
3- ottengo la relazione di dispersione  

**Esempio: discreto**

equazione: $\ddot z_n​=\omega_0^2(z_{n−1}​+z_{n+1}​−2z_n​)$  
ansatz: $z_n(t)=ae^{i(nka-\omega t)}$  

$(i\omega)^2 z_n = \omega_0^2 \, Ae^{i(kna-\omega t)} (e^{-ikna}+e^{ikna}-2)
$  
$-\omega ^2 z_n= \omega_0^2 z_n  \, 2(\cos (ka)-1)$  
$\omega ^2=4 \omega_0^2  \sin(\frac{ka}{2})$  

è dispersiva, ma per piccoli spostamenti $\sin x \sim x$, allora diventa non dispersiva  

**Esempio: continuo**

equazione: $\frac{\partial^2 u}{\partial t^2} = v^2 \frac{\partial^2 u}{\partial x^2}$  
ansatz: $u(x,t) = e^{(kx-\omega t)}$  

$(-\omega^2)=v^2(-k^2)$  
$\omega = vk$  
è non dispersiva  


#### 2. Trova modi normali

**Steps**

1- ho l'equazione di d'Alamebert, assumo sia a var separiabili: $u(x,t)=X(x)T(t)$  
2- ho condizioni al bordo, quindi condizioni su x  
3- le pluggo, vedo cosa mi viene come $X(x)$  
lo potrei scrivere $\phi(x)$, per far capire che è un modo normale  

nota: ad esempio, posso avere ome vincoli la corda finita, semi-infinita o infinita, oppure estremi fissi o mobili  

**Come scrivere condizioni al contorno**

estrmeo fissato: $u(0,t)=0$  
estremo libero: $u_x(0,t)=0$  

**Corda finita con estremi fissi**

condizioni al contorno:  
$u(0,t)=u(L,t)=0$   
cioè $X(0)=X(L)=0$   

risolvo:  
$X(x)=A \cos(kx)+B \sin(kx)$  
1- $X(0)=A =0 $  
2- $X(L)= B \sin (kL) =0$  
per non avere soluzione nulla, devo avere il sin nullo: $kL = n \pi$    

soluzione:  
$X(x)=\sin(k_nx) \quad k_n = \frac{n \pi}{L}, n=1,2...$  

**Corda finita con estremi liberi**

condizioni al contorno:  
$u_x(0,t)=u_x(L,t)=0$  

soluzione:  
$X(x)=\cos (\frac{n \pi x}{L})$  
$u(x,t)= u(x,t) = [a_n \cos(\omega_n t )+b_n \sin (\omega_n t)] \sin(\frac{n \pi x}{L})$  

Significato fisico:  
niente nodo agli estremi  
massima oscillazione possibile ai bordi  
riflessione senza inversione

**Corda semi-infinita con estremo fisso**

condizioni al contorno:  
$u(0,t)=0$  

soluzione:  
$u(0,t)=f(-vt)+g(vt)=0$  
cioè: $f(s)=-g(-s)$  
cos'è: riflessione con cambio di segno  

Interpretazione fisica: quando un’onda arriva al vincolo fisso, la riflessa è invertita (fase π)  
cresta è riflessa come ventre e viceversa 

**Corda semi-infinita con estremo libero**

condizioni al contorno:  
$u_x(0,t)$  

soluzione:  
$u_x(0,t)=f'(-vt)=g'(vt)=0$  
cioè: $f'(-s)=-g'(s)$  
cos'è: riflessione senza cambio di segno  

interpretazione fisica: : a estremo libero la riflessione avviene senza inversione (fase 0)    
 

**Corda semi-infinita**

soluzione generale corda semi-infinita:  
$u(x,t)=f(x-gt)+g(x+vt)$  
f: onda incidente  
g: onda riflessa (dal bordo)   
le condizioni al bordo dicono come devono combinarsi  

spiegazione:  
estremo fisso: quando arriva l'onda incidente, l'estremo non può muoversi, quindi l'onda riflessa deve essere opposta, per cancellare lo spstamento dell'estremo  
estremo libero: quando arriva l'onda incidente, l'estremo può muoversi, quindi lo segue, viene riflesso uguale  
devo solo annullare la pendenza (quella è la condizione al contorno)  

intuizione:  
muro rigido → respinge → cambia segno  
estremità libera → segue il moto → stesso segno


#### 3. Trova evoluzione del moto

**Steps**

1- ho trovato i modi normali, posso scrivere la serie di Fourier che rispetta i vincoli   
$u(x,t) = \Sigma \phi_n(x) (A_n \cos (\omega_nt) +B_n \sin(\omega_n t))$  
2- so condizione iniziale: $u(x,0)=f(x)$  
3- uguagliala alla sommatoria, trova i coefficienti $A_n, B_n$ per ogni componente  

nota: per sistema discreto, serie di Fourier si scrive $x(t)=\Sigma q_n(t) v_n$, come in esercizio con il lagrangiano

### Modi normali e onde piane

**Onde piane e modi normali**

cosa sono: basi di soluzioni  
le onde piane sono una soluzione che vale solo nella spazio infinito  
il sistema fisico (cioè i bordi e i vincoli) seleziona solo alcune combinazioni di onde piane, i modi normali  


. | Onde piane | Modi normali
---|---|---
formula | $e^{ikx}$ | $\phi(x) = \sin(\frac{n\pi x}{L})$  
quando usare | sistema infinito (spazio libero) | sistema finito, con bordi o vincoli  
parametro | k continuo | n discreto, sono uno spettro discreto di modi 

**Dipendenza temporale**

la differenza tra onde piane e modi normali è solo nella dipendenza spaziale  
la dipendenza temporale è uguale in entrambi, è $e^{-i\omega t }$  
quando ho calcolato i modi normali e voglio l'evoluzione del moto, a volte metto $e^{i\omega t}=A_n \cos (\omega_nt) +B_n \sin(\omega_n t)$, per semplificare i calcoli  

**Modi normali**

tutte le parti del sistema si muovono di moto armonico con la stessa frequenza e con una relazione di fase fissa  
un modo normale è dunque una soluzione stazionaria: se
la corda è fatta vibrare con le caratteristiche di un modo normale, nel corso del tempo si manterrà sempre nello stesso modo e nessun altro modo verrà eccitato  

ma se pizzico una corda a caso, non oscillaerà in modo normale, ma con un'onda a caso, sovrapposizione dei modi normali  
che viene stabilita dalle condizioni iniziali, cioè da $u(x,0)=f(x)$ che uguaglio alla serie di Fourier, da cui calcolo i coefficienti per le armoniche (come in esercizio di metodi matematici)   

#### Cosa sono modi normali

**Cosa sono**

discreto: autovalori della matrice di accoppiamento  
continuo: autofunzioni di un operatore differenziale  

**Esempio: discreto**

scrivo l'equazione differenziale con vettori:  
$m\ddot z_n​=k(z_{n+1}​+z_{n−1}​−2z_n​)$  
$\ddot z = A z$  
$ A= \begin{bmatrix} -2 & 1 & 0 & ... \\ 1 & -2 & 1 & ... \\ 0 & 1 & -2 & ... \end{bmatrix}$  

modi normali:  
$Av_n = \lambda_n v_n$  
sono gli autovettori della matrice  
come in esercizio della lagrangiana  

**Esempio: continuo**

$u_{tt}=v^2 u_{xx}$  

separando le var:  
$-\partial^2_x \phi_n = k^2 \phi_n$  
è un problema agli autovalori con condizioni al bordo  

$L=-\partial^2_x$ è un operatore  
è il continuo della matrice di prima???  
i modi sono le sue autofunzioni (si vede dalla formula, dopo che applico l'operatore ottengo sempre $\phi_x$, moltiplicato per un autovalore)    

### Progressive e regressive

**Collegamento**

$\cos (kx) \cos(\omega t)=\frac12 [\cos(kx-\omega t)+\cos(kx + \omega t)]$  
una sinusoide stazionaria è la somma di:  
un’onda che va a destra  
un’onda che va a sinistra  


**Intuizione**

1- dai un colpo al centro di una corda  
si generano due onde, una che si muove verso destra, e una verso sinistra  

2- onda colpisce un muro e torna indietro  
quella che torna è quella regressiva  
nota: prima dell'impatto c'è solo onda progressiva incidente   
poi, l'impatto genera onda regressiva riflessa, che si SOVRAPPONE a quella progressiva  
quindi dopo l'impatto esistono sia progressiva che regressiva  

onda riflessa può avere sfasamento rispetto a onda incidente o no  
dipende se l'estremo è fisso o libero
