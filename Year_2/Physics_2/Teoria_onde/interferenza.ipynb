### Fenomeni di interferenza

#### Battimenti

**Cos'è**

fenomeno he si verifica per inteferenza di due onde con frequenza molto simile  

**Formule**

1- ho due onde, soluzione di d'Alembert:  
$y_1(t) =A \cos(\omega_1 t)$  
$y_2(t) = A \cos (\omega_2t)$  

2- d'Alembert è un'eq lineare, quinid anhe $y_1+y_2$ è una soluzione  
$y_1+y_2 = 2A \cos(\frac{\omega_1+\omega_2}{t})\cos(\frac{\omega_2-\omega_2}{t})$  

3- se le frequenze sono simili:  
$\omega _1 = \omega +\delta \omega$  
$\omega_2 = \omega- \delta \omega$  
$\frac{\delta \omega}{\omega} \ll 1$

4- ottengo:  
$y_1+y_2 = 2A \cos(\delta \omega t) [\cos (\omega t)]$  
il coseno con $\delta \omega$ èmolto più lento dell'altro  
allora immagino che il termine lento sia unito all'ampiezza, formi un'ampiezza modulata nel tempo  

questo genera i battimenti  
$\omega$: l'onda veloce, quindi il fare su e giè velocemente per colorare il grosso del disegno  
$\delta \omega$: il fatto che l'ampiezza sia modulata, l'onda che traccia il contorno

**Intuitivamente**

le onde hanno fase diversa, quindi lungo l'asse orizzontale, in alcuni momenti sono in fase e in altri in controfase, con tutte le versioni intermedie  
se la differenza di fase è poca, si sfasano pian piano, quindi l'ltezza dell'onda risultante cambia pian piano  
questo dàla modulazione della fase come in figura

**Musica**

quando suonate due corde di una chitarra, esse emettono due onde a frequenze diverse che vanno a scontrarsi  
L’effetto è un suono strano che “oscilla” tra un  valore più acuto e uno più grave  

Per accordare la chitarra senza utilizzare strumenti si utilizza questa tecnica:  
quando non si sente più il suono che oscilla significa che le due corde pizzicate hanno la stessa frequenza (perché i battimenti non si verificano se le onde hanno la stessa frequenza)  
quindi che le due corde rappresentano la stessa nota


#### Interferenza

ho due onde $y_1$ e $y_2$, armoniche e sfasate di un angolo $\Phi$  
$y_1 = y_0 \sin (kx-\omega t- \Phi)$  
$y_2 = y_0 \sin (kx-\omega t)$  

**Due onde si incontrano**

si sommano, quindi uso le formule di Prostaferesi per sommare i seni  
$y = y_1+y_2 = 2y_0 \sin (kx-\omega t-\frac{\Phi}{2})\cos \frac{\Phi}{2}$  

casi limite:  
$\Phi=0$: interferenza costruttiva (sono in fase)  
$\Phi = \pi$: interfernza distruttiva (in controfase)
