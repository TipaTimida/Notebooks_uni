### Teoria

**Cosa è**

sovrapposizione di onde piane con diverse frequenze e numeri d’onda  
scritto con la trasformata di Fourier (usata per scrivere una funzione time-dependent come sommatoria infinita di funzioni sinusoidali monocromatiche)  

così, invece di avere una sola onda infinita, hai un’onda localizzata nello spazio  

**Problema delle onde piane**

non sono una rappresentazione realistica, perchè un'onda piana:  

1- è definita in tutto lo spazio ad ogni t  
mentre tipo un laser si trova solo lungo la linea dove lo sparo  
oppue un'onda parte da una sorgente e poi si espande in tutto lo spazio  
e comunque, non ha durata infinita  

2- è monocromatica   
nessuna sorgente è idealmente monocromatica  

soluzione:  
per descrivere fenomeni reali, invece di un'onda piana devo usare un pacchetto d'onda  

**Come si scrive**

onda piana: $e^{ikx}$  
pacchetto d'onda: $f(x)=e^{ikx-x^2/L}$  
la parte aggiunta è una gaussiana (perchè???)  

### Come si descrive

![alt text](https://www.openfisica.com/fisica_ipertesto/openfisica5/immagini/pacchetto_onda.jpg)

**Graficamente**

è come l'immaigne dei battimenti    
oscillazioni veloci dentro: onda portante  
oscillazione lenta che fa il bordo: inviluppo, pacchetto  

se faccio trasformazione di Lorentz, il numero di oscillazioni della portante dentro l'inviluppo non può cambiare (perchè è un conteggio)  
però la frequenza dela portante cambia  
quindi cambia la lunghezza del pacchetto  


**Proprietà**

1- numero d'onda centrale  
$\lambda = \frac {2 \pi}{k_0}$  
determina la lunghezza d’onda dominante  
graficamente: distanza tra due creste delle oscillazioni interne (cioè dell'onda portante)  

2- frequenza centrale  
$\omega_0 = \omega(k)$    
graficamente: la velocità con cui oscillano le creste nel tempo (cioè dell'onda portante)   

3- larghezza in spazio: $\Delta x$  
cioè in che zona dello spazio l'onda è presente/significativa  

4- $v_p = \frac {\omega}k$  
velocità delle creste   
NON trasporta energia! le creste possono anche muoversi più veloci della luce (in certi mezzi)  
ma non significa nulla fisicamente da sole  

5- $v_g = \frac{d \omega}{dk}$   
velocità dell’inviluppo (cioè del pacchetto)  
questa è la velocità dell'energia/informazione  
se $v_g \neq v_p$, componenti viaggiano a velocità diverse, c'è dispersione  

6- energia: $U=IAT$  
I. ampiezza, A: ampiezza della portante, T: periodo della portante  


**Come si disegna**

sezione: $L^2$  
lunghezza: $c \tau$  
quindi $\tau$ è il tempo necessario a tutto il pacchetto per colpire un muro, NON il periodo di un'oscillazione 

**Approssimazione**

se $L \gg \lambda$, approssima come onda piana  

### Dispersione

**Relazione di dispersione**

è la formula:  
$k(\omega)$  

**Definizione dispersione**

posso scrivere un'onda come sovrapposizione di onde con frequenze diverse (serie di Fourier)  
onde con frequenza diversa potrebbero avere una velocità di fase diversa, e quindi separarsi durante il tragitto

c'è dispersione?  
sì, se la velocità di fase dipende dalla frequenza  
quindi, se le componenti dell'onda alle varie frequenza viaggiano con velocità diversa  

**Esempio dispersione**

caso: $\omega \sim k$, cioè hanno dipendenza lineare  
$v_{f} = \frac \omega k$ e $v_f$ è indipendente dalla frequenza, no dispersione  

caso:: $\omega = 2 \omega |\sin (ka)|$   
non è lineare, allora c'è dispersione   

**Velocità**

velocità di fase: $v_f = \frac \omega k$  
la velocità del fronte d'onda nella direzione di propagazione  
può essere diversa per ogni componente, cioè dipendere dalla frequenza  

velocità di gruppo: $v_g = \frac{\partial \omega}{\partial k}$  
più o meno la velocità media con cui si muove il pacchetto  
se $v_g=0$, il pacchetto non si propaga  

### Esercizi

**Notazione**

della specie A:  
$n_A$: densità  
$q_A, m_A$: carica e massa  
$\omega^2_{pA} = \frac{n_Aq_A}{m_A \epsilon_0}$  : frequenza di plasma  
$\Omega_A=\frac{q_AB_0}{m_A}$: frequenza ciclotronica  
$\omega^2$: frequenza dell'onda EM

al posto di A, scrivi il nome della specie, tipo 'e' o 'p' per elettrone o protone  

**Formula: permittività dielettrica del materiale**

1- senza campo magnetico  
$\epsilon(\omega)=1-\Sigma_i \frac{\omega^2_{pi}}{\omega^2}$  

2- con campo magnetico  
$\epsilon(\omega)=1-\Sigma_i \frac{\omega^2_{pi}}{\omega(\omega \pm \Omega_i)}$  
il + o - dipende dalla carica + o - della particella, due cariche di sengo opposto avranno segni opposti, perchè fanno moto cilotronico in direzione opposta  
NON devo togliere il $\pm$ se il testo non mi dice le direzioni, ma se ho cariche ooposte devo metter pm in una e mp nell'altra  


**Se c'è campo magnetico B0**

che succede:  
le particelle NON si muovono più liberamente, ma fanno moto circolare (ciclotrone)  
Il plasma diventa anisotropo  

la risposta dipende da come ruota il campo elettrico dell’onda  
➡️ nascono due modi:  
polarizzazione circolare destra (R)  
polarizzazione circolare sinistra (L)

**Attenzione angoli**

ho dipolo elettrico che emette un'onda   
onda sbatte su superficie tra due mezzi, di normale $ \hat n$    

disegna la traiettoria (linea) che percorre il raggio  
ci sono due angoli diversi nel disegno:  
1- l'angolo tra $\vec p$ del dipolo e la traiettoria del raggio   
2- l'angolo tra $\hat n$ e la traiettoria del raggio  
nota: spesso, $|\vec p$ è parallelo a $\hat n$, così mi risparmio problemi di geometria   

**Intensità radiazione e Fresnel**  
 
voglio intensità della radiazione a distanza L dall'antenna  

1- Il campo emesso dall'antenna dipolare giace nel piano di incidenza per cui occorre prendere il coefficiente di Fresnel per polarizzazione P  

2- $I(L)=\frac {R_p}{r^2} \frac {dP_{irr}}{d \Omega}$  
dove:  
$ \frac {dP_{irr}}{d \Omega} = \frac {3P_{irr}}{4 \pi} \sin^2 \theta$  
$\theta$: l'angolo che il tra la traiettoria del raggio e $\vec p$, indicalo tipo con $\tan \theta =\frac L{2h}$  
$r$: distanza percorsa dal segnale radio / lunghezza della traiettoria, indicala tipo con $r^2=(L^2+h^2)$  
$R_p=|E_r/E_i|$, prese dalle formule di Fresnel, è di quanto diminusice l'ampiezza dell'onda durante la riflessione (nota che dipende da $\theta$)  

ricorda:  
il campo elettrico irradiato da un dipolo è trasverso (⊥ alla direzione di propagazione)  
contenuto nel piano definito da: $p$ e $k$  

**Campo mangetico terrestre e propagazione**

per onde che si propagano in un mezzo magnetizzato si ha tipicamente (a rigore a seconda della polarizzazione e
della direzione relativa tra vettore d'onda e campo magnetico) risonanza per $ω = ω_c$  
dove $\omega _c = \frac {eB}m$  

quindi per onde con lunghezza d'onda $\lambda = 2 \pi c / \omega_c$
