### Pratica: Onda piana

**Come si scrive**

$E(x,t) = Re[\, \hat{y} \, E_0 \, e^{i(kx-\omega t)}] = \, \hat{y}\, E_0 \, \cos(kx-\omega t)$  

x: direzione di propagazione (se cambio x, cambia il modulo di E)  
y: direzione di polarizzazione/oscillazione    

**Semplifica**

se $\lambda \gg a$, dove a è il raggio di una sfera, vuol dire che l'onda è uniforme su tutta la sfera   
quindi scrivo $E=E_0e^{-i\omega t}$  
non metto la parte $e^{i kx}$  

**Formule**

1- dipendenza da spazio e tempo  
$\omega = \frac{2 \pi}{T}$: pulsazione, frequenza dell'onda    
$k = \frac{2 \pi}{\lambda}$: vettore/numero d'onda   

2- velocità di fase:   
$v = \frac{\omega}{k} =  \frac{1}{\sqrt{\epsilon \mu}}= \frac cn$  
per ricordare cosa sta sotto e sopra, ricorda le unità di misura 1/m e 1/s di $k$ e $\omega$ (v e $\omega$ hanno entrambe 1/s, sono velocità)  
nota che $n$ e $\epsilon$ dipendno da $\omega$  

**Note**

1- polarizzazione  
quella sopra è un'onda piana POLARIZZATA LINEARE    
per ottenere un'onda piana con altre polarizzazione, dovo sommare più onde lineari, con polarizzazioni in direzioni diverse  

2- parte reale  
alla fine, voglio come risultato la parte reale  
ma spesso, per fare i calcoli utilizzo l'esponenziale complesso SENZA Re[], e la parte reale la prendo solo alla fine  

**Scrivere due onde con sfasamento**

metti lo sfasamento in una qualunque delle onde    
l'altra parte arbitrariamente da $\phi=0$  
$E_1 = Re(E_0e^{i(kx-\omega t)} \hat{y})$  
$E_2 = Re(E_0 e^{i(ky-\omega t + \phi)}\hat{x})$  

**NOTA**

E e B devono essere paralleli SOLO nelle onde piane, sennò non per forza  
tipo posso avere due campi azimutali, va bene  
tipo, se ho $B \sim r$, sicuro non è un'onda piana, quindi non per forza vale l'ortogonalità  

### Onda generale

**Dipendenze**

E(x,t)  dipende da spazio e tempo, quindi:  
is a 2D surface in 3D space (need 3 axes to plot it: x,t,E)   
so, when i plot a sinusoid in 2D, means i'm keeping fixed one parameter, while the other is on the horizontal axis  
depending on which param is on the horizontal axis, i'm studying temoporal or spatial dependency  

Dipendenza | Cosa descrive | Fenomeni collegati | Come si studia  
---|---|---|---
temporale | come oscilla il campo nel tempo (evoluzione spaziale a punto fisso) | frequenza, periodo, **polarizzazione** (se vettoriale) | (x) fisso, varia (t) |
spaziale | struttura nello spazio (geometria nello spazio a tempo fissato) | lunghezza d’onda, direzione di propagazione, **fronti d’onda** | (t) fisso, varia (x) |
spazio + tempo (fase) | propagazione dell’onda | **velocità di fase**, **velocità di gruppo**, dispersione, trasversale / longitudinale (se vettoriale) | fase costante|


#### Dipendenza temporale

**Onda scalare vs vettoriale**

scalare: non ha direzione di oscillazione (ma ha quella di propagazione)  
vettoriale: ha direzione di oscillazione (può essere trasversale o longitudinale, a seconda che sia parallela o ortogonale a quella di propagazione)  

esempi:    
pressione del suono: $p(x,t)=p_0\cos(kx-\omega t)$ → scalare  
spostamento delle particelle nel suono: $\vec u(x,t)=\hat x \, u_0​ \cos (kx−ωt)​$ → vettoriale longitudinale  
onda EM: $\vec E(x,t)=\hat y \, E_0​ \cos (kx−ωt)​$ → vettoriale trasversale  

**Quando esiste polarizzazione?**

la polarizzazione è lo studio della DIREZIONE dell'onda durante l'evoluzione temporale in punto fissato  

1- scalare: l'onda non ha direzione, quindi non posso studiarla -> non ha polarizzazione  
2- vettoriale longitudinale: per def, l'onda è diretta lungo la direzione di propagazione -> posso dire o che non ha senso la polarizzazione, o che è per forza lineare  
3- vettoriale trasversale: E può esister in tutto il piano ortogonale a $\vec k$, quindi qui ha senso vedere come cambia direzione nel piano durante il tempo -> ha polarizzazione  

**Definizione polarizzazione**

durante l'evoluzione temporale in un punto fissato, come cambia la DIREZIONE dell'onda  

ricordando che, per onde trasversali, questa direzione è per forza ortogonale alla direzione di propagazione  
quindi studio la direzione dell'onda nel piano trasverso  

grafico:  
fissa un punto nello spazio (x fisso, t varia)  
disegna $\vec E$ in quel punto  
guarda che disegno traccia la punta di E nel piano trasverso yz, al passare del tempo  

**Come scrivere polarizzazioni per onda trasversale**

lineare: $E(x,t)=E_0 \cos(kx-\omega t)\hat{y}$  
circolare:  $E(x,t)=E_0 [\cos(kx-\omega t)\hat{y}+\sin(kx-\omega t)\hat{z}]$   
ellittica: $E(x,t)=E_1\cos(kx-\omega t)\hat{y} \, +E_2 \,\sin(kx-\omega t)\hat{z}$  

lineare: lungo una sola asse  
circolare/ellittica: lungo due assi (tramite sovrapposizione di lineari), con ampiezze uguali/diverse (le ampiezze sarebbero gli assi del cerchio/ellisse)  

ricorda che è onda trasversale, quindi non può avere componente lungo la direzione di propagazione  

**Descrizione tipi**

1- lineare: il campo oscilla sempre lungo una retta (E resta sulla stessa linea, cambia solo il modulo)  
2- circolare: modulo di E è costante, ma direzione ruota nel tempo, la punta descrive un erchio (se considero anche la propagazione spaziale, quindi, vedo un'elicoidale)   
3- ellittica: Caso più generale, la punta di E fa un ellisse  

come vedere: plugga un x fisso, ad esempio x=0 per comodità, vedi che figura disegna E(t) variare di t  

#### Dipendenza spaziale

**Definizioni**

1- fronte d'onda  
insieme di punti in cui la grandezza che varia (es. pressione sonora, campo elettromagnetico) ha lo stesso valore, in un dato istante  
cioè dei punti che oscillano in fase  
al contrario della polarizzazione, questo esiste anche per onde scalari  

2- raggio di propagazione   
una retta perpendicolare al fronte d'onda  

3- velocità di fase  
velocità dei fronti d'onda lungo $\vec k$

**Come scrivere fronti d'onda**

onda piana: $E(r,t) = \hat{e}\, E_0 \, \cos(\vec k \cdot \vec r-\omega t)$   
onda circolare: $E(r,t) = \hat{e}\, \frac {E_0} r \, \cos(\vec k \cdot \vec r-\omega t)$  

**Dimostrazione fronti d'onda**

come vedere: plugga un t fisso, ad esempio t=0 per comodità, vedi in che figura E(x) è costante  

1- onda piana   
a tempo fisso: $E \sim \cos(\vec k \cdot \vec r)$  
quindi E è costante in tutti i punti dove $\vec k \cdot \vec r = const$, che è l'equazione di un piano ortogonale a k (r sono tutti i punti del piano che mi interessa, e sono tutti quello il cui prodotto con k, cioè ortogonali a k)  

2- onda circolare  
a tempo fisso: $E \sim \frac 1r \cos(\vec kr)$   
quindi è costante in tutti i punti dove r è costante  
è sparito il prodotto scalare, perchè nell'onda sferica k è sempre parallela a r, quindi è sempre 1, lo tolgo  

**Errore comune: onda sferica vs polarizzazione circolare**

sono cose del tutto separate:  

1- onda sferica: fronti d'onda sferici  
come goccia che cade in un lago, genera fronti d'onda sferici  

2- polarizzazione circolare: fronte d'onda piana, e in punto fisso x, al passare del tempo, E disegna un cerchio   

**Vettore d'onda**

vettore che descrive la propagazione di un'onda, indicandone direzione e verso   

perchè sembra uno scalare, ma è un vettore:  
un'onda piana generale si scrive: $E(r,t) = \hat{y}\, E_0 \, \cos(\vec k \cdot \vec r-\omega t)$   
il prodotto scalare è: $\vec k \cdot \vec r = k_xx+k_yy+k_zz$     
se si propaga lungo un'asse cartesiana principale, tipo lungo x, ho: $\vec k = (k,0,0)$  
quindi è $\vec k$ che contiene la direzione di propagazione dell'onda, non la x  
le coord in teoria appaiono tutte sempre, in $r = (x,y,z)$  
sono i componenti di $\vec k$ accanto che sono 0 e ne cancellano alcune  

#### Fase

**Cosa posso studiare**

1- modulo: fase  
è l'argomento del coseno: $k \cdot r - \omega t$  
dice in che punto dell’oscillazione sei  

2- direzione: traversale/longitudinale  
devo vedere se la direzione di propagazione (dipendenza spaziale) e di oscillazione (dipendenza temporale) sono perpendicolario parallele

#### Piana, trasversale

**Proprietà**

quando: onda EM piana in un mezzo NON assorbente  
$E \perp B \perp k$  
$|E| = v |B|$  

in mezzo assorbente, la prima è vera, ma la seconda no???  

**Rapporti**

1- piana  
obtain by analytically extending the electric field expression into the complex space  

2- piana vs trasversale:  
nel vuoto: piana -> trasversale  
nella materia: non è detto, si possono avere onde con componente longitudinale (es. onde di plasma)  

3- trasversale vs polarizzazione  
trasversale  ⇒  ha polarizzazione  
posso definite polarizzazione solo se l’onda è trasversale, perché riguarda il piano ortogonale a k  

recap: nel vuoto  
piana -> trasversale -> polarizzata  

**Esempio**

$E(x,t)=E_0 [\cos(kx-\omega t)\hat{y}+\sin(kx-\omega t)\hat{z}]$  
trasversale: sì, perchè il campo è solo in direzione y e z, mentre la direzione di propagazione è la x   
linearmente polarizzata: no, perchè c'è un pezzo di campo sia lungo y che lungo z, sono due direzioni, quindi ha polarizzazione circolare  

**Proof: Piana -> trasversale (nel vuoto)**

cosa uso:  
formula onda piana:   $E=E_0 e^{i(k \cdot r-\omega t)}$    
eq Maxwell nel vuoto: $\nabla \cdot E =0$  

$\nabla \cdot E = i(k \cdot E_0)e^{i(k \cdot r - \omega t)}=0$  
l'esponenziale non può essere 0, deve esserlo la parentesi   
e $k \cdot E=0$ è esattamente la definizione di onda trasversale    

nota sul calcolo:  
$\nabla \cdot$ somma tutte le derivate parziali spaziali, qui c'è solo quella per r, che ha derivata spaziale: $ik \cdot$   
dove i non è un vettore quindi lo porto fuori dalla parentesi  

**Cos'è un'onda EM**

È un campo che:
1- si propaga
2- trasporta energia
3- può esistere anche senza sorgente locale

cos’è rispetto alle particelle?
1- Classicamente: È un campo continuo nello spazio.
2- Quantisticamente: Le sue eccitazioni quantizzate sono i fotoni.
