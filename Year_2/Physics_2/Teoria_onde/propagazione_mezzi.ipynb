### Permitittivà dielettrica

**Funzione dielettrica**

$\epsilon(\omega)=1+\frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}$  

cioè come la permittività del mezzo dipende dalla frequenza dell'onda che lo attraversa  

#### Casi

**Casi**

1- Modello di Lorentz  
cosa: elettrone legato (o nube elettronica legata)  
funzione: $\omega_0 \neq 0$ (la funzione dielettrica completa)  
quando: dielettrici e delle transizioni atomiche/molecolari  

2- modello di Drude  
cosa: elettrone libero (non legato)  
funzione: $\omega_0=0$ (no forza di richiamo)  
quando: metalli o plasma elettronico, condutori (perchè gli elettroni di conduzione sono liberi)  

ognuno dei due può essere con o senza dissipazione ($\gamma=0$)

**Casi semplificati**

1- modello di Lorentz con dissipazione  
formula intera    
$\epsilon(\omega)=1+\frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}$  

2- modello di Lorentz senza dissipazione  
$\epsilon(\omega) = 1+\frac{\omega_p^2}{\omega_0^2 -\omega^2}$  

3- modello di Drude con dissipazione  
$\epsilon(\omega)=1- \frac{\omega_p^2}{\omega(\omega+ i \gamma)}$  

4- modello di Drude senza dissipazione  
$\epsilon(\omega)=1-\frac{\omega_p^2}{\omega^2}$  


**Note**

1- nel modello di Lorentz c'è segno +, in quello di Drude segno -    
perchè rimane solo $\omega$, che il segno -, quindi lo porto davanti  

**Altri casi, per esercizi**

1- se ho più di un tipo di cariche:  
$\epsilon(\omega)=1-\Sigma_i \frac{\omega^2_{pi}}{\omega^2}$  
ogni specie contribuisce con la sua frequenza di plasma  
le polarizzazioni si sommano (linearità)  
guarda sotto, quando passo da P a $\epsilon$  

2- con campo magnetico  
$\epsilon(\omega)=1-\Sigma_i \frac{\omega^2_{pi}}{\omega(\omega \pm \gamma)}$  
inizio da: $m \ddot v = q(E + v \times B)$  
cioè: $\ddot x -\dot x \frac {qB}m= \frac{qE}m $  
confronto con l'eq del moto nel modello di Lorentz, ho $\omega=0$ (niente che produce forma elastica di richiamo) e $\gamma = -\frac{qB}{m} = \Omega_c$  
anche qui, se ho più specie sommo l'apporto di ognuna  

#### Trovarla

**Modello di Lorentz: equazione del moto di un singolo elettrone (o di nube elettronica)**

$\ddot x + \gamma \dot x +\omega_0^2 x = -\frac em E$  
  
1- $\gamma$: attrito, dissipazione   
da fonte non precisata, magari attrito con aria    

2- $\omega_0$: frequenza propria di risonanza dell’elettrone legato  
perchè elettrone è attratto dal nucleo con una forza che si comporta come una forza elastica, cioè l'elettrone è un oscillatore armonico con frequenza $\omega_0$    

3- $eE$: force from the external EM wave applied     
ignore the qvB part, since $B \ll E$ when $v \ll c$   


$\omega_0$ → dove il materiale assorbe

ωp → quanto forte è la risposta collettiva

γ → quanto dissipa

**Trovare funzione dielettrica**

1- risolvo eq del moto di singolo elettrone  
voglio $x(t)$ con stessa fase di $E(t)$ (modo per risolvere eq complesse a regime)  
$(-\omega^2 -i\gamma \omega+\omega^2_0)x_0 =-\frac em E_0$  
$x_0 = \frac em \frac{E_0}{\omega_0^2-\omega^2-i \gamma \omega}$   

2- polarizzazione macroscopica  
perchè $\epsilon$ descrive il comportamento dell'intero materiale, non del singolo elettrone  
$P=N p = N \; e\, x(t) = \epsilon_0 \frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega} E_0$   

3- trovo $\epsilon$ dalla definizione di P:    
$P(\omega)=(\epsilon(\omega)-\epsilon_0)E(\omega)$  
$\epsilon(\omega)=1+\frac P{\epsilon_0 E} =1+\frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}$ 

**Polarizzabilità atomica**

$p(\omega)=\alpha(\omega)E(\omega)$  

$α(ω)$: quanto facilmente l’atomo si polarizza  
quanto gli elettroni si spostano rispetto al nucleo quando arriva il campo  
se è grande, l'atomo si deforma molto

**Frequenza di risonanza e plasma**

modello di Lorentz: ci sono sia freq di risonanza che di plasma  
ma quella di risonanza domina (quando vicino alla risonanza), perchè vicino a risonanza il denominatore diverge    
quindi i regimi sono $\omega \ll \omega_0$, $\omega \gg \omega_0$  

modello di Drude: c'è solo freq di plasma  
quindi i regimi sono; $\omega \ll \omega_p$, $\omega \gg \omega_p$  

**COme studiare il grafico**

1- per regimi di basse a alte frequenza, fai il limite di $\epsilon(\omega)$ (matematicamente)  
2- per attorno a risonanza, rifletti sul significato fisico della risonanza, cioè quando l'onda è assorbita  

**Modello di Lorentz**

$\epsilon(\omega)=1+\frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}$  

parte reale  
1- basse freq ($\omega \ll \omega_0$): positiva e costante   
$\epsilon(\omega) \sim 1+\frac{\omega_p^2}{\omega_0^2}$  
2- alte freq ($\omega \gg \omega_0$): tende a 1   
$\epsilon(\omega) \sim 1+\frac{\omega_p^2}{\omega^2} \rightarrow 1$  
3- vicino a risonanza ($\omega \sim \omega_0$): diverge, e può diventare negativa  

parte immaginaria:  
quasi zero ovunque, con solo un picco alla risonanza  

spiegazione:  
alte e basse freq, basta fare i limiti della funzione (matematicamente)  
a risonanza: il sistema “risuona”, l’elettrone assorbe energia, cioè $\epsilon_I$ grande e $\epsilon_R$: cambia velocemente  
lontano dalla risonanza il sistema non risponde molto, non c'è dissipazione  

**Modello di Drude (conduttori)**

$\epsilon(\omega)=1- \frac{\omega_p^2}{\omega(\omega+ i \gamma)}$  

1- bassa frequenza ($\omega < \gamma$)  
$\epsilon \sim 1-i\frac{\omega_p^2}{\omega \gamma}$  
grande parte immaginaria  
cioè forte dissipazione, comportamento conduttivo  

1- media frequenza ($\omega < \omega_p$):  
$\epsilon_R < 0 \rightarrow$ k immaginario puro    
onda evanescente  
cioè non si propaga, il metallo riflette tutto (eg. metalli con luce visibile)   

2- alta frequenza ($\omega > \omega_p$):   
$\epsilon \sim 1 -\frac{\omega_p^2}{\omega_2}$  
come in modello di Lorentz, onda propaga, mezzo è trasparente  
eg. metalli con UV, sono trasparenti per le onde UV   

**Assorbimento**

avviene quando la parte immaginaria è alta  
Lorentz: a freq di risonanza  
Drude: a freq basse  

**Esempio oro**

1- parte Drude  
come tutti i metalli, l'oro riflette tutta la luce visibile  
perchè ha $\omega_p$ maggiore della $\omega$ della luce visibile  

2- parte Lorenze  
Nell’oro ci sono transizioni elettroniche interne (bande d → banda di conduzione) a energie nel blu/UV vicino.  
quindi la luce blu viene assorbita  

conclusione:  
riflette rossa/gialla (bassa freq), assorbe blu (alta freq)  
quindi appare giallo  

**Esempio acqua di mare**

luce visibile è $\omega > \gamma$, alta frequenza  
e anche $\omega > \omega_p$  
quindi k reale, c'è propagazione senza assorbimento  
infatti acqua è trasparente  

**Trasparenza**

trasparente, riferito ad una specifica frequenza, significa che a quella frequenza non c'è assorbimento  
cioè a quella frequenza, la parte immaginaria di $\epsilon$ è piccola  

si dolito quando dico trasparente, assumo che sto parlando della frequenza della luce visibile  
ma posso dire 'trasparente ai raggi UV', cioè i raggi UV lo attraversano senza essere assorbiti  

### Vettore d'onda e  tipi di onde

**Vettore d'onda complesso**

$k=k_R+i k_I$  
$E(x,t)=E_0e^{i(kx-\omega t)}=E_0 e^{-k_Ix} e^{i(k_Rx-\omega t)}$  

1- parte reale di k: propagazione   
ha il termine di fase come un'onda normale  

2- parte immaginaria di k: dissipazione, attenuazione  
ha decadimento esponenziale dell'ampiezza, non oscilla nel tempo  

cioè, onda evanescente:  
$E(x,t)= \hat yE_0 e^{k_I x}\cos (\omega t)$  
oscilla nel tempo, ma non nello spazio  

**Rapporto k e $\epsilon$**

valore di $k$ | valore di $\epsilon$ necessario ($k \sim \sqrt  \epsilon$) | tipo di onda 
---|---|---
reale | reale e positivo | propagante pura (no attenuazione)  
immaginario puro | reale e negativo | evanescente (no propagazione)  
complesso | complesso | propagante smorzata  

**Quando accade**

k reale: propagazione pura  
quando: mezzo senza perdite, $\sigma=0$ (dielettrico, non scorre corrente)    

k complesso: propagazione + attenuazione  
quando: conduttori $\sigma \neq 0$  

k puramente immaginario: solo decadimento (no propagazione)  
quando: sotto frequenza di plasma, riflessione totale

**Relazioni**

$n^2(\omega) = \epsilon_r (\omega) \mu_r(\omega) \sim \epsilon_r (\omega)$    
$k(\omega)=\frac \omega c n(\omega)$  

**Parti reale e complessa di $\epsilon$**

in generale $\epsilon$ è un numero complesso: $\epsilon(\omega)=\epsilon_R(\omega)+i \,\epsilon_I(\omega)$  

1- $\epsilon_R$: propagazione     
perchè: $k=\sqrt \epsilon$, il suo segno stabilisce se k è reale o meno   
$>0$: sì    
$<0$: no (evanescente)  
non può mai essere 0???  

3- $\epsilon_I$: decide se perde energia    
$=0$: no perdita     
$>0$: attenuazione  


**Quando accadono i casi**

guarda la funzione dielettrica per saperlo: 
$\epsilon(\omega)=1+\frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}$    

$\epsilon$ reale: quando $\gamma =0$  
infatti, vuol dire che non c'è attenuazione    

reale positivo: $\omega \ll \omega_0$ (basse frequenze)  


### Altro

**Indice di rifrazione**

$n^2(\omega)=\epsilon_r \mu_0 \sim \epsilon_r$  
perchè $\mu_r \sim 1 $ in quasi tutti i materiali, a meno che non siano proprio ferromagnetici  
mentre $\epsilon$ varia di parecchio  

$n$ → ottica (propagazione)  
$ϵ$ → risposta del materiale

**Frequenza di plasna**

$\omega^2_p = \frac{ne^2}{\epsilon_0 m}$   
n è la densità di elettroni, che è diversa per ogni materiale   
È la frequenza naturale di oscillazione degli elettroni liberi  

si usa soprattuto in modello di Drude (conduttori) dove:  
$\omega > \omega_p \rightarrow \epsilon(\omega)>0$  
quindi onda si propaga  
$\omega < \omega_p \rightarrow \epsilon(\omega)<0$  
quindi k è immaginario, onda non entra, riflessione    

separa:  
trasparenza  
riflessione  

**Onda evanescente**

dove appare:  
1- riflessione totale interna (nel mezzo esterno, onda non propaga → diventa evanescente)  
2- metalli a freq basse, perchè k diventa immaginario  
3- guida d'onda (sotto cutoff)

cos'è:  
il campo esiste ancora, ma non trasporta energia lontano  
resta confinato vicino alla superficie, perchè penetra un po' nel mezzo, ma poi decade esponenzialmente nello spazio  

in pratica serve a rispettare le condizioni di raccordo  

esempio: riflessione totale
nel secondo mezzo non c’è propagazione, ma il campo non è zero  
ho la componente parallela alla superficie di separazione (condizioni al contorno)  
è evanescente (è evanescente, esiste solo lì vicino al bordo)  

energia:  
localmente il campo oscilla nel tempo  
ma il flusso medio (Poynting) non entra nel mezzo  
il campo oscilla nel tempo, ma non oscilla nello spazio, la sua ampiezza diminuisce rapidamente entrando nel mezzo

**Skin depth**

$\delta = \frac{1}{k_I}$    
distanza su cui l’onda si attenua di un fattore e (utile per onda evanescente)  
si usa per scrivere: $E(x) \sim e^{-x/\delta}$ 

formula:  
$\delta \sim \sqrt{\frac 2 {\mu \sigma \omega}}$  
$\sigma$: conducibilità    

alta frequenza → campo cambia veloce → penetra poco  
buon conduttore → schermatura forte → penetra poco

### Poynting

**Dimostra che onda evanescente non trasporta energia**

onda evanescente: $E(x,t)=\hat y E_0e^{-kx}\cos(\omega t)$  

1- trovo campo B indotto (spiegato come in esercizi)  
$\nabla \times E = -\frac{\partial B}{\partial t}$  
$B_z(x,t)=\frac k \omega E_0 e^{-kx} \sin (\omega t)$  
poichè dipendono da sin e cos, E e B sono sfasati di 90°  

2- calcolo S  
$S \sim E \times B  \sim \sin (\omega t) \cos (\omega t) = \sin (2 \omega t)$  
$<S>= \frac 1 T \int_0^T S(x,t) \, dt = 0$  
perchè la media di $\sin (2 \omega t)$ su un periodo è 0    

cioè:  
c’è flusso istantaneo avanti e indietro (NON ho S(t)=0 per ogni t)   
ma nessun trasporto netto di energia (HO $<S(t)>=0$, la media sul periodo)  

**Dimostra che onda propagante trasporta energia**

onda propagante: onda evanescente: $E(x,t)=cos (kx - \omega t)$  

1- trovo B   
$B\sim \cos (kx - \omega t)$   
cioè E e B sono in fase  
davvero??? come si trova???   

2- trovo S  
$S \sim cos^2(kx-\omega t)$  
$<cos^2>=\frac 12$  
quindi $<S> \neq 0$  
 


**Effetto tunnel EM**

in teoria, un'onda evanescente ricade subito  
ma se, prima che sia decaduta del tutto, ci metto un'altro materiale con $\epsilon$ tale che l'onda non è più evanescente, posso riprenderla  
quindi è come effetto Tunnel: anche se in teoria l'onda è evanescente in un mezzo, e quindi non dovrebbe propagarsi in quella direzione, riesce a passare al mezzo dopo, perchè il mezzo problematico era troppo corto per attenuarsi completamente
