ho interfaccia tra due materiali  

1- approssimazione impulsiva:  
cosa: c'è stacco netto tra i due materiali, dscontinuità    
quella che uso sempre in ottica geometrica  
quando: $L \ll \lambda$  

2- approssimazione adiabatica  
cosa: c'è regione di mezzo tra i materiali in cui ci sono proprietà intermedia, che variano in modo continuo, no discontinuità  
quando: $L \gg \lambda$  

L: scala di lunghezza su cui variano le proprietà ottiche del mezzo  
cioè spessore della parte con proprietà intermedie  
$\lambda$: lunghezza d'onda della luce incidente    

noi studieremo solo il caso della discontinuità

**Come disegnare**

per i versi di $k, E, B$, segui la regola della mano destra   
definisci quella ortogonale sempre entrante o sempre uscente, in tutti e tre i raggi   

**Raggi**

1- onda incidente  
arriva dal mezzo 1, sbatte sull'interfaccia  

2- onda riflessa  
torna indietro da dove viene l'onda incidente  

3- onda trasmessa/rifratta  
prosegue dall'altro lato dell'interfaccia, nell'altro materiale  

le frecce disegnate rappresentato $\vec k$ dell'onda  

**Direzioni**

1- $\vec n$: normale all'interfaccia   
direzione perpendicolare alla superficie su cui sbatte onda incidente  

2- piano di incidenza  
piano spannato da $\vec n$ e $\vec k$ (onda incidente)  
è diversa dall'interfaccia, che è il piano che separa i due mezzi  

![](https://www.edmundoptics.co.uk/media/chfcoqox/fig-4-itp.gif?rmode=max)

**Tipi di polarizzazione**

1- polarizzazione P  
quando: E è parallelo al piano di incidenza  
nome: TM (trans magnetic), perchè B è perpendicolar al piano di incidenza  

2- polarizzazione S  
quando: E è ortogonale al piano di incidenza  
nome: TE (trans electric), perchè E è perpendicolar al piano di incidenza   

polarizzazione P e S sono una base per costruire qualunque polarizzazione  
cioè per avere E diretta in qualunque direzione, non solo perfettamente parallela o ortogonale  

**Nota**

nel seguito, usero P e S per indicare le componenti di una un'onda parallela e ortogonale  
non vuol dire che tutta l'onda sia parallela o ortogonale, ma solo che sto considerando quelle componenti  

nota che P e S sono una base, quindi non definiscono la polarizzazione  
la polarizzazione è definita dal DEPHASING tra componenti P ed S  
cioè un'onda che ha sia componenti P ed S, ma sono sfasate di 0°, ha polarizzazione lineare (perchè E è sempre sulla stessa linea, che non è nè parallela nè ortogonale)  
se invece sono sfasate di 90°, è polarizzazione circolare  

**Tipi di riflessione**

$n_1>n_2$: riflessione interna   
$n_1<n_2$: riflessione esterna   

immagina che n minore sia l'aria, m maggiore sia il materiale  
interna significa che onda inizia dentro il materale  
esterna significa che onda inizia fuori dal materiale, cioè nell'aria  

**Recap**

r,t: parlo di raggio riflesso o trasmesso?  
$P,S$: parti di componente parallela o ortogonale a piano di incidenza  


### Angoli

**Formule**

1- onda riflessa: $\theta_r = \theta_i$  
legge di Cartesio  

2- onda trasmessa: $n_1 \sin \theta_i = n_2 \sin \theta_t$  
legge di Snell

**Spiegazione fisica Brewster**

1- La luce fa oscillare cariche nel materiale → dipoli oscillanti (nella direzione di E dell'onda trasmessa)      
2- riemettono luce (riflessione + trasmissione)  
3- ma un dipolo oscillante NON irradia lungo la sua direzione di oscillazione  

all'angolo di Brewster: $k_r \perp k_t$  
in polarizzazione P, campo E è nel piano del disegno, ed è sempre ortogonale a $k$  
dipolo oscilla lungo $E_t$, quindi ortogonale a $k_t$ (sempre)    
ma il $k_r \perp k_t$ (a Brewster), quindi il dipolo è parallelo a $k_r$  

in polarizzazione S invece non succede, perchè???  

quindi l'angolo di Brewster non esiste nel metallo, perchè l'onda trasmessa VISIBILE è evanescente, quindi i dipoli al suo interno non oscillano   
cioè, non c'è perchè i metalli sono mezzi assorbenti, non propaganti, quindi l'onda trasmessa è evanescente  

#### Ricava formule

**Studio preliminare**

1- modellizzo nei mezzi 1 e 2:  
$E_1 = E_i e^{i(kx-\omega t)}+E_r e^{i(k'x-\omega' t)}$  
$E_2 = E_t e^{i(k''x-\omega'' t)}$  
k, k', k'': onda incidente, riflessa, trasmessa 
nel mezzo 1 ci sono onda incidente e riflessa, nel mezzo 2 trasmessa   

2- continuità di componente di E parallela all'interfaccia (a x=0)  
$E_{i \parallel} e^{-i \omega t} + E_{r \parallel} e^{-i \omega' t} = E_{t \parallel} e^{-i \omega'' t}$  
è vera per ogni t solo se: $\omega=\omega' = \omega''$    

3- continuità di componente di E parallela all'interfaccia (a t=0)  
simile a prima, ma ora è $k_y$ che deve essere uguale in tutte le onde  

**Ricava legge di Cartesio**

1- onda incidente e riflessa sono entrambe nel mezzo 1, quindi:  
$n_1 = \frac{k c} \omega=\frac{k' c}{\omega}$   
so che $k'=\sqrt{k_x^2+k_y^2}$, e scelgo la sol $k_x'=-k_x$, perchè vedo che l'onda riflessa deve tornare indietro 

2- conclusione:  
ho: $k'=(-k_x, k_y)$  
graficamente, vuol dire $\theta_r = \theta_i$  

**Ricava legge di Snell**

onda incidente e trasmessa sono in mezzi diversi, quindi:  
$(\frac{n_2}{n_1})^2 = \frac{k_x''^2+k_y''^2}{k_x^2+k_y^2} = \frac{(\frac{k_x''}{k_y''})^2+1}{(\frac{k_x}{k_y})^2+1} = \frac{\frac{1}{\sin^2 \theta_t}}{\frac{1}{\sin^2 \theta_i}} = \frac{\sin^2 \theta_i}{\sin^ 2 \theta_t}$

altro modo:  
luce percorre il cammino che minimizza il cammino ottico (come esercizio di arrivare da A a B passando ad abbeverarsi al fiume)  

#### Altro

**Cammino ottico (OPL: Optical path length)**

$OPL = n \, s = n \, vt = ct$  

cioè, una misura della lunghezza percorsa di un mezzo (s) pesata secondo l'indice di rifrazione di quel mezzo (n)   
in pratica, mezzi con n maggiore sono più faticosi da attraversare  
è anche la distanza che la luce avrebbe percorso nel vuoto in quello stesso tempo ($ct$)


#### Angoli speciali

**Riflessione totale (interna)**

quando: passo da un mezzo più denso a uno meno denso ($n_1 > n_2$)   
cosa: se super l'angolo critico, non c'è onda trasmessa  

spiegazione:  
se vado da mezzo più a meno denso, ho $\theta_t > \theta_i$  
però al massimo posso avere $\theta_t = \frac \pi 2$, che si verfica all'angolo di incidenza $\theta_c$    
oltre questo angolo, secondo legge di Snell dovrei avere $\sin \theta_t > 1$, quindi non ci piò essere onda trasmessa  
quindi l'onda trsmessa è solo un'onda evanescente (deve esserci per rispettare condizioni di raccordo)  

intuitivo:  
onda rimane nell'oggetto più denso, non riesce a passare a quello più rarefatto  
si chiama 'interna' perchè avviene solo se passo da oggetto più denso a uno meno denso (probabilmente l'aria), quindi assumo che l'interno sia il mezzo 1 (dove resta confinato), mentre l'aria sia l'esterno  

**Dimostra che ho onda evanescente**  

1- formule per i due mezzi  
$n_1 = \frac{kc}\omega$  
$n_2 = \frac c \omega \sqrt{k''_x+k''_y}$  

2- riscrivo la seconda  
$(\frac{n \omega}c)^2 = k_x''^2+k_y''^2 = k_x''^2+k^2 \sin^2 \theta_i$  
$k''^2_x = \frac {\omega^2}{c^2} [n_2^2-n_1^2 \sin^2 \theta_i]$  

se $\theta_1 > \theta_c$, la parentesi è negativa, quindi $k_x''$ è immaginario puro, quindi onda evanscente  

### Ampiezze

**Leggi di Fresnel (Frenèl)**

laws about how the amplitude of waves change when they get refracted (passing between 2 mediums)  
easier way than writing the condizioni al contorno sulle componenti di E e B  

essentially, i know an incident wave is a bit reflacted and a bit propagated  
these tell me how much of the wave is reflacted, and how much propagated  

**Come si ricava legge di Fresnel**

con le condizioni di raccordo  
cioè si conserva $E_{\parallel}$ e $H_{\parallel}$  

**Coefficienti di Fresnel**

$r = \frac{|E_r|}{|E_i}$   
$t = \frac{|E_t|}{|E_i}$   
 
both r and t can be $r_S$ or $r_P$, so take this component in all the waves in the formula  
dipendono da $\theta_i$ e $\theta_t$, e il secondo lo posso trovare con la legge di Snell se conosco i materiali     

**Potenze e intensità**

riflettanza: $R = \frac {P_r}{P_i} = r^2$   
trasmittanza: $T = \frac{P_t}{P_i} = \frac{n_2 \cos \theta_t}{n_1 \theta_i}t^2$  

potenza: $P= I \cdot A$ (prodotto scalare tra intensità e area)  
intensità: $I =\frac 12 n \epsilon_0 c |E|^2 \sim n |E|^2$  

conservazione energia: $R+T=1$  
cioè, SE il materiale è non assorbente, tutta la potenza incidente viene o riflessa o propagata  
vale separatamente per le componenti P e S

**Rombo di Fresbel**

che succede:  
1- il raggio entra nel prisma  
2- subisce 2 riflessioni interne totali  
3- ogni riflessione introduce un ritardo di fase tra s e p    

il ritardo di fase introdotto dalla riflessione dipende dall'angolo di incidenza  
quindi, scegliendo adeguatamente gli angoli del raombo, posso fare che entrambe le riflessioni intriducono un ritardo di $\pi/4$  
così in totale dò $\pi/2$, polarizzazione lineare è diventata circolare
