### Effetto Hall

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTIZOOUNfqr3Cexj7FQ9YxlFy3rlXE9Krfd5A&s)

**Situazione**

I e B sono ortogonali  
S=ab è la sezione ortogonale a I  
b è lato parallelo B, a è l'altro  

**Spiegazione**

c'è forza di Lorentz, che crea accumuli di carica sulle superfici  
questi creano un campo elettrico 

a regime stazionario (quando gli accumuli non crescono più):  
$F=qE_H+qv_dB=0$   
$v_d$: velocità di drift (del moto ordinato verso una direzione)  
$E_H = v_D B$: campo generato dagli accumuli di carica  

**Tensione di Hall**

$\Delta V_H = -E_H \, a$  
differenza di potenziale tra le due facce a regime stazionario  

altri modi di scriverlo:  
$I= J \, S = -nqv_d \, ab$  
$\Delta V_H = -(v_d B) \, a = -(-\frac{I}{nq \, ab})Ba = \frac{1}{nq} \frac{IB}{b}$  
$R_H =\frac{1}{nq}$: costante di Hall  

**Dimostra che gli elettroni si muovono**  

nella formula per la tensione di Hall, posso misurare $\Delta V, I , B , b$   
quindi posso misurare $R_H$  

conseguenze:  
1- sperimentalmente, $R_H<0$, e poichè $R=\frac{1}{nq}$, ne deduco che il moto è fatto da elettroni, non protoni  
2- ho un modo per misurare n, la densità dei portatori di carica elettrica  

**Sonda effetto Hall**

strumento che misura il campo magnetico  
in realtà, la componente di B parallela ad una faccia  
poi posso girare lo strumento per misurare tutte le componenti  

come:  
$B=(\frac{b}{R_H})\frac{1}{I} \Delta V_H$   
$\frac{b}{R_H}$: dipende dalla taratura dello strumento, lo setto io  
$I$: è quella inviata dalla sonda, la scelgo io  
$\Delta V_H$: lo misuro

**Cos'è**

correnti generate in blocchi conduttori in campo magnetico variabile (quindi per induzione)  
di solito non sono volute, perchè la corrente dissipa calore per effetto Joule  
quindi per limitarle, invece di un unico blocco si mettono lamine di materiale conduttore intervallare da lamine di matriale isolante  
ma a volte sono fatte appost, tipo nel fornello a induzione o nel freno magnetico  

di solito:  
corrente indotta: quando gira su un circuito  
corrente parassita/di eddy: quand gira su un unico blocco

**Fornello a induzione**

1- si applica B oscillante, che genera una corrente  
con frequenta abbastanza alta da non sentirlo, sennò sentirei un fischio  
su ogni fornello viene applciato separatamente, così li posso accendere separatamente e il resto del piano non è caldo???  

2- i cerchi (e il resto del piano no???) hanno resistività, quindi si riscaldano per effetto Joule (sono una resistenza)  

**Freno magnetico**

treno si muove su una superficie conduttrice  
il freno è un oggetto che applica un campo B verso il basso, quindi verso questa superficie  

quando applico campo B:  
nel binario c'è variazione di flusso, si formano eddy currents  
queste creano a loro volta un campo B indotto per ch si oppone alla variazione di flusso  
1- parte di binario subito avanti al treno:   
il flusso verso il basso sta aumentando, quindi si crea B verso l'alto  
due campi di direzione diversa danno forza repulsiva, quindi la parte di binario avanti è repulsiva, cioè blocca il treno    
2- parte di binario subito ditro al treno:  
il flusso verso il basso sta diminuendo, quindi crea un B indotto verso il basso  
quindi un campo B di stesso verso si quello originale, così c'è forza attrattiva???  

che vuol dire forza attrattiva/repulsiva?  
non serve disegnare il campo indotto, solo la corrente indotta  
poi disegna la forza di Lorentz data da questa corrente e dal campo B originale

### Ciclotrone

**Trova formule**

$m\frac{v^2}R = qv_\perp B$  

$\omega = \frac{v_\perp}R =\frac{q B}m$  
$R= \frac{v}{\omega} = \frac{mv}{2B}$  

momento magnetico (invariante adiabatico) (spiegato in guida a gabbia magnetica):    
$\mu = \frac{mv_ \perp^2}{2B}$      
lo uso per definire:  
$R = \frac{v_\perp}{\omega_c} =\sqrt{\frac{2 \mu m}{q^2 B}}$

come funziona:  
metto unca carica q nel ecrchio

**Radiofarmaci per medicina nucleare**

PET: positron emission tomography  
serve a vedere le zone tumorali all'interno del corpo

**Selettore di velocità**

come funziona:  
sorgente lancia particelle in orizzontale, in una regione con campi E e B

### Effetto pinch

in un cavo, ho un densità uniforme di elettroni che drifta con velocità v, cioè ho una corrente $J=nev$  
la corrente crea un campo B con linee di campo circolari (regola della mano destra con I)  
B causa una forza di Lorentz che spinge gli elettroni verso l'asse del filo  
ma quando si raggruppano sull'asse del filo si forma un conentrazione di cariche, che causa un campo eletrostatico E che li respinge (in pratica si respingono perchè sono dello stesso segno)  
quindi si raggiunge una situazione di equilibrio, dove forza elettrostatica di repulsione è uguale a forza di Lorentz  

conclusione:  
densità di elettroni non è uniforme del filo, ma è maggiore al centro, sull'asse  

**È trascurabile?**

in conduttore ohmico: sì  
in fulmine: no, perchè gli elettroni si muovono troppo veloce   
la loro v è grande perchè l'aria ionizzata è un conduttore non ohmico   
quindi F di lorentz è forte, quindi basta un piccolo spostamento delle cariche per produrre E forti in grado di opporsi alla forza di Lorentz  
quindi effetto pinch è forte, gli elettroni sono molto schiaciati, infatti il fulmine è stretto e sottile  
in pratica il fulmine confina tramite un campo magnetico una colonna di gas ionizzata
