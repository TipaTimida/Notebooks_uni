ho una particella carica  
posso confinarla in una regione dello spazio, usando campi elettrici o magnetici?   

elettrico: sì, ma deve essere variabile nel tempo  
magnetico: sì, ma deve essere variabile nello spazio (non omogeneo)

### Con E

**Tipi**

per il th di Earnshaw non basta campo elettrostatico, ma posso usare:  
trappola di Paul: campo elettrico variabile nel tempo  
trappola di Penning: campo magnetico + elettrico  


**Teorema di Earnshaw**

cosa dice:   
in elettrostatica, non esistono minimi stabili del potenziale  
$\nabla^2 \phi =0$  
cioè un distribuzione di carica (singola carica puntiforme, o insieme di cariche puntiformi) non può stare in equilibrio stabile per il solo effetto di forze elettrostatiche/di un campo elettrostatico  

perchè:  
equilibrio stabile è in un minimo del potenziale, ma non può essistere per conseguenza del th della media (ho detto che V non ha minimo)  
quindi ci sono punti di equilibrio ($\Sigma F=0$), ma non stabili (V è un minimo)  
per equilibrio stabile servono campi dinamici  

**Th della media**

cosa dice:  
la media di V (potenziale) su una superficie sferica nel vuoto è uguale alla V del centro della sfera  
dove V è una funzione armonica (cioè laplaciano=0)  

conseguenze:  
nello spazio che separa le cariche sorgenti, V non può avere max o min  
sennò, centrando una superficie sferica in quel punto, violerebbe il teorema della media  
perchè la media dei punti V della sfera non sarebbe uguale alla V del centro, visto che è un max  

### Con B

### Invariante adiabatico

**Trova invariante adiabatico**

riduco il problema in 1D  
è un oscillatore, quindi $I=T \, E  = \frac{2 \pi}{\omega_c} E = const$  
$I \propto \frac{E}{\omega_c} \propto \frac{v^2_\perp}{B} \propto \omega_c R^2 \propto BR^2$

cosa vuol dire:  
$v^2_\perp/B$: se aumento B, aumenta la velocità della particella (ovvio della def di $\omega_c$)  
$BR^2$: è il flusso del campo magnetico, il raggio dell'orbita cambia il modo che il flusso di B attraverso essa sia costante  

definisco il momento magnetico:  
$\mu = \frac{mv_ \perp^2}{2B}$ è il momento magnetico  
che quindi è conservato (se campo B varia LENTAMENTE), è un invariante adiabatico  


**Conseguenza**

$\mu$ è un ivariante adiabatico  
visto che si conserva, posso riscrivere il raggio ciclotronico (dell'orbita) come:  
$R = \frac{v_\perp}{\omega_c} =\sqrt{\frac{2 \mu m}{q^2 B}}$  

**Campi variabili**

$\mu$, il momento magnetico è costante sia se B varia lungo z, sia se avria lungo t  
gli effetti cambiano:  
varia lungo t: raggio ciclotronico varia  
varia lungo z: crea specchio magentico  

se B varia lungo t: non c'è nessuna direzione preferita nello spazio, niente specchio  

### Bottiglia magnetica, magnetic mirror

**Recap**

campo B maggiore all'estremità, debole al centro  

la particella fa traiettoria elicoidale, attorno ad una linea di campo  
quando entra in zona di campo più forte:
aumenta la componente trasversa
rallenta lungo la direzione del campo
si ferma e torna indietro

risultato: rimbalza tra due “specchi magnetici”

![](https://www.energyencyclopedia.com/data/web/glossary/magnetic-mirror1.svg)

**Specchio magnetico**

1- il campo B produce una forza di Lorentz  
che è ortogonale allo spsotamento, quindi può solo cambiare la direzione della velocità, non il modulo    
quindi l'energia è conservata $E=\frac{1}{2}m(v_\perp^2+v_\parallel^2)$  

2- la particella gira a spirale lungo le linee di campo   
perchè particella sottoposta a campo B fa un cerchio nel piano perpendicolare a B, e moto uniforme lungo l'asse di B  
cioè moto elicoidale  

3- come cambia la traiettoria  
$\mu = \frac{mv^2_\perp}{2B}$ è un invariante adiabatico  
se B cambia lentamente, cioè su un periodo il campo B visto dalla particella è più o meno uguale, allora $\mu$ si conserva  
quindi se B aumenta, $v_\perp$ aumenta  
nel frattempo il raggio diminuisce, quindi fa una spirale  
quindi, per conservare l'energia, $v_\parallel$ diminuisce  
finchè ho un valore di B per ui tutta la velocità deve essere ortogonale, $v_\parallel$ diventa 0, cioè la carica si ferma e torna indietro  

**Bottiglia magnetica**

si crea mettendo due specchi magentici uno di fronte all'altro, così la carica rimbalza tra quei due (tipo ping pong)  

specchio magnetico: B(z) monotona   
bottiglia magnetica: B(z) con un minimo in mezzo

**Magnetosfera terrestre**

cosa: regione di spazio intorno alla Terra dominata dal suo campo magnetico  
funzione: proteggerci dal vento solare (particelle cariche dal Sole)  

**Fasce di Van Allen**

cosa: due regioni toroidali intorno alla Terra (nella magentosfera terrestre)  
sono una realizzazione naturale della bottiglia magnetica  

perchè:  
ha forma come di bottiglia magnetica (linee più fitte = campo più forte alle estremità), quindi ci rimangono intrappole particelle cariche  
perchè il campo magnetico terrestre è quello di un dipolo, fatto proprio ccìome serve ad una trappola magnetica  

**Esempi: aurore boreali**

1- moto delle cariche  
le cariche fanno moto elicoidale intorno ad una linea di campo   
poichè le fasce di Van Allen sono bottiglie magnetiche, la velocità del moto lungo la linea di campo diminuisce, mentre aumenta la velocità con cui fa il moto circolare, e diminusice il raggio  

2- azione della bottiglia magnetica (Fasce di Allen)  
se all'inizio $v_\perp$ è già grande, la bottiglia riesce rallentarla fino a fermarla, la particella rimane intrappolata nella fasca  
se all'inizio è $v_\parallel$ ad essere grande, la bottiglia risce a rallentarla, ma non a fermarla del tutto, quindi la particella segue la linea di campo fino alle zone basse dell'atmosfera  
poichè segue le linee di campo, scende nell'atmosfera ai poli  

3- interazione con atmosfera  
quando una particella carica arriva a quote basse:  
urta molecole di ossigeno e azoto  
eccita i loro elettroni    
quando gli elettroni tornano allo stato normale, emettono luce (verde e rosso per ossigeno, blu e viola per azoto)  
questa è l'aurora
