**Schermare campi**

1- schermare campo E statico  
come: gabbia di Faraday / gabbia elettrica  

2- schermare campo B statico  
come: gabbia magentica  

3- confinare onde EM  
come: guida d'onda  

**Gabbia magnetica vs elettrica**

Cosa | Elettrica | Magnetica
---|---|---
funzionamento | esistono monopoli elettrici, li uso per bloccare E esterno | non esistono monopoli magnetici, posso solo ridirezionare B esterno  
come è costruita | conduttore | ferromagnetici  
principio di funzionamento | E=0 dentro i conduttori (anche in una cavità) | linee di campo di B seguono i ferromagnetici  

### Gabbia di Faraday

**Spiegazione**

cosa fa: isola elettricamente il suo interno dai campi elettrici esterni    
come funziona: effetto di schermo dei conduttori  

![alt text](https://www.chimica-online.it/fisica/immagini/gabbia-di-faraday.jpg)

**Esempi**

può essere una scatola chiusa, una rete metallica, una griglia  
l'oggetto da proteggere si mette al suo interno   

esempi:  
abitacolo di un auto  
cavo coassiale  

**Come funziona effetto di schermo**

1- cariche del conduttore sono libere, si possono muovere all'interno del materiale  
2- se applico un campo esterno, le cariche si ridistribuiscono sulla superficie esterna  
3- questo genera un campo indotto che cancella il campo elettrico all’interno del materiale (o della cavità)  

quando raggiungo situazione di equilibrio di equilibrio, le cariche interne al conduttore non si muovono più, il che vuol dire che il campo interno è diventano E=0  

**Schermaggio elettrostatico: conduttori vs isolanti**

quando soggetto ad un campo elettrico, producono un campo indotto per contrastarlo  
il campo indotto viene prodotto spostando le cariche degli atomi del materiale    

1- conduttori  
le cariche sono libere, si distribuscono sulla superficie del materiale  
campo è forte quanto voglio, annulla completamente quello esterno  
risultato: campo interno nullo  

2- dielettrici  
le cariche sono legate, gli atomi/molecole si polarizzano, cioè il centro della nube elettronica non coincide più col centro dell'atomo  
non c'è separazione netta delle cariche, mi viene un campo più debole  
risultato: campo itnerno diminuito, ma non annullato  

### Gabbia magnetica

**Spiegazione**

cosa fa: devia le linee dei campi B esterni, in modo che non passino al suo interno  
come funziona: linee di campo rimangono dentro i ferromagnetici, condizioni di raccordo

**Perchè campo non esce dal ferromagnetico**

spiegazione 1:  
per minimizzare l'energia: 
$u= \frac{B^2}{2 \mu}$  
quindi nel ferromagnetico, il campo costa meno energia  
simile alla luce che quando passa a un mezzo all'altro tenta di minimizzare il cammino ottico, qui invece il campo tenta di minimizzare l'energia  

spiegazione 2:  
per condiizoni al contorno  
componente normale: si conserva $B= \mu H$  
componente tangenziale: si conserva $H$  
quindi, quale B passa dall'aria al ferromagnetico, la componente tangenziale rimane invariata, quella normale diminuisce di tanto  
cioè il campo si "piega", segue una traiettoria tangenziae, parallela ai bordi del ferromgnetico  

nota:  
le linee di B fanno il tragitto dentro il guscio, senza entrare nella parte centrale, e poi escono nella stessa direzione da cui sono entrate, ma hanno bypassato la parte centrale

**Circuito magnetico**

modellizzando la situazione come un circuito magnetico:  
il ferromagnetico è una zona di bassissima riluttanza (bassa resistenza magnetica)  
legge di Hopkinson: $R = \frac{l}{\mu A}$  
quindi il campo magnetico preferisce entrare nel μ-metal piuttosto che restare nell’aria

### Guida d'onda

**Spiegazione**

cosa: qualunque struttura che costringe un’onda elettromagnetica a propagarsi lungo una direzione  
come: riflessione totale (che è causata dalle condizioni al contorno)  
risultato: l’onda non è libera → esistono modi discreti (TE, TM)  


**Fibra ottica**

un caso speciale di guida d'onda  
non ha bisogno di materiali metallici, ma solo con n grande  

| Guida d’onda metallica                                 | Fibra ottica                              |
| ------------------------------------------------------ | ----------------------------------------- |
| confinamento per **condizioni al contorno su metallo** | confinamento per **indice di rifrazione** |
| tipica: microonde                                      | tipica: ottica                            |
| modi TE/TM puri                                        | modi ibridi (HE, EH)                      |
| campo nullo sul bordo                                  | campo evanescente nel cladding            

**Fibra ottica**

struttura:  
1- core: cilindro centrale, con n grande  
2- cladding: rivestimento cilindrico esterno, con n piccolo  

come funziona:  
l'onda passa dentro, attraverso il core  
non riesce a passare al cladding, perchè c'è riflessione totale  
quindi è confinata dentro il core  

perchè serve il cladding?  
in teoria ci sarebbe riflessione totale anche con l'aria  
ma se metto più fibre assieme rischio che si tocchino e che non ci sia aria di mezzo, quindi che l'onda passi da una fibra all'altra  
così le isolo tra di loro, e poichè un po' protettivo, tipo contro le abrasioni

**Perchè condizioni al contorno causano riflessione totale**

un conduttore emana E normale alla sua superficie  
quindi la componente di E parallela alla sua superficie è nulla  
poichè onda incidente e riflessa hanno componenti parallele opposte, che si cancellano tra loro, l'onda trasmessa non può avere conmponente parallela, o quella totale sarebbe non-0
