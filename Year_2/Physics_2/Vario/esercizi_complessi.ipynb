### Fenomeni (espansione, scattering, oscillazioni)

**Steps**

1. scrivo la legge del moto  

2. devo calcolare quanto vale E o U    
probabilmente è una funzione di r  

**Scrivere equazione del moto**

a seconda di cosa conosco:  

conosco $U(r)$, l'energia potenziale elettrica in funzione della posizione  
$m\ddot{r} = F =-\frac{\partial U}{\partial r}$  

conosco $E(r)$, il campo elettrico:  
$m\ddot{r} = F = qE$  

#### Espansione di un nucleo

**Situazione**

un nucleo è una sfera con carica uniforme  
la immagino come una sfera costellata di cariche puntiformi  
se lascio il sistema libero di evolvere, le cariche si respingono, la sfera si espande  

**Scrivo equazione del moto**

considero una carica $q$   
$r_0$: distanza di q dal centro della sfera all'inizio  
$r(t)$: distanza di q dal centro della sfera al momento t  
$R$: raggio della sfera    

l'equazione del moto per una particella si scrive:  
$m\ddot{r} = F = qE$  

mi serve E, il campo generato dalla sfera  
campo dentro una sfera carica uniforme: $E= kQ_{in} \frac{r}{R^3}$  
$Q_{in}$ è la carica nella sfera interna alla q che sto considerando ora, quindi:  
$Q_{in} =  \rho \frac{r_0^3}{R^3}$  


**Osservazione: carica rimane uniforme**

osservazione:  
E è maggiore sul bordo  
quindi le cariche sul bordo si allontanano più veloce  
e le cariche dentro non possono sorpassarle  
cioè la quantità di carica all'interno di un una shell rimane sempre la stessa  
la densità di carica rimane uniforme, ma cambia il vaore perchè la carica si distribuisce su più spazio  

**Altre richieste**

cose che posso voler sapere:  
forza sentita da una carica q (dentro la sfera) in ogni momento: $F=qE$     
energia di q in un momento preciso: $U = - \int F \, dr $ 


**Non sicura?**

se l'espansione è tale che la densità di carica rimane uniforme  
allora la legge del moto sarebbe:  
$r(t)=r_0\lambda(t)$  
pluggo questo guess nell'equazione, come soluzione (quando lo derivo ottengo $\ddot{\lambda}$)  
se $r_0$ non appare più nell'equazione, si dilata in modo omogeneo  

**Conseguenze sulla densità**

se quella soluzione per l'equazione del moto è corretta:  

Poiché tutte le distanze scalano con lo stesso fattore 
$λ(t)$, il volume di una regione cresce come $λ^3(t)$  
La carica in quella regione resta costante, quindi la densità di carica rimane uniforme durante l’espansione  
cioè la densità cambia valore, resta la stessa ovunque (se all'inizio era la stessa ovunque)

#### Oscillazioni elettroni

**Cosa fare**

scrivo la legge del moto di un elettrone:  
$m_e\ddot{\delta} = eE_{int}$  
  
calcolo cosa è E, il campo elettrico generato dalla separazione di cariche, in funzione della loro distanza  
quindi $E_{int}(\delta)$, dove $\delta$ è la distanza tra le sfere delle cariche + e -  

**Densità elettronica: cosa sono**

$n_e$: densità numerica degli elettroni liberi nel metallo (quanti elettroni di conduzione ci sono per unità di volume)  

$n_i$: densità numerica degli ioni positivi (cioè quanti atomi/nuclei ci sono per unità di volume, a volte chiamato $n_{atomi}$)   

Z: numero di valenza/conduzione, quanti elettroni liberi fornisce ogni atomo  

$\rho_m$: densità massica, kg/m^3, non elettronica
M: la massa molare  
$N_A$: numero di Avogadro

$\rho$: densità di carica

**Densità elettronica**

Per la neutralità elettrica globale del metallo:  
$n_i \; (Ze)=n_e \; e$  
$n_e = Z \; n_i$  

Densità numerica di ioni/atomi:  
$n_i = n \frac{N_A}{V} = \frac{m}{M} \frac{N_A}{V}= \frac{\rho_m}{M} N_A$  

Densità di carica:  
$\rho _e = n_e (-e) $



#### Scattering

per piccoli angoli:  
$\theta \sim \tan \theta = \frac{p_y}{p_x}$  

trucco:  
$p_y = \int F \, dt = \int F \, \frac{dx}{v} = \frac{1}{v}\frac{1}{2 \pi b} \int F \, 2 \pi b \, dx = \frac{1}{v}\frac{1}{2 \pi b} q \int E \; dS = \frac{1}{v}\frac{1}{2 \pi b} q \Phi_S(E) =  \frac{1}{v}\frac{1}{2 \pi b} q \frac{Q}{\epsilon_0}$ 

b è il parametro d'impatto  
in pratica considero un cilindro con asse lungo Q, la carica bersaglio, e largo come b, la distanza tra q, la carica che lancio, e Q  
e lunghissimo, così non devo pensare al flusso delle basi, perchè a distanza infinita il campo elettrico generato da Q è nullo  

#### Piccole oscillazioni

**Steps: armonico**

1. ho l'equazione del moto  
2. approssimo per piccole oscillazioni  
4. riordino per ottenere $\ddot{x}= -\omega^2 x$   

**Approssimo per piccole oscillazioni**

epspando per: $x \ll R$  
cioè se mi muovo lungo l'asse x, e le oscillazioni sono molto più piccole del raggio R dell'oggetto che causa il campo  

come:  
raccogli per trovare $\frac{x}{R} \sim 0$  
poi usa le espansioni di Taylor  

$(1+x)^n \sim 1+nx$  
$\frac{1}{\sqrt{R^2+x^2}}= \frac{1}{R \sqrt{1+\frac{x^2}{R^2}}} = \frac{1}{R}(1+ \frac{x^2}{R^2})^{-1/2} = 1 -\frac{1}{2} \frac{x^2}{R^2} +...$


**Frequenza delle oscillazioni**

ho scritto l'equazione del moto, cerca di capire a cosa corrisponde $\omega ^2$, il coefficiente della x a destra  

**Se so che c'è moto circolare**

scrivo leggi di newton: $\Sigma F_{rad}=m\frac{v^2}{r}$  

in pratica cerco di capire cosa è la forza centripeta  
qui probabilmente è una forza di Coulomb    
o se c'è campo magnetico può essere una forza di Lorentz  

### Teoria energia

**Tipi di energia**

l'energia è fatta di 2 parti:  

1. energia di auto-interazione (self-energy)  
in cariche puntiformi, diverge  
in cariche distribuite su un valore finito, è finita   
energia di una particella causata dal trovarsi nel campo generato da lei stessa   

2. energia di interazione tra cariche diverse  
è finita, ed è l’energia fisica che interessa nella dinamica (lavoro per portare le cariche alla distanza d...)   
energia di una particella causata dal trovarsi nel campo generato da un'altra particella  

cosa calcolano le formule:  
$U=\frac{\epsilon_0}{2} \int E^2 dV$: la somma di entrambe  
$U=k\frac{q_1q_2}{r}$: solo parte di interazione        

cosa chiedono gli esercizi:  

esercizio con distribuzione di carica: voglio quella totale, usa la formula con l'integrale 

esercizio con cariche puntiformi (anche due cariche puntiformi che interagiscono): voglio solo la parte di interazione, usa la seconda formula   
sennò posso usare la parte con l'integrale, ma togliendo l'energia di auto-interazione, che la fa divergere:  
$U=\frac{\epsilon_0}{2} \int (E_1+E_2)^2 dV =  \frac{\epsilon_0}{2} \int E_1^2+E_2^2 \; dV + \epsilon_0 \int E_1 \cdot E_2 \; dV$  
il primo addendo è l'energia di auto-interazione, lo cancello, tengo solo il secondo, cioè quella di interazione  

#### Punti di equilibrio

**Studiarli con piccole oscillazioni**

considero il punto dove x=0  
importante, perchè poi espando per x=0, funziona solo in questo punto  

1- scrivo U=qV  
2- espando per piccole oscillazioni (come nell'eseercizio della legge del moto)   
3- ottengo tipo: $U(z) \sim U(z) +\frac{1}{2}kx^2$  

stabilità:  
Il termine costante è solo un riferimento (energia al centro)  
Il termine quadratico governa il comportamento vicino a x=0  
k>0: il centro è un minimo locale del dell'energia potenziale, equilibrio stabile  
k<0: il centro è un massimo locale dell'energia potenziale, equilibrio instabile  

**Studiarli con metodo di Gauss**

testo:  
ho un anello di raggio R, l'asse chiamata z    
il centro è un punto di equilibrio stabile (se ci mettessi una carica -q)?  


immagino un cilindretto infinitesimale (lungo dz), di raggio r  
lo metto al centro dell'anello, con la sua sse coincidente con l'asse dell'anello  $Q_{in}= \Phi = 0$  

cosa voglio:  
voglio trovare una relazione tra $E_z$ e la componente radiale $E_r$   
cioè cosa fa il campo radiale se la carica q si sposta un po' lungo l'asse, rimane nullo così la carica torna al centro o appare e la tira via?     
così posso capire se il centro è un pozzo di potenziale  

calcolare:  
1- calcolo il flusso del cilindretto (separo le base e la superficie laterale)  
2- so che deve essere 0, li eguaglio  
3- faccio il limite dz->0 (o pensalo come dividere per dz), così a sinistra ottengo la derivata     
4- raccolgo per esprimere il radiale in funzione dell'assiale    
$\Phi = \pi r^2(E_z(z+dz)-E_z(z))+2\pi r \; dz E_r(r)=0$   
$r^2 \frac{dE_z}{dz} + 2r \; \; E(r)=0$   
$E(r)=-\frac{r}{2}\frac{dE_z}{dz}$  
ora se voglio posso sostituire il valore della derivata di E_z, se ho calcolato E_z prima  

Ciò significa che l’equilibrio non è stabile: un flusso entrante deve sempre essere compensato da un flusso uscente in altra direzione  
Vuol dire che se il campo lungo l’asse converge, allora da qualche parte deve esserci un campo divergente, per rispettare Gauss   
Non puoi avere un “pozzo di potenziale” tridimensionale al centro dell’anello   
In altre parole: anche se lungo l’asse il centro sembra un minimo, se permetti alla particella di muoversi anche nel piano, c’è sempre una direzione di fuga  



#### Fuga

**Allontanarsi all'infinito**

conservazione energia  

inizio:   
$K=0$, tutto fermo  
$U$ somma quelle tra tutte le coppie  

fine: $E=0$  
$U=0$, perchè distanza è infinita  
$K=0$, perchè voglio lavoro minimo, quindi finisco ferma  

se invece voglio la velocità di fuga, all'inizio metto: $K=\frac{1}{2}mv_{esc}^2$

### Trovare potenziale con principio di sovrapposizione

**Cosa**

modo di risolvere un esercizio sfruttando la simmetria del sistema  

1- trova tutte le simmetrie del sistema  
2- somma tutti i casi (sovrapponi)  

**Esempio**

È dato un tetraedro regolare di lato L, le 4 facce sono tenute a potenziale V1 , V2 , V3 , V4   
Si determini il potenziale nel centro, detto $V_0$  

1- trova un simmetria del sistema  
osservo che se operiamo una rotazione che manda il tetraedro in se stesso ma scambia i potenziali sulla facce allora il potenziale al centro deve rimanere invariato  

2- somma tutti i casi  
quindi prendiamo ora tutte e 24 le possibili permutazioni dei potenziali sulle 4 facce (supponendo di averle numerate)  
Sovrapponiamo le 24 configurazioni risultanti e otteniamo che, per il principio di sovrapposizione lineare:  
il potenziale al centro sarà $24V_0$  
il potenziale su ogni faccia sarà $6V_1 + 6V_2 + 6V_3 + 6V_4$  
Abbiamo dunque un tetraedro che possiede lo stesso potenziale su tutte e 4 la facce  
Non essendo presente carica all’interno del tetraedro la soluzione sarà campo nullo (E=0) e potenziale costante (V=const) su tutto il volume
da cui otteniamo che per il potenziale al centro vale (potenziale costante):     
$24V_0= 6V_1 + 6V_2 + 6V_3 + 6V_4$



**Caso speciale: tutte le facce allo stesso potenziale**

dentro non ci sono cariche, quindi $\nabla^2V=0$   
Quindi il potenziale deve essere una funzione armonica che si adatta ai bordi  

se $V_1 =V_2=V_3=V_4=V_f$, allora $V(x,y)=V_f$ è una soluzione  
Per unicità della soluzione del problema di Dirichlet, quella è l’unica soluzione  

problema di dirichlet:  
se in un dominio Ω il potenziale V soddisfa l’equazione di Laplace $\nabla^2V=0$   
e ai bordi  ∂Ω V  
allora Esiste al massimo una soluzione che soddisfi queste condizioni  
essenzialmente, se ho una eq differenziale e il giusto numero di condizioni iniziali, c'è un'unica soluzione (o zero, se le condizioni iniziali sono impossibii)

**Note: se l'interno non ha cariche**

eq di Maxwell:  
$\nabla \cdot E = \frac{\rho}{\epsilon_0}$  

se dentro non ci sono cariche, $\nabla \cdot E=0$  
che vuol dire che il campo è SOLENOIDALE  
campo nullo è una soluzione aprticolare di campo solenoidale  
perchè divergenza nulla non vuol dire campo nullo  
eg. un campo costante E=(1,0,0) ha divergenza zero ma non è nullo  

perchè la legge di Gauss dice che se q=0, il flusso netto del campo è nullo  
Il campo può benissimo essere diverso da zero, purché entri e esca in modo da compensarsi  

**Note: eq. per il potenziale se l'interno non ha cariche**

uso l'equaizione di Poisson:  
$\nabla^2 V = - \frac{\rho}{\epsilon_0}$  

se dentro non ci sono cariche, diventa l'eq di Laplace: $\nabla^2 V = 0$  
Quindi dentro il tetraedro il potenziale è una funzione armonica, non necessariamente costante  
Solo in casi simmetrici speciali (tutti i lati al medesimo potenziale) la soluzione che soddisfa Laplace e i vincoli al contorno è  V=costante

recap:  
$ρ=0  ⟹  ∇^2V=0$, ma non implica V=costante
Serve anche che tutte le condizioni al contorno siano uguali perché la soluzione costante sia valida  
In generale, invece, si ottiene una soluzione armonica non banale, e E≠0

**Nota: usare q=0 per trovare il campo**

spesso negli esercizi uso la legge di Gauss, scelgo una superficie, dico q=0 dentro, quindi E=0  
ma q=0 non basta ad implicare E=0  
c'è un'assunzione implicita: per simmetria il campo non può avere direzioni privilegiate  

Esempio classico: una sfera cava conduttrice caricata.  
Per simmetria sferica il campo dentro deve essere radiale e uguale in ogni direzione.  
L’unica possibilità compatibile con “nessuna carica interna” è E=0  

Se invece la geometria o le condizioni al contorno non sono perfettamente simmetriche, non puoi più concludere che  E=0 solo da Q=0
