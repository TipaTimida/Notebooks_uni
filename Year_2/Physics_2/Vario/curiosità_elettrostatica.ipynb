### Concetti base, macroscopici

cosa | formula | unità di misura
---|---|---
forza (di Coulomb) | $\vec{F}=k\frac{q_1q_2}{r^2} \hat{r}$ | N 
campo | $\vec{F}=\vec{E}q$ | V/m
potentiale | $dV = -E \cdot dr$ | J/C
energia | $U = Vq$  | J

**Nomi**

integrale di linea = circuitazione  
$\int _\gamma \vec{E} \cdot d\vec{l}$

#### Info tecniche

RICORDA DI METTERE IL VERSORE DOPO E ED F  

**Scrivere con vettori**

r al denominatore è la distanza tra q e Q  
quindi può anche essere qualcosa di più complicato, basta che poi ci metto la norma intorno  

eg: $|r-r_0|$  

per indicare la direzione, invece di mettere il versore, si può moltiplicare sopra e sotto per il vettore:  
$\vec{F} = k\frac{qQ}{|r-r_0|^3}|\vec{r}-\vec{r_0}|$  

**Carica di prova**

una carica talmente piccola che viene mossa, ma non muove le altre  
quindi non cambia il campo elettrico già esistente, ma lo segue  

**Note**

costante: $k = \frac{1}{4\pi \epsilon_0}$  
$\epsilon_0$: costante dielettrica del vuoto  

segno: +q respinge, -q attrae  

**Formule potenziale**

$dV = -\vec{E} \cdot d\vec{l}$  
$V(B)-V(A)=-\int_A^B E \cdot dl$  
$\hat{r} \cdot d\vec{l} = dl \cos \theta = dr$  

attenzione al segno:  
di solito voglio mettere finale - iniziale   
ma qui mi appare al contrario, quindi mi serve un ulteriore -  
$\int_A^B E \cdot dl = V(A)-V(B) = -(V(B)-V(A))$  

$\vec{E} = -\vec{\nabla} V$  
cioè le derivate parziali di V sono le componenti di E  

**Continuità**

E: campo vettoriale, non per forza continuo  
V: campo scalare continuo  

#### Spiegazioni

**Forza di Coulomb**

forza elettrostatica: forza prodotta da una carica ferma  
la particella che la riceve può essere in moto, ma quella che la produce no  

verso: dipende dalle cariche  
stesso segno respingono, diverso attraggono  

se ci sono 2 cariche:  
principio di azione e reazione, ognuna sente una forza a causa dell'altra, modulo uguale e verso opposto  
$F_{12}=F_{21}$  

se ci sono più di 2 cariche:  
ognuna sente una forza per ciascuna delle altre ariche, fare somma VETTORIALE delle forze per vedere net force  

**Campo e potenziale**

$E_x = -\frac{dV(x,y)}{dx}$  
$E_y = -\frac{dV(x,y)}{dy}$

$E =-grad(V) = -\nabla V$  
cioè le componenti di E sono le derivate parziali di V  

derivando il potenziale si ottiene il campo elettrico  
quindi se V è costante, E è nullo  

**Segno del potenziale**

$\vec{\nabla} V$, il gradiente del potenziale, è un campo vettoriale  
punta nella direzione in cui V aumenta, come una normale derivata  
poichè $E = -\vec{\nabla} V$, il campo elettrico invece punta nella direzione dove V diminuisce  
infatti E punta verso le cariche negative, dove c'è il potenziale minimo???  

pensa il potenziale come un'energia: naturalmente, i corpi si muovono verso la posizione a potenziale minore  
è il caso delle cariche positive, che quindi seguono le linee di E  
mentre le cariche negative fanno l'opposto, vanno in direzione opposta alle linee di E  

**Linee di campo**

dicono cosa succederebbe se una carica di prova positiva (non quelle già lì) fosse posta in quel punto, come si muoverebbe  

come disegnare linee di campo:  
le frecce $\vec{E}$ del campo elettrico vanno da punti a potenziale maggiore a punti a potenziale minore  
come il campo gravitazione va da punti ad altitudine maggiore verso punti ad altitudine minore  
quindi escono da cariche positive, entrano in cariche negative  
cioè campo creato da una carica puntiforme: uscente se q>0 (cioè frecce partono dalla carica e puntano in fuori), entrante se q<0   

proprietà:   
tangenti al campo in ogni punto  
numero di linee proporzionale a intensità del campo  
escono da cariche positive, entrano in negative

**Potenziale**

indica l'energia che devo usare per spostare unca carica di prova dall'infinito fino ad un punto distante $r$ dalla carica Q  
quindi all'infinito il potenziale è nullo, perchè per spostare la carica di prova dall'infinito all'infinito non serve energia, non sto spostando niente  

**Energia**

$U=k \frac{q_1q_2}{r}$ è energia dovuta all'interazione di questa coppia  
se ci sono più di 2 cariche devo sommare energia di tutte le coppie possibili

**Lavoro compiuto dalla forza di Coulomb**

è conservativa, quindi $W = \Delta U$  
dipende solo da punto iniziale e finale, non dal percorso   

perchè se volessi calcolarla:  
$W =\int_A^B F \cdot dl = \int_A^B k \frac{qQ}{r^2} \cos \theta dl = \int_A^B k \frac{qQ}{r^2} dr$  

cioè non dipende più da dl, ma solo dagli estremi  

dl: pezzo della traiettoria che compie davvero  
dr: pezzo radiale rispetto all'altra carica
