### Kirchoff e Ohm

**Leggi**

cosa | macroscopica | locale/microscopica
---|---|---
Ohm | $V=IR$ | $\vec{E}=\rho \, \vec{J}$  
Joule | $P=IV$ | $\frac{dP}{d\tau} = E \cdot J$  

per trovare la prima: plugga $R=\rho \frac{l}{S}$  
poi $V=ed$, $I=JS$  

per la seconda: $\tau$ è un pezzo di volume  
$dW = dq \Delta V$ dove $dq = ne \; d\tau$ e $\Delta V = E \cdot dl = E v_d dt$  


**Legge di Ohm generalizzata**

invece di includere tutti i cali di tensione in $\Delta V$, li esplicito come applicazione di un campo motore o elettromotore  
$V_A-V_B  + \Sigma f_m + \Sigma f_c = IR_{AB}$  

**Resistività vs resistenza**

resistenza: grandezza estensiva  
cioè proprietà di un oggetto in 3D  
dipende da resistività e forma geometrica  

resistività: proprietà di un pezzetto dx di materiale  
dipende da tipo di materiale e temperature  

resistenza: resitività + forma geometrica  
resistività: tipo di materiale + temperatura  

conduttività: $\sigma=\frac{1}{\rho}$  
conduttanza: $G=\frac{1}{R}$  

**Leggi dei circuiti**

Kirchoff: vale per correnti stazionarie o lentamente variabili  
Ohm: vale per regime stazionario, in solidi metallici (detti conduttori Ohmici) 

**Nota acqua minerale**

conducibilità è data dai minerali  
più residuo fisso -> più conduce  

### Generatore di f.e.m

**Definizione**

dispositivo capace di mantenere una d.d.p costante ai capi di un conduttore percorso da corrente (di un filo)  
così mantiene anche la corrente stazionaria  

la mantiene traformando in energia elettrica altre forme di energia  
eg. pila: chimica -> elettrica  

come lo fa:  
in pratica sposta le cariche dal catodo all'anodo del generatore, così possono ricominciare il giro  
lo fa grazie ad un campo elettrico dentro il generatore  
quindi l'effetto del generatore è assimilabile a quello di un campo elettrico non conservativo  
questo campo elettrico è mantenuto grazie all'energia che prende in altre forme (quindi è come se la trasformasse in energia elettrica)  

**Generatore vs condensatore**

generatore: dispositivo attivo, può fornire energia elettrica a un circuito  
condensatore: dispositivo passivo, non produce energia, ma può accumularla e restituirla per un breve tempo  

**Campi elettrici**

$E_s$: campo elettrostatico    
dove: ovunque, nel generatore e nel resto del circuito  
sorgente: le cariche accumulate sui morsetti (le lastre) del generatore, $\vec{E}= \rho \vec{J}$    
quindi è elettrostatico (creato da cariche statiche) -> conservativo  

$E_g$: campo elettromotore  
dove: solo dentro il generatore  
usato per spostare le cariche da catodo ad anodo (i morstti del generatore)  
non è conservativo  

$E_c$: campo elettromotore  
dove: solo dentro carichi non-ionici  

$E=E_s+E_g+E_c=\rho \vec{J}$  
quindi E, quello totale, non è conservativo  

dentro il generatore, $E_s$ e $E_g$ hanno verso opposto  
$E_s$ ha lo stesso verso che ha fuori, cioè lo stesso della corrente, visto che viene dalla formula con la densità di corrente  
$E_g$, dentro il generatore, ha verso opposto  

**Campo contro-elettro-motore**

schematizza la presenza di un carico non-ohmico, dicendo che è una regione di spazio al cui itnerno c'è un campo contro-elettromotore  

se non ce n'è, allora è un circuito in corrente continua (cioè I non dipende, dal tempo, NON I è uguale ovunque)  

**f.e.m.**

un generatore ideale si può schematizzare come: generatore ideale + resistenza interna in serie  
$f$: quella del generatore ideale  
$\Delta V$: d.d.p del generatore ideale, quindi che emette davvero dopo averne persa un po' sulla resistenza interna  

a circuito chiuso:  
$\Delta V = f-r_gI$  
c'è una dipendenza lineare tra $\Delta V$ e $I$  
in realtà vale solo fino ad un certo valore di I, poi $\Delta V$ comincia a scendere velocemente  

a circuito aperto:  
$\Delta V =f$  
perchè a circuito aperto non c'è corrente, la resistenza non causa calo di tensione  


**Nota**

mettere due generatori in parallelo diinuisce la loro resistenza interna  
se sono due generatori uguali, sono equivalenti ad un unico generatore con stessa $\Delta V$ di quello reale, ma resistenza interna dimezzata  

**Nomenclatura**

$\Delta V = \frac{\Delta U}{q} = -\int E \; dl$  

OK: differenza di potenziale, d.d.p, tensione    
NO: voltaggio, è una traduzione inappropriata di voltage  

**Definizioni**

tensione: fa riferimento al lavoro delle forze che muovono le cariche (il lavoro è l'integrale di una forza lungo un certo percorso)   

differenza di potenziale: differenza tra il valore di due punti su un campo scalare, il potenziale 

**Quando tensione = d.d.p ?**

se il campo è elettrostatico:  
tensione e differenza di potenziale coincidono   

se il campo NON è elettrostatico:  
la tensione tra due punti dipende dal percorso  
la differenza di potenziale tra due punti no   

perchè:  
potenziale è un campo scalare, ogni punto ha un valore, quindi differenza tra due punti non dipende dal perorso   
tensione è il lavoro di una forza, quindi se il lavoro non dipende dal percorso (cioè se il campo è conservativo) neanche la tensione dipende dal percorso  

**f.e.m. - forza elettromotrice**

è una tensione, NON una forza  
muove le cariche, in particolare gli elettroni  
indica la tensione applicata dal generatore (batteria)  

### Pila di Volta

**Cos'è**

è un generatore di forza motrice costante   
è una versopne vecchia, oggi si usano altre reazioni, e spesso le pile si fanno ricaricabili  
una pila ricaricabile è un accumulatore  
eg. accumulatore piombo/acido, in un'auto se ne usano 6 in serie, per avere $\Delta V = 12$  

capacità di carica: quanta carica c'è sui morsetti quando è completamente carica  
non si misura in coulomb, ma in Ampere x ora   

**Come funziona**

2 elettrodi: Cu, Zn  
soluzione di acido solforico $H_2SO_4$  

acido si ionizza per affinità elettronica  
cioè $H_2SO_4 +Zn \rightarrow ZnSO_4 +H_2$, perchè così abbassa l'energia libera    
quindi lo zinco ottiene elettroni, perchè neutralizza $SO_4^{-2}$, $Cu$ per de elettroni perchè neutralizza $H^+$  
dopo un po' lo zinco ha troppi elettroni=è troppo carico, la reazione si ferma  
la d.d.p tra gli elettrodi in qeusto momenti è la $\Delta V$ della pila  
quando la uso per produrre corrente, subito dopo ionizzo altro acido, così' mantengo costante questa $\Delta V$  
quindi ho un generatore di forza motrice costante  
ovviamente funziona finchè il sodio non è tuto ionizzato    

**Ricaricabile**

1- l'acido solforisco si esaurisce, la pila è scarica  
2- usando un altro generatorem faccio passare corrente nella pila in senso opposto, il che inverte la reazione di dissociazione facendo riformare l'acido  
3- ora la pila è di nuovo carica
