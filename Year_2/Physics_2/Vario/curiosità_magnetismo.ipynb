### Monopolo magnetico

**Monopolo magnetico**

NON esiste il monopolo magnetico  
il monopolo elettrico invece sì  

cioè può esistere una caria positiva o negativa da sola, ma non un polo magnetico positivo o negativo da solo  
se taglio a metà un magnete ottengo 2 magneti più piccoli, i poli si sdoppiano   

### Poli magnetici terra

**Polo nord magnetico**

immagina come se la Terra avesse una calamaita. un po' inclinata rispetto all'asse geografica  
ciò che chiamiamo il polo nord magnetico, che è vicino al polo nord geografico, in realtà è il polo SUD della calamita  
ma lo chiamiamo polo nord perchè attira il polo nord delle calamite che uso  

### Campo elettromagnetico

**Rapporto tra E e B**

nella legge di Lorentz $F=qvB$, v deve dipendere dal sistema di riferimento  
quindi se cambio sistema di riferimento, cambiano proprio E e B nello spazio  
si trasformano con le trasformazioni di Lorentz  
quindi forze elettriche e magnetiche possono trasformarsi l'una nell'altra  
per questo si chiama campo elettromagnetico  

### Fili in campo magnetico

**Filo (wire)**

$dF = i \, dl \times B$  

cioè F dipende solo dalla parte di lunghezza del filo PERPEDICOLARE alla direzione di B  
è semplicemente la distanza tra punto finale e iniziale, non importa che tragitto fa il filo per arrivarci

**Due fili paralleli**

$F_1=i_1lB_2 = i_1l\frac{i_2 \mu_0}{2 \pi d}$  

direzione:  
se $i_1$ e $i_2$ hanno stesso verso, i fili si attraggono, sennò si respingono  
basta usare la regola della mano destra per ricavare B (curl hands), poi con F=ilB per trovare F  
vedi se F avvicina o allontana i fili  

**Forza elettrica e magnetica in fili paralleli**

fili hanno carica lineare $\lambda$, distanza d  
nota che se il filo ha corrente stazionaria NON C'È campo E  
ma se il filo ha $\lambda$ ed è in movimento a velocità v, ottengo $I=\lambda v$ e c'è E  
voglio forza per unità di lunghezza (perchè se fili sono infiniti, quella totale è infinita)  

1- trova i campi  
campo elettrico generato da un filo: $E_r=2k\frac{\lambda}{r}$  
campo magnetico generato da un filo: $B=\frac{\mu_0 I}{2 \pi r}$  

2- trova le forze  
elettrica:  $\frac{F_E}{l} = \frac{q}{l}E = \lambda E$   
magnetica: $\frac{F_m}{l}= \frac{IB }{l}=IB=\lambda vB$    

3- totale  
$\frac{F}{l} = \lambda (E-vB)=\frac{\lambda^2}{2 \pi r \epsilon_0} (1-\frac{v^2}{c^2})$  
quale è più fprte dipende da v  
la parte magnetica diventa rilevante solo quanfo $v \sim c$, cioè B è la correzione relativistica di E  

**Due fili perpendicolari**

$F = \int_{R_1}^{R_2}i_2 B(r) \, dr =  \frac{\mu_0 i_1i_2}{2 \pi}\ln (\frac{R_2}{R_1})$  

filo 2: sul piano  
filo 1: perpendicolare al piano  

il filo 2 è disposto in modo da essere radiale rispetto al filo 1, nel punto più vicino è a distanza $R_1$; nel più lontano a distanza $R_2$  
anzi non importa che sia radiale, basta che sia sul piano perpendicolare, poi quando integro mi interessano solo le distanze radiali degli estremi tanto  

forza che sente il filo 1: immagino il filo 2 fatto fa tanti piccoli dl, ognuno ad un distana dr dal punto dove il filo 1 interseca il piano  
ogni piccolo dl fa sentire un po' di forza, integro su tutto l, la lunghezza del filo 2

### Quale bacchetta è magnetizzata?

**Domanda**

ho due bacchette:  
una con magnetizzazzione permanente M assiale (cioè un dipolo, una calamita)  
l'altra ferromagnetica (cioè può avere M indotta perchè reagisce ai campoi B, ma non ha M permanente)  
come faccio a sapere quale è la calamita?  

**Risposta**

mettile a forma di T  
se la linea verticale è la calamita: si attaccano  
se la linea verticale è la ferromagnetica: non accade niente  

invece metterle a forma di I non funziona, si attaccano sempre  

**Spiegazione**

calamita: è un dipolo  
produce campo B forte alle estremità/poli, debole al centro   

ferromagnetica: quando immersa in B della calamita, si magnetizza con $M=\chi_m B$  

se la verticale è la calamita:  
il ferro sente il campo forte e si magnetizza  
Il campo del magnete induce nel ferro una magnetizzazione allineata al campo locale, perchè $M=\chi_m H$, e in ferromagnetici $\chi_m >0$    
i poli nel ferro non sono alle estremità della bacchetta, sui lati corti, ma sui lati lunghi, così il lato dove tocca la calamita ha polo oppostoa quello della calamita  

la verticale è il ferro: sente campo quasi nullo, non si magnetizza  

**Note**

se il ferro viene magnetizzato, c'è forza: $F=-\nabla U = -\nabla (\mu \cdot B)$  
F nasce perchè il campo NON è uniforme, quindi il ferro è mosso verso la regione dove B è maggiore, per diminuire l'energia
