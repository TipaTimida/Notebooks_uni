### Energia

**Nota: cosa calcolare**

se ho una conformazione di carica  
e chiede di calcolare l'energia NON della conformazione, MA di una carica posta in un punto, di cui non mi dice il suo valore q  
allora vuole $V$, non $U$, perchè è come se quella fosse una carica di prova  

**Notazione**

$d\tau=dV$ è un pezzo infinitesimo di volume, vuol dire integra sul volume  
$\varphi$ è il potenziale  



**Metodi**

1. integrale sul campo elettrico  
$U = \int u_E \; dV$  
$u_ E = \frac{1}{2}\epsilon E^2 $     
metti $\epsilon$ giusto, se sono nel vuoto è $\epsilon_0$    

2. integrale sulla carica  
$U = \frac{1}{2} \int \varphi \, dq = \frac{1}{2} \int  \varphi  \rho\; dV = \frac{1}{2} \int  \sigma  \rho\; dS   $   
$\rho$: densità di carica LIBERA (anche se è un dielettrico)  
$\varphi$: potenziale (generato da cariche libere+polarizzazione) (di polarizzazione appaiono solo in un dielettrico)  
(per ricordarmela, di solito scrivo U=Vq)  
a seconda se ho un dielettrico o un conduttore, e quindi la carica è volumica o superficiale, devo scegliere in cosa trasformare dq  

3. principio dei lavori virtuali   
lavoro necessario per ottenere la configurazione attuale, portando lì tutti i pezzi dall'infinito   
$U=\int_0^R dW$  

4. $U =\int_r^\infty F \; dr$  
al posto di F, metti quella calcolata dopo il metodo delle cariche immagino, quindi tra tutte le coppie



**Quale usare**

1 e 2 sono modi diversi per interpretare la stessa cosa  
1 associa l'energia al campo, quindi è anche fuori dalla distribuzione di carica   
2 associa l'energia alla carica, quindi è solo dentro  

1 è meglio, non devo pensare alle cariche libere e di polarizzazione  

**Tips: metodi 1 e 2**

entrambi in teoria sono integrati sutto il volume  
ma nel 2 appare la densità di carica, che vale 0 dove non c'è carica, quindi nella pratica lì integro solo la parte dentro la distribuzione di carica  

quando integro, se ho una sfera posso dire $dV = 4 \pi r^2 \; dr$  

prima di iniziare, scrivi E oppure $\varphi$ e $\rho$ a tratti, separa bene le parti dentro e fuori  
così nell'1, quando integro, separa in integrali diversi a seconda dell'equazione di E usata, tipo [0,R] [R, $\infty$]


**Risultato utile**

tutte le figure hanno $U \sim k \frac{Q^2}{R}$, perchè deve ridare dimensionalmente  
poi cambia il coefficiente  

in k metti la $\epsilon$ del dielettrico di quella zona  
conduttore: dentro E=0   
dielettrico: anche all'interno devo calcolare U, quindi potrei doverla spezzare in due, parte dentro e fuori, e mettere epsilon diverse per quelle parti  

il coefficiente è:  
sfera cava conduttrice: $\frac{1}{2}$  
sfera carica uniforme: $\frac{3}{5}$  

**Lavoro per spostare sfera dentro l'acqua**

all'inizio, metà in acqua metà nel vuoto: $U_0 = \frac{1}{2} (U_a + U_v)$  
alla fine, tutta nell'acqua: $U_1 = U_a$  
lavoro: $L=U_1-U_0$  

### Pressione

**Pressione elettrostatica sulla superficie di un conduttore**

1. $p=\frac{1}{2}\epsilon E^2=\frac{1}{2}\sigma E = \frac{\sigma^2}{2 \epsilon}$  
nota che $p=u_E$, la densità di energia, ma dopo non devo fare l'integrale    

2. $p= \frac{F}{S} = - \frac{1}{S(r)} (\frac{dU}{dr})$    

la pressione sulla superficie di un conduttore punta sempre verso fuori  
perchè le cariche sui bordi cercano di spingere il bordi per allontanarsi sempre di più 


**Spiegazioni**

metodo 1:  
E è il campo subito FUORI dalla superficie , cioè sulla superficie  
nota che qui non devo integrare, al contrario del calcolo dell'energia  
l'1/2 è per fare la media tra campo dentro e fuori, perchè tecnicamente sarebbe $P=\frac{1}{2}\epsilon_0(E_{out}^2-E_{in}^2)$   
solo che considero un conduttore, quindi dentro E=0  

metodo 2:  
U è calcolata come voglio  
è il metodo dei lavori virtuali  
S va scritta in funzione di r, la grandezza geometrica variabile del sistema, cioè la dimensione rispetto a cui si fa la variazione virtuale  
è la dimensione geometrica variabile da cui dipende il volume e quindi il lavoro delle forze di pressione

**Esempio metodo 2**

esempio: pressione su sfera conduttrice di raggio a  

$U = \int \varphi \, dq =  \int_0^Q k\frac{q}{a}dq = k\frac{Q^2}{2a}$  

$P=\frac{1}{4 \pi a^2} (-\frac{dU}{da})$

**Pressione magnetica**

pressione elettrica/magnetica: sentita da un conduttore sottposto ad un campo elettrtico/magnetico  
ricorda che pressione è forza per unità di superficie  
ma spesso se chiede la forza, in realtà intende la pressione   
ma se c'è K è forza per unità di lunghezza, perchè K (o $J_s$) ha $dl$ nella formula???

pressione:  
$f_s  = \frac{B^2}{2 \mu}=\frac{1}{2}K B$  
anche qui, come in quella elettrica, 1/2 B è per fare la media tra quello dentro e fuori, quindi B sarebbe la somma tra quello dentro e quello fuori  
quella elettrica era: $f_E =\frac{1}{2}\epsilon E^2=\frac{1}{2}\sigma E = \frac{\sigma^2}{2 \epsilon}$   

### Forza

**Modi per calcolare la forza**

1. $F=qE$  
quando: ho una distribuzione di carica, di cui so E, e una carica puntiforme q    
q: la carica di un oggetto  
E: il campo emanato dall'altro oggetto/distribuzione di carica, SENZA considerare q (NON la sovrapposizione dei campi dei due oggetti)  
se voglio la forza sentita da un conduttore, diventa $F=\frac{1}{2}qE$ con $E=\frac{\sigma}{\epsilon_0}$, perchè è un punto di discontinuità del campo elettrico   
quindi metto 1/2, e poi in teoria sommo campo fuori e dentro (che è nullo)  
per un condensatore, calcolo la forza sentita da ogni piastra e le sommo???

2. $F = p \, S = \int p \, dS$    
se p è diversa in punti diversi, posso separare l'integrale su pezzi diversi della superficie  

3. $F=-\frac{dU}{dr}$  
quando: movimento lungo una direzione r, voglio la forza IN DIREZIONE di r  
r: la grandezza geometrica variabile del sistema, quella su cui c'è il cambiamento (tipo in un condensatore è h, la distanza tra le piastre)     
quando scrivo la formula di U, attenta ad esplicitare tutte le r da cui dipende, potrebbero essere all'interno di E  
se l’energia non dipende dallo spostamento r ne deduciamo che la forza è nulla    

4. $F = \Sigma k\frac{q_1 q_2}{r^2}$  
utile dopo l'esercizio di trovare le cariche immagine, considera tutte le coppie tra q esterna e una carica immagine  
stesso principio per cui forza sentita da una sfera è forza sentita dalla carica puntiforme nel suo centro  

5. $F=-\frac{Q^2}{2} \frac{dC}{dr} = \frac{V^2}{2}\frac{dC}{dr}$  
quando: condensatore  
espando -dU/dr con la formula di U del condensatore  
controlla se V o Q sono fissi nel mio problema, quindi se posso estrarli dalla derivata  
F è diretta verso l'aumentare della capacità, così l'energia si riduce  
attenta al segno, guarda le formule di dU negli appunti di condensatore    

**Se voglio F in una direzione precisa** 

1. se in quella direzione c'è movimento, usa il metodo dU/dr  
attenta che r sia lungo quella direzione  

2. se p su tutta la superficie si annulla per simmetria, e rimane non nulla solo in quella direzio,e usa $p=\int p \, dS$  
se c'è anche in altre direzioni posso considerare solo la sua componente che voglio: $F_z = \int p_z \, dS = \int p \cos \theta dS$   
ma se le altre componenti si annullano per simmetria questo passaggoi non serve, complica solo  

**Calcola forza**

ho una sfera conduttrice con Q, all'esterno c'è dielettrico solo nella metà superiore  
ho calcolato $E=\frac{Q}{2 \pi (\epsilon+\epsilon_0)r^2}$ col teorema di Gauss    
E è uguale su tutta la superficie della sfera (perchè parallelo alla separazione tra mezzi) 

pressione: $p = \frac{1}{2}\epsilon E^2$   
è diversa nella metà superiore e inferiore, perchè c'è $\epsilon$ o $\epsilon_0$    

forza:  
$F=\int_S p \, dS = \int_{S.up} p \, dS + \int_{S.down} p \, dS = \frac{1}{2}\epsilon E^2 (2 \pi R^2) + \frac{1}{2}\epsilon_0 E^2 (2 \pi R^2) =  \frac{1}{2}E^2 (2 \pi R^2) (\epsilon-\epsilon_0)$  


**Momento delle forze**

simile alla formula 3 delle forze:  
$\tau = -\frac{dU}{d\theta}$  
$\theta$ è l'angolo tra E e la superficie del dielettrico  
nota che quando calcolo l'energia di un dielettrico, dovrei calcolar separatemente U dentro e fuori il dielettrico (perchè voglio U di tutto lo spazio)  
U fuori dal dielettrico è teoricamente infinita, perchè lo spazio è infinito, ma è indipendente dall'orientazione del dielettrico, quindi da $\theta$, quindi posso evitare di calcolarla, tanto mi darà 0 quando derivo

### Problema in 1D

posso usare il teorema di Gauss in forma differenziale  

forma integrale:  
$\int_S E \; dS = \frac{1}{\epsilon_0} \int_V \rho dV$  

forma differenziale (in 1D)  
$\nabla \cdot E = \frac{\partial E_x}{\partial x}= \frac{\rho(x)}{\epsilon_0}$

cioè, nelle regioni in cui:  
$\rho = 0$: E non cambia  
$\rho=const$: E varia linearmente  
ricorda di imporre la continuità tra E di tutti i pezzi, e trova un pezzo dove lo conosco, tipo E=0 a $x=-\infty$  

**Quando usare**

quando il campo va in una sola direzione, tipo solo radiale  
e so $\rho$ del materiale attraverso cui passa  

**Esempio**

calcola campo E radiale intorno alla Terra, nell'atmosfera, sapendo che $E=E_0$ alla superficie terrestre e $E=0$ sul bordo dell'atmosfera, ad altezza h  
l'atmosfera ha $\rho$ uniforme  

uso forma differenziale:  
$\partial_z E = \frac{\rho}{\epsilon_0}$  
quindi so che il campo ha derivata costante e deve valere, e so il suo valore in due punti  
è abbastanza per scriverlo:  
$E(z)=-E_0(1-\frac{z}{h})$

più formale:  
$E(z)=\frac{\rho}{\epsilon_0}z + c$  
metti le condizioni al contorno $ E(0)=E_0, E(h)=0$  
dalla seconda ottengo $\frac{\rho}{\epsilon_0}h+E_0=0$, che posso usare per sostituire $\frac{\rho}{\epsilon_0}$, così poi posso raccogliere per $E_0$  

### Problema in 1D

**Quando usare**

quando il campo va in una sola direzione, tipo solo radiale

**Come risolvere**

### Dipoli

**Calcolo momento di dipolo**

$p= \Sigma q_ir_i$   
scegli un'origine degli assi (in mezzo alle particelle?), poi somma il contributo di ogni particella  
un po' come il centro di massa  

tip calcolo:  
in 2D, $r_i$ può essere passato come un vettore colonna  
$p = q_1 L\begin{bmatrix} 1 \\ -1 \end{bmatrix} + q_2 L\begin{bmatrix} 2 \\ 3 \end{bmatrix}$  
spesso possoraccogliere per L e Q, se le q_i sono tutte qualche multiplo di Q  

**Nota**

se $\Sigma q_i =0$, cioè la somma delle cariche è nulla, dipolo non dipende dalla scelta dell’origine  
quindi posso scegliere come origine una delle cariche, così r_i=0 e un addendo sparisce  
come qquando, per calcolare il moemnto, scelgo l'origine per far sparire una tensione  

### Come di distribuiscono cariche

**Situazione**

ho un conduttore cubico con carica Q  
sopra di esso ho uno strato di dielettrico e sopra una lastra conduttrice, senza carica iniziale  
come si distribuisce Q nel cubo?  

risposta:  
all'inizio la carica Q si distribuisce sul bordo del cubo uniformemente (perchè conduttore ha carica solo sul bordo)  
la carica positiva sulla faccia sopra induce una carica negativa nella lastra sopra (carica indotta su un altro conduttore messo lì vicino)  
perchè lo strato isolante (il dielettrico) non fa passare le cariche, ma lascia passare il campo, al suo interno si crea una polarizzazione   
questa carica negativa attira altre cariche positive, si crea un circolo vizioso, tutta la carica del cubo finisce sulla faccia sopra  

cioè:  
Un conduttore libero nello spazio distribuisce la carica uniformemente, perché il campo deve essere nullo dentro e uguale in tutte le direzioni.  
Ma qui c’è una superficie vicina che “disturba” la simmetria: il soffitto conduttore.  
Lo strato isolante (il dielettrico) non fa passare le cariche, ma lascia passare il campo, al suo interno si crea una polarizzazione  

quindi:  
faccia sopra del cubo: cariche libere +Q  
faccia sotto del dielettrico: cariche legate negative (di polarizzazione)  
lastra conduttrice: cariche libere negative indotte

### Altri

#### Resistenza tra sfera conduttrice e infinito

$R=\frac{\rho}{4 \pi a}$  
$\rho$: resistività del dielettrico esterno alla sfera (il mezzo va dalla sfera all'infinito)  
a: raggio della sfera  

come la trovo:   
per simmetria sferica, la corrente è radiale   
$I=J(r) \ dS = \sigma E(r) \, 4 \pi r^2$  
$E(r)=\frac{I \rho}{4 \pi r^2}$  
trovo differenza di potenziale e resistenza:  
$V=\int_a^\infty E(r) \, dr =\frac{I \rho}{4 \pi a}$  
$R=\frac{V}{I}$  


resistenza di una sfera con fuori un dielettrico, tra a<r<b:  
$R=\frac{\rho}{4 \pi} (\frac{1}{a}-\frac{1}{b})$  

**Come risolvere**

per simmetria sferica, la corrente è radiale   
$I=J(r) \ dS = \sigma E(r) \, 4 \pi r^2$  
$E(r)=\frac{I \rho}{4 \pi r^2}$  

trovo differenza di potenziale e resistenza:  
$V=\int_a^\infty E(r) \, dr =\frac{I \rho}{4 \pi a}$  
$R=\frac{V}{I}$  

ogni volta che chiede R, pensa di fare V/I  
quindi inizia scrivendo I, poi calcolaci E, che uso per trovare V    


**Soluzione**

resistenza di una sfera con fuori un dielettrico, tra a<r<b:  
$R=\frac{\rho}{4 \pi} (\frac{1}{a}-\frac{1}{b})$  

$R=\frac{\rho}{4 \pi a}$  
$\rho$: resistività del dielettrico esterno alla sfera (il mezzo va dalla sfera all'infinito)  
a: raggio della sfera
