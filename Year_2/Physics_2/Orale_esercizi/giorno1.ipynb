**Metodo delle immagini**

formula | cosa
---|---
voglio V=0 sul piano, quindi su y=0 <br> quindi $E$ TANGENZIALE al piano deve essere nullo (normale invece può essere non-nullo) <br> la sol è mettere un'altro p, parallelo, a $y=-a$ (punta nella stessa direz di prima)| risolvo con metodo delle immagini


**Calcoli dipolo**

formula | cosa
---|---

formula | cosa
---|---

**Forze, momenti, potenza**

formula | cosa
---|---

$P_E = \tau_E \cdot \omega$ |  potenza sviluppata da E<br> di solito, P=Fv
