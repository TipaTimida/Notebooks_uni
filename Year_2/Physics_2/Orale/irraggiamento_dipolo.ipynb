### Irraggiamento di dipolo

**Risolvi**

proof | cosa
---|---
campi del dipolo in zona di radiazione: <br> $B = \frac{k}{c^3} \frac{C}r$ <br> $E = \frac{k}{c^2}  \frac{C\times \hat r}r$ <br> dove chiamo: <br>  $\vec C = [ \ddot p] \times \hat r$ <br> $C^2 = (\ddot p \times \hat r) (\ddot p \times \hat r)  = \|\ddot p \|^2 \sin^2 \theta$ | situazione: sono nella zona di radiazione di un dipolo <br> le parentesi quadre su p significano calcolato al tempo ritardato  
$r^2(E \times B) = (C \times \hat r ) \times C $ <br> $= C^2\hat r -(C \cdot \hat r)C $ <br> $ = C^2 \hat r$ <br>$S = \epsilon_0 c^2 (E \times B) =  \epsilon_0 \frac{k^2}{c^3} \frac{C^2}{r^2} \hat r = \epsilon_0 \frac{k^2}{c^3}\frac{\| \ddot p \|^2 \sin^2 \theta}{r^2} \hat r$| calcolo $S$, per la trasmissione di energia <br> uso l'identità usata per l'eq d'onda   
$\frac{dP}{d\Omega} = \frac{dP}{d\Sigma} \frac{d\Sigma}{d\Omega} = \|S\| r^2$ <br> $P = \int \|S\| \, r^2 d\Omega $ <br> $ = \epsilon_0 \frac{k^2}{c^3}\|\ddot p\|^2 \| \int_0^{2 \pi} d \phi \int_0^\pi \sin^2 \theta \sin \theta d \theta $ <br> $ =  \epsilon_0 \frac{k^2}{c^3}\|\ddot p\|^2 (2 \pi) ( \frac 43)$ <br> $ =  \frac 83 \frac{k^2}{c^3}\|\ddot p\|^2$    | calcolo la potenza <br> coord polari: <br> $d\Omega = \sin \theta \, d\theta \, d\phi$, angolo solido <br> $d\Sigma = r^2 d\Omega$, elemento di superficie
