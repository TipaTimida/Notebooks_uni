**Angolo di Brewster**

proof | cosa 
---|--- 
$E_r = \frac{n_2^2k_x-n_1^2 k_x''}{n_2^2 k_x + n_1^2 k_x''}E_i = 0$ <br> $n_2^2k_x = n_1k_x''$ <br> $n_2^2 \frac{\omega n_1}c \cos \omega_i = n_1^2 \frac{\omega n_2}c \cos \theta_t$ <br> $\sin (2\theta_i)= \sin (2 \theta_t)$ <br> $2 \theta_i = 2\theta_t - \pi$ <br> $\theta+\theta_t = \frac \pi 2$ | voglio vedere quando la componente P dell'onda riflessa si può annullare <br> quindi metto $E_r=0$ nelle formule di Frenel per polarizzazione P

**Angolo di Brewster: spiegazione**

1- onda trasmessa fa oscillare gli elettroni del materiale     
2- elettroni sono piccoli dipoli oscillanti  
3- irraggiano, ma NON nella direzione in cui oscillano  
(non lungo $\vec p$, cioè non lungo $E$ -nota che p è perpendicolare a k)    
4- quindi se k dell'onda riflessa è parallelo a $p$ dell'onda trasmessa, l'onda riflessa non ha componente P  

4- caso polarizz P  
posso avere $\theta_r + \theta_t = 90^°$ (cioè c'è 90 gradi tra di loro)    

5- caso polarizz S:  
il campo elettrico, quindi $\vec p$, è perpendicolare al piano d’incidenza   
ma $k_r$ deve trovarsi sul piano di incidenza (lo crea)    
quindi non possono essere parallele  
nel disegno, l'onda trasmessa ha le lineette nere entranti nel foglio

**Tipi di polarizzazione**

1- polarizzazione P  
quando: E è parallelo al piano di incidenza  
nome: TM (trans magnetic), perchè B è perpendicolar al piano di incidenza  

2- polarizzazione S  
quando: E è ortogonale al piano di incidenza  
nome: TE (trans electric), perchè E è perpendicolar al piano di incidenza   

polarizzazione P e S sono una base per costruire qualunque polarizzazione  
cioè per avere E diretta in qualunque direzione, non solo perfettamente parallela o ortogonale
