### Th di unicità

fissata la sorgente $\rho(x,y,z)$, l'eq di Poisson ammette solo 1 sol che soddisfi le condizioni al contorno

eq. poisson: $\nabla^2 V = -\frac \rho {\epsilon_0}$  

formula | cosa
---|---
assumo ci siano due sol, $f_1$ e $f_2$ <br> entrambi risolvono Poisson: $\nabla^2 f_1 = \nabla^2 f_2 = -\frac \rho{\epsilon_0}$ <br> stesse condizioni al contorno: $f_1=f_2$ sul bordo S | faccio per assurdo <br> il mio setup ha come contorno la superficie chiamata S
$f=f_1-f_2$ <br> $\nabla^2 f = \nabla^2f_1-\nabla^2 f_2 =0$ <br> $f=0$ sul bordo | faccio la differenza, e la studio: <br> calcolo il laplaciano <br> condizioni al bordo
$\nabla^2 f =0$ -> f è armonica <br> una funzione armonica non può avere massimi o minimi dentro il volume, ma solo sul bordo <br> f è zero su tutto il bordo, quindi f=0 ovunque | osservazioni 
$f_1=f_2$, c'è un'unica sol| conclusione

**Nota**

funzione armonica: $\nabla^2 f=0$    

non ha massimi al centro perchè:  
$\nabla^2 f = \partial_{xx}f + \partial_{yy}f + \partial_{zz} f$  
se f avesse massimo interno, allora in quel punto tutte le derivate seconde sarebbero negative  
quindi avrei $\nabla^2f<0$   

idea fisica:  
il valore di una funzione armonica in un punto è la media dei valori attorno  
Quindi non può essere più grande di tutti i punti vicini, perché una media non può superare tutti i valori da cui è calcolata  

### Th Kirchoff

ho volume V con bordo S  
U rispetta eq delle onde, e la conosco su ogni punto di S   
allora so il valore di U in ogni punto di V (cioè U è unica)  

permette di esprimere U in un punto interno P mediante i valori di U e della sua derivata normale sul bordo

proof | cosa
---|---
$\nabla^2 U + k^2U=0$ |considero due funzioni, G e U, che rispettano eq delle onde
$G(\nabla^2 U + k^2U)=0$ <br> $U(\nabla^2 G + k^2G)=0$| moltiplico un'eq per l'altra funzione 
$F=G(\nabla^2 U ) -U(\nabla^2 G)=0$ <br> vedo che: <br> $F=\nabla \cdot (G \nabla U -U \nabla G)$|sottraggo le eq sopra, e definisco F
$\int_V \, F \, d\tau = \int_S (G \nabla U-U \nabla G) \cdot dS$  <br>  $ 0= \int_S (G \nabla U-U \nabla G) \cdot dS$  <br> perchè F=0 su tutto V |th della divergenza su F
$ \int_S (U \frac{\partial G}{\partial n}- G \frac{\partial U}{\partial n}) \cdot dS=0$ | perchè $\nabla U \cdot \hat n = \frac{\partial U}{\partial n}$, voglio la componente lungo n delle derivate <br> e $d\vec S = \hat n \cdot dS$  
scelgo G onda sferica centrata in P <br> $G= \frac{e^{iks}}s$ <br> $s = \|r-r_p\|$ | scelgo G
per s=0 è singolare, perchè diverge <br> allora applico il th della divergenza, invece che su tutto V, su V meno un pallina centrata in P <br> quindi il volume V' è delimitato da S (bordo esterno) e S' (bordo interno, di piccola pallina che contiene P) <br>  $0 = \int_S (U \nabla G-G \nabla U) \cdot dS +  \int_{S'} (U \nabla G-G \nabla U) \cdot dS $ | problema in P
$U\nabla G-G \nabla U $ <br> $=  U \frac{\partial}{\partial n'} (\frac{e^{iks}}s) -  (\frac{e^{iks}}s)  \frac{\partial U}{\partial n'}$ <br> $= U (\frac{e^{iks}}s) (ik-\frac 1s)-  (\frac{e^{iks}}s)  \frac{\partial U}{\partial n'}$  | riscrivo il secondo addendo mettendo la formula di G <br> n' è la normale alla sferetta che contiene P, cioè a S'
$\int_S ... \, dS =- \int_{S'} [U (\frac{e^{iks}}s) (ik-\frac 1s)-  (\frac{e^{iks}}s)  \frac{\partial U}{\partial n'}] \, dS'$ | rimetto nell'integrale
in coord polari, $dS' = \epsilon^2 d\Omega$ <br> con raggio della sferetta e angolo solido <br> faccio tendere la sferetta al punto P, cioè il raggio a 0, cioè $\epsilon \rightarrow 0$ <br> il lato sinistro non dipende da $\epsilon$, quindi non cambia 
$\int_{\Omega} [U (\frac{e^{ik\epsilon}}\epsilon) (ik-\frac 1\epsilon)-  (\frac{e^{ik\epsilon}}\epsilon)  \frac{\partial U}{\partial n'}] \, \epsilon^2 \, d\Omega$ <br> $ = \int_\Omega (ikU\epsilon -U-\frac{\partial U}{\partial n'} \epsilon) d\Omega$ <br> $ = \int_\Omega U(P) \, d\Omega = U(P) \, 4 \pi$ | calcolo il lato sinistro (scrivo $\epsilon$ invece di s, ogni volta che appare) <br> poi uso Taylor, perchè $\epsilon \rightarrow 0 $
posso scrivere: <br> $ U(P) = \frac 1 {4 \pi} \int_S [U \frac{\partial}{\partial n'} (\frac{e^{iks}}s) -  (\frac{e^{iks}}s)  \frac{\partial U}{\partial n'}] \, dS$ | conclusione
