### Modellizza atomo colpito da onda EM

**Situazione**

onda EM incide su atomo di idrogeno  
lo disegno come un elettrone legato elasticamente al nucleo  
ignoro irraggiamento del protone, perchè p molto più pesante  

nota: se campo elettrico è lungo x (polarizzzione lungo x), allora onda si propaga lungo z  

**Modello generico**

proof | cosa
---|---
$ma = F$ <br> $m(\ddot x +\gamma \dot x + \omega_0^2 x) = q(E+ \dot x \times B)$ | $\gamma$: damping <br> $\omega_0$: forza elastica di richiamo <br> $qE$: forza dell'onda EM sull'elettrone  
$m(\ddot x +\gamma \dot x + \omega_0^2 x) = qE_0 \hat e^{-i \omega t}$ | approssimazioni: <br> 1- cancella $\dot x B$ per condizione di campo debole <br> 2- ipotizzo onda $E=E_0 e^{ikz-i \omega t}$, che studio nel punto z=0  


cosa | dielettrico | conduttore
---|---|---
condizione: eq diff | completa | $\omega_0 = 0$, perchè elettroni non legati  
sol da trovare | $x(t) =  \frac{q}m \frac 1{\omega_0^2-i \gamma \omega-\omega^2} \vec E$ | $v(t) =  \frac{q}m \frac 1{\gamma - i \omega} \vec E$ 
cosa calcolare  <br> $\omega_p^2 = N \frac{e^2}{m \epsilon_0}$, frequenza di plasma  <br> (la parte sinistra della sol)| $\vec P = -Ne \vec x =  \frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}\epsilon_0  E $ | $\vec J = -N e\vec v = \frac{\omega_p^2}{-i \omega+ \gamma} \epsilon_0 E$  
cosa definisco: funzione materiale | suscettività <br> $P= \epsilon_0 \chi(\omega) E$ <br> $\chi(\omega) = \frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega} $  | conducibilità complessa <br> $J =\sigma(\omega) E$ <br> $\sigma(\omega) = \epsilon_0 \frac{\omega_p^2}{\gamma-i \omega}$  


**Note**

$\chi(\omega) = N \alpha(\omega)$  
cioè $\chi$ è per il materiale (quindi contiene $\omega_p$), $\alpha$ per un singolo atomo  
$\chi = \frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega} $  
$\alpha = \frac{e^2}{m \epsilon_0} \frac 1{\omega_0^2-i \gamma \omega-\omega^2}$ 

$\epsilon_r = 1 + \chi$

$\omega$ rappresenta l'onda, $\omega_0$ gli elettroni -> quello più grande vince, trionfa sull'altro

| $\omega \approx \omega_0$ | risonanza           | assorbimento forte                   |
| $\omega \gg \omega_0$     | campo troppo veloce | gli elettroni non riescono a seguire |

#### Funzione dielettrica

**In dielettrico (funzione dielettrica)**

proof | cosa
---|---
$\vec P = (\epsilon- \epsilon_0) \vec E = (\epsilon_r-1)\epsilon_0 E $ <br> $\epsilon_r(\omega)=1+\frac P{\epsilon_0 E} = 1 + \frac 1 {\epsilon_0} \frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}$ <br> cioè $\epsilon_r(\omega) = 1 + N \alpha(\omega)$| funzione dielettrica <br> la ottengo della formula della polarizzazione 

### Cielo blu

**Diffusione**

proof | cosa
---|---
$p(t)= qx(t)$ <br> $\ddot p(t) = \frac {q^2E_0}m  \frac {\omega^2} {\omega_0^2-i \gamma \omega-\omega^2}$ <br> $\|\ddot p \|^2 =\frac {q^2E_0}m  \frac {\omega^2} {\omega_0^2-\omega^2}$| derivata del momento di dipolo <br> caso no risonanza $\omega \neq \omega_0$ e no damping $\gamma =0$
$<P_{tot}> = \frac k {3c^3} \| \ddot p \|^2  $ <br> $=\frac k {3c^3} \frac{q^4E_0^2}{m^2} \frac{\omega^2}{(\omega_0^2-\omega^2)^2}$ <br> $ = \frac k {3c^3} \frac{q^4E_0^2}{m^2} F(\omega)$|  formula della potenza irraggiata da un dipolo
$\sigma = \frac{<P_{tot}>} {\|S\|} = \frac 83 \pi r_e^2 F(\omega)$ | sezione d'urto <br> grandezza che dipende solo dalle caratteristiche della particella <br> dice come è diffusa ogni frequenza <br> $r_e = k\frac{q^2}{mc^2}$, raggio classico dell'elettrone ($mc^2 = k \frac {q^2}r$) <br> $I = \|S\| = \epsilon_0 c^2 \|E \times B \| = \epsilon_0 c \|E^2\| = \epsilon_0 c \frac 12 E_0^2$, Poynting mediato sul periodo
atomi dell'atmosfera diffondono luce del sole  <br> approssimo $\omega_0 \gg \omega$, cioè freq propria degli atomi $\gg$ freq luce solare   <br> quindi semplifico forma della sezione d'urto: $\sigma \sim \frac{\omega^4}{\omega_0^4}$  | approssimazione <br>

###  Equazioni d'onda - Dielettrico vs conduttore

**Dielettrico vs conduttore**

cosa | dielettrico | conduttore
---|---|---
condizione | non ha corrente, quindi $J_{ext}=0 \; \rho_{ext}=0$ | non ha polarizzazione, quindi $P=0$  
eq d'onda | $\nabla^2 E - \frac{1}{c^2} \frac{\partial^2 E}{\partial t^2} = \frac{1}{\epsilon c^2} \frac{\partial^2 P}{\partial t^2}$ |  $\nabla^2 E - \frac{1}{c^2} \frac{\partial^2 E}{\partial t^2} = \frac{1}{\epsilon c^2} \frac{\partial J}{\partial t}$  
formula ottenuta dal modello di diffusione | $\vec P = N \vec p  = \frac{\omega_p^2}{\omega_0^2-\omega^2-i \gamma \omega}E $  | $\vec J = -N ev = \frac{\omega_p^2}{(-i \omega +\gamma )} \epsilon_0 E$  
cosa definisco | polarizzabilità atomica <br> $\vec p = \epsilon_0 \alpha(\omega) E $ | conducibilità complessa <br> $J =\sigma(\omega) E$  
relazione di dispersione | $k^2c^2 = \omega^2 (1 + N\alpha(\omega))$ | $\omega^2 = k^2 c^2 + \frac{i \omega}{(i \omega - \gamma)} \omega_p^2$

**Dielettrico: onde traversali e longitudinali**

proof | cosa
---|---
$\nabla \cdot E = \frac{\rho}{\epsilon} = -\frac{\nabla \cdot P}{\epsilon_0}$ | scrivi eq di Maxwell  
$\nabla \cdot E = -N \alpha(\omega) (\nabla \cdot E)$ <br> $(1+N \alpha(\omega))(\nabla \cdot E)=0$  | in un dielettrico: $P = N\vec p = N \alpha(\omega) \epsilon_0 E$  
$\nabla \cdot E = 0$, onda trasversale | perchè, in onda piana: <br> $E=E_0 e^{i(k \cdot r-\omega t)}$ <br> $\nabla \cdot E = i\vec k \cdot \vec E = 0$ <br> cioà E non ha componente lungo k
$1+ N \alpha(\omega)=0$, onda longitudinale  

io studio solo onde traversali, anche se studio in un dielettrico  

**Ricava eq onde**

1- equazione di rot B  
$\nabla \times B = \mu_0 J_{ext} +\mu_0\epsilon_0 \frac{\partial E}{\partial t} + \mu_0 \frac{\partial P}{\partial t}$    
a seconda se sono in dielettrico o conduttore, metti qualcosa a 0  

2- fai di nuovo il rotore  

2- all'ultimo step, plugga quella giusta  
$-\nabla ^2 E = -\frac{\partial}{\partial t} (\nabla \times B)$   

3- nota che, per passare tra le due eq. d'onda, basta fare $\frac{\partial P}{\partial t} \rightarrow J$   

4- nota: in emtrambi i casi, $\nabla \cdot E=0$  
nel conduttore perchè $\rho_{ext}=0$  
nel dielettrico perchè considero onde trasversali (spiegato sotto)

**Rapporto tra E e B e proprietà**

proof | cosa
---|---
$E=E_0e^{ikx-i \omega t} \hat z$ | immagino di avere questo tipo di sol per E
$\nabla \times E =-\partial_t B$ <br> $-ik E \; \hat y = -i \omega \, B$ | la pluggo nell'eq del rotore di E <br> anche B deve avere stessa forma di E, ma devo trovare ampiezza e direzione  
$B_0 =-\frac{k}\omega E_0$, cioè $\|B_0\|=\frac{\|E_0\|}{c}$ nel vuoto  <br> B è ortogonale a E e k, cioè $E \cdot B =0$ | conclusioni

#### Relazione di dispersione

proof | cosa
---|---
$\nabla^2 E - \frac{1}{c^2} \frac{\partial^2 E}{\partial t^2} = \frac{1}{\epsilon c^2} \frac{\partial^2 P}{\partial t^2}$  | ho eq d'onda <br> dove $\vec P = N \alpha(\omega) \epsilon_0 \vec E$ perchè è un mezzo lineare  
$(ik)^2 -\frac 1{c^2} (-i \omega)^2 = \frac{1}{\epsilon c^2} (-i \omega^2) N \alpha(\omega) \epsilon_0 $ <br> $k^2c^2 = \omega^2 (1 + N\alpha(\omega))$| cerco sol del tipo $ \vec E = E_0 e^{ikx-i \omega t}$  
$n = \frac c{v_f} = \frac{kc} \omega = \sqrt{1 + N \alpha(\omega)} = \sqrt{\epsilon_r(\omega)}$ | introduco indice di rifrazione
$\alpha$ è complesso, quindi anche n, quindi anche k <br> posso scrivere: $k= k_r + i k_i$ <br> $E = E_0 e^{-k_ix} \; e^{ik_r x-i \omega t}$| osservazione: la parte complessa di k è l'attenuazione dell'ampiezza
