### Legge di Ampere da Biot-Savart

**Coords**

r: coord di P, punto dove voglio sapere il campo   
r': coordinata del pezzetto $\tau'$ di sorgente che sto considerando adesso  
$\Delta r = |r-r'|$, distanza tra pezzeto della sorgente che sto considerando e punto P dove voglio il campo  

$\nabla$ e $\nabla'$ agiscono rispettivamente sulle coord x e x'  

**Trova formula di potenziale vettore**

formula | cosa
---|---
$B(r) = \frac{\mu_0}{4 \pi} \int_{\tau'} J(r) \times \frac{\Delta r}{\|\Delta r\|^3} d\tau'$  | Biot-Savart 
$B(r)= \frac{\mu_0}{4 \pi} \int_{\tau'} J(r) \times (-\nabla(\frac 1{\Delta r})) d\tau'$ | trova il gradiente 
$\vec \nabla \times (f \vec v) = (\nabla f) \times v+f(\nabla \times v)$ <br> $J(r) \times \nabla(\frac 1{\Delta r}) $<br>$ = - [\nabla(\frac 1{\Delta r}) \times J(r)]  $<br>$= -[\nabla \times (\frac {J(r)}{\Delta r}) - \frac 1 {\Delta r}(\nabla \times J(r))] $<br>$ -[\nabla \times (\frac {J(r)}{\Delta r})] - 0$ | manipola il gradiente <br> perchè J è funzione della sorgente, quindi di x',y',z', non di x,y,z
$B(r)= \frac{\mu_0}{4 \pi} \int_{\tau'} \nabla \times (\frac{J(r')}{\Delta r}) d\tau'$ | quindi, sono arrivata a:
$B(r) = \nabla \times (\frac{\mu_0}{4 \pi} \int_{\tau'}(\frac{J(r')}{\Delta r}) d\tau')$ <br> $ = \vec \nabla \times \vec A$|inverti operazioni
