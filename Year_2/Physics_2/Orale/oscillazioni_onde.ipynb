### Corda vibrante

#### Deriva l'equazione

**Setup**

corda lunga L  
$u(x,t)$: spostamento in direzione y di un elemento dm della corda centrato in x  
dove $dm = \rho \, dl$  

cioè $u(x,t_0)$ è il profilo della corda all'istante $t_0$

**1. Ricava equazione**

1- trovo le forze  

$\begin{cases}
F_x = T(\cos \alpha_2 - \cos \alpha_1) \sim T(1-1) = \sim 0 \\
F_y = T(\sin \alpha_2-\sin \alpha) \sim T(\tan \alpha_2 - \tan \alpha_1) \sim T (\frac{\partial u}{\partial x}|_2 -\frac{\partial u}{\partial x}|_1 ) \sim T \frac{\partial^2 u}{\partial x^2}dl
\end{cases}$

2- scrivi eq. Newton  
$F=dm \; a$  
$T \frac{\partial^2 u}{\partial x^2}dl =\rho \, dl \; \frac{\partial^2 u}{\partial^2 t}$  
$\frac{\partial^2 u}{\partial^2 t} -v^2 \frac{\partial^2 u}{\partial x^2}=0$  



**1. Note: ricava equazione**

1- assumo $u(x,t) \ll L$, cioè la corda si deforma poco  
allora uso sugli $\alpha$ l'approssimazione di angoli piccoli: $\cos \alpha \sim 1, \; \sin \alpha \sim \alpha \sim \tan \alpha$  
graficamente, le due T hanno verso opposto

2- sottrarre il valore di f in due punti vicini vuol dire fare la derivata tra i due punti, e i punti 1 e 2 sono vicini  

3- pongo $\frac T \rho = v^2$, tanto dimensionalmente va bene  


**2. Risolvi**

1- cambio var:  
$\begin{cases}
a = x-vt \\
b=x+vt
\end{cases}$

2- calcolo le derivate (con chain rule)  
$\frac{\partial u}{\partial x} = \frac{\partial u}{\partial a}\frac{\partial a}{\partial x} + \frac{\partial u}{\partial b}\frac{\partial b}{\partial x} = \frac{\partial u}{\partial a} + \frac{\partial u}{\partial b}$  
$\frac{\partial u}{\partial t} = v(\frac{\partial u}{\partial a} - \frac{\partial u}{\partial b})$  
$\frac{\partial^2 u}{\partial x^2} = (\frac{\partial u}{\partial x})^2 =  (\frac{\partial u}{\partial a} + \frac{\partial u}{\partial b})^2 = \frac{\partial^2 u}{\partial a^2} + \frac{\partial^2 u}{\partial b^2} + 2\frac{\partial u}{\partial a} \frac{\partial u}{\partial b}$   
$\frac{\partial^2 u}{\partial t^2} = v^2 (\frac{\partial^2 u}{\partial a^2} + \frac{\partial^2 u}{\partial b^2} - 2\frac{\partial u}{\partial a} \frac{\partial u}{\partial b})$ 

2- mettile nell'equazione  
ottengo: $\frac{\partial^2}{\partial ab}=0$  
cioè non ci sono termini misti, ho: $u(a,b)=p(a)+q(b)$  
quindi: $u(x,t) = p(x+vt)+q(x-vt)$  
con p,q determinate dalle condizioni al bordo  

#### Battimenti

**Setup**

ho due onde di frequenza simile
metto ampiezza uguale, per semplicità  

**Risolvi**

1- eq delle onde (studio un punto preciso, tipo x=0)    
$u_1(0,t) = A \cos(\omega_1 t)$  
$u_2(0,t) = A \cos(\omega_2 t)$  

2- le sommo con le formule di Prostaferesi:  
$u(0,t) = u_1(0,t)+u_2(0,t) = 2A \cos (\frac{\omega_1+\omega_2}2 t) \cos \cos (\frac{\omega_1-\omega_2}2 t)$  

3- condizione che le frequenze sono simili:  
def: $\frac{|\omega_1-\omega_2|}{\omega_1} \ll 1$  
quindi considero la freq media: $\omega_1 = \omega + \delta \omega \quad \omega_2 = \omega - \delta \omega$  

4- conclusione  
$u(t) = \cos(\omega t)\cos(\delta \omega \, t)$  
il secondo cos è la modulazione dell'ampiezza
