#### Th di Poynting

**Forma generale eq di continuità**

$\frac{\partial}{\partial t} $(densità di energia)$ + \nabla \cdot $(vettore) = (densità di potenza)  

il vettore indica il flusso di energia per unità di tempo attraverso una superficie unitaria  

**1. Risolvo**

proof | cosa
---|---  
$\vec j = \epsilon_0 c^2 (\nabla \times B) -\epsilon_0 \frac{\partial E}{\partial t}$  | inizio: eq di Maxwell per rotore di B <br> isolo la J  
$\vec j \cdot E = \epsilon_0 c^2 E(\nabla \times B) -\frac {\epsilon_0} 2 \frac{\partial E^2}{\partial t}$ <br> nota che: $E \cdot \frac{\partial E}{\partial t} = \frac 12 \frac{\partial E^2}{\partial t}$  | moltiplico per E  
$\nabla \cdot (E \times B) = B \cdot (\nabla \times E)-E \cdot (\nabla \times B)$ | identità vettoriale 
$\nabla \times E = -\frac{\partial B}{\partial t}$ <br> $B \cdot (\nabla \times E) = - \frac 12 \frac{\partial B^2}{\partial t}$  | altra eq del rotore  
$\vec j \cdot E = \epsilon_0 c^2 [-\frac 12 \frac{\partial B^2}{\partial t} - \nabla (E \times B)] -\frac {\epsilon_0} 2 \frac{\partial E^2}{\partial t}$ <br> $-E \cdot j = \frac{\partial}{\partial t}(\frac 12 \epsilon_0 E^2+\frac{B^2}{2 \mu_0}) + \nabla \cdot (\frac 1{\mu_0} E \times B)$ |pluggo tutto e riordino <br>  dò i nomi $u$ ed $S$ alle cose (energia EM per unità di volume e flusso di energia EM) <br> nota che motiplico tutto per -1, così mi riappare la potenza che ho calcolato prima

**Osservazioni su S**

1- S e u non sono unici  
posso sommare una divergenza e una derivata temporale di una stessa quantità vettoriale X, e l'equazione è ancora valida   
$\begin{cases}
u'=u+\nabla \cdot \vec X \\ S' = S-\frac{\partial \vec X}{\partial t}\end{cases}$  
perchè $\frac{\partial}{\partial t}(\nabla \cdot X) = -\nabla \cdot (\frac{\partial X}{\partial t})$  

2- S non è unico  
il teorema mi definisce solo la divergenza di S  
quindi posso sommargli un altro campo a divergenza nulla (quindi uno che si scrive come rotore di un altro)  
posso aggiungere il rotore di un altro campo, e la divergenza non varia  
$S' = S+\nabla \times A \quad \rightarrow \quad \nabla \cdot S = \nabla \cdot S'$   
perchè: $\nabla \cdot (\nabla \times A) = 0$  

**Direzione di S**

1- in un filo:  
E è lungo il filo  
B è azimutale  
quindi S è radiale, punta verso il filo  
strano, vuol dire che l'energia entra nel filo dallo spazio circostante, non “viaggia dentro il filo"  
Comunque, vista l’arbitrarietà del vettore di Poynting, possiamo modificarlo in modo che l’energia scorra lungo il filo come J, il che è molto più intuitivo  

#### Eq. d'onda nel vuoto

**Situazione**

nel vuoto: $J=0, \rho=0$  
P=0: no polarizzazione, no materiale

proof | cosa
---|---  
$\nabla \times E =-\frac{\partial}{\partial t} B$  | inizio: eq di Maxwell per rotore di E
$\nabla \times (\nabla \times E) =-\frac{\partial}{\partial t} (\nabla \times B)$ | moltiplico per $\nabla \times$  
$ -\nabla^2 E + \nabla (\nabla \cdot E)=-\frac{\partial}{\partial t} (\nabla \times B)$ | identità vettoriale <br> $\nabla \times (\nabla \times E) = -\nabla^2 E + \nabla (\nabla \cdot E)$ 
$-\nabla ^2 E = -\frac{\partial}{\partial t} (\mu_0 \epsilon_0 \frac{\partial}{\partial t} E)$ <br> $\nabla ^2 E = \mu_0 \epsilon_0 \frac{\partial^2 E}{\partial t^2}$| altre eq di Maxwell da usare <br> $\nabla \cdot E = 0$ perchè sono nel vuoto <br> $\nabla \times B = \mu_0 \epsilon_0 \frac{\partial E}{\partial t}$

inoltre $\mu_0 \epsilon_0 = \frac 1 {c^2}$  

**2. Giustifica l'identità**

proof | cosa

#### Conservazione carica

**Partendo da Maxwell**

proof | cosa
---|---  
$\nabla \times B =\mu_0 J + \mu_0 \epsilon_0 \frac{\partial E}{\partial t}$  | inizio: eq di Maxwell per rotore di B
$\nabla \cdot (\nabla \times B) =\mu_0 (\nabla \cdot J) + \mu_0 \epsilon_0 \frac{\partial }{\partial t}(\nabla \cdot E)$ | moltiplico per $\nabla \cdot$

**Senza Maxwell**


proof | cosa
---|---

### Conservazione impulso - scrivi T

**Forma generale eq di continuità**

$\frac{\partial}{\partial t} $(densità di impulso)$ + \nabla \cdot $(tensore) = (densità di forza)  

nella conserv energia avevo un vettore, perchè l'energia è scalare  
ma l'impulso è vettoriale quindi ho un tensore  

**1. Parte senza indici**

proof | cosa
---|---
$\rho E + j \times B $| scrivi densità di forza
$(\epsilon_0 \nabla \cdot E) E + [\epsilon_0 c^2 (\nabla \times B)-\epsilon_0 \frac{\partial E}{\partial t}] \times B $ <br> $ =\epsilon_0 E (\nabla \cdot E) + \epsilon_0 c^2 (\nabla \times B) \times B -\epsilon_0 \frac{\partial E}{\partial t} \times B $ | sostituisco in $\rho $ e $J$ i valori dalle eq di Maxwell  
$ =\epsilon_0 E (\nabla \cdot E) \quad + \epsilon_0 c^2 (\nabla \times B) \times B \quad  -\epsilon_0 \frac{\partial E}{\partial t} \times B  \quad -\epsilon_0 E \times \frac{\partial B}{\partial t} \quad + \epsilon_0 E \times \frac{\partial B}{\partial t}$ <br> $ =\epsilon_0 E (\nabla \cdot E) \quad + \epsilon_0 c^2 (\nabla \times B) \times B \quad -\epsilon_0 \frac{\partial}{\partial t}(E \times B) \quad \quad + \epsilon_0 (\nabla \times E) \times E $  | sommo e sottrago $\epsilon_0 E \times \frac{\partial B}{\partial t}$ <br> così unisco addendi 3 e 4 <br> e sull'addendo 5 l'eq di Maxwell $E \times \partial_t B = E \times (- \nabla \times E) = ( \nabla \times E) \times E$
$ = \epsilon_0 \frac{\partial}{\partial t} (E \times B) + \quad [\epsilon_0 E (\nabla \cdot E) + \epsilon_0c^2 (\nabla \times B) \times B + \epsilon_0 (\nabla \times E ) \times E] $ | riordina 

**2. Lavora con indici**

proof | cosa
---|---
$[(\nabla \times E) \times E]_i = \epsilon_{ijk} (\epsilon_{jlm} \partial_l E_m) E_k $| scrivi con indici
$= \epsilon_{jki} \epsilon_{jlm} (\partial_l E_m) E_k $| metti vicini i due $\epsilon$, e falli iniziare con stessa lettera
$= (\delta_{kl}\delta_{im} - \delta_{km} \delta_{il}) (\partial_lE_m) E_k$ | usa proprietà di cambiare due $\epsilon$ in $\delta$ 
$ = (\partial_k E_i)E_k -(\partial_I E_k) E_k$ | usa i $\delta$ per uguagliare la seconda lettera alla prima che appare
$= [\partial_k (E_iE_k)-(\partial_k E_k)E_i] -\frac 12 \partial_i E^2_k$ | usa proprietà della derivata di un prodotto (devo ricordarlo e basta)

**Ricorda**

$\nabla \cdot E = \partial_i E_i$, deve avere indice ripetuto

$\partial$: rappresenta $\nabla$

usare nabla:  
$\partial_i E_i = \nabla \cdot E$  
$\partial_i E_j = \frac{dE_j}{dx_i} = (\frac 1c\partial_t, \nabla) E$    
ma dipende da che coord sto usando, tipo in queste dimostrazioni uso i che indica solo le coord spaziali, non temporali, quindi $\partial_i E_j = \nabla E$

proof | cosa

$\epsilon_0 E (\nabla \cdot E) + \epsilon_0c^2 (\nabla \times B) \times B + \epsilon_0 (\nabla \times E ) \times E$ | voglio esamianre la parentesi quadra con cui avevo finito
$= \epsilon_0 E_i (\partial_k E_k) + \epsilon c^2 [\partial_k(B_iB_k) -\frac 12 \partial_i B_k^2] + \epsilon_0 [\partial_k (E_iE_k) -(\partial_kE_k)E_i-\frac 12 \partial_i E_k^2]$ <br> $ = \epsilon_0 \partial_i [c^2 B_iB_j - \frac 12 c^2 B_jB_j + E_iE_j -\frac 12 E_j E_j]$ <br> ho ottenuto la derivata di qualcosa| riscrivi con indici <br> sono indici muti, li posso rinominare  <be> due addendi sono uguali, si cancellano  
$T_{ij} =\epsilon_0[\frac{E^2+c^2B^2}2 \delta_{ij} -E_iE_j-c^2B_iB_j]$| ho trovato la divergenza di un tensore, il tensore degli sforzi/dello stress di Maxwell

**4. Recap**

proof | cosa
---|---
$f_i = \epsilon_0 \frac{\partial}{\partial t} (E \times B) + \quad \partial _j T_{ij}$ <br> dove: $f_i \rho E_i + (j \times B)_i$| riordina
