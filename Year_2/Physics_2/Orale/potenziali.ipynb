#### Gauge

proof | cosa
---|---
$A'=A+\nabla \psi$ <br> | trasformazione di Gauge su A  
$E= -\nabla \varphi' -\frac{\partial A'}{\partial t}= -\nabla \varphi' -\frac{\partial A}{\partial t} - \frac {\partial}{\partial t} \nabla \psi = -\nabla \varphi -\frac{\partial A}{\partial t}$ <br> quindi: $\varphi' = \varphi -\frac{\partial \psi}{\partial t}$   | come cambiare $\varphi$ per non modificare E   

**Scelta di Gauge**

situazione:  
ho un campo A di cui so solo il valore del rotore  
non basta a definirlo in modo univoco perchè A e A' hanno stesso rotore:  
$A'=A+\nabla \psi$    

perchè:   
$\nabla \psi$ è il gradiente di un campo vettoriale   
quindi è campo conservativo, ha rotore nullo  

conseguenza:  
ho una 'libertà di Gauge', cioè A è definito a meno di un $\nabla \varphi$   
posso sceglierle un valore di $\psi$, significa fare una scelta di Gauge 

**Invarianza di Gauge**

qualunque A scelgo, B non cambia  
questa si chiama 'INVARIANZA DI GAUGE'  

cioè:   
facile: la libertà di cambiare come descrivi un campo fisico, senza cambiare la fisica  
difficile: una simmetria locale di una teoria fisica: significa che puoi cambiare certe quantità “interne” del sistema punto per punto nello spazio-tempo, senza modificare le osservazioni fisiche  

**Trasformazione di Gauge**

$\begin{cases}
A'=A+\nabla \chi \\
\varphi' = \varphi -\frac{\partial \chi}{\partial t}
\end{cases}$

$\chi$ è una funzione, non un campo  

**Che gauge usare**

per sceglierla, mi serve un'altra condizione su A, non mi basta sapere il suo rotore  
qeust'altra condizione mi fa scegliere quanto vale $\varphi$  
posso scegliere la condizione che voglio, quindi ne scelgo una per scrivere le equazioni di Maxwell in modo più facile  

intuitivamente: ho condizione sul rotore di A, ora voglio condizione sulla divergenza  
a seconda della metrica su cui lavoro, scelgo se A è 3-vector o 4-vector  

**Tipi di Gauge**

1- gauge di Coulomb  
quando: caso stazionario   
condizione: $\nabla \cdot A = 0$  
cioè: divergenza di A, con A 3-vettore  
vantaggi: $\varphi$ soddisfa l'equazione di Poisson (e quindi???)  

2- gauge di Lorenz (non c'è la t)  
quando: caso non stazionario  
condizione: $\partial_\mu A^\mu = 0$, cioè $\nabla \cdot A + \frac{1}{c^2} \frac{\partial \varphi}{\partial t}=0$  
cioè: divergenza di A, con A 4-vector  
vantaggi: mantiene invarianza relativistica e le equazioni di Maxwell si disaccoppiano in equazioni d'onda  


**Equazioni nelle nuove Gauge**

1- gauge di Coulomb  
per $\varphi$: come equazione di Poisson, mostra che il potenziale elettrico è istantaneo, non ritardato (che vuol dire???)  
$\nabla^2 \varphi = -\frac \rho {\epsilon_0}$  
per A: equazione delle onde non omogenea:  
$\nabla^2 A -\frac{1}{c^2}\frac{\partial^2A}{\partial t^2}=-\mu_0J +\frac 1{c^2} \nabla (\frac{\partial \varphi}{\partial t})$  

2- gauge di Lorenz  
si disaccoppiano in equazioni d'onda e sono invarianti  
somigliano a equazioni delle onde non omogenee, cioè forzate, con a destra le sorgenti

parte temporale:  
$\nabla^2 \varphi - \frac 1{c^2} \frac{\partial^2 \varphi}{\partial t^2}=-\frac{\rho}{\epsilon_0}$  
parte spaziale:  
$\nabla^2 A - \frac 1{c^2} \frac{\partial^2 A}{\partial t^2}=-\mu_0 J$ 

### Risolvere equazioni d'onda

**Come scrivere**

**Delta di Dirac**

cosa: funzione usata per descrivere cariche puntiformi  
definizione: limite di una gaussiana quando la larghezza va a 0  

proprietà: quando la integro, se $x_0$ appartiene al dominio dell'integrale fa 1, sennò 0  

$S(x,t)= \int S(x_0,t) \delta^3(x-x_0) \, d^3 x_0$  
$\int S(x_0)$: peso/valore della carica nel punto $x_0$  
$ \delta^3(x-x_0)$: considero solo carica su $x_0$  
$d^3x_0$: ogni punto dello spazio su cui integro svolge il ruolo di $x_0$  

**Steps**

#### Eq di Poisson

**1. Scrivo equazione**

proof | cosa
---|---
$\nabla^2 \varphi(x) = -\frac{\rho(x)}{\epsilon_0}$ | scrivo l'eq di Poisson  
$\nabla^2 \varphi_0(x) = -\frac{\rho(x_0) \, \delta^3(x-x_0)}{\epsilon_0} $ | voglio il caso con carica puntiforme, quindi scrivo la sorgente con la Delta di Dirac  
$\nabla^2 \varphi_0(x) = 0$ | voglio calcolare il potenziale lontano dalla sorgente, quindi $x \neq x_0$, quindi $\delta = 0$  

**2. Risolvo eq diff**

proof | cosa
---|---
$\nabla^2 \varphi_0 = \frac 1{r^2} \frac d{dr} (r^2 \frac{d \varphi}{dr}) = 0$ | per simmetria, $\varphi_0$ può dipendere solo dal raggio, non dall'angolo <br> quindi scrivo il laplaciano in coord polari, ma mi interessa solo la parte con la derivata rispetto a r 
$r^2 \frac{d\varphi_0}{dr} = C$ | se la derivata rispetto a r è nulla, e non può dipendere dall'angolo per simmetria, allora è una costante   
$\frac{d\varphi_0}{dr} = \frac C{r^2}$ <br> $\varphi_0(x) = A+\frac Br$ 

**3. Condizioni**

proof | cosa
---|---
$\varphi_0(\infty) = 0$ <br> quindi $A=0$ | uso condizione al bordo 
$\int_D \nabla^2 \varphi_0 \, d^3x= -\frac 1{\epsilon_0} \int_D \rho(x_0) \; \delta^3(x-x_0) d^3x$ | integro l'eq di Poisson (con carica puntiforme) su D, una sfera centrata in $x_0$ 
$\int_D \nabla^2 \varphi_0 \, d^3x = \int_D \nabla \cdot (\nabla \varphi_0) \, d^3x =  \int_{\partial D} \nabla \varphi_0 \cdot dS$ <br>   $\nabla \varphi = \frac{\partial \varphi}{\partial r} = -\frac{B}{r^2}$  | lato sinsitro: uso th della divergenza 
$  \int_D \rho(x_0) \; \delta^3(x-x_0) d^3x = \rho(x_0)$ | lato destro: $\int \delta = 1$, perchè $x_0$ appartiene al dominio su cui integro   
$\int_{\partial D} \nabla \varphi_0 \cdot dS = -\frac{\rho(x_0)}{\epsilon_0}$ <br> $\int_{\partial D} - \frac{B}{r^2} \hat r \cdot dS = -\frac{\rho(x_0)}{\epsilon_0}$ <br> $ -4 \pi B = -\frac{\rho(x_0)}{\epsilon_0}$ <br> $B = k \rho(x_0)$| finisco di risolvere

**4. Soluzione**

proof | cosa
---|---
$\varphi_0(x) = k \frac{\rho(x_0)}{\|x-x_0\|}$| conclusione
$\varphi(x) = k \int \frac{\rho(x_0)}{\|x-x_0\|} d^3x_0$ | integro su tutti i pezzetti

**Recap: nabla in coord polari**

**Approssimazioni**

faccio 2 approssimazioni diverse:  

1- $r \gg d$ (appross in multipolo)

**Note: Nabla**

$\nabla _i = \partial_i = \frac{\partial}{\partial x_i}$  

gradiente: $(\nabla f)_i = \partial_i f$  
dà in output un vettore, quindi devo precisare quale componente voglio  
qui i non è un indice ripetuto, è indice libero, appare anche a sinistra, serve a scegliere la componente, non a fare sommatoria  

divergenza: $\nabla \cdot E = \partial_i E_i$  
dà in output uno scalare, quindi non posso precisare la componente  
l'indice i ripetuto sottintende una somma: $\partial_iE_i = \frac{\partial E_1}{\partial x_1} + \frac{\partial E_2}{\partial x_2} + \frac{\partial E_3}{\partial x_3}$

rotore: $(\nabla \times A)_i =  \epsilon_{ijk} \nabla _j A_k = \epsilon_{ijk} \frac{\partial A_k}{\partial x_j}$  

nota: $\nabla_i = \partial _i$ solo in spazio euclideo cartesiano  
in realtività o in coordinate polari non vale, ci sono termini extra (connessioni)  
infatti le formule con nabla in coordinate polari sono diverse
