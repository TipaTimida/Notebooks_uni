### Interferenza di 2 onde coerenti

**Interferenza di 2 onde**

proof | cosa 
---|---
$E_1 = A_1 \cos(\omega t)$ <br> $E_2 = A_2 \cos(\omega t + \phi)$ <br> $E = E_1+E_2$| situazione: due onde    
$I \propto ⟨(E_1​+E_2​)^2⟩ =  ⟨E_1^2⟩+⟨E_2^2⟩+2⟨E_1E_2⟩$| intensità proporzionale al quadrato di ampiezza
$⟨E_1^2⟩ = A_1^2 ⟨\cos^2 \omega t⟩ = \frac 12 A_1^2$ | termini singoli 
$⟨E_1E_2⟩ = A_1A_2 ⟨\cos (\omega t) \; \cos(\omega t + \phi)⟩ $ <br> $= A_1A_2 \frac 12⟨ \cos \phi + \cos (2 \omega t + \phi)⟩$ <br> $= \frac 12 A_1A_2 \cos \phi$| termine di interferenza <br> uso identità trig: <br> $\cos A \cos B = \frac 12 [\cos(A-B) + \cos (A+B)] $ <br>$ ⟨cos(2ωt+\phi)⟩=0$, calcola l'integrale su un periodo  
$I = I_1+I_2 +2\sqrt{I_1I_2} \cos \phi$| risultato finale, in funzione delle intensità: <br> $I_1 = \frac 12 A_1^2$  

**Nota**

se integro su un periodo:  
$<cos^2>=\frac 12$, pechè è sempre positivo  
$<cos>=0$, perchè le parti positive e negative si cancellano  

la media su un periodo è lineare, quindi la media della somma è la somma delle medie  
$<f>=\frac 1T \int_0^T f\, dt$   
il modulo invece non è lineare, quindi non vale  

**Casi particolari**

proof | cosa
---|---
$I = 2I_0 + 2I_0 \cos \phi = 2I_0(1+\cos \phi) = 4 I_0 \cos^2 \frac\phi 2$| onde coerenti <br> con $A_1=A_2$  
$⟨cos \phi⟩=0$, perchè la fase cambia in modo casuale <br> cioè $⟨E_1E_2⟩ =0$, non c'è termine di interferenza <br> quindi non ci sono frange, non c'è interferenza <br> $I=I_1+I_2$| onde NON coerenti

**Collegamento con differenza di cammino**

proof | cosa
---|---
$\phi = k \delta = \frac{2 \pi}\lambda \delta$ | collegamento differenza di fase $\phi$ e di cammino $\delta$ <br> oerchè per fare un cambio fase di $2 \pi$ serve percorrere un'intero $\lambda$  
$\delta = d \sin \theta$ | nel disegno, con due schermi, cerco la differenza di cammino <br> approssimazione di schermo lontano:  $L \gg d$ <br> cioè vale quando i due raggi sono quasi paralleli  
$\sin \theta \sim \tan \theta = \frac yL$ | approssimazione di piccoli angoli: $\theta \ll 1$

### Diffrazione

proof | cosa
---|---
$E(\theta) \propto \int_{-a/2}^{a/2} e^{ik \, x \sin \theta} \, dx$ | divido la fenditura lungo a in pezzetti dx <br> sommo il campo prodotto da ogni pezzettino nella direzione di angolo $\theta$ <br> nota che: $\phi = k \delta = k \, x \sin \theta$, è la differenza di fase dovuta al diverso cammino per ogni punto  
$E(\theta) \propto [\frac { e^{ik \, x \sin \theta}}{ik \sin \theta}] = ...= \frac{2 \sin (\frac{ka \sin \theta}{2})}{ik \sin \theta}$ <br> definisco $\beta =(\frac{ka \sin \theta}{2})$ | integro <br> uso $e^{ia}-e^{-ia}= 2i \sin a$  
$E(\theta) = E(0) \frac{\sin \beta}\beta$ <br> $I(\theta) = I(0) (\frac{\sin \beta}\beta)^2$| intensità
