### Risolvere eq differenziali

**Quando**

ho equazione differenziale in y, dove y(t)  
uno dei coefficienti è: $A=Re[A_0e^{-i\omega t}]$   

**Steps**
  
1- ipotizzo una soluzione con la stessa fase, quindi della forma:  $y=Re[y_0e^{-i\omega t}]$   
con $y_0$ ignota, è ciò che voglio calcolare   
2- pluggo $y$ e le sue derivate nell'eq differenziale, l'esponenziale dovrebbe semplificarsi perchè è in tutti gli addendi  
se è eq lineare, posso pluggare tutta lìespressione complessa, ignora la parte reale  
3- ora è solo algebra, trova $y_0$  

**Note**

1- ipotizzare una sol della stessa fase significa cerca una sol in 'situazione di regime'  

2- se la fase ha dipendenza sia spaziale che temporale,, metto x=0, mi interessa solo la dipendenza spaziale???  

3- se è equazione lineare, posso invertire l'ordine degli operatori  quindi sposto Re[] alla fine, posso pluggare tutta l'espressione complessa, non serve calcolare la parte reale prima  

4- se il testo mi dà $A=A_0 \cos (\omega t)$, possi comunque dire che espando la soluzione al campo complesso, e riscrivere $A=Re[A_0 e^{-i \omega t}]$  
così ora risolvo nel modo più facile  

**Usi**

spesso si usa in esercizi su:  
induzione  
moto di una carica in presenza di campi    
equazione di continuità della carica  

#### Esempi

**Esempio: induzione**

eq diff in I: $-\frac{d\Phi_B}{dt} =IR+ L\dot I$    
coeff: $B(t)=Re[B_oe^{-i \omega t}]$  

1- assumo, in situazione di regime: $I = Re[I_0 e^{-i \omega t}]$   
2- pluggo dentro: $A \; i \omega B_0 = I_0R-i \omega I_0L$   
3- risolvo, trovando $I_0$  

**Esempio: moto di carica in campo elettrico**  

eq diff: $\ddot r = -\frac em E-v \dot r -\omega^2_0 r$    
coeff: $E=Re[E_0e^{-i(kz- \omega t)}]$  

1- assumo: $r=Re[r_0e^{-i(kz- \omega t)}]$, con $r_0$ da trovare  
2- pluggo in tutti gli r,derivate di r ed E i valori complessi  
inoltre metto x=0, perchè onda piana deve essere polarizzata lungo un'altra direzione, non x, assumo che resti sempre in x=0???  
3- trova $r_0$  

**Esempio: equazione di continuità**

eq diff: $\partial_t \lambda + \partial_z I=0$   
coeff: $I= Re[I_0 e^{-i \omega t}]$  

1- ipotizzo $\lambda = Re[\lambda_0 e^{-i \omega t}]$  
2- pluggo entrambi:  
$-i \omega \lambda_0 e^{-i \omega t} + \frac{dI}{dz} =0$  

NO, I deve dipendere anche da z???  

**Esempio: densità di energia**

$E = Re[E_0d^{i(kx-\omega t)}] \hat y$  

1- calcolo B  
$\partial_t B = -\nabla \times E$  
$B = Re[\frac{E_0}{v_f}e^{i(kx-\omega t)}]\hat x$  

2- formula densità energia  
$U_{em} = <\frac 12 \epsilon_0 E^2+ \frac{B^2}{2 \mu_0}>$  
posso separare la somma dall'inizio  

3- usa formula: $<A^2>= \frac 12 A_0^2$  
$U_{em} = \frac 12 \epsilon_0 <E_2^2> + \frac{1}{2 \mu_0 v_f^2}{E^2} = \frac 14 \epsilon_0 <E_2^2> + \frac{1}{4 \mu_0 v_f^2}{E^2}$  

**Esenmpio: potenza**

$I(z,t) = I_0 e^{i(kz - \omega t)}$  
$V(t) =  V_0 e^{-i \omega t}$  

$<P>=<VI> = \frac 12 V_0 I_0$  
cioè ignora la parte oscillante, prenderne la parte reale e poi fare la media su un periodo mi dà 1/2  

### Media e modulo

Cosa | $\| \cdot \|$ modulo complesso | $Re[\cdot]$ parte reale |$⟨⋅⟩$ media temporale
---|---|---|---
che vuol dire | prendi il modulo di qualcosa di complesso con Pitagora (spesso usato al quadrato) | prendi la parte reale |prendi la media temporale 
uso con $p=p_0e^{-i \omega t}$ | $\|p\| = p_0$ | $ Re[p]= p_0 \cos (\omega t)$ | $<p^2> = \frac 12 p_0^2$  
note | se il quadrato è fuori, prima prendo il modulo, poi faccio il quadrato <br> $\|p\|^2 = p_0^2$ | | $⟨AB⟩ = \frac 12 Re[A_0B_0^*]$  <br> $⟨A^2⟩=\frac 12 A_0^2$ <br>  $<\cos^2(\omega t)> = <\sin^2(\omega t)> = \frac 12 $ <br> $<\cos(\omega t)\sin(\omega t)> = <e^{-i \omega t}>=0$   

**Ordine operatori**

devo usare gli operatori dal più interno al più esterno  
posso invertire l'ordine SOLO nelle equazioni LINEARI (cioè se non c'è elevamento al quadrato)  

#### Altro

**nota**

attenzione a come è definito p, se è complesso o se il vero valore è solo la parte reale:  
$ p=p_0e^{-i \omega t} \quad \rightarrow \quad ⟨p^2⟩=⟨p_0^2⟩⟨e^{-2 i \omega t}⟩ =0$    
$p=Re[p_0e^{-i \omega t}] \quad \rightarrow \quad ⟨p^2⟩= ⟨p_0^2⟩⟨\cos^2 (\omega t)⟩= \frac 12 |p_0|^2 = \frac 12 p_0^2$     

perchè stai mediando una funzione oscillante (coseno + seno) su un periodo  


**Ricorda**

$(1+A)(1+A^*)=(1+A)(1+A)^* = |1+A|^2$  

**Spiegazione: ordine degli operatori**

1- quando posso invertire l'ordine degli operatori?      
solo in EQUAZIONI LINEARI  
cioè non compaiono incognite al quadrato, invece derivate superiori vanno bene  

2- spiegazione  
per linearità (quindi solo nelle eq lineari): $TB(x)=BT(x)$ (con T,B generici operatori)  

3- esempio uso: risolvere eq diff in campo complesso  
se ho come param e ansatz qualcosa definito con la parte complessa: $A=Re[A_0 e^{-i \omega t}]$   
essendo Re[] un operatore, posso spostarlo per ultimo: $ T(Re[x]) = Re[T(x)]$  
invece di calcolare la parte reale prima di pluggare (quella col cos), pluggo tutta la parte complessa, e Re[] lo riscrivo solo nell'ansatz  

4- esempio NON uso: calcola rms  
voglio calcoalre $<I^2>_T$, ma NON è lineare, quindi non posso invertire l'ordine degli operatori  
quindi devo fare le cose nell'ordine giusto, prima elevato al quadrato e poi prendo la media temporale  

5- esempio uso:  
$\vec S=\frac{1}{\mu_0}E \times B$  
per campi scritti in notazione complessa:  
$<S> =\frac{1}{2}\epsilon_0 c^2 Re(E \times B)$  

### Scrivere formule più generiche

**Scrivere formule più generiche**

ho corrente $I(z,t)$ su un resistore di res per unità di lunghezza variabile $r(z)$  
potenza locale istantanea: $p(z,t)=I^2(z,t)r(z)$   
potenza locale media:  $P(z) =  ⟨I^2⟩_T \, r(z)$   
potenza totale istantane: $P(t) \int I^2(z,t) \, r(z) \, dz$  
potenza totale media: $P= \int <I^2(z,t)>_T \, r(z) \, dz$  

nota: se ho un unico resistore R, allora riscrivo togliendo la dipendenza dallo spazio:  
potenza istantanea: $p(t)=I^2(t)R$  
potenza media: $P=<I^2>R$  

recap:   
se c'è dipendenza dal tempo: prendi la media su un periodo $<\cdot>$  
se c'è dipendenza dallo spazio: integra su tutto lo spazio  

**Esempio: potenza totale media**

ho: $I(z,t) = Re[I_0(1-\frac{ |z|}a)e^{-i \omega t}]$  
voglio P dissipata in media su un periodo, su un'asta lungo z da [-a, +a]  

1- formula: $P_{diss}=\int <I^2>_T \, r(z) \; dr $   
integrale: per avere la totale sullo spazio, fa sparire z    
media temporale: per avere la media sul tempo, fa sparire t    

2- media nel tempo:  
$<I^2>_T = <I_0^2 (1-\frac{|z|}a)^2 \cos^2(\omega t) >= \frac 12 I_0^2 (1-\frac{|z|}a)^2 $   

3- integrale nello spazio  
$z = \frac{R}{2a}$    
altro modo di scriverlo: $dR = \frac{R}{2a} dz$ 

4- risolvo:  
$P_{diss}=\int_{-a}^a \frac 12 I_0^2 (1-\frac{|z|}a)^2 \; \frac{R}{2a} dz = \frac{I_0^2R}6$
