### Campo indotto dalla variazione dell'altro

**Steps**

1- guarda che campo conosco e scegli formula  
2- riscrivi derivata spaziale  
3- riordina, per trovare formula di campo non noto  

**1. Guarda che campo conosco**

mi interessa la dipendenza spaziale  
ho due casi  

1- so scrivere campo in forma locale, $E(x,t)$  
cosa: usa eq in forma locale  

2- so scrivere campo solo in forma integrale, $E(t)$, ma so disegnare le linee di campo  
tipo, so è assiale o azimutale  
cosa: usa eq in forma differenziale  

**1. Che campo conosco e come scegliere equazione**

1- $E(x,t)$  
dove: valido in tutto lo spazio (campo locale)  
forma eq: locale/differenziale  
quale eq: derivata spaziale di E (è il più seccante, meglio torglierselo subito)  

2- $E(t)$   
dove: solo in una regione limitata di spazio, con qualche simmetria (eg. dentro un condensatore)  
quale eq: derivata temporale di E 
forma eq: forma integrale (non ha dipendenza spaziale, non posso fare rotore)  

recap:  
$E(x,t)$: locale, derivata spaziale (rotore)   
$E(t)$: integrale, derivata temporale  

**Esempi: che eq scegliere**

1- caso locale  
so: $E=E_y(x,t) \, \hat y$  
uso: $\nabla \times E = -\frac{\partial B}{\partial t}$  (locale, derivata spaziale)  

2- caso integrale (simmetria cilindrica, so assiale)   
so: $E(t)$ (in condensatore)  
uso: $\int B \cdot dl = \frac 1{c^2} \frac{\partial \Phi(E)}{\partial t}$ (integrale, derivata spaziale)  

3- caso integrale (simmetria cilindrica, so azimutale)    
so: $B(t)$ assiale  
uso: $\int E \cdot dl = -\frac {d\Phi(B)}{dt}$ (integrale, derivata spaziale)   

tip: se so qualcosa di assiale, devo fare il flusso di quella cosa  
se so qualcosa di azimutale, devo farci la ciruitazione  

**2. Riscrivi derivata spaziale**

1- caso locale  
se: $E=E_y(x,t) \, \hat y$  
$\nabla \times E = \frac{\partial E_y}{\partial x} \hat z$  
perchè: scrivi formula del rotore, in coord cartesiane, trova l'unica componente non nulla

2- caso integrale  
il campo nella circuitazione (non noto) è azimutale (in entrambi i casi???), quindi:   
$\int B \cdot dl = 2 \pi r \, B_\phi$ 

**Nota: direzione**

se un lato dell'uguaglianza è lungo $\hat z$, anche l'altro lo è  
quindi trova l'unica componente non-nulla da un lato, so la direzione dell'altro  

**Nota: corrente di conduzione e scorrimento**

se il testo mi dice E, e vedo che è variabile, quella è la corrente di scorrimento  
la corrente di conduzione è $J = \sigma E$   
se il testo non specifica che c'è, assumo che non c'è, anche perchè non ho il valore di resistività  

#### Esempi

**Esempio: piana lineare**

so: $E=Re[E_0e^{i(kx-\omega t)}]\; \hat y$   
assumo: $B=Re[B_0e^{i(kx-\omega t)}]$  

1- se assumo (più facile):  
$\nabla \times E = -\partial_t B$   
$(\partial_x E) = i \omega B$  
$B=  \frac1{i \omega} (\partial_x E) \hat z = \frac k \omega E_0 e^{i(kx-\omega t)} \hat z$  

2- se NON assumo:  
$B = \int (- \nabla \times E) dt = \int (\frac{\partial E_y}{\partial x} \hat z) \, dt = [\int \, ik\, E(x,t) \, dt ]\hat z= \frac k \omega E(x,t) \hat z$  

infatti ricorda che, per onde piane, $|E|=c|B|$, e $k, E, B$ sono perpendicolari  

**Esempio: piana con polarizzazione circolare**

se E ha propagazione lungo z, ma polarizzazione circolare (quindi componenti lungo due assi)  
separa E in somma di due piane, trova B generata da ognuna, poi sommale  
cioè, usa superposizione delle onde  

$\partial_tB = -\nabla \times E$  
$-i\omega B = -(\hat x \partial_y-\hat y \partial_x)E_z$  
$B_x = \frac{k_y}{\omega} E_z$    
$B_y = \frac{k_x}{\omega} E_z$    


**Esempio: spira con $I$ variabile**

situazione:  
spira (di raggio r) con I variabile produce B variabile, azimutale  
conosco B(t) e so che è azimutale  

$\int E \, dl = -\frac{d\Phi_B}{dt}$  
$2 \pi r  \; E = - (\pi r^2) \frac{dB}{dt}$   
$E = -\frac r2 \frac{dB}{dt}$    

considero come $\gamma$ una circonferenza di raggio r, cioè la circonferenza della spira  
quindi anche $E$ è azimutale  
nota che sono entrambi azimutali, il che va bene, perchè non sono onde piane  
$\hat \phi$ in B(t) non è ingluenzato dalla derivata temporale, resta uguale    

nota: invece di $\phi_B=BS$, potrei usare $\phi_B=LI$ se è prodotto da un oggetto di cui so L  
di solito si fa nella spira, perchè B non è uniforme nella spira, a meno che non sia molto piccola  

**Esempio: condensatore con Q variabile**

situazione:  
condensatore con Q variabile produce E variabile, assiale  

$\int B \; dl = \frac{1}{c^2} \frac{d\Phi_E}{dt}$  
$2 \pi r \, B = \frac{1}{c^2} (\pi r^2) \frac{dE}{dt}$  
$B = \frac r{2c^2}  \frac{dE}{dt}$  

come $\gamma$ scelgo un cilindro assiale con il condensatore  
quindi B è azimutale  

nota: approssimazione di correnti lentamente variabili  
questo campo B induce un nuovo campo elettrico, $E_1 \sim e \partial_t B_\phi \sim \frac{r^2}{2c^2}\partial^2_t E  $ (leggi di maxwell???)  
quindi l'approssimazione vale se $E_1 \ll E$, cioè quello indotto dalla B indotta è trascurabile  

### Campo in altro frame

**Steps**

1- trova E e B in starting frame, attenta alle direzioni  
2- trova direzione del boost  
3- usa le trasformazioni adatte su ogni componente di E e B  

**Formule**

S' fa boost lungo x, con velocità v  

1- componenti parallele: si conservano  
$E'_x=E_x$ 
$B'_x=B_x$  

2- componenti perpendicolari: cosa che sembra trasformazione di Lorentz  
$E'_y = \gamma(E_y -\beta c B_z)$  
$B'_y = \gamma(B_y +\beta E_z/c)$  

**Come usare le formule**

1- nel prodotto scalare, come secondo termine metti la componente del campo ortogonale sia a $\beta$ sia alla componente che sto calcolando ora  
tipo in formula di $E_x$, se $\beta$ del boost è lungo z, metti $B_y$, legge della mano destra  s

2- ricorda che $\beta = \frac vc$  
con $v$ velocità di boost  

3- è scritta con vettori, quindi disegna le direzioni dei campi per vedere quali sono i segni, a volte il prodotto vettoriale dà un ulteriore -  
specialmente in $v \times B$  

**Tips**

per fare i calcoli, cerca sempre le $\gamma$ per sotituire  
si riconoscono in queste forme: $\gamma = \frac{1}{\sqrt{1- (v/c)^2}} = \frac {c}{\sqrt{c^2-v^2}}$  

infatti, spesso i risultati sono tipo $E'=\frac E \gamma$, quindi anche se nella formula di Lorents c'è $\gamma$ al numeratore, alla fine finisce al denominatore  

#### Altre domande

**Cambio sorgenti**

Q è uguale in ogni frame, solo spazio e tempo cambiano  

1- densità di carica: $\sigma = \frac QA$   
c'è contrazione delle lunghezze: se la superficie dove c'è la carica ha un lato lungo la direzione del boost, quel lato cambia lunghezza  
$A=xy \rightarrow A'=x'y = \frac x \gamma y = \frac A \gamma$  
poi: $\sigma' = \frac Q {A'}=\gamma \sigma$   
cioè, fregatene che sia in moto  

2- densità di corrente  
carica in moto: corrente  
$K=\sigma'v = \gamma \sigma v$   

altro modo, più sicuro:  
fare semplicemente la trasformazione di lorentz sulla quadricorrente  

**Cambio sorgenti: altro modo di risolvere**
  
ho la quadricorrente: $I^\mu = (\lambda c, \vec I)$  
si trasforma come un quadrivettore (attenta alla direzione del boost)   
e ora dalle nuove sorgenti genero i nuovi campi, invece di fare il contratio  

**Forze**

Se la particella è ferma nel sistema iniziale:  
come cambiano:  
componente parallela al boost: $F'_\parallel = F_\parallel$  
componente perpendicolare al boost: $F'_\perp = \frac{F_\perp}\gamma$  

diverso è il caso per la forza per unità di lunghezza  
devo vedere se la lunghezza è lungo il boost, in quel caso come viene scalata, se moltiplicata o divisa per $\gamma$  

nel caso generale, in cui la particella fa qualunque moto nel sistema iniziale, invece è troppo difficile da modellizzare  

**Quadrivettore d'onda**

$K= (\frac \omega c , \frac \omega{v_c} \hat x)$  
x: direzione di propagazione dell'onda  
$ \frac \omega{v_c} = \vec k$  

usa trasf di Lorentz su di esso  
se ottengo $\vec k=0$ significa che è un campo elettrico uniforme, non un'onda (perchè non si muove nello spazio)

#### Teoria

**Find frame where field is easier**

assume have E and B perpendicular, E along y, B along z  
so will do boost along x, the direction of $E \times B$, so don't introduce component parallel to them   
so only need to use the transformation for the perpendiculat components of E and B, not for the parallel one  

1- want frame with B'=0   
can if: $E > cB$  
the trasformation is: $B'_z =\gamma(B-\frac v{c^2}E)$  
to have B'=0, need: $B-\frac{v}{c^2}E=0$, so pick $v=\frac{c^2B}E$  

2- want frame with E'=0  
can if: $cB>E$  
the trasformation is: $E'_y =\gamma(E-vB)$   
to have E'=0 need E-vB=0, quindi $v=\frac EB$  




**Traformazioni di Lorentz**

per passare dai campi E,B da un sistema ad un altro, devo fare trasformazione di Lorentz sul tensore elettromagentico $F^{\mu v}$  

1- scrivo $F^{\mu v}$ nel frame di inizio  
2- lo trasformo: $F^{\mu v '}=\Lambda _\alpha^\mu \Lambda _\beta ^v F^{\alpha \Beta}$  
3- esamuno la forma che deve avere nel frame di uscita, vedo a cosa corrispondono E e B
