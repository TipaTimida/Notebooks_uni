### Pressione di radiazione

**Pressione di radiazione**

caso generale: $p_{rad}=(1+|r^2|) \frac Ic$  
assorbimento completo: $p_{rad} = \frac Ic$  
riflessione completa: $p_{rad} = 2 \frac I c$  

**Equazione del moto**

$m \dot v = p_{rad}A$  
integro e viene: $m(v_f-v_0) = p_{rad} A \tau$  

situazione:  

1- oggetto che colpisce: pacchetto d'onda  
$A$: sezione (del pacchetto d'onda)  
$\tau$: lunghezza (del pacchetto d'onda)  

2- oggetto che viene colpito: specchio / vela solare  
m: massa (dell'oggetto colpito)  
v: velocità (dell'oggetto colpito)    
$p_{rad}$: pressione (sull'oggetto colpito, vedi se è oggetto asssorbente o riflettente) (la I che appare nella sua formula è quella del'l'onda, quindi dell'oggetto che colpisce)  


#### Spiegazione

**Spiegazione formule**

1 è per l'onda incidente  
$|r^2|$ per la potenza dell'onda riflessa (potenza dell'onda scala come ampiezza al quadrato, e ampiezza della riflessa è r per ampiezza dell'incidente)   

nota che il caso assorbimento o riflessione completa è quando $r=0$ oppure $r=1$, cioè la riflessa non è esiste o è tutta l'onda orginaria  

**Spiegazione intuitiva**

se onda colpisce un oggetto, lo spinge  
se l'oggetto riflette l'onda, è sia spinto da quella originaria, sia sente il rinculo di quella che lancia lui 
 

**Spiegazione formale**

uso conservazione quantità di moto  
considero il paccehtto d'onda come un oggetto di quantità di moto $p=\frac Uc $  
$\Delta p$: variazione di momento dell'oggetto colpito    

contro specchio: dopo urto il pacchetto cambia direzione (come pallina contro muro)  
$ \frac Uc = -\frac Uc + \Delta p$  
$\Delta p = 2 \frac Uc$  

contro oggetto assorbente: dopo urto, il pacchetto non c'è più  
$\frac Uc = \Delta p$  

#### Modellizza urto

**Modellizza urto (per onda piana)**

contro oggetto:  
riflettente: $dp = \frac{2 \, dU}c$   
assorbente: $dp = \frac{dU}c$  

**Attenzione: come si scrive la forza**

1- quantità di moto p  
$F = \frac{dp}{dt}$    
oggetto riflettente: $dp = \frac{2 \, dU}c$   
oggetto assorbente: $dp = \frac{dU}c$  

2- pressione P  
$F = PA$  

3- potenza P:  
$F = \frac {P}c$   
dove: $dU = P \, dt$    

in tutti quanti, appare coeff 2 a destra se ho oggetto completamente riflettente  
cioè la forza raddoppia, perchè c'è anche la spinta dell'onda riflessa 

**Spiegazioni**

trova pressione da momento  
$P= \frac FA = \frac 1A \frac{dp}{dt}$    

trova forza in funzione di potenza:  
$F= \frac{dp}{dt} = \frac{dU/c}{dt} = \frac{(P \, dt)/c}{dt}=\frac Pc$  
contro oggetto che assorve  
invece, contro oggetto che riflette: $dp = 2 dU/c$, quindi $F=2\frac Pc$  

#### Specchio (in moto)

**Cosa è**

schermo COMPLETAMENTE RIFLETTENTE, no assorbimento  
cioè riflette tutte le onde EM, oltre di esso E e B sono nulli  

eg. per la luce visibile, un metallo è uno specchio  
perchè la riflette tutta, non è trasparente  

**Specchio fermo o in moto**

specchio fermo: rimanda indietro le onde    
quindi mantiene $\omega$ uguale, cambia segno a $k$  

specchio in moto: non so le formule, devo fare boost fino al frame dove lo specchio è fermo  
poi lo faccio riflettere, e infine torno a frame con specchio in movimento  
quindi, in frame con specchio in moto, $\omega$ dell'onda rifletta è diversa da $\omega$ di quella incidente   
in pratica, sembra un effetto Doppler doppio, $\omega_r = \omega \frac{1- \beta}{1+ \beta}$  

**Onda piana contro specchio**

onda piana ha $\vec k$ ortogonale allo specchio  
come sono riflesse le componenti:  
$k$: cambia segno, perchè onda è riflessa nell'altra direzione  
$E$: cambia segno, perchè devo avere E=0 alla superficie dello specchio (visto che E=0 oltre lo specchio, e devo rispettare le condizioni di raccordo di E ortogonale)  
$B$: NON cambia segno (si trova con regola di mano destra sapendo k ed E)  

specchio fermo: nessuno cambia ampiezza (solo segno)  
specchio in moto: E e B cambiano ampiezza  

**Usi**

uno specchio in moto può variare la frequenza di un'onda   
è il principio su cui si basano gli autovelox  

### Onda dopo riflessione su un mezzo

**Situazione**

onda incidente: $E(x,t)=E_i e^{i(kx-\omega t)}\hat y$  
in x<0 ho aria, in x>0 mezzo con $n=n_1+in_2$  

**Calcola E: onde riflessa e trasmessa**  

1- scrivi formula di ogni onda  
$E_i = (k, \omega)$  
$E_r= (-k, \omega)$  
$E_t = (kn, \omega)$  

spiegazione: 
$\omega$: è sempre uguale    
$k$: nella riflessa diventa $k_r=-k_i$, nella trasmessa $k_t=k_in$ (perchè $\frac k\omega = v = \frac cn $)       
$|E|$: trovala con formule di Fresnel  

2- scrivi tutto insieme:  
primo mezzo: $E_1(x,t)=E_i(x,t)+E_r(x,t)$  
secondo mezzo: $E_2(x,t)=E_t(x,t)$  
cioè separa le zone x<0 e x>0  

**Calcola B e J**

per B: formula per onda piana  
$\partial_t B =-\nabla \times E$  
$-i \omega B_z = -\partial_x E_y$ (formula per onda piana)  

per J:  
$\mu_0J = \nabla \times B -\frac 1{c^2} \partial _t E$  
$\mu_0 J_y = -\partial_x B_z +i\frac{\omega}{c^2} E_y$  
nota che J può esistere solo dentro un materiale, non nel vuoto 

### Approssimazione onda piana

**Casi**
  
1- approssima onda sferica come piana  
formula: $L \ll R$  
cioè guardi una porzione di fronte d’onda molto piccola  
L: dimensione della regione oservata, R: dustanza dalla sorgente  
più corretta: $L^2 \ll R \lambda$, perchè???  

2- pacchetto d'onda  
$L \gg \lambda$  
perchè: $ \theta \sim \frac \lambda L \ll 1$  
così ignoro l'allargamento del fascio, la diffrazione  


**Spiegazione**

in teoria l'onda viaggia dritta lungo x  
ma in realtà si allarga un po' a cono, quindi c'è un allargamento $\Delta L$:  
$\tan \theta = \frac{\Delta L}x$  

per angoli piccoli: (perchè???)    
$\tan \theta \sim \theta \sim \frac \lambda L$  

distanza a cui raddoppia, cioè a cui $\Delta L = L$:  
$x = \frac{L^2}{\lambda}$  

quindi se $x \ll \frac{L^2}\lambda$, il fascio non si è allargato molto, fronti d'onda sono quasi piani  

**Approssimazione uniformità**  

quando studio un punto molto piccolo dello spazio, vuol dire che l'onda è uniforme in tutta quella regione  

esempio:  
se $\lambda \gg a$, dove a è il raggio di una sfera, vuol dire che l'onda è uniforme su tutta la sfera   
quindi scrivo $E=E_0e^{-i\omega t}$  
non metto la parte $e^{i kx}$  

usato spesso con sfere e spire  

### Dispersione nel plasma

**Relazione di dispersione nel plasma**

$\omega^2 = \omega_p^2 +k^2c^2$  

**Trova: rapporto tra empiezze $E_0$ e $B_0$**

impongo  $\nabla \times E = -\partial_t B$    
come per trovare relazione di dispersione 

**Trova: relazione di dispersione**

1- se so espressione locale di E e B  
impongo: $\nabla \times E = -\partial_t B$  
appariranno k e $\omega$, trovo il rapporto che mi interessa  

2- se so $n^2(\omega)$    
uguaglio i modi di scrivere $v_f$  
$v_f = \frac \omega k = \frac cn$  
$k^2c^2 = \omega^2 n^2 =\omega^2(1-\frac{\omega_p^2}{\omega^2}) = \omega^2-\omega_p^2$  

3- so solo E  
uso equazione di Helmholtz: $\nabla^2 u + k^2 u =0$   
doe metto $k^2 = \frac{\omega^2}{c^2}$  

4- in generale:  
$k_x^2+k_y^2 = \frac{\omega^2}{c^2}$  
c'è propagazione se $k_x$ è reale  
isola $k_x$, è uguale a una radice, deve avere argomento positivo  

**Trova: velocità**

$v_f = \frac cn$ $\quad  v_g = \frac{\partial \omega}{\partial k}$  
dove $\omega(k)$ è la relazione di dispersione nel plasma  

**Come trovare forza d'attrito**

1- se ho più di un tipo di cariche:  
$\epsilon(\omega)=1-\Sigma_i \frac{\omega^2_{pi}}{\omega^2}$  
ogni specie contribuisce con la sua frequenza di plasma  
le polarizzazioni si sommano (linearità)  
guarda sotto, quando passo da P a $\epsilon$  

2- con campo magnetico  
$\epsilon(\omega)=1-\Sigma_i \frac{\omega^2_{pi}}{\omega(\omega \pm \gamma)}$  
inizio da: $m \ddot v = q(E + v \times B)$  
cioè: $\ddot x -\dot x \frac {qB}m= \frac{qE}m $  
confronto con l'eq del moto nel modello di Lorentz, ho $\omega=0$ (niente che produce forma elastica di richiamo) e $\gamma = -\frac{qB}{m} = \Omega_c$  
anche qui, se ho più specie sommo l'apporto di ognuna  

### Altro

**Intensità radiazione e Fresnel**  
 
voglio intensità della radiazione a distanza L dall'antenna  

1- Il campo emesso dall'antenna dipolare giace nel piano di incidenza per cui occorre prendere il coefficiente di Fresnel per polarizzazione P  

2- $I(L)=\frac {R_p}{r^2} \frac {dP_{irr}}{d \Omega}$  
dove:  
$ \frac {dP_{irr}}{d \Omega} = \frac {3P_{irr}}{4 \pi} \sin^2 \theta$  
$\theta$: l'angolo che il tra la traiettoria del raggio e $\vec p$, indicalo tipo con $\tan \theta =\frac L{2h}$  
$r$: distanza percorsa dal segnale radio / lunghezza della traiettoria, indicala tipo con $r^2=(L^2+h^2)$  
$R_p=|E_r/E_i|$, prese dalle formule di Fresnel, è di quanto diminusice l'ampiezza dell'onda durante la riflessione (nota che dipende da $\theta$)  

ricorda:  
il campo elettrico irradiato da un dipolo è trasverso (⊥ alla direzione di propagazione)  
contenuto nel piano definito da: $p$ e $k$

dove $\omega _c = \frac {eB}m$  

quindi per onde con lunghezza d'onda $\lambda = 2 \pi c / \omega_c$
