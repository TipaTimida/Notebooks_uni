### Teoria sulla radiazione

**Definizione** 

cosa: è una onda EM libera, che trasporta energia e quantità di moto    
quando: una carica irradia solo se è accelerata (va bene anche accelerazione centripeta, quindi una carica in moto circolare)  

**Frequenza della radiazione**

una carica in moto circolare emette radiazione di tipo dipolare elettrico (classico) alla frequenza del moto     

note:  
1- il moto usa $\omega$, mentre la radiazione usa $f=2 \pi \omega$    
2- emette radiazioni anche dei suoi armmonici 2ω,3ω,... ma per moto uniforme armonico il modo fondamentale è dominante.

**Direzione**

$E_{rad} ∝ \hat n \times ( \hat n \times \ddot r)$  

Una carica accelerata irradia come un dipolo  
cioè la componente dell’accelerazione PERPENDICOLARE alla direzione di osservazione  

**Carica accelerata e dipolo elettrico**

carica accelerata irradia come un dipolo elettrico  

vale l'approssimazione di dipolo?:  
condizione: la dimensione della sorgente è piccola rispetto alla lunghezza   

esempio di approssimazione:  
la sorgente è una particella che fa moto circolare di raggio R  
condizione: $R \ll \lambda$  
dove $\lambda = \frac cf$, con f trovata prima  

### Polarizzazione

**Direzione di osservazione**

osservazione lungo z:   
ti metti molto lontano dalla carica, in direzione z  
cioè se la carica è nell'orogine, mi metto in un punto con coordinate (0,0,R)

perchè:   
la radiazione di una carica accelerata è quella di un dipolo oscillante  
campo radiato è:  
$E_{rad} ∝ \hat n \times (\hat n \times \ddot r)$  
n: direzione di oscillazione  
r: accelerazione della carica  
si vede solo la componente dell’accelerazione perpendicolare alla direzione di osservazione  
è una proiezione, dalla mia direzione vedo solo una parte di moto  


**Esempio: direzione di osservazione**

ho carica che fa moto circolare nel piano xy  

osservazione lungo z:  
sto guardando il piano xy  
vedo l'intero moto, quindi polarizzazione circolare (perchè fa traiettoria circolare)  

osservazione lungo x:  
sto guardando il piano yz  
quindi non vedo l'intero moto circolare, ma solo il moto armonico lungo la direzione y    
quindi polarizzazione lineare lungo y  

osservazione lungo un'altra direzione, non una delle assi principali:  
vedo moto strano, polarizzazione ellittica (per esclusione)  

cioè: ricorda che vedo solo le componenti dell'accelerazione perpendicolari alla mia direzione di osservazione  

**Nota**  

Un’onda piana:  
ha la stessa polarizzazione ovunque  
è una soluzione uniforme di Maxwell

La radiazione di una sorgente (tipo un dipolo o una carica puntiforme):    
è sferica  
la polarizzazione dipende dalla direzione  
è determinata dalla proiezione dell’accelerazione  

### Potenza irradiata

**Domanda**

1- scrivi momento di dipolo  
$|\ddot p|^2$ o $|\ddot m|^2$  

2- scrivi potenza istantanea  
P, formula la dà il testo  

3- calcola potenza irradiata MEDIA su un periodo  $⟨P⟩$   

**1. Momento di dipolo**

1- dipolo magnetico  
$m=IA$, probabilmente con $I(t)$  

2- dipolo elettrico  
$p= qr$, con $r(t)$  
testo mi dà $p(t)$ del dipolo???  

#### Esempi

**Esempio: moto circolare**

elettrone soggetto a forza centrale $f=kr$, fa moto circolare di raggio $a_0$  

1- trova momento  
$m \ddot r = -kr \quad \rightarrow \quad \ddot r = \omega^2 r$   
$|\ddot r|^2 =| \omega^2 r|^2 = \omega^4 a_0^2$    

**Esempio: so r(t) della carica**

1- momento  
trovata da legge del moto di una carica (soggetta a due campi E):    
$r(t) = A [\cos(\omega t +\Phi)\hat x + \cos (\omega t)\hat y]$  
$|\ddot r|=\omega^2 \; r(t)$  

2- momento mediato (per potenza mediata):  
$⟨|\ddot r|^2⟩ = \omega^4 A^2 ⟨[\cos^2 (\omega t +\Phi)+ \cos^2 (\omega t)]⟩ = \omega^2A^2(1/2+1/2)=\omega^4 A^2$   

nota che non dipende dalla fase  
la potenza raddoppia se ci sono due onde invece di una (perchè sommo il 1/2 dei coseni)  

**Calcolo potenza irradiata**

2- accelerazione  
$\ddot z(t) = A \cos \frac{\phi}{2} \cos (\omega t +\frac{\phi}{2})$  

3- magnitudine  
$|\ddot z|^2 = A^2\, |\cos \frac{\phi}{2} | \,  |\cos (\omega t +\frac{\phi}{2})| = A^2 \cos \frac{\phi}{2} \; \frac 12$  
stavolta c'è la dipendenza dalla fase  

### Perdita di energia per irraggiamento, spiraleggiamento

**Che succede**

la P che ho appena calcolato è energia irradiata, quindi persa dalla carica   
$\frac{dE}{dt}=-P$   

cosa | spiraleggiamento | smorzamento 
---|---|---
che succederebbe senza perdere energia | moto circolare | moto lineare 
che succede in realtà |spirale, comincia a cadere a spirale verso il centro, cambia la traiettoria | rallenta la velocità  
tempo di spiraleggiamento / smorzamento | quanto tempo impiega r(t) a cambiare in modo apprezzabile | quanto tempo impiega una grandezza dinamica (ampiezza, velocità, energia) a decadere  

**Risolvere**

1- parto da: $\frac{dE}{dt}=-P$   

2- dovrei ottenere un esponenziale, tipo:  
$\frac{dr}{dt}=\frac r\tau$ per spiraleggiamento  
$\frac{dv}{dt}=\frac v\tau$ per smorzamento 
vedi cosa è $\tau$ 

**1. Come scrivere l'equazione**

1- P: quella calcolata prima   
se moto circolare, metti un generico r invece di $a_0$, il raggio iniziale, perchè ora cambia nel tempo  

2- $\frac{dE}{dt}$ dipende dal tipo di esercizio  
smorzamento: $\frac{dE}{dt} =  \frac{d}{dt}(\frac{1}{2}mv^2)=mv \frac{dv}{dt}$   
spiraleggiamento: $\frac{dE}{dt} = \frac{dE}{dr} \frac{dr}{dt} =  \frac{d}{dr}(m \omega^2 r^2) \frac{dr}{dt} = 2rm\omega^2  \frac{dr}{dt} $  
dove $E=m \omega^2 r^2$ per un oscillatore armonico, perchè: $E= \frac 12 m v^2 + \frac 12 kr^2 = \frac 12 m (\omega r)^2 + \frac 12 (\omega m)^2 r^2 = m \omega^2 r^2$  

**Esempio: spiraleggiamento**

1- P calcolata prima  
$P = \frac{2k}{3c^3} e^2 \omega^2 a^2$  
a(t) è raggio orbita dell'elettrone  

2- perdita di energia  
$\frac{dE}{dt} = \frac{dE}{da} \frac{da}{dt} =  \frac{d}{dt} m\omega^2a^2 = 2 m\omega^2 a\frac{da}{dt}$  

3- totale  
$\frac{da}{dt} = -\frac{k}{3c^3}\frac{e^2}m \omega^2 a$  
che si risolve: $a(t)=a_0e^{-t/\tau}$  

**Quando è valido ???**

approssimazione usata: $\tau \gg T = \frac 1f$  
è la stessa per enrtambi i casi  
la controllo a posteriori  

perchè questa approssimazione:  
devo assumere rispettivamente r e v costanti su un periodo, perchè:  
1- spiraleggiamento  
perchè $U \sim m \omega^2 r^2$ è per un oscillatore armonico, che in teoria questo non è  
ma se $\tau \gg T$ il moto su un periodo assomiglia a quello non dissipativo, continua ad essere un cerchio  
2- smorzamento  
sto assumendo che il tempo caratteristico di perdita di energia per irraggiamento sia lungo rispetto a $f$ della radiazione  
così si può considerare la velocità come approssimativamente costante su un periodo di rotazione
