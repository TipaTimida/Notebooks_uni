**Approssimazione per onde piane**

per le onde piane, applica l'approssimazione $E=cB$  

e ricorda:  
$c^2 = \frac{1}{\epsilon_0 \mu_0}$  
$\frac 1 {\mu_0} = \epsilon_0 c^2$  


cosa | generale | onde piane
---|---|---
densità di energia em| $u=\frac 12 \epsilon_0 E^2 + \frac{B^2}{2 \mu_0}$ | $u = \epsilon_0 E^2$  
densità di quantità di moto | $\vec g= \frac {\vec S}{c^2} = \epsilon_0 (E \times B)$ | $g =\frac {\epsilon_0} c E^2$  
vettore di Poynting / intensità | $I = \|S\| = \epsilon_0 c^2 \|E \times B\|$ |  $I =  \epsilon_0 cE^2$  
pressione | $P = \frac 12 \sigma E + \frac{1}{2}\vec K  \times  \vec B$ <br> $ P= \frac Ic$ | $P=\epsilon_0 E^2 \quad$  (=u)  
densità di potenza dissipata | $W = E \cdot J =\sigma E^2$ <br> in mezzo ohmico | uguale  

**Nota**

nel vuoti, l'energia em è anche energia totale, perchè non c'è materia che immagazzina energia  
se invece ci sono cariche, tipo elettroni in moto, devo sommare anche la loro energia: $u_{tot} =u_{em} + u_{kin}$  
dove $u_{ki} = n_e \frac 12 m_e ^2$  

**Nota**

poi 1/2 viene quando faccio la media nel tempo, quindi $<u>=\frac 12 \epsilon_0 E^2$  
perchè è la media di $E(t) = E_0 \cos(\omega t)$  

**Valori macroscopici**

integra le densità:  
energia: $U= \int u \, dV$    
momento: $ p = \int g \, dV$  
potenza dissipata: $W = \int W \, dV$  

se ho un pacchetto d'onda, $\int dV$ è il volume del pacchetto d'onda, cioè $L^2\, c\tau$  

e ricorda che se oglio tipo la potenza media dissipata, devo fare:   
$<W> = \int \sigma <E^2>  \, dV =  \int \frac 12 \sigma E_0^2  \, dV$   

**Densità di energia**

se voglio energia PER UNITÀ DI LUNGHEZZA in un cilindro con B:  
$U= \int \frac{B^2}{2 \mu_0} dA = \int_0^R \frac{B^2}{2 \mu_0} \, 2 \pi dr \, dr$  

dV: energia totale  
dA: energia per unità di lunghezza  

**Rapporti per onda piana nel vuoto**

forma integrale: $p=\frac U c$   
con densità: $S=uc$ 

nota che sempre u=f (densità di energia = pressione)


**Osservazione**

come scrivere energia di pacchetto (approssimato come onda piana):    
$U = \int u \, dV = \epsilon_0 E^2 \; Ac \tau = I A \tau$  

intuitvamente:  
I: energia per unità di area e per unità di tempo  
$A$: area  
$\tau$: tempo
