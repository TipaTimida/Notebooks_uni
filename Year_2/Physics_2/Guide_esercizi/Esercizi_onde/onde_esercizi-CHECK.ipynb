### Carica soggetta a campo E - aggiungere a radiazione e carica in moto in campo ???  

**Trova equazione del moto**

1- scrivi campi a cui è soggetta  

2- scrivi eq. differnziale:  
$m \ddot r = qE$  

3- integra due volte per il tempo  
tutto ciò che fa è aggiungere $\omega^2$ al denominatore 

nota: spesso considero il moto in una regione di piano piccola, quindi invece di E(x,t) uso E(0,t)  
così è più facile integrare per t, visto che x è sparita  

**Esempio: equazione del moto, due onde con polarizzazioni perpendicolari**

1- campo (somma dei due campi):  
$E(0, t) = E_0(\cos (\omega t +\phi) \hat{x}+ \cos(\omega t)\hat{y})$  

2- equazione differenziale:  
$m\ddot r = qE$  
$\ddot r = \frac{qE_0}{m}(\cos (\omega t +\phi) \hat{x}+ \cos(\omega t)\hat{y})$  

3- integro:  
$r = \frac{qE_0}{m \omega^2}(\cos (\omega t +\phi) \hat{x}+ \cos(\omega t)\hat{y})$  

**Esempio: equazione del moto, due onde con polarizzazioni parallele**

1- campo (somma dei due campi):  
$E = E_0 [\cos (\omega t)+\cos(\omega t +\Phi)] \hat z$   

uso fomula trig:  
$\cos A + \cos B = 2 \cos (\frac{A-B}{2})\cos (\frac{A+B}{2})$  
$E = 2E_0 \cos \frac{\phi}{2} \cos (\omega t +\frac{\phi}{2}) \hat z$  

2- equazione differenziale:  
$m \ddot z= qE$  

3- integro:  
$z(t) = \frac{2q E_0}{m \omega^2}\cos \frac{\phi}{2} \cos (\omega t +\frac{\phi}{2})$  


**Casi di moto**

$r(t) = A [\cos(\omega t)\hat x + \cos (\omega t+\Phi)\hat y]$  

casi:  
$\phi=0$: ottengo x(t) = y(t), cioè il moto della carica è lineare lungo la bisettrice x=y    
$\phi= \pi$: simile a prima, il moto è lungo x=-y    
$\phi = π/2$ il moto della carica è circolare nel piano xy  

**Nota**

nel punto (x,y)=(0,0)  
significa che voglio studiare il moto in un piccolo spazio lì intorno, in cui quindi assumo che quei valori siano abbastanza piccoli da essere 0  
è per facilitare la somma dei campi  

## Gas ionizzato

### Teoria

**Nota**

un gas ionizzato può anche essere chiamato plasma  

**Cosa succede**

plasma è un gas di cariche libere  
quando arriva un campo elettrico:  
1- le cariche si muovono  
2- quindi creano corrente  
3- quindi modificano la propagazione dell’onda   

In pratica il plasma si comporta come un mezzo con permittività ε(ω)  

**Intuizione**

le cariche cercano di schermare il campo → riducono ε  
a bassa frequenza riescono a seguirlo bene → schermano tanto  
a alta frequenza non ce la fanno → il mezzo diventa trasparente  

### Cose da calcolare

**Onda incide su qualcosa**

quell'onda piana porta in sè due campi E e B, ortogonali tra loro e a k  
a seconda dell'esercizio, usa il campo che ti serve  
tipo se è sull'induzione, considera solo campo B  

polarizzazione significa direzione in cui è rivolto E  
sapendo che B è ortogonale a k e E, per esclusione trovo in che direzione è rivolto  

potenza irraggiata:  
formula simile a carica oscillante, ma con $|\ddot m|^2$ invece di p  
si fa uguale, ma trova $m=IS$  

**Forza per unità di lunghezza tra due fili**

se hanno sia carica elettrica che corrente:  
$\frac{dF}{dl} = \frac{dF_E}{dl} + \frac{dF_B}{dl}  = \lambda E \; + I \times B$  

E e D sono il modulo del campo generato da un filo e sentito dall'altro  
cioè, quando è forte il campo del filo 1 nel punto dove si trova il filo 2  

per ricordare:  
$F = qE+IlB$  
$dF = dq \; E+I\,dl \times B$    
e ricorda che $\lambda = \frac{dq}{dl}$  

### Cavo coassiale

**Trova campi**

campi esisotno sulo all'interno dei due cilindri    

E: usa gauss su superficie cilindrica  
$2 \pi r h \, E = \frac{\lambda h}{\epsilon_0}$  
è radiale  

B: usa Ampere su circonferenza coassiale ai cavi  
$B \, 2 \pi r = \mu_0 I$  
è azimutale  

queste formule valgono anche al contrario, tipo se so B azimutale e voglio trovare I  

**Relazione tra cose**

1- tra $\lambda$ e $I$  
dalla relazione di continuità:  
$\partial_t \lambda + \partial_z I=0$    

2- tra $\omega$ e $k$:  
dalla legge di faraday:  
$\nabla \times E = -\partial_t B$  

3- tra $E_0$ e $B_0$:    
dalla legge di faraday:  
$\nabla \times E = -\partial_t B$  

4- tra $V_0$ e $E_0$:  
$V(t) = V_0 e^{-i \omega t} = - \int_a^b E \, dr$
