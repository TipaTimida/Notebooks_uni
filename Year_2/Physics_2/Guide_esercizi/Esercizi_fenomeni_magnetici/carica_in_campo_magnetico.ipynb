### Regola mano destra

**Regola dela mano destra**

$i$: la direzione della corrente, è la direzione in cui si muovono i PROTONI  
se ho un elettrone, i è l'inverso della direzione del suo moto, cioè -v  

regola della mano destra:
1. pollice: i, qv (quindi se q<0, -v)  
2. indice: B
3. medio: F

**Regola della mano destra 2**

curl hands in direction of current i  
pollice punta in direzione di B indotto  

**Forza di Lorentz**

singola carica: $F=qv \land B$  
corrente (generalizzazione di molte cariche): $dF = i \, dl \land B$  

### Carica puntiforme si muove in campo magnetico

**Traiettorie**

di una carica puntiforme in una regione con i campi:  

1- solo campo B  
elicoidale (spirale attorno a linea B)  
chiamato moto di Larmor???  

2- solo campo E  
moto uniformemete accelerato nella direzione di E (il verso dipende dal segno della carica)  
se c'è anche la gravità, fa una parabola  

3- sia E che B  
dipende se sono ortognali o no  

**Teoria: sia E che B**

ho sovrapposizione di 3 moti:  

1. moto ciclotronico intorno a B 

2. drift/deriva (moto a velocità costante) lungo $E \times B$  
lungo l'asse in cui non c'è nè B nè E  

3. accelera lungo la direzione in cui ci sono entrambi i campi(o una loro componente non-0)
quindi solo se non sono ortogonali  
se sono ortogonali, fa moto rettilineo uniforme (cioè crea moto elicoidale)  

se invece E e B sono paralleli:  
fa un'elicoidale intorno a B, ma accelerata perchè c'è anche E  

nota: se E e B sono ortogonali  
non si vede l'accelerazione causata da E  
perchè c'è, ma viene continuamente girata dal campo magnetico e viene convertita in drift a velocità costante   

**Definizioni**

moto elicoidale:  
un'elica è una curva 3D che si avvolge attorno a un asse e avanza lungo di esso  
cioè cerchio (moto ciclotronico) + moto rettilineo  
accade quando c'è solo B, cioè fa moto circolare sul piano ortogonale a B, e rettilineo uniforme lungo B  

moto cicloidale:  
La cicloide è una curva 2D generata dal punto di una ruota che rotola senza slittare su una linea  
quindi nasce da rotazione + traslazione  
È un moto planare, niente asse verticale  
accade se ci sono sia E che B  

moto ciclotronico:  
è solo il moto circolare trasversale causato dal campo magnetico  
quindi non comprende anche il moto rettilineo che compie nel frattempo  

trocoide:  
Una trocoide è una curva piana generata da un punto fissato su un cerchio che rotola lungo una retta…
👉 ma non necessariamente sul bordo.  
se il punto è sul bordo → ottieni una cicloide

| Moto           | Geometria                               | Campi                       | Tipo di traiettoria     |
| -------------- | --------------------------------------- | --------------------------- | ----------------------- |
| ciclotronico   | cerchio                                 | solo (B)                    | circolare               |
| elicoidale     | cerchio + asse                          | solo (B), con (v_\parallel) | elica                   |
| cicloidale     | rotazione + traslazione (caso speciale) | (E \perp B)                 | cicloide                |
| **trocoidale** | rotazione + traslazione (generale)      | (E \perp B)                 | **trocoide (generale)** |


**Lavoro della forza di Lorentz**

per il prodotto vettoriale, F è perpendicolare a v, che è sempre parallela allo spostamento  
quindi forza è perpendicolare allo spostamento, non fa lavoro  

quindi cambia direzione della velocità, ma non modulo

### Guida

**Steps**

1. scrivi le mie info sui campi E e B   
può essere:   
    - un'espressiono per il campo, tipo $E=...$  
    - o il valore di ogni componente: $(E_x, E_y, E_z)$  

2. scrivi eq del moto: $m\ddot{r}=-qE-qv \times B$  
se ho espressione del campo: usa $\ddot{r}$  
se ho valore per ogni componente: separa in $\ddot{x}, \ddot{y}, \ddot{z}$, e attenta alle direzioni  

3. risolvi tutte le eq del moto  

**Risolvi le equazioni del moto**

1- campo ha dipendenza linare dalla coord (r, o quella che sia)  
cioè: $E = Ar$   
è moto armonico, riscrivi l'equazione come $\ddot{r}=-\omega^2 r$, vedi cosa è $\omega$  

2- campo è stazionario, cioè NON dipende dalla coordinata   
magari lo so perchè mi dice $E=...$, e la coord non appare  
scrivi $\dot{v}$ al posto di $\ddot{r}$ ed integra destra e sinistra per trovare $v$  

3- ho due equazioni non disaccoppiate, quindi dipende da una lungo un'altra direzione  

#### So componenti dei campi

**Situazione**

B è lungo asse z: $B=(0,0,B_z)$  
E è in piano xy: $E=(0, E_y, E_z)$  
calcola equazioni del moto di tutte le direzioni  

note:  
dove mettere gli assi lo scelgo io, sceglili in modo che sia così  

**Fissa assi cartesiane**

1- fissane una parallela a B  

2- se c'è anche E:  
fissa la seconda in modo che spanni il piano EB, così lungo la terza asse non c'è nè E nè B  

**Scrivere eq del moto**

devo mettere le forze $qE$ e $qvB$ in tutte le direzioni in cui servono  
prima metti qE in tutte le direzioni necessarie, poi aggiungi qvB  

$qE$: nelle direzioni in cui c'è E, ha sempre segno +   
usa la E di quella direzione  

$qvB$: nelle direzioni ORTOGNALI a B (quindi dove non c'è B), può avere segno + o -  
usa la B nell'unica direzione esistente, e la v della direzione mancante (nè quella di B, nè quella dell'asse su cui sono ora)  
per sapere segno, usa regola della mano destra (ricorda asse x a sinistra, y a destra, z in alto)  


**Esempio**

utile scrivere così:  
$E=(0, E_y, E_z)$  
$B=(0,0,B_z)$  
$v=(v_x, v_y, v_z)$  
$v \times B = (v_yB_z, -v_xB_z, 0)$  

$\begin{cases}
m\dot{v_x} = qv_yB_z \\
m\dot{v_y} = q(E_y -v_xB_z) \\
m\dot{v_z} = qE_z \\
\end{cases}$

#### Risolvere

**Risolvere**

fai in ordine di facilità  

1- eq che non dipende da nesuna v (qui la terza)  
separazione delle var (sposta dt dall'altro lato): 
$v_z(t)=v_z(0)+\frac{qE_z}{m} t$  

2- due equazioni accoppiate (rappresentano un ciclotrone + drift)  
disaccoppiale, risolvine una, poi plugga per trovare l'altra  

**2. Risolvere equazioni accoppiate**

come disaccoppiare:  
fai come con le equazioni di hamilton, che derivo per trovare l'equazione del moto   
1- deriva per t la più lunga, che ha sia E che B (così mi viene omogenea) (ricorda che solo v è una funzione del tempo, il resto sono costanti)  
2- in $\dot{v}$, pluggaci l'altra  
3- ho moto armonico, risolvila trovando $\omega^2$  

come risolvere la seconda, 2 metodi:  
1- come per la prima, ma inizio derivando l'altra  
2- prendi l'equazione con la derivata che non ho ancora trovato, e sulla destra pluggaci la soluzione della v che ho appena trovato, e risolvi  
quindi se mi manca $v_y$, prendo quella dove appare $v_y$, non $\dot{v}_y$ 

**Esempio: equazioni accoppiate**

$\begin{cases}
m\dot{v_x} = qv_yB_z \\
m\dot{v_y} = q(E_y -v_xB_z) \\
\end{cases}$

prima:  
derivo: $\ddot{v}_y = -\frac{q}{m}\dot{v}_xB_z$  
pluggo: $\ddot{v_y} = -\frac{q}{m}  (\frac{q}{m}v_yB_z) B_z = - \frac{q^2 B_z^2}{m}v_y$  
moto armonico: $ \omega = \frac{qB}{m}$  
soluzione: $v_y(t)=A \cos(\omega t+ \varphi)$  

seconda, metodo 1:  
$\ddot{v}_x = \frac{q}{m} \dot{v_y}B = -\omega^2 v_x +\frac{q^2}{m^2}E_x B_y = -\omega^2 (v_x -\frac{E_y}{B_z})$  
è disomogenea, quindi per soluzione sommo soluzione omogenea (come la precedente) e particolare (quando metto $\ddot{v}=0$)  
$v_x(t) = \frac{E_y}{B_z} + A \cos(\omega t + \varphi)$  
i parametri sono gli stessi di prima, perchè l'eqauzione omogenea è uguale a prima  

secondo, metodo 2:
pluggo: $m\dot{v_x} = qv_yB_z$

#### Moto elicoidale

**Cosa fare**

separa la velocità iniziale nella parte parallela e perpendicolare a B  
perchè ricorda che la particella fa una spirale intorno alla linea B  

**Trova R e T di spirale**

1- piano ortogonale a B: moto circolare   
(spirale schiacciata su un piano è un cerchio)  
uguaglia forza lorentz e centripeta per trovare R, il raggio di Larmor   
$F_L = F_C \quad \rightarrow \quad qvB =m\frac{v^2}{R}$  
ricorda che qui, $v$ è $v_\perp$  
poi posso trovare la velocità angolare: $\omega = \frac{v_\perp}{R}$  
(nota che $\omega$ NON dipende da $v_0$ nè R)  

2- lungo B: moto rellitilneo uniforme  
perchè lungo l'asse B non ci sono forze  
posso trovare il passo  (pitch, spostamento orizzontale dopo aver percorso un cerchio)  
$d=v_{\parallel} T$
dove $T = \frac{2 \pi}{\omega}$

#### Risolvere com metodo complesso

**Quando usare**

se uno dei campi si scrive come esponenziale  
tipo:  $E=E_0 \hat y \cos(\omega t)$    

nota:  
Nelle approssimazioni di moto non relativistico si può assumere kx = cost. = 0 nell'argomento del campo dell'onda  
cioè invece di $E=E_0 \hat y \cos(kx-\omega t)$, ho : $E=E_0 \hat y \cos(\omega t)$  

**Come si fa**

1- per ogni cosa dipendente dal tempo, sostituiscila con un tensore:   
$I(t) \rightarrow I_0 e^{-i \omega t}$   

RICORDA IL -, ma perchè? perchè deve cancellarsi con l'esponenziale di B, quindi se B ce l'ha, lo metto anche qui  

2- ho sistema lineare, risplvi per trovare ampiezza $I_0$  

3- torna alla soluzione iniziale con:  
$I_0 = A \rightarrow I(t)=A \cos (\omega t)$  
$I_0=iB \rightarrow I(t)= B \sin (\omega t)$  
perchè a me interessa la parte REALE   
se non è chiaro quale sia l parte reale, razionalizza il denominatore  

**Esempio**

1- scrivere equazioni del moto  
$\dot v_x = -\omega_c v_y$  
$\dot v_y = \omega_c v_x -(e/m) E_0 \cos (\omega t)$   

2- passare in forma complessa:  
$-i \omega \overline v_x = -\omega_c \overline v_y$  
$-i \omega \overline v_y = \omega_c \overline v_x -(e/m)E_0$  

3- risolvo:  
$\overline v_x =-A$  
$\overline v_y = -i B$  

4- torno a sol:  
$v_x = -A \, \cos (\omega t)$  
$v_y = -B \sin (\omega t)$  

**Spiegazione: passare all'esponenziale**

$\dot v = \omega_c v -(e/m) E_0 \cos (\omega t)$   

1- scrivo tutte le cose in forma esponenziale:  
$E(t)=Re[E_0 e^{-i \omega t}]$  
$v(t)=Re[\overline v e^{-i \omega t}]$  

2- le sostituisco tutte, e semplifico l'esponenziale, rimangono solo le ampiezze:    
$-i \omega \overline v= \omega_c \overline v-(e/m) \, E_0$  

3- poi, per tornare dall'ampiezza alla parte reale:  
riprendi la formula usata per trasformare:  
$v(t)=Re[\overline v e^{-i \omega t}]$   
se ho $v=A$, con A reale:  
$v(t)=Re[A e^{-i \omega t}]= A \cos (\omega t)$  
se ho $v=Ai$:  
$v(t)=Re[iB e^{-i \omega t}]= Im[B e^{-i \omega t}] = B \sin (\omega t)$


### Tipi di moto

**Effetto di B**

il campo B produce una forza di Lorentz nelle direzioni perpendicolari a quella di B (perchè cìè prodotto vettoriale)  
quindi il moto lungo B non è accelerato (a meno che in quella direzione ci sia anche E, che lo accelera)
