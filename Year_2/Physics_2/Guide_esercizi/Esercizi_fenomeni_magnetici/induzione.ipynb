### Motionally enduced emf

#### Usa induzione

**Cos'è**

fem causata dal fatto che la sezione immersa in B cambia (cambia S, non B)    

cioè nell'area serve una dipendenza dal tempo:   
o la scrivo con $t$ esplicito, oppure con una var dipendente dal tempo ($x(t)$ o $\theta(t)$)

**Casi**

traslazione di barra:  $dS = lv \, dt = l \ x(t)$    
rotazione di barra:  $dS = \frac{r^2}{2}d \theta(t)$ (area della parte di cerchio di raggio r sottesa dall'angolo $\theta$)  
rotazione di superficie: $B \cdot dS = B \, S \, \cos(\theta) = B \, S \, \cos(\omega t)$

come derivare:  
1- t esplicito: semplicemente fai derivata rispetto a t  
2- var che dipende da t: fai la chain rule, tipo: $\mathcal{E}=  -\frac{d\Phi}{d\theta}\frac{d\theta}{dt}$  
poi, a seconda dei dati del problema, ricorda che $\dot{x}=v$ e $\dot{\theta}=\omega$  
esempio: $\mathcal{E}  = -\frac{d}{dt}(B \, l \, x) = -Bl \dot{x}$

#### Forza magnetica

**Forza magnetica**

casi per calcolarla:  

1- no forze esterne, sistema lasciato evolvere   
allora è la forza di Lorentz  

2- con forze esterne, ad esempio per mantenere v costante  
2 modi, con conservazione energia o approssimazione dipolo  

**Forza magnetica: senza forze esterne**

$F = IlB$  

tip:  
si calcola solo sulle parti di circuito immerse in B  
quelle opposte si cancellano  

**Forza e momento magnetico**

forza magnetica potrebbe non sembrare quella di Lorentz, perchè è in una direzione sbagliata  
cerca momento per oggetto che ruota, forza per oggetto che trasla  

due modi per trovarli:  

1- conservazione energia: $P_{diss}=P_{magn}$    
$P_{diss}=RI^2$  
$P_{magn} = \tau \cdot \omega = F \cdot v$  
nella potenza magnetica, scegli la formula adatta per avere la forza o il momento  
perchè: se c'è $\omega$ o $v$ costante, non può accelerare, qualcosa deve bilanciare l'energia persa per dissipazione  

2- approssimazione dipolo/spira  
si può fare solo se l'oggetto è una spira  
$F = m\times (\nabla \times B)$  
$\tau = m \times B = mB \sin (\omega t)$  
B è il campo esterno, che arriva sulla spira e che produce la corrente indotta  

**Domande varie**

potenza della forza magnetica: $P=Fv$  
potenza disipata per effetto Joule: $P_{d}=RI^2$  
energia dissipata: $U_{diss}= \int P_{diss} dt = \int I^2(t)R \; dt$  
potenza media dissipata in un periodo:  $P_{avg}= \frac{1}{T} \int_0^T P \; dt$  

tensione efficace: $V_{eff} = \frac{V_{max}}{\sqrt{2}}$  
è così solo perché le grandezze in gioco sono sinusoidali  

#### Equazione moto e corrente

**Sistema (circuito con asta)**

$\begin{cases}
m\dot{v}=F  & legge \; Newton \\
 \mathcal{E} = L\dot{I}+RI &  equazione \; circuito \\
\end{cases}$  

note:  
1- I e v sono funzioni del tempo  
2- di solito, $F=IlB$, la forza di Lorentz  
3- poichè qui c'è la L a destra, $\mathcal{E}=Blv$ è la fem dvuta SOLO al campo esterno  

**Trova I(t)**

non mi serve l'eq di Newton, risolvi semplicemente quella del circuito  
risolvi con separazione delle var, viene esponenziale smorzata $\dot{I} = -\frac{I}{\tau}$  

**Trova v(t)**
 
1- isola I dall'eq di Newton (dalla forza di Lorentz) e pluggala in quella del circuito  
(se con induttanza, deriva anche I, così la pluggo in $\dot{I}$)  
2- ora risolvi per v, è di primo o secondo grado     
con induttanza: oscillatore armonico smorzato ($L\ddot{v}+R\dot{v}+\frac{B^2l^2}{m}v=0$)  
senza induttanza: esponenziale smorzato  

massima distanza percorsa:   
$x(\infty) = \int_0^\infty v(t) \, dt = v_0 \tau$  
perchè: $v(t)=v_0 e^{-t/\tau}$  


### Teoria

**Legge di Faraday-Neumann-Henry e di Lentz**

$\mathcal{E} = -\frac{d\Phi_B}{dt}$  

$\Phi_B$: flusso di B su una superficie S  
$\mathcal{E}$: ddp creata sul contorno di S  

da dove viene:  
è una legge di Maxwell (la 3), dove $\mathcal{E}= \int_l E_i \cdot dl$  
si usa se invece di sapere E, voglio solo la fem  

**Direzione: intuizione**

il flusso di B cambia  
allora si genera una fem indotta    
questa crea una corrente che si oppone alla variazione di flusso (questo è il significato del segno “-“)   
-> quindi produce un campo indotto in direzione opposta a quello iniziale  

cioè questa corrente si oppone al cambiamento che l'ha prodotta  
sennò entrambi aumenterebbero all'infinito, si crea energia infinita  

**Cosa vuol dire**

ho flusso di B variabile    
che genera E indotto  
che genera una fem indotta (la fem è generata solo dai campi indotti, non da quelli stazionari)  

come variare il flusso:  
induzione classica: posso variare modulo di B (cioè I), l'area o l'orientazione  
auto e mutua: deve per forza variare il campo B, perchè L e M sono costanti

#### Altri tipi

**Definizioni: tipi di induzione**

L: coefficiente di auto-induzione (induttanza)  
quanto un circuito si auto-influenza con il campo indotto che genera lui stesso  

M: coefficiente di mutua induzione  
quanto due circuiti si influenza tra loro con i campi indotto che generano a vicenda  
due circuiti che si influenzano sono detti "circuiti accoppiati"  

**Perchè posso definire L e M**

per la formula complessa per scrivere B, so che $B \propto I$  
quindi $ \Phi \propto B \propto I$, quindi cerco il coeff  

**Spiegazione: induttanza**

circola I -> si crea B indotto -> B si concatena a S  
il campo magnetico generato si concatena con la sua stessa superficie, quindi il circuito si autoinfluenza  
L mi dice quanto questo auto-influenzamento è forte  

L dipende solo dalla geometria dell'oggetto, come la capacità nel caso dell'elettricità  

**Spiegazione: mutua induzione**

due circuiti percorsi da corrente si influenzano l’uno con l’altro  
uno genera un campo magnetico, che si concatena con l’altro generando un flusso

nell'immagine: la corrente $i_1$ genera un campo magnetico, che si concatena con la superficie 2, generando un flusso  

**Quando posso trascurare induttanza**

di solito sì, perchè L è molto piccola  
ma a volte no, e lo vedo a posteriori:  

disegna il frafico di I(t) o V(t)  
se viene tipo un'onda quadra, quindi con parti verticali, allora in quei tratti ho $\frac{dI}{dt}=\infty$, cioè degenere  
quindi la caduta di potenziale $\mathcal{E}=L\frac{dI}{dt}$ dovuta all'induttanza non è sicuramente trascurabile, perchè per quanto L sia piccola, l'altro è infinito  
quando considero l'induttanza, mi smussa un po' l'onda I(t), non ha più angoli  
cosa che infatti non aveva senso, perchè non si può avere un cambio istantaneo di I, l'informazione ha bisogno di tempo per propagarsi  

**Note**

L e M hanno ha unità di misura H, Henry 

### Calcola L o M

**Come si fa**

usa le formule del flusso, vedi a cosa corrispondono L e M  

**Trova L di solenoide**

campo interno: $B=\mu_0\frac{N}{l}I$  
flusso concatenato: $\Phi_B = B \, S = B \, (N \pi R^2) = \mu_0\frac{N^2}{l}\pi R^2 \; I$  
la cosa moltiplicata a I è L    

**Trova L di toroide**

toroide è tipo una ciambella  
una fetta di ciambella può avere una sezione di qualunque forma  
assumo che abbia una sezione rettangolare  

campo interno: $B = \frac{\mu_0iN_S}{2 \pi r}$  

flusso:  
$\Phi_B = N \; \int_S B \cdot dS = N \; \int_S  \frac{\mu_0iN}{2 \pi r} \; h \; dr =  \frac{\mu_0iN^2 h}{2 \pi} \ln (\frac{b}{a}) $  
dove a,b sono i raggi interno ed esterno del solenoide  

**Trova M: Solenoide con dentro spira**

situazione:  
solenoide, con corrente I  
al suo interno, spira  

campo interno: $B = \mu_0 \frac{N}{l}I$  
flusso concatenato alla spira (o bobina, con B spire): $\Phi = N \, B \, S \cos \theta$  
dove S è area della bobina, B è campo del solenoide  

**Trova M: Spira dentro l'altra**

spira 2 è molto più grande di spira 1, quindi assumo che in spira 1 il campo $B_2$ sia uniforme
$\Phi_1(B_2)= \pi R_1^2 B_2 = \pi R_1^2 \frac{\mu_0 I_2}{2 R_2}$  
quindi: $M_{12} = \frac{R_1^2}{R_2}\frac{\pi \mu_0}{2}$  

### Trasformatore

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/Trasformatore.jpg)

**Struttura**

un oggetto formato da due bobine, avvolte attorno ad un nucleo ferromagnetico  
c'è un generatore collegato alla prima bobina  

circuito con generatore: “circuito primario", tensione $V_p$, spire $N_p$  
circuito senza generatore: "circuito secondario”, $V_s$, spire $N_s$    

**Funzionamento**

1- bobina 1 genera un campo magnetico B  
2- B segue il materiale ferromagnetico e non esce (tutto B rimane confinato nel materiale ferromagnetico)  
2- B arriva al circuito secondario, si concatena con la bobina e genera una tensione indotta (fem)  

regolando il numero di spire delle due bobine, posso cambiare il valore della tensione  
si capisce bene guardando le formule  

**Formule**

1- scrivo la legge di Lentz per le due bobine:  
$V_p = -\frac{d\Phi_p}{dt} = -N_p \frac{\Phi_1}{dt}$  
$V_s = -\frac{d\Phi_s}{dt} = -N_s \frac{\Phi_1}{dt}$  
$\Phi_1$: flusso concatenato con una singola spira, ha valore uguale per entrambi i circuiti  

2- faccio il rapporto:  
$\frac{V_s}{V_p}= \frac{N_s}{N_p}$  

**Nota**

il trasformatore funziona solo con la corrente variabile  
perché se non è variabile il flusso non varia nel tempo e quindi non si crea induzione
