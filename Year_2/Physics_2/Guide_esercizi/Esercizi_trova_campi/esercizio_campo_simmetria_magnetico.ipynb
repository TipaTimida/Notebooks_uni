**Legge di Biot-Savart**

per filo:  
$B = \frac{\mu}{4 \pi} \int_\gamma I(r) \frac{dl \times r}{r^3}\quad $ or $\int \frac{dl \times \hat{r}}{r^2}$

per oggetto con un volume:  
$B = \frac{\mu}{4 \pi} \int_\gamma J(r) \frac{dV \times r}{r^3}\quad $

filo: $I \, dl$  
volume: $J \, dV$    

$\mu_0$: permeabilità magnetica del vuoto  
l'integrale è su una linea chiusa  

#### Usando legge di Biot-Savart

Se c’è una BOBINA, cioè un oggetto formato da N spire, il campo magnetico in un punto è dato dalla stessa formula, moltiplicata però per il numero di spire N   
perché ci sono N spire che generano campo magnetico nel punto P quindi bisogna sommare tutti i contributi

![alt text](https://i.sstatic.net/uUw9p.jpg)

dati:  
$\theta _1, \theta_2$: gli angoli nel punto P, quelli a destra nell'immagine e chiamati $\Phi$, non quelli chiamati $\theta$  
$l$: lunghezza del filo in direzione y  
$x$: distanza orizzontale tra filo e P (OP nell'immagine)   
r: per ogni punto del filo, distanza tra quel punto e P (è una linea diagonale)

**Filo infinito**

let: $l=x \tan \theta$  
$dl = x \sec^2 \theta \, d\theta$

$B = \frac{\mu_0 i}{4 \pi}  \int_{-\theta_1}^{\theta_2}\frac{dl \, \cos \theta}{r^2} = \frac{\mu_0 i}{4 \pi}  \int_{-\theta_1}^{\theta_2}\frac{x \sec^2 \theta \, d\theta}{\frac{x^2}{\cos^2 \theta}} \cos \theta = $  
$=  \frac{\mu_0 i}{4 \pi x}  \int_{-\theta_1}^{\theta_2} \cos \theta \, d \theta = \frac{\mu_0 i}{4 \pi x}  \int_{-\theta_1}^{\theta_2} [\sin \theta ]^{\theta_2}_{-\theta_1} = \frac{\mu_0 i}{4 \pi x} (\sin \theta_1 + \sin \theta_2)$  

oppure: $\frac{\mu_0 i}{4 \pi x} (\cos \alpha_1 + \cos \alpha_2)$ se $\alpha$ sono gli anfoli indicati come $\theta$ nell'immagine, quelli attaccati al filo  

**Spira/loop circolare**

R: raggio spira  
x: distanza orizzontale da centro spira a punto che voglio  
lungo l'asse:  

$B = \frac{\mu_0 i}{4 \pi}\int \frac{dl \cos \theta}{r^2} =  \frac{\mu_0 i}{4 \pi} \frac{1}{R^2+x^2} \frac{R}{\sqrt{R^2+x^2}} \int dl =   \frac{\mu_0 i}{4 \pi} \frac{1}{(R^2+x^2)^{3/2}} 2 \pi R =  \frac{\mu_0 i R^2}{2 (R^2+x^2)^{3/2}}$  

**Centro di un arco**

$\theta$: che angolo è ampio l'arco  

$B = \frac{\mu_0 i R^2}{2R^3} \frac{\theta}{2 \pi} = \frac{\mu_0 i \theta}{4 \pi R}$
