**Recap continuità**  

V: deve essere sempre continuo  

E: 
tangenziale: sempre continuo  
normale: può avere discontinuità SOLO SE c’è carica superficiale  
quindi in una sfera con carica uniforme è continuo, in una cava no  

### Teoria

**Teoremi**

definizione di flusso:  
$\Phi_S(E)= \int_S \vec{E} \cdot d\vec{S}$  

teorema di Gauss:  
$\Phi_S(E) = \frac{1}{\epsilon_0} \int _V \rho \; dV= \frac{q_{in}}{\epsilon_0}$  

teorema della divergenza:  
$\Phi_S(E) = \int \vec{\nabla}\cdot E \; dV$

prima equazione di Maxwell: uguaglio th di gauss e della divergenza  
utile in 1D:  
$\nabla \cdot E = \frac{\rho(x)}{\epsilon_0}$

**Spiegazioni**

c'è flusso?  
a occhio, per vedere se c'è flusso o no, vedi se le linee di E infilzano S  
se no, non c'è flusso  

cosa sono:  
$S$: superficie chiusa (quindi deve racchiudere un volume)    
$\Phi_S(E)$: flusso del campo elettrico concatenato ad S  
$Q_{in}$: la carica racchiusa nel volume DENTRO S (ignora le cariche esterne al volume racchiuso in S)    

**Perchè posso ignorare le cariche esterne ad S?**

perchè sto considerando il flusso  
un carica esterna ha un campo che entra ed esce da S, quindi in totale il flusso è 0  
una interna ha flusso che o entra o esce, c'è uno sbilanciamento, quindi contribuisce al flusso  

**Equazioni locale e non**

Maxwell: locale -> $\nabla \cdot E$ e $\rho$ vanno calcolati in un punto preciso  
Gauss: non locale -> considero Q compreso in tutta S, non in punto preciso  

se Q cambia nel tempo: usa forma locale  
Gauss non va bene, perchè una cambiamento istataneo di Q causa un cambiamento instantaneo di E in tutta la regoine S  
ma non può essere, l'informazione sul cambiamento viaggia alla velocità della luce, non è istantanea  

se E non è differenziabile: usa forma non locale  
perchè per scrivere $\nabla \cdot E$ devo essere in grado di differenziare E  
non ci riesco se ad esempio ci sono salti in E, come nella superficie di contatto tra diversi dielettrici  

### Calcola campo elettrico

**Steps**

1. scegli superficie S    

2. controlla:  
   - le orientazioni dei vettori E ed S  
   - il modulo di E quando tocca S  

3. calcola $\Phi$  

4. calcola $Q_{in}$, la carica racchiusa  

5. scrivi la legge di Gauss  

le soluzioni di $Q_{in}$ e $E$ si scrivono come sistemi  

**Se ci sono dielettrici diversi**

il teorema diventa (f è per free, carica libera):    
$\int D \cdot dS = Q_f$    

separa la superficie S in tante superfici, ognuna con fuori un dielettrico diverso (epsilon diversa)  
e calcola separatamente l'integrale su ogni superficie  

**Scegliere la superficie**

1. deve essere una superficie chiusa (di solito sfera o cilindro)  

2. deve passare per il punto dove voglio calcolare il campo  

3. deve rispettare le properietà di simmetria del sistema  
in modo che il vettore E che tocca la superficie S abbia stesso modulo in ogni punto della superficie  
in pratica, cerca di fare in modo che S sia equipotenziale  
di solito è una forma concentrica con la cosa che voglio studiare  
eg. E ha simmetria sferica, prendo una superficie sferica  


nota che se chiede E sia dentro che fuori, devo farlo 2 volte scegliendo due superfici  
di solito hanno stessa forma, ma una è più piccola e una più grande  

**Controlla le orientazioni**

se E ed S hanno orientazione rispettiva diversa in punti diversi, tipo in qualche parte sono paralleli e in altre perpendicolari:  
separo l'integrale in una somma di integrali   
ognuno deve essere fatto su una superficie su cui S ed E devono avere stessa orientazione rispettiva  

ricorda che $\vec{S}$ ESCE dal solido racchiuso, punta in fuori  

**Calcola la carica interna**

$q_{int} = \int_{V_{int}} \rho \; dV$  
$\rho$: densità di carica volumetrica    

uniforme: non cambia nel tempo  
costante: non cambia nello spazio (qui è la parola sbagliata)  

se $\rho$ è uniforme:  
la porto fuori dall'integrale: $q_{int}=\rho \int dV= \rho V$   
alla fine posso sotituire: $\rho = \frac{q}{\frac{4}{3}\pi R^3}$

se $\rho$ NON è uniforme (di solito è radiale, cioè $\rho(r)$):  
DEVO usare la forma con l'integrale  
per una sfera, $dV = 4 \pi r^2 \; dr$

nota:  
se so la densità $\rho$: moltiplicala per il volume dell'oggetto della dimensione considerata  
se ho la carica totale $q$: moltiplicala per il rapporto tra i volumi dell'oggetto attuale e l'oggetto completo  

**Controlla il modulo**  

se E ha sempre lo stesso modulo quando tocca la superficie, posso portarlo fuori dall'integrale:    
$\int E \cdot S = E  \cos \theta \int dS$



#### Teorema di sovrapposizione

**Quando**

quando ci sono:  

- più oggetti carichi  
eg. due piani paralleli, a formare un conedensatore  

- un oggetto con un buco difficile  
eg. una sfera con una cavità sferica non concentrica  
o cilindro con cilindro cavo dentro, non concentrico  

quando scrivo E è meglio tenermi la densità, non trasformarla in q  

**Steps**

gli oggetti sono semplici, tipo un piano  

1. calcola E di ogni oggetto, come se fosse isolato  
se serve, distingui tra E interno ed esterno, con un sistema

2. dividi lo spazio in regioni dove il campo di scrive allo stesso modo  
eg. 2 sfere che si intersecano individuano 4 zone: dentro entrambe, dentro 1 ma non 2, dentro 2 ma non 1, fuori da entrambe  

3. per ogni zona, somma VETTORIALMENTE i campi di ogni oggetto (scegli le formule dei campi giusti per quella zona)  
specifica il vettore direzione e, dal disegno, cerca di veder se hanno qualche rapporto (somma/differenza) 

**Somma vettoriale campi**

quando scrivo il campo, specifica il vettore direzione (o versore)  
dal disegno, cerca di vedere se questi vettori hanno qualche rapporto (somma/differenza)  

ricorda: in un sfera, r è la distanza tra il CENTRO della sfera e il PUNTO dove voglio sapere il campo    
di solito avrò oggetti chiamati $r_1, r_2, d$, dove d è la distanza tra i centri delle due sfere  

ricorda: $r_1-r_2 =r_1 + (-r_2)$, gira il secondo e somma testa-coda  

**Nota: cariche**

in un piano, il segno di $\sigma$ è il segno di $q$:    
$\sigma >0$: E uscente  
$\sigma <0$: E entrante  
per questo in un condensatore, che ha i piani caricati di segno opposto, il campo elettrico dentro raddoppia e fuori si annulla  

![text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTaBpDtInDy7s7ItsEd9c8Z4jqPGgL4vvDMvg&s)

**Oggetto con buco**

prima di iniziare gli steps, devo individuare gli oggetti  

esempio:  
ho un oggetto con densità $\rho$ uniforme ovunque, eccetto che nella cavità, dove $\rho=0$  

separalo in 2 oggetti: 

1. oggetto pieno, con densità uniforme $\rho$

2. oggetto grande quanto la cavità, con densità uniforme $-\rho$  
le densità hanno modulo uguale e segno opposto, così in totale la cavità non ha densità  
 
ricorda che cambiare il segno di $\rho$ cambia il verso del campo, uscente/entrante  

**Esempio: sfera con cavità non concentrica**

in sfera piena isolante:  
interno: il campo dipende linearmente dalla distanza dal centro  
esterno: come carica puntiforme  

campo generato dalla sfera intera (dentro):  
$E_1=\frac{\rho}{3 \epsilon_0}r$  

campo generato dalla cavità:     
al suo interno:  
$E_2 = \frac{- \rho}{3 \epsilon_0}r'= \frac{- \rho}{3 \epsilon_0}(r-d)$  
al suo esterno:  
$E_3 = k\frac{q}{|r-d|^3}(r-d)$  

il vettore deve essere preso dal centro della sfera che considero fino al punto P  

quindi nella cavità:  
$E= E_1+E_2 = \frac{\rho}{3 \epsilon_0}d$  
cioè è costante  
e diretto lungo la congiungente dei centri  

mentre fuori nella cavità viene   
$E = E_1+E_3$, che è complicato  


**Campo esterno**

se c'è un campo esterno $E_0$ (spesso uniforme), in tutte le zone devono sommare VETTORIALMENTE anche questo  

**Come scrivere superfici**

sfera: $\int dS = 4 \pi r^2$, $\quad $dS= 4 \pi r^2 \; dr$    
semisfera: $\int dS = 2 \pi r^2$ (metà della sfera)  
cilindro: $\int dS = 2 \pi r \, h$, $ \quad dS = 2 \pi r \; dz$ (considera la superficie laterale) (se serve, devo sommare quella circolare delle absi a parte)   
semicilindro (separa lungo il diametro della base): $\pi r h$ (metà della circonferenza)  

#### Calcola potenziale

calcolo V dopo aver calcolato E  
E potrebbe essere distinto tra esterno ed interno  

**Si può fare?**

se so il potenziale in almeno un punto R: 
posso trovare il potenziale in ogni punto r  
$V(r) = V(R) - \int^r_R E \cdot dr$   
$V(r) = V(\infty) - \int_\infty^r E \cdot dr$  


se non so il potenziale in nessun punto:  
non posso trovare il valore del potenziale in nessun punto, solo la differenza di potenziale tra due punti  
$V(B)-V(A)=-\int_A^B E \, dr$ 

altro modo per pensarlo:  
$V = -\int E \,dr$  
quando integro mi viene il +c, quindi mi serve sapere una condizione (il potenziale in un posto)  
trovo +c usando la condizione di quando r ha quel valore  

**Esempi di potenziale in un punto**

se il campo E è diviso in interno ed esterno:  

- esterno: il potenziale all'infinito è 0, $V(\infty)=0$   

- interno: continuità con quello esterno sulla superficie di separazione  
perchè il potenziale è continuo  
non posso dire che all'infinito è nullo, perchè r infinito rientra nel caso esterno  
oppure, tipo in un cilindro, potrei mettere il potenziale nell'asse centrale a 0?

può capitare, come nel filo infinito, che non so il potenziale in nessun punto  

posso mettere il potenziale nullo a infinito solo nelle CARICHE LIMITATE NELLO SPAZIO  
quindi ok una sfera finita, ma non un filo infinito  

### Esempi

#### Sfera

**Cosa ho**

sfera di raggio R  
di diversi tipi  

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/E_superf_sferica.jpg)

**Scegliere S**

uguale per tutti i casi  

SUPERFICIE: sfera concentrica alla mia sfera  
considero separati i casi di $r>R$ e $r<R$  
orientazione: E ed S sempre paralleli  
modulo: E che tocca la superficie ha sempre lo stesso modulo    

$\Phi = \int E \cdot dS = E \int dS = E \, (4 \pi r^2)$  

**Calcolare carica interna**

dipende dal tipo di sfera:  

cava / piena conduttrice: la carica è solo sulla superficie della sfera    
$Q_{in} = \begin{cases}
q & r>R \\
0 & r<R 
\end{cases}$ 

piena isolante: la carica è uniforme in tutto il volume della sfera  
interno: $q_{int}=\int \rho dV= \rho \; \frac{4}{3}\pi r^3 = \frac{q}{\frac{4}{3}\pi R^3} \; \frac{4}{3}\pi r^3 = q\frac{r^3}{R^3} $  
$Q_{in} = \begin{cases}
q & r>R \\
q\frac{r^3}{R^3} & r<R 
\end{cases}$ 

campo:  
$E = \begin{cases}
k\frac{q}{r^2} & r>R \\
\frac{\rho}{3 \epsilon_0}r & r<R \\
\end{cases}$

E isolante detro: $E = \frac{\rho}{3 \epsilon_0}r$  

**Cosa fare**

eguaglio:  
$\Phi = \frac{q_{int}}{\epsilon_0}$  
attenta a scegliere q del tipo giusto  

**Potenziale**

esterno (uguale in entrambe): 
conosco E, e so che V all'infinito è nullo:  
$V = -\int E \,dr$  
uso E del campo esterno, ottengo $V = k \frac{q}{r}$  

interno:  
conosco E interno, e conosco il potenziale all'estremità della sfera  
quindi impongo la continuità del potenziale  

cava:  
$V = -\int E \, dr =  \int 0 = +c$  
quindi V è costante, e poichè deve essere continuo con quello esterno, deve essere uguale a quello sulla superficie  
$V = k \frac{q}{R}$ 

piena isolante:  
$V(r)-V(R) = - \int_R^r E \cdot dr = \frac{\rho}{3 \epsilon} \int_r^R d \, dr =  \frac{\rho}{3 \epsilon} (R^2-r^2)$  
$V(r) = \frac{\rho}{3 \epsilon} (R^2-r^2)  + V(R) = \frac{\rho}{3 \epsilon} (R^2-r^2) + \frac{\rho R^2}{3 \epsilon} = \frac{\rho}{6 \epsilon} (3R^2-r^2)$  

**Sfera con cavità centrale**

c'è $\rho$ uniforme solo in $R_1 < r < R_2$  
usa sempre la stessa superficie, quindi S è uguale  
ma attenta a calcolare la carica interna, cambia in ogni caso:  
devo moltiplicare densità e volume racchiuso, tipo:    
$q_{int}=\rho (\frac{4}{3}\pi r^3- \frac{4}{3}\pi R_1^3)$  

#### Piano infinito

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/E_piano_gaus-1.jpg)

**Campo elettrico**

SUPERFICIE: cilindro (o scatola con facce parallele al piano?)  
orientazioni: diverse, quindi considero sepratamente le superfici delle basi e quella laterale curva  
modulo di E: indipendente dalla distanza dal piano   

$\Phi = \int E \cdot dS = \int_{S_1} E \cdot dS +\int_{S_2} E \cdot dS +\int_{S_3} E \cdot dS = ES_1 + ES_2 = 2ES$  
$q_{int} = \sigma S$

$2E = \frac{\sigma}{\epsilon}$  


**Campo elettrico: conduttore**

simile, ma il lato del cilindro dentro il conduttore/condensatore ha E=0  
perchè dentro un conduttore non c'è campo elettrico  
quindi a sinistra ho E, non 2E  

#### Cilindro infinito / filo infinito

attenta: vale solo per cilindro INFINITO in altezza  
sennò succedono cose strane ai bordi, non so come è diretto E  

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/E_cilindro.jpg)

**Campo elettrico**

SUPERFICIE: cilindro concentrico  
orientazione: diversa, separo le superfici  
modulo di E: su quella laterale, uguale su tutta la superficie  

$\Phi = \int E \cdot dS = \int_{S_1} E \cdot dS +\int_{S_2} E \cdot dS +\int_{S_3} E \cdot dS = 0+0+ES_3 = E \; 2 \pi r h$  
$q_{int} = \int \rho dV = \rho \int dV = \rho  \; 4 \pi R^2 \, h = \lambda h $
ho fatto $\lambda = \rho \; 4 \pi R^2$  

$E \; 2 \pi r h = \frac{\lambda h}{\epsilon}$  



![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/V_cilindro.jpg)

**Potenziale**

non so il potenziale in nessun punto, perchè all'infinito non è nullo  
quindi non posso trovare il valore in nessun punto, solo la differenza di potenziale tra due punti  

$V(B)-V(A)=-\int_A^B E \, dr = ... = -\frac{\lambda}{2 \pi \epsilon} \ln (\frac{r_B}{r_A})$

#### Recap

**Tipi di campo**

oggetto | formula | ricavare | disegna
---|---|---|---
filo | $E=2k\frac{\lambda}{r}$ | $\epsilon_0 (E 2r \pi l) = \lambda l$   | cilindro intorno al filo  
piano | $E=\frac{\sigma}{2 \epsilon_0}$ | $\epsilon_0 (2 E S) = \sigma S$  | cilindro perpendicolare al piano
condensatore | $E=\frac{\sigma}{\epsilon_0}$ | $\epsilon_0 (E S) = \sigma S$  | cilindro con un'estremità fuori, una dentro il condensatore


**Come trovare formula tipo di campo**

$q$ diventa $\lambda l$ o $\sigma S$ a seconda che l'ggetto abbia 1 o 2 dimensioni  

$S$ devo considerare solo la parte parallela ad $E$, non perpendicolare, perchè c'è il prodotto scalare  
filo: solo le superficie laterale del cilindro intorno al filo, non le basi  
nel piano devo considerare entrambe le basi del cilindro, nel condensatore solo quella fuori dal condensatore, perchè dentro $E=0$  


### Dimostrazione th Gauss

**Caso 1:  1 sola carica dentro la superficie S**  

S ha una forma qualunque  
contiene una carica q  

$d \Phi = E \; dS \cos \theta  = k\frac{q}{r^2}dS \cos \theta$  

E ed S non per forza sono parallele  
ma $dS \cos \theta$ è la proiezione di dS sulla sfera di raggio r centrata in q  
perchè $\frac{dS \cos \theta}{r^2} = d \Omega$, dove Omega è l'angolo solido  
quindi:  
$d\Phi = kq d\Omega$  

poi $\int \Omega = 4\pi$ perchè devo girare due volte un angolo completo di $2 \pi$, in direzioni perpendicolari, per creare una sfera  

$\Phi = kq \int \Omega = kq 4 \pi = \frac{q}{\epsilon}$   

**Caso 2: più cariche interne**

usa il principio di sovrapposizione dei campi  

**Caso 3: solo una carica esterna q**

disegna un cono che parte da q, di angolo $d\Omega$, che entra ed esce nella regione S, quindi intercetta il bordo 2 volte    
devo dimostrare che il flusso è nullo, cioè che $\int E(r_1) \cdot dS_1 = - \int E(r_2) \cdot dS_2$  
cioè che il flusso dipende solo dall'angolo solido, quindi quello che entra ed esce è uguale

### Problema in 1D

posso usare il teorema di Gauss in forma differenziale  

forma differenziale (in 1D)  
$\nabla \cdot E = \frac{\partial E_x}{\partial x}= \frac{\rho(x)}{\epsilon_0}$

cioè, nelle regioni in cui:  
$\rho = 0$: E non cambia  
$\rho=const$: E varia linearmente  
ricorda di imporre la continnuità tra E di tutti i pezzi, e trova un pezzo dove lo conosco, tipo E=0 a $x=-\infty$  

### Note

**Se ho q=0**

se l'oggetto NON ha simmetria sferica:  
mi dice solo che il flusso è nullo: $\Phi =\frac{q}{\epsilon}=0$    
ma non mi dice E=0, non è detto, so solo che tanto campo esnrta quanto ne esce    

se l'oggetto ha simmetria sferica:  
posso scrivere: $\Phi=E \int dS$   
perchè E è uguale in ogni punto, se scelgo come S una superficie sferica   
poichè dS non può essere nulllo, allora E=0
