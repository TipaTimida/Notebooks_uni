**Cosa usare**

Gauss:  
- oggetti finiti SENZA BORDO: sfera  
- oggetti con una dimensione infinita: cilindro, piano, filo infinito  

simmetria:  
- oggetti finiti CON BORDO: filo finito, anello, disco  

perchè con Gauss assumo di sapere come è diretto E  
ma non so come funziona vicino al bordo, quindi lì devo usare la simmetria  

### Guida  

**Steps**

1. per sfruttare la simmetria, prendi due pezzi dx opposti, vedi che succede alle componenti dE  
una componente si annulla: $E = \int dE_y = \int dE \cos \theta$  
una componente raddoppia: SE voglio, metti coeff 2, cambia estremi di integrazione  

2. scrivo l'integrale: $E = \int dE = \int k \frac{dq}{r^2} $  
se prima c'era una trig, rimane  

3. cambio variabile di integrazione  
dq non è buona, voglio dx, dS o dV, le ottengo dalle formule di densità di carica elettrica  

4. esprimo le lunghezze utili  
mi serviranno per sostiutire nelle trig  

5. sostituisco tuttoo e calcolo l'integrale  
attenta a scegliere bene gli estremi di integrazione, sono il massimo e minimo valore che la variabile di integrazione può assumere  
a volte, usa il limite all'infinito

**Cambio var di integrazione: densità di carica elettrica**

$dq = \lambda \; dl$  
$dq = \sigma \; dS$  
$dq = \rho \; dV$  

### Filo finito, a metà tra i capi

![alt text](https://i.sstatic.net/UotfW.png)

1- Per sfruttare la simmetria  
prendo due pezze dx equidistanti dal centro  
parte parallela al filo di dE si cancella, ho solo quella perpendicolare   
$E = \int dE_y = \int dE \cos \theta $  

2- Scrivo l'integrale  
$E = \int k\frac{dq}{r^2} \cos \theta$  

3- Cambio variabile di integrazione, non dq  
$dq = \lambda \, dy = \frac{q}{l} dy$  
(carica totale e lunghezza totale del filo)  

4- Esprimo le lunghezze utili  
$r^2=x^2+y^2$  
$\cos \theta = \frac{x}{r}$  

5- Calcolo  
$E(x) = \int _{-l/2}^{l/2} k \frac{\lambda}{x^2+y^2} \frac{x}{\sqrt{x^2+y^2}} dy = ... = 2k \frac{q}{x} \frac{1}{\sqrt{l^2+4x^2}}$


**Per passare da finito a infinito**

faccio un limite:  
$\lim_{l\rightarrow \infty} E$   
$\lim_{l\rightarrow \infty} 2k \frac{\lambda l}{y} \frac{1}{\sqrt{l^2+4y^2}} = 2k \frac{\lambda l}{y} \frac{1}{\sqrt{l^2}} = 2k \frac{\lambda l}{y} \frac{1}{l} = 2k \frac{\lambda}{y}$  

ricorda all'inizio di esplicitare tutti i punti in cui appare $l$, la cosa che va all'infinito, per esempio nella densità di carica  
poi uso comportamento asintotico  

**Integrare sugli angoli**

potrei fare un altro var change, per passare da dx a $d\theta$  

$y = x \tan \theta$  
$dy = \frac{x}{\cos ^2 \theta} d \theta$  
$r = x \cos \theta$  

### Anello

![alt text](https://miro.medium.com/v2/resize:fit:1400/0*bOjlcYzeByHet88X.png)

**Anello: usando dl**

1- Uso la simmetria  
prendo due punti opposti dell'anello  
E perpendicolare all'asse si annulla, rimane solo parallelo  
$E = \int dE_x =  \int E \cos \theta$  

2- Scrivo l'integrale  
$E = \int k \frac{dq}{r^2} \cos \theta$  

3- cambio variabile di integrazione:  
$dq = \lambda \; dl = \frac{q}{2 \pi R}dl$  

4- scrivo lunghezze utili  
$r^2 = x^2+R^2$  
$\cos \theta = \frac{x}{r}$  

5- calcolo  
$E = \int_{0}^{2 \pi R} k \frac{q}{2 \pi R} \frac{1}{x^2+R^2} \frac{x}{\sqrt{x^2+R^2}} dl = \frac{kq}{2 \pi R} \int_0^{2 \pi R} \frac{x}{(x^2+R^2)^{3/2}} \, dl$  

$E = kq\frac{x}{(x^2+R^2)^{3/2}}$  
$V = kq \frac{1}{\sqrt{x^2+R^2}}$  

**Anello: usando $d\theta$**

non confondere gli angoli:  
$\theta$: quello che nell'immagine è chiamato $\theta$, tra E e l'asse  
$\alpha$: è sul piano del l'anello, tipo lancetta dell'orologio, indica quale pezzp dq sto considerando  
di solito si usano al contrario, ma così è coerente con prima  
 
cambio di variabile di integrazione:  
$dq = \lambda \; dl = \lambda \; R \, d\alpha$  

così ottengo:  
$E = k \int_0^{2 \pi} \frac{\lambda \; R \; d\alpha}{r^2}\cos \theta$

nota che r non dipende dall'angolo $\alpha$, perchè tutte le coppie dq sono slla stessa distanza  
quindi posso portarlo fuori  
meglio anche esprimerlo in termini di lunghezze  

$E = \frac{k \lambda R \cos \theta}{r^2} \int_0^{2 \pi} d\alpha = \frac{k \lambda R \cos \theta}{r^2} 2\pi$  

ora sostituisco:  
$\cos \theta = \frac{x}{r}$  
$\lambda = \frac{q}{2 \pi R}$  
e appare la stessa formula di prima  

nota integrazione:  
integro da 0 a $2 \pi$  
il fatto che la componente perpendicolare all'asse si annulla vuol dire che devo moltiplicare per la trig  
ma non devo necessariamente mettere il coeff 2 e dimezzare l'intervallo di integrazione, posso anche lasciare così  
il ragionamento sul prendere una coppia di dq, così una componente si annulla, è solo per notare che si annulla  
non vuol dire che quando integro devo farlo per coppie

**Disco ???**

divido il disco il anelli concentrici, di spessore dr e carica dq   
di un anello:  
$dE_{a} =  k\frac{x \; dq }{(x^2+R^2)^{3/2}}$   

cambio variabile di integrazione:  
$dq = \sigma \, dS = \frac{q}{\pi R^2} dS$  
$dS = 2 \pi r \, dr$  

integro, con estremi per r: $[0, R]$  

**Disco ???**

divido il disco il anelli concentrici, di spessore dr e carica dq   
$dA = 2\pi r \, dr \, z^2$   
$z = \sqrt{r^2+x^2}$  
$\cos \theta = \frac{x}{z}$  
$E= k\int_0^R \frac{\sigma dA}{z^2} \cos \theta = \frac{\sigma}{2\epsilon_0} (1-\frac{z}{\sqrt{x^2+R^2}})$  


**Piano infinito**

è un disco con $R \rightarrow \infty$   
faccio il limite alla formula per E di un disco  

$E = \frac{\sigma}{2 \epsilon_0}$  
quindi campo di un piano infinito NON dipende dalla distanza del punto P dal piano  

### Trova potenziale

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/V_filo.jpg)

**Come fare**

come in esercizio di Gauss, mi serve conoscere formula di E e il valore di V in un punto  

**Potenziale tra due punti, in presenza di campo di un filo infinito**

campo elettrico di filo infinito: $E = 2k \frac{\lambda}{r}$  

$V(B)-V(A) = -\int_A^B E\cdot dr = -\int_{r_A}^{r_B}  2k \frac{\lambda}{r} = -2k \lambda \ln (\frac{r_B}{r_A})$  

devo usare la formula con l'integrale, non solo $V=Ed$, perchè E cambia in ogni punto  

all'infinito il potenziale è infinito, non nullo, quindi non conosco V in alcun punto  
quindi devo esprimere il potenziale come differenza tra due punti, e non come funzione $V(r)$, non posso trovare nessun valore  

**Anello, potenziale sull'asse**  

$V = \int dV = \int k \frac{dq}{r} = \frac{k}{r} \int dq =  \frac{k}{r}q = k\frac{q}{\sqrt{c^2+R^2}}$  
so che per x infinito V diventa nullo  

lo ricalcolo da capo, come per campo elettrico  


![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/V_disco.jpg)

**Disco**

$V = \int dV$  
$dV = k \frac{dq}{\sqrt{x^2+r^2}}$  

$dq = \sigma dS = \sigma w \pi r \; dr$  

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/V_piano.jpg)

**Piano infinito**

forza di Coulomb è conservativa, quindi separo il tragitto in due parti, una perpendicolare e una parallela al piano  
c'è prodotto scalare, quindi quella perpendicolare si cancella, perchè cos(90)=0

integro:  
$V(B)-V(A) = -\int_A^{B'} E \cdot dl -\int_{B'}^B E \cdot dl = -{r_a}^{r_B} \frac{\sigma}{2 \epsilon_0} dx - 0 = \frac{\sigma}{2 \epsilon_0} (x_A-x_B)$
