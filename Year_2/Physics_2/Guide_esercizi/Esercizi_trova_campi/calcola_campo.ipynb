**Casi: da cosa è creato il campo**

1- da una carica/corrente libera  
di cui so la disposizione  
come: legge di Gauss o Ampere  

2- dalla variazione dell'altro campo (E o B)  
di cui so la formula  
come: legge di Faraday o Ampere  

3- da carica/corrente di polarizzazione  
dipende da dove viene la polarizzazione    
cosa: elettrete (so direzione di P o M permanente)   
come: calcola $K_b$ o $\sigma_b$, trova un'analoga distribuzione di cariche libere  
cosa: soggetto a campo esterno  
come: imposta il sistema, fai ansatz su P parallelo a campo esterno  

**Problemi avanzati**

1- oggetto con buco  
piccolo (trascurabile): condizioni di raccordo  
grande: sovrapposizione dei campi   

2- so campo locale in una zona, voglio saperlo in una'altra (con dielettrici diversi)  
come: condizioni di raccordo  

3- voglio E, ma so I  
come: trova Q con  $Q = \int I(t) \; dt$

4- voglio B, ma so Q (e il suo moto)  
come: trova I con $I = \lambda v$ o $K=  \sigma v$  

### Creato da cariche/correnti libere

#### Creato da correnti o cariche libere

**Come**

campo elettrico: usa teorema di Gauss   
$\int_S E \cdot dS = \frac{q}{\epsilon_0}$  

campo magnetico: usa th di Ampere  
$\int_\gamma B \cdot dl = \mu_0 i$  

problema: se ci sono dielettrici diversi  
come: usa le versioni con D e H  
così intanto calcolo il campo ausiliario, poi lo modifico in goni zona  

##### Esempio

**Situazione**

condensatore cilindrico:   
guscio interno di raggio a, esterno di raggio b  
gusci hanno carica o corrente opposta $\pm \lambda$ ($\lambda$ è la carica in una sezione circolare di cilindro)  

**Soluzione campo elettrico**

scelgo S: una superficie cilindrica di raggio r, concentrica/coassiale al mio  

$\Phi = \int E \cdot S = E (2 \pi r L)$  
nota che le basi circolari del cilindro hanno S perpendicolare ad E, quindi non c'è flusso lì (E va da cariche + a -, so il suo verso)    

$Q_{in} = \begin{cases}
+\lambda -\lambda = 0 & r>b \\
\lambda  & a<r<b 
\end{cases}$ 

**Soluzione campo magnetico**

scelgo $\gamma$: un cerchio concentrico di raggio r  

$\int_\gamma B \cdot dl = B \; 2 \pi r$  

$i_{conc} = \begin{cases}
+I-I=0 & r>b \\
I & a<r<b \\
\end{cases}$

$B(r) = \begin{cases} 
\frac{\mu_0 I}{2 \pi r} & r<a \\ 
0 & r>a
\end{cases}$


**Tips**

in conduttore, tutta la carica di si pone sul bordo

**Calcola campo con oggetto con diversi dielettrici**

usa la versione con D, che si adatta $Q=\int D \; dS$  

esempio: cilindro con acqua sotto [0,b], vuoto sopra [h-b, h]    
per il campo radiale, uso:  
scelgo S un cilindro concentrico  
$Q= \int_S D \cdot dS = 2\pi r (\int_0^{h-b}D \; dz + \int_{-b}^0D \; dz)= 2\pi r[\epsilon_0(h-b)+\epsilon_1b]E$  

NON CAPISCO:  
esempio: condensatore, con dentro due dielettrici in parallelo  
scelgo come S un parallelepipedo a cavallo di una faccia del condensatore, così meta è dentro il conduttore (E=0) e l'altra metà è fuori  
la metà fuori è ulteriormente divisa tra i due dielettrici  
$Q = \epsilon_1 E S_1 + \epsilon_2 E S_2 = \epsilon_1 E \, ax +\epsilon_2 E \, a(a-x)$  

i campi E sono uguali perchè paralleli alla linea di separazione tra dielettrici  

### Elettrete

**Esempi comuni di equivalenze**

cariche:  
lastra con $\sigma_b$ uniforme: lastra  
se due lastre: condensatore (cilindro con P assiale uniforme)  
sfera con P uniforme: da sapere, prob scritta nelle note  

cariche:  
cilindro alto con $K_b$ azimutale: solenoide, $NI = K h \quad \rightarrow \quad nI = K$  
cilindro basso con $K_b$ azimutale: spira, $I = Kh$, cioè N=1    
note: solenoide ha $h \gg R$, spira ha $h \ll R$  
e per avere spira devo avere un cilindro PIENO, tanto la corrente scorre solo sul bordo, cioè una sezione piatta di cerchio  

nota: K è la corrente per unità di lunghezza, quindi $K= \frac{dI}{dz} = nI$  

**Cilindro con M assiale**

situazione: cilindro con M assiale, alto h  

1- trova correnti di polarizzazione    
$K_b = M \times \hat{n} =M \hat \varphi$ (azimutale)  

2- trova equivalenza  
correnti sono disegante come quelle in un solenoide    
equivalenza: $nI = K_b$     
viene $B = \mu K_b = \mu M$  

**Spira con M assiale**

situazione: spira (un cilindro basso, di altezza h)    

1- trova correnti di polarizzazione  
anche qui, K=M, come in cilindro  

2- trova equivalenza  
ha K come una spira   
(prima era solenoide, ora è basso quindi una sola spira)   
quindi ho: $I=Kh$, da pliggare in B della spira  

**Eequivalenza tra polarizzazione elettrica e magnetica**

quando: in pratica solo in sfera, o se ho calcolato uno dei due e mi chiede l'altro      
come: se ho oggetto con polarizzazione di M e P in stesso, e so il campo in uno dei due casi, fai cambi E -> H, D -> B, $\frac{P}{\epsilon_0} \rightarrow M$    

esempio: sfera con polarizzazione P o M uniforme   
so che, per P: $E=-\frac{P}{3 \epsilon_0}$  
quindi per M: $H = -\frac{M}{3} \quad \rightarrow \quad B = \mu_0(H + M) = 2/3 \mu_0 M$  

### Campo esterno

**Versioni**

attenzione, ci sono due tipi di esercizi che si assomigliano  

1- so: campo esterno  
quando: se dice di calcolare il campo locale fuori dall'oggetto    
come: metodo spiegato sotto 

2- so: campo locale fuori dall'oggetto  
quando: se devo calcolare solo quello interno all'oggetto    
come: condizioni di raccordo  

**Steps**

1- imposta il sistema:  

$\begin{cases}
E_{l,in}=E_{ext}+E_{ind, in} & sempre \\
E_{l,out}=E_{ext}+E_{ind, out} & sempre \\
E_{l,in}=0 & solo \,se \, conduttore \\  
P=(\epsilon-\epsilon_0) E_{l,in} & solo \, se \,  dielettrico
\end{cases}$

2- assumo che P sia uniforme e nella stessa direzione di $E_{ext}$ (è un ansatz, tanto la sol è unica, se la trovo è giusta)  
trova le formule del campo indotto dato da questa P e le pluggo nel sistema  

3- risolvi prima la parte dentro, per trovare P  
poi calcola $E_{l,out}$    


#### Teoria

**Nomenclatura**

$E_l = E_{ind}+E_{ext}$  

$E_l$: campo locale  
il campo effettivamente presente in una zona, e percepito da un materiale, somma di quello esterno e indotto  

$E_{ext}$: campo applicato dall'esterno, da una sorgente non specificata  

$E_{ind}$: campo generato dalla polarizzazione dell'oggetto, avvenuta in risposta all'eposizione al campo esterno  

ulteriormente, il campo locale e indotto si possono separare in fuori e dentro l'oggetto (in e out), possono avere formule diverse in questi due casi  
ad esempio un conduttore ha campo nullo all'interno, ma non all'esterno  
mentre il campo esterno deve essere uguale dentro e guoti l'oggetto  

#### Esempi

**Esempio: lastra isolante**

situazione: lastra isolante (con spessore non nullo)  
esposta a $E_0$ perpendicolare a essa   

1- polarizzazione dell'oggetto  
ipotizzo P costante, in stessa direzione di $E_0$  
le due facce hanno +$\sigma$ e -$\sigma$, quindi producono due campi uniformi, attenzione alle direzioni    
cioè il campo indotto è come quello di un condensatore

**Esempio: Sfera esposta a E_0**

situazione: sfera con $\epsilon$   
in campo esterno uniforme $E_0$  

1- polarizzazione  
assumo che anche P sia uniforme, come $E_0$  
interno: $E_{ind}= -\frac{P}{3 \epsilon_0}$, ben noto  
esterno: come campo generato da dipolo di momento $\vec{p}$  
quindi devo sapere P e p  

2- imposto il sistema  
come procedere dipende dal tipo di sfera, in generale devo ricavarmi una condizione con P, per trovare P  
poi so che: $p = V \, P = \frac{4}{3}\pi R^3 \, P$  

condizione su P:  
conduttrice: $E_{in}=E_0+E_{ind}=E_0-\frac{P}{3 \epsilon_0}=0$ (campo interno nullo)  
isolante:  $P=(\epsilon-\epsilon_0)E_{in}$ (def di polarizzazione)  
nota che P usa la formula del campo TOTALE interno, mi viene fuori un sistema  

### Sovrapposizione dei campi

**Oggetto con buco**
 
immagina che siano 2 oggetti:  
uno grande come l'oggetto con il buco riempito, con $q, I, P$ o quel che ha  
uno grande come il buco, con $-q, -I, -P$ (cambia di segno, così in quella zona in totale non c'è niente)   

**Differenza tra buco piccolo e grande**

buco piccolo: l'oggetto è quasi intero  
nella zona dentro l'oggetto e fuori dal buco, assumo che il campo locale sia quello generato dal'oggetto intero  

buco grande:   
il campo locale non è mai quello generato dall'oggetto intero  
perchè in realtà non genera quel campo, visto che non somiglia affatto all'oggetto intero  

**Esempio: buco trascurabile**

situazione: una sfera con P, all'equatore manca un disco  

1- campo di oggetto con buco riempito    
$E=-\frac{P}{3 \epsilon_0}$  

2- condizione di raccordo tra oggetto e buco:  
normale alla superficie: $D=\epsilon_0E +P$  
nota che nel buco P=0, perchè è nel vuoto, ed E è la mia incognita    

**Esempio: buco non trascurabile**

situazione: una sfera con P, all'equatore manca un disco  

1- schematizzo la situazione  
sfera intera con polarizzazione P  
cilindro a facce tonde con -P  

2- campi degli ogetti  
sfera con P: formula nota, la stessa di sopra, $E_s=-\frac{P}{3 \epsilon_0}$   
cilindro con -P: come un condensatore con $\sigma$ sulle facce (esercizio di campo sapendo P), $E_C=\frac{\sigma_{tot}}{\epsilon_0}$     

3- sovrappongo i campi  
il buco è all'interno sia della sfera che del condensatore, quindi:  
totale in un punto del buco: $E=E_{s}+E_{c}$  

### Voglio B, so Q in moto

**Esempio: cilindro, Cariche in moto e corrente**

situazione:  
condensatore cilindrico che ruota con $\omega$  
carica $\pm Q$ sulle due armature  
h: altezza   
a,b: raggi delle armature  

1- carica in moto è una corrente  
$K=\frac{I}{h} = \frac{\pm Q}{T \, h}$    
dove $T = \frac{2 \pi}{\omega}$  
carica sulle face interne ed esterne ha segno diverso, cioè ruota in senso opposto  
e ruota in modo da avere una corrente azimutale  

2- trova equivalenza  
somiglia ad un solenoide, dove $nI=K$   
quindi il campo generato da UNA armatura cilindrica è come un solenoide:    
dentro: $B=\mu nI = \mu K$  
fuori: B=0    

3- superposizione dei campi  
in ogni regione  (dentro a, tra e e b, fuori da b) , somma i campi delle due armature   
poichè è un condensatore, se lo faccio giusto il campo deve venire nullo fuori dalle armature

### Tanti conduttori - DA ACCORCIARE

**Quando**

tanti conduttori, e non so come sono disposte le cariche libere  

**Metodi**

1- usa discontinuità di D e E     

2- schematizza come un circuito  
cioè vedi se ci sono dei pezzi (di solito condensatori) in serie o parallelo, e somma la loro C  

**Nota**

$E=\frac{\sigma}{2 \epsilon_0}$: campo emanato da ogni lastra, sia a destra che a sinistra   
è solo quello emaato da UNA lastra, non il campo totale in quella regione, va sommato agli altri   
se $\sigma>0$ uscente, sennò entrante  

$E=\frac{\sigma}{\epsilon_0}$: campo totale nella regione subito fuori da un conduttore  
NON va sommato ad altri, è già il totale  
trovato con la discontinuità del campo elettrico in direzione normale, quando uno dei lati della superficie è l'interno di un dielettrico, che quindi ha $E=0$  

**Guida: metodo 1**

devo scrivere un sistema con tutte le condizioni   
le condizioni possibili sono:  

1. conservazione della carica  
$Q=\sigma_1S_1+\sigma_2S_2$  
ad esempio su un condensatore con carica Q, la densità distribuita sulle due lastre deve essere uguale alla carica totale

2. campo nullo dentro un conduttore  

3. condizioni di continuità del campo (normale e tangenziale)

per scriverle, dà un nome diverso al campo dentro ogni regione / dielettrico


esempio: sistema lineare condensatori in parallelo  
cioè proprio che si toccano tra loro  

$\begin{cases}
Q=\sigma_1S_1+\sigma_2S_2 & conservazione \; della \; carica\\
\frac{\sigma_1}{\epsilon_1} = \frac{\sigma_2}{\epsilon_2}=E & continuià \; campo \; tangenziale   
\end{cases}$

**Guida: metodo 2**

considera ogni dielettrico diverso come un condensatore  
vedi se sono in parallelo o serie

1- calcola C equivalente  
$C=\epsilon \frac{S}{h}$  

2- calcola V tra le piastre  
$V=\frac{Q}{C}$  

3- calcola il campo  
$E=\frac{V}{h}$

**Esempio: metodo 1**

0- intanto dai un nome alle lastre: chiamale A,B,C,D da sinistra a destra  

1- imposta il sistema:  
. conservazione della carica (su ogni conduttore)  
$(\sigma_A+\sigma_B)S=Q$   
$(\sigma_C+\sigma_D)S=Q$   
nota che S dipende dalla forma delle basi, detto dal testo  
. campo all'interno del condensatore è nullo

**Ottenere condizione di campo nullo: metodo 1**

per ogni lastra, disegna le frecce del campo che emana (come quando studio il condensatore)   
di default disegnale uscenti, assumi $\sigma>0$, dal risultato vedrò se è vero   
poi prendi il campo nelle zone che voglio, cioè in A-B e in C-D e mettilo uguale a 0  
ricordando che $E=\frac{\sigma}{2 \epsilon_0}$, semplifica il denominatore e rimangono solo le densità di carica  

esempio:   
._ _A__B__C__D___  
<- | -> | -> | -> | ->    campo emanato da A  
<- | <- | -> | -> | ->    campo emanato da B    
<- | <- | <- | -> | ->    campo emanato da C    
<- | <- | <- | <- | ->    campo emanato da D    

condizioni: basta sommare vettorialmente le frecce nella colonna che mi interessa  
tra le lastre A e B, campo nullo:  
$\sigma_A-\sigma_B-\sigma_c-\sigma_D=0$  
tra le lastre C e D, campo nullo:  
$\sigma_A+\sigma_B+\sigma_c-\sigma_D=0$   

nota che il cmapo emanato da A; disegnato sopra, NON è il campo sentito subito all'esterno della faccia A, E_A,  perchè quest'ultimo viene dalla sovrapposizione di tutti i campi emanati da tutte le facce  

**Ottenere condizione di campo nullo: metodo 2**

dai un nome ad ogni regione, tipo:  
1 | 2 | 3 | 4 | 5   
._A_B_C_D

usa le condizione di continuità del campo:  
$E_2-E_3 = \frac{\sigma_B}{\epsilon_0}$  
$E_3-E_4 =  \frac{\sigma_C}{\epsilon_0}$   
e la condizione di campo nullo nel conduttore:   
$E_2=E_4=0$  

mettendo insieme:  
$E_3 = \frac{\sigma_C}{\epsilon_0} = - \frac{\sigma_B}{\epsilon_0}$  
quindi $\sigma_C = -\sigma_B$
