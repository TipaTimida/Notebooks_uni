### Guida al testo

**Testo**

1- ho dei conduttori volumici di qualche forma   
di cui so $Q$ oppure $V$ sul bordo   

2- nello spazio esterno ai conduttori è presente una densità di carica $ρ(r')$  
ad esempio, c'è una carica q fuori dal conduttore, di solito puntiforme, e so scrivere il potenziale generato da essa  

domanda: devo trovare una configurazione di cariche puntiformi da mettere dentro i conduttori  
tale che: i bordi dei conduttori siano equipotenziali  
cosìuso quelle per trovare il potenziale e il campo in tutto lo spazio  

nota che:  
all’interno del conduttore il campo è nullo e il potenziale costante, è quello della superficie     
quindi vogliamo risolvere solo nello spazio esterno  
quindi la soluzione è valida solo nello spazio esterno  

**Versioni**

posso sapere Q o V di un conduttore  

cosa fare:  
so Q della sfera: metti una carica al centro per avere quella carica totale  
so V sul bordo: metti una carica al centro per avere quel V sul bordo  

come sono detti:  
conduttore messo a terra: V=0 sul bordo  
conduttore isolato scarico: Q=0  

come trasformarli:  
$Q=CV$, dove dipende dalla forma del conduttoe  

**Nota: scalare**

se ho risolto il problema quando il conduttore ha carica $Q$  
e mi chiede di risolverlo con carica $kQ$  
la soluzione $V(x)$ è uguale, basta riscalarla a $kV(x)$

**Piano conduttore e carica puntiforme**

situazione:  
carica puntiforme q  
piano conduttorese senza carica, distante r da q    

soluzione:  
Valore | Cosa | Dove
---|---|---
$q$ | la carica puntiforme, reale | nella carica puntiforme, a $r$ dal piano
$q'=-q$ | la carica immagine | $r'=-r$, cioè simmetrica a q rispetto al piano  

**Sfera conduttrice e carica puntiforme**

situazione:  
carica puntiforme q, a distanza r dal centro della sfera  
sfera con carica Q di raggio a  

soluzione:  
Valore | Cosa | Dove
---|---|---
$q$ | la carica puntiforme, reale | nella carica puntiforme 
$q' = -\frac{a}{r}q$ | la carica immagine di q, per avere V nullo sulla superficie della sfera | $r' = \frac{a^2}{r}$, sulla congiungente tra q e il centro della sfera  
$Q-q'$ | per conservazione della carica all'interno della sfera, o per avere il potenziale richiesto sulla superficie (dipende dalla versione del problema) | nel centro della sfera, fingi sia una carica puntiforme  

**Note sulla sfera**

q e q' annullano la loro V sul bordo  
quindi a dare il potenziale rimane: $Q = CV$  

per calcolare la forza:  
1- utile staccare Q-q' in due addendi, e al posto di q' mettere la sua espressione  
perchè mi cambia la potenza di r al denominatore, mi rende più facile studiare che succede per r diversi  
2- prima di integrare devo esplicitare tutte le r presenti

#### Conduttore sferico con cavità

**Conduttore sferico con cavità con dentro Q**  

in pratica, un guscio con dentro la carica Q   
la carica Q non è per forza nel centro, solo dentro la cavità   
il conduttore, anche se è un guscio, ha uno spessore, quindi ha una superficie interna ed esterna, e tra le due E=0  

soluzione:   

per trovare V dentro il guscio:    
devo aggiungere cariche immagine FUORI dal guscio in modo che il bordo abbia V costante  
tratta la Q come la Q' che aggiungerei come carica immagine di una sfera, e vedi dove sarebbe dovuta essere la Q originale  
nota che per V interno, a seconda di dove è Q, cambia dove metto Q', quindi all'interno il campo cambia, non c'è niente di schermato  

per trovare V fuori dal guscio:   
non servono cariche immagine, basta la discontinuità del campo alla superficie del conduttore       
perchè: usa Gauss su una superficie che passa tutta dentro il conduttore  
cioè E=0 in ogni punto, quindi $\Phi=0$  
allora deve essere $q_{int}=0$, da cui so che sulla superficie interna ci deve essere la carica -q  
quindi per mantenerlo neutro, ci deve essere +q sulla superficie esterna  
quindi quando considero quali sono le cariche in gioco per calcolare V fuori, posso modellizzarlo con solo una carica q nel suo centro  
ignora la carica q originale, dovunque sia (effetto di schermo), e la -q sulla superficie interna  

**Conduttore sferico con cavità con dentro Q: versioni**

ho disco conduttore con Q dentro, non per forza al centro  
voglio sapere il potenziale al suo esterno  

se disco è a collegato a terra:  
sul suo bordo ho V=0  
e anche all'infinito ho V=0 (questo sempre)  
quindi i contorni della zona dove voglio trovare V hanno tutti V=0, cioè vale la soluzione banale V=0 ovunque  
poichè al soluzione è unica, deve essere quella giusta  

se il disco è isolato e scarico (caso spiegato sopra):  
c'è una carica +Q sulla superficie esterna  
quindi fuori sento come se ci fosse una carica +Q al centro

#### Schematizzare a diversi ordini

**Cosa fare**

ordine 0:  
ogni conduttore vede solo se stesso, fai finta che gli altri non esistano, quindi non sommare V pvenineti da altri conduttori

ordine 1:  
tratta il conduttore su cui voglio piazzare le cariche con la forma che ha, ma l'altro rappresentalo solo con una carica puntiforme, come all'ordine 0  
eg. se ho due sfere, A e B, per sapere il potenziale di A considero A una sfera e B una carica puntiforme, così so risolvere con le cariche immagine   
ordine 1: ritrasforma in sfere quelle che in origine lo erano

in pratica devo solo risolvere il problema della sfera adesso, mettendo q sulla sua congiungente  
poi aggiorna le q del piano se serve  

**Esempio: due sfere, con Q e -Q (problema di Coulomb)**

ordine 0:  
entrambe le sfere sono assimilate a puntiformi nel centro

ordine 1:  
assumo che una sfera sia puntiforme, l'altra sferica, risolvo il problema della carica immagine  
quindi ha $Q'$ sulla congiungente e $Q-Q'$ sul centro   
poi, poichè la situazione è simmetrica, copio il risultato anche sull'altra sfera, che avrà $-Q'$ sulla congiungente e $-Q+Q'$ al centro  
sfera 1: Q', Q-Q'  
sfera 2: -Q', -Q+Q'

##### Schematizzare: teoria

**Ordini**

sto separando il problema in 2 'livelli':  

0. effetto principale  
trascuro gli effetti di induzione

1. correzione di primo ordine   
cioè il campo prodotto da un conduttore produce una polarizzazione negli altri conduttori    

**Ordine 0**

non c'è induzione   
la posso trascurare solo se espressamente detto  

teoria:  
vuol dire che sto adottando l’approssimazione di ordine zero nella distanza relativa a/h   
il campo elettrico generato dal secondo conduttore alla superficie della primo è piccolo rispetto al campo dovuto alla sua stessa carica  
quindi ciascun conduttore “vede” solo la propria carica, tratta ogni conduttore come se gli altri non ci fossero   

pratica:  
se ho un conduttore sferico, sostituiscilo con una carica puntiforme nel suo centro  
perchè la sfera ha carica uniforme, non viene polarizzata da campi esterni, visto che li trascura  

**Ordine 1**  

con induzione  

in pratica:  
caso normale, i potenziali generati da ogni conduttore si sommano per principio di superposizione   
posso usare metodo delle cariche immagine completo    
significa tratta il conduttore su cui voglio calcolare V o F con la forma che ha, ma l'altro rappresentalo solo con una carica puntiforme  
eg. se ho due sfere, A e B, per sapere il potenziale di A considero A una sfera e B solo una carica puntiforme, così so risolvere con le cariche immagine  

#### Calcola forza: somma delle forze

**Come fare**

1. fai un recap di tutte le cariche puntiforme e separale in:  
dentro al conduttore su cui voglio calcolare la forza  
fuori da quel conduttore (sia nello spazio fuori che in un altro conduttore va bene)  
2. per ogni coppia di cariche, una dentro e una fuori il conduttore, calcola la forza  
3. somma tutte le forze  
4. approssima il denominatore  

nota: 
1- non considerare la forza tra cariche entrambe dentro o entrambe fuori dal conduttore  
perchè io voglio la forza tra il conduttore e l'esterno   
2- fai attenzione alle distanze da mettere al denominatore  

**Esempio: due sfere, con Q e -Q**

calcolo la forza sentita dalla sfera con carica Q; tanto l'altra è uguale  

ordine 0: (entrambe le sfere sono assimilate a puntiformi nel centro)  
cairche dentro: $Q$  
cariche fuori: $-Q$  
$F_0=k\frac{Q(-Q)}{r^2}$  

ordine 1:  
sfera 1: Q', Q-Q'  
sfera 2: -Q', -Q+Q'  
coppie: [Q', -Q'], [Q', -Q+Q'], [Q-Q', -Q'], [Q-Q', -Q+Q']  

questo è il problema di Coulomb  
la soluzione al primo ordine è: $F=k\frac{Q^2}{r^2}(1-4\frac{r'^3}{r^3})$  

**Esempio: sfera con Q e piano**

il conduttore è tutto il semipiano in cui non c'è la sfera   
non è che sia il conduttore, è la zona dove la soluzione V non vale, quindi dove posso piazzare cariche immagine  
fuori/dentro è riferito alla sfera, perchè voglio la forza sulla sfera  

ordine 0:  
dentro: Q (al centro della sfera)   
fuori: -Q (simmetrica al piano)   
copppie: [Q, -Q]   

ordine 1:  
dentro: q' (sulla congiungente), Q-q' (al centro)   
fuori: Q-q' (Simmetrica al piano)   
coppie:  [Q-q', q], [Q-q', Q-q']

**Approssima (all'ordine 1)**

mi viene detto di approssimare per $\frac{a}{r} \ll 1$  
dove a/r è il fattore che apapre in $q'=\frac{a}{r} \; q$  
a è il raggio della sfera, r la distanza tra centri  

cosa fare:  
1. al denominatore devo far apparire il termine che voglio trascurare  
quindi roba tipo: $r^2(1-(\frac{a}{r})^2)^2$  
2. espando il quadrato:  uso lo sviluppo di taylor su $x=(\frac{a}{r})^2 \ll 1$  
esempio: $\frac{1}{(1-x)^2} \sim 1 + 2x$     

#### Calcola forza: approssimazione in multipolo

**Steps**

dopo avere la lista di cariche dentro e fuori dal conduttore  

1. calcola il momento di monopolo di ogni conduttore   
è semplicemente la carica totale Q di quel conduttore   

2. calcola il momento di dipolo di ogni conduttore  

3. calcolo campi generati dal monopolo e dal dipolo  

4. calcola le forze che voglio  
ogni forza è $F=qE$, dove q è la carica di prova e E il campo di quel mulipolo  
    - ordine 0: solo interazione tra monopoli  
    $F_0 = F_{00}$   
    - ordine 1: anche interazione monopolo-dipolo, ma non dipolo-dipolo: 
    $F_1 = F_{00}+2F_{10}$  


**Calcola momento di dipolo**  

momento va calcolato rispetto ad un polo    
perchè è come se piazzassi lì il dipolo di valore $p$ al posto delle due cariche puntiformi    
attenta a dove lo piazzi, dopo mi servità sapere la sua posizione per mettere r al denominatore  
di solito lo scelgo nel centro di una sfera  

1. approssima fino ad avere 2 cariche nel conduttore  
2. momento si calcola: $p=\Sigma q_ir_i$  
come cariche considera solo quelle dentro il conduttore, ignora tutta la situzione all'esterno  

eg. in una sfera, se scelgo rispetto al centro, ottengo $p=q'r'$, perchè la carica alc entro Q-q' ha braccio nullo  

**Calcolo campi**

monopolo Q: $E=k\frac{Q}{r^2}$   
dipolo $p$ (sferico): $E=\frac{2kp}{r^3}$

**Calcola forze**

il calcolo si semplifica perchè la parte in cui prima dovevo approssimare qui vuol dire solo di non considerare la forza dipolo-dipolo, poichè è di ordine superiore  

nota che 0: monopolo, 1: dipolo  
quindi 00: monopolo-monopolo  

**Esempio: due sfere, con Q e -Q**

momenti di monopolo dei conduttori:   
Q e -Q (semplicemente la carica totale)  

momenti di dipolo dei conduttori:  
lo calcolo sulla sfera con +Q, l'altro è uguale   
posso apporossiamre la sfera con due cariche interne: Q-q' al centro, q' sulla congiungente  
quindi rispetto al centro, il momento è $p=q' \, r'$ (quindi il polo di p è nel centro)  
nota che l'altra sfera avrà invece $-p$  

forza tra monopoli:    
$F_{00}=k\frac{Q(-Q)}{r^2}$    

forza monopolo-dipolo:  
campo dovuto al dipolo di sinistra: $E = \frac{2kp}{r^3}$ (solo per una sfera)  
forza esercitata dal dipolo di sinistra sul monopolo di destra: $F_{10}=(-Q)E$ (metto -Q, perchè quella che riceve la considero un monopolo, quindi la schematizzo con -Q al centro, non -Q+Q' come all'ordine 1)  
il dipolo di destra fa una forza analoga sul monopolo di sinistra, quindi c'è il coeff 2 raddoppio la forza  

ordine 0: $F_0 = F_{00}$  
ordine 1: $F_1 = F_{00}+2F_{10}$

### Configurazioni complesse: Metodo della polarizzazione ?????

**Multipoli**

quando uso il metodo del dipolo indotto, stai separando il problema in due “livelli”  :

1. effetto principale  
questo è il caso in cui trascuro gli effetti del induzione  

2. correzione di primo ordine  
il campo prodotto da un conduttore produce una polarizzazione negli altri conduttori    

**Correzione**

ho 2 conduttori, A e B  

1. il conduttore B genera un campo sul conduttore A  
qui non mi interessa come è fatto il conduttore B, lo semplifico con una carica puntiforme nel suo centro (se è una sfera)  
$E_B=k\frac{Q_b}{h^2}$  
in teoria il campo di B cambia in ogni punto, ma poichè h>>a assumo che sia uniforme su tutta la sfera A, e abbia il valore che assumo nel centro di A  

2. polarizzazione di A sotto l'effetto del campo di B  
se A è una sfera:  
$P_A=2 \pi \epsilon_0a^3E_B$

**Calcola V**

assumo di avere 2 sfere, A e B, di raggi a e b e distanza tra i centri h  
calcolo il potenziale sulla sfera A  

1. effetto principale  
la distribuzione delle cariche non è cambiata dal campo dell'altra sfera  
calcola V cone se i conduttori fossero isolati, poi somma  
$V_0=k\frac{Q_a}{a}+k\frac{Q_b}{h}$  
e considera che $h-a \sim h$  

2. correzione di primo ordine  
potenziale di un dipolo: $V_{dip}=k\frac{\vec{p} \land \hat{r}}{r^2}$  
poi voglio sapere V sulla superficie della sfera A, quindi metti r=a   
attenta al sengo, potrebbe apparire un -  
per saperlo, disegna una carica + in B  
poi disegna la separazione tra le cariche + e - sulla sfera a  
poi disegna le frecce p, da - a +, e r, dal centro di A a B  
se sono in direzione opposta, appare un -  
$V_{dip}=-k\frac{aQ_b}{h^2}$  

se chiede il campo al primo ordine, devo scrivere:  
$V=V_0+V_{dip}$  



**Calcola F**

1. effetto principale  
semplifica ogni conduttore con una carica puntiforme nel suo centro, calcola la forza tra queste due  
$F_0=k\frac{Q_qQ_q}{r^2}$

2. correzione  
forza sentita dal dipolo:  
$F_{dip}=p\frac{dE_B}{dx}$   
la derivata va valutata nel punto dove c'è la sfera, quindi dopo sostituisco  con la posizione del centro della sfera  
poi sostituisci in p il valore di prima  

se chiede la forza totale al primo ordine, rispondo:  
$F=F_0+F_{dip}$  

**Steps**

1. effetto principale  
ogni conduttore ignora l'altro  
fai finta che l'altro non esista proprio, esaminane uno alla volta  
alla fine, NON sommare le V diverse, ognuno vede solo la sua  

2. dipolo indotto  
il conduttore B, genera un campo sul conduttore A  
qui non mi interessa come è fatto il conduttore B, lo semplifico con una carica puntiforme nel suo centro (se è una sfera)  
$E_B=k\frac{Q_b}{h^2}$  

3. polarizzazione di A sotto l'effetto del campo di B  
se A è una sfera:  
$P_A=2 \pi \epsilon_0a^3E_B$  

4. potenziale totale  
A risente del potenziale dato dalla propria carica, dalla carica di B, e dall'effetto della polarizzazzione  
$V_A=k(\frac{Q_a}{b}+\frac{Q_b}{h}-\frac{a Q_b}{h^2})$  
nota che l'ultimo viene dal dipolo, che è fatto di due cariche, quindi immagino sia posizionato nel centro della sfera, in mezzo alle due

1. effetto princiapale

quindi la sfera ha la carica Q distribuita uniforme, la approssimo con una carica puntiforme Q nel suo centro  
risolvo mettendo -Q simmetrico rispetto al piano  
calcola $F_0$ tra Q e -Q  

**Steps: metodo simile dell'energia**

1. risolvi trascurando gli effetti di induzione del piano  
cioè disegno -Q  

2. scrivo l'energia  
tra le coppie ci sono (-Q, Q) e (-Q, p)  
dove p è lo stesso del metodo di prima  
consideralo come posizionata nel centro della sfera, quindi entrambe le coppie sono alla stessa distanza  
ricorda che l'energia di un dipolo è $U=pE$, dove E è quello creato da -Q  

3. forza  
$F=-\frac{dU}{dx}$  
x è distanza tra centro della sfera e -Q

**Come ricavare la soluzione**

Per simmetria cilindrica, possiamo limitarci a studiare il problema bidimensionale  

voglio una soluzione per avere potenziale sulla superficie nullo  
quindi si vuole dedurre la posizione della carica immagine in modo tale che per ogni punto sulla circonferenza valga:  
$k\frac{q}{d}+k\frac{q'}{d'}=0$  
$\frac{q}{q'}=-\frac{d}{d'}$  

quindi d e d' devono mantenere la stessa proporzionalità per ogni punto della circonferenza (perchè q e q' sono costanti)  
è risolto imponendo che il triangolo individuato dai lati r' − a − d' sia simile al triangolo individuato dai lati r − a − d  
Ciò accade se: $r' = \frac{a^2}{r}$
