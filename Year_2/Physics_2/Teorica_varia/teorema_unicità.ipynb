**Problema dell'elettrostatica**

risolvere il problema dell'elettrostatica / problema del potenziale 
vuol dire trovare un potenziale $V(r)$ che rispetti determinate condizioni  

1. equazione di Poisson  
$\nabla^2V(\vec{r})=-\frac{\rho(\vec{r})}{\epsilon_0}$  

2. condizione al contorno  
cioè le informazioni note sul comportamento del
potenziale sul bordo del dominio  

essenzialmente, risolvere un'eq diff con condizioni iniziali  

**Teorema di unicità**

fissata la $\rho(r)$, c'è una sola soluzione dell'equazione di POisson che rispetta le condizioni al contorno  

formalmente:  
il gradiente è unico per tutte le possibili soluzioni  
Dunque una volta trovato un potenziale risolvente esso determina univocamente il campo  
cioè il campo è univocamente determinato dal potenziale e dalla forma del conduttore, ed esso determina univocamente la densità superficiale di carica σ in ogni punto del conduttore  

conseugenza:  
se $\rho(r)=0$ ovunque   
e le condizioni al contorno sono $V=c$ ovunque  
la soluzione costante V(x)=c risolve  l'equazione di laplace e rispetta le condizioni al contorno, quindi è valida  
e poichè deve essere unica è anche quella giusta  

**Condizioni al contorno**

ci sono 2 tipi:  

1. Condizioni di Dirichlet: è noto il potenziale per ogni punto al bordo  
in pratica so V sul bordo di ogni conduttore   
o comunque sui bordi dello spazio che voglio studiare  

2. Condizioni di Neumann: è nota la componente normale (al bordo) del gradiente su ogni spunto del bordo  
in pratica so Q di ogni conduttore

**Equazione di Poisson**

equazione di Laplace: è l'equazione di Poisson in zone con $\rho=0$  
quindi: $\nabla^2 V = 0$
