**Definizione**

a parole:  
risposta di un operatore differenziale lineare a una sorgente puntiforme  

formale:  
$L$: operatore differenziale lineare  
equazione differenziale: $L \, u(x)=f(x)$  
G è la a funzione di ingresso all'operatore che produce per risposta l'impulso elementare (delta di Dirac)  
$LG(x, x')=\delta(x-x')$  


**A che serve**
 
se conosci la risposta alla sorgente puntiforme, puoi ottenere la soluzione per una sorgente qualunque   
basta scriverla come soma o integrare di sorgenti puntuali, per principio di sovrapposizione  

**Come scrivere sorgente puntiforme**

**Formule**

equazione differenziale:  
$L \, u(x)=f(x)$  
L si chiama operatore differenziale  

per definizione di G:  
$L G(x, x')=\delta(x-x')$  
$\delta(x-x')$: sorgente puntiforme nel punto x'  

quindi posso scrivere:  
$f(x)=\int f(x') \delta (x-x') = \int  f(x')LG(x,x') dx'$
