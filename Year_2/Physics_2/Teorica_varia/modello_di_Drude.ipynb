#### Altro

**Teorema di equipartizione**

$\frac{1}{2}m<v>^2=\frac{n_{dof}}{2}k_BT$  
un atomo ha 3 DOF   
perchè il movimento degli elettroni di conduzione è dovuto all'agitazione termica  

è valido quando:  
- è un sistema classico  
quindi no vicino allo 0 assoluto, dove diventa quantistico  
- è all'equilibrio termodinamico   
perchè serve che gli elettroni sono in equilibrio termico con gli atomi del metallo, così T è la temperatura di entrambi   
- è un sistema ergodico  

**Sistema ergodico**

dove la media statistica coincide con la media temporale  

ad esempio, per il valore medio della velocità $<v>$:  
media statistica: osservo quella di diverse particelle in un solo istante   
media temporale: osservo quella di uno stesso elettrone per un po' di tempo   

formalmente: la media si ottiene facendo un sampling di quella variabile, ottengo molti valori, poi faccio la media tra quelli  
a seconda di come faccio questo sampling, cioò da dove prendo i valori:  
Media temporale: quello che ottieni osservando una singola particella o sistema per un tempo molto lungo.  
Media statistica (ensemble): quello che ottieni osservando tanti sistemi identici nello stesso istante, ciascuno in una configurazione diversa.  

**Tipi di velocità**

$\sqrt{<v^2>}$: velocità quadratica media  
ottenuta dal teorema di equipartizione  
anche detta $v_T$, velocità termica, perchè dovuta all'agitazione termica  
fa fare un random walk agli elettroni     

$v_d$: velocità di drift/deriva    
$v_d = -\frac{eE}{m}\tau$ (secondo il modello di Drude)   
velocità del moto ordinato di elettroni sotto l'azione di un campo elettrico  

nota: $v_T \sim 10^5, v_d \sim 10^{-3}$  
quindi quella termica (che va in direzioni a caso) è molto più grande  
quindi gli elettroni si spostano nella direzione indicata dalla velocità di deriva, ma non in linea retta, in modo molto disordinato, tipo random walk ma che pian piano va nella direzione della velocità di drift  

#### Risolvi

**Dati**

$\tau$: tempo medio tra due urti   
$\frac{1}{\tau}$: probabilità di collisione per unità di tempo  

**Situazione**

cosa accade in un periodo dt:  

1- elettroni che hanno già urtato:  
percentuale di elettroni: $\frac{dt}{\tau}$  
velocità media: 0, perchè hanno appena ricevuto velocità random, quindi la media è 0    

2- elettroni che non hanno già urtato:  
percentuale di elettroni: $1-\frac{dt}{\tau}$  
velocità media: $m⟨v(t)⟩ -eE$, perchè hanno avuto la velocità random modificata dal campo elettrico  
v è la velocità media, cioè $⟨v⟩$    

**Trova velocità**

1- media pesata (su già urtato e non ancora), poi ignoro gli infinitesimi di ordine maggiore $(dt)^2$  
$m\; v(t+dt) = \frac{dt}{\tau} \cdot (0) + \frac{dt}{\tau} \cdot (m⟨v(t)⟩ -eE)$  
$m \dot{v} = -\frac{m}{\tau}v-eE$  
perchè eE non ha tau al denominatore???

2- in regime stazionario, con $\dot{v}=0$, calolo la corrente v:  
$\vec{v}=-\frac{e\vec{E}}{m}\tau$  
è in direzione opposta a quella del campo elettrico e a lui proporzionale

**Modello di Drude**

$m\dot{v}=qE-q \alpha v$  
$m\alpha v$: forza viscosa, per modellizzare gli urti  
$\alpha$: quale coefficiente, di unità di misura 1/s
