**Types of potentials**

potentials are solutions to the Maxwell equations  

1- retarded potentials  
for: time-dependant (moving), charge distribution   
it's the most general solution  

2- Liénard–Wiechert potentials  
for: time-dependant (moving), point charge  

**Retarded potentials**

$\phi(r,t)=\frac 1{4 \pi \epsilon_0} \int \frac{\rho(r',t_r)}{|r-r'|} d^3 r'$  
$A(r, t) = \frac{\mu_0}{4 \pi} \int \frac{J(r', t_r)}{|r-r'|} d^3 r'$

**Liénard–Wiechert potentials**

$\phi(r,t)=\frac 1{4 \pi \epsilon_0} [ \frac{q}{(1- n \cdot \beta)R}]_{t_r}$  
$A(r, t) = \frac{\mu_0}{4 \pi}[ \frac{q \vec v}{(1- n \cdot \beta)R}]_{t_r}$  

**Campi ottenuti da Lienar-Wiecher**

$E(r,t) = \frac{q(n-\beta)}{\gamma^2(1-n \cdot \beta)^3R^2}+\frac qc \frac{n \times [(n-\beta)\times \dot \beta]}{c(1-n \cdot \beta^3 R)}$  
$B= [n \times E]_{t_r}$  

tutto calcolato al tempo ritardato

#### Radiation

**Condition for radiation**

there is radiation if:   
$P= \int S \, dA = S \, 4 \pi r^2 = const$  

explanation:  
energy trasported can be expressed by the Poynting vector: $S \sim EB$   
if P=const (and not 0), it means that a finite amount of power crosses every sphere, so energy is truly being carried away   
if it gives 0, energy don't reach infinite, stay closer to the source, so it's not radiating   

when it's verified:  
since $S \sim EB$, to have P=0 must have $E,B \sim \frac 1r$  
if fields scale fast, like $E \sim \frac{1}{r^2}$, i have $P \sim \frac{r^2}{r^4} \rightarrow 0$, so no radiation, power don't arrive at infinity  
essentially, acceleration field scales to 0 slower, so far away from the sources it's the only one left  

**Formulas for power**

1- angular distribution:    
$\frac{dP}{d\Omega}=\frac{dP}{dA} \frac{dA}{d\Omega}= r^2 \; ⟨S⟩$    
tells me in which direction it irradiates more  

2- total power irradiated:    
$P = \int \frac{dP}{d\Omega} d\Omega$  
integrate the power irradiated in all directions  

3- intensity of irradiated power:  
$I=⟨S⟩_T = \frac 1T \int_0^T |S| \, dt$ 
(modulo medio del vettore di Poynting) 
oppure: $dP = I dA$  
so: $\frac{dP}{d\Omega} = r^2 I$
