### Potenziale scalare e vettore

**Definizioni**

posso definire i campi fisici reali in funzione dei potenziali elettromagnetici   

$\begin{cases}
B = \nabla \times A \\
E = -\nabla \phi -\frac{\partial A}{\partial t} 
\end{cases}$  

#### Come ricavarli

**Note di teoria**

1- $\nabla \cdot F =0$  
nome: campo solenoidale   
scrivere come: $F = \nabla \times V$, rotore (di un campo vettoriale)    

2- $\nabla \times F =0$  
nome: campo conservativo  
scrivere come: $F=- \nabla V$, gradiente (di un campo scalare)    

**Come trovare formule dei potenziali**

uso le equazioni di Maxwell DISOMOGENEE (che non dipendono dalle sorgenti)  
$\begin{cases}
\nabla \cdot B =0 \\
\nabla \times E = -\frac{\partial B}{\partial t}
\end{cases}$ 

1- potenziale vettore  
dalla prima, B è solenoidale  
quindi scrivo: $B = \nabla \times A$  

2- potenziale scalare  
nella seconda, pluggo B scelto prima, trovo una cosa conservativa: $\nabla \times (E+\frac{\partial A}{\partial t})=0$  
quindi scrivo: $E = -\nabla \phi -\frac{\partial A}{\partial t}$  

**Come trovare equazioni dei potenziali**

uso le equazioni di Maxwell OMOGENEE (qulle non usate prima)  
pluggo E e B in funzione dei potenziali  

ricorda che:  
$\nabla \times (\nabla \times A) = \nabla (\nabla \cdot A) -\nabla ^2 A$  
e poi plugga la condizione della Gauge che ho scelto  

#### Altro

**Potenziale vettore**

caratteristiche:  
1- definibile ovunque, ache se $J \neq 0$  
2- generalizzabile a casi non stazionari  

osservazione:   
flusso di B = circuitazione di A (vedi con th di Stokes)    

### Gauge

#### Definizione

**Scelta di Gauge**

situazione:  
ho un campo A di cui so solo il valore del rotore  
non basta a definirlo in modo univoco perchè A e A' hanno stesso rotore:  
$A'=A+\nabla \psi$    

perchè:   
$\nabla \psi$ è il gradiente di un campo vettoriale   
quindi è campo conservativo, ha rotore nullo  

conseguenza:  
ho una 'libertà di Gauge', cioè A è definito a meno di un $\nabla \phi$   
posso sceglierle un valore di $\psi$, significa fare una scelta di Gauge 

**Invarianza di Gauge**

qualunque A scelgo, B non cambia  
questa si chiama 'INVARIANZA DI GAUGE'  

cioè:   
facile: la libertà di cambiare come descrivi un campo fisico, senza cambiare la fisica  
difficile: una simmetria locale di una teoria fisica: significa che puoi cambiare certe quantità “interne” del sistema punto per punto nello spazio-tempo, senza modificare le osservazioni fisiche  

#### Trasformazione di Gauge

**Trasformazione di Gauge**

$\begin{cases}
A'=A+\nabla \chi \\
\phi' = \phi -\frac{\partial \chi}{\partial t}
\end{cases}$

$\chi$ è una funzione, non un campo  

**Come trovare trasformazione di Gauge**

1- potenziale vettore  
è lui che fa la trasformazione di Gauge, perchè mi interessa solo il rotore  
quindi posso sommare un campo conservativo, che ha rotore nullo  

2- potenziale scalare  
se cambia A, deve cambiare anche questo, perchè deve rimanere la relazione:   
$E= -\nabla \phi' -\frac{\partial A'}{\partial t}= -\nabla \phi' -\frac{\partial A}{\partial t} - \frac {\partial}{\partial t} \nabla \psi = -\nabla -\frac{\partial A}{\partial t}$  
quindi: $\phi' = \phi -\frac{\partial \psi}{\partial t}$  

**Che gauge usare**

per sceglierla, mi serve un'altra condizione su A, non mi basta sapere il suo rotore  
qeust'altra condizione mi fa scegliere quanto vale $\phi$  
posso scegliere la condizione che voglio, quindi ne scelgo una per scrivere le equazioni di Maxwell in modo più facile  

intuitivamente: ho condizione sul rotore di A, ora voglio condizione sulla divergenza  
a seconda della metrica su cui lavoro, scelgo se A è 3-vector o 4-vector  

**Tipi di Gauge**

1- gauge di Coulomb  
quando: caso stazionario   
condizione: $\nabla \cdot A = 0$  
cioè: divergenza di A, con A 3-vettore  
vantaggi: $\phi$ soddisfa l'equazione di Poisson (e quindi???)  

2- gauge di Lorenz (non c'è la t)  
quando: caso non stazionario  
condizione: $\partial_\mu A^\mu = 0$, cioè $\nabla \cdot A + \frac{1}{c^2} \frac{\partial \phi}{\partial t}=0$  
cioè: divergenza di A, con A 4-vector  
vantaggi: mantiene invarianza relativistica e le equazioni di Maxwell si disaccoppiano in equazioni d'onda  


**Equazioni nelle nuove Gauge**

1- gauge di Coulomb  
per $\phi$: come equazione di Poisson, mostra che il potenziale elettrico è istantaneo, non ritardato (che vuol dire???)  
$\nabla^2 \phi = -\frac \rho {\epsilon_0}$  
per A: equazione delle onde non omogenea:  
$\nabla^2 A -\frac{1}{c^2}\frac{\partial^2A}{\partial t^2}=-\mu_0J +\frac 1{c^2} \nabla (\frac{\partial \phi}{\partial t})$  

2- gauge di Lorenz  
si disaccoppiano in equazioni d'onda e sono invarianti  
somigliano a equazioni delle onde non omogenee, cioè forzate, con a destra le sorgenti  
qui uso il quadrivettore $A = (\frac jc \phi, \vec A)$  
parte temporale:  
$\nabla^2 \phi - \frac 1{c^2} \frac{\partial^2 \phi}{\partial t^2}=-\frac{\rho}{\epsilon_0}$  
parte spaziale:  
$\nabla^2 A - \frac 1{c^2} \frac{\partial^2 A}{\partial t^2}=-\mu_0 J$
