### Come scriverle

**Versioni con H,D**

cambiano le cariche (o correnti, per magentismo) usate:  
E,B: cariche totali (libere + polarizzazione)  
D,H: solo cariche libere (E NON appaiono $\mu_0$ e $\epsilon_0$)

**Forma integrale**  

come ottenerla dalla forma locale:  
1- fai l'integarle da entrambi i lati  
2- usa il teorema della divergenza o di stokes fer spar sparire il $\nabla$  
3- appare o il flusso o la circuitazione  

th divergenza:  
$\int_S E \cdot dS = \int_V (\nabla \cdot E) \; dV$   
th Stokes:   
$\int_\gamma E \cdot dl = \int_S (\nabla \times E) \cdot dS$  

**Usi di nabla**

$\nabla \cdot \quad$: divergenza, parla delle sorgenti del campo   
$\nabla \times \quad$: rotore, parla di ???  
dice come i campi si creano a vicenda  

**Forma locale vs integrale**

forma locale/differenziale: applicata ad un punto preciso dello spazio  
integrale: applicata su uno spazio esteso  

**Derivata parziale e totale**

forma locale: derivata parziale  
perchè sto studiando quel punto x, non voglio cambiarlo  

forma integrale: derivata totale   
perchè può cambiare qualunque fattore, basta che mi cambi il flusso  

speigazione:  
nella forma locale, sto considerando un punto preciso di $B(x,t)$  
quindi voglio vedere se B varia nel tempo, ma NON nello spazio, x deve restare uguale, perchè sto studiando quel punto preciso (perchè è la forma locale)  
derivata parziale mi fa cambiare t ma non x  

nella forma integrale, mi basta che cambi il flusso, per qualunque motivo  
non serve che varia B, può anche variare l'area, l'orientazione...  
si usa derivata totale perchè il flusso è una quantitò globale, non locale  

**Caso statico**

chiamati: Elettrostatica/magnetostatica
vuol dire :niente cambia nel tempo  
quindi: le derivate temporale (sia parziali che totali) sono nulle 
 
poichè nelle leggi di Gauss non appaiono derivate, quelle non hanno una versione diversa per il caso statico  

**Homogeneous vs dishmonogeneous**

homogeneous: represent the inherent structure of fields, do not depend on sources ($\rho, J$)  
dishomogeneous: depend on sources  

### 1,2. Leggi di Gauss

**Legge di Gauss per l'elettricità**

forma / cariche | totali | libere
---|---|---
locale | $\nabla \cdot E = \frac{\rho_{tot}}{\epsilon_0}$ | $\nabla \cdot D =\rho_{free}$  
integrale | $\int_S E \cdot dS = \frac{Q_{tot}}{\epsilon_0}$ | $\int_S D \cdot dS = Q_{free}$


**Legge di Gauss per il magnetismo**

forma / cariche | totali | libere
---|---|---
locale | $\nabla \cdot B = 0$ | $\nabla \cdot H =-\nabla \cdot M$  
integrale |$\int_S B \cdot dS = 0$   | non si usa


**Cosa dicono**

che cosa origina i campi:  
E: ci sono monopoli di carica  
B: non ci sono monopoli di carica (cioè è solenoidale)  
quindi le linee di campo di B sono chiuse  

### 3. Legge di Faraday-Neumann-Lentz

**Forma con E**

questa non si scrive mai con D  
$\nabla \times E = -\frac{\partial B}{\partial t}$  
$∮_c E \cdot dl = -\frac{d}{dt} \int _S B \cdot dS$  


**Legge fisica**  

legge di Faraday-Lentz  
perchè in quella integrale, a sinistra c'è la deifnizione di fem, a destra il flusso di B  

**Caso statico**

$\nabla \times E =0$  
cioè E è conservativo  

### Legge di Ampere-Maxwell

**Con B e H**

$\nabla \times B = \mu_0 J_{tot} + \mu_0\epsilon_0 \frac{\partial E}{\partial t}$  
$\nabla \times H = J_{f} + \frac{\partial D}{\partial t}$  

**Forma integrale**

$∮_c B \cdot dl = \mu_0 I_{tot} + \mu_0 \epsilon_0 \frac{d}{dt} \int _S E \cdot dS$  
$∮_c H \cdot dl = I_{f} + \frac{d}{dt} \int _S D \cdot dS$  

a destra, sono flussi 

**Caso statico**

$∮_c B \cdot dl = \mu_0 I_{tot}$

**Recap correnti**

$J_{tot}$: di conduzione, generata sia da cariche libere che di polarizzazione (comunque moto di cariche)   
$J_f$: di conduzione, generata solo da cariche libere  
$J_s= \epsilon_0 \frac{\partial E}{\partial t}$: di spostamento, generata dalle variazioni del campo elettrico  
in pratica un campo elettrico variabile genera un campo magnetico, esattamente come una vera corrente  

### Nabla

**Recap modi di usare nabla**

3 modi di usare, si distinguono per combinazione di input e output scalare o vettoriale  
per ognuna, si usa un segno di operazione diverso:    
1- div: prodotto scalare (input vettore, output scalare)  
2- rot: prodotto vettoriale (input vettore, output vettore)  
3- grad: normale moltiplicazione (input scalare, output vettore)

input è un CAMPO, campo scalare (V) o vettoriale (E)  


nome |  scritto | formula 
---|---|---
divergenza  | $\nabla \cdot \vec{E}$ | $ \frac{\partial E_x}{\partial x} + \frac{\partial E_y}{\partial y} + \frac{\partial E_z}{\partial z}$    
rotore | $\nabla \times \vec{E}$ | $(\frac{\partial E_z}{\partial y}-\frac{\partial E_y}{dz}, \frac{\partial E_x}{\partial z}-\frac{\partial E_z}{\partial x}, \frac{\partial E_y}{\partial x}-\frac{\partial E_x}{\partial y})$
gradiente  | $\nabla  V$ | $( \frac{\partial V}{\partial x}, \frac{\partial V}{\partial y} , \frac{\partial V}{\partial z})$  

**Tip rotore**

gli addendi col - sono quelli che NON rispettano la regola della mano destra con xyz  
ma devo scrivere la derivata in una sola riga  

$\partial_y E_z\hat x$: yzx la rispetta, ha segno +  
$\partial_z E_y\hat x$: zyx NON la rispetta, perchè punterebbe  a -x, non +x, quindi ha segno -  

**Per ricordare**

è solo un trucco menomonico, NON lo scrivere mai su un compito  
$\vec{\nabla} = (\partial_x, \partial_y, \partial_z)$  
poi fai il prodotto scalare o vettoriale con il campo, a seconda di cosa è richiesto  

sopra la nabla ci va il segno di vettore: $\vec{\nabla}$  

**Definizioni**

continuità della corrente:  
$\frac{d \rho}{dt} + \nabla \cdot J = 0$  

divergenza:  
$\nabla \cdot J = 0$:  corrente stazionaria  
$\nabla \cdot B = 0$:  campo solenoidale  
$\nabla \cdot v = 0$:  fluido incompribmibile    
div=0 significa che non ci sono ammucchiamenti di quel campo in un punto, no sorgenti o pozzi, il flusso attraverso ogni superficie è nullo, le linee di forza sono chiuse  

rotore:  
$\nabla \times E =0$: campo conservativo/irrotazionale  

**Spiegazione fisica**

divergenza: tendenza del campo a divergere o convergere in quel punto  
immagina un fluido, vedi se stagna  
indica: sorgenti / pozzi  
div<0: converge in quel punto

rotore: rotazione del campo sulla superficie  
immagina di fissare una pallina con un puntaspilla in un fluido: se ruota attorno al suo asse, c'è rotore in quel punto  
indica: vortici / rotazioni locali

**Calcola rotore**

tip per ricordare formula**

$\nabla \times F = \begin{bmatrix} \hat{x} & \hat{y} &\hat{z} \\
\partial _x & \partial_y & \partial_z \\
F_x & F_y & F_z
\end{bmatrix}$

ustile sviluppare lungo la prima riga con Lagrange  

**Gradiente in coord polari**

$ \nabla  V = \frac{\partial V}{\partial r} \hat{r} + \frac{1}{r} \frac{\partial V}{\partial \theta} \hat{\theta} + \frac{1}{r \sin \theta} \frac{\partial V}{\partial \varphi}$  

se uso coordinate polari (cilindriche o sferiche) è perchè il problema ha una tale simmetria  
quindi probabilmente V cambierà soltanto in direzione radiale  
quindi per il gradiente di V devo ricordarmi solo la componente radiale, che è il primo addendo ed è normale, non ha coefficienti strani:  
$ \nabla  V = \frac{\partial V}{\partial r}$  


**Laplaciano**

si può applicare sia ad uno scalare che ad un vettore, ridà in output la stessa cosa:  

ad uno scalare:  
$ \nabla^2 V = \frac{\partial^2 V}{\partial x^2} + \frac{\partial^2 V}{\partial y^2} + \frac{\partial^2 V}{\partial z^2}$ 

ad un vettore:  
$\nabla^2 \vec{F} = (\nabla^2 F_x, \nabla^2 F_y, \nabla^2 F_z)$   
essenzialmente separa il vettore in componenti scalari, poi applica il laplaciano su ognuna separatamente  

#### Definire univocamente

**Aggiungere pezzi**

se A e A' hanno lo stesso rotore:  
$A' =A+\nabla f$  
è una trasformazione di Gauge  

se A e A' hanno la stessa divergenza:  
$A'=A + \nabla \times \vec v$ (v è un vettore a caso)  
questa NON è una trasformazione di Gauge

**Scrivere campo come divergenza/rotore di un altro**

posso scrivere come:   

campo solenoidale: rotore di un altro campo vettoriale  
ex:  $B=\nabla \times A$  
dove l'altro campo non è determinato univocamente, ma devo fare una scelta di Gauge  

campo conservativo: gradiente di un altro campo scalare  
ex: $E = -\nabla V$  
cioè il rotore del gradiente di un campo scalare è sempre nullo (perchè consrvativo = rotore nullo)

### Teoremi

**Divergenza e Stokes**

servono a far sparire $\nabla$ nelle dimostrazioni, dopo che ho fatto integrale da entrambi i lati  

th divergenza:  
$\int_S E \cdot dS = \int_V (\nabla \cdot E) \; dV$   
- superficie chiusa -> volume racchiuso  
- flusso -> divergenza    
- Flusso che esce da una superficie chiusa = sorgenti all'interno

th Stokes:   
$\int_\gamma E \cdot dl = \int_S (\nabla \times E) \cdot dS$  
- linea chiusa -> superficie aperta che la ha per contorno    
- circuitazione -> rotore    
- Circolazione lungo un bordo = rotazione del campo sulla superficie, flusso del rotore


**Intuizione sul th della divergenza**

sia il flusso totale che la divergenza indicano quanto il campo è stagnante  
se più flusso entra di quello che ne esce: stagna  
se div<0: stagna  
quindi posso legare questi due concetti  

**Ricavare**

è il modo di definire E univocamente, quindi viene dalle due equazioni di Maxwell che lo riguardano  

1- $\nabla \times E = 0 \rightarrow E=-\nabla V$  
sono due cose equivalenti, perchè campo conservativo -> esprimibile come gradiente di un potenziale  

3- $\nabla \cdot E = \frac{\rho}{\epsilon_0}$  

eq. di Poisson:  
$\nabla \cdot(-\nabla V) = \frac{\rho}{\epsilon_0}$  
$\nabla^2 V = - \frac{\rho}{\epsilon_0}$
