### Operators

**4-gradient**

$\partial^\mu = (\frac 1c \partial _t , + \nabla )$  
$\partial_\mu = (\frac 1c \partial _t , - \nabla )$  


**Recap**

name | input | output| formula | meaning
---|---|---|---|---
gradient | scalar field $\phi$ | 4-vector | $\partial_\mu \phi= (\frac 1c \frac \partial{\partial t} \phi, - \nabla \phi )$ | take derivatives in spacetime directions
divergence | 4-vector $A^\mu$ | scalar| $\partial^\mu A_\mu = \frac 1c \partial_t A^0 +  \nabla \cdot \vec A $ | net flow out (sum of inflow/outflow) → conservation law
d'Alembertian  | scalar field $\phi$ | scalar | $\partial_\mu \partial^\mu \phi= \frac 1{c^2} \frac{\partial^2 \phi}{\partial t^2}-\nabla^2 \phi$ | wave propagation  

**Applying to tensor**
 
if something is applied to a tensor, must separarate the temporal and spatial component  
so have actually more equations  

temporal: first component of 4-vectors, first COLUMN of tensors (element 0)  
spatial: last 3 components of 4-vectors, last 3 columns of tensors (elements 1,2,3)   

### Explanations

**Notes: gradient**


what is it:  
it's the only primitive 4-vector differential operator (like the $\nabla$ in 3D)   
change operation depending on what gets in input  

sign:  
generally, for continuity equations, want the + sign, so use $\partial^\mu A_\mu$, NOT $\partial_\mu A^\mu$  
sign changes like in all other 4-vector, when lowering indices  
1/c at first component is to get rid of the 'c' at the first component in most 4-vectors   

**Explanation: 3D vs 4D**

$\partial^\mu$ in 4D work like $\nabla$ in 3D  
by choosing the correct input and output, can get the same operations  
but adding one more component/addend, the time one  

example:  
in 3D, the gradient is a vector with the derivatives in all directions $\nabla = (\partial_x, \partial_y, \partial_z)$   
in 4D, $ct$ is the first direction, so add that component to the vector  $\partial^\mu = (\frac 1c \partial_t, \partial_x, \partial_y, \partial_z)=(\frac 1c \partial_t, \nabla)$   
so, differentiate in spacetime rather than just in space  


**Contraction**

means: summing over repeated indices  
so happens the contraction of indices, which lowers the rank   
rakn(input)>Rank(output)  

eg. $\partial_\mu V^\mu$, input has rank 1 (vector), output has rank 0 (scalar)  
eg. $\partial_\mu F^{\mu v}$, contraction on 1 index  

**Remember D'alambertian**

$\partial_\mu \partial^\mu$ means i must do a scalar product (between indices are one up, one down)  
like in linear algebra, sum the corresponding components:  

$\partial^\mu = \frac 1c \partial_t + \nabla$  
$\partial_\mu = \frac 1c \partial_t - \nabla$   
$\partial_\mu \partial^\mu = \frac 1 {c^2} -\nabla^2$   

note: like in 3D, can also be applied to a 4-vector, in that case it takes the d'alambertian of each component (which is a scalar)  

**Divergence: usage example**

usage: net flow out -> conservation laws  

1- continuity equation (conservation of charge)  
$\partial_\mu J^\mu = 0 \quad \rightarrow \quad \frac{\partial  \rho}{\partial t} + \nabla \cdot \vec J =0$

2- conservation of energy and momentum   
(applied to tensor, the conservation law is the temporal component)  
$\partial_\mu T^{\mu v}=-f^v$  
temporal component: $\partial _t u + \nabla \cdot S = -J \cdot E$   
note: must multiply everyhting by c to obtain it   

### Maxwell equations

**Maxwell equations**

1-  
general: $\partial_\mu F^{\mu v} = \mu_0 J^v$   
in Lorenz Gauge: $\partial_\mu \partial^\mu A^v= \mu_0 J^v$  

2- $\partial_\mu \tilde F^{\mu v}=0$  


**Explanation Maxwell 1**

is applied to a tensor, split the temproal and spatial component (row 0, rows 1-2-3)  
think about ho F and J are written, what their components are (watch the iamge of F)    

1- temporal: $\partial _i E = \frac{\rho}{\epsilon_0}$  
$F^{0}=(0, \frac{E_x}{c},  \frac{E_y}{c},  \frac{E_z}{c})$    
$J^0 = c \rho$   
so: $\partial _i \frac{E_i}{c}= \mu_0 \;  c \rho$  
normally, use: $\nabla \cdot E = \frac \rho {\epsilon_0}$  

2- spatial: $\frac 1{c^2} \frac {\partial E_i}{\partial t} + (\nabla \times \vec B)_i = \mu_0 J_i$  
$\partial_j F^{ij} = (\nabla \times B)_i$  
normally, use: $\nabla \times B = \mu_0 J + \frac{1}{c^2} \frac{dE}{dt}$  



**Maxwell eq 1 in Lorenz gauge**

always true: $F^{\mu v}=\partial^\mu A^v-\partial^v A^\mu$  
condition of Lorenz Gauge: $\partial_\mu A^\mu = 0$  

therefore, g
the left side:    
$\partial_\mu F^{\mu v} = \partial_\mu \partial^\mu A^v - \partial_\mu \partial^v A^\mu =  \partial_\mu \partial^\mu A^v -  \partial^v(\partial_\mu A^\mu) = \partial_\mu \partial^\mu A^v$
