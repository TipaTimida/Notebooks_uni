#### EM field tensor

![text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcSGoyRzg9rSeecOxxbRe-Vh7EE4S_2usQjs9w&s)

**Definition**

what: rank-2 antisymmetric tensor   
structure: contains the EM field (E,B)   
use: Maxwell equations  

**Structure**

E: time-space components (first row and col)   
B: space-space components (3x3 bottom matrix)  

differences:  
$F^{μν}$: the true electromagnetic field  
E,B: observer-dependent projections  

**Uses**

simplify Maxwell's equations  

1- $\partial_\mu F^{\mu v}= \mu_0 J^v$  
sources (charge/current)   

2- $\partial_j \tilde F ^{\mu v} =0$  
$∂_λ​F_{μν}​+∂_μ​F_{νλ}​+∂_ν​F_{λμ}​=0$   
no magnetic monopoles + Faraday law   

**Notes**

F is antisymmetric, so only has 6 independent components (dim(F)=6)  
3 components from E and 3 from B  
and 0 on diagonal  

**Similar tensors**

1- lowering indices: $F_{\mu v}$  
$F_{\mu v}=g_{\mu \alpha} g_{v \beta}F^{\alpha \beta}$  

2- dual tensor: $\tilde F^{\mu v}$  
$\tilde F^{\mu v}=\frac 12 \epsilon^{\mu v \rho \sigma} F_{\rho \sigma}$  
$\epsilon$ is the LEvi-Civita tensor (totally antisymmetric) (adapted for 4 dimensions???)  
what it mean:  “rotates” the tensor in spacetime, mixes electric and magnetic components (a 90° rotation in the “E–B space” )  
how to write: swap $F^{\mu v} \sim(E,B) \quad \tilde F^{\mu v} \sim (B, -E)$    


**How to use it**

1- in a frame where i kno E and B, build F  
2- transform F to a frame i want to know about  
3- from the F in such frame, extract E and B (look what is written in the cells that contain thei components)  

#### Energy-momentum tensor

![text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcT6bMREI4R1116vDU7XnBg1Z_eu5muxFNKNTg&s)

**Definition**

what: 2-rank symmetric tensor  
structure: contains energy + momentum + stress  
use: conservation laws  

**Structure**

$T^{\mu v} = \begin{bmatrix} 
energy \; density & energy \; flux \\
momentum \;density & stress \;tensor \end{bmatrix}$

1- energy density: $T^{00}=u=\frac 12 (\epsilon_0E^2+\frac 1 {\mu_0}B^2)$  
2- energy flux: $T^{0i}=\frac{S_i}c$    
3- momentum density: $\vec g = \frac {\vec S}{c^2}$  
4- stress tensor: $T^{ij}=\sigma_{ij}= \epsilon_0 (E_iE_j +c^2 B_iB_j -\frac 12 \delta_{ij}(E^2+c^2B^2))$      

**Describe**

describes: pressure from EM fields, forces on surfaces,radiation pressure

how energy and momentum are distributed and flow through spacetime (The EM field carries momentum, not just energy. If light hits something, it can push it (radiation pressure))


**Note: first row and col**

first row and col look the same, buw then i use $T^{\mu v}$ in formulas they become different  
bc shen apply the gradient:  
$\partial_\mu T^{\mu v}$, will get the $\partial_\mu = \frac 1c {\partial}{\partial t}-\nabla \cdot $  
so, the first col gets one more c at denominator, since it's a time component, and becomes the momentum density  
while the first row is space component, don't get another c  

essentially it's momentum not of the movement of an object, but of the movement of energy  
so if there is momentum, it means energy is flowing  

**Uses**

encodes the conservation laws:    
$\partial_\mu T^{\mu v}=-f^v$  

1- time component: energy conservation  
$\partial_tu +\nabla \cdot S = -J \cdot E$  
note that i multiplied everything by c  

2- space components: momentum conservation  
$\partial_t g +\nabla \cdot \sigma = force \; density$  

**How to write it**

can write it entirely starting from F  

### 4-vectors

**Basic ones**

4-position: $x^\mu =(ct, \vec x)$  
4-velocity: $u^\mu =\gamma(c, \vec v)$   
4-momentum: $p^\mu =mu^\mu =(\frac Ec, \vec p)$   

**Nota**

moltiplica per c in modo appropriato, in modo che sia la parte temporale che spaziale abbiamo le stesse unità di misura (dimensionalmente omogenei)  

**4-current**

$J^\mu=(c \rho, \vec J)$   

contains: all sources of EM fields, both charge and current density  

equations:  
1- maxwell: $\partial_\mu F^{\mu v}=\mu_0 J^v$ 
2- continuity equation: $\partial_\mu J^\mu = 0$  

**4-potential**

$A^\mu = (\frac \phi c, A)$  

contains: potentials that generate the EM field  

equations:  
1- write field tensor: $F^{\mu v}=\partial^\mu A^v -\partial^v A^\mu$  
2- condition of Lorenz gauge: $\partial_\mu A^\mu = 0$    
3- maxwell in Lorenx gauge: $A^\mu = \mu_0 J^\mu$   
4- gauge invariance: $A^\mu \rightarrow A^\mu +\partial^\mu \Lambda$  

**4-force**

$f^v = F^{\mu \lambda}J_\lambda = (J \cdot E, \; \rho+J \times B)$  
time component: power density  
spatial component: Lorentz force density  

what it is:  
force density on matter = force per unit volume exerted by the field on charged matter   
so changes in EM momentum produce forces    

equations:  
1- Lorentz force: $f^\mu = q F^{\mu v}u_v$  
non covariant form: $F=q(E+v \times B)$  
2- equation of motion, relativistic force  
$f^\mu = \frac{dp^\mu}{d \tau}$  
3- ???: $\vec f = \frac{\partial g}{\partial t}$  



**Relationships**

field generation:  
J -> A -> F  
field act on particles:  
F + u -> f   
particle motion:  
f -> p -> x  

#### Lorentz invariants of electromagnetism

**What are they**

phsyical quantities that don't change under Lorentz transformation  

1- $F_{\mu v} F^{\mu v}=2(B^2-\frac{E^2}{c^2})$  
intuitively: how strong electric vs magnetic  

2- $F_{\mu v}\tilde F^{\mu v}=-\frac 4c E \cdot B $   
intuitively: how much theya re intertwined  

**Invariant 1**

it's the norm of F  

interpretation: like in relativity, according to the sign of the norm, can define a type of field:  
norm>0: magnetic-like field (exist frame wih only B, where B>0 and E=0)    
norm<0: electric-like field  
norm=0: like an EM wave  

**Invariant 2**

measures how much E and B are aligned  
if E and B are orthogonal in a frame, they are in all others too
