### Vettore di polarizzazione

**Formula**

$M=\mu_0 \chi_m H = \mu_0 (\mu_r-1)H$  
$P=\epsilon_0 \chi_e E = \epsilon_0 (\epsilon_r-1)E$  

**Definizione**

$dq = P \, d\tau$  
$dm = M \, d\tau$  

momento di dipolo elettrico/magnetico per unità di volume   
descrive a livello macroscopico la polarizazione microscopica  

P: dovuto a spostamento di cariche microscopiche  
M: dovuto a orientazione della spira microscopica  

**Altra definizione**

- polarizzazione elettrica:  
$\vec{P} = \lim _{d\tau \rightarrow 0} \frac{\Sigma \vec{p}_i}{d\tau} = \lim _{d\tau \rightarrow 0} \frac{<p> dN}{d\tau} =n <p>$  

- polarizzazione magnetica/magnetizzazione:   
$\vec{M} = n <m>$  

**Note**

M=0 vuol dire che la polarizzazione MEDIA è nulla, cioè i dipoli sono orientati a caso, non che non ci sono dipoli

quando faccio il limite per il volume $d\tau$ tendente a 0, non tendo davvero a 0  
perchè mi serve $<p>$, che è un valore medio  
quindi $d\tau$ deve comunque contenere un numero di molecole $dN$ STATISTICAMENTE SIGNIFICATIVO  
fare il limite a 0 significa trovare la densità di dipoli nel centro di $d\tau$  

### Perchè c'è polarizzazione?

**Perchè c'è polarizzazione?**

pechè ogni atomo è un dipolo elettrico e magnetico  
quindi se viene applicato un campo E o B, si riorientano di conseguenza

**Differenza momento di dipolo**

dipolo magnetico: ogni atomo ha sempre momento $m \neq 0$  
applicare B cambia la distribuzione di in che direzione sono orientati, cioè cambia $<m>$  

dipolo elettrico:   
se la molecola è polare / l'atomo è uno ione: è come per dipolo magnetico (in realtà qui ho un MONOPOLO, non un dipolo ???)  
se la molecola è apolare / l'atomo non è un ione:  senza campo applicato $p=0$, quando viene applciato E  $p \neq 0$

**Perchè è un dipolo magnetico?**

in un atomo, un elettrone che gira intorno al nucleo è una corrente  
quindi un atomo è una spira  
a grande distanza, una spira è un dipolo magnetico  
per questo i materiali reagiscono al campo magnetico  

quindi $m \neq 0$ tutto il tempo, applicare B serve solo a riorientarlo  

**Perchè è un dipolo elettrico?**  

l'atomo è formato da due distribuzioni di carica:  
1- il nucleo (positivo)  
2- la nube elettronica (negativa), che schemaizziamo come una carica puntiforme nel suo centro  

se polare: la molecola è un monopolo, ha $Q \neq 0$  
se apolare: senza campo applicato, $p=0$  
quando gli applico un campo, atomo e nube non sono più uno sopra l'altro, ottengo un dipolo, $p \neq 0$  



### Tipi di polarizzazione

**Tipi**

due modi possibili, che accadono quando applico un campo E/B su un atomo/molecola:  

1. per deformazione  
dove: su molecole con un momento di dipolo indotto   
cosa: ogni molecola aquisice $\vec{p}=\alpha_D E_l$  

2. per orientamento  
dove: su molecole con un momento di dipolo permanente   
cosa: il valore medio diventa $<\vec{p}> = \alpha_O E_l$

nota: in realtà la deformazione agisce su tutto, anche sulle molecole con dipolo permanente  
ma $\alpha _O > \alpha _D$, quindi la deformazione si può trascurare  
$\alpha_O, \alpha_D$ sono la polarizzabilità per deformarzione (elettronica, se parla di un atomo) e orientamento  

**Caratteristiche**

- deformazione  
in E PARALLELO al campo, in B ANTI-parallelo   
NON dipende dalla temperatura  
non satura (non in elettrostatica standard, almeno)  

- orientamento  
parallelo al campo  
dipende dalla temperatura (perchè $\alpha$ ne dipemde)  
ha saturazione (a campi molto intensi, tutti i dipoli diventano completamente allineati, non si può aumentare)  


#### Elettrico

**Come funziona: deformazione**

se su un atomo, è detta 'deformazione elettronica', perchè a deformarsi è la nube elettronica  

come funziona:    
1- atomo/molecola ha $\vec{p}=0$  
2- applico campo esterno  
3- atomo si deforma, ha $\vec{p}\neq 0$  

che vuol dire si deforma:  
un atomo ha nube elettronica, che schematizzo come una sfera con carica volumetrica uniforme negativa, e nucleo positivo nel centro della nube  
quindi all'inizio, no momento di dipolo  
quando applico E, la nube e il nucleo subiscono una forza in direzione opposta (perchè hanno carica di segno diverso)  
quindi il nucleo non è più nel centro della nube (ma è sempre dentro la nube)  
approssimando la nube come una carica negativa puntiforme nel suo centro, si è formato un momento di dipolo  

**Come funziona: orientamento**

come funziona:    
1- tutti i $\vec{p}$ sono orientati a caso  
2- applico campo esterno  
3- i $\vec{p}$ si riorientano, fino a raggiungere l'equilibrio termodinamico (è equilibrio statico o dinamico???)  

come si riorientano: ci sono 2 fattori  
1. energia di un dipolo sottoposto a campo esterno: $U=-\vec{p} \cdot \vec{E_l}$    
quindi per minimizzare energia, il dipolo dovrebbe disporsi parallelo al campo (così $\cos=1$)  
2. ma c'è agitazione termica dovuta alla temperatura, che scombina di nuovo i dipoli  

#### Magnetico

a seconda di come si magnetizzazno, hanno nomi diversi:  
deformazione: diamagnetici (m=0 a riposo)  
orientamento: paramagnetici ($m \neq 0$ a riposo, momento di dipolo permanente)   

**Nota**

ogni elettrone ha un momento magnetico orbitale  
ma sono orientati tuti i caso, quindi quello medio di una sostanza è 0 (in diamagentico)  

**Spiegazione diamagnetico**

2 possibili modello:  

- effetto Zeeman (classico)     
descrive come cambia l’energia (e quindi le frequenze spettrali)  

- processione di Larmor  
descrive come si muove il momento angolare in presenza di 
B  

**Situazione**

L è normale al piano dell'orbita dell'elettrone  
ricorda: $L=m \omega^2 r$  

voglio trovare momento magnetico dell'elettrone:  
$\vec{m} = I \, S = \frac{e}{2\pi / \omega}\vec{\pi r^2} = \frac{e}{2m}\vec{L}$  
$\frac{e}{2m}$: rapporto giromagnetico (dell'elettrone)  

**Effetto Larmor**

applico B, nasce un momento delle forze:    
$\vec{\tau} = \vec{m} \times  \vec{B}$  

so che: $\frac{dL}{dt} = \tau$  

quindi:  
$\frac{dL}{dt} = \frac{e}{2m}\vec{L} \times B = -\omega_L \times L$  
$\omega_L = \frac{e}{2m}\vec{B}$  
cioè il vettore $\vec{L}$ non si inclina, ma ruota con velocità costante attorno a B  

corrente di Larmor: $I = \frac{e}{2 \pi / \omega_L}$  

**Diamagnetismo di Langevin**

Considera un elettrone vincolato su un’orbita circolare di raggio r  
applico campo B, che causa forza di Lorentz   
se non può modificare raggio dell'orbita, modifica velocità angolare dell'elettrone  
ma r può cambiare??? non capisco  

senza B: $F_c = F_e \rightarrow m \omega_0^2 r= k \frac{e^2}{r^2}$  
con B: $$F_c = F_e + F_L \rightarrow m \omega^2 r= k \frac{e^2}{r^2} + e\omega rB$  
sottraggo e approssimo:    
$e \omega B =   m(\omega^2-\omega_0^2)=m(\omega+\omega_0)(\omega-\omega_0)\sim m \, 2\omega \, \Delta \omega$  
$\Delta \omega = -\frac{eB}{2m}$  

variazione del momento amgnetico:  
$\Delta m = \Delta I \, S = -\frac{e}{2 \pi} \Delta \omega \, (\pi r^2) = \frac{e^2r^2}{4m} B$  

ipotesi isotropa (media su tre direzioni spaziali):  
$<r^2>=<x^2>+<y^2>+<z^2>$  
dove ogni componente vale uguale, 1/3 del totale  
se io voglio distanza da B, che assumo diretta lungo la retta r, allora devo prendere 3 componenti su 3, quindi:  
$<r_\perp^2>=\frac{2}{3}r^2$  

quindi:  
$m = -\frac{e^2 r^2}{6m} B$  
all'inizio $m_0=0$, quindi $\Delta m = m$  
nota che il dipolo magnetico è OPPOSTO al campo

Nei paramagnetici, i dipoli permanenti si allineano al campo, ma

### Comportamento meccanico

**Formule**

$U = -\mu \cdot B$  
$F=-\nabla U$  

**Spiegazione**

se $\mu$ è parallelo a $B$: (para, ferro)  
$F= \mu \nabla B$  
cioè F è nello stesso verso di grad(B), che per def punta verso il luogo dove B è maggiore (è una derivata)  
quindi è attirato da B, va dove B è forte  
oppure, $U=-\mu B$, cioè per avere il minimo di U devo aumentare il moduo di B, quindi attirato dopo B è più forte  

se $\mu$ è antiparallelo a $B$: (dia)  
$F=-\mu \nabla B$  
F punta in direzione opposta al gradiente  
è respinto da B, va dove B è debole  

**Nota su mu**

magnetizzazione indotta: $M=\chi_m H$  
dipolo magentico: $\mu = M \, dV$  

i materiali sono classificati a seconda del loro valore di $\chi_m$  
quindi se $\chi_m >0$, $\mu$ è parallelo a $B$

#### Per deformazione

**Come funziona**

se su un atomo, è detta 'deformazione elettronica', perchè a deformarsi è la nube elettronica  

come funziona:    
1- atomo/molecola ha $\vec{p}=0$  
2- applico campo esterno  
3- atomo si deforma, ha $\vec{p}\neq 0$  

che vuol dire si deforma:  
un atomo ha nube elettronica, che schematizzo come una sfera con carica volumetrica uniforme negativa, e nucleo positivo nel centro della nube  
quindi all'inizio, no momento di dipolo  
quando applico E, la nube e il nucleo subiscono una forza in direzione opposta (perchè hanno carica di segno diverso)  
quindi il nucleo non è più nel centro della nube (ma è sempre dentro la nube)  
approssimando la nube come una carica negativa puntiforme nel suo centro, si è formato un momento di dipolo  

**Trovare polarizzabilità**

all'equilibrio, il nucleo sente le forze: $\Sigma F = q_{n} (E_l+E_s)=0$  
$E_l$: campo esterno locale     
$E_a$ (a per atomo): il campo generato dalla nube 
dove $E_a = \rho \frac{r}{3 \epsilon_0}$, campo dentro sfera con carica uniforme  

sostituisco il valore di $E_a$ e riarrangio, trovo distanza di dipolo all'equilibrio:  
$r_0= \frac{3 \epsilon_0}{\rho} E_l = 3 \epsilon_0 E_l \frac{\frac{4}{3}\pi R^3}{Ze}$  

calcolo momento di dipolo:  
$p = q_n \cdot r =4 \pi \epsilon_0 R^2 E_l$  
è il momento di dipolo indotto, acquisito da una sfera uniforme sopposta a campo esterno???  

quindi, poichè $p=\alpha E$  :  
$\alpha_D =4 \pi \epsilon_0 R^2=3 \epsilon_0 V$   

dati:  
$r_0$: distanza (al'equilibrio) tra posizione del nucleo e centro della nube    
$q_n = Ze$: la carica del nucleo, Z il numero atomico  
R: raggio della nube  
V : volume della molecola

#### Per orientamento

**Come funziona**

come funziona:    
1- tutti i $\vec{p}$ sono orientati a caso  
2- applico campo esterno  
3- i $\vec{p}$ si riorientano, fino a raggiungere l'equilibrio termodinamico (è equilibrio statico o dinamico???)  

come si riorientano: ci sono 2 fattori  
1. energia di un dipolo sottoposto a campo esterno: $U=-\vec{p} \cdot \vec{E_l}$    
quindi per minimizzare energia, il dipolo dovrebbe disporsi parallelo al campo (così $\cos=1$)  
2. ma c'è agitazione termica dovuta alla temperatura, che scombina di nuovo i dipoli  

**Equilibrio termodinamico**

un sistema classico (non quantistico) all'equilibrio termodinamico segue la distribuzione di Boltzmann  

distribuzione di Boltzmann:  
$W(U)=Ae^{-\frac{U}{k_BT}}$  
$U$: energia elettrostatica  
$k_BT$: energia termodinamica  

**Trova polarizzabilità**

$W(U) \; d\Omega$  
può avere 2 interpretazioni:  
1- probabilità che il dipolo $p$ sia orientato nell'angolo $d\Omega$  
2- probabilità che il dipolo $p$ abbia energia U  
perchè p ed E sono fissi, quindi U dipende dall'orientazione del dipolo con: $U=-p \cdot E$  

semplificazioni:  
esprimo il differenziale in coord sferiche: $d\Omega = \sin \theta \; d\theta \; 2 \pi$   
uso la condizione di normalizzazione della probabilià (cioè $\int_0^{4\pi} W(U) d \Omega=1$) e ottengo: $A=\frac{1}{4 \pi}$    
assumo $U \ll k_B T$, he è quasis empre vero, e approssimo l'esponenziale per l'esponente tendente a 0: $e^x \sim 1 +x$  

risultato (assumendo il campo E sia diretto lungo l'asse z, e <> significa valore medio):    
$<p_z> = \int p_z \; W(U) d\Omega =... = \frac{p^2 E_l}{3 k_B T}$  
$\alpha_O = \frac{p^2}{3k_B T}$  

**Osservazione**

$<p> \sim E_l$ vuol dire che scalano linearmente  

$<p> \sim E_l$ vale solo per campi piccoli (mentre nella deformazione vale sempre)   
perchè ad un certo punto, quando tutti i dipoli sono allineati al campo E, $<p>$ non può più aumentare   
quindi all'inizio scala linearmente con è, ma poi si satura, si avvicina asintoticamente a $p$, che è il valore massimo, quando tuti i dipoli sono allineati al campo
