### Resistività

**A seconda della resistività**

è un o spettro, quindi:  
tutti i materiali sono in grado di condurre elettricità almeno un po', anche gli isolanti migliori (tranne gli isolanti ideali)   
cambia quanto bene lo fanno  

1. conduttori   
hanno molti elettroni di conduzione (cariche libere)   
resistività $<10^{-8}$  

2. isolanti    
le cariche libere sono assenti o fortemente legate agli atomi  
resistività $>10^{12}$   

3. semiconduttori  
la loro conducibilità elettrica dipende fortemente da temperatura, drogaggio o luce  

4. superconduttori  
sotto una certa temperatura critica, perdono completamente la resistenza elettrica  

**Campo elettrico dentro un materiale**

conduttore:  
campo elettrico interno è nullo in condizioni STAZIONARIE (le cariche si ridistribuiscono fino a schermarlo)  
cioè densità volumica di carica nulla  

isolante:  
campo elettrico può penetrare nel materiale, ma nessuna corrente condotta  
gli elettroni possono solo spostarsi leggermente (polarizzazione)  

**Relazione resistività e temperatura**

conduttori:  
1. se T aumenta, $\rho$ aumenta  
2. anche a $T=0K, \quad$ $\rho>0$, è finita e non nulla    

poichè si comportano in modo strano, questi sono non-ohmici:  
semiconduttori:  se T aumenta, $\rho$ diminuisce    
superconduttori: c'è una temperatura critica $T_c$ a cui di botto la resistività crolla a 0  

#### Sub-tipi

**Semiconduttori**

come funziona:  
se T aumenta, $\rho$ diminuisce    
perchè ci sono degli elettroni nella banda di valenza  
se T aumenta, hanno abbastanza eenergia per fare il salto  
quindi più elettroni conduce di più  

la loro conducibilità elettrica dipende fortemente da temperatura, drogaggio o luce  
A bassa temperatura: isolanti  
A temperatura ambiente o con drogaggio: conduttori controllabili   
usati per controllo e modulazione del flusso di cariche  

**Superconduttori**

proprietà:  
1. effetto Meissner: espellono il campo magnetico (le linee di campo B non possono penetrarci dentro)  
quindi non sono solo “conduttori perfetti”: anche le proprietà magnetiche cambiano radicalmente  
2. visto che R=0, non c'è dissipazione per effetto Joule  
utile per generare campi magnetici intensi, facendo ruotare correnti elevate in un anello superconduttore  

applicazioni:  
per generare campi magnetici intensi: acceleratori, risonanza magnetica nucleare, treni a levitazione magnetica   
per mancanza di dissipazione: risonatori, trasmissione senza perdite  

come si crea:  
per ora, tutte le $T_c<70K$, che è la tempeartura dell'azoto liquido  
quindi devo raffreddarlo con elio liquido, che ha temp più bassa, ma è più costoso e meno abbondante  
quindi si cercano superconduttori che lo diventino a temp superiori dell'azoto liquido  

nota: sono soprattutto solidi, ma non per forza 

**Materiali non ohmici**  

sono materiali che non rispettano la legge di Ohm ($V=iR$)  
per esempio i semiconduttori e i superconduttori  

semiconduttori:  
materiali usati in tutti i dispositivi elettronici, come diodi e transistor (eg. silicio)  
non-ohmici perchè hanno comportamento non-lineare, V/I non è costante, grafico è curvo  
eg. un diodo conduce corrente solo se la tensione supera una certa soglia (tipicamente 0.6–0.7 V per il silicio)  
R dipende da V, curva non lineare  

superconduttori:  
materiali in cui la resistenza è nulla, ma non sono ancora usati perché devono essere a temperature vicine allo zero assoluto (-273.15°C)  
non-ohmici perchè sotto una temperatura critica, la resistenza scende a zero  
quindi, anche con tensione nulla, può fluire corrente  
Oppure, applicando una tensione non nulla, la corrente può aumentare all'infinito senza dissipazione (idealmente)  

R dipende da temperatura, intensità del campo elettrico, direzione della corrente (polarizzazione)  
eg. lampadina a incandescenza (R cambia con T), diodi e led (conducono in un solo verso)  

**Ohmici e non ohmici: dissipazione**

quando attraversato da corrente, l'energia elettrica si trasforma in:  
ohmici: tutta in energia termica, che viene dissipata    
non ohmici: un po' dissipata in energia termica, ma un po' è convertita in altre forme di energia, a spese del generatore  
tipo meccanica (motore elettrico), elettrostatica (condensatore), magnetica (induttore), chimica (per caricare una pila, creo enegia chimica che viene stored nella pila scarica)  

**Isolanti reale e ideali**

isolante ideale: non ha elettroni liberi, non conduce per niente  

isolante reale: ha delle impurità, quindi anche a bassi voltaggi può condurre un po' di corrente  
si chiama 'corrente di fuga/dispersione'   
come nei conduttori, questa corrente aumenta se aumento il voltaggio    

### Ionizzazione, breakdown

**Tensione di breakdown**

la tensione massima che un isolante può sopportare prima che gli elettroni accelerati dal campo elettrico ionizzino il materiale, causando scarica e conduzione improvvisa     
cioè la tensione a cui un isolante smette di esserlo, diventa un conduttore  

spiegazione:  
quando il campo E raggiunge un valore critico, detto 'rigidità dielettrica', c'è un cambiamento di botto        
gli elettroni prendono abbastanza energia dal campo per ionizzare gli atomi del materiale  
così si ottengono elettroni liberi, che sbattendo con altre molecole ne creano altri (valanga di Townsend)   
quindi ora il materiale ha elettroni liberi, è passato all'improvviso all'essere un conduttore

**Ionizzazione di un gas**

ionizzare un gas:  
atomo neutro -> elettrone + ionic core  
un gas ionizzato è detto 'plasma'  

nota che:  
nucleo: atomo senza tutti gli elettroni    
ionic core: atomo senza qualche elettrone (che è stato strappato ed è diventato elettrone libero)  

**Come ionizzare un gas**

devo dare agli elettroni energia sufficiente a superare l'attrazione dello ionic core    
così si allontanano e diventano elettroni liberi  

di solito questa energia è fornita mettendo un campo elettrico locale  
la tensione necessaria per far ionizzare il gas è la 'TENSIONE DI BREAKDOWN'  

**Ionizzazione di un solido**

funzione come nei gas, strappando elettroni agli atomi  
Dopo il breakdown, il matriale non è più isolante, spesso viene danneggiato irreversibilmente  

**Spiegare breakdown ad un bambino**

se ho una concentrazione/mucchio di elettroni, vogliono separarsi perchè cariche si respingono  
ma l'isolante si oppone al passaggio     
pensa l'isolante come una diga   

ma la diga non è forte infinita  
se le cariche aumetano, la sfondano (tensione di breakdown)  
quindi c'è un buco, tutta l'avqua si riversa di botto  
a quel punto, più le cariche sono aumentate fino ad ora, più è distruttivo quando si riversano di botto  

per questo i fulmini sono forti   
l'aria è il migliore isolante a parte il vuoto  
quindi la carica aumenta moltissimo, e quando finalmente riesce a passare è fortissima  

### Note sui conduttori

**Conduttori: argento vs rame ???**

tra i conduttori, il migliore è l'argento, anche più del rame (un pochino di più)  
ma è poco usato perchè:  

1. costa molto di più (mentre conduce solo un pochino meglio, non vale la pena)  

2. ossidazione superficiale   
si ossida a formare solfuro d'argento $Ag_2S$, una patina nera, se esposto all’aria contenente tracce di zolfo $H_2S$ (come l’inquinamento)  
Questa patina non conduce bene, quindi può creare problemi di contatto elettrico.  
Il rame si ossida anche lui, ma l’ossido di rame è aderente e stabile, e si possono usare rivestimenti (come lo stagno) per proteggerlo.  

3. proprietà meccaniche  
Il rame è più duttile e resistente all’usura.  
L’argento è più morbido: i fili d’argento si deformano facilmente, non sono ideali per cavi lunghi o soggetti a vibrazioni.  

quindi l'argento si usa solo per cose molto specializzate, dove serve davvero ottima conduzione  
circuiti microonde, contatti elettrici e connettori di alta qualità, saldature e rivestimenti di componenti elettronici, specchi conduttivi e riflettenti

**Ossidazioni del metallo**

Quando un metallo sta all’aria, reagisce con l’ossigeno (O₂) o altri gas come lo zolfo (H₂S)  
Sulla sua superficie si forma uno strato sottile di composto (ossido o solfuro)  
Questo strato può essere:  
protettivo → blocca ulteriori reazioni, quindi “passiva” il metallo  
non protettivo → si stacca o conduce male, peggiorando il contatto elettrico  

ossido di rame: $Cu_2O$ o $CuO$  
aderisce bene al metallo sottostante: non si stacca facilmente  
È abbastanza sottile e, pur essendo un cattivo conduttore, non rovina troppo la conducibilità nei cavi (dove la corrente scorre all’interno)  
Inoltre si può proteggere facilmente con uno strato di stagno o vernice isolante  
ossidazione più gestibile  

solfuro d'argento  
Questo strato non è protettivo né conduttivo:  
è molto resistivo → ostacola il passaggio di corrente nei contatti;  
può rompersi o staccarsi;  
si riforma continuamente se l’ambiente contiene zolfo.

**Evitare ossidazione**

dove servono contatti elettrici diretti (es. prese, spinotti, saldature), anche l’ossido di rame è fastidioso.  
i fili si stagnano (cioè si ricoprono di stagno) o si proteggono con isolante: così il rame non si ossida nei punti di contatto  

il problema dell'ossidazione non è lungo il cavo, tanto lì la corrente è dentro, ma eni punti di contatto  
la patina di solfuro d'argento è un isolante quasi totale, il contatto non funziona più  
se si stacca, lascia punti scoperti irregolari, quindi la corrente passa solo in punti microscopici, con altissima densità di corrente  
causa surriscaldamento e danni

### Fenomeni

#### Fulmini

**Steps**

1. formazione cariche  
si forma carica libera nella nube per strofinio, e  indotta sul suolo   
aumentano finchè il campo LOCALE (vicino alla superficie della nube e del suolo) superano il valore di breakdown 

2. ionizzazione locale dell'aria  
si creano leader discendente e streamer ascendente  

3. quando si toccano, si crea un canale continuo ionizzato tra unbe e suolo  
ci passa la scarica di ritorno  

**Carica libera  e campo nella nube**

come sono fatte le nubi temporalesche:  
cristalli di ghiaccio piccoli (in alto)  
gocce d’acqua e graupel (grandine molle, più pesante) (in basso)  
perchè ghiaccio ha densità minore dell'acqua,    

cosa succede:  
per effetto triboelettrico (strofinio) e per differenza di temperatura, gli elettroni dell'acqua passano al ghiaccio    

spiegazione:    
si strofininano per via delle correnti turbolente dentro la nuvola  
quindi gli elettroni passano dall'acqua al ghiaccio  
perchè:  
caldi = più energia, più vicino a sfuggire,   
grandi = meno attrazione dal nucleo  

conclusione:  
parte inferiore della nube è negativa, superiore è positiva    
ciò causa un campo elettrico, che induce cariche positive sul suolo (terrestre)  

**Ionizzazione locale dell'aria** 

il valore di breakdown dell'aria è $E \sim 10^5 V/m$  
il campo prodotto dalle cariche della nube (o del suolo) scala con la distanza, quindi:  
il campo medio (valore medio del campo tra suolo e nube) non è abbastanza, quindi non si forma un canale completo nell'aria  
ma il campo locale (valore del campo molto vicino al suolo o nube) è abbastanza alto  
quindi ho IONIZZAZIONE LOCALE  

nelle zone che si sono ionizzate localmente:   
Questo piccolo canale è ora molto più conduttivo dell’aria circostante  
il campo si concentra ancora sulla sua punta  
così il processo continua, 'a gradini'

**Canali ionici**

- Leader discendente  
Parte dalla nube: una successione di piccoli passi di plasma ionizzato    
Crea un canale di plasma debole che cerca il percorso più conduttivo  

- Streamer scendente  
parte da terra, preferibilmente da zone dove il campo è più forte (oggetti appuntiti o metalli)  

quando si toccano, si crea un canale continuo ionizzato dalla nube al suolo (canale conduttore principale)  
attraverso cui passa la scarica, chiamata 'return stroke', 'scarica di ritorno'  

se un lampo 'tremolare' è perchè ci sono più return stroke in uno stesso canale  

**Da dove parte la streamer ascendente**

è più probabile che parta dove è più facile ionizzare l'aria  
quindi da oggetti metallici e appuntiti, perchè:  

effetto punta: campo sulla superficie di un conduttore è più forte nelle zone di raggio di curvatura più piccola, cioè le punte    
campo più intenso -> più facile ionizzare l'aria   

metallo: hanno elettroni liberi → permettono al campo elettrico di distribuirsi e di condurre la corrente del fulmine una volta iniziata   
dovrei ionizzare lo spazio, ma già lo è, ci sono già elettroni liberi, neanche bisongo di farlo  

parafulmini: offrono un punto preferenziale in cui il campo supera più facilmente la soglia di breakdown  
così la scarica si innesca lì (in modo controllato) invece che altrove  

**Pericolosità**

ha energia enorme   
nel corpo umano provoca arresto cardiaco, ustioni profonde e danni nervosi  
il valore della corrente (decine di kA) e della tensione (centinaia di MV) basta per distruggere tessuti, arrestare il cuore e vaporizzare materiali.  

#### Generatore di Van der Graaf

**Cos'è**

un dispositivo elettrostatico reale che accumula cariche elettriche fino a ottenere tensioni altissime  

formato da:  
1- sfera metallica cava (conduttrice) in cima  
2- cinghia isolante che scorre continuamente su due rulli (uno in basso e uno in alto).
3- Due pettini metallici con punte aguzze, uno in basso e uno in alto, collegati a potenziali diversi (usati per aumentare il campo grazie all'effetto punta)  
4- Un motore che fa muovere la cinghia.

**Come funziona**

1- pettine inferiore ionizza l'aria (ha un campo forte per effetto punte)   
2- ioni sono depositati sulla cinghia isolante, mentre sale  
cinghia carica passa vicino al pettine superiore, collegato alla sfera  
3- campo elettrico tra la cinghia e il pettine superiore fa saltare le cariche dalla cinghia alla sfera (è grande per effeto punte)   
4- Le cariche si accumulano sulla superficie esterna della sfera, che essendo conduttrice le distribuisce uniformemente  

sfera accumula cariche finché il campo elettrico sulla sua superficie raggiunge il valore di breakdown dell’aria (3MV/m)   
a quel punto l’aria si ionizza e avviene la scarica — la “scintilla” o il mini-fulmine verso un corpo vicino  

#### Palla al plasma

**Cosa è**

inventata da Nikola Tesla, basata sullo stesso principio delle bobine di Tesla:  
usa alta tensione e alta frequenza per ionizzare il gas all’interno di una sfera di vetro.  

**Come è fatta**

1- sfera di vetro isolante, riempita di gas a bassa pressione(gas nobili)      
2- al centro della sfera, piccolo elettrodo metallico    
3- nella base, oscillatore armonico, che applica alta tensione AC (tensione alternata ad alta frequenza) all'elettrodo  


**Come funziona**

1- tensione alternata genera un campo elettrico fortissimo intorno all’elettrodo centrale  
2- campo ionizza il gas  
3- Quando gli elettroni liberi si ricombinano con gli ioni, si creano le emissione di luce   

**Cosa sono le emissioni di luce**

Le scie sono filamenti di plasma (piccole scariche) che si formano lungo percorsi in cui il campo elettrico è più intenso   
cioè il tragitto fatto da un elettrone per ricongiursi col suo core ionico  

quando un elettrone si ricombina con un ionic core, emette un fotone  
perchè deve abbassare la sua energia per potersi rimettere nell'oritale che usava prima  
quindi il colore dipende da quanta energia deve rilasciare, cioò dal tipo di atomo e quanto forti sono i legami  

come appare:  
La forma ramificata deriva dal comportamento caotico del plasma  
il colore dipende dal gas:    
Neon → rosso-arancio,  
Argon → blu-violetto,   
Krypton/Xenon → azzurrino o bianco-rosato.  

perchè alta tensione AC:  
con DC, si avrebbe una polarizzazione e poi una sola scarica  
poi non c'è più plasma   
perchè tutti gli elettroni si accumulerebbero su un elettrodo  
invece, con AC, il gas rimane ionizzato stabilmente: ogni inversione di campo crea nuove collisioni e ionizzazioni  

**Quando tocco il vetro**

il tuo corpo fornisce un percorso a massa (terra);  
il campo elettrico si concentra sotto il punto di contatto;  
i filamenti di plasma si orientano verso il dito, perché lì il campo è più intenso.  

ma non prendo la scossa, perchè:  
il vetro è un buon isolante, ostacola molto il passagio della corrente  
la corrente è bassisima  
è ad alta frequenza, quindi non penetra in profondità nei tessuti
