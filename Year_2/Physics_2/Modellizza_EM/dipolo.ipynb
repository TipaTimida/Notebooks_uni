### Teoria

**Cos'è**

dipolo è un'approssimazione, tanto più vera quanto l'oggetto è piccolo in confronto alla scala su cui sto operando  
dipolo elettrico: due cariche uguali e opposte $q$ e $-q$, alla distanza costante $\vec{a}$ ($\vec{a}$ va dalla carica - a quella +, per convenzione)  
dipolo magnetico: una spira di sezione S e percorsa dalla corrente I  


**Teorema di equivalenza di Ampere**

una piccola (Così B è uniforme su di essa) spira percorsa da corrente si comporta come un dipolo magnetico di momento $\mu=IS$   
cioè, analogia tra fenomeni elettrici e magnetici   

**Differenza tra dipolo elettrico e magnetico**

linee di campo dei dipoli:  
a grandi distanze sono uguali, ma a piccole no  
guarda la zona tra le due cariche e dentro la calamita, c'è una riga verticale centrale  
nel amgnetico punta su, nell'elettrico giù  
quindi il campo sull'asse è diverso

### Pratica

**Formule**

cosa | elettrico | magnetico 
---|---|---
momento di dipolo | $\vec{p} = q\vec{a}$ | $\vec{\mu} =i\vec{S}$  
momento meccanico/torcente | $\tau = p \times E$ | $\tau = \mu \times B$  
energia potenziale in campo esterno | $U_E = -p \cdot E$ | $U_B = -\mu \cdot B $  
forza in campo esterno non uniforme | $\vec{F}_E=\vec{\nabla}(\vec{p}\cdot \vec{E})$ |  $\vec{F}_B=\vec{\nabla}(\vec{\mu}\cdot \vec{B})$ 

perchè:  $F = -\nabla U$  

**Note**

l'energia potenziale magnetica vale solo se B è uniforme in quella zona (o la spira è tanto piccola che posso assumerlo come uniforme)  
sennà la formula è più difficile  

**Solo per dipolo elettrico**

Cosa | formula
---|---
potenziale | $k \frac{p \cdot \vec{r}}{r^3}$
campo in coord sferiche | $\vec{E} = k\frac{\vec{p} \cdot \vec{r}}{r^3}$  
campo in coord cartesiane | $\vec{E} = \frac{k}{r^3} (3\vec{p} \cdot \hat{r} -\vec{p})$

quelle per campo magentico sono troppo difficili, mai usate


**Calcola momento di dipolo magnetico**

spira: un pezzo di filo chiuso, di qualunque forma, quadrata o tonda  
non importa la forma, le spire si caratterizzano solo per il momento magnetico    
l'importante è che passi corrente solo in un filo sottile, l'interno deve esser vuoto  

se invece è ho un disco pieno:  
$\mu=IS$ vale SOLO per una spira, NON posso usarlo e dire che S è l'area totale  
devo separare in anellini (spire), calcolare dm su ognuno e poi integrare su tutto il disco   

**Calcola momento dipolo elettrico**

$p= \Sigma q_iy_i$  
ogni addendo è una carica puntiforme  

per prendere come origine delle coord un punto preciso:  
calcola la coord di quel punto  
tipo, per il cm di un sistema: $y_{cm} = (\Sigma m_iy_i )/M= 0$  
poi, le $y_i$ di ogni oggetto scrivile in un sistema di riferimento con quella coord come origine  

**Calcola momento di dipolo: generale**

definizione generale di dipolo elettrico:  
$p(t)=\int \vec r \, \rho(r,t) \; d^3r$  

se distribuzione è in 1D:  
$p(t)=\int \vec z \, \int_{-a}^a z \, \lambda(z,t) \, dz $  

cioè, una distribuzione di cariche genera si può approssiamre ad un dipolo (metodo dei multipoli), ha un momento di dipolo elettrico

**Esempio: forza spira**

se ho campo: $B=B(x) \hat z$  
e spira di lati ab e corrente I 

1- momento magnetico di spira: $m=Iab$   
2- forza su spira: $F=\nabla(m \cdot B) = \hat x \; \frac{\partial}{\partial x}(mB)$  
perchè il gradiente dà un vettore con la derivata lungo tutte le componenti, ma qui m è costante e B varia solo lungo x, quindi l'unica non nulla è la componente lungo x  

3- se la spira è piccola:  
posso stimare il campo ai bordi (trattare al primo ordine la variazione del campo magnetico)  
assumendo $B_0$ è il campo al centro, quello sui bordi sinistro e destro:  
$B_{left} = B_0-\frac a2 \frac{\partial B}{\partial x} $   
$B_{right} = B_0+\frac a2 \frac{\partial B}{\partial x} $   
perchè: a/2 è la distanza, dB/dx è la variazione del campo B fratto la distanza (al primo ordine)  


### Multipoli

**Espansione di multipolo**

a cosa serve:  
se ho una distribuzione di carica $\rho(\vec{r})$ limitata nello spazio (di qualche strana forma)  
per calcolare V(r) prodotto da questa distribuzione in punti molto lontani dalla distribuzione, posso approssimare la distribuzione in modi via via più sofisticati  

**Come si fa l'espansione**

come si fa:  
ogni modo dà un addendo per uno sviluppo in serie di V(r)  
generalmente basta il primo non nullo, ma per essere più precisa posso aggiungerne di più  
nota che, con computer, se conosco $\rho(r)$ è possibile calcolare V in ogni punto, basta sommare l'apporto di ogni pezzetto, ma è lungo  
questo metodo mi dà un modo analitico per avere una soluzione appossimata  

cioè:  
prima immagino che la distribuzione sia una sola carica puntiforme  
poi che sia due cariche puntiformi (dipolo)  
poi 4 (quadripolo)...

**Primi termini dell'espansione**

0. termine di monopolo  
assumo che la distribuzione è una carica puntiforme  
$q=\int \rho(r) \; d\tau$ : carica totale della distribuzione  (tau è un pezzo di volume, r la posizione di quel pezzetto)   

1. termine di dipolo  
cioè c'è una separazione di cariche + e -  
$p=\int r \; \rho (r)\; d\tau$: momento di dipolo  

2. termine di quadripolo  
$Q_{ij}=(3x_ix_j-\delta_{ij}r^2) \; \rho(r) d\tau$  

**Espansione in serie**

$V(r) = V^{(0)}(r)+V^{(1)}(r) + ...$   

$V^{(0)}(r) = k\frac{q}{r}$  
$V^{(1)}(r) = k\frac{\vec{p} \cdot \hat{r}}{r^2}$   
$V^{(1)}(r)  \sim \frac{Q_{ij}}{r^3}$    
ogni termine scala con una potenza di r più piccola, così per r molto grande (oggetto molto distante), il primo addendo che appare è il più grande, gli altri li posso trascurare  


### Esempi di poli

**Dipolo**

dipolo elementare: due cariche puntiformi, + e -, a distanza d  

esempio di molecola: acqua, $H_2O$  
q=0 (visto che non è uno ione, non ha carica complessiva), cioè non posso approssimarla come carica puntiforme, il primo termine della serie è nullo  
ma tra H e O c'è differenza di elettronegatività, quindi c'è una separazione di cariche, che geenra un momento di dipolo $p$ su ogni coppia OH  
poichè ha forma piegata, non dritta, se disegno le frecce p vedo che non si cancellano a vicenda, quindi in totale ho un momento di dipolo non nullo   
bisto che è un dipolo, sente le forze dipolo-dipolo  

applicazione: forno a microonde  
sfrutta il fatto che l'acqua è un dipolo   
un dipolo, sottoposto ad un campo elettrico, subisce un momento $M=p \times E$  
il microonde produce un campo variabile  
causa M, che induce delle rotazione nelle molecole d'acqua (cioè l'atomo O gira intorno al proprio centro???)  
per attrito, queste rotazioni producono calore che scalda il cibo      


**Quadripolo**

quadripolo elementare: 4 cariche puntiformi, nei vertici di un quadrato, con le diagonali lunghe 2d (cariche su una diagonale hanno lo stesso segno)  

esempio di molecola: benzene $C_6H_6$  
complessivamente neutra, ha q=0  
le coppie CH hanno dipolo, ma se disegno tutte le frecce si annullano tra loro, quindi in totale p=0  
però c'è comunque un momento di quadrupolo  
posso immaginare la molecola tipo panino, cioè fatta di 3 strati, sopra e sotto con carica - e in mezzo con carica +    

### Campi magnetici da ricordare

**Spira**

SOLO sull'asse: 
$B \, (2R) = \mu_0 I$  
la direzione dipende dalla corrente, legge mano destra

nota: se mi dice che la spira è molto piccola, spesso vuol dire che il campo B concatenato ad essa può ess econsiderato uniforme, e uguale al valore nel centro  

come si trova:  
non si può trovare con teorema di Ampere, devo fare il emtodo complesso di itnegrare con simmetria  

**Solenoide**

su TUTTO l'interno: uniforme e parallelo all'asse  
$ B l = N \mu_0 I$  
fuori: B=0  

come si trova:  
fai un rettangolo a cavallo del bordo del wolenoide e usa th di Ampere  
B è solo dentro  
in un tratto lungo l, ci sono nl spire, quindi la corrente interseca nl volte  
$B \, l = \mu_0 I_{conc} = \mu_0 \, nl I$  -> $B = \mu_0 nI$  

**Filo**

azimutale (cerchi intorno al filo)  
$B \;2 \pi R = \mu_0 I$  

come si trova: th di ampere  
lato sinistra ha lunghezza della circonferenza 

**Toroide**

scelgo $\gamma$: coassiale al toroide  

dentro: $B(2 \pi r)= \mu_0 i N$  
fuori: 0 (no corrente concat, perchè entra ed esce la stessa)  

qui la la densità di coils è: $n=\frac{N}{2 \pi r}$

**Cilindro**

fuori: $B(2 \pi r)=\mu_0 i$, come un filo senza spessore  

dentro:  
pieno conduttore o vuoto: B=0, tutta la carica è sulla superficie, no corrente concatenata  
pieno isolante: $B(2 \pi r) = \mu_0i\frac{\pi r^2}{\pi R^2}$  
perchè la corrente concatenata è: $I_{conc} = U_{tot} \frac{\pi r^2}{\pi R^2}$
