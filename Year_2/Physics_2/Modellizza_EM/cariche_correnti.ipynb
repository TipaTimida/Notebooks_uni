### 3. Continuità della carica

**Equazione di continuità**

forma locale:  
3D: $\partial_t \rho+ \nabla \cdot J = 0$  
2D: $\partial_t \sigma + \nabla \cdot K = 0$  
1D: $\partial_t \lambda + \partial_x I = 0$  

forma integrale:  
3D: $\frac{dq}{dt} + \int J \cdot dA=0$  
2D: $\frac{dq}{dt} + \int K \cdot dl=0$   
1D: $\frac{dq}{dt}+ I = 0$  
secondo addendo sono tutti i modi per esprimere $I$ tramite un'altra corrente  

**Come si trovano**

la forma locale in 3D è quella classica  

1- per passare a forma integrale: integra su dV, dS, dl  
così la carica diventa q, e il secondoè il secondo è integrato in modo avere la formula che dà I  
ricorda che in forma integrale ho la derivata totale, non parziale, perchè adesso Q è solo funzione del tempo, ho tolto la dipendenza dallo spazio integrando  

2- per cambiare dimensione:  
integra sulla direzione cartesiana che mi interessa: dx, dy, dz  

#### Teoria ???

**Versi**

ricorda che dS rappresenta il vettore USCENTE da una superficie, deve puntare in fuori, non dentro il volume racchiuso  
quindi assumo J positiva se uscente (come per campo carica puntiforme, se +q ha linee di campo uscenti)

**Come trovarla: caso stazionario**

1- voglio dire che: la corrente che entra è uguale a quella che esce  
$\int j \cdot dS = 0$  

2- uso il teorema della divergenza:  
$\int j \cdot dS = \int_V \nabla \cdot j \; dV = 0$  

3- equazione di continuità nel caso stazionario:  
$\nabla \cdot j = 0$  

**Come trovarla: caso generale ???**

1- a quella del caso stazionario, aggiungo la corrente di conduzione  
$\nabla \cdot (j_c+\epsilon_0\frac{dE}{dt})=0$  
$\nabla \cdot j_c + \epsilon_0 \frac{d (\nabla \cdot E)}{dt}=0$  

2- uso la legge di Gauss:  
$\nabla \cdot E = \frac{\rho}{\epsilon_0}$  

3- l'equazione di continuità nel caso generale:    
$\nabla \cdot j_c + \frac{d\rho}{dt}=0$

#### 1. Geometria

**Tipi di correnti: geometria**

1- $K$: densità di corrente superficiale  
formula:  $I=\int K \, dl = K \, l$ , dove l è la linea ORTOGONALE alla direzione di K     
grafico: quanta carica passa attraverso una una linea in 1D disegnata su una superficie in 2D  
in pratica: densità lineare   
  
2- $J$: densità di corrente volumica   
formula: $I=\int J \, dS = J \; S$, dove S è la superficie ORTOGONALE alla direzione di J  
grafico: quanta carica passa attraverso una sezione in 2D disegnata nello spazio in 3D   
in pratica: densità superficiale  

il nome si riferisce alla distribuzione spaziale, non alle unità di misura    
ricorda che integrare su una direzione riduce una dmiensione, e I ha 1 dimensione, Q ha 0 dimensioni  

**IMPORTANTE: passarea da K a I**

esempio:   
cilindro di raggio b, con I sulla parte esterna  

rapporto tra I e K: 
ricorda che K si moltiplica per la linea PERPENDICOLARE alla direzione della corrente    
se I assiale: $I = K \, 2 \pi r$ (moltiplico per circonferenza)  
se I azimutale: $I=K \, h$ (moltiplico per altezza)  

perchè: la corrente totale I è la somma della corrente che attraversa una linea 1D disegnata sulla superficie (quindi perpendicolare)  
nota che anche J si moltiplica per l'area perpendicolare a sè per avere I  

a volte potrei dover mettere il segno -, dipende dalla direzione della corrente

### Altra teoria

#### Tipi di campo

**Definizioni**

da cosa è generato il campo:  
1- elettrostatico: da cariche ferme    
2- magnetostatico: da correnti costanti (non variabili)  
3- indotto: da una variazione tempoarale dell'altro campo  

**Proprietà**

1- stazionario: le grandezze non cambiano nel tempo  
2- conservativo: non genera fem  
non conservativo: genera fem  

stazionario -> conservativo  
ma non vale il contrario  

elettrostatico, magnetostatico -> stazionario, conservativo  
indotto -> non stazionario, non conservativo  

**Generazione di campi**

corrente I: genera un campo elettrico B  
carica q: genera un campo elettrico E  

poi, se E o B sono variabili, uno genera l'altro  
quindi se I è variabile, genera B variabile, che genera E  

attenzione: I stazionaria NON genera E  
pensa ad un filo con corrente I, è comunque neutro complessivamente  
quindi ALL'INTERNO del filo c'è il campo $E=\sigma_c J$  
ma ALL'ESTERNO non c'è campo, se ci fosse sarebbe dato da $E=\frac{\sigma_f}{\epsilon}$, ma non c'è un carica libera netta sul filo per continuità della corrente   (guarda la formula di continuità della corrente, se la corrente è stazionaria $\sigma_f=0$, se non lo è un po' si accumula e c'è carica libera, quindi c'è campo)


#### Corrente di spostamento

**Proof**

applico th di Ampere scegliendo una linea $\gamma$ che forma una sezione tra le piastre dei condensatori  
qualunque superficie con questa linea come contorno deve rispettarla  
se prendo superficie che esce fuori ad incrociare il filo, c'è I concantenata  
ma se prendo superfiie piatta che resta tra le lastre dei condensatori, lì non c'è I di conduzione  
quindi deve esserci la corrente di spostamento   

**Che succede in un condensatore**

corrente di conduzione circola solo sul filo  
tra le armature non si muovono cariche, quindi non può circolare  
tra le armature c'è solo corrente di spostamento

#### Condizioni stazionarie

**Condizioni**

- regime stazionario: niente dipende dal tempo  
cioè tutte le derivate PARZIALI del tempo sono nulle  
equazione di continuità diventa: $\nabla \cdot J = 0$, cioè J è SOLENOIDALE    

- lentamente variabile: $\tau \ll \tau_0$  
cioè la scala di tempo in cui varia I è molto più lenta del tempo necessario alla luce per attraversare il circuito  


**Correnti lentamente variabili**

$\tau$: costante di tempo su cui variano le correnti  
cos'è dipende dal tipo di corrente: per $I$ sinusoidale è il periodo, per $I$ esponenziale è la costante di decadimento  
non sempre è definita  

$\tau_0 = \frac{D}{c}$: in pratica il tempo impiegato dalla luce per attraversare il circuito  
D: dimensione caratteristica del circuito  

lentamente variabili: $\tau \ll \tau_0$  
cioè la scala di tempo in cui varia I è molto più lenta del tempo necessario alla luce per attraversare il circuito  
in questo caso, le leggi di Kirchoff sono valide per 'valori istantanei', cioè faccio una foto al circuito  
è il motivo per cui posso usarle in circuiti RC  
cioè, trascuro il ritardo sulla propagazione del segnale E.M. nel circuito  
cioè, se facessi una foto istantanea, tutti i punti in serie hanno la stessa corrente    

**Campo solenoidale**

campo solenoidale = a divergenza nulla  
intuitivamente: no sorgenti o pozzi  
flusso: flusso attraverso qualunque superficie chiusa è nullo (perchè $\int J \cdot ds = \int \nabla \cdot J=0$)  
linee di campo: sono chiuse e continue (formano anelli chiusi o si estendono all’infinito)
