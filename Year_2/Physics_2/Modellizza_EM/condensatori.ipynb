### Cos'è un condensatore

**Cos'è**

è un serbatoio di carica e di energia   
formato da una coppia di conduttori tra cui si instaura il fenomeno dell’induzione COMPLETA  
i conduttori sono chiamati 'armature' del condensatore  

nota: le armature devono essere molto vicine, così posso ignorare gli effetti ai bordi  

#### Induzione

**Induzione**

cos'è:  
proprietà che hanno i corpi carichi di “caricare” altri corpi  
un corpo carico, avvicinato ad altri corpi, cambia la distribuzione di carica nei corpi stessi

come funziona:  
corpo A, carico positivamente, genera campo elettrico tutto intorno, uscente  
avvicino corpo B, esso è attraversato da ALCUNE linee del campo elettrico di A (non da tutte le linee di campo)  

esempio: 
avvicino un oggetto carico a uno neutro (senza farli toccare, così carica non passa):   
quello neutro rimane neutro, ma le cariche si ridistribuiscono, tutte le positive da una parte e tutte le negative dall'altra

**Induzione completa**

un tipo particolare di induzione  
quando B è attraversato da TUTTE le linee del campo elettrico di A  

esempio:  
condensatore è fatto da conduttori, diposti in modo che le linee di forza di uno si chiudano tutte sull’altro 

conseguenze dell'induzione completa:  
1- i due conduttori hanno cariche uguali e opposte, +q e -q  
2- il campo è non-nullo solo nella regne tra le due armature, nullo in tutte le altre (si vede quando disegno il campo elettrico per trovare la capacità)

#### Come funziona in circuito

**Come funziona: carica**

1- all'inizio il condensatore è scarico (entrambe le lastre neutre)    
$\Delta V=0$, quindi per ora non ha voltage drop, entrambe le lastre neutre (scarico, ha $\Delta V=0$, quindi per ora non ha voltage drop)  

2- elettroni cominciano a scorrere nel circuito (da polo - a + del generatore)  
ma si bloccano nel condensatore, non possono passare nel vuoto, quindi si fermano nella piastra che toccano prima (che diventa quella negativa)  
così sto caricando il condensatore, ora ho un $\Delta V$ che aumenta  

3- elettroni continuano a scorrere, e quindi depositarsi sulla lastra, finchè a che il voltage drop de condensatore è uguale al voltaggio della batteria  
ora non passa più corrente  

**Come funziona: scarica**

ora posso rimuovere la batteria, il condensatore agisce come batteria  
come al solito disegna $\Delta V$ accanto al condensatore, il verso è a seconda di quela lastra è positiva e quale negativa

### Capacità condensatori

**Tipi di condensatore**

esistono 3 forme di condensatore: piano, sferico, cilindrico   
capacità dipende da forma geometrica del condensatore, non da quanto carica ci sia in quel momento  
misurata in F, farad  

**Calcolare capacità di un condensatore**

1- trova campo elettrico tra le armature (con Gauss)  
disegna frecce di E generato da ogni armatura, somma vettorialmente  
2- trova ddp TRA LE ARMATURE: $\Delta V = - \int E \cdot dr$  
quindi non mi serve un punto noto da dover far partire la scala    
3- trova capacità: $C = \frac{q}{V}$

#### Condensatore piano

**Forma**

i conduttori sono lastre piane a parallele  
la carica è solo sulle facce che si guardano, non sul resto dell’armatura  

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcTaBpDtInDy7s7ItsEd9c8Z4jqPGgL4vvDMvg&s)

**Trova capacità**

1- trova campo elettrico  
tra le armature, si sommano i campi di due lastre: $E=\frac{\sigma}{\epsilon_0}$  
fuori: le E=0, le frecce si annullano  

2- ddp: $V=Eh$  

3- $C = \frac{q}{V} = \frac{\epsilon_0}{h}A$

#### Cilindrico

**Dati**

$h$: altezza cilindro  
$R_1 < R_2$: raggio esterno parte centrale, raggio interno parte esterna

**Trova capacità**

1- trova campo elettrico  
$\int_{B1} E \cdot dS + \int_{B2} E \cdot dS +\int_{lat} E \cdot dS = \frac{q}{\epsilon_0}$  
$0+0+E \, 2 \pi \, r \, h=   \frac{q}{\epsilon_0}$  

2- differenza di potenziale tra le due armature:  
$\Delta V = - \int_{r_2}^{r_1} E \cdot dr = \frac{q}{2 \pi \epsilon_0 \, h} \ln (\frac{R_2}{R_1})$   

3- capacità  
$C = \frac{q}{\Delta V} = \frac{2 \pi h \epsilon_0}{\ln (\frac{R_2}{R_1})}$

#### Sferico

**Struttura**

visto dall'alto e sezionato, uguale all'immagine del cilindro  

**Capacità**

1- campo elettrico    
$E=k\frac{q}{r^2}$  
le armature sono sfere cave, quindi quella più piccola fa fuori da sè un campo come se fosse una carica puntiforme, quella più grande non fa alcun campo elettrico dentro di sè  

2- ddp:  solito    
3- $C = 4 \pi \epsilon_0 \frac{R_1R_2}{R_2-R_1}$  


**Caso limite: sfera isolata**

quando la sfera esterna è mandata all'infinito, cioè $R_2 = \infty$  
allora $C= \frac{R_1}{k} = R_1 4 \pi \epsilon_0$

### Formule, energia

**Energia**

spiegata nelle note generali sull'energia  

**Forza tra le lastre**

$F = q \frac{E_{cond}}{2} = \frac{1}{2}\frac{q^2}{\epsilon_0} A$  

F: forza sentita da una lastra  
una lastra sente solo il campo elettrico dell'altra, non di sè stessa, che è metà del campo del condensatore  

**Pressione elettrostatica**

c'è una forza elettrostatica, quindi c'è anche una sua pressione  
tra le armature del condensatore c'è una forza che le attrae  
devo usare la forza applicata da UNA armatura sull’altra  

formula solo per condensatore piano:  
$P = \frac{F }{A} = \frac{\sigma^2}{2 \epsilon_0}$  

nota che è uguale alla densità di energia:  
$P =  \frac{\sigma^2}{2 \epsilon_0} = \frac{1}{2} (\frac{\sigma}{\epsilon_0})^2 \epsilon_0= \frac{1}{2}E^2\epsilon_0 = u_e$  

### Energia: V o Q costante, problema coi segni

**Casi**

1- Q costante  
cosa: condensatore isolato, NON c'è attaccato generatore (che quindi non fa lavoro)  

2- V costante  
cosa: c'è attaccato un generatore, che fa lavoro per mantenere V costante  

**Come modellarlo**

se cambia C, c'è sempre:   
1- $dU_C$: variazione die nergia condensatore ($\frac 12 V^2 dC$ oppure $\frac {Q^2}{2C}$, dipende da cosa resta costante)  
2- $dW_{mec}$: lavoro meccanico fatto dall'esterno per muovere le piastre (quindi per cambiare C)  

SOLO in caso a V costante, c'è:  
3- $dW_g$: lavoro fatto dal generatore  
perchè nel caso a Q costante il condensatore è isolato, non c'è generatore  

devo sempre rispettare la conservazione dell'energia   

in entrambi i dW si mette il segno -, perchè voglio il lavoro SUBÌTO sal condensatore, non fatto (visto che il condensatore è il mio sistema)  

se non c'è niente di esterno che faccia lavoro e che stabilisca che succede, ma il sistema è alscialo libero di evolvere:    
invece di dW_{mec} metto K, cioè l'energia cinetica delle armature  

**Casi**

1- caso a Q costante:  
$dU_C - dW_{mec}=0$  
so che $dU_c = \frac {Q^2}{2C}dC$, quindi dW deve essere uguale  

2- caso a V costante:  
$dU_c -dW_{mec}-dW_{gen}=0$  
so che $dU_c = \frac 12 V^2 dC$ e $dW_{gen}= V \, dQ = V^2 dC$, quindi $dW_{mec }=\frac 12 V^2 dC$  
in pratica, $dW_g = 2 dU_c$, perché una metà va al generatore e una metà al lavoro meccanico  
ma se voglio enegia cinetica, che è la versione di dW_{mec} quando è lasciato libero di evolvere, ho $K = dU_c-dW_{gen}=-dU$

**Caso V costante: collegato a generatore**

qui ho sia condensatore che generatore nel circuito  
il mio sistema non include il generatore, ma solo il condensatore  
dal punto di vista del sistema, il lavoro fatto dal generatore è SUBÌTO, quindi cambio segno:  
$U_{tot}=U_c+U_g=U_c-W_g$  

condensatore: facile, usa la formula  
$dU_c =\frac{1}{2}V^2 dC$  

generatore:   
$dW_g = V \, dQ = V^2 dC$  
perchè il condensatore ha: $dQ = V \, dC$  
quindi il generatore emana potenza:  $P = \frac{dW}{dt} = V^2 \frac{dC}{dt}$  
per dC/dt devo usare chain rule con la variabile che varia, tipo la d tra le lastre  
e ricorda che: $dW_g = -2 dU_c$  

totale:  
$dU = -\frac{1}{2}V^2 dC$    

**Spiegazione: caso a V costante**

Quando cambi C:   
una parte dell’energia elettrica torna al generatore  
una parte diventa lavoro meccanico sulle lastre  

$dU_{tot}=dU_c-dW_g = \frac 12 V^2 dC -V^2 dC = -\frac 12 V^2 dC$  
Questa variazione è proprio il lavoro meccanico fatto sul sistema  

quindi:  
$dW_g = 2 dU_c$  
perché una metà va al generatore e una metà al lavoro meccanico  
ma no, qui è quella del generatore che è maggiore???  

**Spiegazione: caso a Q costante**

il condensatore è isolato: non c’è generatore e quindi non c’è scambio di energia elettrica con l’esterno. Tutta l’energia che cambia deve andare nel lavoro meccanico  
cioè: $dU_c = dW_{mec}$  

### Dielettrico

**Dielettrico**

i materiali dielettrici sono materiali che quando vengono inseriti in un condensatore, ne aumentano la capacità  
dielettrico = isolante  

**Che succede se aggiungo un dielettrico**

in entrambi i casi cambia C, ma:  

1- se il generatore è staccato:  
q è costante (perchè condensatore è staccato da tutto, la carica q non può andare da nessun'altra parte, resta costante)  
quindi V cambia  

2- se il generatore (la batteria) è attaccato:   
V è costante (perchè è fissata dal generatore)   
quindi q cambia  

**Come cambia energia**

pensa alle formule dell'energia di un condensatore, con C e la cosa che faccio restare uguale  

**Lavoro svolto mentre aggiungo dielettrico**

1- aggiungo dielettrico con generatore attaccato  
quindi V non cambia, quale è lavoro di generatore?  
$W_g = \Delta q \, V = (q'-q_0)V_0 = (k-1)q_0V_0 = 2(k)U_0$  

2- se il generatore è staccato, chiaramente non può fare lavoro  

### Moto tra armature di condensatore

![alt text](https://www.chimica-online.it/test-fisica/immagini/150.png)

**Moto di una carica tra le lastre**

scrivi formula di Newton:  
$\vec{F}=m\vec{a}=qe\pm mg$  

mg non c'è se è sul piano orizzontale    
attenta al verso della forza di Coulomb, dipende sia da dove sono le cariche sul condensatore, che dalla carica della particella che si muove  

quando esce dalle lastre:  
Coulomb non la sente più  
se mg non c'è: continua con moto rettilineo  
se mg c'è: fa moto parabolico  

**Conservazione energia**

particella lasciata libera ferma in mezzo al condensatore, quale è velocità quando tocca una lastra?  

conserva energiazione:  
all'inizio solo potenziale elettrica, alla fine solo cinetica  
$q\Delta V = \frac{1}{2}mv^2$  

V: differenza di potenziale tra il punto iniziale e finale, quindi il punto dove è ora e la lastra che toccherà alla fine  
quindi sapendo $V_c$ del condensatore, se parte da a metà, equidistante dalle lastre, allora V tra quel punto e la lastra finale è $\frac{1}{2}V_c$  
l'armatura negativa ha V=0, quella positiva $V=V_c$
