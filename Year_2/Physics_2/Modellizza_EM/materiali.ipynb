**Come caratterizzare i materiali**  

1. $\epsilon_r$: permeattività elettrica  
misura: polarizzazione elettrica quando soggetto a campo esterno    

2. $\mu_r$: permeabilità magnetica  
misura: polarizzazione magnetica quando soggetto a campo esterno  

3. $\rho$: resistenza ($\sigma$: conducibilità)   
misura: opposizione al passaggio di cariche   

4. $\chi_e, \chi_m$: suscettività (elettrica o magnetica)  
misura: quanto un materiale risponde a un campo applicato (altro modo di esprimere $\epsilon$ e $\mu$)

### Permettività vs resistività  

cosa | Permettività | Resistività
---|---|---
definizione | quanto il materiale si polarizza quando applichi un campo elettrico | quanto il materiale si oppone al passaggio di corrente
fenomeno | polarizzazione | conduzione  
tipo di fenomeno | RISPOSTA ELETTROSTATICA | FENOMENO DI TRASPORTO
formula | $D = \epsilon E$ | $J = \sigma E$   

**Differenza fisica permetività e resistività**

Permettività: come il materiale si deforma sotto E  
e come scherma i campi???  
polarizzazione → energia immagazzinata

Resistività: come le cariche scorrono sotto E  
conduzione → energia dissipata

**Isolanti, conduttori, semiconduttori**

la definizione corretta è usando la resistività  
perchè sono definizione sulla base di come si comportano al pasaggio di corrente    

valori di resistività:  
isolanti: alta  
conduttori: bassa  
superconduttori: $\rho=0$ (sotto la temperatura critica)  

solo in elettrostatica e magnetostatica, posso usare l'approssimazione:  
$\epsilon_r = \infty$ -> conduttore (perchè conduttore scherma benissimo)   
$\mu_r = 0$ -> superconduttore (perchè ???)  
si usano quando voglio usare il conduttore solo per schermare un campo, non per fargli condurre corrente  

### Permettività vs permeabilità  

**Permettività elettrica: intuitiva**

capacità di opporsi alla formazione di un campo elettrico al suo interno, cioè di SCHERMARE I CAMPI    
perchè polarizzarsi serve a contrastare il campo elettrico esterno:  

1- viene applicato un campo E esterno  
2- le cariche nel materiale si spostano un po'  
3- si crea un campo di polarizzazione che si oppone a E originario  

**Permeabilità magentica: intuitiva**

capacità di farsi attraversare da un campo magnetico  

### Altro

**Differenze $\epsilon$ e $\mu$**

$D = \epsilon E$  
$B = \mu H$  
E,B: campi reali, considerando il materiale  
D,H: campi senza considerare materiale   

$\epsilon$:  
sempre >1 (dielettrici) (quindi E < D)    
quindi aggiungere un materiale può solo indebolire (schermare) i campi al suo interno  

$\mu$:  
può essere sia minore che maggiore di 1  
quindi a seconda se è minore o maggiore, può rafforzare o indebolire il campo al suo interno  
ma è invertito rispetto all'elettricità:  
$>1$: rafforza, un po' o molto (paramegnetico, ferromagnetico)  
$<1$: indebolisce (diamagnetico)  

nota: in etrambi i casi, dia- serve a diminuire il campo  

**Suscettività**

$P=\epsilon_0 \chi_e E$  
$\epsilon_r = 1 + \Chi_e$  

$M=\mu_0 \chi_m H$  
$\mu_r = 1+\chi_m$  

ecco da dove viene la sostituzione M -> P/$\epsilon_0$  

note:  
segno di $\chi$: risposta al campo  
$>0$: risposta opposta al campo (diamagentismo)  
$<0$: risposta allineata al campo  
modulo di $\chi$: più è grande, più risposta forte  

**Elettrete**

oggetto con una P o M permanente  

### Tipi di campi

**Tipi di campi**

E,B: campi reali, effettivamente presenti in quel punto    
D,H: campi senza considerare materiale (cioè senza contare le cariche legate, ma solo quelle libere)   
P, M: campi che descrivono la proprietà elettrica o magnetica della materia  

$D = \epsilon_0 E + P  \sim \epsilon E$  
$B = \mu(H+M) \sim \mu H$   

l'approssimazione vale per: mezzo  lineare, isotropo, omogeneo   
cioè quelli per cui definisco la suscettività (negli esercizi, sempre tranne nei ferromagnetici)  

**Mnemonic trick formule**

E: si scrive con $\epsilon_0$ al denominatore (si trova in k, oppure pensa alla legge di Gauss)  
quindi per togliere l'info sul materiale, devo moltiplicare per $\epsilon_0$  
$D=\epsilon_0 E$  

B: si scrive con $\mu_0$ al numeratore (pensa a campo di Solenoide, o a legge di Ampere)  
quindi per togliere l'info sul materiale, devo dividere per $\epsilon_0$  
$H = \frac{B}{\mu_0}$  

### Spiegazione polarizzazione

**Perchè c'è polarizzazione**

pechè ogni atomo è un dipolo elettrico e magnetico    
quindi se viene applicato un campo E o B, si riorientano di conseguenza

polarizzazione:  
elettrica: spostamento di cariche microscopiche  
magnetica: orientazione della spira microscopica

**Differenza momento di dipolo**

dipolo magnetico: ogni atomo ha sempre momento $m \neq 0$  
applicare B cambia la distribuzione di in che direzione sono orientati, cioè cambia $<m>$  

dipolo elettrico:   
se la molecola è polare / l'atomo è uno ione: è come per dipolo magnetico (in realtà qui ho un MONOPOLO, non un dipolo ???)  
se la molecola è apolare / l'atomo non è un ione:  senza campo applicato $p=0$, quando viene applciato E  $p \neq 0$

**Perchè è un dipolo magnetico?**

in un atomo, un elettrone che gira intorno al nucleo è una corrente  
quindi un atomo è una spira  
a grande distanza, una spira è un dipolo magnetico  
per questo i materiali reagiscono al campo magnetico  

quindi $m \neq 0$ tutto il tempo, applicare B serve solo a riorientarlo  

**Perchè è un dipolo elettrico?**  

l'atomo è formato da due distribuzioni di carica:  
1- il nucleo (positivo)  
2- la nube elettronica (negativa), che schemaizziamo come una carica puntiforme nel suo centro  

se polare: la molecola è un monopolo, ha $Q \neq 0$  
se apolare: senza campo applicato, $p=0$  
quando gli applico un campo, atomo e nube non sono più uno sopra l'altro, ottengo un dipolo, $p \neq 0$  



**Come fa un dielettrico a diminuire E usando polarizzazione**

quando un dielettrico viene inserito in un campo elettrico (verde):  
il dielettrico si polarizza, cioè le sue molecole si riorientano per allinearsi con il campo  
generano il campo rosso, che si oppone a quello iniziale

**Da cosa dipende $\epsilon$**

da cosa dipende quella relativa di un materiale (elettrica):  

1. tipo di materiale  
2. stato di aggregazione (vapore acqueo diverso da acqua liquida)  
3. temperatura  
4. orientazione dei cristalli (per dielettrici non omogenei)  
eg. il quarzo ha un asse ottico, è diverso applicare il campo E parallelo o perpendicolare all'asse   
5. asse di rifrazione  
cioè come si comporta il materiale rispetto alla luce

NON dipende dal campo E applicato
