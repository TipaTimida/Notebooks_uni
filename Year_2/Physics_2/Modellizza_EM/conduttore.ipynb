### Definizione e proprietà

**Definizione**

è un oggetto dentro il quale le cariche elettriche sono libere di muoversi se gli è applicato un campo elettrico  

**Proprietà**

1. campo interno è nullo  

2. campo esterno è perpendicolare alla superficie, ha componente parallela nulla  
$E_\perp = \frac{\sigma}{\epsilon_0}$  
$E_\parallel=0$  

3. potenziale sulla superficie è costante  
se collegato a terra, è nullo  

4. carica interna è nulla, si trova tutta sulla superficie   
cioè no densità di arica volumetrica, solo supericiale ($\rho=0$, $\sigma \neq 0$)

### Spiegazione proprietà

**Campo interno nullo**

visto che le cariche sono libere di muoversi:  
quando applico un campo elettrico, si accumulano sulle superfici esterne   
la condizione di equilibrio è quando le cariche si fermano, che accade se non c'è campo elettrico che le sposti  
quindi all'equilibrio elettrostatico, il campo interno è nullo  

nota che lo spostamento delle cariche ha modificato le linee di ampo esterne al conduttore  

nota: campo indotto  
all'equilibrio, cioè quando le cariche smettono di muuoversi, il campo all'interno deve essere nullo  
quindi, se c'è un campo esterno $E_0$, deve esserci un campo esterno $E_{int}=-E_0$, generato dalla separazione delle cariche  

**Nota: campo interno nullo**

in realtà la condizione vera è che, all'equilirbio gli elettroni al suo interno, gli elettroni al suo interno siano fermi, cioè non sentano nessuna forza netta  
quindi E=0 SOLO SE non considero altre forze  
se tipo considero anche la forza di gravità, allora mi serve una forza elettrica eper bilanciarla, e quindi il campo non è nullo, ma è:  
$F=qE-mg=0$  

**Gabbia di Faraday**

se applico un campo esterno:  
le cariche del conduttore si spostano, accumulandosi sulle superfici esterne  
questo genera un campo indotto che cancella il campo elettrico all’interno della cavità   

si chiama effetto di schermo

ciò è sfruttato per costruire una Gabbia di Faraday  
è un involucro conduttore (può essere una scatola chiusa, una rete metallica, una griglia) che isola elettricamente il suo interno dai campi elettrici esterni  
in pratica per ripararsi da un campo elettrico, ci si rifugia dentro un conduttore, perchè dentro il campo è nullo  

esempi:  
abitacolo di un auto  
cavo coassiale  


**Effetti di schermo**

Schermaggio elettrostatico nei conduttori (gabbia di Faraday):  
Le cariche libere sulla superficie del conduttore si redistribuiscono.  
Generano un campo indotto che annulla quello esterno all’interno del conduttore.  
Risultato: campo NULLO dentro → il conduttore scherma l’interno.

Schermaggio dielettrico:  
In un materiale isolante, gli atomi/molecole si polarizzano.  
La polarizzazione crea un campo opposto a quello applicato.  
Effetto: il campo interno al materiale è RIDOTTO rispetto al campo esterno  

**Campo elettrico esterno (vicino alla superficie)**

in condizione di equilibrio, il campo è perpendicolare alla superficie  
perchè se ci fosse un’eventuale componente del campo parallela alla superficie, questa agirebbe sulle cariche superficiali mettendole in moto e violando quindi la condizione di staticità/equilibrio    
  
si suppone che le cariche superficiali siano vincolate a rimanere comunque ”attaccate” al conduttore, quindi una componente perpendicolare non nulla del campo non avrebbe effetto su di esse  

inoltre posso usare le formule per la discontinuità del campo elettrico, ricordando che all'interno E=0  

**Potenziale sulla superficie**

la componente parallela al superficie del campo è nulla, quindi il potenziale sulla superficie è costante (superfici equipotenziali)   

Il valore di tale potenziale è determinato dalle condizioni statiche del sistema, ad esempio si può imporre:   
* la carica totale presente sul conduttore (in questo caso si dice che il conduttore è isolato)  
* che la superficie del conduttore sia a un certo potenziale V rispetto alla terra (La ”terra” denota l’infinito, per il quale convenzionalmente il potenziale elettrico è fissato a 0)  
in quest’ultimo caso la carica totale del conduttore non è fissata, per cui non è detto che sia isolato  

quindi, i vincoli a cui il sistema è sottoposto
sono:  
* il potenziale sulla superficie del conduttore  
* la forma geometrica del conduttore
stesso

**Potenziale**

$V_i$: il potenziale all'interno del conduttore  
$V_e$: il potenziale sulla superficie del conduttore  
entrambi sono costanti, ma DIVERSI  
$V_i >V_e$  
il perchè è spiegato dalla quantistica  
ma è ovvio, perchè così per estrarre un elettrone bisogna compiere lavoro  

se non specificato, in un esercizio si riferisce a $V_e$  

**Carica interna**

In realtà serve la condizione aggiuntiva che all’interno del conduttore non sia presente carica fissa (non libera di muoversi)    
questo vincolo fa sı̀ che se fossero presenti cariche libere all’interno,queste potrebbero solo depositarsi sulla superficie, per cui in condizioni statiche la carica interna è nulla  

Il fatto che la carica interna sia nulla implica
che il potenziale all’interno del conduttore sia uniforme e pari a quello sulla superficie, per il teorema di unicità

**Effetto delle punte**

campo sulla sueprficie di un conduttore è più forte nelle zone di raggio di curvatura più piccola, cioè le punte  

ad esmepio, vedi che se collego sue sfere conduttrici di raggio diverso  
devo avere stesso potenziale (perchè sono collegate), e so $E=\frac{V}{r}$  
quindi quella di raggio più piccolo emana un campo maggiore

### Come si distribuisce la carica

#### Conduttore con buco

**Recap**

1- se nel buco NON c'è carica:    
le cariche si accumulano sulla superficie esterna dell'oggetto, perchè vogliono massimizzare la distanza tra loro  
quindi il buco è ignorato, NON si accumulano sul bordo del buco    
questo vale per qualunque carica esterna ci sia, basta non ci sia nulla nel buco  

2- se nel buco C'È carica:  
se carica + dentro il buco, la superficie del buco ha carica - per induzione  
per mantenere il conduttore neutro, la superficie esterna del conduttore ha carica +  
tutte q con lo stesso modulo  

3- se ci sono 2 buchi, uno con dentro carica +, uno con carica -:  
la superficie dei buchi ha carica di segno opposto a quella nel buco  
la superficie esterna del conduttore non ha carica???  

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/Conduttore_ideale_buco.jpg)

**Proof: carica su superficie del buco**

1- studio S  
S: una superficie che racchiude la cavità (ma più grande della cavità, il suo bordo è nel conduttore)  
bordo di S ha E=0, perchè è dentro il conduttore  
quindi per Gauus, $q_{int}=0$ all'interno di S  

2- SE dentro la cavità ci sono cariche:  
sulla superficie serve una densità per bilanciarle, e avere carica totale nulla  
nota che il campo fuori ha comunque la simmetria del bordo esterno del conduttore (se conduttore è una sfera, campo esterno ha sferica)  
è un effetto di schermo, non riesco ad accorgermi della presenza della carica nella cavità   

3- SE dentro la superficie NON ci sono cariche:  
scelgo $\gamma$, una curva a cavallo di un bordo della cavità  
zoomo abbastanza che il bordo della cavità è localmente un piano, qundi $\gamma$ ha due alti paralleli e due perpendicoalri  
lati perpendicolari: $E \parallel l$, quindi prodotto scalare è 0  
lati paralleli: quello dentro il conduttore ha E=0  
poichè deve essere $\int_\gamma E \cdot dl=0$, anche l'ultimo lato, quiello parallelo e interno alla cavità, deve dare 0  
poichè posso scegliere qualunque $\gamma$ voglio, l'unico modo per averlo sempre è che $E=0$ dentro la cavità  
su una separazione: $\Delta E_\perp = \frac{\sigma}{\epsilon_0}$  
quindi visto che la componente perpendicolare è uguale, sempre 0 dentro e fuori, non c'è carica superficiale

**Come la carica nel buco viene schermata**

dati:  
q: carica puntiforme al centro  
$q_{in}$: carica su superficie interna guscio  
$q_{out}$: carica su superficie esterna guscio  

1- all'inizio: guscio conduttore neutro  
2- aggiungo una carica q nel centro del guscio  

cosa so:  
all'inizio guscio è neutro: all'inizio $q=0$, quindi q=0 in ogni momento per conservazione della carica (considerando solo le q createsi nel guscio, non quelle provenienti da fuori)   
guscio è conduttore: il flusso all'interno dev'essere nullo   

uso teorema di Gauss:  
$\Phi _E = \int_\gamma E \, dS = \frac{q_{racchiusa}}{\epsilon_0}=0$  
$\gamma$ è una superficie isocentrica al guscio, che comprende la superficie interna del guscio, ma non quella esterna  

quindi devo avere $q_{racchiusa}=0$ (per avere flusso nullo)    
quindi la carica sulla superficie interna del guscio deve essere -q, per contrastare la q al centro  
$q_{in} = -q$  
poi, per farlo rimanere neutro, la superficie esterna del guscio (che non rientra in $\gamma$) deve avere carica +q  
$q_{out}=+q$

**Che succede**

Supponiamo di collegare con un filo le parti di due conduttori:    
le cariche possono passare da uno all'altro   
quindi si compensano, si cancellano a vicenda    

**Problema di collegare conduttori**

imposta il sistema:  
$\begin{cases}
V_1 = V_2 & (sul \; bordo) \\
q_1+q_2 = q_1'+q_2'
\end{cases}$  

la formula di V dipende da che forma hanno i conduttori

**Esempio in figura**

quindi la carica +q è tutta sulla superficie esterna  

### Conduttore ohmico ???

**Conduttore ohmico**

un oggetto materiale che soddisfa la seguente proprietà: la densità di corrente J e il campo elettrico all’interno di esso soddisfano la relazione

$J = \sigma E$  
$\sigma$ è la conducibilità del materiale  

**carica libera all’interno di un conduttore ohmico**

che cosa succede nel tempo?  

mi servono la prima eq di Maxwell e l'eq di continuità:  
$div \, E = \frac{\rho}{\epsilon_0}$  
$div \, J + \frac{\partial \rho}{\partial t}=0$  

conduttore ohmico, quindi $J = \sigma E$  
sostituisco nella seconda:  
$\sigma \, div \, E + \frac{\partial \rho}{\partial t}=0$  
$\frac{\partial \rho}{\partial t}+ \rho \frac{\sigma}{\epsilon_0}=0$  
questa è l'equazione per la densità di carica  

ammette come soluzione la funzione esponenziale decrescente:  
$\rho(\vec{r},t)=\rho(\vec{r},0)e^{-\frac{t}{\tau}}$  
$\frac{1}{\tau}=\frac{\sigma}{\epsilon_0}$  

le cariche libere tendono a respingersi e andare verso i bordi lasciando l’interno del conduttore privo di carica   
Per cui possiamo dedurre che aspettando un tempo
abbastanza lungo verrà realizzata la condizione di conduttore perfetto


### Altro

**Come caricare un condensatore sferico**

1. metto carica +Q in sfera interna  
2. così sull'armatura esterna si distribuiscono le cariche -Q nella superficie interna, +Q in quella esterna per induzione  
3. collega l'armatura esterna al GND, così la superficie esterna perde le cariche positive, diventa una superficie neutra  

ora ho le due superfici che si guardano che hanno cariche opposte, proprio come volevo  

**Su quale lastra è la carica positiva di un condensatore?**

in un circuito, disegna la freccia $\Delta V$, come al solit $- \rightarrow +$    
sulla parte col - c'è l'armatura negativa  


**Applicazioni: touchscreen capacitivo**

lo schermo è fatto di tanti piccoli pixel, ognuno con una $\Delta C$ (tutte uguali  )
quando tocco lo schermo in un punto, in quel punto si sommano in serie la capacità del pixel e del mio dito, quindi la capacità totale cambia  
un circuito lo rivela, così capisce che lo schermo è stato toccato, e anche dove  

c'è anche il touchscreen resistivo, cioè devi premere abbastanza da spostare fisicamente lo schermo, così da far toccare due resistori e far cambiare una reisstenza, che pure verrà rilevata da un circuito  
ma è peggio, perchè lo schermo deve essere flessibile, quindi non di vetro, e poi è più soggetto ad usura
