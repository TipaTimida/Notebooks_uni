### Formule pezzi circuito

cosa / pezzo  | formula generale |  resistore | induttanza | condensatore | generatore
---|---|---|---|---|---
ddp | . | $RI$ | $L \dot{I}$ | $\frac{q}{C}$ | $V$    
potenza | $P=VI$ | $RI^2$ | $LI \dot{I}$ | $\frac{q}{C} I$ |$V I$  
energia | $dU = P \;dt $ | $dU= RI^2 \; dt$ | $\frac{1}{2} LI^2$ | $\frac{1}{2} \frac{q^2}{C}$ | $V \, dQ$  

**Notazione**

indicale con la lettera del componente come pedice  
eg. $P_R = RI^2$  
a volte per il restistore uso il pedice 'd', per 'dissipazione'  

**Calcola energia**

1- usa la formula $dU = P \, dt$  
2- se appaiono $I=\frac{dQ}{dt}$ o $\dot{I}=\frac{dI}{dt}$, semplifica il dt con il denominatore  
se invece conosco $I(t)$, posso lasciare il dt, sostituire l'espressione per I e integrare col dt  

induttanza:  
$dU = P \, dt = LI \frac{dI}{dt} dt = LI \, dI$  
$U = \int_0^I LI \, dI = \frac{1}{2}LI^2$  

condensatore:  
$dU = \frac{q}{C}\frac{dq}{dt} \, dt = \frac{q}{C} \, dq$  
$U =  \frac{1}{2} \frac{q^2}{C}$  

generatore:  
$dU = \mathcal{E} I \; dt  = \mathcal{E} \frac{dQ}{dt} \; dt = \mathcal{E} dQ$    

**Note: cosa succede all'energia**

1- resistore: dissipata per effetto Joule  
2- induttanza: immagazzinata nel campo magnetico  
il solenoide, se attraversato da corrente, crea un campo B al suo interno, che quindi immagazzina energia  
3- condensatore: immagazzinata nel campo elettrico  
se attraversato da corrente, crea un campo E tra le sue piastre, che quindi immagazzina energia  

**Dissipazione**

se non ho cose assimilabili ad un resistore, per l'energia dissipata:  
 $W = \int E \cdot J$ 

### Energia di un campo

**Densità di energia**

| Campo     | Forma generale                                    | Mezzo lineare                           |
| --------- | ------------------------------------------------- | --------------------------------------- |
| Elettrico | $u_E=\displaystyle\int \mathbf D\cdot d\mathbf E$ | $u_E = \displaystyle \frac12 \epsilon E^2$    |
| Magnetico | $u_M=\displaystyle\int \mathbf H\cdot d\mathbf B$ | $u_M = \displaystyle \frac12 \frac{B^2}{\mu}$ |

**Energia**

$U = \int u \; dV$  
vale sia per campo elettrico che magnetico  
metti $\epsilon$ giusto, se sono nel vuoto è $\epsilon_0$    

**Solo per campo elettrico**

1- se V è più facile di E: integrale sulla carica  
$U = \frac{1}{2} \int \varphi \, dq = \frac{1}{2} \int  \varphi  \rho\; dV = \frac{1}{2} \int  \sigma  \rho\; dS   $   
$\rho$: densità di carica LIBERA (anche se è un dielettrico)  
$\varphi$: potenziale (generato da cariche libere+polarizzazione) (di polarizzazione appaiono solo in un dielettrico)  
(per ricordarmela, di solito scrivo U=Vq)  
a seconda se ho un dielettrico o un conduttore, e quindi la carica è volumica o superficiale, devo scegliere in cosa trasformare dq  


$U = \int \varphi \, dq =  \int_0^Q k\frac{q}{a}dq = k\frac{Q^2}{2a}$  



### Tipi: stato e accoppiamento

**Tipi**

sia in energia elettrica che magnetica, esistono due tipi:  

self-energy / energia propria:  
E: energia del campo generato da quella carica  
B: energia del campo prodotto da quella corrente  
ci sarebbero anche se quell'elemento fosse da solo, isolato  

energia di accoppiamento / interazione:  
E: $U_{int} = q_1 \Phi_2(E_1) = q_2 \Phi 1(E_2)$  
$U_{int}=\int \epsilon E_1 \cdot E_2 \, dV$  
B: $U_{int} = MI_1 I_2$  
$U_{int} = I_1 \Phi_2(B_1) = I_2 \Phi 1(B_2)$


il campo prodotto da un elemento si concatena con l'altro elemento, il che produce energia  

**Come scrivere veloce**

energia magnetica:  
$U_M = \frac12 \Sigma_k I_k \Phi_k$  
$\Phi_k$: campo B concatenato al circuito k, fatto dalla sovrapposizione del campo B dato dal circuito k stesso (induttanza) e dai campi creati dagli altri circuiti (murua induzione)

$U_M = \frac12 \Sigma_{i,j} M_{ij} I_i I_j$  
i=j: M sarebbe L, l'autoinduzione  
$i \neq j$: mutua induzione, se scambio i e j rimane uguale, quindi si toglie 1/2   

### Condensatore a V o Q costante, problema coi segni

**Casi**

1- Q costante  
cosa: condensatore isolato, NON c'è attaccato generatore (che quindi non fa lavoro)  

2- V costante  
cosa: c'è attaccato un generatore, che fa lavoro per mantenere V costante  

**Caso Q costante: isolato**

$U=\frac{Q^2}{2C}$  
è solo la formula del condensatore  
$dU = P \, dt = VI \, dt = \frac{Q}{C} I \, dt = \frac{Q}{C} dQ =  \frac 12 \frac{Q^2}C$  

**Caso V costante: collegato a generatore**

qui ho sia condensatore che generatore nel circuito  
il mio sistema non include il generatore, ma solo il condensatore  
dal punto di vista del sistema, il lavoro fatto dal generatore è SUBÌTO, quindi cambio segno:  
$U_{tot}=U_c+U_g=U_c-W_g$  

condensatore: facile, usa la formula  
$dU_c =\frac{1}{2}V^2 dC$  

generatore:   
$dW_g = V \, dQ = V^2 dC$  
perchè il condensatore ha: $dQ = V \, dC$  
quindi il generatore ha emana potenza:  $P = \frac{dW}{dt} = V^2 \frac{dC}{dt}$  
per dC/dt devo usare chain rule con la variabile che varia, tipo la d tra le lastre  
e ricorda che: $dW_g = -2 dU_c$  

totale:  
$dU = -\frac{1}{2}V^2 dC$

### Potenza

**Potenza erogata dal generatore**

è la potenza totale, che fa funzionare tutto  
2 metodi per calcolarla:  

1- $W_{gen}=V_0I$  
quando: se generatore mantiene $V_0$ costante e so scrivere $I(t)$ (SEMPRE MEGLIO)     
come scrivere $I(t)$: se ho condensatore, $$I=\frac{dQ}{dt}$, con Q di condensatore - se ho circuito, legge delle maglie e risolvi per I  

2- uguagliala a tutte le potenze esistenti nel circuito  
quando: ho già calcolato in passato le altre potenze, e non c'è potenza meccanica, che è difficile da calcolare 

**Potenza generatore: esempio condensatore**

metodo 1:  
$W_{gen}=V_0I$  
dove $I(t)=\frac{dQ}{dt}$, dove Q(t) è ad esempio la carica sul condensatore  

metodo 2:  
$W_{gen}=W_{mec}+W_{e}$  
dove: $W_{gen}=V_0I \quad W_{mec}=Fv \quad W_e = \frac{dU}{dt}$  


**Potenza generatore: esempio circuito RL**

metodo 1:  
$W_{gen}=V_0I$  
dove I(t) la trovo risolvendo la legge delle maglie: $V_{gen}=RI+L\dot I$   

metodo 2:  
$W_{gen}=W_{diss}+W_{magn}$  
dove: $W_{diss}=RI \quad W_{magn} =  \frac{d}{dt}\frac 12 (LI^2)  = L I \dot I $  
se ci sono altri oggetti su cui dissipar, aggiugnere anche $W_{diss2}=\int J \cdot E \, dV$  

**Potenza dissipata**

1- oggetto generale (di cui non so R)   
$W_{diss}=\int J \cdot E \, dV$  
$dW_{diss}=J E = \sigma E^2=\frac{J^2}{\sigma}$    
quindi attenta, è potenza per unità di volume  
perchè E è locale, questo è il suo valore in un punto preciso, quindi questa è la potenza dissipata nel punto dove E ha questo valore (oppure nel punto dove J ha quel valore)    
se oggetto è piccolo, assumo E uniforme su tutto l'oggetto   


2- resistore (di cui so R)  
$W_{diss}=RI$

### Forze

**Forza meccanica tra piastre generatore**

se niente accelera (va bene moto rettilineo uniforme), allora $F_{tot}=0$  
somma tutte le forze presenti, di solito sono solo: $F_{mec}+F_{cond}=0$   
dove: $F_{cond}=\frac 12 qE$   
nota che il generatore produce una fem, che NON è una forza  

$F=\frac 12 qE$  
perchè dovuta al campo di una piastra nella posizione dell'altra  

**Non confondere**

$\nabla U = -F$  
$\partial_t U = P$  

ma se chiede la 'variazione temporale dell'energia elettrostatica', vuole U(t), NON P  

**Momento delle forze**

situazione:  
spira di raggio a, con caruca Q  
ruota con $\omega$ variabile, quindi genera B variabile, che genera E  

1- forza infinitesima su un elemento dell'anello  
$df = E \; dq = \frac Q{2 \pi a}\, dl$  

2- momento delle forze  
$\tau_E = \int a \, df$  
dove $\int dl = 2\pi a$  

3- equazione del moto dell'anello:  
$J \dot \omega = \tau_{ext}+\tau_{E}$  
J: inerzia, non uso I perchè si confonde con la corrente $J \dot \omega = I \ddot \theta$  

3- potenza sviluppata  
$P=\tau \cdot \omega$
