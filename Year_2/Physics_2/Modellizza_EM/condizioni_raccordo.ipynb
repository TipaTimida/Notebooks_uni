### Tipi di campi

**E vs D**

D,H: determinati solo da cariche o sorgenti libere -> campo ausiliare  
E,B: determianto da cariche totali (libere + polarizzazione) -> campo fisico reale   
polarizzazione dipende da che materiale c'è davvero in quella zona  

la differenza si vede bene nelle equazioni di Maxwell, dove cambia il tipo di cariche usato e se appare $\epsilon$ e $\mu$  
$\nabla \cdot E = \frac{\rho_{tot}}{\epsilon_0}$  
$\nabla \cdot D = \rho_{free}$  

**Corrispondenza**

$D=\epsilon_0 E +P$  
$B = \mu_0 (H + M)$  

vale la corrispondenza E->H, D->B, M -> P/$\epsilon_0$  

M -> P/$\epsilon_0$  
$\epsilon_0$ al denimonatore serve per cancellare l'$\epsilon_0$ che appare in k, così quando passo a M lo porto via insieme a P  

**Mnemonic trick**

E: si scrive con $\epsilon_0$ al denominatore (si trova in K, oppure pensa alla legge di Gauss)  
quindi per togliere l'info sul materiale, devo moltiplicare per $\epsilon_0$  
$D=\epsilon_0 E$  

B: si scrive con $\mu_0$ al numeratore (pensa a campo di Solenoide, o a legge di Ampere)  
quindi per togliere l'info sul materiale, devo dividere per $\epsilon_0$  
$H = \frac{B}{\mu_0}$

### Formule discontinuità

**Campo elettrico**

$\begin{cases}
E_{1t}=E_{2t} & tangenziale \\
D_{1n}-D_{2n} = \sigma_f & normale, \\  
\end{cases}$  

**Campo magnetico**

$\begin{cases}
H_{1t}-H_{2t}=K_f & tangenziale \\
B_{1n}=B_{2n} & normale, \\  
\end{cases}$  

**Mnemonic trick**

normale: può la superficie essere una sorgente, e quindi cambiare il campo dopo che l'ha attraversata?  
B: no, non ci sono monopoli magnetici -> si conserva  
E: sì, ci sono cariche magnetiche -> $D_1-D_2 = \sigma_f$  

tangenziale: quale è la domanda???  

quando qualcosa si conserva uguale, è il campo fisico reale  
se invece viene modificato, meglio esprimerlo come campo che non considera il materiale  

nota che divB=0, ma divH è diverso da 0  
non la uso mai in questa forma, quindi difficile rendermene conto  

**Nota**

ho una superficie di separazione tra due materiali  
le direzioni tangenziale e normale (parallelo e perpendicolare) li intendo rispetto a questa superficie  

posso usare una condizione di discontinuità solo se il lato è abbastanza lungo  
così posso ignorare l'eggetto degli angoli  

#### Come trovarle

**Componente normale**

disegna: un cilindro attraverso l'interfaccia  
usa legge: di Maxwell col flusso, cioè Gauss (divergenza + th della divergenza)  

steps:  
1- considera solo il flusso lungo le basi del cilindro  
perchè lati sono bassi a piacere  
2- basi sono larghe uguali, raccolgo S a sinistra e lo porto al denominatore a destra  

$\nabla \cdot B = 0 \quad \rightarrow \quad \int_S B \cdot dS \quad \rightarrow \quad  B_1 = B_2$  
$\nabla \cdot D = \rho_f  \quad \rightarrow \quad \int_S D \cdot dS = Q_{free}  \quad \rightarrow \quad D_1 - D_2 = \sigma_f$  

**Componente tangenziale**

disegna: un rettangolo  
usa: la legge di Maxwell sulla circuitazione (rotore + th di Stokes)  

steps:  
1- considera solo la circuitazione lungo i lati paralleli     
perchè gli altri corti piccoli a piacere    
2- i lati paralleli sono lunghi uguali, quindi raccolgo l a sinistra e lo porto al denominatore a destra  

queste sono le caso statico (prendo eq di Maxwell per caso statico):  
$\nabla \times H = J_f \quad \rightarrow \quad H_{1t}-H_{2t}=K_f$  
$\nabla \times E = 0  \quad \rightarrow \quad E_{1t}=E_{2t}$

### Ferromagnetico (Gabbia magentica)  

**B segue il materiale**

il campo B, dentro un ferromagentico, segue il materiale  
cioè, è parallelo all'interfaccia/bordo  

perchè:  
componente normale di B è come fuori, non aumenta  
ma componente tangenziale di B aumenta, perchè parte tangenziale di H si conserva da fuori, ma $\mu$ del ferro è molto maggiore di $\mu_0$, quindi B all'interno del ferro ha la parte tangenziale molto maggiore di quella normale  

è un effetto di 'canalizzazione del campo magnetico'

**A cosa serve**

posso fare uno 'schermo di campo magnetico B'  

prendo una sfera ferromagentica cava  
anche se c'è un campo B fuori, quando colpisce la sfera viene guidato attraverso di essa, e riesce parallelo a dove era entrato, dall'altra parte  
mentre non entra all'interno
