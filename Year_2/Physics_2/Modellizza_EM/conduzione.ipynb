### Continuità della corrente

**Continuità corrente**

equazione di continuità: $(J_2-J_1) \cdot \hat{n} = -\frac{d\sigma_f}{dt}$  
J è considerata normale alla superficie di separazione tra i materiali   

a regime stazionario: $J_1=J_2$   
nello stato transiente: non sono uguali, perchè si sta accumulando carica sull'interfaccia  

**Materiali con conducibilità nulla e non**

$J=\sigma_c E_{n}$   
se in un dielettrico non specifico quanto vale, assumo $\sigma_c=0$  

materiale con $\sigma_c=0$:   
sempre: $J=0$, ma $E$ può essere qualunque valore  
regime stazionari (fine della scarica): E rimane quello iniziale, non è cambiato  

materiale con $\sigma \neq 0$:   
stato transiente: $J \neq 0$, ancora sta scorrendo    
stato stazionario: $J=0$, perchè ha finito,  quindi $E=0$ (perchè $\sigma_c \neq 0$)  

### Scarica di un conduttore

**Steps**

1- calcola campo E dell'oggetto  

2- scrivi equazione di continuità della corrente:  
$\frac{dQ}{dt}=-I(t) = - \int_S J \cdot dS = -\sigma \int E \, dS$    

3- plugga E e cerca di ottenere $\frac{dQ}{dt}=-\frac{Q}{\tau} $  
la soluzione è: $Q(t)=Q_0e^{-t/\tau}$  

**Scarica sfera di raggio R**

1- campo elettrico:  
$E=k\frac{Q}{r^2}$  
nota che Q cambia nel tempo, è $Q(t)$

2- corrente:  
$I(t)= \int J \cdot dS =  \sigma E \; 4 \pi r^2 = \frac{\sigma}{\epsilon_0}Q$  

3- quazione differenziale:  
$\frac{dQ}{dt}=-I(t)=-\frac{\sigma}{\epsilon}  Q$  
sol: $Q(t)=Q(0)e^{-t/\tau} \quad \tau = -\frac{\epsilon}{\sigma}$

**Metodo 2: calcolare integrale**

c'è scarica solo in un materiale con $\sigma _c \neq 0$  
quindi se fuori dal conduttore ci sono materiali diversi, separa l'integrale in addendi, a seconda del avlore di $\sigma_c$

E dovrebbe essere noto, di solito è un conduttore di forma facile    
calcola l'integrale, come con th di Gauss: E è ortogonale alla superficie del conduttore, quindi portalo fuori (perchè ha sempre stesso angolo) e vedi cosa diventa dS  
$dS=4 \pi r^2$ per una sfera e $dS= 2 \pi r h$ per un cilindro  e $S=2 \pi r^2$ per una semisfera  
calcola E rispetto ad un generico r, magari con Gauss, e sostituiscilo  

**Formula per cilindro**

se la formula di E si scrive usando $\lambda$, non $Q$ (ad esempio nel cilindro)  

serve la chain rule:  
$\frac{dQ}{dt} = h \frac{d\lambda}{dt}=-I$    
dividendo per h da entrambe le parti, trovo una soluzione all'equazione differenziale:  
$\lambda(t)=\lambda_0 e^{-t / \tau}$  
da cui posso ritrovare Q(t) moltiplicando per h: $Q(t)=h \; \lambda(t)$

oppure posso pensare che con I si intende la carica per unità di lunghezza   
dove $\int dS=2 \pi r$, solo la superficie esterna, non la sezione  

**Scarica cilindro**

$E=\frac{\lambda}{2 \pi r \epsilon}$  
$I = \int \sigma E \cdot dS = \sigma E (2 \pi r h)$  
dS è la superficie laterale, perchè è quella perpendicolare a E, come nel th di Gauss

uso la chain rule:  
$\frac{dQ}{dt}=-I$  
$h \frac{d\lambda}{dt}=-\frac{\sigma}{\epsilon}\lambda h$  
$\frac{d\lambda}{dt}=-\frac{\sigma}{\epsilon}\lambda$  

trovo le soluzioni:  
$\lambda(t)=\lambda_0 e^{t/\tau}$, con $\tau = \frac{\epsilon}{\sigma}$  
$Q(t)=h\lambda(t) = h\lambda_0e^{t/\tau}=Q_0e^{t/\tau}$  

**Due semifere hanno dielettrici diversi**

conduttore sferico, con Q  
la semisfera sopra ha dielettrico, quella sotto ha aria  

campo elettrico: sono uguali nei due mezzi, perchè sono paralleli alla superficie di separazione  
calcola con Gauss:  
$Q = \epsilon_0 \int_{S1} E \, dS + \epsilon_0\epsilon_r \int_{S2} E \, dS = \epsilon_0(1+\epsilon_r) \; E \frac{1}{2}(4 \pi r^2)$  

corrente:    
$I= \int J \cdot dS = \int \frac{E}{\rho}dS = \frac{1}{\rho} \frac{Q}{4 \pi \epsilon_ (1+\epsilon_r)r^2} \int_{S2} dS = (...) \frac{1}{2}(4 \pi r^2) = \frac{Q}{\rho \epsilon_0 (1+\epsilon_r)}$  
nota che la superficie con l'aria ha $\rho=\infty$, quindi quell'addendo è 0, lì non c'è scarica  

**Due sfere concentriche hanno dielettrici diversi**

conduttore sferico, con Q, fatto da due gusci concentrici con spessore non nullo  
quello dentro ha aria, quello fuori dielettrico con perdite  

dentro il dielettrico devo avere E=0 a regime, perchè si è scaricato  
nell'aria invece $E=k\frac{Q}{r^2}$  
quindi, per continuità, c'è una $\sigma_f$ sulla superficie di separazione (all'interfaccia tra vuoto e dielettrico)

poichè nella regione di vuoto non passa corrente, $\tau=\infty$  

#### Dissipazione

**Dissipazione ed energia**

potenza dissipata per effetto Joule:  
$P(t)=\int J \cdot E \; dV = \sigma \int E^2 dV$  
V è il volume  

energia elettrostatica:  
$U(t)=\int \frac{1}{2}\epsilon_0E^2 \; dV$  

mostra che la variazione è uguale:  
voglio mostrare che $\frac{dU}{dt}=-P(t)$  
quando calcolo il lato sinistro, usa la chain rule $\frac{dU}{dt} = \frac{dU}{dQ}\frac{dQ}{dt}$, e sostituiscici il valore $\frac{dQ}{dt}=-I$, calcolato prima nell'eq differenziale  

**Esempio dissipazione**

sempre sfera di raggio R  

energia elettrostatica: $U(t)=\int _R^\infty \frac{1}{2}\epsilon_0E^2 dV =k\frac{Q^2}{2R}$   
potenza dissipata: $P(t)=\int _R^\infty \sigma E^2 dV = k\frac{\sigma Q^2}{\epsilon R}$  

mostra che sono uguali:  
nota che r non è funzione di t, pchè la sfera non cambia dimensione, ma Q sì  
$\frac{dU}{dt}=\frac{dU}{dQ}\frac{dQ}{dt}=(k\frac{Q}{R}) \frac{dQ}{dt} = k\frac{Q}{R} (-\frac{\sigma}{\epsilon}  Q) = k\frac{\sigma Q^2}{\epsilon R} = -P(t)$
