**Oscilloscopio**

cosa | digitale | analogico
---|---|---
sistematica | per V: 3% lettura <br> per T: trascurabile | 3% lettura (sia per V che T)  
statistica | per V: almeno $VOLTDIV *10/256$, cioè 256/fs, ma se è maggiore usa quella di lettura <br> per T: di lettura | di lettura (non c'è la cosa di 256/fs)

di lettura vuol dire l'incertezza su dove mettere i cursori 

**Oscilloscopio digitale**

incertezza sistematica: 3% (di lettura, perchè sistematico è errore relativo???)      

incertezza di lettura: quanto sono sicura su dove mettere i cursori (probabilmente è più grande, ingloba la sistematica) 
nota che comunque non può essere minore di 1/256 del fondoscala  
il fondoscala è 10 * VOLTDIV  
quindi è almeno: VOLTDIV *10/256    
ma se sono più insicura con i cursori, uso l'incertezza su dove metterli, che li ingloba  

per misure del tempo, considero solo l'errore di lettura, è molto maggiore  

**Oscilloscopio analogico**

incertezza sistematica è 3% sia per volte che per tempi  
mentre nel digitale, per i tempi non c'è quella sistematica, è piccolissima  

sistematica: qui non c'è la cosa di 1/256 del fondoscala, vince sicuro l'incertezza su dove mettere i cursori  

ricorda: dopo il fit, l'errore è la RADICE dei valori sulla diagonale
