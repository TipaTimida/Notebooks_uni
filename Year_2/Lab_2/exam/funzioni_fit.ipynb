### Freq domain

**Ricorda**

$2 \pi f = \omega$  
$\omega/\omega_c = f/f_c$  

**Filtri**

$H=\frac{Z_{out}}{Z_1+Z_2}$  
poi moltiplica per $j \omega C$  

scrivi più bello e aggiungi A:  
$s = j \frac f {f_c} = j \frac{\omega}{\omega_c}$  
$H_{high}(A, f_c) = A\frac{s}{s+1}$  
$H_{low}(A, f_c) = A \frac{1}{s+1}$  

high has s at the top too  

poi: $f_c = \frac{1}{2 \pi RC}$  

**Filtri: gain**

$H_{LP}(\omega)=\frac{1}{1+ j \omega RC}$  

magnitude (attenuation):  
$|H(\omega)| = \frac{1}{\sqrt{1+(\omega RC)^2}}$  
write pretty:  
$G(f, A, f_c)=\frac{A}{\sqrt{1+(f/f_c)^2}}$ 

**Transistor**

ricava:  
$v_{out} = -i_c R_C$   
$v_{in} = [R_B +r_b+(\beta +1) Z_{eq}]i_b$   

formula:  
$H(\omega) = -\frac{R_C}{R_B +r_b+(\beta +1) Z_{eq}} \beta$  
$Z_{eq}=\frac{R_E}{1+j\omega R_EC_E}$  

migliora:  
uso $R'_b = R_B + r_b$   
$\beta \sim \beta +1$  
$\beta_f$ e $R_b'$ sono correlati  
$H = -\frac{R_c}{R'_b + (\beta+1){Z_E}}\beta_f \sim   -\frac{R_c}{\beta(\frac{R_b'}{\beta} +Z_E)} \beta = -\frac{R_C}{\frac{R_b'}{\beta} +Z_E}$   
so params are: $C_E$, $R_b' / \beta$   

**Transistor: gain**

la formula è troppo difficile, non serve ricavarla, calcola solo il gain dalla H  

$Z_{E}=\frac{R_E}{1+j\omega R_EC_E}$  
$T = -\frac {R_C}{A + Z_E}$  
$G(f, A, C_E) = numpy.abs(T)$

il filtro completo, con Z_E, è attivo passa-alto  
togliendo Z_E, e mettenco condensatore tra R_B e v_out, ottengo filtro attivo passa-basso  
RIPASSALI!!! 

**Oscillatore RLC**

otuput across R:  
$H = \frac{R}{R_{tot}+  \frac 1{j \omega C}+ j \omega L} = \frac{j \omega RC}{j \omega R_{tot} C + 1 - \omega^2 / (\omega_0^2)}$  
$\omega_0^2 = \frac 1 {LC}$    

$G= \frac{ 2 \pi RC \; f}{\sqrt{(1-(f/f_0)^2)^2 +(2\pi R_{tot}C \; f)^2}} = \frac{A\; f}{\sqrt{[1-(f/f_0)^2]^2 +(B \; f)^2}}$    
so params are: A, B, f0
