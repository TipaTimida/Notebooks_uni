signal is not perfecly rectified  
'ripple' ia the figure of merit for a power supply, want it as small as possible   
$\Delta V_{ripple} = V_{max}-V_{min}$  

why it appears:

how to minimize it:

**Punto 2**

espressioni:  
$i_d = \frac{V_G-v_d}{R_D}$  
$r_d = \frac{v_d}{i_d}$  
$r_d \sim \frac{dV}{dI}\|_q =  (\frac{dI}{dV})^{-1} =  (\frac{I_s e^{V_q/(\eta V_T)}}{\eta V_T})^{-1} \sim \frac{\eta V_T}{I_q}$  
l'ultimo passaggio è perchè, in regime di conduzione, nel diodo posso approssimare Shockley come: $I= I_s e^{V/(\eta V_T)}$

dove $V_T = 26 mV$  

**Punto 2: Valore atteso: approssimazioni usate**

1- $V_q \gg \eta V_T$   
per approssimare che Shockley sia $I= I_s e^{V/(\eta V_T)}$, cioè togliere il -1 (spiegato in note sul diodo)  

2- $v_d \ll \eta V_T$    
perchè sto linearizzando attorno al punto di lavoro, quindi faccio espansione di Taylor in quel punto  
cioè metto $V = V_q + v_d$, ed espando $I(V)$ con taylor  
$I \sim I_q (1+ x+\frac{x^2}2)$  
dove $x = \frac{v_d}{\eta V_T}$  
$e^{V_q}$ duventa $I_q$, mentre $e^{v_d}$ lo espando con taylor  

quindi:   
$r_d= \frac{v_d}{i_d}=\frac{v_d}{I-I_q} = \frac{v_d}{I_q(x+\frac{x^2}2)}= \frac {r_{d, exp}}{(1+\frac x2)}$  
perchè ricorda che $r_{exp} = \frac {\eta V_T}{I_q} = \frac{v_d}{I_q x}$ 
 
se voglio il valore stimato vicino a quello misurato, deve essere $v_d \ll \eta V_T$  

nota: con l'approssimazione 2, posso anche stimare numericamente l'errore della mia approssimazione   
l'errore è $err_{rel} = \frac x 2$  
esempio: a temp ambiente, $\eta V_T \sim 52mV$, quindi se uso $v_d \sim 5 mV$ sto commettendo un errore relativo del 5%
