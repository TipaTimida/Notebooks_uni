### Wire, cable

**Difference**  

- wire: a single conductor, no insulation  
- cable: two or more conductors twisted around each other, insulating cover in plastic

conductors made of low resistance metal, like copper 

coaxial cable: used to carry high frequency signals (like TV)  

### Switches

**Knife switch**

has a moving rod locked at one end, can connect or not the other end  

types: ?P?T  
in place of ? can be S or D, for single or double   
SPST: single-pole, single-throw  

pole: 
- S: only one locked end  
- D: two locked end, the rods must move together (if connect one open end the second gets automatically connected to)

thrown:  
- S: have 1 endpoint, can only connect or not  
- D: have 2 endpoints, can choose which one to connect to

**Pushbutton**

usually SPST  

2 types:  
- NO: normally open (if i push, it closes)  
- NC: normally closed

#### Relay

**Relay**

electromagnetic, not mechanic switch  
can be opened or closed sending current to it (has a specific pin to receive such orders)  

### Microphones, speakers

**microphone**: convert sound wave variations in electrical current  

conversion process:
1. sound wave variations   
2. back-and-forth movements of a thin film (like a piece of paper), called diaphragm
3. variations in an electrical current  

there are many ways to do step 2 -> 3

**speaker**

convert variations in electrical current into sound waves
