#### Digital operations

**Fixed-point and floating-point fractions**

floating-point: uses scientific notation (a significand 
per a power of 2)
useful with really small or high numbers   
used a lot in gaming systems, its speed is critical to CPU performance

fixed-point: write decimal points between units and decimals


**Two's complement**

to create negative values   
used for subtraction: A - B = A + (-B)   

complement of a number A: smallest nuber you have to add to A to make the digits of the result overlow  
ex: complement of 15 is 85, bc 15+85=100, now three digits, overflow  

in binary: add 2 numbers, if the result is 0, ignoring the overflow, those two numbers are complementary  
overflow is digits in a position not present in neither of the addends

how to find complement:  
1. flip all digits (0 and 1)   
2. add 1

**Take 2's complement**

1. swap all 0 and 1
2. add 1

to go from 5 to -5 and viceversa (always add 1, never subtract it)

- 2 inputs
- 1 output

shapes:

- XOR: like or, but two parallel lines on left

**The morgan's theorem**

explains NAND, NOR: 

1. $\overline{A \cdot B} = \overline{A} + \overline{B}$  
NAND gate is like OR gate on each of the input's complements (obtained with NOT)
2. $\overline{A + B} = \overline{A} \cdot \overline{B}$   
NOR gate is like OR on complements


**Half adder circuit**

- 2 logic gates, both getting inputs A,B:  
  - XOR: gives value A*B
  - AND: gives carry

**Flip flops**

basis of memory circuits   
to retain memory, need something that retain a state (low or high voltage, so 0 or 1) over time  

- 2 inputs
- 2 outputs: $Q, \overline{Q}$
    - Q is the values i'm storing, the other is its inverse  

types:  
- RS (different inputs to set/reset Q to 0/1)
- D (copy D value on rising edge)
- T (toggle Q value on rising edge)

---

**RS flip-flop**

structure:  
- 2 inputs: $\overline{R}, \overline{S}$ (reset, set)
- 2 outputs: $Q, \overline{Q}$  

how it works:  

happens something when set R or S to 1  
then can stop signal, bc change is permanent until another sygnal changes it  
- S=1 -> sets Q=1
- R=1 -> resets to Q=0

- S=R=1, not allowed
- S=R=0, nothing happens


active low: inputs are actually $\overline{R}, \overline{S}$, so to give S=1 need to input $\overline{S}=0$  
so it's activated when input voltage is low instead of high


made of 2 NAND or NOR GATES


---

**D flip-flop**

- 2 inputs: D, C  (data, rising edge clock)  

clock: gives fixed-interval digital sygnal to synchronize operations  
alternates rithmically between 0 and 1 (graph is squares)  
rising edge: when goes from 0 to 1, acts as action signal  
falling edge: when goes from 1 to 0


how it works:  
when C sees a rising edge, input D is copied in Q  
so don't matter what D does in the meantime, value is only copied at rising edge


---


**T flip-flop**

- 1 input: T (toggle, means to flip between 0 and 1)

how it works:  
when T goes from 0 to 1 (rising edge), Q flips state

connecting many T flip-flops gives a circuit that can count, a counter circuit (considering each value of Q as a positional digit in a binary number)
it is an asynchronous counter, bc not all flip-flops are connected directly to same clock, but they understand it has clocked seeing that the previous flip-flop has changed value  
each flip-flop's Q output is the input for the clock of the next one  

synchronous counter all listern directly to same clock, so toggle together

#### CPU architecture

**Buses**

cpu has control over the address space (memory space)  
can read/write anywhere

buses are usually drawn as lines between components  

communications from CPU to memory:  
- address bus: carry number of memory cell requested from CPU to memory
- data bus: carry data found from memory to CPU
- external bus: from CPU to external memory devices (eg. USB; universal serial bus)  
  - its width tells how fast is data sent between these two objects
- internal data bus: stays in the CPU  
  here, paths followed by data can be changed using multiplexers (MUX)
  its width tells how many bits can process with a single operation


bus is made of bundled signal pathways  
each pathways ( a line connecting 2 points) trasnmits one bit    
bus width: number of signal pathways, how big a number it can trasmit  
modern ALU does 64 bit ops, so buses also have 64 bit width

**Address space**

address is communicated with address bus  
so address bus width is the address space size, how many different addresses are there  
if width is 32 bits, i have 2^32 addresses, roughly 4.3GB

**Control signals (R/W, I/O)**

R/W: read/write (related to CPU)  
read/write when from hardware perspective -> it's an electrical operation in the memory, memory don't care where data is going  
can use load/store from software perspective -> deal wit data flow from memory to registers  

R/W are instruction signals  
sent with control bus together with address bus, to tell what to do at that address  


---

I/O: input/output  
I/O port is the gateway to communicate CPU with I/O devices (some devices, like monitor, don't communicate directly with CPUs, others like keyboard do)   
monitor connected to the GPU (graphics processing unit), which communicates to the CPU with a dedicated I/O port  
smaller systems have no GPU, use LCD (liquid crystal display)  
here, CPU sends data thrpugh I/O port to LCD controller, that outputs image

**Sign bits**

sign bit is leftmost  
tells if number is positive (0) or negative (1)  
if start with 1, means that is a two's complement, so to know its value find the complement and know it's negative    
eg: 011=+3, 101=-3, 01 is complement of 11    

101 is -3 or 5?

**Bit shift**

bit shift are for doing some multiplication or divisions (by a power of 2) quickly 

2 types: 
- logical shift (no sign bit)  
bc negative truth values have no sense, what is -1 in logic?  
- arithmetic shift (yes sign bit)  

right shift:  
arithmetic shift: fill blank space created on the left with 1 if sign bit (leftmost) is 1, 0 if sign bit is 0   
logic shift: always fill with 0  

left shift:  
fill spaces on the right with same rule as right shift  
but leftmost bit now change, number could change sign   
if it happens, it's overflow   
means it used more digits that it was allowed, bc it invaded the leftmost bit reserved for the sign  
if it happens, overflow flag of status register is set   

underflow also exists, if right shift make a number so small that becomes zero bc 1s end up outside of alowed number of digits  

circular shift:  
used in criptography  
no bits removed and added, but those taken from an end reappear to the other

**Operand types**

opcodes can have 0,1,2 operands:  
- 0 operans: the set accumulator to 1 opcode  
- 2 operands: add

types of operands: 
1. immediate value processing  
operans is already a value
2. address reference   
  - absolute/direct redressing    
  give address that containd value (effective address)
  - indirect redressing
  give address of reference to value (tell where to find information on where the value is)  
  useful bc operand has not many bits, can't tell long addresses directly  
  - relative redressing
  effective address is result of sum of operand value (address given)and program counter  
  pften used in brach operations  
  - address modification  
  effective address is operand + value in modification register (usually index register)
