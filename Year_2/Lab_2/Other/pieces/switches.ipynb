### Button switch

![alt text](https://www.addicore.com/cdn/shop/products/AD141-2.jpg?v=1689655661)

button switch

**How it look**

from the top looks like a square  
2 side have 2 pins each, 2 sides have none  
in total, 4 pins

**How it works**


```
   ____
1-|   |-3
  |   |
2-|___|-4


```

horizontally are always connected: 1-3 and 2-4 always connected  

the other pairs:  
when not pushed: not connected  
when pushed: connect oins on same side (1-2 and 3-4), so combined with the horizontal which are always connected, everything is connected to everything  

where to put it:  
pins on same side must NOT be connected in the breadbooard, put them in different rows, else the switch changes nothing if pressed or not  

**Types of switch**

push switch: pins on same side are connected only as long i keep pushing at bottom, when i stop they disconnect  

toggle switch: just press the button once, it changes the state, now they will keep being connected until i press it again, non serve tener premuto   
eg. light switch at home  

**Switch-operated led**

to make a switch, the circuit must be done on 2 pins on the same side, on on diagonal ones  
but not on those connected horizontally

### Transistor mosfet

![alt text](https://components101.com/sites/default/files/component_pin/P55NF06-Pinout.jpg)

the front is the side with the plastic cover  
the side all in metal is the back  

**Connect pins**

look from the front, there are 3 pins, start counting from left  

1- gate: if received here voltage changes, connects 2 and 3  
2- drain  
3- source  

what should they be connected to:  
1- to an Arduino pin that sends voltage to control it  
2, 3 are a break in a circuit, that can be closed or not
