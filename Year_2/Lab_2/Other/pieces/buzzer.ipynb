**Tipi di buzzer**

attivo: serve DC (quindi stanrd 5V pin o digital pin)  
suono è creato da disco che vibra  

passivo: serve AC (quindi analog pin o digital pin con PWM)  
usa attrazione e repulsione elettromagnetica per creare suono  

entrambi hanno un pin positivo (più lungo) e negativo (où corto)  
pin positivo si connette al lato positivo della batteria  

**Codice buzzer passivo**

```
passiveBuzz = 9;  

void setup(){
    pinMode(passiveBuzz, OUTPUT);
}

void loop(){
    analogWrite(passiveBuzz, 255);
    delay(5);
    analogWrite(passiveBuzz, 0);
    delay(5);
}

```

per usarlo, devo cambiare V che gli fornisco con un pin analog, così si crea della AC
