**Types**

2 types: LM, DHT  
most common are: LM35, DHT11, DHT22  

used to rilevate temperature  

**How to use LM**

front: the flat side with text  
back: the other  

look at it from the front, has 3 pins:  
ledt: VCC  
center: analog pin for reading  
right: GND  

these 3 pins must NOT be connected, put them in different rows  

just put the sensor anywhere in series in the circuit 

**How to read LM**

middle pin reads voltage  
check the sensor documentation for the 'linear scale factor'  
eg. if it's $10 \frac{mV}{°C}$, mean the sensor sends 10mV on the middle pin for every Celsius degree  
write function to go from value of voltage of middle pin to temperature  
$V = T \lambda$  
rem that voltage read is a fraction of the max one, put max/1023 to have the real one  

**Code LM**

```
int readPin = A2;
float scale_f=0.01;

void setup(){
    pinMode(readPin, INPUT);
    Serial.begin(9600);
}

void loop(){
    readValue = analogRead(readPin);
    voltage = readValue * (5.0/1023);
    temp = voltage/scale_f;
    Serial.println(temp);
    delay(50);
}
```


**How to use DHT sensor**

need DHT sensor library Datafruit


**DHT code**

```
#include <DHT.h>
#include <DHT_U.h>
#define Type DHT11;

int sensePin = 2;
DHT HT(sensePin, Type);

void setup(){
    Serial.begin(9600);
    HT.begin();
}

void loop(){
    humidity = HT.readHumidity();
    temp = HT.readTemperature();
    println(temp);
    delay(50);
}


```


**Notes on reading**

digitalRead: reads 1 bit, 2 values  
analogRead: reads 10 bit, 1023 values  

so analog pins are slower  
so if both are ok, better a digital one
