so, if the resistor is small enough, have enough voltage drop to light up the LED

diode has voltage drop 0f 0.7V

## Transistor gates

the question is: in P, is voltage low or high?  
AND: serie  
OR: parallelo  
VCC -> transistors -> P -> resistor -> GND

transistor allow current to pass (closes the circuit) only if receives voltage from the middle pin 

the inputs are points A and B, whose voltage can be low = 0V or high = 5V  

want to check if at point P, where is the output wire, voltage is high or low  

current can only pass if there is path from VCC to GND  
else there is no current, so resistors cause no voltage drop  

current cannot just decide to flow from VCC to the output point if GND is blocked  
the arrival point MUST be GND  

if a transistor leaves the circuit open, to know voltage at point P:  
see if P is connected through wire to VCC or GND (no breaks in the circuit until there)  
then check if current is allowed to flow: if not, resistor have no voltage drop, ignore them  
P takes same voltage as the point, VCC or GND, that it's connected to  

transistors have no voltage drop  
transistor closed the circuit if recevies HIGH from the gate (middle pin)  

resistors have voltage drop only if traversed by current  


**Steps**

1. check if there is current between VCC and GND in each configuration  
2. check if:  
transistors are closed  
resistors have voltage drop (of there is current)  
3. cehck if P has same olta as VCC or GND  

### NAND gate

![alt text](https://blogger.googleusercontent.com/img/b/R29vZ2xl/AVvXsEjIJL2cqJFfFcnkMnIrvkY2YVoPLYRcu_W_9zPFKtDVLGfiC-QHE8513AwwWuX4PNnQWW5AeT2_QfS6-3e98nZJ3HkOSLiFst86LbFSA-xnJnR_GZ14gYpKSyleBvgRhEPQq5xDSvzikYUvaWVNjTX2o-71Nmk_OsdXfTseQ1jXbNp5yfjgmHpRxBuJ8fk/s746/nand%20gate%20transistor.jpg)

if both A and B high, current flows from VCC to GND  
from GND to output there is no voltage drop, so they're at same voltage, so output is 0V  

if at least one is low, current can't reach GND  
so there is NO CURRENT through the resistor, so the output is at same voltage of VCC, so high  

formula is: $(AB)'$  

A and B can have values of 0 or 1  
multiply them  
then take the 1's complement (essentially swap the value)
