```
int digitalPin = 7;  
int analogPin = A7;
```

then pinMode is the same for both  

how to use them:  
```
digitalWrite(digitalPin, HIGH);
analogWrite(analogPin, 255);
```
digital can just be HIGH or LOW, on or of  
analog gets an integer from 0 (low) to 255 (high)  
where 255 is the max voltage of the pin (max voltage is the pin number, this is relative to that)
eg. if put 255 on the pin A7, i get voltage 7V  

**Pulse width modulation (PWH)**

where:  
some digital pins have a tilde or another symbol before the number on the Arduino borad  

what:  
to obtain partial voltages with digital pins  
this is better than analog pins  

how:  
to obtain 1V on a 5V pin, in a time nterval keep LOW for 4/5 of the time, HIGH for 1/5  

so to get a specific voltage:  
analog: really always otuput that voltage  
digital with PWH: alternates HIGH and LOW, so the average is that voltage  

code:  
in digital pins that support PWH, can use the analogWrite command and write my desired voltage  
then Arduino automatically does the alternating HIGH and LOW thing to obtain it  
such pin is still indicatednormally, with only the number, so symbol before it  

```
int pin = 7;  
analogWrite(pin, 255);
```


### Read input

**Use pins to read an input**

```
int readPin = A7;

void setup(){
    pinMode(readPin, INPUT);
}

    analogRead(readPin);
}
```


**Volt reader**


```
int readPin = A7;

void setup(){
    pinMode(readPin, INPUT);
    Serial.begin(9600);
}

void loop(){
    reader = analogRead(readPin);
    Serial.println(reader);
}
```

**Serial monitor**

sends data from the computer to the Arduino and viceversa  
to se it up, use Serial.begin(9600)  
9600 is the most common transmission rate   

println prints everthing in a newline  
print don't leave space between things, bad, get a single number, tutto attaccato  


**Where to read the serial monitor?**

click on the serial monitor icon, una specie di lente d'ingrandimento  

note that, if a using a physical arduino (not a simulation), the serial monitor must be closed while compiling, or it throws error  

it shows a value between 0-1023 (bc $2^{10}=1024$), which represents the percentage of voltage receiving  
eg. if the reader is A5, and the monitor shows 1023, it's reading the amx voltage, 5V  

analogWrite takes in input number in range 0-255
analogRead outputs numbe in range 0-1023  
but they're both relative scales, the max is always the name of the pin (A5 has 5 max voltage)  
so for A5, 2.5V  is 127 in write and 511 in read  

### RGB led

![alt text](https://projects.arduinocontent.cc/a402677b-ba8b-4441-a052-cda5b140852b.png)

**How to use**

need 3 analog pins, each sending the value of one color, so define a color using the RGB incription  

usually there are resistances already installed inside (like in the image), so it don't fry  

the longer pin must be connected to GND in the Common Cathode (the most comoon version)  
could also be Common Anode, where the longer pin is connected to VCC, but it's really unlikely   


**Color system**

RGB: addictive  
used when NOT starting with sunlight  

RYB: subtractive  
used when starting with sunlight, that's white, and taking away wavelengths  
eg. prisms  

**Choose color with serial monitor**

```

# include <LiquidCrystal.h>

LiquidCrystal led_1(12,11,5,4,3,2);  
// sequence of pins connected to what?  

void setup(){
    led_1.begin(16,2);
    \\(16,2) is size of screen, (cols, rows)
}

void loop(){
    led_1.print("Hello");
    led_1.setCursor(0,1);
    led_1.clear();
}

```

.print() starts printing from where the cursor is now  
if no more space on the screen, don't print  
if there is already text there, it gets overwritten  

setCursor move the cursor to (col, row)  
indexing starts from 0  
so (0,1) is first col of second row
