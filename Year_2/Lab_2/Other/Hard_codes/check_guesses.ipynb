```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
import ipywidgets as widgets

# Your damped oscillator model
def osc(t, A, tau, omega, phi, B):
    return A * np.exp(-t/tau) * np.cos(omega*t + phi) + B

# Example: replace these with your data
# ---------------------------------------
# Suppose you already have x, y, Dx, Dy
x = np.linspace(0, 50000, 200)
y = 5 * np.exp(-x/15000) * np.cos(0.0045*x + np.pi/2) + 0.3*np.random.randn(len(x))
Dx = np.zeros_like(x)
Dy = 0.3*np.ones_like(x)
# ---------------------------------------

# Plotting function with sliders
def update(A=2.5, tau=14000.0, omega=0.0045, phi=np.pi/2, B=0.0):
    plt.figure(figsize=(8,5))
    plt.errorbar(x, y, yerr=Dy, fmt='o', ms=3, alpha=0.6, label="Data")
    xx = np.linspace(min(x), max(x), 500)
    plt.plot(xx, osc(xx, A, tau, omega, phi, B), color='red', lw=2, label="Guess")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)
    plt.show()

# Create interactive sliders
interact(
    update,
    A=FloatSlider(value=(np.max(y)-np.min(y))/2, min=0, max=10, step=0.1),
    tau=FloatSlider(value=14000, min=1000, max=50000, step=500),
    omega=FloatSlider(value=0.0045, min=0.0005, max=0.01, step=0.0001),
    phi=FloatSlider(value=np.pi/2, min=0, max=2*np.pi, step=0.1),
    B=FloatSlider(value=np.mean(y), min=-5, max=5, step=0.1),
)
```
