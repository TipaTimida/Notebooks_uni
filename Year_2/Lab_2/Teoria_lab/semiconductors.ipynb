#### Doping

**Definition**

what it is:  
replacing some atoms in a lattice with another element  
essentially, means creating a very diluted alloy  

why doing it:  
to increase electrical conductivity

**Types: structure**

1- the name tells me what is the majority carrier (mobile charge carrier)  
2- so the ionic core must have the opposite charge (fixed charge)    

n-doped: $D^+ + e^-$  
p-doped: $A^- + hole$

### Bias - How charge carriers and current flow

what | direct bias | reverse bias
---|---|---
cosa è più forte | diffusione | forza del built-in field  
cosa passa | portatori maggioritari (da N vanno via elettroni, da P vanno via holes) | portatori minoritari (da N vanno via protoni, da P vanno via elettroni)  
da che parte spinge gli elettroni | da N a P | da P a N  (opposite direction of built-in field)
how they pass | by diffusion (charges pass bc of gradient, field is too weak to oppose them) | they get sweeped in by the force from the electric field (NOT by diffusion)  

**Example: diode**

what i need:  
bias diretto: diminuisce il built-in field, così riprende la diffusione secondo gradiente da N a P 
N must be close to the GND, so the fem of the battery is in the opposite direction of the built-in field of the diode  

how i works:  
elettroni iniziano da nodo - della batteria, vanno a N, passano a P, e finiscono a polo + della batteria  
ricorda che il moto degli elettroni è in direzione opposta al verso convenzionale della corrente  

**Example: transistor effect**
 
C    -   B     -   E   
N     -   P    -    N   

situation: active mode, so BE in forward, BC in reverse

1- BE junction, forward bias  
quindi vince la diffusione, elettroni passano da N a P  per diffusione  

2- BC junction, reverse bias    
quindi vince il built-in field, elettroni passano da P a N tirati dal field

#### Depletion region (regione di svuotamento)

**Depletion region**

it's the region a cavallo of the junction  
dove cariche mobili si sono ricombinate, quindi sono rimaste solo quelle fisse   
è la regione dove esiste il built-in field      

![](https://upload.wikimedia.org/wikipedia/commons/thumb/f/fa/Pn-junction-equilibrium-graphs.png/960px-Pn-junction-equilibrium-graphs.png)

**Width of depletion region, according to bias**

built-in field is caused by charge separation  

1- reverse bias  -> increase width  
strenghten the built-in field, so strenghten charge separation  
so pushes charges more away from the interface

2- direct bias -> reduce width  
weakens the built-in field, so weakens charge separation  

#### Built-in field

**What**

what: an electric field that exists ONLY IN THE DEPLETION REGION   

caused by: the fixed charges in the depletion region  
consequences: creates a potential barrier that stops further electron diffusion, reach equilibrium

**Direction**

E goes from positive to negative charges    
so from N to P (bc fixed charges are opposite to the name)  

note: field must stop electrons from diffusing  
electrons feel force in the opposite direction of the field  
so if field points from N to P, electrons, which are more abundant in N, feel force from P to N, so get blocked from passing  
exactly what i wanted

**Potential barrier**

the built-in field creates a potential barrier  
that stops diffusion and creates a situation of equilibrium  

potential is higher in the N-doped region, bc it's the one with the positie fixed charges  
and formula of potential means it's higher for positive charges, there is no - in the formula

### Band theory of solids

**What it's for**

visualize difference between conductors, insulator, semiconductors  
by plotting the available energies for electrons in the material  

**Bands**

an isolated atom has discrete energy levels (orbitals)  
in a solid (made of many atoms), these levels broaden into energy bands   
because they split due to interactions and Pauli exclusion???  

assign an electron to a band according to how much energy it has   
if obtains enough energy, it can jump to a higher band  

**Important bands**

valence: the highest energy band that is normally filled with electrons at low temperature  
contains electrons involved in bonding  

conduction: the next higher band above the valence band   
electrons in this band are mobile, cause electrical conduction  
bc these electrons are movile charge carriers, create current  

band gap: the energy gap between them  
so the energy that i need to give to an electron in the valence band to jump to the conduction band  
$E_g = E_{conduction}-E_{valence}$  

**Distinguish materials**

insulator: big band gap ($E_g>4-5 eV$)  
at ordinary temp, no electron can reach the conduction band  
even doping is not enough  

semiconductor: small band gap ($E_g \sim 0.1-0.3 eV$) 
gap can be bridged by small excitations, even just thermal  
doping is relevant  

conductor: no gap, valence and conduction bands ($E_g=0$) overlap    

#### Effects of thermal energy

**At room temp**  

thermal energy: $k_B T$  
potential difference it causes: $V= \frac{k_BT}{e}$   
energy it gives electrons: $U=eV$ (in electronvolt, e=1)  

at room temp, U=26m\, eV   
thermal energy is smaller than what required to “effectively” interact with the matter  
too small to bridge the band gap, even in semiconductors  
but not all atoms have same enrgy, they follow a Boltzmann distribution  
and the higher tail of the distribution can reach the $E_{g}$ value  

so: if $T \neq 0$, expect a small, but non-0 number of electrons to jump the gap  

**Gap and bounds**

if electron in valence band jumps in conductor band  
the bound it was making before is broken  
so jumping the band gap emans breaking a bond  

**Why a missing bond**

in the lattice, a bind might be missing bc of:  
1- presence of impurities  
2- temperature -> too much energy (ionization energy of the atom) can break a bond

### Properties of semiconductors

**Characteristics of semiconductors**

peculiar macroscopic behavior in terms:

1- electrical transport: higher resistivity than conductors
so smaller current  
why: consequence of the smaller density of charge carriers  
  

2- dependence on temperature: NTC (negative temp coeff)   
so I decrease when T increase   
conductors have PTC, the intuitive one, more thermic agitation = more current    
why: if T increase, density of charge carriers increases, so I increase and R decrease 

**Silicon**

1- is an element, non una lega/miscuglio  
2- used in most electronic components, except LEDs  
3- abundant on earth, easy to use it to fabricate stuff  
4- atom arragnement in a regular configuration in the space, called lattice, and the material is said to form a crystal   
Modifications of the regularity, for instance due to the absence of one atom, or to the replacement of one atom with another species, or to any other distortion leading to change the interatomic distance, are called defects  

**Alloys**

semiconductor alloys: made of elements belonging to the III and V columns  
especially useful in optoelectronics  



**Lattice (reticolo)**

lattice: geometrical arrangement (config) of the atoms/ions in the solid
can happen in both conductors, semi, and insulators    

conductors:  
one electron per atom in the lattice is released as a delocalized charge (bc some electrons are in conduction band)   
The density of charge carriers is therefore very large  
so in a metal there is a sea of electrons  

silice (insulator):  
do 4 covalent bonds to hold the grid  
so electrons stay localized (bc are all in valence band)  
all electrons are localized and no charge carrier may exist -> the material is insulator
