### What it is

**What is it**

metodo per esprimere:  
- prima: una funziona periodica  
- dopo: superposizione lineare di funzioni sinusoidali, di frequenza angolare $ω_k = kω_0$ 


why:  
so can study the signal in frequency domain  

**Nomenclatura**

the components of frequencies $\omega_k$ called HARMONICS of order k    
they form an orthogonal base for the space of all signals   

$\omega_0$ si chiama 'armonica fondamentale'  

**Expansion to limited functions**

the signal i want to approximate must be periodic   

trick: if it isn't periodical, but defined in a limited interval  
posso immaginare di estenderla ripetendo questo intervallo all'infinito, così diventa periodica  
si chiama 'estensione implicita'  

**Problem**

se la funzione ha pezzi verticali, cioè divergenti (come la square waveform), viene un po' male, appare il fenomeno di Gibbs  

### How to write

**How to write**

dipende se voglio i coeff nel campo reale o complesso  

1- in real field   
$g(t)=\frac{a_0}{2}+\Sigma_0^\infty b_k \cos(\omega_k t) +\Sigma_0^\infty c_k \sin(\omega_k t)$  

2- in complex field   
$g(t)=\frac{a_0}{2}+\Sigma_{-\infty}^\infty d_k e^{-j \omega k \,t}$  
where: $\omega = 2 \pi f$  
the - at the exponential is a convention  

notes:  
attenta ai limiti delle sommatorie  
nel campo complesso devo usare anche quelle negative, con k<0   

### Find coeffs

**Come si fa**

1- trova periodo della funzione, T  
se funzione è limitata, T è solo l'intervallo su cui è definita  

2- calcola coeff:  
$a_0/2=<g>_T$  
$b_k = \frac{2}{T} \int_0^T g(t) \cos (\omega_k t) \, dt$  
$c_k = \frac{2}{T} \int_0^T g(t) \sin (\omega_k t) \, dt$  
$d_k = \frac{1}{T} \int_0^T g(t') \, e^{j 2 \pi f k \,t'} \, dt'$  

**Note per calcolo**

1- per ogni coeff, moltiplico il segnale per la funzione a cui fa da coeff  

2- l'integrale va fatto su un periodo, vanno bene qualunque estremi che in totale diano il periodo  
quindi se il periodo è $2 \pi$; posso avere $[0, 2\pi]$ o $[-\pi, \pi]$  

3- a è definito così, perchè così è uguale agli altri:  
$a_0 = \frac{2}{T} \int g(t) \, 1 \, dt = 2 <g>_T$  

**Nota: periodicità della funzione**

se la funzione da rappresentare ha periodo T, allora la base trigonometrica è:  
$\cos(\frac{2 \pi n}T x) = \cos (n \omega_0 x)$  

dove x=t se sto usando f(t)  
e $\omega_0 = \frac{2 \pi}{T}$    

a seconda della periodicità della funzione:  
$[-\pi, \pi]=[0, 2 \pi]$: serie completa, di seni e coseni  
$[0, \pi]$: basta o solo seni o coseni, a seconda se faccio estensione implicita su $[-\pi, \pi]$ pari o dispari   
$[0, 4 \pi]$: serie completa, ma seni e coseni sono: $\sin(\frac{nx}{2})$

#### Tips

**Symmetry: delete coefficients**

cosa | simmetria del segnale | simmetria di semionda 
---|---|---
cosa | normale simmetria di una funzione rispetto a asse y (pari o dispari) | disegna un solo periodo, guarda se c'è simmetria (DISPARI!!!) rispetto alla linea verticale T/2    
che fa | cancella o tutti i cos o tutti i sin, perchè funzione pari non può avere componenti dispari e viceversa | elimina le armoniche pari (con k pari)  

**Simmetria del segnale**

cosa fa:   
segnale dispari: elimina i cos (perchè cos è pari)   
segnale pari: elimina i sin   
ma non toglie alcuna armonica tra quelli rimasti  

perchè:  
i coefficienti si ottengono facendo un'integrale su $[-\pi, \pi]$ 
quindi l'integrale di qualcosa di dispari dà 0 (si vede graficamente)  

nota:  
per segnale limitato, definito solo in $[0, T]$ (quindi non definito prima di asse y)  
se è pari o dispari dipende da come scelgo di ripeterlo per renderlo periodico  

**Simmetria di semi-onda**

ce l'hanno: onda quadra, triangolare, a dente di sega simmetrica  
ma cambiare duty cycle a quadra o triangolare la rende non più asimmetrica: appaiono armoniche pari  

#### Other tips coeffs

**Change properties**

change amplitude: multiply all the series by A  
change phase: add a phase inside all sin and cos  

**Other tips**

sometimes can notice by eye if the time integrals on some number of periods is 0  

triangular waveform can be obtained by integrating square waveform

### Theory

#### Theory coeffs

**What coeffs mean**

$a_0/2=<g>_T$: it's the average on a period, so it's 0 for alternate waveforms   
$b, c$: they define the spectral content  

**Spectral content**

what:  
how much each harmonic contributes to the signal (how much that harmonic is present, relatively to the others)  
the distribution of amplitudes across harmonics is the spectrum  
spectra: plots of some physical quantity as a function of frequency  

how to represent:  
the set of all coefficients is the signal’s spectral content  

**Quando usare le armoniche**

un'onda sinusoidale ha solo l'armonica fondamentale, con la sua frequenza e con ampiezza 1  
le altre armoniche appaiono solo per approssimare onde non sinusoidali, tipo quadrate o triangolari  

**Square waveform, Gibbs phenomenon**

problem:  
the derivative of a square waveform in the rise and fall (vertical part) diverges (is infinite)  
while the derivative of sine and cosine functions gets always a finite value  
hard to numerically obtain sharp variations (square) starting from finite slope functions (sin, cos)  
need to use a lot of addends  

Gibbs phenomenon: persistent overshoot or undershoot and oscillation that occurs when approximating a function with a jump discontinuity using a finite number of terms in its Fourier series  

**Nota: strumenti**

alcuni si suonano segando una corda (violino), altre percutendole (piano)  
quelle segate danno suono più simile alle onde triangolari, quelle del piano più simile a onda quadra  

**Decay of coeffs**

decay of coefficients:  
square, sawtooth: $1/k$  
triangular: $1/k^2$

if have bigger coefficients in high-order harmonics, it's sharper  
so the faster the sequence of coefficients decreases, the smoother the function  
so square has sharp parts bc its coeffs decrease slowly, so even gifh harmonics have big coeffs  

**Low filter on square and sinusoid signal**

with a square waveform, gain in the transition region
is larger than with a sinusoidal one  
so, obtain a cut-off frequency larger than the expected value, determined by R and C  

why:  
square waveforms possess higher order harmonics, sinusoidal don't, bc square is sharper  
lower-order ones are attenuated in both by the same amount  
higher-order are attenuated in square and inexistant from the beginning in sinusoidal, so square has bigger signal amplitude in the transition region  

### Hard stuff

**Starting phase**

in the Fourier series, there is no phase inside the trig  
bc it's already encoded in the coeffs $b_k$ and $c_k$  

explanation:  
$\cos (\omega t + \Phi) = \cos(\omega t)\cos \Phi - \sin (\omega t) \sin \Phi$  
cioè moltiplicare una funzione sinusoidale per un peso costante equivale a sfasare quella funzione

why there is phase in pyhon code:  
i wrote the square function with only sin, no cos, so assume that the func is odd, so that it starts with a vertical line  
if it doesn't, would need to use cos too, which meakes it very hard -> so need to change the coeffs, bc those encode the phase  
inasted, easier to leave coffs as is and put a phase inside the trig (or a $t-t_0$ inside of $t$), so an define the starting moment i want and make it appear odd  
i'm talking about macroscopic phase, not fourier phase  

**Amplitude**

macroscopic/signal amplitude: $A$  
tells me height of the waveform  
it is the A that appears in the coeff for square or triangle wave  

Fourier amplitude: $A_n= \sqrt{b_n^2+c_n^2}$  
tells me the height of this harmonic    
actally, tell you how much energy sits in this harmonic (used to draw spectra)  
it is NOT the A that appears in coeffs  

**Phase**

signal/macroscopic phase: $\phi_0$  
initial phase of the whole signal  
related to time shift of the entire waveform: $\phi_0=-\omega t_0$  
so can be swapper with using: $t-t_0$

Fourier basis phase: $\phi_n = \arctan(\frac{c_n}{b_n})$   
the phase of each harmonic    
is NOT due to time shifting, it comes from waveform shape or asymmetry  
note that if use only sin or only cos, this gives 0

each harmonic has the sum of the macroscopic one and the fourier one:  
$w_n =  b_n \sin(\omega t + \phi _n + \phi_0)$  

**Using a filter**

using filter can cause a change in both macroscopic and fourier params  


**Steps to use filter**

1- use the H function  
2- for each harmonic, compute gain and phase  
3- apply them  

`Hn = rc_highpass_H(omega_n, fc)`  
`gain = np.abs(Hn)`  
`phase = np.angle(Hn)`  
`y += (amp_in * gain) * np.sin(omega_n * (t - t0) +( phi_in + phase))`

for amp_in and phase_in, use the Fourier amplitude and phase, not the macroscopic ones  

### Common coeffs

#### Square wave

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRYcghZtJDWWvomg2msDcGJ7p4Qe7LvfYzq0A&s)

ha ampiezza A in un semiperiodo, -A nell'altro  

**Properties**

simmetries:  
1- signal: depends on where it starts  
if start with vertical line, it's odd signal, so only uses sin  
2- half-wave: yes, so drop even harmonics  

coeffs:  
$a$: depends if it's alternated, if yes put 0   
$b_n$: 0, bc said only use sin  
$c_n = \frac{4A}{n \pi}$, with A amplitude, only for n odd  

result:   
$g(t)=\frac{4A}{\pi}(\sin (\omega_0t) + \frac{1}{3} \sin (3 \omega_0t) + \frac{1}{5}\sin (5 \omega_0 t)...)$

**Compute integral**

step | descrizione 
---|---
$b_n = \frac 2T \int_{-T/2}^{T/2} \, f(t) \, \sin (n \omega_0 t) \, dt$  | def di coeff, dove $\omega_0 = \frac {2 \pi}T$
$ = \frac 4T \int_{0}^{T/2} \, A \, \sin (n \omega_0 t) \, dt $ | usa simmetria del segnale  
$= \frac{4A}{T} [-\frac{\cos(n \omega_0 t)}{n \omega_0}]_0^{T/2} = \frac{4A}{T}\frac{1- \cos (n \pi)}{n \omega_0}$ | integra  
$ \frac{4A}{T}\frac{1-(-1)^n}{n \omega_0}$ | poichè $n \in N$, considera separatamente n pari e dispari  
n pari: $=0$ <br>  n dispari: $\frac{4A}{n \pi}$   

**Nota**

la simmetria del segnale la devo usare per arrivare alla risposta  
la simmetria di semionda no, semplicemente è per controllare i risultati  

#### Triangolare

alt text

i rami di salita e discesa sono $f(t)=\pm \frac{4A}{T} t$  
uso 4A perchè vale in $0<t<T/4$  

**Triangle wave**

simmetries:  
1- signal: depends on where it starts  
if start with diagonal line, it's even signal, so only uses cos  
2- half-wave: yes, so drop even harmonics  

coeffs:  
$a$: depends if it's alternated, if not put 0  
$b_k$: 0, bc said only use sin  
$c_k = \frac{8A}{(n \pi)^2}(-1)^{(n-1)/2}$, with A amplitude, only for n odd  

result:   
$g(t)=\frac{8A}{(n \pi)^2}(\sin (\omega_0t) - \frac{1}{3^2} \sin (3 \omega_0t) + \frac{1}{5^2}\sin (5 \omega_0 t)...)$

**Calcolo**

step | descrizione 
---|---
$b_n = \frac 2T \int_{-T/2}^{T/2} \, f(t) \, \sin (n \omega_0 t) \, dt$  | def di coeff, dove $\omega_0 = \frac {2 \pi}T$
$ = \frac 4T \int_{0}^{T/2} \, A \, \sin (n \omega_0 t) \, dt $| usa simmetria del segnale  
$= \frac 4T [\int_0^{T/4} \frac{4A}Tt \sin (n \omega_0 t) \, dt \; + \int_{T/4}^{T/2} (2A - \frac{4A}T t ) \sin (n \omega_0 t ) \; dt]$ | separo parte di salita e discesa  
$= \frac{8A}{(\pi n)^2} \sin (\frac{n \pi} 2)$ | integro e ottengo  
n pari: $=0$ <br>  n dispari: $\frac{8A}{(n \pi)^2}(-1)^{(n-1)/2}$    | poichè $n \in N$, considera separatamente n pari e dispari
