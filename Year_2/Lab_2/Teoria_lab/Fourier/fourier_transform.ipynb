### Definition

**Fourier series vs transform**

both are for passing from time domain to frequency domain  
but Fourier transform is more general, can apply it for more signals  

What | Series | Transform
---|---|---
applied to | PERIODIC signal <br> (or limited, con estensione implicita) | TIME-DEPENDENT signal
can be written as | SOMMATORIA (composizione lineare delle componenti) | INTEGRALE  
what are the components | DISCRETE frequencies (harminonics) <br> in real field have only positive $\omega_k$, in imaginary field also negative <br> $f_ = \frac nT$ | CONTINUOUS frequencies   
what gives me | coeffs $c_n$ for each freq | a continuous func X(f)  
what i plot (in freq plot)|a series of spikes | a continuous function

**Practical**

the fourier transform is the way to compute the coefficients for each frequency  
so, the amplitude of each frequency is the spectra of the signal  

**Recap**

fourier series: go to freq domain for periodic signals  
fourier transform: want to do it for any time-dependant signal  
discrete fourier transform: in reality, can only sample the real signal  

#### When i can do it

**Vars to swap**
  
FT can be defined for any pair of “conjugated” quantities, (that give 1 when multiplied with each other)  

$fT=1$: so i can  
$\omega T = 2 \pi$: so i need to use normalization factors  

**Example**

$t-f: \quad g(t) = \int g̃(f) e^{-j 2 \pi ft} df$  
$t-\omega: \quad g(t) = \frac 1{2\pi} \int g̃(\omega) e^{-j \omega t} d\omega$  

differences:  
1- change what is esplicitato in the exponent of e  
2- appear the normalization factor:  
$f = \frac{\omega}{2 \pi}$  
sometimes, instead of $\frac{1}{2 \pi}$, could put $\frac{1}{\sqrt{2 \pi}}$, to make the transform unitary  

#### Series vs transform

**How to write them**

series:  
$d_k = \frac 1 T \int_0^T g(t') e^{j 2 \pi k ft'} dt'$  
$g(t) = \Sigma_k d_k \, e^{j 2 \pi k f_0 t}$  


transform:  
$g̃(f) = F\{g(t)\} = \int_{-\infty}^{\infty} g(t') e^{j 2 \pi f t'} dt'$  
$g(t) = F^{-1}\{\tilde g(t)\} = \int g̃(f) e^{j 2 \pi ft} df$  

difference:  
series: there is not really a way to go back to g(t), just plug back the coeficients in the series and do the summation    
transform: to go bck to time domain, need to use the inverse transformation, IFT  

**How to write them: explanation**

to go from periodic signal to any time-dependant signal, means i'm doing: $T \rightarrow \infty$  

consequences:  
1- not have harmonic any more, drop k  
2- can't have reference to T anymore, integrate on integrate on $[-\infty, \infty]$ instead of $[0,T]$

#### Inverse Fourier Transform (IFT)

**Definition**

FT: find coeffs to go from g(t) to g(f)  
IFT: go back from g(f) to g(t)  

**Complete formulas**
  
$g(t) = \int g̃(f) e^{j 2 \pi ft} df$  

FT:  
$g̃(f) = F\{g(t)\} = \int_{-\infty}^{\infty} g(t) e^{j 2 \pi f t} dt$  

IFT:  
$g(t) = F^{-1}\{g̃(f)\} = \int _{-\infty}^{\infty} g̃(f) e^{-j 2 \pi t f} df$  

differences:  
1- in the exponential, must have opposite signs  
2- swap from dt to df, according to the var i'm using now  

**Uses**

1- start in time domain  
2- pass in frequency domain (FT)  
3- apply the transfer function (H)   
4- go back to time domain (IFT)   

$g_{out} = F^{-1}\{H(f) g̃_{in} (f )\} $   
where: $g̃_{in}(f ) = F\{g_{in}(t)\}$

### Discrete Fourier Transform (DFT)

**Problem**

i can only measure the signal at discrete values of t, not continuous  
bc the object i use has a sample time (delay between samples)  
don't have a function $g(t)$, but an array with the values of $g$ sampled at some times $t$   

consequence:  
for the transform i cannot do an integral, i must do a finite summation  

what | Continua | discreta
---|---|---
segnale originale | $g(t)$ con t continuo <br> cioè una funzione continua | $g_l$ con l discreto <br> cioè è un array, con l come indice  
trasformata | $g̃(f) = F\{g(t)\} = \int_{-\infty}^{\infty} g(t') e^{-j 2 \pi f t'} dt'$ | $g̃_m = F\{g_ℓ\} = \Sigma^{N-1}_{l=0} \, e^{-j 2 \pi ml/N}$  
output | una funzione continua $g̃(f)$ | un array $g̃_m$  

**Indexing**

can sample a point every $\Delta t$ time  
cioè, ho frequenza di campionamento $f_s = \frac 1{\Delta t}$  
in total, take N points, so $l \in [0, N-1]$  

what is l:  
so $g_0$ is the value sampled at t=0, $g_1$ at $t=\Delta t$, $g_2$ at $r=2 \Delta t$...  

what is m:  
indica le frequenze per cui cerco la ASD  
sono le freq $f_m = \frac mN f_s = 0, \; f_s/N, \; 2f_s/N$...   

note su m:  
1- aumentare N me lo fa diventare più preciso, posso misurare più frequenze   
2- in teoria il massimo è $m=N-1$, come per l, ma la secomnda metà di freqienze è negative  
quindi la massima freq osservabile è per m=M/N, cioè $f_{max}= \frac{f_s}2$    

**Note: actually useful points**

what happens when shift: $m \rightarrow m+  N/2$    
it means doing: $f \rightarrow -f$  

explanation:  
it's a change of $\pi$ in the argument:  
$e^{-j 2 \pi l(m+N/2)/N} = e^{-j 2 \pi lm/N} e^{-j \pi l}  =  e^{-j 2 \pi lm/N}  \, (-1)^{l}$  
so elements with m in [1, N/2 − 1] correspond to positive frequencies, those with m in [N/2, N − 1] to negative frequencies    

consequence:  
all measured data ($g_l$) is real-valued   
so $g_m(f)$ for m and $m+N/2$, meaning $g(f)$ and $g(-f)$, will be complex conjugated  
since i only care about the magnitude, which is the same in complex conjugated, the negative freqs are redundant information, can ignore them  

solution:  
the number of meaningful elements in the transformed array will be the N/2 positive freqs and the m=0  
so only N/2 +1 elements  

**Other explanation**

if V(t) is real, the transform has a simmetry:    
$X(-f)=X^*(f)$  

so negative frequencies are redundant, i don't need the full FFT; can compute only the positive half 


### Fast Fourier Transform (FFT)

**Problem**

in practice, must use the DFT and IDFT  
but they need a lot of calculations, takes long on big datasets  
need $N^2$ opetarions  
FFT is an algorithm that only needs $N \log (N) $ operations 

**Tips: sampling**

condition: $N= 2^p$  
means: the basic implementation of the FFT, the one you are strongly advised to use, requires the number of data point N to be a power of 2   
how: if not, either cut the data or add padding  

tip:  
the transformed array will contain $M = N/2 + 1 = 2^{p−1} + 1$ meaningful elements  
if i tell the software by code that my array is real-valued, it automatically discards redundant values  

### Code

ASD = np.abs(np.fft.rfft(V))
