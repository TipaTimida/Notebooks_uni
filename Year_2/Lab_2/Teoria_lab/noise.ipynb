### Noise

**Noise vs signal**

$V_{in}(t)=S(t)+N(t)$   

noise (N): all those parts of a signal which do not carry information on the physical quantity of interest   
signal (S): the amplitude of the signal  

S/N ratio: Signal-to-noise ratio   
to say how much a signal is 'dirty'  

#### Theory

**Noise sources**

1. electromagnetic coupling (capacitive coupling, E-field coupling)  
External electric fields induce unwanted voltages  
other signals get coupled to the wires connected to the instrument, which act as antennas  
we are always immersed in many electromagnetic fields at radio frequencies, which can be captured and  converted into currents and voltages superposed to the signal of interest  

2. magnetic induction (pick-up noise) (inductive coupling, B-field coupling)  
Time-varying B-fields induce currents in loops  
magnetic coupling with periodical currents, which
produce time-dependent magnetic fields able to induce periodical voltages into the wires connected to the oscilloscope  
Periodical currents are ubiquitous  

coaxial cables  represent a viable method to reduce any form of spurious coupling  

**Flicker noise**

flicker noise / pink noise:  
low-frequency electronic noise with a power spectral density that is inversely proportional to frequency  
caused by random trapping and detrapping of charge carriers in devices like transistors  

Drifts can be induced by: temperature variations, seismic or acoustic noise, mechanical vibrations  
Such effects contribute to flicker noise  
This form of noise cannot be cured by temporal integration, which, on the contrary, enhances its relevance.


rumore flicker / rumore 1/f: più T è grande, più 1/f è grande, e maggiore è il rumore  
quindi rumore flicker è dovuto al drift  



**Ground loop / earth loop**

it's a mistake when building circuits  
occurs when two points of a circuit are intended to have the same ground reference potential, but instead have a different potential between them

cause:  
enough current is flowing in the connection between the two points to produce a voltage drop between them  
current may be produced in a ground loop by electromagnetic induction  

consequence:  
it causes noise, hum, and interference in audio, video, and computer systems  
to avoid it, ensure that all vulnerable signal circuits are referenced to one point as ground  

#### Remove noise

**Temporal integration**

the goal is to decrease N without affecting S too much   
generally do so with temporal integrating (low-pass filter)       

temporal integration works fine when:  
1- the signal is slowly varying  
2- noise and spurious signals vary on much shorter timescales  

else, if S changes fast too, it's also affected by integration   
so must reach a compromise, able to reduce N while negligibly affecting signal amplitude   

**Why integration: time vs frequency domain**

frequency-domain: low-pass filter  
perchè il rumore ha frequenza molto più alta della banda utile del segnale, quindi viene bloccato  

time-domain: integration  
all spurious signals with a fluctuating nature tend to zero upon time integration   
bc positive and negative contributions in the integration interval cancel with each other  
don't need the time to be an integer multiple of the period, since the not-cancelled part is small in percentage

**PSD: phase sensitive detection / Synchronous detection**

un metodo usato per estrarre un segnale molto debole immerso nel rumore  
sfruttando il fatto che il segnale utile è modulato in ampiezza da una funzione periodica nota, mentre il rumore è in genere casuale   

rilevare un segnale modulato usando come riferimento esattamente la stessa funzione periodica che ha modulato il segnale in partenza, in modo da estrarre solo la parte in fase con la modulazione ed eliminare tutto il resto (rumore)

**Homodyne**
