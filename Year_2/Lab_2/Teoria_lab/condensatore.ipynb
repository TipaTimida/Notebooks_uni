### How to build them

**How it's built**

a plane parallel plate capacitor is obtaine by:  
rolling up a sandwich made of alternate conducting (typically, aluminum foils) and dielectric layers  

tolerance:  
fabrication of so-conceived components is inherently not highly controllable, in particular in terms of shape and size   
so often have relatively large tolerances  

materials used:  
dielectrics: sintered ceramic oxides or plastic foils  
including a variety of polymers (e.g., polyethylene, polyester, polypropylene) and even impregnated paper sheets.  
ceramic: an inorganic material with polycrystalline microstructure and an insulating behavior  
sintering: one of the methods to produce ceramics

**Applications**

manipulation of time-dependent signals   
sensor for small displacement (make a plane parallel capacitor whose plates can be displaced)    
sensor for low pressures in gas (use the gas or vapor whose pressure must be measured as the dielectrics)   
position sensitive detectors (like in the ubiquitous touch screens)  

#### Types

**Variable capacitors**

now substituted by solid-state components  
bc var capacitors typically operated using air as the dielectrics, so has too much sensitivity on the operating conditions (temperature, humidity, pressure)  
also has poor mechanical reliability, a small capacitance value, and a relatively large size  

**Electrolytic capacitor**

parts:  
electrolyte: a liquid where ambipolar electrical transport can occur (both negative and positive carriers/ions move)    
electrodes: conductive metal plates  

how it works:  
1- apply a potential difference between two electrodes (anode and cathode)  
2- an electrochemical process leads to ion formation (like in  electrolysis of water)    

anodization / electrochemical etching: if anode is made of alluminium, electrolysis makes its surface rougher, and covered in a layer of aluminium oxide  
rougher surface increases area, and a thin later of ozide means small inter-plate dustance, so very good capacity   

**Problems with electrolytic capacitor**

1- is a polarized component (ha un verso, non posso invertire anodo e catodo)  
so can't be used with alternating signals  
2- has big tolerance (its C varies a lot)  
3- big energy loss, must put a resistor in parallel or risk it exploding  

so mostly used in power supplies, to obtain a quasi-continuous voltage starting from some rectified
waveform  
here, + and - nodes can be picked by design

#### Applications

**Applications**

integrators: reduce sensitivity on fast variations  
why: è come se dà la media di tutto, segue bene la parte in DC  
improve the signal-to-noise (S/N ) ratio  
put a capacitor in parallel with the signal  

differentiators: enhance variations/changes   
why: segue bene i cambiamenti  
when voltage change must be monitored on short time scales  
put a capacitor in series with the signal  
example: coupling capacitor, takes away the DC part, only keeps the variable one  

**Applications: coupling/decoupling capacitor**

coupling capacitor:    
essentially uses a differentiator    
put a capacitor in series with the signal  
differentiators work only for slow signals (frequencies below the cutoff)   
removes the constant (DC) offset and slightly attenuates very low frequencies, while high frequencies pass unchanged
