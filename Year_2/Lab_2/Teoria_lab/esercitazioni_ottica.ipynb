**Reticolo ottico / reticolo di diffrazione**

poichè $ \theta \sim 0$, allora $\sin \theta \sim \tan \theta = \frac xD$  
D: distanza orizzontale col muro  
x: larghezza orizzontale del punto dal centro foglio

descrivere il reticolo:  
g=linee/millimetro (tra 80 e 600 nel lab) (fenditure nel reticolo per millimetro)  
d=1/g (passo reticolare)  

formula:  
$\Delta x = D \frac \lambda d$  
a scuola, D lo chiamavo lab (paralleli al muro = perpendicolari al muro)  

cosa vedo: tanti pallini (massimi di interferenza) equispaziati, di stessa intensità  
misura $\Delta x$ tra di loro  

di solito faccio un grafico $\sin \theta$ berso $I$
