##### Biology

**How the heartbeat works**

The heart is a bio-electrical pump  

1. Each beat is triggered by a tiny electrical pulse generated in the sinoatrial (SA) node   
triggered about once per second: 1 - 1.67 Hz, normal  heartbeat frequency (60-100 beats/min for adults)  

2. That pulse propagates through the heart tissue as an action potential, a voltage spike lasting about 20–100 milliseconds in each muscle cell  
causing coordinated contractions  

**Exciting a cell**

Each cardiomyocyte (heart muscle cell):  
1. resting polarized state: stays at about −90 mV  
2. depolarization: jumps to +20 mV in ~1 ms    
3. repolarization: takes 50–100 ms (refractory period)  

That’s what an ECG records: the electrical changes from millions of cells depolarizing and repolarizing  

**What is polarization for**

In neurons: depolarization → action potential → electrical signal propagation  

In heart muscle, depolarization triggers contraction (via calcium entry)   
Each heartbeat starts with a wave of depolarization spreading across the atria and ventricles.  

**Recap**

Polarized: there is a voltage difference between the two sides (like a charged capacitor)    
This polarization is stored energy — the cell uses it to transmit signals (neurons) or trigger contraction (heart)  

depolarization: reduction (or reversal) of the resting voltage difference across the cell membrane (flipping the polarity)      

**Cell polarization (resting polarization)**

cell membrane: an electrical insulator (lipid bilayer) separating conductive fluids (the cytoplasm and extracellular fluid)  
the ions are concentrated differently inside and outside, bc of selective permeability:  
- inside: potassium $K^+$  
- outside: sodium $Na^+$ and chloride $Cl^-$  

that's literally a capacitor  
So each cell can hold a voltage across its membrane, the resting membrane potential: $V_{rest}=-70mV$ (the inside is negative)  

ion channels:  proteins that let specific charged particles move in and out  

**Depolarization**

if cross the depolarization threshold (≈ −55 mV):  
- triggers opening of ion channels (for example, sodium channels in nerve or heart cells)  
- Na⁺ ions rush in because:  
   - they are more concentrated outside (repelled by positive charge)    
   - they are attracted by the negative interior  
- This inflow of positive charge makes the inside less negative  
the voltage moves toward zero, and sometimes even becomes slightly positive  

**Repolarization**

After depolarization, other channels open:  
- Potassium (K⁺) flows out,  
- Sodium channels close,  
- The sodium–potassium pump restores concentrations.  
- The inside becomes negative again   

Then a short period follows where the cell cannot be excited again, refractory period   

#### ECG

Depolarization: (P and QRS waves)  
→ Cells lose excitability (absolute refractory period).   

Repolarization: (T wave)   
→ Membrane potential goes back toward resting values.   
→ Excitability gradually returns, especially toward the end of the T wave (the relative refractory period).  

After the T wave (flat TP segment):  
→ All cells are fully repolarized, back at resting potential, and completely excitable again — they are simply waiting for the next pacemaker impulse from the SA node.

##### Effects of current

**Danger: DC vs AC**

DC:  
- a single continuous flow of charge in one direction.  
- can let go: When first touched, it can cause a brief muscular jerk, but muscles can then adapt    
- causes: burns and deep tissue heating (if large)   

AC:   
- The direction of current reverses 50–60 times per second.  
- This rapidly changing field repeatedly stimulates nerves and muscles, preventing them from relaxing.  
- causes: continuous muscle contraction and ventricular fibrillation (if at 50-60 Hz) or burns (if higher frequency)    

Roughly, AC becomes as dangerous as DC at half the voltage  

**Normal heartbeat vs excitability**

The signal that makes your heart beat (from SA) is a low-frequency waveform, around 1 Hz    
but the individual nerve and muscle excitations that make up that wave happen on the order of tens of milliseconds, 20–100 Hz   
So the cardiomyocytes are electrically excitable in a frequency range around 30–100 Hz  

recap:  
normal heartbeat: 1 Hz  
cell-level excitability (biophysical reaction speed, electrical responsiveness): 20-100 Hz (10-100 ms)

think about the ECG:  
the cell returns excitable after the repolarization   
the repolarization is the T wave, the small upward hill after the big spike (this is the refractory period, so excitability is regained at the end)  
this is much earlier than when the next heartbeat happen 
see that there is a long flat horizotanl line before the next P wave, the cell is just waiting    

**AC and ventricular fibrillation**

applying AC:  
means applying an external field  
it creates an electric potential gradient across the chest  
what happens depends on frequency:   

if 1 Hz:  
the cells have time to recover (so it's not fibrillating, it's properly pumping, just not at the moments called by the SA node)  
it just causes shocks or muscle contractions. 

if 50-60 Hz:  
The polarity reverses every ~10–20 ms — just within the excitability window of the cells    
Different areas of the heart are depolarized and repolarized out of sync, because not all cells recover at the same time, some may be still in the refractory period and some not    
creating a pattern of asynchronous stimulation, called ventricular fibrillation  

above 1 kHz:   
field oscillates too fast, the membrane potential barely moves before the polarity reverses  
so the ions inside and outside don’t have time to rearrange  
causes surface oscillations and heating on, not real depolarization
the current stays mostly on the skin and doesn’t deeply stimulate the heart  

recap:  
50–60 Hz happens to coincide with the “response speed” of heart tissue, making it the most dangerous frequency range for electric shock   
which is an unfortunate coincidence, because that’s also the frequency chosen for our power grids for technical reasons  

**Not all cells are depolarized at the same time**

because that's how the conduction pathway works  
atria contract first, ventricles later  
So even normally, depolarization spreads gradually, not instantaneously  
so maybe at the first AC shock, the atria have finished, but the ventricules not  
at the second, the ventricules have, but the atria not

look at the ECG:  
atria: depolarize at P wave, regain excitability during/after QRS complex     
vetricles: depolarize at QRS complex, regain excitability at end of T wave  
so atria is excitable before the ventricles  
so if there is a signal in the moment, the atria follows it, the ventricles ignore it  

**After letting go**

When you let go, the external current stops  
but if that current has already triggered fibrillation, the heart does not automatically restart a normal rhythm  
bc the the myocardium is full of small, random, self-sustaining depolarization loops, so the pacemaker in the SA node can't take control anymore  

self-sustained depolarization loop:  
The wave can re-enter an area that has just repolarized — instead of dying out, it keeps circling.  

**Explained easily**

contraction of heart cells is needed to push blood into flow  
like spremere un tubetto di dentifricio  
but after contracting, must wait a bit for new blood to enter the atrium in the heart  
if it contracts continuosly, it's not pushing blood, so such heartbeat is useless  

so, in ventricular fibrillation, the muscle fibers:  
problem 1: contract too fast  
problem 2: contract out of sink  
There’s no coordinated squeeze → no effective pumping → blood flow stops → cardiac arrest.

**Defibrillator**

what it does:  
it stops the chaotic electrical activity so that the heart’s natural pacemaker (SA node) can resume control.  

How it works:  
During fibrillation, many random loops of current circulate in the ventricles  
The shock from a defibrillator delivers a large, brief current that depolarizes every cell at once, resetting them all to the same resting potential  
If the heart’s conduction system is still intact, it will spontaneously restart in a normal rhythm within a few seconds.

and a stopped heart?  
If the heart is in asystole (flatline):  
The cells are no longer excitable — their ion gradients have collapsed.  
Applying a shock won’t do anything, because there’s nothing to reset, it’s already electrically silent.  
so on flatline patients (ECG is a flatline, there's no electrical activity) using defibrillation is useless, need:  
- CPR (to maintain blood flow and oxygen)   
- drugs like epinephrine (to chemically stimulate the heart’s electrical system).

**How defibrillator works**

the membrane has a resting potential  
so the membrane stores a small electric field across it  

When a defibrillator pulse or any electric shock passes through the body:  
1. it sets up an electric field across tissues    
2. each cell, being a small conductive sphere with an insulating shell, experiences an induced voltage across its membrane due to this external field  
3. That’s enough to shift the membrane potential by tens of millivolts, comparable to what happens naturally during depolarization  
4. if cross the depolarization threshold (≈ −55 mV), triggers opening of ion channels and thus depolarizes the cells   

**Types of defibrillation**

depends on what defibrillates (don't pump effectively): atrial or ventricular defibrillation (AF, VF)   

- AF  
the ventricles still contract, although irregularly, so blood keeps moving: reduced but not null output  
not fatal but long-term risks, person stays conscious    

- VF  
no coordinated contraction, so no blood flow at all: no cardiac output  
pass out, fatal within minutes  

**Explained types of defibrillation**

P wave: atria contraction  
QRS complex: ventricular contraction  

the P wave only push a little extra blood, like 10% (called 'atrial kick')  
most of the job is done in the QRS complex (in the ECG, it's the biggest)  

in AF: no atrial kick, blood just passively flows from the atria into the ventricles (thanks to pressure differences)  
in VF: the ventricles quiver, so no blood is ejected at all
