### Photodiodes (rivelatore)

si chiama rivelatore, non rilevatore  

**What it is**

transducers for the light power  
a device able to produce an electric signal proportional (or at least dependent on) the power PL of the light impinging onto its active area  

**Formula**

$I_{ph}=k_{ph} P_L$  

I: current produced by the diode  
$P_L=\frac{I_{em}}A$ : incident power  
k: calibration coeff, depends on $\lambda$  

**How it works**

photoelectric effect:  
when hit by light, happens electron-hole pair generation  
bc photon gives energy to the pair, so breaks their bond  
since in the depletion region there is $E_{built-in}$, if the pair is created in that region, electron and hole are pulled to the correct sides, thus generating a current  

so, only works if light hits pairs in the depletion region  

**How to strenghten it**

1- use reverse bias  
widens the depletion region  
strengthens the electric field  

2- add an intrinsic (non doped) region between N-doped and P-doped  
so the depletion region is bigger  

### Laser

**Properties**

its radiation is (ideally):  
1- monochromatic   
bc photons all have same eneergy, and $E=hf$  
2- collimated  
bc all photons have same momentum    
3- polarized  
bc all photons have the same angular momentum  

bc photons thrown by the laser are all equal to each other

**Speckles**

what it is:  
shine the coherent light (laser) on a rough surface    
vedo pattern granulosi tipo “puntini”   

perchè:  
se uso onda coerente su superficie ruvida, la luce viene riflessa da tanti punti diversi  
ogni punto introduce una fase diversa (perché i cammini sono diversi)  
le onde si sommano (nella retina dell'osservatore) → interferenza casuale  
quindi: interferenza di tantissime onde con fase casuale ma coerente  

nota:  
nelle onde non coerenti non accade  
perchè le fasi cambiano continuamente, le interferenze si “mediano”
