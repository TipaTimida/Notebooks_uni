### Segnali

#### Forme d'onda

**Forma d'onda**

forma d'onda, waveform, segnale: la funzione $V(t)$    
NON è un'onda, quelle sono funzioni di TEMPO E SPAZIO, cioè hanno anche propagazione nello spazio, i segnali no  

segnale monocromatico: ha una sola frequenza  
segnale periodico: c'è un periodo  
segnale alternato: la sua media temporale su un periodo è 0, $<g(t)>_T=0$  

**Attenzione ai nomi**

sinusoidale: analiticamente, si scrive con una sinusoide  
alternata: media su un periodo è 0  
AC: varia nel tempo (in qualunque modo va bene)  
DC: non varia nel tempo  

esempio:      
una sinusoide con un offset è AC e periodica, ma NON alternata  
se l'offset è ababstanza grande, potrebbe addirittura non invertire mai la polarità (cioè non toccare mai valori negativi di V)  

**Waveform types**

dipende da in che modo si scrive $g(t)$ analaticamente, o dal plot V-t  

- sinusoidale: $g(t)= g_0 \cos(\omega t + \phi)$  
$g_0$: amplitude, ampiezza ($g_{pp}=2g_0$: peak-to-peak amplitude)      
$\phi$: constant phase term, termine di fase costante (spesso setto $\phi=0$, tanto dice solo da dove inizia l'onda)  

- quadrata  
non ha funzione analitica semplice  
se è asimmetrica: pulse train  

- triangolare  
non ha funzione analitica semplice  
se è asimmetrica: sawtooth, dente di sega    

asimmetrica: i low e high non durano uguale, cioè cambio il duty cycle  
duty cycle:  percentuale di tempo in cui è su HIGH (se simmetrico, è 50%)    

#### RMS

**Media temporale**

$⟨g(t)⟩_T = \frac{1}{T}\int_{t}^{t+T} g(t) \, dt$  

gli estremi di integrazione possono essere qualunque cosa spanni un periodo, tipo[0, T] o [-T/2,+T/2]  

**Rms amplitude, valore efficace**

nomi:  
rms = root mean square   
o anche 'corrente efficace'  

definizione:  
1- media di $g^2(t)$ su un periodo: $g_{rms}=\sqrt{⟨g^2(t)⟩_T}$  
ricorda di elevare al quadrato, sennò ottengo 0 se è alternata  
2- RMS è la corrente continua equivalente che dissipa la stessa potenza

esempio:  
$I_{eff}^2 = \frac{1}{T} \int_{t}^{t+T}I^2(t) \, dt$  


**Tipi comuni**

tipo | formula | perchè
---|---|---
sinusoidal |  $g_{rms}=\frac{g_0}{\sqrt{2}}$  | valor medio di $\sin^2$ sun un periodo è 1/2  
square |  $g_{rms}=g_0$ | perchè faccio di $g^2$, quindi anche la parte a -g sembra sia a +g
triangular | $g_{rms}=\frac{g_0}{\sqrt{3}}$  

### Domains

**How to represent the signal**

TIME domain:

- plot: amplitude vs time

FREQUENCY domain:  
- signal: decomposed in sum of sinusoids (broken into its frequency components with Fourier Transform)

### Phasor

**NOTA**

i fasori implicano che siamo in regime sinusoidale  
cioè tutte le grandezze scalano con $e^{i\omega t}$  
anche se poi, con Fourier, grazie alle sinusoidi posso trovare tutte le altre   

**How to write**

segnale reale:  
$g(t)=g_0 \cos (\omega t+ \varphi)$  

fasore:  
$g_\omega(t)=g_0 [\cos (\omega t+ \varphi)+j\sin (\omega t + \varphi)]=g_0e^{j\omega t}e^{j \varphi}$  

to tell them apart, the phasor has a pedice $\omega$, or a tilde over it


**Phasor: definition**

a mathematical description of the real signal, add an imaginary part  
estensione analitica di un segnale reale $g(t)$, o 'METODO SIMBOLICO'   

why:  
in frequency domain, for each frequency component $\omega$, the quantity V (or I or Q) can be represented as a complex number   
complex number = vector in the complex plane  
essentially a way to draw the frequency spectra in a carthesian plane, differently from a normal histogram    

**Properties**

for each frequency component:   
magnitude: amplitude of that component  
argument/angle: dephasing 

#### Other

**Does a phasor rotate?**  

a wave with more frequency component is made up of more phasors  
in the drawing, all the phasors are still, they're just a way to visualize the spectra of the wave  

then, in reality, obviously each component would be rotating with its own $\omega$  
so if i would make an animation, can see at each time what is their current dephasing  
usually i draw at t=0, so the dephasing is  the relative dephasing between components at the start, not the phase that each component has at each moment (for that, just make it rotate with $\omega$)  

**Multiply phasors**

rem the formulas:  
$⟨AB⟩ = \frac 12 Re[A_0B_0^*]$  <br> $⟨A^2⟩=\frac 12 A_0^2$  
$<\cos^2(\omega t)> = <\sin^2(\omega t)> = \frac 12 $   
$<\cos(\omega t)\sin(\omega t)> = <e^{-i \omega t}>=0$   

example: power  
in DC, P=IV, both I and V are both phasors, so:  
$<P>_T=\frac{1}{2}Re\{V_\omega^* \, I_\omega ^*\}=V_{rms}I_{rms}\cos (\Delta \Phi) = \frac{V_0 I_0}{2} \cos(\Delta \Phi)$  
cioè 1/2 è lì per trasformare entrambi i fasori in RMS  


**Note**

the complex conjugate is to take the PHASE DIFFERENCE, else would take the same sum  
bc multiplying phasors means adding angles, and taking the conjugate means chaning the sign of the angle, so:  

$\phi_{AB^*} = \phi_A +\phi_{B^*} = \phi_A + (-\phi_B)$  



**Operazioni coi fasori** 

Derivare: moltiplicare per jω   
Integrare: dividere per jω   

perchè:  
$V = e^{j\omega t}$  
$\int V \; dt= \frac{e^{j \omega T}}{j\omega}$  
$\frac{dV}{dt} = j\omega e^{j\omega t}$  

### Symbolic method

**What is it**

method to model systems, using Z and H  

**Operate on phasors**

in a circuit, each component causes a:  
1- dephasing (phase shift) between V and I  
2- gain  
which are different for each frequency component  

can use 2 operators that act on phasors to describe it:  

1. $Z(\omega)$: impedance  
describes action of one component (object): $V = Z(\omega) I$     

2. $H(\omega)$: transfer function    
describe action of whole circuit: $V_{out}=H(\omega)V_{in}$        

**Properties**

for both:    
- magnitude: gain (attenuamento)  
$|H(\omega)| = G(\omega)$  

- argument: dephasing (phase shift)  
$arg(H)= \phi(\omega) =\arctan(\frac{Im(H)}{Re(H)})$  

the magnitude is called gain only for H, for Z it's a bit unproper  

**Apply them**


how to apply:  
same way, changes change what things i'm comparing   

transfer function:   
$g_{in} = a \cos(\omega t )$  
$g_{out}= H(\omega) g_{in} =  G(\omega) \, a \cos (\omega t + \phi (\omega))$   

impedance:  
$I(t)=I_0 \cos (\omega t)$  
$V(t)=G(\omega) I_0 \cos(\omega t + \phi(\omega))$  


#### Theory

**What are those operators**

what they are: matrices, which act on phasors     
but just in complicated systems  
in easy ones, it's simply a complex number (written in exponential form)  

for each $\omega$, $H(\omega)$ is a vector in the complex plane???  
so matrix or vector???  

what they do: rotate and scale each frequency component of the spectra of the input wave (wach phasor) (each frequenyc is modified in a different way), to obtain the spectra of the output wave  

in both:  
magnitude: gain  
angle: dephasing  


**IMPORTANT difference**

but they create 2 different types of dephasing and gain: 

Z: describes one component, so relationship between V and I  
component phase shift: shift between V and I (across a component)  
impedance gain: power/flow conversion factor between voltage and current

H: describes the whole signal, so relationship between input and output V

### Impedance

**What is it**

$Z=R+jX$   

$Z$: impedenza    
a complex quantity that can act on phasors  
substitute the resistance in AC  

$R=Re(Z)$: resistenza   
opposizione al passaggio di corrente (DC e AC)  
dissipazione

$X=Im(Z)$: reattanza   
opposizione che dipende dalla frequenza, dovuta a condensatori e induttori  
determina sfasamento tra corrente e tensione introdotto dal componente   
non dissipa energia, la immagazzina e poi rilascia, questo causa ritardi, cioè lo sfasamento  
if it's 0, the argument is 0, so no depashing (resistor has no dephasing)  

**Recap impedance of components**

- resistore: $Z_R=R$  
no dephasing 

- condensatore: $Z_C=\frac{1}{j \omega C}= -\frac{j}{\omega C}$   
Z punta verso il basso (verso -j), quindi rotazione di -90 gradi  
poichè $V=ZI$, allora V è indietro di 90 gradi rispetto a I, cioè è in ritardo  
this dephasing makes sense, bc in un condensatore V is at its max when I is at its min  

- induttore: $Z_I=+j \omega L$  
qui invece la corrente è in ritardo rispetto a V  

**Note**

definizione operativa: $V_{\omega} = Z I_\omega$  
è simile alla legge di Ohm, quindi continua a far  valere le leggi di Kirchoff  

e serie e parallelo di impedenze è come per le resistenze  
impedenza si misura in Ohm, in tuti i casi
