**Misurare la frequenza: frequenziometro**

it's easy, bc measuring frequencies entails measuring time intervals   
which is is a naturally digital operation, since it can be reduced to counting periods produced by a clock    

therefore:  
statistical uncertainty can in principle be
as small as a single period   
if the clock is sufficiently fast, the relative contribution of statistical errors can in principle turn out negligible in many practical applications  

rephrase with noise:  
naturally digital measurement enables getting rid
of “analog fluctuations” (or noise), which are often a relevant source of uncertainty  

how to do it:  
1- the clock produces a pulse train  
2- counting entails discriminating the passage from low to high level in the pulse train  
set a certain threshold and identify as high and low levels those over or under  
Analog noise, which would manifest itself as a fluctuation of high and low levels, does not produce effects in the discrimination, unless its amplitude is comparable to the threshold (quite an uncommon condition)  

gate time: time needed to accomplish the frequency measurement  
auto-oscillator/time-base: a part of the clock (generally a piece of quartz)  

### Oscilloscope

#### Come funziona

**What is it**

made of 2 parts:  
digitizer: continuously sampling the voltage   
computer: manipulate the digitized voltage and show it
as a function of time on a display  

there is latency, don't really work in real time 


**How it works**

1. a cathode ray tube (CRT) emits a beam of electrons  
perchè è un filamento di materiale ultrafondente, se attraversato dalla corrente rilasciata elettroni per effetto termoionico  

2. $e^-$ are extremely light, so they can be deflected (moved) very quickly by electric fields  
so are deflected at wish in a controlled manner, depending on potential differences applied to the pairs of deflection plates  

3. $e^-$ hit a phosporescent screen (phospor layer)    

4. phsoporence: the phospor layer emits photons, visible light, from where it got hit  

5. phosphor continues to glow for a short time even after the beam moves away, then disappears  


**Sweep**

sweep: horizontal motion of the electron beam and consequent formation of a luminous trace on the screen   
so imagine a crayon moving horizontally across the screen  

how to do it:  
the horizontal deflection is controlled by a pair of deflection plates, to which i apply potential difference linearly proportional to time (with a ramp generator)  
when the screen is finished, the beam is brought back to the starting position, so the waveform is actually a sawtooth  
also wait a bit, to be sure that the phosporence is gone  

note that, when goin back to the start (restarting the sweep), there is a 'dead time' (time where it's not drawing)  


**Explanation: movement of the electrons**

i movimenti sono controllati da due condensatori  
cioè che importa è a che waveform è collegato il condensatore  
codensatore con piastre orizzontali: ha V verticale, controlla movimento verticale  


movimento verticale: $V_{in}$  
ogni piastra è collegata ad una probe, così $\Delta V$ è quella misurata dal circuito, così l'elettrone si spsota su e guù seguendolo  

movimento orizzontale: $V$ è un dente di sega  
durante la salita, $x \sim V \sim t $, si muove a velocità costante  
durante la parte verticale, V si spegne un attimo, l'elettrone torna bruscamente a sinistra 

#### Phosporescence

**Why phosporescence**

phosporescence's lifetime is usually a few tens or hundreds of milliseconds (how long it keeps glowing)  
its kinda matched with the persistence time of images in human retina, so even a very fast trace can be seen as a stable line  
(actually it's refreshing really fast the scrren, like a videogame a scorrimento laterale)  

**Persistance of vision in human eyes**

When i see something bright, the light-sensitive cells in the retina (rods and cones) keep sending signals to your brain for a short time, even after the light is gone  
That short time is called persistence of vision (or retinal persistence), around 1/10 of a second  
So if something flashes faster than that i can’t distinguish the individual flashes, you see a continuous image  
The principle is adopted in all two-dimensional displays, including the flat displays found in any computer or smartphone  

**Brightness of the light**

depends on:  

1. kinetical energy of electrons  
this can be modified with a knob on the oscilloscope  

2. dose of impinging particles, intended (number of incident electrons per unit surface)  
it's the reason why in a square waveform the vertical segment are barely visible, since the transition between high and low is really fast, so not many electrons hit in that time

#### Funzionalità

**Zeroing**

before using it:  
must manually set the x=0 line in the picture (rest position) using knobs  

why:  
can't always keep it in the middle, bc of spurious magnetic fields that lead to unwanted deflections, due to cyclotron motion  
Spurious magnetic
fields, either static or time-dependent, are ubiquitous (like the terrestrial field)

**Ground connection**

if want the absolute value of V, must connect the oscilloscope to GND, so he know what it is  
if want just to read the amplitude, don't need to do it  

**Uncertainty**

non-statistical uncertainty always prevails  

**Channels**

can do multichannel operation (have 2 channels at once) in 2 ways:  
1. ALT: alternating sweeps (each sweep is one signal)  
2. CHOP: chopping the sweep (alternate signals continuosly, so have a broken sequence of segments)    

automatically selects the best mode according to the TIME/DIV setting  

#### Coupling

**Attenzione a cosa dico**

il bottone si chiama AC/DC  
NON leggerlo corrente alternta/corrente continua, è sbagliato  
AC e DC sono nomi dati ai tipi di coupling  

**Coupling**

$V(t)=V_{DC}+v_{AC}(t)$  
$V_{DC}$ is a constant offset, superposed on a time-dependent (typically small) signal $v(t)$  
 
problematic to see on oscilloscope:  
bc must show the x=0 line, so there is a big empty space which fills the screen  

coupling: how the oscilloscope input circuit is connected (or coupled) to the signal source
- DC coupling  
The input passes everything, both DC and AC parts of the signal  
When you want to see the true voltage, including offsets  

- AC coupling  
A capacitor is inserted in series with the input, which blocks DC but passes AC -> constant component of the signal is removed  
When the DC part is large but you only care about the oscillations  

**Coupling vs decoupling**  

decoupling or bypass capacitor: allows DC to pass, while blocking AC  
A decoupling or bypass capacitor is placed in parallel with the source and the load  

coupling capacitor: allows AC to pass, while blocking DC  
a coupling capacitor is placed in series with the load

fare il coupling vuol dire fare la derivata rispetto al tempo  
quindi se ottengo 0 vuol dire che è indipendente dal tempo, non che vale 0  

#### Trigger

**Trigger**

2 types of operation:  

1. automatic trigger mode (asynchornous operation)  
starts the sweep automatically at regular intervals, no matter what the signal does (quindi segnal e sweep non sono sincronizzati)    
Simple but can miss transients or show unstable waveforms  

2. normal trigger mode (Synchronous operation)  
Starts a sweep when the signal crosses a chosen voltage level (quindi segnale e sweep sono sincronizzati)  
Stable display, synchronized with signal, used for accurate measurement  

stuff that works bad with automatic trigger mode:  
1- transient signals = events that happen only once, and last really little    
eg. discharge of a capacitor   
if the oscilloscope sweeps automatically, it might refresh in the middle of it, or catch it in the dead time, and miss it  
2- periodic signals  
if starts sweeping at random times, then each sweep will start at a different phase of the wave  
the result on the screen looks unstable, like the waveform is drifting or moving sideways continuously  

**Trigger options**

- slope: + or -  
whether to trigger when the signal goes up (+) through the threshold or down (-)  

- source:  
which signal's amplirude i must monitor  
pick between an input channel (CH1, CH2), ax external source (EXT), the electric power line (LINE), switch between the two input channels when doing multichannel operation (VERT)    

- level  
the exact voltage value where the sweep starts  
in an analog oscilloscope it cannot be known a priori, since the knob is not calibrated, so there is uncertainty in setting it  

- filters  
able to increase/decrease sensitivity on a
specific frequency range  
can be called 'trigger coupling' (which is different from 'input coupling')  
possibilities: AC, LFR, HFR (all normal, low/high frequency rejection)
