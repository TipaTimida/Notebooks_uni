### Anisotropia

**Anisotropia**
  
proprietà di un materiale per cui le sue caratteristiche (meccaniche, ottiche, elettriche) variano a seconda della direzione di osservazione o misurazione   
dicroismo e birifrangenza sono tipi di anisotropia ottica  

**Tipi di anisotropia**

cosa | dicroismo | birifrangenza 
---|---|---
quale è la proprieà anisotropica | coefficiente di assorbimento | indice di rifrazione (quindi velocità di fase) (cioè $n = \sqrt{\epsilon_r}$)   
efficienza | bassa, abbassa intensità | alta, non assorbe luce, quindi non abbassa intensità
nome | an artificial system that exhibits dichroism is called 'polaroid' | se è intenzionale, 'retarding waveplate' 

#### Polaroid

**Modellizza**

$E_{0, out} \; \hat e_{out} = E_{0,in} (\hat p \cdot \hat e_{in}) \; \hat p$  

notes:  
1- wave is described by amplitude and polarization $E= E_0  \, \hat e$  
2- often call $\theta$ the angle between $\hat e_{in}$ and $\hat p$

##### Struttura

**Cosa è p**

1- structure  
have a matrix, with embedded some rods  
when hit by light/radiation, they start obscillating, behave like obscillating dipoles  
so they absors more the radiation coming ortogonal to the axis, not at all the radiation parallel to the axis  

2- what is p  
is the direction of the axis of the dipole  
cioè la direzione lungo cui l'assorbimento è nullo, lascia passare quella componente del campo    

**Teoria vs realtà**

theory: radiation is absorbed (so the wave amplitude attenuated) only along one direction   

reality: it is weakly absorbed even in the other dicretions, but much less  
also there is aborption, so some portion of the impinging wave intensity is employed in dissipative mechanisms, not involving emission of light and in principle similar to material heating    

**Note: circular polarization**

if input wave has circular polarization  
transmitted intensity IS diminished, but how much it is diminished by is NOT dependent on the angle  
bc, geometrically, the projection of the circle is the same in every direction  

so, to see if a wave has circular polarization, check if intensity is independent of the polaroid angle

**Malus law**

intensity is the thing i observe in experiments  
$I_{out} \sim |E_{0, out}|^2 \sim I_{in} \cos^2 \theta$  

intensity is proprotional to the square of the amplitude    
the unspecified proportionality accounts for residual absorption by a non ideal polarizer  

why:  
$E=E_0 \cos \theta$  
$I \sim E^2 = I_0 \cos^2 \theta$  

**Diminuzione intensità**

$I_{out} = I_{in} \cos^2 \theta$  

per luce non polarrizata, $<I_{out}>=\frac 12 I_{in}$

**2 Polaroids**

$E_o \hat p_o \sim E_i (\hat p \cdot \hat e_i) (\hat p_2 \cdot \hat p_1) \hat p_2 = [ E_i \cos \theta_1 \cos(\theta_2-\theta_1) ] \hat p_2$  

i: input, o: output    
$\theta_1, \theta_2$: angles of $p_1$ and $p_2$ with $\hat e_i$  

case: cross-polarized configuration  
what: $\hat p_1 $ and $\hat p_2$ are orthogonal  
$E_0 =  E_i \cos \theta_1 \cos (\pi/2 -\theta_1) = E_i \cos \theta_1 \sin \theta_1 \sim E_i \sin (2 \theta_1)$  
$I_{out} \sim I_{in} \sin^2 (2 \theta_1)$   
so no intensity passes, bc take away both components x and y

#### Flat screens

**What is it**

display method after the cathode ray tubes  
use an LCD: liquid crystal display  

**LCD**

is similar to the matrix with rods in a polaroid, but:  
the matrix is fluid  
the rods can rotate, and are mafe of dielectric material (so have a static dipole moment)    

this config is called 'optical valve'  
the system is called 'nematic crystal'  

**How LCD works**

1- rods are randomyl distribited in the liquid matrix  

2- a static electric field is applied  
so the rods rotate and align to it (bc a rod is a dipole, so aligned means it minimizes energy)  
can put many electrodes on the screen, so each region receives a different electric field, thus rods align in a different direction  

3- rod alignment gives the liquid crystal an anisotropic structure  
the normally isotropic optical properties of the liquid crystal will acquire a dichroic character  

4- when light hits the screen:  
it gets absorbed and passed differently is different parts, so can control each part's color  

**Problem**

to control each part, need many electrodes, which is a problem for miniaturization  
solution: use 'twisted nematic crystals'

#### Birefringence

**Model**

$n_{slow}$, $n_{fast}$: indice di rifrazione nelle due direzioni  

ritardo tra le componenti dell'onda:  
$\Delta t_{ret} = \frac{d}{v_{slow}}-\frac {d}{v_{fast}} = \frac dc(n_{slow}-n_{fast})= \frac dc \Delta n$  

phase shift:  
$\Delta \Phi = 2 \pi \frac{\Delta t_{ret}}T = \frac \omega c d \Delta n = \frac{2 \pi} \lambda \Delta n \, d$  

**Retarding waveplate**

1- what is it:  
an object that uses birifrangence to give a specific phase shift to the components of the wave     
like $\pi/2$ o $\pi/4$  

2- uses  
circular polarization has $\pi/2$ dephasing between the two components  
so, if have input wave with $\pi/4$ shift and i put in a $\lambda/4$ waveplate, i get circular polarizarion

**Magic colors**

setup:  
white light -> polaroid 1 -> birifrangent amterial (tape) -> polaroid 2 -> observer  

steps:  
1- polaroid 1: linearly polarizes the light, it's still white    
2- tape: retards components, so get ellyptical polarization    
$\Delta \phi \sim \frac 1 \lambda$  
ogni colore acquisisce una fase diversa, qundi una polarizazione diversa  
3- polaroid 2: project along one direction, depending on its angle  
in that direction, depending on the polarizations obtained at the previous step, some colots have more intensity, others have less  
so, rotating the second polaroid, can change the direction i'm looking at, and so the color i see  

to change color, rotate polaroid 2 or change thickness of tape
