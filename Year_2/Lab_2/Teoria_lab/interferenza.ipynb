### Definizioni

**Principio di Huygens**

ogni punto di un fronte d'onda si comporta come una sorgente secondaria puntiforme di onde sferiche, in fase con la primaria  

note:  
1- si comporta da sorgente di onde sferiche, anche se l'originale era piana (basta immaginare un disegno di propagazione dell'onda da due fenditure)    
ma tanto l'intereferenza mi interessa su uno schermo molto lontano dalle slit, e a grande distanze un'onda sferica è localmente piana   
2- le sorgenti secondarie sono in fase  
quindi $I_1=I_2$, e $\hat e_1$, $\hat e_2$ sono paralleli  

**Fenomeni**

cause: field superposition   
sources come from Huygens principle, and depending on the type of sources, can have:  

What | Interference | Diffraction
---|---|---
sources | discrete radiation sources | sorgente continua
example | two slits, reticolo (each slit is a source) | single slit (each point in the slit is a source)  

![](https://cdn.britannica.com/04/5704-050-A28FFCBE/waves-wave-principle-plane-point-Huygens-BB.jpg)

**Practical tip**

problema:  
difficile l'interferenza negli esperimenti usando due sorgenti  
perchè è difficile sincronizzare due laser per sparare allo stesso istante, perchè serve un tempo di propagazione del segnale  

soluzioni:  
1- interferometri basati sul beam splitting (come esperimento di Michelson e Morley per trovare l'etere)    
2- interferometri di tipo waveform splitting (cioè reticolo, roba che sfrutta il principio di Huygens) -> migliore  

**Principio di reciprocità**

reticolo (muro con due split vuoti) è uguale al suo opposto (superficie trasparente con due slit opachi)  

### Generale

**Come studiarli**

la cosa migliore è scrivere $I(\theta)$  
perchè mi dice sia dove sono i max/min, che quale è la loro luminosità, se tutti uguali o no  



**Setup**

Quando | lettera | cosa
---|---|---
per INTERFERENZA |  $d$ <br> N| distanza tra fenditure (se reticolo, passo del reticolo) <br> numero di slits del reticolo (conto solo quelle illuminate dal fascio originale, non tutte quelle esistenti)  
per DIFFRAZIONE | $a$ | larghezza fenditura  
sempre | $D$ <br> $\lambda$ | distanza fenditura-schermo (lunghezza dell'asse ottico) <br> lunghezza d'onda , $k = \frac{2 \pi}\lambda$    
per osservare schermo |  $x$ <br> $\theta$ | distanza di un punto dall'asse ottico  <br>  angolo di un punto con l'asse ottico  


immagina che il disegno sopra sia visto dall'altro, cioè è tutto su un piano orizzontale  

**Setup: per reticolo**

$g= \frac 1 d$: slit linear density (lines/mm)  
$N=b/d$: number of illuminated slits  
b: transverse size of the beam  

**Modellizza superposizione**

lettera | cosa | note  
---|---|---
$\delta$ |  differenza tra gli OPTICAL PATH LENGTH delle due onde | NON è la distanza tra le sorgenti <br> diisegna i due raggi da ogni sorgente al punto che mi interessa, e calcola la differenza tra le loro lunghezze
$\Phi$ |  differenza di FASE INTRINSECA tra le onde (cioè tra le sorgenti)  | spesso 0, se le due sorgenti sono due punti del fronte d'onda, quindi emettono in fase  
$\Delta \phi=(k \delta+ \Delta \Phi)$ | differenza di FASE TOTALE tra le onde |  come il termine della fase in onda EM, l'esponente dell'esponenziale  
$\alpha = \frac{\Delta \phi}2$ | parametro angolare | in interferenza lo chiamo $\alpha$, in diffrazione $\beta$

**Modelliza superposizione: formule**

formule: quasi uguali, ma nella diffrazione scambio tutte le $d$ con $a$  
inteferenza: $\delta =d   \sin \theta$  
diffrazione: $\delta = a \sin \theta$


$\alpha = \frac{\Delta \phi}2 = \frac {kd \sin \theta}2=\pi \frac d \lambda \sin \theta$  

### Cosa succede

**Confronto**

What | cosa è | formula intensità | cosa vedo
---|---|---|---
interferenza (reticolo) |caso generale inteferenza | $I(\alpha) = I_0 \frac{\sin^2 (N\alpha)}{\sin^2 \alpha}$  | max principali (alti e stretti) + secondari (tipo ripple) 
interferenza (double slit) | reticolo con $N=2$ |  $I(\alpha) = 4I_0 \cos^2 \alpha$ |solo max principali
diffrazione (single slit) | single slit con lunghezza finita di $a$ |  $I (\alpha) =I_0 \frac{\sin^2 \alpha} {\alpha^2}$  | un max principale + ripple
interferenza + diffrazione | entrambi | reticolo: $I(\theta) = \frac{\sin^2 (N\alpha)}{\sin^2 \alpha} \; (\frac{\sin \beta}{\beta})^2$  <br> double slit: $I(\theta)=I_0 \cos^2 \alpha \; (\frac{\sin \beta}{\beta})^2$  | inviluppo

#### Modellizzare

**Interferenza: reticolo vs double slit**

What | reticolo | double slit 
---|---|---
struttura | qualunque N , caso generale| N=2, caso particolare
intensità max principali | $I_{max} = N^2 I_0$  <br> più luminoso, si vede anche segna spergnere la luce| $I_{max} = 4 I_0$  
larghezza frange | massimi assoluti molto stretti, $\sim 1/N$ | frange equispaziate, non dipende da nesusn parametro  

la formula dell'intensità del double slit la trovo pluggando N=2, e poi usando identità trig  

intuitivamente:  
nel reticolo ci sono più sorgenti, quindi più onde, se sono tutte in fase quella risultate è molto più luminosa  
e posso avere massimi locali, perchè alcune possono essere in fase e altre no  

#### Coord max e min

**Interference**

when | costructive | destructive
---|---|---
general case | $\Delta \phi = 2m \pi$ | $\Delta \phi =(2m+1)\pi$  
se $\Phi=0$ (dephasing intrinseco tra sorgenti) | $\delta = m\lambda$  | $\delta = (m+\frac 12) \lambda$  

m=0,1,2...  

**Coord dei massimi**

per piccoli angoli:  
$x \sim D \tan \theta \sim D \sin \theta$  

dalla tabella sopra, coord dei max sono:  
$\sin \theta = \frac{m \lambda}{d}$

quindi:   
max-max: $x_m = m \frac{D \lambda}{d}$  
max-min: $\Delta x \sim \frac 12 D \frac \lambda d$

questi sono per l'interferenza, per la diffrazione è uguale ma con a invece di d   

**Wuando ignoro diffrazione**  

la diffrazione c'è sempre, ma:  
$a \sim \lambda$: non trascurabile  
$a \gg \lambda$ trascurabile  

perchè trasuro se $a \gg \lambda$:   
allora $\Delta x $ è molto piccolo, quindi il puntino si allarga di pochissimo   
quindi è come se il raggio continuasse e dare immagine puntiforme  
se ogni raggio dà immaigne puntiforme, ottengo come zona illuminata esattamente la fenditura, la luce non si estende oltre  
cioè, interferenza è negligible  

**Nota**

sommare onde di freq diversa: serie di Fourier, posso ottenere qualunque segnale, anche onoda quadra o triangolare  

moltiplicare onde di freq diversa: inviluppo, ottengo una sinusoidale con fase modulata  
graficamente, l'onda di periodo più grande è l'inviluppo, cioè il bordo esterno, quella di periodo corto è la portante, viene schiacciata veticalmente per stare dentro l'inviluppo  

### CD and DVD

**Struttura**

disco con tracce a spirale  
lungo la traccia: pit (buchi) e land (piatto)  
un laser lo illumina, e a causa della diffrazione, il raggio riflesso è diverso a seconda se colpisce un pit o land  
c'è un fotosensore che legge il raggio riflesso, vedse è 0 o 1 (pit o land)  

quindi:  
informazione = variazione di riflettività  

| parametro                   | CD                   | DVD             |
| --------------------------- | -------------------- | --------------- |
| lunghezza d’onda laser      | ~780 nm (infrarosso) | ~650 nm (rosso) |
| pitch (distanza tra tracce) | 1.6 µm               | 0.74 µm         |
| dimensione bit              | ~0.83 µm             | più piccoli     |
| capacità                    | ~700 MB              | ~4.7 GB         |

cioè il DVD si legge con laser con $\lambda$ più piccola, quindi la dimensione dei pit può essere più piccola  
quindi posso metterci più informazione
