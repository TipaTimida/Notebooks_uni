### Eddy/Foucalt's current (correnti parassite)

**What is it**

variable B has variable flux  
flux produces a fem $\mathcal{E}$, bc of Lenz's law  
if circuitazione is computed on a conductor, the fem creates a current  

**Consequences**

1- parasitic dissipation  
2- shielding of the induction magnetic fields  

**1. Parasitic dissipation**

what:  
eddy current flow in a conductor, so a bit of energy is dissipated via Joule effect  
called parasitic when it's not intentional  

applications:  
inductive ovens, magnetic brakes, magnetic dampers  
these are explained in physics 2 notes (should put them together)  

avoid it:  
use wires with a small section, so the flux is smaller, so the fem is smaller, so the current is smaller  
or break the original plate in lots of small plates side by side (like the objects to insert in the inductors in the lab) -> called using a stack of layers, electrically insulated from each other  

**2. Shielding**

what:  
eddy currents produce a B'(t)  
bc of the - in Lenz's law, it's in the opposite direction as the one that created the eddy currents in the first place   
(this is simply magnetic induction)  
so in the conductive material, the total field B+B' will be smaller than the original one (bc B' is in the opposite direction as B)  
meaing B is shielded  

applications:   
coaxial cable    

**Coaxial cable**

in general, have a box made of a conductive material:   

1- have B(t) outside the box  
inside, get B=0, it's perfectly shielded   

2- have B(t) inside the box  
outside, not produce any effect  

it's like inside and outside are magnetically separated  
similar to Faraday's cage, but taht was for shielding electrical fields (how was it??? but that didn't need time dependant fields, unlike here)

### Coaxial cable

**Problem: coupling**

if there is an alternated current, there is field B(t)  
eg. the 50 Hz signal from the electrical power network, has a STRAY field B(t)  
stray fields can easily couple to wires, and create fem (bc of Lentz law)  
so, often collect spurius signals from somewhere   

**Nomenclature**

1- shielded caple:   
a system of two (or more) conductors, typically arranged in a coaxial
configuration  

2- coaxial cable:  
system of two conductors arranged in a coaxial configuration AND
featuring a specific behavior in terms of signal transmission, owing to their specific construction (materials and geometries)  

so the opposite of the intuitive one  

#### Why it works

**Situation**

have a coaxial wire:  
a cilindrical capacitor, with a generatore $V(t)$ between the the plates, and a load $Z_L$ to close the circuit  
plates have radius b>a, are tall l

could have a material between the two plates, so $\epsilon= \epsilon_0 \epsilon_r$  

**Compute E and B inside the cable**

1- E  
direction: radial (bc perpendicular to the surface inside plate, which is a conductor)  
compute: $\int D \, dS = Q_{free}$  
$\epsilon E \; (2 \pi rl)= Q_{free} \quad \rightarrow \quad E = \frac{Q(t)}{2 \pi \epsilon l} \frac{1}{r} \hat r$ 

2- V between plates  
$V(t) = \int^{out}_{in} E \cdot l = \frac{Q(t)}{2 \pi \epsilon l} \ln (\frac ba)$  
thus:  
$E = \frac{V(t)}{\ln (b/a)}\frac 1r \hat r$  

3- B  
direction: caused by the current I in the inside plate, so like a cable, make circles around it  
compute: $B = \frac{\mu_0 I(t)}{2 \pi} \frac{1}{r} \hat \theta$  

**Poynting vector**

$S = E \times\frac  B {\mu} = E \times H = \frac{V(t)I(t)}{2 \pi \ln (b/a)} \frac 1 {r^2} \hat z$  
it's axial, so the flux is across the section of the cable:  
$\Phi(S) = \int S \cdot dS = \int_a^b S(r) \; 2 \pi r \, dr = V(t)I(t)$  

meaning:  
$\Phi(S)$ has dimension of a power, is energy crossing the considered surface per unit time  
$\Phi(s)=V(t)I(t)$ means that ALL power instantaneously provided by the generator is transferred thorugh the cable  
so, no power will leak out of the system and no coupling will occur with external circuits, obviously except the load  
we can also assume that no coupling will also occur with the inner conductor due to external time dependent field

if S is contained inside the cable  
Energy flows only inside → guided wave  
so no leakage of power occurs out of the cable
