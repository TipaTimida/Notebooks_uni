### Independent stages

**Condition of independent stages**

2 methods to write it:  
$Z_{in,2} \gg Z_{out,1}$  
$i_{in,2} \ll i_{out,1}$  

**Condition: intuitively**

if i want to add a stage (stage 2), its impedance must be really big, so current from stage 1 doesn't go to it  
so it doesn't disturb stage 1, which contineues to behave as if stage 2 wasn't there  
so it's easier to model  

exemple:  
stage 2: multimetro o un oscilloscopio  
devo mettergli una resistenza interna grande   


**Explanation condition**

situation:  
have 2 stages, 1 and 2  

what i want:  
$i_{out,1}$, la corrente che esce verso la prossima maglia, è trascurabile  
so minimize the fraction of current flowing in the first stage which is branched offto the second stage   
else the first stage gets disrupted, it's harder to model  

### Resistenza di input/output

#### How to compute

**Formule**

1- input impedance:  
$V_{in} = Z_{in}I_{in}$  

2- output impedance:   
$V_{out}=Z_{out}I_{out}$  

**Steps: for DC**

1- draw the circuit    

2- identify the nodes $V_{in}$ and $V_{out}$   
(where i take input and output)  

3- measure $I_{in}$ (current on wire starting from node $V_{in}$)  

4- turn off sources + measure $I_{out}$ (current on wire starting from node $V_{out}$)  

**Tips: measure I**

1- parti dal nodo $v_{in}$ o $v_{out}$  
2- vai fino al GND, segnando tutte le cadute di voltaggio   
(ONLY IF doing output resistance, $v_{in}$ counts as GND, else no)  
3- esprimi tutte le correnti in funzione di quella che parte dal nodo che mi interessa, $I_{in}$ o $I_{out}$    
4- riordina per trovare la formula, vedi cosa è Z  

**Notes**

1- in pratica sto immaginando di chiudere i terminali di $V_{in}$ con una batteria, così ho creato un loop, su cui usare la legge delle maglie  
ma più facile vederlo come parti dal punto e arriva a GND  

2- for OUTPUT impedance  
sostituisci $v_{in}$ con GND     
note that i have 2 GNDs now, must consider both paths in parallel !!!   
maybe connect the $V_{in}$ to the old GND, so i have a single arrival point  
the output resistance is the thevenin resistance  

3- if the battery is a DC, must put $\omega=0$ in all the Z of the components  
so the formula of $Z_{in}$ and $Z_{out}$ is technically the same, just computed with a specific $\omega=0$  
also, in that case the DC batteries are NOT shortcuted to GND???

**Example: voltage divider**

1- input resistance:  
$v_{in}=I(R_1+R_2)$  
$R_{in}=R_1+R_2$  

2- output resistance:  
$v_{out} =I(R_1 || R_2)$    
$R_{out}=R_1 || R_2$  

here there is only one I, so in both resistances i just divide by that I, bc it is the current at both nodes $v_{in}$ and $v_{out}$  

![](https://media.geeksforgeeks.org/wp-content/uploads/20240219120854/image-120.webp)

**Example: RC circuit**

1- input impedance    
follow $V_{in}$ to GND (the bottom line)  
$v_{in}=(Z_C + Z_R)I$   
$Z_{in} =Z_C+Z_R$  

2- output impedance  
shortcut $V_{in}$ to GND, follow $V_{out}$ to GND  
$v_{out} = (Z_C || Z_R)I$  
$Z_{out}=Z_C || Z_R$  

note that, if i swap C and R, i get the same results  

![](https://upload.wikimedia.org/wikipedia/commons/thumb/b/b8/NPN_emitter_follower.svg/250px-NPN_emitter_follower.svg.png)

ci sono anche $R_B$ e $R_C$  

**Example: transistor**

1- input resistance  
è la solita legge delle maglie del transistor, uguale in common collector ed emitter (segui percorso fa $v_{in}$ a GND)  
$v_{in} = [R_B +r_b+(\beta +1) R_E]i_b$   
$R_{in} = R_B +r_b+(\beta +1) R_E$  

2- output resistance  
cambia per common collector e emitter  
$V_+$ va al GND, $V_{out}$ è preso sotto E (common collector) o sopra C (common emitter)  
per arrivare al GND, non posso attraversare il transistor, perchè la giunzione CB è in reverse durante il regime attivo, quindi ha resistenza infinita  
$v_{out, ce} -R_c i_c = 0$  
$R_{out, ce}=R_c$   
$v_{out, cc} - R_Ei_e=0$  
$R_{out, cc}=R_E$  

### Chaining filters

**Compute impedances**

Z can be examined at different frequencies, depedning on what my objective is:  
1- not chaning the baising of a transistor  
only check what happens at f=0, bc baising is set with DC  
2- chanining filters  
for each filter, check what happens at the frequence that it lets pass   

**Compute impedances**

for each filter, Z is computed in the frequency that it lets pass  
approximate with $f=0$ or $f =\ \infty$, so $Z_C$ as well is 0 or infinity  

1- $R_{in}=Z_R+Z_C$  
low frew: $Z_{in} = R + \infty = \infty$  
high freq: $Z_{out} = R + 0 = R$  

2- $R_{out}= R_1 || R_2$  
low freq: $Z_{out} = R \, || \, \infty = R$  
high freq: $Z_{out} = R \, || \, 0 = 0$  

**Chain filters**

rem my condition is:   
$Z_{out, 1} \ll Z_{in,2}$  

to make it always true, can use: $0 \ll \infty$  
so: 1=high-pass, 2=low-pass  

if i do the other order, low -> high:  
i get as condition: $R_L \ll R_H$  
so i works only if the resistor in the high-pass filter is much bigger than that in the low-pass  

**Note** 

The output impedance changes the most around cutoff.  
This region is where loading from the next filter causes the biggest effect.  
so must compute the impedance arounf cutoff (so in the region that it lets pass)  

### Oscilloscope

**How to sketch (and input impedency)**

made of a C and R in parallel  
$Z_{in} = R_{in} \, || \, C_{in}$  
$R_{in}$: internal resistance  
$C_{in}$: capacità spuria/parassita  

**Impedance**

1- input impedance:  
$Z = R \, || \, Z_C$  
bc it's made of a capacitor and resistor in parallel

2- output impedance:  
don't care, since will only use it nella seconda maglia  

problem with input impedance:    
at high freq, $Z_C \rightarrow 0$, and since in a parallel wins the smallest component, get $Z = 0$  
so oscilloscope start affecting the circuit, called 'loading effect'  
(loading effect = the measuring instrument becomes part of the circuit)  
in this case, the oscilloscope acts as a low-pass filter???   

solution:  
C must be small enough that, even for high freqs, $Z_C$ is still big enough  

**Filters in an oscilloscope**

upper bound: 50 MHz  
lower bound: 0 when used in DC (lets everything pass, even DC with f=0), between 5 and 20 when used in AC (bc don't let pass DC anymore, and kinda blocks really slow ones too)   

so there is always a low-pass filter (the upper bound)  
if put AC mode, means adding a high-pass filter  

even in AC mode, might still say bandwidth is 50 MHz, to refer to the upper bound
