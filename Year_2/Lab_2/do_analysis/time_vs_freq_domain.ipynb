**Nota**

spesso cambio nome al circuito per dire se lo sto studiando in time o freq domain  

circuito RC:  
time: integrator/derivator, o carica/scarica condensatore  
freq: filtro passa-alto, passa-basso  

circuito RLC:  
time: oscillatore armonico smorzato  
frequency: dove plotto la risonanza  

per passare dall'uno all'latro dominio, si usa la trasformata di Fourier  

**Come si fa**

in entrambi devo scrivere la legge delle maglie, ma:  

1- time domain  
disegna il circuito chiuso  
scrivi: legge delle maglie, $\Sigma V = 0$, con tutto espresso in funzione di I(t) o Q(t)  
descrizione: risolvi eq differenziale, per avere I(t) o Q(t)  
calcola: V del componente in funzione di I(t) o Q(t), a seconda di dove prendo l'output (quindi solo qui importa il componente su cui prendo il drop) (tipo $V_R(t) = RI(t)$)

2- freq domain  
separa la batteria in VCC e GND, disegna il circuito aperto    
scrivi: $V_{out}$ e $V_{in}$ (qui importa il componente su cui prendo il drop)      
descrizione: $H=V_{out}/V_{in}$  
calcola: gain e dephasing  

**Come scrivere legge delle maglie**

ogni componente si scrive con:  
1- time domain: V, il drop su di esso  
2- freq domain: Z, la sua impedenza  

. | time: V | freq: Z 
---|---|---
resistore | $RI$ | $R$
condensatore | $\frac QC$ | $\frac{1}{j \omega C}$  
induttanza | $L \dot{I}$ | $j \omega L$  

### Esempio: circuito RC

#### Time domain

**Risolvere**

legge della maglie in funzione di t:  
$\mathcal{E}-IR -\frac{q}{C}=0$  

risolvo:  
$\dot{I} = -\frac{I}{RC}$   
carica: $i(t) = i_0e^{-\frac{t}{RC}}$  
carica: $i(t) = -i_0e^{-\frac{t}{RC}}$  

che cambia tra carica e scarica???  

**Drop su qualcosa**

drop su R: $V_R(t) =RI$  
drop su C: $V_C(t)=\frac QC = \frac IC$  

#### Freq domain

**Risolvere**

scrivi voltaggi in funzione di $\omega$:  
$v_{in} = (Z_R+Z_C)I$  
$v_{out} = Z_C I$  

risolvi:  
$H=\frac{Z_C}{Z_R+Z_C}$
