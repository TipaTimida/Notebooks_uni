### Tipi di errori

. | Errore statistico | Errore sistematico 
---|---|---
altro nome | di digitalizzazione/lettura (per digitale/analogico) | di calibrazione
dove lo trovo | digitale: nel manuale - analogico: stimato da me (eg. 1 tick spacing) | nel manuale  
distinguerli | è un valore assoluto fisso da sommare, tipo $\pm 1 \; digit$ | è un valore relativo da sommare, tipo 5% della lettura o del valore nominale  
what | the random scatter you get when you repeat the measurement many times | the possible bias or scale error of the instrument itself  
example | your voltmeter fluctuates between 5.08 V and 5.10 V, → ±0.01 V | your voltmeter might read 0.5% too high or too low for all measurements, because its internal reference isn’t perfect → ±0.5%

#### Spiegazione errori

**Spiegazione: errore di lettura**

reflects fluctuations in the measurement due to: misalignments, parallax issues, mechanical vibrations...  

caused by the finite size of:  
- the needle (or the trace displayed on the oscilloscope screen) 
- the ticks marked on the quadrant (or on the oscilloscope graticule)  

**Spiegazione: errore di calibrazione**

assume i can neglect fluctuations (neglect statistical error): thus, by repeating many times the same measurement, i would obtain always the same result, which would be always an overestimate or an underestimate of the true value    
Therefore, in principle no distribution would occur, being the uncertainty related with non statistical (often called systematic in the jargon) mechanisms     
is conceptually similar to the tolerance the manufacturer attributes, for instance, to the resistance value of a resistor

### Propagare incertezza

**Esempio veloce: quando ho solo moltiplicazione/divisione**

$f=\frac{1}{2 \pi RC}$  

1- trova cosa con incertezza (va bene nominale, tanto è circa)  
2, $\pi$: non hanno incertezza  
$R$: incertezza 5% nominale  
$C$: incertezza 20% nominale  

2- somma incertezze percentuali    
f ha incertezza di 5%+20%=25%  

3- calcola f, e vedi auanto è quella percentuale  
f si calcola con i valori centrali dei parametri  
poi calcola 25% di quella f  

nota:  
non sommare in quadratura, perchè non sono errori statistici

**Propagazione per moltiplicazioni/divisioni**

$f=x^ay^bz^c$  

errore sistematico, o non sono sicura: somma errori relativi  
$\frac{\Delta_f}{f}= |a \frac{\Delta_x}{x}| + |b \frac{\Delta_y}{y}| +|c \frac{\Delta_z}{z}|$  

errore statistico: somma in quadratura  
$(\frac{\sigma_f}{f})^2= (a \frac{\sigma_x}{x})^2 + (b \frac{\sigma_y}{y})^2 +(c \frac{\sigma_z}{z})^2$  

parametri correlati:  
$\sigma_f^2=(\sigma_x \frac{\partial f}{\partial x})^2+ (\sigma_y \frac{\partial f}{\partial y})^2+2 \sigma_{xy} \frac{\partial f}{\partial x} \frac{\partial f}{\partial y}$  

**Propagazione errori per somma/sottrazione**

semplicemente somma gli errori, non sono pesati in nessun modo  

### Fit: trattare gli errori

#### Test chi quadro

**Quando si può fare?**

condizione: l'errore deve essere la DEVIAZIONE STANDARD  
perchè solo così $\chi^2$ è una variabile aleatoria  

come averle la dev standard:  
1- time domain: durante l'acquisizione da computer, dì al programma di fare la media su tante sweep  
così mi ridà la stddev come errore  
2- freq domain: misura tutti i valori nell'oscilloscopio sula stessa scala, così non ho errore di calibrazione  
questo è errore statistico, ma per avere stddev devo dividere per $\sqrt{12}$ (assumo distribuzione uniforme dell'errore)  

**Cosa fare**

1- errore prevalentemente statistico:  
calcola valore atteso di $\chi ^2$ e deviazione della sua distribuzione  
vedi entro quante barre di errore è dal valore ottenuto  

2- errore prevalentemente sistematico:  
chi quadro non ha alcun significato statistico, per evidenziare ciò lo chiamo kappa quadro  
è comunque un estimatore, ha sempre un significato e va sempre riportato, è un risultato del best fit  
voglio solo che sia più basso possibile    
quindi riporta il chi quadro ridotto, ma non dire entro quante sigma è, perchè qui non so la deviazione della sua distribuzione  

**Perchè facciamo il fit?**

non ci interessa valutare se il modello è giusto  
viene dalla fisica, assumiamo che lo sia  
voglio solo sapere i parametri ottimizzati  

#### Absolute sigma

**Absolute sigma**

di default, è messa su false  
cambia solo il valore delle incertezze dei parametri ottimali, non a che parametri converge  

’False’ or ’True’ corresponding to the asymptotic and standard error, respectively:  

- True -> standard error: se l'errore è assimilabile alla deviazione standard  
usa se le incertezze sono prevalentemente statistiche  

- False -> asymptotic: accounts for the possible over or underestimation of $κ^2$ due to under or overestimated errors   
according to: $\Delta p_a = \sqrt{k^2_{red}} \Delta p_s$  
usa se le incertezze sono prevalentemente sistematiche  

in pratica: absolute sigma = gli errori sono statistici?  

#### Altro

**Quando NON è un errore statistico?**

se tutte le misure hanno un ERRORE RELATIVO COSTANTE: probabilmente è un'incertezza di calibrazione, non un errore statistico  
cioè nel grafico, le barre di errore sono più lunghe per i valori più in alto, con y maggiore  

**Sigma**

$\sigma$ è il simbolo della DEVIAZIONE STANDARD, non di un errore in generale      
posso usarlo come simbolo e parlare di deviazione e varianza solo se sono sicura che sia un errore sattistico, non sistematico  

**Errori effettivi**

posso passarlo a curve_fit, invece dell'errore normale:  
$\Delta {eff}_i ^2 = (\Delta y_i)^2 +(\Delta x_i \frac{\partial f}{\partial x})^2$  

the derivative must be computed on the optimal parameters  
so first must compute them with the normal error, then can repeat the fit with the effective error to find the uncertainty  

**Normalized covariance**

$c_{xy} = \frac{\sigma_{xy}}{\sigma_x \sigma_y}$  
c is a number between -1 and +1  

**Covariance**

Finally, the sign of covariances can represent an additional check of the best-fit  
for instance, assuming  
a linear model function with two parameters (intercept and slope), one should expect a negative covariance  
since the increase of one parameter could be compensated by decreasing the other one.
