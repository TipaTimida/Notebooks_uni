1- write $V_{in}$ geometrically, as a function  
2- then integrate or derivate it, to get $V_{out}$

quadra: $g=x$ (horizontal line)  
triangolare: $g=x^2$ (linea obliqua)  
$\frac{d}{dt}$ triangolare = quadra    
$\int $ quadra = triangolare  

$\int$ triangolare = rami di parabola (è più bassa e cicciotta di una sinusoidale)  
$\int$ sin = cos (quindi sto solo sfasando)  
$\frac{d}{dt} sin = cos$

### Physical explanations

**How to remember topology and condition**

1- where to take output across:  
must pass from one voltage drop to the other, so if take output across R, so V=IR:  
in RC circuit, must pass to Q (in drop of C), so must integrate I -> integrator  
in LC circuit, must pass to I' (in drop of L), so must derivate I -> derivator  

2- what is condition  
the thing i'm NOT taking the drop across must have no drop  
if it's L or C, pick condition to make it an short  
if it's R, pick condition to make the other an open, so no current passes  

tip condition: think about the name of the filter  
f that the filter BLOCK: it's integrating or derivating, it works   
f that the filter lets pass: leave mostly unchanged, just a bit attenuated   

**Condizione su f o T**  

ricorda come si comporta C in questi casi:  
1- low f: C è un aperto, non scorre I -> quindi no drop sul resistore  
2- high f: C è un corto, ha Z=0 -> quindi no drop su C     

**Come vedere se integra/deriva**

1- scrivi legge delle maglie  
2- esprimi il V attraverso cui prendo l'output    
3- passa da Q a I o iceversa, così ottengo il drop sull'altro componente  
4- usa condizione su f per uguagliare questo drop al generatore    

cioè, la condizione su f o T è fatta in modo da minimizzare il drop sul componente che NON sto guardando  

**Circuiti RC**

$V_G(t)-V_R(t)-V_C(t)=0$   

- integrator: C as output, small T  
$V_C(t)=\frac{Q(t)}{C} = \frac{\int I(t) \; dt}{C}=\frac{1}{RC} \int V_R(t) \, dt = \frac{1}{RC} \int V_G(t) \, dt$ 

- differentiator: R as output, big T    
$V_R(t)=I(t)R=R\frac{dQ}{dt} = RC \frac{dV_C}{dt}= RC \frac{dV_G}{dt}$  


**Circuiti LR**

$V_G(t)-V_R(t)-V_L(t)=0$  

- integrator: R as output, big T        
$V_R = RI = \frac{R}{L}\int V_L(t) \, dt =  \frac{R}{L}\int V_G(t) \, dt$  

- derivator: L as outputs, small T   
$V_L = L\frac{dI}{dt} = \frac{L}{R}V_R= \frac{L}{R}V_G$  

**Osservazione: RC vs RL**

have opposite topology:  
RC integrator: output across C  
LR integrator: output across R  

mathematically:  
bc $V_L = L \frac{dI}{dt}$, $V_C = \frac{Q}{C}= \frac{\int I}{C}$  
so integration and derivation are inversed  

physically:  
inductors and capacitor become shorts or open (so Z=0 or $Z=\infty$) at opposite frequencies   
so to minimize the drop on the component i'm not observing, need the opposite condition on T 

### Altro

**Prove filter=integrator/derivator using transfer func**

low-pass for high f = integrator  
$H_L = \frac{1}{1+j\frac{f}{f_c}}$ with $f / f_c \gg 1$  
$H = \frac{1}{1+j\frac{\omega}{\omega_c}} \sim \frac{1}{j\frac{\omega}\omega_c} = -j\frac{f_c}{f}$   
modulo: 1/f  
fase: -pi/2 (perchè punta in -j, cioè verso il basso)  

high-pass for low f = derivator  
$H_H = \frac{j\frac{f}{f_c}}{1+j\frac{f}{f_c}}$ with $f / f_c \ll 1$  
$H = \frac{j\frac{f}{f_c}}{1} = j\frac{f}{f_c}$  
modulo: f  
fase: +pi/2    

**Integrator and derivator: conseguenze non volute, little imperfections**

if give in input a square wave:  

high-pass = derivator:  
takes well the sudden spike (vertical line), but once start the part with DC it starts attenuating it  
so in the horizontal line, the wave starts decreasing (bc it's being attnuated), even if it's technically not derivating, bc it don't have the condition on T  

low-pass = integrator:  
when it integrates, appears an offset  
perchè l'onda di input non è esattamente simmetrica (alternata), ma ha un po' di tolleranza sul duty cycle, quindi il suo valore medio (che è ciò che calcola l'integratore) è diverso da 0  
per risolvere, lavora a frequenze alte, tipo 20KHz, così posso accoppiare in AC  

**How wave changes graphically, in freq domain???**

1- integrator = low-pass filter:  
low freq, cioè $T \gg \tau$: mostly unchanged, sharp spikes a bit attenuated  
high freq, cioè $T \ll \tau$: integra, kinda takes the average value  
so: smooths down the spikes, get a line a bit traballante around the average value  

2- derivator = high-pass filter  
high freq, cioè $T \ll \tau$: mostly unchanged  
low freq, cioè $T \gg \tau$: deriva  
so: it outputs a sudden spike when there is a fast change in V, and outputs V=0 when V stays constant???  

**Explain: frequency of signal**

charge and discharge of capacitor:  

beginning: suddenly close the circuit  
spike in voltage, high frequency
so $Z_C \sim 0$, acts like a corto        
  
end: start teaching reaches stationary state, voltage changes slowly, until it don't change anymore, low frequency  
low frequency: $Z_C \sim \infty$, acts like a break in the circuit (shortcircuit)   

**Frequency of signal**

in a periodical signal, can define $f=\frac{1}{T}$  
cut then, it can be a superposition of waves of different frequencies  

eg. 
pure sinusoidal: monochromatic  
square, triangle: polychromatic (made with Fpurier)
