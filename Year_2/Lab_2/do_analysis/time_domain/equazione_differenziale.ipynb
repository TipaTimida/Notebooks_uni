### 1. Scrivi legge delle maglie

**Steps**

1- trova i poli + e - su ogni componente  
2- percorri la maglia in un senso qualunque, sommando i drop su ogni componente  
ATTENTA AL SEGNO del drop  

**1. Poli + e - dei componenti**

1- generatore: dati dal disegno  

2- altri pezzi: dipende dal verso della corrente, corrente entra nel polo +   
in pratica il polo più vicino al polo + del generatore è quello positivo, come in lab  

attenzione: se NON c'è il generatore (tipo in scarica condensatore, o in oscillatore autonomo), allora il condensatore agisce da generatore  
cioè usa quello per stabilire quale è il VCC, e da quello individua i poli + e - degli altri componenti  
ma poi $I = -\dot Q$, c'è un -  

in pratica:  
1- stabilisci il VCC e il GND (se c'è generatore è ai suoi capi, sennò è ai capi del condensatore che agisce da batteria)     
2- per ogni componente, individua polo + e - (più vicino a VCC è GND)  

tip:  
metti + al drop dell'oggetto che agisce da generatore (il generatore stesso o, se non c'è, un condensatore)  
tutti gli altri componenti, metti - al drop  

**2. Percorri la maglia**

il segno del drop di ogni componente dipende in che senso lo attraverso, cioè quale polo tocco per prima:    
| attraversamento | segno  |
| --------------- | ------ |
| da **− → +**    | **+V** |
| da **+ → −**    | **−V** |

in pratica, vedi se sto scendendo o salendo di voltaggio  

**3. Riscrivi tutto in funzione di Q o I**

a volte appaiono sia $I(t)$ che $Q(t)$   
se voglio tutto in q: metti $I = \dot{q}$  
se voglio tutto in I: dividi tutto per dt, così q sparisce (così anche $\mathcal{E}$ sparisce se è in DC, il suo valore mi tornerà quando metterò le condizioni iniziali alla fine)  

attenzione ai segni:   
se non c'è generatore, è C che ha la carica iniziale, allora $I=-\dot Q$, c'è il segno -   
perchè la corrente scorre quando la carica di C diminuisce, visto che va in circolo  
mentre se c'è un generatore, è $I=\dot Q$   
e NON si mette il segno -, perchè nei circuiti assumo che la corrente sia data dai protoni, non dagli elettroni  

**Intuizione**

percorri la maglia, e sottrai il drop di ogni componente incontrato  
se incontro un C, controlla se agisce da batteria (cioè se manca il generatore), allora mettici il segno + al drop e poi ricorda $I = -\dot Q$

#### Risolvi eq differenziali: lungo

**Esempio: separazione var**

1- legge della maglie: $\mathcal{E} -L \frac{dI}{dt}-RI=0$  
$\frac{dI}{dt}=-\frac RL I + \frac{\mathcal{E}}{L}$  

2- sol omogenea: $\frac{dI}{dt}=-\frac{R}{L}I$  
$I_h(t) = Ae^{-t/\tau} \quad \tau = \frac{L}{R}$  

3- sol particolare: $\mathcal{E} -RI=0$   
$I_p = \frac{\mathcal{E}}{R}$  

4- sol generale e condizioni iniziali:  
per avere I=0 al tempo t=0 devo mettere: $A = -I_0$   
$I(t) =    \frac{\mathcal{E}}{R} (1- e^{-t/\tau})$

per avere Q=Q_0 al tempo t=0 devo mettere (se avessi equazione diff su Q)???

**Versione solo omogenea: separazione var**

legge delle maglie:  
$-V_c + V_R = 0$  
$\frac{q}{C}+R\frac{dq}{dt}=0$  
$\frac{dq}{q} = -\frac{1}{RC} dt$  
$[\ln q]^q_{q_0} = -\frac{1}{RC} [t]_0^t$  
metti direttamente gli estremi di integrazione  

$q(t) = q_0 e^{-\frac{1}{RC}t}$  
$q_0 = V_0C$  


### Esempio: circuito RC

**Carica vs scarica**

l'equazione differenziale è diversa, perchè nella scarica ho tolto il generatore  
inoltre cambiano le condizioni iniziali  

è più facile dare le condizioni su Q che su C  
perchè su Q devo considerare se è 0 o $Q_0$ mentre su I devo pensare al segno  

**Scrivi eq differenziale**  

1- carica:  
gen -> R -> C -> GND  
gen (-+) -> R (+-) -> C (+-)  
attraverso da sinistra a destra, trovo:  
$\mathcal{E}-IR-\frac QC=0$   
in funzione di Q: $\mathcal{E}-\dot Q R-\frac QC=0$  
in funzione di I: $\frac{d\mathcal{E}}{dt}-\dot I R-\frac IC=0$  

2- scarica:    
C -> R -> GND  
C(-+) -> R(+-)   
attraverso da sinistra a destra, trovo:  
$\frac QC -RI=0$   
in funzione di Q: $\frac QC +R \dot Q$ (ricorda $I = -\dot Q$, perchè non c'è generatore)  

**Risolvila**

1- carica:  
eq diff: $\mathcal{E}-IR -\frac{q}{C}=0$   
sol generale: $Q(t)=Ae^{-t/\tau}+C{\mathcal{E}}$      
condizione iniziale: $Q(0)=0$  
quindi: $Q(t)=C{\mathcal{E}} (1-e^{-t/\tau})$  

2- scarica  
eq diff: $\frac{q}{C}+IR=0$   
condizione iniziale: $Q(0)=Ae^{-t/\tau}$  
condizione iniziale: $Q(0)=Q_0$   
quindi: $Q(t) = Q_0 e^{-t/\tau}$   

### Energia

**Cosa fare**

in time domain, posso calcolare l'energia di tutti i componenti  
per farlo, trova Q(t) e I(t) del circuito, e pluggali nelle formule dell'energia dei componenti  

resistore: $P_R=I^2R$  
condensatore: $E_C=\frac Q{2C^2}$   
induttore: $E_L=\frac 12 LI^2$   
generatore: $P_g = VI$  

$E = \int_0^\infty P(t) \, dt$  

**Esempio: energia circuito RC**

scarica:   
energia iniziale del condensatore: $E_C=\frac 12 CV^2$  
energia dissipata nella resistenza: $E = \int_0^\infty RI^2 \, dt  = R \frac{Q_0^2}{RC^2} \int_0^\infty  e^{-2t/RC} = R \frac{Q_0^2}{RC^2} \frac{RC}2 = \frac{Q_0^2}{2C}$  
dove: $Q_0 = CV$  

carica:  
fornita dal generatore: $E_g = \int_0^\infty V \, I(t) \, dt = \frac{V^2}R \int_0^\infty e^{-t/RC} \, dt = \frac{V^2}R \, RC = CV^2$  
finale nel condensatore: $E_C = \frac 12 CV^2$  
dissipata nella resistenza: $E_R = E_g-E_c=\frac 12 CV^2$    

risposta:  
scarica: tutte l'energia del condensatore è dissipata sulla resistenza  
carica: l'energia del generatore va metà al condensatore, metà dissipata sulla resistenza  

**NOTA IMPORTANTE**

se non ci sono resistenze, non c'è energia dissipata  
quindi E(t)=const  
allora calcola in un momento facile, tipo all'inizio, quando C è carico e I=0  
coì c'è solo energia su C, non su L
