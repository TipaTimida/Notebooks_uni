2- $\dot{I} = - \frac{I}{L/R} -\frac{\mathcal{E}}{L}$

**Scrivi eq differenziale**

1- $\mathcal{E}-IR-\frac QC=0$

$Q(t)=C{\mathcal{E}} (1-e^{-t/\tau})$

1-  $\frac{Q}{C}-IR=0$

$Q(t) = Q_0 e^{-t/\tau}$

quindi come generatore metto un'onda quadra alternata

semionda LOW: adesso è settata Q=I=0 da raggiungere, cioè il circuito si scarica tutto, sto tipo resettando per ricominciare con un altro ciclo

### Circuito LC

![alt text](https://mathtube.altervista.org/wp-content/uploads/2020/09/Circuito_LC.jpg)

#### Behaviour

**Spiegazione**

il condensatore e l’induttanza si scambiano corrente tra loro, dando origine a correnti e tensioni PERIODICHE    
energia passa da immagazzinata nel campo elettrico di C a immagazzinata in quello magnetico di L  
cioè c'è oscillatore armonico, alla frequenza naturale perchè non c'è R  

#### Ricava formule

**Soluzione**

1- $\frac{q}{C}-L\dot{I}=0$  
2- $\ddot{I}=-\frac{I}{LC}$  
3- moto armonico: $i(t)= i_0\sin(\omega t + \varphi)$  
dove $\omega^2 = \frac 1{LC}$  

$\varphi$ è lo sfasamento iniziale  
siccome la corrente prima di chiudere l’interruttore è nulla, lo sfasamento è nullo: φ=0  

#### Energia

**Energia**

calcola $E_L(t)$ e $E_C(t)$, ma come, con le formule sopra???  
vedo che periodo di E è la metà del periodo di I   

lavoro totale: somma dei alvori di condensatore e induttanza  
$W_{tot}=W_C+WL_ = \frac{1}{2}CV_C^2+\frac{1}{2}Li^2$  

i e $V_c$ le conosco, posso sostituirle  
viene:  
$W_{tot}= \frac{1}{2}CV_0^2$  
all’inizio l’unica cosa carica è il condensatore, quindi l’energia che circola nel circuito sarà quella del condensatore

#### Struttura: generatore

**Presenza del generatore**

il circuito RLC è un oscillatore smorzato a frequenza naturale  
quindi all'inizio segue la sua oscillazione naturale smorzata, ma poi si ferma  
prchè $V \rightarrow 0$, cioè la soluzione omogenea dell'equazione tende a 0  

se metto un generatore:  
un generatore mi dà una soluzione particolare dell'equazione, la rende non omogenea  
quando decade la sol omogenea, il circuito va a regim secondo quello che dice il generatore 

**Come lo voglio io**

generatore di onda quadra seguito da un diodo  
così posso osservare il circuito oscillare senza generatore durante la semionda negativa, e poi ricaricare il circuito durante la semionda positiva  

perchè:  
semionda positiva: diodo in conduzione, c'è generatore, setta condizioni iniziali (cioè carica Q)   
semionda negativa: diodo in interdizione, quindi il generatore non c'è, ma C è carico, quindi comincia a oscillare il solo circuito RLC

#### Ricava formule (senza generatore, cioè lasciato oscillare)

**Ricava soluzione**

1- $\frac QC -IR-L \dot I = 0$  
2- $\frac QC + \dot Q R + \ddot Q L=0$  
non c'è generatore, quindi: $I =- \dot Q \quad$ $\dot I = \ddot I = - \ddot Q$  
3- è oscillatore armomico smorzato  

$L \ddot Q + R \dot Q + \frac Q{C}=0$   
$Q(t) = Ae^{-\gamma t} \cos(\omega t + \phi)$  

**Oscillatore damped**

1- confronta l'equazione con la forma standard per oscillatore damped:  
$L \ddot Q + R \dot Q + \frac Q{C}=0$   
$x'' + 2 \gamma x' +\omega_0^2 x =0$  

2- vedo:  
$\gamma = \frac{R}{2L}$  
$\omega_0^2 = \frac 1 {LC}$  

3- poi le formule sono (spiego dopo perchè):    
$\tau = \frac 1 \gamma$  
$\omega = \sqrt{\omega_0^2 -\gamma^2}$  
dove pluggo gamma e omega_0 trovate prima  

**Damped harmonic obscillator: exaplanation**

general formula:  
$x'' + 2 \gamma x' +\omega_0^2 x =0$  

example: sistema massa-molla smorzatore  
$m\ddot{x}+\beta \dot{x}+kx=0$  
$m=L$: L plays the role of the mass m, in accordance with the concept of inertia attributed to inductance  
$\beta = R$: dissipative term, in accordance with the occurrence of friction-like dissipation when charge carriers move in a conductor   
$C = k$: C is essential to enable oscillations, being the capacitor a reservoir of energy, analogies between the electrostatic energy of a capacitor and the elastic energy of a spring   

**Come si scrive soluzione**

$L \ddot Q + R \dot Q + \frac Q{C}=0$   
$Q(t) = Ae^{-\gamma t} \cos(\omega t + \phi)$  

1- risolvo eq diff di secondo ordine sostituendo $\lambda^2$ in $Q''$ ecc..  
ottengo radici: $\lambda = -\gamma \pm \sqrt{\gamma^2-\omega^2_0}$  
nel caso sottosmorzato, cioè oscillatorio, dove $\gamma < \omega_0$, la radice è negativa, quindi:  
$\sqrt{\gamma^2-\omega^2_0} = -i\sqrt{\omega_0^2-\gamma^2}$  
cioè: $\lambda = -\gamma \pm i\omega$  
cioè $\omega$ è un parametro che introduco qui, per scrivere più semplicemente l'interno della radice

2- sol generale:   
$Q(t)=e^{-\gamma t}(c_1e^{i \omega t}+c_2e^{-i \omega t}) = e^{-\gamma t} (A \cos (\omega t)+B \sin(\omega t)) =  A e^{-\gamma t} \cos (\omega t + \phi)$  

3- poi pongo $\tau = \frac 1 \gamma$   
lo scelgo io, perchè mi fa capire meglio il decadimento esponenziale:  
$Q(t)=  A e^{-t/\tau} \cos (\omega t + \phi)$  

pulsazioni:  
$\omega_0$: pulsazione naturale, se non ci fosse smorzamento  
$\omega=\sqrt{\omega_0^2-\gamma^2}$: pulsazione reale, quando c'è lo smorzamento

$\frac{d^2V}{dt^2}+\frac{1}{RC}\frac{dV}{dt}+\frac{1}{LC}=0$
