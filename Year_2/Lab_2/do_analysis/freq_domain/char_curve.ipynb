### Resistenza

What | effective/static resistance | dynamic/differntial resistance
---|---|---
defined for| any device with an I–V curve | nonlinear devices 
formula | $R_{eff}=\frac{V_q}{I_q}$ | $r_d = \frac{v_d}{i_d} \sim \frac{dV}{dI}\|_q$  
useful for |  large-signal behavior and DC bias | time-dependent and small signals  

**Rapporto**

dinamica < statica    

perchè:  
dinamica: relativa ad una perturbazione  
stazionaria: in una situazione stazionaria  
è più facile applicare una perturbazione che raggiungere una situazione stazionaria  
più facile rispondere alle variazioni che creare una situzione stazionaria da zero  

cioò stesso motivo per cui coefficiente di attito dinamico è minore di statico

#### Misurarle

**Perchè è difficile**

in componente non lineari, dipendono dal working point   
quindi sono difficili da misurare  

**Problema: resistenza statica diodo**

come funziona un multimetro:  
non misura la resistenza direttamente, ma inietta una corrente nota I, misura la V (o il contrario)  
così ricava R=V/I  

problema:  
quanto è grande I di test dipende dal fondoscala  
fondoscala alto → I_test bassa (se la tenessa alta, non riuscirebbe a misurare la V)  
in pratica, la corrente di test stabilisce il working point del diodo  
e quindi la resistenza a quel punto   

conseguenze:  
I_test alta -> working point nella parte di conduzione del grafico -> conduce bene, R bassa  
I_test bassa -> working point nella parte di interdizione -> non conduce, R alta  

**Come misurare resistenze**

1- fai un ciruito a due maglie, una di bias e una con la piccola oscillazione (il segnale vero)   

2- per dinamica:   
misura $v_d$ dell'oscillazione sul componente e $i_d$ sulla maglia non di bias (prendendo il drop su un resistore noto)    

3- per statica: misura $I_q$ nella malia di bias e $V_q$ sul componente  

nota che $V(t)=V_q+v_d(t)$, con $V_q \gg v_d$   
quindi per avere $V_q$ guarda con oscilloscopio, neanche la vedo l'oscillazione  
per $v_d$ devo mettere filtro passa-alto nell'oscilloscopio  

#### Resistenza dinamica: valore atteso

**Valore atteso della resistenza dinamica: diodo**

1- linearizzo la formula della resistenza dinamica: $r_d = \frac{v_d}{i_d} = \frac{dV}{dI}|_q$  
in pratica passo agli infinitesimi   
significa che invece della char curve, considero la sua tangente nel punto di lavoro  

2- ora posso usare come I(V) l'equazione di Shocley, che derivo:  
$\frac{dV}{dI}  =  (\frac{dI}{dV})^{-1} =  (\frac{I_s e^{V_q/(\eta V_T)}}{\eta V_T})^{-1} \sim \frac{\eta V_T}{I_q}$  
l'ultimo passaggio è perchè, in regime di conduzione, nel diodo posso approssimare Shockley come: $I= I_s e^{V/(\eta V_T)}$

**Valore atteso: approssimazioni usate**

1- $V_q \gg \eta V_T$   
per approssimare che Shockley sia $I= I_s e^{V/(\eta V_T)}$ (spiegato in note sul diodo)  

2- $v_d \ll \eta V_T$    
perchè sto linearizzando attorno al punto di lavoro, quindi faccio espansione di Taylor in quel punto (cioè, $I_qe^x$ dove $x \sim 0$)   
$I \sim I_q ( 1 + \frac{v_d}{\eta V_T} + \frac{v_d^2}{2(\eta V_T)^2})$  
quindi: $r_d^{-1} = \frac{i_d}{v_d}=\frac{I-I_q}{v_d} = \frac {I_q}{n V_T} (1+\frac{v_d}{2 \eta V_T})   = r^{-1}_{d, exp}(1+\frac{v_d}{2 \eta V_T})$  
cioè nella mia linareizzazione ho trascurato il termine $\frac{I_q}{n V_T}$  
se voglio il valore stimato vicino a quello misurato, deve essere $v_d \ll \eta V_T$  

nota: con l'approssimazione 2, posso anche stimare numericamente l'errore della mia approssimazione   
esempio: a temp ambiente, $\eta V_T \sim 52mV$, quindi se uso $v_d \sim 5 mV$ sto commettendo un errore relativo del 5%

**Problem**

the transition from the diffusion-limited regime to the series-resistance-limited regime  

total voltage drop of junction:  
$\Delta V = \Delta V_{junc} +IR_{ohmic}$    
$ R_{ohmic} I$: comes from majority carrier transport in bulk semiconductor  
$\Delta V_{junc}$: comes from diffusion of minority carriers (Shockley exponential)   

**What happens**

small foward current: Ohmic term negligible  
$V \sim V_{junc} = \eta V_t \ln (I/I_s)$  
why???  

big forward scurrent: Ohmic term wins, Shockley term saturates in comparison  
$V \sim IR_{ohmic}$  
so the curve deviates from the exponential one, tends to be linear like in ohmic component
