### Find transfer function

**Steps**


1- draw the circuit    

2- identify the nodes where i take input and output (where i attach the oscilloscope)  

3- scrive le leggi delle maglie, per le maglie in cui appaiono $v_{out}$ e $v_{in}$  
cioè parti dal nodo dove lo inietto e arriva al GND, scrivendo tutte le cadute di potenziale    
(cioè immagina di unire il nodo dove lo metto e il GND per fare una maglia, dove usare legge delle maglie)  
se puoi, semplifica esprimendo tutte le correnti in funzione di una sola  

3- usa: $H(\omega)=\frac{v_{out}}{v_{in}}$  

**Shortcut**

invece di scrivere le leggi della maglie per $v_{in}$ e $v_{out}$, vedi se riesco a scriverne una sola che li comprende entrambi  
es. nel circuito RC 

**Per scrivere meglio**

1- se c'è un condensatore, dopo aver scritto $H=\frac {V_{out}}{V_{in}}$, moltiplica per $j \omega C$ sopra e sotto  
così non ho frazione nella frazione  

2- se è un filtro, cerca di sostituire da qualche parte $f_c$  
di solito, viene tipo $f_c/f$  

3- spesso, sostituisco $\omega = 2 \pi f$  

**Nota**

situazione:  
se H non ha parte immaginaria (perchè il circuito non ha C nè L)   
allora H non dipende neanche da $\omega$  

conseguenze:  
posso chiamare: 
$g=\pm |H|$: un gain che può essere negativo  
il - indica l'inversione di 180 gradi    
quindi $\pm g$ corrispondono a dephasing di 0 o $\pi$, gli unici possibili senza parte immaginaria  
quindi diverso da $G = |H|$, perchè G è per forza positivo

#### Circuito in AC

**Variazioni**

qui mi interessano le variazioni, quindi:  
1- le correnti e voltaggi di scrivono con la lettera minuscola  
2- i voltaggi delle giunzioni si scrivono con la resistenza dinamica: $v_{be}=r_bi_b$ (e la corrente di dove entra???)  
3- le sorgenti di DC sono considerate GND  

attenzione al punto 3:    
se OLTRE A $v_{in}$ e $v_{out}$, nel circuito ci sono ALTRE sorgente di voltaggio, e sono in DC -> quelle contano come massa    
ma, quando calcolo $v_{out}$, NON devo mandare al GND anche $v_{in}$, quello avviene solo per la resistenza di output  
eg. $V_{cc}$ è considerato massa (AC ground), in esercizio con transistor  

**Variazioni e non**

un segnale è composto da:  
$I(t)=I_q+i(t)$  
1- parte in DC, stabilisce il working point  
2- parte in AC, piccole oscillazioni  

se voglio scrivere la legge della maglie completa, con anche la aprte in DC, fai il contrario di quello che dice nella guida per le oscillazioni:  
scrivi I e V con lettera maiuscola    
non togliere sorgenti di DC    

**Modello small signal**

quando:  
trova guadagno e impedenza di ingresso  
(invece il modello DC si usa tipo per trovare il bias del transistor)  
(potrei calcolare l'impedenza di ingresso anche in DC, ma di solito metto un condensatore di coupling, per bloccare il DC, quindi sarebbe semplicemente $Z_{in,DC}=\infty$, visto che il condensatore diventa un corto)

che succede:  
Le sorgenti DC diventano massa (ground AC)
Si studia solo la parte variabile, $i$ invece di $I$  

**Esempio: modello DC vs AC nel circuito RC**

1- in DC:  
C è un aperto, non c'è corrente, non c'è drop su R  
legge delle maglie: $V_{in}=V_{out}$  
funzione di traferimento: $H=1$   

2- in AC:  
leggi delle maglie:   
$v_{in} = (Z_R+Z_C)I$  
$v_{out} = Z_C I$  
funzione di traferimento: $H=\frac{Z_C}{Z_R+Z_C}$  
quindi si comporta da filtro  

cioè, il comportamento cambia in DC e AC  
il comportamento da filtro ha senso solo in AC

### Esempi

#### RC circuit

![](https://static4.arrow.com/-/media/images/research-and-events/articles/1018/1018_rc_circuits_inset_3.jpg)

il nodo nero in basso sarebbe il GND  

**Circuito RC**  

1- metodo normale  
$v_{in} = IZ_C + I Z_R$  
$v_{out} = IZ_R$  
$H(\omega) = \frac{Z_R}{Z_C+Z_R}$  

2- shortcut: voltage divider  
usa la formula del voltage divider  
$V_{out} = V_{in}\frac{Z_C}{Z_R+Z_C}$  

3- altra shortcut: KCL  
pick as node between R and C, where there is the voltage i want, V_C    
$\frac{V_{out}-V_{in}}{Z_R} + \frac{V_{out}-0}{Z_c}$  


**Scrivere più bello**

$H_{HP}= \frac{R}{R+\frac{1}{j \omega C}} = \frac{j\omega C R}{j \omega R C+1} = \frac{j {\omega / \omega_c}}{j \omega / \omega_c +1} = \frac{j {f / f_c}}{j f / f_c +1}$  

perchè: $\omega_c = \frac{1}{RC}$  
e $f/f_c = \omega / \omega_c$

### Compute gain and dephasing from H  

**Compute argument**

1- $arg(1/A)=-arg(A)$   
bc $\frac{1}{j}=-j$, and $\tan(-x)=-\tan (x)$  
 
2- $arg (a+ib)=\arctan(b/a)$  

3- if have fraction:  
with imaginary part only at bottom: use $\arg(1/A)=-arg(A)$, then compute arg of A   
with imaginary part both up and down: RAZIONALIZZA  

**Compute magnitude**

1- if have fraction: compute separately magnitude of numerator and denominator (|AB|=|A||B|)  

2- if one is a complex number, use Pytagora theorem: $|a+ib|= \sqrt{a^2+b^2}$, there is no $i^2$   

**Tip**

1- frequency when there is dephasing of 0 or $\pi$:  
must pick $\omega$ so H has NO IMAGINARY PART  
so dephasing is necessarily 0 or $\pi$, which one depends on the system, it's obligated  

2- intuitive dephasing  
rememeber H is a phasor, so dephasing is the angle it points at  
if have $H=-j\frac{f_c}{f}$, H point downward, so dephasing is $-\pi/2$  

#### Examples

**Low-pass filter**

$H_{LP}(\omega)=\frac{1}{1+ j \omega RC}$  

magnitude (attenuation):  
$|H(\omega)| = \frac{1}{\sqrt{1+(\omega RC)^2}}$  
used pytagora on denominator  

argument (dephasing):  
$arg(H) = - arg(1+ j \omega RC)  = -\arctan(\omega RC)$  
used the two formulas written above  


**High-pass filter**

$H_{HP}=\frac{j\omega RC}{1+ j \omega RC}$  
chiamo $\omega RC = x$  

1- razionalizzo:  
$H=\frac{jx}{1+jx} = \frac{jx}{1+jx} \frac{1-jx}{1-jx} = \frac{x^2+jx}{1+x^2}$  
$Re(H) = \frac{x^2}{1+x^2}$  
$Im(H) = \frac{x}{1+x^2}$  

2- argomento:  
$\phi= \arctan(\frac{x}{x^2})= \arctan (\frac 1x)$  

3- guadagno:  
$|H| = \frac{x}{\sqrt{1+x^2}}$  

**Behaviour at different frequencies**

low frequency: $\omega RC \ll 1$  
$|H(\omega)| \sim 1$ -> no attenuation  
$arg(H) =0$  

high frequency: $\omega RC \gg 1$  
$|H(\omega)| \sim \frac{1}{\omega RC}$ -> strong attenuation  
$arg(H)=-90$
