### Amplificazione

**Definizione**

amplificazione: guadagno in POTENZA  
cioè $|P_{out}/P_{in}|>1$ per almeno qualche frequenza  

atenzione:   
il gain $G=|H(\omega)|= |V_{out}/V_{in}|$ misura il guadagno in AMPIEZZA  
che NON è amplificazione  
quindi avere G>1 non vuol dire che il circuito stia amplificando  

graficamente:  
amplificare vuol dire aumentare l'energia, cioè aumentare l'AREA SOTTESA a V(t) in un periodo, NON aumentare l'ampiezza picco-picco (cioè istantanea)   
eg. circuito LRC può avere G>1 vicino alla risonanza, ma comunque non sta amplificando  

**Attivo vs passivo**

componente:  
1- passivo: non può generare energia, può solo immagazzinarla (L, C) o dissiparla (R)  
2- attivo: può fornire energia al circuito, perché è collegato a una alimentazione esterna  

circuito passivo: fatto solo di componenti passivi  
quindi non può amplificare  

**Passive vs active filters**

| Property     | Passive                                            | Active             |
| ------------ | -------------------------------------------------- | ------------------ |
| components   | R, L, C                                            | op-amp, transistor |
| need external power supply? | no                                                 | yes                |
| power gain   |  $\le 1$                                          | $>1$ possible      |
| voltage gain | usually $\le 1$, but can exceed 1 near resonance | arbitrary          |


**Perchè G>1 at resonance, anche se non amplifica**

In un RLC vicino alla risonanza, L e C si scambiano energia tra campi E e B    
L’energia immagazzinata può essere grande rispetto a quella fornita istante per istante dal generatore  
Questo fa sì che le tensioni su L o su C possano “risuonare” e diventare molto più grandi di Vin  
quindi se prendo l'output su C o L (solo in questi casi???, su R no???), potrei vedere una tensione grande, sembra che G>1  
Il generatore continua però a fornire solo la potenza necessaria a compensare le perdite su R

#### Attenuazione

**Gain vs attenuation**

gain and attenuation represent the same thing, but expressed in different scales:  

voltage gain (linear scale) -> attenuation/decibels (log scale)  
1 -> 0 dB  
0.1 -> -20 dB  
10 -> 20 dB  

**How to write in decibels**

2 definitions, depending if the quantities between which i compute the gain X are amplitude or power  
1- amplitude (eg. V, I): $dB = 10 \log(X)$  
2- power: $dB=20 \log(X)$

why:  
decibels are simply the gain in log scale, so take the base 10 log of the gain    
but change the coefficient, bc usually power is quadratic wrt amplitude, $P=IV^2$, so this way it's consistent  

**Computation**

-20 dB attenuator means:  

$-20dB = 10 \log (P'/P)$  
$-2 = \log (P'/P)$  
$P'/P=10^{-2}$  
$P'=\frac{P}{100}$  

so the new amplitude is 1/100 of the old one  
if use other formula, find $V'=\frac{1}{10}V$  

**What are decibels**

what is it:   
a unit of measurement for dimensionless quantities   
such as those coming out from ratios $X=\frac{x_1}{x_2}$  
useful to express stuff that naturally behaves on a logarithmic scale  

use in acoustics:  
in order for the decibel to be used
as an absolute, rather relative, unit of measurement, must pick a specific value for $x_2$ from wich start the scale (reference for the gain)   
in acoustics, for sound pressure level, it's the minimum
intensity able to produce an effect in a normal human ear    
in acoustics i'm talking about powers, so use the 10 coefficient

### Plot

#### Explanation: plot

**Description of plot**  

low-pass filter:  
$H(\omega)=\frac{1}{1+j\omega / \omega_C}$    
low frequency: $H(\omega) \sim 1 \rightarrow |H| \sim 1 \rightarrow 0 dB$  
high frequency:$H(\omega) \sim \frac{\omega_C}{\omega} \rightarrow |H| \sim  \frac{\omega_C}{\omega}  \rightarrow 20 \log (\omega_C) - 20log(ω)=const-20 \, dB$  

high-pass filter:  
$H(\omega)=\frac{j\omega/ \omega_C}{1+j\omega / \omega_C}$    
low frequency: $H(\omega) \sim \frac{\omega}{\omega_C} \rightarrow +20 \, dB$    
high frequency: $H(\omega) \sim 1 \rightarrow \rightarrow 0 dB$  


**Perchè pendenza di 20dB?**

$H(\omega) \sim \frac{\omega_C}{\omega} \rightarrow |H| \sim  \frac{\omega_C}{\omega}  \rightarrow 20 \log H =  \log (\omega_C) - 20log(ω)=const-20 \, dB$  

20dB è espresso in db/decade  
$\log \omega$ significa una decase  
$\log H$ significa decibel  
quindi è come una retta, solo attenta ad individuire x e y  

**CHECK !!! Study bode plot**

in transition band, can approximate to the first order with:  
$H(f) \sim \pm \frac{f_T}{f}$   
choose sign according if it's high-pass ot low-pass  

the horizontal line of the pass-band and the diagonal line of the transition band (se disegno i loro prulongamenti) cross at the cut-off frequency   

**Examine plot**

when plot in bilog, see kinda zig-zag, there is a mostly horizontal line and a diagonal line  
extimate pendance of diagonal line for LPF: when $f \gg f_c$  
$|H| = \frac{1}{\sqrt{1+(f/f_c)^2}} = \frac{1}{f_c/f\sqrt{(f/f_c)^2+1}} \sim \frac{f}f_c$  

due frequenze differiscono di un'ottava: sono una il doppio dell'altra  
due frequenze differiscono di una decade: sono una dieci volte l'altra   

quindi in una decade ci sono circa 3 ottave: 1->2->4->8
