**Sfasamento**

$\frac{\Delta \phi}{2 \pi}= \frac{\Delta t}{T}$  

visualizza le due onde sui due canali  

come fare:  
1- metti accoppiamento in AC (se posso)  
2- setta i GND uno sopra l'altro, su una linea che chiamo 0  
3- misura $\Delta t$: con i cursori verticali, misura la distanza tra i momenti in cui le due onde toccano la linea x=0  
4- misura $T/2$: misura la distanza tra i momenti in cui la stessa onda tocca l'asse x  
5- usa la formula e calcola $\Delta \phi / \pi$  

note:  
1- basta misurare T, non T/2, tanto poi nella formula per avere $\Delta \phi$ c'è un fattore 2 che divide T  
2- spesso la risposta si indica come $\Delta \phi / \pi$, quindi non moltiplicare per $\pi$, lascialo come unità do misura  
3- se due onde dello stesso tipo, posso misurare la distanza tra i momenti in cui raggiungono il picco, invece di quando toccano l'asse x  
così non serve mettere accoppiamento AC, se non posso  

**Sfasamento: XY mode**

ho due segnali:  
$\begin{cases}
x(t)=A \sin (\omega t) \\
y(t)=B \sin (\omega t + B) \end{cases}$  

at t=0, have x=0 and $y=B \sin \theta$  
so $\sin \phi = \frac{y_0}B$    
$y_0$: coordinate of the intercept with the y axis (so when x=0)  
B: vertical semi-amplitude of the ellipse, half-height of ellypse (bc would be radius in a circle) (cerca la mezza-larghezza massima, è B, perchè si ha quando sin=1)  

or at y=0, would get: $\sin \phi = \frac{x_0}A$  
$x_0$: coord of the itnercept with the x axis   
$A$: ezza-larghezza dell'ellisse  

**Trova $\tau=RC$**

formula: $\tau = \frac{\Delta t}{\ln (V_1/V_2)}$  
quando: se V(t) segue un decadimento espoenziale   

perchè:   
assumo: $V(t)=Ae^{-t/\tau}$  
misuro V in due punti: $\frac {V_1}{V_2} = e^{\Delta t / \tau}$  
$\tau = \frac{\Delta t}{\ln (V_1/V_2)}$  

tip: circuito RC  
misura $\Delta t$ tra le linee 90% e 10%, così $\ln = 2.2$  
non posso usare le linee 0% e 100%, la prima fa divergere il log, e la seconda è un asintoto, non la tocca a nessun t    

tip: circuito RLC  
posso trovare la forma $V(t)$ con decadimento esponenziale se considero SOLO i picchi (perchè pluggo cos=1 nella formula completa)   
quindi V1 e V2 vanno misurati su due picchi  

**Transistor va in saturazione, quindi c'è spike sul voltaggio**

quando misuro ampiezza del voltaggio, NON devo misurare la spike

**Intuitive meaning of figures and dephasing**

generally consider on CH1 and CH2 two signals with SAME FREQUENCY  
so can write (rem that dephasing is only in one, the other i arbitrately set to 0):     
$V_1=A_1\cos (\omega t)$   
$V_2 = A_2\cos (\omega t + \Delta \Phi)$  

cases:  
$\Phi=0$: V1 and V2 increase together, have always same value, so a line with m>1  
$\Phi=\pi$: V1 and V2 have opposite values, so  a line with m<1  
$\Phi=\pi/2$: V1 and V2 are a sin and a cos, so when one is max, the other is min (0), like in an ellypse with axes parallel to the x and y axes  
più aumenta lo sfasamento, più diventa ellittico/circolare, e meno rettilineo (l'ellisse diventa più cicciottella)  

**Attenuation**

decides:  
1- if ellypse: length of axes   
if $A_1=A_2$, it's actually a circle  
se le ampiezze sono uguali, ottengo un cerchio centrato (è come plottare la traiettoria di sin e cos, che sono sfasati di pi/2, ottengo un cerchio)

2- if line: angle with x axis  
if $A_1=A_2$, it's a bisector  

3- se A_1 è 0, ottengo una retta verticale, cioè il segnale X è attenuatissimo, al punto che è trascurabile rispetto al segnale Y

#### Teoria

**Profondità dinamica di digitalizzazione**  

anche detta: bit depth, profondità di bit  
indica il numero di bit usati per rappresentare ogni campione (audio) o pixel (immagine), determinando la gamma dinamica e la precisione del segnale

analogico:  
in teoria la deflessione del fascio può avere qualsiasi angolo, posso avere qualunque disegno  

analogico:  
la sensibilità è limitata dalla bit depth  
per ogni punto campionato, può restituire un valore di V a 8bit, cioè tra 0 e 255  
per arduino: era a 12 bit, quindi tra 0 e 4096  
sensibilità: 1/256 del fondoscala  

**Campionamento e modalità di acquisizione**

il campionamento avviene a $\Delta t = 1 ns$  
ogni nanosecondo, campiona un punto  
ma non tutti riescono ad apparire sullo schermo dell'oscilloscopio, gli altri sono scartati  
quindi in realtà l'intervallo $\Delta t_1$ tra due punti registrati nel record è più grande  

4 modalità di acquisizione:  
1- sample: campiona molti punti, ma nel record ne plotta solo uno ogni $\Delta t_1$ 
2- peak detect: aspetta un $\Delta t_1$, poi plotta max e min che ha visto in quel tempo, non semplicemente il primo punto che gli arriva  
usata per vedere i glitch (problema transitorio, che succede una volta sola) (a noi non serve)   
3- hi-res (resolution): per aumentare sensibilità e risoluzione dello strumento  
aspetto un $\Delta t_1$, poi plotta la media dei punti presi in quel tempo  
problema: riduce banda passante dello strumento (per noi non è grave)  
4- average: fa una media su un certo numero di spazzate consecutive  
miglioro sensibilità, ma devo aspettare un po' di tempo  
la stessa cosa che fa arduino, quando mi dice anche l'errore statistico  
qui, se uso uno script python, posso trovare l'errore anche dall'oscilloscopio  
per ridurre il rumore, 'average' è più forte di hi-res  

**Media mobile**

vuol dire, per ogni nuovo punto che campiono, nel record non viene salvato quel punto, ma la media tra gli ultimi N punti campionati (questo compreso)  
in pratica, campiona tanti punti  
poi usando una sliding window che si muove di 1 alla volta, nel record salva la media di ogni sequenza all'interno della sliding window
