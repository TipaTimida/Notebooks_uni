### Procedimento: misurare V,I,R

ALWAYS put incertezza  

**Voltage**

which tool: oscilloscope  
how: visualize the signal at that node, compute the peak-to-peak amplitude of the signal shown  
peak-to-peak bc it's easier to measure, can write that one directly nella scheda  
but in formulas, want peak-to-peak or the half???  

tips:  
1- if only care about amplitude of the oscillation, can use AC coupling  
eg. in diode or transistor, i don't care about the DC, it only sets the working point  

**Current**

2 methods:  

1- DC current: multimeter   
put it in series to the component on which i want to know I  

2- AC current: oscilloscope  
use it to measure voltage (drop) across a resistor, then compute $i=v/R$ (so, get the amplitude of the oscillating current)  
so, to measure a current, must put a resistor in that place  
example: in diode dynamical resistance, put an R in the AC mash, so i measure drop on it, so find $i_d$ of the perturbation

#### Obiettivo: misurare C,L,F

**Misurare C**

1- circuito RC (time domain)  
fitta la scarica, è più facile (perchè la funzione di fit è più facile?)  

2- filtro RC in dominio frequenza (freq domain)   
trova $f_{1/2}$, so il rapporto con $f_c$, che ha C nella formula  
$f_{1/2} = \sqrt 3 f_t = \frac {\sqrt 3}{2 \pi RC}$    
nota che questa cosa vale solo per circuito RC, perchè dipende dalla transfer function 

**Misurare L**

1- circuito LRC   
(o LC, in teoria c'è res interna di L, ma in formla di f0 tanto non appare R, chissene frega)    
trova freq di risonanza (quando dephasing=0, è una linea in mode XY), usa la sua formula per trovare L ($\omega_0= \frac{1}{\sqrt{LC}}$)   
ricorda $2 \pi$ al denominatore, sennò misuro $\omega$  

2- filtro LR  
trova freq di taglio, poi $f_{1/2}= \sqrt 3 f_t = \sqrt 3 \frac {R+r}{2 \pi L}$     
ma difficile, perchè qui nella formula di f di taglio appare R+r, ma r dipende dalla freq, e non la so determinare  
mentre prima apparivano solo C e L, quindi l'altro metodo è meglio  

**Misurare freq**

check what happens at that frequency (py plygging it in H(f)):  

if have a specific dephasing: use YX mode until i see the correc thing   
eg. resonance frequency. when have displacement=0  

if have max gain: use normal osscilloscope, check when Vout is max    
eg. quando voglio $f_{1/2}$

quindi con arduino viene meglio

**Come misurare resistenze**

1- fai un ciruito a due maglie, una di bias e una con la piccola oscillazione (il segnale vero)   

2- per dinamica:   
misura $v_d$ dell'oscillazione sul componente e $i_d$ sulla maglia non di bias (prendendo il drop su un resistore noto)    

3- per statica:  
misura $I_q$ nella maglia di bias e $V_q$ sul componente  

nota che $V(t)=V_q+v_d(t)$, con $V_q \gg v_d$   
quindi per avere $V_q$ guarda con oscilloscopio, neanche la vedo l'oscillazione  
per $v_d$ devo mettere filtro passa-alto nell'oscilloscopio
