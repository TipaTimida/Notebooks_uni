### How to do a measurement

**What i want to measure**

$\Omega$: resistance  
V: voltage  
A: current  

voltage and current might appear 2 twice, bc both have 2 options:  
--- DC  
$\sim$: AC  

**Probes / boccole**

1- black: GND (common)  
2- red: VCC (live, hot wire)  
3- green: ground wire  
si collega al GND, per rendere sicuro il contenitore degli apparecchi elettrici, c'è solo nelle prese a 3 buchi  

**How to measure stuff**

1- resistance: on de-energized components  
2- current: in series  
3- voltage:   
voltage drop pf something: in parallel

**Quale multimetro usare**

### Come usare

**Come scrivere il risultato**

1- scegli fondoscala    
2- leggi il valore  
3- metti la giusta unità di misura  
4- aggiungi le incertezze  

#### Digitale

**1. Scegliere fondoscala**

voglio leggere 3 cifre significative  

casi:  
ne lego di meno -> abbassa il fondoscala (esempio: leggo 0.03)  
leggo 1 -> alza il fondoscala  
vuol dire che il valore è talmente grande che non lo legge nemmeno  

**2. Leggi il valore**

scrivi esattamente ciò che leggi  
cioè scrivi SOLO le 3 cifre significative date, lasciando il punto dov'è  

**3. Unità di misura**

il nome del fondoscala mi dice anche l'unità di misura, cioè quale è la potenza della cifra subito prima del punto mobile  
eg. fondoscala 200kV ignifica che il punto segna i kV    

**4. Incertezza**

devo considerare 2 errori:  
quanto valgono di preciso lo trovo nel manuale del multimetro, nella colonna 'precisione'    

1- errore di digitalizzazione : carattere prevalentemente statistico ($\pm $ 1 digit)  
per noi equivale all'errore di lettura  
digit è l'ultima cifra che appare, cioè l'ultima cifra significativa  

2- errore di calibrazione: carattere sistematico  
un errore che il costruttore mi deve dire quanto vale, è scritto nel manuale del multimetro ($\pm 0.8% $ della lettura)  

$\pm 0.8% $ della lettura $\pm 1$ digit  

**Esempio**

uso fondoscala da $200k\Omega$  

1- leggo $3.30$  
2,3- devo scrivere: $3.30 \; k\Omega$  
4- dal manuale leggo: $\pm 0.8% $ della lettura $\pm 1$ digit   
cioè: $3.21 k\Omega * 0.08 + 0.01 = 0.03+0.01 = 0.04$

### Tips to use multimeter

**Caution**

don't let the probes touch each other while touching a circuit  
bc creates a short-circuit, can cause a spark or fire  

**Check meter fuse**

1- set to measure resistance mode  (Ω)  
2- touch the probes together  
3- what i get:  
$R=0$: good fuse  
$R=\infty=O.L.=1.$: blown fuse  

**Check wire continuity**

what happens if:  
probes touch each other: R=0  
probes touch nothing (the air) or touch an open circuit: $R = \infty$  
written O.L. for open loop

### Leggere resistenza dalle bande

**Bande sulle resistenze**

4 fasce colorate  
la prima è più vicina all'estremità (al filo)    
l'ultima è più distante dall'estremità, ed è di solito argento o oro   

cosa indicano:  
1,2: mantissa (cifre significative)  
3: potenza di 10  
4: tolleranza  

**Come scrivere il risultato**

1. lascia la tolleranza indicata come PERCENTUALE  
non calcolare quanto vale quella percentuale in questo caso   

2. usa solo le cifre significative indicate sulle bande  
per farlo, usa il giusto multiplo di $\Omega$    

**Esempio: letto dalle bande**

mantissa: 68    
potenza: 4  
tolleranza: 10%  
cioè: $68 \cdot 10^4$

posso, perchè ogni 3 ce n'è 1  
potrei scivere 68 o 6.8 o 0.68

come trovare quello giusto  
scrivo con tutte le cifre, con uno 0 in più all'inizio, e posiziono i multipli  
poi seleziono quello che appare in corrispondenza di una delle cifre o dello 0 subito prima

M__K__1  
qui devo scegliere M, che corrisponde allo 0 subito prima  

poi scrivo la tolleranza, lasciata in percentuale:  
$0.68 \; M\Omega \pm 10\%$

### Teoria

#### Voltmeter

**Come funziona**

dove metterlo:  
voglio misurare $\Delta V$, quindi devo collegarlo in modo che legga la tensione originale  
cioè in parallelo (perchè tutti i rami hanno la stessa tensione)

che resistenza interna:  
per non disturbare il circuito originale devo mettere $r_{V}=\infty$  
così nel ramo aggiuntivo non passa corrente, come se ci fosse un break nel nuovo ramo, così sembra il circuito originale  
$r_V$: resistenza interna / di input del voltmetro  


**Digitale e analogico**

multimetro digitale: $r_V>10 M\Omega$  
multimetro analogico: $r_V = 20 k\Omega/V_{Fs}$ (dipende dal fondoscala)  
oscilloscopio: $r_V=1 M\Omega$  

quindi quello digitale è il migliore  

#### Amperometer

**Come funziona**

dove metterlo:  
voglio misurare $i$, quindi devo collegarlo in modo che legga la corrente originale  
cioè in serie (perchè tutti i componenti sono sttraversati dalla stessa corrente)  

che resistenza interna:  
per non disturbare il circuito originale, serve $r_A=0$   
così non ci sono cadute di potenziale, il pezzo con l'amperometro sembra solo un pezzo di filo, come nel circuito originale    

**Digitale e analogico**

il manuale risporta il voltage drop causato dallo strumento, $\Delta V_{ins}$, a seconda del full-scale range $I_{fs}$   
posso usarlo per calcolare la resistenza interna: $r_A = \frac{\Delta V_{ins}}{I_{fs}}$  

multimetro digitale e analogico sono piò o meno uguali  
l'oscilloscopio non può misurare la corrente  

**Se cambio il fondoscala ???**

se cambio il fondoscala sul multimetro che legge A, cambia la sua resistenza interna  
quindi V letta dall'altro multimetro cambia pure  

aumento fondoscala dell'amperometro:    
diminuisce sua resistenza interna

#### Multimetro

**Analogico vs digitale**

analogico: ha reistenza interna bassa  
buono per misurare corrente  
chiamato galvanometro  

digitale: ha resistenza interna alta  
buono per misurare voltaggio  

**Problems**

analog:  
low internal resistance -> if connected in parallel, changes current in the circuit, so measure the wrong $\Delta V$   
because less current passes in the wire i want to measure (with respect to the hypothetical current that would pass there if there were no multimeter)  
so V=IR, if in reality I is smaller than estimated, then V as well is smaller than extimated  

digital:  
high itnernal resistance -> if connected in series, cause a voltage drop  

**Quale errore prevale**

in multimetro digitale, errore di calibrazione prevale  
in multimetro analogico, errori di  calibrazione e digitizzazione sono simili  

quindi, entrambe le incertezze sono NON STATISTICHE  

**Transducers**

strumenti in grado di convertire quantità elettriche in qualcosa di visibile a occhio (needle of a quadrant or digit on a screen)

**Analogico vs digitale**

Electric quantities of our interest are analog in nature (all of them, if we exclude effects due to the discrete nature of the electric charge that are too tiny to be detected with conventional tools)  
cioè ignoriamo che la carica sia quantizzata, tanto è meccanica quantistica  

analog device:  
would in principle appreciate infinitesimal (as
small as you like) variations of such quantities, obviously provided that sensitivity is large enough.  
to increase sensitivity, could use a longer needle, or a beam of light instead of a needle and refleci it with a mirror, so can see an even longer path   

digital device:  
will convert the analog quantity into an integer number, being therefore unable, by definition, to follow infinitesimal variations due to its finite resolution
