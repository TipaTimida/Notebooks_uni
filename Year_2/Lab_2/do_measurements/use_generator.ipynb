### Guida generatore DC

per accenderle, premere pulsante in basso a sinistra   
non posso usarlo come strumento di misura perchè il costruttore non ha indicato l'errore  

**Cambiare V o A erogate**

1- premi il pulsante voltage  
comincia a lampeggiare la cifra più a sinistra del voltaggio  
2- gira la manopola finchè è la cifra che voglio  
3- ripremi il pulsante voltage, comincia a lampeggaire la seconda cifra  
vai avanti così  

stessa cosa per la corrente  
ma è la corrente massima che può essere erogata, non si cambia mai  

l'altro alimentatore, più piccolo, che sembra un caricatore, può dare solo 5V, non posso cambiarlo  

#### Teoria

**Sketches**

generatore reale di voltaggio:  
V e R_in in serie  
per renderlo ideale, voglio R_in piccola  

generatore reale di corrente:  
V e R_in in PARALLELLO!!!  
per renderlo ideale, voglio R_in grande
