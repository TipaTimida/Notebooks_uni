**To smooth it**

la versione base del circuito dà un output molto a onde  
per avere un po' più dritto posso aggiungere un condensatore in parallelo alla resistenza  

per renderla ancora più piatta, posso aggiungere un altro elemento in parallelo agli altri due, un diodo zener  

**Altri aggiustamenti**

la corrente originale è AC, quindi grafico è sinusoidale, ha picchi  
se un picco è troppo alto, un diodo può bruciarsi  
per evitarlo, posso aggiungere un induttore in serie alla resistenza e al condensatore (nel filo orizzontale, prima che si separi nei loro due fili)  
l'induttore tiene la corrente più costante

### Measuring bridges

**What are**

ways to emasure C and L  
today, not used anymore  

**Structure**

alt text

**Config**

in the middle: voltmeter or amperometer  
there are 4 impedances:  
1- the unknown one i want to emasure  
2- a variable, calibrated one (so can change it, and know its value at each moment, there is scale on it)  
3,4- known ones (to select the full-scale range of the measurement, so to adjust the measurement sensitivity) (3 must be the same type as 1)

**How it works**

1- balance the bridge  
change the variable impedance until the voltemeter is the center to read $\Delta V=0$ (or the amperometer reads i=0)  

2- write formula for the bridge balanced to find the unknown impedance  
$Z_1Z_4 = Z_2 Z_3$  
it's the product of OPPOSITE impedances  

**Types**

1- Wien's bridge  
for: measuring R  
how: put all Z=R  

2- de Sauty's bridge  
for: measuring C  
how: 1 and 3 are C, 2 and 4 are R, also must use sinusoidal current from generator  

3- Maxell's bridge  
for: measuring L  
how: 1 is L, with its internal resistance, 2 is varaible R, 3 is R, 4 is parallel of R and variable C  
uso una C nota, non L nota, perchè due L vicini si perturbano per mutua induzione, meglio evitare (tipo trasformatore)

1- in DC:

use it to find $r$, the internal res of L  

2- in AC:  
$j \omega L_1R_3 +r_1R_3=R_3R_4 + j \omega...$  
parte reale a immaginaria devo essere essere uguali, separatamente

#### Notes

**Nota: impedenze variabili**

C variabile si crea con due settori semicircolari che posso ruotare, posti paralleli, così a seconda di quanta parte delle armature sono una davanti all'altra, cambia la C    
mentre R variabile si crea con un cerchio di materiale resistivo e una sbarra, in modo da fare due res in parallelo come voglio io  

**Find formula for bridge balanced**

want condition on impedances when bridge is balanced  
$\Delta V=V_D-V_C =R_3 I_D - R_1 I_C=0$   

scrivo tutto in funzione di una sola corrente, sosituendo:  
$(R_1+R_2)I_C=(R_3+R_4)I_D$  

ottengo: 
$Z_1Z_4 = Z_2 Z_3$  

**Nota on balance formula**

assume 4 resistances:  
$Z_1= Z_2 \frac{Z_3}{Z_4}$  

$Z_1$: unknown  
$Z_4$: variable  
$R_3/R_4$: enables to adjust the full-scale range, so can always be resistances
