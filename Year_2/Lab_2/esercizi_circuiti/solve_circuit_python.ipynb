### Explanations

**Solving a voltage divider**

1- write Z of all the components, C and R  

2- write the formula for combining them in series or parallel  

3- compute the amplitude of Z     

```python
# Example
import numpy as np

freqs=[100, 1000, 10000, 10**6]

for f in freqs:
    R = 10**6
    CA = 1/(1j * 2 * np.pi * f * 0.1e-6)   # 0.10 μF = 0.1e-6 F
    C0 = 1/(1j * 2 * np.pi * f * 25e-12)   # 25 pF  = 25e-12 F

    Z = CA + (R*C0)/(R+C0) 
    print(f'f = {f} Hz -> Z = {abs(Z)} Ω')
```

**How to deal with complex values**

write j in a formula: multiply the thing by `1j`, remeber the 1  
it's from numpy  

compute:  
amplitude: `abs(Z)`   
angle: `angle(Z)` 

**Using filters**

1- write the formula for the filter

2- write values of components (C, R, f)  

3- compute cut-off frequency of the filter  
and also omega of current, if used that in filter formula  

4- pass my frequency and the cut-off one to the filter  

5- on the filter output, compute gain and phase  

```python
# Example

def rc_highpass_H(omega, fc):
    """
    RC high-pass transfer function H(jw) = (j*w/wc) / (1 + j*w/wc)
    """
    wc = 2*np.pi*fc
    s = 1j * (omega / wc)
    return s / (1.0 + s)

f = 305
R = 3000
C = 0.1e-6

omega = 2 * np.pi * f
fc = 1/ ( 2 * np.pi * R *C )


H = rc_highpass_H(omega, fc)
gain = np.abs(H)**3
phase = np.angle(H)*3

print(gain)
print(phase)
```

```python
# Punto 11

freqs=[100, 1000, 10000, 10**6]

for f in freqs:

    # oscilloscopio
    R = 10**6
    CA = 1/(1j * 2 * np.pi * f * 0.1e-6)   # 0.10 μF = 0.1e-6 F
    C0 = 1/(1j * 2 * np.pi * f * 25e-12)   # 25 pF  = 25e-12 F
    Z_osc = CA + (R*C0)/(R+C0) 

    # circuito ora
    R1 = 1000
    C1 = 1/(1j * 2 * np.pi * f * 0.1e-6)   # 0.10 μF = 0.1e-6 F
    V0 = 10 

    # C || oscilloscopio
    Z_load = (C1*Z_osc)/(C1+Z_osc)
    V = V0 * Z_load /(R1+Z_load)

    print(f'f = {f} Hz -> V = {abs(V)}V')
```
