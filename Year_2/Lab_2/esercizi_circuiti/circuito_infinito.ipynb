**Situazione**

ho circuito SEMI-INFINITO maglie quadrate, numerate da 0 a n (con n infinito)      
considero sue impedenza generiche, $Z_1$ e $Z_2$ (rispettivamente, L e C)  

risolvere:  
poichè è semiinfinita, aggiugnere un'altra maglia a sinistra non cambia niente  
assuendo che il circuito abbia un'impedenza totale $Z_0$:   
$Z_1 + (Z_2 || Z_0) = Z_0$   
che ha soluzione: $Z_0 = \frac{z_1} 2 (\frac{Z_1^2}4 +Z_1Z_2)^{1/2}$  

plugga i valori di $Z_1$ e $Z_2$ a seconda di cosa sono, reisteori, condensatori o induttori  

**Costruirlo realmente**

per costruire un circuito che davvero si comporti così, mi basta terminarlo con un'impedenza $Z_0$
