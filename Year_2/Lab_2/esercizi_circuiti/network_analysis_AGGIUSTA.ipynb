### Superposition theorem

**What** 

1. replace all batteries except for one with a piece of wire (essentially, shortcircuit them)  
2. in this easier circuit, find current in each wire, pay attention to direction  
3. repeat steps 1 and 2, each time leaving a different battery  
4. for each wire, add all currents found using each battery (pay attention to sign)  

**Easier version**

if i only care about current or voltage drop on a spcific wire:  
in each repetition, instead of fully solving the circuit, i just need to find the value i'm interested in (current or voltage drop)
then i sum all the values from the different repetitions  

### Thevenin, Norton theorems

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRVkeDvlccRooCX78PCE5iidKx0wH2a7pSe3Q&s)

**What it means**

take a complex circuit, and rewrite it as:  
Thevenin: real voltage generator (V and R in series)  
Norton: real current generator (I and R in parallel)  

Thévenin: guarda la tensione quando non scorre corrente  
Norton: guarda la corrente quando non c’è tensione  

**When can use**

funziona solo se applicato a circuiti con componenti lineari  
no: componenti non lineari, come transistor  
ok: condensatori e induttori  

**Important**

Thevenin is not a way to solve a circuit  
it's a way to simplify a circuit before attempting to solve it

### KCL

**Risolvere con KCL: steps**

1. dà il nome ai nodi  
cioè a tutti i punti che separano due componenti  
se due punti, anche se sembrano nodi dal disegno, sono equipotenziali, sono uno stesso nodo, dagli un'unico nome  

2. per ogni generatore, scrivi il rapporto tra i due nodi  
ad esempio, se ho batteria da 5V e il lato + è verso il nodo A: $V_A -V_B = 5$  

3. uso KCL su un nodo o supernodo  

4. scelgo un GND (ground)  
cioè un nodo a cui dare $V_A=0$    
usa quello che appare di più nella KCL, o nelle relazioni tra nodi, così semplifico molto  

5. plugga info 2 e 4 nella KCL, dovrebbe essere equazione in funzione di un solo nodo  
risolvi per trovare V in quel nodo  

il valore trovato al punto 5 è valido come valore assoluto solo se il nodo scelto come GND lo è davvero  
mentre in generale, quel valore è la ddp tra tale nodo e il nodo scelto come GND (che ho settato a 0 per semplificare i conti)  

la ddp ha senso solo come differenza tra due punti, non mi interesserà mai il valore assoluto  
tipo, per calcolare una corrente, mi serve la ddp tra gli estremi del filo  

**3. Applicare KCL**

1- scegli un nodo $V_A$   
(possibilmente toccato da quante pi resistenze possibile???)  

2- per ogni suo nodo vicino (di primo grado, cioè in ogni direzione mi fermo al nodo più vicino):  
$\Sigma i_k =0 \quad \rightarrow \quad \Sigma \frac{V_{chosen}-V_{neighbor,k}}{R_k}$  
$R_k$ è la resistenza tra quei due nodi  
nella differenza, il nodo scelto viene sempre primo, così le correnti hanno il segno giusto col verso  

**3. Applicare KCL con SUPERNODO**

cos'è supernodo:  
2 o più nodi collegati da una (o più) batteria   

come fare:  
immagina di fonderli in un'unico nodo  
quindi ignora i fili con batteria che li uniscono, non metterli nella KCL  
usa tutti i vicini lontani di 1 per OGNI nodo (quindi qualche nodo potrebbe essere contato due volte, una per ognuno dei nodi del supernodo, se è un vicino di entrambi)  

perché: 
è come usarle la KCL su entrambi i nodi, e poi sommarle  
i termini del filo che li unisce si cancellano a vicenda, restano gli altri vicini  
anche se uso un supernodo, devo comunque mettere il rapporto tra questi due nodi al punto 2, e poi sostituirlo nella KCL   
così ho tutto in funzione di un solo nodo  

esempio, con supernodo A-D (A ha 3 vicini, D ne ha 1, ignorando il tratto AD):    
$\frac{V_A-V_B}{R_1} + \frac{V_A-V_C}{R_2} + \frac{V_A-V_C}{R_3} +\frac{V_D-V_C}{R_4}$  

**Tip: generatore di corrente**

se nella sommatoria $\Sigma i$  so il valore di una I  
perchè viene da un generatore costante di correte  
allora quell'addendo lo lascio come I, non ah senso sostituire come V/R   
attenta al segno della correntte  

perché:  
un altro modo di interpretarlo è di scrivere la KCL con tutte le frazioni come al solito  
e al punto 2, oltre alle relazione tra nodi dei generatori di tensione, metto anche la relazione tra nodi dei generatori di corrente  
tipo: $V_D -V_A = I_{gen} R_1$  
dove tra D e A ci sono solo il generatore di corrente e la resistenza 1  
e poi pluggo questa cosa nel numeratore della frazione, e mi rimane $I_{gen}$ 

se invece il generatore non mi dà una corrente che arriva al supernodo, devo aggiungere la sua condizione alle condizioni sui geenratori di voltaggio al punto 2:  
generatore di tensione: $V_A-V_B = V_{gen}$    
generatore di corrente: $V_A-V_B = I_{gen}R$  

**Verso corrente**

$V_B > V_A$  
allora corrente scorre da B verso A (immagina sia protoni, scorre erso il punto con carica -)

### $\Delta$ - Y conversion

![alt text](https://www.allaboutcircuits.com/uploads/articles/three-terminal-network-diagram.jpg)

**Convert resistances between $\Delta$ and Y**

from $\Delta$ to $Y$:  
$R_x = \frac{\Sigma R_{xi}}{\Sigma R_{ij}}$  
sopra: somma tutte le resistenze che toccano quel nodo in $\Delta$  
sotto: somma tutte le resistenze in $\Delta$   

from $Y$ to $\Delta$:  
$R_{xy}= \frac{\Sigma R_iR_j}{R_z}$  
sopra: somma tutti i doppi prodotti in Y  
sotto: la resistaneza con la lattera che non appare in xy
