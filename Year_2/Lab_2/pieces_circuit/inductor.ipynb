### Behaviour

**Cosa fa**

si oppone alla variazione di corrente nel circuito  
quindi quando la corrente fovrebbe cambiare, rallenta questa variazione  
non cambia a quale I finale arrivo, ma la raggiunge più lentamente  

**Come lo fa**

quando scorre I nell'induttore, si crea un campo magnetico indotto B (assiale)  

3- se I resta costante: 
B crea una corrente indotta in direzione opposta a I originale  
quindi I attraverso L diminuisce, c'è un drop di potenziale  

4- se I cambia:  
allora B cambia, si crea un voltaggio che si oppone alla variazione di B (lgge di Lentz)  
cioè una corrente indotta per compensare la variazione della I originale  
così si oppone al cambiamento di I  

#### Entra-voltage, extra-current

**Cosa sono**

quando apro o chiudo il circuito, la corrente:  
1- in teoria: passa istantaneamente da 0 ad un valore finito  
2- in pratica: ci arriva pian piano (a causa di L che si oppone al cambio di corrente), quindi il circuito è temporaneamente in uno stato in cui succedono cose strane  

per evitarle, devo abbassare l'ampiezza dell'onda del generatore???  

**Formalmente**

$\mathcal{E}=-L\dot{I}$  

quindi quando I passa da 0 ad un valore finito, la sua derivata è infinita  
quindi $\mathcal{E}$ è divergente, succedono cose strane  

| situazione                    | cosa succede                  | perché                            |
| ----------------------------- | ----------------------------- | --------------------------------- |
| apri circuito con induttore   | **tensione enorme**           | corrente non vuole fermarsi       |
| chiudi circuito con induttore | **corrente grande richiesta** | corrente non vuole partire subito |


**Extra voltage**

1- quando: apro il circuito (cioè rompo il filo)  

2- in teoria: da i=finita a i=0  
perchè in circuito aperto non scorre corrente  

3- problema: la corrente non può andare a 0 di botto, sennè $\dot I =\infty$, e quindi servirebbe $\epsilon$ infinita  
la l'induttore cerca dia vvicinarsi a $\epsilon$ infinita, quindi genera molta tensione, cerca di far continuare a scorrere la corrente  
QUINDI PER UN PO' LA CORRENTE CONTINUA A SCORRERE UN PO', NON VA IMMEDIATAMANTE A 0  

4- conseguenza:  
se il circuito è aperto, la tensione cresce molto  
finché l’aria si ionizza → scintilla  

5- esempio: quando spegni: un motore, un relè, un contatto elettrico, spesso si vede una scintilla.

**Extra-current**

1- quando: chiudo il circuito 

2- in teoria: da i=0 a i=finita   
perchè quando chiudo il circuito, inizia a scorrere corrente  

3- problema: l’induttore si oppone al cambiamento, perchè $V_{gen}-L \dot I =$  
quindi se all'improvviso l'induttore riceve una $V_{gen}$, la corrente I aumenta lentamente     
l’induttore richiede molta corrente dalla rete

4- esempio: quando accendi aspirapolvere o un motore  
per un attimo succede:  
il motore richiede molta corrente  
la rete non riesce a darla subito  
le lampadine si abbassano di luminosità  

5- dove è corrente extra  
in realtà non c'è corrente extra (cioè corrente è sempre minore del valore finale di regime)  
ma c'è potenza extra richiesta alla batteria per costruire B all'interno dell'induttore  
ho circuito RL: potenza sui componenti sono $P_{gen}=V_0I$, $O_R = RI^2$, $P_L = V_L I$  
quindi $P_L$ è maggiore quando I cambia molto, cioè appena acceso, mentre a regime (quando I fissa), L non riceve più potenza  
cioè quella potenza è usata da L per costruire il campo B al suo interno per opporsi alla variazione di I  
è quella potenza che viene richiesta all batteria, che fa sì che richieda un po' di tempo a partire

**Spiegazione: ferromagentico**

formule per calcolare L:  
$\Phi(B)=LI$   
$ B = \mu H$  
$\int H \, dl = \mu_0 I$  

spiegazione:  
per una I fissa, produco una H fissa  
quindi il materiale dentro cambia la B prodotta, quindi il flusso di B che si concatena con il componente

più energia dissipata vuol dire più smorzamento, quindi $\tau$ diminusice

### Oscillator with diode 

**Schematic**

generator -> diode -> (C || (L+R)) -> GND  

the generator creates a square waveform, with:  
1- horizontal lines at $+V_G$ and $-V_G$  
2- $T_G  \gg T = \frac {2 \pi}\omega$ (cosa è $\omega$??? qualeche var del circuito RLC???)  

**How it works**

1- positive halfwave (of the square wave)  
diode is in forward, lets it pass (but causes a bit of drop across it, so it's $V_0 \le V_G$)  
quindi i due rami con C e (L+R) è come se avessero un generatore  
ramo con C: il condesantore si carica, come al solito, fino a condizione stazionaria $Q_0=CV_0$    
ramo con L+R: l’induttore si oppone ai cambiamenti di corrente, quindi I cresce lentamente  
dopo un po', regime stazionario: C è un aperto e L un chiuso (cioè C ha finito di caricarsi, L si è opposto quanto potevo, ma la corrente è arrivat al massimo)  

2- negative halfwave  
diode is in reverse, generator voltage non passa più   
ora ho circuito RLC senza generatore, che si comporta come damped oscillator

**Come sfasano C e L**

hanno come impedenza -j e +j

**Effetto pelle**

cosa:  
R di un filo aumenta all'aumentare di f   

perchè:  
filo cilindrico di raggio a, percorso da corrente J  
assumo J sia uniforme nel filo, ma non costante, varia nel tempo  
J produce un campo B tangenziale (è un filo attraversato da corrente), anche questo che varia nel tempo  
calcolo la variazione del flusso di B: $\frac{d \Phi(B)}{dt}$   
se B varia, si genera un controcampo indotto che gli si oppone, che dà luogo a delle correnti nel cilindro, J_ind (prodotte da induzione magnetica)  
disegna J_ind: nella regione esterna è concorde con la corrente J, mentre nella regione interna ha verso opposto a J    
quindi la corrente effettiva, che è la somma di J e J_ind, è maggiore al bordo, e tende a 0 all'interno  

J non è uniforme su tutta la sezione del filo, è maggiore alla circonferenza esterna  

formula:   
|J| aumenta esponenzialmente in funzione di r, e raggiunge il massimo sul bordo del filo, r=R  

$\delta_{skin}=\sqrt {\frac{2 \rho}{\omega \mu_0 \mu_r}}$  S
profondità effetto pelle  
ovviamente, vedo che l'effetto pelle si veririca solo se $\omega \neq 0$ (sennò le correnti indotte non si generano)   

conseguenza: se corrente circola solo nella parte esterna, la resistenza del filo cambia, perchè cambia la sezione del resistore, e $R=\rho \frac{h}{S}$  
se f aumenta, S diminuisce, quindi R aumenta  
se il raggio del filo è abbastanza piccolo, ho sempre $r < \delta_{pelle}$, quindi l'effetto pelle non si vede

### Mutua induzione

**How to model it**

if there are 2 inductors, on each one there is voltage drop:    
$V_1 = j \omega L_1I_1 +j \omega M I_2$  

example: equation of 2 coupled circuits RL, only the primary one with generator    
$\begin{cases} 
V_1 = (R_1+j \omega L_1)I_1 +j \omega M I_2 \\
V_2 = (R_2+j \omega L_2)I_2 +j \omega M I_1
\end{cases}$

other way to model mutual induction:    
$V_1 = (R'+j \omega L')I_1\omega$

#### Circuito magnetico

**Tipi di circuito**

1- elettrico  
J è confinata nel circuito, e il suo flusso è costante  
fatto di materiale conduttore  

2- magnetico  
B è confinata nel circuito, e il suo flusso è costante  
fatto di materiale ferromagnetico     esempio: trasformatore  

**Circuito magnetico**

è un materiale ferromagnetico a forma di tubo che si richiude su se stesso  
cioè un circuito che, invece di essere fatto di un filo di rame, è fatto di un filo di materiale magentico (chiamato 'tubo di flusso')  

è una linea di campo B, perchè essendo ferromagnetico, se c'è B all'interno lo segue  

**Semplificazioni**

se materiale omogeneo e di sezione S uniforme e raggio del filo piccolo rispetto al raggio dell'anello (il circuito ha un forma tonda)  

per semplificare come si scrive:  
flusso del campo B in una qualunque sezione del tubo è uniforme  
in circuito elettrico, è $I=\Phi(J)$ che rimane corrente, perchè non ci sono sorgenti nè pozzi per le cariche di corrente

**Hopkins law**

ho un circuito magnetico a forma di anello, lungo l  

nel circuito magnetico, ci avvolgono intorno una bobina di N spire, con corrente I  
uso: $\nabla \times H = J$ (assumo non c'è D)  
$\int_\gamma H \cdot dl = I_{conc}$  
come $\gamma$, scelgo il percorso attraverso l'anello (lungo il circuito)  
quindi $I_{conc}$ è la corrente che attraversa il circuito, che è NI    
quindi: $Hl = NI$  
ora calcolo: $B=\mu H$  

inoltre:  
$B= \frac{\Phi_S(B)}{S}$  
posso scrivere: $\Phi(B) \frac{l}{\mu S}=NI$   
$\frac{l}{\mu S}$: riluttanza R (perchè dipende solo dalla geometria del problema) (è simile alla resistenza, in elettrico)    

legge di Hopkinson:  
$\Phi(B) \, R = \mathcal{V}$  
simile alla legge di Ohm  
metto flusso di B perchè prima c'era I, il flusso di J, che è la cosa che si conserva nel circuito elettrico   
quindi serie e parallelo di riluttanze è come serie e parallelo di resistenze, e posso usare le legge di Kirchoff (scambiando I con flusso di J)   

analogia formale:  
cioè se ho un circuito magentico, posso sostituirlo con uno elettrico:  
le bobine avolte intorno sono i generatori  
ogni pezzo di circuito è una resistenza  
risolvi il circuito  

**Traferro**

è un circuito magnetico con un buco lungo il circuito (il buco è come una resistenza con una R diversa da quella lungo il circuito)  


a cosa serve: ad avere accesso al campo B dentro il materiale (perchè nel buco B è uguale a dentro il materiale, perchè componente normale di B si conserva)   
ad esempio, per fare la risonanza magnetica, è il buco dove si mette il corpo della persona   

**Nota**

quando scrivo: $L=\frac{ \Phi(B)}{I}$, Phi è il flusso concatenato con l'avvolgimento (con tutte le spire)  
mentre sopra, come $\Phi$, ho usato il flusso concatenato con una SOLA SPIRA  
cioè $\Phi(B)=N \Phi$  

quindi: $L=\frac{N^2}{R}$
