### Real vs ideal components

**Cosa vuol dire**

tutti i pezzi (eccetto le resistenze), oltre ad essere sè stessi, hanno:  
1- resistenza interna  
2- capacità spuria  

di solito posso assumere:  
capacità: sempre trascurabile  
resistenza: trascurabile solo se nel circuito c'è una resistenza molto più grande  

**Come modellizzare**

ho un generico pezzo Z   
capacità spuria: in parallelo a Z  
res interna: in serie al parallelo di C e Z  

esempio: induttore  
ideale: L  
reale: (L || C) + r  
più usato: L + r  

**Note**

C e r sono difficili da modellizzare perchè NON sono indipendenti dalla frequenza  
nei fit assumo che lo siano per semplicità, ma non è vero  

#### Spiegazioni fisiche

**Perchè appare C spuria**

appare quando nella struttura del componente sembra di vedere un condensatore (due conduttori vicini e in mezzp un dielettrico)  

1- diodo  
reverse biased: plates = p and n regions, dielectric = depletion layer  
forward biased: due to stored charge carriers in the junction ???  

2- transistor  
ogni coppia di piastre (BC, BC, CE) somiglia ad un condensatore, come nel diodo  

3- induttore  
piastre: two adjacent turns of the coil  
dielectric: the air in between   
the turns are really close, quindi anche se l'aria non conduce molto, è abbastanza per avere una C abbastanza grande  

**Self-resonance**

poichè l'induttore ha anche una C, è come se avessi sempre un circuito LC, quindi può avere risonanza  

### Generator

**Model**

non ha C spuria, ma solo R interna  

reale: gen + r    
ideale: gen  

**Conseguenze: voltaggio in output**

un po' di voltaggio è perso nel voltage drop su r, quindi dà in output un voltaggio più basso che previsto   
se dopo ho un'altra resistenza (un load), ho un partitore di tensione  
quindi la vera V data in output (cioè misura dopo r interna) sarebbe:  
$V_{out}=V_{gen}\frac{R_L}{R_L+r_{int}}$    

**Problema: onda quadra distorta**

perchè ci sono:  
r: resistenza interna del generatore  
C: capacità spuria dell'oscilloscopio (due to the probes)  

quindi ho un crcuito RC  
è un passa-basso (perchè il mio output è all'oscilloscopio, che è vicino a C)   
quindi le parti verticali dell'onda quadra vengono distorte, un po' arrotondate  

nota: stesso problema se invece dell'oscilloscopio uso oggetti con capacità spuria, tipo un diodo o transistor

### Impedance mismatch  

**What happens**

When a signal travels along a cable, it behaves like a wave.  
If the wave reaches a load with the same impedance, the energy is absorbed.   
But if the load impedance is different, part of the wave reflects back.

reflection coefficient:  
$Γ=\frac{Z_L−Z_0}{Z_L+Z_0}$  
if $\Gamma \sim 1$, almost the entire signal reflects  

**When happens**

when the output impedance of the source, the impedance of the transmission line (cable), and the load impedance are not equal  

Example in many labs:  
function generator output impedance → 50 Ω  
coaxial cable characteristic impedance → 50 Ω  
oscilloscope input → 1 MΩ  

here, the problem is the oscilloscope  
so, must put a 50Ohm resistance in parallel to the 1MOhm one, so the 50 one wins, and all is well  
but then, isn't the input resistance too low, and the maglie non indipendenti???  

**At high frequency**

easier to think about wavelength, than frequency: $v = f \lambda$  

at 1kHz, or 200km:   
A 1 m cable is negligible compared with the wavelength  
So the signal behaves like a simple circuit, not a wave → reflections don’t matter  

at 100 MHz, or 2m:    
Now a 1 m cable is half a wavelength.  
The signal behaves like a transmission line, and reflections strongly affect the waveform.

 


**Guida d'onda**

Una guida d’onda è una struttura fisica che serve a guidare la propagazione delle onde elettromagnetiche lungo una direzione, confinandole nello spazio.   
Una guida d’onda è una struttura fisica che confinando i campi elettromagnetici permette a un’onda di propagarsi lungo una direzione controllata.  

esempio:  
La forma più semplice è un tubo metallico cavo (spesso rettangolare).   
Le pareti metalliche impongono condizioni sui campi elettrici e magnetici, quindi l’onda resta confinata   
Questo è usato molto nelle microonde.  

La fibra ottica è una guida d’onda per la luce.  
Qui il confinamento avviene tramite riflessione totale interna tra:  
core   
cladding.
