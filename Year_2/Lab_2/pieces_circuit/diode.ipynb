### Behaviour

**Characteristic curve**

**Terminology**

forward/reverse current: names of current in forward/reverse bias  

$V_b \sim -80V$: breakdown voltage   
$V_t \sim 0.7$: threashold voltage, built-in voltage  
potential difference above which current intensity gets “remarkable”  
it's NOT the max voltage the diode can withstand  

| Regime                 | Bias             | Behavior of I                     |
| ---------------------- | ---------------- | ---------------------------- |
| Breakdown          | large negative V, $V < V_b$ | avalanche / Zener conduction |
| Reverse leakage   | small negative V, $V > V_b$ | small, $I \sim I_S$, don't depends on voltage     |
| cut-off regime / interdizione | $0 < V < V_t$ | nearly null  
| Forward conduction / conduzione | $V > V_t$   | exponential, Shockley, $I \propto e^{V/(nV_T)}$  |


**Saturation**

unlike BJT, the diode has no saturation regime  
bc saturation regime means a setting of combination of biases that causes saturation    
here there is only one bias, can only change its strenght, so it's not a regime technically  
but Shockley is an exponential with saturation, so for really big I it can get saturated

### Model

**Equation of the char curve: Shockley Equation**

$I = I_s (e^{\Delta V /(\eta V_T)}-1)$  

**Note on Shocley Equation**

1- works for any PN junction, so it's the same as the transistor one 

2- se plottato, questo dà il grafico senza la parte di breakdown, finirebbe asintoticamente a $I_S$ per tutti i V negativi    

3- durante il regime di conduzione, posso assumere $\Delta V \gg \eta V_T$  
quindi la posso semplificare come: $I =I_s \; e^{V/(\eta V_T)}$  
cioè tolgo il -1, tanto l'esponenziale ha un argomento grande, quindi è molto maggiore di 1  

**Cosa sono i valori**

$I_S$: corrente di saturazione inversa (after this, it saturates???)  
$\Delta V$: applied voltage across diode  
$\eta$: ideality factor, (coefficiente di idealità) if ideal put 1 (depends on the junction material, geometry, and size)  
nella pratica, circa 2    
$V_T = \frac{kT}{e}$: thermal voltage, ddp termica, circa 25-26 mV  

**Nota: dipendenza di I dalla temperatura**

in direct bias:  
se T aumenta, I aumenta  
dall'eq. di Shocley sembra che dovrebbe essere il contrario, perchè $T$ è all'interno di $\eta$  

ma c'è $I_S$ è dovuta ai portatori di carica minoritari  
e, se aumento T, anche $I_S$ aumenta  
$I_S$ aumenta più di $\eta$, quindi in totale I aumenta  

se aumento T, aumenta l'agitazione termica, quindi aumenta la probabilità di avere processi di pair generation, quindi aumenta il numero di cariche minoritarie e maggioritarie  
ma le cariche maggioritarie sono già tante, non si vede mlto l'aumento, mentre su quelle minoritarie si vede bene   

### Reverse breakdown

**Definition**

general phenomenon where a diode suddenly conducts a lot of current under reverse bias, when passing a critical voltage (the breakdown voltage)  
lo stesso del fulmine  

can be caused by 2 different physical mechanisms:  
1- Zener effect (at low voltage)  
2- Avalanche effect (at high voltage)   

**Zener effect**

what is it:  
quantum mechanical tunneling of electrons across a very thin depletion region  
(field-induced quantum tunneling across a reverse-biased junction)  

happens when:  
doping is very high  
depletion region is very narrow  
electric field is very strong  

why:  
The strong field bends the bands enough that electrons tunnel directly from valence band of p-side to conduction band of n-side  

**Avalanche effect**

what is it:  
classical (not quantum) effect  
due to impact ionization: a minority carrier, accelerated by the field, collides with lattice and knosk loose another electron, then chain reaction  

What | Zener | Avalanche
---|---|---
cause | quantum tunneling | impact ionization
doping | high | low/moderate  
depletion region width | narrow | wide
breakdown voltage | low, 2-6V | high, 6-1000V  
temp coefficient | negative | positive  

**Zener diode**

a diode intentionally designed to operate safely in reverse breakdown  
it's made so the reverse breakdown happens earlier, maybe even at positive voltages  

how it works:  
$V \le V_z$: open circuit (so blocks current???)  
$V \ge V_z$: mantiene ai suoi capi la tensione $V_z$, called the Zener voltage  
used to provide a constant oltage  

**A che serve un diodo?**

a mantenere un riferimento per il GND  
cioè un componente che mantiene la stessa $\Delta V$, qualunque sia la corrente che lo attraversa  

### Practical applications

**How diode works**

essentially, diode applies max(0, V)   
so only lets pass when voltage has positive sign, the positive half-wave  
ampliude drops a bit even here, bc of potential drop across diode  

**Rectifier**

gen -> diode -> circuit  

gen: square waveform of HIGH=5, LOW=-5  
diode: square waveform of HIGH=5, LOW=0  
so it's for trasforming an alternate wave in a NON alternate one  
it's easier than setting the amplitude to make the low stay exactly at V=0  

usage example:  
in RLC oscillator, want to have HIGH to set conditions, and LOW=0 to delete the generator and leave it to obscillate  
so hide the generator behind a diode  

#### Half-wave rectifier (raddrizzatore a singola semionda), livellatore

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRHjsUu8FGZAxe5ygxNigS5PvLAa0kcMPXcow&s)

**What**

have alternate (sinusoidal) waverform  
want (quasi-)continuous voltage  

**Circuit**

gen -> diode -> (R || C) -> GND  

1- generator  
produce alternate sinusoidal waveform  

2- diode   
change into non-alternate sinusoidal waveform, always non-negative    
translates alternate waveform in non-alternate periodical signal  
essentially, apply max(0, V)   

2- (R || C):  
R is the load, is only there to take the voltage across something (second stage)  
so output is taken in parallel to R  
C is a decoupling capacitor, use as a smoothing capacitor, to reduce ripples -> produce quasi-continuous voltage  

**Ripple**

signal is not perfecly rectified  
'ripple' ia the figure of merit for a power supply, want it as small as possible   
$\Delta V_{ripple} = V_{max}-V_{min}$  

why it appears:  
happen during the negative halfwave, when C fa in tempo a scaricarsi un po'   
so V taken across C decreases a bit, don't stay stable  

how to minimize it:  
1- reduce period of the waveform (increase f), so negative halfwave lasts less, C si scarica meno  
2- increase $\tau$ (so increase R or C), so decay of Q is slower, so in same time discharges less   

**Examine regimes**

conduction regime (positive half wave):   
gen + r_g -> C -> GND  
diode is useles here, does nothing  
all current passes through C, ignore wire with R, (in a parallel, wins the lowest resistance)  
so $\tau_{charge} = r_g C$   

cut-off regime (negative half wave):    
C -> R -> GND  
the diode blocks the wave, so generator is canceled  
so only have a circuit with C and R   
now: $\tau_{discharge}=RC$  

usually, $R_{eff} \ll r_g$  
so: $\tau_{dis} > \tau_{ch}$   
the biggest is $\tau$, the more time it needs to do that action (charge or decharging)

### Resistenza interna

**Problema**

come funziona un multimetro:  
non misura la resistenza direttamente, ma inietta una corrente nota I, misura la V (o il contrario)  
così ricava R=V/I  

problema:  
quanto è grande I di test dipende dal fondoscala  
fondoscala alto → I_test bassa (se la tenessa alta, non riuscirebbe a misurare la V)  
in pratica, la corrente di test stabilisce il working point del diodo  
e quindi la resistenza a quel punto   

conseguenze:  
I_test alta -> working point nella parte di conduzione del grafico -> conduce bene, R bassa  
I_test bassa -> working point nella parte di interdizione -> non conduce, R alta  

**Soluzione**

per misurare la resistenza dinamica, fai un ciruito a due maglie  
una per settare working point, altra per dare piccola oscillazione  
così posso calcolare resistenza dinamica ad un punto noto
