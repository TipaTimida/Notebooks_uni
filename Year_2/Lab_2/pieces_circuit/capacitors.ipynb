### Struttura

**Tipi**

poliestere: non polarizzati, non hanno verso  
elettrolitici: polarizzati, hanno verso  

**Come sapere il verso**

guarda le basi circolari del cilindro  
1- se sono uguali: poliestere  
2- se sono diverse: elettrolitico  
nera, di plastica: nodo +  
lucida, di metallo: nodo -  

**Quale usare**

poliestere:  
sono più costosi  
funzionano se si inverte la polarità  

elettrolitici:  
sono più economici, possono raggiugnere capacità più alte  
ma NON si possono usare se si inverte il segno della polarità (se la tensione cambia segno), perchè è polarizzato   

**Cosa vuol dire invertire la polarità**

NON significa solo usare corrente in AC  
ma usare corrente in AC che abbia valori sia positivi che negativi  
tipo, se fosse sinusoidale ma con un offset così alto da essere sempre V>0 (quindi non alternata), allora potrei usare quello elettrolitico  

esempio: livellatore a mezza semi-onda (in guida diodo)  
generatore usa corrente in AC  
ma dopo il diodo, al condensatore arriva un'onda con sempre V positiva, non cambia mai polarità  
quindi posso usare anche condensatore polarizzto  

esempio: studio resistenza dinamica diodo (esercitazione 9)  
il condensatore è in serie ad una sorgente alternata, quindi riceve V di entrambe le polarità  
quindi devo usare per forza quello non polarizzato  


### Types

**Names**

capacitors are assigned names according to what job they are performing  
so not according to topology, that is a consequence (to make it to do a job, must put it in a specific point)  
a seconda se studio time o frequency domain, il nome potrebbe cambiare  

coupling capacitor: condensatore di DISACCOPPIAMENTO (block DC)  
names are confusing  

**Recap**

Name | Job | Topology 
---|---|---
Coupling | block DC between stages | series with signal path
Decoupling | stabilize power rail | parallel to power rail (to GND)  
Filter | Shape signal bandwidth | parallel to signal path (to GND)

**Mnemonic trick: IMPORTANTE!!!**

C ha come impedenza: $Z_c = \frac{1}{j \omega C}$  
alta f: Z=0, è corto  
bassa f: Z=infinito, è un aperto  

quindi:  
C fa passare solo alte frequenze e blocca le basse  
mettilo nel percorso che vuoi che facciano le alte frequenze  

se voglio bloccare le alte, mandalo al GND  
se voglio far passare solo le alte (bloccare il DC), mandalo al prossimo pezzo del circuito  

ricorda nomi:  
decoupling: separa pezzi di circuito (de-), impedisci che il rumore (AC) si propaghi  
coupling: l'altro

**Behaviour a frequenze diverse**

2 modi di spiegarlo:  

1- matematicamente: con l'impedenza  
low f: $Z_C \sim \infty$, è un aperto  
high frequencies: $Z_C \sim 0$, è un chiuso  

2- fisicamente:  
low T (high f): T small means the capacitor has not enough time to charge, so its Q stay small, so its V does too    
so there is a lot of current  
high T (low f): there is enough time for the capacitor to charge, so at one point its voltage is the same as the battery, so there is no current thorught the circuit anymore  


**Do not confuse**

accoppiamento nell'oscilloscopio:  
AC coupling: block DC  
means use a coupling capacitor  

DC coupling: do not block anything, show everything, both DC and AC  
so do not use any capacitor, do NOT mean to use a decoupling capacitor

### Decoupling capacitor

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/_1024x534_crop_center-center_82_line/Decoupling-Capacitor.png.webp) 

**Topology**

VCC -> C || load  
in parallel in parallel with signal path (power source and load)  

**Purposes**

1- local electrical energy reservoir, to oppose changes of voltage    
2- filter AC noise  

**Local electrical energy reservoir**

what:  
provides energy to keep the voltage stable whenever the input voltage suddenly drops (nattery is switched off)    
so provides a stable supply to a circuit or a component  

eg. smoothen ripples   
when used for this function, also called 'SMOOTHING' or 'reservoir' capacitor  
shows its work in TIME domain  
VCC -> IC || (C + GND)  

**Filter AC noise**

how:  
Z is lower for high frequency  
so high freq pass through the capacitor and get routed to GND     

what:  
protect both the circuit from the electrical noise of the power supply, and the power source from electrical noise generated within the circuit (decouples the two, they don't share noise)  
This way, the circuit or the component being supplied is accepting only a pure DC signal  
preventing one circuit block from affecting another, isolates power-supply noise  
when used for this function, also aclled 'BYPASS' capacitor  

**Note**

Commonly, two capacitors are placed in parallel to act as decoupling capacitors  
One is a smaller value and the other is a larger one  
The larger one stores most of the energy in the circuit and filters the lower frequency noise
