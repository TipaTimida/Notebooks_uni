### Definizione

#### Cos'è

**Risonanza**

generale: the condition in which energy is transferred and stored back-and-forth between two forms with maximum efficiency   

in LC crcuit: condizione che può verificarsi dove sono presenti sia un effetto capacitivo, che un effetto induttivo, quando le due impedenze hanno lo stesso valore ($Z_C=Z_L$)

**Non confondere**

Cosa | che succede | formula |
---|---|---
$\omega_0$: freq naturale/propria di risonanza | massimizzato trasferimento di energia, è freq dell'oscillatore LC ideale, cioè se non ci fosse resistenza/damping  | $\omega_0= \frac{1}{\sqrt{LC}}$  
$\omega$: freq delle oscillazioni libere smorzare | quando ho circuito RLC | $\omega = \sqrt{\omega^2_0 -\gamma^2}$  
$\omega_{peak}$: frequenza di picco in regime forzato | il guadagno è massimo | $\omega_{peak} = \sqrt{\omega^2_0 -2 \gamma^2}$  

**Non confondere: note**

se $V_{out}$ è preso across C o L (qualcosa che dipende da f): $\omega_{peak} \neq \omega_0$  
se è preso across R: sono uguali   
scrivi T per vederlo, prova a massimizzarlo, se ho preso Vout su T il modo di massimizzarlo è togliere la parte immaginaria di Z  

#### A cosa serve

**Uso**

rispondere in modo selettivo su ogni frequenza    cioè amplificare (relativamente) una frequenza rispetto alle altre  
è più semplice graficamente, guardando il bode plot del gain 

**Risposta selettiva**
 
1- uso: seleziono una frequenza tra tante  
esempio: radio  

2- uso: filtro il rumore fuori banda  
usempio: ho segnale di f nota, ma con molto rumore  
lo ricevo con un circuito RLC con f come frequenza di risonanza, così amplifico solo quella    

3- uso: a ottenere una sinusoide da un'onda EM quadra o triangolare  
per farlo, devo togliere le armoniche alte, cioè selezionare solo la frequnza più bassa  

nota: basterebbe LC, ma nella realtà un po' di R c'è sempre, anche solo sui fili   

**Come funziona la selezione di una frequenza**

disegna il grafico di G-f del circuito risonante RLC    

vedi che il gain è:  
$G \sim 1$ su $f_0$, quella risonanante   
$G \ll 1$ sulle altre  
perchè il grafico ha un picco a $f_0$  
  
quindi in pratica mi rimane solo $f_0$  
ho ottenuto una sinusoide

$\omega L = \frac1{\omega C}$

#### Formule

**Descrivi risonanza**

cosa | formula | def
---|---|---
banda passante, bandwidth | $\Delta \omega = \frac{R}{L} \sim \frac 1 \tau$  | quante frequenze considero come "passanti"; cioè con un G abbastanza alto  
fattore di qualità/merito della risonanza | $Q = \frac{\omega_0}{\Delta \omega} =\frac 1R \sqrt{\frac LC}$ | quanto bene il sistema conserva energia rispetto a quanto la perde

se invece uso come larghezza la FWHM (quella del lab), allora tutte le formule (Q, BW) hanno un fattore $\sqrt{3}$, sempre al numeratore (perchè???)

**Cosa dice Q**

1- def:  
quanto bene il sistema conserva energia rispetto a quanto la perde  
$Q = \frac{energy \; stored}{energy \; dissipated} \; per \; cycle$  
cioè Q è maggiore se l'oscillatore è poco smorzato   

2- forma del picco di risonanza  
Q grande → risonanza molto selettiva      
Q piccolo → risonanza larga   

3- nell'oscilloscopio RLC, Q è il numero di pseudoperiodi che si vedono durante l'oscillazione, prima che diventano invisibili  

**Trova formula di Q**

calcolo le due energie nel momento del ciclo dove Q=0, quindi dove I è massima   
quindi l'energia totale è solo la parte magnetica, $U_{stored}=\frac12 L I_{max}^2$  
sempre ipotesi che sia molto sottosmorzato, così energia è effettivamente costante in un periodo  
trascuro irraggiamenti, perchè le frequenze sono troppo basse ???  

energia dissipata = potenza dissipata x periodo   
$U_{diss} = \frac 12 RI^2 \; T$   

quindi: $Q = 2 \pi \frac{U_{stored}}{U_{diss}} = \omega \frac LR = \omega \frac \tau 2 =\omega \tau_u$  


### Spiegazione: fasori

**Spiegazione con fasori: L e C in serie**

esamino i loro drop: $V_L=Z_L I$ e $V_C=Z_C I$  

1- fase:   
I è uguale in trambi, perchè sono in serie  
$Z_L$ e $Z_C$ sono fasori verticali, verso su e giù, sfasano di +90 e -90   
quindi $V_L$ e $V_C$ sono sfasati di 180, sempre  

2- modulo  
c'è una frequenza omega per cui il loro modulo è uguale  
quindi a questa omega, $V_L=-V_C$    
cioè ognuna ha un drop di potenziale, ma in totale il ramo non ne ha, perchè queste differenze di potenziale sono sfasate di pi  

**Spiegazione con fasori: L e C in parallelo**

adesso V è in comune, mentre le I sono diverse  
$V=Z_C I_C$ e $V=Z_L I_L$ 

1- fase  
le I sono sfasate di 180 gradi (uno +90, uno -90)   

2- modulo  
esiste una freq (la stessa di prima, omega0, la frequenza propria dell'oscillatore) dove i moduli delle due I sono uguali  
quindi $I_C=-I_L$
quindi corrente c'è, ma rimane in quel quadrato, non esce  

**Spiegazione fisica: antirisonanza**

quindi:  
a questa $f_0$, ho che $ I_L=-I_C $  
dalla legge dei nodi: $I_{gen}=I_C+I_L=0$    
Il generatore non deve quasi fornire corrente, è come se il circuito non assorbisse nulla  
quindi questo parallelo tra L e C sempra avere impedenza enorme   
quindi il circuito “si oppone” al passaggio di corrente    

a parole:  
a $f_0$, la corrente che “vorrebbe” passare nel ramo induttivo e quella nel ramo capacitivo si compensano   
come se l’energia oscillasse tra campo magnetico dell’induttore e campo elettrico del condensatore
