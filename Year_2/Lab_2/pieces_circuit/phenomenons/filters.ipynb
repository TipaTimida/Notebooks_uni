### Recap

What | low-pass | high-pass 
---|---|---
output across | C | R 
mnemonic trick | because capacitor voltage cannot follow rapid changes (inteGrator = C) | because resistor voltage follows the current, which flows only when voltage changes (derivator more R in the name)
in time domain | integrator | derivator
topology | capacitor in parallel with the signal | capacitor in series with the signal path


**Schematics**

VCC -> R + C -> GND  
low pass  

VCC -> C + R -> GND  
high pass

in both, take output across the second component  

**When it works**

freq domain: lets pass freq in name, block the other  
time domain: does the action on freq blocked, leave unchaged freq that is let passed

**To chain them**

$ Z_{out, 1} \ll Z_{in, 2}$    
pernsa che: se in seconda amglia c'è oscilloscopio, deve avere resistenza interna grande per non modificare il circuito, così I non passa lì  

so the best order is: high-pass -> low pass    


**To remember H for RC circuits**

$s = j\frac{f}{f_c}$, where: $f_c = \frac{1}{2 \pi RC}$   

low-pass:  $\frac{1}{1+s}$   
high-pass: $\frac{s}{1+s}$ (also has s on top, on the high part of the fraction)  

**To remember**

imagine the alphabet as a stairs, with A at the bottom  
low pass: output across C, which is lower  
high pass: output across R, which is higher  

### Theory

#### Definitions

**Definition of filter**

ideally: a system able to completely “cut off” all frequency components of a signal above or below a certain frequency  

realistically: this is only approximated    
rejection of frequency components is not truly sharp, bc $H(f)$ is not a step function (so can't have only 2 states, complete accept and complete reject, but has a transition region, where stuff is a bit attenuated)  

intuitively: un circuito è un filtro se la sua funzione di trasferimento dipende dalla frequenza

**Take output across a component**

high-pass and low-pass differ in across which component i take the output  

taking the output across something means:  
1- put that component as the closest to the GND  
2- take the signal from the node just before that component   
so the signal at that node is actually the voltage drop between that point and GND  

take output across C:  
VCC -> R -> C -> GND  

take output across R:  
VCC -> C -> R -> GND  

#### Time vs freq domain

**How it works in frequency domain**

1- separate the signal in its frequency components  
2- compute the cut-off frequency of the filter  
3- for EACH freq component, pick the case:    
f in the name: pass   
f NOT in the name: attenuated    

example: high pass lets pass high f, blocks low f  

**How it works in time domain**

1- working in time domain:  
need T or f of the SIGNAL 

2- compute the time constant $\tau$  

3- pick the case (just once here)  
f that the filter BLOCK: it's integrating or derivating, it works   
f that the filter lets pass: leave mostly unchanged, just a bit attenuated  

#### Types

**Types of filters**

- single-pole (a un polo): made of 2 components (like an RC circuit)   
cioè c'è un solo nodo da cui posso scegliere di prendere l'output, ed è tra i due componenti  

- higher-pole: more components  
have steeper slopes in the transition region  

**Band-pass, notch filters (filtri passa-banda, sopprimi-banda)**

low-pass, high-pass: let pass frequencies lower or higher than a threshold  
band-pass, notch: let pass frequencies inside or outside an interval (so has both low and high threshold)  

how to make band-pass (passa-banda):    
just put a low-pass and high-pass in series  
but attention, they interact with each other a bit  

hot to make a notch:  
cannot connect a low-pass and high-pass in parallel  
because would connect their output nodes, which must never be done, creates undesired coupling  
to avoid, do a similar thing, but using a schematic called 'bridge-tee filter' (HOW TO DO???)    

**Bandwidth**

the size (width) of the frequency interval that the system lets pass  
$BW = f_{high} -f_{low}$  
but sometimes, use to mean only $f_{high}$, the upper bound of the interval, implicitly meaning that the upper bound is 0  

**Practical: making pass-band, dimensionamento**

to make the pass-band work:  
need a region where both filters allow passage, so: $f_L > f_H$ (a band below the low-pass and above the high-pass threshold)  

to have the condition of maglie indipendenti:  
if order low -> high: need  $R_L \ll R_H$ , explain above why (so don't care about C)  
if order high -> low: it's always fine, as explained above

**Applications**

integrators/low-pass/decoupling: pass low f, block high f  
does: block noise 
formally: reduce sensitivity on fast variations, improve the signal-to-noise (S/N) ratio  

differentiators/high-pass/coupling: pass high f, block low f  
does: blok DC, offset  
formally: enhance variations/changes (use when voltage change must be monitored on short time scales)  
this is the AC coupling of the obscillascope

**Integrating noise**

hypothesis: noise is alternated and periodic, with period $T_N$  
while $T_S$ is period of signal    

if have a lot of noise ($S/N \ll 1$):  
pick $\tau$ of the integrator such that: $T_N \ll \tau \ll T_S$  
integrator=high-pass, so integrates for high freqs, so for low T, T << tau  
noise: it's T << tau, gets integrated, since it's alternated it gives mean=0   
signal: it's T >> tau, so doesn't integrate, it's left mostly unchanged, just un po' addolcito  

problem:  
the longer the measuration, the bigger S/N after integration  
but if it's too long, i have the problem of the flicker noise 

**Derivator: really noisy**

Il derivatore:  
amplifica le componenti ad alta frequenza  
quindi qualsiasi piccola imperfezione esce come rumore o distorsione  
per questo è molto più difficile ottenere una forma pulita in output
