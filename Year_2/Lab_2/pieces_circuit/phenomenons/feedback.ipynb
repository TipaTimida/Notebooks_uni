**Feedback: what it is**

when the output affects the input  

note: there can be a wire from $v_{out}$ a $v_{in}$, in which case it's called 'feedback loop'   
but it's not necessary, can be even without wire

**Struttura feedback loop**

the 'feedback loop' is made of:  

1- amplificatore  
eg. transistor ad emettitore comune  

2- ramo di feedback  
serve un condensatore perchè non deve far passare corrente in DC, sennò cambia il working point  

### Model

**Model**

transfer function:  
$v_{out}=H_a(v_{in}+H_fv_{out})$  
$H_{cl}=\frac{v_{out}}{v_{in}} = \frac{H_a}{1-H_aH_f}$  

notes:  
1- cl = closed loop  
2- if don't depend of frequency, can call it g instead of H  

**Attention**

1- closed-loop gain:  
$H_{cl} =\frac{H_a}{1-H_aH_f}$    
overall system gain from source to output  
used: to study gain (so what exactly happens WHEN obscillation happens)  

2- loop gain:  
$H_l = H_a H_f$  
signal after one round in loop  
used: to study obscillation condition and stability (so just IF obscillation happens)  

**Tipi di feedback**

dipende da $g_f$, e quindi cosa mi viene nella formula  

1- feedback positivo  
$|H_{cl}|> |H_a|$  
same sign  
uses: increases gain, may oscillate (bc it cannot diverge to infinity and create infinite energy)  

2- feedback negativo  
$|H_{cl}|< |H_a|$    
the portion of output variation re-injected into the
input has an opposite sign with respect to the input variation   
uses: reduces gain, stabilizes  

#### Example

**Situation**

amplifier: transistor ad emettitore comune, senza R_E  
feedback branch: a capacitor C  

**What happens**

$H_a=g_a$, perchè non dipende dalla frequenza  
ma $H_f$ dipende dalla frequenza  

1- f bassa: C è un aperto, no feedback  

2- f alta: C è un corto, quindi $v_{in}$ e $v_{out}$ sono collegati  
è un feedback negativo, riduce sempre di più il guadagno  
in totale, si comporta come un filtro passa-basso  

**Trova transfer function**

1- legge dei nodi sul nodo sopra $C_F$    
$i_{out}+i_F=i_c$ (i segni si vedono dal diesgno del circuito)  
$-\frac{v_{out}}{R_C} +\frac{v_{in}-v_{out}}{Z_F} = \beta_f i_b = \beta_f \frac{v_{in}}{R_B+r_b}$  

2- funzione di trasferimento:  
$v_{out}(\frac{1}{Z_F}+\frac{1}{R_C})=v_{in} (\frac{1}{Z_F}-\frac{\beta}{R_B+r_b})$  

$T(\omega) = -\frac{R_C}{R_B+r_b}\beta_f (\frac{1-j\omega(R_B+r_b)C_F/\beta_f}{1+j\omega R_C C_F})$  
nota: il primo fattore è $g_A$ dell'amplificatore senza feedback  

note:  
1- qui $i_v \neq i_{out}$  
2- $v_{in}=(R_b+r_b)i_b$  
scrivendo la legge della maglie sulla maglia di input  

### Auto-obscillator

#### Definition

**Cosa fa**

un circuito che:  
prende energia da una sorgente costante (DC)  
e produce un oscillazioni costanti  

quindi, senza bisogno di un segnale interno oscillatorio  

**Come funziona**

1- rumore iniziale  
2- Amplificazione
il feedback positivo la amplifica  
3— Selezione della frequenza
il risonatore seleziona una frequenza precisa
4— Saturazione   
una non linearità stabilizza l'ampiezza, inpedisce di divergere  

**Obscillation condition**

$H_aH_f = (1,0)$  

cioè:  
gain = 1  
dephasing = 0  

why:  
the signal that gets reinputted must be as strong as the original one and with same dephasing, so nothing changes, keep going like before  


**How to start it**

auto-oscillations:  
require an initial trigger to be activated  
then, after i remove the trigger, will continue to obscillate by itself  

what is the trigger:  
usually, the trigger is not intentionally created, but it
rather consists of fluctuations, or noise  
Since in the transient stage of the process a rapid growth of the auto-oscillation amplitude occurs, even tiny perturbations of the stationary state may result sufficient to trigger auto-oscillations  

**Why gain must be larger than losses**

You always have: resistive losses, 
energy dissipation, signal attenuation  
So a signal circulating in the loop tends to shrink  
so the branch with the amplifier must give a gain big enough to counter them  

| parte                  | ruolo                            |
| ---------------------- | -------------------------------- |
| transistor             | amplificatore (fornisce energia) |
| alimentazione          | sorgente di energia              |
| rete RC (3 passa-alto) | feedback + selezione frequenza   |
| rumore iniziale        | “innesco”                        |


la rete RC seleziona la frequenza  
perchè solo ad una frequenza precisa ho la condizione che lo sfasamento sia di 180 (per compensare i 180 dati dal transistor)  
quindi, quella è la frequenza dell'oscillazione prodotta  

#### Model

**Model it**

note that $v_{in}$ and $v_{out}$ are functions of time  

1- rewrite formula for feedback:  
$v_{out}(t)=g_a[v_{in}(t)+g_fv_{out}(t-\tau)]$  
since need time for the signal to travel, the output signal received is actually one that was outputed a bit ago  

2- expand for $\tau \sim t$ (now ignore the (t), and derivatives are wrt dt)    
$v_{out}=g_a[v_{in}+g_f(v_{out}-\tau \dot v_{out})+\frac{\tau^2}2 \ddot v_{out}]$  
$\ddot v_{out} -\frac 2 \tau \dot v_{out} +\frac 2{\tau^2} (1-\frac 1 {g_ag_f}) v_{out}=-\frac 2 {\tau^2} v_{in}$  

3- rename cofficients, it looks like a driven harmonic obcillator  
$\ddot v_{out} -\gamma_+ v_{out} +\omega_a^2 v_{out}=-\frac 2 {\tau^2} v_{in}$  

4- problem: for damping, the first derivative should have sign +, not -, so add a $\gamma_-$ to add damping (not well modeled, but whatever)  
$\ddot v_{out} -(\gamma_+ \gamma_-)v_{out} +\omega_a^2 v_{out}=-\frac 2 {\tau^2} v_{in}$  

5- $v_{in}$ is the trigger, which is not needed to sustain the obscillation, so put $v_{in}=0$  
means picking a time twhere $v_{in}(t)=0$, and imagining to set the initial condition (the trigger that caused it)  
also, $\gamma^+$ depends on $\tau$, which changes depending on the path taken, so it's possible that $\gamma_-=\gamma_+$, in which case:  
$\ddot v_{out} +\omega_a^2 v_{out}=0$   
it's an harmonic obscillator

**Situation**

**Model the phase shift network**

IMPORTANT: the three loops are not independent (current branched off is not negligible)  
so must study them all together, with a loop equation  
$I_i$: current in the i-esim loop (so under the the C_i capacitor)  

1- equazioni delle maglie:  
$V_{in}=Z_CI_1 +R(I_1-I_2) $   
$0 = Z_CI_2 +R(I_2-I_1) +R(I_2-I_3)$  
$0 = Z_C I_3 +R'I_3 + R(I_3-I_2)$  

2- the output is:  
$V_{out} = R'I_3$  
now can write $H(f)$ 

**Auto-obscillation**

problem:  
H depends on frequency  
to have $H_aH_f=(1,0)$, since $H_a$ is a transistor with 180 dephasing, $H_f$ must also have 180 dephasing  

the 180 phase shift happens at the frequency:  
$\omega_\pi= \frac{1}{\sqrt{3R(R+R')C^2}}$  
who is approzimately like $\omega_a$  


**How to choose feedback branch**

what i want:  
transistor gives 180 dephasing  
so to have positive feedbak, the feedback branch too must give 180 dephasing  
capacitors have at most 90 shift, but achieveing 90 gives it a very low gain, better to use 3 of them and give 60 shift to each  

use low-pass or high-pass?:   
non voglio modificare il bias del transistor, che è fissato dalla corrente DC   
quindi guardo che succede a $f=0$, voglio che la corrente non passi

low-pass (opposto del disegno): mi viene il filo orizzontale con 3R  
high-pass: mi viene resistenza infinita  
quindi meglio high-pass

#### Larsen effect

**Larsen effect**

what:  
have a microphone connected to an audio amplifier  
if you point the microphone towards the loudspeaker, there is a high probability of activating an oscillating behavior, resulting in a
continuous (and annoying) note emitted by the loudspeaker  

how:  
the microphone (system input) collects a portion of the sound produced by the loudspeaker (output)   
so output gets sent back to the input  
if the output and input are in phase, i activate an obscillating behaviour  
Since the continuous note does not require any oscillating signal purposely connected to the amplifier, we will define the produced oscillations as auto-oscillations, aka self-oscillations  
meaning: once it's started, it will continue, don't need to keep giving it power (assuming no dissipation)  

notes:  
feedbak branch: the path traveled by the sound  
in closed rooms, have many possible paths, bc sound rimbalza on the walls  
so have big chance that in at least one path, input and output are in phase   
some signal must initially exist in order to trigger the oscillation: the trigger can
be represented by any sort of spurious signal, or, if you prefer, of noise   
a temporal delay is certainly involved in the effect, since sound propagation is not instan-
taneous

### Nyquist plot

**Cosa è**

to study the stability of a feedback system  
shows how a system responds to all frequencies by plotting its complex gain  

**Plot**

plot the complex gain:  
x-axis: Re[G], cioè il gain  
y-axis: Im[G], cioè il dephasing  

#### Use it

**Nyquist criterion**

the system gets unstable when:  
the point (1,0) is encyrcled by the Nyquist plot (line plotted)  

why:   
at point (1,0) have gain=1 and dephasing=0  
so it keeps obscillating forever, the signal don't change    


**How to do it**

1- compute $H_{l}(\omega) = H_a(\omega) H_f(\omega)$  

2- for each value of $\omega$ between $[-\infty, \infty]$, draw a point: $(Re\{H_{l}\}, Im\{H_{l}\})$  

i got a line  

**Note**

changing $g_a$ changes the line plotted    
so this plot also proves that a too small amplifier gain (in absolute terms) does not lead the system to be unstable
