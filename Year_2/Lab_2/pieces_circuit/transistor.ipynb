### Structure

![alt text](https://cdn.sparkfun.com/assets/learn_tutorials/1/9/3/transistor-active-electrons.png)

**Structure**

3 parts:  
1- E: emitter ($n^+$ doped, more heavily doped than C)    
2- B: base (p-doped)   
3- C: collector (n-doped)  

BJT = bipolar junction transistor  

**Physically**

collector: the biggest part, connected to heat sink (dissipates most heat)   
base: thin, to facilitate passage  
emitter: more heavily doped and smaller than C  

note:  
E and C cannot be swapped  
bc have different structure (one is outside, one inside) and strength of the doping  

**How it works**

has 3 wires, one in each part:   

1- emitter: input signal, the one to be modified  

2- base: controller, the one who decides what the transistor does   
$\Delta V_{BE}$, base-emitter voltage  

3- collector: output signal / controlled: it's a modification of the input one  
$\Delta V_{CE}$ collector-emitter voltage (the whole thing)   

**Types**

1- BJTs: current-controlled  
2- MOSFETs: voltage-controlled  

so mosfets are better for miniaturization and large scale integration, used in CPUs and RAMs  
bc currents have dissipation via Joule effect, so if i tried to put a lot of BJTs together, they would all burn (thermal runaway)  

**The lab one**

B and E are indicated written in the bottom  
green: attach to C WITHOUT any resistance, so node between C and $R_C$  
the two reds: attacch to C WITH the resistance (middle one is $2.2 k\Omega$, external is $1 k\Omega$)

#### Transistor effect

![alt text](https://www.newtoncbraga.com/images/sampledata/artigo2020/q0079e_0001.png)

**Formulas**

remember, only works in ACTIVE REGION 
essentially: $|I_C| \sim |I_E| \gg |I_B|$   

$\begin{cases}
I_C = \alpha_F I_E & \alpha_F \sim 0.95 \\ 
I_C=\beta_F I_B & \beta_F \sim 100 \\
\beta_F  = \frac{\alpha_F}{1-\alpha_F} \\
\alpha_F  = \frac{\beta_F}{1+\beta_F}
\end{cases}$  

altre relazioni ricavabili da lì:    
$I_E=(β_F+1)I_B$  
$I_B = (1-\alpha_F)I_E$  


**What it is**

the phenomenon why, IN ACTIVE REGION, a little base current can produce big variations in the collector current  
se non ci fosse l'eggetto transistor, il transistor si comporterebbe solo come due diodi back-to-back  

**Explanation**

1- config: active region (only happens in active region)    
so BE forward bias, BC reverse bias  

2- BC junction  
it's in reverse, so the depletion region gets bigger  
so, since the head is small, most of it is occupied bu the depletion region  

3- BE junction  
it's in forward, so electrons cross the junction to recombine (diffusion)  
but they don't find enough holes in B (perchè già la zona B è molto piccola, e poi most of it is depletion region caused by the BC junction)    
so most of the electrons get caught by the strong field across the BC junction, and get pulled in C  

4- regarding currents  
electrons recombining with holes in B: $I_B$  
electrons being sucked in C: $I_C$   
since most of the electrons go to C, $I_C \gg I_B$  

5- consequence:  
$I_C =-\alpha_F I_E$  
con $\alpha_F$ poco meno di 1 (quasi tutta $I_E$ va a C, solo poca va a B)  
se scambio cavi di E e C, e quindi scambio le bias delle giunzioni, cioè uso la modalità inutile, al posto di $\alpha_F$ metto $\alpha_R$, che è molto meno di 1   

**Rapporto tra alpha e beta**

1- leggi dei nodi, dentro il transistor:  
$I_C+I_B-I_E=0$  

2- effetto transistor:  
$I_C = \alpha_F I_E$ 

3- quindi:  
$I_C(1 -\frac1{\alpha_F}) +I_B=0$  
$I_B = I_C(\frac1{\alpha_F}-1 ) = I_C (\frac{1-\alpha_F}{\alpha_F}) = \beta_F I_B $  

$\beta_F  = \frac{\alpha_F}{1-\alpha_F}$: current gain of the transistor    
it's really big, around 100, so a small change in $I_B$, the control current, produces a big change in $I_C$, the controlled one  

**Spiegazione grafica relazione $I_C$ e $I_B$**

maglie:  
Vcc -> R_C -> V_{CE} -> GND  
Vcc -> R_p -> R_B -> V_{BE} -> GND   
poi aggiungo una v variabile alla giunzione BE, per modificare $I_B$  

che succede:  
pensa alla char curve di collettore, $V_{CE}- I_C$  
la load line resta il fissa  
che succede se aumenta $I_B$:  
1- cambia la curva di collettore da usare, devo usarne una più alta  
2- cambia l'intersezione con la load line, diventa più alta    
cioè il punto di lavoro scorre su e giù sulla load line  
3- il punto di lavoro ha coord $(V_{CE}, I_C)$     
quindi se è più in alto, vuol dire che $I_C$ è aumentato  

se invece $I_B$ diminuisse, si abbasserebbe la curva di collettore da usare, e quindi si abbasserebbe il working point, la coordinata di $I_C$  

#### Early effect

**What it is**

$I_C$ increases slightly with $V_{CE}$  
not just with $V_{BE}$  

gradicamente: nella char curve di collettore, in regime ativo le righe non sono proprio orizzontali, ma un po' inclinate verso l'alto  

**Explanation**

1- when increase $V_{CE}$, the BC junction gets more reverse biased  
why: 


Microscopic interpretation of the Early effect is straightforward: at increasing VCE , the
BC junction gets more and more reverse biased. Therefore, the width, or thickness, of the
depletion region across the BC interface, see Sec. 12.2.1 and Fig. 12.3, increases, reducing the
probability of recombination for the electrons coming from the emitter, eventually leading to
an enhancement of the transistor effect, hence an increase of the gain.  

#### Saturation

**Saturation**

problem:  
have circuit: $V_{CC}-V_{CE}-I_CR_C=0$  
if send current $I_B$, transtistor should output current $I_C=\beta I_B$  
but must be $I_C \le \frac{V}{R}$, it's a constraint of the circuit (legge delle maglie se il tranistor fosse un corto)  
if it's bigger (so if i send $I_B$ tale che darebbe $I_C$ maggiore di quella ammessa), the transistor must limit the current sent, means it's in saturation  

consequences:  
collector voltage collapses to a small value  
more base current does not change the collector current     

what happens:  
$I_C$ è limitata dal circuito, non vale più  $I_C = \beta I_B$  
anche la giunzione base-collettore diventa forward  

perchè il voltaggio del transistor va a 0:  
la corrente massima possibile si ottiene facendo finta che il transistor sia un corto, quindi $V_{CE}=0$  
basta guardare la legge delle maglie

#### Tranconduttanza

**Definizione**

$g_m = \frac{i_c}{v_{be}}$  
misura la sensibilità del transistore  
cioè quanto varia la corrente di collettore al variare della tensione base-emettitore  

**Stimalo**

1- linearizzo (cioè trasformo in derivata, come per calcolare $r_d$, la resistenza dinamica)  
$g_m = \frac{dI_C}{dV_{BE}}$  

2- uso la legge di Shockley:    
$I_C = I_S e^{V_{BE/(\eta V_T)}}$  
ho usato la versione senza il -1, che fare bene in regime attivo  

3- viene:  
$g_m = \frac{1}{\eta V_T}I_S e^{V_{BE/(\eta V_T)}} = \frac{I_C}{\eta V_T}$   


**Graficamente**

$g_m$ si calcola simile alla resistenza dinamica, ma è la derivata $\frac{dI_C}{dV_{BE}}$  
quindi è la tangente nel grafico $I_C-V_{BE}$    
che non disegno mai, non è nessuna delle due curve caratteristiche importanti  
e $r_e$ è la cotangente   

**Altro**

1- altro modo di scriverlo:  
$g_m = \frac{i_c}{v_{be}} = \frac{i_c}{r_\pi i_b} = \frac\beta{r_\pi}$  

2- resistenza quivalente:  
$r_e = \frac{1}{g_m}$  

3- usi: posso riscrivere $v_{out}$   
$v_{out}=-i_cR_C =-g_mv_{be}R_C$   


### Voltage gain

**Modi di scriverlo**

due modi per scriverlo:  

1- gain dello stadio: $A_v = \frac{v_{out}}{v_b}$  
il guadagno del transistor stesso, senza considerare come il segnale arriva alla base  
in pratica ignora $R_B$, immagina di prendere $v_{in}$ tra il transistor e $R_B$     

2- gain del circuito completo: $g_{ce} = \frac{v_{out}}{v_{in}}$    
misura quanto amplifica l’intero circuito visto dall’esterno  
in pratica prendi $v_{in}$ prima di $R_B$, così misuro anche il suo drop  

io di solito uso il secondo, che si può ottenere dalla transfer equation H  
visto che H non ha parte immaginaria, posso definire $g_{ce} = |H|$, e il segno - indica un'inversione di 180 gradi  

**Formule: common emitter**

common emitter:  
$v_{in} = [R_b+r_b+(\beta_f+1)Z_{eq}]i_b$  
$v_{out} =- R_c i_c$   
$H(\omega)=\frac{v_{out}}{v_{in}} = -\frac{R_C}{R_b+r_b+(\beta_f+1)Z_{eq}}\beta_f$  
dove $Z_{eq}=\frac{R_E}{1+j\omega R_EC_E}$  

note:  
se non c'è C all'emettitore, $Z_{eq}=R_E$  
H non ha parte immaginaria, la chiamo solo $g_{ce}$  

a cosa tende $g_{ce}$  
$R_E$ bassa: $g_{ce} \sim-\frac{R_C}{R_B+r_b}{\frac{i_c}{i_b}} \sim -\frac{R_C}{R_B} \beta$  
$R_E$ alta: $g_{ce} \sim -\frac{R_C \beta}{(\beta-1)R_E} \sim -\frac{R_C}{R_E}$  

**Formule: common collector**

common collector:  
$v_{in}$: come in common emitter  
$v_{out} = -R_Ei_e = -R_E(1+\beta)i_b$  
$g_{cc} = -\frac{-R_E(1+\beta)i_b}{[R_B +r_b+(\beta +1) R_E]i_b} = \frac{R_E(1+\beta)}{R_B +r_b+(\beta +1) R_E}$  

note: rispetto al common emitter    
1- il - sparisce, infatti qui non c'è sfasamento  
2- $g_{cc} <1$, è un buffer, non un amplificatore  

**Formule più facili da ricordare**

1- scrivi con transduttanza:  
$A_v = -\frac{b_m R_C}{1+g_mR_E}$  

2- a cosa tende:  
$A_v = -\frac{g_m R_C}{1+g_mR_E} = -\frac{R_C}{R_E + 1/g_m} \sim - \frac{R_C}{R_E}$    

3- se $R_E=0$:  
$A_v = -g_m R_C$  

**To simplify**

1- ricorda che il valore stimato per una giunzione (diodo) è:  
$r_b \sim \frac{\eta V_T}{I_B}$  

2- so che $i_C = \beta_f i_b$  
posso apprssimare $\beta_f = \beta_F = \frac{I_C}{I_B }$  
visto che le correnti in DC sono più facili da misurare    

**Types of gain**

1- voltage gain: $g_{ce}$  
2- current gain: quello dato dal fenomeno del transistor  
3- power gain: 
$G_{ce,P} = \frac{|v_{out} \, i_c|}{|v_{in} \, i_b|} = A_v \beta_f$  
perchè $P=IV$
