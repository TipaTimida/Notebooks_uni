#### Teoria

**Biasing**

cosa: scegliere il punto di lavoro del componente, e quindi del regime  
quando: ho componenti con una PN junction (diodo, transistor)  
come: devo decidere la I o la V da inviare 

**Biasing con V o I?**

ogni componente ha una char curve, I(V)  
posso invertirla e trovare I(V)  
cioè stabilire la corrente o il voltaggio è la stessa cosa, ovvero scegliere il working point  
cioè segnare sulla char curve il punto con quella coordinata, che può essere una I o una V  

esempio: diodo  
rispetta la legge di Schocley: $I=I_s e^{V/(\eta V_T)}$  
posso invertire: $V=V_T \eta \ln (\frac{I}{I_s})$  
cioè, mandare una precisa corrente mi setta un preciso voltaggio, quindi stabilisce il regime  

quale delle due dipende da cosa è più facile:    
diodo: più facile scegliere V  
transistor: più facile scegliere $I_B$

### Advanced biasing for transistor

#### Add resistor at emitter

**Why**

why: prevents thermal runaway, so makes the circuit less sensitive to tempreature  
consequence: stabilizes the operating point, linearizes the amplifier (meaning???)  
how: introducing negative feedback  

negative point: lowers the gain  

**Problem**

1- positive feedback loop (circolo vizioso)    
current and temperature of the junction si fanno aumentare a vicenda  
bc I causes dissipation via Joule's effect, which increase T of junction, which increases I (property of PN junction)  

2- thermal runway  
finisce con 'thermal runway': rottura del componente per aumento incontrollato di corrente  

note: why increasing T increases I in a PN junction  
in Shockley equation, both $e^{1/T}$ nd $I_S$ depend on temperature    
$I_S$ does bc, the higher T; the more thermic agitation, so the more electron-hole pairs are generated  
$I_S$ increases more than the exponential, so in totale if T increase, I increase  

**Solution: add $R_E$**

breaks the positive feedback loop (by adding to it a negative feedback, thatn when felles I increase it lowers it)     
I ↑ → drop on $R_E$ ↑ → $V_{BE}$ ↓ → I ↓   
bc in char curve, see that if V goes down, then I goes down too, i change the biasing  

**Graphical explanation**

consider the load line $I_B(V_{BE})$_  

$I_B = -\frac{1}{R_B'}V_{BE} +\frac{V_{cc}-R_EI_{Cq}}{R_b'}$   

how to find load line:  
write loop law: $V_{cc}-R_b'I_B-V_{BE}-R_EI_E =0$   
use that: $I_E \sim I_C$ e $R'_B = R_B+r_b$ 

easy | graph
---|---
$I$ ↑ | $I_C$ ↑  
drop on $R_E$ ↑ | intercetta di $I_B(V_{BE})$ gets lower  
$V_{BE}$ ↓ | the load line is translated down, so changes the working point, which is now lower, so its coords $(I_B, V_{BE})$ are lower  
$I$ ↓  | since have $I_C=\beta I_B$, $I_C$ gets lower too

#### Add capacitor at emitter

**Resistor and capacitor at emitter**

structure:  
they're in parallel (so wins the lowest Z)  

why:  
it behaves like a high-pass on gain  
not filtering the signal, but filtering the feedback  

wxplanation:  
high f: $Z_C = 0$, so like $R_E$ is not there, so high gain  
low f: $Z_C = \infty$, so $R_E$ is there, lower gain  
look like high pass filter, bc attenuates more the low freqs (rather then attenuates more, here it amplifies less)  

**Capacitor at collector**

structure:  
a wire with a capacitor between nodes $v_{in}$ and $v_{out}$  

why:  
low f: $Z_C = \infty$, like the wire is not there, normal amplifier  
high f: $Z_C=0$, normal wire, so a portion of the output gets inhected in the input  
the amplifier inverts the signal, so the signal reinputed is dephased wrt the original one -> negative feedback
