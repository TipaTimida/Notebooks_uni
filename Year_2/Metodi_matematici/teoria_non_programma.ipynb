### Taxonomyof mathematical objects

**Linear functional**

input: vector  
output: scalar 

$f= V \rightarrow R$  

example: $f(v)=a_iv^i$  

If it’s not linear, it's just called a:  
Scalar-valued function on a vector space

**Bilinear form**

input: 2 vectors  
outputs: a scalar  

$B: V \times V \rightarrow R$  

Examples:  
Dot product  
metric tensor  


**Binary operator**

input: 2 numbers  
output: a number  

$R \times R \rightarroe R$  

example: +, -, x, /

**Operator**

$T: F \righarrow F$  

input: function  
output: function  

example:  
Derivative operator  
Fourier transform  
Laplacian
