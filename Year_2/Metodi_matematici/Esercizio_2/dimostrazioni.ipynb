#### Convergenaz dominata

**Teorema della convergenza dominata**

ho una funzione $h_n(x)$ tale che  
$h_n(x) \rightarrow h(x)$ quasi ovunque (cioè i punti in cui non accade sono un insieme numerabile)  
$|h_n(x)| \le g(x)$, dove g è integrabile in $L^2$ e indipendente da n  
g si chiama dominante integrabile  

allora:  
$\int h_n(x) dx \rightarrow h(x) dx$  
$\lim_{n \rightarrow \infty} \int h_n = \lim _{n \rightarrow \infty} h_n$  
cioè posso invertire integrale e limite  


**Esempio: applicare convergenza dominata**

ho: $h_n(x) = |\cos(2x)|^{2n} |d(x)|^2$  
voglio provare che: $h_n(x) \rightarrow 0$  

esercizio | cosa
---|---
per quasi ogni x (eccetto un insieme numerabile di punti): <br> $\|\cos(2x)\| < 1$ <br> quindi, quasi ovunque: <br> $h_n(x) =  \|\cos(2x)\|^{2n} \|d(x)\|^2 \rightarrow 0$ | ipotesi 1: convergenza quasi ovunque  
$ h_n(x) \le \|d(x)\|^2$  <br> perchè $\cos \le 1$ <br>quindi d(x) è la dominante, ed è integrabile perchè $\|d\|^2 \in L^1$| ipotesi 2: funzione dominante integrabile
$h_n \rightarrow 0$ | conseguenza

#### Teoria

**Tipi di convergenza**

1- successione di funzioni:  
puntuale: $f_n(x)\rightarrow f(x) \quad \forall x \in X$  
uniforme: $sup_{x \in X}|f_n(x)-f(x) | \rightarrow 0$  

2- succesione di operatori:   
forte degli operatori: $|T_nf -Tf| \rightarrow 0$  
in orma operatoriale: $|T_n-T| = sup_{|f|=1} |(T_n-T)f| \rightarrow 0$

nella convergenza puntuale fissi x  
nella convergenza forte fissi f  
e DOPO verifichi la convergenza  

nella convergenza uniforme e in norma il controllo vale contemporaneamente per tutte le x o tutti i vettori di norma 1 (cioè tutte le direzioni, i versori)

$Af = \Sigma_{n=0}^\infty \psi_{[n,n+1]}(x) \, e^{-nx} \, f(x) = 0$

richiesta:

2- trova T (fai il limite)    
$lim \, A^kf = \psi_{[0,1]}(x) \;  f(x) \; + \;  \Sigma_{n=1}^\infty e^{-\infty} $  
$= \psi_{[0,1]}(x) \;  f(x) $    
$=T$  

3- scrivo la richiesta   
$A^kf-Tf =  \Sigma_{n=1}^\infty \psi_{[n,n+1]}(x) \, e^{-knx} \, f(x)$

4- prendo la norma   
(porto fuori la sommaotoria, $\chi$ cambia gli estremi di integrazione)  
$|A^kf-Tf|^2 =$  
$ = \int_0^\infty |A^kf-Tf|^2 \, dx$    
$= \int_0^\infty  |\Sigma_{n=1}^\infty \psi_{[n,n+1]}(x) \, e^{-knx} \, f(x)|^2 \, dx $  
$= \Sigma_{n=1}^\infty \int_0^\infty  | \psi_{[n,n+1]}(x) \, e^{-knx} \, f(x)|^2 \, dx $
