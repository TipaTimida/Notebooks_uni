### Teoria

**Come è scritto**

$Te_n = ...$  

tips:  
se c'è una sommatoria, prova ad espanderla per capire meglio cosa fa  

esempio:  
$Te_n = \frac 1 {\sqrt n} \Sigma_{k=1}^n e_k$   
$Te_1 = e_1$  
$Te_2 = \frac 1 {\sqrt 2} (e_1+e_2)$  
$Te_3 = \frac 1 {\sqrt 3} (e_1+e_2+e_3)$    

**Come si usa**

invece di applicarlo ad una funzione generica f, si applica ad un vettore generico x  

$x = \Sigma x_n e_n$  
$Tx = \Sigma x_n Te_n$  
e sostitusco lìoperatore dato nel testo a a $Te_n$  


**Spazio**

spazio di dimensione n si descrive scrivendone la base:  
$V_n = \{v_n := \Sigma_1^n x_n e_n\}$    


### Norma

**Come si fa**

uso Parseval  
ricorda che è base canonica, ha norma 1  

**Esempio**

$Te_n = \frac n{n+1}e_{n+1}$   

1- limitatezza  
$|Tx|^2 = \Sigma |x_n \frac n{n+1} e_{n+1}|^2 $  
$ = \Sigma |x_n|^2 | \frac n {n+1} |^2 $   
$\le \Sigma |x_n|^2   
$|x|^2$  
quindi $|T|\le 1$  

2- dimostro che è il sup  
se uso il vettore $v=e_n$, ottengo: $Tv = |\frac n{n+1} e_{n+1}| = \frac {n}{n+1}$   
$|T| = sup \frac{|Te_n|}{|e_n|} = sup |Te_n| =\ge |Tv| = \frac n{n+1} \rightarrow 1$  
cioè $|T| \ge 1$  
combinato a prima, $|T|=1$ 

### Autovalori

**Come si fa**

1-imposta la condizione: $Tx = \lambda x$   
x è scritto come un vettore    

2- è un sistema, separa le condizioni su ogni $e_k$  

**Esempio**

$Te_n = \frac 1n e_n$  

$x = \Sigma_1 x_n e_n$    
$Tx = \Sigma_1 x_n [ \frac 1n e_n]$    
$\lambda x = \lambda x_n e_n$  

$Tx = \lambda x$  
per ogni e_n:  
$\frac{x_n}n = \lambda x_n$    
$(\frac 1n -\lambda)x_n=0$  
per avere $x \neq 0$, almeno un coeff x_n deve essere non-0  
quindi devo avere $\lambda = \frac 1 n$, per un n qualunque  
quindi gli autovalori sono: $\lambda_k = \frac 1k$ con k=1,2...   
e l'autospazio relativo è $E_{\lambda_k} =span\{e_k\}$  

**Esempio**

$Te_n = \frac{n}{n+1} e_{n+1}$  

$Tx = \lambda x$  
$\Sigma_1 x_n \frac{n}{n+1} e_{n+1} = \lambda \Sigma_1 x_n e^n$  
$\Sigma_2 x_{k-1} \frac{k-1}{k} e_{k} = \lambda \Sigma_1 x_n e^n$    

$\begin{cases}
0 = \lambda x_1 & e_1 \\
x_{k-1}  \frac {k-1}{k} =\lambda x_k & e_k, \; k \ge 2 
\end{cases}$  
 
caso $\lambda \neq 0$: $x_i=0$, non è un autovalore    
caso $\lambda =0$: prendo la norma  
$|Tx|^2 = 0$  
$\Sigma _1 |x_n|^2 |\frac n{n+1}|^2 |e_{n+1}|^2 = 0$  
unica sol è $x_i=0$, non è un autovalore  
faccio la norma così tutti gli addendi sono non-negativi, così per avere =0 devono essere tutti nulli      



**Esempio**

$Te_n = \frac 1 {\sqrt n} \Sigma_{k=1}^n e_k$   
trova gli autovalori, quando n=2  

se disegno la amtrice, vedo che gli autoaloi sono 1, $\frac 1{\sqrt 2}$  

1- calcola T sui valori che mi interessano  
$Te_1 = e_1$  
$Te_2 = \frac 1 {\sqrt 2} (e_1+e_2)$  

2- calcolo T su un vettore generico    
$x = x_1e_1 + x_2e_2$    
$Tx = x_1 Te_1 + x_2Te_2 $  
$=x_1[e_1] + x_2[\frac 1 {\sqrt 2} (e_1+e_2)]$  
$= e_1(x_1-\frac{x_2}{\sqrt 2}) + e_2(\frac{x_2}{\sqrt 2})$  

3- scrivo la condizione: $Tx = \lambda x$   
nota che gli $e_i$ si possono togliere, sono sono per chiarezza  
$\begin{cases}
e_1[x_1-\frac{x_2}{\sqrt 2}] = \lambda x_1 e_1 \\
e_2[\frac{x_2}{\sqrt 2}] = \lambda x_2 e_2   
\end{cases}$

4- risolvo per $\lambda = 1$  
ottengo $x_2=0$  
non ho condizioni su $x_1$, quindi va bene uno qualunque  
$v=x_1e_1$ sono gli autovettori  ù

5- risolvo per $\lambda = \frac 1 {\sqrt 2}$  
ottenfo $x_2 = x_1(1- \sqrt 2)$  
gli autovettori sono: $v = x_1e_1 -x_1(\sqrt 2 -1) e_2$  
se ne voglio uno in particolare, posso mettere $x_1 = 1$  

**Tip**

se so disegnare la matrice di $T_n$:  
se è triangolare superiore, i valori sulla duagonale sono gli autovalori  

### Prodotto scalare

**Note**

la base canonica è ortonormale, quindi: $<e_i, e_j> = \delta_{ij}$  

**Esempio**

$Te_n = \frac 1 {\sqrt n} \Sigma_{k=1}^n e_k$   

$<e_, Te_n>  = \frac 1 {\sqrt n} <e_k, \Sigma_1^n e_i>$  

caso: k>n  
a destra non appare $e_k$  
$<e_, Te_n> =0$  

caso: k<n  
a destra appare $e_k$  
$ <e_, Te_n> = \frac {1}{\sqrt n} <e_k, e_k> = \frac 1 {\sqrt n}$ 

**Norma**

sono in $L^2$, quindi $|f| < \infty$    
quindi, con Parseval:    
$|f|^2 = \Sigma _1^\infty |<f_k, e_n>|^2 < \infty$

### Suriettivo vs denso

**Suriettivo**

$Tx=y$  

per dimostrare che NON è suriettivo:  
trova un $y \in L^2$ che non si può raggiungere, perchè:  
1- nessun vettore x porta a quella  
2- c'è un vettore x che porta a quello, ma $|x|^2 = \infty$, quindi $x \notin L^2$  


**Denso vs suriettivo**

suriettivo: $Im(T) = H$  
denso: $\overline {Im(T)} = H$  

T suriettivo -> Im(T) densa  
ma non per forza il contrario  
cioé:  
può essere denso e NON suriettivo  
ma non può essere suriettivo e non denso  

**Nomenclatura**

vettori a supporto finito: hanno solo un numero finito di componenti non nulle    
so che sono densi in $l^2$  

**Chiusura**

$I = (0,1)$  
$\overline I = [0,1]$  
cioè, la chiusura aggiunge all'insieme tutti i punti che puoi ottenere come limite di successioni di elementi dell'insieme  

$\overline {Im(T)}$  significa:   
1- prendi tutti i vettori che stanno nell'immagine di T  
2- aggiungi anche tutti i vettori che possono essere approssimati arbitrariamente bene da vettori dell'immagine  


**Esempio: NON suriettivo**  

$Te_n = \frac 1n e_n$  

scelgo il vettore: $y = (1, \frac 12, \frac 13...)$  
(nota che è un vettore a supporto infinito, cioè ha infiniti componenti non-0)  
vedo che $ y \in L^2$, perchè $|y|^3 = \Sigma \frac 1{n^2} < \infty$    
y è raggiunto da: $x(1,1,1...)$  
ma  $x \notin L^2$, perchè $|x|^2 = \Sigma |1|^2 = \infty$  
quindi non è suriettivo  

**Esempio: denso**

$Te_n = \frac 1n e_n$  

considero i vettori a supporto finito: $y^{(N)}= (y_1, ..., y_N, 0, 0,...)$  

1- ogni $y^{(N)}$ appartiene a Im(T)  
basta scegliere: $x^{(N)} = (y_1, 2y_1, ... N y_N , 0 ,0 ...)$  
così: $Tx=y$  

2- i vettori a supporto finito sono densi in $L^2$  
quindi $\overline{Im(T)} = L^2$
