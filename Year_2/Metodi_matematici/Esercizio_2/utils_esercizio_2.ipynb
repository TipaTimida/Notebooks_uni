### Supporto

$H = \{f(x): supp_f \subseteq [1, \infty)\}$  
significa l'insieme delle funzioni che hanno come dominio $[1, \infty)$, cioè da dove posso prendere x   
assumo che f sia definita a tratti in modo che per le altre x vale sempre 0, così se mi dà problemi perchè appaiono in un integrale posso tagliare la parte fuori dal supporto  

**Esempio**

$\int _0^\infty f = \int_1^\infty f$  
è vero solo se $f$ ha come supporto $[1, \infty)$  
cioè nell'intervallo $[0,1)$ vale 0, lo posso togliere dall'integrale  

**Esempio**

$Tf = f(e^x)$  
vedo che l'argomento $e^x$ appartiene sempre a $[1, \infty]$  
quindi se scelgo $f_1$ con supporto in [0,1], cioè tale che $f_1(x)=0$ se $ x \in [1, \infty)$, allora $Tf_1 = 0$   
quindi T è un operatore nullo per tutte le f con supporto [0,1]  

### Proietori

**Definizione**

un operatore che mi ha come immagine un certo insieme   
si dice che proietta su quell'insieme  

atenzione:  
T è unitario su H: se dò in input una funzione f dentro H, T ha norma 1, cioè la Tf in output ha la stessa norma di f    
T proietta su H: qualunque f gli dò, l'output è una funzione dentro H  

**Esempio**

$Tf =f$  
$T=I$, è il proiettore identità  

$Tf = \chi_{[1, \infty](x)} f(x)$  
T è il proiettore su $H= \{f(x): supp \subseteq [1,\infty]\}$

### Funzioni periodiche

**Definizione**

funzione con periodicità $\pi$:  
$f(x) = f(x + \pi)$  

me lo dice il testo se sto lavorando in uno spazio di funzioni periodiche  

**Proprietà**

se il testo dice che una funzion ha periodicità $2\pi$:  

1- posso sommarere o sottrarre $2\pi$ all'argomento, e rimane uguale:    
cioè $f(x-\pi) = f(x-\pi + 2 \pi) = f(x+\pi)$  

2- negli interali, ogni intervallo lungo $2 \pi$ va bene  
cioè, $\int_{-\pi}^{\pi} = \int_0^{2\pi}$  
utile quando faccio cambio var per trovare l'aggiunto, e mi cambia gli estremi  

**Esempio: usando periodicià**

$Tf= e^{ix}f(\pi-x) + e^{-ix}f(\pi+x)$   
e sono nel dominio dove $f$ ha periodicità $2 \pi$ 

esercizio | cosa
---|---
$Tf= e^{ix}[p(\pi-x)+d(\pi-x)] + e^{-ix}[p(\pi+x)+d(\pi+x)]$ | f=p+d 
$= e^{ix}[p(-\pi-x)+d(-\pi-x)] + e^{-ix}[p(\pi+x)+d(\pi+x)]$  | uso periodicità di $2\pi$ sul primo addendo
$= e^{ix}[p(\pi + x)-d(\pi + x)] + e^{-ix}[p(\pi+x)+d(\pi+x)]$ | uso che p e d siano pari e dispari  
$ = p(\pi+x)[e^{ix}+e^{-ix}] + d(\pi+x)[-e^{-ix}+e^{-ix}] $<br> $=  p(\pi+x) \, \cos x - d(\pi+x) \, 2i \sin x$ | trasformo esponeniali in trig  
$P(Tf) =p(\pi+x) \, \cos x - d(\pi+x) \, 2i \sin x$ <br> $D(Tf)=0$ | risultato


### Separa in parte pari/dispari

**Come si fa**

1- sostituisci: $f(x)=p(x)+d(x)$  
dove p,d sono funzioni pari e dispari    

2- usa le proprietà $p(-x)=p(x)$ e $d(-x)=d(x)$ per semplificare l'argomento dentro   
come argomento dentro voglio avere solo x  
shortcut: $f(-x) = p(-x)+d(-x)=p(x)-d(x)$  

3- se so la periodicità, utile usare anche quella    

**Attenzione**

quando divido in parte pari e dispari, non sempre p = parte pari  

ricorda che:  
pari x pari = dispari x dispari = pari  
pari x dispari = dispari  
la moltiplicazione di funzioni corrisponde all'ADDIZIONE, non alla moltiplicazione di numeri pari e dispari  

**Rimettere le f**

se devo calcolare operatori composti, devo far riapparire f     
basta fare: $p=Pf$, $d=Df$  

esempio:  
$p-\cos(2x) d =Pf-\cos(2x)Df$  

#### Esempi

**Esempio**

$Tf(x) = \sin^ 2 (x)\, f(x) + \cos ^2 (x) \, f(-x)$  
$Tf(x) =  \sin^2 (x) (p+d)+\cos^2 (x) (p-d) = p-\cos(2x) d$  

quindi:  
$P(Tf) = p$  
$D(Tf) = \cos 3x \, d \quad$  (perchè pari x dispari = dispari)  

**Esempio**

$Tf = \frac{1}{2}(e^{ix}f(-x)+e^{-ix}f(x))$  
$Tf = \frac{1}{2}[p(e^{ix}+e^{-ix}) +d(-e^{ix}+e^{-ix})] = \frac{1}{2}[p (2\cos x) +d (-2i \sin x)] = p \cos x -id \sin x$    

quindi:  
$P(Tf) = p \cos x -id \sin x \quad$ (perchè dispari x dispari = pari)   
$D(Tf) = 0$  

**Esempio: quando non si può usare parità**

$Tf=e^{ix} f(-x)$  
$Tf = e^{ix} \, p - e^{ix} \, d$  

non posso separare in parte pari e dispari, perchè $e^{ix}$ non è ne pari nè dispari
