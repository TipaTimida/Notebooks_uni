### Limitatezza e norma

#### Teoria

**Limitatezza vs norma**

1- limitatezza: $||T|| \le A$   
cioè solo un upper bound  

2- norma: $||T||=A$    
è il SUP, secondo la definizione  
cioè non per forza un valore davvero raggiunto, ma il più basso valore maggiore o uguale di tutti quelli possibili   
$||Tf|| = sup_{f \neq 0} \frac {||Tf||}{||f||} = sup_{||f||=1} ||Tf||$  

**Definizione norma**

1- T applicato ad una funzione  
$||Tf||^2 = \int |Tf|^2 \, dx$  

2- T applicato ad un vettore $v=x_ne_n$:  
$||Tv||^2 = \Sigma |x_n|^2 |Te_n|^2$  

**Trovare norma**

come dimostrare che A è davvero raggiunto:  

1- trova funzione f per cui $||Tf|| = A||f||$  
cioè: la norma è una MAX  
poichè per def la norma è la maggiore delle norme possibili, allora $|T| \ge A$  
combinalo con $T \le $ della limitatezza, ottengo $T=A$  
  
2- trova successione $f_n$ per cui: $\frac{||Tf_n||}{||f_n||} \rightarrow A$    
cioè: la norma è una SUP (non viene effettivamente mai raggiunta da nessuna f)  
cioè quando $\le$, sii più stringente, fai il sup  invece di upper bound-> poi scegli $f_n$ concentrata nella x scelta  

#### Per calcolare

**Inequalities**

what | formula | equality when | notes
---|---|---|---
schwarz inequality |  $\|<A,B>\|^2 \le <A,A> \, <B,B> = \|A\|^2 \, \|B\|^2$ | $Af=\lambda Bf$, cioè sono linearmente dipendenti <br> (if i need a specific f for which there is equality, i put $\lambda =1$) <br> se ottengo f=0: probabilmente il termine misto è sempre nullo (calcola l'integrale e controlla) | usa con norma quadrata
triangle inequality |  $\|A+B\| \le \|A\| + \|B\|$  | come Schwarz | usa ocn valore assoluto  
trig | $ A (\cos x)\le A$ | 1- f con $A=0$  <br> 2- f_n centrata nella x_0 che mi dà $\cos(x_0)=1$  
upper bound del dominio | $\int_a^b x \le \int _a^b b$ <br> $x \, f(x) \le b \, f(x)\quad $ if  $L^2(a,b)$  | $f_n$ centrata in $x=b$ |
cambio supporto | $\int_1^\infty \|f\|^2 \le \int_0^\infty \|f\|^2$ | f con supporto in  $[1, \infty)$ <br> cioè f=0 nell'intervallo [0,1]  
parità | $\|d\| \le \|p\| + \| d \| = \| f\|$ <br> $\|Df\| \le  \|f\|$ | $p=0$

valgono sia per $|f|$ che per $|f|^2$    
$|fg| = |f| \, |g|$  
$|e^{ix}|=1 $ (perchè è solo una rotazione) (!!! deve apparire i all'esponente)  
$|sign(x) \, f | = |-f| = |f|$

esempio no:

shortcut:

#### Impostazione: T=A+B

**Generale**

$||Tf||^2 = ||(A+B)f||^2 $   
$= <Af, Af> + <Bf, Bf> + ⟨Af,Bf⟩+⟨Bf,Af⟩$   
$=||Af||^2 +||Bf||^2 + \,  \int Af \, \overline{Bf} +  \int Bf \, \overline{Af}  $     
$=||Af||^2 +||Bf||^2 + 2Re⟨Af,Bf⟩$  

**Limitatezza: facile**

$||Tf|| = ||(A+B)f|| $  
$ = ||Af+Bf ||$  
$\le ||Af|| + ||Bf||$  

**Limitatezza: difficile (non usare)**

$||Tf||^2  =  $  
$\le  ||Af||^2 +||Bf||^2 + 2 ∥Af∥∥Bf∥  $    
$\le  ||Af||^2 +||Bf||^2 \; + ||A||^2 + ||B||^2 $   
$ \le 2(||Af||^2+\|Bf||^2)  $  

come: cauchy + disug quadrato binomio 

#### Esercizi

**1. Disug triang, trovo f=0**

$Tf = f(-x) + sgn(x) \, f(x)$ 
$L^2(-1, 1)$  

1- limitatezza facile  
$|Tf| \le |Af|+|Bf| = |f| + |f| = 2 |f|$  
$||T|| \le 2$  

2- norma: trovo per f per cui Af=Bf  
l'unica sol è f=0, questo metodo non va bene, calcolo il prodotto misto

4- norma: calcola prodotto misto  
$|Tf|^2 = |Af|^2 +|Bf|^2 +⟨Af,Bf⟩+⟨Bf,Af⟩$   
$ = 2|f|^2 + \int f(-x) \overline{sgn(x) \, f(x)} \, dx  \; +  \; \int sgn(x) \, f(x) \overline{f(-x)} \, dx$  
$ = 2|f|^2 + \int sgn(x) [f(-x)\overline{f(x)} + f(-x)\overline{f(x)}] \, dx$  
[] è pari rispetto a x=0 (perchè g(-x)=g(x)), sgn(x) è dispari  
quindi l'integrale dà sempre 0  
quindi: $|Tf|^2 = 2|f|^2 $  



**2. Disug triang**

$Tf(x) = \frac{1}{2}(e^{ix}f(-x)+e^{-ix}f(x))$  

1- limitatezza  
$|Tf| \le |Af|+|Bf|$  
$ = \frac 12 |f| + \frac 12 |f| $  
$=|f|$  
quindi: $|T| \le 1$  

2- norma: cerco f per cui Af=Bf  
$e^{ix}f(-x) = e^{-ix} f(x)$   
$d(-x)=d(x)$  
cioè: d(x) è dispari 
dove ho definito: $f(x)=e^{ix}d(x)$   

per questa f ho:  
$|Tf| = 2|Af| = 2 \frac 12 = 1$  
$|T| \ge  1$, perchè è il sup  
quindi: $|T|=1$  

**3. Upper bound del dominio**

$Tf = x \, f(x)$  
$L^2(0,1)$  

1- limitatezza:  
$||Tf||_2^2 = \int_0^1 |x \, f(x) |^2 \, dx = $  
$= \int_0^1 x^2 \, |f(x)|^2 \, dx $     
$\le \int_0^1 |f(x)|^2 \, dx = ||f||_2^2$  
$|T| \le |f|$  

2- norma: shortcut    
scegliendo funzioni sempre più concentrate vicino a x=1, ottengo |T|=1  

3- norma: fatto bene  
scelgo successione concentrata in x=1: $f_n(x) = \sqrt{n} \chi_{(1-\frac 1n, 1)}(x)$  
che ha: $||f_n||^2_2 = \int_{1-1/n}^1 n \, dx = 1$    
$||Tf_n||_2^2 =  \int_0^1 |x \, f_n(x) |^2 \, dx =$  
$=  n \int_{1-1/n}^1 x^2 \, dx = $  
$\ge n(1-\frac 1n)^2 \frac 1n \quad $ ( min^2 x interval_length)  
$ = (1-\frac 1n)^2 \rightarrow 1$  
quindi $||T|| \ge 1$   
unendo le disuguaglianze, $||T||=1$   

**4. Trig**
  
$Tf = p + \cos (2x) \; d$  
$L^2(-1,1)$   

esercizio | cosa
---|---
$\|Tf\|^2 = \|p\|^2 + \|\cos (2x) \; d\|^2$ | separa in pari e dispari 
$ \le \|p\|^2 + \|d\|^2 = \|f\|^2$ <br> quindi $\|T\| \le 1$| upper bound  
con uguaglianza per f pari (d=0) <br> quindi $\|T\|=1$ | norma 

**5. Cambio supporto**

$Tf = f(x) \chi_{[1, \infty]}$   

1- limitatezza  
$|Tf|^2 = \int_1^\infty |f(x)|^2 dx $  
$\le \int_0^\infty |f(x)|^2 dx$  
$= |f|$  

2- norma  
l'upper bound diventa uguaglianza per le funzioni f che hanno supporto (dominio) in $[1, \infty)$ (cioè valgono 0 fuori dal quell'intervallo)  

**6. Trig**

$Tf =(\sin x + \cos x)f(\pi-x)$  
$L^2(0, \pi)$  

1- limitatezza  
$|Tf| = |\sin x + \cos x| \; |f(\pi-x)|$  
$ =\sqrt{|\sin x + \cos x|^2} \; |f(x)|$    
$= \sqrt{\sin^2x + \cos^2 x + 2 \sin x \cos x} \; |f(x)|$  
$= \sqrt{1+ \sin (2x)} \; |f(x)|$   
$ \le \sqrt{2} |f|$  

2- norma  
$sup \|1+\sin(2x)\| = 2$     
raggiunto quando $x=\frac \pi 4$  
quindi scegli f_n concentrata in $x=\frac \pi 4$    

**t. Var change**

esercizio | cosa
---|---
$\|Tf\|^2 = \int_0^\pi \|(\sin x + \cos x)\|^2 \|f(\pi-x)\|^2 \, dx $| serve cambio var, scrivi come integrale  
$\int_0^\pi (\sin^2x + \cos^2x + 2\sin x\cos x) \|f(\pi-x)\|^2 \, dx$ <br> $= \int_0^\pi (1+ \sin(2x)) \|f(\pi-x)\|^2 \, dx$ <br> $ \le \int_0^\pi 2 \|f(\pi-x)\|^2 \, dx$ | upper bound
$= \int_0^\pi 2 \;  \|f(y)\|^2  \, dy$ | $y= \pi-x$ <br> ricorda di scambiare i segni di dy e gli estremi  
$\|T \|= \sqrt 2$ | soluzione  
$ sup \|\cos x + \sin x \| ^2 $ <br> $ = sup \|1+\sin(2x)\| = 2$ | norma: trova il sup dove ho fatto l'upper bound

**Trig**

1- metodo poco elegante:  
$Tf = p + \cos (2x) \; d$  
$|Tf|^2 = |p|^2 + |\cos (2x) \; d|^2 \le |p|^2 + |d|^2 = |f|^2$  
quindi $|T| \le 1$  


**Parità**

$Tf=sign(x)[f(-x)-f(x)] = -2SDf$  
$|Tf|^2 = 4 |S|^2 |Df|^2 = 4 |Df|^2 \le 4 |f|^2$   
quindi $|T|\le 2$  

uguaglianza se f è dispari, così Df=f  


### Aggiunto

**Generale**

$<g, Tf> = \int g \, \overline{Tf} $  
$ = ... = \int T^*g \, \overline f = <T^*g, f>$  
guarda cosa è $T^*$  

note: 

1- integrale finale deve avere:  
stessi estremi di integr di quello iniziale    
$\overline {f(x)}$ deve avere solo x come argomento  
mentre $T^*g$ può essere qualunque espressione, con qualunque argomento    

2- il coniugato può stare al primo o al secondo membro, ma sii consistente  
bisogna metterlo sulle $f$ e sulle $i$ ($\overline i = -i$)  

**Attenzione: sostituzioni**

se faccio un cambio var, ricorda di cambiare:  
1- estremi di integrazione  
2- differenziale  

**T somma di autovalori: T=A+B**

$T^* = (A+B)^*=A^*+B^*$  
viene di conseguenza dal separare l'integrale in una somma  

se A è unitario, allor a $A^*=A^{-1}$, che è più facile da calcolare  

**Usando parità**

meglio scrivere il prodotto scalare con la bilinearità, invece che con l'integrale 

ricorda che (p,d)=0 solo se solo pari e dispari rispetto al punto medio di L^2  

#### Esempi

**Esempio: aggiunto**

$Tf = e^{ix} f(-x)$  
su $L^2(-\infty, \infty)$  

esercizio | cosa
---|---  
$⟨g,Tf⟩ = \int_{-\infty}^\infty g(x) \; e^{-ix} \overline {f(-x)} \; dx$ | scrivo come integrale   
$ = \int_{\infty}^{-\infty} g(-y) e^{iy}\overline{f(y)} \; (-dy)$ | cambio var: y=-x 
$\int_{- \infty}^{\infty} g(-y) e^{iy}\overline{f(y)} \; dy  =  ⟨T^*g, f⟩$ | scambio estremi 
$T^*g = e^{ix} \; g(-x)$  <br> cioè $T^*=T$    | soluzione


**Esempio: aggiunto**

$Tf(x) = \sin^ 2 (x)\, f(x) + \cos ^2 (x) \, f(-x)$  

esercizio | cosa
---|---
$Tf(x) =  \int _a^b g(x) \; \sin^ 2 (x)\, \overline{f(x)} + \cos ^2 (x) \, \overline{f(-x)}$  <br> | scrivo come integral <br> separo in una somma 
$= \int _a^b g(x) \; \sin^ 2 (x)\, \overline{f(x)} \quad + \int_a^b g(x) \;  \cos ^2 (x) \, \overline{f(-x)}$ |  separo in una somma  
$ = \int _a^b g(x) \; \sin^ 2 (x)\; \overline{f(x)} = ⟨\sin^2 (x) \; g(x),f(x)⟩$  | primo addendo (già pronto)  
$=  \int_b^a  g(-y) \; \cos ^2 (-y) \; \overline{f(y)} \, (-dy)$ <br> $= \int_a^b g(-y) \; \cos ^2 (-y) \; \overline{f(y)} \, dy$ <br> $= ⟨\cos^2 (x) \;  g(-x),f(x)⟩$    | secondo addendo  
$⟨g,Tf⟩ = ⟨\sin^2(x)g(x)+\cos^2(x) g(-x),f⟩$ <br> cioè $T^* = T$  | totale

**Esempio: aggiunto, arg complicato**

$Tf=e^{x/2}f(e^x)$   

esercizio | cosa
---|---
$⟨f, Th⟩  = \int_0^\infty f \, e^{x/2}h(e^x)$  | scrivo integrale
$= \int_1^\infty \frac{f(\ln z)}{\sqrt{z}} h(z) \, dz$ | cambio var: $z=e^{x}$   
$T^*f(x)=\frac{f(\ln z)}{\sqrt{z}} \chi_{[1, \infty)} (z)$ | soluzione <br> $\chi$ si usa per cambiare gli estremi di integrazione nella risposta <br> perchè non sono più quelli iniziali di $L^2$, sono cambiati nel cambio var

**Esempio shortcut: aggiunto SENZA prodotto scalare**  

$Tf= e^{ix}f(\pi-x) + e^{-ix}f(\pi+x)$  

esercizio | cosa
---|---
$Tf= Af(x)+Bf(x)$ <br>  $A$ e $B$ sono unitari  | separo in una somma  
$g(x)=Af =e^{ix}f(\pi-x)$ <br> $g(\pi-y) = e^{i(\pi-y)} f(y)$ <br> $g(\pi-y)  = (-1) e^{-iy} f(y)$  <br> $f(y) = -e^{ix}g(\pi-y) $  <br> $A^*g=A^{-1}g =  -e^{ix}g(\pi-y)$ | primo addendo <br> cambio var $y=\pi-x$  
... <br> $B^*g= B^{-1}g =e^{-iy}g(\pi+y)$  | secondo addendo  
$T^*f=A^*f(x)+B^*f(x) = -e^{ix}f(\pi-y) + e^{-iy}f(\pi+y)$ | totale

$⟨g, Tf⟩$ | testo

### Chaining

**Steps**

se devo calcolare $T^*Tf$:  

1- parti dall'esterno: scrivi $T^*f$   
2- ogni f scambiala con $Tf$ (mettendoci dentro l'argimento giusto)  


**Esempio**

$Tf = x f(x)$  
$T^*f= 2 x^2 f(x^3)$  

$T^*(Tf)= 2 x^2 [Tf(x^3)] = 2x^2[x^2 f(x^3)]=2x^5 f(x^3)$  

### Inverso

**Generale**

$Tf(x) = ... = g(x)$    
manipola algebricamente per isolare $f(x)$ (con arg x), così trovo:  
$f(x) = T^*g(x)$  

in pratica, devo risovlere eq ricorsiva per f, fingendo che g(x) sia qualcosa di noto 

**Esempio**

$Af = (T-\lambda I)f = f(-x)+\epsilon(x)f(x)-\lambda f$   
$\epsilon(x)=sign(x)$  

esempio | cosa
---|---
$ Tf = f(-x)+\epsilon(x)f(x)-\lambda f =g(x)$ | condizione 
$\begin{cases} f(-x)+\epsilon(x)f(x)-\lambda f =g(x) \\ f(x)-\epsilon(x)f(-x)-\lambda f =g(-x) \end{cases}$  | devo risolvere per f, appaiono sia +x che -x <br> uso il metodo del sistema +x,-x <br> $\epsilon(-x)=-\epsilon(x)$  
$  f(-x) =g(x) - \epsilon(x)f(x)+\lambda f$ <br> $(2-\lambda^2)f(x) =g(-x)+(\epsilon(x)+\lambda)g(x)$   | risolvo, sostituendo f(-x)  
$f(x) = \frac{g(-x)+(\epsilon(x)+\lambda)g(x)}{2-\lambda^2} $ <br> $\lambda \neq \pm \sqrt 2$| soluzione <br> perchè $f(x)=A^{-1}g(x)$  

**Trova inverso**

vai a tentativi  
voglio inverso di $T-\lambda I$, so che $T^2 = 2I$  

provo come inverso: $T+\lambda I$  
$(T-\lambda I) (T+\lambda I) = T^2 -\lambda^2 I = (2-\lambda^2)I$  
quindi l'inverso: $(T-\lambda I)^{-1} = \frac{T+\lambda I}{2-\lambda^2}$  
ed esiste per $\lambda \neq \pm \sqrt 2$ (sennò viola la C.E.)  

volendo, posso scrivere la risposta come $(T+ \lambda I)f$, cioè calcolandolo proprio  

### Autovalori

**Generale**

$Tf = \lambda f$  

è eq ricorsiva per f  
devo vedere se c'è una f non-0 che la rende valida  
e in caso:  
1- autovalori: che $\lambda$ servono per renderla valida?  
2- autofunzioni: quali f hanno questa $\lambda$?   

**Tips per impostare**

1- spesso utile separare $\lambda=0$ e $\lambda \neq 0$  

2- nota che l'uguaglianza deve valere per OGNI x, quindi in pratica x NON deve apparire  
quindi se appare una x devo mettere a 0 qualcosa che gli è moltiplicato, così sparisce  
    
3- spesso la domanda è provare che non ci sono autovalori diversi da 0


**Casi speciali**

1- so una particolare formula per $T^2$ facile  
cosa faccio: $T^2f = \lambda^2f$     

2- T è unitario o isometrico    
cosa faccio: $\pm 1$ sono gli unici autovalori  
perchè $|Tf|^2=|f|^2$ per OGNI x  

3- se f è f(x)=f(x+c) ???

**Attenzione** 

$T^2f = \lambda^2f$  
$\lambda$ è autovalore di T  

$T^2 f = \lambda f$  
$\lambda$ è autovalore di $T^2$  

### Spazio ortogonale

**Cosa è**

lo spazio formato da tutte le funzioni ortogonali ad un'altro  

$\int f \, g dx = 0  \quad \forall g \in H$  
f è l'insieme di tutte le funzioni dello spazio $H^\perp$  

**Esempio**

ho: $H=\{e^{ix}d(x) \; | d \; dispari\}$   

il suo ortogonale è: $H^{\perp} = \{ e^{ix}p(x) \; | \; p \, pari\}$   
perchè: $\int e^{ix}d(x) \; \overline{e^{ix}p(x)} = \int p(x) d(x) = 0$  

nota:  
l'esponenziale serve, perchè $e^{ix}=\cos x + i \sin x$, cioè ha sia una parte pari che una dispari  
quindi quando calcolo $\int e^{ix} d\, p$, se ci appare anche l'esponenziale la funzione integrale non è dispari  
se metto l'esponenziale anche nello spazio ortogonale, poichè faccio il coniugato del secondo membro si cancellano
