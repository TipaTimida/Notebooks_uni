### Definizione

**Operatore integrale**

un operatore in $L^2(a,b)$ che si scrive:  
$(Tf)(x) = \int_a^b K(x,y) \; f(y) \, dy$  


**Kernel**

K(x,y): kernel  
la funzione che pesa quanto il valore di f(y) contribuisce al valore di Tf(x)  

come si trova:  
se è già scritto come un unico integrale, facile  
sennò è spiegato negli utils  

#### Esempi

**Esempio: trova kernel**

$(Tf)(x)= \int_{-1}^{1} \theta(x-y)\theta(x+y) f(y) \, dy$  
$K(x,y) =  \theta(x-y)\, \theta(x+y)$  

**Esempio: trova kernel**

$Tf=\int_0^x f(y) \, dy -\int_x^ \pi f(y) \, dy$  

lo definisco a tratti:  
$K(x,y) = \begin{cases} 
+1 & 0 \le y < x \\ 
-1 & x < y \le \pi 
\end{cases}$  
oppure: $K(x,y) = sign(x-y)$  

ignora i punti sulla diagonale x=y, tanto sono pochi, non influenzano tanto il valore del kernel  

### Limitato e compatto

**Operatore di Hilbert-Schmidt**

condizione:  
$\int \int |K(x,y)|^2 \, dx \, dy  < \infty$    

gli estremi di itnegrazione sono gli stessi di L, al solito, sia per x che y  
cioè, la richiesta è che: $K \in L^2( (a,b) \times (a,b))$


**Conseguenze**

1- implicazioni:  
operatore di Hilbert-Schmidt -> compatto -> limitato   

2- posso trovare una stima superiore per la norma: $|T| \le |T_{HS}| = |K|_{L^2}$    
dove l'ultimo è risultato dell'integrale di $K$ su $L^2$, fatto sopra per provare che è di Hilbert-Schmidt  
questa è veramente una stima superiore, più del fatto che la norma è già una stima in sè   

#### Esempi

**Esempio: prova che è limitato**

$K(x,y) = sign(x-y)$  
sono su $L(0, \pi)$  
 
$\int_0^\pi \int_0^\pi 1 \, dx \, dy = \pi^2$  
perchè K=sign è o +1 o -1, quindi $K^2=1$  

### Aggiunto

**Come si fa**

$K(x,y)\quad  \rightarrow \quad \overline{K(y,x)}$  

1- scambia x e y nel kernel  
2- metti il coniugato nel kernel  
di solito il kernel è reale, quindi il coniugato non fa niente  

**Spiegazione**

posso anche usare il metodo solito di scrivere $<Tf,g>$, mi verrà esattamente questa cosa  
nota che dentro l'integrale del prodotto scalare c'è un altro integrale, quello di Tf  

$<g,Tf> = \int g(x) \; \overline{[\int K(x+y) \, f(y) \, dy ]}  \; dx$  
$= \int [\int g(x) \overline{K(x,y)} \; dx] \; \overline{f(y)} \; dx $    
quindi ho preso l'aggiunti di K e scambiato le var x,y  

#### Esempi

**Esempio: aggiunto**

$K(x,y)=sign(x-y)$  
sono su $L^2(0, \pi)$  

$(T^*f)(x) = \int_0^\pi \overline K(y,x) f(y) \, dy =  \int_0^\pi  sign(y-x) \, dy = -  \int_0^\pi K(x,y) dy$  
quindi: $T^*=-T$  

**Esempio: integrale**

$Tf= \int^{1}_{-1}K(x,y) f(y) \, dy$  
$T^*f= \int^{1}_{-1}K(y,x) f(y) \, dy$  


### Autovalori e autofunzioni

**Come si fa**

1- scrivi la condizione $Tf = \lambda f$    
2- deriva da emtrambi i lati  

**Compute derivative of an integral**

$F'(x) = f(upper bound) \cdot (upper bound)' - f(lower bound) \cdot (lower bound)'$

where f integrand function    
obv, if one bound has termine noto, its derivative is 0, so that term don't appear  

eg:    
$F(x) = \int _{x^2}^{x^4} \frac{1}{ln(t)} dt$  
$F'(x) = \frac{4x^3}{ln(x^4)} - \frac{2x}{ln(x^2)}$  

#### Esempi

**Esempio: Trova autovalori e autofunzioni**

$Tf = \int_0^x f(y) \, dy -\int_x^\pi f(y) \, dy$  

esercizio | cosa
---|---
caso: $Tf = 0$ <br> $f(x) -(-f(x))=0$ <br> $2f = 0$ <br> quindi non ci sono autofunzioni (quindi neanche autovalori) | caso: $\lambda = 0$    
caso: $Tf = \lambda f$  <br> $2f = \lambda f'$  <br> $f(x)= Ae^{(2/\lambda) \, x}$ <br> ho trovato le autofunzioni, ma non so ancora quali n sono autovalori | caso: $\lambda \neq 0 $ 
condizione: $f(0)+f(\pi)=0$  <br> $e^{2 \pi / \lambda}=-1$ <br> $\frac {2 \pi} \lambda = i(2n+1) \pi \quad$ con $n \in Z$   <br>  $\lambda_n = -\frac{2i}{2n+1}$   <br> sostituisco nelle autofunzioni e tolgo la A davanti: <br> $f_n(x) = e^{i(2n+1)x}$   | trova autovalori nel caso $\lambda \neq 0 $ <br> per farlo, devo ricavare una condizione dalla formula di Tf <br> ad esempio $f(0)+f(\pi)=0$    
  

**Esempio: integrale**

$Tf(x) = \int_{-x}^xf(y) \, dy$   

per $\lambda = 0$  
derivo: $f(x)+f(-x)=0$  
cioè f è dispari (quindi le f dispari sono autovettori di autovalore 0)  

per $\lambda \neq 0$   
Tf è dispari, perchè $Tf(-x)=-Tf(x)$   
ma $Tf=\lambda f$, quindi anche f deve essere dispari  
ma per f dispari, Tf=0 (integrale di cosa dispari su dominio simmetrico)  
quindi f=0  
non ci sono altri autovalori 

**Autovalori di $T^*T$**

$Tf= 2 \int_0^x p(y) \, dy$   
$T^*f = \int_{|x|}^1 p(y) \, dy$  
$T^*Tf = \int_{|x|}^1 2 \int_0^x p(y) \, dy$    
ho anche usato la parità  

caso: $\lambda = 0$  
$ker \; T^*T = \; ker T$  
perchè: $<T^*Tf, f> = <Tf, Tf> = |Tf|^2$ che è una norma, quindi è 0 solo se Tf=0  
$Tf=0$ quando f ha solo parte dispari, quindi queste sono le autofunzioni di 0 (le f tali che $f(-x)=-f(x)$)  

caso: $\lambda \neq 0$  
$T^*T$ è pari, perchè $T^*T(-x)=T^*T(x)$  
poichè $T^*Tf = \lambda f$, anche f è pari  
derivo entrambi i lati:    
$ -2 \int_0^x p(y) \, dy = \lambda p'(x) $  
$-2p(x) = \lambda p''(x)$  
$p''(x)=-\frac{2}{\lambda}p(x)$  
soluzione: $p(x)=A \cos (k x) + B \sin (k x)$ con  $k^2 = \frac{2}{\lambda}$    
ricorda che p deve essere pari, quindi B=0, non può esserci il seno  

ora so le autofunzioni, devo cercare gli autovalori:  
vedo la condizione:  $T^*Tf(1)=0$  
$T^*T \, p(1)=\lambda p(1) = 0$    
quindi la parte pari di f deve essere nulla quando pluggo x=1: $A \cos(k)=0$   
non posso avere A=0, o non avrei sol, quindi $\cos k = 0$, cioè $k = (n+\frac{1}{2})\pi$  
$\lambda _n = \frac{2}{k^2} = \frac{8}{(2n+1)^2 \pi^2}$ per n intero, da 0 in poi  


### Altro

**Parità**

$F(x) = \int f(x)$  

if $f(x)$ is even, $F(x)$ is odd and viceversa  
means taking the derivative or the integral changes the parity
