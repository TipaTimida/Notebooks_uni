### Funzioni comuni

**Funzione di Heavyside**

$\theta(t) = \begin{cases}
1 &  t>0 \\
0 &  t<0 \end{cases}$  

**Funzione segno**

$sign(t) = \begin{cases}
1 &  t>0 \\
-1 & t<0 \end{cases}$  

**Funzione caratteristica di un intervallo**

$χ_{[1,∞)}(x)​ = \begin{cases}
1 &  x \in [1, \infty) \\
0 & else
\end{cases}$   

funzione caratteristica (o indicatrice) dell’intervallo $[1,∞)$  
si usa per restringere l'intervallo di integrazione in un integrale  

**Nota**

1- in entrabi i casi, $t=0$ lo considero irrilevante in $L^2$  
tanto l'insieme dei punti con t=0 ha misura nulla in $L^2$  

2- l'argomento può anche anche essere qualcosa di più lungo, tipo $\theta(x-y)$  
basta vedere quando l'arg è positivo o negativo

**Nota: funzione caratteristica di un intervallo**

$\chi_{[1, \infty](e^x)} = \chi_{[0, \infty](x)}$  
perchè per avere $e^x$ da 1 a infinito devo avere x da 0 a infinito  

accade per cambio var $e^x \rightarrow x$  
o quando pluggo qualcosa dentro la x, per fare funzione composta   

### Come lavorarci

**1. Scrivere prodotto scalare (norma, aggiunto)**

separa l'integrale in una somma  
in ogni tratto ci metto la funzione come è definita per quel tratto  

1- norma:  
l'upper bound è che immagino che su tutto il dominio sia definita come sul tratto maggiore  
si raggiunge usando una f che ha supporto solo sul tratto che ho scelto  
se faccio dei var change, ricorda ci cambiare gli estremi di integrazione, probabilmente alcuni tratti si scambiano  

2- aggiunto:  
anche alla fine otterrò una funzione a tratti  

**2. Autovalori**

scrivi la definizione $Tf(x) = \lambda f$ per x in ogni tratto  

1- se non ci sono var change:   
la valuto nei due tratti, da cui ottengo due euqazioni slegate  
facile, trova un autovalore diverso per ogni tratto   
gli autovettori sono le f con supporto solo in quel tratto  

2- se ci sono var change:   
devo risolvere un'equazione ricorsiva dove Tf è una funzione a tratti  

**Esempio: senza var change**

$Tf = \begin{cases}
-f(x) & [-1,0] \\
2f(x) & [0,1] \\
 \end{cases}$  

1- norma  
$|Tf|^2 = \int_{-1}^0 |-f(x)|^2 \, dx + \int_0^1 |4f(x)|^2 \, dx$   
$ = \int_{-1}^0 |f|^2 \, dx + 4\int_0^1 |f|^2 \, dx$   
$\le 4 (\int_{-1}^1 |f|^2 dx) = 4|f|^2$   
raggiunta con una f che ha sipporto in $[0,1]$  

2- aggiunto:  
$(g, Tf) = \int_{-1}^0 \overline{g(x)} [-f(x)] \, dx  + \int_{0}^1 \overline{g(x)} [2f(x)] \, dx $  
$ = \int_{-1}^0 [-\overline{g(x)}] \, f(x) \, dx  + \int_{0}^1 [2 \overline{g(x)}] \, f(x) \, dx $    
$T^*g = \begin{cases} 
-g(x) & [-1, 0] \\
2g(x) & [0,1] \end{cases}$  

3- autovalori  
$\begin{cases}
-f(x) = \lambda f(x) & [-1,0] \\
2f(x) = \lambda f(x) & [0,1] \end{cases}$  
$\lambda = -1$, autovettori: $f=\chi _{[-1,0]}(x)$, cioè f con supporto in [-1,0]  
$\lambda = -2$, autovettori: f con supporto in [0,1]  


**Esempio: con var change**

$Tf = \begin{cases} 
f(x+1) & [0,1] \\
f(x-1) & [1,2] \end{cases}$  

1- norma  
$|Tf|^2 = \int_0^1 |f(x+1)|^2 dx + \int_1^2 |f(x-1)|^2 dx$  
$ =  \int_1^2 |f(y)|^2 dy + \int_0^1 |f(t)|^2 dt$  
$=\int_0^2 |f|^2 dx = |f|^2$   

2- aggiunto  
$(g, Tf) = \int_0^1 \overline{g(x)} f(x+1) dx + \int_1^2 \overline{g(x)} f(x-1) dx$  
$= \int_1^2 \overline{g(y-1)} f(y) dy + \int_1^2 \overline{g(t+1)} f(t) dt$   
$= (T^*g, f)$  
$T^*g = \begin{cases}
g(x+1) & [0,1] \\
g(x-1) & [1,2] \end{cases}$   

**Esempio: autovalori con var change (metodo 1)**

equazione ricorsiva: $Tf(x) = \lambda f(x)$  
dove Tf è una funzione a tratti   
$Tf = \begin{cases}
f(x+1)  & x \in [0,1] \\
f(x-1) & x \in [1,2] \end{cases}$  

1- separo gli autovettori in tratti uguali   
$f(x) = \begin{cases}
h(x) & [0,1] \\
g(x) & [1,2] 
\end{cases}$  
con h arbitrario, voglio sapere g in funzione di h  

2- valuto l'equazione nel primo tratto, $x \in [0,1]$  
$Tf(x) = \lambda f(x)$  
$f(x+1) = \lambda f(x)$  
$g(x+1) = \lambda h(x)$  
$g(y) = \lambda h(y-1)$  

3- autovettori (pluggo g):  
$f(x) = \begin{cases}
h(x) & [0,1] \\
\lambda h(x-1) & [1,2] 
\end{cases}$  

4- autovalori: valuto l'equazione nel tratto $x \in [1,2]$  
$Tf(x) = \lambda f(x)$  
$f(x-1) = \lambda f(x)$   
$h(x-1) = \lambda \lambda h(x-1)$  
$\lambda^2 = 1$  
$\lambda = \pm 1$  

$f(x)+f(-x)=\lambda f(x)$  

**Esempio: autovalori con var change (metodo 2, brutto???)**

3- autovalori  
$\begin{cases}
f(x+1) = \lambda f(x) & x \in [0,1] \\
f(x-1) = \lambda f(x) & x \in [1,2] \end{cases}$  

1- var change -> per portare la var nel secondo tratto ad avere lo stesso dominio del primo tratto    
quindi nel SECONDO tratto, faccio $y=x-1$, così $y \in [0,1]$, come nel primo tratto  
$f(y)=\lambda f(y+1)$  

2- questa nuova che ho trovato, e il primo tratto dell'originale, hanno entrambi dominio [0,1]  
quindi queste due lo posso sostituire una nell'altra, come nel solito nel sistema con var change  
cioè: poichè $y \in [0,1]$, devo sostituirci f(x+1) presa dal PRIMO tratto, dove $x \in [0,1]$  
$f(y) = \lambda [\lambda f(y)]$  
quindi: $(a-\lambda^2)f(y)=0$  
$\lambda = \pm 1 $  

tip:  
sarebbe stato più veloce notare che $T^2=I$, quindi $\lambda = \pm 1$  

4- autovettori  
caso $\lambda = 1$  
dalla def di Tf vedo che, fissato il primo tratto, il secondo è obbligato  
quindi scelgo arbitrariamente una funzione $h \in L^2(0,1)$, che sta nel primo tratto  
e sul secondo, ci va: 
$f(x) = f(x-1)$, poichè $x \in [1,0]$ (f(x-1) quindi è quella definita sul primo tratto)  
$f(x) = \begin{cases}
h(x) & x \in [0,1] \\
h(x-1) & x \in [1,2] \end{cases}$

caso: $\lambda = -1$  
anche qui, fissato il primo tratto, il secondo è obbligato  
perchè sul secondo tratto ho $f(x) = -f(x-1)$, dove $x \in [1,2]$  
$f(x) = \begin{cases}
h(x) & x \in [0,1] \\
-h(x-1) & x \in [1,2] \end{cases}$


### Kernel con funzione a tratti

**1. Riscriverla**

1- definisci T a tratti  
2- disegna le zone dove T ha i vari valori  
3- cerca di descrivere con un integrale la zona dove ha +1  

#### Esempi: trova kernel

**Graficamente**

$Tf = \int \theta(x+y)\theta(x-y) f(y)\, dy$  
dominio $L^2(-1,1)$   

esercizio | cosa
---|---
$Q = \{(x,y): \|x\|<1, \; \|y\|<1\}$ <br> è un quadrato con i veritici nei punti (1,1) , (1-,1)...| disegno il dominio $L^2(-1,1)$
$K(x,y) = \begin{cases} 1 & if \;  x+y>0,\; x-y>0 \\ 0 & else \end{cases}$ | scrivo il Kernel (o la funzione) a tratti  
$x+y>0 \rightarrow y>-x$ <br> $x-y>0 \rightarrow y<x$ <br> quindi: $-x<y<x$  <br> $A= \{(x,y)\in Q : \quad -x<y<x\} $ | descrivo la regione A, dove K=1
disegna le rette $y=-x$ e $y=+x$ <br> per la prima considera la parte sopra, per la seconda quella sotto <br> quindi mi viene il triangolo a destra tra le bisettrici | disegno A 
$Tf = \theta(x) \int_{-x}^x f(y) dy$ <br> $= \begin{cases} 0 & if -1<x<0 \\ \int_{-x}^xf(y) \, dy & 0<x<1 \end{cases}$  | scrivo analiticamente <br> A è spannata da linee verticali (dy) che stanno tra le due bisettrici (y=-x -> y=+x) considerando solo la parte a destra ($\theta(x)$, ancora una heavyside)    

**Riscrivi a tratti**

$Tf = \int_0^x f(y) \, dy -\int_x^\pi f(y) \, dy$  
$L^2(0, \pi)$  

esercizio | cosa
---|---
$Q = \{(x,y): 0<x<\pi, 0<y<\pi\}$ <br> è solo il primo quadrante, un quadrato di lato $\pi$ | disegna il dominio  
regione A, dove K=1: <br> tra le righe y=0 e $y= x$, quindi sotto la bisettrice <br> $A = \{0<y<x\}$ <br> regione B, dove K=-1: <br> tra le righe y=x e $y=\pi$, quindi sopra la bisettrice <br>  $B = \{x<y<\pi\}$ <br>| disegna regioni
$K(x,y)=\begin{cases} +1 & 0≤y<x \\ -1 & x<y≤π  \end{cases}$ | scrivi a tratti <br> le righe le prendo dagli estremi dell'integrale 
$K(x,y) = sign(y-x)$ | osservazione <br> nota che y=x è la bisettrice
