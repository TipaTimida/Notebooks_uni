### Equazioni ricorsive con f(x) e f(-x)  

**Quando appaiono**

1- condizione per trovare gli autovalori  
$Tf = \lambda f$   

2- condizione di uguaglianza della disuguaglianza di Schwarz (quando cerco la norma)  
$Af = Bf$  

nota: sposta tutto a sinistra, così ho $Tf-\lambda f=0$

metodo | quando | cosa | note 
---|---|---|---
sistema pari-dispari | se riesco trovare parte pari e dispari dei due lati dell'equazione | $\begin{cases} P(Tf) = P(\lambda f) \\  D(Tf)=D(\lambda f)  \end{cases}$  
sistema +x, -x | se gli unici arg sono +x, -x | $\begin{cases} Tf(x) = \lambda\, f(x) \\ Tf(-x)=\lambda f(-x)  \end{cases}$ <br> poi plugga f(-x) da una nell'altra | 1- per trovare autovettori, plugga i valori di $\lambda$ trovati -> probabilmente otterrò un'altra eq ricorsiva su f, che posso risolvere separando f in tratti x>0 e x<0 (ma posso anche lasciare l'eq ricorsiva?) <br> 2- se le due equazioni vengono uguali, prova a individuare cose opposte  
individua cose opposte | se vedo una sotto-funzione che si ripete | definisci questa sottofunzione in rapporto ad f, studia l'equazione su di essa  
funzioni a tratti | se appare una funione a tratti (Tf, sign...) | 1. gli autovettori sono funzioni a tratti, con il primo un arbitrario h(x) <br> 2. valuta l'equazione nel primo tratto, per trovare il secondo tratto in funzione del primo | se poi voglio gli autovalori, devo valutare l'equazione del secondo tratto  
deriva entrambi i lati | se ho operatore integrale  

**Valutare un'equazione in un punto**

se l'equazione contiene una funzione A TRATTI (Tf, sign, ...)  
devo specificare in che punto la sto valutando  
1- guarda con che argomento la sto chiamando    
e trova a che tratto corrisponde quell'argomento        
2- pluggaci la definizione di quel tratto, MA usando come argomento quello con cui la sto chiamando (al posto della x)  

esempio:   
$f(x) = \begin{cases}
h(x) & x \in (0, \infty) \\
2x h(-x) & x \in (-\infty, 0)\end{cases}$   

$f(x)+f(-x)=0$    

calcolata in x>0:  
f(x) è presa dal primo tratto (perchè arg is x>0), con arg x  
f(-x) è presa dal secondo tratto (perchè arg is -x<0), con arg -x  
$h(x) + 2(-x)h(x) = 0$  

calcolata in x<0:  
f(x) è presa dal secondo tratto  (perchè arg is x<0), con arg x  
f(-x) è presa dal primo tratto (perchè arg is -x>0), con arg -x  
$2x h(-x) + h(-x)=0$  

#### Esempi

**1. Sistema pari-dispari**

$f(x)+f(-x)=\lambda f(x)$  

cosa | esercizio
---|---
parti pari e dispari | $(p+d) +\; (p-d)  =\; \lambda(p+d)$  
sistema pari e dispari | $\begin{cases} 2p = \lambda p \\ 0 = \lambda d \end{cases}$  
risolvi |  $\begin{cases} (2-\lambda)p = 0 \\ 0 = \lambda d \end{cases}$  
risposta | ho due possibili soluzioni: <br> 1- $\lambda =0$, $p=0$, cioè f dispari <br> 2- $\lambda = 2$, $d=0$, cioè f pari  

**2. Sistema +x, -x**

cosa | esercizio
---|---
scrivi sistema  | $\begin{cases} f(-x)+f(x)=\lambda f(x) \\ f(x)+f(-x)=\lambda f(-x) \end{cases}$  
da una delle due, isola $f(x)$ | $f(-x) = f(x)(\lambda-1)$ <br> dalla prima
pluggalo nell'altra | $f(x) + f(x)(\lambda-1) = \lambda f(x)(\lambda-1)$ <br> nella seconda 
risolvi |  $f(x)(-\lambda^2+\lambda)=0$ <br> $f(x)\lambda (\lambda-2)=0$  
risposta | ho due possibili soluzioni: $\lambda = 0$ e $\lambda = 2$    
plugga sol per trovare formule di f(x) | se $\lambda = 0$: $f(-x)+f(x)=0$, cioè f dispari <br> se $\lambda = 2$: $f(-x)=f(x)$, cioè f pari  

**3. Individua parti opposte**

equazione: $ e^{ix}f(\pi-x)=-e^{-ix}f(\pi+x)$   

che faccio:  
chiamo il lato sinistro: $e^{ix}f(\pi-x)=d(x)$  
la condizione diventa: $d(x)=-d(-x)$, cioè d deve essere dispari  

soluzione:  
$f(\pi-x) = e^{-ix}d(x)$  
$f(y) = e^{-i(\pi-y)}d(\pi-y)$  
$f(y) = -e^{iy} d(\pi-y)$ con d dispari  

**4. Funzioni a tratti**

$f(x)+f(-x)=\lambda f(x)$  
$f(-x) = f(x)(\lambda-1)$  

1- osservazione  
se stabilisco la metà di f con x>0, la metà x<0 è obbligata  
separo gli autovettori in tratti:   
$f(x) = \begin{cases}
h(x) & x>0 \\
g(x) & x<0 \end{cases}$   
con h arbitrario, voglio sapere g in funzione di h  

2- valuto l'equazione in x>0 (nel primo tratto)   
f(-x): dal secondo tratto, con arg -x  
f(x): dal primo tratto, con arg x  
$g(-x) = h(x) (\lambda -1)$  
io voglio sapere g(x), per scriverla in f(x) = ..., quindi trovala con un var change  
$g(x) = h(-x) (\lambda -1)$  

3- sostituiscilo nella funzione scritta a tratti:  
$f(x) = \begin{cases}
h(x) & x>0 \\
h(-x) (\lambda-1) & x<0 \end{cases}$  

4- per autovalori: calcolo in x<0 (l'altro tratto, quello che non ho usato prima)   
quindi f(-x) appartiene al primo tratto (ma con arg -x), f(x) al secondo  
$f(-x) = f(x)(\lambda-1)$  
$h(-x) = h(-x)(\lambda -1)(\lambda -1)$  
$1 = (\lambda -1)^2$  
$\lambda = 0$ oppure $\lambda = 2$

**Pari-dispari ???**

$Tf = \frac{1}{2}(e^{ix}f(-x)+e^{-ix}f(x))$  

1- separo in parte pari e dispari:  
$Tf = \cos x \, p -i \sin x \, d$   
è completamente pari  

2- scrivo sistema:  
$ \begin{cases}
 \cos (x) \, p -i \sin (x) \, d = \lambda p \\
 0 = \lambda d
 \end{cases}$  
quindi d=0, e poi $(\cos x \, -\lambda )p=0$, quindi non c'è soluzione per $\lambda \neq 0$

**4. Funzione a tratti**

equazione ricorsiva: $Tf(x) = \lambda f(x)$  
dove Tf è una funzione a tratti   
$Tf = \begin{cases}
f(x+1)  & x \in [0,1] \\
f(x-1) & x \in [1,2] \end{cases}$  

1- separo gli autovettori in tratti uguali   
$f(x) = \begin{cases}
h(x) & [0,1] \\
g(x) & [1,2] 
\end{cases}$  
con h arbitrario, voglio sapere g in funzione di h  

2- valuto l'equazione nel primo tratto, $x \in [0,1]$  
$Tf(x) = \lambda f(x)$  
$f(x+1) = \lambda f(x)$  
$g(x+1) = \lambda h(x)$  
$g(y) = \lambda h(y-1)$  

3- autovettori (pluggo g):  
$f(x) = \begin{cases}
h(x) & [0,1] \\
\lambda h(x-1) & [1,2] 
\end{cases}$  

4- autovalori: valuto l'equazione nel tratto $x \in [1,2]$  
$Tf(x) = \lambda f(x)$  
$f(x-1) = \lambda f(x)$   
$h(x-1) = \lambda \lambda h(x-1)$  
$\lambda^2 = 1$  
$\lambda = \pm 1$  

**Esempio: sistema +x, -x**

$Tf = \frac{1}{2}(e^{ix}f(-x)+e^{-ix}f(x))$  

1- scrivo sistema:  
$\frac{1}{2}(e^{ix}f(-x)+e^{-ix}f(x)) = \lambda f(x)$  
$\frac{1}{2}(e^{-ix}f(x)+e^{ix}f(-x))=\lambda f(-x)$  

2- risolvo:  
i lati sinistri sono uguali, li uguaglio:  
$\lambda f(x) = \lambda f(-x)$  
due casi:  
$\lambda =0$, che lo verifica  
$\lambda \neq 0$, che implica $f(x)=f(-x)$, cioè f pari  

3- caso $\lambda =0$  
basta isolare:  
$f(-x)=-e^{2ix} f(x)$  
le autofunzioni sono quelle con queste proprietà  
se voglio, posso risolvere l'eq ricorsiva per scrivere f a tratti, x>0 e x<0  

4- caso $\lambda \neq 0$  
ho detto che f deve essere pari, quindi riscrivo la condizione di autovettore:  
$e^{ix}f(x)+e^{-ix}f(x) = 2 \lambda f(x)$    
$f(x)(e^{ix+e^{-ix}})=2 \lambda f(x)$  
$2 \cos (x) \, f(x) = 2 \lambda f(x)$  
$(\cos x -\lambda)f(x)=0$  
non ha soluzione, perchè cos varia seconda x, qualunque $\lambda$ scelgo non può valere per tutti gli x    

**Esempio: cambio-var**

$Tf = e^{ix} f(-x)$  
non posso individuare parte pari e dispari in Tf, quindi uso questo metodo  

1- sistema  
$ e^{ix} f(-x) = \lambda f(x)$  
$e^{-ix} f(x) = \lambda f(-x)$  

2- nella prima, evidenzio f(-x)  
$ f(-x) = \lambda f(x) e^{-ix}$  

3- lo plggo nella seconda:  
$e^{-ix} f(x) = \lambda^2 f(x) e^{-ix}$  
$f(x)e^{-ix}(1-\lambda^2)=0$  
l'unica cosa che può essere 0 è $\lambda = \pm 1$  
sono gli autovalori
