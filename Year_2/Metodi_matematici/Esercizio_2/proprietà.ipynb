cosa | definizione 
---|---

anti-autoaggiunto |  $T^* =-T$

**Proprietà di $T^*T**

1- è autoaggiunto  

2- $ker \, (T^*T)=ker \, T$  
perchè:  
$<T^*Tf, f> = <Tf, Tf> = |Tf|^2$   
che è una norma, quindi è 0 solo se f=0

### Spettro

#### Teoria

$A=T-\lambda I$  

domande:  
1- iniettivo? (= $\lambda$ è un autovalore?)   
2- suriettivo?  
3- immagine è densa?  

tipo | puntuale | continuo | rimasto
---|---|---|---
iniettivo | NO | SÌ | SÌ
suriettivo | | NO
immagine densa | | SÌ | NO
nota | è l'insieme degli autovalori | | se T NORMALE, non ha spettro rimasto  

**Iniettivo**

siggnifica: esiste una $f \neq 0$ per cui $Af=0$?  
cioè, esiste una f per cui $Tf = -\lambda f$ ?  

quindi:  
è iniettivo: $\lambda$ è un autovalore (cioè la condizione non ha soluzione, oppure: è invertibile)  
NON è initettio: $\lambda$ non è un autovalore  

**Immagine densa**

l'immagine è densa se:   
$ker (T^*-\overline \lambda I)= \{0\}$  

perchè:  
$\overline {Im(A)} = (ker A^*)^\perp$  
Im(A) densa $\leftrightarrow$ $ker A^* = \{0\}$  

**Spettro residuo (spiegazione immagine densa)**
 
se $\lambda$ è nello spettro residuo di T, allora $\overline \lambda$ è un autovalore di $T^*$    
non vale l'opposto    

conseguenza:  
se T è normale (autoaggiunto, antiautoaggiunto, o unitario), allora non ha spettro rimasto   

#### Come risolvere

**Osservazione**

trova l'inverso di $(T-\lambda I)f=g$  
1- se un valore di $\lambda$ viola la C.E. sempre, appartiene allo spettro puntuale, è un autovalore   
2- se un valore di $\lambda$ viola la C.E. SOLO per un preciso valore di x, appartiene allo spettro continuo, non è un autovalore
