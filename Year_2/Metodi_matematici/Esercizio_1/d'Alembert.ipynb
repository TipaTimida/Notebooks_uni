### Teoria

**Equazione di d'Alembert / delle onde**

$u_{xx}-\frac {1}{c^2} u_{tt}=0$    

1- c è la velocità, ma di solito c=1  
2- è in 1D, quindi il laplaciano è la derivata di x

**Come risolvere**

ci sono 2 modi, la formula generale e la formula di d'Alembert  
quando usare la formula di d'Alembert lo dice esplicitamente il testo

se posso scegliere:   
usa sempre la formula di d'Alembert, è più facile trovare g ed f  
ECCETTO se ho f=g=0, allora scegli la formula generale    


### Soluzione generale

**Soluzione**

$u(x,t)=p(x+ct)+q(x-ct)$  
anche qui, c=1 di solito   
posso anche invertire, scrivere q(t-x)    

come trovare p,q:  
plugga le condizioni iniziali in u(x,t)   
ottengo un sistema da risolvere per trovare p,q   

**Cosa sono p,q**

p(x+t): viaggia in direzione $\leftarrow$ (impulso che viene dal movimento dell'estremo destro)  
q(x-t): viaggia in direzione $\rightarrow$ (impulso che viene dal movimento dell'estremo sinistro)

**Derivate**

f e g hanno solo un argomento, quindi c'è solo f', non f_x ed f_t  

**Tip: eliminare una costante ???**

cosa devo risolvere:  
p(x)+q(x)=f(x)  
p'(x)=q'(x)  

come faccio:  
integro la seconda, ottengo:  
p(x)-q(x)=c   
posso cancellare il +c usando, invece di p e q, P e Q:    
$P(x)=p(x)-c/2$  
$Q(x)=q(x)+c/2$  
anche queste rispettano che: u(x,t)=p(x,t)+q(x,t)=P(x,t)+Q(x,t)  

ma osservale bene:  
P(x)-Q(x)=(p(c)-c/2)-(q(x)+c/2)=p(x)-q(x)=c-c =0  
quindi, senza perdere di generalità, posso assumere:  
p(x)=q(x)  

ora uso questa cosa per semplificare il resto dell'esercizio, devo trovare solo p invece di p e q  

nota: questo vale per qualunque argomento, anche t, non solo x  
perchè???  
attenzione, p e q sono funzioni di x+t e x-t, quindi dopo aver trovato la soluzione p(x), devo traformarla in p(x+t) 



### Formula di d'Alembert

**Soluzione**

$u(x,t) = \frac12 [f(x-t)+f(x+t)] +\frac12 \int_{x-t}^{x+t} g(s) \, ds$  

cosa sono f,g:   
le condizioni iniziali (posizione e velocità)   
$f(s) = u(x,0)$   
$g(s) = u_t(x,0)$  

##### Definita a tratti

**Che succede**

spesso ho:  
f=0  
g definita solo nel tratto $[0, \pi]$  

quindi:  
1- faccio un'estensione di g su tutta la retta (che tipo dipende da condizioni al bordo di v)  
2- poi, a seconda di che caso sto considerando, devo separare l'integrale in una somma, con il tratto giusto di g in ognuna 

**Quali sono i tratti**

2 domande da farmi:  
$x-t \ge 0 \; ?\quad$ cioè $t \ge x \; ?$  
$x+t \ge \pi  \; ? \quad$ cioè $t \ge \pi-x \; ?$      

così ho 4 casi  

**Separa in casi**

1- separa integrale  
se g è definita a tratti, e c'è la separazione dei tratti a x=0 e $x=\pi$, anche l'integrale va separato in quei tratti  
$u(x,t) = -\frac 12 [\int_{x-t}^0 g(s) \; ds + \int_{0}^\pi g(s) \; ds  + \int_{\pi}^{x+t} g(s) \; ds ]$  

2- quale g(s) mettere in ogni tratto  
guarda dove ho scritto g a tratti, metti quello corretto per quell'intervallo (sotto 0, tra 0 e $\pi$, sopra $\pi$)  

a volte il testo mi specifica quale caso devo considerare  
se non attraversa nessun salto di g, non serve separare in casi  

**Fisicamente**

considero un punto x sulla corda:  
dista x dal margine sinistro, $\pi-x$ dal margine destro  
ho c=1, quindi il movimento si propaga a velocità 1   

domanda 1:   
se $t\ge x$: l'impulso dal movimento dell'estrmeo sinistro ha raggiunto il punto  
se $\le$, ancora non l'ha raggiunto  

domanda 2:  
se $t \ge \pi-x$: l'impulso dal movimento dell'estrmo destro ha raggiunto il punto  
se $\le$, ancora non l'ha raggiunto  

### Esercizi formula di d'Alembert

**Esempio: formula di d'Alembert**

dati:  
equazione delle onde  
dominio: $[0, \infty]$ 
iniziali:    
$g(x) = u(x,0)=0$   
$f(x) = u_t(x,0)=\frac{x^2}{1+x^2}$   
bordo: $u(0,t)=0$  

1- prolungo su tutta la retta  
poichè $u=0$ sul bordo x=0, faccio prolungamento dispari oltre x=0  
g=0 su tutta la retta    
$f(x)$ diventa $-f(-x)$ sul tratto x<0    

2- uso formula di d'Alembert  
f va scritta a tratti, a seconda se l'argomento è positivo o negativo   
l'addendo x+t è sempre positivo   
ma, nel secondo tratto, il secondo addendo è negativo, quindi lo sostituisco con -f(-x)   
$u(x,t) = \begin{cases} 
\frac 12[f(x+t)+f(x-t)] & x-t \ge 0 \\
\frac 12[f(x+t)-f(t-x)] & x-t<0 \\
\end{cases}$  

**Esempio: formula di d'Alembert**
