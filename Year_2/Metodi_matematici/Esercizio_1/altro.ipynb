#### Lavorare con prodotto scalare

**Lavorare con prodotto scalare**

le parti reali e quelle immaginarie sono indipendenti  
quindi, in un prodotto scalare, le posso separare usando la multilinearità  
ricorda di dare il coniugato al primo membro (o al secondo, ma si consistente)  

esempio:  
$<v_n, f> = <c_n + is_n, p+iq>$  
$= <c_n, p> + <c_n, iq> +<s_n, ip> + <is_n, iq>$    
$= <c_n, p> + i <c_n, q> + \overline i <s_n, p> + \overline i i <s_n, q>$   
$= <c_n, p> +i <c_n, q> - i <s_n, p> -i^2 \overline i <s_n, q>$     
$ <c_n, p> + <s_n, q> + i( <c_n, q> - <s_n, p> )$  

$\begin{cases}
<c_n, p> + <s_n, q>=0 \\
<c_n, q> - <s_n, p> \end{cases} =0$   


#### Introdurre v

**Come è fatto**

mi dice: $\Delta u(x,y) = 2xy$   
introduce: $\Delta v(x,y) = u(x,y) -\frac{x^3y+xy^3}{6}$  
vuol dire che sono legate in qualche modo   

**Trova eq diff che rispetta v**

probabilmente è la stessa (o simile) a quella che rispetta u   
prova a pluggare v nella stessa  differenziale, vedi se ridà  

esempio:  
so $v=u -\frac{x^3y+xy^3}{6}$ e $\Delta u = 0$  
allora anche $\Delta v=0$, perchè $\Delta (\frac{x^3y+xy^3}{6}) =0$ (laplaciano è $\Delta u = u_{xx}+u_{yy}$ )  


**Trova condizioni al bordo per v**

$\Delta v(x,y) = u(x,y) -\frac{x^3y+xy^3}{6}$  
cerca le condizioni sugli stessi bordi di u  
quinid nel primo addendo plugga l condizione su u, nel secondo i valori di (x,y) del bordo dove cerco la condizione
