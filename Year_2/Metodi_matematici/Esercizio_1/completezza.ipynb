### Formula

**Teoria**

base: $\{ \phi_n(x) \}_0^\infty$  
dominio: $L^2(a,b)$  

scrivo l'equazione:  
$\int_a^b f(x) \; \phi_n(x) \, dx= 0$  

se l'unica sol è f=0:  
la base è completa su questo dominio    

se invece esiste una sol $f \neq 0$:  
questa f è ortogonale a tutti i vettori della base, e la base non è completa  

**Nomenclatura**

completo = denso in questi esecizi, la domanda è la stessa

### Metodi per risolvere

**1. Simmetria**

se: ogni componente della base ha una simmetria rispetto al punto medio del mio dominio   
cosa: scrivi $f=p+d$, e separa l'integrale in una somma  
cerca di fare annullare separatamente p e d  

**2. Ridurmi a una base completa**

con un cambio var, riconduci la funzione integrale ad un prodotto tra f e una base che so essere completa in quell'intervallo  
quindi f deve essere 0 (perchè so che quella base è completa)

ho:  
$\int_a^b \, g(x) \; x_n \; dx =0$  

se so che:  
$x_n$ è completo in [a,b]  
$g(x) \in L^2(0, \pi)$  

allora g=0 (quasi ovunque, che è ciò che voglio sapere)  

#### Esempi

**Esempio: usa simmetria**

$L^2(-\frac \pi 2, \frac \pi 2)$  
$B = X_n(x) = \sin (n(x+\frac \pi 2))$   

1- studio parità della base (rispetto a 0, il punto medio dell'intervallo)    
$X_{2k}(x) = \sin (2kx)$, antisimm a 0  
$X_{2k+1}(x) = \cos((2k+1)x)$, symm a 0    

2- condizione completezza: ortogonalità a X_2k    
formula | cosa
---|---
$\int_{-\pi/2}^{\pi/2} f(x) \, \sin(2nx) \, dx =0$  | condizione  
$\int_{-\pi/2}^{\pi/2} p(x) \, \sin(2nx) \, dx + \int_{-\pi/2}^{\pi/2} d(x) \, \sin(2nx) \, dx = 0$  | separo f in parte pari e dispari, f=p+d  
$0 + 2 \int_0^{\pi/2} d(x) \, \sin(2nx) \, dx = 0$  | primo integrale ha integranda dispari, secondo pari
so che $ \sin(2nx)$ è completa su $[0, \pi/2]$ quindi l'unica sol è d(x)=0  

3- condizione completezza: 2k+1  
come prima, ma $X_{2k+1}$ invece di $X_{2k}$   
ottego che p(x)=0  

4- conclusione  
f=0, quindi non ci sono f ortogonali a tutto  
quindi B è una base compelta di $L^2(-\frac \pi 2, \frac \pi 2)$  


**Esempio: trova fourier**

base: $\{1, \cos(nx) , \sin(nx)\}_0^\infty$    

condizioni: f ortogonale a tutta la base  
$\int 1 \, f \, dx = 0$    
$\int \cos(nx) \, f \, dx = 0$    
$\int \sin(nx) \, f \, dx = 0$   

queste sono le formule per i coefficienti della serie di Fourier (non importa il $1/\pi$ davanti, tanto a me interessa che siano 0)  
quindi f=0 (perchè ha tutti i coeff 0), come volevo, quindi è uno spazio completo   

OPPURE: posso usare anche qui il metodo della parità  

**Esempio: ricondurmi a qualcosa di completo**

base: $\{e^x \sin(nx)\}$  
dominio: $[0, \pi]$  

esercizio | cosa
---|---
$\int _0^\pi\, f(x) \; e^x\sin(nx) \; dx =0$  | scrivo condizione  
$\int _0^\pi\, h(x) \; \sin(nx) \; dx =0$ <br> $h(x)=e^x  \, f(x)$  | riscrivo per usare qualcosa di completo   
 $h(x) \in L^2$, perchè $f \in L^2$ e $e^x$ è limitato nel dominio che uso <br> $\sin(nx)$ so che è completo in qeusto dominio <br> quindi, $h(x)=0$ | uso completezza
$h(x)=e^x  \, f(x) = 0$ <br> l'esponenziale non può essere 0, quindi deve essere $f(x)=0$ <br> è ciò che volevo, allora la mia base è completa | torno alla f originale

altro modo: potrei dividere per $e^x$, tanto non può essere 0  
così evito di introdurre h(x)  

**Esempio: ricondurmi a qualcosa di completo - dimostrare che f è integrabile**

base: $\{e^{-ny}\}_1^\infty$  
dominio: $L^2(0, \infty)$  

esercizio | cosa  
---|---
$\int_0^\infty e^{-ny} \, g(y) \, dy $ | condizione 
$ = \int_1^0 g(-\ln z) \, z^n \; (-\frac{dz}z) $ <br>$= \int_0^1 g(-\ln z) z^{n-1} \, dz$   | cambio var per trovare una base completa, i polinomi <br> $z= e^{-y}, \quad dz = -e^{-y} dy$  
$g(-\ln x) \in L^2(0,1)$, perchè: <br> $\int_0^1  \|g(-\ln x)\|^2 \, dz = \int_0^\infty \|g(y)\|^2 e^{-y} \, dy < \int_0^\infty \|g(y)\|^2 \, dx< \infty$ <br> perchè $0<e^{-y}\le 1$ in questo intervallo <br> e $g(y) \in L^2(0, \infty)$ per come l'ho introdotta | prima condizione per usare completezza <br> per integrabilità di g, faccio di nuovo cambio var, ma da z a y  
$g(-\ln x) \in L^2(0,1)$, spiegato sopra <br> $\{z^{n-1}\}_1^\infty = \{z^n\}_0^\infty$, quindi è completo in questo dominio <br> quindi $g=0$ | uso completezza 

**Esempio: cambio dominio**

base: $\sin (nx)$  
dominio: $[5 \pi, 6 \pi]$  

esercizio | cosa
---|---
$\int _{5\pi}^{6\pi}\, f(x) \; \sin(nx) \; dx =0$  | scrivo condizione  
$= \int _0^\pi f(y + 5 \pi) \; \sin (ny + 5n\pi) \, dy $ <br> $=  \int _0^\pi f(y + 5 \pi) \; (-1)^n \sin(ny) \, dx$ <br> $=  (-1)^n \int _0^\pi f(y+5  pi) \sin(ny) \, dx$  <br> | cambio var <br> $\sin (nx)$ è completo su $[0, \pi]$    
$f(y+5 \pi)  \in L^2 $, (dovrei mostrarlo come nell'ultimo esrcizio) <br> $\sin(nx)$ so che è completo in questo dominio <br> quindi, $h(x)=0$ | uso completezza
l'unica cosa che può essere 0 all'interno di h(x) è $f(x)$ | torno alla f originale

### Esempi: trova funzioni ortogonali

**Come**

spesso si usa la parità  

**Esempio: trova funzioni ortogonali**

base: $\sin(nx)$  
dominio: $L^2(\frac \pi 2 , \frac{3\pi}2)$      

1- studio simmetria delle funzioni della base rispetto a $\pi$ (punto medio del dominio)  
$\sin nx $ è antisimmetrica rispetto a $ \pi$  
perchè $\sin(2\pi-x)= ... =- \sin(x)$  

2- conclusione  
$\int f \, \sin nx =0$  
se f è simmetrica rispetto a $\pi$, l'integrale dà 0, ma $f \neq 0$  
quindi la base non è completa, e tutte le funzioni pari sono ortogonali  

#### Tips

**Cose complete**

dominio | base completa
---|---
$[-\pi, \pi]$ | $\{1, \cos (nx), \sin(nx)\}_1^\infty$  
$[0, \pi]$ | $\{\sin x\}_1^\infty$ o<br> $\{1, \cos x\}_1^\infty = \{cos(nx)\}_0^\infty$ <br> perchè posso espandere qualeunque funzione con prolungamento pari o dispari, a scelta, quindi rappresentarle con solo una base pari o una dispari (solo cos o solo sin)
$[0,1]$ | $\{z^n\}_0^\infty$ (polinomi)


**Dimostra che la base dei seni è completa**

base: $\{\sin(nx)\}_0^\infty$   
dimostra sia denso su $L(0, \pi)$  

problema: la condizione f ortogonale a tutto mi mette a 0 solo i coefficienti di Fourier del seno, non del coseno  

come risolvere: estendo f con prolungamento dispari  
$F(x) = \begin{cases} f(x) & x>0 \\ -f(-x) & x<0 \end{cases}$  
coeff del seno di F: sono 0, perchè è la condizione di f ortogonale alla base  
formalmente: $\int_{-\pi}^\pi F \, \sin(nx) \, dx = 2 \int_0^\pi f \, \sin (nx) \, dx = 0$   
coeff del coseno di F: sono 0, perchè F è dispari, non può avere componenti pari (pari x dispari = 0 in un dominio simmetrico)   

soluzione:  
quindi F=0 in ogni punto  
poichè f è una restrizione di F in $[0, \pi]$, in quei punti deve avere lo stesso valore, cioè 0  
questo è proprio il dominio dove volevo dimostrare che è completo  

**Cambiare estremi**

espandere estremi: $(0,1) \rightarrow (0,b)$  
come: cambio var da x a z: $bx =z$  

resringere estremi: $(0,b) \rightarrow (a,b)$, con a>0  
è ovvio, se è completo su qualcosa di più grande lo p anche su qualcosa di più piccolo  
ma per farlo preciso, posso definire l'estensione di f a 0
