### Simmetria

**Formula**

ho un intervallo $L^2(a,b)$, con punto medio c  

una funzione f è simmetrica (+) o asimmetrica (-) rispetto al punto c se:  
$f(c+x) = \pm f(c-x)$   
$f(a+x) = \pm f(b-x)$     

tip: se il punto medio dell'intervallo è 0, usa la prima    
se uno dei due estremi dell'intervallo è 0, usa la seconda  

**Come sfruttare la simmetria**

d e p essere simmetriche/antisimmetriche RISPETTO A C, il punto medio dell'intervallo  

1- integrali  
$\int_a^b d(x)  = 0$  
$\int _a^b p(x) = 2 \int_c^b p(x)$  

2- se moltiplico funzioni, la parità si trasmette come quando SOMMO (sommo, non moltiplico) numeri pari e dispari  
d x p = dispari + pari = dispari  
d x d = dispari + dispari = pari  

3- una funzione f pari, scritta come serie, può avere solo componenti pari (e viceversa)  
perchè, quando calcolo i coeff di fourier, $<f, \phi(x)>$ è nullo se f e la base sono una pari e una dispari  

**Nomenclatura**

se studio la simemtria rispetto ad una linea diversa da x=0, invece di pari/dispari dico che le funzioni sono simmetriche/antisimmetriche  

### Controlla simmetria

**Esercizio: separa n pari e dispari**

$X_n(x)=\sin (nx)$  
$L^2(0, \pi)$   

simmetria rispetto a $\pi/2$  
$X(\pi-x) = \sin(n(\pi-x)) = \sin(n\pi -nx) = (-1)^{n+1} \sin nx =  (-1)^{n+1} X(x)$   
$X_{2n}$ (n pari): antisimmetrica   
$X_{2n+1}$ (n dispari): simmetrica  

### Recap

**Simmetria rispetto a 0**

$X_n(x) = \sin (n(x+\frac \pi 2))$  
$X_{2k}(x) = \sin (2kx)$, antisimm a 0   
$X_{2k+1}(x) = \cos((2k+1)x)$, symm a 0      


**Simmetria rispetto a $\pi/2$**

$X_n(x)=\sin (nx)$  
$X_{2k}$ (n pari): antisimmetrica   
$X_{2k+1}$ (n dispari): simmetrica  

### Come sfruttare la simmetria

**Componenti di una funzione**

1- dati:  
$L^2(0, \pi)$   
$f= \sin^2 x$    
$B = \{\sin nx\}_{n \ge 1}$   

2- trova simmetria rispetto $\pi/2$    
f: simmetrica  
$X_{2k}$: antisymm  
$X_{2k+1}$: symm  

3- conclusioni  
f è simmetrica, quindi può avere solo componenti simmetriche  
cioè come componenti ha solo i seni con n dispari, $\sin[(2k+1)x]$  
cioè, quando calcolo i coeff di Fourier, per gli n pari mi verrà che i coeff sono 0
