1- leggi il testo e organizza le info  
2- separa in due ODE disaccoppiate  
3- risolvi le ODE  
4- usa condizioni al bordo per trovare $\lambda(n)$ e A e B

### 1. Leggi il testo

**Cosa mi dice il testo**

1- equazione differenziale su u(x,t)   

2- dominio S su cui u è definita: $x \in [0, \pi]$ 

3- condizioni AL BORDO: $u(0, t), u(x, t)$  
cioè cosa accade ad ogni istante nei bordi del dominio

4- condizioni INIZIALI: $u(x,0), u_t(x,0)$  
cioè posizione e velocità all'inizio   

**1. Equzione differenziale**

spesso dà il laplaciano, che cambia forma a seconda se $u$ è in coord cartesiane o polari    

cart: $\Delta u = u_{xx}+u_{yy}$    
polari: $\Delta u = u_{rr}+\frac{1}{r}u_r+\frac{1}{r^2}u_{\phi \phi}$

**3. Condizioni**

utile riscriverli in forma univar    

esempi:  
$u(0,y)=0 \quad \rightarrow \quad X(0)=0$   
$u_\phi(r, \pi)=0 \quad \rightarrow \quad \Phi'(\pi)=0$  

nota: se ho =0 su un bordo, quello lo uso subito per trovare X(x) e Y(y)  
se invece su un bordo ho una funzione, tipo f(x), allora quello lo userò DOPO aver scritto u(x,y), per calcolare i coeff di fourier, quindi non usarla quando sto risolvendo le funzioni univar  

### 2. Separa in due ODE disaccoppiate

**Steps**

1- faccio un ansatz: u è il prodotto di funzioni univar  

2- pluggo l'ansatz nell'equaz differenziale di u  
per ogni addendo, ricorda di mettere tutte le univar, e poi derivare quelle indicate  

3- divido per l'intera u (espressa come prootto delle funzioni univar)  

4- separo le var dai due lati  

5- uguaglio entrambi i lati a $\lambda$ (detta 'costante di separazione')  
ora ho due ODE  

**Segno di $\lambda$**

può avere segno + o -   
solo una è corretta, ma non so a priori quale  
devo vedere quale delle due mi dà una soluzione non banale (cioè no X=0)    
il che dipende dalle condizioni al contorno  

come scegliere:  
guarda condizioni al contorno di X(x) e Y(y)  
se ce n'è una che si annulla in due punti, probabilmente quella è un oscillatore armonico  
in generale, $X'' = -\lambda X$ è più facile da risolvere di $+\lambda$, prima prova questa  

esempio:  
$\frac{\ddot X}{X} = -\frac{ \ddot Y}{Y} = \pm \lambda$  
codizioni al contorno: $Y(\frac{\pi}2 ) = Y(-\frac{\pi}2 ) =0$  
Y(y) si annulla in due punti, quindi è probabilmente un oscillatore armonico (l'altra sarebbe un esponenziale, o con le trig esponenziali, non si annulla in due punti)  
quindi metti $+ \lambda$, così ho $\ddot Y = -\lambda Y$  


#### Esempi

**Esempio: laplaciano cartesiano**

ansatz: $u=X(x)Y(y)$  
eq diff: $\Delta u = u_{xx}+u_{yy}=0$  
  
$X''(x)Y(y) +X(x)Y'(y)=0$  
$\frac{X''(x)}{X(x)}+\frac{Y''(y)}{Y(y)}=0$  
$\frac{Y''(y)}{Y(y)} = -\frac{X''(x)}{X(x)} = \lambda$  

ho due problemi ODE:  
$\begin{cases}
X''+\lambda X=0 \\
Y''-\lambda Y=0
\end{cases}$  

**Esempio: laplaciano polare**

ansatz:  $u(r, \phi)=R(t) \Phi(\phi)$    
eq diff: $\Delta u = u_{rr}+\frac{1}{r}u_r+\frac{1}{r^2}u_{\phi \phi}$

$R'' \Phi + \frac{1}{r}R' \Phi +\frac{1}{r^2}R\Phi''=0$  
$\frac{r^2R''+rR'}{R} + \frac{\Phi''}{\Phi}=0$  
$\frac{r^2R''+rR'}{R} = - \frac{\Phi''}{\Phi}=\lambda$   

nota: qui ho dovuto moltiplicare anche per r, perchè la voglio avere nella parte con R, non con $\Phi$  

ho due ODE:  
$\begin{cases}
\Phi'' + \lambda \Phi =0 \\
r^2R''+rR'-\lambda R =0 
\end{cases}$  

**Esempio random**

ansatz: $u(x,t)=X(x)T(t)$    
equazione: $u_{tt}=u_{tx}$   

$T''X = T'X'$  
$\frac{T''}{T}=\frac{X'}{X}=\lambda$  

$\begin{cases}
T'' = \lambda T' \\
X' = \lambda X 
\end{cases}$

### 3. Risolvi le ODE  

equazioni differenziali | soluzione | come si fa | note
---|---|---|---
$X''=-\lambda X$ | $X(x)=A\cos(nx)+B\sin(nx) $ <br> $ = C\sin(nx+\phi) $ <br> $ = D \cos(nx+\phi)$ <br> $n = \sqrt{\lambda}$ | oscillatore armonico| quale sol scegliere dipende da dove si annulla: <br> 1- X o X' nulla in due punti, di cui uno è 0: sin + cos, senza fase <br> 2-  X=0 in un punto non-0 (tipo $\pi$ o $\pi/2$): sin con fase <br> 3- X'=0 in un punto non-0: cos con fase
$Y'' = +\lambda Y$ | $Y(y) = Ae^{ny}+Be^{-ny} = $ <br> $ = C \cosh(ny)+D\sinh(ny ) $<br> $= E \sinh(ny + \phi) $ <br> $=  F \cosh(ny + \phi) $ <br> $ n = \sqrt{\lambda}$ | come normale disequazione di secondo grado <br> sostituisci $Y'', Y', Y$ con $x^2, x, 1$, trova le radici e mettile come coefficiente all'esponente della e  | meglio usare le trig iperboliche <br> scegli come per oscillatore armonico, dipende da dove si annulla X o X' (sinh(0)=0, cosh non si annulla mai) <br> eccezione: se ho $Y''=0$, sol è $Y(y)=Ay + B$
$X' = +\lambda X$  | $X(x)=A e^{\lambda x}$  | 1- a occhio, quando derivo moltiplico la funzione per $\lambda$ <br> 2- oppure separazione delle var, X'=dy/dx, X=y
$X' = -\lambda X$  | $X(x) = Ae^{-\lambda x} = Be^{-x/\tau}$ <br>  $\tau = \frac{1}{\lambda}$ | per far apparire $\tau$, cioè il tempo di decadimento, devo mettere il segno - | caso particolare della riga prima 
$T''=\lambda T'$ | $T(t) = A + Be^{\lambda t}$ | 1- sostituisco T'->X, T''->X', ottengo la sol già vista sopra <br> 2- poi integro per riavere T, e rinomino le costanti<br> $T(t)= \int Ce^{\lambda t} \, dt = \frac{C}{\lambda}e^{\lambda t} + D = A + Be^{\lambda t}$  
$r^2R''+\alpha rR'+\beta R=0$ | $R(r)=Ar^{n}+Br^{-n}$ | prova soluzioni a potenza, del tipo $R(r)=r^k$  | equazione di Eulero <br> appare in laplaciano polare  
$aX''+b \dot X + c X=0$ | $\Delta = b^2-4ac \quad s_{1,2} = \frac{-b\pm \sqrt{\Delta}}{2a} = \alpha \pm i \beta$ <br> $\Delta> 0 \, : \quad X(x)= Ae^{s_1x}+Be^{s_2x}$ <br> $\Delta = 0 \, : \quad X(x)= (A+Bx)e^{sx}$ <br> $\Delta < 0 \, : \quad X(x)= Ae^{s_1x}+Be^{s_2x} = e^{\alpha x}(Ae^{i \beta x} +Be^{-i\beta x}) = e^{\alpha x}(A\cos (\beta x) + B\sin (\beta x))$ | normale diseq di secondo grado, sostituisci | se X si annulli in due punti: serve $\Delta <0$, quindi metti $\pm \lambda$ accordingly quando separo le var <br> $\Delta <0$ si impone scrivendo: $\Delta = -n^2$ <br> in $\Delta$ dovrebbe apparire $\lambda$, così questa è anche la relazione $\lambda(n)$, poi i valori di n accettabili li trovo con le condizioni al bordo  

**Nota: secondo grado generica**

searare in parte immaginaria e reale  
$s_{1,2} = \frac {1 \pm \sqrt{1-4 \lambda }}2$    

1- separa parte immaginaria e reale  
$s_{1,2} = \frac 12  \pm \sqrt{\frac{1-4 \lambda }4} = \frac 12 \pm \sqrt{-n^2} =  \frac 12 \pm in$  
quindi $\Delta = \frac{1-4 \lambda }4 = -n^2$ è la relazione tra $\lambda$ e $n$  
poi i valori accettabili di n li troverò con le condizioni al bordo  

2- scrivi la sol  
$X(x) = e^{x/2}(A \cos(nx)+B\sin(nx))$  

**Equazione di Eulero**

quando: laplaciano in coord polari  

esempio:  
eq diff: $r^2R''+rR'-n^2R=0$  
ansatz: $R(r)=r^k$    
calcolo derivate: $R' = kr^{k-1} \quad R'' = k(k-1)r^{k-2}$  
le pluggo nell'equazione: $r^2 \, k(k-1)r^{k-2} + r \,  kr^{k-1} -n^2 r^k= r^k(k(k-1)+k-n^2)=r^k(k^2-n^2)=0$   
quindi $k=\pm n$  
sol generale: $R(r)=Ar^{n}+Br^{-n}$  

nota: u LIMITATA  
è una condizione data nel testo (anche se non lo dice espressamente nel testo, si sottintende)    
vuol dire che u non può mai andare ad infinito (nel dominio dove è definita)  
quindi, se è definita per $r \in [0,1]$, sopra non ci può stare l'esponenziale con $r^{-n}$, perchè dà infinito  
quindi la sol generale sarebbe solo: $R(r)=Ar^n$, togliendo la A ottengo $R(r)=r^n$  
nota che in questo caso, per n=0 ottengo $R(r)=A =const$  

**Trig iperboliche**

sinh e cosh sono dispari e pari, come sin e cos  
cioè $\sinh(-x)=-\sinh (x)$

### 4. Usa condizioni al contorno

**Cosa fare**

voglio trovare:  

1- $\lambda(n)$   
già la so, quando ho introdotto n devo sapere che legame aveva con $\lambda$  

2- che valori di n ammette  
di solito $ n \in N$   

3- A e B, che sono apparse quando ho risolte lo equazioni differenziali  

**1. Trova $\lambda(n)$**

1- se come equaz diff avevo un laplaciano: l'ho già trovata  
o nell'oscillatore armonico o nell'equazione di Eulero  

2- sennò: la devo trovare da una delle condizioni   
di solito, una che mi uguaglia il valore su due rette   

**2. Valori di n**

di solito, $n \in Z$ o $n \in N $  
dipende se pluggare +n e -n dà due soluzioni indipendenti o no   

esempi:  
$e^{nx}$: $e^{x}$, $e^{-x}$ sono indipendenti, usa Z  
$\sin (nx)$:  $\quad \sin(-x)=-\sin (x)$, NON sono indipendenti, usa N  

**3. Trova A e B**

trovale applicando le condizioni al bordo e iniziali  
(scrivile come unviar, per sapere quali applicare)  

nota: dopo scriverò le soluzioni nella serie, dove saranno moltiplicate da $a_n$  
quindi non mi interessa se rimane un coeff davanti alla soluzione, lo posso ignorare, sarà inglobato in $a_n$  
esempio: $-A \cos(nx) \rightarrow \cos(nx)$  

#### Esempi

**Esempio: trova $\lambda$**

ho risolto: $X(x)=Ae^{\lambda x}$  

caso 1: $X(\pi)=X(0)$    
$e^{\lambda \pi} =1$  
$\lambda \pi = i2 \pi n$   
perchè l'esponenziale è un vettore complesso nel piano complesso, dà 1 (cioè solo parte reale) quando l'angolo è un multiplo dell'angolo giro, quindi $2 \pi n$  
con $n \in Z$, perchè i valori negativi di n qui danno soluzioni diverse da quelli positivi, ci devono stare  
e usare valori negativi di n valida comunque l'uguaglianza sopra  

caso 2: $X(\pi)+X(0)=0$  
$e^{\lambda \pi} =-1$  
$\lambda \pi =  i\pi (2n+1)$  
oppure basta dire 2n-1 = k dispari  
anche qui, $n \in Z$    

**Esempio: trova $\lambda$**

equaz diff: $X'' = -\lambda X$   
condiz iniziali: $X'(0)=X'(\pi)=0$   

$X(x)=A \cos(nx)+B \sin (nx)$  
$X'=-An \sin (nx)+Bn \cos (nx)$  

$X'(0)=Bn =0$  
$X'(\pi)=-An \sin(n \pi)$  
NON può essere A=0, sennò ho soluzione banale  
quindi devo avere $\sin (n \pi)=0$, che si ha quando $n \in N$  
mi è rimasta A, metto A=1  

soluzione: $X'(x) =\cos (nx)$ con $n^2 =\lambda$ e $n \in N$  
ho $n \in N$, e non in Z, perche $\sin(-x)= -\sin (x)$, quindi non sono indipendenti, sarebbero soluzioni ridondanti, mentre io voglio una base (quindi solo coi vettori indipendenti)  

**Esempio: trova A e B**

ho: $\Phi(\phi)=A\cos(n\phi)+B\sin(n\phi)\quad n = \sqrt{\lambda}$  
condizioni al contorno: $\Phi'(0)=\Phi'(\pi)=0$  

quindi:  
$\Phi(\phi)=\cos(n \phi)$
