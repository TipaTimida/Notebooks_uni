### Basi complete

**Come trovare la base**

1- ho $X''= -n^2 X$    
2- risolvi  
3- applica condizioni al bordo  
come sol, trovo la base  

ricprda che se c'è un coeff davanti alla trig (tipo un -), lo posso ignorare   
tanto è una base   

#### Basi comuni

intervallo | condizioni bordo | base completa ortogonale
---|---|---
$L^2(0, L)$  | $X(0)=X(L)=0$ | $\sin(\frac {n\pi}L x) \quad n \ge 1$  
$L^2(0, L)$  | $X'(0)=X'(L)=0$ | $\cos(\frac {n\pi}L x) \quad n \ge 0$  
$L^2(0, L)$  | $X'(0)=X(L)=0$ | $\cos(\frac {(2n+1)\pi}{2L} x) \quad n \ge 0$  
$L^2(0, L)$  | $X(0)=X'(L)=0$ | $\sin(\frac {(2n+1)\pi}{2L} x) \quad n \ge 0$

### Teoria

**Problema di Sturm–Liouville**

cosa:  
risolvere eq diff: $-\frac{d}{dx} [p(x)\, X'(x)] \; + \;q(x)\, X(x) = \lambda \, w(x) \, X(x)$  
date le condizioni al bordo  
caso più semplice:: $-X''(x)=\lambda X(x)$   
cioè l'oscilllatore armonico   

**Note**

su [-L, L] mi servono sia sin che cos, per esprimere funzioni pari e dispari  
cioè, usa entrambe quelle possibili per [0, L]      

**Condizioni**

condizioni di Dirichlet: valore di X in un punto  
condizioni di Neumann: valore di X' in un punto  

### Serie di Fourier

**Come trovare coeff**

1- scrivi f come serie  
usando una base ortogonale  
$f= \Sigma a_n \phi(x)$   

2- calcola norma della base  
$||\phi(x)||^2 = \int|\phi(x)|^2 \, dx$  
integrale calcolato nell'intervallo $L^2$ dove mi trovo  

3- formula:  
$a_n = \frac{⟨f,ϕ_n⟩​}{||\phi_n||^2}$   
integrale calcolato in $L^2(a,b)$ su cui sono  

**Coeff comuni**

dominio | base completa | serie | coeff 
---|---|---|---
$L^2(-\pi, \pi)$ | $\{1, \cos(nx), \sin(nx)\}_1^\infty$ |  $f(x) \sim \frac{a_0}2 +\Sigma_1 [a_n \cos(nx)+b_n \sin (nx)]$ | $a_n = \frac{1}{\pi} \int^\pi_{-\pi} f \, \cos(nx) \, dx$ <br> $b_n = \frac{1}{\pi} \int^\pi_{-\pi} f \, \sin(nx) \, dx$ <br> $a_0 = \frac{1}{\pi} \int^\pi_{-\pi} f \, dx$ 
$L^2(0, \pi)$ | $\{\sin (nx)\}_1^{\infty}$ | $f(x) \sim \Sigma_1 b_n \sin(nx)$ | $b_n = \frac{2}{\pi} \int_0^\pi f(x) \, \sin(nx) \, dx $
$L^2(0, \pi)$ | $\{1, \cos (nx)\}_1^{\infty}$ | $f(x) \sim  \frac{a_0}2 + \Sigma_1 a_n \cos(nx)$ | $a_n = \frac{2}{\pi} \int_0^\pi f(x) \, \cos(nx) \, dx $ <br> $a_0 = \frac 2 \pi \int_0^\pi f(x) \, dx$   
$L^2(-\pi, \pi)$ | serie esponenziali |  $f(x)=\Sigma ^\infty_{-\infty} c_n e^{inx}$ | $c_n = \frac{1}{2 \pi} \int_{-\pi}^\pi f \, e^{-inx} \, dx$ 

#### Note

**Tips**

1- altro dominio  
fai un cambio var  
esempio: $[0, 2 \pi]$ ha le stesse formule di $[-\pi, \pi]$, ma ha come estremi di integrazione gli estremi di questo intervallo   

2- f a tratti   
separo l'integrale in somma, ognuno con un tratto

**Nota norma**

su $[0,L]$: tutte le basi complete ortogonali hanno norma $||\phi_n|| = \frac L2$ (tranne 1, che ha L)    
mentre su $[-L,L]$, hanno norma L (tranne 1, che ha 2L)   

per questo separo $a_0$:  
è moltiplicato per $\cos (0)=1$, che ha norma diversa  
ma per fare tutto uniforme, così lo ricordo meglio, scrivo direttamente $\frac {a_0}2$ nella serie    
così ad esempio in $[-L,L]$,, invece di avere $\frac 1 {2 \pi}$ come coeff dell'integrale, ha solo $\frac 1 \pi$ come gli altri, perchè $\frac 12$ l'ho tirato fuori ed esplicitato nella serie  
è simile a quando normalizzo
