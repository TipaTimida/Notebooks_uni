### Steps

**Testo**

ho scritto una u(x,t) generica, con $a_n$    
ora mi dice f(x) su un lato  
-> trova la u specifica che lo rispetta  

**Steps**

1- scrivi la soluzione generale u(x,t)  
(cioè con $a_n$)  
se comodo, separa pari-dispari o n=0

2- calcola $u$ sulla linea dove c'è la f  
(basta pluggare le coord richieste)   

3- f=u(x,t), e calcola i coeff di Fourier $a_n$  

4- controlla se le simmetrie delle componenti hanno senso  

**1. Soluzione di u generale** 

1- generale:  
$u(x,y)= \Sigma_{n=0}^{\infty} a_n X_n(x)Y_n(y)$  

2- separa n=0:  
$u(x,y) = a_0 X_0(x) Y_0(y) + \Sigma_{n=1}^{\infty} a_n X_n(x)Y_n(y)$  
quando: X_0 o Y_0 hanno una forma molto facile, tipo 1, o 0  
mi sarà più facile calcolare il coeff dopo   
DEVO farlo se quella n viola la C.E. della formula generale di $a_n$ che ho trovato 

2- separa pari dispari:  
$u(x,y)= a_0 X_0(x) Y_0(y) + \Sigma_{k=1}^{\infty} a_{2k} X_{2k}(x)Y_{2k}(y) +  \Sigma_{k=1}^{\infty} a_{2k+1} X_{2k+1}(x)Y_{2k+1}(y)$  
quando: se la simmetria di X_n nel mio dominio dipende da se n è pari o dispari  
ricorda di pluggare n->2k, n->2k+1 in entrambe, sia X che Y   
passo anche farlo dopo, quando calcolo i coeff  

nota: se la base è sin, parto da n=1, perche sin(0)=0  
se è cos, serve anche n=0, perchè cos(0)=1

**Esempio: separa pari-dispari**

$L^2(0, \pi)$  
$\phi(x) = \sin(nx)$  

$\phi(\pi-x) = \sin(n(\pi-x)) = \sin(n\pi)\cos(nx)-\cos(n\pi) \sin (nx) = -(-1)^n \sin (nx) = (-1)^{n+1} \phi(x)$  
n pari: $\phi_n$ anti-symm  
n dispari: $\phi_n$ symm  
quindi bisogna separare in pari e dispari  

**2. Calcola u sulla linea: trucchi per semplificare**

se dopo che ho calcolato u sulla linea mi viene altro oltre a $a_n$ e a $\phi(x)$, posso fare:  

1- sostituire  
$u(x)=\Sigma a_n y^2 \cos (nx) = \Sigma b_n \cos (nx)$   
dove $b_n = a_n y^2$  
$b_n$ si trova con formula di Fourier, poi scrivi $a_n$ in funzione di $b_n$   

2- razionalizzare  
originale: $u(x,y)= \Sigma a_n \sinh(n(\frac \pi 2-x)) \sin(ny)$   
razionalizzata: $u(x,y)= \Sigma a_n \frac {\sinh(n(\frac \pi 2-x))}{\sinh(n \pi)} \sin(ny)$    
perchè sulla linea che voglio: $X (\frac \pi 2) = \sinh(n \pi)$   
così, sulla linea che mi interessa, ho solo: $u(\frac \pi 2, y) = \Sigma a_n sin(ny)$    


**3. Coeff di Fourier**

$f = \Sigma a_n \phi_n(x)$  
$a_n = \frac{<f, \phi_n(x)>}{|\phi_n(x)|^2}$  
non scordare la norma della base al denominatore, dipende da in che dominio sono  
se $\phi$ è complesso, ricorda di fare il coniugato  

**4. Osservazioni coeff**

cosa ha f | condizione | conseguenza
---|---|---
simmetrico | $f(-x)=f(x)$ | può avere solo componenti pari
antisimmetrico | $f(-x)=-f(x)$ | può avere solo componenti dispari

cosa | formula
---|---

#### Potenza trig

**Come trovarle**

formule di Eulero:  
$2i \, \sin x = e^{ix}-e^{-ix}$  
$2 \cos x = e^{ix}+e^{-ix}$  

1- eleva alla n le formule di Eulero  
2- sviluppa con binomio di newton  

$\cos^n x = (\frac{e^{ix} + e^{-ix}}2)^n =\frac{1}{2^n} \Sigma_{k=0}^n \begin{pmatrix}n \\ k \end{pmatrix} e^{i(n-k)x}e^{-ikx} =\frac{1}{2^n} \Sigma_{k=0}^n \begin{pmatrix}n \\ k \end{pmatrix} e^{i(n-2k)x}$  
poi raggrupi i termini con esponenti opposti, perchè: $e^{inx}+e^{-inx} = 2 \cos (mx)$  

**Esempio**

$\cos^3 = \frac 18 [1e^{3ix} + 3e^{ix} + 3e^{-ix}+1e^{-3ix}] = \frac 18 [2 \cos(x)+ 6 \cos(3x)] = \frac{3\cos + \cos(3x)}4$  

**Potenze di cos**

$\sin^nx = \frac{1}{(2i)^n}\Sigma_{k=0}^n \begin{pmatrix}n \\ k \end{pmatrix} (-1)^k e^{i(n-k)x}e^{-ikx} = \frac{1}{(2i)^n}\Sigma_{k=0}^n \begin{pmatrix}n \\ k \end{pmatrix} (-1)^k e^{i(n-2k)x}$

**Binomio di Newton**

$(a+b)^n = \Sigma^n_{k=0}  \begin{pmatrix}n \\ k \end{pmatrix}  a^{n-k} b^k$
