### Formula

**Come trovo formula**

1- scrivi f come serie  
usa una base ORTOGONALE $\phi(x)$    
$f= \Sigma a_n \phi(x)$   

2- calcola norma della base  
$||\phi(x)||^2 = \int|\phi(x)|^2 \, dx$  
integrale calcolato nell'intervallo $L^2$ dove mi trovo  

3- formula:  
$||f||^2 = \Sigma |a_n|^2 \, ||\phi(x)||^2$  

**Formule comuni**

dominio | base | formula
---|---|---
$[0, \pi]$ | seni | $\int_0^\pi f^2(x) \, dx =\frac \pi 2 \Sigma_1 b_n^2$  
$[0, \pi]$ | coseni | $\int_0^\pi f^2(x) \, dx =\frac \pi 2 [\frac{a_0^2}2+\Sigma_1 a_n^2]$ <br> dove $f(x) \sim \frac{a_0}2+...$   
$[-\pi, \pi]$ | seni e coseni | $\int_{-\pi}^\pi f^2(x) \, dx =\frac \pi 2 [ \frac{a_0^2}2+\Sigma_1 (a_n^2 + b_n^2)]$ <br> dove $f(x) \sim \frac{a_0}2+...$  

ricorda che cos(nx) ha norma diversa per n=0, quindi separalo  

#### Esempi: trova formula

**Esempio: serie di sin + cos**

1- scrivi f come serie  
$f(x) = \Sigma a_n \sin (2nx) + \Sigma b_n \cos ((2n+1)x)$  
deve essere ortogonale, quindi $\sin (2nx)$ e $\cos ((2n+1)x)$  devono essere ortogonali  

2- calcola norma  
$L^2(-\pi/2, \pi/2)$  
$||\sin (2nx) || = \int _{-\pi/2}^{\pi/2} | \sin (2nx)|^2 \, dx = \frac \pi 2$    
$||\sin ((2n+1)x) || = \frac \pi 2$  

3- scrivi formula  
$||f||^2 = \Sigma |a_n| ^2 \, | \sin (2nx)|^2 + \Sigma |b_n|^2 | \cos ((2n+1)x)|^2 $  
$=    \frac \pi 2 \Sigma |a_n| ^2  + \frac \pi 2 \Sigma |b_n|^2 |  $ 

**Esempio: serie di cos**

1- scrivi f come serie  
$f(x) = \frac {a_0} 2 \cdot 1+ \Sigma_1 a_n \cos(nx)$  

2- trova norme   
$L^2(-\pi/2, \pi/2)$  
$||1|| = \pi$  
$||\cos nx || = \frac \pi 2$  

3- scrivi formula  
$||f||^2 = (\frac {a_0} 2)^2 ||1||^2 + \Sigma_1 |a_n|^2 ||\cos nx||^2 $  
$= \pi \frac{a_0^2}{4} + \frac \pi 2 \Sigma_1 |a_n|^2$  

**Esempio: serie di sin**

1- scrivi f come serie   
$f = \Sigma_1 b_n \, \sin(nx)$   
$ =  \Sigma_1 b_{2k} \, \sin(2nx) \; + \; \Sigma_0 b_{2k+1} \, \sin((2n+1) x)$  
nota che tutti i sin sono ortogonali tra loro perchè hanno arg diversi, quindi è base ortog  

2- trova norme  
$L^2(-\pi/2, \pi/2)$  
$||\sin nx || = \frac \pi 2$  

3- scrivi formula  
$||f||^2 = \Sigma |b_{2k}|^2 \, \frac \pi 2  + \; \Sigma_0 |b_{2k+1}|^2 \, \frac \pi 2$  

### Esercizio

**Che devo fare**

1- scrivo la formula giusta   
2- lato sinistro: svolgo l'integrale  
3- lato destro: scrivo semplicemente i coeff trovati  

probabilmente avrò un'uguaglianza tra una sommatoria e qualcos'altro, nella sommatoria devo lasciare solo le cose con la k

**Calcola serie di Parseval**


dati:   
$L^2(-\pi/2, +\pi/2)$  
base di seni  
$f(x) = \cos x \chi_{[0, \pi/2]}(x)$  
$b_1 = \frac 12 \quad b_{2k} = \frac{4k(-1)^k}{\pi(4k^2-1)} \quad b_{2k+1} = 0 \quad k \ge 1$

1- scrivi formula per base di seni

2- lato sinsitro

$\frac \pi 2 \Sigma |b_n|^2 = \frac \pi 2 |b_1|^2 + \frac \pi 2 \Sigma_1 |b_{2k} |^2 + \frac \pi 2 \Sigma_1 |b_{2k+1} |^2  $

$\Sigma \frac{k^2}{(4k^2-1)^2} = \frac {\pi^2}{64}$  

**Esempio: stima della norma**

voglio dimostrare che: $\int_0^\pi u^2(x,y) dy \le ||f(y)|^2$    

dati:  
$[0, \pi]$, base dei coseni, $Y_n(y)=\cos(ny)$  
$u(x,y) = \Sigma b_n X_n(x) Y_n(y) $    
$f(y)= \Sigma b_n X_n(\pi) Y_n(y) = \Sigma b_n Y_n(y)$  
f ha x fissato, u invece no  

cosa fare:  
1- trova norme secondo parseval (sono entrambi scritti in base dei cos $Y_n(x)$, identifica i coeff):  
$||u(x, \cdot)||^2 = \frac \pi 2 \Sigma b_n^2 X_n^2(x)$  
$|f|^2 = \int_0^\pi |f|^2 \, dy = \frac \pi 2 \Sigma b_n^2$  

2- voglio dimostrare la disuguaglianza $||u||^2 \le ||f||^2$   
quindi devo dimostrare che $X^2(x) \le 1$  

### Teoria

**Identità di Parseval**

energia della funzione = somma delle energie dei coefficienti di Fourier  

quindi se ho una funzione espressa in una base trig, i suoi coeff sono legati alla norma  
basta capire quali sono questi coeff  

**Identità di Parseval**

$||v||^2 = \Sigma_1^\infty ||<e_i, v>||^2$  
cioè la norma di un vettore è uguale alla somma dei quadrati delle sue componenti  
estensione di pitagora con tante dimensioni  
$<e_i, v>$: componente di v lungo $e_i$
