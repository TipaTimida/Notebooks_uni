### Teoria

**Var proprie**

$u=u(a,b) =u(\,a(x,y), \, b(x,y) \,)$  

a,b sono le var proprie di u  
cioè le variabili indipendenti della funzione   
quindi, quando calcolo le derivate, posso scrivere solo $u_a, u_b$, non altre cose tipo $u_x$  

**Funzione univar (1 argomento)**

$\frac {d}{dx} f(\, a(x) \, ) = f'(\, a(x)\, ) \; \frac{d}{dx} a(x)$  

usa derivata totale d/dx  
posso cambiare rispetto a che var definisco, ma NON ci sono f_x o f_y, solo f'  

esempio:  
$u(x,t) = f(x+t) + g(x-t)$  
$\frac {du}{dt} = f'(x+t) \cdot 1 + g'(x-t) \cdot (-1) =  f'(x+t) - g'(x-t)$

**Funzione multivar (2 argomenti)**

$f(x,y) = g(\theta(x,y), r(x,y))$  
$\frac{\partial f}{\partial y} = \frac{\partial g}{\partial r}\frac{\partial r}{\partial y} + \frac{\partial g}{\partial \theta} \frac{\partial  \theta}{\partial y}$  
$f_y =g_r r_y + g_\theta \theta_y$  

usa derivata parziale $\partial$  

### Cambio coords

**Cambio coords**

cosa: voglio descrivere la stessa funzione, ma usando coord diverse  

1- funzione originale: $u(x,y)$  
le sue var proprie sono: x,y  
quindi posso calcolare solo le derivate $u_x, u_y$  

2- voglio passare a coord polari   
quindi faccio il cambio base: $x=r \cos \theta, \quad y= r \sin \theta$  

3- la stessa funzione u, in coord polari, la chiamo col nome v:  
$v(r, \theta)= u(x,y) = u(r \cos \theta, r \sin \theta)$  
cioè u e v sono uguali, descrivono la stessa quantità fisica/geometrica, ma usando coordinate diverse  

### Esercizi: derivate

**Da carth a polari**

$f(x,y) = g(\theta,r) = g(\theta(x,y), r(x,y))$   

1- scrivi il cambio var, in entrambi i versi    
$\begin{cases}
x = r \cos \theta \\
y = r \sin \theta
\end{cases} \quad$
$\begin{cases}
r^2 = x^2+y^2 \\
\theta = \arctan (\frac yx)
\end{cases}$

2- esprime derivara rispetto alle nuove var, con chain rule  
$\frac {\partial f}{\partial y} = \frac {\partial g}{\partial r}\frac {\partial r}{\partial y}    + \frac {\partial g}{\partial \theta} \frac {\partial \theta}{\partial y}$  

3- calcola le derivate tra le var (che appaiono nella formula allo step 2)  
esprimile in funzione delle var finali  
$\frac {\partial r}{\partial y} = \frac {\partial }{\partial r}\sqrt{x^2+y^2} = \frac y{\sqrt{x^2+y^2}} = \frac yr = \sin \theta$  
$\frac {\partial \theta}{\partial y} = \frac {\partial}{\partial y} \arctan (\frac yx) = \frac{1}{1+(y/x)^2} \frac 1x = \frac x{x^2+y^2} = \frac {r \cos \theta}{r^2} = \frac {\cos \theta}r$  

4- metti insieme   
$f_y =g_r\frac {\partial r}{\partial y}    + g_\theta\frac {\partial \theta}{\partial y}$    
$ = \sin \theta \, g_r + \frac{\cos \theta}r  \, g_\theta$  


**Shortcut per trovare derivate al punto 3**

$\tan \theta = \frac yr$  

derivo rispetto a y entrambi i lati (a sinistra chain rule $\frac{d}{dy} = \frac{d}{d \theta} \frac{d \theta}{dy}$)  
$\sec^2 \theta \; \frac{\partial \theta}{\partial y} = \frac 1x$  
$ \frac{\partial \theta}{\partial y}  = \frac{\cos^2 \theta}{r \cos \theta} = \frac{\cos \theta}r$  


$r^2 = x^2+y^2$  

derivo rispetto a y entrambi i lati  
$2r \, \frac{\partial r}{\partial y} = 2y$   
$\frac{\partial r}{\partial y} = \frac{y}r = \sin \theta$  

**Al contrario**

$g(r,\theta) = f(x,y)$  

$\frac {\partial x}{\partial r} = \cos \theta$  
$\frac {\partial y}{\partial r} = \sin \theta$  

$\frac {\partial g}{\partial r} = \frac {\partial f}{\partial x}\frac {\partial x}{\partial r} + \frac {\partial f}{\partial y} \frac {\partial y}{\partial r}$  
$g_r = \cos \theta f_x + \sin \theta f_y$

**Attenzione ai nomi**

x,y sono usati sia come:  
1- COORDINATE (tra parentesi tonde)  
2- ARGOMENTI (come pedice)  

come argomenti, x significa la direzione orizzontale, y la direzione verticale  
ma come coord, possono essere scambiare  

posso cambiare la lettera sritta nelle coordinate, tipo mettere x=a, ma NON posso cambiare gli argomenti, cioè i pedici, quelli restano x e y  

**Intuizione per coordinate e argomenti**

scrivo una funzione u(a,b)  
che prende in input due numeri a,b e mi plotta il punto di coord $a\hat x + b\ \hat y$  

a,b: variabili/coordinate che passo alla funzione  
potrei dargi qualunque nome  
sono semplicemente numeri 

x,y: proprietà dello spazio (o della funzione???), mi dicono cosa rappresentano i due valori presi in input/argomenti  
non cambiano mai, farei confusione (a meno di passare alle coord polari)  
quindi si è deciso che il primo argomento è la posizione orizzontale, il secondo la posizione verticale  

problema: a volte invece dei nomi a,b uso x,y  
ma queste x,y sono coordinate, cioè le var passate, e sono diverse dagli argomenti x,y che indicano sempre le direzioni destra e alto  

esempio: u(y,x)  
passo due var x,y  
la posizione orizzontale (primo argomento) è decisa dalla var y  
la posizione verticale (secondo argomento) è decisa dalla var x  


**Errore comune**

definisco la funzione riflessa rispetto alla bisettrice:  
$v(x,y)=u(y,x)$  

voglio calcolarne la derivata:  
$\frac \partial {\partial x} u(y,x) = u_y$   
d/dx si riferisce alla var x  
$u_y$ si riferisce all'argomento y, cioè il secondo    
cioè, se cambio il valore della var x che gli passo, cambia il valore indicato dell'argomento y, cioè la posizione verticale del punto plottato  

**Calcolare derivate**

devo usare la chain rule, che a parole dice:  
$\Sigma$ derivata rispetto ad un ARGOMENTO x derivata rispetto alla COORD in quell'argomento  

esempio:  
$\frac{\partial}{\partial x} u(y,x) = u_x\frac{\partial y}{\partial x} + u_y \frac{\partial x }{\partial x} = u_y$  
$u_y, u_x$ sono le derivate rispetto agli argomenti (x,y indicano gli argomenti, perchè sono nel pedice)  
poi devo fare la derivata rispetto al valore in coordinate che trovo scritto dentro quell'argomento  

esempio:  
$\frac{\partial}{\partial } u(\pi-y, \pi-x) = u_x \frac{\partial}{\partial x}(\pi-y) +  u_y \frac{\partial}{\partial x}(\pi-x) = -u_y = -u_y(\pi-y, \pi-x)$  

alla fine le coord le copio uguali a come erano all'inizio, quelle non cambiano     

**Simmetria per scambio coord**

1- ho $u(x,y)$  

2- voglio studiare la simmetria per scambio coord  
quindi definisco $v(y,x)$  

**Trova la normale**

voglio la derivata in direzione normale alla bisettrice del primo quadrante  

1- il versore normale alla bisettrice è: $\hat n= (-1,1)$  
volendo, posso razionalizzarlo (non serve): $\hat n = \frac{(-1,1)}{\sqrt 2}$  

2- derivate sulle bisettrice  
so che: $u(x,y)=u(y,x)$  
derivo rispetto a x (alla coord x, non al primo arg) entrambi i lati: $u_x(x,y) = u_y(x,y)$  
sulla bisettrice x=y: $u_x(x,x)=u_y(x,x)$, quindi $u_y-u_x$  

3- definizione di derivata lungo una direzione  
$\frac{\partial u}{\partial n} = \nabla u \cdot \hat n = (u_x,u_y) \cdot (-1, 1) = -u_x + u_y$  
$u_x$ e $u_y$ sono in realtà $u_x(x,x)$ e $u_y(y,y)$, cioè calcolate sulle bisettrice  

4- conclusione  
$\frac{\partial u}{\partial n} = 0$ sulla bisettrice, come volevo
