### Teoria

**Cosa ho**

1- funzione u(x,t)  

2- CONDIZIONI INIZIALI: $u(x,0) \quad u_t(x,0)$  
definite su un dominio $x \in [0, \pi]$  

3- CONDIZIONI AL BORDO (del dominio): $u(0,t) \quad u(\pi, t)$

**Cosa voglio fare**

voglio estendere le CONDIZIONI INIZIALI su un dominio più grande  
per scegliere che tipo di estensione fare, devo guadare le CONDIZIONI AL BORDO  
!!! NON devo calcolare le condizioni iniziali nel punto del bordo, ma guardare le condizioni al bordo ad ogni tempo t  

cosa ottengo:  
le condizioni iniziali sono una funzione definita a tratti  
la chiamo $\tilde f(x)$ oppure $F(x)$ (se f(x) era la condizione iniziale originale)  

**Nomenclatura: condizioni al bordo**

1- Dirichlet: valore di u al bordo  
eg. $u(0,t)=0$
 
2- Neumann: valore di della derivata di u NORMALE al bordo  
eg. $u_x(0,t)=0$  

**Nota: graficamente**

se disegno tutti i tratti di $\tilde g(x)$, NON vedo la stessa simmetria che ho usato per estendere, e possono anche esserci dei salti sulle righe oltre cui ho esteso    
perchè per estendere ho usato la simmetria delle condizioni al bordo di u, NON delle condizioni iniziali  

esempio:  
condizioni iniziali: $g(x)=\cos x$ su $[0, \pi]$  
estensione dispari (oltre x=0):   
$\tilde g(x) = \begin{cases} -\cos x & [-\pi, 0] \\ \cos x & [0, \pi] \end{cases}$   
in entrambi i tratti ho l'ondina che scende (perchè nel tratto negativo ribalto la funzione, ho -cos)  
infatti nel punto x=0 c'è un salto: passa da -1 a +1  
(per corretteza, quindi, dovrei definire g(0)=0)

### Tipo di estensioni

**Recap**

tipo di estensione | simmetrica, pari | antisimmetrica, dispari |
---|---|---
condizione su bordo VECCHIO <br> (retta L) | $u_y(L,y)=0$ <br> derivata normale nulla <br> condiz Neumann | $u(L,y)=0$ <br> u è nulla sulla retta<br> condiz Dirichlet  
condizione su bordo NUOVO | condizioni al bordo nuovo simmetriche a L | condizioni al bordo nuovo anti-symmsimmetriche a L
come scriverla | $F(x) = \begin{cases} f(2c-x) & x<c \\  f(x) & x>c \end{cases}$ | $F(x) = \begin{cases} -f(2c-x) & x<c \\  f(x) & x>c \end{cases}$ 

il 'come scriverla' è se ho definito f nello spazio x>c, voglio definirlo in x<c  

**Esempio**

u(x,y) è deifinito solo sul semicerchio superiore  
voglio espandere su tutto il cerchio  
ho 2 modi per scegliere il tipo di estensione  

1- condizioni al BORDO VECCHIO  
cioè: condizioni sulla linea x=0, quella oltre cui voglio estendere  
se u(0,y)=0 -> dispari  
se u_y(0,y)=0 -> pari  

2- condizioni al BORDO NUOVO  
cioè: $\phi(x,t)$ su tutto il bordo del cerchio    
cosa fa: le condizioni iniziali copiano la simmetria delle condizioni al bordo  
se $\phi$ è simmetrico rispetto a x=0 -> dispari  
se $\phi$ è anti-symm rispetto a x=0 -> pari  

**Trovare soluzione dopo prolungamento**

1- che base posso usare nella sommatoria  
se v è un'estensione pari rispetto alla linea L, allora può avere solo componenti pari rispetto alla linea L  

2- attenzione  
in questo genere di esercizi mi toglie le condizioni iniziali con cui ho trovato le sol generali, ricomincia da capo  
quindi, se alla separazione di variabili avevo $X''=-\lambda X$, la sol generale è somma di seni E coseni, non ho più condizioni da pluggare  
PERÒ, adesso ho la simmetria del prolungamento che mi fa cancellare seni o coseni  
quindi, RISCRIVI BENE TUTTE LE CONDIZIONI CHE HO ATTUALMENTE  

### Esempi

**Esempio BENE**

condizioni iniziali: $u(x,0)=0 \quad u_t(x,0)= \cos x$    
condizioni al bordo: $u(0, t)=0 \quad u(\pi, t)$  

1- cosa voglio:  
sono sul dominio $[0, \pi]$  
voglio estendere le condizioni INIZIALI su tutta la retta 

2- che estensione:  
dipende dalle condizioni al bordo: sono 0 su entrambi bordi, quindi estensione dispari rispetto a entrambi i bordi   
inoltre, $\cos x$ è $2\pi$-periodico, mantengo la periodicità, quindi mi basta estendere su $[-\pi, \pi]$, e poi si ripete  

3- calcola  
ho $g(x) = \cos x$ su $[0, \pi]$  
estensione dispari su $[-\pi, 0]$ (attraverso la retta a: x=0):   
$g(x)=-g(2a-x) = -g(-x) =-\cos (-x)=-\cos x$    
poi c'è l'estension $2\pi$-periodica
soluzione:   
$g(x) = \begin{cases} -\cos x & [-\pi, 0] \\ \cos x & [0, \pi] \\ -\cos x & [\pi, 2\pi] \end{cases}$ 

**Esempio: condizioni su L**

condizioni iniziali: $u(x,0) = f(x,t)= \sin^2 x$  
condizioni al bordo: $u(0, t)=0 \quad u(\pi, t)=0$  

1- cosa voglio:  
estendere sul dominio $[-\pi, 3 \pi]$   
f(x) è $2\pi$-periodica, quindi basta estendere su $[-\pi, \pi]$, poi uso la periodicità  

2- quale estensione  
la condizione al bordo su x=0 è u(0,t)=0, quindi estensione dispari  

3- calcolo: estensione oltre a=0   
$g(x)=-g(2a-x) =-g(-x) = -\sin^2(-x)=-\sin^2 x$  

$\tilde g (x) = \begin{cases} 
-\sin^2 x  & [-\pi, 0] \\
\sin^2 x & [0, \pi] \\
-\sin^2 & [\pi, 2\pi] \\
\sin^2 x  &[2 \pi, 3\pi]
\end{cases}$  

gli ultimi 2 tratti vengono dalla $2\pi$-periodicità  

**Esempio: condizioni su B o L?**

A: semicerchio positivo, dominio di u   
B: cerchio totale, dominio di v  
L: linea y=0, oltre cui voglio prolungare  
voglio estendere u a v     

1- caso: condizioni su L  
$u_\phi=0$, cioè la normale alla linea è nulla      
quindi devo usare prolungamento pari  

2- caso: condizioni su B   
condizioni sul bordo del di B: $g(\phi)=\sin (2 \phi)$  
è una condizione al bordo DISPARI rispetto alla linea $\phi=0$ e $\phi=\pi$, perchè $g(-\phi)=-g(\phi)$  
quindi anche u è dispari rispetto a quella linea  


### Proofs

**Spiegazione: derivata normale alla linea nell'estensione pari**

ho f simmetrica pari a x=0  
$f(x,y)=f(-x,y)$  

calcolo le derivate da entrambi i lati, usando la chain rule  
$f_x(x,y)=-f_x(-x,y)$  

quindi in x=0:  
$f_x=-f_x$, cioè $f_x=0$

### Esempi facili

**Recap: come si scrive l'estensione**

$\begin{cases}
f(x) & x>0 \\
f(-x) & x<0, pari \\
-f(-x) & x<0, dispari \\
\end{cases}$

**Esempio facile**

$f(x)=x$  

1- estensione pari:  
$f(x) = \begin{cases}
x & x>0 \\
-x & x<0 \\
\end{cases}$

2- estensione dispari:  
$f(x) = \begin{cases}
x & x>0 \\
-(-x) & x<0 \\
\end{cases}$

**Esempio facile**

$f(x)=x^2$  

1- estensione pari:  
$f(x) = \begin{cases}
x^2 & x>0 \\
x^2 & x<0 \\
\end{cases}$

2- estensione dispari:  
$f(x) = \begin{cases}
x^2 & x>0 \\
-(x^2) & x<0 \\
\end{cases}$
