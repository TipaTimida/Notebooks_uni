### Spazi

**Spazio vettoriale**

generalizzazione dell'insieme dei vettori  
dotato delle operazioni di:  
1- somma tra vettori   
2- moltiplicazioni per scalare  

esempio:  
Q è uno spazio vettoriale su Q, NON su R  
perchè su R non vale la chiusura per la moltiplicazione per scalare  

**Tipi principali**

1- spazio di Banach  
completo + normato  

2- spazio di Hilbert  
completo + prodotto scalare  

**Norma vs prodotto scalare**

di norma e prodotto scalare ce ne sono tanti tipi possibili  

a cosa serve:  
norma: limiti, convergenza, continuità  
prodotto scalare: norma, ortogonalità, proiezioni

il prodotto scalare implica la norma, perchè $<v,v>= ||v||^2$  
cioè prodotto scalare è norma + altre cose  
quindi spazi di Banach sono tipo particolare di spazi di Hilbert  

**Separabile**

spazio di Hilbert SEPARABILE: ha base ortonormale numerabile (cioè composta da un numero finito di vettori)  
se lo spazio ha dimensione finita, è ovviamente separabile  
se ha dimensione infinita, non è detto  

### Operatore

**Definizioni**

sono simboli o enti funzionali, che agiscono su uno o più elementi (chiamati operandi o funzioni) per trasformarli o combinarli, producendo un nuovo valore o una nuova funzione  
si distinguono cosa prendono in input e output:  

1- funzione: prende in input e output numeri  
eg. f(x)=2x è una funzione  

2- operatore: input e output funzioni  
eg. d/dx è un operatore  

3- funzionale: input funzione, output numero  
eg. funzionale che gli dai una curva e ti restituisce la sua lunghezza

ognuna di quelle può essere lineare  
cioè rispetta la lnearità: T(2f+g)=2f'+g'

T: operatore  
prende in input f(x), dà in output Tf(x)  

esempio: $(Tf)(x)=\sin^2 x \, f(x) + \cos^2 x\, f(-x)$

### Completezza e densità

#### Definizioni

**Completo vs denso**

$x_n$ è completo in $L^2(a,b)$:   
$span(x_n)=L^2(a,b)$  

lo span di $x_n$ è denso in $L^2(a,b)$:   
ogni funzione in $L^2(a,b)$ è approssimato quanto bee voglio in base $x_n$  

quindi denso è un po' meno forte, non c'è proprio ogni funzione, ma solo un'approssimazione???  

**Intuitive**

1- completo:   
proprietà di uno spazio metrico   
non ha buchi  
esempio: Q non è completo, ha un buco in tutti i numeri reali, R è completo  

2- denso:  
proprietà di un sottospazio (A è denso rispetto a B, non in assoluto)  
A può essere usato per approssimare B quanto bene voglio (con precisione infinita)    
esempio: Q è denso in R, ma non completo, perchè ad ogni numero reale posso avvicinarmi quanto voglio (approssimarlo quanto bene voglio), però è un buco, non ci arriverò mai  

**Formali**

1- completo:  
ogni successione di Cauchy converge al suo interno  

2- denso:  
tra due punti qualunque di B c’è sempre almeno un punto di A, cioè $\overline{A}=B$ (che vuol dire???)  

**Attenzione: completo vs denso**

sia denso che completo vivono in uno spazio ambiente H (per me, $L^2(a,b)$)  

1- completo: proprietà della base     
le “direzioni” fornite da A bastano a non lasciare nessuna direzione ortogonale (definizione per esercizi)  

2- denso: proprietà dello spazio  
con ciò che posso costruire a partire da 
A, riesco ad approssimare tutto  

proprietà:  
in uno spazio di Hilbert:  
A completo $\leftrightarrow span(A)$ è denso in H  

esempio:  
la base di Fourier ${sin⁡(nx),cos⁡(nx)}$ è completa in $L^2(−π,π)$  
quindi le serie di Fourier possono approssimare ogni funzione in $L^2$  

**Definizione per esercizi**

è completo in $L^2$ se: l'unico vettore ortogonale a tutto lo spazio (cioè a tutti i vettori della base) è il vettore nullo  

una famiglia $\{ϕ_n\}$ è completa in $L^2$ se:  
$f∈L^2$ e $⟨f,ϕ_n⟩=0 \; ∀n \quad ⇒ \quad f=0$

#### Come scrivere cose

**Nomi per indicare spazi di funzioni**

$C[a,b]$: funzioni continue in questo intervallo [a,b]  
$P[a,b] = \{x^n\}$: polinomi in questo intervallo  
$\tau =\{e^{inx}\}= \{1, \cos(nx), \sin(nx)\}$: serie trigonometrica (polinomi trigonometrici)  
$L^2(a,b)$: funzioni quadrato-integrabili su questo intervallo (cioè: $\int_{a}^b |f(x)|^2 dx < \infty$)  

la cosa tra parentesi graffe è la BASE dello spazio  
$\tau$ è un nome arbitrario, nel compito devo dire cosa indico con quello  

**Come scrivere una base**

come per scrivere uno span, ma non scrivere span davanti  

quindi:  
1- parentesi graffa  
2- all'interno metti tutti i tipi di vettori della base  
3- può esserci una var n che può assumere diversi valori, allora fuori metti gli estremi  

esempio: $\{1, \cos(nx) , \sin(nx)\}_0^\infty$ 

**Nota**

$L^2(0,\infty) \subset L^2(0,1)$  
perchè una funzione f è in un $L^2$ se è integrabile in quell'intervallo  
quindi potrebbero esserci funzioni integrabili in [0,1], ma non in $[0, \infty]$, magari perchè hanno un punto critico all'infinito  

**A cosa serve $L^2(a,b)$**

$L^2(a,b)$: significa che f è quadrato-integrabile in [a,b]  
ogni volta che appare un integrale (anche quello per prodototto scalare), devo mettere [a,b] come estremi di integrazione, così l'integrale è finito   
questo mi dice che la norma deve esistere finita  

$||f||^2=\int_{a}^b |f(x)|^2 $

#### Teoremi

**Identità di Parseval**

$||v||^2 = \Sigma_1^\infty ||<e_i, v>||^2$  

cos'è: estensione di Pitagora con tante dimensioni  
cioè la norma di un vettore è uguale alla somma dei quadrati delle sue componenti  
$<e_i, v>$: componente di v lungo $e_i$, la dimensione i-esima  

**Teorema di Munts-Sastz**

data una successione crescente di numeri reali positivi, ${λ_i}$:  
l'insieme $A= \{x^{\lambda_i}\}$ è completo in $L^2(0,1)$ se e solo se:  
$\sum_{i=1}^{\infty} \frac{1}{\lambda_i} = \infty$  

spiegazione:  
A: insieme spannato da alcuni polinomi, ma non tutti (uso solo alcuni esponenti, non tutti i numeri interi)   
condizione: se sommo gli inversi di tutti gli esponenti usati, ottengo infinito  
cioè ho usato abbastanza monomi per spannare lo spazio

### Fourier serie: simmetria di semionda

**Perchè spariscono gli n pari**

situazione:  
f(x) simmetrica rispetto a $\pi/2$  
cioè $f(\pi -x)=f(x)$  

l'integrale di $b_n$ è:  
$b_n = I_n = \int_0^\pi f(x) \sin(nx) \, dx$  

sostituisco $x \rightarrow \pi -x$  
$I_n = \int_0^\pi f(\pi-x)\sin(n\pi-nx) \, dx = \int f(x) \sin(n\pi-nx) \, dx$  
inoltre: $\sin(n\pi-nx) = \sin(n\pi)\cos(nx)-\cos(n \pi) \sin(nx) =0- (-1)^n\sin(nx)$  
quindi:  
$I_n = \int_0^\pi f(x) [-(-1)^n \sin (nx)] \, dx = -(-1)^n I_n$  

cioè:  
$(1+(-1)^n) I_n=0$  
se n è pari: $2I_n = 0 \rightarrow I_n=0$
