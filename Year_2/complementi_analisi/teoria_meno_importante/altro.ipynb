### Spazio duale

**Spazio duale**

$(R^n)^* = \{L: R^n \rightarrow R\}$  
è lo spazio di tutti i funzionali (applicazioni lineari) applicabili a quello spazio  

**1-forma differenziale**

$\omega: \Omega \rightarrow (R^n)^*$  
$\omega$ è una 1-forma differenziale  

$dx_1, ... dx_n$ sono una base dello spazio duale  

**Campo vettoriale**

$F: \omega \rightarrow R^n$  
F è un campo vettoriale  

**Th**

$f: \Omega \rightarrow R^n$  
f è differenziabile in $p_0$ se e solo se tutte le sue componenti $f_i$ sono differenziabili in $p_0$  
e si ha:  
$Df(p_0)=\begin{bmatrix}\nabla f_1(p_0) \\ ... \\ \nabla f_m (p_0) \end{bmatrix}$ 

### Derivate funzione composta

**Situazione**

$f: R^2 \rightarrow R^3$  
$g: R^3 \rightarrow R$  

vuol dire:  
$f(x,y)=(f_1(x,y), f_2(x,y), f_3(x,y))$  
$g(u_1, u_2, u_3)$  

funzione composta:  
$h(x,y)=g(f(x,y))=g(f_1(x,y), f_2(x,y), f_3(x,y))$  

**Derivata con regola della catena**

$\frac{\partial}{\partial x}h(x,y)=\Sigma \frac{\partial g}{\partial u_i}(f(x,y)) \cdot \frac{\partial f_i}{\partial x} (x,y)$  

**Esempio numerico**

$f(x,y)=(x^2+y, xy, y^2)$  
$g(u_1, u_2, u_3) = (u_1-2u_2-u_3)$  
$h(x,y)=x^2+y +2 xy -y^2$  

derivate parziali di g, rispetto a $u_1, u_2, u_3$:  
$\nabla g = (1,2,-1) = (\partial _{u1}, \partial_{u2}, \partial_{u3})$    

derivate di f rispetto a x:  
$(2x, y, 0)=(\partial_x f_1, \partial_x f_2, \partial_x f_3)$  

formula:  
$\frac{\partial h}{\partial x} = (1,2,-1) \cdot (2x, y, 0) = 2x + 2y$
