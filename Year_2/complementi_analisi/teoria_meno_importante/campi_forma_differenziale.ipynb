### Definizioni

**Forma differenziale**

in 2D:  
dato un campo $F: \Omega \subseteq R^2 \longrightarrow R^2$  
la forma differenziale è:  
$\omega = F_1dx + F_2 dy$  

in 3D:  
dato un campo $F: \Omega \subseteq R^3 \longrightarrow R^3$  
la forma differenziale è:  
$\omega = F_1dx + F_2 dy+F_3 dz$  

essenzialmente è un altro modo di scrivere un campo, al posto di mettere le componenti $F_i$ in un vettore  

**Campo vettoriale**

un campo vettoriale è una funzione F:  
$F: \Omega \subseteq R^n \longrightarrow R^m$  

cioè associa ad ogni punto $x \in \Omega$ un vettore  

è un vettore in cui i componenti sono funzioni:  
$F = \begin{bmatrix} F_1(x_1, ..., x_n) \\ ... \\ F_m(x_1, ..., x_n) \end{bmatrix}$

**Esempio: Passare tra campo e forma differenziale**

forma differenziale: $\omega = x^2y \, dx + y^2x \, dx$  
campo: $F=(x^2y, y^2x)$

**Nota**

se f è una funzione scalare, allora $\nabla f: \Omega \rightarrow R^n$ è un campo ettoriale, detto campo gradiente   

### Proprietà campo

**Veloce**

campo conservativo: se esiste un potenziale V  
se lo è, l'integrale curvilineo di seconda specie è la differenza tra i punti iniziale e finale  

**Def: Campo conservativo**

ipotesi:  
spazio $\Omega$ aperto e semplicemente connesso  
campo $F: \Omega \subseteq R^3 \longrightarrow R^2$  
F è $C^1(\Omega)$  

F è conservativo se esiste:  
$V:  \Omega \subseteq R^3 \longrightarrow R^3$  
di classe $C^2$ tale che:  
$F = \nabla V = (\frac{\partial U}{\partial x},\frac{\partial U}{\partial y}, \frac{\partial U}{\partial x})$  

U si chiama potenziale del campo vettoriale  


**Campo conservativo e integrale curvilineo**

campo conservativo: campo $F: \Omega \subseteq R^3 \longrightarrow R^3$  
curva $r: [a,b] \rightarrow R^3$ regolare, semplice, sostegno contenuto in $\Omega$  
$r(a)=P, r(b)=Q$ sono i punti iniziale e finale della curva  

vale:  
$\int_r F = U(Q)-U(P)$  

come in fisica, la differenza di energia potenziale è il lavoro, che è l'integrale della forza per lo spazio   
F, il campo, cos'è? forza per spazio?  

formalmente: l'integrale curvilineo di seconda specie tra due punti dipende solo dagli estremi

**Proprietà dei campi conservativi**

campo $F: \Omega \subseteq R^3 \longrightarrow R^3$ di tipo $C^1(\Omega)$  

le 3 proposizioni sono equivalenti:  
1- F è conservativo in $\Omega$  
2- l'integrale curvilineo di seconda specie tra due punti dipende solo dagli estremi
cioè prese due curve $\gamma_1$ e $\gamma_2$ regolari a tratti, semplici, con sostegno in $\Omega$ ed ESTREMI COINCIDENTI, $\int_ F{\gamma_1}=\int_{\gamma_2}F$  
intuitivamente, il lvaoro non dipende dalla traiettoria, path (che rebbe la curva $\gamma$), ma solo dal punto di inizio e fine  
3- per ogni curva CHIUSA $\gamma$ regolare a tratti, semplice, con sostegno in $\Omega$, vale $\int \gamma F =0$  
intuitivamente, il potenziale è una funzione di stato
