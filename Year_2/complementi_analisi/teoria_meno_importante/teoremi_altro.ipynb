### Permanenza del segno

ipotesi:  
$f: A \rightarrow R$  
f continua in z  
f(z)>0  

tesi:  
$\exists \delta >0 : \quad f(x)>0 \; \forall x \in A \cap B(z, \delta)$  
cioè: posso disegnare un intorno che contiene solo numeri positivi  

**Proof**

1- suppongo sia falsa    
quindi esiste una palla di raggio $B(z, \delta)$ che contiene un numero non-positivo x  

2- costruisco una successione  
per ogni k, prendo $\delta = \frac{1}{k}$  
poichè x è in una palla di raggio 1/k centrata in z:  
$|z-x_k| < \frac{1}{k}$  

3- conseguenza  
la successione tende a z, perchè se $k \sim \infty$; allora $x_k \rightarrow z$  

4- continuità  
poichè f è continua in z, $f(x_k) \rightarrow (z)$  
per ipotesi f(z)>0, quindi anche $f(x_k)$ deve esserlo  
non può essere non-positivo, come sarebbe se la tesi fosse falsa  
quindi contraddizione, la tesi è falsa  

### Teorema C2

ipotesi:  
$f,g : A \rightarrow R^m$  
$h: B \supset R^m \rightarrow R^k$  
f,g continue in z  
h continua in f(z)  

tesi:  
1- $\lambda g + \mu g $ è continua in z, qualunque siano i coeff  
2- se m=1, $f \cdot g$ è continua in z  
3- se se m=1 e $g(z) \neq 0$, esiste un palla centrata in z per cui f/g è ben definita e continua in z   
4- $h(f)$ è continua in Z  

**Tip**

si risolvono tutte sfruttando la proprietà che f è continua in z  

definizione: f continua in z  
se $x_k \rightarrow z$, allora $f(x_k) \rightarrow f(z)$  

in tutte le proof userò questa successione  


**Proof 1**

uso la successione:  
$\lambda f(x_k)+\mu f(x_k) \rightarrow \lambda f(z)+\mu g(z)$  

**Proof 2**

uso la successione:  
$f(x_k)g(x_k) \rightarrow f(z)g(z)$  

**Proof 3**

assumo g(z)<0 (con >0 sarebbe uguale)  
permanenza del segno: esiste una palla $B(z, \delta)$ dove g(x)<0  
in questa palla g non è mai 0, quindi f/g è ben definita  
e poichè f è continua, prendendo la stessa successione di prima:  
$\frac{f(x_k)}{g(x_k)} \rightarrow \frac{f(z)}{g(z)}$  

**Proof 4**

prendo la stessa successione di prima  
per continuità di f in z: $f(x_k) \rightarrow f(z)$   
per continuità di h in f(z): $h(f(x_k)) \rightarrow h(f(z))$
