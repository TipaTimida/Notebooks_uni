### Moltiplicatori di Lagrange

**Estremi vincolati**

$f: A \subseteq R^2 \rightarrow R$  
$\Gamma$, il sostegno di una curva, è detto vincolo  
$f|_\Gamma$ è la funzione vincolata  

significa la funzione $f(x,y)$ prendendo in input i punti di quella curva  

estremi vincolati: massimi e minimi di $f|_\Gamma$, cioè possono essere scelti solo tra le immagini dei punti della curva  

**Trovare gli estremi vincolati**

2 metodi:  

1. parametrizzare la curva   
voglio $x(t), y(t)$, così possono plug in $f(x,y)$    
ora ho solo una funzione in t, univariabile, i minimi sono dove df/dt=0  

2. moltiplicatori di Lagrange  

**Esempio: parametrizzare il vincolo**

$f(x,y)=xy$  
$\Gamma:$ circonferenza di centro (0,1) e raggio 1  

parametrizzo il vincolo:  
$\Gamma = \begin{cases} 
x(t)= \cos(t) \\
y(t) = 1+ \sin (t) \end{cases}$  
$t \in [0, 2\pi]$  

sostituisco:  
$f|_\Gamma = \cos(t) (1+ \sin(t)) =g(t)$  

trovo max e min:  
$g'(t)=0$  

**Th dei moltiplicatori di Lagrange**

$f,g \in C^1(R^2)$  
$P^*$ estremo vincolaro di $f|_\Gamma$  
$\Gamma = \{(x,y) \in R^2 : g(x,y)=0\}$  

se $\nabla g(P^*) \neq 0$:  
$\exists \lambda: \quad \nabla f(P^*)=\lambda \nabla g(P^*)$ 

se $\nabla g(P^*) = 0$: non so  

**Recap moltiplicatori Lagrange**

per trovare max e min vincolati, basta risolvere:  

$\begin{cases}
g(x,y)=0 \\
\nabla f = \lambda \nabla g \end{cases}
$

### Funzione Lagrangiana

riassume il sistema dei moltiplicatori di Lagrange  

Lagrangiana:  
$L(x,y, \lambda) = f(x,y) - \lambda g(x,y)$  

il sistema diventa:  
$\nabla L = 0$  

**Trovare estremi vincolati**

steps:  

1. trovare punti singolari, dove $\nabla g(P^*)=0$, e studiarli a parte  

2. studiare i punti critici della Lagrangiana, dove $\nabla L =0$  

**Esempio**

1--- Testo

f: maximize  
$\Gamma$: constraint  

fai parallelepipedo con maggior volume di area 12  
$f(x,y,z)= xyz$  
$\Gamma: g(x,y,z) = 2xz+2yz+2xy-12 =0$  

2--- punti singolari

$\nabla g(x,y,z)=0$  
tutte le derivate parziali a 0, sistema  
poi guarda se geometricamente hanno senso  

3--- lagrangiana  

$L(x,y,z)= f-\lambda g$  

$\nabla L = 0$  
risolvi il sistema  
devo trovare $\lambda$, e poi x,y,z saranno espresse in funzione di $\lambda$
