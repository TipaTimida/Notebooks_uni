### Quadratica

**Def: quadratica**

insieme Q con le proprietà:  

$Q = \{(x,y,z)\in R^3 : f(x,y,z)=0\}$  

$f(x,y,z) = ax^2+by^2+cz^2+dxy+exz+ fyz+gx+hy+iz+m$  

cioè f è un polinomio di secondo grado  

tramite rotazioni e traslazioni una qualsiasi quadrica può essere scritta in uno dei seguenti due modi:  

$aX^2+bY^2+cZ^2 = d$   
$aX^2+bY^2 = cZ$  

$x, X$ sono legate ma diverse  
eg, potrebbe essere $X = x-1$  

**Studiare una quadratica**

prendo un piano perpendicolare ad un asse, tipo $Z=k$, quindi parallelo al piano xy   

sostituisco $Z=k$ nella formula  
k è una costante, lo sposto nel membro a destra, dove sono le costanti, ora ho equazione in 2 variabili, non 3

### Tipi

**Ellissoide**

$\frac{X^2}{a^2}+\frac{Y^2}{b^2}+\frac{Z^2}{c^2}=1$  

$a,b,c >0$  

**Iperboloide ad una falda**


$\frac{X^2}{a^2}+\frac{Y^2}{b^2}-\frac{Z^2}{c^2}=1$  

$a,b,c >0$  

**Iperboloide a due falde**


$\frac{X^2}{a^2}-\frac{Y^2}{b^2}-\frac{Z^2}{c^2}=1$  

$a,b,c >0$  

### Forma quadratica

**Def: forma quadratica**

H matrice simmetrica  
forma quadratica associata ad H:  

$Q: R^n \rightarrow R$  
$Q(v) = <Hv,v> \quad \forall v \in R^n$  

forma quadratica associata ad  una matrice è il prodotto scalare che descrive, usato per calcolare la norma    

**Definita positiva**

definita positiva: $Q(v)>0 \quad \forall v \in R^n, v \neq 0$  

come algebra lineare, definita negativa, semidefinita positiva, indefinita
