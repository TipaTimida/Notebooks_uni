con K insieme compatto e misurabile  

$\sigma K$ frontiera dell'insieme K  
formata da un'unione di curvd $\gamma _j$ chiuse, semplicei, regolari a tratti e orientate positivamente  

campo vettoriale: $F: \Omega \subseteq R^3 \rightarrow R^3$  
che sia $C^1(\Omega)$ con $\Omega \subseteq \sigma(K)$  

allora vale:  
$\Sigma \int _{\sigma(\gamma_j)} F = \int _{\sigma(K)} <rot(F), N>$

ho una superficie $\sigma$

$\gamma_j$ sono i controtni della proiezione del foglio, quindi sono linee disegnate sul tavolo, appartengono a K  
$\sigma(\gamma_j)$ sono i contorni del foglio, appartengono al foglio stesso, a $\sigma$
