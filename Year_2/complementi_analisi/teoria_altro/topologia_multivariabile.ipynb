**Massimo vs minimo**

per un punto dove il gradiente è nullo:  
 
max: $<H(x-x_0), (x-x_0)> \le 0$  
min: $...\ge 0$  

cioè, se H in $x_0$ è:  
- definita positiva: minimo  
- definita negativa: massimo  
- indefinita: sella  

se è semidefinita non si sa, potrebbe esserci qualunque cosa  

come univariabile: se derivata seconda è positiva, è un minimo  


**Recap: trova max e min su TUTTO R**  

1. calcola le derivate parziali rispetto a tutte le variabili  

2. poni le derivate parziali = 0  

3. trova x,y che risolvono il sistema (punti critici) 

4. scrivi la matrice Hessiana (calcolando tutte le derivate parziali seconde)

5. Calcola la Hessiana in punto critico, controlla come è definita, così so che tipo di punto è (max, min, sella)  
per vedere come è definita trova gli autovalori, come in algebra lineare

**Forma quadratica**

è un prodotto scalare in algebra lineare  
cioè un polinomio che può essere espresso come prodotto scalare tra le variabili (x,y,z) e una matrice simmetrica A:  
$q(x)$ (il polinomio, come quando espando il prodotto scalare) tale che:  
$\exists A=A^t \; : q(x)=<x, Ax>$  

**Approssima con Taylor**

primo ordine:  
$f(x)=f(p)+<\nabla f(p), x-p> = f(p)+df(p)(x-p)$  

secondo ordine:  
$f(x)=f(p)+<\nabla f(p), (x-p)> + \frac{1}{2}<Hf(p)(x-p), x-p> = f(p)+f_x(p)x+f_y(p)y +\frac{1}{2}(f_{xx}(p)x^2+2f_{xy}(p)xy+f_{yy}(p)y^2)$  

nota che x=(x,y), è un vettore

voglio approssimare f attrno a p

2- calcola $\nabla f(p)$  
3- calcola $Hf(p)$  
4- scrivi l'approssimazione
