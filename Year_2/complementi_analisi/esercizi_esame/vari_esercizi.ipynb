### Misurabile

**Def**

La misura di Lebesgue m generalizza lunghezza/area/volume  
Un insieme è misurabile se è “compatibile” con la costruzione di Lebesgue (spesso: aperti/chiusi/insiemi con frontiera “piccola” sono misurabili)  

**Quali insiemi sono misurabili**

- tutti gli insiemi aperti o chiusi di $R^n$ sono misurabili  
- intersezione (finita/numerabile) di misurabili è misurabile  

**Esempio**

$E = \{(x, y, z) ∈ R^3 : \quad x^2 + y^2 + z^2 < 1, \quad x ≥ 0,  \quad y ≥ 0\}$  

è intersezione di:  
- B(0,1)  
una palla aperta centrata in (0,0,0) di raggio 1  
è un aperto di $R^3$, quindi misurabile  

- $[0,+∞)×[0,+∞) \times R$  
che viene dalla altre condizioni, che limitano i valori di x e y  
è un sottoinsieme chiuso di $R^3$, quindi è misurabile  
nota che è chiuso perchè il prodotto cartesiano di insiemi chiusi è chiuso   
e $[0,+∞)$ è chiuso perchè il suo complementare $(-\infty, 0)$ è aperto  
intuitivamente, da $[0, \infty)$ una successione non può uscire (tendere a un punto non nell'insieme)  

E è intersezione di misurabili, quindi è misurabile  

### Integrabile

**È integrabile?**

se la funzione integranda è non-negativa (non può mai essere negativa, in tutto l'intervallo di integrazione scelto):  
l’integrabilità di g corrisponde alla finitezza di $\int _E g$   

perchè:  
g non negativa: non può oscillare  
g finita: non può divergere  
quindi deve convergere  

formalmente: in quanto tale integrale è ben definito per la non negatività dell’integranda  
quindi se la prossima domanda è di calcolare l'integrale posso semplicemente calcolarlo, ed è una dimostrazione che è integrabile  

oppure ricorda che: continua sul dominio -> integrabile  

calcolare l'integrale e non dà infinito non è sempre una dimostrazione che sia integrale??? come faccio a dimostrare che sia finita invece???

**Se g non è definita in un punto**

non importa se g non è definita in un solo punto (o anche su un insieme naturale di punti)  
perchè un singolo punto ha area nulla, quindi non impatta il fatto che ci sia o no  

formalmente: 
poiché un singolo punto ha misura nulla  
ogni funzione h ≥ 0 definita su $P = {(x_0, y_0, z_0)}$ è misurabile su Z  
inoltre $\int _P h = 0$ essendo $m_3(P)=0$  

Possiamo quindi scrivere $\int_E g = \int _{E-P} g + \int _P g = \int_{E- P} g$, ignorando il valore di g su P    
ovvero un qualunque valore assegnato a g su P non cambia il risultato  

ricorda che la condizione per l'integrabilità è:  

$f:I \rightarrow R$ is integrable if and only if:  

1. $f$ bounded  (sup, inf are not infinite)  . 

2. $disc(f)$ is negligible (has measure zero)   
$disc(f)$: set of all points of discontinuity  

quindi va bene che sia discontinua in un punto, che ha misura 0   

### Superficie regolare e di livello

**Condizioni**

sono nomi che si danno ad un insieme S  

- superficie di livello: se posso scrivere $S=g^{-1}(c)$   
dove g sarebbe il solito vincolo  
in pratica chiede che sia g=0 e non $g\le 0$  

- superficie regolare: se è una superficie di livello E ha $\nabla g \neq 0$ in tutti i suoi punti (quindi nei punti che rispettano g)  
il teorma che lo dice è il 'Teorema delle funzioni implicite'   

quindi:  
superficie regolare -> superficie di livello  

**Esempio**

$S = \{(x,y,z) \in R^2 :\quad x^2+y^2+z^2-4=0\}$  
dimostra che è una superficie regoalre  

condizione 1:    
$S=g^{-1}(0)$  

condizione 2:  
$\nabla g(x,y,z)=(2x,2y,2z)$  
si annulla solo in (0,0,0)  
ma questo punto non appartiene alla sfera, perchè non soddisfa g  
quindi $\nabla g\neq 0$ su tutto S  

per il teorema delle funzioni implicite, S è una superficie regolare  

**Attenzione**

$S=\{(x,y,z) : x^2=0\}$  
è una superficie regolare?  

se facessi il metodo verrebbe che $\nabla g=(2x,0,0)$, che è nullo nell'origine, quindi non lo è  
ma nota che S è semplicemente il piano yz, che è ovviamente regolare, perchè in ogni punto si comporta come un piano  
quindi il problema è che ho scelto la funzione definente (cioè g) sbagliata  
qualla correta sarebbe $x=0$  

funzione definente: è il vincolo g che definisce una superficie REGOLARE  
cioè il suo gradiente è nullo in tutti i punti della superficie  

**Esempio**

$C_0 = \{(x,y,z): y^2+(z-1)^2=x^2 \quad 0<x<1\}$  
$C$ stesso g, ma $0 \le x \le 1$  
sono superfici regolari?  

$\nabla g = (-2x, 2y, 2(z-1))$  
si annulla per $(0,0,1)$, che si trova in C ma non in $C_0$  
quindi $C_0$ è una superficie regolare, C no  

graficamente, C e $C_0$ sono lo stesso cono (infatti g è simile a formula per parametrizzazione cono), ma $C_0$ è senza il vertice, che è l'unico punto di non regolarità di un cono  

### Th di Green

**Cosa è**

è una versione bidimensionale del teorema del rotore   
che considera superfici appartenenti a $R^2$  

**Teorema di Green**

condizioni:  
$\Omega \subset R^2$ aperto regolare  
$P(x,y), Q(x,y)$ funzioni di classe $C^1$ su $\Omega$  

th:  
$\int _{\partial^+Ω} P \, dx + Q \, dy = \int \int _\Omega (\frac{\partial Q}{\partial x}- \frac{\partial P}{\partial y}) \; dx \; dy$  

note:  
$\partial^+Ω$ significa che il bordo di $\Omega$ è orientato in senso antiorario (come negli angoli, è quello positivo per default)  
se il bordo fosse in senso orario, si cambia segno:  
$\int _{\partial^-Ω} P \, dx + Q \, dy = - \int \int _\Omega (\frac{\partial Q}{\partial x}- \frac{\partial P}{\partial y}) \; dx \; dy$  


**Come ricordarlo**

metodo 1:   
$ - det \begin{bmatrix} \partial_x & \partial_y \\ P & Q    \end{bmatrix}$  
ricorda il segno -  

metodo 2:  
$F=(P,Q,0)$  
$\nabla \times F = (0,0,\frac{\partial Q}{\partial x}- \frac{\partial P}{\partial y})$ (rot F)  
pouchè voglio la parte perpendicolare alla superficie:  
$(\nabla \times F) \cdot \hat{n}$


**Esercizi con teorema di Green**

risolvi:  
$\int _{\partial^+Ω} x \; dy$  

trova la forma:  
$\int _{\partial^+Ω} P \, dx + Q \, dy$  
$P(x,y)=0$  
$Q(x,y)=x$  

ora usa la trasformazione  
nota che, se mi viene dato il dominio, devo anche anche trovare i suoi estremi, come al solito, per integrare  

**Se il dominio non è regolare**

ad esempio, se il dominio ha un buco  
tipo è un anello tra i raggi $r_1$ e $r_2$  
i bordi dell'anello sono $C_1$ e $C_2$, e chiamo $C=C_1 \cup C_2$  

non posso usare Green sui cerchi perchè non sono regolari  
ma posso separare l'anello in due parti, $D_1$ e $D_2$, la metà sopra e la metà sotto, e fare Green su ognuno di loro  
poi separo l'integrale di linea in addendi, ognuno su un pezzo di curva $C_i$  
nota che le parti che ho aggiunto per separare, $C_5$ e $C_6$, si cancelleranno  
quindi ottengo $\int _D (Q_x-P_y) \; dA = \int_C P \, dx + Q \, dy$

### Piano tangente

**Iperpiano: def**

piano con una dimensione in meno della dimensione dello spazio  
posso individuarlo con:  
- vettore perpendicolare ad esso: $\vec{n}$  
- un punto appartenente al pian: $p$   

comprende tutti i punti $\vec{x}=(x,y,z)$ del tipo:  
$(\vec{x} -\vec{p} ) \cdot \vec{n}=0$  

**Esercizi: cosa è $\vec{n}$**

per descrivere piano tangente a:  

1. a superficie di livello con vincolo g: $S=\{g(x,y,z)=0\}$  
$\hat{n}=∇g(p)$  
cioè il gradiente nel punto di tangenza  

2. a grafico di f  
costruisci superficie di livello con $g=f(x,y)-z=0$  
fai metodo di prima, con $p_z=f(p_x,p_y)$   

3. a superficie parametrizzata:  $\phi(u,v)=(x(u,v), y(u,v), z(u,v))$   
costruisci superficie di livello, cercando una g che funzioni  

**Esempio: superficie di livello**

$g= x^2+y^2+z^2-4=0$    
$p=(1,1, \sqrt{2})$  

$\nabla g(p) \cdot (x-p_x, y-p_y, z-p_z)=0$  
$(2,2,2\sqrt{2}) \cdot (x-1, y-1, z-\sqrt{2})=0$  
$2(x-1)+2(y-1)+2\sqrt{2}(z-\sqrt{2})=0$  

**Esempio: grafico di f**

$z=f= \sqrt{4-x^2-y^2}$  
$p=(1,1)$  

trasformo in superficie di livello, mi verrebbe come l'esercizio sopra  
risolvo uguale  

**Esempio: superficie parametrizzata**

$\phi(u,v)=(u,v,u^2+v^2)$  
$p=(u_0, v_0, u_0^2+v_0^2)$    

trasformo in superficie di livello:  
$g=z-x^2-y^2$  

faccio il gradiente su g in coord cartesiane, poi ci pluggo il punto p:  
$(-2u_0, -2v_0, 1) \cdot  (x-u_0, y-v_0, z-u^2-v_0^2)$  


**Altro metodo: superficie parametrizzata**

invece di cercare superficie di livello, per trovare n posso fare:  
calcola, con il metodo del det: $\vec{n}=\phi_u(p) \times \phi_v(p)$  

$\phi(u,v)=(u,v,u^2+v^2)$  
$p=(u_0, v_0, u_0^2+v_0^2)$    

trovo n:  
$\phi_u = (1,0,2u_0)$  
$\phi_v=(0,1,2v_0)$  
$|\phi_u \times \phi_v | = (-2u_0, -2v_0, 1)$ (facendo il det)  

ora risolvo:  
$(-2u_0, -2v_0, 1) \cdot (x-u_0, y-v_0, z-u^2-v_0^2)$  



**Piano tangente**

è il GRAFICO DEL DIFFERENZIALE  

perchè:  
il differenziale è la migliore approssimazione di f in un punto: $f(p+h) \sim f(p)+df(p)(h)$   
quindi l'insieme:  
$A=\{(p+h, f(p)+df(p)(h)) : \; h \in R^n\}$  
è l'iperpiano tangente al grafico in $(p,f(p))$
