### Trova punti critici

**Dati**

un insieme, con un vincolo g e altre condizioni:    
$S=\{(x,y,z): g(x,y,z)=0, \; altre \; condizioni\}$  

le altre condizioni sono disuguaglianze, di solito del tipo $x \ge 0$    
servono a sapere alcune proprietà dell'insieme  

una funzione (per dare un valore ad ogni punto dello spazio):  
$f(x,y,x)$    

**Trova tutti i punti critici**

1. controlla che il vincolo g sia g=0  

2. scrivi ordinatamente f, g, e le altre condizioni di S  
se necessario, riscrivi bene S   

3. scrivi sistema con i moltiplicatori di lagrange:      
$\begin{cases} 
\nabla f = \lambda \nabla g \\
g(x,y,z)=0  \\
altre \; condizioni \; di \; S  \end{cases}$ 

4. trova punti (x,y,z) che risolvono il sistema  


**1. Controlla il vincolo g**

preliminari:  
il vincolo g potrebbe non essere già tutto a sinistra, in caso spostalo  
esempio: $\quad x+y=z-1 \quad \rightarrow \quad g=x+y-z+1=0$  

generalmente il vincolo g è di 2 tipi:   

1. $g=0$  
è una superficie  
ok, fai sistema di Lagrange  

2. $g \le 0$  
è un volume o un'area    
difficile, guarda guida sotto  
esempio: $B(0,1)$ fa parte di questo tipo  
oppure $x^2+y^2-1\le z \le 1$, che combina pure una delle altre condizioni  

#### Quando $g \le 0$  

**Nota**

$E = \{g \le 0 \quad altre \, condizioni\}$  

se g(x,y,z) ha 3 variabili:  
$g = 0$: definisce una superficie (uguaglianza)   
$g \le 0$: definisce un volume racchiuso nella superficie (diseguaglianza)  

**Cosa fare**  

1. dividi l'insieme in frontiera e parte interna, cioè in $g=0$ e $g< 0$  
$S= Int(S)+Fr(S)$  

2. trova i punti critici su ognuno dei due, separatamente  

    - Fr(S) : ha $g=0$  
    usa sistema di Lagrange, come al solito  

    - Int(S) : ha $g<0$  
    trova tutti i punti con $\nabla f=0$, poi controlla quali rispettano $g<0$   

esempio:   
$S = \{(x, y, z) ∈ R^3 : g \le 0\}$  
$S=Fr(S)+Int(S)$   
$Fr (S) =  \{(x, y, z) ∈ R^3 :g= 0\}$   
$Int(S)= \{(x, y, z) ∈ R^3 : g < 0\}$   

**Trucco: SOLO quando $g \le 0$**

1. può avere punti critici interni? cioè in $Int(S)$  
solo se $\nabla f$ può annullarsi (f, NON g, far annullare g serve a vedere se è una superficie regolare)    
cioè per qualche valore di x,y,z posso ottenere (0,0,0)  
quindi se ad esempio $∇f=(2x,1)$ la seconda componente non potrà mai annullarsi, non ha punti critici interni  

2. può avere punti critici di bordo? cioè in $Fr(S)$   
solo se S è chiuso  
cioè se tra le 'altre condizioni' c'è $\le$, not $<$  

se non ha punti critici nè interni nè di bordo, non ha punti critici    

**Teoria: punti critici interni e di bordo**

ha senso fare la distinzione solo se $g\le 0$  
sennò non ha senso distinguere interno e bordo (anche se l'insieme è chiuso, qualunque cosa ci sia nelle altre condizioni???)

una singolarità / punto critico può essere:  
- punto interno: dove la derivata è nulla (dove il gradiente è nullo, se ci sono più variabili)   
- punto di bordo (cioè almeno una derivata parzialie non esiste???)    

quindi per trovare i punti critici in un insieme posso cercare separatamente punti interni e di bordo  
nota che il sistema ci vincoli di lagrange mostra TUTTI i punti critici, sia interni che di bordo   

#### Quando altre condizioni hanno $\le$

**Esempio**

$S= \{g=0, 0 \le z \le 2\}$  

allora per trovare i punti critici di S devo studiare separatamente la parte con i valori centrali di z e quelle con i valori di bordo, perchè quelli non appaiono nel sistema dei moltiplicatori di lagrange, cioè:  
$A = \{g=0, 0<z<2\}$   
$B=\{g=0, z=0\}$  
$C=\{g=0, z=3\}$   
$S=A+B+C$  

**Recap: se sono una diseguaglianza**

se $g \le 0$: separa in parte interna e frontiera, e poi fai:    
interna: deve avere nabla f nullo  
frontiera: moltiplicatori di lagrange  

se $condizioni \ge 0$: separa nei valori ammessi centrali e di bordo  
valori centrali: moltiplicatori di lagrange  
valori di bordo: riduci la dimensionalità, poi moltiplicatori di lagrange (o a occhio)  

### Trattare gli insiemi

**Come scriverli**

1. intersezione o sottoinsieme  
mantenere tutti i vincoli di entrambi gli insiemi (g e le 'altre condizioni')   
nota: l'ordine dei termini del prodotto vettoriale è importante, sono nell'ordine le variabili x,y,z    
    - se mi fissa una var: posso definire un nuovo insieme e una nuova funzione per diminuire la dimensionalità, togliere una var 

2. somma  
di solito non lo scrivo come insime, lo lascio espreso come somma  

3. l'insieme è grafico di una h(x,y)  
usa 3 var, dove $z=h(x,y)$, e come vincolo $g=z-h(x,y)=0$      

**Come trovare max e min**

ricorda che unione e intersezione di compatti è compatto  

1. intersezione  
semplicemente scrivilo e applica il sistema di Lagrange  

2. somma  
controlla se max e min esistono su ognuno dei due separatemente (sì se f è continua su un compatto)  
trova max e min per ognuno degli insiemi addendi   
e poi tra tutti max, trova il più max, e uguale per il min  
(è il metodo usato quando $g \le 0$, quindi $S=Int(S)+Fr(S)$)  


**Esempio: intersezione/sottoinsieme**

$S = \{(x, y, z) ∈ R^3 : x^2 + y^2 + z^2 = 1, z ≥ 0\}$  
$f(x,y,z)=x+y+x$  

$Σ=S∩(R^2×(0,+∞))$  
$\Sigma = \{(x, y, z) ∈ R^3 : x^2 + y^2 + z^2 = 1, z > 0\}$     
cioè non può più avere z=0  

$C = S ∩ (R^2 × {0})$  
$C = \{(x, y, z) ∈ R^3 : x^2 + y^2 + z^2 = 1, z = 0\}$  
$C = \{(x, y, 0) ∈ R^3 : x^2 + y^2 = 1\}$  

posso anche definirne uno nuovo per ridurre la dimensionalità:  
$D = \{(x, y) ∈ R^2 : x^2 + y^2 = 1\}$  
$h(y,z)=f(x,y,0) = x+y$  


**Esempio: grafico**

$f(x,y)=x^2-y$  
$D = \{(x,y) \in R^2: x^2+y^2 \le 1\}$  
$f: D\rightarrow R$  

$S = gr (f) = \{(x,y,f(x,y)): (x,y) \in D\}$  
$S = \{(x,y, x^2-y): x^2+y^2 \le 1\}$  
$S = \{(x,y, z): (x,y) \in D, \; g(x,y,z)=z-x^2+y =0 \}$  


**Insieme compatto**

è chiuso se:   
SIA g che le altre condizioni hanno segni $=$ oppure $\ge$, non $>$   
cioè nessun vincolo può avere una disuguaglianza stretta   

è limitato se:  
nessuna varibile può essere $\infty$   
quindi guarda g(x,y,z) scritta dentro l'insieme, controlla se qualche variabile può essere $\infty$ e comunque rispettare g   
eg. $g= x^2+y^2+z^2-9=0$, nessuna var può essere $\infty$ o non è possibile rispettare g, quindi è limitato    

compatto: limitato + chiuso  

**Nota frontiere**

la frontiera potrebbe essere unione di due insiemi  
se in g appare $\le$ due volte, scrivi due insiemi, ogni volta sostituendo uno con un uguale lasciando $\le$ l'altro  

$A=\{x^2+y^2-1\le z\le 1\}$  
$Fr \, A = B + C$  
$B = \{ x^2+y^2-1\le z, \; z=1, \} = \{ x^2+y^2-2\le 0, \; z=1, \} $  
$C= \{x^2+y^2-1=z, \;  z\le 1\}$   
(nella prima, in z sostituisci la cosa a cui è uguale)  
nota che g è quello che unisce più coordinate, non quello che fissa z=1

**Come trattare $\lambda$**

un punto (x,y,z) risolve il sistema solo se esiste un valore di $\lambda$ per cui il sistema è corretto  
quindi in pratica devo cercare una soluzione di tipo $(x,y,z, \lambda)$, di cui poi mi interesseranno solo i primi 3  

per il calcolo:  
il valore di $\lambda$ non mi interessa, quindi se $\lambda$ appare sempre con uno stesso coefficiente, posso fare un cambio di variabile  
tipo se ho sempre $2 \lambda$ posso togliere il 2, è un cambio di variabile $\lambda \mapsto 2 \lambda$  

per risolvere sistema:  
spesso, esamina a parte i casi $\lambda = 0$ e $\lambda \neq 0$  
utile se ho una riga tipo $0=\lambda z$  
così so se settare z necessariamente a 0 o no  
per ognuno, vedi quali var devono essere 0 o non-0  
magari dividi equazioni diverse, ricorda che quello vale se il denominatore è diverso da 0, controlla  parte la soluzione denominatore=0  
ricorda di controllare se i punti trovati rispettano g=0  

**Trova tuti i punti critici se S=R**

mi dice solo di calcolare i punti critici di f  
non ho un insieme su cui calcolarlo, cioè $S=R^n$  
allora è come se fosse $g=0$, quindi $\nabla g=0$, quindi la condizione per i punti critici diventa solo $\nabla f =0$  
posso anche pensarlo che l'insieme non ha bordi, quindi ci sono solo punti critici interni, che sono quelli con gradiente nullo  

1- trova tutti i punti critici, quelli con gradiente nullo  
2- scrivi la hessiana e calcolala in ogni punto, per vedere che tipo di punto è

quindi rispetta l sistema dei moltiplicatori di lagrange  
nota che un punto critico vincolato potrebbe NON essere un punto critico libero

#### Max/min e sup/inf

**Tips: esistono max e min?**

se f dipende dalle var x,y,z ma in g appaiono solo le var x,y, sicuramente S non è limitata, perchè z può essere infinito  
il contrario invece va bene, so ho g(x,y) e f(x) può essere limitato  

se f è una somma di addendi:  
max e min esistono se esistono per ogni addendo, posso considerarli separatamente  
se un addendo ha solo sup/inf, non max/min, allora f non ha max/min  
se in un addendo di f appaiono solo x,y, allora il dominio con g con solo x,y è compatto rispetto a quell'addendo, quindi quell'addendo ha max e min   

esempio:  
$f=e^{xy}+e^{-z^2}$  
$S=\{(x,y,z): \, (x-1)^2+y^2=3\}$  
S non è compatto rispetto ad f  
$e^{-z^2}$ ha max 1, quando z=0, e inf 0, quando $z=\infty$  
S è compatto rispetto a $h(x,y)=e^{xy}$, che quindi ha max e min  
quindi f su S ha max e inf (no min, perchè il secondo addendo non ha min)  

**Esempio**

A=B+C  
se A,B sono compoatti, ma C no   
A,B hanno max e min  

il max e min di A possono essere uno di questi:  
1- il max e min di B   
2- un qualunque punto critico di C  
nota che C, anche se non ha per forza max e min, potrebbe comunque averlo  

**Max/min, sup/inf**

max, min non esistono sempre  
sup, inf esistono sempre  

in sup, inf posso suare come coord $\pm \infty$, in max/min no  
quindi $e^x$ ha sup e inf, ma non max e min  
perchè sup e inf sono raggiunti a $\pm \infty$, posso trovarli a occhio  

per vedere se ha max/min:  
vedi cosa succede mandando ogni coord a $\pm \infty$, se f aumenta/diminuisce sempre di più o no   

**Dimostra sup/inf**

dimostra che ha sup/inf, non max/min:  
se serve, trova upper/lower bound di f (per semplificarla)  
e mostra che quando tendo una var a $\infty$, f va ad $\infty$  

dimostra che max/min:  
se serve, trova upper/lower bound di f (per semplificarla)  
trova inf/sup  
mostra che quel valore è vermanete raggiunto pluggando un punto P, non è ottenuto all'infinito
