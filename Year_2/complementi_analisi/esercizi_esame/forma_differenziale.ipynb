**Dati**

$ω=P \, dx+Q \, dy+R \, dz = f_x \; dx + f_y \;dy  + f_z \; dz$   
definita su $\Omega$  

**Definizione**

Una 1-forma è una combinazione lineare di differenziali  
$\omega = A_1(x) \, dx_1 + ... + A_n(x) \, dx_n$  


### Esatta / chiusa

**Condizioni**

- esatta: esiste V tale che $\omega = dV$  
campo conservativo  

- chiusa: $rot=\vec{0}$    
campo irrotazionale  

**Rapporti**

esatta -> chiusa  
se $\omega$ è continua ($F \in C^1$, dove $\omega = dF$)  

chiusa -> esatta   
se $\Omega$ è semplicemente connesso

**Come dimostrare**

dimostra che è CHIUSA: facile, usa la formula e controlla se rot=0   
nota che se un campo è totalmente radiale, è per forza irrotazionale, quindi chiuso  
basta usare la def di rotore con le coordinate sferiche  

controlla se ESATTA: faticoso, di solito 3 opzioni:  

- se dopo chiede di calcolare V: fallo direttamente e poi dì che quella è la prova che è esatta  

- se $\Omega$ è semplicemente connesso: dimostra che è chiusa (facile, con la formula di rot), e chiusa -> esatta  

- se $\Omega$ è chiaramente scelto per non essere semplicemente connesso, e dopo non chiede di calcolare V: probabilmente non è esatta  
dimostralo trovando un curva chiusa con circuitazione non nulla (con integrale non nullo)  

**Condizione irrotazionale/chiusa**

$rot(F) = det \begin{bmatrix} i & j & k \\ \partial_x & \partial_y & \partial_z \\ f_x & f_y & f_z \end{bmatrix}$

espandi il det con Lagrange rispetto alla prima riga  
tutti gli addendi devono essere 0  
l'ultima riga sono le componenti di $\omega$, P,Q,R  

quindi in $R^2$ è solo det=0: $\quad \frac{∂Q}{∂x}​=\frac{∂P}{∂y}$  

**Semplicemente connesso**

in $R^2$: non ha buchi  

in $R^3$: non ha un'asse mancante (o un tunnel)  
 
semplicemente connessi:  
- tutto $R^3$
- palla aperta o chiusa, tipo $x^2+y^2+z^2<1$ (o con $\le$)    
- $R^3$ privato di una retta finita  
-  $\Omega_1 = (0, \infty) \times R$  
un semipiano in $R^2$ è sempl connesso  

#### Dimostra che NON è esatta

**Come fare**

trova un curva chiusa con circuitazione non nulla (con integrale non nullo)  

steps:  
1. dati: ho $\omega$  
2. scelgo $\gamma$, una curva chiusa
3. calcolo l'integrale $\int_\gamma \omega$  
se è non-0, non è esatta  
se è 0, non so dire niente (perchè deve esserlo su ogni curva per essere esatta)    

**Come scegliere la curva $\gamma$**

1. guarda dove il dominio è bucato  
se faccio questo metodo, vuol dire che $\Omega$ non è semplicemente connesso  
esempio: $\Omega = R^3- \{(0,y,0): y \in R\}$    
manca tutta l'asse y, cioè i punti dove x=z=0  

2. scegli una curva che avvolge il buco  
qui, la curva che avvolge il buco nel piano y=0  
$\gamma(t)= (r \cos t, 0, r \sin t) \quad t \in [0, 2\pi]$  
per r va bene un valore a caso, metti 1  

regole generali:  
buco è un punto in $R^2$ -> scegli un cerchio attorno a quel punto  
buco è una retta in $R^3$ -> scegli un cerchio intorno a essa  
cioè serve qualcosa che contenga il buco  

attenta, deve essere chiusa  
quindi di solito scelgo un cerchio  

**Esempio: dimostra che NON è esatta**

1. dati  

$\omega = -\frac{z}{x^2+z^2}dx + y \, dy + \frac{x}{x^2+z^2}dz$  
$\Omega = R^3- \{(0,y,0): y \in R\}$  
 

2. scelgo $\gamma$  

a $\Omega$ manca l'asse y, quindi scelgo un cerchio lì attorno  
$\gamma :[0, 2 \pi] \rightarrow \Omega$  
$\gamma(t)= (\cos t, 0, \sin t)$  

3. calcolo  

quando scrivo $\omega$: devo sostituire tutte le x,y,z e dx,dy,dz con quelle definite dalla curva scelta  
nota che $dx = d(\cos t)= - \sin t \; dt$  
ogni componente di $\omega$ è un integrale separato  

$\int _\gamma \omega = \int_0^{2\pi}  -\sin t \; (- \sin t \, dt) + \int_0^{2\pi} 0 dy + \int_0^{2\pi}\cos t (\cos t \; dt)= 2\pi \neq 0$  

quindi non è esatta

#### Metodo base

**Steps: 2 componenti**

$\omega = f_x \, dx + f_y \, dy$  

1. $f(x,y)=\int f_x dx = ... + c(y)$     
scegli una componente, integrala per avere f    
la costante c è funzione dell'altra variabile, quella che è rimasta fissa  

2. $\partial _y f(x,y)= ... c'(y) = f_y$    
differenzia rispetto all'altra variabile, quella dentro la costante, e uguagliala alla componente che non ho usato  
così trovo c'  

3. $c(y)= \int c'(y) \, dy= ... +c$  
integra c' per troavare c, pluggala e ho finito  
nota che, alla fine, rimane comunque un +c  
perchè V è un potenziale, è deifnito a meno di una costante  


**Steps: 3 componenti**

$\omega = f_x \, dx + f_y \, dy + f_z \, dz$  
 
1. $f(x,y,z)=\int f_x \, dx=...+c(y,z)$    
scegli una componente, e integrala per ottenere f(x,y,z)  
la costante è funzione di tutte le variabili non usate  

2. $\partial _y f(x,y,z) = ... + c_y(y,z) = f_y$  
differenzia f rispetta ad una delle variabili nella costante c, ed uguagliala alla componente giusta  
nota che così ottengo una derivata parziale di c  

3. $c(y,z)=\int c_y(y,z) \, dy = ... + g(z)$  
integra la derivata parziale di c rispetto alla variabile per cui ho appena derivato, così ottengo g in funzione dell'unica variabile che non ho ancora usato  
così, in f = ..., posso sostituire c con g  

4. $\partial _z f(x,y,z) \, dz = ... g_z(z)$  
$g(z) =\int g_z(z) \, dz$  
ora è come l'esercizio con due sole componenti, deriva per la var non ancora usata e trova c  


**Esempio: 2 componenti**

$\omega = 2x^3y^4 dx \; + 2x^4y^3 dy$  

1- integro $f_x$    
$f(x,y) = \int f_x \; dx = \frac{1}{2}x^4y^4 +\frac{1}{2}x^2+c(y)$   

2- derivo  
$\partial _y f = 2x^4y^3+c'(y)=f_y$  
quindi $c'(y)=y$  

3- trovo c   
$c = \int c'(y)= \frac{1}{2}y^2 +c$  

$V=f(x,y)=  \frac{1}{2}x^4y^4 +\frac{1}{2}x^2 + \frac{1}{2}y^2 +c$  

**Esempio: 3 componenti**

$\omega = P \, dx + Q \, dy + R \, dz$  

1- integro rispetto a x  
$f(x,y,z)= \int P \, dx = ... + c(y,z)$  

2- derivo rispetto a y  
$\partial_y f = ... + c_y(y,z) = Q$  
trova $c_y$  

3- semplifico c:  
$c(y,z)=\int c_y (y,z) dy= ... +g(z)$   
$f= ... + g(z)$  

4- derivo rispetto a z:  
$\partial_y f = ... +g_z(z) = R$   
trova $g_z$  

5- integro per trovare g  
e la pluggo nella soluzione  

#### Trucchetti

**Componenti univariabile**

se in P,Q,R appare in ognuno solo una variabile, posso trovare le componenti di V integrando ogni pezzo separatamente e poi sommandoli  
si può fare solo se P,Q,R solo univariabili, sennò rimane +c(x,y,z) in funzione delle altre variabili  
così invece si cancellano, rimane solo +c come numero    

cioè:  
$V = \int P \; dx + \int Q \; dy + \int R \; dz$  

poi posso mettere +c=0 se nonho richieste particolari???

**Coordinate polari**  

una funzione potrebbe non essere univariabile in coord cartesiane, ma esserlo in polari  

quando passo alle polari succede:  
$r^2=x^2+y^2+z^2$  
$r \, dr = x \, dx + y \, dy + z \, dz$ (attenzione, serve la r davanti, non si può togliere, la ottengo deruvando la sostituzione con r sopra)          
comodo perchè magari p tutto funzione di r, non degli angoli    
sennò devo usare anche $x=r \cos \theta, \quad y = r \sin \theta$, ma non ne vale la pena se appare anche $\theta$, fallo solo se ci sono una x o una y in ogni componente, che spariscono inglobate dal dr  

1. passo alle coordinate polari  
2. faccio l'integrale indefinito  
(uso c=0, se non ho particolari richieste)
3. torno alle coordinate cartesiane, pluggando in r la formula in cartesiane   

nota che se un campo è totalmente radiale, è per forza irrotazionale, quindi chiuso  
basta usare la def di rotore con le coordinate sferiche  
ed è fisicamente ovvio, perchè se il campo punta in direzione radiale non può generare una rotazione  

se lo faccio, e poi devo calcolare integrale su una curva, dopo aver trovato le coord x,y,z dei punti di inizio e fine, trova le coord r di quei punti (co pitagora)  

**Esempio: coordinate polari**
 
$\omega = x \ln (x^2+y^2+z^2) dx + y \ln (x^2+y^2+z^2) dy + z \ln (x^2+y^2+z^2) dz$  

1- passo alle coordinate polari  
$\omega = r \, \log (r^2) dr$  

2- integrale indefinito  
$V(r)= \int r \, \log(r^2) \; dr = \frac{1}{2}(r^2 \log (r^2-r^2))+c$  
ora è funzione solo di r, quindi posso integrare facilmente (qui, usando integrazione per parti)  
mi serve una primitiva qualunque, quindi posso mettere c=0   

3- plug in r l'espressione in coordinate cartesiane  

**Colpo d'occhio**

a volte è più veloce guardare $\omega$ e cercare di indovinare la sua primitiva  
ad esempio, se le componenti di $\omega$ al denominatore hanno qualcosa al quadrato, probabilmente V è una frazione con quella cosa al denominatore (senza quadrato)

### Calcola un integrale su una curva $\gamma$

**Se $\omega$ è esatta**

esatta vuol dire che il campo è conservativo   
l'integrale è uguale su tutte le curve che partono e finiscono nello stesso punto

steps:  
1- trova gli estremi della curva  
2- calcola V in quei punti  

quindi:  
$\int_\gamma \omega = V(\gamma(B))-V(\gamma(A))$  
V è una primitiva di $\omega$ (quindi la formula che definisce il potenziale)  
V è calcolata nei punti di estremità della curva

**Se $\omega$ NON è esatta**

calcola l'integrale sulla curva normalmente, niente trucchetti  
cioè scrivi l'integrale e sostituisci x,y,z con quello indicato da $\gamma$, poi devi davvero risolvere l'integrale  
come estremi di integrazione, metti gli estremi di integrazione di t  

esempio;  
$\gamma(t)= (\cos t, \sin t, \sin t)$  
$\gamma : (0, \pi) \rightarrow R^3$  
come differenziale usa: $dr = \dot{\gamma(t)} \; dt$  
quindi: $\int_\gamma \omega = \int_0^\pi \omega(\gamma(t)) \; \dot{\gamma(t)} \; dt$   

sostituisci x,y,z, con i valori della prima riga  
come estremi di integrazione usa $0$ e $\pi$  

**Esempio se $\omega$ è esatta**

la curva è:  
$\gamma(t)=(3 \cos 2t, 8 \sin 2t)$  
$t: [0, \pi] \rightarrow R^2$  
ho già calcolato V(x,y)  

trovo i punti all'estremità:  
$\gamma(0) = (3,0)$  
$\gamma(\pi)=(3,0)$  

poichè è esatta, posso calcolare l'integrale così:  
$\int _\gamma \omega = V(\gamma(\pi)-V(\gamma(0)))=V(3,0)-V(3,0)=0$
