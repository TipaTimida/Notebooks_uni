**Dati**

funzione: f(x,y,z)  
se dice calcola l'area di un grafico, vuol dire f=1  

superficie: S  
data come insieme, a me interessa guardare g per vedere che forma ha  

dominio: D   
viene dopo aver parametrizzato S, è la cosa su cui integro  

**Modi di risolvere**

ci sono due metodi: 

1- parametrizzazione standard $r$  
2- se la superficie è scrivibile come grafico di una funzione bivariabile $u$, cioè $z=u(x,y)$  

NON scegliere l'1 per una sfera, è troppo difficile   
sempre meglio il metodo 2  

**Steps**

$ \int_S f \, d\sigma$  

1. parametrizza  
1- scegli un vincolo $u$  
2- scegli parametrizzazione $\vec{r}$    

2. sostituisci il differenziale  
1- $d \sigma = \sqrt{|\nabla u|^2 +1} \, dA$    
2- $d \sigma = ||r_x \times r_\theta|| \, dA$   

3. cambia il dominio del dell'integrale a D, trova i suoi estremi di integrazione  

4. sostituisci la parametrizzazione in f (se c'è f)

**Note su f**  

la presenza di f influisce solo sullo step 4:  
devo sostituirci dentro la parametrizzazione, e probabilmente complica l'integrale  
ma non influisce sulla scelta della parametrizzazione  
perchè non cambia quali punti dello spazio voglio considerare, ma solo quanto valgono  

#### Esempi

**Metodo 1**

$C=\{(x,y,z): y^2+(z-1)^2=x^2 \quad 0 \le x \le 1\}$  
f=1  

1- parametrizza  
$u(y,z)=x=\sqrt{y^2+(z-1)^2}$  
D: dominio di U è una sfera di raggio 1 (perchè è il massimo di x) centrata in (0,1)  

2- sostituisci differenziale  
$|\nabla u|^2 = (\partial_x u)^2+(\partial_y u)^2 = 1$  
$d \sigma = \sqrt{2} \, dA$  

3- calcola integrale  
$\int_D \sqrt{2} dA= \sqrt{2}  \int_D \, dA$  
D è un cerchio di raggio 1, quindi $\int_D \, dA = \pi$  

**Metodo 2**

$S= \{(x,y,z): y^2+(z−1)^2=x^2, 0≤x≤1\}$

1- parametrizzazione (è un cono)  
$r(x, \theta)=(x, x \sin \theta, 1+ x \cos \theta)$  
$x \in [0,1]$  
$\theta \in [0, 2\pi]$  

2- sostituisco il differenziale    

$r_x = (1, \sin \theta, \cos \theta)$  
$r_\theta = (0, x \cos \theta, - x \sin \theta)$  
$r_x \times r_\theta= (-x, x \sin \theta, x \cos \theta)$  
$||r_x \times r_\theta||^2=2x^2$ (prodotto scalare tra i due vettori)   
$||r_x \times r_\theta||=x \sqrt{2}$  

3- calcola integrale (qui lo faccio lungo e bene)  
$\int_D  x \sqrt{2} \, dx \, d\theta = \int_0^1 \int_0^{2\pi} x \sqrt{2}\; d\theta \; dx = \sqrt{2}(2 \pi)\frac{1}{2}= \pi \sqrt{2}$  
qui non devo fare la sotituzione col diff polare moltiplicando per r, ma dA è già espresso come moltiplicazione delle due var nella parametrizzazione  

#### Parametrizza una superficie: metodo 1

**Steps**

1. scegli una var  
2. scrivila come funzione delle altre  
3. descrivi il suo dominio D  

**1. Quale variabile scegliere**

1- se c'è una variabile di cui mi dice gli estremi, tipo $0<x<1$, scegli quella (sarà utile per trovare il dominio)  

2- sennò scegli una variabile che si isola facilmente  

**3. Descrivi il suo dominio**

1. trova valore min e max che può avere la funzione u  
(se so gli estremi della variabile isolata è facile)   

2. sostituisci u con la sua espressione delle altre var  
questa è la g del dominio D  

**Esempio**

$S=\{(x,y,z): \quad y^2+(z-1)^2=x^2, \quad 0\le x \le 1 \}$   

1- scegli una var  
x, perchè so gli estremi  

2- la isolo  
$x=u(y,z)=\sqrt{y^2+(z-1)^2}$   

3.1 - trovo min e max di u  
$0 \le u \le 1$, perchè $x=u$  

3-2 - descrivi dominio  
$D= \{(y,z) \in R^2: \quad 0 \le u(y,z) \le 1\} = \{(y,z) \in R^2: \quad y^2+(z-1)^2 \le 1\}$  

interpretazione:  
è un disco del piano yz, centrato in (0,1), di raggio 1  
dovrei mettere $0\le u \le 1$, ma u è fatta di quadrati e per forza positiva, quindi posso omettere che è maggiore di 0  
questa informazione mi aiuta a trovare gli estremi di integrazione   

**Esempio 2**

2- trova e isola var:  
$y=u(x,z)=\sqrt{9-x^2-y^2}$   

3.1 - min e max di u  
$u \ge 0$ perchè è una radice (che espresso in funzione delle altre, sarebbe $x^2+y^2 < 9$)  
nota che: se dalle altre condizioni mi viene detto anche $x^2+z^2 \le 5$, questa è più forte e vince  

$D = \{(x,y) : x^2+y^2\le 5\}$   
cioè un disco di raggio $\sqrt{5}$  

**Nota: S vs D**

S: superficie nello spazio 3D su cui integro  
cioè sommo il valore di f su tutti i punti appartenenti ad S  

D: dominio della funzione u  
pezzo del piano xy  

come scrivere: passa da S a D quando inserisco la parametrizzazione u  
$\int \int _S f \; dS = \int \int _D f  \; \sqrt{1+|\nabla u|^2} \; dA$  

#### Parametrizzare una superficie: metodo 2

**Steps**

1. riconosco la forma geometrica  

2. per ogni var, vedo se c'è accanto un termine noto (o coefficiente) 

3. scrivo il sistema per parametrizzare  

**Esempi comuni**

| Superficie               | Parametrizzazione tipica                                    | Dominio                              |
| ------------------------ | ----------------------------------------------------------- | ------------------------------------ |
| Cilindro $x^2+y^2=R^2$   | $(R\cos\theta,\, R\sin\theta,\, z)$                         | $\theta\in[0,2\pi),\, z\in[a,b]$     |
| Cono $x^2+y^2 = (k z)^2$ | $(kz\sin\theta,\, kz\cos\theta,\, z)$                       | $z\in[0,h],\ \theta\in[0,2\pi)$      |
| Sfera $x^2+y^2+z^2=R^2$   | $(R\sin\theta \cos\varphi,\, R\sin\theta\sin\varphi,\, R\cos\theta)$ | $\theta\in[0,\pi],\ \varphi\in[0,2\pi)$ 

**Sistema con termine noto**

cosa:  
tutte le variabili al quadrato posso in realtà essere una sottrazione o una somma tra variabile e costante  

come: scrivi il sistema lineare  
sinistra: le variabili nella mia equazione  
se c'è un termine noto ci va messo, poi devo rielaborare in modo da avere solo la var a sinistra  
destra: la parametrizzazione standard per quella figura, scritta in tabella  

esempio: $y^2+(z−1)^2=x^2$ è un cono  
(in pratica prima fai sostituzione $z \rightarrow z-1$ e lavora su quella, trovaci la parametrizzazione, poi rimetti la z-1)    
$\begin{cases}
x =x \\
y = x \sin \theta \\
z-1 = x \cos \theta \\
\end{cases}$  

poichè la z ha un termine noto:  
$z=1+x \cos \theta$  
quindi la parametrizzazione è:  
$(x, x \sin \theta, 1+x \cos \theta)$   

gli estremi della x cambiano se la z ha un termine noto???  
ma se è una sotituzione z->z-1, non dovrebbe esserci una parentesi a (1+x), prima di moltiplicarlo per il coseno???  

**Riconoscere e interpretare forma geometrica**

parto da $f(x,y,z)=0$  

cilindro: una variabile è assente  
quindi la sezione ortogonale è sempre la stessa  
esempio: $x^2+z^2=1$, cilindro circolare con asse y  
la var che NON appare è quella che nella parametrizzazione resta uguale  

sfera: $x^2+y^2+z^2=R^2$   
ben noto   

cono: rapporto lineare tra raggio della sezione e distanza da un punto/asse  
esempio: $y^2+z^2=(kx)^2$  

**Esempio con interpretazione**

1. riconosco la forma geometrica  

L’equazione $y^2+(z−1)^2=x^2$ dice:  
per ogni x, la sezione in $(y,z)$ è un cerchio di raggio ∣x∣ e centro $(y=0,z=1)$   

2. un cerchio si parametrizza in modo standard  

$y = x \sin \theta$  
$z = 1+ x \cos \theta$  
$\theta \in [0, 2\pi)$  

3. aggiungo la coordinata x  

quindi ottengo:  
$r(x, \theta)=(x, x \sin \theta, 1+x \cos \theta)$  


#### Aiuti

**Se il modulo del gradiente è costante**

se il modulo è costante, cioè un termine noto, che non dipende dalle var:  
$\int \int _D \, dy \; dz = Area(D)$  
poi l'area di D la calcolo con le formule di geometria, cercando di capire che forma ha  

esempio: usando la D definita sopra  
$\sigma(C)=\int \int_D \sqrt{1+1} \; dy \; dz = \sqrt{2} \; Area(D) = \sqrt{2} \times (\pi 1^2) = \pi \sqrt{2}$  

**Se la funzione $u$ non è definita in un punto**

cioè se viola la C.E. in un punto, ad esempio in (0,1)  
Possiamo eliminare il punto (0, 1) da D, il dominio di $u$, perchè ha misura di area nulla  

quindi invece di $\int_D$ scrivo $\int_{D \backslash \{(0,1)\}}$ in basso nell'integrale  

### Coordinate vs parametrizzazioni

**Cosa sono**

coordinate: per descrivere tutto lo spazio 3D  
si usa come sostituizione negli integrali di volume  
voglio usare un set di coordinate che mostri meglio alcune simmetrie dell'oggetto che studio  

parametrizzazione: per descrivere una superficie (nello spazio 3D, ma è un oggetto 2D)  
si usa negli integrali di superficie  
viglio diminuire la dimensionalità (Rappresentazione di un luogo geometrico, di una legge, di una funzione, mediante un numero conveniente di parametri)  

**Rapporto**

le coordinate non dipendono dall'oggetto, quindi tipo non dipendono dal raggio della sfera specifica che voglio studiare  
mentre la parametrizzazione sì  

in pratica:  
cambio coord: ha una generica r  
parametrizzazione: ha la r specifica del solido che studiando  

infatti cambio coord, visto che deve descrivere un volume e non una superficie, ha bisogno di una var in più (dimensione 3, non 2)  

**Esempio: cilindro**

per volume e superficie, cambia g (uguaglianza o disuguaglianza)  

volume: $E = \{x^2+y^2 \le 4, \quad 0 \le z \le 3\}$  
coord: $(x,y,z) = (r \cos \phi, r \sin \phi, z)$  
si usa per: $dV = r \, dz \, dr \, d\phi$  

superficie: $S = \{x^2+y^2 = 4, \quad 0 \le z \le 3\}$  
parametrizzazione: $\Phi (\theta, z): (2 \cos \theta, 2 \sin \theta, z)$  
si usa per: $d \sigma = ||\Phi_\theta \times \Phi_z|| d\theta \, dz$
