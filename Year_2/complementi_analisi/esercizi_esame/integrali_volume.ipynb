### Che esercizio

**Tipi di esercizi**

2 tipi, e come riconoscerli dal testo:  

1. di volume  
vincolo g è una DISUGUAGLIANZA ($g \le 0$)  

2. di superficie  
vincolo g è un UGUAGLIANZA ($g=0$)  

**Nota su g**

$E = \{g,  \quad altre \, condizioni\}$  

se g(x,y,z) ha 3 variabili:  
$g = 0$: definisce una superficie (uguaglianza)   
$g \le 0$: definisce un volume racchiuso nella superficie (diseguaglianza)  

### Guida

**Steps**

$\int_E \, dV$  

1. capisci che solido è, scegli le coord polari più adatte  
di solito, sferiche o cilindriche  

2. sostituisci le coord scelte in g  

3. trova gli estremi di integrazione  
come si fa dipende dal tipo scelto  

#### Scegli coord adatte

**Scegli coord più adatte**

a seconda di g (i termini noti possono essere qualunque numero):    
1. $x^2+y^2+x^2 \le 9 \quad$: sfera  
2. $x^2+y^2 \le z^2 + 4,  \quad x^2+y^2 \le 4 \quad$: cilindriche (sono un cono e un cilindro, ma usano entrambe coord coniche)  
3. solido di rotazione: cilindriche  

**Recap**

vedi da che lato appare z per distinguere tra sferiche e cilindriche  
se è un solido di rotazione, me lo dice espressamente  

#### Cambio coord

**Cambio coord**

cilindriche:  
$f(x,y,z) = f(r \cos \varphi, \quad r \sin \varphi, \quad z=z)$  
$dV = r \; dz \; dr \; d\varphi$   
$r^2 = x^2+y^2$  

sferiche:  
$f(x,y,z)=f(r \; \sin \theta \; \cos \varphi, \quad r \; \sin \theta \; \sin \varphi , \quad r \; \cos \theta )$  
$dV = r^2 \; \sin \theta \; dr \; d\varphi \; d\theta$  
$z^2=x^2+y^2+z^2$   

polari in 2D:  
$f(x,y,z) = f(r \cos \varphi, \quad r \sin \varphi)$  
$dV = r \; dr \; d\varphi$   


#### Trova estremi di integrazione: sfera

**Sfera: Teoria**  

estremi standard:  
$ r \in [0, R]$      
$\varphi \in [0, 2\pi]$ : angolo azimutale (nel piano xy)  
$\theta \in [0, \pi]$  : angolo polare (da z-asse)  

$\varphi$: come l'angolo in coordinate polari, l'angolo sul piano orizzonatale  
parte dall'asse x e poi in senso antiorario verso l'asse y  

$\theta$: l'angolo verticale, l'altezza del punto  
$\theta = 0$ è in cima, $\theta = \pi /2$ è sul piano xy, all'equatore, $\theta=\pi$ è sul fondo, al polo sud

**Sfera: trova estremi**

r: il minimo è 0, il massimo lo vedo da g  
esempio: $r^2 \le 4^2 \rightarrow r \in [0,2]$  

angoli: modo intuitivo, pensa a cosa voglioni dire gli angoli  
esempio: solo nel primo quadrante ($x>0, y>0$) $\rightarrow \varphi \in [0, \pi/2]$  
esempio: solo semisfera superiore $\rightarrow \theta \in [0, \pi/2]$

#### Trova estremi di integrazione: cilindro

**Cilindro: teoria**

il vincolo è: $a(r) \le b(z) \le c$  

se c NON contiene r, cioè è un termine noto: posso scegliere se integrare prima per z o r  
per r è più facile, perchè estremo inferiore è 0, quindi primo integrale dà un solo addendo  

se c contiene r: integra prima per z, è molto più facile  

**Come trovare estremi di integrazione: integra prima per z**

1- quelli di z sono i due lati del vincolo  
ricorda di mettere la radice quadrata  
es. $r^2 \le z^2 \le 8 -r^2 \rightarrow z \in [r, \sqrt{8-r^2}]$  

2- per r, innanzitutto so $r \ge 0$, poi unisci componenti a e c del vincolo:  
$r^2 \le 8-r^2 \rightarrow r \le 4$  
ricorda la radice  
nota che r deve essere positivo, prendi soluzione positiva della radice  

3- per phi, usa il default $\phi \in [0, 2 \pi]$  

**Come trovare estremi di integrazione: integra prima per r**

fallo SOLO se in c NON appare r  

1- quelli di r:  
so che $r \ge 0$   
e poi prendi i componenti a e b del vincolo, dove appaiono r e z:    
$r^2+4 \le z$  

2- quelli di z:  
nel vincolo, setta r=0, vedi cosa ottengo   
$r^2+4 \le z^2 \le 9 \rightarrow z \in [2, 3]$  
ricorda la radice  

3- per phi, di default   

**Spiegazione del metodo NON sicuro: integrare prima per r e phi, e alla fine per z**

in pratica mi sto chiedendo: che succede per ogni specifico z? com'è la sezione ad ogni z?  

$x^2+y^2+4 \le z^2 \le 9 \; \rightarrow \; x^2+y^2 \le z^2-4$  
cioè per ogni z ho un disco di raggio $z-4$      
allora posso:  
1- trovare gli estremi di z (in assoluto, non quelli dipendenti da r, quindi come se z fosse la seconda variabile), che sono $z \in [2,3]$  
2- integro prima sulla parte orizzontale, di cui so calcolare l'area perchè è un cerchio, e dà $\pi (z^2-4)$  
o posso semplicemente fare l'integrale  

sennò per scriverlo, formale, potrei scrivere:  
$m(E)= \int_2^3 dz \, (\int_{E_z} dx \, dy) = \int_2^3 dz \; B(0, \sqrt{z^2-4}) = \int_2^3 dz \, \pi (z^2-4)$

**Esempio**

insieme:  
$E=\{ x^2+y^2+4 \le z^2 \le 9, \quad z\ge 0 \}$  

metodo con calcoli:  
1- passo a coord cilindriche  
$r^2+4 \le z^2 \le 9$  

2- trovo estremi di integrazione (ordine dei differenziali è come voglio io, quindi scelgo la versione più facile)   

se integro prima per dr:  
$r \in [0,\sqrt{z^2-4}]$  
$z \in [2,3]$  
$\phi \in [0, 2 \pi]$   

se integro prima per dz:  
$z \in [\sqrt{4+r^2}, 3]$  
$r \in [0, \sqrt{5}]$  
$\phi \in [0, 2 \pi]$  

3- integro  
scegli l'ordine giusto di differenziali a seconda della scelta fatta prima  

#### Solido di rotazione

**Come fare**

mi descrive la sezione che faccio ruotare  
disegna questa sezione in un piano xy, poi rinominalo $rz$  

estremi di:  
$z$: sono funzioni di r, i lati diagonali  
nota che in questo piano, y=ax diventa z=ar  
$r$: gli estremi possibili di r, guarda a occhio (le linee erticali che delimitano)  
$\phi$: standard, $[0, 2 \pi]$  

NON posso scrivere gli estremi in x,y e poi fare la sostituizione per passare alle coord cilindriche, ma devo vederli graficamente  

**Esempio**

1- figura sul piano yz: triangolo con vertici: (1,2), (1, -2), (3,0)  
noto che è simmetrico rispetto al''asse, quindi posso integrarne metà e poi moltiplciare per due  

2- la descrivo  
$z \in [3-r, 0]$, dove 3-r è il lato sopra diagonale, 0 è l'asse orizzontale, tanto è simmetrico, poi metto un coeff 2  
nota che considera il piano xy ove ho disegnato la figura come il piano rz  
$r \in [1,3]$  
$\phi \in [0, 2 \pi]$  

3- integro  
$V = 2 \, \int_B  dV = 2 \, \int _0^{2 \pi} \int_1^3 \int_0^{3-r} r \, dz \, dr \, d\phi = $   
$=2 \,(\int_0^{2 \pi} \, d \phi) (\int_1^3 \int_0^{3-r} r \, dz \, dr) = \ \, (2 \pi) ( \int _1^3 r(3-r) dr) = 4 \pi [3/2r^2-r^3/3]^3_1= 40/3 \pi$  
2 all'inizio è per come ho definito la figura, è simmetrica rispetto all'asse x quindi ne integro metà  

**Scrivere in modo formale**

descrivo con degli insiemi, dove scrivo di volta in volta gli estremi di integrazione  

descrivere la sezione (figura piana)  
(ha estremi di r e z, ma le chiamo ancora y e z, perchè sono sul piamo yx)   
$S = \{(y,z) \in R^2 : \, 1 \le y \le 3, \, y-3 \le z \le 3-y\}$  

descrivere il solido di rotazione (in coord cilindriche, contiene estremi dell'angolo):  
$E = \{r \, \cos \phi, r \, \sin \phi, h\}: (r,h) \in S, \, 0 \le \phi \le 2 \pi$  
questo contiene la condizione su $\theta$ e le altre due (per quello dico che è in S, così non le riscrivo)  
nota che, per mostrare che sono coord diverse, z posso soprannominarla h qui  
ma poi devo fare dh nell'integrale, meglio di no e chiamarlo z  

#### Tips calcolo

**Variabili separabili**

se:  
- nell'integrale posso separare in due FATTORI (non addendi), uno che contiene tutte le $r$, uno tutte le $\varphi$   
- ho un dominio prodotto  
cioè $D= [a,b] \times [c,d]$, NON ho funzioni negli estremi di integrazione  

allora posso separare l'integrale in prodotto di due integrali, è più elegante   
o anche se una variabile non appare proprio, allora l'integrale di quella variabile avrà 1 come funzione integranda    

cioè:  
$f(x,y)=g(x)h(y)$  
$\int f \; dx \, dy = (\int g \, dx)(\int h \, dy)$  

esempio:  
$\int_0^{2\pi} \int_0^1 \rho \cos \theta \; d\rho \; d\theta = (\int_0^{2\pi} \cos \theta \; d\theta) \; (\int_0^1 \rho \; d\rho)$  


**Proprietà**

se D può essere spezzato in due parti disgiunte, posso separare l'integrale in due (come quando in 1D spezzo l'intervallo di integrazione)
