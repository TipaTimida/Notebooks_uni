### Definizioni

**Insieme**

tutte le cose definite come tutti i punti che rispettano qualche caratteristica, il vincolo g  
a seconda del vincolo g, posso distinguere cose diverse, tipo un volume o una superficie  

**Funzioni vettoriali CONTINUE**

a seconda delle componenti di dominio e codominio:  

- curva: $r:I \subseteq R \rightarrow R^n$   
cambiando il valore in input, mi muovo lungo la curva, avanti e indietro  

- superficie : $\sigma: D \subseteq R^2 \rightarrow R^3$   
dando 2 valori in input, individuo un punto  

- campo: $F: \Omega \subseteq R^n \longrightarrow R^m$  
le componenti in output sono funzioni associa ad ogni punto dello spazio un vettore  

il dominio deve essere continuo in 1D, o aperto e connesso in 2D    

**Nota**

una curva e una superficie sono parametrizzazioni e funzioni vettoriali   
funzione vettoriale: dà tanti valori in output quante sono le coordinate del punto, non un solo valore  
parametrizzazione: Rappresentazione di un luogo geometrico, di una legge, di una funzione, mediante un numero conveniente di parametri  

funzione vettoriale: o il dominio o il codominio ha più componenti  

**Arco di una curva**

$r|_{[a,b]} \quad$ con $[a,b] \in I$  

intuitivo: un pezzo della curva, ricavato utilizzando come dominio solo un pezzo dell’intervallo I  

si può scrivere anche: 
$r: [a,b] \rightarrow R^n$  


#### Sostegno

**Sostegno di curva/superficie**

è l'immagine della funzione (il disegno)  
l'insieme dei punti effettivamente percorsi dalla curva, ovvero la sua immagine geometrica nel piano o nello spazio   
in pratica il disegno, formato da tutti i vettori ottenuti facendo variare il parametro indipendente all'interno del dominio

se $\gamma(t), \; t \in I$ è la curva, il sostegno è $A = \{\gamma(t), \; t \in I\}$

**Disegna il sostegno di una curva**

disegno nel piano (x,y)  

$r(t)=(0, 2t) \quad t \in [-2,0]$  
è un tratto verticale da (0,-4) a (0,0)  
ovvio, perchè x è sempre 0, y ha tutti i numeri da -4 a 0

$r(t)=(2 \sin(\pi t), -t^2) \quad t \in (0, 2]$  
trasforma in un sistema, poi sostituisci per avere una funzione implicita  
$\begin{cases}
x=2 \sin (\pi t) \\
y = -t^2 & t = \sqrt{-y} \end{cases}$   
$x = 2\sin(\pi \sqrt{-y})$  
non la so comunque disegnare  

### Come si scrive

**Curva**

$r: [0, \pi] \rightarrow R^2$  

metodi di scrivere:  

$t \mapsto (\cos(t), \sin(t))$  

$r : \begin{cases} x(t)=\cos(t) \\ y(t) = \sin(t) \end{cases}$  

$r(t)=(\cos(t), \sin(t))$  

è un semicerchio  

**Superficie**

superficie in forma cartesiana:  

$\sigma = \begin{cases}
x = x(u,v) \\
y=y(u,v) \\
z=z(u,v) \end{cases}$   

di solito fissane 2, tipo y=u,  z=v, quindi in z appare l'equazione cartesiana del piano, ma con u,v invece di y,z  

o comunque, scrivi l'equazione cartesiana nella variabile più facile  

**Esempio**

$10x+2y+5z+3=0$  

$\sigma = \begin{cases}
x = \frac{1}{10}(-2u-5v-3) \\
y=u \\
z=v \end{cases}$  

**Esempio 2**

$z=x^2+y^2$  

$\sigma = \begin{cases}
x = u \\
y=v \\
z=u^2+v^2 \end{cases}$

### Tipi

**Come scrivere definizioni**

una curva ... continua è una funzione $r: I \rightarrow R^n$, con $I = [a,b]$, tale che:  

**Definizioni** 

chiusa: primo e ultimo punto coincidono   
cioè $r(a)=r(b)$  

continua: r continua su I  
problematico solo se definita a tratti, controlla che sia vero ai punti di separazione di tratti diversi  

derivabile/ di classe $C^1$: OGNI COMPONENTE DEL VETTORE $\vec{r}(t)$ deve essere derivabile   
stessa definizione di una funzione vettoriale  

semplice: non tocca due volte lo stesso punto (eccetto inziale e finale, quello va bene)   
cioè  $r|_{[a,b]}$  è iniettiva, eccetto che cagli estremi  

**Curva/superficie semplice**

una curva tale che $r|_{[a,b]}$ è iniettiva  
(o per la superficie, $\sigma$)  

graficamente: se il suo sostegno non ha lacci   
(il disegno non ha incroci, non tocca due volte lo stesso punto)  
laccio: punto toccato almeno 2 volte  

analiticamente: presi due valori diversi di t, si ottengono due diversi vettori r (anche una sola componente diversa basta)  
dalla definizione, vedo che è l'analogo dell'iniettività delle funzioni: se $r(t_1)= r(t_2)$, allora $t_1 =t_2$   
$t_1$ e $t_2$ non possono essere i punti iniziali e finali, quello vuole solo dire che è chiusa  

### Regolare

#### Definizione

**Curva/superficie regolare**

r, $\sigma$ è una parametrizzazione   
definisce una curva, superficie regolare se:  

condizione curva:   
$r \in C^1(I)$  
$r'(t)\neq 0 \quad  \forall t \in I$  

condizione superficie:  
$\sigma \in C^1(D)$  
$\frac{\partial \sigma}{\partial u}$ e $\frac{\partial \sigma}{\partial v}$ sono linearmente indipendenti  

**Definizione formale: superficie regolare**

per ogni punto p, esiste un intorno U dove:  
S (la superficie) può essere rappresentata come grafico di una funzione differenziabile o come immagine regolare di una mappa parametrica σ(u,v) 

**Definizione intuitiva: superficie regolare**

Una superficie regolare in $R^3$ è un oggetto bidimensionale che, localmente, si comporta “come un piano”  

cioè, in ogni punto:  
1- assomiglia a un piano (niente spigoli o cuspidi)  
2- puoi descriverla localmente con due coordinate (u,v)  
3- è possibile definire un piano tangente e quindi una direzione normale  
cioè posso zoomare ovunque e vedere un piano  

esempio:  
sfera, cilindro: ok, sempre liscio  
cono: no, il vertice non ha piano tangente, non è regolare  

**Definizione pratica**

$\nabla r \neq 0$ in ogni punto della curva/superficie     
fai il gradiente e non posso avere tutte le componenti nulle contempoaneamente su un punto appartenente alla curva/superficie  

**Nota**

per la curva, la base è fatta da un solo vettore, quindi non ha senso chiedere l'indipendenza, basta che sia non nullo  

spesso si dimostra $C^\infty$, tanto con le trig e le funzioni polinomiali è sempre vero, ma non è necessario 

**Th delle funzioni implicite / th delle superfici regolari come insiemi di livello ???**

se $g: R^2 \rightarrow R$ è $C^1$  
e ricorda che chiamo $S=g^{-1}(0)$  

superfici regolari:  
se $\nabla g(p) \neq 0 \quad \forall p \in S$  
allora S è una superficie regolare  

funzioni implicite:  
se in un punto p ho $g(p)=0$ e $\frac{\partial }{\partial z}g(p) \neq 0$  
allora esiste localmente una funzione z=f(x,y) di classe $C^1$ che descrive la soluzione g(x,y,z)=0  

Il teorema delle superfici regolari come insiemi di livello è proprio una conseguenza diretta del teorema delle funzioni implicite  
perchè quest'ultimo mi dice che la superficie è localmente assimilabile ad un piano, che è la definizione di superficie regolare  


#### Definizioni di altre cose

**Definizione: superficie di livello**

è un insieme definito come: $S=\{(x,y,z)∈R^3∣g(x,y,z)=c\}$  
dove:  
$g:R^3→R$ è una funzione qualsiasi (almeno continua o derivabile)   
c è una costante  

in pratia è uguale alla condizione pratica  

**Controllare independenza lineare**

1- scrivere la jacobiana: 
ogni colonna ha le derivate di tutte le componenti rispetto ad un parametro (u,v)

2- u,v sono indipendenti se la jacobiana ha rango 2  

$J = \begin{bmatrix} 
\partial_u \sigma_1 & \partial_v \sigma 1 \\
\partial_u \sigma_2 & \partial_v \sigma 2 \\
\partial_u \sigma_3 & \partial_v \sigma 3 \\
\end{bmatrix}$


**Curva regolare a tratti**

se $I$ (il dominio di r) può essere scritto come unione finita di sottointervalli, all’interno dei quali r è regolare  
quindi i punti dove non è regolare sono gli estremi dei sottointervalli  

cioè è regolare tranne in un numero finito di punti  

**Aperto regolare nel piano**

Un aperto $Ω⊂R^2$ si dice regolare (o dominio regolare a tratti) se: 
il suo bordo è una curva regolare a tratti   
(cioè il suo bordo ∂Ω è un’unione finita di curve semplici di classe $C^1$, che si incontrano solo agli estremi)    
in più ∂Ω può essere orientato positivamente (in senso antiorario) in modo che la regione resti sempre alla sinistra della curva  

in pratica una versione in 2D per la def di curva regolare a tratti  

#### Esempi dimostrazione: curva e superficie regolare

**DImostro che è una curva regolare**

1- arco di cerchio  
2- segmento  

le parametrizzazioni sono:  
segmento: $\sigma(t)=(1-t)A+tB=((1-t)x_A+tx_B, \; \; (1-t)y_A+ty_B), \quad t \in [0,1]$  
cerchio: $\sigma(t)=(R \cos t, R \sin t), \quad t \in [0, 2 \pi]$   

controllo la classe di regolarità:  
segmento: $C^\infty$  
perchè posso derivare quante volte voglio, avrò 1 la prima volta e poi sempre 0, e f=0 è una funzione continua, quindi tutte le derivate sono continue  
tra l'altro, le funzioni polinomiali sono sempre derivabili   
cerchio: sin e cos sono $C^\infty$, per quanto le derivo non si annullano mai, e sono funzioni continue  

la condizione per una curva regolare è che abbia una paramtrizzazione $c^\infty$  
quindi sono regolari  

**Esempio: dimostra che cilindro è una superficie regolare**

cilindro: $x^2+y^2=1$  

1- parametrizzazione  
$\sigma(u,v)=(\cos u, \sin u, v)$  
$u \in (0, \pi), v \in R$  
quella solita, guarda la guida alla parametrizzazione   

2- Regolarità della mappa  
$\cos⁡ u$ e $\sin u, v$ sono funzioni $C^\infty$ (lisce, hanno tutte le derivate esistenti e continue)  
quindi $\sigma(u,v)$ è puna mappa $C^\infty$  

3- vettori tangenti  
calcolo le derivate parziali  
$\sigma_u = (-\sin u, \cos u, 0)$  
$\sigma_v = (0,0,1)$  
sono linearmente indipendenti (perchè non sono multipli)  
sennò si prova con: $\sigma_u \times \sigma_v \neq 0$   

**Perchè T è regolare ???**

T è un triangolo di vertici A, B, C  

Il bordo ∂T è costituito da tre segmenti ( AB, BC, CA )  
ciascuno è il grafico di una parametrizzazione lineare $C^∞$  
i segmenti si incontrano solo ai vertici  
Dunque ∂T è una curva a tratti $C^1$ semplice e chiusa  
quindi T è un aperto regolare.


### Lunghezza curva e area superficie regolare

**Recap**

lunghezza: $l(r) = \int_I dl$  
superficie: $A(\sigma)= \int_D \, d\sigma$   

differenziali:  
$dl = \sqrt{|\nabla r|} \; dt$ (quello dell'integrale di linea di prima specie)  
$d\sigma =\sqrt{|\nabla u|^2 +1} \; dA$  
dove $dA = dx \; dy$, o comunque le due var che appaiono in u  

$\nabla r = ||r'(t)||  =  \sqrt{(\partial_t x(t))^2+(\partial_t y(t))^2}$  
$|\nabla u|^2 = (\partial_x u)^2+(\partial_y u)^2 =  ||r_v \times r_v||^2$  

**Def: lunghezza arco di curva**

separo il dominio in unione di tanti intervalli infinitesimi  
in ogni intervallo, approssimo curva come segmento che unisce gli estremi  
lunghezza della curva è sup della sommatoria delle lunghezze dei segmenti  

**Teorema: lunghezza di una curva regolare**

$r:[a,b]\rightarrow R^n$  

$l(r) = \int _a^b ||r'(t)|| \; dt$  

se è regolare a tratti:  
la formula si applica su ciascun intervallo di regolarità e si sommano i contributi  



**Lunghezza curva vs lunghezza sostegno**

possono essere diverse  
tipo una curva che perccore due volte il cerchio ha lunghezza doppia del sostegno  

**Curva rettificabile**

se la sua lunghezza esiste ed è finita  
unione di curve rettificabili è rettificabile  

### Parametri ???

Curva originale:  
$r: [a,b] \rightarrow R^n$  

Cambio di parametrizzazione:  
$\phi:[c,d]\rightarrow [a,b]$  
$u \mapsto \phi(u) = t$  
funzione derivabile e biettiva  

$r_1:[c,d]\rightarrow R^n$  
$u \mapsto r(\phi(u))$  

**Spiegazione**

$\Phi$: è una funzione che prende in input un numero $u$ e dà in output un altro numero $\Phi(u)$, che chiamerò t    
esempio: $\phi(u)=2u+1 =t$  
sempre a cambiare il nome che dò ad ogni punto, essenzialmente sposto il sistema di assi  
si chiama cambio di parametrizzazione perchè invece di indicare i punti dicendone il loro valore u, li indico dicendone il loro valore t  

se $\phi'(u)>0$: le parametrizzazioni si chiamano equivalenti  
sennò, è presente un cambio di orientazione (cioè il disegno è uguale, ma la curva è percorsa nel verso opposto, scambia fine e inizio)

**Proprietà**

- $r, r_1$ hanno lo stesso sostegno  
perchè cambia il modo in cui la percorro, non quale strada percorro  
quindi lunghezza della curva resta uguale, integrale di prima specie è invariante  

- le derivate sono diverse: $r_1'(u)=r'(u)\phi'(u)$  

- se r è regolare e $\Phi' \neq 0$, allora $r_1$ è regolare  
perchè def di regolare è che $r'$ non sia nullo in nessun punto della curva  

**Th: kunghezza della curva non cambia con cambio di parametrizzazione**

proof:  

prendo   
$r: [a,b]\rightarrow R^n$  

cambio di parametrizzazione:  
$\phi: [c,d] \rightarrow[a,b]$  
$r_1(u)=r(\phi(u))$

calcolo quest'altra lunghezza:  
$l(r_1)=\int _c^d||r_1'(u)|| \; du = \int _c^d ||r'(\phi(u))\phi'(u)|| \; du$  

assumo $\phi'<0$, l'altro caso è uguale  
$\phi'$ è uno scalare, non un vettore, quindi posso separare  
faccio la sostituzione $\phi(u)=t$  
$= - \int _c^d ||r'(\phi(u))|| \;  \phi'(u) \; du = -\int _{\phi(c)}^{\phi(d)} ||r'(t)|| \; dt = -\int _{b}^{a} ||r'(t)|| \; dt = \int _{a}^{b} ||r'(t)|| \; dt = l(r)$    

$\phi(c)=b, \phi(d)=a$ perchè $\phi'<0$, quindi la funzione è decrescente  
nell'altro caso sono invertiti, $\phi(c)=a$  

### Ascissa curvilinea

**Definizione**

ascissa curvilinea:  
$s(t)=\int_{t_0}^t ||r'(u)|| \; du$  
con $t, t_0 \in [a,b]$  

è l'integrale del differenziale che appare negli integrali curvilinei di prima specie  
rappresenta la lunghezza del tratto di curva da $t_0$ a $t$  

**Intuitivamente**

posso parametrizzare la curva tramite s o t   
cioè per individuare un punto, dire quanto spazio o quanto tempo ci vuole a raggiungerlo  

quindi s è una misura della posizione di un punto lungo una curva   
definita come la lunghezza dell'arco di curva compreso tra quel punto e un punto fisso, detto origine  
in pratica, come sistema di coord definisco un asse esattamente sovrapposto alla curva, su cui piazzo un'origine a caso  

$s(t)$: lunghezza percorsa fino a t (coord nel sistema di coord dell'ascissa curvilinea)  
$r(s)$: curva parametrizzata tramite lunghezza d'arco

**Proprietà**

$||r'(s)||=1$  

se parametrizzo la curva in funzione di s, cioè dell’ascissa curvilinea, il vettore tangente alla curva in ogni punto sarà un versore   
cioè avrà modulo 1   

intuitivamente:  
parametrizzare in funzione di s mi setta la velocità, cioè so quanto veloce percorro la curva, percorro 1 unità di lunghezza per 1 unità di parametro  
cioè: $r'(s) =1$   

per scrivere $r(s)$:  
1- calcola $s(t)$ (è un integrale da risolvere)  
2- trova l'invers $t(s)$   
3- pluggala in $r(t)$, così ho $r(t(s))$  


**Dimostrazione proprietà**

$r'(s)=\frac{dr}{ds}=\frac{dr}{dt} \frac{dt}{ds} = \frac{r'(t)}{s'(t)}$   

derivo la definizione di ascissa curvilinea, così ottengo:    
$s'(t)=||r'(t)||$  

quindi nella formula di prima:  
$r'(s) =1$  

### Integrali curvilinei

**Tipi**

si differenziano a seconda di:  
cosa è il dominio: curva, superficie, volume  
cosa è la funzione integranda: scalare, vettoriale 

tutti i metodi sotto valgono solo per curve/superfici regolari (o almeno regolari a tratti)   

**Tipi**

prima specie: sommare una quantità scalare (f) lungo una curva NON orientata  
seconda specie: sommare un campo vettoriale (F) lungo una curva orientata  

il dominio è sempre una curva parametrizzata da $\gamma(t)$  
ma cambia il differenziale e l'integranda (F è un campo vettoriale, f una funzione scalare)  

**Tipi di integrali di linea**

- integrale di linea di PRIMA SPECIE:   
$\int_\gamma f \; ds$  
$ds = ||\gamma'(t)|| \; dt$  

- integrale di linea di SECONDA SPECIE:  
$\int_\gamma F \cdot d\vec{r} = \int _r P \, dx + Q \, dy$    
$F=(P, Q)$  
$d\vec{r} = \gamma'(t) \, dt = (dx, dy) = (x' \, dt, y' \, dt)$    
usa la chain rule pr il differenziale  

**Integrali di liea rispetto a x o y**
 
$\int_C f \; dx$  
$\int_C f \; dy$  

sono rispettivamente l'integrale di seconda specie con campo $F=(f,0)$ o $F=(0,f)$  

**Proprietà**

sono invarianti sotto parametrizzazioni equivalenti  
quello di prima specie è invariante sotto cambi di orientazione, quello di seconda invece inverte il segno (perchè la curva è orientata)  

### Tipi di integrali

**Note**

è uguale:  
$f$: associa un valore ad ogni punto dello spazio, perchè alcuni sono più importanti e pesano di più  

cambiano:  
il differenziale e il dominio (cambia proprio la dimensionalità del dominio)    

**Integrale singolo vs di linea**

singolo: $\int_D f(x) \; dx$   
linea: $\int_C f(x,y) \; ds$ (può essere di prima o seconda specie) 

dominio:  
$D \subset R$: è un intervallo dell'asse x  
$C$: è una curva nello spazio (lo spazio può avere qualunque dimensione, 2 o 3 o più)  

differenziale:  
$dx$ va bene così  
$ds$ dipende se prima o seconda specie  

cosa calcola:  
singolo: l'area compresa tra l'asse x e la curva $y=f(x)$ (la lunghezza dell'intervallo, pesata per il valore di f)  
linea: la lunghezza della curva, pesata per il valore di f   

**Integrale doppio vs di superficie**

integrale doppio: $\int_D f(x,y) \; dA$   
integrale di superficie: $\int_\Sigma f(x,y) \; d\sigma$  

dominio:  
$ D \subset R^2$: è una superficie piatta sul piano xy (tipo una macchia)    
$\Sigma$: è una superficie nello spazio 3D, ed è generalmente parametrizzata con:  
$\Phi: U \subset R^2 \rightarrow R^3$ (cioè con 2 var decsrivo qualcosa in 3D)

differnziale:  
$dA = dx \, dy$  
$d\sigma = |\Phi_u \times \Phi_v| du \, dv = \sqrt{1+|\nabla u|^2} dA$  

cosa calcolano:  
doppio: il VOLUME tra il piano xy e la superficie $z=f(x,y)$, nella zona del dominio D, cioè l'estensione di D pesata da f   
superficie: l'ESTENSIONE della superficie z, nella zona del dominio D (nei punti di z dove sotto c'è la macchia del dominio)  

**Integrale triplo**

integrale triplo = integrale di volume:     
$\int_D f(x,y,z) \; dV$   

$D \subset R^3$: un volume nello spazio
