### Palle / intorni

$B(P,r) = \{x \in R^n : |x-P|<r|\}$  

$B((x_0,y_0), r) = \{(x,y) \in R^2: \; (x-x_0)^2+(y-y_0)^2 \le r\}$  

palla centrata in $P=(x_0, y_0)$ , di raggio r  

$<r$: aperta  
$\le r$: chiusa (la B si scrive con una doppia stanghetta verticale)  

**Intuizione**

è un intorno del punto P, di raggio r  
in 1d: un pezzo di segmento  
in 2d: un disco  
in 3d: una sfera  

### Insieme

$A \subset R^n$ è l'insieme  
dominio è $R^n$ perchè usiamo funzioni multivariabile 

#### Aperto

$\forall y \in A \quad \exists r > 0 : \quad B(y, r) \subset A$  

**Intuizione**

in qualunque punto, posso fare una palla centrata in quel punto abbastanza piccola da essere contenuta nell'insieme  
in pratica, ciò vuol dire che non deve avere il bordo, sennò se prendo un punto del bordo qualunque palla andrà immediatamente fuori  

**Proprietà**

A chiuso -> $A^c$ aperto (e viceversa)  

se f(x,y) continua -> gr(x,y,f(x,y)) chiuso  
(cioè il grafico di una funzione continua a 3 variabili è sempre chiuso)  

**Problematici aperto/chiuso**

nota che R è sia chiuso che aperto (quindi $\infty$ può essere sia un'estremità chiusa che aperta, è un jolly)  
(0, 1]: nè chiuso nè aperto  
$(1, \infty)$: aperto  
$[1, \infty)$: chiuso  

**Per dimostrare**

NON è aperto: se contiene un punto che non ha un intorno tutto appartenente all'insieme  
NON è chiuso: se non ha un suo punto di accumulazione  

eg: (0,1]  
1 è nell'insieme, ma non ha un intorno tutto nell'insieme: non è aperto  
0 è un punto di accumulazione, ma non è nell'insieme: non è chiuso  

##### Aperto in A

$S \subseteq A$ è aperto in A se:  
$\exists U : \quad U \, aperto \quad \land \quad S=A \cap U$  


**Intuizione**

cioè prendo U, un insieme APERTO    
la sua parte 'aperta in A' è la parte che si trova dentro A  
vuol dire semplicemente la parte dentro A di un insieme aperto U    
quindi è solo una parte di U, l'originale, e se considero solo questa parte potrebbe non essere aperta nel senso normale    

si chiama 'aperto relativo' (ad un sottoinsieme)

**Esempi**

A=[0,1]  
S=[0,1)   

posso scegliere $U=(1,-1)$  
S è l'aperto di U in A perchè: $S = A \cap U$   
ma S non è aperto in R    



#### Chiuso

**Varie definizioni**

1- ogni successione dell'insieme converge in un punto appartenente all'insieme  
2- la sua aderenza è uguale a sè stesso  
3- contiene tutti i suoi punti di accumulazione   

**Proprietà**

A, B chiusi (o aperti) -> $A \cap B, \quad A \cup B, \quad A \times B$ chiusi (o aperti)  
ma per intersezione aperti e unione chiusi, la mantiene solo se gli insiemi sono in numero finito  

chiuso e connesso non hanno niente a che fare, posso presentarsi in qualunque combinazione  


**Esempio**

$[0,+∞)×[0,+∞) \times R$  
 è chiuso perchè il prodotto cartesiano di insiemi chiusi è chiuso   
e $[0,+∞)$ è chiuso perchè il suo complementare $(-\infty, 0)$ è aperto  
intuitivamente, da $[0, \infty)$ una successione non può uscire (tendere a un punto non nell'insieme)  


#### Limitato

$\exists R >0 : \quad A \subset B(O, R)$  

se è compreso in una palla di raggio finito centrata nell'origine   


**Definizioni alternative: per dimostrazioni**

per dimostrare che NON è limitato:  
ogni sequenza fatta da punti di A deve convergere, non può divergere  
nota che può anche convrgere fuori dall'insieme, tipo sul bordo, perchè può essere aperto  
converge nell'insieme solo se è chiuso, ma a quel punto chiuso+limitato = comapatto  

usata per dimostare che è limitato:      
ogni vettore in A ha norma finita  

**Dimostra formalmente che un insieme è limitato**

$A = \{(x,y,z): x^2+y^2+1< z <10\}$  

intuitivamente: perchè nessuna delle var x,y,z può essere $\infty$, o g(x,y,z) non è rispettata  

formalmente:  
esprimo la norma di un generico vettore, $(x,y,z) \in A$      
$|(x,y,z)|^2 = x^2+y^2+z^2-1+1 < z+z^2-1 < 109$    
quindi: $A \subset B(O, \sqrt{109})$  
quindi A è limitato  


**Dimostra formalmente che un insieme è NON limitato**  

$A = \{(x,y) \in R^2: \quad x^2y^2<1\} = \{(x,y) \in R^2: \quad |xy|<1\}$

intuitivamente: |xy|<1 è parte interna di un'iperbole, disegnala e vedi che può andare ad infinito, gli assi sono asintoti  

formalmente:  
prendo $(\frac{1}{2k}, k) \in A$  
per ogni k rispetta g(x,y), quindi tutta la successione è in A, ma la sua norma tende a infinito:  
$|(x_k,y_k)| = |\frac{1}{4k^2}+k^2| \rightarrow \infty$  
quindi A contiene un vettore di norma infinita, è illimitato  

come scegliere la sequenza:  
prova a disegnarla, è un'iperbole  
quindi il punto problematico, che si allontana all'infinito, è quando uno dei due è x, l'altro 1/x con x tendente a infinito  

#### Compatto

$\forall x_k \subset A \quad \exists x \in A, \; \exists \alpha :N \rightarrow N \; | \quad x_{\alpha(k)}\rightarrow x \in A$  

Every sequence in K has a convergent subsequence whose limit is also in K (stronger than closed, force the subsequence to converge)  

**Proprietà: teoremi**

Heine-Borel:  
compatto = chiuso + limitato  

th di Weierstrass: se A compatto e f continua, f ha max e min su A   
utile per trovare l'immagine di una funzione  
se so che è continua e ha max e min, trovo quali sono e l'immagine sono tuti i numeri compresi  

**Proprietà**

A,B compatti -> $A \cap B, \quad A \cup B, \quad A \times B$ compatti  
f continua, A compatto -> f(A) compatto  

### Insieme: connessione

#### Rapporti

**Rapporti**

semplicemente connesso -> connesso per archi -> connesso    

questo vale se A è chiuso  
ma se A è aperto, connesso per archi = connesso  

**Esempio: connesso, ma non connesso per archi**

'topologist's sine curve'   
$S = \{(x, \sin \frac{1}{x}): x \in (0,1)\} \cup \{(0,0)\}$  
chiamo A la prima parte, B la seconda  
A, B sono connessi  
quindi la chiusura di A è connessa  
la chiusura di A e B sono non disgiunti, quindi la loro unione è connessa  

problema: qui ho A, non la chiusura, va bene lo stesso???

#### Connesso

**Definizione**

NON si può scrivere come unione di insiemi disgiunti non vuoti  

**Proprietà**

queste valgono anche per connesso per archi:  
A connesso, f continua -> f(A) connesso  
A,B connessi e non disgiunti -> $A \cup$ connessi  

A connesso -> la chiusura di A connesso  
A,B connessi -> A x B connesso

#### Connesso per archi (e convesso)

$\forall x,y \in A \quad \exists \gamma:[0,1]\rightarrow A \; | \quad \gamma \; continua \land \gamma(0)=x \land \gamma(1)=y$



**Intuizione**

comunque scelgo due punti appartenenti all'insieme, posso unirli con un cammino (che curva come voglio) che rimane tutto dentro l'insieme (non esce mai fuori)  

ok: una sfera  
non ok: due sfere separate  

**Proprietà**

f continua, C connesso per archi -> f(C) connesso per archi  
proof: la curva $f(\gamma)$ ha tutte le proprità richieste, cioè mappa i punti estremi ed è continua  

C coneesso, per archi, $C \subset R$ -> C è un intervallo  

f continua, C connesso per archi   
se esistono max e min di f in C, allora f(C)=[min f, max f]  
se esiste il min di  f in C, ma non il max, allora f(C)=[min f, sup f]  
in pratica per trovare l'immagine di qualcosa continuo, devo trovare gli estremi, se ci sono metto max e min, sennò sup e inf  

#### Semplicemente connesso

**Definizione**

ogni curva chiusa in A può essere contratta in un punto senza uscire da A   

oppure, dal teorema di Poincaré:    
In un insieme semplicemente connesso, ogni 1-forma chiusa è anche esatta

**Proprietà**

semplicemente connesso -> conesso per archi 

**Esempi**

piano:  
sempl connesso: disco aperto, semipiano    
NON sempl connesso: disco (o piano) con un punto (o più) mancante nel suo mezzo, con un buco     
il buco deve essere in mezzo all'insieme, se è tipo un seimipiano invece non daà problemi  

3D:  
sempl conn: sfera aperta, spazion 3D cun un punto amancante  
NON: spazio 3D con un'asse mancante, spazio con un tor/tunnel mancante  

cioè non lo è se ha:  
2D: un buco  
3D: un tunnel, gli manca una retta intera    

**Perchè no tunnel o linea intera**

perchè se prendo una curva attorno a quell'asse/tunnel (tipo anello attorno al dito)  
quella curva non la posso sfilare  
quindi quando la faccio convergere ad un punto, deve passare nella linea che non è compresa nell'insieme  
se fosse una linea finita invece potrei sfilarla, non avrei problemi  

#### Convesso

$\forall x,y \in A : \quad [x,y] =\{x+t(y-x), \; t\in[0,1]\}$  

**Conesso per archi vs convesso**

comunque scelgo due punti appartenenti all'insieme, posso unirli con un cammino che rimane dentro l'insieme, che è:  
connesso per archi: una curva fatta come mi pare  
convesso: un segmento, quindi in linea d'aria  

convesso -> connesso  
ma non necessariamente il contrario  
due punti potrebbero essere uniti da una linea curva, ma non da una dritta  

### Parti di un insieme

x è un punto [nome] di A se:  

nome | formula | nome della parte  
---|---|---
di frontiera | $\forall r > 0 \quad B(x,r) \cap A \neq \empty \land B(x,r) \cap A^c \neq \empty$ | Fr A   
di aderenza / chiusura | $\forall r >0 \quad B(x,r) \cap A \neq \empty$ | $\overline{A}$    
interno  | $\exist r>0 : \quad B(x,r) \subset A$   | Int A
di accumulazione | $\forall r >0 \; B(p,r) -\{p\} \neq \empty$  | D(A)

**Come scrivere una parte**

$Fr \, A = \{x \in R^n | x \, di \, frontiera\}$  

**Punto di frontiera**

se disegno una palla in un punto di frontiera, deve avere sia una parte dentro l'insieme che fuori (cioè dentro il suo complementare)  
cioè deve intersecare sia l'insieme che il suo complementare  

**Punto di chiusura**

un punto in cui ogni suo intorno (palla) interseca l'insieme  
ma il punto stesso può non appartenere all'insieme   
tipo, in un insieme aperto, la chiusura è il corrispondente insieme chiuso (cioè considero anche i bordi)   
cioè l'insieme chiuso appena più grande, giusto abbastanza per chiudere le estremità  

la chiusura di (0,1) è [0,1]  

attenta ai simboli:  
$\overline{A}$: chiusura di A  
$A^c$: complementare di A  

**Punto interno**

se posso disegnare una palla centrata lì, tutta dentro l'insieme  
quindi l'insieme meno il bordo  
in un insieme aperto, tutti i punti sono interni  


**Punto do accumulazione**

$p$ punto di accumulazione se:  
cioè posso fare una palla centrata in p piccola quanto voglio, c'è almeno un altro punto dentro  

quindi esiste una successione in A che tende a p, ma non diventa mai p

D(A): derivato di A, insieme di tutti i punti di accumulazione di A  

**Esempi**

$A = \{(x,y,z): \quad y^2 \ge z^2+x^2, \quad y>0\}$  

$Int(A) = \{(x,y,z): \quad y^2 > z^2+x^2, \quad y>0\}$  
$\overline{A}=\{(x,y,z): \quad y^2 \ge z^2+x^2, \quad y\ge 0\}$  
$Fr(A)= \{(x,y,z): \quad y^2 = z^2+x^2, \quad y>0\}$  

cambia il segno di g:  
Int: >  
chiusura: $\ge$  
Fr: =  


### Grafico

$gr \, f = \{(x,y, f(x,y)) \in R^3 \; | \; (x,y) \in A\}$  

**Dimensione**

ho una funzione $f: A \rightarrow R$, con quante variabili voglio  
il grafico di f ha dimensione 1 in più del dominio, perchè devo aggiungere una dimensione con il valore di f  

**Leggere un insieme come grafico**

ho un insieme A  
posso provare a rielaborarlo per trasformarlo nel grafico di una f  

vedi se riesco a isolare z, in modo da vederlo come grafico di qualcosa  
$A=\{(x,y,z)∈R^3: \quad z=\sqrt{4x^2+y^2}\}$  
è il grafico di $f(x,y)=\sqrt{4x^2+y^2}$  

proprietà che quindi posso usare:    
se f è continua, A (cioè $f(R)$) è chiuso e connesso per archi   
ricorda che f è continua se composizione di funzioni continue
