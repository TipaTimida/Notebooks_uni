### Th del Dini

**Teorema del Dini (delle funzioni implicite)**

cosa ho:  
$F: \, A \subseteq R^2 \, \rightarrow \, R$  

ipotesi:  
1- $F(x_0,y_0)=0$  
2- $\frac{\partial F}{\partial y} |_{(x_0,y_0)} \neq 0$   
3- $F \in C^1(A)$  

th:  
esistono un intorno U e V di $x_0$ e $y_0$ e un'unica funzione $f: U \rightarrow V$ tale che  
$F(x, f(x))=0 \quad \forall x \in U$ e $f(x_0)=y_0$  

cioè una funzione che sostituisce y e che è traslata della stessa quantità  

**Note**

1- vuol dire scrivi F in forma implicita  
2- derivata parziale rispetto a y, calcolata nel punto richiesto   
3- A può essere quanto piccolo voglio, è un intorno di $(x_0, y_0)???  


**Nomenclatura**

forma esplicita: $y = f(x)$   
forma implicita: $F(x,f(x))=0$  

f è definita implicitamente da F  

#### Corollari

**Corollari - LEGGILI**

f è della stessa classe di regolarità di F, cioè è $C_1$    
il suo differenziale è: $f' = -\frac{\partial_x F(x, f(x))}{\partial _y F(x, f(x))}$, cioè come al solito per F, ma pluggandoci dentro f   
la ottengo derivando l'indentità $F(x, f(x))=0$  

**Corollario 1**

per calcolare la derivata:  
$f'(x) = -\frac{F_x(x, f(x))}{F_y(x, f(x))}$  

$F_x, F_y$ sono le derivate parziali  

quindi f è della stessa classe di regolarità di F  

**Proof 1**

$F(x,f(x)) = 0 $ perchè è una forma implicita  

derivo con la chain rule:  
$\frac{\partial F}{\partial x}+ \frac{\partial F}{\partial y}f'(x)= 0$  

ricorda $f=y$ 
rearrange, trovo ciò che voglio


**Corollario 2**

$f''(x)=-\frac{1}{F_y}(F_xx+2F_{xy}(f')^2)$

### Più dimensioni

**Teorema del Dini in 3 dimensioni**

ipotesi:  
$F: A \subset R^3 \rightarrow R$   
A aperto  
$F(x_0,y_0, z_0)=0$  
$F \in C^1(A)$  
$\frac{\partial F}{\partial z}(x_0,y_0, z_0) \neq 0$  

th: esiste un intorno U di $(x_0, y_0)$ e un'unica funzione $y=f(x)$ con $g: U \rightarrow R$ tale che:  
$F(x, y, g(x,y))=0 \quad \forall x \in I , z_0 = g(x_0, y_0)$  

come prima, ma tutto scritto in 3 var  

**Corollario 1**

$\frac{\partial g}{\partial x} = -\frac{F_x(x, y, g(x,y))}{F_z(x, y, g(x,y))}$  

$\frac{\partial g}{\partial y} = -\frac{F_y(x, y, g(x,y))}{F_z(x, y, g(x,y))}$  


**Th delle funzioni implicite**

generealizzazione del th del Dini a qualunque numero di dimensioni  
vale per $F: R^{n+m}\rightarrow R^m$  
mentre quello di Dini era solo per $R^3$ o $R^2$  

condizioni:  
$F(p)=0$  
la jacobiana in p è invertibile  

th: esiste f tale che: $y=f(x), \quad f \in C^1, \quad F(x,f(x))=0$ localmente  

### Studiare funzione implicita

**Studiare eq in due variabili**

ho $F(x,y)$  
studia monotonia e convessità intorno a $x_0$  

come:  

1. verifica le ipotesi del teorema del dini, così posso scrivere $F(x,y)=F(x,y(x))=0$  
cioè controlla in quale/i $y_0$ vale $F(x_0, y_0)=0$ e la derivata è non-nulla  
di seguito, studia questi punti  

2.  $\frac{\partial F}{\partial x} = 0$  
deriva entrambi i lati rispetto a x (ricorda la chain rule), $\frac{dF}{dx} = \frac{\partial F}{\partial x} + \frac{\partial F}{\partial y} \, y'$  
poi inverti per trovare $y'(x,y)$  
e plugga $(x_0,y_0)$ per trovare i valori  

3. derivo di nuovo per x da entrambi i lati  
ottengo $y''(x,y,y')$  

**Esempio**

$F(x,y)=x^2+y^2-1=0$  
studia $x_0=0$  

1- verifica ipotesi teorema di Dini  
trovo che per avere F=0 serve il punto (0,1)  

2- calcolo di y'(x)  
deriva per x entrambi i lati, usa chain rule  
$\frac{dF}{dx}=0$  
$2x + 2y \, y' =0$  
$y' = -\frac{x}{y}$  
così ho y'(x,y), posso calcolare y' nel punto (0,1)  
cioè, trova $y'$ con la formula $y' = -\frac{F_x}{F_y}$  

3- calcolo di y''(x)  
derivo di nuovo destra e sinistra rispetto a x  
$2+2 (y')^2+2y \, y''=0$  
così ho $y''(x,y,y')$
