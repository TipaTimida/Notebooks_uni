### Cosa

$\int _D \;  f   \; dA$    

f: il valore della funzione in ogni punto, ciò che devo integrare  
quanto vale ogni punto dello spazio, alcuni sono più importanti e valgono di più  

D: dominio, la regione di spazio che considero  

dA: il differenziale  

**Visualizzarlo in 2D**

f(x,y) è una superficie z(x,y), quindi che fluttua sopra il piano xy    
D è una macchia sul piano xy, che mi dice quale parte devo considerare   

quindi calcolo il volume compreso tra il piano xy e la superficie z, ma solo nella zona colorata da D  

#### Dominio

**Tipi**

ci sono 3 tipi:
  
1. insieme prodotto (definito con tutti estremi)    
$D=[a,b] \times [c,d]$ o anche $D=[r_1, r_2] \times [\theta_1, \theta_2]$   
facile, basta metterli come estremi di ogni integrale multiplo (attenta all'ordine, dopo aver integrato per x metti gli estremi di x)   
$\int_a^b \int_c^d f(x,y) \, dy \, dx$

2. y-semplice: (con $y(x)$ e delimitato da linee verticali)   
$D=\{(x,y): \; a \le x \le b , \; g_1(x)\le y \le g_2(x)\}$  
$dA= dy \; dx$  

3. x-semplice (con $x(y)$ e delimitato da linee orizzontali)    
$D=\{(x,y): \; h_1(y) \le x \le h_2(y) , \; c \le y \le d\}$  
$dA= dx \; dy$  

nel differenziale, integra prima rispetto alla variabile con delle funzioni negli estremi  
così alla fine integro rispetto a quella con i numeri, e tutto diventa un numero  
per ricordare: x-semplice vuol dire che gli estremi della x sono facili, sono solo coord delle rette  

**Disegnare il dominio in 2D**

nota che puoi ignorare f, non importa se è complicata  
mi importa solo che ordine usa dA e gli estremi di integrazione  

1. disegna le funzioni dell'integrale più interno  
se è dy, allora sono $y=g_1(x), y=g_2(x)$  
se è dx, sono $x=h_1(x), x=h_2(x)$  
ovvio, perchè devo pluggare dentro una funzione della var per cui ancora non ho integrato  
colora l'area tra di loro  

2. disegna le righe dell'integrale più esterno  
se è dx, allora sono $x=a$ e $x=b$, cioè verticali  
se è dy, allora sono $y=c$ e $y=d$, cioè orizzontali    
servono da limitatori dell'area trovata prima  

interpretazione:  
al punto 1, dx significa: se disegno una riga orizzontale, deve partire dalla funzione $x=h_1(x)$ e finire alla funzione $x=h_2(x)$  

nota:  
gli estremi corrispondenti a dx si leggono $x=...$, quelli corrispondeti a dy invece $y=...$


**Suggerimento: triangolo**

ho un triangolo con lato orizzontale, parallelo all'asse x  
2 modi per integrare:  

1. scrivi le funzioni y=g(x) dei suoi lati  
e devo separare in due triangoli, quindi devo fare 2 integrali  
cioè definisci $D=D_1 \cup D_2$, la separazione sta a livello del vertice non sull'asse x  

2. scrivi le funzioni x=g(y) dei suoi lati  
basta fare y(x) e poi invertire  
così il lato orizzontale diventa una boundary per y (linea orizzontale), non una funzione     
e ho solo 2 funzioni, devo fare un solo integrale

##### Note

**Nomenclatura insieme**

semplice: either x-semplice o  y-semplice  
formale: se esistono due curve, g_1 e g_2, tali che D si può scrivere come mostrato sopra  

regolare: unione di un numero finito di insiemi semplici  

condizioni di integrabilità: f continua e D regolare

**Proof coordinate polari**

prendi un arco $d\theta$ e un raggio r, e prolungalo di $dr$   
la zona delimitata è una specie di rettangolo, di area dA, e lati $dr$ quelli radiali, $r d\theta$  quelli tangenziali (dr è così piccolo che assumo che questi lati tangenziali siano uguali)  
area di un rettangolo è moltiplicare i lati, quindi: $dA = r \; dr \; d\theta$  

#### Differenziali

$dA = dx \; dy$   

**Differenziare per una sola var**

ho una funzione con due var  
ma la integro rispetto ad una var sola, faccio un integrale singolo  
allora la costante è una funzione dell'altra variabile  

$\int 2xy \, dx = x^2y + c(y)$   

**In che ordine**

quando sono più di uno (tipo $dx \; dy$), si fa prima quello più interno (immagina di mettere delle parentesi)  
ovviamente, se faccio prima dx, gli estremi di integrazione della x devono essere più interni  

### Metodi per calcolare

#### Inverti differenziali

**Cambia ordine**

si può fare?  
th di Fubini: sì, se f è integrabile    
ricorda che continua -> integrabile  

come si fa: dipende dalla forma di D   

- D è un insieme prodotto  
semplicemente scambia gli estremi, quelli interni diventano esterni  

- D NON è un insieme prodotto  
devo calcolare i nuovi estremi  

**Steps: D NON è insieme prodotto**

come trovare i nuovi estremi di integrazione  
inizia dai nuovi estremi interni  

nuovi estremi interni:  
se è dx, significa: se disegno una riga orizzontale, deve partire dalla funzione $x=h_1(x)$ e finire alla funzione $x=h_2(x)$   
cioè cerca due funzioni verticali che delimitano l'area (i lati verticali dell'area)  
probabilmente sono le stesse funzioni che avevo prima nell'integrale più interno, ma manipolate algebricamente per trovare l'inversa, tipo: $y=x^3 \rightarrow x=\sqrt[3]{x}$   
e spesso cambia l'ordine, quella che era sopra, dopo averla invertita, diventa sotto  

estremi per l'integrale esterno:  
se è dy, vuol dire: trova linee y=a, cioè orizzontali, che delimitano l'area tra le curve di prima nei punti dove voglio io  
spesso sono i punti di intersezione delle due funzioni (nello stesso punto di prima, perchè le funzioni non sono cambiate), ma se prima avevo preso le coord x ora mi servono le coord y  

**Teoremi**

th Tonelli:  
f(x,y) è non negativa e misurabile -> posso scambiare l'ordine dei differenziali dx e dy    

th Fubini:  
f integrabile -> posso scambiare l'ordine dei differenziali  
più generale di quello di Tonelli, f può avere anche valori negativi  
tanto se l'integrale si può fare, allora si può fare e basta, non può dipendere dall'ordine  

#### Cambio di variabili

**Jacobiana**

$J =\frac{\partial(x,y)}{\partial(u,v)} =det \begin{bmatrix} \partial_u x & \partial_v x \\ \partial _u y & \partial_v y \end{bmatrix}$  

prima ho l'integrale in variabili (x,y)  
voglio averlo in variabili (u,v), con $x=g(u,v), y=h(u,v)$  
$\int_D f(x,y) \, dA = \int_S f(g(u,v), h(u,v)) \;  J \, dA $
