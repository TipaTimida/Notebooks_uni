### Definizioni

**Derivata direzionale**

$ \partial_v f = \frac{\partial f}{\partial v} = \lim_{t \rightarrow 0} \frac{f(p_0+tv)-f(p_0)}{t}$  
la derivata direzionale esiste se questo limite tende ad un valore finito  

oppure:  
$\lim_{h \rightarrow 0} \frac{f(x_1, x_i+h, x_n)-f(x_1, x_n)}{h}$    

**Derivata direzionale vs parziale**

derivata direzionale rispetto a v:  
come cambia f se mi sposto seguendo il vettore v  
cioè la variazione di f se mi sposto in una certa direzione sul dominio   

derivata parziale:   
un caso particolare di derivata direzionale, dove il vettore che seguo è un versore (cioè fa parte della base ortogonale)  
se uso come v il versore dell'asse x, sto facendo la derivata parziale rispetto a x

**Differenziabile**

f è differenziabile se esiste L lineare tale che:  
$ \lim_{t \rightarrow 0} \frac{f(p_0+tv)-f(p_0)-L(tv)}{|tv|}=0$  

differenziabile = se esiste il differenziale    
il differenziale è L, a volte chiamato anche df  

altro modo di scriverlo, con tv -> h:  
$\lim _{h \rightarrow 0} \frac{f(\vec{x}_0+h)-f(\vec{x}_0)-L(h)}{|h|} = 0$  

**Differenziale**

L è l'applicazione lineare/mappa lineare che approssima f intorno a x  
mappa lineare: prende in input spostaento h, dice la variazione di f  
$f(x_0+h)=f(x_0)+L(h)+o(|h|) \quad$  per $h \rightarrow 0$   
$f(x)=f(x_0)+L(x-x_0)+o(x-x_0) \quad$ per $x \rightarrow x_0$  

un teorema?  
$df_x(h)=\Sigma \frac{\partial f}{\partial x_i} \, h_i$  

#### Matrici

**Gradiente**

$\nabla f = (\frac{\partial f}{\partial x_1}, ..., \frac{\partial f}{\partial x_n})$  

**Vettore gradiente**

indica la direzione dove la funzione ha la massima pendenza  

$v = \frac{\nabla f(p)}{|\nabla f(p)|}$

**Scrivere Jacobiana ed Hessiana**

hessiana:  
ogni riga ha più a sinistra al denominatore (quindi il secondo che uso effettivamente) $x_i$ con 'i' numero di quella riga  
più a destra metti la lista, quindi fin lì una riga è un gradiente  

jacobiana:  
anche qui in un ogni riga i denominatori sono spannati tutti, ogni riga è un gradiente  
e in ogni riga metti al numeratore $f_i$ con 'i' di quella riga   

recap: ogni riga è un gradiente (spanna tutte le var)  
ogni col cambia: la componente f (jacobiana), la seconda derivata (hessiana)  


**Hessiana vs jacobiana**

hessiana: per funzione scalare (codominio è R)  
mostra tutte le derivate di ordine 2, quindi tutti gli ordini con cui posso derivare    

jacobiana: per funzione vettoriale (codominio è $R^n$)  
mostra tutte le derivate di ordine 1, quindi tutte le combinazioni di componente di f e var per cui derivo   

**Matrice Hessiana**

contiene le derivate seconde  
sulla diagonale le derivate seconde non miste  
grazie al teorema di Schwarz, so che è simmetrica se $f \in C^2$  

posso scrivere:  
$d^2f(p)(x)= <x, Hf(p)x>$ ???  

**Derivate di ordine superiore**

indica, nell'ordine in cui sono state effettuate, tutte le derivate  
secondo ordine: è composta da due derivate    

attenta all'ordine:  
$\frac{\partial^2 f}{\partial xy}$: prima deriva per y, poi per x (immagina di mettere una parentesi)  
$\partial_{xy} f$: prima deriva per x, poi per y (qui la metto subito nell'ordine giusto)  


posso usarle per scrivere il differenziale di ordine superiore:  
$\frac{\partial^k}{\partial v^k}=d^kf(v)$  
esempio:  
$d^2f(p)(x,y)= f_{xx}(p)x^2+f_{xy}(p)xy + f_{yx}(p)yx+f_{yy}(p)y^2$

### Come scrivere cose

**Nomenclatura**

$L_p(v), \quad df(p)(v)$: differenziale  
$\partial_v f(p)$: derivata direzionale  
$\nabla f(p)$: gradiente    
$D_v f(p)$: jacobiana  

tutto calcolato nel punto p  

**Dire che qualcosa è differenziabile**

SOLO SE f è differenziabile in p, lungo v (esiste derivata direzionale):      
sono modi diersi di dire la stessa cosa  
$df(p)(v) = L_p(v) = D_vf(p)=\nabla f(p) \cdot v$  

se voglio posso togliere (p) da tutte quante, se non mi interessa uno specifico punto  
$df(v)=L(v) = D_vf=\nabla f \cdot v$

v must be normalized before being used (its norm must be 1)

**Derivata parziale e direzionale**

$v=(a,b,c)$  
$\partial_v f(x,y,z)  = f_x \, a + f_y \, b + f_z \, c$  
$\partial_vf(x,y,z)=(f_x,f_y,f_z) \cdot (a,b,c)= \nabla f \cdot v$

**Differenziale e gradiente e jacobiana**


$df(v)=f_x \, dx + f_y \, dy$  
posso aggiungere (p) a tutti per calcolarlo in un punto preciso  

$df(v) = \frac{\partial f}{\partial x_1} dx_1 + .. + \frac{\partial f}{\partial x_n} dx_n$  
$\nabla f = (\frac{\partial f}{\partial x_1}, ..., \frac{\partial f}{\partial x_1})$  
$\vec{v}=(dx_1, ..., dx_n)$  

**Esempio: differenziale e gradiente**

$f(x,y)=x^2+y^2$  
$p_0=(1,2)$  
$v = (v_1, v_2)$ perchè sono in $R^2$  

gradiente:  
$\nabla f(1,2)=(2,4)$  

differenziale:  
$L_{(1,2)}(v) = df(1,2)(v)=2v_1+4v_2$  

cioè, posso usarlo per approssimare con Taylor:  
$f(1+v_1, 2+v_2) \sim f(1,2) + 2v_1 +4v_2 = f(1,2) + df(1,2)(v_1, v_2)$  
in pratica ad ogni ordine metto un addendo per ogni variabile  


**Differenziale e derivata direzionale**

$df(p)(v) = df(p)(\Sigma \xi_i e_i) = \Sigma \xi_i \; df(p) e_i = \Sigma \xi_i \partial_{e_i} f(p)$  
uso la linearità del differenziale  
e dopo riconosco una derivata direzionale

### Teoremi

#### Rapporti

**Rapporto: derivate parziali e differenziabilità**

la 1 implica tutte le altre, etc   
1- P-cont: derivate parziali continue  
2- f-diff: f differenziabile  
3- D-exist: derivate direzionali esistono    
(nota che, visto che f è diff, allora P-exist->D-exist)  

**DF2 - Teorema del differenziale totale (1->2)**

se le derivate parziali sono continue in p, allora f è differenziabile in p  

anche detto: Teorema di differenziabilità delle funzioni $C^1$  

se $f:\Omega \subset R^n \rightarrow R$ è di classe $C^1$  
(cioè f e tutte le sue derivate parziali di primo grado esistono e sono contnue in $\Omega$)  
allora f è differenziabile in ogni punto di $\Omega$  
e la sua differenziale in un punto $x_0$ è data dal gradiente in $x_0$:  
$df(x_0)(h)=\nabla f(x_0)\cdot h$


**DF1 : differenziabile -> esistenza derivate direzionali (2->3)**

testo:  
se f è differenziale in $p$  
allora in p esiste la derivata direzionale rispetto ad ogni v, e vale:  
$\partial_v f(p_0) = df(p_0)(v)$  


Cioè:  
se f è differenziabile in un punto, allora in quel punto coincidono:   
$\partial_v f(p_0)$: derivata direzionale (definita via limite, è un limite)    
$df(p_0)(v)$: differenziale applicata al vettore v (è un oggetto lineare)  

come scrivere:  
differenziabile: $df(p_0)(v)$  
esistenza derivata direzionale: $\partial_v f(p_0)$  
metti l'uguaglianza  

**Rapporto: continuità derivate direzionali e differenziabilità**

D-cont: derivate direzionali continue  
f-diff: f differenziabile  

non hanno nessun rapporto di implicazione, una può essere vera e l'altra no  
D-cont, ma non f-diff: $f(x,y)=|(x,y)|=\sqrt{x^2+y^2}$  
f-diff, ma non D-cont: $f=x^2 \sin \frac{1}{y}$  

**Derivate parziali e direzionali**

SOLO SE f è differenziabile:   
la derivata direzionale è una combinazione lineare delle derivate parziali:  
$df(v) = \nabla f \cdot v = f_xv_1+f_yv_2$  
quindi f-diff implica l'esistenza sia delle parziali che delle direzionali  

ESISTENZA: esistono direzionali -> esistono parziali   
perchè parziali sono un tipo di direzionali   
il contrario SOLO se f è differenziabile, in quel caso le parziali formano una abse come mostrato sopra  
sennò no, se f non + differenziale, allora non posso usare le parziali come base delle direzionali    

CONTINUITÀ: parziali continue -> direzionali continue    
perchè combinazione lineare di continue è continua (e parziali continue implica f differenzibile)    
non vale l'implicazione opposta  
cioè che avere tutte le direzionali TRANNE le parziali continue non implica avere le parziali continue  

**Nota**

differenziabile -> continua   
è ovvio da analisi 1  

**Funzioni vettoriali**

una funzione vettoriale $f=(f_1, ..., f_n)$ è differenziabile SE E SOLO SE tutte le sue componenti lo sono  

**Derivabile vs differenziabile**

derivabile: le derivate parziali (o la derivata, se in una var sola) esistono  
differenziabile: esiste L lineare che la approssima al primo ordine

in 1 var sono la stessa cosa, in multivar no:

1 var:  
derivata continua -> differenziabile = derivabile  

multivar:  
derivate parziali continue -> differenziabile ->  derivabile  

**DF1 : proof**

proof:  
1- scrivo df (in modo simile alla derivata direzionale, che poi vorrò)  
$ \lim_{t \rightarrow 0} \frac{f(p_0+tv)-f(p_0)-L(tv)}{|tv|}=0$  

2- tolgo v dal denominatore, tanto non può essere 0, non è quella a portare il limite a 0  
3- tolgo abs value dal denominatore  
4- uso la linearità di L per portare fuori t, che semplifico col denominatore  
$ \lim_{t \rightarrow 0} \frac{f(p_0+tv)-f(p_0)}{t} -L(v)=0$  
$ \lim_{t \rightarrow 0} \frac{f(p_0+tv)-f(p_0)}{t} = L(v)$   

ho ottenuto la mia tesi, perchè il lato sinsitro è la definizione di derivata direzionale  
quindi ho dimostrato che esiste, e poichè è un limite anche che è unica  


#### Altri

**DF3: Differenziabilità composte**

$h(x)=g(x)f(x)$  
se g e f sono differenziabili, anche h lo è  
e: $d(g(f)) = dg(f) \; df$

quindi se f è fatta da composizione di funzioni di tipo $C^1$ o $C^\infty$, allora è differenziabile  

**Th di Schwarz**

se $f \in C^2$:  
cioè le derivate di secondo ordine sono continue:  
  
$f_{xy}=f_{yx}$  
cioè le derivate miste sono uguali, posso permutare l'ordine in cui derivo  

il contrario non vale, cioè se sono uguali non vuol dire che sono continue  

**Th di Schwarz di ordine superiore**

le derivate fino all'ordine k sono continue  
allora in tutte le derivate di tale ordine (e inferiori) posso modificare come voglio l'ordine con cui faccio le derivate, e avrò lo stesso risultato (cioè posso permutare la lista delle derivate)  

**Th di Taylor**

sono le approssimazioni di Taylor per funzioni multivaribile

approssimazione di f intorno a $x_0$: 

$f(x)=f(x_0)+<\nabla f(x_0), (x-x_0)> + \frac{1}{2} ...$  

**Th di Fermat**

$f: \Omega \rightarrow R$  
se $x_0$ è un punto di massimo o minimo locale e f è differenziabile in $x_0$  
allora $\nabla f(x_0)=0$  
oppure $df(x_0)=0$  

ed è detto punto critico  

cioè, in punti critici il gradiente ha tutte le componenti nulle

**Definire una funzione**

$f: \Omega \rightarrow R$  
di solito richiedo che $\Omega$ sia aperto    
così non devo preoccuparmi se sono sul bordo, posso parlare di limite ovunque, ed avvicinarmi ad un punto da ogni direzione  
inoltre, la definizione di derivabile necessita che sia aperto, perchè se scegliessi un punto sul bordo non funzionerebbe  

**Teoremi vari: gradiente**

il valore massimo di $df$ (cioè la maggior velocità di cambiamento di f) è $||\nabla f||$ e accade della direzione data da $||\nabla f||$  
in pratica, il gradiente punta nella direzione che cambia più veloce, e più è grande più veloce cambia  

$\nabla f(x_0,y_0)$ è ortogonale alla curva di livello $f(x,y)=k$ al punto $(x_0, y_0)$  


### Esercizi: controlla le cose

#### Esistenza derivate direzionali

**Come fare**

esiste se:  
la definizione di derivata direzionale come limite dà un valore finito (non oscilla o diverge)  

**Steps**  

prendo:   
una direzione generale: $v=(\alpha, \beta)$  
il punto che voglio studiare: $p=(x_0, y_0)$ -> qui ci sono dei numeri   

1. scrivo la definizione di derivata direzionale  
$ \partial_v f = \lim_{t \rightarrow 0} \frac{f(p_0+tv)-f(p_0)}{t}$  

2. sostituisco il punto p  
così il primo addendo è la funzione ATTORNO a quel punto  
il secondo addendo è la funzione IN quel punto  

3. sostituisco v e scrivo la formula della funzione  

4. risolvo il limite  
quindi visto che t tende a 0 devo tenere gli addendi con l'esponente più basso  

5. alla fine, vedo se ho problemi di esistenza  
se sì, controlla quelli a parte  

**Esempio**

$f(x,y) = \begin{cases}
\frac{xy^2}{x^2+y^4} & (x,y) \neq (0,0) \\
0 & (x,y)=(0,0)
\end{cases}$

voglio vedere se esistono in $p=(0,0)$  

$ \partial_v f = \lim_{t \rightarrow 0} \frac{f(p_0+tv)-f(p_0)}{t}$  
$ \lim_{t \rightarrow 0} \frac{f(tv)-f(0)}{t} $  
$\lim_{t \rightarrow 0} \frac{f(t\alpha, t\beta)}{t} =  \lim_{t \rightarrow 0}  \frac{1}{t} (\frac{\alpha \beta^2 t^3}{\alpha^2 t^2 +\beta^4t^4}) $  
$ \lim_{t \rightarrow 0} \frac{\alpha \beta^2}{\alpha^2}  = \frac{\beta^2}{\alpha}  $  

l'unico problema è $\alpha =0$, lo controllo a parte:  
$\lim_{t \rightarrow 0} \frac{f(0, t\beta)}{t} = \lim_{t \rightarrow 0} \frac{0}{t} = 0$  
ok, tende a 0 che è un numero finito, va bene     
quindi tutte le derivate direzionali esistono    

**Se voglio solo quelle parziali**

se mi interessano solo quelle, posso usare $v=(x,0)$ e $v=(0, y)$ invece  
cioè calcolo subito i casi specifici, non quello generale

#### Differenziabilità

**Come fare**

è differenziabile se:  
$\lim _{(x,y)\rightarrow (p)} \frac{f(x,y)-f(p)-\nabla f(p)\cdot (x,y)}{\sqrt{x^2+y^2}}=0$  

perchè vuol dire che esiste un'approssimazione lineare:  
$f(h,k)=f(p)+\nabla f(p) \cdot (h,k) + o(\sqrt{h^2+k^2}) \quad$ per $(h,k)\rightarrow (p)$  

spesso si usa p=(0,0)  

**Steps**

1. scrivi il limite  
2. calcola e sostituisci $f(0,0)$ e $\nabla f(0,0)$  
3. se ho una trig, fai upper bound  
4. sennò, scegli una direzione, tipo $y=mx$  
5. il limite deve tendere a 0 

**Trucchi**

posso vedere se è differenziabile fa quando ho calcolato le derivate parziali  

è differenziabile se:  
se le derivate parziali sono continue  

NON è differenziabile se:  
quando controllo se esistono le derivate direzionali, alla fine ho ottenuto $f=\frac{\alpha^2}{\beta}$  
vedo che f NON È LINEARE in $\alpha$ e $\beta$  
cioè non è lineare in v  
quindi non può essere differenziabile, che è una mappa lineare  
e io so che, se è differenziabile, la derivata direzionale deve coincidere con il differenziale (DF1)      


**Esempio**

$f=\frac{xy}{x^2+y^2}$

$\lim _{(x,y)\rightarrow (0,0)} \frac{f(x,y)-f(0,0)-\nabla f(0,0)\cdot (x,y)}{\sqrt{x^2+y^2}}=$  
$ =  \lim _{(x,y)\rightarrow (0,0)} \frac{f(x,y)-0-(0,0)\cdot (x,y)}{\sqrt{x^2+y^2}} = $  
$= \lim _{(x,y)\rightarrow (0,0)} \frac{f(x,y)}{\sqrt{x^2+y^2}} =$  

scelgo un cammino per tendere a (0,0), tipo la retta $y=x$:  
$f(x,x) = \frac{x}{2}$   

quindi:  
$\lim _{(x,y)\rightarrow (0,0)} \frac{f(x,y)}{\sqrt{x^2+y^2}}= \lim \frac{x/2}{|x|\sqrt{2}} \lim = \frac{1}{2 \sqrt{2}} sgn(x)$  
il limite non è 0, e comunque non esiste  
quindi non è differenziabile    


**Esempio**

$f=x^2 \sin \frac{1}{y}$

$\lim _{(x,y)\rightarrow (0,0)} \frac{f(x,y)-f(0,0)-\nabla f(0,0)\cdot (x,y)}{\sqrt{x^2+y^2}}=$  
$ =  \lim _{(x,y)\rightarrow (0,0)} \frac{f(x,y)-0-(0,0)\cdot (x,y)}{\sqrt{x^2+y^2}} = $  
$= \lim _{(x,y)\rightarrow (0,0)} \frac{f(x,y)}{\sqrt{x^2+y^2}}=0$  

per $y \neq 0$:  
$|f(x,y)| = |x^2 \sin (1/y) | \le x^2$  

quindi:  
$ \lim \frac{|f(x,y)|}{\sqrt{x^2+y^2}}\le \lim \frac{x^2}{\sqrt{x^2+y^2}}\le \lim |x| =0 $   
perchè la frazione è minore di x ( x quadro fratto una cosa più grande di x), ed è positiva  

tende a 0 come richiesto, quindi è differenziabile  

#### Continuità

ricorda che la continuità di una funzione dipende dal dominio su cui la sto studiando  

**Come dimostrare che NON è continua**

metodo 1:  
trova due successioni che tendono allo stesso punto  
ma la funzione calcolata su quelle successioni ha limiti diversi  
nota che una successione di solito è parte dell'altra, ma con una variabile settata già a dove voglio fare convergere  

metodo 2:  
calcola i limiti attraverso due cammini diversi, devono essere diversi  

**Nota metodo 2**

metodo generale:  
scegli una retta, tipo $y=mx$    
calcola $f(x, mx)$  
il risultato NON deve dipendere da m  
se dipende da m non è continua, perchè seguendo strade diversi ho valori diversi  

altri metodi:  
cammini comuni da usare sono l'asse x (y=0), l'asse y (x=0) e la bisettrice (x=y)  

**Esempio: metodo 1**

$f(x,y) = \begin{cases}
\frac{xy^2}{x^2+y^4} & (x,y) \neq (0,0) \\
0 & (x,y)=(0,0)
\end{cases}$

voglio studiare continuità in (0,0)  
scelgo come successioni, che tendono a (0,0):   
$a_n = (\frac{1}{k}, \frac{1}{\sqrt{k}})$  
$b_n =  (0, \frac{1}{\sqrt{k}})$  

calcolo f in quelle successioni:  
$f(a_n)= \frac{1}{2}$, per ogni k  
$f(b_n) = 0$, per ogni k  
sono diversi, quindi f non è continua in (0,0)  

**Esempio: metodo 2**

$f(x,y)=\frac{xy}{x^2+y^2}$  
$g(x,y)=\frac{xy^2}{x^2+y^2}$  

voglio studiare la continuità in (0,0)  

scelgo $y=mx$  
$f(x, mx) = \frac{m}{1+m^2}$  
dipende da m, quindi non è continua in (0,0)  

$g(x,mx)= \frac{m}{1+m^2}x \rightarrow 0$  
qualunque strada prendo, tende comunque a 0  
quindi g è continua in (0,0)  

#### Continuità derivate parziali

$f(x,y)=\begin{cases}

0 & (x,y)=(0,0)

derivate seconde non sono continue, quindi $f_{xy} \neq f_{yx}$  

**Derivate miste del secondo ordine**

nota che non è necessario mettere t al posto di una variabile, ma si fa solo per visualizzare bene quale sto cambiando e quale no  

1- calcolo derivate del primo ordine, nel punto p=(0,0):     
$f_x=\frac{f(t,y)-f(0,y)}{t} = \frac{t^2-y^3}{t^2+y^2} \rightarrow -y$  
$f_y=\frac{f(x,t)-f(x,0)}{t} = \frac{x^3-xt^2}{x^2+t^2} \rightarrow x$  
per trovarle, mi avvicino al punto in questione lungo l'asse che voglio, muovo solo una coordinata  

miste:   
$f_{xy}=\frac{f_x(0,y)-f_x(0,0)}{y} \rightarrow -1$  
$f_{yx}=\frac{f_y(x,0)-f(0,0)}{x} \rightarrow 1$  
sono diverse
