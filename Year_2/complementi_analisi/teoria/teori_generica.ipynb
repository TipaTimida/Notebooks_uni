### Punti critici, massimi e minimi

#### Defs

$\Omega$ è sempre aperto  
f è definita come: $f: \Omega \subset R^n \rightarrow R$   

**Punto critico**

$x_0 \in \Omega$ è un punto critico di $f: \Omega \subset R^n \rightarrow R$   
se $\nabla f(x_0)=0$  

note:  
i punti dove f non è derivabile possono essere detti punti critici o no  
a volte chiamati punti critici di bordo  

**Massimo/minimo locale**

$x_0$ è un un massimo locale di $f: \Omega \subset R^n \rightarrow R$   
se $\exist r > 0$ tale che $f(x) \le f(x_0) \quad \forall x \in B(x_0, r)$  

cioè tutti i valori in un intorno piccolo quanto voglio sono minori  

**Punto di sella**

$x_0 \in \Omega$ è un punto di sella se:  
$\forall U$ intorno di $x_0, \quad \exist x_1, x_2 \in U$ tali che $f(x_1)<f(x_0)<f(x_2)$  

punto di sella -> punto critico  
ma non viceversa  

cioè è un punto critico che non è ne max nè min (no estremo locale)
