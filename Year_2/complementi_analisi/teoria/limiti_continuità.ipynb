### Limiti multivar

**Limite in multivar**

problema: non esiste il concetto di infinito, perchè è uno scalare, non un vettore  
quindi non posso fare quando $x \rightarrow \infty$  
ma devo fare quando $\vec{x}$ tende ad un vettore, la cui norma tende ad infinito,  $||\vec{x}|| \rightarrow \infty$  

per il resto sono uguali alle definizioni univariabile  

**Prova che ha un limite**

trova un upper bound per $|f(x,y)|$  
mostra che, quando (x,y) tende a (x_0, y_0), l'upper bound di f tende a l  
???  

hint:  
1- metti valore assoluto a tutta la funzione    
2- |ab| =|a||b| per seprare il valore assoluto tra numeratore e denominatore, e anche tutte le moltiplicazioni che appaiono  
tipo |ab/c| diventa |a||b|/|c|  
3- se uno dei due è chiaramente positivo, togli il valore assoluto   
4- se ho diversi valori assoluti da moltiplicare, tipo |x| |y|, posso sostituire: $|x|\le \sqrt{x^2+y^2}$, stessa cosa per y  
altrimenti scritto: $x < (x,y,z) = \sqrt{x^2+y^2+x^2}$  
così sia x che y diventano la stessa cosa, posso raccogliere  
è la disuguaglianza: $|x_i| \le |x| = \sqrt{\Sigma x_i^2}$

**Prova che un punto NON ha limite**

metodo facile:  
prendi due successioni che tendono allo stesso punto  
mostra che hanno limite diverso  

metodo avanzato:  
invece che due successioni ne prendo una sola, con un parametro   
la successione deve sempre tendere al punto di cui voglio investigare il limite, qualunque sia il parametro  
mostra che, se cambio il parametro, cambia il valore a cui tende $f(x_n)$  


#### Trova limiti

**Esempio: prova che NON ha limite**

metodo avanzato:  
   
$f(x,y)=\frac{(x-1)(y-2)}{(x-1)^2+(y-2)^2}$ 
 
prendo la succesione: $p_k=(1+\frac{\alpha}{k}, 2+ \frac{\beta}{k})$  

a cosa tende la successione:  
qualunque i parametri alpha e beta, tende a (1,2) quando k tende a infinito  

a cosa tende la funzione della successione:  
$f(p_k)= \frac{\alpha \beta}{\alpha^2+\beta^2}$  
il limite può variare, quindi f(1,2) non ha limite  


### Funzioni continue

**Def**

se $x \rightarrow x_0$, allora $f(x)\rightarrow f(x_0)$  
usata nelle dimostrazioni  

**Funzione vettoriale**

una funzione vettoriale (cioè il codominio è pure multivar) è continua se tutte le componenti (che dà in output) sono continue

#### È continua?

**IMPORTANTE**

la continuità di una funzione f DIPENDE DAL DOMINIO, che va specificato    

**Esempio**

$f=\frac{1}{|x|}$ è continua?  
dipende dal dominio:  
ha un punto di discontinuità in x=0  
quindi nel dominio R è discontinua, nel dominio R-{0} è continua  

**Funzione definita a tratti**

se la definisco a tratti, qualunque valore gli dò nel punto non sarà mai continua  

$f=\begin{cases} 
\frac{1}{|x|} & x \neq 0 \\
\lambda & x=0 
\end{cases}$

**Come scrivere il dominio**

spesso lo scrivo solo, non lo risolvo  

tipo se ho: $f= \frac{1}{x+y^2}$  
posso dire che è continua sul dominio:  
$A=\{(x,y)\in R^2 : \;x+y^2 \neq 0\}$  

#### Proprietà

**Se f continua**  

preserva le proprietà degli insiemi: compatto, connesso, connesso per archi  

NON preserva: aperto, chiuso  

eg: $f=|x|$  
(-1,1)->[0,1)  
non è più aperto  

eg: $f=\arctan x$  
R -> (-pi/2, pi/2)  
non è più chiuso  

**Proprietà**

composizione di funzioni continue è continua  
usata per giustificare che la f originale è continua  

restrizione di funzione continua è continua  
usata per gisutificare che la restrizione allo spazio che sto studiando è continua  

### Massimi e minimi

**Tipi**

se la funzione lì è maggiore/minore di:    
 
- assoluto: in ogni altro x nel dominio   

- locale: in ogni x di una palla centrata lì, piccola a piacere    
$\exists R>0: \quad f(p)=max _{A \cap B(p, R)}f$

**Come scrivere**

$max_A \; f = max \:f(A)=f(p)$  

massimo: il valore f(p), è unico  
punti di massimo: le coordinate dove il massimo è raggiunto, possono essere più di una, tipo $p_1$ e $p_2$  

**Come scrivere l'immagine**

vedi se max e min esistono, mettici parentesi quadre  
se uno dei due non esiste, rimpiazzalo con sup o inf, con parentesi tonde  

esempio:  
$Im \; f = (inf_A \, f, max_A \,f)$   

#### Teoremi su max e min

**Th Weierstrass**

A compatto, f continua -> esistono massimo e minimo di f su A  
(cioè non sono inf e sup, sono davvero raggiunti in A)  

A può essere qualunque cosa: R^n, una curva strana...  

**Altri teoremi**

generalizzazione weierstrass:  
f continua, A compatto -> f(A) compatto  

corollario:  
f continua, A compatto -> gr(A) compatto  
perchè $gr(f)=(x, f(x))$  
x->x mappa continua  
x->f(x) mappa continua  
combinazione di mappe continue è continua  

**Come usarlo: dimostra che ci sono max e min di S**

f continua, ristretta ad un compatto A -> f ha max e min su A  

come dimostrare le ipotesi:    
S compatto: chiuso + limitato     
f continua in A: se f è continua, lo è anche in tutte le sue restrizioni  

### Tipi di funzioni

**Functioni in campo complesso**

f è una funzione:  

- a variabili complesse: dominio è sottoinsieme di C    

- a valori complessi: codominio è sottoinsieme di C  
o anche solo f è complessa  

**Funzione vettoriale**

cioè il codominio è pure multivar, non solo il dominio  
quindi dà in output un vettore, con delle componenti  

esempio:  
$f: R^2 \rightarrow R^3$  
$(x,y) \rightarrow (u_1(x,y), u_2(x,y), u_3(x,y))$  
quindi f prende due valori in input e manda in ouput un vettore a tre componenti
