$v = k[A]^2[B]$

**Affinità elettronica lungo un periodo**

affinità elettronica = elettronegatività  

def: capacità di un atomo di attrarre a sé gli elettroni condivisi in un legame chimico  

lungo un periodo aumenta il numero di protoni nel nucleo, mentre i nuovi elettroni sono sempre nello stesso orbitale, quindi non si allontanano e non schermano  
quindi aumenta la forza di attrazione del nucleo sugli elettroni di valenza, e quindi l'elettronegatività

### Equazione di Nernst (non CAPISCO ???)

**Domanda**

Ricavare l’equazione di Nerst usando esclusivamente i concetti di 
lavoro ed energia libera  

**Risolvo**

1- in un processo reversibile:  
$\Delta G = -W_{max}$  

2- lavoro elettrico è dato da:  
$W_{max} = qV = nFE$  
F: costante di faraday, carica di una mole di elettroni  
E: fem della cella, quella calcolata con formula di Nernst  
n: numero di moli  

3- in condizioni standard:  
$\Delta G^° = -nFE^°$  

4- altra formula di termodinamica   
$\Delta G = \Delta G^° + RT \ln Q$  
Q è il quoziente di reazione, quello che si scrive come K, ad esempio:  
$Zn_{s}+CU^{2+} \rightarrow Zn^{2+} + Cu_s$  
$Q = \frac{[Zn^{2+}]}{Cu^{2+}}$  

5- metto tutto insieme  
$-nFE = -nFE^° +RT \ln Q$   
$E = E^° +\frac{RT}{nF}\ln Q$  
a 25°C, $RT/F = 0.0592$  


### Reazione

**Testo**

CoCl2(s) + H2(g) ⇆ Co(s) + HCl(g)  

all'inizio:  
1 mol di CoCL2  
2 mol di H_2  

K_p = 0.196 atm  
T = 570°C 
V= 10L  

calcola la pressione parziale di HCl all'equilibrio

**Risolvo**

1- bilancia la reazione  
$CoCl_2(s) + H_2(g) ⇆ Co(s) + 2HCl(g)$  

2- scrivo K_p  
$K_p = \frac{P_{HCl}^2}{P_{H_2}}$  
dove $P_{X} = n_{X} \frac{RT}V$  

3- scrivo la tabella per la variazione   
$n_{H_2} = 2-x$   
$n_{HCl}=2x$    

4- risolvo  
$K_p = \frac{4x^2}{2-x} \frac{RT}V$  
$x = 0.16$  

ricorda quanto vale R, costante dei gas
