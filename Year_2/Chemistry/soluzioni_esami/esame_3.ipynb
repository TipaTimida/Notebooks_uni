### Raggio atomico/ionico e elettronegatività

**1- Motivazioni della di erenza fra raggio atomico e raggio ionico**

cioè, cambia il raggio a seconda che sia un atomo neutro o uno ione? (K o K^+)  

Un catione è più piccolo dell’atomo neutro perché perde elettroni e spesso anche un livello elettronico.  
Quindi c'è meno effetto di schermo, gli elettroni rimasti sentono maggiormente la carica nucleare.  
Un anione è più grande perché acquista elettroni, aumentando repulsione elettronica e schermaggio.

**2- Elettronegatività**

Capacità di un atomo di attrarre a sé gli elettroni condivisi in un legame chimico.  
Aumenta lungo un periodo e scende lungo un gruppo.

### Clausius-Clapeyron

**Trova eq Clapeyron per solido->gas**

1- eq di Clapeyron (per una generica transizione di fase):    
$\frac{dP}{dT} = \frac{\Delta H}{T \Delta V}$    

2- nella sublimazione:  
$\Delta V \sim V_{gas}$  
perchè il volume nello stato gassoso è molto maggiore di quello nello stato solido  
$V_{gas} =\frac{RT}{P}$ perchè gas ideale (considero n=1???)

3- quindi:  
$\frac{dP}{dT} = \frac{\Delta H}{T} \frac{P}{RT} $   
$\frac{dP}P = \frac{\Delta H}{RT} \frac{dT}{T^2}$

4- integro:  
$\ln P = -\frac{\Delta H}{R} \frac{1}{T} +const$
