### Espansione dell'ottetto ed energia di ionizzazione

**Espansione dell'ottetto**

È la possibilità per alcuni atomi di avere più di 8 elettroni nel guscio di valenza  
Può avvenire per elementi dal terzo periodo in poi, come S, P, Cl, perché possono coinvolgere orbitali d  
Un esempio è SF6, in cui lo zolfo forma 6 legami  

**Energia di ionizzazione**

energia necessaria per rimuovere un elettrone da un atomo isolato in faso gassosa:  
$X_{(g)} \rightarrow X^+_{(g)}+e^-$  

È sempre positiva, perché bisogna fornire energia  

Esistono prima, seconda, terza energia di ionizzazione, ecc.  

aumenta lungo un periodo e diminuisce lungo un gruppo.

### Equazione di Clausius-Clapeyron

**Ricavarla per transizione solido-solido**

si ricava sempre uguale, non importa che transizione  

condizione: sistema all'equilibrio di fase  
$dG_1 = dG_2$  

$V_1dP - S_1dT =  V_2dP - S_2dT$  
$ (V_2-V_1)dP = (S_2-S_1) dT$  
$\frac{dP}{dT}=\frac{\Delta S}{\Delta V}$  
infine: $\Delta S = \frac{\Delta H}{T}$  

Per fasi condensate, cioè solido-solido o solido-liquido, ΔV si può considerare circa costante  

nota: è un cambiamento di fase, quindi $\Delta N_i=0$, non cambia la composizione delle specie  

### Esercizi pH

**Testo**

a- 0.1 di NaF  
b- 0.1 di NaF e 0.05 di HF  
c- 10mL della soluzione B diluiti a 20mL  

sapendo K_a(HF)  

**Caso 1**

NaF è acido forte, si scompone completamente in Na^+ e F^  

cosa reagisce:  
HF è acido debole, quindi F^- è base forte, reagisce    
NaOH è base forte, quindi Na^+ è acido debole, non reagisce  

$F^- + H_2O⇌ HF + OH^-$   

1- sistema  
$K_b = \frac{K_w}{K_a} = \frac{[HF][OH^-]}{[F^-]}$  
tabella variazioni: $[F^-]=0.1-x \quad [HF]=[OH^-]=x$  
qui appare $c_0-x$ perchè non è scomposizione di un sale, quindi c'è anche il denominatore nella formula di K, qndi c'è qualcosa che esiste all'inizio  

2- risolvo  
$K_b = \frac{x^2}{0.1-x}$  
calcolo x, poi $pH = 14 + log(x)$  

nota: se K_b è molto piccola, ci sono pochi reagenti, quindi x è piccola (x è la concentrazioe dei prodotti)  
allora $c_0-x \sim c_0$  

**Nota: soluzione tampone**

nel caso A, $K_b = 5.8 \, 10^{-11}$ è molto piccolo, ci sono quasi solo reagenti, quindi quasi solo $F^-$, poco HF  
quindi non è proprio un tampone  

nel caso B invece sono presenti in quantità confrontabile, F^- è solo il doppio di HF, non vari oridini di grandezza superiore  
all'inizio, o all'equilibrio  

per avere una soluzione tampone, il rapporto tra F?- e HF deve essere dell'ordine dell'unità, tra 0.1 e 10  

**Caso 2**

NaF si dissocia completamente, ho $0.1$ di F  
e ho $0.05$ di HF, cioè una soluzione tampone  

$pH = pK_a + \log \frac{F^-}{HF}$  
ho già tutto, K e le concentrazioni, basta metterle nella formula  

altro modo:  
se non ricordo la formula, posso rifare semplicemente la formula della K, ma con le nuove concentrazioni  

$K_a = \frac{[H^+][F^-]}{[HF]} = \frac{x \, (0.05+x)}{0.1-x}$   
ora calcolo x, ma poichè $K_a$ è molto piccola, approssimo $K_a = \frac{0.05}{0.1}x$  
ora posso calcolare: $pH = 14 + \log x$  

questa è la stessa cosa della formula con pH e pK del tampone  

**Caso 3**

è una soluzione tampone, diluirla non cambia il pH  

si vede, se moltiplico lo stesso numero 0.05 e 0.1, x non cambia
