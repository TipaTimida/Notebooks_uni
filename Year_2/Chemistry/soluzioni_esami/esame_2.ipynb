### Funzione di stato ausiliaria

1- cos'è  
una funzione termodinamica costruita a partire da altre funzioni di stato (S, T) per descrivere meglio un sistema in certe condizioni   

2- a cosa serve  
un sistema lasciato libero di evolvere tenta di minimizzare la propria energia  
a seconda di quali variabili è in grado di modificare, e di quali sono fissate dall'ambiente circostante (var naturali), la formula dell'energia da minimizzare è diversa  
  
3- esempio  
ad esempio, in chimica usiamo l'energia interna U, l'entalpia H, l'energia libera di Helmholtz Fe l'energia libera di Gibbs 0  
mi fanno capire se una trasformazione è spontanea o no, e a che equilibrio tende il sistema  

### Traformata di Legendre

1- cos'è  
serve a cambiare le variabili naturali di una funzione termodinamica   
cioè a passare da un porenziale termodinamico all'altro    

2- esempio:  
dall'energia interna U(S,V), si ottiene l'energia libera di Helmholtz F(T,V)=U-TS

### Reazione

**Testo**

$I_2+H_2 ⇆ 2HI$    
all'equilibrio: $n_{HI}=0.3 \quad n_H=n_I=0.8$  
V=0.5L  

1- trovare le $K_p, K_x, K_c$  
2- che succede se V'=2V   
3- che succede se aggiungo 0.5 moli di azoto   

**1- Trova le K**

1- scrivo dormula con frazioni molari:  
$K_x = \frac{n_{HI}^2}{n_I^2 \; n_H^2} = 0.14$  
nota che $n_{tot}$ al denominatore di ogni frazione molare si semplifica 

2- passo a concentrazioni molari  
$[C]=\frac nV$  
nota che, anche qui, V si semplifica  
quindi $K_C = K_p$  

3- passo a pressioni parziali  
$p_i = \frac{n_i}{n_{tot}} \,p_{tot}  = \frac{n_i}{V} RT$    
anche qui si semplifica tutto, $K_P = K_x$  
si vede anche perchè $\Delta n = 2-1-1 = 0$  

**Altre formule**

$K_c$: usa concentrazioni molari  
$K_p$: usa pressioni parziali $p_i = p_{tot} \frac{n_i}{n_{tot}} = \frac{n_iRT}{V}$  
$K_x$: usa frazioni molari $ x_i=\frac{n_i}{n_{tot}}$  

le formule si scrivono allo stesso modo di K_c, con gli stessi esponenti, ma invece di [C] metto p oppure x  

esempio: $K_p = \frac{P^2_{HI}}{P_{I^2}P_{H^2}}$  

passaggi:  
$\Delta n $: la differenza tra la somma dei numeri stechio prima e dopo, NON tra i valori delle moli all'equilibrio   
$K_p = K_c (RT)^{\Delta n}$  
$K_p = K_x P_{tot}^{\Delta n}$   
per ricordarle, guarda la formula delle pressioni parziali, riconoscici la concentraazioni e la frazione molare, al resto metti $\Delta n$ come esponente   

**2- Raddoppio V**

l'equilibrio non cambia, perchè nella formula di K_c il volume si semplifica    

**3- introduco dell'azoto**

L’azoto è inerte, non succede niente  

perchè:  
La pressione parziale, o la concentrazione dipende soltanto dalle moli di quella specie  
aggiungere azoto non la cambia, PERCHÈ il volume del contenitore V è mantenuto costante

**4- Introduciamo una soluzione in cui è solubile HI**

HI viene sottratto alla fase gassosa  
L’equilibrio si sposta verso destra per produrre altro HI

### Precipitazione

**Nomenclatura**

nitrato d'argento: $AgNO_3$ (N +5)  
nitrito d'argento: $AgNO_2$ (N +3)   
il nome si riconosce dall'anione  

-ato: numero di ossidazione maggiore per l'atomo centrle  
-ito: numero di ossidazione minore (quindi gli attacco meno ossigeni, ito è simile a ino, qualcosa di piccolo, con poche cose attaccate) 

iodato: $IO_3^-$, I +5  
iodito: $IO_2^-$, I +3  

iodato di potassio: $KIO_3$  
ioduro di potassio $KI$  

URO: anioni senza ossigeno  

recap:  
$I^-$ ioduro  
$IO_2^-$ iodito  
$IO_3^-$ iodato  

**Testo**

$AgNO_3$: $V=40mL$, $[C]= 0.0125 \; M$     
$KIO_3$ :  $V=60mL$, $[C]= 0.0217 \; M$   
$K_{AgIO^3} = 1.8 \, 10^{-5}$  

1- massa di iodato precipitato  
2- concentrazione delle oecie  


**Risolvo**

1- reazioni di dissociazione  
$AgNO_3 ⇌ Ag^+ + NO_3^-$  
$KIO_3 → K^{+} + IO_3^-$   

2- moli iniziali  
$n_{Ag} = [AgNO_3]_0 \;  V_{AgNO_3} = 5 \, 10^{-4} \, mol$  
$n_{IO} = [KIO_3]_0 \, V_{KIO_3} = 1.3 \, 10^{-3} \, mol$   

3- concentrazioni iniziali dopo il mescolamento  
$V_{tot} = 100 mL = 0.1 L$  
$[Ag^+]_0 = {n_{Ag}}{V_{tot}} = 5 \; 10^{-3} \; M$  
$[IO_3^-]_0 = {n_{IO_3^-}}{V_{tot}} = 1.3 \; 10^{-2} \; M$  

4- quoziente di reazione  
si è formato:  
$AgIO_3 ⇌ Ag^+ + IO_3^- $  
all'inizio, sono nella situazione  
$Q = [Ag^+]_0[IO_3^-]_0 = 6.51\, 10^{-5}$  
vedo che $Q > K_{ps}$, quindi si forma precipitato    

5- quantità di precipitato  
ne precipita abbastanza per avere K_ps  
$K_{ps} = \frac{(n_{Ag}-x)(n_{IO} -x)} {V_{tot}^2}$  
$x = 3.17 \, 10^{-4} mol$  
utile scriverla così, in modo che x sia già in moli, non una concentrazione  

6- trovo la massa  
massa molare di $AgIO_3$: M = 282.77 g/mol  
somma masse molari di ogni atomo  
massa: m=xM$

6- concentrazioni finali  
$[Ag^+]=[Ag]_0 -\frac{x}{V_{tot}}$  


**Attenzione**

devo stare attenta se x la definisco come concentrazione o quantità di moli  
entrambi vanno bene, ma sii consistente
