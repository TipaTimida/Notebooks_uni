### Temperatura critica e energia di ionizzazione

**Temperatura critica**

temperatura al di sopra della quale un gas non può essere liquefatto per sola compressione, qualunque sia la pressione applicata   
(una sostanza non può più esistere allo stato liquido, indipendentemente dalla pressione applicata)  

**Andamento energia di ionizzazione lungo un periodo**

sposandosi a destra lungo un perido, aumenta il numero di protoni, quindi aumenta la forza con cui il nucleo attira gli elettroni di valenza  
pertanto è più difficile strapparli, e l'energia di ionizzazoine aumenta  

### Calcola pH

**Testo**

soluzioni:  
A. HCl 0.1 M  
B. $NH_2$ 0.2 M  
$K_b(NH_3) = 1.8 \, 10^{-5}$  

trova pH di:  
10 mL di A + 5mL di B.  
50 mL di A + 50 ml di B.  
75 mL di A + 50 ml di B.  

**Come mescolare**

1- $V_A$ e $V_B$, i volumi delle soluzioni usate    

2- $V_{tot} = V_A+V_B$  

3- $n_A = [A] \, V_A$  
moli delle soluzioni  

4- concentrazioni dopo aver mescolato  
$[A] = \frac{n_A}{V_{tot}}  = [A_{prima}] \frac{V_A}{V_{tot}}$ 

**Note**

$HCl+NH_3​→NH^4​+Cl^−$  

l'ammoniaca è una base, accetta una H (ovviamente, non può rilasciare OH, perchè non ne ha)  
se passo da base a acido coniugato, devo passare da $K_b$ a $K_a$, dove $K_a = \frac{K_w}{K_b}$  
HCl è acido forte, quindi l'equilibrio è tutto spostato verso i prodotti 

ci sono due reazioni consecutive, la prima ha l'equilibrio tutto spostato a destra, la seconda ha K_ps:    
1- $\quad HCl+NH_3​→NH^4​+Cl^−$  
2- $\quad NH_4^+ ​⇆ NH^3 + H^+$  
dove $K_a = \frac{K_w}{K_b}$  

**Risolvo: caso 1**

1- prima reazione, tabella moli  
fase | HCl | NH3 | NH4+ 
---|---|---|---
inizio | 0.001 | 0.001 | 0 | 0 
variazione | -x | -x  | +x | +x
fine | 0.001 -x | 0.001 -x | x 
fine calcolata | 0 | 0 | 0.001   

quindi so che alla fine ho: $n_{NH_4^+} = 0.001$, da cui posso trovare la concentrazione  
 
2- seconda reazione. sistema    
$NH_4^+ ​⇆ NH^3 + H^+$  ​ 
$[NH^3]= [H^+]=x$     
$K_a = \frac{[H^+][NH^3]}{NH_4^+} = \frac{x^2}{[NH_4^+]}$   
dove $K_a = \frac{K_w}{K_b}$  
risolvo per x e trovo:  
$pH = -log[x] = 5.22$  

**Risolvo: caso 2**

1- prima reazione, tabella moli  
fase | HCl | NH3 | NH4+ 
---|---|---|---
inizio | 0.005 | 0.010 | 0 | 0 
variazione | -x | -x  | +x | +x
fine | 0.005 -x | 0.010 -x | x | x
fine calcolata | 0 | 0.005 | 0.005   | 0.005 

cioè sono rimasti sia $NH_3$ che $NH_4^+$, acido e base coniugata, ho una soluzione tampone  

2- calcolare pH  
uso Henderson–Hasselbalch: $pH = pK_a + log \frac{[A^-]}{{HA}}$   
poichè nella riga 'fine calcolata', acido e base coniugata hanno le stesse moli, diventa  
$pH = pK_a$  
cioè formula del pH, ma usando il valore K_A invece di [H^+]   
$pK_a = -log[K_a]$ 

**Risolvo: caso 3**

1- prima reazione, tabella moli  
fase | HCl | NH3 | NH4+ 
---|---|---|---
inizio | 0.0075 | 0.050 | 0 
variazione | -x | -x  | +x
fine | 0.0075 -x | 0.050 -x | x
fine calcolata | 0 | 0.0025 | 0.075 

cioè sono rimasti sia $NH_3$ che $NH_4^+$, acido e base coniugata, ho una soluzione tampone  

2- calcolare pH  
uso Henderson–Hasselbalch: $pH = pK_a + log \frac{[A^-]}{{HA}}$   
qui, $log \frac{[A^-]}{{HA}} = \log (0.0025/0.0075) = log(1/3)$  

**Nota**

in questo esercizio si vede come la quantità un reagenti, in questo caso HCl, sia limitante  

### Arrhenius

**Equazione di Arrhenius**

$k=Ae^{-E_a/RT}$  

1- significato dei parametri:  
k: costante cinetica (quella delle legge cinetica, velcotià di reazione)    
T: temperatura  
R: costante dei gas  
$E_0$: energia di attivazione  
A: fattore pre-esponenziale (probabilità di collisione “giusta”) (sezione d'urto, misura la probabilità che le molecole possedendo l’energia richiesta diano luogo ad un urto reattivo)  

2- dipendenza di k da T: 
se T aumenta, k aumenta   
perchè se T aumenta, aumenta il contenuto energetico delle molecole, ovvero aumenta il numero degli urti efficaci, rendendo più veloce la trasformazione

**Temperatura di inversione**

Se una reazione diretta e una inversa hanno dipendenze diverse dalla temperatura, può esistere una temperatura in cui le due velocità diventano uguali: temperatura di inversione  

È la temperatura alla quale un gas reale, soggetto a espansione adiabatica (a entalpia costante) senza compiere lavoro, passa da un processo di raffreddamento a uno di riscaldamento.

### Reazione

**Testo**

$CO + H_2O ⇆ CO_2 + H_2$  
$K_c = 5.06$  

V=1 L  
concentrazioni in ordine: 0.1, 0.2, 1, 0.5 M  

trova composizione all'equilibrio  

**Risolvo**

1- quoziente di rezione:  
$Q = \frac{[CO_2][H_2]}{[CO][H_2O]} = 25$  

2- ho che $Q> K_c$, quindi devo abbassare Q, cioè l'equilibrio si sposta verso i reagenti    

3- trovo concentr all'equilibrio  
$K_c = \frac{([CO_2]-x)([H_2]-x)}{([CO]+x)([H_2O]+x)} $  
$x=0.11$
