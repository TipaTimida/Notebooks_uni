### Definizioni

**Energia di ionizzazione**

l’energia richiesta per rimuovere un elettrone da un atomo isolato (neutro, se è la prima ionizzazione)    
aumenta lungo un periodo e diminuisce lungo un gruppo  

**Orbitale ibrido**

orbitale ottenuto dalla combinazione di orbitali atomici dello stesso atomo  
permette all'atomo di formare più legami

### Energia libera di Helmholtz

**Trovare differenziale, partendo dall'energia interna**

1- definizione:  
$F = U -TS$  

2-energia interna:  
$dU=TdS−PdV+∑​μ_i​dN_i​$   

3- pluggo  
$dF = dU - TdS - SdT = -PdV -SdT + ∑​μ_i​dN_i​$   


**Ruolo nell'equilibrio chimico a V,T costanti**

un sistema a V,T costanti evolve per minimizzare F  
cioè, se dF<0, quella reazione avviene spontanemante  
quindi, nell'equilibrio chimico a V,T costanti, F è minima  

### Pila

**Testo**

elettrodi: AgCl, $Ag_2SO_4$   
so K_ps di entrambi  

trova ddp  

**Trova ddp**

1- trova le concentrazioni di Ag  

1.1 AgCl  
$AgCl ⇌Ag^+ + Cl^-$  
$[Ag^+]=s  \quad [Cl^-]=s$  
$K_{AgCl} = [Ag^+][Cl^-] =s^2$  
$s =\sqrt{K} = 1.3 \; 10^{-5}$  
$[Ag^+] =  1.3 \; 10^{-5} \,M$     

1.2 AgSO  
$Ag_2SO_4 ⇌ 2 Ag^+ + SO_4^{2-}$  
$[Ag^+]=2s  \quad [SO^-]=s$  
$K_{AgSo} = [Ag^+]^2[SO^{2-}] = 4s^2 \, s = 4s^3$  
$s = (K/4)^{1/3} = 1.62 \; 10^{-2}$  
$[Ag^+] =  3.24 \; 10^{-2} \,M$     

2- usa formula per ddp  
$\Delta E = \frac{0.059}{1}\log\frac{[Ag^+]_{catodo}}{[Ag^+]_{anodo}} = 0.2 V$  

note:  
usa log, non ln (trovalo sulla calc)  
ddp deve essere positiva, metti concentrazione maggiore sopra, minore sotto  

**Aggiungo elemento**

nel secondo elemento, aggiungo $[Na_2SO_4]= 10^{-3} M$  

$Na_2 SO_4 \rightarrow 2Na^+ + SO_4^{2-}$   
$Ag_2SO_4 ⇌ 2 Ag^+ + SO_4^{2-}$  

1- sistema  
$2[Na_2SO_4]= [Na^+]$  
$K_{AgSo} = [Ag^+]^2[SO_4^{2-}] = (2s)^2 (s+10^{-3})$   

perchè qui SO viee da entrambe le reazioni, quindi devo somamre il valore prodotto da entrambe  
quello prodoto dalla prima è fisso, quello prodotto dalla seconda è variabile (s)  
cioè è come quando chiamo x e y le parti prodotte da reazioni diverse
