### Scrivere sistema

#### Tabella di variazione

**Come si scrive**

1- se non specifica altrimenti, solo i reagenti hanno una concentrazione iniziale, i prodotti no  

2- variazione: i reagenti hanno -x, i prodotti +x, ogni specie ha come coeff il suo coeff stechiometrico  

non importa se sono solidi o gassosi qui  
quello mi importerà solo per scrivere K, alcune celle non mi interesseranno  

**Come si usa**

la tabella si scrive uguale, poi per trovare le concentrazioni all'equilibrio:   

1- reagente è debole:   
scrivi K, pluggaci i valori all'equilibrio, e risolvi per x  

2- reagente è forte:    
usa tutto il reagente possibile, portalo a 0 (attenta al reagente limitante)  
in questo caso, la situzione di equilibrio è quando la reazione ha finito di avvenire  
è come avere $K=\infty$  

se il testo non mi dice il valore di $K_{ps}$ per una molecola, probabilmente è perchè è forte (cose forti sono basi con X(OH) e acidi con alogeni HCl, eccetto qualcosa???)  

##### Esempi

**Esempio: reagente debole**

$Ag_2​SO_4​(s)⇌2Ag(aq)+SO_4^{2−}​(aq)$   

$Ag_2​SO_4​(s)$ ⇌ | $2Ag(aq)$ | $+SO_4^{2−}​(aq)$ 
---|---|---
$c_0$ -x | +2x | +x  

$K_{ps} = [Ag^+]^2 [SO_4^{2-}] = (2x)^2 \, x$  

**Esempio: reagente debole**

$HA +H_2O_{(l)} ⇌ H^+ + A^-$  


$HA$ | H2O  ⇌ | H+ | A- 
---|---|---|---
$c_0$ -x | $b_0$ -x | +x | +x   

$K_a = \frac{[H^+][A^-]}{[HA]} = \frac{x \, x}{c_0-x}$  

note:  
$b_0$ è enorme, è per questo, che i liquidi li ignoro quando scrivo K  
se $K_a$ è molto piccolo, allora ci sono pochi prodotti, cioè $x$ è piccolo, quindi posso semplificare $c_0 -x \sim c_0$  

**Esempio: reagente forte**

cioè $[Na]=2[NaSO]_0$ e $[SO]=[NaSO]_0$  
di solito chiamo $[NaSO]_0 = c_0$

#### K di reazione

**How to write**

$aA + bB  ⇆ cC + dD$   
$K= \frac{[C]^c[D]^d}{[A]^a[B]^b}$  

**IMPORTANTE**

considera solo le sostanze:  
1- GASSOSE (con (g) in basso)  
2- o IN SOLUZIONE (con (aq) in basso)   

ometti le altre:  
considerale per bilanciare la reazione, ma poi non metterle nella formula per K  
bc if the concentration of something is constant, can delete it from the equality, assume it's inglobated in the K  

per questo in $K_{ps}$ dei sali non c'è denominatore, perchè il reagente è il sale, che è solido  
stessa cosa per $K_w$, prodotto ionico dell'acqua  

**Esempi**

esempio 1:  
$A_{(l)} + 2B_{(g)} \rightarrow C_{(g)} + 3D_{(g)}$  
$K_c = \frac{[C][D]^3}{[B]^2}$

esempio 2:  
$2CaSO_{3(s)} + O_{2(g)} ⇆ 2CaO_{(s)} + 2SO_{3(g)}$  
$k = \frac{[SO_3]^2}{[O_2]}$


**Tipi di K**

$K_c$: usa concentrazioni molari $C_i = \frac{n_i}{V_{tot}}$  
$K_x$: usa frazioni molari $ x_i=\frac{n_i}{n_{tot}}$  
$K_p$: usa pressioni parziali $p_i = p_{tot} \frac{n_i}{n_{tot}} = \frac{n_iRT}{V}$  

le formule si scrivono allo stesso modo di K_c, con gli stessi esponenti, ma invece di [C] metto p oppure x  
esempio: $K_p = \frac{P^2_{HI}}{P_{I^2}P_{H^2}}$  

passaggi:  
$\Delta n $: la differenza tra la somma dei numeri stechio di reagenti e prodotti SOLO GASSOSI (o AQ???), NON tra i valori delle moli all'equilibrio   
$K_p = K_c (RT)^{\Delta n}$  
$K_p = K_x P_{tot}^{\Delta n}$   
per ricordarle, guarda la formula delle pressioni parziali, riconoscici la concentraazioni e la frazione molare, al resto metti $\Delta n$ come esponente    

**K_a e K_b**

sono costanti di equilibrio per acidi e basi  
K_a: reagente è un acido  
K_b: reagente è una base  

per una coppia acido-base coniugata, il loro prodotto è K_w  

a volte in un esercizio me ne viene dato uno, ma in realtà mi serve quell'altro, posso trasformarli  

#### Condizione di elettroneutralità

**Esempio**

$PbCl_2 ⇌ Pb^{2+}+ 2Cl^{-}$  
$PbBr_2  ⇌ Pb^{2+}+ 2Br^{-}$  

$2[Pb^{2+}]=[Cl^-]+[Br]^-$  

IMPORTANTE: il 2 viene perchè Pb^2+ ha una carica 2+ in alto, NON perchè Pb appare in due equazioni di dissociazione  

**Condizione di elettroneutralità**

cosa:  
In una soluzione elettricamente neutra, le cariche positive devono bilanciare quelle negative  

come si scrive:  
$\Sigma q \,  [ione +]​ = \Sigma q \, [ione -]$  
i coeff $q$ NON sono i coefficienti stechiometrici, ma sono il numero di ionizzazzione, cioè il numero di cariche in più o in meno, scritto in alto a destra   
considera solo gli ioni a destra, cioè nella parte dissociata  


### Domande particolari

#### Sale binario

**Cosa**

voglio sapere il pH di una soluzione  
ma la cosa che sciolgo non ha tra i suoi ioni nè H nè OH  
invece, si scioglie in due elementi, che poi reagiscono con l'acqua e cambiano il pH  

**Esempio**

sciolgo il sale:   
$NaF \rightarrow Na^+ + F^-$  

entrambi gli ioni interagisocno con l'acqua:    
$Na^+ + H_2O \leftarrow NaOH + H^+$  
$F^- + H_2O ⇌ HF + OH^-$  

so che NaOH è una base forte, quindi l'equilibrio della prima è tutto a sinistra, dove non c'è niente che cambia il pH  
invece HF è una base debole, quindi ho $⇌$, quindi ci sono un po' di $OH^-$  

quindi il mio esercizio è semplicemente di calcolare il pH creato dalla seconda reazione  
e so che $[NaF]=[F^-]$, è il sale sciolto all'inizio  
$K_b = \frac{[HF][OH^-]}{[F^-]} = \frac{x^2}{c_0-x}$


**Recap**

interagiscono con l'acqua solo gli ioni debole  
cioè quelli il cui coniugato è forte  

#### Precipitazione

**Domande**

1- si forma precipitato?  
guarda se $Q > K_{ps}$ (cioè troppa roba)  
dove Q, il quoziente di reazione, è la formula di K con le concentrazioni al MOMENTO ATTUALE (cioè quelle iniziali)   

2- quanto se ne forma?  
scrivi tabella di variazione in modo che all'equilibrio ci sia $K_{ps}$  
la varazione è il numero di moli di precipitato  

**Quantità di precipitato**

$AgIO_3 ⇌ Ag^+ + IO_3^- $  
$K_{ps} = \frac{(n_{Ag}-x)(n_{IO} -x)} {V_{tot}^2}$  

x è la quantità di moli percipitata  
posso anche scrivere la formula con la concentrazione, e poi passo alla quantità di moli  

concentrazioni finali:  
$[Ag^+]=[Ag]_0 -\frac{x}{V_{tot}}$  

**Definizione**

se ho una soluzione con dentro un sale, in concentrazione maggiore del suo limite di solubilità $K_{ps}$ (cioè in condizioni di oversaturazione)  
allora tutto il sale non può essere disciolto (perchè supera il limite), un po' è allo stato solido, e si chiama 'precipitato' o 'corpo di fondo'  

formale:  
precipitazione: il fenomeno della separazione sotto forma di solido di un soluto (detto precipitato) che si trova in una soluzione in concentrazione maggiore del suo limite di solubilità  
per cui la soluzione, prima che sopraggiunga la precipitazione, si trova in condizioni di sovrasaturazione rispetto a tale soluto  

#### Autoprolisi dell'acqua

**Cosa**

cosa:  
diventa l'esercizio di due sostanze che si dissociano insieme  
la seconda è l'acqua, e ha:   
$H_2O ⇌ H^+ + OH^-$  
$K_w = [H^+][OH^-] = 10^{-14}$  

quando:  
se sto studiando il pH, e l'altra sostanza che si dissocia ha una molarità molto bassa, dell'ordine di $10^{-7}$  
quindi i protoni ottenuti dalla dissociazione dell'acqua non sono trascurabili  

#### pH

**Domande**

Quanto soluto per avere uno specifico pH? = quale pH per sciogliere questa quantità di soluto?  

se voglio un valore preciso di pH, allora so il valore di $[H^+]$ che voglio  
risolvi come per trovare concentrazioni all'equilibrio, ed eguagliala al valore richiesto  

**Calcolare pH**

$pH = - log[H^+]$  
$pOH = -log[OH^-]$  
$pH = 14- pOH = 14 + log[OH^-]$  


**ATTENTA**

in chimica usa sempre LOG, non ln  

#### Tampone

**Cosa**

una soluzione con acido e base coniugata  
tipo, $NH_3$ e $NH_4^+$, oppure $HF$ e $F^-$       
cioè, due cose che si differenziano per una H  

così, posso scrivere una reazione in cui uno dei due reagisce con l'acqua, e ottengo l'altro come prodotto  
(attenta a cosa meto come reagente, e quindi se scegliere K_a o K_b)  

**Come si risolve**

come al solito, con la tabella di variazione  

sennò, c'è una shortcut, che sarebbe la legge di Henderson–Hasselbalch:  
$pH = pK_a + log \frac{[A^-]}{{HA}}$   
dove $pK_a = -log[K_a]$  

comunque, se scrivo la tabella di variazione, ottengo lo stesso risultato  

**Nota**

considero tamponi solo quelli dove HA e A^- hanno più o meno la stessa concetrazione, cioò 0.1 <[HA]/[A^-] < 10  

se uno è molto maggiore dell'altro non è un tampone   
comunque il procedimento non cambia, fai sempre la tabella di variazione  

**Diluzione**

se la diluisco, non cambia il pH  

si vede, se moltiplico per lo stesso numero 0.05 e 0.1, x non cambia  

##### Esempio

**Esempio**

$HF ⇌ H^+ + F^-$  

HF| H+ |  F- 
---|---|---
0.1-x | +x | 0.05+x  

$K_a = \frac{[H^+][F^-]}{[HF]} = \frac{x \, (0.05+x)}{0.1-x}$    
uso K_a perchè ho scelto l'acido come reagente (conosco il valore di K_a$)  

poichè $K_a$ è molto piccola, anche x lo è, approssimo: $K_a = \frac{0.05}{0.1}x$  

ora posso calcolare: $pH = - \log x$  

**Nota**

in teoria, la reazione di un acido con l'acqua è:  
$HF+H_2​O⇌H_3​O^++F^−$  
ma metto sempre $[H^+]=[H_3O^+]$  

**Note**

1- come ritrovo Henderson:  
$pH = - \log x = -\log(K_a \frac{0.1}{0.05}) = -log (K_a) + \log(\frac{0.05}{0.01})$  

2- ho usato la reazione con l'acido come reagente  
per usare quella della base,  quindi $K_b$, avrei dovuto fare:  
$F^- + H_2O⇌ HF + OH^-$   
e quindi calcolare la concentrazione di OH-

#### Elettrodi

**Domanda**

trova la ddp  

come:  
le due dissociazioni sono SEPARATE, non nello stesso liquido, quindi non c'è condizione di elettroneutralità  
trova le due concentrazioni e mettile nella formula di Nernst  

**Esempio**

ho pila basata su Ag, i sali sono $AgCl$ e $AgI$   

1- trovo concentrazioni:     
$[Ag^+]_{Cl} = s_{Cl} = \sqrt{K_{ps}(AgCl)}$  
$[Ag^+]_I = s_I = \sqrt{K_{ps}(AgI)}$    
 
2- formula di Nernst  
$\Delta E = \frac{0.059}{1}\log\frac{[Ag^+]_{catodo}}{[Ag^+]_{anodo}}$   

nota: devo avere $\Delta E$ positivo, quindi nel log metto sopra la concentrazione maggiore  
cioè quella con esponente più alto (considera che sono negativi, quindi più baddo in valore assoluto)  

**Altra versione**

mi dà una sola molecola che si dissocia  
e mi dà $E_0$, il 'potenziale di riduzione standard dello ione ...'  

usa quest'altra formula:  
$E = E_0 - \frac{0.059}{1}\log[Ag^+]$    

**Spiegazione**

def:  
misura la tendenza di un elemento a ridursi od ossidarsi rispetto all’idrogeno standard    

esempio:  
$Cu + 2H^+ \leftarrow Cu^{2+}+H_2$  
$Zn + 2H^+ \rightarrow Zn^{2+}+H_2$  
nella prima, è Cu che tende a ridursi di più di H (tende al lato dove Cu ha numero di ossidazione minore), quindi $E_0>0$ per Cu, si riduce meglio di H  
nella seconda, H si riduce meglio di Zn, quindi $E_0 <0$ per Zn  

nota:  
non importa che entrambi prendano carica positiva, conta se fanno meglio o peggio di H  

**Elettrolisi**

elettrolisi: Ora voglio forzare la reazione al contrario, cioè fare depositare Ag sul catodo

Per far avvenire l’elettrolisi devo superare la forza elettromotrice opposta della pila  
ho calcolato $\Delta E$ della pila, serve una tensione maggiore di quella:  
$E_{applicata} > \Delta E$  

**Attenzione**

$pH = pK_a + log \frac{[A^-]}{{HA}}$  
quando: tampone  
nome: Henderson–Hasselbalch  
A e HA sono base e acido coniugati  

$\Delta E = \frac{0.059}{1}\log\frac{[Ag^+]_{catodo}}{[Ag^+]_{anodo}} =  E_0 - \frac{0.059}{1}\log[Ag^+]$    
quando: pila  
nome: Nernst  
le cocentr sono anodo e catodo    

la pila ha -, il tampone +

**Steps**

domanda: Tempo per depositare 1g di Ag, con una corrente I nota

### Risolvere sistema

**Steps**

1- sempre, preliminari  
per ogni reazione, trattala come fosse da sola  
scrivi la tabella di variazione  
poi se è forte calcola valori all'equilibrio, se è debole scrivi K (SOLO con nomi delle specie per ora, non sostituire la x della variazione)  

2- SENZA elettroneutralità  
nelle formule di K, come concentrazione delle sostanze metti la somma delle concentrazioni che appaiono nelle due tabelle (per specie che appaiono in entrambe le reazioni)  

3- CON elettroneutralità  
chiama $s$ la concentrazione della sostanza in comune tra le due reazioni  
le altre concentrazioni, dei PRODOTTI non in comune, trovale invertendo la K (se reazione debole) o direttamente dalla tabella (se reazione forte)  
nota: se nella K appare anche un REAGENTE, scrivilo come $c_0-[...]$, usando la concentrazione della specie che apapre SOLO in questa reazione, non in quella in comune  
scrivi la formula di elettroneutralità e ostituisci queste cose  

**Quando usarlo**

la condizione di elettroneutralià è una scorciatoia per calcolare la quantità totale dello ione in comune  
mi dà un sistema più semplice per calcolare (x+y), invece di risolvere per x e y, e poi sommare  

cioè, se ho due reazioni deboli e con UN SOLO REAGENTE, con la condizione di elettroneutralità è molto più facile  
se almeno una è forte, vanno bene entrambe  

#### Esempio

**Preliminari**

$PbCl_2 ⇌ Pb^{2+}+ 2Cl^{-}$  
$PbBr_2  ⇌ Pb^{2+}+ 2Br^{-}$  

$K_{ps}(PbCl_2) = [Pb^{2+}][Cl^{-}]^2$   
$K_{ps}(PbBr_2) = [Pb^{2+}][Br^{-}]^2$  

**CON elettroneutralità**

1- specie in comune: $[Pb^{2+}]=s$    

2- altre concentrazioni  
$[Cl] =  \sqrt{\frac{K_{ps}(PbCl_2)}{s}}$  
$[Br] = \sqrt{\frac{K_{ps}(PbBr_2)}{s}}$    

3- risolvo  
$2[Pb]=[Cl]+[Br]$  
$2s = \sqrt{\frac{K_{ps}(PbCl_2)}{s}} +\sqrt{\frac{K_{ps}(PbBr_2)}{s}}$

$HCl → H^+ + Cl^-$

reazione forte: calcola valori finali  
HCl →|  $H^+$ | $Cl^-$

1- metto le concentrazioni delle specie (H appare in entrambe, sommo le concentrazioni)  
$K_w = [H^+][OH^-] = (c_0+y)y$  
 
2- risolvo:  
$y = \frac{-c_0+\sqrt{c_0^2+4K}}2$  

#### Esempio 3

**Preliminari**

$H_2O_{(l)} + HA  ⇌  H^+ + A^-$  
$H_2O_{(l)}  ⇌  H^+ + OH^-$  


H2O | HA  ⇌|  $H^+$ | A
---|---|---|---
$a_0$ -x| $c_0-x$ | x | x   

$K_a = \frac{[H^+][A^-]}{[HA]}$  

$H_2O$ ⇌ | $H^+$ | $OH^-$
---|---|---
$b_0$ -y | y | y

$K_w = [H^+][OH^-]$  

**SENZA elettroneutralità**

$\begin {cases} 
K_a  = \frac{(x+y)x}{c_0-x} \\
K_w = (x+y)y \\
\end{cases}$  

è difficile da risolvere  

**CON elettroneutralità**

1- specie in comune: $[H]=s$  

2- concentrazioni degli altri prodotti:    
$[OH] = \frac{K_w}{s}$  
$K_a = \frac{s [A]}{c_0 - [A]}\quad  \rightarrow \quad [A] = \frac{c_0K_a}{s+K_a}$  
esprimo il reagente HA in funzione di A, il prodotto che è SOLO in queta reazione  

3- sostituisco  
$[H]=[A]+[OH]$  
$s =  \frac{c_0 K_a}{s+K_a} + \frac{K_w}{s}$
