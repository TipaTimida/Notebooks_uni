### Arrhenius

**Equazione di Arrhenius**

$k=Ae^{-E_a/RT}$  

1- significato dei parametri:  
k: costante cinetica (quella delle legge cinetica, velcotià di reazione)    
T: temperatura  
R: costante dei gas  
$E_0$: energia di attivazione  
A: fattore pre-esponenziale (probabilità di collisione “giusta”) (sezione d'urto, misura la probabilità che le molecole possedendo l’energia richiesta diano luogo ad un urto reattivo)  

2- dipendenza di k da T: 
se T aumenta, k aumenta   
perchè se T aumenta, aumenta il contenuto energetico delle molecole, ovvero aumenta il numero degli urti efficaci, rendendo più veloce la trasformazione

**Temperatura di inversione**

Se una reazione diretta e una inversa hanno dipendenze diverse dalla temperatura, può esistere una temperatura in cui le due velocità diventano uguali: temperatura di inversione  

È la temperatura alla quale un gas reale, soggetto a espansione adiabatica (a entalpia costante) senza compiere lavoro, passa da un processo di raffreddamento a uno di riscaldamento.    

È la soglia termica che determina se un gas si riscalda o si raffredda quando viene fatto espandere liberamente.
