### Struttura di Lewis

**Steps**

1- disegnare lo scheletro della molecola (senza le H)  
2- conta elettroni rimasti  
3- usali per completare ottetti di atomi più elettronegativi (esterni)  
se necessario, fai doppi legami per finire ottetto di atomo centrale  

**Come gestire le H**  

non attaccarle all'inizio, disegna solo lo scheletro della molecola senza un protone (se ho $HCO_3$, disegna $CO^-_3$)  
costruisci bene la molecola, vedi dove sono legami multipli, quindi dove posso attaccare H  

**1. Disegnare scheletro molecolare**

cosa:  
cioè attaccare tutti gli atomi tra loro, con un legame singolo (no doppi)  
quindi solo vedere chi è connesso a chi  

tips:  
1- atomo centrale di solito è il meno elettronegativo, cioè quello a cui mancano più elettroni di valenza  
2- simmetria: le strutture sono spesso simmetriche, quindi se ho un atomo di cui c'è una sola copia probabilmente sta al centro  (ma elettronegatività ha precedenza, come in $N_2O$)  
3- non deve essere chiusa, devo solo attaccare tutti gli atomi  
4- trascura gli alogeni (7 elettroni di valenza) e l'idrogeno, aggiungili alla fine, tanto possono avere un solo legame, non estendono la catena  

**2. Conta elettroni**

1- elettroni di valenza    
come: somma degli elettroni di valenza di tutti gli atomi (per ogni atomo, numero di valenza è il numero di colonna nella tavola)  
se la molecola è un ione, aggiungi anche quello (tipo se ha carica 1-, aggiungi un elettrone)  

2- elettroni usati nello scheletro  
sottrai 2 per ogni legame singolo nello scheletro 

**3. Completa ottetti**  

1- assegna doppietti ad atomi più elettronegativi (esterni)  
2- se non ne ho abbastanza per completare ottetto di atomo centrale, crea quanti legami doppi servono  
un doppietto di un atomo confinante può diventare un legame doppio, così conta come 2 elettroni per ENTRAMBI gli atomi  
quindi trasformare doppietto in legame non modifica l'ottetto dell'atomo proprietario del doppietto, semplicemente aggiunge due elettroni all'atomo che riceve  

posso trasformare un doppietto assegnato ad un atomo in un legame  
in tal caso vale come un doppietto per entrambi gli atomi che unsice  

#### Note

**Attenzione: quanti elettrone conta un legame**

dipende da cosa sto contando:  

1- elettroni totali della molecola:  
legame singolo=2e, li conto una sola volta perché appartengono alla molecola  

2- controllo ottetto di un atomo  
legame sigolo=2e, PER OGNI ATOMO  
perchè: ogni atomo considera entrambi gli elettroni del legame come se fossero suoi  
cioè gli stessi elettroni vengono contati per entrambi gli atomi, ma non sono duplicati fisicamente  

3- carica formale  
legame singolo=1e   
perché: nella carica formale si assume che il legame sia diviso equamente tra i due atomi

### Polarità

![alt text](https://www.researchgate.net/publication/357151895/figure/fig5/AS:1102290060611596@1639817838412/Structural-formulae-of-carbon-dioxide-on-the-left-hand-and-water-on-the-right-hand-The.png)

**Recap**

una molecola è apolare solo se la somma vettoriale dei dipoli di legame è zero, cioè:  
1- tutti i legami A–X sono uguali  
2- la geometria è perfettamente simmetrica (considerando anche gli angoli di legame)

**Metodo lungo**

1- disegna bene la geometria molecolare, considerando gli angoli di legame  
non serve ricordare il nome della struttura geometrica, posso disegnarla a occhio ricordando le regole sugli angoli di legame (dppietto libero > legame multiplo > legame singolo)   

2- per ogni coppia di atomi legati, disegna il dipolo di legame (freccia)  
nota che coppie uguali hanno frecce di lunghezza uguale, coppie diverse di lunghezza diverse, non ben specificata  

3- fai la somma vettoriale dei dipoli di legame, se viene 0 è apolare  

**Shortcut**

se tutti atomi ugauli (a parte il centrale):  
apolare se E=0, sennò polare  
altri casi di apolare: lineare ($X_2E_3$), non so nome ($X_4E_2$), 

se ci sono atomi diversi:  
polare  

**Soluzioni**

biatomiche: polare solo se sono atomi diversi (se il legame è polare)  

triatomiche ($AX_2$):  
angolata ($E=1,2$): polare  
lineare ($E=0,3,4$): se legami sono uguali (le due molecole X sono dello stesso tipo) è apolare, se diversi polare   

tetraatomiche ($AX_3$):  
triangolare piramidale o T ($E=1,2,3$): polare  
triangolare planare ($E=0$): sd tutti gli X sono uguale apolare, sennà polare  

pentaatomiche ($AX_4$):   
se non tutti legami uguali: polare  
se tutti legami uguali (tutti X uguali):   
$E=0,2$: apolare  
$E=1$: polare   

esaatomiche ($AX_5$):  
se non tutti legami uguali: polare
se tutti legami uguali (tutti X uguali):  
$E=0$: apolare  
$E=1$: polare  

eptatomiche ($AX_6$): 
se non tutti legami uguali: polare
se tutti legami uguali (tutti X uguali):  
$E=0$: apolare 


### Check: carica formale

**Calcola la CF - carica formale**

per ogni atomo, calcola la sua CF così:  
CF = elettroni di valenza - elettroni assegnati  
CF = (numero del gruppo) - 2(numero doppietti) - (numero legami)  
CF =  (numero del gruppo) - elettroni non condivisi - 1/2 elettroni condivisi

spiegazione:  
1-  gli elettroni di legame soono assegnati metà a ciascuno dei due atomi legati  
quindi 1 legame = 1 elettrone  

2- entrambi gli elettroni di un doppietto libero sono associati all'atomo su cui si trovano  

**Come usare la CF**

CF della molecola = somma delle CF di tutti i suoi elettroni  
deve essere uguale alla carica della molecola  
cioè se è neutra 0, sennò alla carica dello ione  

è giusto? o è meglio, ma non necessario???  

**Indecisione**

se più versioni rispettano la CF della molecola, quale scegliere:  

1- quella in cui più atomi hanno CF=0  
o se non c'è, quella con le CF di modulo minore  

2- le cariche formali di atomi adiacenti dovrebbero essere di segno opposto  

3- quella dove le cariche formali più negative sono sugli atomi più elettronegativi, e viceversa  

### Esempi

#### CO_2

**Struttura di Lewis**

1- disegno scheletro  
O - C - O  
perchè rispetto simmetria  

2- conto elettroni:  
di valenza: 4+(6x2)=16  
disponibili: 16-2x2=12  

3- completo ottetti:  
O esterni hanno 2 elettroni, gliene dò 6 a testa  
ho finito elettroni disponibili, ma C ha ancora solo 4 elettroni  
quindi gli creo due legami doppi (uno con ogni O), così tutti gli ottetti sono completi     
O = C = O  
perchè se facessi O–C≡O avrebbe cariche formali più grandi, non va bene  

**Geometria molecolare**

atomo centrale: C  
doppietti liberi: 0  
atomi legati: 2  

geometria: $AX_2$   
nome: lineare  
numero sterico: 2+0=2  
ibridazione: sp  

**Polarità**

poichè è lineare, e i due legami sono uguale, i momenti di dipolo dei due legami si annullano  
quindi è apolare  

#### HNO_3

**Struttura di Lewis**

1- scheletro:  
O-N-O-H  
O /

2- conto elettroni:

in totale ne ho usati 16, e N ha ancora solo 6 elettroni  
quindi gli devo creare un legame doppio con uno degli ossigeni  
uno qualunque va bene, ci sono 3 forme di risonanza

### TIps ottetto

**Troppi o pochi elettroni**

1- ottetto incompleto   
quand: atomo centrale appartiene ai gruppi 1,2,3, quindi non può fare abbastanza legami da completare l'ottetto    
eg. $BF_3$, non posso creare legami doppi perchè F è alogeno, può fare solo un legame singolo???  

2- ottetto espanso    
quando: atomo centrale appartiene al periodo 3 o più grande, quindi può ricevere più di 8 elettroni di valenza  
eg: $PCl_5$, $SF_6$   
se possono, lo fanno, infatti il fosforo allo stato elementare è $P_4$, non $P_2$ 

3- numero dispari di elettroni  
un atomo deve finire con 7 elettroni, non 8  
è l'atomo meno elettronegativo, perchè non riesce ad ottenere l'ultimo  
questo singoletto lo conto come un doppietto quando scrivo $AX_nE_m$  
eg. NO, $NO_2$, l'atomo con 7 è l'azoto  

**Risonanza**

non cambia lo scheletro della molecola, ma solo dove si trovano i legami multipli  
eg. $O_3$, $N_2O$ è particolare  
però ad esempio $CO_2$ non ha risonanza, posso solo dividerli equalmente tra gli atomi vicini ???

casi strani:  
1- HCNO e HNCO non sono forme di risonanza, ma isomeri (stessi atomi ma disposti diversamente)  
2- $N_2O$, la forma col legame triplo con O è più rara, la posso ignorare (ricorda che N-N fa legame triplo, va bene anche doppio, ma non singolo)  

**C'è risonanza?**  

1- se un atomo ha 2 vicini e deve fare 3 legami (dispari), c'è per forza  
eg: $NO_2$  

2- se un atomo ha 2 vicini e deve fare 4 legami (pari, può esserci o no):  
c'è: se è attaccata a due atomi diversi, come $N_2O$ (N-N-O)  
non c'è: se attaccata a due atomi uguali, come $CO_2$, loh a bilanciato O=C=O  

**Elettroni totali dispari**

meglio finire con un elettrone in meno che uno in più  
consideralo quando conto quanti legami fare  
si vede ad esempio in NO, non fare un terzo legame, o O avrebbe un elettrone in più  
invece, è meglio che N ne abbia uno in meno  

l'elettrone singolo va su N, perchè è meno elettrenegativo di O, quindi O ha la precedenza per completare l'ottetto  
nota: per il calcolo del numero sterico, l'elettrone spaiato vale come un doppietto 

### Tips scheletro

**Stella o catena?**

guarda quale è l'atomo meno elettronegativo, da mettere al centro  

1- se ce n'è due o tre copie: probabilmente avrò una catena, per farlo simmetrico   
eg. $N_2O_2$ (O-N-N-O)  
rg. $C_2O_5$ (O-C-C-C-O)  

2- se ce n'è una sola copia:    
ci attacco tutti gli altri a stella  
eg. $CO_3$, $COF_2$, C al centro e gli altri 3 attaccati a C   
eg. $SOCl_2$, S al centro e il resto a stella  

eccezione: $S_2O_3$, metti un S al centro e ci attacco tutto intorno a stella  

**Note stella**

1- ordine degli atomi intorno  
non importa se gli atomi uguali sono contigui o opposti, la geometria non cambia  
eg $CCl_2F_2$ ha C al centro, ma gli altri 4 intorno possono stare in qualunque permutazione  

**Tanti alogeni (penultima gruppo)**

eg. $IF^{-4}, IF^3, I^-_3, SeF_4$  

cosa: se ce n'è uno diverso, mettilo al centro, e attacca a lui tutti gli alogeni, non importa quanti sono  
se non c'è n'è uno diverso, metti al centro uno degli alogeni, e attacagli tutti gli altri (sempre a forma di stella/asterisco)  

poi completa gli ottetti degli alogeni (6e ad ognuno), gli elettroni rimasti mettili sull'atomo centrale, anche se va molto oltre l'ottetto  

caso: se la molecola ha $^+$ oppure $^-$  
conta -1 o +1 elettroni di valenza  
e dopo che ho finito di disegnare e assegnare gli elettroni (anche qeusto qui), aggiungi parentesi quadre con $[]^+$ o $[]^-$ intorno alla molecola  
sembra di contarlo due volte questa carica, perè è così  
questa cosa della parentesi vale solo per gli alogeni, non per le altre molecole, tipo $N_2^-$, perchè??? anzi, in $NO_3^-$ sì???  

isolettronica:  
se molecola ha $^+$ o $^-$: assegnalo all'atomo al centro, vedi cosa diventa (P con $^+$ diventa $Si$, va indietor di una casella, oppure $C$, va bene uno qualunque della colonna giusta) (I con $^-$ divetna Xe, fa avanti di una casella)    
se molecola NON ha  $^+$ o $^-$: fai il contrario, se gliela assegno torno indietro di una casella ($IF_3$ isoelettronico a $SF^-_3$)  

gli alogeni NON fanno doppi legami  

**3 alogeni**

catena di 3  
quello al centro ha 2 legami singoli (uno con ugnuno degli altri) e 3 doppietti ( quindi espande l'ottetto)  
quindi $AX_2E_3$  

anche in quelli a stella, assegna abbastanza elettroni da finire l'ottetto di quelli esterni, poi dai i rimanenti a quello centrale, espandigli l'ottetto, così anche quello ha dei doppietti  
una forma comune è $AX_4E$, altalena, oppure $AX_4E_2$, planare quadrata  

**Combinazioni N,O**

1- $N_2 O_x$  
x pari: N attaccati tra loro, metti gli O simmetrici (attaccati tutti alle N, non tra loro) (basta ricordare che in $N_2O$ sono attaccati)   
x dispari: fino a x=3 N attaccati, x=5 in poi N separati da una O (così diventa simmetrico)  

**Combinazioni S,O**

1- $S_2O_x$  
x è sempre dispari???  
fino a x=5: S attaccati (mentre con N, già a x=5 sono separati)   
da x=7: separati da un O   

nota:   
a x=5 la S resta unita, perchè 5,S sono simili     
mentre la N è già separata a 5, con una O in mezzo alle N  

**Tanti O**

1- non si legano mai tra loro (a meno che non ci sia solo O)  
quindi si attaccano alle estremità della molecola, se possibile in modo simmetrico  
eg. $N_2O_3$, tutti gli O somno legati alle N 
eg. $NO_3$: tutti gli O sono leati alla N  
eg. $C_2O_4$: tutti gli O sono legati alle C

1- costruisci la molecola SENZA attaccare la H  
conta comunque il suo elettrone, come se la molecola avesse $^-$  

2- se c'è un ossigeno con un legame singolo, attaccaci la H   
solo una, perchè un ossigeno fa al massimo 2 legami 

3- sennò, attaccalo a C o N, quello più ESTERNO  

eccezione: HCO_2  
la H è attaccata alla C a stella  

se ci sono tanti O a cui attaccarsi: si attaccano alle estremità dell catena più lunga   
eg. $H_2S_2O_7$ (se c'è una sola H, mettila sempre all'estremità della catena lunga, ma ad una sola estremità)

### Esempio

**$HCO_3^-$**

1- ignoro la H, ho $CO_3^{2-}$

distribuendo 
assegno elettroni di valenza: un ossigeno ha doppio legame, due hanno singolo  

2- attacco la H:  
va ad uno degli ossigeni con legame singolo  
ho due opzioni, quindi strutture di risonanza  

**HNCO**

1- inizio da $NCO^-$  
C è il meno elettronegativo, quindi N-C-O  
ho due strutture di risonanza, N≡C-O oppure N=C=O   
in teoria, O non dovrebbe poter fare legame doppio, ma lo fa??? è un dativo??

2- casi:
N≡C-O, allora H si attacca alla O  
N=C=O, allora H si attacca alla N (perchè C ha già fatto i suoi 4 legami, mentre N, che potrebbe farne 3, ne ha fatti solo 2) 
sono due molecole diverse, due isomeri, NON strutture di risonanza

### Altro modo

**Steps**

conta:  
1- Ev = elettroni di valenza della molecola  
come metodos corsa, somma tutti i numeri di colonna    
2- Etot = elettroni per completare tutti gli ottetti (o duetto per H)  

calcola:  
3- $N = \frac{E_{tot}-E_v}2$  
numero di legami nello scheletro  
4- $doppietti = \frac{Ev-2N}2$  
numero di doppietti da sistemare  

disegna:  
4- scheletro con legami singoli (anche le H, poi aggiungi legami doppi fino ad arrivare a N)  
5- doppietti rimasti mettili intotno agli atomi   

**Esempio $HNO_2$**

1- conta:  
$E_v = 1 + 5 + 6x2 = 18$  
$E_{tot} = 2 + 8x3 = 26$  

2- calcola:  
$N = \frac{26-18}2 = 4$  
$dopp = \frac{18 - 2x4}2 = 5$    

3- disegna:  
metti i legami singoli per fare O-N-O-H  
manca un legame, fa un doppio: O=N-O-H  
attacca i 5 doppietti per completare gli ottetti
