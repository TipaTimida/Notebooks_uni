### Gas perfetti

**Ipotesi**

1- molecole puntiformi  
volume delle molecole trascurabile rispetto al volume occupato dal gas (bassa densità)  

2- particelle indipendenti  
ignoro gli urti tra le molecole, tanto c'è bassa densità  

le ipotesi valgono solo per V alta e P bassa (bassa densità e pochi urti)  


**Equazioni**

equazione dei gas perfetti:  
PV=nRT  

equazione di Van der Walls (gas reali):  
$(P+a\frac{n^2}{V^2})(V-nb)=nRT$   

a: probabilità di urti fra le molecole  
corregge le forze attrattive tra le molecole  
Gli urti fra le molecole riducono gli urti con le pareti, quindi riduce P   

b: covolume, volume occupato fisicamente da una mole di gas  
corregge il volume occupato realmente dalle molecole  
Ogni gas ha un suo valore di b, non é costante universale  

**Isoterma critica**

plotta $P(V)$ di una sostanza, a varie temperature T  
quando lo plotto alla temperatura $T_c$ (diversa per ogni sostanza), detta temperatura critica, appare un flesso a tangente orizzontale, detto "punto critico"    
la linea si chiama "isoterma critica", separa due regioni del piano PV  

regione sopra: $T>T_C$, supercritica  
gas non può liquefare per compressione  

regione sotto; $T< T_c$, subcritica   
il gas può liquefare per compressione  
quando è in questo stao, lo chiamo "vapore", non gas  

il punto critico è la coordinata $(T_c, P_c V_c)$ del punto a tangente orizzontale

**Ricorda**

nel modello di van der Waals:  
$V_{mc}=3b$  

$V_{m,c}$: volume molare critico  
volume occupato da una mole della sostanza nel punto critico  

**Cosa succede in una subcritica: passaggio vapore -> liquido**

guarda il grafico, da destra verso sinsitra 

1- comprimo il vapore  
cioè diminuisco V, e aumenta P    
tratto da destra a B  

2- avviene la liquefazione  
è il tratto orizzontale (da B ad A)  
durante passaggio di stato, V diminuisce (V_liq < V_gas), ma T e P restano costanti  

3- comprimo il liquido    
Il liquido è quasi incomprimibile, quindi per diminuire ancora un poco il volume serve un forte aumento della pressione  
Per questo la curva diventa molto ripida  

problema: secondo l'equazione di Vand er walls non dovrebbe esserci il tratto orizzontale, ma quello tratteggiato  
ma lì c'è un tratto in cui sia P che V diminuiscono, cioè comprimendo il sistema la pressione diminuirebbe, il che non ha senso

### Velocità molecole

**Pressione**

molecola sbatte sulla parete:  
$\Delta p = mv_x$  

frequenza degli urti sulla parete:  
se le molecole NON urtano tra di loro, ma solo sulle pareti di un cubo di lato L, allora il tempo tra due urti è:    
$\Delta t = L/v_x$   

forza sulla parete di una particella:  
$F = \frac{\Delta p}{\Delta t}= \frac{mv_x^2}L$  

forza media sulla parete di tutte le particelle:  
$F = nN_A \frac{<v>^2}{3L}$  
$N=nN_A$: numero di particelle  
non tutte le molecole si muovono alla stessa velocità, $<v>$ è quella media  
$<v_x^2> = \frac{<v^2>}3$, perchè v è ugualmente distribuita nelle 3 coordinate  

pressione:  
$P = N\frac{F}{S} = nN_A \frac{<v>^2}{3L^3}$  
$N=nN:_A$: numero di particelle

**Osservazioni: Energia cinetica**

energia cinetica di una mole di gas:  
$E = \frac 12 \, N_am \, <v^2>$  

sostituisco la v:  
$P = \frac 23 n E$  
$ E = \frac 32 RT$  
cioè E è proporzionale a T  

**Distribuzione di Maxwell-Boltzmann**



se aumento T, la curva si abbassa e si slarga  
perchè le molecole raggiungono v più alte, e può esserci più variazione  

**Velocità**

picco: $v_p$, velocità più probabile  
$<v^2>$ è il quadrato di $v_{rms}$, che è più a destra di $v_p$  

a T uguale, due gas hanno la stessa E  
poichè $E = \frac 12 mv^2$, quello più pesante si muove più lentamente  
"legge di Graham": la velocità di diffusione è inversamente proporzionale alla radice quadrata della massa molare

### Atomo di H e funzione d'onda

**Spettri atomici**

eccito un gas -> non emette tutte le lunghezze d’onda possibili, ma solo alcune righe discrete  
si chiama "spettro caratteristico" dell'elemento  
L’esistenza di righe discrete indica che l’elettrone può avere solo determinati valori di energia  

quando l'elettrone passa da un livello all'alto, emette un fotone di energia uguale alla differenza di energia tra i due livelli:  
$\Delta E _{ph}= hf$, quindi a seconda della sua energia sarà una linea di colore diverso 

homo: orbitale occupato più alto in energia (dove ho messo l'ultimo elettrone di valenza)  
lumo: orbitale libero più basso in energia (il prossimo da riempire)  
La maggior parte delle transizioni elettroniche avvengono fra HOMO e LUMO


**Modello di Bohr**

l'elettrone può occupare solo alcuni livelli energetici  
cioè deve trovarsi in un orbitale, non può essere in mezzo tra due  

energia di un elettrone:  
$E_n = -\frac{Z^2R}{n^2}$
n: numero quantico principale     
Z: numero atomico dell'atomo (quindi lo stesso orbitale in atomi diversi ha diversa energia)
R: costante di Rydberg  

problema: modello di bohr va bene solo per l'idrogeno  
per tutto il resto, serve la dualità onda-particella dell'elettrone  

**Funzione d'onda**

l’elettrone è descritto da una funzione d’onda, $ψ(r
)$  
le grandezze osservabili sono sue funzioni, tipo il suo modulo al quadrato è la densità di probabilità di trovare l'eettrone nel punto r  

quindi:  
un orbitale non è un’orbita  
è una funzione d’onda  
oppure una regione dello spazio in cui è elevata la probabilità di trovare l’elettrone 


**Funzione d'onda di particelle**

principio di indeterminazione di Heisenberg  
$\Delta x \Delta p \ge \frac h2$   

massa maggiore -> minore incertezza sulla velocità  
quindi il comportamento ondulatorio del protone è molto meno evidente  
considero il protone come quasi immobile, senza funzione d'onda  

**Stato fondamentale**

un atomo è allo stato fondamentale se tutti gli elettroni occupano i livelli energetici più bassi possibili  
cioè non c'è niente di eccitato  

H è allo stato fondamentale se il suo unico elettrone si trova in 1s  
la prob di trovare l'elettrone è massima alla distanza $a_0$, chiamata "raggio di borh"  

**Sali colorati**

di solito tutti gli orbitali d sono degeneri  
ma in un metallo, circondato da altri atomi, questi altri atomi creano un campo eletricco che colpisce ogni orbitale d in modo divreso, visto che hanno orientazione diversa  
quindi alla fine hanno energie diverse, così posso fare transizione tra due orbitali d  

ligandi: una molecola o ione (acqua, carbonato, ammonio...) che si lega ad un catione metallico  
eg. $[Cu(H_2O)_6]^{2+}$  
ligandi diversi producono $\Delta E$ diverse tra gli orbitali  
quindi cambia il colore  


### Forze intermolecolari

**Cosa sono**

intermolecolari: tra molecole diverse  
stabiliscono il punto di ebollizione  

intramolecolari: tra atomi di una stessa molecola  

**Tipi**

1- ione-dipolo  
dipolo vuol dire molecola polare  
eg. $Na^+$ e $H_2O$  

2- dipolo-dipolo   
più la molecola è polare, più è forte  
tra dipoli permanenti  

3- forze di dispersione di London   
tra molecole NON polari  
Sono dovute a dipoli istantanei e indotti  
I nuclei di una molecola attraggono gli elettroni della molecola vicina, creano un diplo indotto  
è più forte quanto più la molecola è polarizzabile (cioè quanto più è grande, perchè c'è più prob che appaia un dipolo indotto)  

4- legame a idrogeno:  
È un’interazione dipolo-dipolo particolarmente intensa  
quando H è legato a F, O o N   
e interagisce con una coppia elettronica libera di F, O o N di un’altra molecola  

5- dipolo permanente–dipolo indotto 
tra molecola polare e apolare  

le "forze di Van der walls" è l'insieme di: dipolo-dipolo, dipolo permanente-dipolo indotto, London  

c'è anche una forza di repulsione, conseguenza del principio di Pauli, ma agisce solo a corto raggio  

**Curiosità: densità del ghiaccio**

di solito i solidi sono più densi dei liquidi  
ma il ghiaccio è molto ordinato per ottimizzare i legami ad idrogeno  
quindi è meno denso, galleggia sull'acqua  

**Polarizzabilità**

polarizzabilità è massimo in basso a sinistra, come il raggio  

fattori:  
1- dimensione (peso molecolare)  
quindi i cationi sono meno polarizzabili dell'atomo neutro, perchè più piccolo   
e gli anioni sono più polarizzabili  

2- superficie molecolare disponibile a contatti intermolecolari (forma della molecola)  
molecole lunghe hanno forze maggiori di quelle sferiche, perchè superficie maggiore  

**Punto triplo**

punto triplo: particolare stato termodinamico in cui le tre fasi di aggregazione (solida, liquida e gassosa) coesistono in perfetto equilibrio  

punto critico: in corrispondenza delle quali una sostanza può esistere come miscela bifase gas-liquido    
il limite estremo oltre il quale la distinzione tra liquido e gas scompare, dando origine a un fluido supercritico
