### Configurazione elettronica

#### Numeri quantici

Lettera | numero quantico ... | Cosa indica | che valori | Note
---|---|---|---|---
n | principale | dimensione orbitale (distanza media nucleo-elettroni di valenza), energia elettroni | 1 in su | coefficiente dell'orbitale (2s ha n=2)   
l | angolare | forma orbitale | 0 - (n-1) | ad ogni numero è associata una lettera, che dè il nome all'orbitale (s,p,d,f = 0,1,2,3)  
m | magnetico | orientazione orbitale nello spazio | -l .. +l, compreso 0 | eg. quale dei 3 orbitali p  
$m_s$ | di spin | rotazione dell'elettrone attorno al proprio asse | +1/2 o -1/2 | principio di Pauli, due doppietti devono avere spin diverso

**Nota**

il numero m cambia l'energia SOLO SE sono presenti campi magnetici (quindi come è orientato l'orbitale rispetto al campo cambia la sua energia)  

**Esempio numeri quantici**

n,l,m: sono per descrivere un quadratino  
se aggiungo $m_s$ indico se voglio l'elettrone con spin verso su o giù di quel quadratino  

esempio: primo elettrone nell'orbitale 2p:  
n=2 (è il coefficiente del livello)  
l=1 (corrispondete al tipo p)  
m=-1 (è nel primo dei 3 quadratini, chiamati -1,0,+1)  
m_s=1/2 (quello verso su è positivo, verso giù negativo)    

**Scrivere configurazione elettronica**

per velocità, si può scrivere il simbolo del gas nobile precedente tra parentesi, poi aggiungere gli elettroni di valenza  

$1s^22s^22p^63s^23p^3 = [Ne]3s^23p^3$ 


#### Riempimento orbitali

**Regole di riempimento**

1. elettroni iniziano a riempire gli orbitali a più bassa energia  

2. principio di esclusione di Pauli  
in ogni orbitale possono stare al massimo due elettroni, ma con spin contrapposto  

3. Hund's rule  
nel caso di orbitali degeneri, come i tre orbitali p, prima si riempiono tutti con un elettrone a spin = +1/2 (parallelo), successivamente si aggiungono gli altri elettroni a spin = -1/2 (antiparallelo)  

**Energia degli orbitali**

n,l influenzano l'energia  
m no, indica solo quale quadratino usare, ma sono tutti uguali  
orbitali con la stessa energia si dicono degeneri (come i 3 orbitali p) 

per metterli in ordine di energia: calcola n+l per ogni orbitale, mettili in ordine  
il risultato si vede nel diagramma di Aufbau, l'immagine sotto  
mostra l'ordine di riempimento degli orbitali atomici con gli elettroni, in base al loro livello energetico crescente

#### Note

**Difficoltà nel rimuovere elettroni**

sono più difficili da rimuovere gli elettroni che:  

- hanno un nucleo con èiù protoni  

- hanno altri elettroni nello stesso strato  

- hanno molti elettroni negli strati più interni  
Gli elettroni interni schermano gli elettroni esterni più efficacemente di quanto non facciano gli elettroni presenti nello stesso sottolivello  

### Metalli

**Proprietà metalli**

- Riflettono la luce
- Sono malleabili (lamine sottili)
- Sono duttili (fili sottili)
- Sono conduttori di corrente e calore
- I loro ossidi sono solidi basici altofondenti
- Tendono a formare cationi in soluzione
acquosa (quando vengono ossidati formano cationi)

Il carattere metallico aumenta in un gruppo  
e diminuisce in un periodo  
max: in basso a sinistra  

**Ossidi**

metalli: sono solidi e basici (cioè, si fondono con l'acqua per creare una base???)

$Na_2 O_{(s)} + H_2O_{(l)} \rightarrow 2NaOH_{(aq)}$  

non-metalli: sono acidi  

$P_4O_{10(s)} + H_2O_{(l)} \rightarrow 4H_3PO_{4(aq)}$  


metalli: tendono a cedere elettroni  
perchè sono nei primi gruppi, meglio tornare all'ottetto precedente  

non-metalli: tendono a guadagnare elettroni  
perchè sono negli ultimi gruppi, meglio finire l'ottetto  

**Metalli alcalini (gruppo 1)**

- morbidi (si tagliano con un coltello)  
- formano cationi monovalenti, 1+  
- reattività aumenta lungo il gruppo  
- reagiscono con l'acqua per creare una base e idrogeno gassoso  
$Na_2 O_{(s)} + H_2O_{(l)} \rightarrow 2NaOH_{(aq)}+ H_{2(g)}$  
- reagiscono con $O_2$ per formare ossidi basici    
- emettono colori particolari se posti
in una fiamma ad alta temperatura

**Metalli alcalino-terrosi**

- più duri e densi dei metalli alcalini  
- formano cationi positivi bivalenti, 2+ 
- reattività cresce lungo il gruppo

**Idrogeno**

$H_2$: allo stato elementare, un gas biatomico incolore  
$H^-$: ione idruro, se acquista un elettrone  
$H^+$: protone, se perde un elettrone  

in realtà $H^+$ non esiste come ione isolato ma solo
legato a molecole d’acqua, $H^+_{(aq)}$ 

**Gruppo 6**

Scendendo lungo il gruppo cresce il carattere metallico (O2 è un non-metallo, Te è un metalloide, Po è un metallo)  

l'ossigeno di solito ha numero di ossidazione -2, è -1 solo nel perossido di idrogeno  

**Alogeno**

allo stato elementare sono molecole
biatomiche  
tendono ad acquistare un elettrone per
diventare ioni alogenuro:  
$X_2 + 2e^- → 2X^-$   

fluoro: una delle sostanze più reattive conosciute  
cloro: l’alogeno industrialmente più utilizzato., prodotto per idrolisi dal salgemma (NaCl)  
cloro + acqua = acido ipocloroso (HOCl), usato per la potabilizzazione dell’acqua  

I composti binari degli alogeni con l’idrogeno sono

### Redox

ossidazione: perdita di elettroni  
riduzione: acquisto di elettroni  

agente ossidante: viene ridotto, serve a far ossidare l'altro  
e viceversa  

perchè si riferiscono al numero di ossidazione, quello in alto a destra negli ioni  
quindi ridurlo vuol dire acquistare elettroni, così da 0 va a -1  

**Numero di ossidazione**

carica formale attribuita a ciascun atomo in un composto   
Rappresenta il numero di elettroni che un atomo guadagna, perde o condivide rispetto al suo stato fondamentale   

carica formale:  
carica teorica assegnata a un singolo atomo all'interno di una molecola     
Si calcola ipotizzando che gli elettroni di legame siano condivisi in modo equamente paritetico tra gli atomi, indipendentemente dalla loro reale elettronegatività  

**Ricorda**

cosa | catodo | anodo 
---|---|---
polo | + | - 
attira | elettroni | protoni 
ci avviene | riduzione | ossidazione
ricorda | catodo e riduzione iniziano per consonante | anodo e ossidazione iniziano per vocale  

### Stechiometria

**Legge della composizione definita**

Un composto contiene gli stessi elementi
nello stesso rapporto di massa  

cioè: l'acqua, $H_2O$, contiene sempre 2H per ogni O  
viato che $M_H =1$ e $M_O=16$, H rappresenta il 2/18 della massa totale, cioè il 12% della massa totale  

**Legge delle proporzioni multiple (legge di Dalton)**

Quando due elementi si combinano per formare più composti diversi:  
considera l'elemento che cambia di quantità nei due composti e fai il rapporto tra la sua massa nei due composti  
il rapporto darà un numero intero piccolo  

cioè: $CO, CO_2$  
in CO, stanno 1 mol di C (12g) e 1 mol di O (16g)  
in $CO_2$, 1 mol di C e 2 moli di O (36g)  
rapporto multiplo: fissata la massa di carbonio (12g), le masse di ossigeno stanno in ratpport 16 : 32 = 1 : 2  
in pratica, in un cpmposto posso mettere solo un numero intero di atomi di un tipo  

**Recap**

1- legge della composizione definita  
ogni composta ha una formula molecolare, va rispettata  

2- legge delle proporzioni multiple (legge di Dalton)  
per scrivere una formula molecorae, un elemento deve avere un numero intero di copie  

### Colori dei sali

**Perchè un oggetto è colorato**

elettroni possono assorbire fotoni  
così ottengono la loro rnergia e saltano a un livello energetico più alto: ΔE=hν  
la E che serve per fare il salto, e quindi il quali fotoni sono assorbiti, dipende dalla molecola  
se assorbe dei fotoni con frequenza della luce visibile, allora quei colori sono assorbiti e non riflessi  
quindi a me non arriva il bianco, ma qualche colore (bianco meno quello assorbito, cioè il suo complementare)   

cioè: Se un composto è colorato, sta avvenendo una transizione elettronica nel visibile   

**Salto di orbitali**

HOMO = orbitale occupato più energetico  
LUMO = orbitale vuoto meno energetico  
La transizione più probabile è HOMO → LUMO  

cioè elettrone ottiene più energia, salta in un orbitale più alto  

**Colori molecole organiche vs metalli**

1- molecole organiche (idrocarburi)   
hanno solo legami sigma   
le transizioni elettroniche possibili richiedono molto energia  
quindi vengoon presi da fotoni dello spettro UV, non visible  
sono incolori  

2- metalli di transizione  
hanno orbitali d parzialmente riempiti  
le transizioni d−d cadono nel visibile  

| materiale | assorbimento | scattering  | risultato   |
| --------- | ------------ | ----------- | ----------- |
| vetro     | quasi nullo  | quasi nullo | trasparente |
| carta     | basso        | molto       | bianco      |
| pigmento  | selettivo    | sì          | colorato    |

oggetto bianco: riflette tutte le lunghezze d’onda, diffonde la luce in tutte le direzioni  
oggetto transparente: non assorbe nel visibile, non diffonde la luce  
la differenza è lo scattering
