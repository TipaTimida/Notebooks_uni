#### Velocità di reazione

**Cos'è**

cos'è:  
quanto velocemente varia la concentrazione dei reagenti o dei prodotti    
è una grandezza intensiva (quindi non dipende dalla massa del sistema)

a cosa serve:  
studia la velocità delle reazioni chimiche  
il fatto che una reazione possa avvenire dal punto di vista termodinamico non significa che questa avvenga in un tempo ragionevole

**Come si scrive**

2 modi, basandomi sulla:  
1- definizione stechiometrica    
2- legge cinetica  

posso uguagliarli, devono darmi lo stesso risultato  

**Def basata sulla stechiometria**

$aA+bB→cC+dD$  
$v= -\frac{1}{a}\frac{d[A]}{dt} = -\frac{1}{b}\frac{d[B]}{dt} = \frac{1}{c}\frac{d[C]}{dt} = \frac{1}{d}\frac{d[D]}{dt}$  

note:  
i reagenti hanno segno -, perchè la concentrazione diminuisce, quindi sennò avrei velocità negativa  
ognuno è diviso per il suo coeff stechiometrico  

**Def usando la legge cinetica**

$aA+bB→cC+dD$  
$v = k[A]^m [B]^n$  
k: costante cinetica (dipende da T, dalla specifica reazione, e altro, non è una costante universale, la trovo con la legge di Arrhenius)       
m,n: ordini parziali (sono sperimentali, non i coeff stechio)    
m+n = ordine totale della reazione

**Nota: ordini parziali**

in generale vanno ricavati sperimentalmente   
ma in una reazione elementare, ciò che avviene in un unico stadio molecolare (cioè ha un solo passaggio), coincidono con i coeff stechio   

cioè in una reazione elementare, l'ordine corrisponde alla molecolarità (numero di molecole reagenti)

#### Ordini delle reazioni

**Cosa sono**

come trovarlo: è la somma degli esponenti con cui le concentrazioni compaiono nella legge cinetica  

a cosa serve: dice come la velocità scala con la concentrazione delle specie chimiche  

| ordine | legge cinetica           | tipico di                          | soluzione          |
| ------ | ------------------------ | ---------------------------------- | ------------------ |
| 0      | $v = k$                  | processi superficiali, saturazione | $[A]=[A]_0-kt$     |
| 1      | $v = k[A]$               | decadimento, unimolecolari         | $[A]=[A]_0e^{-kt}$ |
| 2      | $v = k[A]^2 = k[A][B]$ | collisioni bimolecolari            | $1/[A]=1/[A]_0+kt$ |
| 3      | $v = k[A]^2[B]$          | rarissimo                          | complesso          |


**Riassunto ordini**

1- dipende da un solo oggetto (atomo o molecola) che ha una probabilità di trasformarsi  
esempio: decadimento

2- dipende dall’urto fra due oggetti  

3- esiste, ma raro che tre oggetti si incontrano nello stesso punto  

0- quando il reagente è in grandissimo eccesso, oppure quando la velocità è limitata da un fattore diverso  
quindi la velocità non dipende dalla concentrazione dei reagenti  
ad esempio, se la reazione ha bisogno di un sito reattivo dove
avvenire, la velocità dipende dal numero di siti  
esempio: la saturazione di un catalizzatore o una superficie metallica, enzimi  

**Pseudo primo ordine**

Molto spesso (per non dire sempre) le reazioni del secondo ordine sono studiate con un largo eccesso di uno dei reagenti  
Queste sono dette condizioni di pseudo primo ordine  

eg. $HCl+H_2O \rightarrow H_3O^+ +Cl^-$  
acqua é solvente la sua concentrazione é altissima e praticamente non varia  
è concentrazione di pseudo primo ordine  

**Decadimento radioattivo**

teoria:  
è un processo puramente probabilistico  
cioè ogni nucleo ha una probabilità costante per unità di tempo di decadere (indipendente dagli altri nuclei)   
quindi la velocità di decadimento è proporzionale al numero di nuclei presenti    

conclusione:  
tutti i decadimenti nucleari spontanei sono di primo ordine  
è la natura probabilistica del fenomeno che lo fa sempre risultare di primo ordine   

**Primo ordine: note**

1- scrivo eq differenziale, uguagliando le due formule per v:  
$v=-\frac 1n \frac{dN}{dt}$  
$v = k N$  
quindi: $\frac{dN}{dt} =- \lambda N$  
$\lambda$ è un nome a caso che gli ho dato per semplificare  

2- si risolve con separazione delle variabili:  
$N(t)=N_0e^{-\lambda t}$  

3- ora posso calcolare:  
tempo di dimezzamento: $t_{1/2}=\frac{\ln(2)}{k}$    
$\tau = \frac{1}{\lambda}$  

#### Legge di Arrhenius

**Formula**

$k=Ae^{-E_0/(RT)}$  

k: costante cinetica (quella delle legge cinetica)  
T: temperatura  
R: costante dei gas  
$E_0$: energia di attivazione  
A; fattore pre-esponenziale (probabilità di collisione “giusta”) (sezione d'urto, misura la probabilità che le molecole possedendo l’energia richiesta diano luogo ad un urto reattivo)  

k non è sempre uguale, tipo se T aumetna k aumenta, quindi la velocità di reazione aumenta   
perchè se T aumenta, aumenta il contenuto energetico delle molecole, ovvero aumenta il numero degli urti efficaci rendendo più veloce la trasformazione

**Contatore Geiger???**

Il contatore Geiger misura il numero di decadimenti radioattivi per unità di tempo, cioè la attività:  

$A(t)=-\frac{dN}{dt}=kN(t)$  

con N(t) numero di nuclei radioattivi presenti al tempo t  


#### Stadi di una trsformazione

**Nomi**

reagenti, prodotti: molecole stabili cioè che posso isolare e conservare   
intermedio: specie che esiste per un tempo molto breve e che non é possibile isolare

**Energia di attivazione**

energia che serve ad una specie stabile o ad un intermedio per raggiungere lo stato di transizione  
cioè: energia minima che occorre ai reagenti per rompere alcuni dei loro
legami e per iniziare una reazione  

proviene da:  
energia termica (reazione termica)   
luce (reazione fotochimica) 

**Stadi di una trasformazione**

stato: un punto preciso  
stadio: un periodo di tempo delimitato da intermedio, reagenti o prodotto   

1- stadio lento/limitante: nell'immagine, il periodo tra i reagenti e gli itnermedi  
perchè è quello con energia di attivazione maggiore  
determina la velocità dell’intero processo e la sua equazione cinetica  
il dislivello energetico tra i reagenti e i prodotti corrisponde alla variazione di entalpia ΔH  

2- stato di transizione: il punto sulla cima dello stadio lento  
il punto con la più alta energia  
la fase della reazione in cui si stanno rompendo i legami dei reagenti e sono in via di formazione i legami tra le molecole dei prodotti, con la formazione di un composto intermedio detto complesso attivato

**Catalisi**

catalizzatori: sostanze che permettono alla reazione di avvenire con una più bassa energia di attivazione  

non compare nella stechiometria nella reazione  
Può comparire nei passaggi intermedi come reagente ed un passaggio successivo come prodotto (viene rigenerato)  

Praticamente tutte le reazioni che avvengono negli
organismi viventi sono catalizzate



**Tipi di catalizzatori**

omogenei: nella stessa fase dei reagenti (entrambi liquidi, o entrambi gassosi)   
alta attività ma sono difficili da separare  

eterogenei: in fase diversa rispetto ai reagenti  
la reazione avviene sulla loro superficie, sono più facili da recuperare ma la selettività è legata ai siti attivi

**Rezioni concorrenti, selettività**

reazioni concorreni: processi in cui i reagenti possono trasformarsi in prodotti differenti attraverso percorsi di reazione distinti e simultanei, ciascuno con la propria costante cinetica  

$A \rightarrow B+C$  
a me interessa B, C è un sottoprodotto inutile  
Un catalizzatore selettivo è quello che favorisce uno di questi prodotti rispetto agli altri  

$s$: selettività, rapporto tra le velocità di reazione e costanti cinetiche    
$s = \frac{v_B}{v_C}$  

temperatura di inversione: la T alle quali le due reazioni concorrenti hanno la stessa costante cinetica $k_1=k_2$  
cioè per T maggiore o minore prevale una diversa delle due  

**Molarità**

il numero di molecole reagenti che vi partecipano  
monomolecolari e dimolecolari; frequenti  
trimolecolari: rare, poco probabile che tre molecole si urtino contemporaneamente e in modo efficace
