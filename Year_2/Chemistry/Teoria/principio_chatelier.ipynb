### Le Chatelier

**Principio di Le Châtelier**  

se un sistema all’equilibrio viene disturbato, il sistema si muove in modo da contrastare il disturbo

**Esempio di perturbazione**

1- concentrazione  
2- pressione  
3- volume  
4- temperatura  

**Varia concentrazione**

si interpreta tramite il quoziente di reazione Q  
se aggiungo reagenti, reazione va a destra, cioè si sposta verso i prodotti (trasforma più reagenti in prodotti)  

quindi per ottimizzare un processo si deve fornire continuamente reagenti e rimuovere i prodotti man mano che si formano

**Varia volume/pressione**

moli di gas di un lato: somma di tutti i coefficienti stechiometrici di quel lato (SOLO DELLE SPECIE GASSOSE, $\Delta n$)    
aumento P = diminuisco V ->  L’equilibrio si sposta verso il lato con meno moli di gas, per ridurre la pressione  
perchè meno moli vuol dire meno palline, cioè meno urti e meno pressione  
e viceversa  

**Varia T**

la costante di equilibrio K dipende da T (non da volume e pressione)  

per comodità, aggiungo il calore 'cal' alla formula della trasformazione  
reazione endotermica: il calore viene assorbito, è un reagente  
reazione esotermica: il calore viene liberato, è un prodotto  

se abbasso T, voglio alzarla, cioè liberare cal, quindi mi sposto verso il lato in cui appare 'cal'  
se alzo T, voglio abbassarla, quindi mi sposto dal lato senza 'cal', cioè la direzione che ha cal come reagente  

**Attenzione: varia pressione**

la variazione di pressione che influenza l’equilibrio è quella determinata da una variazione di volume, che determina una variazione delle pressioni PARZIALI di reagenti e prodotti gassosi  

cioè:  
devono variare le pressioni delle cose che partecipano alla reazione  
Una variazione di pressione ottenuta, a volume costante, per aggiunta di un gas inerte (per uno ione, si dice spettatore) non ha effetti sull’equilibrio  
ovvio, perchè $K_p$ resterebbe uguale  

**Catalizzatore**

abbassa l’energia di attivazione della
reazione diretta e di quella inversa nello stesso modo  
quindi rende più veloce il raggiungimento dell’equilibrio  
ma non cambia la composizione della miscela di equilibrio  

non compare nella formula stechiometrica della reazione
