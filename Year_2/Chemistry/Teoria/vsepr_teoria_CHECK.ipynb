### Geometria molecolare: definizioni

#### Definizioni

**Teoria VSEPR**

VSEPR - Valence Shell Electron Pair Repulsion   

dice che:  
la forma delle molecole può essere prevista sulla base
delle repulsioni tra gli elettroni di valenza  
la forma che assumono le molecole è quella che minimizza queste
repulsioni


**Struttura di Lewis**

è la geometria delle coppie elettroniche  
cioè la posizione nello spazio di TUTTE le coppie elettroniche, sia doppietti liberi che di legame    

### Teorie per legami chimici

**Tipi di teorie**

sono progressivamente più raffinate   
descrivono come si formano i legami chimici 

teoria | Lewis | legame di valenza | orbitali molecolari

dove si trovano gli elettroni | tra una coppia di atomi, cioè a fare un legame, o come doppietto su un atomo | Principalmente localizzati tra due atomi | Distribuiti sull’intera molecola

### Angoli di legame

![alt text](https://scienzedidattica.wordpress.com/wp-content/uploads/2015/03/immagine_1-vsepr.jpg)

**Angoli di legame**

cosa respinge di più:  
1- doppietto libero (di più)  
2- legame multiplo  
3- legame singolo  

conseguenze:  
la molecola con più doppietti liberi ha angolo di legame più piccolo

**Spiegazione repulsione**

1- elettroni di legame:  
sono attratti da due nuclei  
quindi si trovano più distanti dall’atomo centrale  
quindi danno minore respulsione  

2- doppietto libero  
è più localizzato vicino al nucleo e occupa più spazio  
quindi maggiore repulsione  

### Polarità

**Spiegazione**

se due atomi diversi sono legati:  
la differenza di elettronegatività genera un dipolo di legame  
ogni dipolo ha un momento di dipolo, $\vec{\mu} = Q\vec{r}$ (carica per distanza tra le molecole)

il momento di dipolo della molecola è dato dalla somma VETTORIALE di tutti i momenti di legame  
se il momento di dipolo della molecola è non-0, è detta molecola polare  

### Teoria del legame di valenza

**Teoria del legame di valenza**

quando due nuclei si avvicinano, i loro orbitali si sovrappongono  
maggiore è la sovrapposizione, più forte è il legame formato  

**Distanza di legame**

distanza tra i due nuclei a cui l'energia raggiunge un minimo   
quindi se gli atomi sono più vicini o lontani, hanno energia più alta, cioè si sviluppa una forza di repulsione o attrazione  

A distanza di legame le attrazioni nucleo-elettroni prevalgono sulle repulsioni nucleo-nucleo ed elettrone-elettrone   

### Orbitali ibridi

**Problema**

considera la molecola $BeCl_2$, col berillio al centro  
il berillio ha configurazione elettronica $2s^2$  
quindi non ha singoletti con cui fare legami covalenti  

stesso problema che ha il carbonio  
vuole fare 4 legami, ma ha 2 singoletti e 1 doppietto

**Ibridazione**

gli orbitali $2s$ e $2p$ si mescolano per dare un orbitale ibrido  
l'ibridazione richiede energia  
ma gli orbitali ibridi danno legami più forti, perchè hanno sovrapposizioni più estese  
ciò compensa la perdita di energia  

**Orbitali ibridi**

numero di orbitali ibridi ottenuti = numero di orbitali atomici mescolati  
in pratica conta il numero di lettere nel suo nome (ognuna tante volte quanto il suo esponente), quello è il nuemero di orbitali ibridi ottenuti, quindi di legami che consente di fare  

tipo di orbitale varia a seconda dei tipi di orbitali usati per crearlo (il nome)  

**Orbitali sp**

Quando si mescolano n orbitali atomici, si devono
ottenere n orbitali ibridi  
s + p = 2 sp  
s + p +p = 3 $sp^2$  
s+p+p+p = 4 $sp^3$

hanno tutti energia intermedia tra s e p  
maggiore è l'esponente della p, maggiore l'energia ???  

forme:  
sp: due lobi asimmetrici (mentre orbitale p ha due lobi simmetrici)   
$sp^2$: tre lobi
$sp^3$: tetraedro, con angolo di 109.5 -> tutte le molecole con numero sterico 4 sono ibridate $sp^3$  

**Esempio orbitali sp**

$CH_4$: $sp^3$  
perchè il carbonio ha 2 singoletti e un doppietto  
di solito il carboio vuole fare 4 legami, quindi si ibrida creando 4 orbitali $sp^3$  

$NH_3$: $sp^3$  
$H_2O$: $sp^3$  
questi in teoria potrebbero, hanno abbastanza singoletti  
ma le osservazioni sperimentali mostrano angoli di legame che possono solo apparire con l'ibridazione, quindi devono essere ibridati  
ma perchè, non potrebbe essere dovuto alla repulsione dei doppietti liberi???

**Orbitali spd**

se un atomo vuole fare 5 o 6 lgeami, non gli bastano i 4 orbitali $sp^3$  
invece, usa gli orvitali ibridati spd

per fare 5 legami: 5 orbitali $sp^3d$  
per fare 6 legami: 6 orbitali $sp^3d^2$  

#### Orbitali molecolari

**Cosa sono**

Dalla combinazione di due orbitali atomici (di atomi diversi) si formano due orbitali molecolari

cosa | legante | antilegante 
---|---|---
nome | $\sigma_{1s}$ | $\sigma_{1s}^*$
energia | inferiore all'orbitale atomico iniziale | superiore...  
densità elettronica | maggiore tra i due nuclei | nulla tra i due nuclei  
cosa fa | favorisce il legame | ostacola il legame

**Ordine di legame**

$ol = \frac{n_l-n_a}2$   
ol: ordine di legame, misura della forza e stabilità di un legame  
nl, na: numero di elettroni negli orbitali leganti, antileganti  

ol = 0, no legame  
ol = 1, singolo  
ol = 2, doppio  

**Esempio: H2**

ogni H ha config: $1s^1$   
quindi la molecola ha 2 elettroni di valenza  

se sommo due orbitali s, ottengo, gli orbitali $\sigma$ e $\sigma^*$ (legante e antilegante)   
comincio a riempire $\sigma$, ho finito gli elettroni  
quindi $ol = \frac{2-0}2 =1$  
c'è legame singolo tra le H  

**Esempio: Li2**

ogni Li ha config: $1s^22s^1$     
quindi la molecola ha 2 elettroni di valenza   

come nel caso H2

**Proprietà magnetiche**

La teoria degli orbitali molecolari permette di prevedere le proprietà magnetiche  

1- paramagnetico   
quando: ci sono elettroni spaiati (singoletto)   
cosa: attratta dal campo  

2- diamagnetico  
quando: no elettroni spaiati  
cosa: respinta dal campo (dia = via, respinta)  

**Proprietà magnetiche di $O_2$**

sperimentalmente è paramagnetico  
eppure, la struttura di Lewis non mostra singoletti  

però la teoria degli orbitali molecolari mostra due elettroni spaiati in $\pi^*_{2p}$

**Legame tra atomi diversi**

i due orbitali NON sono alla stessa energia  
non tutti gli orbitali 2s hanno la stessa energia, dipende dall'atomo  
tipo, l'1s di H ha energia maggiore del 2p di F  

tip: più l'atomo è elettronegativo, più i suoi orbitali sono stabili, e la loro energia bassa  
F è più elettroneg di H, quindi il suo orbitale di valenza ha energia minore di quello di H  

stessa cosa con NO: il 2p di O ha energia minore del 2p di N  

conseguenza:  
se si combinano di orbitali di energia differente, gli orbitali molecolari che si formano non hanno una forma simmetrica rispetto ai due nuclei   
Gli orbitali leganti contengono più del 50% dell’orbitale atomico di energia minore e quindi avranno una distribuzione spostata verso l’atomo corrispondente  
(guarda nel disegno, gli orbitali leganti sono più vicini all'atomo con energia minore, quelli non leganti all'atomo con enegia maggiore)  

se la molecola è stabile, o.l.>0, cioè ci osno più elettroni negli orbitali leganti 
Il legame quindi è polarizzato con una carica negativa
sull'atomo di bassa energia (che ricorda, è quello più elettroneg, ha senso)

### Tipi di legami

#### Covalenti

**Legame covalente**

modi diversi di definirlo, a seconda di quale teoria uso  
si forma quando:  
1- gli orbitali di due atomi si sovrappongono   
2- la regione di sovrapposizione, situata tra i due nuclei, è occupata da una coppia di elettroni

le due funzioni d’onda sono in fase   
l’ampiezza aumenta nella regione internucleare

**Tipi di legame**

Legame covalente: si realizza mediante la sovrapposizione degli orbitali dei due atomi e messa in comune degli elettroni che occupano questi orbitali.  

Legame ionico: un atomo cede elettroni e un altro li
acquista, con formazione di ioni di carica opposta che
vengono tenuti insieme da forze elettrostatiche.

legame metallico: compartecipazione di elettroni tra un gran numero di atomi

**Energia di legame**

l’energia che occorre fornire per rompere il legame  
un po' come l'energia di reticolo è l'energia che serve per separare il reticolo (è l'equivalente dell'energia di legame, ma per il legame ionico)  
viene misurata come entalpia $\Delta H$, cioè il calore assorbito a P costante  

legami multipli hanno:  
1- energia di legame maggiore  
2- lunghezza di legame minore  

un legame doppio ha un’energia di legame minore di 2 legami semplici  
quindi è favorta una reazione in cui quello doppio si rompe, e si formano 2 semplici  

**Tipi**

1- legame sigma:  
densità elettronica: massima sull’asse che unisce i due nuclei  
quando: va bene sempre, s-s, p-p, s-p, ibrido-obrido  
direzione: la congiugnete tra i due nuclei  

2- legame pi:  
densità elettronica: massima sopra e sotto l’asse che unisce i due nuclei  
quando: solo da sovrapposizione di due orbitali NON ibridizzati   
direzione: ortogonale agli altri legami  

semplice: 1 sigma  
doppio: 1 sigma, 1 pi  
triplo: 1 sigma, 2 pi  

**Triplo legame**

In ogni triplo legame (es. N2):   
un legame π è sempre sopra e sotto il piano dei nuclei   
mentre l’altro è davanti e dietro tale piano   

immagina piano xyz, i due legami pi devono essere perpendicolari tra loro e al legame sigma (la congiungente dei nuclei)  

**Nota**

obritali ibridi danno sempre legami sigma, perchè sono direzionali  
quindi numero sterico indica sia l'ibridazione dell'atomo, che il numero di legami sigma e di doppietti liberi ???

**Legame ionico**

gli elettroni sono molto più localizzati sull'atomo che ha maggiore elettronegatività   

#### Delocalizzati / Risonanza

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQMtojUtp0Vcs9dnamCUAgEGmRewks8T5Or6w&s)

**Legami pi delocalizzati: benzene**

esempio: benzene  
6 legami sigma C-C  
6 legami sigma C-H  
6 orbitali p non ibridizzati, uno su ogni C  

per completare gli ottetti, ogni C ha bisogno di un doppio legame  
ma ci sono 2 modi per farlo, dette strutture di risonanza  
quindi il benzene è un ibrido di risonanza  
la sua forma vera non è nessuna delle due, perchè i legami pi non sono localizzati tra nessuna coppia di atomi, ma sono delocalizzati su tutti e 6 i carboni  
infatti, sperimentalmente tutti i legami C-C hanno la stessa lunghezza,  metà tra un legame singolo e un doppio  

**Legame pi delocalizzato: ione nitrato**  

$NO_3^-$  
anche qui non so con quale coppia fare il doppio legame  
però qui non c'è un modo veloce di scriverlo, come prima col cerchio, devo solo disegnarle tutte  

nota che solo i legami pi possono essere delocalizzati, non i sigma  

### Eccezioni regola dell'ottetto

1. Molecole con un numero dispari di elettroni.
2. Molecole con un atomo che ha l’ottetto
incompleto.
3. Molecole con un atomo che espande l’ottetto.
4. Ioni che non hanno la configurazione del gas
nobile

**Proprietà magnetiche delle sostanze**

diamagnetiche: no elettroni spaiati  
debolmente respinte da campo magnetico  

paramagnetiche: elettrone spaiato (perchè molecola ha un numero dispari di elettroni, come NO)     
attratte da campo magnetico, spesso colorato  

ferromagnetiche: ferro e poco altro  
fortemente attratte da campo magnetico  

**Problema**

l'ossigeno molecolare, $O_2$, è paramagnetico  
quindi deve avere degli elettroni spaiati, per simmetria uno su ogni atomo  
però la struttura di Lewis non ne ha, perchè vengono usati per fare un legame doppio???

**Risonanza**

non cambia lo scheletro della molecola, ma solo dove si trovano i legami doppi  
esempio. $N_2O$
