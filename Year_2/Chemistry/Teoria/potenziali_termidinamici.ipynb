### Potenziali termodinamici

Formula | Nome | variabili naturali | significato fisico
---|---|---|---
U | energia interna (energia totale microscopica del sistema)| S,V | sistema isolato (no scambio calore nè lavoro) 
H = U + pV | entalpia | S,P | solo teoria    
F = U − TS  |energia libera di Helmhholtz | T,V | contenitore rigido a T costante  
G = H − TS | energia libera di Gibbs  | T,P | sistema reale (T fissato da ambiente, p da atmosfera)   

#### Spiegazioni

**Quale usare**

usa quella le cui variabili naturali rimangono fisse  
tipo, in un esperimento a T,P costanti, usa G  
così so che all'equilibrio ho un minimo di G  

**Potenziali termodinamici: cos'è**

un sistema lasciato libero di evolvere tenta di minimizzare la propria energia  
a seconda di quali variabili è in grado di modificare, e di quali sono fissate dall'ambiente circostante (var naturali), la formula dell'energia da minimizzare è diversa  
queste energia si chiamano 'potenziali termdinamici' (potenziali perchè sono funzioni di stato)

cioè:  
sistema a T,P costante evolve verso un minimo di G  
sistena a T,V costante evolve verso un minimo di F  

**Potenziali termodinamici: a cosa serve**

ognuno, per il caso in cui mantengo fisse le var naturali, mi dice:  
1- cosa succede spontaneamente (cioè come evolve il sistema, perchè deve minimizzare il potenziale)  
2- quanta energia utile puoi ottenere (la variazione di potenziale tra due stati l'energia utile, che può essere o lavoro estratto o calore scambiato)     

**Variabili naturali**

gli altri potenziali termodinamici si trvano con la trasformata di Legendre  
che cambia da quali var dipende  
il secondo addendo è la moltiplicazione tra ver vecchia e var nuova  

esempio:  
H = U + pV 
quindi se U(S,V), allora H(S,p)  

**Note**

per ricordare i segni nelle formule:  
di solito -, come sempre nella trasformata di Legendre  
in H uso +, perchè scaldare a pressione costante richiede più energia che a volume costante, visto che devo obbligatoriamente compiere un lavoro espansivo (H e U sono calori)  

**Spontaneità**

ricorda che un sistema cerca di minimizzare la sua energia:      
ΔG < 0 → processo spontaneo (perchè G_fin < G_iniz, sto minimizzando G)    
ΔG > 0 → processo non spontaneo  
ΔG = 0 → sistema all’equilibrio  

non vale solo per G, ma per ogni potenziale termodinamico, anche per H  
H<0: spontaneo, rilascia calore  
H>0: non spotaneo, serve calore  
i segni sono come nel calore normale, Q>0 vuol dire assorbito  

#### Energia, lavoro, calore

**Energia, lavoro, calore**

1- i potenziali (U,F,G) sono energie, cioè variabili di stato  

2- la loro variazione (variazione di energia), a seconda dei casi, può corrispondere a calore o lavoro  

**Formule per calore scambiato / lavoro estratto**

prima legge della dinamica: $Q= \Delta U+W$ 

1- calore scambiato   
a V costante: $Q=\Delta U + 0= \Delta U $  
a P costante: $Q = \Delta U +p \Delta V =\Delta H$  
nel mondo reale, spesso ho condizioni a p costante, quindi H è più utile di U  

2- lavoro estratto   
a V costante: $W = -\Delta F$  
a P costante: $W = -\Delta G$  
perchè??? 

**Energia/lavoro**

1- energia totale: U  
include tutta l'energia microscopica, ma non tutta è utilizzabile  
perchè l'entropia mi impedisce di usare tutta l’energia

2- energia degradata: TS  
la parte di energia che NON può diventare lavoro  
perché il sistema deve mantenere un certo disordine (entropia), quindi un po' di energia viene dispersa come calore

3- energia utile: F, G  
energia che può diventare lavoro   
tolgo la parte persa nell'entropia (energia degradata)  

4- energia estraibile, lavoro estraibile: $-\Delta F$ o $- \Delta G$   
variazione dell’energia utile  
cioè F e G mi dicono quanto lavoro posso estrarre, nel caso in cui mantengo quelle var naturali fisse  
si chiama lavoro perchè è una variazione di energia  

**F vs G**

$\Delta F$ mi dice il lavoro SOLO se a V costante, cioè se non c'è espansione  
perchè in teoria comprende anche il lavoro di espansione, cioè spingere l'aria per potersi espandere  
ma questo non dà energia sfruttabile  
il lavoro utilizzabile davvero è quello elettrico o chimico

3- $dN_i​=0$: significa che non cambia la composizione delle specie

### Equazione di Claypeiron

**Equazione di Clapeyron**

$\frac{dP}{dT} = \frac{\Delta H}{T \Delta v}$   

$\Delta H$ e $\Delta v$ è la differenza tra le due fasi ($v$ è il volume MOLARE, cioè $v=V/mol$, ma nel modo in cui la ricavo c'è V normale???)  
T: temperatura assoluta, in quel momento (se le fasi sono inequilibrio, hanno stessa T)    

#### Teoria

**Cos'è**

legge di equilibrio di fase: descrive come cambia la pressione di equilibrio tra due fasi, al variare della temperatura  

cosa fa:  
ci sono dei punti sul grafico (T,P) dove il passaggio di fase è in pausa, stanno coesistendo entrambe le fasi, ma nessuna si sta trasformando nell'altra  
se sposto un po' T o P, il passaggio di fase riprende  
ma io invece voglio tenerlo bloccato  
quindi se vario un po' T, come devo variare P per non far avanzare il passaggio di fase?  

graficamente: descrive come avanzare sul piano T-P per rimanere sulla linea dell'equilibrio di fase  

**Equilibrio di fase**

cos'è:  
è una configurazione di (p,T) per cui:  
il sistema è alla temperatura e pressione a cui dovrebbe avvenire un passaggio di fase  
il passaggio di fase è iniziato, ma non sta procedendo, il sistema è fermo all'equilibrio di fase  
quindi ci sono contemporaneamente entrambe le fasi (tipo un po' di ghiaccio e un po' di acqua)  
fasi: parti diverse di un miscuglio, che si distinguono a occhio nudo  

come si scrive:  
all'equilibrio di due fasi, $G_{fase \,1} = G_{fase \; 2}$  
così $\Delta G=0$, il che vuol dire che il sistema è all'equilibrio, nessuna delle due trasformazioni avviene spontaneamente  
uso quello di Gibbs, perchè è il più comodo in condizioni reali  


**Plot**

posso fare un plot T-P, su cui disegno una linea  
è la linea dell'"equilibrio di fase"  
cioè in tutti i punti su quella linea, la trasformazione è bloccata durante il passaggio di fase  
mentre sopra o sotto la linea esistono solo la fase 1 o la fase 2 (se dò tempo al sistema di evolvere fino all'equilibrio)   

quindi l'equazione di Clapeyron mi dà la tendenza di questa linea  

#### Ricavarla

**Ricavarla**

condizione: sistema all'equilibrio di fase  
$dG_1 = dG_2$  

$V_1dP - S_1dT =  V_2dP - S_2dT$  
$ (V_2-V_1)dP = (S_2-S_1) dT$  
$\frac{dP}{dT}=\frac{\Delta S}{\Delta V}$  
infine: $\Delta S = \frac{\Delta H}{T}$  

nota: è un cambiamento di fase, quindi $\Delta N_i=0$, non cambia la composizione delle specie  

#### Casi specifici

**Caso: fusione**

caso specifico della legge di Clapeyron, quando il passaggio di fase è solido->liquido  

sostituisci:  
$\Delta H = L_f$: entalpia di fusione  
$V=\frac{M}{\rho}$  (M: massa molare), V : volume molare  

diventa:  
$\frac{dP}{dT} = \frac{L_f}{T (\frac{M}{\rho_l}-\frac{M}{\rho_s})}$   

**Caso: evaporazione (legge di Clausius-Clapeyron)**

1- semplifico    
$\Delta V \sim V_v$ perchè $V_v \gg V_l$ (v=vapore, l=liquido, cioè il volume di quando era liquido è quasi 0 al confronto)    
$V_v =\frac{RT}{P}$, perchè gas ideale (e molare, quindi non divido per n)  
$\frac{dP}{dT} = \frac{P \Delta H}{RT^2}$  

2- integro:  
$\ln P = -\frac{\Delta H}{R} \frac{1}{T} +const$  

$\ln(\frac{P_2}{P_1})=-\frac{\Delta H}{R}(\frac{1}{T_2}-\frac{1}{T_1})$  

**Caso: acqua**

durante la fusione, M non cambia, $\rho$ sì  
di solito, $\rho_l < \rho_s \rightarrow V_l > V_s \rightarrow \Delta V > 0$  
quindi la cirva dP/dT>0 ha pendenza positiva    
cioè se aumento P, devo aumentare anche T per avere la fusione  

ma il ghiaccio è meno denso dell'acqua, $\Delta V>0$  
quindi aumentando la pressione, l’acqua fonde a temperature più basse  


### Capacità termica

**Definizione**

capacità termica: quantità di calore necessaria per aumentare di un grado la temperatura di un certo sistema  

$C = \frac{\Delta E_{cin}}{\Delta T}$  

**Usarla per scrivere calore**

variazione di energia: calore scambiato a volume costante  
$\Delta E = Q_V = nC_v \Delta T$  

vaiazione di entalpia: calore scambiato a pressione costante  
$\Delta H = Q_p = nC_p$  

$C_p=C_v+R$    
Scaldare a pressione costante richiede più energia che a volume costante perché devo obbligatoriamente compiere un lavoro espansivo  
cioè a volume costante W=0, a pressione costante è non-nullo

**Principio di equipartizione dell'energia**

Energia termica è la forma di energia più
disordinata e si ripartisce in modo uguale fra tutti i gradi di libertà  
Ad ogni grado di libertà della molecola corrisponde una capacità termica molare di 1/2RT  
$C= \frac{DOF}{2}R$  

gas monoatomico: può solo traslare nello spazio, in 3D (3 degree of freedom)  
gas poliatomico non lineare: può traslare e ruotare attorno a 3 assi (6 DOF)  
gas poliatomico lineare: può traslare e ruotare attorno a 2 assi (5 DOF) (perchè ruotare attorno alla linea della molecola non è davvero ruotare, non conta)  

**Legge di Dulong e Petit**

vale per solidi cristallini (regolari, tipo sali)  
non per soldi amorfi (tipo vetro, che si comportono più come liquidi)  

$C_p \sim C_v \sim C \sim 3R$  
perchè un solido si espande poco, quindi la differenza fra pressione costante e volume costante é meno importante
quindi la chiamo genericamente capacità termica molare  

**Reticolo cristallino e DOF**

Una molecola (od uno ione) in reticolo cristallino presenta 6 gradi di libertà vibrazionali  

in un cristallo (reticolo), ogni ione ne ha altri 6 vicini  
In questa struttura molto rigida gli unici gradi di libertà sono vibrazionali, ovvero gli ioni oscillano
intorno alla posizione di equilibrio  
È come se ogni coppia di vicini fosse collegata da una molla   
ne ho 6, perchè uno ione ha 6 vicini    

**Equazione di Kirchoff**

La capacità termica a pressione costante é la
derivata dell’entapia rispetto alla temperatura  

$\Delta H_{T2} = \Delta H_{T1} + \int \Delta C_p \, dT$  
nota che $\Delta H$ si per indicare l'entalpia in uno stato preciso, non una variazione  

**Bomba di Maler / bomba calorimetrica**

cosa: strumento usato per misurare il calore di combustione  
cioè $\Delta H$ rilasciato durante la combustione (cioè le calorie di un alimento)  

come funziona:  
1- ho una quantità nota di materiale  
2- lo metto in un contenitore chiuso, detto "bomba" (quindi sono a V costante, uso U come energia, non G)  
3- immergo il contenitore in una massa nota d'acqua    
4- con una scarica elettrica faccio scattare la combustione  
5- il calore emesso alza la temperatura dell'acqua  
quindi misurando T dell'acqua, so il calore di combustione  

$\Delta U = C_{to } \Delta T_{water}$   
$C_{tot}$ è la C del sistema formato da acqua e calorimetro  
$\Delta H = U +\Delta n RT$
