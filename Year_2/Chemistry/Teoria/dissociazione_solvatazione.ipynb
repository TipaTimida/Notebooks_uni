### Definizioni composti

**Miscela vs composto**

insieme di 2 o più sostanze:  

1- miscela: unite SENZA REAZIONE CHIMICA (no legami chimici)  
esempio: verso olio e acqua insieme, non reagiscono a formare altro, semplicemente ora li ho nella stessa ciotola  

2- composto: unite TRAMITE LEGAMI CHIMICI  
esempio: $H_2O$ (qualunque cosa con specifica formula molecolare)  


**Miscela vs composto: proprietà**

cosa | miscela | composto 
---|---|---
composizione | variabile (posso versarne quanto ne voglio di ogni sostanza) | fissa (componenti sono in proporzioni definite, c'è una formula)    
proprietà chimiche | ogni componente mantiene le proprie | cambiano rispetto ai componenti, sono specifiche per questo composto  
separabile | con metodi fisici (filtrazione, distillazione, ecc.) | solo con reazioni chimiche  

**Tipi di miscele**

miscuglio: miscela eterogenea, più fasi presenti  
composizione non uniforme  

soluzione: miscela omogena, una fase  
composizione uniforme, componenti non distinguibili

che vuol dire:  
fase = componente  
Una fase è una porzione di materia omogenea, con composizione e proprietà fisiche uniformi, separata dalle altre da una SUPERFICIE DI SEPARAZIONE     
eg: acqua pura, acqua + sale = 1 fase (perchè sale è disciolto omogeneo in acqua, fa soluzione, quindi non vedo due porzioni diverse separate da un'interfaccia)     
acqua + olio, acqua + sabbia = 2 fasi  

### Mettere sostanza in acqua

**Dissociazione vs solvatazione**

avvengono entrambe:  

1- dissociazione: separazione del composto in ioni  
processo strutturale  
$NaCl \rightarrow Na^+ + Cl^-$    

2- solvatazione: interazione con acqua (il solvente) 
processo di stabilizzazione energetica, non di separazione  
$Na^+ \rightarrow Na^+_{(aq)}$ 
stabilizza gli ioni  
riguarda l’interazione tra soluto e solvente  


**Dissociazione**

rompi interazioni ioniche (cioè rempo reticolo ionico), ottieni particelle libere  

**Solvatazione/idratazione**

1- le molecole di $H₂O$ circondano gli ioni Na⁺ e Cl⁻, tipo formandogli delle bolle attorno    
2- orientano i dipoli nel verso corretto (perchè O ha carica parziale negativa, H carica parziale positiva)  
questo si chiama 'guscio di idratazione'  
3- così stabilizzano le particelle e abbassano l’energia del sistema  
uno ione circondato da questa bolla di acqua si dice ione idratato  

ione idratato: è un catione o anione circondato da molecole d'acqua polari, formando un "guscio di idratazione" stabilizzato da attrazioni elettrostatiche (attrazione tra ione e carica parziale di un atomo dell'acqua)  
si indica col pedice (aq): $Na^+_{(aq)}$ è uno ione idratato  

#### Tipi di solvatazione

**Tipi di interazione**

cosa | aspecifica | specifica | reattiva
---|---|---|---
cosa è | interazioni generiche (interazione fisica) | interazioni direzionali (interazione chimica) | reazione chimica  
che fa il solvente | non si sposta in una configurazione particolare per circondare il soluto, semplicemente interagisce in modo elettrostatico con il soluto | si orienta attorno al soluto a formare un guscio di idratazione  |non si limita a circondare il soluto, ma reagisce chimicamente con esso (non solo interazione, ma proprio trasformazione chimica)  
cosa viene creato | interazioni deboli e non direzionali, comuni a molte sostanze    (forze di Van der Walls, forze dipolo-dipolo)  | Interazioni direzionali e selettive, legate alla struttura chimica delle molecole (legami a idrogeno, interazione ione-dipolo)    
esempio | gas disciolti in liquidi, molecole apolari in solventi apolari  | Na⁺ e Cl⁻ in acqua (ione–dipolo), alcol in acqua (legami H)   | HCl in acqua → H₃O⁺ + Cl⁻    

**Quando avvengono**

in una reazione può avvenire più di un tipo, magari da parti diverse del solvente  
ad esempio, nel bulk avviene sempre l'interazione aspecifica, ma nella cibotassi potrebbe avvenire interazione specifica o anche reattiva  

**Parti di una soluzione**

soluto: La sostanza che viene disciolta  
solvente: La sostanza che scioglie il soluto e costituisce la parte maggiore della soluzione  

cibotassi: parte di solvente che circonda da vicino il soluto, che interagisce specificatamante col soluto  
bulk: parte di solvente più lontana dal saluto, interagisce in modo non specifico con il soluto  

#### Ciclo termodinamico della solvatazione

$ΔG_{sol}$: libera energia di solvatazione   
variazione di energia libera quando porti il soluto dallo stato isolato (gas o solido) allo stato solvato  

nota: uso G, che è a T,p costanti (e non F, che è a T,V costanti) perchè il volume può cambiare????

si scompone in contributi: energia di  
1- cavitazione  
2- elettrostatica  
3- dispersione  
4- repulsione  
5- Variazione geometrica del soluto    

**Interazioni soluto-solente**

nome | cosa | da cosa dipende | pro o contro dissoluzione?
---|---|---|---
cavitazione | occorre creare nel bulk del solvente uno spazio (cavità) per contenere il soluto <br> fare spazio costa energia, perchè bisogna rompere le interazioni tra molecole del solvente, cioè battere la tenzione superficiale, quindi $G_{cav}>0$  | grandezza del soluto <br> tensione superficiale del solvente | pro
elettrostatica | la distribuzione di carica elettronica del soluto interagisce con il solvente <br> cioè il campo elettrostatico del soluto polarizza il solvente | distribuzione di carica del soluto <br> costante dielettrica del solvente | contro
dispersione | interazione fra le fluttuazioni istantanee della densità di carica elettronica del soluto e del solvente | dimensioni molecolari (più l amolecola è grande, più è probabile che ci siano fluttuazioni) | pro
repulsione | repulsione fra le nubi elettroniche del soluto e del solvente <br> è ciò che impedisce agli atomi di penetrarsi (principio di Pauli) | dimensioni molecolari | pro
variazione geometrica | Il solvente può modificare la distribuzione elettronica e perfino la geometria del soluto <br> Solventi polari tendono ad aumentare il dipolo dei soluti | polarità del solvente | contro


**Osservazione**

step per sciogliere un soluto:  
1- separare le particelle del soluto (costa energia)  
2- creare spazio nel solvente (costa energia)  
3- formare nuove interazioni tra soluto e solvente (rilascia energia)  

conseguenza:  
una sostanza si scioglie bene quando le nuove interazioni soluto–solvente sono abbastanza simili, per natura e intensità, a quelle che bisogna rompere nel soluto e nel solvente  
cioè, "Il simile scioglie il simile"  

un soluto polare o ionico si scioglie bene in un solvente polare, perché può formare interazioni dipolo-dipolo, legami a idrogeno o interazioni ione-dipolo  
un soluto apolare si scioglie meglio in un solvente apolare, perché entrambi interagiscono soprattutto tramite forze di dispersione  

### Elettroliti

**Definizione**

elettrolita: sostanza che in soluzione acquosa (o quando è fuso) si dissocia in ioni, e quindi conduce corrente elettrica  
non-elettrolita: può dissolversi in acqua, ma dà molecole neutre, non ioni  

eg: $NaOH \rightarrow Na^+_{(aq)}+OH^-_{(aq)}$   
quindi NaOH è un elettrolita, perchè si dissocia in ioni (SE messo in acqua)  


**Tipi di elettroliti**

elettroliti si distinguono in: acidi, basi, sali  
acidi: in acqua libera $H^+$  
basi: ina cqua libera $OH^-$  
sali: non liberano nessuno dei due, si dissocia in anione + catione  

#### Acidi e basi

**Definizioni**

1- def di Bronsted-Lowry  
acido: dona $H^+$  
base: accetta $H^+$  

2- def di Lewis:  
acido: accetta un doppietto  
base: dona un doppietto  


**Acqua**

acqua può essere sia base che acido, è anfiprotica  
anfiprotico: SOLVENTE che può essere sia acido che base  
anfotero: sostanza che può essere sia acido che base 

effetto livellante del solvente:  
ogni acido o base più forte di $H^+$ o $OH^-$ reagisce con l'acqua per dare questi ioni  

**Tipi**

idruri: H-X  
ossiacidi: X-OH  

cos'è: dipende dall'elettronegatività di X  
bassa (metallo): base  
intermedia: acido debole  
alta (non metallo): acido forte  

**Forza**

un acido è forte se riesce a perdere facilmente H⁺  
che è più facile se X-H è molto polare, perchè:  
1- c'è quasi già separazione di cariche (parziali, per ora), il che favorisce la dissociazione  
2- i legami ionici sono più deboli di quelli covalenti  

**Forte o debole**

acido forte: X molto elettronegativo (così ha legme molto polare)  
base forte: X molto poco elettronegativo  
il legame deve comunque essere polare, sennò non è nè acido nè base  

per ossiacidi: più O, più forte  
perchè aumenta anche la polarità del legame O-H  

**Coppie acido base coniugate**

base coniugata dell'acido: ciò che rimane dell’acido dopo che il protone è stato
donato  
esempio: $HA, A^-$  

acido coniugato della base: ciò che si forma dalla base dopo che ha accettato il protone   
esempio: $H_2O, H_3O^+$  

quindi devono differire solo per un protone   

**Forza relativa**

Più forte è l’acido, più debole è la base coniugata  
quindi la base coniugata di un acido forte ha proprietà basiche trascurabili, è uno ione spettatore  
eg. $HCl$ è forte, quindi $Cl^-$ è debole  

$H^+$: l'acido più forte che può esistere in equilibrio in soluzione acquosa  
$OH^-$: la base più forte che può esistere in equilibrio in soluzione acquosa  
Ogni acido o base che è più forte di H+ oppure OH- reagisce con l’acqua, per dare H+ oppure OH  
Si chiama effetto livellante del solvente  

**Lista**

acidi forti:  
HCl, HBr, HI (idruri con alogeno, tranne HF)  
HNO₃ (perchè acido coniugato di una base debole)   
HClO₄, H₂SO₄   

acidi deboli:  
HF (attenzione, sembra debba essere forte, ma è debole!)  
H₂CO₃, H₃PO₄  

basi forti:  
NaOH, KOH, Ca(OH)₂, Ba(OH)₂ (tutti ossiacidi metallici) 

basi deboli: 
NH3, AgOH  

#### Idrolisi (sali)

**Idrolisi**

cos'è: reazione chimica in cui le molecole vengono scisse in parti più piccole attraverso l'aggiunta di acqua  
cosa fa: permette a un sale di cambiare il ph di una soluzione, anche se è un composto neutro   
quinsi 'sale acido' è comunque un composto neutro, ma dopo l'idrolisi rende il ph più acido  

come funziona:  
1- metto sale in acqua, si dissocia in ioni  
per ora la soluzione non ha cambiato ph, perchè nella dissociazione non ci sono ioni H+  
2- ioni del sale reagiscono con l'acqua  
e solo ora liberano $H^+$ o $OH^-$, alterando il pH della soluzione  

quando avviene:  
solo gli ioni che vengono da basi/acidi deboli idrolizzano  
cioè reagiscono con l'acqua per rilasciare ioni $H^+$ o $OH^-$  

**Tipi di sali**

ricorda che solo gli ioni che vengono da basi/acidi deboli idrolizzano  
posso usare questa informazione per classificarli  

- sale neutro = Acido forte + Base forte   
Nessuno ione idrolizza -> pH ≈ 7  
Esempi: NaCl, KNO₃, Na₂SO₄  

- sale acido = Acido forte + Base debole  
solo catione idrolizza, liberando H⁺ -> pH < 7  
Esempi: NH₄Cl, AlCl₃, NaF  

- sale basico = Acido debole + Base forte  
solo anione idrolizza, liberando $OH^-$ -> ph > 7  
Esempi: Na₂CO₃, CH₃COONa

- soluzione tampone = acido + base coniugata  
pH rimane costante anche se aggiungo cose o diluisco
