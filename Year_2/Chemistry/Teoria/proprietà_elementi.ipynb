### Generale

**Termini**

gruppo: colonna (muoversi in un gruppo: scendere)  
scendere lungo una colonna vuol dire aumentare il numero quantico n  

periodo: riga (muoversi in un periodo: andare a destra)  

come ricordarlo:  
period = period = first period, la prima ora di scuola  
nell'orario scolastico, the first period is the first horizontal line  

**Che succede**

1- lungo un periodo (riga):   
aumenta il numero di protoni -> carica nucleare efficace aumenta  
gli elettroni aggiunti vanno nello stesso livello energetico -> schermaggio cambia poco  
quindi: AUMENTA l'attrazione del nucleo sugli elettroni, soprattutto su quelli più esterni  

2- lunga un gruppo (colonna):  
aumenta il numero di livelli elettronici -> gli elettroni di valenza sono più lontani dal nucleo, quelli interni schermano il nucleo   
quindi: DIMINUISCE l'attrazione del nucleo sugli elettroni più esterni  

quindi il max dell'attrazione è in alto a destra, il min in basso a sinistra  

**Recap max**

1- raggio: max su MIN attrazione  
perchè: gli elettroni esterni possono stare più lontani    

2- energia di ionizzazione: max su MAX attrazione  
perchè: elettroni più difficili da tirar via   

3- elettronegatività: max su MAX attrazione  
perchè: più attrae forte più è bravo ad attrarre gli elettroni del legame  

tip: guarda la tavola periodica  
in alto a destra c'è He, l'elio, con la E di elettronegatività ed energia  
in basso a sinistra c'è Fr, con la R di raggio

**Definizioni**

raggio atomico: metà della distanza minima tra i nuclei di due atomi adiacenti dello stesso elemento???  

elettronegatività: capacità di un atomo di attrarre a sé gli elettroni condivisi in un legame chimico  

energia di ionizzazione: l’energia richiesta per rimuovere un elettrone da un atomo neutro isolato

**Energia di prima ionizzazione**

l’energia richiesta per rimuovere un elettrone da un atomo neutro isolato:  
Na(g) → Na+(g) + e-  

l'energia di seconda ionizzazione è maggiore della prima, perchè ora ci sono meno elettroni, ognuno è attirato di più dal nucleo  

**Energia di ionizzazione: eccezioni**

di solito, lungo un periodo aumenta  
ma nel primo e il quarto elettrone p diminuisce  
Questo indica che sono le configurazioni ottenute togliendoli, cioè $ns^2$ e $np^3$, sono stabili  

perchè:  
$ns^2$: Gli elettroni s risentono di più dello schermo generato dagli elettroni interni  
$np^3$: se mettessi un secondo elettrone in un orbitale p, la repulsione interelettronica aumenta, quindi aggiungerne uno non è vantaggioso  


**Affinità elettronica / elettronegatività**

capacità di un atomo di attrarre a sé gli elettroni condivisi in un legame chimico  

energia in gioco quando un atomo neutro isolato acquista un elettrone  
esotermica: si sviluppa energia  
endotermica: viene assorbita energia  

$Cl_{(g)} + e^- → Cl^-_{(g)}$

massima nel gruppo 7  
lungo un periodo: cresce  
eccetto nelle configurazioni stabili: $s^2$, $s^2p^3$, $s^2p^6$, qui può esserci anche il segno +, non -  

lungo un gruppo: diminuisce  

**Nota: config stabili**

In generale l’affinità elettronica cresce da sinistra a destra  
MA è bassa o addirittura endotermica (il processo è sfavorito, nella tavola mi dà elettronegatività positiva invece che negativa), per le configurazioni già trovate stabili  
cioè la s2 e la s2p3, e oprattutto per la s2p6 (quelle dove l'energia di ionizzazzione diminuisce se aggiungo un altro elettrone)    

**Carica nucleare effettiva**

L’attrazione del nucleo che un elettrone effettivamente sente è una combinazione di:  
1- attrazione del nucleo   
2- repulsione degli elettroni dei gusci più interni (effetto di schermo)   

$z_{eff}=S-S$  
$z_{eff}$: carica effettiva  
Z: numero atomico  
S: numero di elettroni più interni  

### Reticolo ionico

**Reticolo ionico**

cosa: struttura tridimensionale ordinata e ripetitiva   
dove: si forma nei solidi ionici (cristalli ionici)    
perchè: a causa dell'attrazione elettrostatica tra ioni di carica opposta  

cioè la stabilità del reticolo ionico dipende da interazioni coulombiane tra ogni ione e tutti gli altri del cristallo    
In un solido ionico ogni ione sente la forza coulombiana di tutti gli altri ioni, non solo del vicino più prossimo  

**Studiare stabilità reticolo**

nome | energia reticolare | costante di Madelung   
---|---|---
cosa | energia per separare un solido ionico nei suoi ioni gassosi a distanza infinita | numero adimensionale  
dipende da | cariche, raggi, struttura | solo dalla geometria del reticolo, non dagli elementi usati 
misura/quantifica | la forza del legame ionico | le interazioni coulombiane in un reticolo ionico infinito  

**Perchè si crea solido ionico**

perchè è energeticamente conveniente:  
crare ioni costa energia (perchè devo strappare elettroni)  
ma impacchettarli nel reticolo ne libera molta

**Fattori energetici alla base della sua formazione**

1- Energia di ionizzazione  
energia necessaria per rimuovere un elettrone dall’atomo che diventerà catione  

2- Affinità elettronica  
energia liberata quando un atomo acquista un elettrone per formare l’anione  

3- Energia reticolare   
energia liberata quando i cationi e gli anioni gassosi si dispongono nel reticolo cristallino  
questa energia liberata è molto più grande delle altre due  

quindi:  
Il legame ionico si forma perché la forte energia reticolare liberata nel cristallo compensa l'energie necessaria per ionizzare l’atomo   

$\Delta E_{tot}=E_i -A_e-U_{ret}$  
se $\Delta E<0$, c'è formazione spotanea del reticolo  

#### Formule

**Energia elettrostatica di uno ione nel cristallo**

$E=-M \frac{1}{4 \pi \epsilon_0 }\frac{q_-q_+}{r_0}$    
è come forza elettrostatica tra due ioni, ma moltiplicata per M  
M è una sommatoria che diventa sempre più piccola, ha il compito di considerare l'apporto sempre più piccolo degli ioni man mano più lontani 

tips: posso scrivere $q_+=-z_+e$, $q_- =z_-e$    
$z_+, z_-$: cariche di cationi e anioni   

**Costante di Madelung**

$M=\Sigma_i \pm \frac{r_0}{r_i}$   

$r_i$: distanza dell'i-esimo ione più lontano  
$r_0$: distanza tra ioni adiacenti (cioè la più piccola distanza tra due ioni del reticolo)      
in questo modo, in E si cancella $r_0$, e sotto ci si trova invece $r_i$, così considero gli ltri ioni, per cui cambia solo la distanza  
$\pm$: il segno dipende se l’ione è di carica opposta o uguale all’ione di riferimento  

**Esempio: costante di Madelung di NaCl**

considero uno ione del reticolo, ha:  
6 ioni opposti a distanza $r_0$    
12 ioni uguali a distanza $\sqrt{2} r_0$  
...  
quindi: $M= 6-\frac{12}{\sqrt{2}}+ ... \sim 1.74$  

**Formula di Born-Landé**

$U= N_A E (1- \frac 1n) = -\frac{N_AMz_+z_-e^2}{4\pi\epsilon_0r_0} (1-\frac{1}{n})$  

E: energia di uno ione del cristallo    
U: energia di una mole di cristallo (energia reticolare)   
$N_a$: numero di avogadro    

cosa è l'ultima parte:  
Se esistesse solo l’attrazione coulombiana, gli ioni collasserebbero uno sull’altro  
Quando gli elettroni si sovrappongono compare una repulsione quantistica (principio di Pauli)  
Questa repulsione viene modellata come: $\frac B {r^n}$  
n : esponente di Born (repulsione a corto raggio)
