#### Teoria distribuzioni

**Important ones**

name | formula | params
---|--- |---
binomial |  $B(k; n,p)=  \binom{n}{k}p^k(1-p)^{n-k}$ | p: prob of one ty of the event succeding (having outcome 1) <br> n: how many tries  
exponential |  $\epsilon(x;, \lambda) = \begin{cases} \lambda e^{-\lambda x} & x \in [0, \infty] \\ 0 & else \end{cases}$ | $\lambda= \frac 1 \tau$: costante di decadimento (1/emivita)
poissonian | $P(k;\mu) = \frac{\mu^k}{k!} e^{-\mu}$ | $\mu$: media degli eventi nell'intervallo  

**Poisson**

ipotesi:  
1- indipendensa degli eventi (prob $\lambda$ non dipende da N, il numero di atomi rimasti)  
2- stazionarietà ($\lambda$ è costante nel tempo)  
3- non simultaneità  

formule:  
$P(k;\mu) = \frac{\mu^k}{k!}e^{-\mu}$  
$E[k] = Var[k] = \mu$  

dove trovo $\mu$:  
$\mu = \int_t^{t+\Delta t}A(t') \, dt'$  
perchè $\mu$ è la media di eventi nell'intervallo $\Delta t$, è A(t) è il numero di eventi nell'intervallo t  

**Esponenziale**

è una distribuzione continua  
quindi la densità di prob $f_T(t)$ mi dice la prob che la variabile T (tempo sopravvissuto, tempo a cui il nucleo decade) si trovi nell'intervallo (t+dt)  

a me interessa la prob che la var T sia maggiore o minore di qualche valore:  
$P(T> t) = \int_t^\infty f_T(t) \, dt = e^{-\lambda t}$   
$P(T < t) = \int_0^t f_T(t) \, dt = 1- e^{-\lambda t}$   

la media è:  
$E[T] = \frac 1 \lambda = \tau$ (che chiamo emivita)  

**Binomiale**

$E[N(t)] = np = N_0e^{-\lambda t}$    
quanto atomi mi aspettano siano rimasti al tempo t, in media

##### Memory-less

**Memory-less: examples**

intuitive example: flip a coin (it's memory-less, but NOT exponential distribution)  
prob of getting head does not depends on what happened on previous throws  
even if before got all tails, prob is always 0.5  

harder example: decay (it's memory-less AND has exponential distribution)  
a radioactive nucleus has 30% prob of not decaying in the first 1s  
after 10s, i check and it's still there: again, the nucleus has 30% prob of not decaying in the successive 1s (come se riazzerassi il tempo)  
don't know if it's the starting 1s or not, don't care  

essentially only care about how long is the time interval before something happens, not when it started  
make it start at the last time i checked and was sure it didn't happen yet  
don't matter how much time passed before that, it's ininfluent  

**Memorylessness**

probability of an event happening don't depend on how much i have already waited  

written formally:  
$P(x \ge s+t | x \ge s) = P(x \ge t)$  

bc in this case, writing the cumulative function gives:  
$P(x \ge s + t) = P (x \ge s) P(x \ge t)$

### Nel decadimento

**Distributions appearing**

distribution | variabile aleatoria | formula  | argomenti 
---|---|---|---
binomiale | $N(t)$: numero di nuclei non ancora decaduti al tempo t <br> $D(t)$, quanti eventi accadono in un tempo t (quanti nuclei decaduti entro il tempo t)| $N(t) \sim B(N_0, e^{-\lambda t})$ <br> $D(t) \sim B(N_0, 1-e^{-\lambda t})$| $n=N_0$: numeri di nuclei iniziali  <br> $p=p(T> t)=e^{-\lambda t}$ prob che un nucleo non sia ancora decaduto al tempo t (segue distrib esponenziale)  
Poisson | $K=D(\Delta t)$ : numero di eventi in un BREVE intervallo ($\Delta t \ll \tau$) | $K  \sim P(A \Delta t)$  | $\mu \sim \lambda N \Delta t =A \Delta t$ : numero medio di decadimenti per unità di tempo in quel momento <br> A: attività, numero medio di decadimenti per unità di tempo  
esponenziale | $T$ : tempo di decadimento di un nucleo preciso, con costante di dec $\lambda$ | $T \sim  \epsilon(\lambda )$ |$\lambda$: costante di decadimento    
esponenziale | $T_{next}$ : tempo fino al prossimo decadimento, di un qualunque nucleo edl campione | $T_{next} \sim \epsilon(N\lambda) = \epsilon(A)$ | N: numero di atomi del campione <br> $\lambda$: decay constant di ogni atomo del campione  

evento = decadimento   

**Osservazione**

$N(t) \sim B(N_0, e^{-\lambda t})$    
cioè segue una binomiale  

di solito, a me interessa in media quanti N sono rimasti al tempo t, cioè:  
$E[N(t)] = N_0 e^{-\lambda t}$   
spesso questo lo chiamo direttamente N(t), omettendo E[]  

**Tasso di decadimento**

$\lambda$: per un singolo nucleo  
$A=\lambda N$: per l'itnero campione di N nuclei  

se N cambia poco:  
$\mu \sim A \lambda \Delta t$  
$t = \epsilon(N\lambda)$ (t: tempo da due decadimenti del campione)

**Poisson vs binomiale**

1- condizioni:  
$t \rightarrow 0$  
$N_0 \rightarrow \infty$  
cioè considero un intervallo piccolo e tanti nuclei  

2- conseguenze:  
$p=1-e^{-\lambda} t \rightarrow 0$ (prob nucleo già decaduto al tempo t)  
$N(t) \sim N_0$ (perchè troppo poco tempo per diminuire vistosamente, vistoche sono tanti)  

3- conclusione  
binomiale -> poisson  
con $\mu = np$  

4- calcola la media  
$\mu = N_0 (1-e^{-\lambda \Delta t}) \sim N(\lambda \Delta t) = A \Delta t$    
poichè $t \sim 0$,  uso Taylor sull'esponenziale    
e ricorda che posso scrivere $N(t) \sim N_0$  

**Binomiale vs Poisson: quando usarle**

1- binomiale  
nuclei deaduti in lunghi periodi  
dove ne decadono tanti  

2- Poisson    
nuclei ne caduti in corti periodi  
il numero di decaduti deve essere negligible rispetto al numero di nuclei iniziale

**Incertezza sui conteggi**

voglio incertezza statistica dellì1%, cioè:  
$\frac{\sigma_N}{N}= 1%$    
poichè per una poissoniana, $Var(N) = N$, posso fare:  
$\frac{\sigma_N}{N} \sim \frac{\sqrt{N}}N = 1%$  
$N = 10^4$  

Questo \(1\%\) è una deviazione standard; non significa che il risultato sia garantito entro l’1%.
