**Modello: decadimento**
  
$N \xrightarrow{\lambda} M$  

$\begin{cases}
\frac{dN}{dt}=-\lambda N \\
\frac{dM}{dt} = -\frac{dN}{dt} = \lambda N
\end{cases}$

N=genitore, M=figlio  
per ogni N che decade, si forma un M  
N decade, quindi diminuiscono, ha derivata negativa     

### Tipi di decay

**Tipi di decadimento**

decadimento: quando qualcosa scompare, e si crea qualco'altro  
decadimento radioattivo: il processo di decadimento fatto da nuclei radioattivi   

non tutti i decadimenti sono esponenziali  
lo sono solo quelli che sono processi poissoniani (eg. decadimenti radioattivi)  
alcuni invece non rispettano l'indipendenza (ad esempio quando ho A+A->B, quindi se la quantità di A aumenta, è più facile che si combinino per far avvenire l'evento)  

**Formule decay**

1- vale per OGNI DECAY:    
$dP = \lambda \; dt$  
dP: prob che fino al tempo t non sia ancora decaduta, e nel successivo tempo dt decada  

2- ONLY RADIOACTIVE DECAY: $\lambda$ è COSTANTE   
cioè, la prob di un nucleo di decadere NON dipende da quanti nuclei ci sono (in chimica, reazione di ordine 0)

### Modello: casi comuni

**Come risolvere**

1- disegna lo schema dei decadimenti (cosa decade in cosa)  

2- scrivi le condizioni iniziali (quantità di ogni elemento al tempo 0)  

3- per ogni elemento, scrivi l'eq diff $dN/dt$   
a destra, metti:  
$-\lambda N$ per tutti i decadimenti di cui è reagente  
$\lambda N_{reagente}$ per tutti i decadimenti di cui è prodotto     
come quando a dinamica, ogni forza, la aggiungenvo alla ma di entrambi gli oggetti coinvolti, con segno diverso  

4- risolvi il sistema   
inizio risolvendo quella del reagente, poi pluggo la N(t) trovata nell'eq diff del prodotto   

**Facile: un solo decadimento**

$N \xrightarrow{\lambda} M$  
condizioni: $N(0)=N_0 , M(0)=0$  

modello:  
$\begin{cases}
\frac{dN}{dt}=-\lambda N \\ 
\frac{dM}{dt} = \lambda N
\end{cases}$

risolvo:   
$\begin{cases} 
N(t) = N_0e^{-\lambda t} \\
M(t)=N_0-N(t) = N_0(1- e^{-\lambda t})   
\end{cases}$

**Catena**

$N_1 \xrightarrow{\lambda_1}N_2 \xrightarrow{\lambda_2}N_3\xrightarrow ... \xrightarrow{\lambda_n} N_n$  
$N_1(0)=N_0, \; N_2(0)=N_3(0)=N_{n} =0$   

sistema:  
$\begin{cases}
\frac{dN_1}{dt}=-\lambda_1N_1 \\
\frac{dN_i}{dt}=\lambda_{i-1}N_{i-1} -\lambda_iN_i & i \in [2, n-1]\\
\frac{dN_n}{dt}=\lambda_{n-1} N_{n-1} \\
\end{cases}$  
il sistema di chiama: BATEMAN equations  

risolvo:  
$\begin{cases}
N_1(t) = N_0e^{-\lambda_1 t} \\
N_i(t) = N_0 \Sigma_{j=0}^i \; [(\prod_{k=0, k \neq j}^i \frac{\lambda_k}{\lambda_k-\lambda_j}) e^{-\lambda_j t}] & i \in [2, n]
\end{cases}$  

**Parallelo**

$N = \begin{cases}
\xrightarrow{\lambda_1} M_1 \\
\xrightarrow{\lambda_2} M_2 \\
\xrightarrow{\lambda_n} M_n
\end{cases}$
$N(0)=N_0, M_i(0)=0$  

sistema  
$\begin{cases}
\frac{dN}{dt}=-(\lambda_1 + ... + \lambda_n)N \\ 
\frac{dM_i}{dt} = \lambda_i N
\end{cases}$  

risolvo  
$\begin{cases} 
N(t) = N_0e^{-\lambda_t t} \\
M_i(t)=\frac{\lambda_i}{\lambda_T} N_0 (1-e^{-\lambda t})  
\end{cases}$   
$\lambda_T = \Sigma \lambda_i$  
$b_i = \frac{\lambda_i}{\lambda_T}$ branching ratio, rapporto di decadimento (prob che il nucleo decada nell' i-esimo modo)  

### Equilibrio secolare

**Cosa**

la quantità e l'attività ($N(t)$ e $A(t)$) di un isotopo radioattivo figlio rimangono costanti nel tempo  
perché il suo tasso di produzione (per decadimento del nucleo padre) è uguale al suo tasso di decadimento  

formule:  
$A_1(t)≃A_2(t)$   
$N_1(t) = const$  
$N_2(t) = const$    
il rapporto tra la quantità degli elementi, in ogni momento, dipende dai rapporti tra le $\lambda$  

**Quando avviene**

1- $\lambda_2 \gg \lambda_1$    
cioè: in una catena, il figlio decade molto più veloce del padre 
cioè il padre vive molto più a lungo del figlio   

2- considero un tempo lungo rispetto all'emivita del figlio  
così $N_P(t) \sim const$, quindi $A_P(t) \sim const$  

**Conseguenze**

il rapporto tra la quantità degli elementi, in ogni momento, dipende dai rapporti tra le $\lambda$   
in generale, $N_F \ll N_P$  

esempio: decay uranio  
$U→X→Y→Pb$   
ciascuno ha pochi atomi rispetto a $N_U$, cioè $N_u \gg N_i \quad \forall i$  
e può essere trascurato nel bilancio del numero di atomi    
quindi: $A_u = A_{Pb}$   
cioè: atomi U decaduti = atomi Pb creati -> fingo che non ci siano atomi U già decaduti, ma che non hanno ancora fatto tutta la catena fino a Pb   
intuitivamente: poichè X decade più veloce di U, appena X viene creato decade subito tutto

**Spiegazione con grafico**

all'inizio c'è solo padre, non figli  

1- attività del padre:   
considero un tempo $t << \tau_1$  
quindi il padre rimane quasi costante: $N_P(t) \sim N_0$    
quindi la sua attività rimane costante: $A(t) = N_P(t) \lambda \sim N_0 \lambda$   

2- attività del figlio:  
all'inizio il numero di figli $N_F(t)$ cresce (prodotti dal padre)  
ma anche $A_F(t)=N_F(t) \lambda$ cresce, finchè $A_P(t) = A_F(t)$  
cioè il numero di figli creati al secondo è uguale al numero di figli decaduti al secondo  
quindi $N_F(t)$ rimane costante da lì in poi, quindi anche $A_F(t)$  

**Proof: equilibrio secolare**

$N_1 \xrightarrow{\lambda_1}N_2 \xrightarrow{\lambda_2}$  
$N_1(0)=N_0, \; N_2(0)=0$  

modello:  
$\begin{cases}
\frac{dN_1}{dt}=-\lambda_1N_1 \\
\frac{dN_2}{dt}=\lambda_1N_1 -\lambda_2N_2
\end{cases}$  

risolvo:  
$\begin{cases} 
N_1(t) = N_0e^{-\lambda_1 t} \\
N_2(t)=\frac{\lambda_1}{\lambda_2-\lambda_1} N_0 (e^{-\lambda_1 t}-e^{-\lambda_2 t})  
\end{cases}$   

confronto le attività: $A_i(t) = \lambda_i N_i(t)$    
$\frac{A_2(t)}{A_1(t)} = \frac{\lambda_2}{\lambda_2-\lambda_1}(1-e^{-(\lambda_2-\lambda_1)t})$  

condizione 1: $\lambda_1 \ll \lambda_2$   
allora: $\lambda_2-\lambda_1 \sim \lambda_2$  
$\frac{A_2(t)}{A_1(t)} ≃ 1 (1-e^{-\lambda_2t})$  

condizione 2: $t \gg \tau_2 = \frac 1 {\lambda_2}$  
allora: $e^{-\lambda_2t} \sim 0$   
$\frac{A_2(t)}{A_1(t)} ≃ 1 $  

**Proof più veloce**
    
1- c'è un momento in cui $\frac{dN_2}{dt}=0$  
cioè: $\lambda_2 N_2 =\lambda_1N_1$     
cioè $A_1(t) = A_2(t)$  

2- per tempi lunghi: $N_1(t) = N_0e^{-\lambda_1t}$   
quindi $N_1$ decade tutto  

3- per tempi brevi (rispetto all'emivitia di $N_1$): $\lambda_1t = t/\tau \ll 1$  
quindi approssimo l'exp con Taylor $e^{-\lambda_1t} \sim 1$   
quindi: $N_1(t) \sim N_0$  
cioè $N_1$ è costante, quindi anche $N_2$ deve essere constante, come da osservazione 1
