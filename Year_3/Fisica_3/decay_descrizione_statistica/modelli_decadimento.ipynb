### Modello base

**Modello: decadimento**
  
$N \xrightarrow{\lambda} M$  

$\frac{dN}{dt}=-\lambda N$  
$\frac{dM}{dt} = -\frac{dN}{dt} = \lambda N$  

N(t): nuclei NON ancora decaduti al tempo t (ha il $-\lambda$ perchè il numero di nuclei rimasti diminuisce nel tempo dt)  
M(t): nuclei prodotti del decadimento al tempo t ( ha il $+\lambda$ perchè il numero di nuclei decaduti/creati aumenta nel tempo dt)  

**Come risolvere**

1- disegna lo schema dei decadimenti (cosa decade in cosa)  
2- scrivi le condizioni iniziali (quantità di ogni elemento al tempo 0)  
3- per ogni elemento, scrivi l'eq diff $dN/dt$   
a destra, metti:  
$-\lambda N$ per tutti i decadimenti di cui è reagente  
$\lambda N_{reagente}$ per tutti i decadimenti di cui è prodotto     
come quando a dinamica, ogni forza, la aggiungenvo alla ma di entrambi gli oggetti coinvolti, con segno diverso  
4- risolvi il sistema   
inizio risolvendo quella del reagente, poi pluggo la N(t) trovata nell'eq diff del prodotto   

#### Teoria

**Ipotesi**

vale per ogni decadimento  
$dP = \lambda \; dt$  
dP: prob che fino al tempo t non sia ancora decaduta, e nel successivo tempo dt decada  

per un decadimento radioattivo: $\lambda$ è COSTANTE  

**Tipi di decadimento**

decadimento: quando qualcosa scompare, e si crea qualco'altro  
decadimento radioattivo: il processo di decadimento fatto da nuclei radioattivi   

non tutti i decadimenti sono esponenziali  
lo sono solo quelli che sono processi poissoniani (eg. decadimenti radioattivi)  
alcuni invece non rispettano l'indipendenza (ad esempio quando ho A+A->B, quindi se la quantità di A aumenta, è più facile che si combinino per far avvenire l'evento)  

**Spiegazione modello**

1- modello N  
$dN = -N \, dP = -N \; \lambda \, dt$  
attenzione:   
$dN$: variazione di N, il numero di nuclei rimasti, nel tempo dt -> se N diminuisce, <0   
$N \lambda dt$ : numero di nuclei che decadono nel tempo dt -> se N diminuisce, >0  
quindi cose opposte, per questo c'è il -    

2- modello M  
per ogni N che decade, si crea un M  
quindi dN = dM  

nota: per farlo più formale  
invece di dN, si può scrivere N(t+dt)-N(t)  
poi dividere per dt e fare il lim dt->0  

**Come studiare decadimento**

non posso predirre quando un nucleo preciso decadrà  
ma posso predirre statisticamente, quale frazione di un campione sarà decaduta in un intervallo di tempo

è il nucleo che decade, non l'intero atomo   
The electrons can subsequently rearrange, so atomic radiation may also be emitted, but the radioactive transformation originates in the nucleus.

**Activity**

A: expected number of decays per unit time (decay rate)   
$A(t) = -\frac{dN}{dt} = \lambda N(t)$  

A: si misura con contatore Geiger -> meglio sperimentalmente  
N: si misura studiando il nucleo con tecniche elettromagnetiche -> difficile  

$A = -\frac{dN_{rimasti}}{dt} = \frac{dN_{decaduti}}{dt}$  
di solito con N intendo quelli rimasti, quindi ci va il -  

unità di misura Bq (Bequerel), numero di deadimenti al secondo  


A: numero di decadimenti al secondo  
N: numero di campioni NON ancora decaduti  


### Modello: casi comuni

**Facile: un solo decadimento**

$N \xrightarrow{\lambda} M$  
condizioni: $N(0)=N_0 , M(0)=0$  

modello:  
$\begin{cases}
\frac{dN}{dt}=-\lambda N \\ 
\frac{dM}{dt} = \lambda N
\end{cases}$

risolvo:   
$\begin{cases} 
N(t) = N_0e^{-\lambda t} \\
M(t)=N_0-N(t) = N_0(1- e^{-\lambda t})   
\end{cases}$

**Catena**

$N_1 \xrightarrow{\lambda_1}N_2 \xrightarrow{\lambda_2}N_3\xrightarrow ... \xrightarrow{\lambda_n} N_n$  
$N_1(0)=N_0, \; N_2(0)=N_3(0)=N_{n} =0$   

sistema:  
$\begin{cases}
\frac{dN_1}{dt}=-\lambda_1N_1 \\
\frac{dN_i}{dt}=\lambda_{i-1}N_{i-1} -\lambda_iN_i & i \in [2, n-1]\\
\frac{dN_n}{dt}=\lambda_{n-1} N_{n-1} \\
\end{cases}$  
il sistema di chiama: BATEMAN equations  

risolvo:  
$\begin{cases}
N_1(t) = N_0e^{-\lambda_1 t} \\
N_i(t) = N_0 \Sigma_{j=0}^i \; [(\prod_{k=0, k \neq j}^i \frac{\lambda_k}{\lambda_k-\lambda_j}) e^{-\lambda_j t}] & i \in [2, n]
\end{cases}$  

**Parallelo**

$N = \begin{cases}
\xrightarrow{\lambda_1} M_1 \\
\xrightarrow{\lambda_2} M_2 \\
\xrightarrow{\lambda_n} M_n
\end{cases}$
$N(0)=N_0, M_i(0)=0$  

sistema  
$\begin{cases}
\frac{dN}{dt}=-(\lambda_1 + ... + \lambda_n)N \\ 
\frac{dM_i}{dt} = \lambda_i N
\end{cases}$  

risolvo  
$\begin{cases} 
N(t) = N_0e^{-\lambda_t t} \\
M_i(t)=\frac{\lambda_i}{\lambda_T} N_0 (1-e^{-\lambda t})  
\end{cases}$   
$\lambda_T = \Sigma \lambda_i$  
$b_i = \frac{\lambda_i}{\lambda_T}$ branching ratio, rapporto di decadimento (prob che il nucleo decada nell' i-esimo modo)

#### Equilibrio secolare

**Equilibrio secolare**

cosa:  
la quantità (e l'attività) di un isotopo radioattivo figlio rimangono costanti nel tempo  
perché il suo tasso di produzione (per decadimento del nucleo padre) è uguale al suo tasso di decadimento  

quando:  
1- in una catena, il figlio decade molto più veloce del padre  
cioè il padre vive molto più a lungo del figlio  
$\lambda_2 \gg \lambda_1$  
2- in un tempo lungo rispetto all'emivita del figlio  

in formule:  
NON vuol dire $N_1=N_2$  
ma vuol dire che le loro ATTIVIT° sono simili: $A_1(t)≃A_2(t)$
dopo un po' di tempo, si instaura l'equilibrio  
cioè: $\frac{dN_1}{dt=0}$  
$\lambda_1N_1 = ... = \lambda_n A_n$  
il rapporto tra la quantità degli elementi, in ogni momenti , dipende dai rapporti tra le $\lambda$  

**Nota**

il figlio decade più veloce del padre  
ma c'è in quantità minore (perchè all'inizio c'è solo padre, non figlio), quindi:  
relativamente al suo totale decade più veloce (cioè ha $\lambda$ maggiore)  
ma in valore assoluto decadono meno nuclei  

**Intuitivo**

$U→X→Y→Pb$  
se U decade molto più lentamente degli altri:  
in 1s si formano 100 X, e ne decadono pure 100  
quindi il numero diaotmi di X rimane costante, anche se i singoli atomi vengono continuamente sostituiti  

se ciascun intermedio decade molto più rapidamente dell’uranio:  
ciascuno ha pochi atomi rispetto a $N_U$, cioè $N_u \gg N_i \quad \forall i$  
e può essere trascurato nel bilancio del numero di atomi    
poichè X decade più veloce di U, appena  X viene creato decade subito tutto  
quindi: atomi X creati = atomi X decaduti  
in formule: $A_u = A_x$  
perchè $A_i$ è il numero di decadimenti avvenuti alla specie i  
e ogni atomo x creato corrisponde al decadimento di un atomo U      

**Con numeri**

dati:  
figlio decade più veloce del padre: $\lambda_1 = 0.0001 \quad \lambda_2 = 1$      
all'inizio c'è solo padre, $N_0 = 10000$  

che succede:  
il padre ha un'emivita grande: se considero $t \ll T_{1/2}$, allora rimane quasi costante $N_P(t) \sim N_0$  
produce figli: $M=\lambda_1 N_0$ = 10 figli/s  
all'inizio ci sono figli, ma pian piano iniziano a crescere  
poichè i figli decadono più veloce, quando ci sono solo 1o figli, allora le attività di padre e figlio sono uguali  
cioè numero di figli creato è numero di figli distrutto  


**Proof: equilibrio secolare**

$N_1 \xrightarrow{\lambda_1}N_2 \xrightarrow{\lambda_2}$  
$N_1(0)=N_0, \; N_2(0)=0$  

modello:  
$\begin{cases}
\frac{dN_1}{dt}=-\lambda_1N_1 \\
\frac{dN_2}{dt}=\lambda_1N_1 -\lambda_2N_2
\end{cases}$  

risolvo:  
$\begin{cases} 
N_1(t) = N_0e^{-\lambda_1 t} \\
N_2(t)=\frac{\lambda_1}{\lambda_2-\lambda_1} N_0 (e^{-\lambda_1 t}-e^{-\lambda_2 t})  
\end{cases}$   

confronto le attività: $A_i(t) = \lambda_i N_i(t)$    
$\frac{A_2(t)}{A_1(t)} = \frac{\lambda_2}{\lambda_2-\lambda_1}(1-e^{-(\lambda_2-\lambda_1)t})$  

condizione 1: $\lambda_1 \ll \lambda_2 = \tau_2 $   
allora: $\lambda_2-\lambda_1 \sim \lambda_2$  
$\frac{A_2(t)}{A_1(t)} ≃ 1 (1-e^{-\lambda_2t})$  

condizione 2: $t \gg \frac 1 {\lambda_2}$  
allora: $e^{-\lambda_2t} \sim 0$ 
$\frac{A_2(t)}{A_1(t)} ≃ 1 $  

**Proof più veloce**
    
1- c'è un momento in cui $\frac{dN_2}{dt}=0$  
cioè: $\lambda_2 N_2 =\lambda_1N_1$     
cioè $A_1(t) = A_2(t)$  

2- per tempi lunghi: $N_1(t) = N_0e^{-\lambda_1t}$  
quindi $N_1$ decade tutto  

3- per tempi brevi (rispetto all'emivitia di $N_1$): $\lambda_1t = t/\tau \ll 1$  
quindi approssimo l'exp con Taylor $e^{-\lambda_1t} \sim 1$   
quindi: $N_1(t) \sim N_0$  
cioè $N_1$ è costante, quindi anche $N_2$ deve essere constante, come da osservazione 1
