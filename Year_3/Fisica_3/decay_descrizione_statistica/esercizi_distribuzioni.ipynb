#### Stima vita media protone (con corpo umano)

**Ipotesi**

un protone che decade all'interno del corpo umano rilascia energia, assorbita in parte dal corpo umano 
se durante la vita assorbissimo $D_{max}$, saremmo morti  

domanda: trova upper bound per $\tau_p$, emivita di un protone 

**Cosa sono**

D(T): dose assorbita nel tempo T  
$ T = 100 y$ : tempo della vita umana  
M: massa corpo umano (si semplifica)  
$m_u=931 MeV$: massa nucleone  
f=1 : frazione dell'energia liberata dal decay assorbita dal corpo umano (metto 1 per semplificare i conti)  
$N_p$: numero di protoni nel corpo all'inizio della vita (assumo che poi non se ne formino di nuovi nel corso della vita)  
$\gamma_p$: rapporto tra numero di protoni e neutroni $N_p = \gamma N_{nucleoni}$ 
$m_p$: massa di un protone (non è uguale a m_u??? sì???)

**Modello**

$\begin{cases}
D(T) = \frac{E_{abs}}{M} = \frac{N_{dec}(T) E_{dec}}M \\
E_{dec} = f \, m_pc^2 \\
N_{dec}(T) = N_p-N_pe^{-T/\tau} = N_p(1-e^{T/\tau}) \sim N_p \frac T \tau & T/\tau \ll 1 \\  
N_p = \gamma N_{nucleoni} = \gamma_p \frac M {m_u} 
D(T) \le D_{max}  
\end{cases}$

quindi:   
$D(t) = \frac{N_{dec} E_{dec}}M = \frac{(\gamma_p \frac M {m_u}\frac T \tau) \; (f m_p c^2)}M = \frac{\gamma_p f m_pc^2}{m_u} \frac T \tau \le D_{max}$  
$\tau > 10^{18} y$  

#### Stima vita media protone (con rilevatore)

**Proprietà dell'apparato**

1- FONDO  
ci sono altre cose/particelle/eventi che potrei confondere con quella che mi interessa?   
eg. raggi cosmici che colpiscono il rilevatore possono sembrare decay di protoni  

2- EFFICIENZA  
quanti degli eventi che accadono mentre misuro vengono davvero a rilevare?  
eg. 40%: rileva solo il 40% dei decay che avvengono davvero  
Frazione degli eventi che vengono rilevati  
$\varepsilon = \frac{N_{rilevati}}{_{eventi}}$  

3- ESPOSIZIONE  
$\varepsilon = N_P \; T$   
Massa attiva osservata per un certo tempo  
numero di protoni osservati x tempo di osservazione  
statisticamente: osservare 1 protone per 100 anni è uguale ad osservare 100 protoni per 1 anno    

4- SENSIBILITÀ  
Variazione del segnale \(S\) per variazione della grandezza \(x\)  
$s = \frac {dS}{dx}$  
non mi interessa, la ignoro  

**Cosa sono**

T=1y: tempo di osservazione  
$P(k \,u)$: poissoniana, la distrib che seguono i decay  
K=0: ho osservato 0 decay  

$\mu_{teo}$: quanti decay in media nel tempo T (secondo la teoria)  
$\mu$: numero medio di decay registrati davvero dal rilevatore  
$\mu_{90}$: limite superiore su $\mu$, ricavato dall’osservazione di zero eventi al 90% di confidenza    

NO fondo    
$\varepsilon = 40%$

**Confidence level**

devo stimare $\mu$  
non lo so stimare, ma so trovare un limite superiore (cioè $\mu_{90}$)  

ho osservato 0 eventi  
per mettere un limite superiore, cerco una media abbastanza grande da rendere questo risultato POCO probabile  
cioè: $P(K \ge 1, \mu_{90})= 0.90$  
Se quella media fosse vera, nel 90% degli esperimenti vedresti almeno un evento   
io invece ne hai visti zero, quindi lo uso come soglia superiore (perchè è improbabile che $\mu$ sia più estrema si così, già mi sembra molto distante dalle mie osservazioni)  

so che $\mu < \mu_{90}$, percjè è stima superiore  

**Sistema**

$\begin{cases}
K \sim P(k; \mu) = \frac{\mu^k e^{-\mu}}{k!} & P(0; \mu) = e^{-\mu} \\
\mu_{teo} =N_p (1-e^{-T/\tau}) \sim N_p \frac T \tau \\
\mu = \varepsilon \mu_{teo} \\
P(K \ge 1, \mu_{90}) =  1-e^{-\mu_{90}} = 0.90 \\
\mu < \mu_{90} 
\end{cases}$

quindi:  
$\mu_{90}=2.3$  
$\tau > 10^{33} y$  

**CL - confidence level**

per gli upper o lower bound, di solito si usa CL=90  
per i $\pm$ (cioè gli errori), di solito si usa CL=68
