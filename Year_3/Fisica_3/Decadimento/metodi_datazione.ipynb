### Catene uranio-piombo

#### Modello

**Dati**

t : tempo attuale (oggi)  
t=0: quando si è formata la roccia / l'uranio ha cominciato a decadere  

**Ipotesi**

1- $N_{pl}(0)=0$  
non c'era piombo spontaneamente nella roccia, c'è solo quello dovuto al decadimento dell'uranio

2- sistema chiuso  
cioè il piombo creato non può uscire dalla roccia  

conseguenza: $N_0= N_{pl}(t)+N_u(t)$  

**Modello**

$\begin{cases}
{}^{235}U \xrightarrow{\lambda_5} {}^{207}Pl \\   
{}^{238}U \xrightarrow{\lambda_8} {}^{206}Pl \\   
\end{cases}$

$\begin{cases}
N_u(t) = N_0e^{-\lambda t} \\
N_{pl} (t) = N_0-N_u(t) = N_0(1-e^{-\lambda t}) \\
\end{cases}$  
$R(t) = \frac{N_{pl(t)}}{N_u(t)}= e^{\lambda t}-1$  
quindi il ratio oggi non dipende da $N_0$; il numero di atomi di uranio all'inizio

#### Plot

**Utilizzo**

misuro il rapporto tra U e Pl oggi nella roccia, cioè $R(t) = \frac{N_{pl(t)}}{N_u(t)}$    
da questo riesco a trovare t, il tempo passato dalla formazione della roccia

**Controllo**

ho due catene uranio-piombo possibili   
se le mie ipotesi sono esatte, devono darmi lo stesso risultato t  

**Cosa plotto**

plotto i rapporti delle due catene  
ottengo una curva parametrizzata  
nella prima trovo t(x), la pluggo nella seconda per avere y(x)  

$\begin{cases}
x(t) = \frac{N_{206}(t)}{N_{238}(t)} = e^{\lambda_8t} -1 \\
y(t) = \frac{N_{207}(t)}{N_{235}(t)} = e^{\lambda_5t} -1 \\
\end{cases}$  

$y(x) \sim (x-1)^{\lambda_5/\lambda_8}-1 \sim (x-1)^6$     
si chiama CURVA DI CONCORDIA

**Come usarla**

misuro i rappori tra gli elementi delle due catene  
mi dà un punto di coordinate (x,y) = (ratio_1, ratio_2), come mostrano sopra  

se le ipotesi del moedllo sono giuste:  
questi punti appartengono alla curva di concordia  

se una non è vera:  
i punti sono su una retta, detta RETTA DI DISCORDIA  


**Retta di discordia**

di solito, l'ipotesi non vera è quella di sistema cbiuso  
vuol dire che, temporanemante, non è stato un sistema chiuso (ma poi lo è ridiventato)  
ad esempio, un evento ha fatto ha fatto perdere del piombo al materiale 

la retta taglia la curva in due punti  
se ipotizziamo che la perdita di piombo sia dovuta ad un unico evento:  
intersezione superiore: l’età di formazione del minerale  
intersezione infeiore: l’età dell’evento che lo ha alterato

#### Problema

**Problema**

in teoria ci sono stati intermedi del decadimento tra uranio e piombo  
quindi dovrebbe essere  $N_0= N_{pl}(t)+N_u(t) + \sum N_{intermedi}(t)$  

**Soluzione: usare eq. secolare**

condizione:  
$\lambda_u \ll \lambda_i$, dove $\lambda_i$ sono tutti quelli delle trasformazioni seguenti    
cioè l'uranio decade più piano di tutti i suoi figli   

conseguenza:  
si instaura eq. secolare:  
$A_i = A_u$      
$\lambda_i N_i = \lambda_u N_u$  
$N_i = \frac{\lambda_u}{\lambda_i} N_u \ll N_u$  

conclusione:  
quindi posso trascurare gli stati intermedi, ognuno ha pochissimi atomi  
posso dire comunque $ $N_0= N_{pl}(t)+N_u(t) $  

**Versione sbagliata**

NON bisogna farlo misurando il ratio oggi tra i due isotopi dell'uranio  
perchè nel modello non si cancella $N_0$  
quindi dovrei aggiungere l'ipotesi forte che all'inizio erano 50:50, che non è vero (erano 70:30)  

### Carbonio 14

#### Modello

**Come funziona**

raggi cosmici creano ${}^{14}C$ nell'atmosfera  
il rapporto tra $R_{atm}(t)={}^{12}C/{}^{14}C$ varia nel tempo  

organismi viventi:  
finchè vivono, scambiano carbonio con l'atmosfera, quindi hanno anche loro $R=R_{atm}$  
dopo che muoiono, ${}^{12}C$ rimane uguale perchè è stabile, mentre ${}^{14}C$ decade  

**Modello**

t=0: momento in cui l'organismo è morto    
t: oggi  

il rapporto tra i carboni nell'rganismo ad un tempo t successivo è:  
$R(t) = \frac{N_{14}(t)}{N_{12}(t)} = \frac{N_{14}(0)e^{-\lambda t}}{N_{12}(0)} = R_{atm}(0) e^{-\lambda t}$    

**Utilizzo**

misuro il rapporto tra i carboni in un organismo oggi, cioè $R(t)$  
se sapessi $R_{atm}(0)$, cioè qual'era il rapporto nell'atmosfera nell'istante in cui è morto, potrei calcolare t (tempo passato dalla morte)  

problema: $R_{atm}(t)$ non è costante  
serve una calibrazione  

come lo calibro:  
dato degli alberi contando i cerchi del tronco  
poi misuro il rapporto in questi  
così so che tal rapporto corrisponde a tale vecchiaia, mi costruiscono un grafico tra anno - $R_{atm}$  

R_atm is a function of t  
so the whole thing is a function of t, can find t  

#### Teoria

**Nota**

perchè $R_{atm}$ varia:  

1- vento solare  
devia i raggi cosmici, li fa entrare più o meno nell'atmosfera  

2- test nucleari  
"BOMB PEAK", 1955-1963 : picco nella curva di calibrazione  
quindi in questo periodo c'è molta differenza di $R_{atm}$ in poco tempo, è facile datare con precisione  

3- rivoluzione industriale  
modifica molto la composizione dell'atmosfera  
STRADIVARI GAP : un tratto in cui la cura di calibrazione è quasi piatta  
difficile datare in questo punto, un sacco di tempo ha la stessa $R_{atm}$  


#### Grafico

**Grafico**

dal numero di C14 ricavo un'età radiocarbonica  
espressa come $2500 \pm 30\, {}^14C \, BP$    
BP = before present  
non vuol di 2500 anni fa, perchè la quantitàdi C14 nell'atmosfera è variata nel tempo  

cioè: se $R_{atm}$ non fosse variata, sarebbe morto 2500 anni fa  
ma è variata, quindi devo considerare anche quello  

curva di calibrazione: prendo materiali di cui so l'età, e ci misuro C14  
così ottengo dei punti da fittare per fare la mia curva di calibrazione
