### Modellizza decadimeto

#### Ndof

$X \rightarrow Y_1 + ...è Y_n$  

quanti sono i dof:  
$\begin{cases}
cosa & ndof \\
 p^\mu & 4N \\
(p^\mu)^2 = m^2 & -N  \\
p^\mu_{X,i} = \sum^N p^\mu_i & -4 \end{cases}$

ndof = 3N-4    
quindi un oggetto deve decadere in almeno 2, non può decadere in un solo oggetto  

ndof: quanti cood servono per descrivere univocmanet un sistema  
per ogni particella serve il 4-impulso, quindi 4 coord  
ma poi devo togliere i constraint (eq. di conservazione): ogni particella ha quello sulla massa invariante, e ci sono le 4 componenti del 4-impulso di X da conservare  

conseguenza: n un decay a 2 corpi l'energia è fissata, ma in uno a 3 ha uno spettro   
in decay a 2 corpi: ho 2 dof, mi serono per decidere gli angoli di emissione, non ne rimangono per le energie, che sono fisse  


**Energia**

1- nel decay a 2 corpi:  
è obbligata: $E_1 = \frac{m_x^2+m_1^2-m_2^2}{2m_x}$ (nel frame di X)  
(m sono le masse invarianti della particelle, cioè la norma del 4-impulso)    
si trova coservando il 4-impulso $p_x^\mu =p_1^\mu + p_2^\mu$   
eg. decay alpha, gamma     

2- nel decay a 3 corpi:  
non è obbligata, ha spettro di valori possibili, con un min e un max  
non è obbligata perchè ci sono troppi ndof  
eg. decay beta  

#### Esempi

**Q-value**

1- $Q= \sum_n m_{iniz} - \sum_m m_{fin}$    
energia liberata in un decadimento, grazie alla differenza tra le masse a riposo iniziali e finali   
(m: massa a risposo, non massa invariante)

2- $Q = \sum_n T_{fin} - \sum_n T_{iniz}$     
grazie alla differenza tra energie cinetiche  
utile mettermi nel frame dove tutti i reagenti sono fermi (facile se ce n'è solo 1)  
nota che iniz e fin si scambiano  
perchè: consrvazione componente energia del 4-impulso: $m_x = E_1+E_2 = m_1+T_1+m_2+T_2$  

se Q>0, decadimento spontaneo  
sennò no, serbe energia cinetica iniziale  

**Classica o relativistica?**

se $M \gg Q$, allora posso usare la meccanica classica  
perchè le particelle non sono veloci rispetto alla loro massa  
dove stimo: $m = A \, u$ (numero di mazza x energia di un protone)  

classica: $T=\frac{p^2}{2m}$  
relativistica: $T = E-m = \sqrt{p^2-m^2}-m$  

intuizione: se classica, pensalo come un urto tra palline  
la più leggera acquista più velocità, la più pesante rincula solo un pochino (per conservare il momento)  

### Spettro dell'energia

**Plottare lo spettro**

asse x: T (energia cinetica, posso anche chiamarla E, ma si intende comunque cinetica)  

asse y: $\frac{dN}{dT}=p(T)$: numero di elettroni per unità di energia, cioè la densità dello spettro
lo trovo usando la regola d'oro   di fermi    
dN: numero di elettroni con energia nel range (T, T+dT)    

**Regola d'oro di Fermi**

$\Gamma _{i\rightarrow f} = \frac{2\pi}\hbar \;  |\langle f |H| \rangle i|^2 \; p(E)$     
$\Gamma _{i\rightarrow f}$: prob di transizione dallo stato i a f, per unità di tempo   

|H| : dinamica -> se la forza è forte, fa avvenire più spesso la transizione (ignoro questa parte, se integro su tutto diventa una costante)  
dinamico, interazione potenziale che tiene insieme il nucleo  

$p(E)$: densità degli stati FINALI, coiè nello stato f (numero di stati finali accessibili)  
spazio delle fasi, numero degli stati accessibili con energia E 

E: energia FINALE (cioè della particella emessa, NON di quella che decade)

**Densità degli stati**  

1- con cosa scala  
$N$: numero di elettroni che hanno fatto la transizione da i ad f  
$N_0$: quanti elettronia vevo all'inizio, allo stato i  
$N = N_0\, \Gamma_{i \rightarrow f} \, \Delta t \propto \Gamma_{i \rightarrow f} \propto p(T) = \frac {dN}{dT}$  

2- calcolare  
dn: numero di stati di sponibili    
dN: numero di elettroni osservabili  
dn/dE: Densità degli stati quantistici disponibili  
dN/dE: Spettro degli elettroni prodotti o osservati     
$d^{(1)}n = \frac{d^3p}{(2 \pi \hbar)^2}$  
non appare d^3q, forse perchè voglio la densità per unità di volume???  
in decay beta, producono 3 particelle, cioè: $dn = dn_{e-} +dn_Y +dn_\nu$    
questi sono gli stati totali, non quelli disponibili (cioè che rispettano le regole di conservazione con,lo stato iniziale i, la particella prima del decay)  

3- ipotesi:  
$m_\nu \sim 0 $ (neutrini hanno massa piccola)  
$T_Y \sim 0$ (energia cinetica del nucleo generato dal decay è più bassa di quelle di elettrone e neutrini, perchè è più grosso, immagina una collisione classica)  
quindi, il volume sullo spazio delle fasi delle particelle:  
$dn \propto d^3p \, d^3 q \, d^3P$  
momenti di: p=elettrone, q=neutrino, P=Y  

4- so che: $d^3p = p^2 \, dp \, d\Omega$  
non mi interessa l'angolo, integro su tutti, quindi ho $2 \pi p^2 \,dp$  
posso farlo per elettrone e neutrino, ma non per Y  
perchè devo conservare il momento $\vec P + \vec p + \vec q=0$, quindi se ne fisso 2, il terzo è obbligato  
$dn \propto d^3p \,  d^3q \, d^3P \propto p^2 dp \, q^2 dq \, 1$  

5- cambio var:  
$dp \, dq = J(\frac{\partial p,q}{\partial T,Q}) \, dt \, dQ$  
$q=Q-T$  
$p=\sqrt{2mT}$   
J=jacobiano (è il determinante della matrice, p facile da calcolare)    
$dn \propto p^2 \, q^2 \, dp \, dq  = wmT (Q-T)^2  \frac{\sqrt {2m}}{2 \sqrt T} dT \, dQ$  
$\frac{dn}{dQ} \sqrt T (Q-T)^2 \, dT$
