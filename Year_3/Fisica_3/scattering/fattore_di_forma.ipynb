### Studiare con onde

**Scattering con onde vs particelle**

lncio un elettrone contro il bersaglio  
(scelgo l'eletrone perchè è leggero, è facile da acccelerare)  

dualità onda-particella:  
$\lambda=\frac \hbar p$  
$\lambda$: lunghezza d'onda di De Broglie dell'elettrone  
$\vec p$: momento dell'elettrone  

casi:  
$r_0$: lunghezza caratteristica del bersaglio    
se $\lambda \sim r_0$: considero e- come onda piana $e^{i q \cdot k}$ -> scattering con onde    
se $\lambda \gg r_0$: considero e- come particella -> scattering tra particelle  

**Cosa voglio fare**

voglio studiare un bersaglio lanciandogli addosso un'onda EM    

gli lancio contro un'onda o particella, vedo come è diffusa  
perchè sciascuna parte del bersaglio, cioè della distrib di carica, contribuisce a diffondere l'onda  
vedendo com'è l'onda diffusa, capisco come è fatto il bersaglio  

**Disegno**

è come il disegno dell'inteferenza  
mi interessa come i raggi si incontrano in un punto a distanza infinita, quindi $k$ da ogni centro scatteratore deve essere parallelo  
voglio dephasing tra i raggi riflessi da ogni centro  

### Fattore di forma

**Def**

descrive come la distribuzione di carica modifica l’onda diffusa  
dipende da forma e dimensioni del bersaglio    

**Spiegazione**

ciascun punto del bersaglio, cioè della distrib di carica, agisce da centro scatteratore, contribuisce a diffondere l'onda  
ma questi contributi arrivano dopo aver percorso distanze diverse, quindi con fasi diverse, quindi devo vedere fanno interfereenza costruttiva o distruttiva   

nota:  
è un caso più generale dell'interferenza col reticolo a fisica 2  
qui i buchi non sono in una linea verticale, ma possono essere sparsi nello spazio 3D  


**Formula**

$F(\vec q) = \frac 1 Q \int \rho(\vec r) e^{i \vec q \cdot \vec r} d^3r$  
$Q = \int \rho(\vec r) \, d^3 r$   

**Modellizzazione**

Q: la distrib di carica = il bersaglio   

$ \vec q = \Delta \vec k = \vec k_f- \vec k_i$: il trasferimento di vettore d’onda tra onda finale e onda incidente    
in urto elastico, le k hanno lo stesso modulo, cambia solo l'angolo, quindi graficamente:   
$|q| = 2k \sin \frac{\theta }2$  

$k= \frac{2 \pi}{\lambda}$

#### Altro

**Nota**

q=0 -> F(0)=1  
tuttele parti del bersaglio contribuiscono in fase  

**Altre formule**

$A_{tot}(\theta) = A_{puntiforme}(\theta) F(q)$   
$\frac{d \sigma}{d\Omega} |_{composto}  = \frac{d \sigma}{d\Omega} |_{puntiforme} |F(q)|^2$   

### Calcola F(q)

**Steps**

1- scrivi $\rho(r)$ del bersaglio  

2- calcola $F(q)$ con la formula sopra   
$F(\vec q) = \frac 1 Q \int \rho(\vec r) e^{i \vec q \cdot \vec r} d^3r$  

#### Distrib simmetria sferica

**Formula generale F(q) - distrib simmetria sferica**

$F(q)  = \frac 1 Q \int \rho(\vec r) e^{i \vec q \cdot \vec r} d^3r$  
$= \frac 1{Q} \quad \int_0^\infty r^2 \, dr  \quad \int_{-1}^1 dcos \theta \quad \int_0^{2\pi} d \varphi  \quad \rho(r)  e^{i q r \cos \theta}$  
$= \frac 1Q (2\pi) \quad \int \rho(r) \, r^2 \; dr \quad \int dcos\theta \, e^{iqr \cos \theta}$   
$= \frac 1Q (2\pi) \quad \int \rho(r) \, r^2\; dr  \quad \frac 1{iqr} (e^{iqr}-e^{-iqr})$     
$ = \frac {4\pi}Q \int dr \;\rho(r) \;  r^2 \frac{\sin(qr)}{qr}$

come ho fatto ogni integrale sferico:  
$d\varphi$: integro su tutti gli angoli, quindi $2 \pi$  
$dcos \theta$: integrale dell'esponenziale, facile -> poi $e^{ix}-e^{-ix} = 2i \sin x$   
$dr$: non fatto ancora, dipende da $\rho(r)$  

**Coord sferiche**

$d^3 r = r^2 \, \sin \theta \, dr \, d\theta \, d\varphi$  
$= r^2  \, dr \, (d\cos \theta)  d \varphi$  

se non mi interessa l'angolo, ma solo il raggio (simmetria sferica):  
$\int _{-1}^1 d\cos \theta \; \int_0^{2\pi} d\varphi = \, 2 ; 2 \pi = 4 \pi$  
  
lo chiamerò: 
$d^3 r = r^2 dr \, d\Omega$  
con: $\int d\Omega = 4 \pi$  

**Proprietà della Delta**

$\int _I f(r) \delta(r-R) dr$  
se I contiene il punto r=R: $int = f(R)$    
se I NON contiene quel punto: $int =0$  
è il risultato dell'INTEGRALE, non è una trasformazione dell'integranda  

#### Guscio sferico

**1. Scrivo densità di carica**

guscio di raggio R, spessore infinitesimo, carica Q   
$\rho(r) = A \delta(r-R)$  

trovo A con condizione di normalizzazione:  
$Q = \int_V \rho(r) \, d^3r$  
$ = A \;  \int_0^\infty \delta(r-R) r^2 \, dr \; \int d \Omega \; $    
$ = A \; R^2 \; 4 \pi$  

densità di carica normalizzata  
$\rho(r)= \frac Q{4 \pi R^2} \delta(r-R)$  

**2. Fattore di forma**

parto da F a simmetria sferica:  
$F(q) = \frac {4 \pi }Q \pi \int dr \; r^2  \,   \rho(r) \frac{\sin(qr)}{qr}$  
$ =  \frac {4 \pi }Q \;  \frac {Q}{4 \pi R^2} \,  \int dr \; r^2  \,   \delta(r-R) \frac{\sin(qr)}{qr}$   
$= \frac 1 {R^2} R^2 \frac{\sin (qR)}{qR}$  
$ = \frac{\sin (qR)}{qR}$    

come: proprietà della $\delta$  

#### Sfera intera (non fatta a lezione)

**1. densità di carica**

sfera di raggio A e carica Q  
se $r \ge a$: $\rho(r) = \frac{3Q}{4 \pi a^2}$    

**2. Fattore di forma** 

anche qui, plotta $\rho(r)$ nella formula di F(q) per distrib sferica   
come: fai integrale per parti  

$F(q) = 3 \frac{\sin(qa)-qa\cos(qa)}{(qa)^3}$

#### Metodo 2: RMS

**Steps**

1- scrivi F(q) della forma del bersaglio  

2- usa Taylor per approssimare a $q \sim 0$   
ottengo $F(q) \sim 1  +f(R)...$  

3- uguaglialo a: $F(q) \sim 1 +\frac 16q^2 R^2$  
ho la funzione: $\langle r^2 \rangle = g(R)$, dove R l'ho calcolato col metodo prima    
quindo ho trovato il RMS  


**Teoria**

def di RMS (raggio quadratico medio):  
$\langle r^2 \rangle = 4 \pi \int_0 ^\infty r^4 \rho(r) \, dr$    

espando formula generale di F(q) intorno a 0  
$F(q) = 1 + \frac 16 q^2 \langle r^2 \rangle$  
$\langle r^2 \rangle = -6 \frac{d}{dq^2} F(q)|_{q=0}$
