**Modellizzare un rilevatore / detector**

disegnalo come un faretto di area A, che punta verso il centro scatteratore  

simbolo | nome | significato
---|---|---
A | sezione di accettanza del rilevatore  | l'area che, se colpita, registra un evento
R | | distanza del rilevatore dal centro scatteratore  
$d\Omega = \frac{dA}{R^2}$ | angolo solido che sottende il rilevatore | il rilevatore vede solo le particelle scatterate dentro questo angolo solido  

**Cose importanti**

1- distribuzione angolare - sezione d'urto differenziale: $\frac{d\sigma}{d\Omega}$    
dice quante delle particelle della coroncina iniziale $d \sigma$ vengono scatterati dentro l'angolo solido $d\Omega$ (cioèquante ne vede il rilevatore)  
quindi può dipendere da $\theta$ -> cioè ad al uni valori di $\theta$ possono essere scatterate  più particelle, in altri meno  
It measures how strongly particles scatter into a particular direction (per unit solid angle)  
if it's bigger, more particles reach that solid angle    

2- $\theta(b)$   
dato un b, quanto viene deviata la particella   
assumo che, se b fsso, anche $\theta$ sia fisso (approccio deterministico)  

**Cose che ottengo**

tutte sono dà come funzioni di $\theta$    

$\frac {d \sigma}{d\Omega}$  
area efficace per angolo solido  
domanda: Quanto è efficace il bersaglio nel deviare particelle verso questa piccola regione di direzioni?    
dà come risposta un valore relativo tra le direzioni, ma in valore assoluto non significa niente  

$\frac {d P}{d\Omega} = \frac 1 {\sigma_{tot}} \frac {d\sigma}{d\Omega}$   
prob per angolo solido  
domanda: Qui consideriamo soltanto le particelle che hanno urtato la sfera e chiediamo: Quale frazione esce verso questa piccola regione di direzioni?  
la risposta è una prob, è stata normalizzata  

$\frac {d P}{d\theta}$    
probabilità per intervallo di angolo polare  
domanda: Quale frazione esce tra $\theta$ e $\theta+d\theta$, contando tutte le direzioni attorno all’asse z?
invece di considerare solo un angolo solido, considero tutta la corona (se c'è simemtria circolare è uguale, non dice niente di più di prima)


**Definizione: sezione d'urto differenziale**

cosa faccio: incido con un flusso di particelle  
$dN = J \frac{d \sigma}{d \Omega} d\Omega$  

dN: numero di particelle scatterate nell'angolo solido $d\Omega$    
cioé, registrate dal rilevatore  

J: flusso (numero particelle / (unità di area * unità di tempo))  

### Caso: simmetria circolare

![alt  text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcR141-Fuid6LZiD3KodsxtPDVs3-HSv66DaA3DTP1KdkV30yoHihnd128Q&s=10)

**Teoria**

ipotesi:   
1- simmetria rispetto a asse z    
2- V(r) è repulsivo, allontana la partiella dal centro scatteratore  

conseguenza:    
le particelle in una coroncina $d \sigma$ (quella a sinistra) vegono scatterate lungo un'altra corona  
cioè, lungo la parete di un cono con angolo $\theta$ rispetto all'asse z (il cono nell'immagine in alto)

**Formule**

area di una corona = circonferenza * spessore  
$\begin{cases} 
d \sigma = 2 \pi  b \, db \\  
d\Omega = 2 \pi \sin \theta \, d\theta \end{cases}$

trovo le formule (la seconda la trovo invertendo la prima):  
$\begin{cases}
\frac{d\sigma}{d \Omega} = \frac{b}{\sin \theta} |\frac {db}{d \theta}| \\
b^2(\theta) = 2 \int_\theta^\pi f(u)\sin(u) du &  f(u) = \frac{d\sigma}{d \Omega}(\theta) \end{cases} $    
la u è una var muta  


### Cosa fare

**Indagare la materia con gli urti**

$\frac{d\sigma}{d\Omega}(\theta) \quad \rightarrow \quad b(\theta) \quad \rightarrow \quad V(r)$

$V(r) \quad \rightarrow \quad b(\theta)  \quad \rightarrow \quad \frac{d\sigma}{d \Omega}(\theta)$  

### Esercizi

#### Scatteratore: sfera non carica

**V -> distrib angolare**

1- V  
non ha proprio una formula, ma so che la particella fa urto elastico  
viene riflessa come su specchio tangente al punto di urto (la sfera è localmente piana)  

2- b  
dal disegno, sapendo come riflette uno specchio:    
$b = R \sin \alpha = R \cos \frac{\theta }2$    
vale se $b \le R$, sennò non interagisce  

3- distrib ang  
$\frac{d \sigma}{d \Omega} = .. = -\frac{R^2}4$  

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQ8HUIOxtawmcruNfdT-LqZ4pIv96XEU6VPDbq0MTwimDmP2A0EiNQqrGk&s=10)

**Distribuzione isotropa**

isotropa: NON dipende dall'angolo $\theta$  
cioè in $\frac{d\sigma}{d\Omega} $ non apapre $\theta$  
cioè il numero di particelle riflesse per unità di angolo solido è uguale  

se mando un fascio su una sfera rigida, genero una distrib isotropa   

**Distrib ang -> V**

1- $\frac{d\sigma}{d\Omega} = L^2$   
è l'osservazione sperimentale che ho fatto, che ho distrib isotropa, non dipende dalla direzione  

2- b      
$b^2(\theta) = 2 \int_\theta^\pi f(u) \sin(u) du$  
$b(\theta) = 2L \cos \frac \theta 2$   

3- V  
la $b(\theta)$ ottenuta è la legge di deviazione prodotta da una sfera di raggio $R = 2L$ (la vedo sopra, dall'ex nel verso opposto)        

nota:  
una distribuzione isotropa è compatibile con la sfera rigida  
ma da sola non dimostra che il potenziale sia proprio quello (potrei avere altri V che producono pure una distrib isotropa)

#### Scatteratore: sfera carica (V coulombiano)

**V -> distrib angolare**

1- V  
$V = \frac {Ze}r$  
nota: il percorso è un iperbole, perchè il potenziale coulombiano devia l'orbita   

2- $b(\theta)$  
$b(\theta) = \frac{mzZe^2}{p^2} \frac{\cos(\theta/2)}{\sin(\theta/2)}$  

3- distrib ang  
$\frac{d \sigma}{d\Omega} = \frac 1 {\sin^4 (\theta/2)} (\frac {mzZe^2}{p^2})^2 = \frac 14 r_0^2 \frac 1 {\sin^4 (\theta/2)}$    

**Risolvo: step 2**

conservo momento angolare:  
$L_0 =mV_0b$  
$L= I \omega = -m^2 \dot \alpha$  

considero la variazione di momento nell'asse verticale  
all'inizio è nullo, perchè la particella si muove verso destra  
ma alla fine si muove di angolo $\theta$, quindi è non-nullo   
$m \, dv = F \, dt$  
variazione di quantità di moto = impulso  

in questo caso:   
$mv_0 \sin \theta - 0 = \int_{-\infty}^\infty
F_y(t) \, dt = \int_{-\infty}^\infty -\frac{dV}{dy} dt$    
molti calcoli...

**Osservazioni**

ONLY for repulsive Coulomb scattering

1- ad angolo fissato: $\frac{d\sigma}{d\Omega} \sim \frac 1 {p^4}$    
quind una particella più lenta (p piccolo) può essere deviata più facilmente anche a grandi angoli???  
Fisicamente, per ottenere la stessa deviazione, una particella lenta può passare più lontano dal centro diffusore: cresce l’area delle traiettorie incidenti che contribuiscono a quell’angolo.

2- ad impulso fissato: $\frac{d\sigma}{d\Omega} \sim \frac 1 {\sin^4 \frac \theta 2}$   
quindi più particelle vengono scatterate ad angoli $\theta$ piccoli, e poche ad angoli grandi  

3- small b produces large deflection, while large b produces small deflection     

**Altro modo di scriverlo**

$|\varphi(q)|^2 = \frac 1 {q^4}$  
prob di trovare una particella con impulso scatterato q  

$\vec q = \vec p_{fin}- \vec p_{in} = 2p \sin \frac \theta 2$  
q: impulso trasferito durante l'urto  

**Studiare un sistema**

$\Delta r \sim \frac \hbar q$  
$\Delta r$: dimensione caratteristica dell'oggetto da studiare  
quindi: per studiare sistema su scala piccola, devo usare impulsi grandi (cioè particelle molto veloci)
