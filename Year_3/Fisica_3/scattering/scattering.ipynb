### Bombardamento

**Bombardare con un neutrone**

cosa potrebbe succedere:  
$n + X \rightarrow \begin{cases}
n+X \\
n+X^* \\
\gamma + Y \\
fissione \, in \, frammenti \end{cases}$  

1- diffusione elastica  
2- diffusione anelastica (X prende un po' energia, per passare a stato eccitato)  
3- fotoproduzione (crea fotone)

### Modellizza

decay | scattering
---|---
$N(t)=N_0e^{-t/\tau}$ | $N(x) = N_0 e^{-x/\lambda}$  
$\tau = \langle t \rangle$: vita media | $\lambda = \langle x \rangle$ cammino libero medio  
$\frac 1 \tau = \lambda$ <br> $\lambda$: larghezza di decay (DIVERSO DA QUELLA A DESTRA) | $\frac 1 \lambda = n^* \sigma$ <br> $\sigma$: sezione d'urto 
se molti possibili decay:  <br> $ \lambda = \sum_i \lambda_i$ | se molte èpossibili reazioni: <br> $\sigma = \sum_i \sigma_i$ 

**Sezione d'urto**

$R= \Phi N_t \sigma$  

R (ratio): # di interazioni /secondo    
$\Phi$: flusso di particelle incidenti  
$N_t$ numero di bersagli/targets    
$\sigma$: sezione d'urto  

sezione d'urto: probabilità di interazione, in unità di area  

**Unità di misura - sezione d'urto**

b = barn, pagliaio    
$1 \, b = 10^{-28} m$  
quindi: $1b = 100 fm^2$ ($1 fm = 10^{-13}cm$)  
ricorda: $r_0 = 1.25 fm$ (raggio protone)  

**Centri scatteratori / diffusori**

cosa sono: diffusori: oggetti che, se ci sbatte una particella, rimbalza via    
di solito i nuclei che compongono il materiale  

**Caratterizza un materiale**

$n^*$: numero di diffusori per cm^2   
$n^* = \frac{\rho}{MM}N_{Av}$  
in teoria: quanti atomi (o molecole) ci sono in un certo volume di materiale

$\rho$: densità  
MM: massa molare, g/mol  
$N_{av}$: numero di avogadro  

**Intensità**

cosa è:  
lancio un fascio di particelle verso destra  
tra x e x+dx ci sono centri scatteratori  
voglio vedere quante ne passano  

$I(x+dx)=I(x)-N_{sch}$  
$N_{sch} = I(x) \frac \sigma A n^* \, A dx$ numero di particelle schiantate, non passano  
considero un fascio di particelle incidenti di forma cilindrica, con area A, vedo quante sbattono sui diffusori  

questa eq diff mi serve a trovare $N(x) = N_0 e^{-x/\lambda}$, la legge dello scattering

**Parametro di impatto**

stessa def di fisica classica  
a seconda di b, la particella viene deviata in modo diverso  
assumo che, a b fissato, il percorso fatto è fisso (deterministico)  

#### Rutherford vs Rayleigh

sono tutti scattering, cioè deviazioni di qualcosa a causa di qualcos'altro  

scattering | Rutherford | Rayleigh | Compton
---|---|---|---
particella/onda  incidente | particella carica | onda EM (luce) | fotone ad alta energia (X o gamma)
bersaglio (centro scatteratore) | nucleo atomico (oggetto carico) | particelle/molecole molto più piccole della lunghezza d'onda della luce | elettrone libero o debolmente legato
esempio | particelle $\alpha$ lanciate contro nucleo atomico <br>  ha fatto scoprire che esiste il nucleo | intensità della luce diffusa è $\sim \frac 1 {\lambda^4}$, quindi il blu è diffuso più del rosso <br> quindi il cielo è blu
primary force / mechanism | forza di coulomb | radiazione di dipolo elettrico
natura | elastica <br> energia  cinetica totale si conserva | elastica <br> fotone diffuso ha stessa enrgia e freq, cambia solo direzione e fase | anelastica <br> fotone cede energia e momento all'elettrone, quindi fotone diffuso ha $\lambda$ maggiore e f minore
