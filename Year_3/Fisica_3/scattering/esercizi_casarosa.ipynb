### Sezione d'urto

cosa incide | formula
---|---
singola particella | $\sigma$
flusso | $R$  

$R = \Phi_{in} N_t \sigma$  

R: rate (numero di eventi/catture per unità di tempo)  
$\Phi_{in}$: flusso incidente   
$N_t$: numero di centri scatteratori (numero di NUCLEONI, non molecole, ogni nucleone è un centro scatteratore)  
$\sigma$: seione d'urto  

intensità del flusso:  
$I(x) = I_0 e^{-x/\lambda}$   
$\lambda$: cammino libero medio  
$\frac 1 \lambda = n^* \sigma$  
$n^*$: densità di centri scattertaori (nel materiale)   

lamina sottile:   
quando $d \ll \lambda$  
cioè la lunghezza della lamina (d) è molto più piccola del cammino liebro medio ($\lambda$)  
allora assumo che I(x) non diminuisca dentro la lamina, $I(x) \sim I_0$ DENTRO LA LAMINA   
 

sezione d'urto differenziale  
$\frac{d\sigma}{d\Omega}$ : numero di particelle scatterate??? in un angolo solido $d\Omega$  
$\frac{d\sigma}{d\Omega}$ : delle particelle appartenenti alla corona $d\sigma$, quante vengono scatterate nall'angolo solido $d\Omega$  


per avere la $\sigma$; integro in tutto l'angolo solido  
$\sigma = \int \frac{d\sigma}{d\Omega} \, d\Omega$    

**Scontro tra flussi di particelle**

$R = L_{ist} \sigma$  
$L_{ist}=f \, n \, \frac{N_A \, N_B}{A}$  
$L_{int} = L_{ist} \, \Delta t$   
$N= R \, \Delta t = L_{int} \sigma$  

$L_{ist}$: luminosità istantanea (una proprietà dell'acceleratore) [cm^-2 s^-1]  
$f$: frequenza di rivoluzione dei bunch nell'anello dell'acceleratore  
n: numero di bunch nell'anello (bunch = nuvoletta di particelle, perchè non sono mandate come flusso continua, ma a burst di gruppi di particelle)  
$N_A, N_B$: quante particelle di tipo A/B in un bunch di tipo A/B  
$A$: sezione perpendicolare di attraversamento delle particelle (sezione dei bunch che si scontra)  

$\Delta 4$: tempo di osservazione  
$L_{int}$: luminosità istantanea, qyanti dati ha preso l'acceleratore  


### Ex. flusso contro lamina

cosa: lancio fascio di neutroni contro lamina d'ro  
alcuni vengono catturati dall'oro, che diventa un isotopo con un protone in più  
quanti vengono catturati??   

reazione di cattura:  
$n+Au^{197} \rightarrow Au^{198} + \gamma$  

**Dati**

$E_n$: energia cinetica dei neutroni  
$\digma$: quanto spesso ol neutrone biene catturato dall'oro (fai finta di lanciare una singola particella)  
$I$: intensità fascio incidente  
$\Delta t$: tempo di osservazione  
$d$: larghezza lamina  

**Altre var**

$N_A$: numero di avogadro  
$M_{lam}$: massa della lamina  
$m_{Au}$: massa di un atomo di oro (o nucleone???)  
$MM_{Au}$: massa molare di oro (di una mole)  
$\rho_{Au}$: densità dell'oro 

**Risolvo**

1- rate  
$\begin{cases}
R = \Phi_{in} N_t \sigma \\
\Phi_{in} = \frac {I}{A} \sim \frac{I_0} A & lamina \; sottile \\
N_t = \frac{M_{lam}}{m_{Au}} = \frac{\rho_{Au} \, Ad }{MM_{au}/N_A}
\end{cases}$

$R = \frac{I_0} A \frac{\rho A d}{MM} \, \sigma$  
$= I_0 (\frac{\rho_{Au}\, }{MM}N_A) \; \sigma \, d$   
$= I_0 (n^* \sigma) \, d$  
$ = I_0 \frac d {\lambda}$    

2- numero di eventi  
$N = R \, \Delta t$  

**Problema: lamina sottile**

devo controllare se davvero lo è, per poter dire $I(x) \sim I_0$ dentro la lamina  
quindi devo calcolare $\lambda$  

$\begin{cases}
\lambda = \frac 1 {n^* \sigma} \\
N_t = n^* (Ad) \\
N_t = \frac{M_{lam}}{m_{Au}} = \frac{\rho_{Au} \, Ad }{MM_{au}/N_A} & come \; sopra
\end{cases}$

$n^* = \frac{\rho_{Au}N_A}{MM}$  
$\lambda = \frac{MM}{\rho N_A} \frac 1 {\sigma}$  

d è un dato, controlla se:  
$\lambda \ll d$  

## Ex. Scontro tra flussi di particelle

**Cosa**

scontro tra due particelle:  
$e^+e^- \rightarrow \mu^+ \mu^-$  

**Problema**

problema: il mio rilevatore non ha tutta l'accetazione angolare, cioè non vede le particella prodotto emesse in tutte le direzioni, ma solo in qualcuna  
perchè: alcune direzioni sono occupate dall'acceleratore, dai magneti, dall'anello... non posso metterci il rilevatore  
quindi devo calcolare prima la sezione d'urto differenziale, poi passare a quella totale  
  
quindi posso calcoalre due cose  
1- eventi totali avvenuti: integro su tutto l'angolo solido $\int_{-1}^1 dcos \theta$  
2- eventi visti dal rilevatore: integro sull'angolo solido che può vedere, eg $|\cos \theta | < 0.8$   
quindi:  $\int_{-0.8}^{0.8} dcos \theta$  

**Risolvo**
  
il testo mi dà la formula della sezione d'urto differenziale, che è funzione di $\cos \theta$  
$\frac{d \sigma}{d \Omega} = \frac{\alpha^2}{2S} (1+ \cos^2 \theta)$  

$\sigma = \int \frac{d \sigma}{d \Omega} d\Omega = \int _0^{2\pi} d\phi \int_{-1}^1 d\cos \theta  \; \frac{\alpha^2}{2S} (1+ \cos^2 \theta)$    
$=  \frac{\alpha^2}{2S}  \, 2 \pi \int_{-1}^1 du \, (1+u^2)$  
$ = 1.99 \, 10^{-6} \, Gev = 0.76 mb$   

$N = L_{int} \sigms$  
nota: $L_{int}$ me lo danno i dati  

**Unità di misura**

$\frac{d \sigma}{d \Omega} $ è scritat in unità $\hbar = c = 1$, quindi la sol mi viene con le unità di un'energia  
ma io la voglio in barn, con le unità di un'area  
si trasforma così:  
$1 = 197 \, Mev \,fm$    

### Ex. quanti neutrini interagiscono

**Cosa**

i neutrini hanno prob di interazione bassa, ma non zero  
quindi, mentre uso un rilevatore per studiare qualcosa, vedo anche dei falsi eventi, che in realtà sono i neutrini che reagiscono con l'acqua  
bisogna contare quanti sono nel tempo di misura (1y = 1 anno), per toglierli dal totale   
$\nu + H_{2}O \rightarrow ...???$  

**Dati**

$E_\nu$: energia di un neutrino (è diversa per ogni neutrino)  



**Problema**

$\sigma$ dei neutrini dipende dalla loro energia $E_\nu$   

soluzione:  
$\frac{d\Phi_{in}}{dE_\nu}$: differenziale di E, cioè mi interessano solo i neutrini del flusso con questa specifica E  
$\sigma(E)$: dipendenza da E, cioè si applica solo ai neutrini con questa specifica energia  

nota:  
nel range di energie [0.5, 10] Gev trovo che $\sigma(E)$ è lineare  
$\sigma(E_\nu)= \sigma_0 (\frac{E_\mu}{E_0})^{2.7}$  
quindi: $\frac{d\Phi}{dE_\nu} = \phi_0 ( \frac{E_\nu}{E_0})^{2.7}$  
quindi poi faccio $\int_0.5 ^{10} dE_\nu$  

**Risolvo**

$\frac{dR}{dE_\nu} = \frac{d\Phi_{in}}{dE_\nu} N_t \sigma(E)$  
$R = \int \frac{dR}{dE_\nu} dE_\nu$  

nota: $N_t = N_{H_2O} \, 18$  
18: numero di nucleoni in una molecola d'acqua
