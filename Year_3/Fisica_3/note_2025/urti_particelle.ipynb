**Processo di urto**

lo scrivo come una reazione:  
$a+b \rightarrow f$  

cosa misuro: quantità di moto di ogni particella in stato iniziale e finale  
come in fisica 1, studio prima e dopo urto, non cerco di modelizzare durante l'urto  

**Tipi di urti**

tipo | def| esempio
---|---|---
elastico | NON cambia: tipo di particelle, stato interno delle particelle (fondamentale/eccitato) <br> cambia: impulso, spin delle particelle | $a+b \rightarrow a+b$ 
anelastico | cambia: tipo, stato interno delle particelle | $a+b \rightarrow a + p_1$

**Metodo di misura/osservazione**

var cinematiche: quantità di moto, polarizzazione (se presente), spin (se presente)

metodo | def
---|---
esclusivo | misuro tutte le var cinematiche dei prodotti  
inclusivo | misuro solo alcune quantità var cinematiche dei prodotti

perchè usare modo inclusivo:  
1- impossibile sperimentalmente misurare p per alcune particelle prodotte  
2- inutile misurare p per alcune particelle
