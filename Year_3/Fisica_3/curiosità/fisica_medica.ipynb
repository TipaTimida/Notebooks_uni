1- dose: $E = \frac{E_{assorbita}}{M}$  
cosa: quanta radiazione/energia viene ASSORBITA da un corpo, per unità di massa    
unità: 1 G = 1 J/kg (gray)   

2- dose equivalente: $D_{eq} = \alpha D$  
cosa: usata per gli effetti sul rivelatore (perchè si danneggia)  
unità: 1 Sv   = 1/Kg (Sievert)   

3- dose efficace: $D_{ef} = \alpha \beta D$  
cosa: usata in radioprotezione (effetti sul corpo umano)   
$\alpha$: tiene conto del tipo di radiazione  
$\beta$: tiene conto del tipo di tessuto (del corpo umano)  
unità: 1 Sv   = 1/Kg (Sievert)
