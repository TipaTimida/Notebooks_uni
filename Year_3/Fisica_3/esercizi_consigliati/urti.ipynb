**1. sezione d'urto (in fisica nucleare)**

def: la probabilità che avvenga un'interazione (urto, assorbimento, fissione) tra un fascio di particelle incidenti e un bersaglio    
cioè, misura l’efficacia di un processo d’urto

geometricamente:  
non è per forza la sezione del nucleo, ma è una "superfcie efficace"    
un nucleo può avere una sezione d'urto molto grande o molto piccola a seconda del tipo di reazione e dell'energia della particella incidente  

unità: area (1 barn = $10^{-24 }cm^2$)

a sezione di cosa misura la prob, ha nomi diversi:   
sezione d'urto... | def
---|---
totale $\sigma$|  Comprende la somma di tutti i possibili canali / processi d'interazione (elastici, anelastici, di assorbimento...)  
elastica | urto elastico <br> NON cambia: tipo di particelle, i loro stati interni (fondamentale/eccitato) <br> cambia: momenti 
anelastica | urto anelastico <br> bersaglio viene eccitato o si generano nuove particelle
di estinzione | misura quanto viene ridotta la componente del fascio che prosegue nella direzione incidente, cioè il fascio non deviato e non modificato <br> include tutti i processi che che lo sottraggono a quella componente (diffusione, assorbimento, produzione di particelle)

**2. Osservare un urto**

per urto tra 2 particell:  

osservazione | def | esempio
---|---|---
esclusiva | si identifica l’intero stato finale (particelle prodotte e il loro stato interno) | $p+p \rightarrow p+p+\pi_0$
inclusiva | si osserva solo una parte dello stato finale e si somma su ciò che non è misurato |  $p+p \rightarrow p+X <br> misurare un protone a una certa energia e direzione, c'è anche generico X che ignoro

**3. Q-value**

differenza tra l’energia di massa iniziale e quella finale, nel sistema del cm  
$Q = (M_{in} - M_{fin})c^2$  

processo:   
esotermico: Q>0 (libera K)  
endotermico: Q<0 (occorre fornire energia per renderlo possibile)  
eso, endo: dice VERSO DOVE è il moto dell'energia (verso fuori, o verso dentro la particella)  

**4. densità di corrente di particelle**

J: densità di corrente di particelle   
numero di particelle che attraversa, per unità di tempo, una superficie unitaria (perpendicolare alla loro velocità)    
(cioè, def come un flusso)  
$J = \frac{dN}{dA \, dt}$  

se facio uniforme di velocità v, e densità n:  
$J = nv$  

**5. sezione d'urto**

caso | formula 
---|---
particelle incidente su un bersaglio singolo | $P = \frac \sigma A$ <br> A: area su cui è distribuito il fascio incidente (su cui si trovano le particelle incidenti)  <br> P: prob che avvenga l'interazione | area efficace associata al processo 
Fascio su un bersaglio singolo | $ N' = J \sigma$ <br> J: densità di corrente della particelle <br> $\hat N$: freq degli eventi, rate [1/s]
fascio sottile su lamina | $N'= J N_s \sigma$ <br> $N_s$: numero di bersagli per unità di superficie <br> cioè, una lamina contiene molti bersagli / nuclei
Urti nel volume tra due specie | $r = \frac{dN}{dt \, dV} = n_1n_2 \sigma(v_{rel} v_{rel})$ <br>$n_i$: concentrazione <br> $v_{rel}$: velocità relativa

**6. Sezione d'urto - onda**

$N' = \phi_\gamma \sigma$  
$\phi_\gamma = \frac{I_{inc}}{E_\gamma} = \frac{I_{inc}}{\hbar \nu}$ (flusso di fotoni)  
quindi: $P= \sigma I_{inc}$  



differenziale $\frac {d \sigma}{d \Omega}$ | probabilità che una particella venga deviata in un particolare angolo solido $\Omega$
