**Cambiare unità di misura**

devo moltiplicare per una frazione =1, in cui numeratore e denom hanno diversa unità di misura   
scegli numeratore e denominatore in modo da cancellare l'unità vecchia e lasciare quella nuova  
cioè devo dire quante unità del numeratore ci sono nel denominatore  

esempio:    
$1 \, J = 6.24 \, 10^{12} \, MeV$   
per passare da J a Mev, moltiplico per $1 = 6.24 \, 10^{12} \frac{MeV}J$    
cioè, per quante Mev (cosa nuova) ci sono in un Joule (cosa vecchia)

1.10 Denire la lunghezza d'onda di DeBroglie e calcolarne il valore numerico per un elettrone di energia
1keV .
  
La lunghezza d'onda di De Broglie è associata a una particella materiale e vale: $\lambda = \frac h p$  
uso approssimazione NON relativistica: $K = \frac{p^2}{2m_e}$  
da cui: $ p = \sqrt{2m_e K}$  

$\lambda = \frac h p = \frac {2 \pi \hbar}p = (2 \pi) \; (\hbar c) \frac 1{\sqrt {2} \sqrt K \sqrt {m_ec^2}}$  
$= 2 \pi \; (197 \, Mev \, fm) \; \frac {1. 44 \; 10^{-3/2}Mev \; 0.511^{1/2} }$  
$= 38.023 \, fm \quad 10^{-15} \frac m{fm} \; 10^{9} \frac {nm}m $  
$ = 0.038 \, nm$  

L'approssimazione non relativistica è adeguata perché 1 keV è molto minore dell'energia a riposo dell'elettrone, pari a circa 511 keV.
