#### 1.0 Effetti delle radiazioni   - IGNORALO

**Setup**

1- fialetta con dentro del radio e il vuoto  

**Effetti**

effetto | cosa | conclusione 
---|---|---
fotografico | avvicino la fiala a lastra fotografica -> la impressiona <br> " a oggetto fluorescente -> causa fluorescenza
fluorescenza indotta | la fiala stessa diventa leggermente fluorescente <br> di più dei punti più vicini al radio <br> se pareti sottili, anche gli oggetti esterni hanno fluorescenza indotta | c'è una parte di radiazione poco penetrante
elettrostatico | fiala ottiene carica negativa (pian piano) | la radiazione penetrante ha carica +, quella poco penetrante ha carica -  
ionizzazzione
sviluppo di calore | siluppa 160 cal/g per hour  
sostanze gassose | nella fiala sia ccumulano 2 gas: <br> elio: aumenta linearmente col tempo <br> radon: all'inizio aumenta, ma poi si asintotizza

radon: è radioativo e pericoloso  
si accuula in grotte sotterane, cantine...

#### 2.0 Tipi di radiazioni

**Setup**

1- fiala con radon e vuoto  
2- lastre di Al o Pb, di vario spessore  
3- rilevatore  

**Obiettivo**

1- quantificare il decadimento  
2- dividere i diversi tipi  

come:   
uso la ionizzazione dell'aria, l'effetto più facile da quantificare  
il rilevatore è un tubo a vuoto, quanto più l'aria viene ionizzaata, tanto più passa corrente  

![](graphs/radiation_absorption.png)

**Grafico**

asse x: spessore della lastra messa tra sorgente di radiazione e rilevatore  
questo è il grafico per la lastra di Pb  
quello per l'alluminio è simile, ha pure 3 parti, ma tutte e 3 sono un po' più ripide  

asse y: numero di particelle che colpiscono/raggiungono il rilevatore al secondo (ln A)  

**Outcome**

nel grafico, posso distingere 3 parti  

numero | disegno | cosa  | penetrazione | spiegazione
---|---|---|---|---
1 | linea verticale | radiazione $\alpha$| bassa | anche una lastra piccolissima, quasi inesistente, le blocca <br> quindi fa diminuire subito il numero di particelle che raggiungono il rilevatore
2 | arco di cerchio <br> fino a 2.5 quadretti | radiazione $\beta$ | media | diminusice sensibilmente man mano che aumento lo spessore
3 | linea quasi orizzontale | radiazione $\gamma$ | alta | per quanto aumento lo spessore della lastra, passa comunque


**Formula importante**

$p[MeV] = 3 \, B[T] \, R[cm]$  
mnemonic: Per ME 3 BoTtiglie di Rum in CoMune  

per:  
calcolare l'impulso di una particella in MeV  
quado soggetta ad un campo magnetico in Tesla  
sapendo il raggio della sua traiettoria in cm  

nota che E,p,m usano tutte l'unità di misura MeV  
perchè considero c=1, come in relatività, quindi E=m+p, hanno tutte le stesse unità di misura
