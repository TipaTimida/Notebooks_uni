### Q-value

**Defs**

$Q = [M_{iniz}-M_{fin}]c^2$    
M è la massa NUCLEARE   

$Q = T_f -T_i$  
energy released (nota che scambio iniziale e finale)    
perchè: consrvazione componente energia del 4-impulso, in cm frame: $m_x = E_1+E_2 = m_1+T_1+m_2+T_2$  

**Significato**
 
1- Q>0: decay permesso   
MA non so se accade, quello dipende anche della dinamica, ci devono essere le interazioni giuste (tunneling attraverso barriera di potenziale) 

2- Q<0: decay non permesso   
di sicuro non può accadere spontaneamente

### Energia cinetica

**Modello**

esempio: alpha decay    
X -> Y + $\alpha$    

mi metto in frame di CM, cioè di X  

leggi di conservazione:  
$\begin{cases}
M_xc^2=M_yc^2+M_\alpha c^2+T_y+T_\alpha \\
0= \vec p_y + \vec p_\alpha
\end{cases}$

calcolo Q:  
$Q = (M_x-M_y-M_\alpha)c^2 = T_y+T_\alpha = \frac{p^2}{2M_Y}+  \frac{p^2}{2M_\alpha}$    
$  = \frac{p^2}{2M_\alpha} (1+ \frac{M_\alpha}{M_y})$  

**Energia particelle**

come faccio:  
1- per $\alpha$: $T_\alpha = \frac{p^2}{2M_\alpha} $ è il primo fattore di Q    
2- per Y: faccio $T_y = \frac{p^2}{2M_\alpha} \frac{M_\alpha}{M_y}$  

$T_\alpha = \frac Q{1+M_\alpha/M_Y} \sim  Q (1- \frac{M_\alpha}{M_Y}) \sim Q(1-\frac {4}{A})$   
$T_Y = Q \frac{M_\alpha/M_Y}{1+M_\alpha/M_Y} \sim Q \frac 4 A$   
quindi: $T_\alpha \gg T_Y$  

approssimazione per nuclei pesanti:   
$\frac {M_\alpha}{M_y } \sim \frac 4{A-4} \sim \frac 4 A$  
cioè, tolgo $\overline B$ da entrambe le masse???

**Osservazioni**

in alpha decay, la particella $\alpha$ prende quasi tutta l'energia cinetica, perchè è più piccola     
come durante un urto normale, palla contro muro, il muro è così più pesante che rimane fermo  

infatti: $Q= T_f-T_i$   
ma X non ha energia cinetica perchè sono nel suo frame  
e Y è molto più pesante, quindi l'urto con $\alpha$ non lo muove    
quindi tutta l'energia va a $\alpha$    

infatti, se $M_y \gg M_\alpha$, ho:  
$Q \sim \frac{p^2}{2M_\alpha}$  
cioè Q è l'energia cinetica ottenuta dalla particella $\alpha$    

**Classica o relativistica?**

se $M \gg Q$, allora posso usare la meccanica classica  
perchè le particelle non sono veloci rispetto alla loro massa  
dove stimo: $m = A \, u$ (numero di massa x energia di un protone)  

classica: $T=\frac{p^2}{2m}$  
relativistica: $T = E-m = \sqrt{p^2-m^2}-m$  

intuizione: se classica, pensalo come un urto tra palline  
la più leggera acquista più velocità, la più pesante rincula solo un pochino (per conservare il momento)  

**Shortcut: energia in decay**

1- decay a 2 corpi:  
è obbligata: $E_1 = \frac{m_x^2+m_1^2-m_2^2}{2m_x}$ (nel frame di X)  
(m sono le masse invarianti della particelle, cioè la norma del 4-impulso)    
come si trova: conservando il 4-impulso $p_x^\mu =p_1^\mu + p_2^\mu$   

2- decay a 3 corpi:  
non è obbligata: ha spettro di valori possibili, con un min e un max  
non è obbligata perchè ci sono troppi ndof  

### Binding energy e masse

**Massa di un sistema legato**

$M_{sys}= \sum_{componenti} m -B/c^2$    

nucleo e atomo sono sistemi legati  

che scala | nucleo | atomo
---|---|---
massa del sistema legato| $M_n(A,Z) = Zm_p +(A-Z)m_n-B$ | $M_{at}(A,Z) = M_{n}(A,Z) +Z m_e -\frac{B_e}{c^2}$  
componenti del sistema legato | Z protoni, A-Z neutroni | nucleo, Z elettroni  
interazione responsabile del legame | nucleare forte | EM
energia di legame | B: energia di legame nucleare <br>  Energia necessaria a separare il nucleo in protoni e neutroni liberi | $B_e$: energia di legame degli elettroni <br> Energia necessaria a separare il nucleo e gli elettroni liberi  
si trascura B? | no <br> o la scrivo in funzione di $\Delta M$, che prendo dai database<br> o, per A grandi, $B(A,Z) = \overline B \, A$ | sì  

| | $\Delta M$: difetto di massa nucleare | $\Delta$: eccesso di massa |
|---|---|---|
| **Formula** | $ \Delta M=Zm_p+(A-Z)m_n-M_n$ |  $ \Delta=(M_{at}-Au)c^2$
| **Cosa confronta?** | Massa dei nucleoni **separati** meno massa del **nucleo legato** | Massa dell’**atomo** meno il riferimento convenzionale **Au** |
| **Unità** | Massa: kg, \(u\), MeV/\(c^2\) | Energia: J, MeV |
| **Segno** | Positivo per un nucleo legato | Può essere positivo, negativo o zero |
| **Rapporto con \(B\)** | $B=\Delta M \; c^2$| Non coincide con $ B_e$ (che comunque si trascura) |

di solito, quello tabulato nei databse è l'ECCESSO DI MASSA  

**Energia di separazione**

sistema: un intero nucleo  
componenti: due nuclei più piccoli  
B: energia per separare il nucleo in questi due più piccoli (NON per separare ogni nucleone)  

eg. C -> D  
$M_n(C) = 2M_n(D) -B$
