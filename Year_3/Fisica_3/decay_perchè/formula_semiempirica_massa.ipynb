### Binding energy

![](./graphs/binding_energy.png)

**Forma**

1- prima parte: picchi  
è la fissione nucleare, la reazione va verso destra, i nuclei aumentano A  

2- seconda parte: plateau     
è la fusione nucleare, la reazione va verso sinistra, i nuclei diminuiscono A  


**Scala: da ricordare**  

il picco è a (50, 8)    
asse x arriva a 250, asse y arriva a 8  

asse x: A  
asse y: B/A   
quindi: $B/A  \sim 8MeV$  nel plateau  

**Plateau**

per A grandi: B/A = const  
quindi: $B(A,Z) = A \frac{B(A,Z)}{A} = A \, \overline B$  
dove: $\overline B \sim 8 MeV$, l'altezza del plateau  

**Forze a corto vs lungo raggio**

raggio | corto | lungo
---|---|---
esempio | coulombiana | nucleare 
che vuol dire | funziona anche a distanze grandi | decade subito, funziona solo a distanze piccole
formula | $B= A \frac{v_{eff}}V$ <br> interagisce solo con gli oggetti molto vicini, all'interno di un volume efficace| $B = N_{coppie} = \frac 12 A(A-1)$ <br> ogni coppia di oggetti interagisce
come sarebbe il plot B/A - A  | plateau <br> $\frac B A = const$  |  retta <br> $B/A \sim A$  

#### Modello a goccia

**Perchè**

per avere $\frac BA=const$, poichè $v_{eff}$ è fisso e uguale per ogni nucleo, devo avere: $\frac AV = const$ (guarda la formula)  

cioé: $V \propto A$  
vuol dire che il volume del nucleo (V) aumenta col numero di nucleoni (A)   
quindi il nucleo si comporta come un liquido: se aggiungo pezzi, aumenta il volume, non la densità (come invece è il caso dei gas)  

**Proof: A $\propto$ V**

$B = uN_{coppie}$   
u: energia di legame di una coppia  
$N_{coppie}$: quante coppie interagiscono dentro il nucleo  

conta le coppie:  
in EM, ogni elettrone interagisce con tutti gli altri  
qui no, ognuno interagisce solo con quelli vicini, dentro un "volume effettivo", perchè la forza nucleare è a corto raggio 

quanti nucleoni contiene il "volume effettivo": $\rho V_{eff} = \frac AV V_{eff}$ 
$\rho$: densità di nucleoni

quante sono le coppie: 
$N_{coppie}= \frac 12 A_1 A_2 = \frac 12 \, A \, \frac AV V_{eff}$    
il primo nucleone è uno qualunque del nucleo, il secondo è nel volume effettivo del primo  
1/2: perchè la coppia A1-A2 e A2-A1 sono la stessa, sennò le conto due volte  

conclusione:  
$B \sim \frac u2 \; A \; \frac AV V_{eff}$    
so che: $\frac BA = \frac u2 \frac AV V_{eff} = const$  
poichè $V_{eff}$ e $u$ sono costanti, allora $\frac AV = const$   
Dunque il volume cresce in proporzione al numero di nucleoni

### Quando nucleo è stabile?

**Grafico**

per A fissato, $B(Z) \sim Z^2$  
quindi è una parabola  

un nucleo Z può decadere verso destra, quindi verso uno con Z+1, tramite $\beta_-$  
mentre verso sinsitra, cioè verso Z-1, con $\beta_+$ oppure CE    

**Cosa devo fare**

1- assumo A fisso in $M_n(A,Z)$  
così: $M_n(A_{fixed},Z) \sim Z^2$  
è una parabola (con buco veso l'alto)  

2- voglio trovare l'elemento con M minore, cioè quello al vertice  
è il più stabile, non decade, mentre gli altri sì  
cioè: dato questo A, quale Z dà l'elemento più stabile? quale è la miglior proporzione di protoni e neutroni per avere un nucleo sttabile?  

3- calcolo lo Z per cui: $\frac{\partial M(A,Z)}{\partial Z} =0$  
nota che la formula cambia per i casi PP, DD, DP (la parte $B_P$)  

**Nota**

$\frac{\partial M(A,Z)}{\partial Z} = (m_p-m_n) - \frac{\partial B(A,Z)}{\partial Z}$    
$\sim - \frac{\partial B(A,Z)}{\partial Z}$  
perchè protone e neutrone hanno quasi stessa massa  

**Caso: A dispari (caso PD)**

$Z = \frac{A}{2+\frac{a_c}{2a_a}A^{1/3}} = \frac{A}{2+0.0157A^{1/3}}$    

1- A piccoli: $Z \sim \frac A2$  
cioè N=Z, sta sulla bisettrice del piano N-Z  

2- A grandi: $Z <\frac A2$  
non posso trascurare la correzione  
scende un po' sotto la bisettrice  

grafico: solo una parabola  
quello stabile è quello al vertice  

**Caso: A pari**

ci sono due parabole, per i casi PP e DD  
sono una sopra l'altra, separate da $2\frac{a_p}{\sqrt A}$  
cioè il doppio delsl'ultimo termine, $B_P$, quello che può essere + o -  

solo il vertice della parabola sotto è stabile  
quello della parabola sopra può decadere nella parabola sotto, con $\beta$ decay  

**Osservazione**

maggiore è B/A, più quel nucleo è stabile  
cioè, Fe è il nucleo più stabile, non fa fissione nè fusione nuclare  

### Formula semi-empirica di massa 

![](./parabole_formula_semi-empirica.png)

**Cosa è**

dice i componenti binding energy  
per approssimare l'energia di legame B(A,Z) e la massa di un nucleo atomico   
basandosi sul modello a goccia di liquido

**Nome**

semi-empirica perchè unisce parte teorica e sperimentale  

teoria: per scrivere la funzione di fit  
cioò l'approssimazione a goccia di liquido  
e quando considero i vari effetti, tipo la superficie, la repulsione elettrica e l’appaiamento dei nucleoni  

sperimentale: i punti che plotto sul grafico, e che uso per fittare i coeff  

**Componenti**

formula | nome | che vuol dire
---|---|---
$B_v = a_vA$ | volume | nucleone interagisce solo con nucleoni vicini <br> forza nucleare è a corto raggio  
$B_S = -a_S A^{2/3}$ | superficie | nucleoni sulla superficie hanno meno volume efficace, sono più facili da rimuovere
$B_c = -a_c \frac{Z(Z-1)}{A^{1/3}}$ | coulombiana | i protoni hanno carica uguale, si respingono <br> se ce n'è tanti, il nucleo è più fragile <br> scala su 1/r, come Coulomb <br> $Z-1 \sim Z$ (ogni protone interagisce con tutti i protoni tranne sè stesso), perchè forza coulombiana è a lungo raggio    
$B_A = a_A \frac{(A-2Z)^2}A$ | asimmetria | a differenza tra il numero di protoni e neutroni  
$B_P = \pm \frac{a_p}{\sqrt A}$ <br> PP=+, DD=-, PD=0 (NZ, numeri di protoni e neutroni)| parità | se gli spin sono allineati o disallineati, l'energia cambia  

nota: se B diminuisce -> il nucleo è più fragile, più facile staccare nucleoni  

i coefficienti a si trovano con un fit  
$a_r \sim 1.75 Mev \quad a_s \sim 17.8 \quad a_c\sim 0.71   \quad a_a \sim 23.7 \quad a_p \sim 11$

**Scala dimensioni nucleo**

$V \sim A$  
quindi (assuendo nucleo sferico):  
$S \sim A^{2/3}$  
$R \sum A^{1/3}$  

### Spiegazione coeffs (modello a goccia)

**Av**

graficamente: plateau

così B/A è costante     

che vuol dire: considero solo i nuclei abbastanza vicini da risentire della forza nucleare, che è a corto raggio 

**Aa**

protoni e neutroni sono fermioni (particelle a soin semi-intero) (per bosoni, intero)  
th di spin: la waveform di un sistema di fermioni è dispari per scambio di aspin (per bosoni, pari)  
-> quindi, due fermioni non possono occupare lo stesso stato quantico (principio di Pauli???)  

$N \propto V p_F^3 = A F_p^3$  
$p_F$: impulso di Fermi (massimo impulso raggiunto da una particella quando sono impilate nello spazio delle fasi)   
N: numero di neutroni  
quindi: $p= (\frac NA)^{1/3}$    
Possiamo immaginare gli stati occupati come una sfera nello spazio degli impulsi  
Il numero di stati occupati dipende dal volume di questa sfera: $N \propto V p_f^3$  
N è spiegata sotto
analogamente, posso trovare Z; per i protoni  

energia cinetica di neutroni: $T_N \propto N \frac{p_N^2}{m_n}$  
$T_N \propto \frac {N^{5/3}}{A^{2/3}}$  
$T_P \propto  \frac {Z^{5/3}}{A^{2/3}}$   
$T_P+T_N =  \frac {Z^{5/3} + N^{5/3}}{A^{2/3}}$    

cosa voglio: vedere come cambia l'energia se, invece di Z=N, ho $Z \gg N$ oppure $Z \ll N$  
$N= \frac A2 + \Delta \quad Z = \frac A2 - \Delta$  
$T_P+T_N \propto \frac{(\frac A2 - \Delta)^{5/3} +  (\frac A2 + \Delta)^{5/3}}{A^2/3} $  
$= \frac {(\frac A 2)^{5/3}(1-\frac {2 \Delta}A)^{5/3} + (\frac A 2)^{5/3}(1+\frac {2 \Delta}A)^{5/3}}{A^{2/3}}$  
$= \sim A [1-\frac {10} 3+ \epsilon \frac{\Delta^2}{A^2} +1 +\frac {10}3 \frac \Delta A + \epsilon  \frac{\Delta^2}{A^2}]$  
$ = A(2+\epsilon \frac{\Delta^2}{A^2})$  
$\sim  \frac{\Delta^2} A$  
$ = \frac{(N-Z)^2}A = \frac{(A-2Z)^2}A$  
ho usato che $\Delta \rightarrow 0$, uso Taylor  
$\Delta$: asimmetria tra numero di neutroni e protoni nel nucleo  

**Aa - intuizione**

se Z+N=A è fissato:  
un nucleo con più neutroni che protoni deve “riempire” più stati disponibili per i neutroni  
Per il principio di Pauli, i neutroni in eccesso finiscono in stati con impulso maggiore e quindi energia cinetica maggiore  
Lo stesso vale, al contrario, se ci sono più protoni


$T_p+T_n$: energia di simmetria    
misura l’aumento di energia cinetica rispetto al caso equilibrato N=Z    
È positiva perché uno sbilanciamento aumenta l’energia cinetica
MA, nella formula semi-empirica, scrivo B, quanto è legato

**Stati per particelle**

1- protoni e neutroni hanno due buche di potenziale diverse  
perchè protoni sentono anche forza coulombiana, mentre neutroni solo quella forte  

2- Immaginiamo di mettere un certo numero di protoni nella buca di potenziale.  
per prinipio di Pauli, non possono occupare lo stesso stato  
occupano prima gli stati a energia più bassa, poi quelli progressivamente più alti  
a temperatura=0, tutti gli stati sono occupati fino a un certo impulso massimo, $p_f$ = impulso di Fermi    
L'impulso di Fermi rappresenta quindi il massimo impulso occupato nello stato fondamentale del nucleo   

3- numero di stati disponibili    
1-  
particella confinata in volume V  
ogni stato corrisponde ad un volume $(2 \pi \hbar ^3)$ nello spazio delle fasi (visto a meccanica classica)     
quindi nello spazio degli impulsi, gli stati quantistici sono distribuiti con densità: $\frac V{(2 \pi \hbar)^3}$    
2- 
considero impulsi compresi tra p e p+dp  
Nello spazio degli impulsi, questi punti formano un guscio sferico di volume: $dV_p = 4 \pi p^2 \, dp$  
quindi il numero di stati disponibili è: $dn = \frac V{(2 \pi \hbar)^3} 4 \pi p^2 dp\quad $ (densità per volume)

4- numero di particelle  
integriamo tutti gli stati occupati, da impulso zero fino all'impulso di Fermi: $N = \int_0^{p_F} dn = .. \int_0^{p_F} p^2 dp = \frac{V p_F^3}{3\pi^2 \hbar^3}$    


### Ricavare raggio nucleo

**Metodo 1- usare ac**  

1- $U = a_c\frac {Z^2}{A^{1/3}}$  
U: energia del nucleo, modellizzato come una sfera di raggio $R_0$ con Z protoni (cioè, carica $Q=Zm_e$)  
$a_c$ è un valore noto, ottenuto da un fit    

2- $R_0 \sim r_0 A^{1/3}$  
voglio trovare $r_0$, l'ordine di grandezza del raggio nucleare  

**Sol: metodo 1**

1- carica sfera  
$Q= Zm_e$
$\rho(r) = \begin{cases}
\frac Q {\frac 43 \pi R_0^3}  & r< R_0 \\
0 & r> R_0 
\end{cases}$  

2- campo elettrico sfera  
gauss: come quell di carica puntiforme, con carica uguale a quella dentro la sfera di raggio r  
$E = \begin{cases}
k \frac Q {r^2} \frac{r^3}{R^3}  & r< R_0 \\
k \frac Q {r^2} & r> R_0 
\end{cases}$ 

3- energia  
$U = \int u \; dV$  
$ = \int_{r <R_0} \frac 12 \epsilon E^2 \; 4 \pi r^2 dr \quad \int_{r >R_0} \frac 12 \epsilon E^2 \; 4 \pi r^2 dr $   
$= \frac 35 \frac{e^2}{4 \pi \epsilon_0 \, R_0}$  

4- uguaglio U e coeff di coulomb  
$r_0 =  \frac 35 \frac{e^2}{4 \pi \epsilon_0} \frac 1 {a_c}$  
$= \frac 35 \alpha \, \hbar c \frac 1 {a_c}$  
$= \frac 35  \frac 1 {137} \, 197 MeV \,  \frac 1 {0.71 \, MeV}$  
$ = \frac 35  \; {1.44 \, MeV \, fm} \; \frac 1 {0.71 \, MeV} = 1.22 \, fm$

**Metodo 2 - mirror nuclei (nuclei speculari)**

cosa sono:  
una coppia di nuclei con A=2Z+1  
si scambiano il numero di protoni e neutroni, che è diverso di 1  
${}^{2Z+1}_Z X = {}^{2Z+1}_{Z+1} Y$

come usarli:  
$\Delta B = B(X)-B(Y) = \frac {a_c}{A^{1/3}}[(Z-1)^2-Z^2] $  
$ =  \frac {a_c}{A^{1/3}} A = a_c A^{2/3}$  
cioè tutti gli altri componenti di B si cancellano, rimane solo quello coulombiano  
dove $a_c = \frac 35 \frac{\alpha \hbar c}{r_0}$, calcolato col metodo 1  

cosa fare:   
plot con $A^{2/3}$ su asse z, Q su asse y  
(scrivi formula di Q, ottengo Q in funzione di $\Delta B$, quindi è come plottare $\Delta B$ - però Q è più facile da calcolare)  
calcola Q per tante coppie di nuclei mirror, dovrei avere una retta  
fai fit lineare, dalla pendenza ricavo $r_0=1.39 \, fm$  

B si misura   
(o meglio si misura Q, e poi scrivo la sua formula per esprimerlo in funzione di B)

### Quando è possibile un decay?

**Come saperlo**

1- scrivi Q della reazione  
2- se Q>0, è possibile  

**Quando è possibile $\alpha$ decay?**

esercizio | cosa
---|---
$Q = M(A,Z) -M(A-4, Z-2) -M(4,2)$ <br> $= -B(A,Z) +B(A-4,Z-2) +B(4,2)$  | scrivo la Q dell'$\alpha$ decay
$B(A,Z) = A\frac{B(A,Z)}{A} = A \overline B$  <br> $B(A-4,Z-2) = (A-4) \frac{B(A-4,Z-2)}{A-4} = (A-4) \overline B$  |per nuclei grandi (nel plateau): B/A=const
$ Q = -A\overline B +(A-4) \overline B +B(4,2) >0$ | scrivo disequazione  
$\overline B< \frac{B(4,2)}4 $ <br> $\frac{ B(A,Z)}A < 7 MeV$  | risolvo <br> perchè He si trova su un picco del grafico, ha B/A=7 MeV   
i nuclei in cui è vera sono quelli con A grande <br> serve circa cioè A > 160  <br> cioè solo i nuclei pesanti fanno $\alpha$ decay | conclusione

v⚠️ Un'ultima precisazione importante

Avere Q>0 significa che il decadimento è energeticamente possibile, ma non necessariamente che avvenga rapidamente.

Il decadimento alfa è infatti regolato anche dall'effetto tunnel: la particella alfa deve attraversare la barriera di potenziale coulombiana che la tiene confinata nel nucleo.

Quindi:

    Q>0: decadimento energeticamente consentito.
    Effetto tunnel: determina la probabilità e quindi il tempo di dimezzamento.

Questa distinzione è molto importante in fisica nucleare.
