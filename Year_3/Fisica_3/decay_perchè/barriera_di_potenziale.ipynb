### Condizioni per avvenire

**1. Decay/fissione spontaneo**

mi dice che il proceso/reazione è ENERGETICAMENTE PERMESSO    
ma non mi garantisce che avvenga rapidamente  

in realtà, per avvenire, sia alpha decay che fissione devono attraversare una barriera di potenziale   
quindi quanto veloce avviene dipende dalla prob di avere effetto tunnel  

1- Q>0: rende energeticamente possibile il decadimento  
2- barriera: determina quanto è probabile e quanto a lungo bisogna aspettare  

**Barriera di potenziale**

alpha decay: barriera di potenziale coulombiana che la respinge dal nucleo  
fissione: barriera di fissione, legata all’energia necessaria per deformare il nucleo  

**Perchè barriera di potenziale**

Q>0 dice che stato finale ha energia minore di stato iniziale  
e so che un sistema evolve verso uno stato ad energia inferiore  

MA: per arrivarci devo attraversare un punto ad energia maggiore, la barriera di potenziale  
pensa ai grafici di chimica  

**Nota: regola d'oro di Fermi**

descrive il decay spontaneo  
la prima parte dice quanto l’interazione favorisce la transizione, la seconda quanti stati finali sono disponibili  

il Q>0 positivo rende possibili stati finali  
ma l’interazione e la probabilità di raggiungerli determinano quanto spesso avviene il processo  

**2. Fissione indotta**

anche se NON è vero Q>0  

come: bombardo il nucleo con un neutrone (che non è carico, quindi non viene sente la barriera coulombiana)   
il neutrone dà energia al nucleo (lo fa passare allo stato eccitato), così ora ha abbastanza energia per superare la barriera

#### Effetto tunnel

**Tunnel effect**

$\lambda = P \nu$  
$\lambda$: prob di decadere con $\alpha$-decay  
P: prob che avvenga l'effetto tunnel (si calcola sotto, in funzione di G)
$\nu$: freq degli urti di $\alpha$ con la barriera di potenziale  

cioè, immagino la particella $\alpha$ come una particella pre-formata all'interno del nucleo (nel pozzo di potenziale)  
in fisica classica non può superare la barriera  
ma qui sì, con effetto tunnel  

**Effetto tunnel: esempio fisica classica**

1- setup  
due prismi di vetro separati da un sottile strato d'aria  

2- cosa succede  
tra il primo prisma e l'aria, c'è riflessione totale  
cioè il campo E nell'aria, perpendicolare, alla superficie di separazione, è evanescente  

MA, evanescente vuol dire che decade veloce, come un esponenziale, non all'istanet  
se lo strato d'aria è abbastanza corto, il campo arriva all'altro prisma che non è ancora decaduto, e quindi lo vedo comunque nell'altro prisma  
(anche se in teoria c'è riflessione totale, quindi a prima occhiata non dovrebbe esserci raggio trasmesso)

#### MQ

**Barriera di potenziale**

c'è una barriera di potenziale coulombiana  
la approssimo come una serie di barriere di potenziale rettangolari  
con la MQ, calcolo la prob per la wavefunction di superare la barriera  

prob di arrivare a distanza r dal centro del nucleo:  
$\phi(r) \sim exp[-\int_a^b \frac{\sqrt {2m [V(r)-E]}}{\hbar} dx ]= exp[-G]$    
$P \sim |\phi|^2 \sim e^{-2G}$  
r=a: quandola barriera inizia  
r=b: quando la particella $\alpha$ esce dalla barriera   
cioè a,b sono le due intersezioni della riga orizzontal e della barriera  
G: fattore di Gamov (per scrivere più veloce)  

risolvo G, quando barriera coulombiana $V(r)= \frac{zZe^2}r$:  
$G= \frac{\sqrt {2m}} \hbar \int_a^b \sqrt{\frac{zZ^2e^2}r-E} dr$    
$ = b \frac{\sqrt{2mE}} \hbar \int_{a/b}^1 \sqrt{\frac 1 y -} dy$  
$ = \frac {zZE^2} E \frac{\sqrt{2mE}} \hbar A$    
$ = \frac {zZe^2 \sqrt{2m}}{\hbar \sqrt E} \, A$  
A è un integrale, un numero  


**Trovo legge di Geiger-Nuttel**

dalla formula di G; posso scrierlo:   
$G= \frac d{\sqrt E} + f$  
$P \sim e^{-2G}$  
$\tau = \frac 1 \tau = \frac 1{\nu P} \frac 1 \nu exp[\frac d{\sqrt E} + f]$  

quindi:  
$\log \tau = \frac{a}{\sqrt E}+b$   

nota: Geiger-Nuttel è osservazione sperimentale  
ma poi, la posso spiegare con la teoria  

### Fissione

**Come avvengono le cose**

1- alpha decay:  
la particella alpha è preformata all'interno del nucleo  
con effetto tunnel, esce fuori, superando la barriera di potenziale coulombiana  

2- fissione nucleare  
inizialmente nucleo è tondo  
se riceve energia diventa ellissoide (con stesso volume), finchè non è più stabile e si divide in due (tipo mitosi)  

**Perchè ellissoide non è stabile**

cambiano alcune energie  

1- modellizzo la deformazione:  
ellisse con asse lungo b, gli altri due a  
$\begin{cases} 
a = r(1+ \epsilon) \\
b = \frac r {\sqrt{1 - \epsilon}} \sim r(1-\frac \epsilon 2 + \frac 38 \epsilon^2) \end{cases}$  

2- energia di superficie    
intuitivo: a parità di volume, sfera ha superficie minore  
$E_s = a_s A^{2/3} (1+ \frac 25 \epsilon^2)$  
l'ultimo termine è di quando è cambiata la superficie dell'ellissoide, rispetto a quella della sfera  

3- energia coulombiana  
intuitivo: nell'ellise, i protoni sono più lontani tra loro, si respingono di meno  
$E_c = a_c \frac{Z^2}{A^{1/3}}(1-\frac 1 5 \epsilon^2)$  

4- somma delle variazioni  
$\Delta E = \frac{ \epsilon^2}5 a^{2/3} (2a_s -a_c \frac{Z^2}A)$  
se $\Delta E <0$, il nucleo deformato non è più stabile -> c'è fissione in due nuclei  
quindi, fissione può accadere se: $\frac{Z^2}{A} > \frac{2a_s}{a_c} \sim 48$  

**Modello a goccia vs guscio**

secondo modello a goccia: Uranio dovrebbe fare fissione in due nuclei uguali (la metà di lui)  
invece nella realtà no, si separa in uno grande e uno piccolo  
quindi il modello a goccia ha dei problemi  

il perchè è spiegato dal modello a guscio, che studia i gusci elettronici, e dice he alcuni sono più stabili, quindi sono preferiti
