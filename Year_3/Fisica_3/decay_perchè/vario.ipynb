### Geiger-Nuttal

**Cosa**

osservazione sperimentale/empirica    
dice la relazione tra l'energia di una particella alpha emessa in un decay e l'emivita dell'elemento che decade  

**Cosa**

$\ln T_{1/2} = \frac A {\sqrt Q} +B$  
A,B are coeffs, just to say posso fittarlo come una retta  

quindi:  
piccola variazione di Q (energia dell'$\alpha$ paricle) -> grande variazione di $T_{1/2}$
