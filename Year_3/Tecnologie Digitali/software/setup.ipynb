#### Moving a venv

**Moving the project dir**

do not move the ./venv, cause trouble with paths of packages  

what to do  
1- save all pacakges downloaded in it  
2- move the project files  
3- recreate the environment (with the same name)  
4- download all packages   


```python
# save and deactivate current one
source .venv/bin/activate
python -m pip freeze > requirements.txt
deactivate

#...move the project dir...

# remake one in new location
python3 -m venv .venv
source .venv/bin/activate
python -m pip install -r requirements.txt
```

### Install things

```python
# Python modules

# download
pip install packet_name
pip install -r requirements.txt   
python -m pip install package_name   # in current venv  

# check if there is stuff
pip show packet_name
pip list
pip check  # check if packages have incompatible or missing dependencies
pip freeze  # produce list of dependencies, save it to requirements.txt

# delete
pip uninstall packet_name
```

### Kill programs

```python
# find id of thing to kill  
ps aux | grep tina

# normal kill, with id
kill 1234

# normal kill, kills all processes who match (contain) this name
pkill firefox

# normal kill, with this EXACT process name
pkill -x firefox

# forceful kill  
kill -9 1234
kill -SIGKILL 1234
pkill -9 -x firefox
```
