**Advanced search**

param | what
---|---
NPeaks=4 | only return 4 peaks (the biggest ones)  
SortStr=descend  | returns the peaks from the highest to the lowest, NOT in the order they appear in  
MinPeakProminence=0.2 | quanto deve svettare almeno tra i punti attorno a sè  
MinPeakHeight=0.366 | what value of y it must reach  
MinPeakDistance = 100 | minimum distance in the locs (so, in the x coords if i specified the array x) of two peaks  

**What it returns**

name | what
---|---
pks | the local maximums in the array y <br> they're returned in the order they appear in (unless i use SortStr)
locs | if there is no array x -> the indexes of y where the peaks appear <br> if there is array x -> the value of x at the indexes where the peaks appear in y -> essentially, x(locs)
widths | the width of the peak, $\sigma$ of the gaussian
