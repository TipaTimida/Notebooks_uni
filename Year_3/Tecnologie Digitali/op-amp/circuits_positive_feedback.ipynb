### Come risolvere

**Cosa fa**

cosa succede: se c'è uno squuilibrio tra gli ingressi V+ e V-, queato viene incrementato, fino alla saturazione   
per semplificare, diciamo che satura a $V_{s\pm}$, anche se è veroolo per i rail-to-rail, e anche lì è un'approssimazione   

**Come risolvere**

devo usare IPOTESI AUTOCONSISTENTI    
cioè, ipotizzo che Vout sia ad una saturazione  
cerco la condizione ai morseti perchè l'op-amp continui a spingere $V_{out}$ verso questa stessa saturazione, non l'opposta  

$V_{out} = A (V_+-V_-)$   
assumo $A = \infty$, l'op-amp usa la differenza tra gli input per estremizzare $V_{out}$   
in feedback positivo:  

caso | cosa succede
---|---
$V_+ > V_-$ |  $V_{out}=V_{s+}$ <br> output spinto verso in alto, fino a saturare
$V_+ <V_-$ |  $V_{out}=V_{s-}$ <br> output spinto verso in basso, fino a saturare  

### Schmitt trigger (inverting)

![alt text](https://hackatronic.com/wp-content/uploads/2024/11/Schmitt-Trigger-Circuit-Types--1210x642.webp)

**Schematics**

è come un inverting amplifier  
ma con retroazione positiva invece che negativa (cioè scambio V+ e V-)  

**Cosa fa**

guarda il plot di VIn e VOut  
zona instabile: la zona compresa tra le righe orizzontali rosse (thresholds: $\beta V_{s-}$, $\beta V_{s+}$)    
quando $V_{in} > \beta V_{s+}$, cioè attraversa la linea rossa in alto -> viene settato $V_{out} = V_{s-}$  
quando $V_{in} < \beta V_{s-}$, cioè attraversa la linea rossa in basso -> viene settato $V_{out} = V_{s+}$  


**Intuizione: elettronica digitale**

se Vin sale sopra la linea in alto, setta Vout=LOW  
se Vin scende sotto la linea in basso, setta Vout=HIGH  
finchè Vin non fa queste due cose, rimane lo stato precedente, quindi Vout dipende dalla storia precedente (è una memoria)  

**Modello**

circuito è come un inverting amplifier, quindi ai nodi di input:   
$V_- = V_{in}$  
$V_+ = \beta V_{out} = \frac{R_1}{R_1R_2}V_{out}$  

stato attuale | condizione perchè si mantenga | condizione qui
---|---|---
$V_{out}=V_{s+}$ | output spinto in su: $V_+> V_-$ | $\beta V_s >V_{in}$  
$V_{out}=V_{s-}$ | output spinto in giù: $V_+ < V-$ | $\beta V_{s-} <V_{in}$  

se la condizione viene violata, passa all'altro stato

### Multivibratore astabile

**Circuito**

al trigger di Schmitt, aggiungi un'altra rete di retroazione   
cioè una rete RC (tra Vout e Vin)  

**Cosa fa**

Vin: pinna di squalo (carica e scarica di C)  
Vout: un'onda quadra, tra $V_{s+}$ e $V_{s-}$

genera un'onda quadra senza alcun segnale di ingresso  

**Modello**

cambia il voltaggio ai nodi:  
$V_- = V_c$  
$V_+ = \beta V_{out} = \frac{R_1}{R_1+R_2}V_{out}$  

stato attuale | condizione perchè si mantenga | condizione qui | cosa succede nel circuito
---|---|---|---
$V_{out}=V_{s+}$ | output spinto in su: $V_+> V_-$ | $\beta V_s >V_c$ | C si carica, quindi $V_c$ aumenta <br> finchè supera la soglia
$V_{out}=V_{s-}$ | output spinto in giù: $V_+ < V-$ | $\beta V_{s-} <V_c$ | C si scarica, quindi $V_c$ diminusce <br> finchè scende sotto la soglia

osservazione: se $V_{out}=V_{s+}$ oppure $V_{s-}$, cambia il verso della corrente che va verso C, quindi si carica o scarica  

**Periodo**

il condensatore si carica e scarica per stare tra le due threshold:  
$\beta V_{s-}< V_c < \beta V_{s+}$  
un periodo è il tempo per partire da sotto, arrivare sopra, e tornare sotto (nota che il tempo per andare sa giù a su o da su a giù e lo stesso)  

tempo per caricarsi da $-V_{s}$ a $+V_{s}$ (assumo hanno stesso modulo)  
$V_c(t) =V_{s}+(V_c(0)-V_{s})e^{t/\tau}$    
$V_c(0)=-\beta V_s$, V del condes quando inizia la carica  
quindi: $V_c(t)=V_s[1-(1+\beta)e^{t/\tau}$  
$t= \tau \ln (\frac{1+\beta}{1-\beta})$  

periodo: $T = 2t$
