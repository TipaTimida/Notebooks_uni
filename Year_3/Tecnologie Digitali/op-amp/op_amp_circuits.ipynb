### Read schematic

**How to model it**

$\begin{cases}
V_+=V_- \\
KCL \, at \, V_+ \\
KCL \, at V_-
\end{cases}$  

what i want: $G= \frac{V_{out}}{V_{in}}$  
or if there are more input voltages, just Vout as a function of all the input: $V_{out}=\sum \beta_k V_{in,k}$  

**Write KCL**

1- find the node i'm doing it at   
here, V+ means the node just in front of it   
bc it draws no current, so the small wire has no voltage drop  

2- consider all the wires starting from that node, and do $ i = \frac{V_i-V_+}{R}$  
can put before the current or other node, just be consistent  
$V_+$ is the voltage my node, $V_i$ is the voltage after the resistor R   

3- se ci sono più input (sorgenti):  
th di sovrapposizione. spegnile tutte tranne una, e calcola Vout (col solito metodo della KCL)    
poi somma tutti i Vout ottenuti con le varie sorgenti   
avrò: $V_{out}=\sum \beta_k V_{in, k}$  

note: if the node is directly connected to a known voltage, don't do KCL  
its eq is just: $V_- = 0$ or $V_+ = V_{in}$  
se è $V_+ = GND$, it's called VIRTUAL GROUND (terra virtuale)  

**Current on a wire**

component | time domain | freq domain
---|---|---
resistor | $I = \frac VR$ | ""
capacitor | $I = \frac d {dt} Q = C \frac {d}{dt}V$  | $I= \frac V{Z_C} = V \, j \omega C$  
inductor | $I = \frac 1 L \int V$ <br> perchè: $V = L \dot I$ | $I= \frac V{Z_L} = \frac V {j \omega L}$
diodo | $I = I_S [e^{-V/(\eta V_T)}-1] \sim  I_S e^{-V/(\eta V_T)}$ | ?


**Da time domain a fq domain**

considera: $V_{in}(t) \rightarrow V(\omega) = e^{j \omega t}$  
quindi: $\int V = \frac 1{j \omega}V \quad \frac d{dt}V = j \omega V$  

## Examples

### Voltage follower

![alt text](https://www.allaboutcircuits.com/uploads/articles/Intro-to-Op-Amps.png)

**What it does**

It does not increase the voltage  
It acts as a buffer between circuits, bc has $R_{in}= \infty$  
so can chain circuits, while keeping them independent stages  

why i can?  
$R_{in} = V_{in}/I_{in} = V_{in}/0 = \infty$  
(i=0 bc of rule 2)  

why i want?  
if i chain two filters normally, i get $H \neg H_1 H_2$, bc they modify each other by drawing current  
if i chain them with a buffer in between: $H = H_1 H_2$, they're independent stages  

**Formula**

$\begin{cases}
V_+=V_{in} \\
V_- = V_{out}\\
\end{cases}$  

gain  
$G = \frac{V_{out}}{V_{in}} = \frac{V_+}{V_-} =1$  

### Inverting amplifier

**What it does**

amplifies an input signal while “inverting” (giving it dephasing of 180°, or flipping it ???)  

![alt text](https://www.allaboutcircuits.com/uploads/articles/Intro-to-Op-Amps_(1).png)

**How it works**

$\begin{cases}
V_+=0 \\
\frac{V_{in}-V_-}{R_{in}} +\frac{V_{out}-V_-}{R_f} = 0 & KCL \, at \, V_- 
\end{cases}$  

$V_{out} = R_f\frac{V_{in}}{R_{in}}$  
$G  = -\frac{R_f}{R_{in}}$    

**Tip to remember**

inverting: Vin va a V-, inverting input  
NON-inverting: Vin va a V+, the NON-inverting input 

### Non-inverting amplifier

![](https://www.allaboutcircuits.com/uploads/articles/one-op-amp-two_resistors.jpg)

**Formula**

$\begin{cases}
V_+ = V_{in} \\
\frac{0-V_{-}}{R_1} + \frac{V_{out}-V_{-}}{R_2}=0 & KCL \, at \, V_-
\end{cases}$

$ V_{p} (-\frac{1}{R_1}+\frac{1}{R_2}) + V_{out}   (\frac {1}{R_2}) = 0$    
$\frac{V_{out}}{V_p} = \frac{1/R_1+1/R_2}{1/R_1} = 1+\frac {R_1}{R_2}$   
$G = \frac{V_{out}}{V_{in}} = \frac{V_{out}}{V_p} = 1+\frac {R_1}{R_2}$


**Shortcut**

instead of $\Sigma = i$ on node -, i could do:  
R1 and R2 are a voltage divider, and V- is the voltage in their middle    
$V_{-} = \frac{R_1}{R_1+R_2} V_{out}$  

**Voltage-to current**

also called voltage-to-current amplifier  
converts input voltage in a known output current:  

calling $V_{out}$ the current across $R_1$  

node -  
$\Sigma i = I_{out} + \frac{V_{out}-V_p}{R_2}=0$    

together:  
$I_{out} = \frac{V_{in}}R$  

### Summing amplifier

**What it does**

does a weighted sum of all the voltage inputs  
with the resistors as weights  

![text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQ_qcrBsFdeOpRdHPRGpgqQy4axIE5rCvL9s9jw9YgI6w-AtZQ66fMo9y0&s=10)

**Formula**

node +   
$V_p = 0$  

node -  
$\frac{V_{out}-V_p}{R_f} +\frac{V_{in1}-V_p}{R_1} + ... = 0$  

together:  
$\frac{V_{out}}{R_f} +\frac{V_{in1}}{R_1} + ... = 0$  

gain:  
$V_{out} = -R_f (\Sigma \frac{V_{in,i}}{R_i})$ 

### Non-inverting summing amplifer

![text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcT56m9rKN0R8k_2JyWfyATkPAbC380BELnPPL0VHUmjykuc6HT_bVUbGTdi&s=10)

**Formula**

node +  
already seen in node - of the inverting summing  
$\frac{V_{out}-V_p}{R_f} +\frac{V_{in1}-V_p}{R_1} + ... = 0$  
$V_p = \frac{\Sigma \frac{V_i}{R_i}}{\Sigma \frac 1 {R_1}}$  
eg. with 2 inputs: $V_p = \frac{V_1R_2+V_2R_1}{R_1+R_2}$  

node -  
already seen on node - of the non-inverting amplifier  
$\frac{V_{out}}{V_p} = 1+\frac {R_i}{R_f}$  

eg. $R_1 = R_2$ and 2 inputs  
$V{out} = V_1 + V_2$

### Schmidt trigger

![alt text](https://hackatronic.com/wp-content/uploads/2024/11/Schmitt-Trigger-Circuit-Types--1210x642.webp)

**Cosa**

sfrutto retroazione positiva per fare un trigger  
io sto facendo quello invertente   

**Cosa fa**

guarda il plot di VIn e VOut  
la zona compresa tra le righe orizzontali rosse (thresholds: $\beta V_{s-}$, $\beta V_{s+}$) è detta "zona instabile"  
quando $V_{in} > \beta V_{s+}$, cioè attraversa la linea rossa in alto -> viene settato $V_{out} = V_{s-}$  
quando $V_{in} < \beta V_{s-}$, cioè attraversa la linea rossa in basso -> viene settato $V_{out} = V_{s+}$  


**Cosa fa, in parole semplici**

se Vin sale sopra la linea in alto, setta Vout=LOW  
se Vin scende sotto la linea in basso, setta Vout=HIGH  
finchè Vin non fa queste due cose, rimane lo stato precedente, quindi Vout dipende dalla storia precedente (è una memoria)  

Perchè??? COme si modellizza il circuito???

**Modello**

retroazione positiva: $V_+ = \beta V_{out}$  


**Nota**

se aggiungo un rete RC tra Vout e Vin, ottengo che Vin si muove come una pinna di squalo  
e Vout è un'onda quadra  

perchè:  
C si scarica -> Vin supera la soglia in alto -> setto Vout = LOW  
allora C si carica -> Vin scende sotto la soglia in alto -> setto Vout=HIGH  
allora C si scarica...

osservazione: se $V_{out}=V_{s+}$ oppure $V_{s-}$, cambia il verso della corrente che va verso C, quindi si carica o scarica
