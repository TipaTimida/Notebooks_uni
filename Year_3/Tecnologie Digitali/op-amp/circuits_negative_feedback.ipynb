### Read schematic

**How to model it**

$\begin{cases}
V_+=V_- \\
KCL \, at \, V_+ \\
KCL \, at V_-
\end{cases}$  

what i want: $G= \frac{V_{out}}{V_{in}}$  
or if there are more input voltages, just Vout as a function of all the input: $V_{out}=\sum \beta_k V_{in,k}$  

**Write KCL**

1- find the node i'm doing it at   
here, V+ means the node just in front of it   
bc it draws no current, so the small wire has no voltage drop  

2- consider all the wires starting from that node, and do $ i = \frac{V_i-V_+}{R}$  
can put before the current or other node, just be consistent  
$V_+$ is the voltage my node, $V_i$ is the voltage after the resistor R   

3- se ci sono più input (sorgenti):  
th di sovrapposizione. spegnile tutte tranne una, e calcola Vout (col solito metodo della KCL)    
poi somma tutti i Vout ottenuti con le varie sorgenti   
avrò: $V_{out}=\sum \beta_k V_{in, k}$  

note: if the node is directly connected to a known voltage, don't do KCL  
its eq is just: $V_- = 0$ or $V_+ = V_{in}$  
se è $V_+ = GND$, it's called VIRTUAL GROUND (terra virtuale)  

**Current on a wire**

component | time domain | freq domain
---|---|---
resistor | $I = \frac VR$ | ""
capacitor | $I = \frac d {dt} Q = C \frac {d}{dt}V$  | $I= \frac V{Z_C} = V \, j \omega C$  
inductor | $I = \frac 1 L \int V$ <br> perchè: $V = L \dot I$ | $I= \frac V{Z_L} = \frac V {j \omega L}$
diodo | $I = I_S [e^{-V/(\eta V_T)}-1] \sim  I_S e^{-V/(\eta V_T)}$ | ?


**Da time domain a fq domain**

considera: $V_{in}(t) \rightarrow V(\omega) = e^{j \omega t}$  
quindi: $\int V = \frac 1{j \omega}V \quad \frac d{dt}V = j \omega V$  

## Examples

### Buffer (inseguitore) (Voltage follower)

![alt text](https://www.allaboutcircuits.com/uploads/articles/Intro-to-Op-Amps.png)

**What it does**

It does not increase the voltage  
It acts as a buffer between circuits, bc has $R_{in}= \infty$  
so can chain circuits, while keeping them independent stages  

why i can?  
$R_{in} = V_{in}/I_{in} = V_{in}/0 = \infty$  
(i=0 bc of rule 2)  

why i want?  
if i chain two filters normally, i get $H \neg H_1 H_2$, bc they modify each other by drawing current  
if i chain them with a buffer in between: $H = H_1 H_2$, they're independent stages  

**Formula**

$\begin{cases}
V_+=V_{in} \\
V_- = V_{out}\\
\end{cases}$  

gain  
$G = \frac{V_{out}}{V_{in}} = \frac{V_+}{V_-} =1$  

### Inverting amplifier

**What it does**

amplifies an input signal while “inverting” (giving it dephasing of 180°, or flipping it ???)  

![alt text](https://www.allaboutcircuits.com/uploads/articles/Intro-to-Op-Amps_(1).png)

**How it works**

$\begin{cases}
V_+=0 \\
\frac{V_{in}-V_-}{R_{in}} +\frac{V_{out}-V_-}{R_f} = 0 & KCL \, at \, V_- 
\end{cases}$  

$V_{out} = R_f\frac{V_{in}}{R_{in}}$  
$G  = -\frac{R_f}{R_{in}}$    

**Tip to remember**

inverting: Vin va a V-, inverting input  
NON-inverting: Vin va a V+, the NON-inverting input 

### Non-inverting amplifier

![](https://www.allaboutcircuits.com/uploads/articles/one-op-amp-two_resistors.jpg)

**Formula**

$\begin{cases}
V_+ = V_{in} \\
\frac{0-V_{-}}{R_1} + \frac{V_{out}-V_{-}}{R_2}=0 & KCL \, at \, V_-
\end{cases}$

$ V_{p} (-\frac{1}{R_1}+\frac{1}{R_2}) + V_{out}   (\frac {1}{R_2}) = 0$    
$\frac{V_{out}}{V_p} = \frac{1/R_1+1/R_2}{1/R_1} = 1+\frac {R_1}{R_2}$   
$G = \frac{V_{out}}{V_{in}} = \frac{V_{out}}{V_p} = 1+\frac {R_2}{R_1}$


**Shortcut**

instead of $\Sigma = i$ on node -, i could do:  
R1 and R2 are a voltage divider, and V- is the voltage in their middle    
$V_{-} = \frac{R_1}{R_1+R_2} V_{out}$  

**Voltage-to current**

also called voltage-to-current amplifier  
converts input voltage in a known output current:  

calling $V_{out}$ the current across $R_1$  

node -  
$\Sigma i = I_{out} + \frac{V_{out}-V_p}{R_2}=0$    

together:  
$I_{out} = \frac{V_{in}}R$  

### Sommatore (Summing amplifier)

**What it does**

does a weighted sum of all the voltage inputs  
with the resistors as weights  

![text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQ_qcrBsFdeOpRdHPRGpgqQy4axIE5rCvL9s9jw9YgI6w-AtZQ66fMo9y0&s=10)

**Modello**

KCL at node -  
$\sum I_k + I_f=0$  
$\frac{V_{in,k}}{R_k} + \frac{V_{out}}{R_f} = 0$  
$V_{out} = -R_f (\Sigma \frac{V_{in,k}}{R_k})$ 

#### NOT seen: Non-inverting summing amplifer

![text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcT56m9rKN0R8k_2JyWfyATkPAbC380BELnPPL0VHUmjykuc6HT_bVUbGTdi&s=10)

**Formula**

node +  
already seen in node - of the inverting summing  
$\frac{V_{out}-V_p}{R_f} +\frac{V_{in1}-V_p}{R_1} + ... = 0$  
$V_p = \frac{\Sigma \frac{V_i}{R_i}}{\Sigma \frac 1 {R_1}}$  
eg. with 2 inputs: $V_p = \frac{V_1R_2+V_2R_1}{R_1+R_2}$  

node -  
already seen on node - of the non-inverting amplifier  
$\frac{V_{out}}{V_p} = 1+\frac {R_i}{R_f}$  

eg. $R_1 = R_2$ and 2 inputs  
$V{out} = V_1 + V_2$  

### Sottrattore (Difference amplifier)

![](https://www.allaboutcircuits.com/uploads/articles/differential-op-amp-circuits.jpg)  
Rf in alto a destra, Rg in basso a destra  
R1, R2 a sinitra  

**Modello**

KCL ai nodi V+ e V-  
$\begin{cases}
\frac{V_{in,1}-V_+}{R_1} + \frac{V_{out}-V_-}{R_f} = 0 \\
\frac{V_{in,2}-V_+}{R_2} + \frac{V_{out}-V_+}{R_g} = 0 \\
V_+=V_- \end{cases}$
 
risolvilo con principio di superposizione  

$G =  A_1V_{in,1}+A_2V_{in,2} = \frac{R_g(R_1+R_f)}{R_1(R_2+R_g)}V_{V_{in, 2}}-\frac {R_f}{R_1}V_{in,1}$


### Differentiator

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/_1024x432_crop_center-center_82_line/op-amp-differentiator-circuit-1.jpg.webp)

**Modello**

1- time domain  
KCL at node -  
$I_c+I_R=0$  
$C V'_{in} + \frac{V_{out}}R = 0$    
$V_{out}= -V_{in}'RC = -\tau V_{in}'$  

2- freq domain  
è amplifier invertente, con:  
$G= -\frac{Z_R}{Z_C} =-\frac R{1/j \omega c} = -j \omega \tau$  

**Tipi di modello**

1- freq domain  
$V_{out} = - j \omega \tau V_{in}$  
è un derivatore (in frequency domain, moltiplicare per $-j \omega$ significa derivare)  

2- time domain  
$V_{out} = -\tau \frac {dV_{in}}{dt}$  

**Bode plot**

il bode plot di un derivatore è una linea diagonale, che parte da 0 e va verso in alto a destra  
ma deve essere reale, non può arrivare a |G| = $\infty$, dopo un po' deve riscendere  
quindi il derivatore funziona solo a freq basse -> RILEGGI LE NOTE, PERCHÈ SOLO A FREQ BASSE??? 

### Integrator amplifier

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/_1200x630_crop_center-center_82_none/op-amp-integrator-tutorial-thumbnail.jpeg?mtime=1715165858)

**Formulas**

kCL at node -  
$I_R+I_C=0$  
$\frac{V_{in}}{R} + \frac{dQ}{dt}=0$    
$\frac{V_{in}}{R} + C\frac{dV_{out}}{dt}=0$    
$V_{out} = -\frac{1}{RC} \int_0^t V_{in}\, dt$  



### Log ed esponenziale

**Modello: esponenziale**

nota: $W_1 = V_{in,1}$  

$I_d+I_R = 0$  
$I_s e^{-W_1/\eta V_T}+ \frac{V_{out}}R=0$  
$V_{out} = RI_s e^{-W_1/\eta V_T}$  

$RI_s$: voltaggio scala da amplificatore esponenziale  

**Modello: log**

scambia R e diodo  
$I_d+I_R = 0$  
$I_s e^{-V_{out}_1/\eta V_T}+ \frac{V_{d}}R=0$  
$V_{out} = -\eta V_T \log (\frac{W_1}{RI_s})$  

$\eta V_T$: voltaggio scala  

**Applicazione**

circuito a stadi:  
1- trasformo in log  
2- sommatore  
3- esponenziale  

è un moltiplicatore  

### Convertitori d'impedenza

**Utilizzi**

1-  convertire una corrente in un voltaggio   
perchè misurare corrente è difficile, ma voltaggio è facile  

2- controllare una corrente con un voltaggio  
cioè come un transistor, aumentare di molto una corrente  
per questo si chiama "amplificatore a trans-impedenza"  

**Modello 1: convertitore corrente–tensione**

KCl at V-    
$I_d+I_R=0$  
$I_d+ \frac{V_{out}}R = 0$  

**Modello 2: convertitore tensione-corrente**


KCl at V-    
$I_d+I_R=0$  
$I_d+ \frac{V_{in}}R = 0$  

### Giratore (induttanza simulata)

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcQVvhEwFE3Si9xIoqNVXiyOb4pHtdZNsy9ny6hQBlY6pA&s=10)

**Modello**

V+ is middle node of voltage divider (with C and R)   
$V_+ = \frac R{R+1/j \omega C}V_{in} = \frac{j \omega \tau}{1+j \omega \tau}$ 

R2 is betwee Vin and V+, bc the op-amp is a buffer, so $V_{out}=V_2$  
$I_2 = \frac{V_{in-V+}}{R_2} = \frac{V_{in}}{R(1+j\omega \tau)}$  

$\frac{V_{in}}{I_2} = R_2+j \omega R_2RC = R_2+ j \omega L$  
è come un0induttanza simulata  

**Utilizzo**

per avere un circuito con L, senza usare induttanze reali  

problemi delle induttanze reali:  
1- ingrombranti  
2- costose  
3- sporche (si accoppiano tra loro, hanno res interna)

### Filtro risonante

**Cosa fa**

è un circuito risonante (senza bsìisogno di usare induttanze)  

nota: un cicuito risonante è, in pratica, un passa-banda con una banda strettissima, un solo punto  

**Modello**

B: node between R_in e i due C    
R_in: after Vin  
R: res between V- and Vout   

KCL su V-  
$V_B j \omega C+ \frac {V_{out}}R = 0$  
$V_{out}=-j \omega RC V_B$  

KCL su B  
${V_B-V_{in}}{R_{in}} +(V_B-V_{out}) j \omega C+(V_B-0) j \omega C=0$  
$V_{in} = [2V_b j \omega C-V_{out}j \omega C]R_{in}+V_B = V_B[2 j \omega CR_{in}-\omega^2C^2RR_{in}+1]$  

gain  
$G = \frac{V_{out}}{V_{in}} = \frac{-j \omega/(R_{in}C)}{1/(R R_{in}C^2)-\omega^2 + j \omega (2/RC)}$  

**Funzione di fit**

$G(j \omega) = A \frac {j \omega \omega_0}{\omega_0^2-\omega^2+j \omega \gamma}$  
$\omega_0 = \frac 1{C \sqrt{RR_{in}}} \quad \gamma = \frac 2{RC} \quad A=-\sqrt{\frac R{R_{in}}}$  
$\omega_0$: freq di risonanza  
$\gamma$: damping  

**Note**

gain a risonanza: $|G(\omega_0)| = |A \frac {\omega_0} \gamma| = |-\frac R{2 R_{in}}|$  
bandwidth: $\Delta \omega = \omega_h-\omega_L = \gamma$ (high-low, distanza tra freq di taglio)     
quality factor: $Q = \frac{\omega_0}{\Delta \omega}$  

cose note di risonanza:  
$\omega_L-\omega_H= \gamma \quad \sqrt{\omega_L \omega_H} = \omega_0$ 

in freq:  
$f_0 =\frac{\omega_0}{2 \pi}$  
$\Delta f = \frac {\Delta \omega }{2 \pi} = \frac \gamma {2 \pi}$  
$Q = \frac{f_0}{\Delta f}$
