---|---|---

### Modello ideale  

**Properties: ideal**

$A \rightarrow \infty$  
$R_{out}=0$  
$R_{in}= \infty$ (cioè $i_+=i_=0$)  

**Conseguenze**

$G=\frac A{1+\beta A} \sim \frac 1 \beta$   
il $\sim$ vale solo per gli ideali, quando $A \rightarrow \infty$     
così il guadagno non dipende più da A, ma solo da $\beta$ (la retroazione)   

#### Regole d'oro

**Golden rules**

number | 1 | 2 
---|---|---
name | virtual short | virtal open 
regards | voltage | current 
formula | $V_+=V_-$ | $i_+=i_-=0$  
explanation | the op-amp tries to make them the same, using negative feedback  | V+ and V- draw no current, have $R_{in} = \infty$  

**Quando valgono**

solo se c'è retroazione NEGATIVA   
perchè: feedback negativo sopprime la differenza fra i due ingressi V+ e V- -> limita il voltaggio di uscita  
nel limite con A infinito, V+ = V-  

### Op-amp reale

**Limitazioni non-lineari**

1- saturazione: $V_{s-} < V_{out} < V_{s+}$  
se lo supera, c'è clipping del segnale, cioè viene tagliato a V+ o Vs-  
se è di tipo 'rail-to-rail', si avvicina molto alle tensioni di supply  

2- A dipende dalla frequenza  
$A(\omega) = |A(\omega)|e^{j \phi(\omega)}$  
come: se f aumenta -> |A| diminuisce, dephasing aumenta  
perchè: per costruire A, ci sono C interni, che dipendono dalla frequenza  

3- la SLEW RATE, velocità di variazione del voltaggio $dV_{out}/dt$, ha un limite   
quindi non posso riprodurre linee verticali, come quelle di un'onda quadra, ma vengono un po' diagonali    
questo è un limite non-lineare, ha un impatto sulla risposta in freq  

#### Meno importanti

**Rin, Rout**

non sono davvero infinita e nulla  
ma, la retroazione le estremizza nel modo che fa comodo a noi  
$R_{in}$: una res DENTRO il triangolo, tra V+ e V-, su cui scorre $i_{in}$  
$R_{out}$: una res DENTRO il triangolo, attaccata a $V_{out}$  

$i_{in} = \frac{V_+-V_-}{R_{in}} = \frac{V_{out}}A \frac 1{R_{in}} = \frac{GV_{in}}{A R_{in}} \sim \frac{V_{in}}{\beta AR_{in}}$   
$\frac{V_{in}}{i_{in}} = R_{in}(1 + \beta A) \sim R_{in} \beta A$  
cioè, la res in entrata è in realtà maggiore del componente $R_{in}$ (SE c'è retroazione)  

$\frac{V_{out}}{i_{out}} = \frac{R_{out}}{1+\beta} \sim \frac{R_{out}}{\beta A}$  
e la res in uscita è minore di quella del componente che c'è davvero  

**Voltaggio di offset**

$V_{out}=A(V_+-V_-+V_{off})$  

**Corrente di bias**

$i_+=i_- = i_b \neq 0$  
è fissa, non dipende dai voltaggi V+, V-  

**Uscita**

$i_{out}$ ha un limite, al massimo poche decine di mA

### Compensazione in frequenza

**Recap**

1- situazione: A dipende dalla freq  
2- problema: op-amp può diventare instabile, se $\phi=\pi$ e $\beta |A| > 1$        
3- soluzione: la 'compensazione in frequenza' fa in modo che non possa accadere
4- conseguenze: devo scegliere tra gain alto o banda larga  

**1. Situazione**

op-amp reale con feedback negativo:  
$G=\frac A{1+\beta A}$  
$A(\omega) = |A(\omega)| e^{j \phi(\omega)}$  

**2. Problema**

quando $\phi=\pi$   
se $|\beta A| < 1$: disturbance is smaller after each trip through the loop -> stable, decaying obscillation  
se $|\beta A| = 1$: substained obscillation  
se $|\beta A| > 1$: obscillations grows -> INSTABLE

so, it's determined by the pole of the closed-loop system   
poles with positive real part produce growing responses 

**3. Soluzione**

cosa voglio:  
voglio impedire che diventi instabile  
quindi devo costruire l'op-amp in modo che, quando $\phi=\pi$, sia $\beta |A(\omega_1)| <1$  

come si fa: compensazione del polo dominante  
tecnica che consiste nell'aggiungere un C nell'op-amp, per farlo comportare come un passa-basso 
così: $A(\omega) = \frac{A_0}{1+j \omega \tau_0}$  
$A_0$: guadagno in DC (a f=0) (dura fino a $f_c$, poi cala, come nel bode plot di un passa-basso)  
così, alla freq di taglio (larghezza di banda) $f_c = \frac{1}{2 \pi \tau_0}$, |A| comincia a calare (come nel bode di un passa-basso)  
e quando arrivo a $\omega_1$, |A| è già calato tanto

**4. Conseguenze**

se il mio op-amp ha |A| alta:  
devo cominciare a farla diminuire subito, a una $\omega$ bassa, sennò non fa in tempo ad arrivare a $\beta|A| <1$ prima di $\omega_1$  

quindi con lo stesso op-amp puoi ottenere:  
1- un guadagno alto su una banda stretta  
2- un guadagno basso su una banda larga  
banda: l'insieme delle freq su cui funziona bene, cioè prima che A cominci a scendere  

#### Stabilità

**Recap: stabilità**

il problema è $\phi = \pi$ e $A >1$   
mi chiedo: quando si verifica una condizione, quanto manca al verificarsi dell'altra?  
se ancora tanto, bene  

**Stabilità di un op-amp**

definisco:  
$\omega_1$: quando $\phi(\omega_1)=\pi$  
$\omega_2$: quando $|A(\omega_2)|=1$  

posso quantificare la stabilità in 2 modi:  

1- gain margin (margine di guadagno)  
$k_m = \frac 1{|A(\omega_1)|}$  
Quando la fase arriva a $\pi$, quanto è sotto 1 il guadagno d’anello?  

2- margine di fase  
$\phi_m = \pi-\phi(\omega_2)$  
Quando il guadagno d’anello arriva a 1, quanti gradi mancano a $\pi$?    

un op-amp è ben smorzato se:  
$k_m > 1 \quad \phi_m > 0$   

**Stabilità**

ho Vin e Vout costante  
un disturbo modifica Vout  
1- Circuito stabile: la retroazione corregge il disturbo e l’uscita torna al valore iniziale  
2- Circuito instabile: la risposta al disturbo cresce e l’uscita si allontana dal valore iniziale  

può succedere che l’uscita inizi a oscillare con ampiezza crescente, anche se l’ingresso è costante  
ampiezza non cresce all’infinito: viene infine limitata dall’alimentazione e dalla saturazione  

perchè: ritardo della retroazione  
la correzione arriva nel momento sbagliato e rinforza la variazione

**Nota**

la condizione $1+\beta A = 0$ describes the ideal boundary of sustained oscillation  

$G=\infty$ don' mean $V_{out} = \infty$, perchè c'è saturazione ($V_{out} <V_{s+}$)    
quindi, se $G=\infty$, il modello lineare non funziona più  

#### Prodotto guadagno-banda

**Bode plot**

ho detto che l'op-amp agisce come un passa-basso   

Bode plot:  
1- inizia con linea orizzontale ad $A_0$   
2- fino a freq di taglio/banda $f_c$  
3- poi scende, con linea obliqua    

**Prodotto guadagno-banda**

definizione:  
prodotto fra guadagno a frequenza nulla e banda (è costante):    
$GBWP = G_0 \; f(\omega) = \frac{A_0}{2 \pi \tau_0} = const$  
(intuitivamente ha senso, perchè se voglio aumentare uno, devo diminuire l'altro)  

a cosa serve:  
è una figura di merito del componente  
più è alto, più posso avere un op-amp con gain alto e con una banda larga  

tip calcolo:  
se è costante, allora è comodo calcoalrlo quando $G_0=1$  
e quindi è uguale alla banda che si ha alla freq dove $G_0=1$  

**Trova GBWP**

scrivo G di un op-amp, che ho fatto con la compensazione in freq:  
$G(\omega) = \frac{A(\omega)}{1+\beta A(\omega)} = \frac{A_0}{1+j \omega \tau_0} \frac 1 { 1+ \frac{A_0}{1+j \omega \tau_0}} = G_0 \frac 1 {1+ \tau}$  
dove:   
$G_0=\frac {A_0}{1+\beta A_0}$  
$\tau = \frac {\tau_0}{1+\beta A_0}$   
quindi, il circuito con rete di retroazione è ancora un passa-basso (come A da solo)  
sia il guadagno sia la costante di tempo vengono divisi per lo stesso fattore $1+\beta A_0$  

ma la freq di taglio (o "larghezza di banda") è $f_t = \frac 1 {2 \pi \tau}$  
quindi questa è moltiplicata di un fattore $1+\beta A_0$    

quindi $G_0 ft = const$  
perchè uno aumenta, e uno diminuisce dello stesso fattore
