**LED vs diode**

cosa | diodo | LED
---|---|---
Main function | control current flow <br> Any light produced is usually tiny and unintended| emit light 
material (semiconductor) | indirect band gap  | direct band gap
example material | silicon (Si),  germanium (Ge) | gallium arsenide (GaAs), gallium phosphide-based alloys, gallium nitride (GaN)  
why material | radiative recombination is inefficient <br> cioè: nella ricombinazione, emettono energia soprattutto TERMICA (phonons)| radiative recombination is efficient <br> cioè: nella ricombinazione, emettono soprattutto EM (fotoni)  
cosa succede all'energia rilasciata nella ricombinazione | convertita in energia EM (luce) | dissipata come calore  

### LED

#### Uso in circuito

**Struttura**

due pins:  
1- anodo (+): lungo  
2- catodo (-): corto  
oppur, il guscio colorato ha una parte piatta, non tonda, dal lato del catodo  

**Attenzione uso**

1- sono polarizzati  
mettilo nel verso giusto  

2- non superare la corrente massima (10-20ms in alimentazione continua)  
sennò fonde per effetto Joule  

#### Modello

**Equazione**

$\begin{cases}
I = I_s (e^{V/V_T}-1) & come \, diodo \\
V_T = k_BT/e \sim 26mV & voltaggio \, termico \\   
I_s \propto e^{-E_g/(k_BT)} \\
\end{cases}$ 

quindi: $I \propto e^{(V-V_g)V_T}$  
quando $B \gg V_T$  

**Nota**

spiegazione:  
$I_S$ dipende dall'oggetto, non nota  
ma assumiamo che si auguale per tutti i led uguali (cioè, della stessa marca)    
perchè: $I_s \propto n_i^2 \propto E_g$

**Come funziona: bande**

1- forward voltage: rende più facile agli elettroni passare dalla banda di conduzione del lato N al lato P   
cioè, passaggio di portatori maggioritari  

2- nella zona di svuotamento: ricombinazione  
cioè: an electron meets a hole  
A hole is a missing electron in the valence band  
The electron drops into that empty state, so the electron–hole pair disappears  

3- The electron loses energy in the process  
In an LED, much of that energy is released as a photon, a particle of light  

**Material**

a photon change a lot the energy of an electron, but not much its momentum  

for a led, if have indirect band gap, the electron recombining (falling into valence band) don't just need a photon, but also a phonon (a vibration of the crystal, who can change its momentum)   
That extra step makes light emission much less likely; more of the energy becomes heat  
so silicon, which has an indirect gap, is a poor led material  

in a photodiode i don't care, an indirect gap is ok too  

recap:  
direct-gap material is especially important for efficient light emission  
while a photodiode can use either type

#### Spettro frequenza

**Energia fotone emesso**

fotone emesso ha energia uguale al band gap   
(più o meno, dipende dalla k dell'elettrone, ma facciamo finta che abbia quella minima)    
$E_\gamma \sim E_g = \frac{hc}{\lambda}$  

$E_\gamma = hf = \hbar \omega = \hbar \frac{2 \pi}T = \frac{\hbar c}{\lambda}$  

T: periodo di oscillazoine (del fotone???)


**Spettro regione visibile**

violetto: 400 nm, 1.7 eV    
rosso: 700 nm, 3 eV    

energie in eV:   
$E[eV] = \frac{1240}{\lambda [nm]}$    


### Fotodiodo

#### Uso in circuito

**Regimi di operazione**

regime | fotovoltaico | fotoconduttivo
---|---|---
come ottenerlo | carico ad ALTA impedenza <br> eg. lascia aperto | carico a BASSA impedenza <br> eg. fai un corto  
cosa fa |le cariche generate si accumulano sui lle estremità del fotodiodo (non riescono a scorrere) <br> si crea una DIFFERENZA DI POTENZIALE | le cariche generate scorrono fuori dal componente <br> si crea una FOTOCORRENTE

**Cose non ideali**

1- capacità spuria (C in parallelo al diodo)  
2- resistenza in serie e di Shunt (in parallelo a V e C)  
3- corrente di buio (offset in $I_L$, generata anche se non c'è luce)  
4- rumore (NEP = noise equivalent power, cioè quale W in ingresso darebbe una rilevazione con rapporto S/N=1)  


#### Modello

**Formula**

$I=I_s(e^{V/V_T}-1)-I_L$  

come diodo ideale, ma traslata da una foto-corrente $I_L$   
sorgente monocromatica di potenza W, freq $\nu$:   
$I_L = \eta(\nu) \frac{W}{h\nu} e$     


**Nomi**

$\eta(\nu)$: efficienza quantica  
quanti fotoni generati (che formano $I_L$) per fotone incidente  
la sua dipendenza da $\lambda$ si chiama "risposta spettrale" del fotodiodo  

$R= \frac{\eta e}{h \nu}$: responsività [A/W]   
(spettrale, perchè ha dipendenza da $\nu$. quindi da $\lambda$)

$\frac{W}{h\nu}$: rate di fotoni generati  


#### Spiegazione bande

**Come funziona: bande**

1- Light (photons) hits the photodiode.
2- If the photon energy is sufficient, it generates electron–hole pairs.  
another way to say: it promotes an electron from valence band to conduction band  
3- The electric field in the depletion region separates these charge carriers.
4- Their movement produces a photocurrent.

photon energy sufficient if:  
$E_\gamma > E_{g}$

**Recap**

1- grandi $\lambda$:  
band gap (troppo grande, fotone non ha abbastanza energia)  

2- piccole $\lambda$:  
1- cause interne  
ricombinazione con difetti del materiale (soprattutto su superficie)  
assorbimenti lotani da SCR   
2- cause interne  
risolte con anti-reflection coating  

**Spiegazione cadute**

1- short wavelength (high photon energy)  
very short-wavelength photons may be absorbed near the semiconductor surface, where some carriers recombine before being collected  
essentially, they recombine before getting separated by the field(which is what creates the current)  

2- high wavelenght (low energy)  
no enough energy to pass the band gap  
so, no pair generation, nothing can produce current  

**Absorption far from SCR (depletion region)**

generating an electron–hole pair is not enough to produce a photocurrent. The electron and hole must also be collected before they recombine  

pair generation happens where the photon is absorbed  

1- in depletion region:  
there is strong field, they get separated  

2- far from junction (near surface of diode)  
no field  
The minority carrier must move toward the depletion region mainly by diffusion.  
During this movement, it might recombine with another carrier.   
If recombination occurs before the carrier reaches the depletion region, it will not contribute to the collected photocurrent.
