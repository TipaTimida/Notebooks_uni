**What to do**

1- diode clipping   
limits (cuts) part of a waveform above or below a chosen voltage level    
so do a max() or min() between v_in and a specific level  

2- diode clamping  
shifts the entire waveform upward or downward, without ideally changing its peak-to-peak amplitude  

### Clipping

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRZ_PAATbI0Dsi3cGe57nCkntrVLlFkdz65hB1HZDy7Zg&s=10)

**How it works: postive clipping diode**

1- if v_out > 0.7  
the diode is conducting  
a conducting diode is a wire with a voltage drop $\Delta V = 0.7$  
so the point where i take v_out is kept at 0.7V from GND    

2- if v_out < 0.7  
the diode is not conducting  
so the circuit is open, no current passes, so no drop over R  
the point where v_out is taken is at the same voltage as v_in    

$v_{out} = \begin{cases}
0.7 & v_{out} > 0.7 \\
v_{in} & v_{out} < 0.7
\end{cases}$

**Different circuits**

v_out is always taken across the diode  

positive clipping (2 versions):  
VCC -> R -> (+-) diode -> GND    
VCC -> (-+) diode -> R -> GND    


negative clipping:  
VCC -> (+-) diode -> R -> GND    

clipping both half cyles:  
VCC -> R -> [(+-) diode || (-+) diode]  -> GND    

### Clamping

![](https://www.tutorialspoint.com/electronic_circuits/images/positive_clamper_circuit.jpg)

changing the direction of the diode creates a negative clamper circuit (trasla la waveform verso il basso)

**How it works**

1- negative half cycle (so when the + is on top of the battery)  
diode is forward-biased, so a shortcircuit    
so R is short-circuited too, drop across C is $V_C = v_in$  
so C charges until the the max value of v_in

2- positive half cycle  (so when the + is on bottom of the battery)  
diode is reverse-biased, so an open circuit    
so current passes through R  
have RC circuit, C discharges on R  
discharge time of capacitor: $\tau = RC$  
if pick a waveform with period T that $T \ll \tau$, C isn't discharged much  
so C behaves a bit like a battery of DC, so add DC to my signal  
bc have:  
v_in -> C -> v_out -> GND    
so drop across the fiode is v_in + the DC from C (which is = v_in, as seen in previous half cycle)

note: works bc the charging time of C (time needed to charge until ???) is much quicker than the decharging time  
$\tau = RC$  
C is always the same  
R for decharging is the R  
R for charging is the internal r of the diode (and of the generator)

### Recap diode

**Biasing**

bias | forward | reverse
---|---|---
when | $\Delta V > 0.7$ | $\Delta V < 0.7$ 
how to draw | a wire, with a drop of $\Delta V = 0.7$ on it <br> or a shortcircuit, consider $\Delta v \sim 0$ | an open wire   
consequence | its two nodes are kept at a $0.7V$ difference | current does not pass through that wire, check if current actually runs in the circuit  

note:  
$\Delta V$ is the voltage drop across the diode     
reverse bias: when $\Delta V$ across it is <0.7  
if working with big voltages, can ignore this 0.7, just consider it forward and verse biased when $\Delta V$ >0 or <0   
so just check which node has higher voltage  

**Charge**

formulas:  
charging: $V_c(t) =V_0 (1-e^{-t/\tau})$ 
discharging: $V_c(t) =V_0 e^{-t/\tau}$  
so $\tau$ is the time to charge from 0% tp 63.2%  
or to discharge from 100% to 36.8%  

charging time:  
1τ: about 63.2% charged  
3τ: about 95%  
5τ: about 99.3%, usually treated as “fully charged”
