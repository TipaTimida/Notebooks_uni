**Cosa**

elementi di un circuito logico che combinano dei segnali in entrata ( input ) fornendo in uscita ( output ) un altro segnale.

**Come funziona**

usa l'algebra boolena  
come input e output considera solo 1 o 0  
fai l'operazione logica sugli input, vedi cosa dà in output  

### Flip flops

**Cosa**

circuiti sequenziali usati come memoria elementare   

è fatto collegando tra loro porte logiche    
ma spesso lo considero un unico componente, lo disegno tipoblack box  

**Nomenclatura**

circuiti logici: sistema digitale che elabora segnali elettrici basandosi sull'algebra di Boole  

circuito sequenziale: l'output non dipende soltanto dallo stato degli input correnti, ma anche dallo stato precedente

circuito combinatorio: output determinato in modo univoco dallo stato corrente degli input  


**Nome**

i primi flip flop furono realizzati in passato usando dei relé meccanici che emettevano un rumore (o doppio rumore) a ogni cambiamento di stato  

**Types**

SR: set reset  
JK  
R: toggle  
D: delay  

#### Flip-flop SR

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/_1024x246_crop_center-center_82_line/555-timer-SR-flip-flop-input-output-test.jpg.webp)

**Internal structure**

![alt text](https://www.andreaminini.org/data/andreamininiorg/flip-flop-funzionamento-1.gif)

**Struttura**

due porte NOR con retroazione  

NOR outputs 1 only if both inputs are 0   
so if even just a 1 as input, outputs 0

$\overline U$ is called 'U bar'    

**Funzionamento**

1- set: S=1, R=0 
$\overline U =0$, perchè uno dei due input è 1  
U must be the opposite of $\overline{U}$, so 0  

2- reset: R=1, S=0  
U=0, set by R  
so $\overline U = 1$ 

3- invalid: S=1, R=1  
invalid, bc both U and $\overline{U}$ would be 0, since both have a 1 as input  
but must be opposites  
 
4- memory: S=0, R=0  
l'output dipende dalle retroazioni, cioè dallo steto precedente del crcuito  
mantiene gli output dello stato precedente  
esempio: se segue un reset avevo $\overline U =1$, quindi la porta sotto riceve un 1 in input, quindi ottengo U=0  

**Intuitivo**

mi interessa l'uscita U  
set: S=1, cioè setto U=1  
reset: R=1, cioè resetto U=0  

se entrambi 0, continua a fare ciò che faceva prima, cioè agisce da memoria  
cioè sta appendo di ricevere comandi  
così devo solo premere S una volta, non tenerlo premuto, per avere U=1

#### Spiegazione tipi

**Come usarli**

considera il tempo discretizzato  
ad ogni $\Delta t$, chiamato clock, il valore degli input può variare (ma non ll'interno del $\Delta t$)  
vedi cosa succede al segnale in uscita  

**Tipi**

due input:  

SR: spiegato sopra  

JK: come SR, ma nel caso 1-1 inverte lo stato logico corrente delle uscire (1 diventa 0 e viceversa)  


un input:  

D: copia sull'uscita il valore ricevuto  

T: se T=0 rimane com'è, se T=1 cambia output

**Latch vs flip-flop**

latch:  
ha un terzo input: E (enable)  
se E=1, funziona come flip-flop  
se E=0, non funziona, è corto-circuitato  

flip-flop:  
ha un terzo input: CK (clock)  
il cambio di stato avviene in base al segnale di clock  
In genere, si verifica sul fronte di salite del clock ossia quando il segnale di clock passa da 0 a 1  
essenzialmente, scandisce il tempo, quando CK passa da 0 a 1 il flip-flop legge gli input (si attiva)
