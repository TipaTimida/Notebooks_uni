### Wiring

**Pieces**

1- voltage divider   
with 3 equal resister of $5k\Omega$  
split VCC in 3 equal parts  

2- comparators   
output 1 (HIGH) if V+ > V-  
else output 0  

3- flip-flop + driving circuit
gets set or reset by the comparators (S=1 sets to Q=1, R=1 resets to Q=0, S=R=0 nothing happens)  
the IC's output is Q, bc take Qbar and then go through a NOT door (drivinf circuit)  
pin RESET resets to Q=0 when it receives a LOW (not a HIGH, unlike pin R)  

4- discharge transistor   
if the base receives 1, it's on, so closes the discharge wire  
if base receives 0, it's off, so discharge wire is open  
in this schematic, it's on (wire close) when Q=0, (so Qbar=1)

**Internal working**

1- HIGH state    
when: receives TRIGGER > 1/3 Vcc   
trigger comparator: S=1    
flip flop: set Q=1  
transistor: off (bc Qbar=0)
discharge wire: open  

2- LOW state  
when: receives THRESHOLD > 2/3 Vcc  
threshold comparator: R=1    
flip flop: reset Q=0  
transistor: on (bc Qbar=1)
discharge wire: close (working)    

opening and closing the discharge wire is what modifies the external circuit  

how to remember steps:  
1- check if the 555 is currently in the HIGH or LOW state  
2- see what the external circuit looks like  
to see if some capacitor is charging/decharging, and so if a comparator can be activated withouthuman input  
3- find the conditions for when a comparator outputs 1  
(trigger or threshold as titles means what happens when such comparator outputs a 1)

### Monostable Multivibrator Configuration

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/555-timer-monostable-operation-schematic-diagram.png)

**What it does**

when close the switch, the output becomes HIGH for a time $T = 1.1 \times RC$  
(sends a pulse that lasts T)  

while sending the pulse, it's unresponsive, ignores other requests to send pulse   

#### Initial conditions

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/_1024x551_crop_center-center_82_line/1368878/555-timer-monostable-capacitor-discharge.png.webp)

**Initial state**

1- switch is open  
2- the flip flop is set to Q=0, by sending a 0 through RESET
(which is different from pin R, that is inside the 555)  

consequences:  
1- $\overline Q = 1$, so the discharge transistor is on, so $C_1$ can discharge to GND (it's the path wih less resistance)  
2- the upper comparator gives 0, bc node + receives from C, which is connected to GND, while node - from the voltage divider, which is >0  
3- the lower comparator gives 0, bc node + is connected to the voltage divider, so lower than VCC, and node - is connected to the VCC itself (bc switch is close, so no current in that wire, no drop on R2)  
4- the flip-flop receives both 0, so acts like a memory, keeps curent state Q=0  

#### Operating state

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/_1024x551_crop_center-center_82_line/1368879/555-timer-monostable-triggered.png.webp)

**Steps**

1- trigger  
when: close the switch (so TRIGGER=0, bc connected to GND), so S=1    
555 state: HIGH   
outside circuit: C1 is in RC circuit, charging of a capacitor (no longer connected to GND through the discharge wire)  

2- threshold  
when: C1, who is charging, reaches 2/3VCC  
555 state: LOW  
outside circuit: C1 is again connected to GND thught the discharge wire, so it discharges to 0    

**Formula**

T: time needed for C1 to go from 0 to 2/3 VCC  
so time until the threshold comparator outputs R=1  


RC circuit charging:  
$V_c(t) = V_{cc}(1-e^{-t/RC})$  

time to charge to 2/3Vcc:  
$\frac 23 V_{cc} = V_{cc}(1-e^{-t/RC})$   
$-\frac{t}{RC}= \ln (1/3)$  
$T = RC \ln 3 \sim 1.1 \, RC$

### Astable multiivibrator

**What it does**

no stable state  
it continuously oscillates between the two states HIGH and LOW  
don't need any external triggering, it obscillates by itself

**Stetps**

0- initial state  
use RESET to gave OUT = 0  
C1 uncharged  
TRIGGER and DISCHARGE wires always are at same V, so only one comparator can output 1 at a given instant (but both can output 0 at same time)  

1- trigger  
when: at start, bc TRIGGER=0  
555 state: HIGH  
outside circuit: C1 is in RC circuit (no more discharge wire), start charging    
VCC -> R1 -> R2 -> C1 -> GND    

2- threshold  
when: while charging, C1 reaches 2/3 VCC, so R=1  
555 state: LOW  
outside circuit: C1 is in RC circuit, start discharging  
VCC -> R2 -> C1 -> GND   

3- trigger  
when: hile charging, C1 reaches 1/3 VCC, so S=1  
555 state: HIGH  
outside circuit: like in previous high state, C restarts charging  

4- looping  
continue like this, C charging and decharging between 1/3 and 2/3 VCC  
the waveform is pinna di squalo, that of a RC circuit charging  
also notice that Tc > Td

![alt text](https://dwma4bz18k1bd.cloudfront.net/tutorials/_1024x467_crop_center-center_82_line/1000811/astable-multivibrator-555-timer-output-waveform.jpg.webp)

**Time Tc**

Tc: time to charge from 1/3 VCC to 2/3 VCC    
Td: time to decharge from 2/3 VCC to 1/3 VCC   

find formula in same way explained in monostate:  
$T_c = RC \ln 2 \sim 0.693 \, (R_1+R_2)C$  
$T_d = RC \ln 3/2 \sim 0.693 \, R_2C$      
the charge is thourgh R1+R2, while discharge just thourgh R2  
so dischargig is faster
