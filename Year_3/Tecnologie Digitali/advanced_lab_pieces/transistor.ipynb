### Definizione

**Definizione**

componenti elettronici che permettono di controllare una corrente (grande) attraverso un segnale elettrico (piccolo)  

**Famiglie**

| | BJT | FET |
|---|---|---|
| Nome completo | *Bipolar Junction Transistor* | *Field-Effect Transistor* |
| Tre terminali | Base, Collettore, Emettitore | Gate, Drain, Source |
| Idea di funzionamento | Una piccola CORRENTE nella base controlla una corrente maggiore tra collettore ed emettitore | La TENSIONE tra gate e source crea un campo elettrico che controlla la corrente tra drain e source |

cioè cambia se a controllare è la corrente o la tensione  

**Pros**

1- FET  
la tensione $V_{GS}$ modifica, tramite un campo elettrico, la facilità con cui la corrente attraversa il percorso tra source e drain, chiamato canale  
assorbe poca corrente -> ha $R_{in}$ alta  
tipo comune: MOSFET, JFET
