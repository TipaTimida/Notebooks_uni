### Serie vs transform

**Continuous signal: Fourier series vs transform**


What | Series | Transform
---|---|---
applied to | PERIODIC signal <br> (or limited, con estensione implicita) | TIME-DEPENDENT signal
can be written as | SOMMATORIA (composizione lineare delle componenti) | INTEGRALE  
what are the components | DISCRETE frequencies (harminonics) <br> in real field have only positive $\omega_k$, in imaginary field also negative <br> $f_ = \frac nT$ | CONTINUOUS frequencies   
what gives me | coeffs $c_n$ for each freq | a continuous func X(f)  
what i plot (in freq plot)|a series of spikes | a continuous function

**How to write it**

what | serie | transform
---|---|---
from time to freq | func: $g(t)=\sum_{k=-\infty}^\infty \hat g_k e^{j \omega_kt}$ <br> coeffs: $\hat g_k = \frac 1T \int_0^T g(t) e^{-j \omega_k t}dt \quad \omega_k = \frac {2 \pi}T k$ |  $g̃(f) = F\{g(t)\} = \int_{-\infty}^{\infty} g(t') e^{j 2 \pi f t'} dt'$ 
from freq to time | just plug back the coeficients in the series above, and do the summation | $g(t) = F^{-1}\{\tilde g(t)\} = \int g̃(f) e^{j 2 \pi ft} df$  <br> antitrasformata (IFT) 

the trasformata is indicated by a hat or tilde  

**How to write them: explanation**

to go from periodic signal to any time-dependant signal, means i'm doing: $T \rightarrow \infty$  

consequences:  
1- not have harmonic any more, drop k  
2- can't have reference to T anymore, integrate on integrate on $[-\infty, \infty]$ instead of $[0,T]$  

#### Formula trasformata

**Vars to swap?**


FT can be defined for any pair of “conjugated” quantities (that give 1 when multiplied with each other)     

$fT=1$: so i can  
$\omega T = 2 \pi$: so i need to use normalization factors  

**Example**

$t-f: \quad g(t) = \int g̃(f) e^{-j 2 \pi ft} df$  
$t-\omega: \quad g(t) = \frac 1{2\pi} \int g̃(\omega) e^{-j \omega t} d\omega$  

differences:  
1- change what is esplicitato in the exponent of e  
2- appear the normalization factor:  
$f = \frac{\omega}{2 \pi}$  
sometimes, instead of $\frac{1}{2 \pi}$, could put $\frac{1}{\sqrt{2 \pi}}$, to make the transform unitary

### Trasformata discreta

**Parametri da scegliere**

nome | lunghezza dell'acquisizione | campionamento  
---|---|---
def 1 | T: durata totale dell'acquisizione in secondi | $f_s$: frequenza di campionamento (punti al secondo)  
def 2 | N: numero di campioni acquisiti | $\Delta t $: intervallo temporale tra due campioni   
rapporto defs | $T = N \Delta t$ | $f_s = \frac 1{\Delta t}$     
cosa determina | risoluzione in frequenza / risoluzione spettrale (quanto precisamente distinguiamo le frequenze) <br> quindi, i tick sull'asse x | la massima freq che possiamo osservare senza aliasing
cosa determina: formula | $\Delta f = \frac{f_s}N = \frac 1T$ <br> $f_m = m \, \Delta f \quad m =0,1, ...$  | $f_{max} = \frac{f_s}2$  
cosa determina: def | $\Delta f$: risoluzione della trasformata in freq (larghezza dei bin della trasformata) | $f_{max}$: massima freq rilevabile senza aliasing
cosa fa se aumento il param | distinguo freq più vicine | vedo freq più alte  

**Recap: cosa ottengo**

quante spike: N, come il nuemro dei campioni  
distanza tra le spike: $\Delta f = \frac 1 T$, dipende da quanto a lungo misuro (NON dal numero di campioni)

##### Leakage

**Explanation: leakage**

to use the transform: i assume my recorded signal repeats forever (assume it to be periodic)  
so if it records exactly 5 periods, ok  
but if it records 5.5 period, that's bad  

why it's bad:  
when tries to extedn it to be periodic, i get a weird signal, with a jump at the join  
fitting that jump is hard, must use various frequencies, whose amplitude gets wrongly raised  
it'w wrong bc this is not the signal i actully wanted to fit, there should not be that jump  

**Example: leakage**

params:  
T = 0.09 s  
fs = 1000  
f= 50 Hz (signal i want to fit)    

problem:  
1- in time T, it makes 50/0.09 = 4.5 cycles    
2- the freq resolution is: $\Delta f = \frac {f_s} N = 11.11 Hz$   
3- have bins for 44.44 Hz and 55.56 Hz, but not for 50 Hz   
so it gets sprad in both bins (and a bit in the further ones too)  

solution:  
record for T = 0.10 s    
in that time, it completes 5 cycles   
freq resolutoin is $\Delta f = 10 Hz$, so i have a bin at 50 Hz  
so the signal shows a single spike  

recap: to choosing T and fs , must have f of the signal among the set of $f_m$  
so signal must make integer number of cycles  
either compute it, or check the picture on the oscilloscope  

##### Aliasing

**Explanation: aliasing**

massima freq rappresentabile: $f_{max} = \frac {f_s}2$   
se una componente ha f maggiore, viene riflessa, cioè vedo un riga sbagliata, il suo alias   
$f_{alias} = | f_{real}-f_s |$  

**Esempio**

dati:  
$f_s = 7.5 \, kHz$  
f= 4 kHz  
la massima freq che posso vedere è 3.75    
vedo la sua riflessione: $f_{alias} = | f_{real}-f_s | = | 4-7.5| = 3.5 \, kHz$  

**Teorema di Nyquist**

per rappresentare correttamente una componente di freq f, mi serve campionamento almeno:  
$f_S > 2f$  
quindi, la freq maggiore che posso conoscere correttamente è: $f_{max} = \frac {f_s}2$  

$ \frac {f_s}2$ : frequenza di Nyquist  

**Graficamente**

ho una sinusoide a freq alta, che oscilla molto   
ma prendo solo pochi punti  
quei punti si potrebbero unire con una sinusoide a freq più bassa, che oscilla di meno  
erroneamente, penso che la sinusoide a freq bassa sia il fit giusto  
quindi, nel mio spettro, mi trovo una componente a freq più bassa di quella corretta    

![alt text](https://atx7006.salland.com/img/fft/sampled_signal_aliasing.jpg)

**Proof**

ipotesi:  
$f_s = 10 \, Hz$ (freq di campionamento)    
$f= 9\, Hz$ (freq del segnale)  

prendo i campioni:  
$x_n = \sin(\omega t_n)= \sin(2 \pi f \, \frac{n}{f_s}) = \sin (2 \pi \frac 9{10}n)$  

osserva che, poichè 9=10-1:  
$\sin (2 \pi \frac 9{10}n) = \sin(2 \pi n-2 \pi \frac 1{10}n) = -\sin (2 \pi \frac 1 {10}n)$  
cioè, nei punti campionati, una sinusoide da 9Hz produce gli stessi valori (a meno di una fase o di un segno) di una sinusoide da 1Hz  
quindi invece della componente da 9Hz, vedo l'alias da 1Hz  

#### Modello, proofs

**Formula**

$t_m = m \, \Delta t \quad (m = 0, 1, ... N-1)$  
$\hat f_k = \Sigma _{m=0}^{N-1} f(t_m) e^{-j \omega_k t_m}$  
$f(t_m)=f_m$, per scrivere veloce    
cioè tutte le t diventano $t_m$, le uniche che ho  
e le armoniche non spannano tutto R, ma sono solo gli indici dei punti misurati  
NOTA IMPOERANTEs: $\omega_k t_m = 2 \pi \frac {km}T$  

antitrasformata:  
$f_m = \frac 1N \Sigma _{m=0}^{N-1} f(t_m) e^{-j \omega_k t_m}$  

METILLA INSIEME ALLA TRASFORMATA NORMALE E ALLA SERIE

**Proof indici periodici**

i coeff $\hat f_k$ non sono infiniti come nella serie continua  
dopo un po', sono identici  
$\hat f_{k+N} = \sum_{m=0}^{N-1} f_m e^{-2j \pi(k+N)m/N} = \sum_{m=0}^{N-1} f_m e^{-2j \pi km/N} =\hat f_k$  
se ho N campionamenti, ottengo N ampiezze  diverse  

le altre sono uguali, facendo $mod(f_s)$ sul valore della freq, o mod(N) sull'indice della freq 

conseguenze:  
1- alias: una freq $f_{k+N}=f_k+f_s$ apparirà con l'alias $f_k$   
cioè è tutto modulo $f_s$  
per vedere se sono alias, cambia la freq di campionamento e vedi se si spostano  
2- posso usare le freq in un range centrato in 0  

**What freqs in spectrum**

1- se uso la DFT: $\hat f_{k+N}=\hat f_k$   
cioè: indici periodici  
quindi: di solito scelgo di centrare le freq utilizzate attorno a 0, così sono tutte le range  $[-f_s/2, f_s/2]$, invece di $[0, f_s]$  
cioè posso cambiare le freq che uso con $mod(f_s)$,  

2- se il segnale è reale (in lab, sempre): $\hat f_{-k}= \hat f_k^*$   
cioè: i coeff a freq opposta sono complessi coniugati (stesso modulo)   
quindi:  l’informazione delle frequenze negative è ridondante, basta considerare solo le freq positive (le altre non danno altre info), quindi ho range $[0, f_s]$  


**Example: what freqs in spectrum**

$f_s = 1000 \, Hz$  
$N=10$  

1- calcolo le freqs  
$\Delta f = f_s/N = 100 \, Hz$   
$f_m = 0, 100, ..., 900$  

2- uso la DFT: ho indici periodici, quindi li centro a 0  
(wrap arounf, facendo modulo $mod(f_s)$, cioè $mod(100)$)  
$f_m = -400, -300, ... 0,... , 500$  

3- ho segnale reale: uso solo metà positiva  
$f_m= $0, 100, ... 500$  

**Ampiezza delle componenti spettrali**

$\hat f_k$ ha unità strane, non lo plottare su asse y   
invece, plotta l'ampiezza: $A_k = \frac {2|\hat f_k|}N$  
1/N_: normalizzazione dell'antitrasformata   
2: raccoglie i contributi a $\omega_k$  
quindi l'ampiezza zero è_ $A_0 = |\hat f_0|/N$  

per la potenza, uso ampiezza quadratica media:  
$\langle x^2 \rangle_k = \frac {2 |\hat f_k|^2}{N^2}$

### FFT - Fast Fourier Transform

**Problem**

the discrete algorithm (DFT and IDFT) has complexity $O(N^2)$  
cannot use on big datasets  
te FFT is an algorithm with $O(N \log (N))$  

**Tips: sampling**

condition: $N= 2^p$  
means: the basic implementation of the FFT, the one you are strongly advised to use, requires the number of data point N to be a power of 2   
how: if not, either cut the data or add padding  

tip:  
the transformed array will contain $M = N/2 + 1 = 2^{p−1} + 1$ meaningful elements  
if i tell the software by code that my array is real-valued, it automatically discards redundant values
