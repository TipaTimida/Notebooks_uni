### Misuare H($\omega$)

**Metodi**

metodo | CW | IR
---|---|---
nome | continuous wave | impulse response
input (to probe the circuit) |  a cos at a chosen freq  |  a short impulse 
input formula |  $f(t)=\cos (\omega_0 t)$ <br> fasori: $f(t)=e^{j \omega_0 t}$| $f(t)=\delta(t)$   
output | a cos changed in phase and amplitude  | the response in time dmoain (the Green dunction)   
output formula | $r(t) = A \cos (\omega_0 t + \phi)$ <br> fasori: $r(t) = Ae^{j\omega_0 t}e^{j\phi}$ | $G(t)= e^{-t} \theta(t)$  
next step | have $H(\omega_0)$, only at the chosen freq <br> must take many points at different freqs, then fit them, to find $H(\omega)$ | have $G(t)=H(t)$ <br> Fourier-trasform it to freq domain, so have the full $H(\omega)$  
notes | | faster, but harder in practice

f: forzante  
r: risposta  

**Example: CW**

i send: $f(t)=\cos(2t)$  
i measure: $r(t)= \frac 12 \cos(2t-\pi/4)$  
i found ONE value of the trnasfer func: $H(2) = \frac 12 e^{-j \pi/4}$   

must do it for many more points, then fit them  

**Example: IR**

i send $f(t)=\delta(t)$  
i measure: $G(t)= e^{-t} \theta(t)$  

fourier-transform the response:  
rem that $\hat G(\omega) = H(\omega)$ (see above)  
$H(\omega) = \int_{-\infty}^\infty G(t) e^{-j \omega t} \, dt $  
$ = \int_0^t e^{-(1+j\omega)t} \, dt$  
$= \frac{1}{1 + j \omega}$  

i found the whole transfer function  

### Bose plot

$H(\omega)$ is a multiplication of factors  
in bi-log scale, multiplication becomes a sum  
so plot the Bose plot of each factor, and then add the values of the lines at all points
