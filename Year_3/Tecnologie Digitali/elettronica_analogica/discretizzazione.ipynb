### Metodi

**Nomenclatura**

ADC: analog-digital converter  
DAC: digital-analog converter  
DAQ : digital acquisition -> uno strumento per misurare fenomeni fisici che restituisce un segnale digitale  

**Tipi di ADC**

Nome | Come funziona | Pro | Con
---|---|---|---
parallelo / flash| tanti comparatori <br> comparano $V_{in}$ a dei valori noti | veloce (operazioni in parallelo) | non miniaturizzabile, circuito ha tanti pezzi
iterativo / SAR | un solo comparatore <br> procede come ricerca binaria, vede in quale metà è del range, poi dimezza il range e procede | 
pipeline | un mistodi SAR e flash | il migliore <br> buon tradeoff tra velocità e miniaturizzabilità
Sigma-Delta | utilizza la frequenza??? | il più preciso | il più lento
Dual ramp | utilizza il tempo???

### Problemi

**Problemi**

1- potrebbe amplificare o sopprimere le fluttuazioni
dipende da che valore ha il misurando (se dopo la discretizzazione quel valore esiste preciso, o è in mezzo a due opzioni)    

2- l'errore potrebbe essere sistematco o statistico  
dipende dal tipo di dato  

3- non-idealità  
la risoluzione potrebbe non essere uguale in tutto il range dellos trumento (eg. nei multimetri)  
quindi se misuro un segnale che in origine è una retta nel grafico, dopo la quantizzazione sembra un po' un log  

### Dithering

**Cosa è**

aggiunta di rumore controllato  
per mascherare errori di discretizzazione

esempio: se ho un segnale analogico  
ma per leggerlo devo usare uno strumento che discretizza  

**Steps**

1- analog signal  
2- aggiungi del rumore controllato (PRIMA di discretizzare)  
3- discretizza    
4- ripeti i primi 3 step N volte, e fai una media  

**Come funziona**

setup:   
ho un segnale analogico  
lo strumento ha risoluzione 1, quindi arrotonda all'intero più vicino  
come viene letto 2.3?

1- senza rumore: sempre 2  
ma 2 è il risultato che mi dà nel range [1.5, 2.5], non posso sapere di più  

2- con rumore  
aggiungo un rumore piatto $\pm 0.5$  
quindi se ripeto tante volte la misura, trovo una distrib uniforme nel range [1.7, 2.8]  
quindi ottengo 1 il 30%, e 2 il 70% delle volte  
vedendo con che probabilità ottengo le risposte, posso sapere di preciso che numero era inizialmente
