### Rete lineare

**Nomenclatura**

rete: circuito  

lineare: grafico V-I è una retta, cioè ci sono solo componenti ohmici (forzanti, R, C)  
non lineare: appaiono componenti non ohmici (diodi, bobine, transistor)

forzanti: generatori/sorgenti di V o I (sorgenti indipendenti)  


**Porta**

un coppia di nodi, collegati ad una rete      

si caratterizza con:  
1- voltaggio di porta (V tra i nodi)  
2- corrente di porta (I che entra ed esce dai due nodi, cioè I she scorrerebbe in un corto che mettessi tra di loro)  

**Res equivalente**

cosa è:  
quantifica come una variazione di corrente i modifica il voltaggio di porta v  

formula:  
generale: $R_{eq} = \frac {dv}{di}$  
per reti lineari: $R_{eq} = \frac vi$  

nomi:  
Resistenza equivalente / di ingresso / di uscita  
il nome dipende dal ruolo della porta

#### Thevenin e Norton

**Thevenin e Norton**

una rete lineare si può rappresentare come:  

1- Thevenin  
generatore ideale $V_{th}$ in SERIE a $R_{th}$  

2- Norton  
generatore ideale $I_{N}$ in PARALLELO a $R_{th}$  

**Calcolare cose**

simbolo | cosa | come si calcola
---|---|---
$V_{th}$ | voltaggio di porta |  open-circuit voltage <br> cioè stacca il load, calcola $\Delta V$ tra i nodi della porta  
$I_n$ | corrente di porta | shorcircuit il load (metti uno short tra i nodi) <br> calcola la corrente che scorre in questo short (quindi tra i nodi)  
$R_{th}=R_n$ | resistenza equivalente |  spegni le forzanti, poi calcola la res tra i nodi

tip:  
posso calcolarne 2 a scelta, e po la terza con $V_{th}=R_{th}I_n$    
quindi calcola le due più facili  
I potrebbe essere più facile di V, perchè la corrente passa tutta nei rami senza res    


**Spegni le forzanti**

V->corto, I->aperto  
così lo steso tratto ha V=0 e I=0

### Retta di carico

**Definizione**

cosa è:  
la posso tracciare per ogni porta  
descrive la relazione che lega la corrente e voltaggio della porta  

a cosa serve:  
determinare facilmente voltaggio e corrente effettivi in presenza di un load connesso alla porta  

**Come disegnare**

prima, rappresenta la porta come Thevenin   

la retta deve passare per i punti (V,I):   
1- ($V_{th}$, 0)  
cioè se non c'è un carico tra i nodi, quindi circuito aperto    
2- (0, $V_{th}/R_{th}$)  
cioè se c'è un cortocircuito tra i due nodi (non piò esserci voltage drop su un cortocitcuito, tutot il drop deve essere su $R_th$)

poichè siamo in rete lineare, traccia la retta tra i due punti

**Usarla**

nello stesso grafico, disegna:  
1- retta di carico  
2- retta caratteristica I(V) del componente  

grafico ha V su asse x, I su asse y  

### Reti non-lineari

**Studio circuito non lineare**

studialo due volte:  
1- in DC, per trovare il working point  
2- in AC, nel regime di piccole oscillazione, per trovare il gain  

sostituisci ogni pezzo con:  

pezzo | DC (continua) | AC (piccole oscillazioni)
---|---|---
batteria DC | normale | GND 
batteria AC | aperto <br> (la ignoro, non diventa gnd) | normale
C | aperto | $Z_C = \frac 1{j \omega C}$ <br> se f è alta, un corto
bobina | corto | $Z_L = j \omega L$ <br> se f è alta, un aperto 
giunzione BE transistor / diodo | $\Delta V_{BE}=0.7$ <br> fissa, come diodo | $v_{be} = i_b r_\pi$  <br> ci sono altri modi

#### Transistor

**Diodo - per piccoli segnali**

simbolo | cosa | formula per modellare | stima
---|---|---|---
$r_\pi$ | resistenza dinamica vista dalla base |$v_{BE}= r_\pi i_b$  |  $r_\pi = \frac{V_T}{I_{B,Q}}$ 
$r_e$ | resistenza dinamica vista dall’emettitore |$v_{be}=e_ei_e$ | $r_e = \frac{V_T}{I_{E,Q}}$ 
$g_m$ | transconduttanza (rapporto tra variazione di $v_{BE}$ e di $i_C$) | $i_c = g_m v_{be}$  |  $g_m = \frac{I_C,Q}{V_T}$

cioè la resistenza dinamica di BE la posso scrivere o in funzione di i_b, o di i_e, o di i_c  

$\eta \sim 1$  
$V_T \sim 23mV$ a temp ambiente  

**Rapporti tra correnti**

$\begin{cases}
i_c = \beta i_b \\
i_E = (\beta +1)i_b \\ 
\end{cases}$

valgono sia per correnti DC (maiuscolo) che AC (minuscolo)  
scrivi tutto in funzione di $i_B$, èerchè la corente di base è quella che controlla il BJT  

**Studia**

1- scrivi legge delle maglie su:  
ramo della corrente di base (parte da $V_{cc}$ o $v_{in}$)  
ramo del nodo che mi interessa ($V_C$ per working point, $v_{out}$ per oscillazioni)  

2- scrivi tutte le correnti in funzione di quella di base  
perchè il BJT è controllato dalla corrente di base   
posso trovare $I_B$ dalla prima equazione, e poi pluggarla nella seconda  
oppure, nella seconda lascia $i_C$, e poi nel gain trovo $i_c/i_b = \beta$  

**Regione**

la regione attiva richiede:  
$V_B> V_E$ (BE polaizzata diretta)   
$V_C > V_B$ (BC polarizzata inversa)   


**Working point**

di solito si intende:  
$Q=(V_{CE,Q}, I_{C,Q})$  
V è il drop tra C ed E  
se dopo E c'è subito il GND, sarebbe quindi V_C      

ma tutti i nodi hanno una V e I di working point  

### Trasferimento di potenza

adattamento di impedenza:  
potenza media assorbita dal carico è massima quando: $R_L = R_{out}$  

**Perchè massimizzare il trasferimento di potenza**

cosa | massima potenza sul carico | massima efficienza energetica
---|---|---
condizione | $R_{load} = R_{out}$  | $R_{load} \gg R_{out}$    
quando lo voglio | la sorgente ha una resistenza d’uscita (out) che non puoi ridurre, e vuoi ottenere dal carico la massima potenza possibile | la sorgente deve alimentare realmente un utilizzatore e vuoi limitare: energia sprecata, consumo di batteria
esempio uso | antenne, lineee di trasmissione | rete elettrica domestica, alimentatori, motori, batterie, circuiti digitali
che succede | il carico ha la prestazione migliore (a parità di alimentazione Vcc) <br> ma c'è molto spreco di energia | il carico lavora col minor spreco di energia possibile <br> ma la prestazione non è la migliore ottenibile (eg. lampadina non raggiunge luminosità massima)
