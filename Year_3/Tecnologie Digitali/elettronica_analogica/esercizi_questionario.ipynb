$V_{th}-IR_{th}-V_{out}=0$

$I_{out} = \frac{V_{th}-V_{out}}{R_{th}}$

$I_{out}=0$ ->  non esce corrente, quindi è come se il circuito non fosse diturbato  
è semplicemente Vout, la V di quel punto

### Trasferimento di potenza

adattamento di impedenza:  
potenza media assorbita dal carico è massima quando: $R_L = R_{out}$  

**Calcolalo**

v - R_L - R_out - GND  

$v(t) = V \sin(\omega t)$  
$i(t) = \frac{v(t)}{R} = \frac{ V \sin(\omega t)}{R_L+R_{out}}$  
$p_L(t) = i^2(t) R_L = \frac{V^2}{(R_L+R_{out})^2} \sin^2 (\omega t) R_L$  
$\langle P_L \rangle = \frac {V^2} 2 \frac{R_L}{(R_L+R_{out})^2}$  
la media di sin^2 è 1/2  
così la media dipende dal valore efficace del voltaggio, non dall'ampiezza di picco  

massimo:  
impongo: $\frac{dP}{dR_L}= 0$  
mi dà: $R_L = R_{out}$  
con questa condizione, la potenza è: $\langle P_L \rangle = \frac {V^2}{ 8 R_L}$  


### BJT transistor

#### Esempio 1

![](pics/questionario_bjt.png)

**Punto di lavoro**

1- legge della maglie  
$\begin{cases}  
V_{cc} -I_B (R_P+R_B) -V_{BE}  = 0 & percorso \; I_B \\
V_{cc}-I_CR_C = V_{C} & node \, i \, want
\end{cases} $ 

2- trova $I_B$, poi tutto in funzione di $I_B$  
$I_B = \frac{V_{cc}-V_{BE}}{R_P+R_B}$  
$V_C = V_{cc}-I_CR_C = V_{cc} \beta I_B R_C$  

**Gain**

1- legge delle maglie  
$\begin{cases}
v_{in} - (R_B+r_\pi)i_b =0\\ 
0-R_Ci_c = v_{out} \\
\end{cases}$  

2- gain  
$A = \frac{v_{out}}{v_{in}} = \frac{-R_Ci_c}{ (R_B+r_\pi)i_b}= -\frac{R_C}{R_B+r_\pi}\beta$  

nota: ho scritto le stesse eq del working point, ma col sistema scritto per piccole oscillazioni  

#### Esempio 2

![](pics/questionrio_bjt_2.png)

nota: posso ignorare R2, è molto più piccola rispetto a R1???

**Punto di lavoro**

$\begin{cases}
V_{th}-I_BR_{th}-V_{BE}-I_ER_E = 0 \\
V_{cc}-I_CR_C = V_{out} \\
\end{cases}$  

**Gain**

$\begin{cases}
v_{in}- v_{be} -I_ER_E = 0 \\
v_{be} = i_b r_\pi = i_b \frac{V_T}{I_B,Q} \\
0-R_Ci_C = v_{out} \\
\end{cases}$

tutto in funzione di $i_b$:  
$v_{in} -r_\pi i_b -(\beta+1)i_b R_E=0$ 

gain:  
$A = \frac{v_{out}}{v_{in}} = -\frac{R_C}{r_\pi+(\beta+1)R_E}\beta \sim -\frac{R_C}{R_E}$    
dove approssimo facendo il limite per $\beta \rightarrow \infty$  


**Approssimazione di partitore imperturbato**

se fosse imperturbato (cioè il transistor non assorbe corrente, $i_b=0$):  
$V_B = V_{th} = 1.091 V$  

se invece considero la corrente assorbita (cioè il calcolo che ho fatto cercando il working point), ho:  
$V_B = 1.050V$  

differenza relativa è:  
$\frac{1.091-1.050}{1.091} = 3.8%$  
è meel 10%, quindi ok, facciamo finta che sia imperturbato  

**Drift**

1- $I \uparrow$   
non importa quale, tanto sono tutte legate  
2- drop su $R_E \uparrow$  
3- $V_{BE} \downarrow$  
perchè $V_B$ è tenuto fisso dal biasing (dalla corrente di base)  
4- il transistor conduce meno e contrasta l’aumento iniziale di corrente

quindi $R_E$ stabilizza il punto di lavoro rispetto alla temperatura, drifta di meno  

**Graphical explanation**

consider the load line $I_B(V_{BE})$  

$I_B = -\frac{1}{R_B'}V_{BE} +\frac{V_{cc}-R_EI_{Cq}}{R_b'}$   

how to find load line:  
write loop law: $V_{cc}-R_b'I_B-V_{BE}-R_EI_E =0$   
use that: $I_E \sim I_C$ e $R'_B = R_B+r_b$
