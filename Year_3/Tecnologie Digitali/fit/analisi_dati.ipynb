### Incertezze

**Come si scrive un risultato**

1- incertezza ha 1 (max 2) cifre significative  
2- valore centrale e incertezza hanno stesso unmero di cifre significative  
3- ricorda le unità  

**Propagazione errori statistici**

se incertezze indipendenti (no correlazione tra i parametri):  

$y=f(x)$  
$\sigma^2_y = \Sigma_i (\frac{\partial f}{\partial x_i}) \sigma_i^2$

**Precisione e accuratezza**

1- precisione: conrcordanza tra misura ripetute  
(quanto lo strumento è sicuro del risultato)  
dà incertezza STATISTICA -> diminuisce facedo più misurazioni  

2- accuratezza: concordanza tra valore misurato e valore vero del misurando  
(quanto lo strumento è corretto)  
dà incertezza SISTEMATICA -> lo strumento è proprio sbagliato, fare più misure non aiuta  

### Fit

#### Definizione

**Cosa dice**

cosa ho:  
coppie di punti $(x_k,y_k)$    
per ogni coppia, un'incertezza $\sigma_k$   
modello $f(p,x)$  

il valore più probabile dei paramtri $p$ è quello che minimizza:  
$\chi^2(p) = \Sigma \frac{[y_k-f(p,x)]^2}{\sigma^2}$ 

**Nomenclatura**

p: vettore dei parametri  
$\omega_k = \frac 1 {\sigma_k^2}$ : pesi statistici  
$r_k = y_k-f(p,x_k)$ : residui  

**Ipotesi per fare il fit**

1- errore SOLO sulle y (sulle x è trascurabile)  
2- errore sulle y è di tipo STATISTICO e GAUSSIANO  
3- errori dei vari punti spetimentali sono SCORRELATI  

#### Covarianza

**Rapporti tra var**

due var $x_i, x_j$ sono:   

indipendenti: $P(x_i, x_j)=P(x_i)P(x_j)$   
sapere x_i non dà info su x_j  

scorrelate: $\sigma^2_{ij}=0$  
???  

indip -> scorrelate : sempre  
scorrelate -> indip : solo con gaussiana  

**Formule distribuzioni**

valore atteso:  
$E[f(x)] = \int f(x) \, p(x) \, d^nx$  

media:  
$\mu = E[x]$  

matrice di covarianza (si può chiamare C o $\sigma^2$):  
$C_{ij} = E(x_i-\mu(x_j-\mu_j))$  
var diverse: $C_{ij}=cov(x_i,x_j)$    
stessa var (diagonale): $C_{ii}=\sigma^2_i = E[(x_i-\mu_i^2)]$  

**Significato cov**

$cov(x_1, x_2)>0$ : tendono ad aumentare insieme    
se $x_1\ge \mu_1$, generalmente anche $x_2 > \mu_2$  
pensa ai due fattori della formula della cov  

$cov(x_1, x_2)>0$ : se uno aumenta l'altro tende a diminuire
