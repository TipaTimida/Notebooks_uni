### Proofs

#### Minimizzare chi = best params

**Cosa dice**

cosa ho:  
coppie di punti $(x_k,y_k)$    
per ogni coppia, un'incertezza $\sigma_k$   
modello $f(p,x)$ (cioè i parametri p scelti) 

**Nomenclatura**

p: vettore dei parametri  
$\omega_k = \frac 1 {\sigma_k^2}$ : pesi statistici  
$r_k = y_k-f(p,x_k)$ : residui  

**Cosa voglio dimostrare**

ipotesi:  
1- errore SOLO sulle y (sulle x è trascurabile)  
2- errore sulle y è di tipo STATISTICO e GAUSSIANO  
3- errori dei vari punti spetimentali sono SCORRELATI  

tesi:  
il valore più probabile dei paramtri $p$ è quello che minimizza:  
$\chi^2(p) = \Sigma \frac{[y_k-f(p,x)]^2}{\sigma^2}$ 

**Proof: minimizzare chi = best params**

M: il modello (cioè i parametri p scelti) è corretto  
D: ho ottenuto tutte queste coppie di dati (x,y)  

1- se il modello è corretto, prob di ottenere una singola coppia di dati  
uso: ipotesi di errore gaussiano  
$p(r_k) = \frac{1}{\sqrt{2 \pi \sigma^2_k}} exp[-\frac{r_k^2}{2 \sigma_k^2}]$     
$p(x_k, y_k | M) = P_K = \frac{1}{\sqrt{2 \pi \sigma^2_k}} exp[-\frac{[y_k-f(p,x_k)]^2}{2 \sigma_k^2}]$  

2- prob di ottenere tutti i dati D:  
uso: prob di errori indipendenti (mi fa moltiplicare le prob)   
$p(D|M) = \prod P_k \sim \prod exp[-\frac{[y_k-f(p,x_k)]^2}{2 \sigma_k^2}] = exp[-\Sigma \frac{[y_k-f(p,x_k)]^2}{2 \sigma_k^2}] =  e^{-\chi^2/2}$  

3- ribalto col th di Bayes  
$p(M|D) = \frac{p(D|M)p(M)}{p(D)} \sim p(D|M) \sim e^{-\chi^2/2}$  
p(M|D): prob di che il modello sia corretto, sapendo che ho ottenuto questi dati  
p(D|M): prob di ottenere questi dati, se so che il modello è corretto  

4- devo massimizzare p(M|D), la prob che il modello sia corretto  
si fa diminuendo il valore di $\chi$  

#### Rislvi fit analiticamente

**Fit generale: steps**

1- espando $\chi^2(p)$ intorno al punto di best fit  
2- calcolo l'hessiana che appare nell'espansione  
vedo che H si può scrivere inn funzione di J  
3- pluggo l'espansione di $\chi^2(p)$ nellla funzione per la prob p(M|D)  
4- riconosco una gaussiana multivar  

**Fit generale**

1- espando $\chi^2(p)$ intorno al punto di best fit  
$p= p_{bf}+\delta p$  
$ \chi^2(p) \sim  \chi^2(p_{bf}) + [\nabla \chi^2(p_{bf})]^T \, \delta p + \frac 12 (\delta p)^T \, H_{\chi^2} \, \delta p $    
$ = \chi^2(p_{bf})+ \frac 12 (\delta p)^T \, H_{\chi^2} \, \delta p $  
derivata prima nel punto di best-fit nulla perchè è un punto di minimo    
quindi svliluppo al secondo ordine
 
2- calcolo hessiana  
$\chi^2(p) =  \Sigma \frac{[y_k-f(p,x)]^2}{\sigma^2}$  
$\frac{\partial \chi^2}{\partial p_i} = \Sigma_k \frac{-2r_k}{\sigma_k} \frac{\partial f_k}{\partial p_i} = -2 \Sigma_k \frac 1{\sigma_k^2} \, [r_k \, \frac{\partial f_k}{\partial p_i}] $  
$\frac{\partial^2 \chi^2}{\partial p_i p_j} = (-2) \Sigma_k \frac {1}{\sigma_k^2} [(-\frac{\partial f_k}{\partial  p_j}) \frac{\partial f_k}{\partial  p_i} + r_k \frac{\partial^2 f_k}{\partial p_i \partial p_j}] = 2 \Sigma_k \frac{1}{\sigma_k^2} \frac{\partial f_k}{\partial  p_i} \frac{\partial f_k}{\partial  p_j} -2 \Sigma_k \frac{r_k}{\sigma_k^2} \frac{\partial^2 f_k}{\partial  p_ip_j} $   
dove $f_k(p)=f(x_k, p)$  

3- se il fit è buono, i residui sono piccoli, quindi $r_k \sim 0$  
trascuro il secondo temine  
$[H_{\chi^2}]_{ij} \sim 2 ∑_k \frac{1}{\sigma_k^2} \frac{\partial f_k}{\partial  p_i} \frac{\partial f_k}{\partial  p_j}$   
nota che, se è un fit con funzione lineare, allora tutte le derivate seconde sono 0, quindi il secondo termine è proprio 0  
mentre in generale ci tende soltanto, è un'approssimazione  

4- osservazione sul jacobiano pesato   
$J^*_{ki}= \frac{1}{\sigma_k} \frac{\partial f_k(x)}{\partial p_i}$   
$J^*_{ki}J^*_{kj} = \frac{1}{\sigma_k^2} \frac{\partial f_k(x)}{\partial p_i} \frac{\partial f_k(x)}{\partial p_j}$   
$\Sigma_k J^*_{ki}J^*_{kj} = [(J^*)^T \, J^*]_{ij}$  
posso pluggarlo sopra:  
$[H_{\chi^2}]_{ij} \sim [(J^*)^T \, J^*]_{ij}$  

5- ritorno alla formula  
$P(M|D) \sim e^{-\chi^2(p)/2}$  
$ \sim exp[-\frac{\chi^2(p_{bf})}2] \, exp [-\frac 12 (\delta p)^T \, (J^*)^TJ^* \, \delta p]$  
il primo fattore è costante rispetto a p, quindi lo ignoro quando vedo come scala     
$\sim  exp [-\frac 12 (\delta p)^T \, (J^*)^TJ^* \, \delta p]$  

6- gaussiana multivariata  
ricorda che è: $P(p) \sim  exp[-\frac 12 (p - \mu)^T \; C^{-1}(p -\mu)] $     
confrontandola con l'espressione che ho trovato sopra:  
media: $\mu=p_{bf}$ (perchè $\delta p = p- p_{bf}$)   
matrice di covarianza: $C^{-1}=(J^*)^TJ^*$ 

**Esempio: retta**

dati:  
$f(m,q,x) = mx+q$  
$p=(m,q)$     
$\sigma_k=\sigma$

1- jacobiana  
$J= \frac{\partial f}{\partial p_j}|_{x_i} \begin{bmatrix} x_1 & 1 \\ ... & ... \\ x_n & 1 \end{bmatrix}$  
prima col: derivata per m  
seconda col: derivata per q  
$J^* = \frac 1 {\sigma} J$  
grazie alla linearità: $J \, p = \begin{bmatrix} f(p,x_1) \\ ... \\ f(p, x_n) \end{bmatrix}$  

2- chi  
$\chi^2(p) = \frac{|y-Jp|^2}{\sigma^2} = \frac 1 {\sigma^2} (y-Jp)^T(y-Jp)$    
dove $y = (y_1, ...,y_n)$  

3- best fit  
il best fit è quello col chi minore  
quindi devo imporre: $\nabla_p \chi^2=0$  
$ \chi^2 =\frac 1 {\sigma^2} (y-Jp)^T(y-Jp)$  
$=\frac 1 {\sigma^2} (y^T-p^TJ^T)(y-Jp)$  
$=\frac 1 {\sigma^2} (y^Ty-y^TJp-p^TJ^Ty+p^TJ^TJP)$   
$= \frac 1 {\sigma^2} (y^Ty-2p^TJ^Ty+p^TJ^TJp)$  
per fare la derivata, $p^T$ lo considero come una p (e ricorda chain rule)  
$\nabla_p \chi^2 =  \frac 1 {\sigma^2} (0 -2J^Ty + [J^TJp +pJ^TJ]) $  
$= \frac 1 {\sigma^2} (-2J^Ty +2J^TJp)$    
$= -\frac {2} {\sigma^2} J^T(y -Jp)$    
perchè $J^TJ$ è simmetrica e $j^TJp = p^TJ^Ty$, perchè sono scalari, e uno scalare è uguale alla sua trasposta  
risolvo:  
$J^Ty-J^TJp_{bf}=0$  
$J^TJp_{bf}=J^Ty$  
$p_{bf}=(J^TJ)^{-1}J^Ty$  

4- calcolo $(J^TJ)$  
$J^TJ = \begin{bmatrix} x_1 & ... & x_n \\ 1 & ... & 1 \end{bmatrix} \begin{bmatrix} x_1 & 1 \\ ... & ... \\ x_n & 1 \end{bmatrix} = \begin{bmatrix} \Sigma_i x_i^2 & \Sigma_i x_i \\ \Sigma_i x_i & n \end{bmatrix} =  \begin{bmatrix} S_{xx} & S_x \\ S_x & n \end{bmatrix} $  
i nomi S glieli ho dati io    

5- la inverto  
$A^{-1} = \frac 1{ad-bc} \begin{bmatrix} d & -b \\ -c & a\end{bmatrix}$  
$(J^TJ)^{-1} = \frac 1{\Delta} \begin{bmatrix}n & -S_x \\ -S_x & S_{xx} \end{bmatrix}$

6- calcolo l'altro pezzo  
$J^Ty = \begin{bmatrix} \Sigma_i x_iy_i \\ \Sigma_i y_i \end{bmatrix} = \begin{bmatrix} S_{xy} \\ S_y \end{bmatrix}$

7- totale: param best-fit  
$p_{bf} = \frac 1{\Delta} \begin{bmatrix}n & -S_x \\ -S_x & S_{xx} \end{bmatrix}  \begin{bmatrix} S_{xy} \\ S_y \end{bmatrix} $  
ottengo:   
$m_{bf} = \frac{nS_{xy}-S_xS_y}{\Delta}$  
$q_{bf} = \frac{S_{xx}S_y-S_xS_{xy}}{\Delta}$  

8- covarianza  
$C^{-1} = (J^*)^T \, J^* = \frac 1 {sigma^2} J^TJ  = \frac 1 {sigma^2} \begin{bmatrix} S_{xx} & S_x \\ S_x & n \end{bmatrix} $  
$C = \frac{\sigma^2}{\Delta} \begin{bmatrix}n & -S_x \\ -S_x & S_{xx} \end{bmatrix} = \begin{bmatrix} \sigma_m^2 & Cov(m,q) \\ Cov(q,m) & \sigma_q^2 \end{bmatrix}$  

9- correlazione  
$Cov(m,q) = -\frac{\sigma^2}\Delta S_x$  
$\Delta>0$, quindi il segno è l'opposto di quello di $S_x=\Sigma x_i$  
quindi se le $x_i$ sono per lo più positive, Cov<0  
cioè per un punto x>0, se m aumenta (retta più pendente), per mantenere la retta vicino ai dati devo diminuire q  

### Definizioni

#### Generale

**Vettori**

di default, sono vettori colonna  
per trasformarlo in riga, devo fare il trasposto  
ma quando voglio scrivere un vettore al computer, di solito lo scrivo in riga, per fare più veloce  

**Quadrato di un vettore**

$|x| = x \cdot x = x^T x$  
il primo deve essere un vettore riga, per rispettare le regole della moltiplicazione matriciale   

ricorda:  
$(AB)^T = B^TA^T$

#### Funzioni

$f: R^n \rightarrow R^m$  

input:  
univar: n=1  
multivar: n>1  

output:  
scalare: m=1  
vettoriale: m>1    

**Esempio: calare vs vettoriale**

f scalare: 
$f(x,y)=2x+y$  

f vettoriale.  
$f(x,y) = (x+2y, -x+1, y+1)$  


#### Matrici con derivate

**Recap - per funzioni MULTIVAR**

cosa | scrittura | cosa contiene | si applica a
---|---|---|---
gradiente | $\nabla f$ | derivate prime | f scalare
jacobiano | $Jf$ | derivate prime | f vettoriale  
hessiana | $Hf$ | derivate seconde | f scalare

**Gradiente**

$\nabla f = (\frac{\partial f}{\partial x_1},  \frac{\partial f}{\partial x_2})$  

come utilizzare: (prodotto scalare)    
$(\nabla f)^T \; (\delta x) = \frac{\partial f}{\partial x_1} \delta x_1 +  \frac{\partial f}{\partial x_2} \delta x_2$  

in un punto di minimo, il gradiente è nullo  

**Jacobiano**

$J_{ij}= \frac{\partial f_i(x)}{\partial x_j}$  
cioè: come cambia il valore di $f_i$, se modifico il parametro $x_j$  

$J^*_{ij}= \frac{1}{\sigma_i} \frac{\partial f_i(x)}{\partial x_j} =\begin{bmatrix} \frac 1{\sigma_1} \frac{\partial f_1}{\partial x_1} & \frac 1{\sigma_1} \frac{\partial f_1}{\partial x_2} \\ \frac 1{\sigma_2} \frac{\partial f_2}{\partial x_1} & \frac 1{\sigma_2} \frac{\partial f_2}{\partial x_2} \end{bmatrix}$   
jacobiano pesato  
come cambia il valore di $f_i$, se modifico il parametro $x_j$, misurando il cambiamento in unità di $\sigma_i$  
il peso dà maggior peso ai dati misurati con più precisione  


**Hessiana**

$[H_f]_{ij} = \frac{\partial^2 f}{\partial x_i \partial x_j}$  

esempio: $R^2$  
$H_f  =\begin{bmatrix} \frac{\partial^2 f}{\partial x_1^2} & \frac{\partial^2 f}{\partial x_1x_2}  \\  \frac{\partial^2 f}{\partial x_2x_1}& \frac{\partial^2 f}{\partial x_2^2}   \end{bmatrix}$  

come utilizzare:  
$(\delta x)^T \, H_f \, \delta x$  
$=\begin{bmatrix} \delta x_1 & \delta x_2 \end{bmatrix} \begin{bmatrix} H_{11} & H_{12} \\ H_{21} & H_{22}  \end{bmatrix} \begin{bmatrix} \delta x_1 \\ \delta x_2 \end{bmatrix}$  
$ = H_{11} (\delta x_1)^2+ H_{12} \delta x_1 \, \delta x_2 +  H_{21} \delta x_1 \, \delta x_2 +H_{22} (\delta x_2)^2$  
$ = H_{11} (\delta x_1)^2+ 2H_{12} \delta x_1 \, \delta x_2  +H_{22} (\delta x_2)^2$  
cioè vicino al minimo, f sembra una parabola multidimensionale  

**Th di Schwarz**

ipotesi: $f \in C^2(U)$  
cioè f, f', f'' sono continue  

tesi: $f_{xy}=f_{yx}$  
cioè H è simmetrica $H_f=F_f^T$  

**Taylor**

f scalare:  
$g(x+dx) \sim g(x) +g'(x) \, dx + \frac 12 g''(x) (dx)^2$  

f vettoriale:  
$g(x+\delta x) \sim g(x)+\nabla f(x)^T \; \delta x + \frac 12 (\delta x )^T H_f \delta x$  
$\delta x = (\delta x_1, \delta x_2)$  
cioè derivata prima e seconda sono diventati gradiente ed hessiana  
la traposta serve per per le regole del prodotto matriciale  

**Moltiplicazione matrici**

$\Sigma_k J^*_{ki}J^*_{kj} = [(J^*)^T \, J^*]_{ij}$  

dimensione:  
(rows, cols)  
(1xm)(mxn)(nx2) = (1x2)  
i numeri vicino devono essere uguali  
il risultato ha i numeri al bordo come dim  

#### Altro

**Gaussiana multivariata**

f univar:  
$P(x) = \frac 1{\sqrt{2 \pi}\sigma} exp[-\frac 12 (\frac{x-\mu}{\sigma})^2]$  

f multivar:  
$P(\vec x) = \frac 1{\sqrt{(2 \pi)^n \, det \, C}} exp[-\frac 12 (\vec x - \vec \mu)^T \; C^{-1}(\vec x -\vec \mu)] $  
estensione della gaussiana a più dimensioni  
si usa se sto fittando più di un parametro, cioè se p e x sono vettori di dimensione n  

l'esponente dice quanto il punto $\vec x$ è lontano dalla media $\vec \mu$, tenendo conto di incertezze e correlazioni  


**Matrice di covarianza**

$C = \begin{bmatrix} \sigma_1^2 & Cov(x_1,x_2) \\ Cov(x_2,x_1) & \sigma_2^2 \end{bmatrix}$

C: matrice di covarianza  
($C^{-1}$: matrice di precisione)  
a dim=1, diventa: $C=\sigma^2$ e $C^{-1}= 1/\sigma^2$  

**Correlazione**

per quantificare la CORRELAZIONE si usa:  
$Corr(x_i, x_j) = \rho_{ij} = \frac{C_{ij}}{\sqrt{C_{ii}C_{jj}}} = \frac{Cov(x_i, x_j)}{\sigma_i \sigma_j}$  
è come la covarianza, ma normalizzata, così ho un riferimento per capire il suo valore  
$\rho = 0$: scorrelati  
$\rho =1$: se uno aumenta, anche l'altro lo fa  
$\rho=-1$: se uno aumenta, l'altro diminuisce
