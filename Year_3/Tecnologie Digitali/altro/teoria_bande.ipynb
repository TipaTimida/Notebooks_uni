### Orbitali molecolari

**Come si formano**

due orbiatali atomici di atomi vicini si sovrappongono  
diventano due orbitali molecolari, uno legante e uno non-legante (di energia un po' minore e maggiore di quella degli orbitali atomici)

**Continuo di energie**

in un cristallo ci sono tanti atomi vicini  
quindi tanti orbitali si sovrappongono ripetutamente, quindi creo tanti orbtali atomici con piccolissime differenze di energia  
quindi la $\Delta E$ tra gli orbitali diventa piccolissima, possiamo dire che E è quasi un valore continuo  

ma generalmenete si sovrappongono leganti con leganti e non-leganti con non-leganti, quindi la parte in mezzo non viene riempita???

### Struttura a bande

**Vettore d'onda**

k: vettore d'onda    
tells us how rapidly the phase changes in space  

relationship with:  
1- wavelength: $k = \frac{2 \pi}\lambda$    
2- momentum of the particle: $\dot P \psi = -i \hbar \frac {d}{dx} (Ae^{ikx}) = \hbar k \psi$  
$p = \hbar k$   


**Band gap**

1- indiretto:   
se il minimo  della banda di valenza e il massimo della banda di conduzione sono a k diversi  
quindi, per fare il passagio, must supply the required momnetum (p) -> a photon don't give momentum, but a PHONON (quantum of lattice vibration) does  
però così ll'emissione di luce è meno efficiente  
eg. silicio    

2- diretto:  
se il minimo  della banda di valenza e il massimo della banda di conduzione sono allo stesso k   
eg. germanio

$E_g = min_k E_c(k)-max _k E_v(k)$  
il gap è la fascia orizzontale tra il minimo della conducition band e il massimo della valence band  
cioè quanto costa almeno passare dall'una all'altra

**Banda**

Il cristallo determina le possibilità: le funzioni $E_k$, cioè le curve del grafico  
Lo stato dell’elettrone determina quale possibilità è occupata: in quale banda e con quale k si trova  

in ogni banda, l'energia di un elettrone è fissata a seconda del valore di k  
k può essere modificato da urti, vibrazioni, campi magnetici...

### Bloch

**Th di Bloch**

in un cristallo, posso scrivvere le autofunzioni (QM) per ogni elettrone così:  
$\phi_{nk}(r)=e^{ik\cdot r}u_{nk}(r)$  
dove: $u_{nk}(r+R)=u_{nk}(r)$   
dove la traslazione r->R+r è una simmetria del cristallo (quindi, distanza tra due atomi vicini)  

cioè:   
1- plane-wave phase factor  
è la H di una particella libera   
2- modulazione   
appare solo negli elettroni del cristallo

quindi la funzione cambia di una fase da un atomo all'altro:    
$\phi_{nk}(r) = e^{ikR}\phi_{nk}(r)$  
quindi, la densità è periodica: $|\phi_{nk}(r+R)|^2 = |\phi_{nk}(r)|^2$
