### 1. Esistenza bande

1- cristalli hanno SIMMETRIA TRASLAZIONALE DISCRETA   
$\psi(r+R)=\psi(r)$  

2- uso th di BLOCH  

3- if interatomic distance is small, orbitals  have small energy difference: create bands

#### Bloch

**Th di Bloch**

ipotesi:   
simmetria traslazionale discreta: $\psi(r+NR)=\psi(r) \quad \forall N \in N$    
R: distanza tra due atomi del cristallo  

tesi:  
posso scrivere gli autostati di H (i cui autovalori sono le energie possibili) così:  
$H \psi_{nk}(r) = E_n(k) \psi_{nk}(r)$   
$\psi_{nk}(r)=e^{ik\cdot r}u_{nk}(r)$  
dove: $u_{nk}(r+R)=u_{nk}(r)$   

n: numero di banda (valenza, conduzione...)  
k: wave-vector dell'elettrone  

**Osservazione**

1- l'autostato è un'onda piana  

2- l'autostato, quando passa da un atomo all'altro del cristatllo, riceve solo una fase    
perchè $u(r)$ è periodica di periodo R, quella non cambia     
$\phi_{nk}(r+R) = e^{ikR}\phi_{nk}(r)$  

3- quindi, anche la densità è periodica:  
$|\phi_{nk}(r+R)|^2 = |\phi_{nk}(r)|^2$

**Conseguenze: esistenaz bande**

Consideriamo un cristallo generico, dove N atomi hanno orbitali s e p  

1- distanza infinita: N livelli degeneri, con energia $E_s$ e $E_p$  
cioè, gli orbitali non si uniscono a formare quelli molecolari  

2- diminusco la distanza: si creano bande  
bands are omposed of closely spaced orbitals

**Moto elettrone su edge della banda**

graficamente:  
se gli applico una forza, gli cambio p, cioè k -> quindi lo sposto su un'altro punto, della STESSA curva, del grafico E-k  
se prima era nel minimo: ora E è aumentata  
se prima era nel massimo: ora E è diminuita  

formule:  
$\vec a = \vec F \frac 1 \hbar \frac{d^2E}{dk^2}= \frac{\vec F}{m_eff}$  
definisco una massa efficace, positiva per gli elettroni in conduzione, negativa per quelli invalenza  
così, quando sono sottoposti ad una forza, quelli in conduzione aumentano energia, quelli in valenza la diminuiscono  

**Moto di un elettrone sul fondo di una banda**

$E_c(k)$, nel suo punto di minimo, è localmente una parabola  
se applico un campo elettrico costante:  
$F = -e\phi$    
$v_g = \frac 1 \hbar \frac{dE(k)}{dk} = 0$ (velocità di gruppo deve essere nulla, ma perchè???)  

accelerazione:  
$\vec a = \frac{ d\vec v}{dt} = \frac 1 \hbar  \frac{d}{dt} \frac{dE(k)}{dk} = \frac 1 \hbar  \frac{d\vec k }{dt} \frac{dE^2}{dk^2} = \frac{\vec F} \hbar  \frac{dE^2}{dk^2} >0$  

teorema dell'accelrazione:  
$\frac{dE}{dt} = F \cdot v = 0$  
$\frac{d^2E}{dt^2} = F \cdot a > 0$    
Si applica una forza -> l’energia aumenta

### 2. Drift-diffusione

**Sistema per modellizzare**

$\begin{cases} 
J_n = \frac{ne^2 \tau}{m^*_c}E - \; (-e) D_n \nabla n \\
J_p = \frac{ne^2 \tau}{m^*_c}E - \; (+e) D_p \nabla p \\   
\frac{\partial n}{\partial t}+\frac 1 {-e} \nabla \cdot J_n = G-R \\
\frac{\partial p}{\partial t}+\frac 1 {+e} \nabla \cdot J_p = G-R \\
-\varepsilon \nabla^2 \phi = e[p(r)-n(r)+N_d^(r)-N_a^-(r)]
\end{cases}$  


#### Spiegazione equazioni

**Equazione drift-diffusione**

$J_n = \frac{ne^2 \tau}{m^*_c}E - \; (-e) D_n \nabla n$  
$J_n$ = drift - diffusione   


posso definire:  
conduttività: $\sigma = \frac{ne^2 \tau}{m^*_c} = ne \mu$  
così il drift è il solito: $J = \sigma E$   

1- drift    
causa: campo elettrico   
E esercita forza sugli elettroni  
e- hanno carica negativa, quindi si muovno in verso oppposto a E  
ma la corrente è la direzione opposta al moto degli elettroni, quindi la stessa di E  
$n$: concentrazione di elettroni  
$\tau$: tempo medio tra urti  
$\mu$: mobilità  

2- diffusione   
causa: distribuzione non uniforme degli elettroni  
si muovono secondo gradiente, tendono a spostarsi verso la zona dove sono meno concetrati  
$\Phi =-D_n\nabla n$  ($\Phi$: flusso degli elettroni: quanti elettroni attraversano una superficie, in unità di area e tempo)  
$J_n = e \Phi$ (J_n: densità di corrente, quanta carica attraversa una superficie, per unità di area e di tempo) (quindi, moltiplico il flusso per la carica di un elettrone)  

stessa cosa per le buche, ma hanno carica positiva +e  

**Equazioni di continuità**


$\frac{\partial n}{\partial t}+\frac 1 {-e} \nabla \cdot J_n = G-R$    

la solita eq. di continuità di fisica 2  
G: rate of generation (quindi, creazione di una carica)  
R: rate of recombination (quindi, sparizione di una carica)

fisica 2:    
$\frac{\partial \rho}{\partial t} +\nabla \cdot J =0$  
$\rho$: densità di carica  
qui, cariche non si creano nè spariscono  

**Eq. di Poisson**

$-\varepsilon \nabla^2 \phi = e[p(r)-n(r)+N_d^(r)-N_a^-(r)]$    

the distribution of charge in the semiconductor determines the electric potential and therefore the electric field  

a fisica 2:  
$\delta \phi^2 = f$  
f: funzione sorgente
