**States of an electron**

in a crystal, an electron can occupy only certain quantum states  
each state has an energy  

$E_n(\vec k)$  

we identify a state with:  
1- wave-vector: k    
2- band number: n  

**How do you read this graph**

1- Choose a value of k  
2- Find where that vertical line meets the band curves  
3- Each intersection gives an ALLOWED ENERGY at that k  
so have several allowed energies for a same k, one for each band  
4- choose the curve/band, that gives me the energy $E_n(k)$ 

**Bande**

band: a group of allowed states    
graphically: one of the curves  
i give a name to each band (valenza, conduzione...)  

there is a band gap between bands  
means that no electron can have that energy, no matter its k  
$E_g = min_k E_c(k)-max _k E_v(k)$  
il gap è la fascia orizzontale tra il MINIMO della conduction band e il MASSIMO della valence band  
cioè quanto costa ALMENO passare dall'una all'altra   


**How to read this graph. 2**

#### Dualismo

**Dualismo onda-particella**

at SMALL SIZES, a particle shows wave-like behaviour  
behaves like a wave of $\lambda$ decided by the De Broglie equation  
so, identifying a particle with k means identifyig its momentum p, at small sizes  


$\begin{cases}
\lambda = \frac h p & De \, Broglie \\
|k| = \frac{2 \pi}\lambda
\end{cases}$  
$\vec k =\vec p \hbar$     

**Vettore d'onda**

k: vettore d'onda    
tells us how rapidly the phase changes in space  

relationship with:  
1- wavelength: $k = \frac{2 \pi}\lambda$    
2- momentum of the particle: $\dot P \psi = -i \hbar \frac {d}{dx} (Ae^{ikx}) = \hbar k \psi$  
$p = \hbar k$

### Fermi level

**Fermi level**

Fermi level $E_F$: a reference energy describing how electron states are occupied  
It is an occupation marker     

At zero temperature:  
1- Allowed states below $E_F$ are occupied  
2- Allowed states above $E_F$ are empty  

At ordinary temperatures:  
this boundary becomes gradual  
1- states sufficiently below $E_F$ are mostly occupied    
2- states sufficiently above it are mostly empty    

**How to move it**

unbiased: the two sides are in equilibrium, electrons have no preferred direction of flow  
they share one Fermi level, so the dashed line is continuous and horizontal      

biased: battery it maintains different electron energy references on the N and P sides  
so, draw two separate dashed levels, one for each side    
Their vertical separation corresponds to the applied voltage  


**Consequences of moving it**

If $E_F$ moves closer to the conduction band:  
more conduction-band states tend to be occupied  
so, more conduction electrons    

If $E_F$ moves closer to the valence band:  
more valence-band states tend to be empty  
so, more holes. ???

#### Biasing

**Biasing**

the band edge is an energy $E_n(x)$  
adding a potential changes the energy of all particles in that region, so translates the edge  
$E_c(x) = E_{c,0}-e\phi(x)$  
$E_v(x) = E_{v,0}-e\phi(x)$  

$E_c(x_R)-E_c(x_L) =e\int_{x_L}^{x_R} \phi(x)\,dx $  

applying a potential in a spatial region changes by the same amound the energy of all states in the that region, of all bands  
so cannot change the band gap  
can only change the height of the curve compared to other spatial positions x  

**Explanation: equilibrium**

N side has many conduction electrons    
P side has many holes (holes are drawn as valence electrons)

1- line edges  
SCR: built-in field changes changes the energy of the electrons   
so, E(x) is curved, not straight, bc of the built-in field    
E(x) has a slope: $\frac{E(x)}{dx} = e\phi(x)$  
QNR: no field, so energy stays constant  

2- Fermi level  
same height on both sides, because the whole junction is in thermal equilibrium  

so, anode and cathode have E(x) at different height bc they are at different voltages, so have different E  
la differenza di energia tra i lati di una stessa banda è 0.7V, cioè il voltaggio attraverso il diodo    

**Motion of particles**


an electron can move to another region if it has enough energy to stay in the conduction band of the new region (cannot go to valence band, bc it's full)  

the conduction-band edge rises when you go from N to P  
for an electron in the N-side, this represents an energy barrier -> it cannot pass    

![](./bias2.jpg)


**Applying bias**

unbiased: majority carries (electron on N side) see a potential barrier, cannot cross over to P  
while minority carriers (electrons on P sides) can cross over to N   

FORWARD BIAS: N gets higher, P gets lower, so conduction bands on the two sides get closer (weakens the built-in field, which is the one who separated them originally)    
now the barrier is weak enough that electron from N can go to P -> have current (remember that current is in opposite direction to electron movement, conventionally)  

REVERSE bias: N gets even lower, P gets even higher (strengthen the built-in field)   
so, e- in conduction band in N sees an even bigger barries, cannot go to conduction band in P    
but, e- from valence band in P can go to condution band in N, bc it's so low they almost cross: it's QUANTUM TUNNELING    
so, electrons from P go to N -> reverse current (in opposite direction)  

se metto un bias $\Delta V$, creo una distanza $\Delta V$ tra i due lati del Fermi level  

**Quantum tunneling**

Classically: the intervening region is forbidden    
Quantum mechanically: the electron’s wavefunction can extend through it, giving a nonzero probability of finding the electron on the other side  
Strong doping makes the relevant barrier sufficiently thin for this to occur readily.
