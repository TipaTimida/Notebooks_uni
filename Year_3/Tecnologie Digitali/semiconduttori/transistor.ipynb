### Diodo

**Formula**

$p(x) = \frac{n_i^2}{N_D} +\frac{n_i^2}{N_D}(e^{V/V_T}-1)e^{x(L_p)}$  
p(x): hole concentration in the N region (at distance x from the junction)   
$N_d$: donor concentration in n-type paterial  
$L_p$: hole diffusion length, the typical distance holes travel in the n-side before recombining  

1-first addend: concentration of holes in N-doped material at equilibrium  
come from:  
$n_0p_0 = n_i^2$  
$n_0 \sim N_D$  

2- seconda addend: extra holes injected by forward bias (bc takes away electrons at extremity, so it's like it injects holes)
