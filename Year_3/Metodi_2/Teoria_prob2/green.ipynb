### Teoria

**Cos'è**

1- Green function: a description of what a system does after a single, idealized, very brief input at one place and time  
so model how it responds to a brief impulse  

2- convolution: calculation that combines many shifted copies of that simple response   
so imagine the input signal as a serie of impuelses, at different times (so shift in time)    
combine the response to each impulse to have the reponse to the full input signal

**1. Definizione**

G(t)=0 per t<0  

cioè: comincia ad esistere la risposta G solo dopo il tempo 0, è stata causata da qualcosa che è accaduto a t=0 

**2. Th di Titchmarsh**

una funzione G(t) è causale se:  
la sua trasformata $\tilde G(\omega)$ è il bordo di una funzione analitica nel semipiano superiore, con una certa stima di integrabilità  
usala se: ho $\tilde G(\omega)$, voglio sapere se G(t) è causale senza calcolarla  

steps:  
1-passa al piano complesso, $\omega = x+iy$  
2- controlla che non ci siano poli nel semipiano superiore (cioè: è analitica nel piano superiore)   
3- verifica la stima integrale sul semipiano superiore (y>0)  

### Proprietà

**Relazione di Kramers-Kronig (relazioni di dispersione)**

cosa: relazione tra parte reale e immaginaria di una funzione analitica  

ipotesi:   
$G(\omega) = G(\omega)+ i G(\omega)$  
G analitica sul semipiano superiore (nota che causale -> analitica sul piano superiore, per Titchmarsh)  
G scala più veloce di $1/|\omega$| per $|\omega| \rightarrow \infty$

tesi:  
$Re[G(\omega)] =\frac 1 \pi P \int_R \frac{Im [G(u)]}{u-\omega}du$  
$Im[G(\omega)] =-\frac 1 \pi P \int_R \frac{Re [G(u)]}{u-\omega}du$   

**G a supporto compatto**

se scritta a tratti (su tratti di t), le parti dove $G \neq 0$ sono un tratto finito

### Teoria: Convoluzione vs moltiplicazione

convolution can become multiplication (and vice-versa), if used Fourier transform

**Convolution**

$(g * f) (s) = \int_R g(u) f(s-u) \, du $  


cosa | time | come diventa in freq
---|---|---
time conv -> freq mult| $x(t) = (g * f) (t)$ | $\tilde x(\omega) = \tilde  g(\omega) \, \tilde f(\omega)$   
freq conv -> time mult | $x(t) =g(t) f(t)$ | $\tilde x(\omega) = \frac 1 {2 \pi} (\tilde g * \tilde f)(\omega)$   

**Don't confuse: Chaining vs convolution**

chaining:  
$(f∘g)(t)=f(g(t))$    
used to change the thing on the x axis  
if know a as func of b and b as func  of c, want a as func of c  

convolution:  
$ (f*g)(t)=\int_{-\infty}^{\infty} f(\tau)\,g(t-\tau)\,d\tau$   
in a circuit:  
f: input signal  
g: response to one impulse  
i'm consider the input signal as a sum of impulses, and adding the response to all of them
