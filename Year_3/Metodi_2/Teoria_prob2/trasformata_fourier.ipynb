### Definizione

**Convenzione che uso**

$\tilde f(\omega) = \int_R f(t) e^{i \omega t} \, dt$  
$f(t) = \frac 1 {2 \pi} \int_R \tilde f(\omega) e^{- i \omega t} d\omega$  


**Convenzioni**

$\tilde f(\omega) = c \int_R f(t) e^{\sigma \; i \omega t} \, dt$  
$f(t) = \frac 1 {2 \pi c} \int_R \tilde f(\omega) e^{-\sigma \; i \omega t} d\omega$  

1- sign choice: $\sigma = \pm 1$    
i usually use $\sigma=1$  

2- normalization: $c$, può essere ogni valore  
i usually use $c=1$  

**Usarla più volte**

se ho: $F[f(t)] = \tilde f(\omega)$  

1- antitrasformata  
$F^{-1}[\tilde f(\omega)] = f(t)$  
utile per: tornare al time domain dopo aver risolto in freq domain  

2- duality theorem  
$F[\tilde f(t)] = 2 \pi c^2 f(-\omega)$  
utile per: ricordarsi le formule comuni, a coppie

esempio:  
$F[\delta(t)] = 1$  
se uso F: $F[1] = 2 \pi \delta(-\omega) = 2 \pi \frac 1 {|-1|} \delta(\omega)$  
se uso $F^{-1}$: $F^{-1}[1] = \delta (t)$  

### Esercizio

**Steps**

1- separa con linearità  
2- applica le proprietà: freq shift, derivative  
3- calcola le F di tutti i pezzi   
o cose note, o calcola l'integrale (con th residui)

**1.Linearità**

perchè: la trasformata è un integrale, posso separarl in addendi  
$F[af(t)+bg(t)]= aF[f(t)]+bF[g(t)]$  

**2. Proprietà**

cosa | formula f(t) | formula Fourier Transform
---|---|---
| Time shift | $f(t-t_0)$ | $e^{\sigma \; i\omega t_0}\tilde f(\omega)$ |
| Frequency shift |$e^{i\omega_0t}f(t)$ | $\tilde f(\omega+\sigma \, \omega_0)$ |
| Time scaling | $f(at)$, $a\ne0$ | $\dfrac{1}{\|a\|}\tilde f\!\left(\dfrac{\omega}{a}\right)$ |
| Time derivative |$ f'(t)$ | $-\sigma \, i\omega\,\tilde f(\omega)$ |
| Multiply by time | $t f(t)$ | $-\sigma \, i\,\dfrac{d\tilde f}{d\omega}$ |

la prima riga si usa se, nel freq domain, mia ppare $e^{i \omega t_0}$ moltiplicato a qualcosa, cioè per fare l'anti-trasformata  

ricorda che, di defalut, $\sigma=1$  
se invece, ho messo $c \neq 1$, allora moltiplica la formula di Fourier per c (come si vede nella def della convenzione)  

**2. Properietà: come si scrive**

freq shift: $F[e^{it}](\omega) = F[1](\omega+1)$  
derivata: $F[x']= (-i \omega) F[x]$  

**3. Common ones**

Function | transform
---|---
 $\delta(t)$ | $1$ |
$1$ | $2\pi \delta(\omega) $
$PV  \frac 1t$ | $\sigma i \pi sgn(\omega)$
$sgn(t)$ | $2 \sigma i PV \frac 1 \omega$ 
$\theta(t)$ | $\pi\delta(\omega)+\sigma i\,\operatorname{PV}\dfrac1\omega$ |
$e^{-at}\theta(t),\ a>0$| $\dfrac{1}{a-\sigma i\omega}$ |

se invece, ho messo $c \neq 1$, allora moltiplica la formula di Fourier per c (come si vede nella def della convenzione)  

$F^{-1}[\delta(\omega)] = \frac 1 {2 \pi}$


**3. Less common ones**

Function | transform
---|---
$e^{-a\|t\|}, a>0$ | $\dfrac{2a}{a^2+\omega^2}$ |
$\cos(\omega_0t)$ | $\pi \,[\delta(\omega-\omega_0)+\delta(\omega+\omega_0)]$ |
$\sin(\omega_0t)$ | $\sigma i\pi \,[\delta(\omega-\sigma\omega_0)-\delta(\omega+\sigma\omega_0)]$ |
$e^{-t^2/(2\tau^2)}, \tau>0$ | $\sqrt{2\pi}\,\tau\,e^{-\tau^2\omega^2/2}$

per trig e cos, prima trasforma in somma di esponenziale, poi usa FOurier su quelli, che danno la delta  


### Spiegazione proprietà

**Da dove vengono le proprietà**

si vedono facendo la trasformata su quella cosa  
eg. derivata -> integration by parts  
$\int_R f'(t) e^{i \omega t} \, dt = -i \omega \int_R f(t) e^{i \omega t} dt$    
if the boundary term vanishes, and it does if $f \in L^2$, which i assume  

**Gaussiane**

Fourier trasforma gaussiana in gaussiana  
ma, se prima era larga, diventa stretta  
prima: $e^{-t^2/(2\tau^2)}, \tau>0$   
dopo: $\sqrt{2\pi}\,\tau\,e^{-\tau^2\omega^2/2}$

nota: per la larghezza, uso $\tau$ invece di $\sigma$, per non confonderla con il segno della convenzione    

### Distribuzioni

#### PV - valore principale di Cauchy

**Cosa fa**

formalmente:  
dice di togliere un intervallo simmetrico attorno a zero e poi farlo restringere  
così tolgo solo 0, il punto problematico

graficamente:   
invece di passare per il punto O, il bordo fa unacurva di raggio r, con $ r \rightarrow 0$  

**Utilizzo**

quando:  
se ho integrale con una singolarità nel suo intervallo di integrazione  

se non lo uso: devo staccare l'intervallo nei punti singolari, e vedere che succede  
se lo uso: ignora le singolarità interne, non staccare, semplicemente calcola  valori agli estremi come sempre

**Esempio**

$\frac 1x$ ha punto singolare x=0  

1- NON lo uso  
$\int_{-1}^{1} \frac 1x = \int_{-1}^0 \frac 1x+ \int_{0}^{1} \frac 1x$  
e devo studiare se a x=0  diverge o converge  

2- lo uso  
$PV \int_{-1}^{1} \frac {dx}x = \lim_{\epsilon \rightarrow 0^+} [\int_{-1}^{-\epsilon} \frac {dx} x+ \int_{\epsilon}^{1} \frac {dx}x]$  
$P \int_{-1}^{1} \frac {dx} x = \ln x|_{-1}^1 = 0$   
cioè ingoro il punto problematico  


#### Delta di Dirac

**Proprietà**

nome | cosa
---|--
time scaling | $\delta(ax) = \frac {1}{\|a\|}\delta (x)$  
derivata | $f(\omega) \delta'(\omega-a)= f(a)\delta'(\omega-a)-f'(a)(\delta(\omega-a))$ 
derivata (caso particolare)| $(\omega-a)\delta'(\omega-a) = -\delta(\omega-a)$
Prodotto con una funzione | $f(\omega) \delta(\omega-a) =f(a) \delta(\omega-a)$ <br> perchè è non-0 solo se $\omega=a$
