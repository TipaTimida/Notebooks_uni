### Laplace transform

**Transform**

go from a func of time to a func of something else  


**Laplace transform**

$s= \sigma + j \omega$  
s: a var that describes how the signal changes over time  
$\sigma$. exponential growth or decay  
$\omega$ obscillation  

**Why using it**

turns a diff equation into an algebric one  
then, transform back the answer  

**Definizione**

$L\{f(t)\}=\int_0^\infty e^{-st} \, f(t) \, dt$   
alternate notation: $L\{f(t)\}=F(s)$  
in pratica cambio var, da t a s  

c'è una tabella con le più comuni  

#### Formule

**Proprietà**

lineare:  
$L\{af(t)+bg(t)\}=aF(s)+bG(s)$  

vale anche per l'inversa:  
$L^{-1}\{aF(t)+bG(t)\}=aL^{-1}\{F(s)\}+bL^{-1}\{G(s)\}$  


**Common pairs**

inizio da: y(x) (ma potrebbe anche essere V(t), o qualunque altro) 
passo a: Y(s)  

$L\{y''\} =s^2Y(s)-sy(0)-y'(0)$  
$L\{y'\} =sY(s)-y(0)$  
$L\{y\} = Y(s)$  
$L\{x\}= \frac 1{s^2}$  
$L\{1\} = \frac 1s$     
$L\{\theta(t)\} =L\{1\} = \frac 1s$   

to transform back:  
$L\{e^{-at} \cos (bt) \} = \frac{s+a}{(s+a)^2+b^2}$    
$L\{e^{-at} \sin (bt) \} = \frac{b}{(s+a)^2+b^2}$  
$L\{\frac 1{s+a}\}= e^{-at}$  

they work in both senses    
they form a laplace pair, so if L(a)=b, then L(b)=a  

### Esempio

**Metodo: Laplace**

1- diff eq del circuito  
$\tau v_{out}' +v_{out}=V_0 $  

2- usa Laplace (servono codizioni iniziali)  
$\tau[sY(s)-v_{out}(0)] + Y(s) = \frac{V_0}{s}$  
$Y(s) =\frac{V_0}{s(\tau s +1)}=V_0(\frac 1s -\frac 1{s+1/\tau})$  

3- transform back  
$V_{out}(t)=V_0(1-e^{-t/\tau})$  

**Use Laplace transforamtion to solve diff eq**

example: $y''-10y'+9y=5t$  

1- take Laplace on both sides  
$L\{y''\} -10L\{y'\} +9L\{y\} = L\{5t\} $  

2- usa formule:  
$L\{y''\} =s^2Y(s)-sy(0)-y'(0)$  
$L\{y'\} =sY(s)-y(0)$  
$L\{y\} = Y(s)$  
e quella a destra, dalla tabella: $L\{5t\} = \frac{5}{s^2}$  
$s^2Y(s)-sy(0)-y'(0) -10(sY(s)-y(0)) +9Y(s) = \frac{5}{s^2}$  

3- plugga condizioni iniziali, quindi y(0) e y'(0)diventano numeri  
poi raccogli Y(s)  
$(s^2-10s+9)Y(s)+s-12 = \frac{5}{s^2}$  

4- risolvi per Y  
$Y(s) = \frac{5}{s^2(s-9)(s-1)}+\frac{12-s}{(s-9)(s-1)} = \frac{5+12s^2-s^3}{s^2(s-9)(s-1)}$  
separa in una somma, con partial fractions:  +
$Y(s) = \frac{A}{s}+\frac{B}{s^2}+\frac{C}{s-9}+\frac{D}{s-1}$    
come si fa: ugugalia i numeratori  
scegli valori appropriati di s, tipo s=0, s=1, s=9... per trovare tutti i valori di A,B,C,D 

5- dai la trsformazione di laplace inversa per avere $y(t)$

### Laplace vs Fourier transform

| | Fourier transform | Laplace transform |
|---|---|---|
| Variable | Real frequency $\omega$ | Complex frequency $s=\sigma+i\omega$ |
| Kernel (one common convention) | $e^{-i\omega t}$ | $e^{-st}=e^{-\sigma t}e^{-i\omega t}$ |
| Best for | Frequency content and steady sinusoidal behavior | Transients, initial conditions, and system stability |

The Laplace transform adds the factor $e^{-\sigma t}$, which can make an otherwise non-convergent integral converge  
For example, a signal that grows exponentially may not have an ordinary Fourier transform, while its Laplace transform can exist for some values of $s$

so fourier is a special case of laplace, where:  
1- sigma=0  
2- estremi di integrazione sono $(-\infty, \infty)$, invece di $(0, \infty)$
