**Power law, serie geometrica**

$\sum z^n = \frac 1 {1-z}\quad |z|<1$  
posso comunque trovarla con taylor, facendo $z \sim 0$ a destra  

**Radius of convergence**

$\lim_{n \rightarrow \infty} \frac{a_{n+1}}{a_n} = L$  
$\frac 1L$: radius of convergence

### Mandelbrot set

**Cos'è**

Mandelson set M: un insieme di punti del piano complesso  
contiene i punti $z_0$ tali che:  
$z_{n+1}=z_n^2+z_0$  
$z_n$: orbita di $z_0$ (n parte da 0)  
se $z_n \rightarrow \infty \quad z_0 \notin M$  
sennò: $z_0 \in M$  

**Esempio**

$z_0=1$  
$z_1 = 1^2+1=2$  
$z_2 = 2^2+1 = 5$  
$z_n \rightarrow \infty$  
quindi $1 \notin M$  

$z_0 =i$  
$z_1 = -i+i$  
$z_2 = -i$  
...
$z_n$ non andrà mai all'infinito  
quindi $ i \in M$  

**Note**

it is connected (è un pezzo unico, non due separati)

#### Julia set

**Cos'è**

regole simile al Mandelbrot set, ma:  
$z_{n+1}=z_n^2+c$  
dove "c" è un numero complesso scelto  
per ogni c scelto, posso disegnare un diverso set  

**Julia vs Mendelson**

se $c \in M$: Julia set è connected  
se $c \notin M$: Julia set è disconnected
