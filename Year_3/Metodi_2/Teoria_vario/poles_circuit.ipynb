#### Definizione

**Cosa sono**

$f(s) = \frac {N(s)}{D(s)}$  

zeri: radici di $N(s)$    
They mark inputs or frequencies the system suppresses  

poli: radici di $D(s)$   
They describe the system’s natural responses—such as decaying or growing oscillations  

s è una var complessa:  
$s= \sigma + i \omega$  
nota: a volte, uso $z=x+iy$  

**Come usarlo (in un circuito)**

f(s) is the CLOSED-LOOP gain, o la tranfer function, so Vout/Vin,    
use it to find the beahaviour of the circuit  

#### Studia comportamento

**A cosa servono**

trovali, e plottali in un plot $Re[s], Im[s]$  
a seconda di dove sono, mi dice cosa succede al circuito a quella frequenza  

circuit behaves like: $e^{st}$  
$Re[s]$: decay/grow      
$Im[s]$: obscillation  

**Studia comportamento**

| Pole                  | Corresponding behaviour                                   | What you observe              |
| --------------------- | --------------------------------------------------------- | ----------------------------- |
| $s=-\sigma$, $\sigma>0$     | $e^{-\sigma t}$                                        | Decays                        |
| $s=+\sigma$, $\sigma>0$     | $e^{\sigma t}$                                                | Grows                         |
| $s=\pm j\omega_0$   | $\cos(\omega_0t), \; \sin(\omega_0t)$                  | Oscillates without dying away (resonance) |
| $s=-a\pm j\omega_0$ | $e^{-at}\cos(\omega_0t)$ and $e^{-at}\sin(\omega_0t)$ | Oscillates while decaying     |

**Stabile**

che vuol dire: no growth (but decay is ok)    
quindi: $Re[s]<0$  
cioè, tutti i poli sono sel semipiano sinistro del plot

#### Studiare circuito

1- Transient behavior  
the temporary part of the output right after something changes  
eg. switching the circuit on, suddenly changing its input  
it fades away as the circuit settles  

2- Steady-state behavior   
what remains after that temporary part has faded   
it depends on the input  

**Diff eq**

1- solo omogenea: transiente   
chiamala transiente solo SE decade  
se invece cresce, it's an undamped system, keeps obscillating  
2- sol particolare: steady-state  

eg. RC circuit  
$\tau V' +V =V_0$   
omogena: $V(t) = C e^{-t/\tau}$ -> transiente, è il caricamento, che è temporaneo   
particolare: $V(t) = V_0$ -> totale, è cosa resta dopo aver finio di caricarsi  
$V(t) =V_0(1-e^{-t/\tau})$  


**Esempio: high-pass**

$H(\omega) = \frac{s \tau}{1+ s \tau}$  

1- poli: $s=-1/\tau$  
come cade il transiente (il caricamento o scaricamento di C)  

2- zeri: $s=0$  
quali freq vengono bloccate, qui $\omega=0$
