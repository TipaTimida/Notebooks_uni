### Principal argument

$arg(z_1)=arg(z_2)$  
if $\phi_1 = \phi_2+2k \pi$  

In order to make the argument of a well-defined number:  
it is sometimes restricted to the interval $[- \pi, \pi]$  
This special choice is called the principal value or the main branch of the argument and is written as Arg(z) (con la maiuscola)

### Complex funcs

**Multiple-valued**

$w=f(z)$  
single-valued: each input has exactly one output   
multiple-valued: each input has more than one output  

injective: each output has at most one input (different inputs always have different outputs)

### Infinity

**Only one**

in R, have two infinities, $-\infty$ and $+\infty$  
in C, only have one, $\infty$  
bc C has no natural ordering, unlike R  

formally:  
we add a $\infty$ symbol to C to obtain the "extended complex plane£, called $C^*= C \cup \{\infty \}$ 

**Riemann sphere**

The extended complex plane can be mapped onto the surface of a sphere where:  
south pole = origin  
north pole = $\infty$  

**Stereographic projection**

the surface of the sphere is the complex plane  
so, has no ordering, unlike R  

how to pick a point:  
1- pick a point z it in the carthesian plane C (the normal one)  
2- draw line from it to the north pole of the Riemman sphere (a sphere with the south pole on O)  
3- the point of the surface it intersects is the stereographic projection of z  

### Derivata complessa

**Pratica**

per fare le derivate, posso farle direttamente su z  
non preoccuparti di passare a z=x+iy  

**Equazioni Cauchy-Riemann**

f ha una derivata complessa:  
$f(z) =u(x,y)+iv(x,y)$  
$f'(z) = u_x+iv_x=v_y-iu_y$  

quindi esiste se:  
u e v sono differenziabili, e soddisfano le equazioni di Cauchy-Riemann  
$u_x=v_y$  
$u_y=-v_y$  

**Analitica**

funzione analitica in $z_0$= differenziabile in $z_0$   
cioè, rispetta le equazioni di Cauchy-Riemann in $z_0$  

### Riemann surface

**Definition**

connected one-dimensional complex manifold  
easy: a geometric surface that locally looks and acts just like the standard complex plane C, even though its shape can be more complex  
can be a sphere, a ulti-sheeted spiral (log)...  
and there must be a way to map ever point in C to a point of the surface  

**Manifold**

uno spazio topologico che localmente è simile a uno spazio euclideo N-dimensionale   
ma che globalmente può avere proprietà geometriche differenti   
eg. può essere "curvo" contrariamente allo spazio euclideo  

eg. la superficie di una sfera è un manifold, perchè è curva, non la posso rappresetare (spianare) perfettamente su un piano euclideo (piatto)

### Integrale su una cura complessa

**Tip**

separa parte reale e immaginaria  

$\int_C f(x) \, dz = \int_a^b f( \, z(t) \,) z'(t) \, dt$  
dove: $z(t)=x(t)+i(t)$ è la parametrizzazione della curva C  

$\int_C f(z) \, dz$  
$=\int_C(u+iv) \, (dx+ i \, dy) $  
$ = \int_C (u \, dx-v \, dy) + \; i\int_C (v \, dx + u \, dy)$  

**Esempio**

$f(z)= \overline z$  
$z(t) = 3t + it^2$ (parametrizzazione)    
$-1 \le t \le 4$ (bounds di C)  

$\int_C \overline z \, dz $  
$\int_{-1}^4 (3t-it^2) \, (3 +2it) \, dt$  
$\int_{-1}^4 (2t^3+9t) dt + i \int_{-1}^4 3t^2 dt$
