### Trig

NON usare le trig  
trasformale in esponenziali complessi  

$\begin{cases}
\sin z = \frac {e^{iz}-e^{-iz}}{2i} & \cos z = \frac{ e^{iz}+e^{-iz}}2 \\
\sinh z = \frac {e^{z}-e^{-z}}{2} & \cosh z = \frac{ e^{z}+e^{-z}}2
\end{cases}$  

#### Trig iperboliche

$\cosh^2z -\sinh^2z = 1$  
$\cos z = \cosh(iz)$  
$\sin z = -i\sinh(iz)$    
$\cosh'z = \sinh z$ (non ha il -)   
$\sinh'z = \cosh z$ 

### Esponenziali

**Equazioni con exp complessi**

cerca di portare tutto a sinistra, e uguaglio a 1 o -1, che so scrivere

esempio 1  
$e^z=e^{-z}$  
$e^z e^z = e^{-z}e^z$  
$e^{2z} = 1 = e^{i 2 \pi k}$   
$z_n = 2 \pi k \quad n \in Z$  

esempio 2  
$e^{z_n} =-1 = e^{i (2n+1)\pi} \quad n \in Z$   
$z_n = i \pi (2n+1)$  

**-1 oscillante**

$e^{\pm i\pi n} = (-1)^n$

**Rapporto nei limiti**

$\frac{e^{aR}}{e^R} = e^{(aR-R)} = e^{R(a-1)}$ <br> if a<1, then it diverges when $R \rightarrow \infty$  

#### Log

**Log di numero complesso**

usa la formula sopra: 
$\log z= \log |z| + i \, arg(z)$  

esempio:  
$i = 1 e^{i \pi/2}$  
$\log i = \log 1 + i \pi = 0 + i \frac \pi 2$

per Residuo: lascia log z  
per integrale in campo complesso con log z in integranda: usa trasformazione  

### Radici

**De Moivre's formula (potenza)**  

$z^n = r^n e^{i , n\theta}$  
cioè: $z^n = r^n (\cos (\theta)+i \sin(n\theta))$  

**Roots**

$z_k^{1/n} = r^{1/n} e^{i \theta_k/n} =  r^{1/n} e^{i(\theta+ 2 k \pi)/n}$    
devo aggiungere $2k \pi$ per avere tutte le radici
