### Th residui

**Come si scrive**

$\oint_C f(z) \, dz = \pm \; 2\pi i \sum_{k=1}^n Res(f, z_k)$    

$z_k$: singolarità racchiuse DENTRO C (anche sul bordo va bene)  

segno a destra dipende dal verso di percorrenza C:  
orario: - (orientato negativo)   
antiorario: + (orientato positivo, come gli angoli)     

**Steps**

1- identifica contorno C e integranda f(z)  
2- calcola poli di f(z) racchiusi DENTRO C  
3- usa formula  

**Definizione**

permette di calcolare un integrale:  
1- lungo un percorso chiuso   
2- nel piano complesso  
$\int_C f(z) \; dz$   

usando soltanto il comportamento della funzione vicino alle singolarità racchiuse nel percorso    

### Ex. integrale solo sulla retta reale

**Tipi di esercizi**

1- quando: ho già calcolato l'integrale su un percorso chiuso dato dal testo (eg. rettangolare), che comprende la retta reale  
cosa: separa quel percorso in tratti    

2- quando: else  
cosa: completa tu il percorso con un arco/semicerchio di raggio R (attenta se sul semipiano sopra o sotto) 

poi, ogni tratto guarda a cosa tende 

**Come vedere a cosa tendono i tratti**

scrivi l'integrale giusto:  
1- invece di z, metti solo la var che cambia  
2- con i giusti estremi  
3- e ricorda ci cambiare il differenziale  

che tratto | come si scrive integrale | a cosa tende di solito
---|---|---
asse reale | $\lim _{R \rightarrow \infty} \int_{-R}^R f(z) dz$ | a $I$ stesso  
linea orizzontale <br> a $y=i\pi$, con $x \in (-R, R)$| $\lim _{R \rightarrow \infty} \int_{-R}^R f(x+i\pi) dx$ | a $\alpha I$, oppure a $I + \alpha$  
linea verticale <br> a $x=R$, con $y \in [0, \pi]$ | $\lim _{R \rightarrow \infty} \int_0^\pi f(R+iy) \, i \, dy$ | a 0 
arco <br> di raggio R | $\lim _{R \rightarrow \infty} \int_{C_R} f(z) \, dz$ | a 0 
arco piccolo intorno a singolarità $z_0$| parametrizza la curva: <br> $z= z_0+\epsilon e^{i \theta} \quad 0 \le \theta \le \pi$ |valore finito  

assumo che I sia l'integrale sull'asse reale, quello che mi interessa

#### Dimostra che un tratto tende a 0

**Steps**

1- fai upper bound della NORMA dell'integranda, $max|f(R)|$  
2- moltiplicalo per lunghezza del percorso  
3- vedi se tende a 0     

**Metodi**

quando | condizione | cosa dimostrare
---|---|--
arco di raggio R | $\|\int_{C_R} f(z) \, dz\| \le \; \pi R \; \max_{C_R} \|f(R)\| \ \rightarrow 0$ | $R \, \max_{C_R} \|f(R)\| \rightarrow 0$  
tratto verticale in $x= R$  | $ \|\int_{0}^{2\pi} f(z) \, dz\| \le \; 2\pi  \; \max \|f(R)\| \ \rightarrow 0$ | $\max \|f(R)\| \rightarrow 0$  
se l'integranda ha esponenziale con arg IMMAGINARIO (eg. trasformate Fourier) <br> "lemma di Jordan" |  $\|\int e^{i \omega z} g(z) \, dz\| \le \frac \pi \omega \max_{C_R} \|g(R)\| \rightarrow 0$ | $\max_{C_R} \|g(R)\| \rightarrow 0$  
integranda è dispari | se intervallo è simmetrico, $\rightarrow 0$

max significa fare upper bound  
quindi moltiplico l'upper bound dell'integranda per la lunghezza del tratto ($\pi R$ per un arco, $2 \pi$ per un segmento lungo $2 \pi$)

**Dimostra che tende a 0**

se sospetto che tenda a 0, meglio calcolare la norma   
perchè se $|I_2 |\rightarrow 0$, allora $I_2 \rightarrow 0$  
e la norma è utile, mi toglia roba  

**Arco di cerchio $\epsilon$**

quando si usa: quando un polo è sul bordo, per non passarci sopra  
devo aver fatto serie di TL su quel polo  

1- disegno un piccolo semicerchio attorno al polo  
2- lo parametrizzo   
3- ottengo che il suo integrale dipende dal residuo del polo  

esercizio | cosa
---|---
parametrizzo la curva: <br> $z= z_0+\epsilon e^{i \theta} \quad 0 \le \theta \le \pi$ <br> $  dz = i \epsilon e^{i \theta} d \theta$ <br> così, $z-z_0 = \varepsilon e^{i \theta}$ <br> attenta se l'integrale è $\int_0^\pi$ oppure $\int_\pi^0$, dipende dal verso di percorrenza, mi cambia il segno $\pm$ alla fine| parametrizzo la curva (piccolo semicerhio intorno a $z_0$) 
$D(z) = D(z_0) + D'(z_0)(z-z_0) + O((z-z_0)^2)$ <br> $ =0 + -2 (\varepsilon e^{i \theta}) +o(\varepsilon^2)$| parametrizza denom, approssimando vicino al polo <br> $z_0$ è un polo, quindi D(z0)=0
$N(z) = 2z = 2(i \pi+\epsilon e^{i \theta})= 2i \pi +O(\varepsilon)$| parametrizza numeratore
$I_4 = \int_0^\pi \frac{2i \pi}{-2\varepsilon e^{i \theta}} i \varepsilon e^{i \theta} \, d\theta$  <br> $ = \int_0^\pi \pi \; d\omega = \pi^2$| sostituisci la parametrizzazione nell'integrale

se $z_0$ è un polo singolo, ottengo: $I_r = \pm i \pi R[f, z_0]$    
cioè, è come mettere nel th rei residui anche i poli sul bordo  


#### Chiudi il percorso con un arco

**Dove fare l'arco**

posso farlo sul semipiano sopra o sotto  
scegli quello dove l'integranda decade   
cioè quello dove la NORMA dell'integranda sull'arco fa 0 se $R\rightarrow \infty$, così l'integrale pure fa 0  

**Su che piano decade la funzione**

cosa contiene la funzione | norma di f| caso 1 | caso 2
---|---|---|---
$e^{i \omega z} = e^{i \omega(x+iy)}= e^{i \omega z}e^{-\omega y}$  <br> eg. trasformata | $e^{-\omega y}$ | caso: $\omega>0$ <br> decade su y>0 <br> quindi chiudo sul semipiano superiore | dcaso: $\omega <0$: <br> decade su y<0 <br> chiudo sul semipiano inferiore
$e^{-izt} = e^{-ixt}e^{yt}$  <br> eg. antitrasformata | $e^{yt}$  |caso: $t >0$ <br> decade su w<0, semipiano inferiore  | caso: $t<0$ <br> decade su semipiano superiore   
no esponenziali | | guarda comunque la funzione, guarda dove decade <br> se decade in entrambi, scegli quello pià facile (cioè con meno poli)
 
z è la var su cui NON sto integrando, cioè quella che voglio ottenere  
quindi durante una trasformata è sigma, in un'anti-trasformata è t  

**Verso di percorrenza**

voglio percorrere la retta reale da $-\infty$ a $+ \infty$  
per avere integrale $\int_{-\infty}^\infty$  
quindi vedi se, col percors che ho costruito, è un senso orario o antiorario  

### Tips calcoli

FACCIO PRIMA NORMA O UPPER BOUND???

**Prendere la norma**

1- fasi hanno modulo 1  
$|i| = |e^{iy}| = 1$  

2- portare modulo dentro integrale  
$|I_2| =  |\int f(z) \, dz | \le \int |f(z)| dz$   

3- se z è complessa, specificalo:   
$|e^z| = |e^{x+iy}| = |e^x e^{iy}| = |e^x|$ 

**Fare upper bound (dopo norma)**

$max f(R)$ significa fare l'upper bound, e poi pluggare z=R  

se ho una frazione:  
1- separa numeratore e denom: $|ab|=|a| \; |b|$   
2- numeratore: $|a+b| \le |a|+|b|$  
3- denominatore: $|a + b| \ge |\,|a| - |b| \,|$   
stessa cosa per a-b, semplicemente considero |-b|, ma tanto la norma toglie il -  

esempio:  
numer: $|R+iy| \le |R|+ |i y| = R + y$  
poi fai upper bound, $y \le \pi$  
denom: $|e^R e^{iy}+1| \le |\,\|e^Re^{iy}| -|1| \,| = |e^R -1| = e^R-1$  (perchè R>0)  
$\frac 1{|e^Re^{iy}+1|} \le \frac 1 {e^R-1}$  


**Altre proprietà modulo**

$| \, |z_1|-|z_2| \, | \le |z_1+z_2| \le |z_1|+|z_2|$   
$| \, |z_1|-|z_2| \, | \le |z_1-z_2| \le |z_1|+|z_2|$
