### Definizioni

**Poli**

definizione: punti che violano la C.E.  
cioè, gli zeri del denominatore  

singolarità | isolata | non isolata
---|---|--
cosa | punto singolo che viola la C.E. <br> f non è analitica lì, ma lo è in un suo intorno | qualunque intorno del punto prendo, ce n'è un altro singolare      
esempio | zeri del denom | $x= \pm \infty$  <br> $\sqrt x \rightarrow x > 0$  

### Singolarità isolate

**Steps: come studiarle**

1- trova in che punti $z_n$ sono   
2- trovo tipo  
3- calcola residuo (o guardalo dalla serie)  
4- trova dominio di convergenza (cioè distanza dalle altre singolarità) 

#### 1. Dove sono

negli zero del denominatore:  
$D(z_0)=0$  

#### 2. Tipo di singolarità

tipo | come appare nella serie di TL | come appare in una funzione
---|---|---
Eliminabile | Nessuna potenza negativa | $f(z) = \frac {\sin z}z \sim \frac {z+ z^3...}{z}$
Polo di ordine m | Un numero finito di potenze negative, fino a $a_{-m}$ | $1/z^m$ |  $\|f(z)\|\to\infty$ lungo tutti i percorsi
| Essenziale | Infinite potenze negative non nulle | $e^{1/z}$ |Nessuna delle due condizioni precedenti


**Come trovare tipo: metodi**

1- linearizza in quel punto con serie TL, vedi quale è la prima potenza  
2- fattorizza il denominatore, vedo che molteplicità ha quel polo    
se so fattorizzare D(z), è il miglior modo  
3- studia ordine di N e D    
fallo se non so fattorizare, c'è roba difficile tipo trig o esponenziale al denom    

**Metodo 3: studia ordine di N e D**

N ha uno zero di ordine p  
D ha uno zero di ordine q  
polo ha ordine: q-p    

zero di ordine p: significa che   
$\partial^p N(z_n)=0$   
$\partial^{p+1}N(z_n) \neq 0$ e finito    
cioè, devo arrivare alla p-esima derivata per non avere 0 quando ci pluggo il polo  

nota: se $q \le p$, è una singolarità eliminabile

**Metodo 3: esempio**

1- studio numer  
$N(z_0)=0$   
N ha ordine 0  
 
2- studio denom  
$D(z_0) =0$ e $D'(z_0) \neq 0$ e finito  
D ha ordine 1  

3- conclusione  
il polo ha ordine D-N = 1-0 = 1  

**Metodo 2: formalmente**

calcola il limite: $\lim_{z\rightarrow z_0}(z-z_0)^m f(z)$  
se finito e non-0: ordine m  
se 0: ordine minore di m  
se infinito ($\pm$): ordine maggiore di m  

attenzione:  
$a_{-m} = \lim_{z\rightarrow z_0}(z-z_0)^m f(z)  $  
$a_{-1} = Res[f, z_0] = \frac 1 {m-1!} \frac {d^{m-1}}{dz^{m-1}} [(z-z_0)^m f(z)]\|_{z=z_0}$   
per sapere l'ordine, a me interessa $a_{-m}$, quindi faccio la prima formula, non quella del residuo 

**Nota**

polo semplice = polo di ordine 1  

##### Eliminabile

**Nota**

essenzialmente, è un polo di ordine 0  

**Singolarità eliminabile**

1- quando lo è: se non ci sono potenze negative  
cioè $a_n=0$ se $n<0$  

2- come prolungare in quei punti  
calcola  $\lim_{z \rightarrow z_0}f(z)$ in quel punto, usando la funzione linearizzata   
oppure usa l'Hopital per calcolare il limite (deriva sopra e sotto f(z), se ho F.I.)

cioè: questa sing. si può eliminare prolungando f in quei punti, con un valore che scelgo 

#### 3. Residui

**Residui**

cosa: il coeff $a_{-1}$  

ordine polo | formula
---|---
m | $Res[f, z_0] = \frac 1 {m-1!} \frac {d^{m-1}}{dz^{m-1}} [(z-z_0)^m f(z)]\|_{z=z_0}$  
1 | $Res[f, z_0] = \lim_{z \rightarrow z_0} (z-z_0)f(z)$
2 | $Res[f,z_0] = \frac {d}{dz}[(z-z_0)^2 f(z)]\|_{z=z_0}$

**Esempio** 
  
$f(z) = \frac{e^z}{(z-z_0)^2}$  
$Res(f, z_0)= \frac{d}{dz}e^z|_{z=z_0}= e^{z_0}$  

**Osservazioni**

1- se singolarità eliminabile: Res=0   
perchè non può avere potenze negative  

2- se f è pari, ha solo potenze pari  
quindi $c_{-1}=Res = 0$   

#### 4. Convergenza

 $z_0$ è singolarità? | nome serie | dove converge la serie | formula
---|---|---|---
no, oppure eliminabile | Taylor | in un disco centrato in $z_0$ <br> cioè, in un intorno di $z_0$ | $\|z-z_0\| < R$   
sì (non eliminabile) | TL | in una corona centrata in $z_0$ <br> cioè, in un intorno di $z_0$, tranne nel punto stesso | $r < \|z-z_0\| < R$ <br> con $r =0$ 

**Come scriverla**

1- trova tutti i nodi di $f(z)$    
2- vedi se $z_0$ è singolarità, scegli formula giusta  
3- R è la DISTANZA (norma) tra il nodo su cui sto facendo la serie ($z_0$) e l'altro più vicino   

### Infinito

#### Tipo

**Tipi**

| Tipo | Comportamento quando $\|z\|\to\infty$ 
|---|---|
| Eliminabile |  $f(z) \rightarrow c$  
| Polo di ordine m | $\frac {f(z)}{z^m} \rightarrow c$  <br> quindi, $f(z) \rightarrow \infty$ 
| Essenziale | Nessuna delle due condizioni precedenti  

c: un valore finito non-0, uguale lungo tutti i percorsi   

**Essenziale**

def: non c'è un limite finito comune, e f(z) non tende a 0 lgo tutti i percorsi  
in pratica, per dimostrare che lo sia: mostra che, su due percorsi diversi, ottengo limit diversi  

è spesso essenziale se in f(z) appaiono gli esponenziali  
perchè lungo asse reale o immaginaria pu o decadere/crescere o oscillare, ottengo cose diverse  

**1. Limiti per capire il tipo**

che limiti posso fare:  
1- lungo asse reale f(x) o immaginaria f(iy)   
2- verso $+\infty$ o $-\infty$  

quale scegliere:  
di solito, è sempre essenziale, quindi devo ottenere due risultati diversi  
se ho $e^z$, per fare il limite devo dare in modo che decada/cresca, non oscilli -> sere esponente reale  
quindi scegli asse reale o immaginaria a seconda di csa mi dà esponente reale, e fai a $\pm \infty$ sulla stessa asse  

**Altro modo per capire il tipo**  

se ho scritto serie di TL in z=0:  
fai var change w =1/z, così guardo f(w) =f(1/z)  
il tipo di singolarità di w è quello di $\infty$ (e lo trovo guardando quante potenze negative ci sono)  

attenta se uso questo metodo per calcolare il residuo, la formula è un po' strana, guardala sotto

#### Residuo

**Residuo**

quando: $\infty$ è sing. isolata  
come: Teorema interno/esterno dei residui:   
$Res(f, \infty) -= \sum_j Res(f, z_j)$  
$z_j$: tutte le singolarità finite della funzione (cioè, tutte tranne $\infty$)  

**Infinito è singolarità isolata?**

lo è se:  
f è olomorfa per ogni $|z| >R$  
cioè: al di fuori di un cerchio di raggio R, NON ci sono altre singolarità (eccetto $\infty$)  

hint: se f ha un numero finito si dingolarità, è olomorfa fuori da un cerchio -> è sing. isolata    

nota: qualunque tipo di singolarità infinita può essere isolata o no  
quindi posso avere essenziale isolata ed essenziale non isolata  

**Tip: Punto all'infinito**

voglio studiare la singolarità: $z = \infty$  
fai cambio var: $w = \frac 1 z$    
così studio w=0  

tipo: il tipo di w=0 è lo stesso di $z=\infty$   
residuo:  
$Res[f(z), \infty] = -Res[\frac {f(1/w)} {w^2}, 0]$  
non puoi studiare solo f(0), devi dividere per w^2 (viene dal cambio var del differenziale)
