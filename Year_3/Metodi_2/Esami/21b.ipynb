### Prob 1 - Th residui e log branches

$f(z) = \frac {\log z}{(1+z^2)^2}$  
$\Gamma$: corona superiore, di raggi R->infty e r->0  
calcola integrale su Gamma    

**Gestire log**

esercizio | cosa
---|---
per non intersecare $\Gamma$, scelgo $\arg(z) \in (-\frac \pi 2, \frac {3 \pi}2)$ | taglio di log z  
$\log z = \log \| z\| +i \, arg(z)$ | scrivo log z con determinazione principale (n=0)  

**Calcola integrale**

esercizio | cosa
---|---
denom: $(1+z^2)^2=(z-i)^2(z+i)^2$ <br> poli: $z= \pm i$, entrambi si secondo ordine <br> ma solo z=i è racchiuso in $\Gamma$| cerco poli
$Res(f,i) = \frac {d}{dz}[(z-i)^2 f(z)]\|_{z=i} $ <br> $= \frac d{dz} (\frac {\log z}{(z+i)^2}) = \frac 1 {z(z+1)^2} - \frac{2\log z}{(z+1)^2}$ <br>  $Res[f,i] = \frac 1{i(2i)^2}-\frac{2(i\pi/2)}{(2i)^3} = \frac i4  +\frac \pi 8$|  calcolo residui
$I = -\frac \pi 2 + i \frac{\pi^2}4$ | th dei residui

**Log di numero complesso**

usa la formula sopra: 
$\log z= \log |z| + i arg(z)$  

$i = 1 e^{i \pi/2}$  
$\log i = \log 1 + i \pi = 0 + i \frac \pi 2$

**Calcola altri integrali**

$I_1 = \int_0^\infty \frac {\log x}{(1+x^2)^2} dx$   
$I_1 = \int_0^\infty \frac {1}{(1+x^2)^2} dx$  


esercizio | cosa
---|---
$C_R$: curva di raggio R (esterna) <br> $C_r$: curva di raggio r (interna) <br> $I_-$: asse reale negativa <br> $I_+$: asse reale positiva | separa il contorno in parti dritte e curve
$\|\int_{C_R} f(z) \, dz \|$<br> $ \le \int_{C_R} \frac{\|\log z\|}{\|1+z^2\|^2} \, \|dz\| $ <br> $\le \pi R \frac{\sqrt{(\log R)^2+\pi^2}}{\|R^2-1\|^2}$ <br> $\sim \frac {R}{R^4}\rightarrow 0$ <br> quindi: $\int_{C_R} \rightarrow 0$| curva esterna ha integrale nullo, quando $R \rightarrow \infty$
$\|\int_{C_r} f(z) \, dz \|$<br> $ \le \int_{C_r} \frac{\|\log z\|}{\|1+z^2\|^2} \, \|dz\| $ <br> $\le \pi 2 \frac{\sqrt{(\log 2)^2+\pi^2}}{\|1-r^2\|^2}$ <br> $\sim r \rightarrow 0$ <br> quindi: $\int_{C_r} \rightarrow 0$| curva interna ha integrale nullo, quando $r \rightarrow 0$
$\int_{I-} f(z) dz = \int _{-R}^{-r} \frac{\log \|x\| + i \pi}{(1+x^2)^2} $ <br> $ = \int_r^R \frac{\log u+i \pi}{(1+x^2)^2} du $ <br> $ = I_1+i \pi I_2$| retta reale negativa, z=1x+0iy <br> ma x è negativo in questo integrale, quindi ha arg=$\pi$ <br> poi var change u=-x
$I_+=I_1$ | retta reale positiva
$I_{\Gamma} = I_+ + I_- +C_R+C_r = 2I_1+i \pi I_2 = -\frac \pi 2 + i \frac{\pi^2}4$ <br> $I_1 = \frac 12 Re[I_\Gamma]= -\frac \pi 4$ <br> $ I_2 =\frac 1 \pi Im[I_\Gamma]= \pi 4$| sommo tutto

### Prob 2

**Trasformata di Fourier**

esercizio | cosa
---|---
$F[x(t)] = \tilde x(\omega)$ <br> $F[x'(t)] = (-i \omega) \,  \tilde x(\omega)$ | trasformata di Fourier lato sinsitro  
$F[\theta(t) e^{-\epsilon t}] =\int_R \theta(t) e^{-\epsilon t} \, e^{i \omega t} \, dt $ <br> $= \int_0 ^\infty e^{(i \omega-\epsilon)t} \, dt$ <br> $=[\frac {e^{(i\omega-\epsilon)t}}{i \omega - \epsilon}]^\infty_0 $ <br> $ = \frac 1 {\epsilon - i \omega} = \frac {i}{\omega + i \epsilon}$  | trasformata di Fourier lato destro <br> l'esponenziale a $\infty$ fa 0, perchè la parte reale $e^{-\epsilon t}$ decade, quella immaginaria non mi interessa???
$i(\omega+1) \tilde x(\omega) = \frac{i}{\omega + i \epsilon}$ <br> $ (\omega+1) \tilde x(\omega) = \frac 1 {\omega + i \epsilon}$| insieme   

**Soluzione in dominio freq**

esercizio | cosa
---|---
$ (\omega+1) \tilde x(\omega) = \frac 1 {\omega + i \epsilon}$ | equazione
$\tilde x_p(\omega) = P \frac 1{(\omega +1)(\omega + i \epsilon)}$ | sol particolare <br> semplicemenet inverto, e metto P perchè viola la CE
omogenea: $(\omega+1) \; \tilde x_o(\omega) = 0$ <br> la sol x può essere non-0 solo nel punto dove il primo fattore è 0, cioè in $\omega=-1$ -> è una sol concentrata in $\omega=-1$, una delta<br> $\tilde x_o(\omega) = c \delta(\omega +1)$ | sol omogenea <br> c è un coeff complesso dell'omogenea, $c \in C$
$\tilde x(\omega) = x_p + x_o$ |sol generale

**Antitrasformare**

esercizio | cosa
---|---
$\tilde x_p(\omega) = \frac 1 {1+i \epsilon}[\frac i {\omega + i \epsilon}-P\frac 1 {\omega+1}]$ <br> $\tilde x(\omega) = \frac 1 {1+i \epsilon}[\frac i {\omega + i \epsilon}-P\frac 1 {\omega+1}] +c\delta(\omega+1)$| sol in freq domain, da antitrasformare <br> riscrivo la sol particolare in modo più facile
$F^{-1}[\frac 1 {\omega + i \epsilon}] = -i \theta(t) e^{-\epsilon t}$ (inversa dell'iniziale) <br> $F^{-1}[\delta(\omega)] = \frac 1 {2 \pi}$ <br> $F^{-1}[P\frac 1 \omega] = \frac 1 {2i} sign(t)$|trasformate note
$F^{-1}[\delta(\omega+1)] = \frac{1}{2 \pi}e^{it} $<br> $F^{-1}[P\frac 1 {\omega+1}] = \frac 1 {2i} sign(t) \, e^{it}$| usando la traslazione  
$x(t) = \frac 1{1 -i \epsilon}[F^{-1}(\frac 1 {\omega+i\epsilon})-F^{-1}[\frac 1 {\omega+1}]]+c F^{-1}[\delta(\omega+1)]$ <br> $ = \frac 1{1 -i \epsilon} [-i \theta(t) e^{-\epsilon t}+\frac 12 sign(t) e^{it}] + \frac c{2 \pi} e^{it}$| tutto insieme
$x(t) = \frac 1{1 -i \epsilon} [e^{it}-e^{-\epsilon t}] \theta(t)+ [\frac c{2 \pi} -\frac i {2(1-i\epsilon)}]e^{it}$ | passa da sign a $\theta$
devo avere $\frac c{2 \pi} -\frac i {2(1-i\epsilon)}=0$ <br> perchè il primo addendo, grazie alla theta, è già 0 per t<0 <br> quindi scelgo c apposta| quando è causale, cioè x(t)=0 per t<0
