### Prob 1 - Serie

$S(z)=\sum_{n=0}^\infty (\frac z4)^n + \sum_{n=0}^\infty (\frac 4{z^2})^n$

**Convergenza**

esercizio | cosa
---|---
converge se: $\frac {\|z\|} 4 < 1 \quad \rightarrow \quad \|z \|  <4$ <br> converge a: $\frac 1{1-z/4} = \frac{4}{4z}$| prima serie
converge se: $\frac 4{\| z^2\|}  < 1 \quad \rightarrow \quad \|z \| >2$ <br> converge a: $\frac 1{1-4/z^2} = \frac{z^2}{z^2-4}$| seconda serie
converge se: $2 < \|z\| < 4$ <br> converge a: $S(x) = \frac{4}{4-z} +  \frac{z^2}{z^2-4} = -\frac 4 {z-4} +  \frac{z^2}{(z-2)(z+2)} $ | serie totale

**Singolarità**

esercizio | cosa
---|---
poli: $z=4, \pm 2$ <br> tutti di ordine 1, perchè hanno molteplicità 1 al denom | trova poli
$Res[S, 4] \lim_{z \rightarrow 4} (z-4)S(z) = -4+0 = -4$ <br> $Res[S, 2]|=0+1$ <br> $Res[S,-2]=-1$ | residui

**Singolarità all'infinito**

esercizio | cosa
---|---
$S(x) = \frac{4}{4-z} +  \frac{z^2}{z^2-4} $ <br> $\lim_{z \rightarrow \infty}S(z) = 1+0=1$ | limite (non ho dubbi su roba che oscilla)
è una singolarità reimovibile, perchè ha limite finito e uguale in tutti i percorsi| conclusione
$Res[S, \infty] = -[-4+1-1]=4$| residuo

**Serie geom**

$\sum_{n=0}^\infty q^n$  
converge se: $|q|<1$  
converge a: $\frac 1{1-q}$ 

somma parziale (formula più generale):  
$\sum_{n=0}^N q^n = \frac {1-q^{N+1}}{1-q}$  

deve sempre partire da n=0

### Prob 2 - eq diff con Fourier

$x'+x=\theta(t)$  

risolvi con trasformata di Fourier

esercizio | cosa
---|---
$(-iw) \tilde x(\omega) +\tilde x(\omega) = iP\frac 1 \omega +\pi \delta(\omega)$ <br> $(-i\omega +1)\tilde x(\omega) = iP\frac 1 \omega +\pi \delta(\omega)$ | trasformata   
$\tilde x(\omega) = \frac 1 {1-i \omega}P \frac 1 \omega +\frac \pi{1-i \omega} \delta(\omega)$ |sol particolare
$ \tilde x(\omega)  =-\frac 1 {\omega +i} P \frac 1 \omega +\pi \delta(\omega)$ | proprietà della delta e razionalizzo
$\tilde x(\omega) = i P \frac 1 \omega -\frac i {\omega+i} +\pi \delta(\omega)$ | scomposizione $\frac 1{\omega(\omega +1)}= \frac 1 i (\frac 1 \omega -\frac 1 {\omega+i})$  

**Anti-trasformo**

esercizio | cosa
---|---
$x(t) = i F^{-1}[P\frac 1 \omega]-i F^{-1}[\frac  1{\omega+i}] +\pi F^{-1}[\delta(\omega)]$ |linearità
$F^{-1}[P \frac 1 \omega] = \frac 1 {2i} sign(t) $ <br> $F^{-1}[\delta(\omega)] = \frac 1 {2 \pi}$ <br> $sign(t)= 2 \theta(t)-1$| note
$x(t) = \theta(t) -F^{-1}[\frac 1 {\omega+i}]$| dove sono arrivata
$F^{-1}[\frac 1 {\omega+i}] =\int_R  \frac 1 {\omega+i} e^{-i \omega t} d\omega$ | difficile, usa formula di anti-trasformata
polo in $\omega=i$ <br> integrale decade su piano positvo se t<0, negativo se t>0 <br> usa lemma di Jordan per dire che la curva ha integrale nullo | risolvi integrale
$F^{-1}[\frac 1 {\omega+i}] = \begin{cases} 0 & t<0 \\ -ie^{-t} & t>0 \end{cases} $ <br> $F^{-1}[\frac 1 {\omega+i}] = \theta(t) \, -ie^{-t}$ | soluzione anti-trasformata
$x(t) = \theta- \theta F^{-1}[\frac 1 {\omega+i}] = \theta(t) (1- e^{-t})$| sol particolare in time domain
$x'+x=0$ <br> $x_(t)=Ae^{-t}$ | sol omogenea (dall'eq omogenea in time domain)
$x(t) = \theta(t) (1- e^{-t}) + Ae^{-t}$<br> vedo che $A=x(0)$ | sol totale
