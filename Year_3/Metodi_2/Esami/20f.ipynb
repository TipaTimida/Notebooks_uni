### Prob 1

$f(z)=\frac {e-e^{iz}}{z+ia}$

**Serie di TL**

nella singolaritò: $z=-ia$  
vuole l'intera serie, non solo due termini, quindi quando linearizzo devo usare l'intera serie  

esercizio | cosa
---|---
$h= z+ia \rightarrow 0$ <br> $iz = ih+a$ <br> $f(h) = \frac {e -e^{ih+a}}{h} = \frac {e-e^{a}e^{ih}}h$ | cambio var 
$e^h = \sum_0^\infty \frac{h^n i^n}{n!}$ <br>$ =1 +  \sum_1^\infty \frac{h^n i^n}{n!}$|  linearizzo e (con serie intera)
$f(h) = \frac {e-e^a}h -e^a \sum_1 \frac {i^nh^{n-1}}{n!} $ <br> $ = \frac {e-e^a}h -e^a \sum_0 \frac {i^{k-1} h^{k}}{(k-1)!} $| tutto insieme
$f(z+ia) = \frac {e-e^a}{z+ia} -e^a\sum_0 \frac {i^{k-1} (z+ia)^{k}}{(k-1)!}$| torno a var originale
$0 < \|z+ia\| < \infty$ <br> perchè non ci sono altre singolarità in f | convergenza 
è un polo di ordine 1 (perchè il minore è $c_{-1}$)| polo
$Res[f,-ia] = c_{-1} = e-e^a$| residuo

esercizio | cosa
---|---
intera quando non ha potenze negative <br> cioè $a_{-1}=e-e^a=0$ <br> cioè $a=1+i 2 \pi k \quad k \in Z$ | intera
$f(z=-ia)=f(h=0) = c_0 = -ie^a=-ie$ COME???| calcola limite in quel punto
$f(z=-ia)= \lim_{z \rightarrow -ia} \frac{e-e^{iz}}{z+ia}= \lim \frac{-ie^{iz}}1 = -ie^a=-ie$| calcola limite con l'Hopital

**Punto all'infinito**

esercizio | cosa
---|---
$\lim_{i \rightarrow \infty} f(iy) = \frac{e-e^{-y}}{iy+ia}= \sim e^{-y} = 0$  <br> uso asse immaginario così exp ha esponente reale, decade invece di oscillare| limite su asse immaginaro positivo
$\lim_{i \rightarrow -\infty} f(iy) = \frac{e-e^{-y}}{iy+ia} \sim  e^{-y}= \infty$  | limite su asse immaginario negativo
sono diversi -> singolarità essenziale 
$Res[f, \infty] = -Res[f, -ia] = e^a-e$| residuo

### Prob 2 - Green

$\tilde G_\epsilon(\omega) = i\frac{e-e^{i \omega}}{\omega+i \epsilon}$  

**Causale: Titchmarsh**

esercizio | cosa
---|---
$\omega = x +iy$ <br> $\tilde G(\omega) = i \frac{e-e^{ix}e^{-y}}{x+i(y +\epsilon)}$| scrivo $\omega$ nel piano complesso
$\| \tilde G(\omega) \|^2 = \frac{\|e-e^{i \omega}\|^2}{\omega^2+\epsilon^2}$ | calcolo norma
$\omega = -i \epsilon$ <br> quindi non ce n'è sul piano superiore <br> quindi sul piano superiore G è analitica, come richiesto| analitica
mostra che norma sul semipiano superireo non è infinita ???


esercizio ALTRO??? | cosa
---|---
$\|G\|^2  \le \frac{\|e-e^{iu}e^{-v}\|}{u^2+(v+\epsilon)^2} \le   \frac{(e+1)^2}{v^2+\epsilon^2}$ <br> quindi: analitica nel semipiano superiore | sul semipiano superiore, cioè se v>0
$\int _{-\infty}^{\infty} \|G(u+iv)\| dv < \infty$ | sul semipiano superiore, cioè se v>0
$\|G\|^2  \le \frac{(e-1)^2}{\omega^2+\epsilon^2}$ <br> quindi: $\tilde G \in L^2(R)$ | su asse reale, cioè se $\omega \in R$ (NON SERVE???)


**Antitrasformata**

esecizio | cosa
---|---
$G(t) = F^{-1}[\tilde G(\omega)] = ie F^{-1}[\frac 1 {\omega +  \epsilon}] -i F^{-1}[\frac {e^{i \omega}} {\omega +  \epsilon}]$| formula antitrasforamta (uso linearità)
$= ie F^{-1}[\frac 1 {\omega +  \epsilon}](t) -i F^{-1}[\frac {1} {\omega +  \epsilon}](t-1)$ | th di traslazione
$F^{-1}[\frac 1 {\omega +  \epsilon}](t) = \frac 1 {2 \pi} \int_R \frac 1 {\omega +  \epsilon} e^{-i \omega t} d \omega$  | antitrasformata dei pezzi
ha solo un polo in $\omega = -i \epsilon$ <br> $Res[f, -i \epsilon] = \lim (\omega+ \epsilon) \, \frac 1 {\omega +  \epsilon} e^{-i \omega t} = \lim e^{-i \omega t} = e^{-\epsilon t}$ | th dei residui
se t<0: decade sul piano superiore <br>  chiudo il cerchio lì, dove on ci sono residui <br>  $\int = 0$ | caso t<0 
se t>0: decade sul piano inferiore <br> dove c'è un residuo <br> $\frac 1 {2 \pi} \int f(\omega) \d \omega =-2 \pi i Res[f, -i \epsilon] $ <br> $\int = \frac{-2\pi i}{2 \pi} e^{-\epsilon t} = -i e^{-\epsilon t}$ | caso t>0
$F^{-1}[\frac 1 {\omega +  \epsilon}] = \begin{cases} 0 & t<0 \\ -ie^{-\epsilon t} & t>0 \end{cases}$| soluzione 
$F^{-1}[\frac 1 {\omega +  \epsilon}] = -i \theta(t) e^{-\epsilon t}$| passo a $\theta$
$G(t) = e\theta(t)e^{-\epsilon t} -\theta(t-1) e^{-\epsilon(t-1)}$| G totale
$\lim _{\epsilon \rightarrow 0} G(t) = e\theta(t)-\theta(t-1)$| limite 

**Limite, e poi antitrasforma**

esercizio | cosa
---|---
$\lim _{\epsilon \rightarrow 0^+} \frac 1 {\omega + i \epsilon} = P \frac 1 \omega -i \pi \delta(\omega)$| usa il limite notevole
$\lim_{\epsilon \rightarrow 0^+} \tilde G(\omega) = \lim_{\epsilon \rightarrow 0^+} i \frac{e-e^{i \omega}}{\omega + i \epsilon}$ <br> $=(e-e^{i \omega}) [iP \frac 1 \omega + \pi \delta(\omega)]$ <br> $ = F[e \theta(t)-\theta(t-1)]$| calcola il limite
$G(t) = e \theta(t)-\theta(t-1)$ <br> come visto dallo scorso punto| conclusione
