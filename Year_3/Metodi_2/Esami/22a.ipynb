### Prob 1

$K(a) = \lim_{R \rightarrow \infty} \oint_\Gamma \frac {e^{az}}{1+e^z} \, dz$   
$a \in R$  

$\Gamma_R$ è il rettangolo $\{z=i+iy \; | \; -R \le x \le R, \quad 0 \le y \le 2 \pi\}$, in sens orario

**1 - risolvi con th residui**

esercizio | cosa
---|---
$f(z) = \frac {e^{az}}{1+e^z}$| passa a piano complesso <br> esponenziale è reale, quindi NON è oscillante 
$e^{z_n} =-1 = e^{i (2n+1)\pi} \quad n \in Z$ <br> $z_n = i \pi (2n+1) \quad n \in Z$ | trova poli
$D(z_n)=0 \quad D'(z_n) =-1 \neq 0 \quad N(z_n) \neq 0$ <br> quindi, ordine 1 <br> $Res(f, z_n) = \frac{N(z_n)}{D'(z_n)}=-e^{ia\pi(2n+1)}$| ordine dei poli
$\oint_{\Gamma_R} f(z) \, dz = 2 \pi Res(f, z_0)$  <br> $ = 2 \pi i (-e^{ia \pi})$ <br> perchè $z_0$ è l'unico polo contenuto dentro il rettagolo (disegnalo) | scrivo th residui
