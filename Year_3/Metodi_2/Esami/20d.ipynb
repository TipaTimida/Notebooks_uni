### Prob 1

$f(z) = \frac{e^{iz}-a}{z^2+1}$  
$a \in C$  

**Punti singolari isolati**

esercizio | cosa
---|---
$f(z) = \frac{e^{iz}-a}{(z-i)(z+i)}$ <br> poli: $z=\pm i$, di primo ordine | poli
$Res[f, i] = \lim_{z \rightarrow i} \frac {e^{iz}-a}{z+i} = \frac {e^-{1}-a}{2i}$ <br> se $a=e^{-1}$, è eliminabile <br> $Res[f, i] = \lim_{z \rightarrow i} \frac {e^{iz}-a}{z-i} = \frac {e^-{1}-a}{-2i}$ <br> se $a=e$, è eliminabile <br>  |residui 

**Singolarità infinito**

esercizio | cosa
---|---
devo farlo decadere, non oscillare, quindi studio asse immaginaria <br> $f(iy)= \frac{e^{-y}-a}{-y^2+1}$ | cosa fare
$\lim_{y \rightarrow \infty} f(iy) = 0$ <br> $\lim_{y \rightarrow \infty} f(iy) = - \infty$ <br> sono diversi -> singolarità essenziale 
$Res[f, \infty] = -Res[f,i]-Res[f, -i] = \frac{e-a}{2i}-\frac{e^{-1}-a}{2i} = \frac {e-e^{-1}}{2i}$| residuo

**Serie di TL a z=0**

esercizio | cosa
---|---
$e^{iz} = 1 +(iz)+ \frac 12 (iz)^2+ o(z^3)$ <br> $ = 1+iz-\frac {z^2}2$ | linearizza numeratore  
$\frac 1 {1+z^2}= 1-z^2+o(z^4)$| linearizza denominatore 
$f(x) = (1-a + iz - \frac {z^2}2+ o(z^3)) (1-z^2+o(z^4))$ <br> $ = (1-a) + iz + (a-\frac 32)z^2 + o(z^3)$| insieme 
$\|z\| < 1$ <br> perchè i poli sono $z= \pm i$, e z=0 non dà problemi, quindi non devo escluderlo| convergenza
