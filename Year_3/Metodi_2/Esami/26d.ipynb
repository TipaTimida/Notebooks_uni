### Prob 2

$\frac {db}{dt} +ikb(t) =a(t)-a(-1)$  
k reale

**1. Trova funzione di Green in freq domain**

esercizio | cosa
---|---
$a(t) = \delta(t)$ <br> $b(t) = G(t)$| setup funzione di Green
$F[G']+ikF[G] = F[\delta] -F[\delta(t-1)]$ <br> $(-i \omega) \tilde G + ik \tilde G = 1 - e^{i \omega} $  <br> $i(k- \omega) \tilde G(\omega) = 1- e^{i \omega}$| trasformo
$\tilde G_p (\omega) = i(1-e^{i \omega}) P \frac 1 {\omega -k}$ | sol particolare
$(k- \omega) \tilde G_h(\omega) $ <br> $\tilde G_h(\omega) = \tilde c \; \delta (\omega-k)$| sol generale (setto $a(t)=0$, cioè tolgo parte a destra)
$\tilde G(\omega) = i(1-e^{i \omega}) P \frac 1 {\omega -k} +\tilde c \; \delta (\omega-k)$  | sol generale
$G(t) = i F^{-1}[P \frac 1 {\omega -k}](t) -  F^{-1}[P \frac 1 {\omega -k}](t-1)+ \tilde c F^{-1} [\delta(\omega-k)]$ <br> $F^{-1}[P \frac 1 {\omega -k}] = e^{-ikt} F^{-1}[P \frac 1 \omega] = e^{-ikt} (-\frac i2 sgn(t))$ <br> $F^{-1}[\delta-k] = e^{-ikt}F^{-1}[\delta] = \frac 1 {2 \pi} e^{-ikt} \quad \frac{\tilde c}{2 \pi} = c$ | trasformo
$G(t) = \frac 12 e^{-ikt} sgn(t) - \frac 12 e^{-ik(t-1)}sgn(t-1) + c e^{i kt}$ |  soluzione

**2. Trova G in time domain**

esercizio | cosa
---|--
se $t \neq 0,1$: equazione omogenea <br> $G'+ikG =0$ <br> $ G(t) = \begin{cases} Ae^{-ikt} & t<0 \\ Be^{-ikt} & 0<t<1 \\ De^{-ikt} & t> 1 \end{cases}$| omogenea
al punto t=0: $\epsilon \rightarrow 0$ <br> $\int_{-\epsilon}^\epsilon G'(t) \,  dt + ik \int_{-\epsilon}^\epsilon G(t) \, dt = \int_{-\epsilon}^\epsilon  \delta(t) \, dt - \int_{-\epsilon}^\epsilon \delta(t-1) \, dt$ <br> $ [G(\epsilon)-G(-\epsilon)] + 0 = 1 -0 $ <br> $G(0^+)-G(0^-) = 1$ <br> ultimo integrale fa 0, perhè quella delta è non-0 solo in t=1| prima continuità: integro l'eq diff attorno a t=0
al punto t=1: $\epsilon \rightarrow 1$ <br> $\int_{-\epsilon}^\epsilon G'(t) \,  dt + ik \int_{-\epsilon}^\epsilon G(t) \, dt = \int_{-\epsilon}^\epsilon  \delta(t) \, dt - \int_{-\epsilon}^\epsilon \delta(t-1) \, dt$ <br> $ [G(\epsilon)-G(-\epsilon)] + 0 = 0 -1 $ <br> $G(1^+)-G(1^-) = -1$ | prima continuità: integro attorno a t=1
$G(0^+)-G(0^-) = 1$ <br> $G(1^+)-G(1^-) = -1$  | | recap: integrando l'e attraverso i punti di salto, ho ottenuto le condizioni
condizione t=0: $B-A = 1$ <br> condizione t=1: $(D-B) e^{-ik} = -1$ <br> quindi: $D=B_e^{ik} = A+1-e^{ik}$ | risolvo condizioni
$ G(t) = \begin{cases} Ae^{-ikt} & t<0 \\ (A+1)e^{-ikt} & 0<t<1 \\ (A+1-e^{ik})e^{-ikt} & t> 1 \end{cases}$ <br> i valori nei singoli 0,1 non cambiano la distribuzione|  soluzione  

**3. Proprietà di G**

esercizio | cosa
---|---
è causale se A=0, così G=0 quando t<0 | causalità
devo annullarla negli intervalli ifniniti, cioè t<0 e t >1 <br> cioè, oltre ad A=0, serve $1-e^{ik}=0$ <br> quindi, $k= 2 \pi n \quad n \in Z$ <br> quindi: <br> $ G(t) = \begin{cases} e^{-ikt} & 0<t<1 \\ 0 & else \end{cases}$ | supporto compatto
se G a supporto compatto: la sua trasformata (in freq domain) si può estendere a una funzione intera <br> $\tilde G(\omega) = \int_0^1 e^{-ikt} e^{i \omega t} dt $ <br> $ = \int_0^1 e^{i(\omega-k) t} dt$ <br> $ = \frac{e^{i(\omega-k)-1}}{i(\omega-k)} = \frac{e^{i \omega}-1}{i(\omega-k)}$ <br> in $\omega = k$ ho singolarità eliminabile, quindi la trasformata $\tilde G(\omega)$ è una funzione intera | proprietà di G a supporto compatto
