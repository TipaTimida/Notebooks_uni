### Prob 1

$I = \int_R \frac {\sin x}{x(x-i)}dx$  

esercizio | cosa
---|---
$I = P \int_R \frac{e^{iz}-e^{-iz}}{2i \, z (z-i)} dz$ <br> $ = \frac 1{2i} P\int_R \frac{e^{iz}}{z (z-i)} dz - \frac 1 {2i}P\int_R \frac{e^{-iz}}{z (z-i)} dz$ <br> $=\frac 1{2i}[I_1-I_2]$ | separo
$e^{iz} = e^{ix-y}$ <br> $I_1$ decade sul semipiano superiore, $I_2$ su quello inferiore | dove decade 
$g= \frac{e^{iz}}{z(z-i)}$ <br> poli: $z=0, \; z=i$ <br> $Res[g,0]=\frac 1{-i}$ <br> $Res[g, i] = \frac 1 {ie}$ | poli
$\frac{1}{\|z(z-i)\|} \le \frac 1{R \|R-1\|}\rightarrow 0$ <bR> quindi integrale sul semicerchio è 0 | lemma di Jordan 
$I_1 = 2 \pi i (Res[g, 0]+ Res[g, i]) $ <br> $ =\frac {2\pi}e$ |th residui
$I_2 = - \pi i Res[g,0] = \pi$ | th residui
$I =\frac 1{2i}[I_1-I_2]$ | metto insieme (formula iniziale)

### Prob 2

$\dot x + ix = e^{-i a t}$  
$a \in R$  

**Trasformata di Fourier**

esercizio | cosa
---|---
$F[x]=\tilde x (\omega)$ <br> $F^[\dot x]=(-i \omega) \tilde x(\omega)$ <br> $F[e^{-iat}](\omega) = F[1](\omega-a) = 2 \pi \delta(\omega-a)$ | pezzi
$(-i \omega)\tilde x(\omega) + i \tilde x(\omega) = 2 \pi \delta(\omega-a)$ <br> $(\omega-1) \tilde x(\omega) = 2\pi i \delta(\omega-a)$| intero 

**Risolvi**

esercizio | cosa
---|---
proprietà: $f(\omega) \delta(\omega-a) =f(a) \delta(\omega-a)$ <br> perchè è non-0 solo se $\omega=a$| usare delta 
ansatz: $\tilde x_p(\omega)= k \delta(\omega-a)$ <br> voglio trovare k| sol particolare, PERCHÈ???
$(\omega-1) [k \delta(\omega-a)] = 2\pi i \delta(\omega-a)$ <br> $(a-1) [k \delta(\omega-a)] = 2\pi i \delta(\omega-a)$ <br> $k= \frac {2\pi i}{a-1}$| plugga ansatz
$(\omega-1) \tilde x(\omega)=0$ <br> $\tilde x_o(\omega) = \tilde c \delta(\omega-1)$ <br> può essere non-0 solo se l'altro fattore $\omega-1=0$| sol omogenea
$\tilde x(\omega) = \tilde x_p(\omega) +\tilde x_o(\omega)  | totale

**Antitrasformata**

esercizio | cosa
---|---
$F^{-1}[\delta(\omega -a)] = \frac 1 {2 \pi} e^{-i at}$ |noti (inverso di quello iniziale)  
$x(t) = \frac 1{a-1}e^{-iat}+ \frac {\tilde c}{2 \pi}e^{-it}$ <br> $ \frac 1{a-1}e^{-iat}+ ce^{-it}$ |insieme <br> la costante arbitraria c ingloba anche $2 \pi$

**Caso: a=1**

esercizio | cosa
---|---
$\tilde x_p$ non va bene, k viola la CE per a=1 | problema
proprietà delta: $(\omega-1)\delta'(\omega-1) = -\delta(\omega-1)$
$(\omega-1)\tilde x(\omega) = 2 \pi i \delta(\omega-1) $ <br> $ = -2 \pi i(\omega-1) \delta'(\omega-1)$ <br> $\tilde x_p (\omega) = -2 \pi i \delta'(\omega-1)$ | sol particolare
$F^{-1}[\delta'(\omega-1)]= itF^{-1}[\delta(\omega-1)]$| antitrasformo
