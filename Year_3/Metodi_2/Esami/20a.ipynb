### Prob 2 - Green function

$RI'+\frac 1C I + V'$  
V: inpt  
I: output  

**Trova funzione Green**

se non specifica, basta la particolare  

esercizio | cosa
---|---
$R(-i \omega)\tilde I(\omega) + \frac 1 C \tilde I(\omega) = (-i \omega) \tilde V(\omega)$ <br> $(-i \omega R + \frac 1 C)\tilde I(\omega) = -i \omega \tilde V(\omega)$ | trasforma 
$\tilde I(\omega) = \tilde G(\omega) \tilde V(\omega)$ <br> $\tilde G(\omega) = \frac 1 R -\frac 1{R^2C}  \frac 1 {RC- i\omega}$ | trova funzione di Green (particolare) <br> uguale a mettere $V(t)=\delta(t)$ e $I(t) = G(t)$ 
$G(t) = \frac 1R F^{-1}[1] -\frac 1 {R^2C} F^{-1}[ \frac 1 {RC- i\omega}] $ <br> $ = \frac 1R \delta(t) - \frac 1 {R^2C} \, \theta(t) e^{-t/RC}$ | antitrasforma
