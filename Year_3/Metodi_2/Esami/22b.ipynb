### Prob 1

$f(z) = \frac z {\sin^2 z}$  

**Punto 1 - singolarità isolati**

esercizio | cosa
---|---
$\sin(z_n) = 0 \rightarrow \quad z_n = k \pi \quad k \in Z $ | trova singolarità  
$Res(f,0) = \lim_{z\rightarrow 0}(z-0) \frac{z}{\sin^2(z)} = \lim_{z\rightarrow 0} \frac{z^2}{(z+o(z^3))^2} = \lim_{z\rightarrow 0} \frac{z^2}{z^2+O(z^4)} = 1$ <br> finito e non-0, quindi polo di ordine 1 <br> si vede anche: $f(z) \sim \frac z {z^2} \sim \frac 1 {z}$ | caso z=0  
$\lim_{z\rightarrow k \pi}(z-k \pi)^2 \frac{z}{\sin^2(z)} = \lim_{z\rightarrow k \pi}(z-k \pi)^2 \frac{z}{(z-k\pi)^2 +O(z-k\pi)^4} = \lim _{z \rightarrow 4\pi} \frac{z}{1+O(z-k \pi)^2} = k \pi$   <br> finito e non-0 (se k non-0, quindi il caso z=0 è uno speciale), quindi poli di ordine 2 | caso: $z= k \pi$  
ovviamente, $z=\infty$ è un punto singolare, ma non isolato

### Prob 2

$\tilde G_n(\omega) = i^n \frac{e-e^{i\omega}}{(\omega+i)^n}$

**1. Antitrasformata di Fourier**  

so $\tilde G_n(\omega)$, voglio $G_n(t)$

esercizio | cosa
---|---
$\tilde G(\omega) = i^n e\frac{1}{(\omega+i)^n} - i^n \frac{e^{i \omega}}{(\omega+i)^n}$ <br> $G(t) = i^n eF^{-1}[\frac{1}{(\omega+i)^n}] - i^n F^{-1}[\frac{e^{i \omega}}{(\omega+i)^n}]$ <br> $ = i^n eF^{-1}[\frac{1}{(\omega+i)^n}](t) - i^n F^{-1}[\frac{1}{(\omega+i)^n}](t-1)$ <br> $ = i^n ef(t) - i^n f(t-1)$ | linearità e time shift
$f(t) = \frac{1}{2\pi} \int_R \frac{e^{-i \omega t}}{(\omega+i)^n} d \omega$  | scrivo antitrasformazione  
$g(z) = \frac{e^{-izt}}{(z+i)^n}$ <br> ha un solo polo, in $z=-i$, di ordine n | passo a funzione complessa ($\omega \rightarrow z$)
$Res(f, -i) = \frac 1{(n-1)!} [\frac{d^{n-1}}{dz^{n-1}} e^{-izt}]_{z=-i}$  <br> $= \frac 1{(n-1)!}  (-it)^{n-1} e^{-izt}\|_{z=-i}$  <br> $ = \frac  {(-i)^{n-1}}{(n-1)!} t^{n-1} e^{-t}$ | residuo del polo di ordine n
$e^{-izt} = e^{-ixt}e^{yt}$ <br> se t>0: decade se y<0, semipiano inferiore <br> se t<0: secade se y>0, semipiano superiore | in quale semipiano f(t) decade  
per $\|z\|=R$ <br> $\frac 1 {\|z+i\|^n}  \le \frac1 {(R-1)^n}\rightarrow 0$ | tratto di arco è nullo (lemma di Jordan)
non ha poli all'interno: <br> $\int_R g(z) \, dz  = 0$ <br> quindi $f(t)=0$| caso t<0  
c'e dentro il polo <br> $\int_{-\infty}^\infty g(z) d\omega = -2 \pi i Res(f, -i)$ <br> $f(t) = \frac 1 {2 \pi} \int g(\omega) \, d\omega$ | caso t>0 
$f(t) = \begin{cases} 0 & t<0 \\ \frac {(-i)^{n-1}}{(n-1)!}t^{n-1}e^{-t} & t>0 \end{cases}$  | sol trasforamta
$G(t) = \frac 1{(n-1)!} e^{-(t-1)} [t^{n-1}\theta(t) -(t-1)^{n-1} \theta(t-1)]$| sol totale

**Domande**

1- causale   
condizione: $G(t)=0$ per $t<0$   
vera se: $n \ge 1$  

2- a supporto compatto   
devo mettere G=0 nel tratto infinito t>1, il che accade per n=1  

**Punto 2 - trova output, usando funzione di Green**

situazione | esercizio
---|---
caso n=1 <br> input: $a(t)=\delta(t-a)$ | setup
$b(t) =(G*a)(t)$ <br> $\tilde b(\omega) = \tilde G\omega) \tilde a(\omega)$ | scrivi output
$\tilde G_1(\omega) \;$: dal testo <br> $\tilde a(\omega) = F[\delta(t-a)] = e^{i \omega a} F[\delta t] =e^{i \omega a}$ |trasformate Fouier
$\tilde b(\omega) = e^{i \omega a}\tilde G_1(\omega)$| calcolo output 
$b(t) = F^{-1}[\tilde G_1(\omega) e^{i \omega a}](t) =  F^{-1}[\tilde G_1(\omega)](t-a) = G_1(t-a)$ | anti-trasformo output
