### Derivata complessa

#### Definizioni

**Funzione complessa vs reale**

continuità, derivabilità... si trovano avvicinandosi al punto con un limite: $\lim_{h \rightarrow 0}f(z_0+h)$  

differenza: come h può avvicinarsi a zero:  
1- su R: lungo la retta reale, cioè solo da destra o sinistra  
2- su C: da qualsiasi direzione nel piano  

il limite su C esiste se è indipendente dalla direzione di h   
(cioè ottengo lo stesso risultato qualunque sia la direzione del vettore h)  

#### Derivata complessa

**Analitica e olomorfa (in C)**

cosa | olomorfa | analitica
---|---|---
viewpoint | derivata | serie di potenze
cosa esiste |  derivata <br> $f'(z) = \lim_{h \rightarrow 0} \frac{f(x+h)-f(z)}h < \infty$  | serie di potenze convergente (serie di Taylor) <br> $f(z) = \sum_{n=0}^\infty a_n (z-z_0)^n < \infty$ 


analitica = derivabile in senso complesso

**Teorema**

su C: olomorfa = analitica

poly e sin sono analitici in tutto C

#### Cauchy-Riemann

**Pratica**

per fare le derivate, posso farle direttamente su z  
non preoccuparti di passare a z=x+iy  

**COndizioni di Cauchy-Riemann**

$f_x+if_y = 0$  
nel punto $z_0=(z_0,y_0)$  

f analitica = rispetta ondizioni CR   

nota che: rispetta CR -> è derivabile in senso reale (perchè devo avere $f_x, f_y$ per scriverle)

**Altro modo di scrivere CR**

f ha una derivata complessa:  
$f(z) =u(x,y)+iv(x,y)$  
$f'(z) = u_x+iv_x=v_y-iu_y$  

quindi esiste se:  
u e v sono differenziabili, e soddisfano le equazioni di Cauchy-Riemann  
$u_x=v_y$  
$u_y=-v_y$  

**CR in coords polari**

$\frac{\partial f{\partial r}} +\frac 1r \frac{\partial f}{\partial \varphi}=0$  
come si trovano: fai cambio var, a var polari

### Logaritmo complesso

**Formula**

all branches:  
$\ln z = \ln(re^{i(\theta+2n \pi)}) = \ln r + i(\theta+2n\pi)$  

ramo principale (n=0):  
$\ln z = \ln(re^{i \theta}) = \ln r + i \theta$  
dove: $r=|z| \quad \theta= arg(z)$  


**Branches**

1- scegli il taglio di log z  
cioè: scegli un dominio di lunghezza $2 \pi$ a cui appartiene $\Theta$  
eg. $\Theta \in [-\pi, \pi]$, oppure $\Theta \in [-\frac \pi 2, \frac {3 \pi}2]$  
come sceglierlo: fai in modo che non taglio l'area su cui sto facendo l'integrale

2- scrivi determinazione principale  
cioè n=0

a cosa serve scegliere il branch:  
quando faccio il limite, devo sapere a che arg tende  

#### Branches

**Perchè branches**

perchè log è una funzione multiple-valued  
quindi devo scegliere un preciso output tra quelli possibili  

graficamente: il plot è un'elica infinita  
quindi ogni punto sul piano C ha infiniti punti a cui viene mappata dalla funzione

**Branch**

f: multiple-valued function  
F: branch, a single-valued function, analytics  
at each point, F(z) is one of the possible resulst of f(z)  

branch cut: a portion of a line or curve that is introduced in order to define a branch F   
singular points for F: points on the branch cut for F  
branch points: points that appear in all branches
