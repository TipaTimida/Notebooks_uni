### Teoria

**Cos'è**

1- Green function: a description of what a system does after a single, idealized, very brief input at one place and time  
so model how it responds to a brief impulse  

2- convolution: calculation that combines many shifted copies of that simple response   
so imagine the input signal as a serie of impuelses, at different times (so shift in time)    
combine the response to each impulse to have the reponse to the full input signal

### Teoria: Convoluzione vs moltiplicazione

convolution can become multiplication (and vice-versa), if used Fourier transform

**Convolution**

$(g * f) (s) = \int_R g(u) f(s-u) \, du $  


cosa | time | come diventa in freq
---|---|---
time conv -> freq mult| $x(t) = (g * f) (t)$ | $\tilde x(\omega) = \tilde  g(\omega) \, \tilde f(\omega)$   
freq conv -> time mult | $x(t) =g(t) f(t)$ | $\tilde x(\omega) = \frac 1 {2 \pi} (\tilde g * \tilde f)(\omega)$   

**Don't confuse: Chaining vs convolution**

chaining:  
$(f∘g)(t)=f(g(t))$    
used to change the thing on the x axis  
if know a as func of b and b as func  of c, want a as func of c  

convolution:  
$ (f*g)(t)=\int_{-\infty}^{\infty} f(\tau)\,g(t-\tau)\,d\tau$   
in a circuit:  
f: input signal  
g: response to one impulse  
i'm consider the input signal as a sum of impulses, and adding the response to all of them
