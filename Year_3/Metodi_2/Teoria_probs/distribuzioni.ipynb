### Distribuzione 

**Definizione**

oggetti che generalizzano il concetto di funzione   
utili per: derivare anche funzioni che hanno salti o non sono derivabili nel senso usuale
 
cosa | funzione | distribuzione
---|---|---
cosa è | funzione | funzionale lineare
input | punto x | funzione di prova $\varphi(x)$ 
output | numero | numero
come si scrive | $f(x)$ | $\langle T, \varphi \rangle$
cosa fa | fa delle modifiche (seguendo una formula) ad un numero (il punto x) | fa delle modifiche ad una funzione di test 

**Da funzione a distribuzione**

come:  
$\langle T, \varphi \rangle = \int_{-\infty}^\infty u(x) \varphi(x) \, dx$  
u: funzione  
$T_u$: distribuzione  

quando T può diventare u:    
non tutte le distrib $T_u$ si possono esprimere come funzione u  
si dicono 'disribuzione singolare'  
eg. la delta non si può  

quando u può diventare T:    
per avere $T \in S'$:  
u deve essere $L^1(R), \, L^2(R),$, oppure un poly   

**Attenzione**

NON confonderle con le disrtibuzioni di probabilità  
qui parliamo di distribuzioni nel senso dell'analisi matematica  

**Condizioni**

per essere una distribuzione, T deve essere:  
1- lineare  
2- continua (se $\varphi$ cambia poco, il numero in output cambia poco)

### Proprietà distribuzioni

**Derivata distribuzionale**

$ \langle T', \varphi  \rangle = -\langle T, \varphi' \rangle$  

eg. nel senso delle distribuzioni, $\theta' = \delta$   
Il gradino non ha una derivata ordinaria nel punto del salto, ma la derivata distribuzionale descrive quel salto con una delta  

**Convergenza debole**

$\langle T_u, \varphi \rangle \rightarrow \langle T, \varphi \rangle \quad \forall \varphi$  

proprietà:  
1- se ho $u_n(x) \rightarrow u(x)$, e $u_n(x) \in L^2$, allora $T_n \rightarrow T$ in S'  

#### Spazi

**Tipi di distribuzioni**

a seconda di su che funzioni test può agire, ci sono diversi tipi di distribuzioni   
lo spazio delle distribuzioni (cioè dei lineari) è lo SPAZIO DUALE di quello delle funzioni  

spazio funzioni test | spazio distribuzioni
---|---
$\xi =C^\infty$ infinitamente derivabili <br> cioè non-0 solo in un intervallo limitato | $\xi'$ : disrtibuzioni a supporto compatto  
$S$: funzioni $C^\infty$ a decrescenza rapida <br> (funzioni infinitamente derivabili che, insieme a tutte le loro derivate, decrescono rapidamente all'infinito) | S', distribuzioni temperate  
D: funzioni $C^\infty$ e a supporto compatto | D': distribuzioni di Schwartz

rapporto:  
$D \subset S \subset \xi$  
$\xi ' \subset S' \subset D'$

### Spazi

**Derivata distribuzionale**

uso def di derivata distribuzionale

### Esempi distribuzioni

**Delta**

$\langle \delta , \varphi \rangle = \varphi(0)$  
appartiene a $\xi'$  

**Theta**

la sua derivata è la delta:  
$\langle \theta', \varphi \rangle = - \langle \theta, \varphi' \rangle = -\int_0^\infty\varphi'(t) \, dt = -[\varphi(\infty) -\varphi(0)] = \varphi(0)$  
so che: $\langle \delta, \varphi \rangle = \varphi(0)$  
quindi $\theta' = \varphi$  

perchè $\varphi(\infty)=0$, per definizione di spazio S  
e perchè posso scambiare la derivata nel prodotto scalare da un lato all'altro???

**Sign**

$sgn' = 2 \delta$  
$2\theta= 1 + sgn$  

**PV**

$\langle PV \frac 1t, \varphi \rangle  = \lim _{\epsilon \rightarrow 0^+} \int _{|t| > \epsilon} \frac {\varphi(t)}t dt$  

proprietà:  
$t P\frac 1 t = 1$  
cioè, posso usarla per semplificare come al solito  
Vuol dire che, dopo aver moltiplicato la distribuzione per t, la singolarità scompare  

**Limite notevole**

$\lim_{\epsilon \rightarrow 0^+} \frac 1 {x\pm i \epsilon} = P \frac 1x \mp i \pi \delta(x)$

### PV - valore principale di Cauchy

**Cosa fa**

formalmente:  
dice di togliere un intervallo simmetrico attorno a zero e poi farlo restringere  
così tolgo solo il punto problematico  
e in quel punto/intervallo, setto f=0, così non mi dà problemi  

graficamente:   
invece di passare per il punto O, il bordo fa una curva di raggio r, con $ r \rightarrow 0$  

**Utilizzo**

quando:  
se ho integrale con una singolarità nel suo intervallo di integrazione  

se non lo uso: devo staccare l'intervallo nei punti singolari, e vedere che succede  
se lo uso: ignora le singolarità interne, non staccare, semplicemente calcola  valori agli estremi come sempre

**Esempio**

$\frac 1x$ ha punto singolare x=0  

1- NON lo uso  
$\int_{-1}^{1} \frac 1x = \int_{-1}^0 \frac 1x+ \int_{0}^{1} \frac 1x$  
e devo studiare se a x=0  diverge o converge  

2- lo uso  
$PV \int_{-1}^{1} \frac {dx}x = \lim_{\epsilon \rightarrow 0^+} [\int_{-1}^{-\epsilon} \frac {dx} x+ \int_{\epsilon}^{1} \frac {dx}x]$  
$P \int_{-1}^{1} \frac {dx} x = \ln x|_{-1}^1 = 0$   
cioè ingoro il punto problematico  


**Utilizzo in eq diff**

ignora la P  
posso usare il suo argomento come al solito, si può semplificare con altre cose

formalmente, vale:  
$\omega P \frac 1 \omega = 1$
