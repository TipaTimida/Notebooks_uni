### Spazi

S: spazio dele funzioni Schwartz ($\varphi$)  
funzioni infinitamente derivabili che, insieme a tutte le loro derivate, decrescono rapidamente all'infinito

S': il suo spazio duale  
anche detto: lo spazio delle distribuzioni temperate ($\delta$)    
funzionali lineari e continui che agiscono sulle funzioni di S  
ci appartengono, ad esempio, la delta di Dirac e la sua derivata   
Una distribuzione T agisce sulle funzioni di prova $\varphi$ tramite un numero $\langle T,\varphi\rangle$

**Derivata distribuzionale**

$ \langle T', \varphi  \rangle = -\langle T, \varphi' \rangle$  
ottenuta tramite l'integrazione per parti (perchè so che i termini al bordo si annullano per $\varphi$, quindi il fattore uv|_bounday = 0 )

**Proving stuff**

$F[\delta '(\omega)]  = \langle \delta', e^{i \omega t} \rangle = - \langle \delta, i \omega e^{i \omega t} \rangle  = -i \omega$  
uso def di derivata distribuzionale  
e poi uso proprietà di $\delta$

### Esempi distribuzioni

$\langle \delta, \varphi \rangle = \phi(0)$  
$\delta$ è la delta di Dirac, intesa come distribuzione   
significa che, quando la delta agisce sulla funzione di prova $\varphi$, restituisce il valore di $\varphi$ nel punto 0  

**Theta**

la sua derivata è la delta:  
$\langle \theta', \varphi \rangle = - \langle \theta, \varphi' \rangle = -\int_0^\infty\varphi'(t) \, dt = -[\varphi(\infty) -\varphi(0)] = \varphi(0)$  
so che: $\langle \delta, \varphi \rangle = \varphi(0)$  
quindi $\theta' = \varphi$  

perchè $\varphi(\infty)=0$, per definizione di spazio S  
e perchè posso scambiare la derivata nel prodotto scalare da un lato all'altro???

**Sign**

$sgn' = 2 \delta$  
$2\theta= 1 + sgn$  

**PV**

$\langle PV \frac 1t, \varphi \rangle  = \lim _{\epsilon \rightarrow 0^+} \int _{|t| > \epsilon} \frac {\varphi(t)}t dt$  

proprietà:  
$t P\frac 1 t = 1$  
cioè, posso usarla per semplificare come al solito  
Vuol dire che, dopo aver moltiplicato la distribuzione per t, la singolarità scompare  

### PV - valore principale di Cauchy

**Cosa fa**

formalmente:  
dice di togliere un intervallo simmetrico attorno a zero e poi farlo restringere  
così tolgo solo 0, il punto problematico

graficamente:   
invece di passare per il punto O, il bordo fa una curva di raggio r, con $ r \rightarrow 0$  

**Utilizzo**

quando:  
se ho integrale con una singolarità nel suo intervallo di integrazione  

se non lo uso: devo staccare l'intervallo nei punti singolari, e vedere che succede  
se lo uso: ignora le singolarità interne, non staccare, semplicemente calcola  valori agli estremi come sempre

**Esempio**

$\frac 1x$ ha punto singolare x=0  

1- NON lo uso  
$\int_{-1}^{1} \frac 1x = \int_{-1}^0 \frac 1x+ \int_{0}^{1} \frac 1x$  
e devo studiare se a x=0  diverge o converge  

2- lo uso  
$PV \int_{-1}^{1} \frac {dx}x = \lim_{\epsilon \rightarrow 0^+} [\int_{-1}^{-\epsilon} \frac {dx} x+ \int_{\epsilon}^{1} \frac {dx}x]$  
$P \int_{-1}^{1} \frac {dx} x = \ln x|_{-1}^1 = 0$   
cioè ingoro il punto problematico  


**Utilizzo in eq diff**

ignora la P  
posso usare il suo argomento come al solito, si può semplificare con altre cose

formalmente, vale:  
$\omega P \frac 1 \omega = 1$
