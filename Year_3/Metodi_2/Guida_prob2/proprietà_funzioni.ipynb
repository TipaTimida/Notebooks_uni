### Analitica, intera

**Def**

analitica in D: non ha poli (singolarità) nel dominio D  
intera: è analitica (non ha poli) in tutto il piano complesso C  

**Th di identità per le funzioni analitiche**

funzione analitica che si annulla si un intervallo è identicamente nulla    

esempio (19f):  
so che: $b(\omega)= 0 \quad -\omega_0 < \omega < \omega_0$   
allora: $b(\omega)=0$ ovunque  

### Causale

**Def**

G(t)=0 per t<0    
cioè: comincia ad esistere la risposta G solo dopo il tempo 0, è stata causata da qualcosa che è accaduto a t=0 

**Th di Titchmarsh**

una funzione G(t) è causale se:  
la sua trasformata $\tilde G(\omega)$ è il bordo di una funzione analitica nel semipiano superiore, con una certa stima di integrabilità  
usala se: ho $\tilde G(\omega)$, voglio sapere se G(t) è causale senza calcolarla  

steps:  
1-passa al piano complesso, $\omega = x+iy$  
2- controlla che non ci siano poli nel semipiano superiore (cioè: è analitica nel piano superiore)   
3- verifica la stima integrale sul semipiano superiore (y>0)  

**Dimostra che G è causale**

se ho $G(\omega)$, 2 metodi:  
1- antitrafsorma in G(t), vedi se rispetta la def -> MEGLIO  
2- th di Titchmarsh, così non devo anti-trasforamre a G(t) 

### Supporto compatto

**Def**

se scritta a tratti (in time domain), le parti dove $G(t) \neq 0$ sono un tratto finito


**Proprietà**

G a supporto compatto -> trasformata intera   

cioè:  
a(t) supporto compatto -> $\tilde a(\omega)=F[a(t)]$ è intera (analitica su tutto il piano)  

### Relazioni di dispersione

**Def**

cosa: relazione tra parte reale e immaginaria di una funzione analitica    

quando la posso usare: f è analitica sul semipiano superiore (cioè non ha poli lì, nemmeno sull'asse reale)  
nota che causale -> analitica sul piano superiore, per Titchmarsh   

**Relazione di Kramers-Kronig (relazioni di dispersione)**

ipotesi:   
$G(\omega) = G(\omega)+ i G(\omega)$  
G analitica sul semipiano superiore   
G scala più veloce di $1/|\omega$| per $|\omega| \rightarrow \infty$

tesi:  
$Re[G(\omega)] =\frac 1 \pi P \int_R \frac{Im [G(u)]}{u-\omega}du$  
$Im[G(\omega)] =-\frac 1 \pi P \int_R \frac{Re [G(u)]}{u-\omega}du$   

**Da dove si ricavano**

formula di Sohotski-Plemelj:  
$PV \int_{-\infty}^\infty \frac{\chi(v)}{v-\omega}dv = i \pi \chi(\omega)$

per ottenere le relazioni di dispersione:  
1- scrivo $\chi=\chi_I+i\chi_R$   
2- uguaglio le parti reali e immaginarie (a destra c'è i, quindi $\chi_R$ diventa immaginaria, $i \chi_I$ diventa reale, con un -, che infatti vedo che appare)
