### Definizione

**Convenzione che uso**

$\tilde f(\omega) = \int_R f(t) e^{i \omega t} \, dt$  
$f(t) = \frac 1 {2 \pi} \int_R \tilde f(\omega) e^{- i \omega t} d\omega$  


**Convenzioni**

$\tilde f(\omega) = c \int_R f(t) e^{\sigma \; i \omega t} \, dt$  
$f(t) = \frac 1 {2 \pi c} \int_R \tilde f(\omega) e^{-\sigma \; i \omega t} d\omega$  

1- sign choice: $\sigma = \pm 1$    
i usually use $\sigma=1$  

2- normalization: $c$, può essere ogni valore  
i usually use $c=1$  

**Usarla più volte**

se ho: $F[f(t)] = \tilde f(\omega)$  

cosa | antitrasformata | duality theorem
---|---|---
formula  | $F^{-1}[\tilde f(\omega)] = f(t)$  | $F[\tilde f(t)] = 2 \pi c^2 f(-\omega)$  
utile per | tornare al time domain dopo aver risolto in freq domain | ricordarsi le formule comuni, a coppie
esempio: $F[\delta(t)] = 1$ |$F^{-1}[1] = \delta (t)$  | $F[1] = 2 \pi \delta(-\omega) = 2 \pi \frac 1 {\|-1\|} \delta(\omega)$  


### Guida trasformata

**Steps (trasformata e anti-trasformata)**

1- separa con linearità  
2- applica le proprietà: freq/time shift, derivata  
3- calcola le F di tutti i pezzi   
o cose note, o calcola l'integrale (con th residui)  
4- dopo aver trasformato, usa proprietà delle distribuzioni che appaiono   
(e razionalizza, se "i" al denom)

#### 1. Linearità

$F[af(t)+bg(t)]= aF[f(t)]+bF[g(t)]$  

perchè: la trasformata è un integrale, posso separarla in addendi  

#### 2. Proprietà transform

cosa | time | freq 
---|---|---
| Time shift | $f(t-t_0)$ | $e^{i\omega t_0}\tilde f(\omega)$ |
| Frequency shift |$e^{i\omega_0t}f(t)$ | $\tilde f(\omega+ \, \omega_0)$ |
| Time scaling | $f(at)$, $a\ne0$ | $\dfrac{1}{\|a\|}\tilde f\!\left(\dfrac{\omega}{a}\right)$ |
| Time derivative |$ f'(t)$ | $(- \, i\omega)\tilde f(\omega)$ |
| Multiply by time | $t f(t)$ | $- \, i\,\dfrac{d\tilde f}{d\omega}$ |

fai attenzione agli ESPONENZIALI COMPLESSI, danno gli shift (in time o freq)

**Come si scrivono**

posso usarli o nel verso time->freq, o in quello opposto:  
$F[e^{it}](\omega) = F[1](\omega+1)= \delta(\omega+1)$  
$F^{-1}[P \frac 1 {\omega -k}] = e^{-ikt} F^{-1}[P \frac 1 \omega] = e^{-ikt} (-\frac i2 sgn(t))$

come scriverli:  
freq shift: $F[e^{it}](\omega) = F[1](\omega+1)$  
derivata: $F[x']= (-i \omega) F[x]$

#### Comuni, ma da calcolare

$F[\frac 1 {a^2+ t^2}] = \frac \pi a e^{-a |\omega |}$ 

passo al piano complessso, t=z  
$F = \int_R \frac 1{(z-ia)(z+ia)}e^{i \omega z} dx$  

vedo dove decade:  
$\omega >0$: semipiano positivo -> polo $z=ia$  
$\omega <0$: semipiano negativo -> polo $z=-ia$  

lenna di Jordan: l'arco con $R= \infty$ ha integrale nullo  

th residui:  
$\omega >0$: $F = 2 \pi i Res[ia] = \frac \pi a e^{-a \omega}$  
$\omega <0$: $F = 2 \pi i Res[-ia] = \frac \pi a e^{a \omega}$  


se cambio var di integrazione, è uguale:  
$F[\frac 1 {a^2+\omega^2}] = \frac \pi a e^{-a \|t\|}$

**Attenzione**

le formule non si differenziano a seconda se ho come arg t o $\omega$, ma se uso traformata o anti-trasformata  
le due sopra, sono entrambe trasformate, quindi va bene scambiare  
ma quelle note no, perchè io voglio trasforamre epoi ANTI-trasformare, per trovare la sol del mio problema originale  
quindi, la seconda volta, devo usaer le proprietà delle anti-trasformate

#### 4. Proprietà distrib

**Delta**

nome | cosa | esempio
---|---|---
time scaling | $\delta(ax) = \frac {1}{\|a\|}\delta (x)$  
derivata | $f(\omega) \delta'(\omega-a)= f(a)\delta'(\omega-a)-f'(a)(\delta(\omega-a))$ | $(\omega-a)\delta'(\omega-a) = -\delta(\omega-a)$
Prodotto con una funzione | $f(\omega) \delta(\omega-a) =f(a) \delta(\omega-a)$ <br> perchè è non-0 solo se $\omega=a$ | $ f(\omega) \delta(\omega) = f(0) \delta(\omega)$


**PV**

$\omega P \frac 1 \omega = 1$   
in pratica, ignora il PV, semplifica come sempre 

**Scomposizione**

$\frac 1{\omega(\omega +1)}= \frac 1 i (\frac 1 \omega -\frac 1 {\omega+i})$    
nota, il lato sinistro può apparire così: $\frac 1 {\omega+1}P\frac 1 \omega$  


**Sign e theta**

$sign(t) = 2\theta(t)-1$  
$sign(t)=\theta(t)-\theta(-t)$

**Theta e funzione a tratti**

$G(t) = \begin{cases} 0 & t<0 \\ -ie^{-\epsilon t} & t>0 \end{cases}$  
$G(t) = -i \theta(t) e^{-\epsilon t}$

### Spiegazione proprietà

**Da dove vengono le proprietà**

si vedono facendo la trasformata su quella cosa  
eg. derivata -> integration by parts  
$\int_R f'(t) e^{i \omega t} \, dt = -i \omega \int_R f(t) e^{i \omega t} dt$    
if the boundary term vanishes, and it does if $f \in L^2$, which i assume  

**Gaussiane**

Fourier trasforma gaussiana in gaussiana  
ma, se prima era larga, diventa stretta  
prima: $e^{-t^2/(2\tau^2)}, \tau>0$   
dopo: $\sqrt{2\pi}\,\tau\,e^{-\tau^2\omega^2/2}$

nota: per la larghezza, uso $\tau$ invece di $\sigma$, per non confonderla con il segno della convenzione
