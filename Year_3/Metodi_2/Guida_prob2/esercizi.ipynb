### Come risolvere

#### Freq domain

**Sol particolare**

semplicemente isola $\tilde G(\omega)$    
se c'è qualcosa con $\omega$ al denominatore, metti PV davanti, tipo: $i e^{i \omega} PV \frac 1{k-\omega}$  


**Sol omogenea**

sarà G moltiplicato per qualcos'altro, uguale a 0  
quindi non posso dividere per quella cosa  

G può essere non-0 solo quando l'altra cosa è 0  
quindi è una delta  
$(k-\omega) \tilde G_h(\omega) =0$  
$\tilde G(\omega) = \tilde c \, \delta(\omega +k)$  

nota: la trasforamta di G è $\frac{\tilde c}2 e^{-ikt} = ce^{-ikt}$ ??? controlla  
questa la potrei risolvere anche direttamente in time domain, tanto è facile (a meno che il testo non specifichi di risolvere il freq domain)

**1. Trova funzione di Green in freq domain**

esercizio | cosa
---|---
$a(t) = \delta(t)$ <br> $b(t) = G(t)$| setup funzione di Green
$F[G']+ikF[G] = F[\delta] -F[\delta(t-1)]$ <br> $(-i \omega) \tilde G + ik \tilde G = 1 - e^{i \omega} $  <br> $i(k- \omega) \tilde G(\omega) = 1- e^{i \omega}$| trasformo
$\tilde G_p (\omega) = i(1-e^{i \omega}) P \frac 1 {\omega -k}$ | sol particolare
$(k- \omega) \tilde G_h(\omega) $ <br> $\tilde G_h(\omega) = \tilde c \; \delta (\omega-k)$| sol generale (setto $a(t)=0$, cioè tolgo parte a destra)
$\tilde G(\omega) = i(1-e^{i \omega}) P \frac 1 {\omega -k} +\tilde c \; \delta (\omega-k)$  | sol generale
$G(t) = i F^{-1}[P \frac 1 {\omega -k}](t) -  F^{-1}[P \frac 1 {\omega -k}](t-1)+ \tilde c F^{-1} [\delta(\omega-k)]$ <br> $F^{-1}[P \frac 1 {\omega -k}] = e^{-ikt} F^{-1}[P \frac 1 \omega] = e^{-ikt} (-\frac i2 sgn(t))$ <br> $F^{-1}[\delta-k] = e^{-ikt}F^{-1}[\delta] = \frac 1 {2 \pi} e^{-ikt} \quad \frac{\tilde c}{2 \pi} = c$ | trasformo
$G(t) = \frac 12 e^{-ikt} sgn(t) - \frac 12 e^{-ik(t-1)}sgn(t-1) + c e^{i kt}$ |  soluzione

#### Time domain

**Se ci sono le DELTA**

1- dividi in intervalli temproali, separati da impulsi  
cioè, l'ar delle $\delta$ divide gli intervalli temporali  

2- per ogni intervallo:    
trova sol omogenea (setta a=0)  

3- integra sui punti che separano gli intervalli  
così ho condizioni di continuità, posso scrivere i coeff degli intervalli tutti in funzione di uno solo

**2. Trova G in time domain**

esercizio | cosa
---|--
se $t \neq 0,1$: equazione omogenea <br> $G'+ikG =0$ <br> $ G(t) = \begin{cases} Ae^{-ikt} & t<0 \\ Be^{-ikt} & 0<t<1 \\ De^{-ikt} & t> 1 \end{cases}$| omogenea
al punto t=0: $\epsilon \rightarrow 0$ <br> $\int_{-\epsilon}^\epsilon G'(t) \,  dt + ik \int_{-\epsilon}^\epsilon G(t) \, dt = \int_{-\epsilon}^\epsilon  \delta(t) \, dt - \int_{-\epsilon}^\epsilon \delta(t-1) \, dt$ <br> $ [G(\epsilon)-G(-\epsilon)] + 0 = 1 -0 $ <br> $G(0^+)-G(0^-) = 1$ <br> ultimo integrale fa 0, perhè quella delta è non-0 solo in t=1| prima continuità: integro l'eq diff attorno a t=0
al punto t=1: $\epsilon \rightarrow 1$ <br> $\int_{-\epsilon}^\epsilon G'(t) \,  dt + ik \int_{-\epsilon}^\epsilon G(t) \, dt = \int_{-\epsilon}^\epsilon  \delta(t) \, dt - \int_{-\epsilon}^\epsilon \delta(t-1) \, dt$ <br> $ [G(\epsilon)-G(-\epsilon)] + 0 = 0 -1 $ <br> $G(1^+)-G(1^-) = -1$ | prima continuità: integro attorno a t=1
$G(0^+)-G(0^-) = 1$ <br> $G(1^+)-G(1^-) = -1$  | | recap: integrando l'e attraverso i punti di salto, ho ottenuto le condizioni
condizione t=0: $B-A = 1$ <br> condizione t=1: $(D-B) e^{-ik} = -1$ <br> quindi: $D=B_e^{ik} = A+1-e^{ik}$ | risolvo condizioni
$ G(t) = \begin{cases} Ae^{-ikt} & t<0 \\ (A+1)e^{-ikt} & 0<t<1 \\ (A+1-e^{ik})e^{-ikt} & t> 1 \end{cases}$ <br> i valori nei singoli 0,1 non cambiano la distribuzione|  soluzione  

### Proprietà (di G)

#### Causale

**Dimostra che G è causale**

metodi:  
1- calcola G(t) (anti-trasformando $\tilde G(\omega)$), vedi che rispetta la def di causale  
2- th di Titchmarsh, se non voglio anti-trasforamre a G(t)    
3- 

**1. Definizione**

G(t)=0 per t<0  

cioè: comincia ad esistere la risposta G solo dopo il tempo 0, è stata causata da qualcosa che è accaduto a t=0 

**2. Th di Titchmarsh**

una funzione G(t) è causale se:  
la sua trasformata $\tilde G(\omega)$ è il bordo di una funzione analitica nel semipiano superiore, con una certa stima di integrabilità  
usala se: ho $\tilde G(\omega)$, voglio sapere se G(t) è causale senza calcolarla  

steps:  
1-passa al piano complesso, $\omega = x+iy$  
2- controlla che non ci siano poli nel semipiano superiore (cioè: è analitica nel piano superiore)   
3- verifica la stima integrale sul semipiano superiore (y>0)  

#### Parte Re e Im

**Relazione di Kramers-Kronig (relazioni di dispersione)**

cosa: relazione tra parte reale e immaginaria di una funzione analitica  

ipotesi:   
$G(\omega) = G(\omega)+ i G(\omega)$  
G analitica sul semipiano superiore (nota che causale -> analitica sul piano superiore, per Titchmarsh)  
G scala più veloce di $1/|\omega$| per $|\omega| \rightarrow \infty$

tesi:  
$Re[G(\omega)] =\frac 1 \pi P \int_R \frac{Im [G(u)]}{u-\omega}du$  
$Im[G(\omega)] =-\frac 1 \pi P \int_R \frac{Re [G(u)]}{u-\omega}du$   

#### Supporto compatto

**Definizione**

se scritta a tratti (su tratti di t), le parti dove $G \neq 0$ sono un tratto finito


**Proprietà**

se G a supporto compatto: la sua trasformata (in freq domain) si può estendere a una funzione intera   
(cioè, se faccio la sua trasformata, ottengo una funzione intera)
