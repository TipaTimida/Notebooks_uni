### Eq diff in time domain: sol particolare e homo


**Metodo 1 (senza funzione di Green)**

QUANTE SOL:  
sol particolare: plugga $a(t)$ dato dal testo   
sol homo: plugga $a(t)=0$    
sol generale: $b(t) = b_p(t)+b_h(t)$  

COME RISOLVERE:  
l'eq diff potrei risolverla anche direttamente in time domain (come ad analisi 1), per entrambe le sol  
in pratica, è facile da fare solo per la sol homo   
ma non è obbligatorio, posso anche risolvere la sol homo ripetendo il metodo 1  

nota:  
questo va fatto ogni volta che risolvo un sistema col metodo 1  
quindi anche quando lo uso per trovare la funzione di Green  

**Sol particolare**

1- isola $\tilde b(\omega)$   
(che potrebbe essere $G(\omega)$, se sto cercando G)  

2- se il denominatore si annulla per una $\omega$ REALE, metti PV davanti  
eg. $i e^{i \omega} PV \frac 1{k-\omega}$  

ricorda di usare le proprietà delle disrtib, tipo di $\delta$  

**Sol omogenea in freq domain**

1- sarà G moltiplicato per qualcos'altro, uguale a 0  
quindi non posso dividere per quella cosa  

2- G può essere non-0 solo quando l'altra cosa è 0  
quindi è una delta  
$(k-\omega) \tilde G_h(\omega) =0$  
$\tilde G(\omega) = \tilde c \, \delta(\omega +k)$  

3- poi, quando anti-trasformo:     
$F^{-1}[\delta-k] = e^{-ikt}F^{-1}[\delta] = \frac 1 {2 \pi} e^{-ikt} \quad \frac{\tilde c}{2 \pi} = c$  

#### omogenea in time domain

**Se ci sono le DELTA**

1- dividi in intervalli temporali, separati dagli impulsi $\delta$  
cioè, l'arg delle $\delta$ divide gli intervalli temporali  

2- per ogni intervallo:    
trova sol omogenea (setta a(t)=0)  

3- integra sui punti che separano gli intervalli  
così ho condizioni di continuità, posso scrivere i coeff degli intervalli tutti in funzione di uno solo

**2. Trova G in time domain**

esercizio | cosa
---|--
se $t \neq 0,1$: equazione omogenea <br> $G'+ikG =0$ <br> $ G(t) = \begin{cases} Ae^{-ikt} & t<0 \\ Be^{-ikt} & 0<t<1 \\ De^{-ikt} & t> 1 \end{cases}$| omogenea
al punto t=0: $\epsilon \rightarrow 0$ <br> $\int_{-\epsilon}^\epsilon G'(t) \,  dt + ik \int_{-\epsilon}^\epsilon G(t) \, dt = \int_{-\epsilon}^\epsilon  \delta(t) \, dt - \int_{-\epsilon}^\epsilon \delta(t-1) \, dt$ <br> $ [G(\epsilon)-G(-\epsilon)] + 0 = 1 -0 $ <br> $G(0^+)-G(0^-) = 1$ <br> ultimo integrale fa 0, perhè quella delta è non-0 solo in t=1| prima continuità: integro l'eq diff attorno a t=0
al punto t=1: $\epsilon \rightarrow 1$ <br> $\int_{-\epsilon}^\epsilon G'(t) \,  dt + ik \int_{-\epsilon}^\epsilon G(t) \, dt = \int_{-\epsilon}^\epsilon  \delta(t) \, dt - \int_{-\epsilon}^\epsilon \delta(t-1) \, dt$ <br> $ [G(\epsilon)-G(-\epsilon)] + 0 = 0 -1 $ <br> $G(1^+)-G(1^-) = -1$ | prima continuità: integro attorno a t=1
$G(0^+)-G(0^-) = 1$ <br> $G(1^+)-G(1^-) = -1$  | | recap: integrando l'e attraverso i punti di salto, ho ottenuto le condizioni
condizione t=0: $B-A = 1$ <br> condizione t=1: $(D-B) e^{-ik} = -1$ <br> quindi: $D=B_e^{ik} = A+1-e^{ik}$ | risolvo condizioni
$ G(t) = \begin{cases} Ae^{-ikt} & t<0 \\ (A+1)e^{-ikt} & 0<t<1 \\ (A+1-e^{ik})e^{-ikt} & t> 1 \end{cases}$ <br> i valori nei singoli 0,1 non cambiano la distribuzione|  soluzione
