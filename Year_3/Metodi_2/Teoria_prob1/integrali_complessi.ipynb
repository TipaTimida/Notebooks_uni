### 1. Setup

---|---

**3. Verso di percorrenza**

voglio percorrere la retta reale da $-\infty$ a $+ \infty$

quindi vedi se, col mio percorso, è un senso orario o antiorario  

#### Integrale su retta reale

**Come fare**

cosa: chiudi il percorso con un semicerchio di raggio $R \rightarrow \infty$, sul semipiano sopra o sotto    
scegli quello dove f(z) decade    

quando: eg. trasformate e antitrasformate di Fourier

**Su che piano decade la funzione**

cosa contiene la funzione | caso 1 | caso 2
---|---|---
$e^{i \omega z} = e^{i \omega(x+iy)}= e^{i \omega z}e^{-\omega y}$  <br> eg. trasformata | caso: $\omega>0$ <br> decade su y>0 <br> quindi chiudo sul semipiano superiore | dcaso: $\omega <0$: <br> decade su y<0 <br> chiudo sul semipiano inferiore
$e^{-izt} = e^{-ixt}e^{yt}$  <br> eg. antitrasformata | caso: $t >0$ <br> decade su <0, semipiano inferiore  | caso: $t<0$ <br> decade su semipiano superiore   
no esponenziali | guarda comunque la funzione, guarda dove decade <br> se decade in entrambi, scegli quello pià facile (cioò con meno poli)
 
ricorda che il decay dipende solo dalla prate reale dell'esponente, quella immaginaria lo fa oscillare  s

### 3. Integrale sull'arco

**Definizione**

quando: se voglio l'integrale su R, ma ho esteso il contorno con un semicerchio    
cosa: devo dimostrare che l’integrale su una grande semicirconferenza tende a zero quando il raggio $R \rightarrow \infty$  

cioè per dimostrare che:  
$\lim_{R \rightarrow \infty} \int_{C_R} f(z) dz =0$  

**Recap**

metodo | quando | condizione | cosa dimostrare
---|---|---|--
generale: stima dell'arco $\pi R$ | sempre | $\|\int_{C_R} f(z) \, dz\| \le \; \pi R \; \max_{C_R} \|f(R)\| \ \rightarrow 0$ | $R \, \max_{C_R} \|f(R)\| \rightarrow 0$  
lemma di Jordan | se l'integranda ha esponenziale con arg IMMAGINARIO |  $\|\int e^{i \omega z} g(z) \, dz\| \le \frac \pi \omega \max_{C_R} \|g(R)\| \rightarrow 0$ | $\max_{C_R} \|g(R)\| \rightarrow 0$  

**Nota**

se l'integranda ha esponenziale con arg immaginario:  
posso anche usare il metodo generale, mettendo $|e^{i \omega z}|=1$  
ma poi devo moltiplicare per R, quindi a volte potrebbe darmi dei falsi nefgativi (dirmi che non va a 0, ma invece ci va)   
in questi casi, controlla con Jordan  
però se il metodo generale dice che va a 0, ci va sicuramente, non devo controllare con Jordan  

**2. Tips per norma di |f(z)|**

$\max_{C_R}$ significa fare l'upper bound  

1- fasi hanno modulo 1  
$|i| = |e^{iy}| = 1$  

2- portare modulodentro integrale  
$|I_2| =  |\int f(z) \, dz | \le \int |f(z)| dz$  

3- maggiorare una frazione   
separa numeratore e denom: $|ab|=|a| \; |b|$   
numeratore: $|a+b| \le |a|+|b|$  
denominatore: $|a+b| \ge |\,|a| - |b| \,|$  

esempio:   
denom: $|1+e^R e^{iy}| \le |\,\|e^Re^{iy}| -|1| \,| = |e^R -1| = e^R-1$   
$\frac 1{|1+e^Re^{iy}|} \le \frac 1 {e^R-1}$  


### 4. Th dei residui

**Come si scrive**

1- generale:   
$\oint_C f(z) \, dz = \pm \; 2\pi i \sum_{k=1}^n Res(f, z_k)$    
$z_k$: singolarità racchiue dentro C  

2- se mi interessa integrale sull'asse reale, e ho esteso il contorno:    
$\int_{-\infty}^{\infty} f(z) \, dx + \int_{C_R} f(z) \, dz = \pm 2 \pi i \, \sum Res(f, z_k)$  

segno a destra dipende dal verso di C:  
orario: - (orientato negativo)   
antiorario: + (orientato positivo, come gli angoli)     

**Definizione**

permette di calcolare un integrale lungo un percorso chiuso nel piano complesso  
usando soltanto il comportamento della funzione vicino alle singolarità racchiuse nel percorso
