**Poli**

definizione: punti che violano la C.E.  
cioè, gli zeri del denominatore  

### Punti di singolarità

singolarità | isolata | non isolata
---|---|--
cosa | punto singolo che viola la C.E. <br> f non è analitica lì, ma lo è in un suo intorno | qualunque intorno del punto prendo, ce n'è un altro singolare      
esempio | zeri del denom | $x= \pm \infty$  <br> $\sqrt x \rightarrow x > 0$  

#### Singolarità isolate

**Steps: come studiarle**

1- trova in che punti sono  

2- trovo tipo  
o linearizzo la funzione in quel punto con serie TL  
oppure fattorizzo il denominatore, vedo che molteplicità ha quel polo  

3- trova dominio di convergenza   
cioè distanza dalle altre singolarità  

4- calcola residuo (o guardalo dalla serie)

tipo | come appare nella serie di TL | come appare in una funzione
---|---|---
Eliminabile | Nessuna potenza negativa | $f(z) = \frac {\sin z}z \sim \frac {z+ z^3...}{z}$
Polo di ordine m | Un numero finito di potenze negative, fino a $a_{-m}$ | $1/z^m$ |  $\|f(z)\|\to\infty$ lungo tutti i percorsi
| Essenziale | Infinite potenze negative non nulle | $e^{1/z}$ |Nessuna delle due condizioni precedenti


**Trova ordine di polo: metodi**

1- linearizza f(z), come con che potenze scala la singolarità  
2- calola il limite: $\lim_{z\rightarrow z_0}^m f(z)$  
se finito e non-0: ordine m  
se 0: ordine minore di m  
se infinito (+ o -): ordine maggiore di n  

$a_{-m} = \lim_{z\rightarrow z_0}^m f(z)  
$a_{-1} = Res[f, z_0] = \frac 1 {m-1!} \frac {d^{m-1}}{dz^{m-1}} [(z-z_0)^m f(z)]\|_{z=z_0}$   
per sapere l'ordine, a me interessa $a_{-m}$, quindi faccio la prima formula, non quella del residuo 

**Nota**

polo semplice = polo di ordine 1  

**Residui**

cosa: il coeff $a_{-1}$  

ordine polo | formula
---|---
m | $Res[f, z_0] = \frac 1 {m-1!} \frac {d^{m-1}}{dz^{m-1}} [(z-z_0)^m f(z)]\|_{z=z_0}$  
1 | $Res[f, z_0] = \lim_{z \rightarrow z_0} (z-z_0)f(z)$
2 | $Res[f,z_0] = \frac {d}{dz}[(z-z_0)^2 f(z)]\|_{z=z_0}$

nota: se singolarità eliminabile, Res=0 (perchè non può avere potenze negative)

esempio:  
$f(z) = \frac{e^z}{(z-z_0)^2}$  
$Res(f, z_0)= \frac{d}{dz}e^z|_{z=z_0}= e^{z_0}$  

**Singolarità eliminabile**

1- quando lo è: se non ci sono potenze negative  
cioè $a_n=0$ se $n<0$  

2- come prolungare in quei punti  
calcola  $\lim_{z \rightarrow z_0}f(z)$ in quel punto, usando la funzione linearizzata   
oppure usa l'Hopital per calcolare il limite (deriva sopra e sotto f(z), se ho F.I.)

cioè: questa sing. si può eliminare prolungando f in quei punti, con un valore che scelgo 

#### Isolate - altro modo

**Se f è una frazione**

$f(z) = \frac {N(z)}{D(z)}$  

singolarità: $D(z_n)=0$  

espando con Taylor sulla singolarità:  
$N(z)=N(z_n)+N'(z_n)(z-z_n)...$  
$D(z)=D(z_n)+D'(z_n)(z-z_n)...$  

se:  
1- $D'(z_n) \neq 0$ (cioè lo sviluppo di Taylor di D(z) inizia con un termine lineare)    
2- $N(z_n) \neq 0$   
allora $z_n$ è un polo di ordine 1    
perchè posso scrivere: $f(z)\sim \frac{N(z_n)}{D'(z_n)(z-z_n)}$  
nota che il residuo è: $Res(f, z_n) = \frac{N(z)}{D'(z_n)}$  

**Polo di ordine m**

$f(z)=\frac{g(z)}{(z-z_0)^m}$  
dove $g(z_0) \neq 0$  

| Metodo | Cosa controllare |
|---|---|
| Fattorizzazione | Dopo le cancellazioni, resta \((z-z_0)^m\) al denominatore e un numeratore non nullo in \(z_0\). |
| Limite | \(\displaystyle\lim_{z\to z_0}(z-z_0)^m f(z)\) è finito e diverso da zero. |
| Laurent | Il termine con la potenza più negativa è \(\displaystyle c_{-m}/(z-z_0)^m\), con \(c_{-m}\neq0\). |
| Derivate di numeratore e denominatore | Si confrontano gli ordini dei loro zeri. |

**Metodo: derivate di N e D (per POLO SEMPLICE)**

caso:  
$D(z_0)=D'(z_0)=0$  
$D''(z_n) \neq 0$  
$N(z_0) \neq 0$  
allora: polo di ordine 2  

caso generale:  
N ha uno zero di ordine p  
D ha uno zero di ordine q  
polo ha ordine: q-p  
nota: se $q \le p$, è una singolarità deliminabile

spiegazione:  
$Res[f, z_0] = \lim_{z \rightarrow z_0} \frac {N(z)}{D(z)} = \lim_{z \rightarrow z_0} \frac{N(z)}{D'(z)} = \frac{N(z_0}{D'(z_0)}$   
vale solo per polo semplice  

quando si usa: se non mi va di linearizzare al denominatore, così faccio semplicemente la derivata e poi calcolo nella singolarità

### Definizioni

nome | cosa
---|---
analitica su $\Omega$ | non ha poli su $\Omega$
intera | analitica in tutto C (non ha nessuna singolarità sul piano complesso)  

nota: singolarità eliminabili non contano come singolarità  
quindi, se chiede di renderla intera, devo rendere tutte le singolarità eliminabili  

**Funzione complessa vs reale**

La differenza fondamentale è come h può avvicinarsi a zero:  
- su R, lungo la retta reale, cioè solo da destra o sinistra
- su C, da qualsiasi direzione nel piano  

dico che f'(z) esiste se:  
$f'(z) = \lim_{h \rightarrow 0} \frac{f(x+h)-f(z)}h$  
il limite è indipendente dalla direzione di h (ottengo lo stesso risultato qualunque sia la direzione del vettore h)

**Analitica e olomorfa (in C)**

cosa | olomorfa | analitica
---|---|---
viewpoint | derivata | serie di potenze
cosa esiste |  $f'(z) = \lim_{h \rightarrow 0} \frac{f(x+h)-f(z)}h < \infty$ <br> derivata | $f(z) = \sum_{n=0}^\infty a_n (z-z_0)^n < \infty$ <br> serie di potenze convergente (serie di Taylor)  

analitica su un insieme R: è analitica in ogni punto di R   

**Teorema**

su C: olomorfa = analitica  
su R: essere infinitamente derivabile non implica essere analitica  

poly e sin sono analitici in tutto C

### Logaritmo complesso

**Formula**

all branches:  
$\ln z = \ln(re^{i(\theta+2n \pi)}) = \ln r + i(\theta+2n\pi)$  

ramo principale (n=0):  
$\ln z = \ln(re^{i \theta}) = \ln r + i \theta$  
dove: $r=|z| \quad \theta= arg(z)$  


**Branches**

1- scegli il taglio di log z  
cioè: scegli un dominio di lunghezza $2 \pi$ a cui appartiene $\Theta$  
eg. $\Theta \in [-\pi, \pi]$, oppure $\Theta \in [-\frac \pi 2, \frac {3 \pi}2]$  
come sceglierlo: fai in modo che non taglio l'area su cui sto facendo l'integrale

2- scrivi determinazione principale  
cioè n=0

a cosa serve scegliere il branch:  
quando faccio il limite, devo sapere a che arg tende  

#### Branches

**Perchè branches**

perchè log è una funzione multiple-valued  
quindi devo scegliere un preciso output tra quelli possibili  

graficamente: il plot è un'elica infinita  
quindi ogni punto sul piano C ha infiniti punti a cui viene mappata dalla funzione

**Branch**

f: multiple-valued function  
F: branch, a single-valued function, analytics  
at each point, F(z) is one of the possible resulst of f(z)  

branch cut: a portion of a line or curve that is introduced in order to define a branch F   
singular points for F: points on the branch cut for F  
branch points: points that appear in all branches
