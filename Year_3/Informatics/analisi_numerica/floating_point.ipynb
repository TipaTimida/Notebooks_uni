**Precisione**

def: quanto spazio allochiamo epr salvare un numero  
single: 32 bit  
double: 64 bit  

**Unit randoff**

def: l'ultima cifra della mantissa, cioè fino a dove posso salvare i numeri  
cioè fin dove sono sicura che il numero sia giusto, dopo quello ho trocanto qualunque cosa ci fosse  

single: u=$2^{-23}  \sim 19^{-7}$  
double: $u=10^{-16}$  

**Errore in operazioni**

$x\; op\; y = (x \, op \, y) (1+\delta)$   
$\delta \le u$   
quindi sono sicura che l'errore relativo ($\delta$) sia minore dell'unit randoff (u)

**Cancellazione**

cosa: problema che sorge se faccio molte operazioni in serie  
eg. 2 ops in serie  

$x+y+z = [(x +y) (1+\delta_1) +z] (1+ \delta_2)$    
$= (z+y+z)(1+\delta_2+\deòta_1\frac{x+y}{x+y+z})$  

se x,y,z >0: $|\frac{x+y}{x+y+z}| \le 1$, quindi ok  
se z<0: non vero, potrei avere errore grande  

eg. z=1, y=0.00001, z=-1 -> $|\frac{1.000001}{0.00001}|  \gg 1$  
quindi ho un errore grande
