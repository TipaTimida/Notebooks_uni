**Types**

**Big-O notation**

the stuff in parenthesis tells me how the complezity scaled  
remember that it ignores constants: O(2n)=O(n)  

O(1): The amount of work doesn't depend on the size of the input  
O(n): amoun of work scales with n

### Find complexity

**How**

1- identify all loops  
2- for each loop, find its complexity  
3- combine loops, adding or multiplying the complexities  

follow rules of big-O maths

**2. Type of loops**

have array of size n  

what | code | complexity
---|---|---
no loop | arr[3] <br> for i in range(3) | O(3)  
normal loop | for i in range(n) | O(n)
loop that halves something | while n > 1: <br> n = n // 2 | O($log_2$ n)  

essentially, the complexity of a loop is how many times it runs, the number inside range()

**3. Multiple loops**

1- nested loops  
must multiply complexity  
O(n) x O(n) = O($n^2$)   

2- one loop after the other  
must add complexity  
O(n) + O(n) = O(2n) = O(n)   

remember, big-O tells me the scaling  

**Big-O maths**

keep the thing it scales with  
O(n^2+n+1)=O(n^2)  

### Examples

// Subsequent loops  
// complexity: O(n)+O(m) = O(n+m)  

for i in range(n):  
    ...  
for j in range(m):  
    ...  

// Weird nested loop  
// O(n^2)  

for i in range(n):
    for j in range(i):


how many times runs the inside loop:  
0+1+2+...+(n-1) = n(n-1)/2 ~ n^2  
i care about how many times the nested loop runs, that's the complexity  

other way;  
if the inner loop runs n(n-1)/2 times, its average length is (n-1)/2, since it's made to be run n times by the outer loop  
so: O(n) x O((n-1)/2) = O(n) x O(n) = O(n^2)

// Fake nested loops  
// complexity: O(n) x O(5) = O(5n) = O(n)

for i in range(n):
    for j in range(5):

// Loop + halving   
// O(n) x O(log n) = O(n log n)

for i in range(n):
    j = n

    while j > 1:
        j = j // 2
