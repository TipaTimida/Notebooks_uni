```python
main:
    push    rbp
    mov     rbp, rsp
    mov     eax, 0
    pop     rbp
    ret
```

push rbp  
Saves the old value of the rbp register on the stack. rbp is commonly used as the function’s frame pointer.

mov rbp, rsp  
Sets rbp equal to the current stack pointer rsp. This establishes a new stack frame for main.

mov eax, 0  
Places 0 in eax. In the usual x86-64 calling convention, a function’s integer return value is stored in eax.  
so, it means: return 0  

pop rbp  
Restores the old value of rbp from the stack.    

ret  
Returns to the code that called the func
