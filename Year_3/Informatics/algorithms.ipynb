### Types

divide and conquer: break into smaller problems, recombine result (binary search)  
greedy: make best local choice at each step (Djikstra)   
dynamic programming: store previous results to avoid recalculation (Fibonacci)  
brute force  



#### Greedy

make best local choice at each step, hope will lead to global optimum  
don't backtrack or reconsider past decisions  

so can only use in problems with an optimal substructure and greedy choice property (so when local optimum leads to global optimum)  

eg:  
Dijkstra’s Algorithm, finds the shortest path in a graph  
Prim’s Algorithm, find the Minimum Spanning Tree (MST)


#### Dynamic programming (DP)

solve problems by breaking them into overlapping subproblems  
use memoization (top-down) or tabulation (bottom-up) to store previous results and avoid redundant calculations  

eg:  
Fibonacci  
Knapsack problem (0/1)   
Bellman-Ford algorithm

#### Brute-force

Tries all possible solutions and picks the best one  
Simple but inefficient for large inputs

eg:  
Traveling Salesman Problem (TSP)

#### Divide and conquer

algorithmic paradigm used to solve complex problems   

- divide: break problem into smaller subproblems  
- conquer: recursively solve each subproblem  
- combine: combine the results

**Sorting algorithms**

merge sort and quick sort are of this type  
merge: split array in 2 halves, sort recursively each half,  merges sorted halves  
quick: Partitions around a pivot, then sorts subarrays

**Search agorithms**

binary serach: find middle element, recursively search left or right

### Bubble sort

![alt text](https://favtutor.com/resources/images/uploads/mceu_61632030011682402256084.png)

- compare 2 consecutive elements, if needed swap them  
- loop for all consecutive pairs, so N-1  
- repeat N times the cycle  

N is size of array  
if during a run nothing is swapped, stop early, it's already done  
just check it with a bool var

**Complexity**

in each loop, N-1 comparisons  
repeat loop N times  

$O(\,n(n-1)\,) = O(n^2)$  

### Selection sort

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcS_fHJsmypE4EvccWfS0RX_85zmcRezRgWsfg&s)

- check all values to find lowest  
- move it to start of array, before the first one, or swap it with first one    
- repeat N times, so collocate all values  

1- travel throught the array to find the minimum, swap it with A[0]  
2- travel through all A[1] until A[N-1] to find the minimum, put it in A[1]  

**Shifting problem**

if remove a value and put it at the start, before the first one the other values must do many shifts, which is time consuming   
so better to swap the value i picked with the first one, less shifts   

**Complexity**

to find smallest number must compare all those currently still in the unsorted array  
on average, those are $\frac{N}{2}$  
repeat N times, to build a sorted array of same size  

$O(\frac{n^2}{2}) = o(n^2)$  

**Bubble vs selection sort**

have same complexity, but  
selection: best and worst nearly same time  
bubbloe: best case runtime, when it's already ordered, is only $O(n)$

### Insertion sort

![alt text](https://miro.medium.com/v2/resize:fit:765/0*1zi2XtjiLXa3LYZh.PNG)

- take first value from unsorted part
- put it in correct place in sorted part (by moving it one place at a time)
- repeat N times, until unsorted part is empty

**Shifting problem**

same problem as selection sort  
indstead of removing and inserting the value i want to move:  
copy in a var the value i want to move  
shift all values on its left to the right, so gets overwritten, until spot where i want to place it is free  
then place is there

**Complexity**

worse casestill $O(n^2)$  
but average is $O(n \log n)$  

### Counting sort

conditions:  

- non-negative integers only  
- better if many repeated number (small range of values)  
to be effective there must be repeated numbers, unique values must be less then length of array  

- create an array for counting occurrences of each value  
in $a_i$, store how many times i appears  
- create the sorted array from this information

**Complexity**

k: unique values  
n: array length  

best: $O(n)$
worst: $O(n^2)$  
avg: $O(n+k)$  

### Radix sort

- check right-most digit, store in buckets from 0 to 9  
- reconstruct array by putting in sequence the stuff from from bucket 0, then 1 etc...
- move a digit left and repeat  

to work, first number to enter a bucket must be first number pulled out  

![alt-text](https://www.researchgate.net/publication/291086231/figure/fig1/AS:614214452404240@1523451545568/Simplistic-illustration-of-the-steps-performed-in-a-radix-sort-In-this-example-the.png)

**Complexity**

n: array length  
k: number of digits of higher number (often $k=\log n$)  

avg: $O(nk)$  
best: $O(n)$  
wrst: $O(n^2)$  

### Merge sort

- split array in single values (each value is an subarray of legnth 1)  
- merge subarrays, keeping them ordered

**Complexity**

$O(n \log n)$

### Quick sort

a divide-and-conquer algorithm  

- choose a pivot element (any one, don't care which)  
- partition the array: check all elements, put those smaller than pivot on its left, bigger on its right  
- pick 2 new pivots, one in left and one in right subarray, repeat

**Complexity**

best, avg: $O(n \log n)$  
worst: $O(n^2)$  

## Search algorithms

3 cases:  
1. unordered array: linear search  
2. ordered array: binary search
3. graph/tree: many types (DFS, BFS)  

**Array search**

ordered array: binary search, $O(\log _2 n)$  
check middle value: if value i want is higher, repeat with subarray higher and viceversa  

unordered array: linear search, $O(n)$  
check all values in order, brute-force  

interpolation search  
optimal for uniformly distributed data  
smarter version of binary search that estimates where to search using a linear interpolation formula, not just checks the middle    
$O(\log _2 \log _2 n)$  

in algorithm complexity, unless specified otherwise, log is in base 2  


**Tree search**

V: vertices  
E: edges  

DFS: depth first search, $O(V+E)$    
go as down and left as you can, then go up and as soon as you can go down again  
used to find paths in mazes, puzzles, or AI decision trees  

BFS: breadth first search, $O(V+E)$  
see all neighbours of current nodes, then pick non-finished leftmost node and see all neighbors of that
