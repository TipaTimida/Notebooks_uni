**Components**

1- editor  
wrire source files  

2- compiler  
translate in machine code, to getn an .o or .obj

3- linker  
join code with libraries, to get an .exe  

an IDE contains all these and a graphical interface

**Build**

start when click 'compile and run'  

1- preprocessing  
read preprocessing directives  

2- compilation  
translate in assembly   
then translate in binary object code 
if syntax error, build stops here  

3- linking  
link all object files  
if a definition is missing, get a linker error
