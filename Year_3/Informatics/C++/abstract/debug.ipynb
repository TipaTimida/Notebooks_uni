### Commands

**Graphical interface**

there are the lines of source code  
in Code:Blocks, there is arrow pointing to the first statement YET to be executed (so the next one it will execute)  

#### Stepping

execute the code one statement at a time  

1- step into: execute next statement  
if it's a func call, just call it, move to the header of that func  

2- step over: xecute next statement  
if it's a func call, execute it completely, don't step through it  
in Code Blocks, it's called 'Next line'

3- step out: finishes executing all the func i'm currently in    

#### Breaks

1- Run to cursor: only exist in graphical interface, instead of putting a break, click with the cursor where i want it to stop  

2- continue: runs until next breakpoint, or end of program  

3- start: start running, until next breakpoint

4- set next statement: change the point of execution to some other statement (jumping)    
so skip some code, don't execute it  
it changes next line to execute, not the current state of the program (current vars)  


**Jumps**

DO NOT jump into a different function  
bc i'm skipping setting up the stack and registers appropriately (like putting the return address in RET, or putting the params passed in the appropriate registers)  
so when it will try to do something or return, it will have indefined behaviour  
instead, jump to the line on the func i'm currently one that CALLS the func i want to see  

**Breakpoints**

set it: right click in the line, 'Toggle breakpoint'  
it's shown by a red circle on that line  
stops BEFORE executing that line (so when arrow is on top of the circle)

#### Stack

**What is it**

The stack is a region of memory used to manage active function calls  
When a function runs, it usually gets a stack frame containing things such as: local vars, registers, ret address

### Various tips

**std::cerr - validating your code flow**


what:  
when program crashes, print lines in part of the program to see if i got there  
like at the top of a func, to see if it was called  

how:  
use cerr to print, not cout, bc the first one is NOT buffered, so it prints in the line where it's called  
while cout is buffered, so if the program crashes before the buffer is full, it prints nothing, even though it reached that point 

other method:   
add the following statement to the top of your main() function:   
`std::cout << std::unitbuf;`   enables automatic flushing for std::cout, but only keep it for debugging, it diminishes performance  
use it when using a debugger, or it might not print stuff when i need it to  
maybe wrap it in proprocessor macros  
`#ifdef DEBUG`  
`std::cout << std::unitbuf;` 
`#endif`


**Logging**

to output log can use std::clog, but it uses the std error stream  
i could write it to file instead, but better to use an external lib  

example with the plog logger  

1- include the logger headers 
`#include <plog/Log.h>`  
`#include <plog/Initializers/RollingFileInitializer.h`

2- initialize the logger (at the start of main)
`int main() {` 
`plog::init(plog::debug, "Logfile.txt");` 

3- output to the log as if you were writing to the console  
`PLOGD << "main() called";`   
PLOGD is defined by the plog library

4- then, to turn off logging, change the initializing to:  
`plog::init(plog::none , "Logfile.txt");`
