### Binary system

**Decimal -> binary**

1- divide by 2 at each step, while storing the remainders (essentially, mod 2)  
2- repeat until get 0  
3- read the remainders them from the last to the first  

**Example**

10 / 2 = 5 resto 0  
5 / 2 = 2 resto 1  
2 / 2 = 1 resto 0  
1 / 2 = 0 resto 1  

$10_{10} = 1010_2$   

**Shortcut**

1- subtract the greatest power of 2 that fits in the number     
2- repeat on the remanider until i get 0  

for fractions the same thing, but subtract 2^-1, 2^-2, 2^-3 ... instead   

example:  
10 - 1(2^3) -0(2^2) -1(2^1)-0(2^0) =    
10 - 1x8 - 0x4 - 1x2 - 0x1 =   
10 - 8 - 2 = 0    

**Binary -> decimal**

sum all the powers of 2 where there is a 1  

#### Fractions

**Decimal -> binary**

1- multiply (not divide) the fraction by 2  
2- store the integer part (as remainder), continue on the fractional part until get 0  
3- read the integer parts from first to last (order opposite than for integers)  

**Example**

0.625 x 2 = 1.25 -> 1  
0.25 x 2 = 0.5 -> 0   
0.5 x 2 = 1 -> 1  

$0.625_{10} = 0.101_2$   

**Binary -> decimal**
  
like for integers, but using negative powers of 2  

example:  
$0.101_2 = 1 \frac 12 + 0 \frac 14 + 1 \frac 18 = 0.625$   

**General number: example**

$10_{10} = 1010_2$  
$0.625_{10}=0.101_2$  
so: $10.635_{10} = 1010.101_2$  

**Problem with writing 0.1 in binary**

0.4 x 2 = 0.8  
0.8 x 2 = 1.6  
0.6 x 2 = 1.2  

so restart, it never ends

### 2's complement

**What is it**

complement of a number A: smallest nuber you have to add to A to make the digits of the result overlow  
ex: 10s complement of 15 is 85, bc 15+85=100, now three digits, overflow  

in binary: add 2 numbers, if the result is 0, ignoring the overflow, those two numbers are complementary  

**Take 2's complement**

1- swap all bits (0 with 1)  
2- add 1

to go from 5 to -5 and viceversa (always add 1, never subtract it)

**Example**  

using 6 bits  
range signed: [0, 63]  
range unsigned: [-32, 31]

### Mobile comma / binary floating-point notation

#### How to write

1- convert a number to floating point  

$N = (−1)^s × 0.m × 2^{e−b}$  

2- write a floating point number  

$0b \; [s][e+b][m]$  

letter | name | meaning | values
---|---|---|---
s | sign bit | if N is positive or negative | 0 or 1 
m | mantissa | the decimal digits of N | anything 
e | exponent 
b | bias | used to represent number <1 without needing to add sign to exponent | if write a number with k bits, the bias is $b=2^{k-1}-1$  
e-b | real exponent

| Format               | Total bits | Sign (bits) | Exponent (bits) | Mantissa (bits) | Mantissa (decimal places) |
| -------------------- | ---------: | ---: | -------: | -------: | ------------------------: |
| **Half precision**   |         16 |    1 |        5 |       10 |               ~3–4 digits |
| **Single precision** |         32 |    1 |        8 |       23 |                 ~7 digits |
| **Double precision** |         64 |    1 |       11 |       52 |             ~15–16 digits |


**Floating-point arithmetic**

IEEE 754 Standard: defines how floating-point numbers are stored and processed

**Note**

most significant bit: leftmost bit  
like most significant digit in significant digits  

#### Example

**Decimal -> binary floating point**

m=1011

### Floating point: weird errors

```python
x = 0.1 + 0.1
print(f'{x:.30f}')
print(x == 0.2)

x = 0.1 + 0.2
print(f'{x:.30f}')
print(x == 0.3)
```

**Explanation**

Computers represent decimal numbers in binary  
using IEEE 754 double-precision floating-point format  

Some decimal numbers cannot be exactly represented in binary, like 0.2 or 0.3  
bc the mantissa has an infinite binary representation 

$0.200000000000000011102230246252$  
$0.200000000000000011102230246252$  


$0.300000000000000044408920985006 $ (Stored x)  
$0.299999999999999988897769753748 $ (Stored 0.3)

```python
# e means 10^

x1 = 1.0e25
x2 = 1.0

print(x1 + x2 - x1)
print(x1 - x1 + x2)
```

**Explanation**

$x_1=10^{25}$  
$x_2 = 1$  
in first case, since $x1 >> x2$, <u>the mantissa (sequence of digits after the comma) is not long enough to hold the $1$</u>      
so it's lost  
essentially $x1 +x2 = 10^{25}$  
so, after, deleting x1, it doesn't remember that 1    

**Note mantissa**

the mantissa has 52-bit precision (it's long 52 bits)  
but this is expressed in binary  
thinking itin decimal, it's 
$2^{52} \sim 10^{15.9}$  

meaning, when thinking about a decimal number, the mantissa would be ling 15 digits  

after that, can't represent smaller values  
which is the reason why $10^{25}$ cannot hold the 1, the mantissa is not long enough
