### What it is

**Namespace**

provides another type of scope region (called namespace scope), that allows you to declare or define names inside of it for the purpose of disambiguation  
bc the names declared in a namespace are isolated from names declared in other scopes, allowing such names to exist without conflict  

essentially: add a prefix to the names, so they're technically different names  
an example is the std namespace, used for I/O

**What it contains**

A namespace can only contain declarations and definitions  
Executable statements are only allowed as part of a definition (eg. of a function)  

**Global namespae**

any name that is not defined inside a class, function, or a namespace is considered to be part of an implicitly-defined namespace, called the global namespace (global scope)  
that's why funcs have a global scope

best practice: never define vars as globals, unless they're const  

### Code

**Access stuff in a namespace**

different ways to do it:  

1- scope resolution operator (explicitate the namespace as a prefix)  
`std::cout << "Hello world!";`   
The :: symbol is an operator called the 'scope resolution operator' (put the namespace on the left)  
to mean the global namespace, put :: without anything on its left  

2- using statements  
`using namespace std;`    
allows us to access names in the std namespace with no namespace prefix  
not good, avoid it  

**Using statements**

1- bad way  
`using namespace std;`   
can always skip std   

2- good way  
`using std::cout;`  
can use cout instead of sd::cout  



**Create a user-defined namespace**

in namespace identifier, sometimes the first letter is capitalized too  

when defining funcs inside a namespace, if call a func WITHOUT specifying its namespace:  
will first check if such a func exist in the same namespace i'm writing in (if yes, used that one)  
and if not, cheks if it exists in the global namespace  

useful, bc when writing in a namespace can easily use other funcs defined in that same namespace, without specifying the namespace (writing ::) everytime   

can nest namespaces inside another  
to call a func in it, must specify the path, using all the namespace parents    
Foo::Goo::add(1, 2)  

**Namespace aliases**

if a namespace has a long name, can define an alias for callying it  
the alias is defined OUTSIDE of the mainspace, in a normal func, for exmaple in main()  
obv, the alias is an obj, only has scope where i define it  

namespace Active = Foo::Goo;  
std::cout << Active::add(1, 2)

#### Declare namespace in multi-file

**How**  

in the file with the forward declarations, must also put the forward declarations of all the funcs defined in a namespace  
and put them in a namespace with same name  

```python
add.h

#ifndef ADD_H
#define ADD_H

namespace BasicMath
{
    // function add() is part of namespace BasicMath
    int add(int x, int y);
}

#endif
```

```python
add.cpp

#include "add.h"

namespace BasicMath
{
    // define the function add() inside namespace BasicMath
    int add(int x, int y)
    {
        return x + y;
    }
}
```

```python
main.cpp  

#include "add.h" // for BasicMath::add()

#include <iostream>

int main()
{
    std::cout << BasicMath::add(4, 3) << '\n';

    return 0;
}
```

**Note**

the definitions inside a namespace can be split in more files  
this is used by the standard library std, it's why i can import just a piece that i need, for example <iostream>, instead of all of it
