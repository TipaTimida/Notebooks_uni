**Recap**

directive | what | when to use
---|---|---
#include | paste another file in this point | to import libs and header files 
#define | define a macro, a text substitution | no: for text substitutions <br> yes: for conditional compilation (set up a property)  
#undef | revoke #define 
#ifdef / #ifndef | conditional compilation 
#endif | end conditional compilation block
#pragma | specific instruction for the compiler  

### Preprocessing directives

**Preprocessing**

a stage before compiling  
doesn't change the code file  
all changes made by the preprocessor happen either temporarily in-memory or using temporary files  

**General**

1- how to write  
start with #  
end in a newline (NOT a semicolo,)  
have different syntax, not C++ syntax  

2- where to write  
can be defined wherever, but generally at top of the file (unless i need them in some particular place)  
are processed BEFORE compiling  

**Scope**

directives are only valid from the point of definition to the end of the file in which they are defined (not in all file)  
Directives defined in one file do not have any impact on other files (unless they are #included into another file)  

### Include

**How it works**

#include simply takes the file i want to import, and pastes its contents instead of the #include line  

so don't write definitions in header file, or risk to have it defined twice, and give error 

**Header file names**

name of the file to #include is written inside:  

1- in angled brackets:  
for files i didn't write myself  
so look for them in specific IDE directories  
`#include <iostream>`    

2- in double quotes:   
for files i wrote myself  
look for them in current dir (if don't find it, look in IDE directories)  
`#include "miofile.h"`    

**Libraries**

iostream is part of C++ standard library, don't need to download it   
and don't end in .h, bc iostream.h is the old version, which don't use the std namespace (do not use it)  

### Define

**What is it**

this statement is called a MACRO  
it's a rule that defines how input text is converted into replacement output text  

is MACRO the whole statement, or just the args??

usually the name are in uppercase, with words separated by underscores  

**Types of usage**

1- object-like macros WITH substitution text   
#define OLD_TEXT new_text  
every time OLD_TEXT appears in the code, replace it with new_text   
it's risky, perchè non fa controlli di tipo  
to do text substitutions, use: 'constexpr' or const vars (assignments of vars known at compile time) or 'using' (for type aliases)  

2- object-like macros WITHOUT substitution text
#define IDENTIFIER  
technically, substitutes IDENTIFIER with nothing (except when IDENTIFIER is in another preprocessor director, or #ifdef would not work)  
pratically, set a property called IDENTIFIER  
used for: conditional compilation  

#### Conditional compilation

**Conditional compilation**

what is it:  
specify under what conditions something will or won’t compile  
better method than commenting out pieces of code, since nested comments don't work  
instead of commenting out, just set a macro  

how to do:  
code | what
---|---
#if IDENTIFIER   | compile only if i set this macro  
#ifnot IDENTIFIER   | compile if i didn't set this macro  
#if 0   | comment a block out  
#if 1   | reenable it  
#endif | close the block on where i'm doing conditional compilation <br> like closing brackets of 'if'
