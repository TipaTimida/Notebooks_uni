### Static

**Stateful**

an algorithm (function) can be:  

1- stateful: retains some info beteween calls   
state: current values held in stateful variables  

2- stateless: not store any information   so, must be given all the information it needs to work with whenever it is called  

**Static /stateful var**

the var s_state has:  
1- local scope: can only be accessed inside plusOne()  
2- static lifetime: it exists until the program terminates  

so, 'static' gives it static lifetime  
it's not destroyed when the function closes  
so if call again the func, the same s_state as before returns in scope  
and obv, it's initialized only the first time it's assigned  
the next times, the line with static is ignored, so s_state is NOT reassigned 3  

```python
#include <iostream>

int plusOne()
{
    static int s_state { 3 }; // only initialized the first time this function is called

    // Generate the next number

    ++s_state;      // first we modify the state
    return s_state; // then we use the new state to generate the next number in the sequence
}

int main()
{
    std::cout << plusOne() << '\n';
    std::cout << plusOne() << '\n';
    std::cout << plusOne() << '\n';

    return 0;
}

prints: 4,5,6  
```

### PRNG - Pseudo-random number generators

**How it works**

1- define a func with a state var    
2- in the func, do some big operation that make the state var overflow  
3- return the modulo of the state var (so have a number in an acceptable range)  

**How many squences can generate**

depends on the state  
if the state has 128 bits, can generate up to 2^128 different sequences  

**Seed**

the func is deterministic  
so if the initial value of the static var stays the same, the sequence of numbers generated stays the same   

seed: first value of the static var  

**Seeding**

seeding = picking a seed 

a seed must be different every time the program is run, or always get the same sequence  
but can't use th random number generation to produce it, or i'm in a catch-22  

there are 2 ways to find a seed:  
1- use the system clock  
2- se the system's random device (faster to code)  
is an implementation-defined PRNG, store in the random library  
implementation-defined means that the way it works changes on different OS: might take the number from a special hardware, the OS itself, a system entropy pool...

**Why not just using random device?**

because std::random_device is implementation defined (so, changes depending on my OS)  
so we can’t assume much about it  
it may be expensive to access, or cause our program to pause while waiting for more random numbers to become available  
or cause issues for other programs requesting random numbers via the same method  

so, just use it once for seeding 

**When to seed**

only create and seed the random generator ONCE in the program  
if the gnerator is part of a function that i'll call more times, make its declaration 'static'  

```python
Generate 16-bit pseudo-random numbers


void seedPRNG(unsigned int seed)
{
    g_state = seed;
}

unsigned int LCG16() // our PRNG
{
    g_state = 8253729 * g_state + 2396403; 
    return g_state % 32768;
}
```
