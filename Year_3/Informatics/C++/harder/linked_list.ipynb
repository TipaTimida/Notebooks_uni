**What is it**

dynamic data structure that uses pointers for its implementation    
an array that can grow and shrink as needed, from any point in the array  

**Pros**

1- Items can be added or removed from the middle of the list  
2- don't need to know size at the start  


**Cons**  

1- no random access: to get a value in the middle, must traverse all the items before it    
2- need dynamic allocation and pointers, so code is complex   
3- Linked lists have a much larger overhead over arrays, since linked list items are dynamically allocated (which is less efficient in memory usage) and each item in the list also must store an additional pointer ??? 

**Structure**

a set of dynamically allocated nodes  
each node (a struct) contains:  
1- the value of that node 
2- a pointer to the next node  

then, to use it, need 2 other pointers:  
head: pointer to first node  
tail: pointer to last node
