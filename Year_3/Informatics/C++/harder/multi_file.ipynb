### Header files

**What files in a c++ program**

1- code files  
extension cpp

2- header files  
extension .h or .hpp  
used write forward declarations in one place and import them in a code file (so has list of all dependencies)  
for example, iostream is an header file (has the declaration, not definition, those are stored where??? in the compiler itself???)  
don't define them here, or will be like they're defined multiple times, one in each code files where it's imported  

so declare funcs in header file  
define them in a code file with same name as header file (#define here too the header file, makes it easier to debug, if made a mistake)    
and put the main program in another code file, called main  

could aldo have mutiple header files, each qith all the funcs needed for a specific funcitonality  

#include simply takes the file i want to import, and pastes its contents instead of the #define line  

so don't write definitions in header file, or risk to have it defined twice, and give error 

do not #include cpp files  
those should be added to the directory and compiled  

**Note**

in header files, might need to include other headers  
that is ok  
but better to explicitly include them in the code files too, just inc ase the header files get changed  

**List of stuff to include**

1- the paired header file for this code file (e.g. add.cpp should #include "add.h")    
2- Other headers from the same project (e.g. #include "mymath.h")  
3- 3rd party library headers (e.g. #include <boost/tuple/tuple.hpp>)  
4- Standard library headers (e.g. #include <iostream>)  

#### Header guards

**Problem**

i wrote function definitions in header files (.h)  
each function can be defined only once on a file  
so if i include twice the same header, get error  

solution:  
put a header guard, to prevent including the header file twice

**What is header guars**

to prevent defining a func twice in the same file 
check if this header file has already been included
if not, import it, and check off that it's been defined  

the name to define is generally the full filename of the header file, typed in all caps, using underscores for spaces or punctuation  

just put the declaration in the header, not the definition, bc it will be included in all files, and would give link problems  

**Methods to write it**

1- conditional compilation  
check if it's alady been included by checking a flag  
in case, include it  

2- pragma once   
but #pragma are defined by the compiler, so might not work on all compilers  


// Header guards

#ifndef SQUARE_H
#define SQUARE_H

// declarations  

#endif


// NEW METHOD

#pragma once    
// declarations  


### Linkage ???

**Linkage**

determines if a var or func defined in a source file is visible to other files during the linking phase  

**Types of linkage**

what | internal | external
---|---|---
visibility | local | global
who can access | only this file | other files too, if they import it with 'extern'
how to do | 1- const global vars  <br> 2- keyword static <br> 3- anonymous namespace | 1- non-const global vars 


**Global vars**

non-costn global vars, have external linkage  
const ones, have internal linkage  

but they are shared the same way, using extern  

**ODR - one definition rule**

when: for vars with oexternal linkage  
what: must be defined only once  

consequence:  
don't put vars in headers (or if you do, use header guards)  
ese, risk including it twice, and linker throws an error  

**Inline var**

can be defined in a header file and included in multiple .cpp files without causing a “multiple definition” linker error  
so can omit header guard, even if i include it twice it's ok  

// Globals.h     
inline int score = 0;
