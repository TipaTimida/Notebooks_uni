### Dynamic allocation

**What is it**

allows to allocate data whose size or lifetime is not known at compile time  

**Code explanation**

new int;  
dynamically allocate an integer, meaning request an integer’s worth of memory from the OS   
return value ia a pointer containing the memory address of the newly allocated object  
so, use it to initialize a pointer  

new (std::nothrow) int  
value will be set to a null pointer if the integer allocation fails  
but don't use it, better to thrown an catch an error     

delete  
returns the memory allocated to the OS  
doesn't actually delete anything   
it does NOT change the scope of my pointer, nor destroys, just turns it into a dangling pointer      
deleting a null pointer: nothing happens  
deleting a pointer pointing to the stack, not heap: causes trouble, don't do  

**Pointers and arrays**

int* ptr;  
ptr = new int[100];  

creates an array of 100 int objects on the heap  
and and returns the address of the first element  
so, the ptr points to the first element of the array  
so, can call ptr[0], ptr[1] ...  
then, must deallocate with:  
delete[] ptr;  

**For arrays**

works the same wave, but the commands new and delete take []  
inside the [] put the size of the array  
The length of dynamically allocated arrays has type std::size_t  
remember [] after delete, or get undefined behaviour  

size_t size {5};  
int* array { new int[size] {} };  
delete[] array;  

note: this lets me make an array whose size is only known at run-time, but CANNOT make an array that changes size  
for that, use a vector  

**For classes**

put () after the type, bc means call the constructor  
so i'm initializing it too, i can call funcs on it  

Animal* a = new Animal();   
a->sound();  


### Common problems

### Dangling pointer

**Dangling pointer**

a pointer that points to deallocated memory    
causes undefined behaviour  

solution:  
when deallocating memory, rememeber to set the pointer to nullptr  
except if it's going out of scope right after, in this case it's fine  

**Null pointer**

useful to see which memory is not allocated  

#### Memory leak

**Memory leak**

what: the program loses the address of some bit of dynamically allocated memory before giving it back to the OS  

problems: eat up free memory while the program is running    
Quella memoria rimane occupata (e inutilizzabile) fino alla chiusura del programma. 


note: when the program ends, all memory gets deallocated  

how can happen:  
pointer holding the address goes ot of scope ot gets overwritten  

**Explanation**

lifetime of:  
dynamically allocated memory: until explicitely deleted  
pointers that hold it: until they're in scope, like normal vars  

problem: if pointer goes out of scope, but i haven't deallocated yet  
i lost the address of the memory, so now i cannot deallocated it  
it's a waste, bc i cannot use it, and the os cannot give it to any other program  
it's a memory leak

### Memory errors

**1. Memory leak**

memory allocated in the Heap don't get released with delete  
so cannot use it, nor reassign it  
it wastes RAM  

cause: forgot to use 'delete' bc pointer to allocated memory went out of scope  

**2. Buffer overflow**

writing more than an array can contain  
corrupts data in subsequent memory cells  

**3. Stack overflow**

no more space in the stack  

cause: allocated arrays too big, or bc of an infinite recursion  

**Segmentation fault**

illegal access to memory  

cause: dereferencing nnullptr or dangling pointer, not initializing pointer,
