**Unit testing**

Testing a small part of your code in isolation to ensure that “unit” of code is correct   
eg. check that each function works, before adding the call to them in the program  

**How to do unit testing**

for each func i write, write also another function that tests it  
make it automated, so it has no params, test the function on some literals  
if the func i want to test ahas in 'if' inside, put enough asserts to check all possible outcomes  
and use assert in this test  

the asserts can be enclosed in the preprocessor definition DEBUG  
so i can turn it off while writing code, so it don't slow my program by running tests each time  

**Unit testing frameworks**

are third-party libs    
help simplify the process of writing, maintaining, and executing unit tests  


**Function coverage**

means how much code is being tested  

1- statement coverage  
means each statement must be run  

2- branch coverage  
means, if have an 'if', must do a test for each possible outcome  

3- loop coverage  
if have a loop (for, while), should check that it works properly if run 0,1,2 times  
if works 2, will probably work more than 2  
valled the '0, 1, 2 test'  

**Testing input**

if the func to test needs input / params, test many categories 

1- if arg is an int  
test a positive one, 0, and a engative one  
and if possible, also overflow  

2- if arg is float  
check nunmbers with precision issues  
like 0.1 and -0.1 (slightly larger than expected)  
or 0.7 and -0.7 (slightly smaller than expected)  

3- if arg is string  
check empty string, alphanumeric string, string with whitespace (leading, trailing, and inner), string that is all whitespace  


**Types of errors**

syntax error: the grammar is wrong, it doesn't run, compiler gives error    

semantic error: it technically runs, but don't work well, don't do what i wanted it to, compiler don't give error (so harder to catch)  
often causes undefined behaviour  

**Defensive programming**

trying to anticipate all of the ways software can be misused (either by end-users, or by developers)  

i need to handle tha bad paths:  
1- when a function return as a failure, rather than succesful  
2- when get input or args which are wrong semantically  
eg. get input of wrong type  
eg. try to open file that don't exist  

### Input validation

**Input validation**

check that the input entered is ok  

types:  
1- Inline (as the user types)  
Prevent the user from typing invalid input in the first place  
(have a checking function that activate each time the user presses a key, if it returns false it don't show the result of pressing that key)  
2- Post-entry (after has types)  
try to extract with >>, and handle the errors    

**How to handle invalid input**

1- flush the buffer (optional)   
or if extraction failed, need to make cin working again  
2- have the user try again  

**Tip**

at the start of a func, use an if to check the WRONG cases, and if so, do something  
do not check for the right case, or becomes more nested  

### Assertion

**What it does**

if condition = true, nothing happens   
if condition = false, give error  
and tells me the line where the failed assertion was, useful for debugging  

**Usage**

1- at start of function to validate function params (instead of try...except)    

2- to test functions  

3- to signal unimplemented features  

**Signal unimplemented features**

assert(moved && "Need to handle case where student was just moved to another classroom");    

document cases that were not implemented because they were not needed at the time the programmer wrote the code  
That way, if a developer encounters a situation where this case is needed, the code will fail with a useful error message  
and the programmer can then determine how to implement that case.


**Using it**

asserts take some ime to run  
and should only be used when debugging, not in the production code  

how to disable asserts:  
`#define NDEBUG`  
must put BEFORE all #include  
NDEBUG is a special word, the code already knows it mean 'ignore the asserts'  
if want them to run again, comment this line  

how to enable asserts:  
`#undef NDEBUG`  

**Static_assert**

an assertion that is checked at compile-time rather than at runtime  
needs no lib, unlike assert  

when i can, static_assert is better than assert  

**Assert vs excception**

assert: for stuff that can't solve, only solution is to exit the code  
and stuff that is the fault of the programmer  
so, can use it to test that functions work  

exceptions: for stuff that can solve  
eg. validate input  

### Excpetions

**Commands**

1- try  
define the block where the code might fail  

2- throw  
signal a failure  
give a name to the error  

3- catch  
if receives an error with such name, do something  
catch(...) handles error with every name
