### As-if rule

**Nomenclature**

optimization: making software more efficient  

optimizer: program that optimizes another program   

as-if rule:  the compiler can modify a program however it likes in order to produce more optimized code, so long as those modifications do not affect a program’s “observable behavior”.  
so doesn't change the source code, just what happens in the background, without changing observable behaviour  

**Compile-time evaluation**

by default, expressions are avaluated at run-time  

but can fully or partially evaluate certain expressions at compile-time (rather than at run-time)  
this way, run-time is shorter  

there are many techniques  
rem, these are executed by the compiler, not by me changing the source code  

must be enabled in the IDE  
by default it's not used, bc makes it hard to use the debugger  

**1. constant folding**

replaces expressions that have literal operands with the result of the expression  
int x { 3 + 4 };  
is computed at compile.time, so at runtime see directly:  
int x { 7 };  


**2. constant propagation**

if a var has constant value, replace all its apppearances with a literal with its value  
so during run-time don't need to fetch the value every time  
a var which don't have the qualifier const, but that in practice it's const, might be propagated too   
but it definitely is if i call it const, so do it  

**Dead code elimination**

eliminate code that has no effect on the program's behaviour  
eg. vars initialized but not used  

### Compile-time programming

**What**

specify that something must be evaluated at compile-time  
so the program runs faster  

1- constexpr vars  
2- constexpr funcs  
3- func templates 
4- assert, static_assert
