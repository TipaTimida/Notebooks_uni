**Size and decay**

a statically declared array maintains its size only until it's in scope  
when it's not in scope (eg. when passed as arg to another function): the array decays to a POINTER to the memory address of the first element of the array  
using sizeof now gives me the size of the pointer

Se vuoi passare un array a una funzione e conoscerne la dimensione, devi
passare la dimensione come secondo parametro   

!!!  
a vector doesn't decay to a pointer when passed as an argument  

**Reassignment**

CAN change the values stored in an array, one by one  
CANNOT reassign an entire built-in array at once  

std::array can, instead  

**Array in stack vs heap**

what | static | dynamic
---|---|---
declaration | char str []; | char * str = new char [20];
strlen | how long is the string actually stored in the array | ""
sizeof | if array has decaded to a pointer: size of the pointer, 8 <br> bc passing an array as arg to sizeof() make it decay to a pointer | total size of the memory allocated for the array, 20  

**Array and pointer equivalence**

an array is also a pointer to its first element  
soto get an element, can write both (rememeber pointer arithmetic):  

int arr[5];   
arr[3];  
*(arr + 3);
