### Whitespace

**whitespace-independent language**

does not enforce a formatting, can put as many tabs as i want  
python, instead, throws error if indentation is not correct  

1- inside a string, all whitespaces are taken literally  
ma non posso andare a capo  

2- Quoted text separated by nothing but whitespace (spaces, tabs, or newlines) will be concatenated (so don't need to put << or + to concatenare, but do it for clarity)

3- Preprocessor directives must be placed on separate lines  


**Format best practices**

1- for functions, or other text inside two braces:  
the opening brace should be either after the function declaration, or in the next line  
the body of the function should be indented one tab to the right of the braces  

2- to split a statement on two line, the part on the second line should be indented with one more tab (to understand it's a continutation of the previous line, not a new statement)
