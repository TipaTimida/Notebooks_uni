### What is it

**What it is**

a var used to store memory addresses  
so its value is a memory address  


it's a reference to a var in memory    
They're "pointing" to a memory-address where a variable is stored  

**Type**

needs a datatype that matches the referenced variable  
bc the data in memory is just byte, is 

why:  
bc stuff in memory is binary, is read differently depending on what i want to read it as (int, char)  
so knowing the type it's necessary for deferencing  

### Code 

**Declare**

`int *b {};`

declare:  
the * could be anywhere, attached to int, b, or separated from both  
bc c++ don't care about spaces   

initialize:  
is initialized it like normal vars   
generally initialized with 0, a non accessible address, for safety reasons, or to nullptr

**Nomenclature**

*p, * is called the dereference operator  
bc a pointer is a reference to a var, and * takes the var it is referenceing too  
so take away the reference, go back to the original var  

dereference a pointer: access the value stored at the memory address contained in the pointer

**Attention: dereference or type?**

when initializing, the * is not for deferencing, but it's a synonim of 'pointer', so it's saying the type  
so on the right i need to know the value i want to store in the pointer, so the address i want to point to, not the value i want to put at that address    
so attach the * to int instead of the pointer name, so make less confusion, and remember it's a part of the type, NOT the dereference operator

### Void pointers 

**What**

can point to a var of any type (can be typecasted to any type)  

problem:  
cannot directly dereference it (cannot access the value at the address it points to)  
bc i don't know what type this value it points to is, so don't know how to read it (a byte can have different meaning if interpreted as an int, char...)  

solution:  
to dereference it, first need to cast it (convert it to the correct pointer type)  
only in C, not in C++: can also equalize the void pointer with one of the type i want

### Other types of pointer

**Null pointer**

means: points to address 0x0  

how to use: 
1- in 'if' conditions, to check if (p=nullptr)  
2- initialize pointer as null, instead of leaving them dangling  

NOT to do: dereference it  
bc cannot write at address 0x0  
if try, i get segmentation fault  

`int* p = nullptr;` 


**Dangling pointer**

contains an address, but the object at that address was destroyed (if it's out of scope it's not dangling, that's just passing by reference)  

so the value at that address ia a random one, don't mean anything  
if dereference it, undefined behaviour  

out of scope vs destroyed:  
out of scope: currently you cannot call it, but later i will go back to a func where it's in scope again  
destroyed: it won't be in scope ever again (bc i closed the func or loop that created, or maybe i destroyed an instance of a class)

### Lvalue and rvalue

**What they are**

they're types of expressions  
depends on what is left after evaluating the expression  
(since they're expressions, they can be sued aniqhere an expression is allowed)  

1- l-value:  
evaluates to: an identifiable object (that persist past the end of the expression) (var, func WITH AN IDENTIFIER)   
this can also be modifiable or non-modifiable, depending if the obj it evaluates to is const or not  
so something i can save a value in  

2- r-value:   
evaluates to: a temporary value (literal)  

an l-value can be implicitly converted to r-vale, but not the opposite  

**Usage**

the operand = requires:  
left operand: l-value (cioè, must be a var)  
right operand: r-value (cioè, must be a value)   

this is why x=5 is ok, but 5=x is not    
x=y is ok too, bc y undergoes conversion from l-value to r-value  

**Examples**

1- Compilation error  
(a - b) = 7;   
bc the operator - gives an r-value (a temporary value, a literal)  
which don't have a persistent memory address, where i can assign the value 7   

2- Valid code  
(a -= b) = 7;   
bc the operator -= returns the left operans, a, after it's been decreased, so an l-value  
a has a memory address, so i can assign it the value 7

### Arithmetic

**Pointer arithmetic**

since they represent addresses, arithmetic takes it into account  
arithmetic ops take into the account the dimension of the data it's pointing to (scaling)  

**Explanation**

if add +1 to a pointer, i'm saying go to the next cell, NOT add 11 to the address  
so must add to the address the size of a memory cell for this type, NOT just 1  

**Operations**

p + N  
add N × sizeof(type) to the address  
so, N is the number of var cells to skip, increase the pointer accordingly  

p-q  
return number of cells between pointers
