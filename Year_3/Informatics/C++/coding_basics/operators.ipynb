### Order

**How to do an expression**

1- evaluate the operands   
in random order, depends on architecture  
except if have && or ||, those evaluate first the left operand, bc might do short-circuit  

2- evaluate the operators  
order according to their precedence and associativity, or brackets   
for each operation, attention to what it returns, if the left or right operand (usually the left)  

#### Order of operators

**Rules**

1- precedence  
operators with a higher precedence level are grouped with operands first    
eg. multiplication happens before addition  
1 + 2x3 = 1 + (2x3)

2- associativiy  
If two operators with the same precedence level are adjacent to each other in an expression, the operator’s associativity tells the compiler whether to evaluate the operators (not the operands!) from left to right or from right to left  
eg. subtraction has associativity from left to right, so the leftmost '-' works first  
7-4-1 = (7-4) -1  
eg. extraction has associativty to the left  
(std::cout >> a ) >> b  

**Precedence**

can assign a level of precedence to an operator, from 1 to 17  
1 is highest precedence  

1- POST increment  ++, --  
2- PRE increment ++, --  
3- maths  
order: like in normal algebra, % at same time as * and /    
4- relational:  >= and <=, == and !=    
5- bitwise:  ==, !, &, ^, |  
not, and, xor, or      
6- logic: !, == and !=, &&, ||  
7- assignment operator =, +=, -= ...
has the lowest precedence, so i know that an expression will fully evaluate before assignment  

note: unary operators, like -x and !x (minus and not), have really high precedence  
in maths and logic expressions, are done first  

relational has more precedence than logic, so first do statements like x>y to give true or  (1 or 0), then use AND, OR... to combine the true and false  

note:  
&&, ||: logical operators  
&, |: bitwise operators  

**Associativity**

nearly all operators have left ro right  
the only ones with right to left are:  

1- those that contain =  
like =, +=, -=, *=  ...   

2- increments  
++ , --    

3- logical negations  
!, not

example: x  = y *= z;  
is read as: x = (y *= z);

**Tip**

the order is always univocally decided even if i don't put parenthesis  
but put them anyway, so it's easier to read and remember  

if there is only one assignment operator, don't need to put parenthesis on its right side, since i know it will evaluate completely, since assignment has lowest precedence  
but if there are 2 assignment operators, maybe put parenthesis, it's easier to read  

**Example**

assume there are no parenthesis, what is the precedence order in these expressions?  

logic:  
(a || (b && c)) || d;  
&& has highest precedence  
|| has left to rigt associativity, so start with using left operand  

logic + relational:  
write: (value == 0 || value == 1)
evaluate to: ((value == 0) || (value == 1))

maths:  
z *= ((++y) + 5);  
increment ++ has high precedence  
then *= has really low precedence, so finish evaluating all the stuff on the left

exception:

#### Short-circuit

**What**

only work with LOGICAL operator  
if result is obvious from the first operand, don't even evaluate the second  

cases:  
A && B : if A=false, don't evaluate B  
A || B : if A=true, don't evaluate B  

**Notes**  

rem: AND and OR must evaluate forst the left operand, then the right (while in other operators it's random)  
so can short circuit only if the operand that make the result sure is the FIRST one  
true || x : yes shortcircuit  
x || true : no shortcircuit   

**Problem**

if the operator was an expression that contained operators with a side effect, like an increment or a function call, it doesn't happen 

**Note on shortcircuit**

if ( n2 != 0 && n1 % n2 == 0) { ... }  

must put the check that n2 is non-0 as first operand  
so if gives false, it short-circuits and doesn't try to do the divsion by n2, which would cause errors

### Returns and side effects

has some observable effect beyond producing a return value  

example of side effects:   

1- modify global state (modify vars)

2- doing input or output

3- modify param (by using ref or pointer)

### Particular operators

#### Division

**Types of division**

1- floating point division  
when: at least one of the operands is floating point value (float or double)  
returns: a floating point value  

2- integer division  
when: both operands are int   
returns: the FLOOR of the division   
bc wants to return an int, so it truncates the decimal part  

**Why**

arithmetic operators must have operands of matching types  
so if have an int and a double, the int gets converted to double (not the opposite, bc that would be a narrowing conversion)

**Problem: how to do floating point division with integers**

cast (static or implicit) at least one integer as floating point   
`int x{ 7 }, y{ 4 };`   
`z = static_cast<double>(x) / y` 

**Error: late casting**

double y = double (10/3) ;   

this returns 3.0, not 3.33  
bc 10/2 is integer division, so return 3  
then, cast 3 as double   

correct methog:  
double y = 10.0 / 3

**Problem 2: divide by 0**

divide by 0 : undefined behaviour   
divide by 0.0 : get NaN or Inf, depend on the architecture

#### Increment

increment | x++ | ++x
---|---|---
is operand x incremented? | yes |yes 
what is return value? | x BEFORE increment | x AFTER increment 

so there is difference only if i'm saving the result (return value) somewhere  
if not, and i just want to increment x, they do the same thing  

**Performance**

x++ is slower: must create a temporary copy of x to return it, wasting resources  
but for small var types, don't care  

**Pre vs post**

1- prefix-increment: ++x  
Increment  the operand x, then return x   
what increments: the operand  
what returns: the operand (after being incremented)  

2- postfix-increment: x++  
create a temmporary copy of x, then increment the operand x, then return the copy   
modifies the operand, but returns the original value of the operand (so the operand stays modified, i'll see it if i call it later, but in this line i still get the original value)  
what increments: the operand  
what returns: the copy (has not been incremented)  


if can choose, prefix is more efficient, since don't need to create a temporary copy  

**Example**

1- Pre    
int x { 5 };  
int y { ++x };   
first increment x to 6, then return it / evaluate it    
so get x=5, y=6    

2- Post   
int x { 5 };  
int y { x++ };      
first make a copy of x, then increment the operand (which is why i get x=6, i'm modifying the operand)  
the, return the copy, who has not been incremented, so is still 5   
so get x=6, y=5  
so the original operand x is modified to 6, i'll see if i try to call it, but y gets the original value, so 5  

#### Comma

**What is it**

it has the lowest precedence, even lower than assignment  

what it does: x,y  
evaluate x, then evaluate y, then return y  

**Usage**

std::cout << (++x, ++y) << '\n';   
increments both operands, but return (so print) only y  

**Good practice**

do not use the comma operator, it's bad to read  
just write the two evaluations in different lines  

the , used to separate vars is NOT the comma operator, it's the comma separator, it's ok to use  
example of comma separator:  
add(x, y);  
int x {1}, y {2};
