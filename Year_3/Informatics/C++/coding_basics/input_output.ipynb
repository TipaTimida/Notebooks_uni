### Input

**Commands**

command | `std::cin >> x >> y` | `std::getline(std::cin >> std::ws, name); `  
---|---|---
discard LEADING spaces | yes | no (yes, if give it `>> std::ws`)

#### Explanation

**How >> works**

steps:  
1- discard leading whitespaces  
2- check if buffer is empty  
ifyes, ask user for input  
if no, don't ask for input  
3- extracts as many consecutive characters as it can, until it encounters either a newline character (representing the end of the line of input) or a character that is not valid for the variable being extracted to  

result of the extraction:  
1- if at step 3 extracted at least a character: success  
2- if a step 3 didn't extract anything: failure  
the var i'm writing to gets assigned 0, and any future extractions will immediately fail, until i flush the cin   

**Buffer**

when arrive at an input command:  
1- if buffer is empty: wait for me to write in it   
2- if buffer is non-empty: do not give me the chance to write in it, just extract from it  
(don't check what it's in it, so the extraction could fail)  

write in it:    
can write whatever i want, until i press Enter  
which gets saved as an escape sequence  
all this stuff, \n compreso, gets saved in the input buffer  

extract from it:  
according to command used, check if must read leading whitespaces, and until when to extract  
essentially, extract as many consecutive characters as it can, until a char that cannot be stored in this type of var (any whitespace for cin, a \n for getline())   

if extraction fails:  
means buffer is non-empty, but the first char cannot be put in the var (not the right type for the var)    
what happens: put var=0    
until i flush the input buffer, cannot do any more extractions   
flush means clean it, remove all data in it  

why buffering:  
can do multiple extractions on same input, is quicker  

**Leading whitespace**

what it means:  
when receive input command, FIRST removes leading spaces, THEN checks if buffer is now empty  

problem:  
after i extracted everything all i could from it, a buffer is left with \n (the enter key)  
so \n is a leading space  
if next input request is `cin >>`: \n is removed, so sees buffer empty, asks me for input  
if next input is `getline()`: \n is left there, so sees buffer non-empty, so stores '\n' in my var, don't ask me for input  
solution: use input manipulator std::ws, to tell getline() to ignore leading spaces     
std::getline(std::cin >> std::ws, name); 

note: technically it's >> than ignores the leading spaces, not cin  
which is why just putting cin inside getline() don't ignore the leading spaces 

**Examples input buffering**

std::cin >> x;

---|---|---

#### Explanation

**Buffering**

The data from each output request (all things concatenated after cout) is appended to an output buffer  
Later, data from (the front of) the output buffer is flushed to the output device (the console) (FIFO, first in firdt out)  

problem: if program crashes, data still in buffer never gets read  

**Note**

std::cerr is the error stream, it's NOT buffered, appear right away  

#### Format

**Syntax**

where i want to insert a var, put {}, a placeholder  
after the string: put a comma and the vras to replace in  
inside the palceholder: format specs  
(different from in python, where inside the palceholder put the var i want in it)  

placeholder syntax:    
{argument-index:format-specification}
if don't put argument-index, the vars go in the order i wrote them after the comma    
format specification means some letter (b = lowercase binary)  
if don't need to specify index not format, don't need the : either  

format is a function that returns a string  
then, i can pass it to other stuff, like to cout, or to a var

### Manipulators

**What**

command that change he way input/output is processed  

if put them on a stream (input or output), they persist that stream (don't also go over to the other stream) until i explicitly take them of  

exception: std:setw(16)   
it only applies to the next thing you print  

some can only be used as input manipulator, those only last in the line i wrote them in, don't persist  

**Types**

1- base manipulators  
need no params  
don't need libs  
cout << hex << 42;

2- param manipulators  
cout << setw (10) << " Ciao " ;  
need library `<iomanip>`    

**Persistance of manipulators**

1- sticky (persistente)  
must explicitly remove them  

2- non-sticky (temporaneo)  
only work for the output operation  

**Examples**

`std::left` align on the left (persists)  
`std::setw(16)` use padding so the next thing printed is long AT LEAST 16 spaces (don't persist, only apply to next thing printed)   
see an example in section 'print pretty'  

How they work  

how to evaluate <<: 
start from the left (put brackets to understand it)  
does what it must  
then leave the operand on its left  

example:  
( std::cout << std::setw(16) ) << "col1" << "col2";
( std::cout << "col1" ) << "col2";   \\ with padding mode activated  
step 1: activates padding mode, the returns std::cout, the left operand, to continue to evaluate the statement

### Files

**What to do**

there are 3 classes in fstream  
1- ofstream: create and write to file (output file stream)  
2- ifstream: read from file (input file stream)  
3- fstream: can do both  

to use a file, define the file as an obj with one of these 3 classes  
remember to close the file at the end, to clean up unnecessary memory space

**Append**

ofstream overwrites existing files  
to append to one, use the flag ios::app when opening the file  

ofstream MyFile ( "filename.txt" , ios::app ) ;  

**Reading**

**Reading safely**
