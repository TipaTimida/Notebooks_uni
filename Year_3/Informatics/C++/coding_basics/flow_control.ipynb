### If

**Tips condition**

if (x)  
x is var, means false if 0, true if any other number (evaluate int like a bool)  
so, this line means: if x is non-zero / non-empty

**Shortcut: skip if condition**

can be used to skip an if condition thath checks if that is true or not  

`bool zero { (x == 0) }; `  

`bool isEven(int x)`   
`{ return (x % 2) == 0; }`    
or, could write the func as: return !(x % 2);  

essentially, if need to return a true/false value, return the evaluation of the logic expression itself, since that will be a 0 or 1  

**If-else vs if-if**

if-else: the next checks are done only if the ones before were false  
if-if: always do all the checks, they're different if statement  

else is not obligatory  

**Semicolon**

do not put a semicolon after the 'if' condition  
acts like 'pass' in python  
so terminates my 'if', that thus doesn't work   

tip:  
if intentionally want to skip something, it's useful to define pass as a preprocessor directive transalted as null text  
`#define PASS`

`PASS;`  
the preprocessor directive translates the PASS to nothing, so only ; is left, which tells c++ to do nothing   
but having a big PASS makes it easier to read

### For

**Keywords**

break: get out of the loop, go to first statement after it     
continue: skip this step of the loop (apply update and go to next step)

**Tip writing condition**

if want something to happen from i=0 to i=9, how to write test condition:  
yes: i <= 10   
no: i != 10  
using != is risky, bc if for some reaason i=10 is skipped, the loop will never stop  

the loop runs for all times the test condition gives true:

runs from i=1 to i=5  **

**Omitted expression**

**Multiple counters**

### Switch

**When to use**

test EQUALITY (==, cannot do <=)  
of an expression, against a small set of values  

**Explained**

1- switch (x)   
x is the conditional expression that i want to evaluate  
must evaluate to an integral type or an enum  
cannot use expression whose result is a float, string...

2- case 1  
1 is the case label (the name of this case)  
must be a constant expression  
generally, this line is not indented  

if the conditional expression is equal to a case label:  
execution of the program continue from that point (so, means a made a jumo fom the line 'switch' to the line 'case')  
so the program continue to run normally form there onward   
which is why i dont' include the code after each case in block, but instead return at the end of it (to get out from the {} in which i deined the cases)

**Default case**

onlt execute if no other case matches  
usually put as the last case  

if no case matches, and there is no default case, simply don't execute anything  
exit the switch block without doing anything, continue running the program after it  

**Why the condition need integral values?**

because switch statements are designed to be highly optimized  
Historically, the most common way for compilers to implement switch statements is via Jump tables  
and jump tables only work with integral values

a jump table works much like an array  
an integral value is used as the array index to “jump” directly to a result

**Return vs break**

return: exits the current function  

break: exits the current (most inner) loop or switch    
so just exots the switch block, continue running the commands after it, in the same function  

fallthrough:  
if at the end of a case, i don't put either return nor break, the program will continue running the subsequent cases   
bc the switch is only a jump to a line, but after that point continues running normally  

if the fallthrough is intentional, at the end of the case, instead of return or break, point the attribute: `[[fallthrough]]` 

so:  
break is a jump to another line in this func  
return takes the ret stored in the stack, and uses it to exit this function  

continue:  
skip this iteration, but don't exit the loop  
it's a jump to the bottom of the loop  

**Attributes**

allows the programmer to provide the compiler with some additional data about the code    
maybe, to tell it that something was intentional, so don't output a warning about it   
is placed between double brackets  

other example: [[maybe_unused]], for vars

**Grouping cases**

Remember, execution begins at the first statement after a matching case label  
and continue until a break or turn  

Case labels aren’t statements (they’re labels), so they don’t count, nothing happens on them  

this is no considered fallthrough  

**Vars inside the switch**

declare or define (but not initialize) variables inside the switch, both before and after the case labels   

cases aren't urrounded by {}, they aren't each a different scope  
they're all part of a common scop,e the one of the switch (made by the {} enclosing all the cases)  
so a var defined in a case can be used in the successive cases (even if the line that declares is skipped it's ok, bc declarations are processed at compile.time, not runtime)

**Using continue: example**

to skip iterations where there is some condition  
so, instead of doing an if that executes if the conditionn is not met, do an if that skips if the condition is met  
so, reduce number of nested blocks  

bad version:  
#include <iostream>

int main()
{
    for (int count{ 0 }; count < 10; ++count)
    {
        // if the number is not divisible by 4...
        if ((count % 4) != 0) // nested block
        {
                // Print the number
                std::cout << count << '\n';
        }
    }

    return 0;
}  



good version:  
#include <iostream>

int main()
{
    for (int count{ 0 }; count < 10; ++count)
    {
        // if the number is divisible by 4, skip this iteration
        if ((count % 4) == 0)
            continue;

        // no nested block needed

        std::cout << count << '\n';
    }

    return 0;
}


usage:  
break: reduce number of vars  
continue: reduce nested loops  

some thing for early returns, bc careful using them  

#### Exit

**What happens when retunring from main?**

what the return statement do:  
1- destroy all vars  
2- call std::exit(), passing it the return value (status code)  
which performs cleanup  
3- return control to the OS, passing it the status code  

std::exit(0); // terminate and return status code 0 to operating system

std::exit() can also be called explicitly, to make it terminate early  

The std::exit() function does not clean up local variables in the current function or up the call stack  
so can be dangerous if your program relies on any local variables cleaning themselves up (second time i run, will work different???)  

std::atexit()   
allows you to specify a function (or more) that will automatically be called on program termination via std::exit()  
so, both exits and does clean up  
these funcs must take no args and have no return value  

so in a code, would write:   
void cleanup() {} 
std::atexit(cleanup);   \\ pass the func name, NOT a call to the func, don't put ()  
...  
std::exit(0);


**Ways to terminate**

1- normal termination    
std::exit()  
does not mean succesfull, the status coed could still indicate failure, but the termination is intentional   

2- abnormal termination  
std::abort()   
means the program had some kind of unusual runtime error and couldn’t continue to run  
eg. trying to divide by 0  
din't do any cleanup  

3- ???  
std::terminate()  
used together with 'exceptions' 

try not to use either of these  
they're dangerous, bc don't clean up local vars  
if there is an error, better to write code to handle exceptions than forcefully closing the program  

std::exit make us lose our call stack and any debugging information that might help us isolate the problem (bc we assume we have no problem)  
std::abort give us the option to start debugging at the point where the program aborted  

### Goto

**What it is**

unconditional jumo  
execution jumps to another spot in the code  

1- goto identifierName;  
goto statement  
where jump start  

2- identifierName:  
statement label  
where jumo ends  

note that the statement label is in scope of all the file, even before the point of declararion  
so can jumo to a line that i haven't reached yet    
this is said that 'statement labels' utilisze 'function scope'  

statement label must be followed by a statement  
if there's nothign to do after it, bc the function is over and just want to close the braces, put a ; in an empty line  
which means 'do nothing'  

**Limitations to jump**

1- can only jumo to a line in the same function i'm in now  
bc going to another one means that args and return valu and not put in the stack properly, will have trouble  

2- if jump forward, cannot skip over the initialization of a var that is in scope in the point i jumped to  

**When to use**

try to avoid it, makes the path of execution hard to understand  
use instead if and loops  

only case when it's ok:  
if have many nested blocks, and have to exit just a few of them  
can jumo directly to the right spot, instead of counting how many break to use
