### Public vs private

what | public  | private
---|---|---
access | ok even outside the class | only from inside the class <br> so using public funcs defined in the class
access a var | `myDog.name = "Max";` | using a public class function: <br> `myDog.getAge()` <br> `myDog.setAge(3);`
pros for vars| faster to write | 1- the code doesn't accidentally update them <br> 2- can make restrictions in what values i'm allowed to give (writing them in the func setValue)
access a func | just call it <br> `myDog.bark()` | must be called from a public func of the same class 
usage for funcs | interface (with the user)| internal helper

**Protected**

public - members are accessible from outside the class  
private - members cannot be accessed (or viewed) from outside the class  
protected - members cannot be accessed from outside the class, however, they can be accessed in inherited classes  

**Note**

if don't wrote an access specifier, by default it's private  

**Friend function**

#### Constructor

**Types of constructor**

1- parametric constructor  
can initialize params  
using an initialization string is faster than doing the initialization in the body   
if have nothing to initialize, can omit the constructor  

2- copy constructor  
to make an obj by copying an already existing object  
note: the other obj must be passed by ref (const, so i don't accidentally modify it), not value   
bc passing by value needs a copy, create a recursive problem  

**Member Initialization List**

it's faster than assigning in the body  
bc the obj is built directly with the final value  
while, with assignment in the bofy, first must build, than assign, are two steps  

### Destructor

**When to do**

when one of the params is a pointer, which in the constructor (or in another func) gets dinamically allocated  
so, in the destructor must deallocate it

### Inheritance

**Nomenclature**

1- derived class (child)   
the class that inherits from another class  

2- base class (parent)   
the class being inherited from  

**Code**

use the : symbol  
a class can be a child of more than one class  

**Polymorphism**

when we have many classes that are related to each other by inheritance  
we can perform a single action in different ways  
means a child class has a function that overrides one of the parent class (so, with same name)  
so that function has does different thing depending on the child class of the obj  

**Virtual**

give this keyword to the function in the parent class that's being overridden    
means: this function only exists if there isn't a better one (one belonging to the child, so more specific) 

1- without virtual  
C++ decides which function to call based on the pointer type, not the actual object type.  
what happens: even if call the func on a child class, runs that of the parent class -> wrong

2- with virtual  
it checks the actual object the pointer is pointing to.  
what happens: if call the func on a child class, runs that of the child class, like i want  

**Override**

does nothing, is optional  
just to remember than that function is overwriting another  
write it in the function of the child class  

**Types of polymorphism**

1- static polymorphism (early binding)  
what: templates or overloading   
the call resolution happens at compile time  

2- dynamic resolution (late binding)  
what: virtual and override funcs  
meaning: define funcs with same name and belonging to parent/child classes (using override and virtual)  
the call resolution happens at runtime (must see who called it)  


### Static

**Recap**

usually, the keyword static means that it just goes out of scope at the end of its function, doesn't get destroyed  

**Static property**

that property has the same value for all members of the class, doesn't change for each object  

consequences:  
must be defined and initialized OUTSIDE of the class (bc need to allocate outsde memory, it's not store inside the objects)  

**Static class function**

funcs that belong to the class, not a specific object  

consequences:  
1- to call it, don't need to use an object, just classname::func()   
2- cannot call it using this->  
3- can only use static properties (or they wouldn't know which object's value to use)
