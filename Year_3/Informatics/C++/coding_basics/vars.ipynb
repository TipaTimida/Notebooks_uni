#### Overflow

**Integer overflow**

what:    
if try to store a number outside of range, get overflow  

what i get:
the number gets wrapped aroung, using the modulo operation (imagine the range closed like a circle, like the clock)      
this is undefined behaviour  

**Explanation: unsigned**

if unsigned:  
overflowing means putting a bit in the nex byte  
which is not read, so that 256 looks like it was lost  

example:  
280 = 00000001 00011000  
but only read the second byte, which is 24  
as a matter of fact, 280 mod(256) = 24  

can also wrap in the other direction  
if i give x=-1, but range is [0, 255], then gives me 255   
like in array indexing, where -1 is the last element 


**Explanation: signed**

if go above the upper range, i'm modifying the first bit, which is used in used in the 2's complement for negative nums 
2's complement: swap all 0 and 1, then add 1  

example 
129 = 1000 0001  
127 = 0111 111  
if write 129 in a signed var, i see the first bit is 1, so must take the 2's complement, which is -127 (pu the - bc i said i'm doing the 2's complement)

note that: 129 - (-127)= 256  
129 = 127+2, but 126 is the max, so wrap around 2 spots: 127 > -128 > -127

### Type conversion

**Types conversions**

type conversion | implicit | explicit
---|---|---
who decides to do it | the compiler, without asking me | me, and i write it clearly in the code  
gives error? | yes, if narrowing | no, not even if narrowing, bc i told him to do it  
initialize | `int x = 5`<br> `y=(double) x` <br> gets converted to 5.0, a double value  | `static_cast<int>(5.5)` 

**Narrowing conversion**


a property of the conversion  
when i'm losing some data, the conversion is not perfect  

int -> double: can convert perfectly, don't lose data, is NOT narrowing  
double -> int: must round it, so lose data, it IS narrowing    
double -> float: float has less memory than double, so it IS narrowing  
remember than brace initialization forbids narrowing conversion  

**Types of conversion**

1- safe conversion (value-preserving conversion):    
every possible source value can be converted into an equal value of the destination type  
so it's alwats ok, cannot be narrowing  
eg. int -> float  

2- lossy conversion    
at least one value of the source type cannot be converted into an equal value of the destination type  
COULD BE narrowing (not neecssarily: 5.0->5 is not narrowing, but 5.1->5 is)  
eg. float-> int  

3- Reinterpretive conversions   
the converted value may be different than the source value, but no data is lost  
eg. signed -> unsigned  


**Explicit type conversion**

tell the compiler i do want that conversion, so it doesn't output an error even if it's narrowing  

`static_cast<new_type>(expression)` 
angled brackets are used either in preprocessor directives (like for libs) or to hold the name of a type (it's how to treat a type as a parameter)    
note that it don't change the expression i'm passing it, but creates a new object (temporary, unless i save it somehwere)  

usage:  

1- print the ascii value of a char:  
`static_cast<int>(charVar)`  

2- convert signed in unsigned and viceversa:    
`int s2 { 5 };` 
`unsigned int u2 { static_cast<unsigned int>(s2) };`  
if the value can't be represented in the destination type, the value gets modulo-wrapped  

**Static vs dynamic**


static_cast: “I know this conversion is valid.”  
dynamic_cast: “Please check whether this conversion is valid.”  

compile-time conversion between compatible types    
dynamic: runtime-checked conversion between polymorphic class types??? (NOT YET SEEN)  

**Why is type convertion important?**

bc the same bits are read differently, according to the type of var it must be  

eg. the same bits for 'int 3' can be read as 'float 4.2039e-45'  
bc for int, just translate from binary to decimal, while for float assume it's written with a mantissa and exponent, and read appropriately  

**How a conversion can fail**

1- there isn’t a standard conversion between those two types  
eg. cannot convert string to int  
int x { "14" };

2- that conversion is disallowed    
eg. brace initialization disallows narrowing conversion  
int x { 3.5 };
