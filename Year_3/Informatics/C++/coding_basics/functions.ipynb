### Nomenclature

**Main func**

1- there MUST a function called int main()  
or it will fail to link   

2- cannot be called explicitly inside the code  

3- must have return type int  

4- it's not always the first func that executes:   
global vars are initialized before calling main  
if their initializer invokes a function, this is called before main  

**How to call them**

normal funcs: func(var)   
member funcs (belong to a class): object.func()  

**If forget ()**

and since this pointer should never have address 0, that bool value will always be true  

this is a 'function pointer'

#### Forward declaration  

**Forward declaration**

tells the compiler about the existence of an identifier, before actually defining the identifier  
just need to tell what params wants and what it returns, so can check that the code that calls the func is ok    
for now, ignore what it does with such params, i'll define it later  

how to do it:  
the function header, terminated with a SEMICOLON (which is not there in the definition)  
only need to tell the param type, not their names (for clarity i should put their names too, it's best practice)  
`int add(int, int);`  
`int add(int x, int y);`   

**Why using forward declaration**

func is called before its definition  
or if it's defined in another code file  

### Scope

**Is it in scope?**

each function is a different scope  
so a var defined in a func is NOT in scope in another func  
not even if the func that created it is still ongoing, just paused bc called for another  

eg. i'm in main(), i call funcA()  
a var declared in main() is not in scope in funcA(), and one declared in funcA() is not in scope in main()  

**Passing vars between funcs**

1- passing it whe i'm calling a func: put it in the args  

2- passing it whe i'm closing a func: put it in the return value  
make sure the return type is correct

#### By value or by reference

**Recap: by ref**

to pass by reference, it's like passing by value, but put & after the param type  
to say it's an alias of the args, not a new var  

modifiche si ripercuotono sulla var originale  
non serve allocare nuova memoria, buono per var grandi  

if don't need to modify the obj passed, pass a const reference  

**Types**

1- by value  
the func creates a copy of the var i'm passing to it, in its local scope  
so, during the func i'm modifyig the copy in its local scope, but the original one  out of scope stays unchanged  

2- by reference    
the func defines an alternative name (a reference) for the var i'm passing it  
so now i can access that same memory cell, just calling it differently  

3- pass by pointer  
a special case of pass by value  
pass as value the address of the var i want to change  
and inside the func, dereference the pointer to change the original var, in the outside scope


**Which to use**

by ref and pointer can be used to modify the original var  
by ref is easier to write: simply put a reference as type in the header, then use the var name as usual  
always one by ref, unless the value nullptr is meaningful (reference cannot be null)

always use by refrence (so avoid creating a copy, which is costly)  
and i i don't want to accidentally change the value i'm passing it, put is 'const'  
why should i use by pointer??? 

**Examples**

a: original var, defined in main()  
want to call the func increase(), which modifies a  

**Const**

defining a param as const means that its value won't change during the function  

void funcA(const int num)

### Return

**What it does**

when evaluating a function call, i get the return object  
which is a 'temporary object'  
means something will null scope (its value exist just this once)  
so if i don'tsave it somewhere, it gets lost, it don't exist anymore  

std::cout << getValueFromUser() << '\n';   

can only return a single value  

**Mistake: return pointer to local var**  

what:  
return type: pointer  
return value: address to a local var  

it's a prob, bc when the func returns, that var gets destroyed, so i'm returning a dangling pointer  

**Possible solutions**

1- statica var  
make that var whose address i'm returing a static var    
static int ris ;  

2- dynamic allocation    
dynamically allcate that var, so don't get destroyed when function ends, but when i want  
but then, must remember to destroy it myself  
int* ris = new int ;  

3- instead of returning a pointer, returns the value of the local var itself  


#### Returning a reference

**Why**

1- creare an l-value (accesso in scrittura)  
the function is an access to a class member, that i want to modify from the outside  

2- to chain functions    

3- to prevent deep copy when reading

#### Types

**Value or ref**   

the return statement is the same: `return valueToReturn`     
but changse the return type  

1- pass by value  
`int getValue() {}`  

2- pass by ref  
`int& getValue() {}`  
return an address (value that can be assigned to a pointer)  
when to use: a class method returns one of its members  
do not return a reference to a local var, since it will be destroyed when the function returns 



**Types of copy**

when you copy an object:  

1- deep copy    
you also create a separate copy of the data it owns  

2- shallow copy    
making both objects point to the same memory  

pass by reference avoids doing a deep copy  

**Notes**

1- if the return type is void, don't need a return statement, can just close the braces  

2- i can call a function who happens to be returning a value and not store it anywhere, it's fine    

2- at the first return statement, the func returns  
if there is other stuff written after (even other return statements), they're ignored  
so can put many return statements, combined with flow control  

### Params

**Default parameter value**

next to the param, give it its default value   
so if call it without args, the param takes that value  

**Function overloading**

define functions with same name, but different type or number of params  
according to the args i call it with, the code chooses the correct one  

funcs are distinguished based on number of type of the PARAMS, not on the return type  

problem: can fail if don't know which one to pick (often bc of implicit conversion)  
eg. if have f(int) and f(double), calling f(3.5f), with a float, don't know what to do
