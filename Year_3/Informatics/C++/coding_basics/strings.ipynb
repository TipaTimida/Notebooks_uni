### C-style string literal

**What is it**

it's the type of the string literals in double quotes: "Hello world"  

CANNOT be the type of a var  
can assign it to a var pf type std::string or std::string_view var  

it's null terminated  
it's a fundamental type  

**Literals**  

to write a string literal of type std::String os std::string_view, use suffixes 's' or 'sv'  
only useful for auto type deduction  

using namespace std::string_literals; // easy access to the s suffix  
std::cout << "foo\n";   // no suffix is a C-style string literal  
std::cout << "goo\n"s;  // s suffix is a std::string literal  
std::cout << "goo\n"sv;  // sv suffix is a std::string_view literal

### String vs string_view

#### Properties explained

**String and string_view**

they're both class types  
so they can store more than one var (can have many properties to set)  

string: store many char  
string_view: store a pointer to the first char of the string it's viewing, and the length of such string  

**Constexpr**

string_view allows constexpr  
while string doesn't  

bc string_view is read-only, so essentially an implementation of: 'const std::string'  

#### Initializing

**Implicit conversions**

implicit conversion allowed:  
C-style -> string, string_view  
string -> string_view (the opposite is NOT allowed)  

accepted values:  
view_string type var accepts C-style and string too   
bc both do implicit conversion to string_view  
which is why, when funcs ack for a string_view parameter, can pass a string arg, and gets implicitly converted to the string_view type    

**What can initialize with**

can initialize with:  

1- a C-style string literal    
bc literals exist all the time  

2- an std::string var that OUTLIVES THE VIEW   
bc cannot use the view outside the scope of the var i'm viewing  


CANNOT initialize with a temporary obj, like:    
1- an std::string literal (the literal with 's' suffix)  
2- return value of a func   

**Passing between string and string_view**

to go between view_string and string:  
1- can initialize a string using a string_view, and viceversa    
2- can use static cast to turn srting_view into a string, and viceversa  
3- can do implicit conversion from string to string_view, but NOT the opposite  
why: to prevent mistakes, accidentally passing a std::string_view argument to a std::string parameter, and inadvertently making an expensive copy where such a copy may not be required

#### Using string_view

**The value it's viewing**

string_view is just a view of an existing string (the last one assigned to it)   
provides read-only access to an existing string, without making a copy  

so, for it to work, the existing string must be:  
1- ALIVE    
so can be out of scope, but cannot have been destroyed (or the view would be dangling, like a pointer)  
2- UNCHANGED   
changing the existing at causes undefined behaviour of string_view (from this point of view, not like a pointer)  

**What can i modify**

1- CANNOT change the existing string  

2- CAN change the string_view  
means make it point to something else   
assigning a new string to a string_view causes it to view the new string  
it does not modify the prior string being viewed in any way 

**Pros and cons of std::string**

pro: flexible

con: slow:  
long to copy, so bad to pass as an arg

### Owner vs viewer

**Meaning**

1- owner: owns an object  
it's safe, as long as the owner exists, the obj does too  
can modify the obj  
when destroy the owner, it destroys the obj  

2- viewer: views an obj  
it's not safe, it works only as long as the owner exists -> if owner is destroyed, the obj is too, while if viewer is destroyed, nothing happens to the obj  
cannot modify the obj  
don't destroy the obj when don't want it anymore, just stop watching it  

**When to use a viewer?**

1- when the original obj cannot be modified or destroyed until my job with the viewer is done  
2- when don't need to modify the obj  

example: a view is useful to pass a read-only arg to a func() (an arg i just need to read, not modify)    
bc the original obj is in the scope of main(), so it's out of scope while i'm in func(), so i'm sure it can't be modified  
and my viewer, which is in func(), gets destroyed at the end of func(), so as long as i use my viewer, the obj cannot be modified  
this way, don't need to pass the obj as arg, making an expensive copy  

**Why a viewer is better to pass to a func?**

owner: store the whole obj   
viewer: store a pointer to the first char of the obj, and the lenght of the obj  

so, the viewer has less stuff to copy  
and rem that a func makes a copy of the args passed to it, to a var in its scope  

**Why is modifying the obj while using the viewer a problem?**

if i lengthen the obj, it may be moved to an other address in memory, bc needs more space to be stored   
but view stores still the pointer to the original location, so can't find the obj anymore  
so, have a dangling pointer  
so, undefined behaviour  

or if i lengthen/shorten and it's not moved, the view don't know the length changed, keeps showing me the original length  
so cannot see th new chars, or see random chars in place of the ones that aren't there anymore  

same prob if obj is destroyed: it gets cancelled to free the memory cell, so now who knows what the viewer it's reading  

**Errors: initializing viewer with temporary obj**

the obj i'm initializing the viewer with is the object i want to view  
a viewer cannot be initialized with a temporary value (return value from a func, user-inputted text)  
bc these things are temporary obj, are not saved anywhere  
so the obj i'm trying to view don't exist  
these things must be initialized in an owner
