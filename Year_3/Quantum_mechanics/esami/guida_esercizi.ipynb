### Esercizi su stati

**Evol temporale di uno stato**

step | formula
---|---
cosa mi serve | 1- $\hat H$ del mio sistema <br> 2- $\|\psi_0 \rangle= \| \psi(t=0)\rangle $  
1- trova autostati di H (normalizzati) | $\hat H \|\phi_n \rangle = E_n \| \phi_n \rangle$  
2- scrivi stato iniziale su base di autostati di H <br> cioè, trova i $c_n$ | $\|\psi_0 \rangle = \sum c_n \|\phi_n \rangle$  
3- evolvi ogni autostato (mettendo la sua fase)| $\|\psi(t) \rangle = \sum c_n exp[-\frac i \hbar E_n t] \|\phi_n \rangle$  

**Valore di aspettazione**

mi serve:   
A (una matrice / un operatore in qualche rappr)  
$|\psi, t \rangle$: uno stato   
(può essere dipendente dal tempo o in un momento preciso, anche il valore di aspett sarà in funzione tempo o no)

$\langle A \rangle = \langle \phi, t | A | \phi, t \rangle $  

note:  
1- fai moltiplicazione tra matrici  
2- il bra prende l'aggiunto (cioè il coniugato, cambia segno di i) -> quindi la fase globale si cancella

**Tips elemento di matrice**

cosa | formula
---|---
base è (di solito) ORTONORMALE | $\langle n_1 \| n_2 \rangle = 0$  
quadrato di bnomio |  $(a-b)^2 = a^2-ab-ba+b^2$  
LINEARITÀ prod scalare | $\langle n \| a+b \|n  \rangle = \langle n \| a \|n  \rangle  + \langle n \| b \|n  \rangle $  
se sistema SEPARABILE <br>  separa gli stati e gli operatori che agiscono su oguno | $\langle n_1, n_2 \| X_1 X_2 \| 0,1 \rangle = \langle n_1 \| X_1\| 0 \rangle \langle n_2 \| X_2 \| 1 \rangle $ <br> eg. $\langle 1,0 \| x̂_1 p̂_2^2 \| 0,0 \rangle = \langle 1 \| x_1 \| 0 \rangle \langle 0 \| p_2^2 \| 0 \rangle$  
se OSC ARMONICO: <br>se ho valore atteso di x o p, scrivili in funzione di a, che so come agisce  | $\hat a \| 0 \rangle = 0$  <br> $\hat a^\dagger \| 0 \rangle = \|1 \rangle$ <br> $\hat q = \hat Q l_\omega = \frac {l_\omega} { \sqrt 2}(a+a^\dagger)$ <br> $\hat p =   \hat P  p_\omega= -i \frac {p_\omega}{ \sqrt 2}(a-a^\dagger)$

**Elemento di matrice: usa simmetria**

se bra e ket contengono lo stesso stato:  
il prodotto scalare è l'integrale della matriec per la densità dello stato:   
$\langle n | M | n \rangle = \int M \, |n|^2$   
quindi rispetta le regole di simmetria  

se lo stato è pari o dispari (quindi, il suo quadrato è pari)
e se M è dispari (ad esempio, X)  
allora il loro prodotto è 0  
cioè: $\langle 0 \hat X | 0 \rangle = 0$   
MA ATTENZIONE: $|\hat X | 0 \rangle \neq 0$   

altro modo di dirlo:   
posizione e impulso cambiano segno sotto parità $\hat \Pi \hat X\hat \Pi = -\hat X$    
se $\psi\rangle$ ha parità definita, il valore medio di ciascuno di questi operatori deve essere uguale al suo opposto; dunque è zero

**Trova stati (fondamentale, eccitati)**

mi serve: spettro $E_n$  

come fare:  
fondamentale: energia minima  
primo eccitato: energia penultima minima   

nota: se ho stato prodotto, potrei avere più livelli con la stessa energia  
quindi potrei avere più livelli primi eccitati  

**Trova stati: esempi**

$E_n = \hbar \omega (n_1+n_2+1)$  
stato fondamentale: $|0,0 \rangle$ (energia: $\hbar\omega$)   
primi stati eccitati: $|1,0 \rangle  \quad |0,1\rangle$ (energia: $2\hbar\omega$ )   

$E_n = \hbar \omega (n_1+2n_2+1)$  
stato fondamentale: $|0,0 \rangle$ (energia: $\hbar\omega$)   
primo stato eccitato: $|1,0 \rangle$  (energia: $2\hbar\omega$ )  
secondi stati eccitati: $|2,0 \rangle  \quad |0,1\rangle$  (energia: $3\hbar\omega$ )
