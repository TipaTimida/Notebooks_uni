### 2. Harmonic obscillator ???

$H = \frac{p_1^2}{2m}+\frac{p_2^2}{2m} +\frac m 2 \omega^2 (x_1^2+x_2^2)$   
una particella che si muove su un piano (quindi $x_2$ si intende l'asse y)  

**1. Prob di essere in un punto**

dove la voglio: $x_1^2+x_2^2 \le \frac{\hbar}{m \omega}$   
la particella è in ground state  

1- osservazione  
H è un sistema complesso: sono due oscillatori aromonici indipendenti, con stessa freq, uno lungo x e uno lungo y  
quindi, wavefunction del ground state: $\Phi_{00}(x_1, x_2)=\phi_0(x_1)\phi_0(x_2)$  
maiuscola = spazio totale, minuscola = sottospazio      

2- scrivo wavefunction  
l'osc armonico è un caso noto, so la sol del suo ground state (me la dà il testo)    
$\phi_{0}(x)= \frac 1 {\sqrt l \pi^{1/4}}exp[-\frac{x^2}{2l^2}]$  
$\Phi_{00}(x_1, x_2) = \frac{1}{\pi l} exp[-\frac{(x_1^2+x_2^2)}{2l^2}]$  

3- prob di essere in un punto  
$|\Phi_{00}|^2 = \frac 1 {\pi l^2} exp[-\frac{(x_1^2+x_2^2)}{l^2}]$  

4- cosa voglio:  
prob che sia un punto che rispetta:  $x_1^2+x_2^2 \le \frac{\hbar}{m \omega} = l$    
cioè, dentro un disco di raggio l  
quindi, coord polari:  
$r^2=x_1^2+x_2^2$  
$dx_1 \, dx_2 = dA =  r \, dr \, d\phi$  

5- prob di essere nel disco:  
integro la prob di essere in un punto in tutti i punti che mi vanno bene (disco di raggio l)  
$P = \int_0^{2 \pi} \;\int_0^l |\Phi_{00}|^2 \, r \, dr \,  d \phi  $  
$=  \int_0^{2 \pi} \, d\phi \; \int_0^l \frac 1 {\pi l^2} exp[-\frac{(r^2)}{l^2}] \,r \, dr$    
$= \frac 2{l^2} \int_0^l re^{-r^2/l^2} \,r \, dr$   
$=\int_0^1 e^{-u} du = 1-e^{-1} \sim 0.63$     
var change: $u=r^2/l^2$   
quindi: prob è circa $63\%$

**Elementi di matrice oscillatore**

per un oscillatore in 1D, i valori sono noti:  
ricorda che:   
$\hat a | 0 \rangle = 0$  
$\hat a^\dagger | 0 \rangle = |1 \rangle$    
$x = \frac l {\sqrt 2} (\hat a + \hat a^ \dagger)$  

quindi:  
$\langle 0 | \hat x | 0 \rangle = \frac  {l}{\sqrt 2} \langle 0|a| 0 \rangle +  \frac  {l}{\sqrt 2} \langle 0|a^\dagger| 0 \rangle$   
$ 0 + \langle 1| 0 \rangle  = 0$
