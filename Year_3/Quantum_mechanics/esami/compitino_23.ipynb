### Osc armonico

$\hat H = \frac {\hat p^2}{2m} + \frac m2 (\omega^2 \hat x_1^2 + 4 \omega^2 \hat x_2^2)$  
$\hat p^2 = \hat p_1^2 + \hat p_2^2$   

cioè particella su piano 2D  
c'è oscillatore armonico su entrambe le direzione, ma con freq diversa  

**1. Spettro**

1- riscrivi con Ladder operators  
$\hat H = \hbar \omega (\hat a_1^\dagger \hat a_1 + \frac 12 ) + 2\hbar \omega (\hat a_2^\dagger \hat a_2 + \frac 12 )$  
vedi dalla H che, lungo asse x, la frequenza è doppia   

2- spettro di uno stato $|n_1, n_2 \rangle$:  
$E_{n_1, n_2} = \hbar \omega (n_1 + 2n_2+\frac 32)$    
perchè ricorda che $a^\dagger a = N$ ha come spettro solo numeri interi  

3- degenerazione:  
$|0, 1 \rangle $ e $|2, 0 \rangle$ hanno stessa energia   

**Theory: Ladder operators**

per ogni direzione j=1,2 sono diversi    
$\hat a_j  = \frac 1 {\sqrt 2}(\hat x_j + i \hat p_j)$   
dove x_j, p_j sono OPERATORI RISCALATI  

**Domanda**

elemento di matrice di $x̂^1 p̂_2^2$ tra lo stato fondamentale e il primo stato eccitato  
cioè, $M= x̂^1 p̂_2^2$ è la matrice

primo stato eccitato: $|1,0 \rangle$ (è il primo perchè si mettono in ordine di energia)   

voglio l'elemento: $\langle 1,0 | x̂^1 p̂_2^2 | 0,0 \rangle$    
possono anche invertire gli stati, tanto la matrice è hermitiana
