### 1. Oscillatore armonico

$\hat H = \frac{\hat p^2}{2m} + \frac 12 m \omega?2 \hat x^2$  

**1. Scrivere con ladder operators**

$H = \hbar \omega (a^\dagger a + \frac 12)$  
$a, a^\dagger$ definiti nel solito modo  

autovalori:  
$E_N = \hbar \omega (n+\frac 12)$  
$n = 0,1,2...$  

**2. Elementi di matrice**
 
matrice: $\hat x$ \quad \hat x^2$  
stati: $|0 \rangle$ e $|1 \rangle$   

1- scrivi operatore in funzione di ladder operator:  
$\hat x = \frac{l_\omega}{\sqrt 2}(a+a^\dagger)$  
$\hat x^2 = \frac{l_\omega^2}{2}(a^2 + aa^\dagger + a^\dagger a +(a^\dagger)^2)$   

2- calcola, sapendo cosa fanno i ladder operators  
$\langle 1 | \hat x |0 \rangle =\frac{l_\omega}{\sqrt 2}  \langle 1 |a+a^\dagger|0 \rangle = \frac{l_\omega}{\sqrt 2}$    
fallo per tutti gli elementi della matrice:  
$\hat x = \begin{cases} 0 & l/\sqrt 2 \\ l/\sqrt 2 & 0 \end{cases}$  
$\hat x^2 = \begin{cases} l^2/2 & 0\\ 0 & 3l^2/2 \end{cases}$

**3. Deviazione nello stato fondamentale**

def: $\Delta x= \sqrt {\langle \psi | \hat x^2 | \psi \rangle}$   
dove $\psi = | 0\rangle$ (lo chiede il testo)  

1- deviazione posizione  
$\Delta x = \sqrt {\langle 0 | \hat x^2 | 0}$   
$= \sqrt{l^2/2} = \frac l{\sqrt 2}$  
$x^2$ calcolato al punto prima  

2- deviazione momento:  
$\Delta p = {\langle 0 | \hat p^2 | 0}$   
$= \frac {\hbar} {\sqrt l}$    

3- principio di indeterminazione:  
$\Delta x  \Delta p = \frac \hbar 2$  
lo satura (infatti lo stato fondamentale è una gaussiana)
