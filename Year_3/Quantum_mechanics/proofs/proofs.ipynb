**Conseguenze (Schrodinger)**

stato misto:  
$\rho(t) = |\psi(t) \rangle \langle \psi(t) $  
$= U^* \langle \psi(t) | \;  U| \psi(t) \rangle $  
$U^* \rho_0 U$  
cioè si trasforma come un operatore  
e quindi rispetta l'eq dell'evoluzione temporale degli operatori  

**Problema**

ma so scrivera solo in teoria, non in pratica, perchè appare $e^H$ che non so come calcoalre  

**Risolvere**

assumo:  
$t_0=0$  
H non dipenden dal tempo 
H ha spettro discreto  

1- trovo autostati di H  
$\hat H |\phi_n \rangle = E_n |\phi_n \rangle$

2- scrivo stato iniziale e finale nella base di autostati (ortonormale)    
$|\psi(t_0) \rangle = \sum_n c_n | \phi_n \rangle  \quad c_n = \langle  \phi_n |  \psi(t_0)  \rangle $  
$|\psi(t) \rangle = U \, |\psi(t_0) \rangle = \sum c_n \; exp[-\frac i \hbar E_nt] |  \, \psi_n \rangle$    
cioè: se applico ad un autostato, $H \rightarrow E_n$   
(anche se H è all'esponenziale, tanto lo porto giù scrivendo l'esponenziale come serie di Taylor, def di esponenziale di un operatore)     

3- posso misurare un osservabile:  
$|\psi \rangle = \sum_n c_n \;  exp[-\frac i \hbar E_nt] \; |\phi_n \rangle$   
$\langle \psi | = \sum_m c_m^* \; exp[+\frac i \hbar E_nt] \;  \langle \phi_m|$  
$\langle \hat O \rangle = \langle \psi(t) | \hat O | \psi(t) \rangle = \sum_{n,m} c_n^* c_m \, exp[\frac i \hbar (E_n-E_m)t ] \;  \langle \phi_n |\hat O | \phi_m \rangle$  

4- se lo stato è un autostato di H, cioè $\psi(0)\rangle = |\phi_k \rangle$  
solo $c_k \neq 0$  
allora l'osservabileO  è costante (ma non per forza $E_k$, perchè è un autostato di H, non di O)    
$\langle \hat O \rangle = \phi_k | \hat O | \phi_k \rightarrow$   

**Osservazione**

su H.. 
$\langle \phi_n | \hat H| \phi_m \rangle = E_m | \langle \phi_n| \phi_m \rangle  = E_m \delta_{nm}$   
quindi:  
$\langle \hat H \rangle = \sum_{n,m} c_n^* c_m \, exp[\frac i \hbar (E_n-E_m)t ] \;  \langle \phi_n |\hat H | \phi_m \rangle$    
$= \sum_n c^*_nc_n \, exp[0] \, \langle \phi_n | \hat H | \phi_n \rangle$   
$= \sum_n |c_n|^2 E_n \langle \phi_n | \phi_n \rangle$  
$= \sum_n |c_n|^" E_n$  


**Moto confinato**

quando: se H ha spettro discreto  

cosa: non si espande troppo nello spazio delle fasi  
$P_\Omega(t_0 >0) \quad \rightarrow \quad \lim_{t \rightarrow \infty} \frac 1T P_\Omega (t) \, dt > 0$  
in una regione $\Omega$, dove all'inizio c'era prob non nulla di trovarsi, anche dopo un tempo infinito c'è prob non nulla di trovarsi  
quindi la waveform di probabilità non può spari dalla zona da dove è partita, solo espandersi un po'  

$P_\Omega(t) = \int_\Omega dx\; |\psi(x,t)|^2$   
prob di trovarsi nella regione $\Omega$ al tempo t  

**Trovare U (evoluzione temporale)**

cosa voglio:  
voglio la U con cui scrivere l'evoluzione temporale  
so che uno stato soggetto a evol. temporale deve rispettare l'eq di Schrodinger   
mi serve un caso noto (per risolvere problema di Cauchy)  

$\begin{cases}
|\psi(t) \rangle = \hat U(t,t_0) | \psi(t_0) \rangle & evol.\, temporale \\
i \hbar \frac{d}{dt} |\psi(t) \rangle= \hat H |\psi(t) \rangle & Scrodinger \\
\hat U(t_0, t_0)= \hat I
\end{cases}$  

cosa faccio:  
plug the first in the second  (must be true for every starting state, so delete $|\psi(t_0) \rangle$)    
$\frac{\partial \hat U}{\partial t} =-\frac i \hbar \hat H \hat U$  
$\hat U(t,t_0)= exp[-\frac i \hbar \hat H(t-t_0)]$  
 
nota: spesso uso $t_0=0$    

nota: per $t \sim t_0$, un esponenziale si definisce come:  
$\hat U(t,t_0) = \hat I -\frac i \hbar \hat H(t-t_0) + O((t-t_0)^2)$  
cioè applico H o sue potenze  
quindi, se applico ad un autospazio DI H, ho:  
$\hat U(t,t_0)= exp[-\frac i \hbar \hat E_n(t-t_0)]$    
dove: $H |\phi_n \rangle = E_n |\psi_n \rangle $  


### Immetria: generatori

**Generatori**

1- trasformazione infinitesima, cioè $s \sim 0$  
$\hat U(s) = I + s \frac{d}{ds}\hat U(s) + O(s^2)$  

2- so che $U^\dagger U = I$  
$I + s [\frac{dU}{ds}+\frac{dU^\dagger}{ds}] + O(s^2)=I$  
$\frac{dU}{ds}=-\frac{dU^\dagger}{ds}=0$  

3- soluzione  
$\frac{dU}{ds}=i \hat K$  
con K autoaggiunto 

**Th di Stone: proof**

risolvi il problema di Cauchy:  
$\begin{cases}
\frac{dU}{ds}=i \hat K \\
\hat U(0)=I
\end{cases}$

ottengo: $\hat U(s) = exp[i \hat K s]$  

th: se $\hat U$ è associata ad una simmetria -> K commmuta con H   
cioè: posso fare prima la trasformazione o l'evoluzione temporale, il risultato non cambia  
cioè K è una costante del moto, il suo valore d'aspettazione non cambia col tempo    
variante in MQ del th di Noether  


**Th di Wigner**

una trasformazione che rispetta le simmetrie deve essere rappresentata da un operatore unitario (o antiunitario)  

nota:  
osservabile O: autoaggiunto $O^\dagger = O$  
trasformazione A: unitario: $A^\dagger A = A A^\dagger = I$  

**Operatore impulso**

è il generatore delle traslazioni:  
$T(a) = exp [-\frac i \hbar a \hat P ]$  

perchè:  
l'impulso (= la velocità) genera uno spostamento (cioè una traslazione)
