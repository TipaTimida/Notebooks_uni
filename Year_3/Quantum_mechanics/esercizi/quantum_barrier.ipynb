##### Simmetrie

quindi $[\hat H, \hat \Pi]=0$, c'è la simmetria di parità  

voglio una base di autovetori in comune tra H e $\Pi$

**Base comune H e $\Pi$**

essendo H con spettro non degenere, una base qualunque va bene  
quindi, nella base trovata, cerco la parità di ogni autovettore  

parità | pari (+1) | dispari (-1)  
---|---|---
left | $\phi_L(x) = A_+ e^{\lambda x}$ |  $\phi_L(x) = A_- e^{\lambda x}$
right | $\phi_R(x) = A_+ e^{-\lambda x}$ |  $\phi_R(x) = A_- e^{-\lambda x}$
inside | $\phi_I(x) =C_+\cos(kx)$ |  $\phi_I(x) =C_- \sin(kx)$

come l'ho trovata: alla base generale (bound states), applico la condizione:  
1- $\pi=1$ -> $\phi_R(x) = \phi_L(-x) \quad \phi_I(x)=\phi(-x)$    
2- $\pi=1$ -> $\phi_R(x) = -\phi_L(-x) \quad \phi_I(x)=-\phi(-x)$    
da cui scrivo i coeff in funzionedegli altri o cancello adendi

### Quantum barrier

**1. Potenziale**

box potential / quantum well:  
$V(x) = \begin{cases}
0 & else \\
V_0 & [-\frac a2, \frac a2] 
\end{cases}$  

3 regioni: left, inside, right

**2. Sol eq autovalori**

caso | $E<V_0$ | $E>V_0$
---|---|---
inside | $E>0$: decade | $E>0$: oscilla  
outside | $E<V_0$: oscilla | $E>V_0$: oscilla  
tipo | SCATTERED <br> perchè decade a $\pm \infty$, se scelgo bene i coeff | SCATTERED <br> perchè non decade a $\pm \infty$, non normalizzabile  

cioè ci sono solo stati scattered  

**Stati scattered**

caso: $E< V_0$, dove dentro decade    
è stato scattered, studio lo scattering  

inside: $\phi_I(x) = \alpha_i e^{-k'x} + \beta e^{k'x}$    
outside: $\phi_{L/R}(x) = \alpha_{L/R}e^{ik x}+\beta_{L/R}e^{-ik x}$  

nota: $\beta_R=0$  
perchè l'onda arriva da sinistra e sbatte sulla barriera, o rimabalza indietro o prosegue a destra  
ma non può muoversi verso sinistra nella regione di destra  

**3. Condizioni al contorno**

così esprimo i coeff tutti in funzione di uno:    
$\alpha_L = [\cos(\lambda a)-i \frac{\lambda^2+k^2}{2 \lambda k} \sin(\lambda a)]e^{ika}\alpha_R$  
$\beta_L =  i \frac{\lambda^2+k^2}{2 \lambda k} \sin(\lambda a) \alpha_R$  

quindi:  
$T = \frac{4E(E-V_0)}{4E(E-V_0)+V_0^2 \sin[\frac 1 \hbar \sqrt{2m(E-V_0)}a]}$  
T oscilla tra $T_{max}=1$ (quando sin=0 ???) e $T_{min}$ (quando sin=1 ???)  
max T: interferenza costruttiva  
min T: distruttiva  

**Significato coeff**

$\alpha$: onda va verso destra  ->
$\beta$: onda fa verso sinistra <-  
L/R/I: in che regione si trova l'onda  

assumo che l'onda iniziale sia in L, viaggia verso destra: ->  
$r = \frac{\beta_L}{\alpha_L}: reflecion coeff  
a sinistra, quanta torna indietro, di quella incidente sulla barriera?  
$R = |r|^2$: reflection amplitude    
$t = \frac{\alpha_R}{\alpha_L}$ transmission coeff  
quanta onda è passata a destra e continua a viaggiare a destra, di quella incidente sulla barrier?  
$T = |t|^2$: transmission amplitude
