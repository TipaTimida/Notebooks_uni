**2. Scattering eigenstate**

caso: $E> V_0$, dove oscilla sia dentro che fuori (ma con k diverso, perchè cambia la V)  
è stato scattered, studio lo scattering  

$\phi(x) = \begin{cases}
\alpha_L e^{ikx} +\beta_L e^{-ikx} & left \\
\alpha_I e^{i\lambda x} +\beta_I e^{-i \lambda x} & inside \\
\alpha_R e^{ikx} & right \\
\end{cases}$

nota: $\beta_R=0$    
perchè assumo che l'onda incidente iniziale sia in L (left), viaggia verso destra: ->    
quindi sbatte sulla barriera, o rimabalza indietro o prosegue a destra  
ma non può muoversi verso sinistra nella regione di destra  

**Significato coeff**

$\alpha$: onda va verso destra  ->  
$\beta$: onda fa verso sinistra <-    
L/R/I: in che regione si trova l'onda  


**Coeff per onde**


$r = \frac{\beta_L}{\alpha_L}$: reflection coeff  
a sinistra, quanta torna indietro, di quella incidente sulla barriera?  
$R = |r|^2$: reflection amplitude 
   
$t = \frac{\alpha_R}{\alpha_L}$ transmission coeff  
quanta onda è passata a destra e continua a viaggiare a destra, di quella incidente sulla barrier?  
$T = |t|^2$: transmission amplitude    
intuitively, it's the tunneling throught the barrier (bc i could have picked the case where $\phi$ decades inside the barrier)  


**3. Condizioni al contorno**

continuità di $\psi$ e $\psi'$ dove si toccano le regioni  
così esprimo i coeff tutti in funzione di uno:    
$\alpha_L = [\cos(\lambda a)-i \frac{\lambda^2+k^2}{2 \lambda k} \sin(\lambda a)]e^{ika}\alpha_R$  
$\beta_L =  i \frac{\lambda^2+k^2}{2 \lambda k} \sin^2(\lambda a) \alpha_R$  

quindi:  
$T = \frac{4E(E-V_0)}{4E(E-V_0)+V_0^2 \sin[\frac 1 \hbar \sqrt{2m(E-V_0)}a]}$  


**Osservazioni su T**

T oscilla tra:  
1- $T_{max}=1$

max T: interferenza costruttiva

min T: distruttiva
