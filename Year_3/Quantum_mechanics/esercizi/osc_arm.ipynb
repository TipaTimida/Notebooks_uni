**Scrivere H dell'oscillatore**

step |cosa | H
---|---|---
1|normale, plugga potenziale armonico | $\hat H = \frac{\hat P^2}{2m} + \frac 12 m \omega^2 \hat X^2$    
2|definisci operatori RISCALATI: <br> scale: $E_\omega = \hbar \omega \quad P_\omega = \sqrt {\hbar \omega m} \quad l_\omega = \sqrt{\frac \hbar {m \omega}}$ <br> operatori: $\hat q = \frac {\hat Q}{l_\omega} \quad\hat p = \frac{\hat P}{p_\omega}$ | $\hat H = \hbar \frac \omega 2(\hat p^2+\hat q^2)$
3|definisco operatore di innalzamento: <br> $\hat a = \frac 1 {\sqrt 2}(\hat q + i \hat p)$ | $\hat H = \hbar \frac \omega 2(1+2a^\dagger a)$  
4|definisco nuovo operatore: <br> $\hat N = a^\dagger a $ | $\hat H = \hbar \frac \omega 2(1+\hat N)$

**Base di autovettori**

1- so che: $[\hat \Pi, \hat X^2]=0$  
quindi il sistena è parity-invariant  

2- spettro di H è discreto  
quindi: no degenerazioni, ogni autovettore di $\hat H$ lo è anche di $\Pi$  

3- cerco gli autovalori per $\Pi$ della base di autovettori $\hat H$    
pari (autov = +1): $|0 \rangle, |2 \rangle...$   
dispari (autov = -1): $|1 \rangle, |3 \rangle...$  

ha senso, $|0 \rangle$ è quello a energia minore, quindi ha 0 poli  
allora deve essere pari  

**Normalizzo la base**

$|n \rangle = \frac 1{\sqrt n!} (\hat a^\dagger)^n |0 \rangle$  
$a^\dagger$, l'operatore d'innalzamento, è per passare agli altri autovettori  
$ \frac 1{\sqrt n!}$ è il coefficiente di normalizzazione, per cancellare l'autovalore che ottengo quando applico $\hat a^\dagger$  

### Spettro di H

**Come trovarlo**

$\hat H = f(\hat N)$  
quindi se trovo lo spettro di N, so anche quello di H  

**Osservazioni**

1- discreto  
2- equispaziato   
3- il ground state, cioè $| 0 \rangle$, ha $E_0 = \frac 12 \hbar \omega \neq 0$ 

**Nomi**

$| 0 \rangle$: ground state -> quello con energia minore  
altri: stati eccitati -> con energia maggiore della minima    

**Principio di indeterminazione**

è saturato solo dal ground state  
quindi il ground state è una gaussiana, gli stati eccitati no  

intuivamente: se aumento l'energia, aumento l'incertezza $\Delta_Q \Delta_P$ delle particelle

no: $\hat a$

**Formule**

---|---|---

#### Parità

definizione: $\Pi | x \rangle = |-x \rangle$  

**Proprietà**

1- autoaggiunto -> autovalori reali $\pm 1$
2- unitario -> può essere una simmetria (parity-invariance)  
3- $\Pi^2=I$   

4- $\Pi \hat X \Pi^{-1} = - \hat X$   
$\Pi \hat P \Pi^{-1} = - \hat P$  
$\Pi \hat P^2 \Pi^{-1} =  \hat P^2 \quad \rightarrow \quad [\hat \Pi, \hat P^2]=0$      
$\Pi \hat X^2 \Pi^{-1} =  \hat X^2 \quad \rightarrow \quad [\hat \Pi, \hat X^2]=0$      
quindi: $\Pi$ commuta sempre con la parte cinetica di $\hat H$  
per commutare con $\hat H$, devo solo controllare che commuti con $V(\hat X)$, cioè che V(X) rispetti la simmetria  

**Come usarlo**

se $[\hat H, \hat \Pi]=0$  
allora posso trovare una base di $\hat H$ di autovettori, in cui ognuno è anche autovettore di $\hat \Pi$ (cioè ha una parità)  

attenzione:   
se $\hat H$ ha pettro degenere: esiste una base di autovettori comune a $\hat H$ e $\Pi$, ma non tutti gli autovettori di H lo sono anche di $\Pi$ -> quindi scegli bene la base
se H ha spettro NON degenere: ogni autovettore di H lo è anche di $\Pi$, e viceversa -> quindi scegli una base qualunque  
quindi devo sceglierla bene  

ho base che;  
$\hat H |\psi \rangle = E_n | \psi\rangle $   
$\hat \Pi |\psi \rangle = \pm | \psi\rangle $ 

**Proprietà**

condizioni:  
1- $\hat H$ ha spettro discreto  
2- sono in 1D   
3- $[\hat H, \hat \Pi]=$  

allora: no degenerazioni nello spetro di $\hat H$  
quindi ogni autostato di $\hat H$ è anche autostato di $\Pi$ (non devo scegliere bene la base)
