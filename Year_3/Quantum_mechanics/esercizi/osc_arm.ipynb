#### Parità

$\Pi \hat P \Pi^{-1} = - \hat P$

#### Parità

definizione: $\Pi | x \rangle = |-x \rangle$  

**Proprietà**

1- autoaggiunto -> autovalori reali $\pm 1$
2- unitario -> può essere una simmetria (parity-invariance)  
3- $\Pi^2=I$   

4- $\Pi \hat X \Pi^{-1} = - \hat X$   
$\Pi \hat P \Pi^{-1} = - \hat P$  
$\Pi \hat P^2 \Pi^{-1} =  \hat P^2 \quad \rightarrow \quad [\hat \Pi, \hat P^2]=0$      
$\Pi \hat X^2 \Pi^{-1} =  \hat X^2 \quad \rightarrow \quad [\hat \Pi, \hat X^2]=0$      
quindi: $\Pi$ commuta sempre con la parte cinetica di $\hat H$  
per commutare con $\hat H$, devo solo controllare che commuti con $V(\hat X)$, cioè che V(X) rispetti la simmetria  

**Come usarlo**

se $[\hat H, \hat \Pi]=0$  
allora posso trovare una base di $\hat H$ di autovettori, in cui ognuno è anche autovettore di $\hat \Pi$ (cioè ha una parità)  

attenzione:   
se $\hat H$ ha pettro degenere: esiste una base di autovettori comune a $\hat H$ e $\Pi$, ma non tutti gli autovettori di H lo sono anche di $\Pi$ -> quindi scegli bene la base
se H ha spettro NON degenere: ogni autovettore di H lo è anche di $\Pi$, e viceversa -> quindi scegli una base qualunque  
quindi devo sceglierla bene  

ho base che;  
$\hat H |\psi \rangle = E_n | \psi\rangle $   
$\hat \Pi |\psi \rangle = \pm | \psi\rangle $ 

**Proprietà**

condizioni:  
1- $\hat H$ ha spettro discreto  
2- sono in 1D   
3- $[\hat H, \hat \Pi]=$  

allora: no degenerazioni nello spetro di $\hat H$  
quindi ogni autostato di $\hat H$ è anche autostato di $\Pi$ (non devo scegliere bene la base)   

### Ex. osc arm spostato

**Setup**

t=0:   
V(x) centrato in $x=x_0$    
sono nello stato fondamentale  

t>0:   
V(x) centrato in $x=0$    
stato: non è più autostato di H, come evolve?  
cioè, cambio il potenziale (usando un campo elettrico)  

**Setup**

1- t=0: V centrato in $x_0$  
$H_i = \frac{\hat p^2}{2m} + \frac 12 m \omega^2 (\hat x-x_0)^2$  
$\hat a_i = \frac{1}{\sqrt 2} [\frac{l_0}\hbar\hat p - \frac i {l_0} (\hat x-x_0)]$    

2- t>0: V centrato in $0$  
$H = \frac{\hat p^2}{2m} + \frac 12 m \omega^2 (\hat x)^2 = \hbar \omega (aa^\dagger + \frac 12)$    
$\hat a= \frac{1}{\sqrt 2} [\frac{l_0}\hbar\hat p - \frac i {l_0} \hat x]$    

nota che: $\hat a = \hat a_i -\frac i {\sqrt 2 l_0} x_0$  

**Risolvo: scomponi in base di H**

1- trovo $|\psi_0\rangle$  
lo stato fondamentale è quello che: $\hat a_i |\psi_0 \rangle = 0$  

2- lo scrivo nella base di autostati di H: $|\psi_0 \rangle = \sum c_n|  \phi_n \rangle$     
l'eq diventa:  
$\hat a_i \sum c_n |\phi_n \rangle = 0$    
$\sum c_n (\hat a+ \frac i {\sqrt 2 l_0}x_0)|\phi_n \rangle =0$  
$\sum c_n (\hat a- \alpha) |\phi_n \rangle =0$  
$\sum c_n \sqrt n |\phi_{n+1} \rangle = \sum c_n \alpha |\phi_n \rangle =0$   (proprietà del ladder operator su un autostato)  
$\sum c_{n+1} \sqrt {n+1} |\phi_{n} \rangle = \sum c_n \alpha |\phi_n \rangle =0$

3- ricorrenza:  
$c_{n+1} = c_n \frac{\alpha}{\sqrt{n+1}}$  
sol: $c_n = \frac{\alpha^n}{\sqrt{n!}} c_0$   
$c_0$ lo trovo con la condizione di normalizzazione: $|c_0|^2 = exp[-|\alpha|^2]$  

4- sol:  
$|\psi_0 \rangle = \sum \frac{\alpha^n}{\sqrt{n!}} c_0 |\phi_n \rangle$   

nota:  
quindi $|\phi_0\rangle$ è autovalore di entrambi i ladder operator:  
$\hat a_i | \phi_0 \rangle = 0$  
$\hat a | \phi_0 \rangle = -\frac i {\sqrt 2 l_0}x_0  |\phi_0 \rangle = \alpha| \phi_0 \rangle$ (definisco $\alpha$)  

**Risolvo: evol temporale**

gli autostati di un osc armonico hanno energia $E_n = \hbar \omega (n + \frac 12 )$     
$|\psi_0 \rangle = \sum \frac{\alpha^n}{\sqrt{n!}} c_0 |\phi_n \rangle$     
$|\psi(t) \rangle = \sum \; exp[-\frac i \hbar E_n t]  \; \frac{\alpha^n}{\sqrt{n!}} c_0 |\phi_n \rangle$   
$ = \sum \; exp[- i\omega (n+ \frac 12) t]  \; \frac{\alpha^n}{\sqrt{n!}} c_0 |\phi_n \rangle$   
$=exp[-i \omega t \frac 12 ] \,  exp[-|\alpha|^2 \frac 12] \; \sum \frac{\alpha(t)} {\sqrt n!} | \phi_n \rangle$  
tiro fuori un fattoredi fase globale  

ho definito:  
$\alpha(t) = \alpha e^{-i \omega t}$  
nel modulo quadro la fase non importa, prendo la norma $\alpha$  

**Osservazioni: forma**

$|\phi_0 \rangle $ è una gaussiana, perchè è stato fondamentale di un osc armonico (centrato in $x_0$)  

vedo che:   
$\hat |\psi(t) \rangle = \alpha(t) |\psi(t)$    
la uso per scrivere $|\psi(t) \rangle$ in rappr delle coords (come ho usato $\hat a |0 \rangle = |0 \rangle$ per la precedente)  
come: scrivo $\hat a $ in funzione di $\hat x $ e $\hat p$, in rappr delle coords, ottengo eq diff  
ottengo: $\psi(x,t) = (\frac{1}{\sqrt \pi l})^{1/4} exp[-\frac{x-\langle x \rangle (t)}{2l^2}- i \frac{\langle p \rangle (t)}\hbar x]$  
che è una gaussiana  

è un pacchetto gaussiana che fa un MOTO RIGIDO:    
cambia $x_0$, la coord del centro  
ma la larghezza, $l$, resta uguale (non si squaglia)
