### Teoria

**2-levels system**

cosa è: sistema con solo due livelli nello spettro, $E_0$ e $E_1$    
come si crea: se il gap tra $E_1$ e $E_2$ è molto grande, posso assumere che $E_2$ non esista, le particelle non riusciranno mai a saltare lassù, anche se ricevono molta energia  

un sistema a fue livelli è un q-bit (bit quantistico)  

### Setup

#### Stati

$|0 \rangle  \quad |1 \rangle$  
sono normalizzati e ortogonali  

stato generico:  
$|\phi \rangle = \alpha |0 \rangle + \beta | 0 \rangle$  
$\alpha^2 + \beta^2 = 1$  

#### Operatori

**Operatori**

$|0 \rangle \langle 0 | $: proiezione su $|0 \rangle$  
$|1 \rangle \langle 1 | $: proiezione su $|1 \rangle$  
$|0 \rangle \langle 1 | $: ladder operator, passa da 1 a 0  
$|1 \rangle \langle 0 | $: ladder operator, passa da 0 a 1  

**Matrici di Pauli**

$\{I, \hat \sigma^x, \hat \sigma^y, \hat \sigma^z \}$

ottenuti combinando gli operatori sopra:  
...  

sono una base completa  
il loro osservabile è il momento angoalre  

**Autostati**

ogni matrice di Pauli ha autostati diversi  

**Proprietà**

1- commutatore: $[\sigma^i, \sigma^j]= 2i \varepsilon_{ijk} \sigma^k$  
cioè ottengo la terza (quella che non appare a sinistra)  
$\varepsilon \neq 0$ se ijk sono tutti diversi  
e vale $\pm 1$ a seconda se ij sono in permutazione pari o dispari  

2- anticommutazione:  
$\{\sigma^i, \sigma^j\} = \sigma^i \sigma^j + \sigma^j \sigma^i = 2I \delta_{ij}$  

3- proprietà ottenuta dalle due prima:  
$\sigma^i \sigma^j = I \delta_{ij} + i\varepsilon_{ijk} \sigma^k$    

#### Hamiltoniana

**H generale**

so che H deve essere autoaggiunta, quindi la forma generale è:  
$\begin{cases}
\varepsilon_0 & w^* \\
w & \varpsilon_1 \end{cases}$  

nota: $|0 \rangle, |1 \rangle$ NON sono autostati di H  

$E_0 =\frac{\varepsilon_0  +\varepsilon_1 }2$   
$\Delta = \frac{\varepsilon_0  -\varepsilon_1 }2$  

w: costante di accoppiaento  
può essere complessa: $w = w' + i w''$ 
dice quale è la scala di energia che accoppia gli stati    
se $\Delta \gg |\omega|$: 0 e 1 sono troppo distanti, w non è abbastanza forte per mischiarli  

**Autostati**


tip: scrivi  _{
$\hat H = E_0I- w'\sigma?x + w'' \sigma^y +\Delta \sigma^z$  
$= U H_{diag} U^*$  
usando proprietà di amtrici di Pauli    

$H_{diag} = E_0 I + E \sigma^z$   
è una matrice diagonale:  
$H_{diag}  |0 \rangle = E_0 + E |0 \rangle $  
$H_{diag}  |1 \rangle = E_0 - E |1 \rangle $    

$U = exp[i \theta \sigma^y /2] exp[i \varphi \sigma^z /2]$    
è una matrice di cambio di base   

so gli autostati di $H_{diag}$ -> per sapere quelli di H, devo cambiargli la base con la stessa matrice U   
autostati di H:  
$|\phi _+ \rangle = U|1 \rangle \quad E_+ = E_0 + \sqrt{\Delta^2+|\omega|^2}$  
$|\phi _- \rangle = U|0 \rangle \quad E_- = E_0 - \sqrt{\Delta^2+|\omega|^2}$
