### Teoria

**2-levels system**

def: $dim H = 2$  

cioè:  
un spazio di Hilbert con dim=2   
quindi ha solo due livelli nello spettro, $E_0$ e $E_1$    
come si crea: se il gap tra $E_1$ e $E_2$ è molto grande, posso assumere che $E_2$ non esista, le particelle non riusciranno mai a saltare lassù, anche se ricevono molta energia  

un sistema a due livelli è un q-bit (bit quantistico)  

### Setup

#### Stati

**Base**

$|1 \rangle=(1,0)$  (up)    
$|0 \rangle = (0,1)$  (down)   
sono normalizzati e ortogonali  

attenta, quando scrivo le matrici, il vettore 1 viene prima dello 0 nella base  

**Stato generico**

$|\phi \rangle = \alpha |0 \rangle + \beta | 0 \rangle$  
$\alpha^2 + \beta^2 = 1$  

#### Operatori

**Operatori**

$|0 \rangle \langle 0 | $: proiezione su $|0 \rangle$  
$|1 \rangle \langle 1 | $: proiezione su $|1 \rangle$  
$|0 \rangle \langle 1 | $: ladder operator, passa da 1 a 0  
$|1 \rangle \langle 0 | $: ladder operator, passa da 0 a 1  

**Matrici di Pauli**

$\{I, \hat \sigma^x, \hat \sigma^y, \hat \sigma^z \}$

ottenuti combinando gli operatori sopra:  
...  

sono una base completa  
il loro osservabile è il momento angoalre  

**Matrici di Pauli**

formula | matrice | significato
---|---|---
$I = \|0 \rangle \langle 0 \| + \|1 \rangle \langle 1 \| $  | $\begin{bmatrix} 1 & \\ & 1 \end{bmatrix}$ | identità <br> lascia lo stato attuale
$\sigma_x = \|0 \rangle \langle 1 \| + \|1 \rangle \langle 0 \| $  | $\begin{bmatrix}  & 1\\1 &  \end{bmatrix}$ | toggle <br> swappa stati 0 e 1 
$\sigma_y = i \|0 \rangle \langle 1 \| -i \|1 \rangle \langle 0 \| $  | $\begin{bmatrix}  & -i \\ +i &  \end{bmatrix}$ | 
$\sigma_z = \|1 \rangle \langle 1 \| - \|0 \rangle \langle 0 \| $  | $\begin{bmatrix} 1 & \\ &-1  \end{bmatrix}$ | 

ricorda che la prima è la colonna 1, la seconda 0  
$|a \rangle \langle b |$ indica la cella row=a, col=b  

**Autostati**

ogni matrice di Pauli ha autostati diversi  

**Proprietà**

1- commutatore: $[\sigma^i, \sigma^j]= 2i \varepsilon_{ijk} \sigma^k$  
cioè ottengo la terza (quella che non appare a sinistra)  
$\varepsilon \neq 0$ se ijk sono tutti diversi  
e vale $\pm 1$ a seconda se ij sono in permutazione pari o dispari  

2- anticommutazione:  
$\{\sigma^i, \sigma^j\} = \sigma^i \sigma^j + \sigma^j \sigma^i = 2I \delta_{ij}$  

3- proprietà ottenuta dalle due prima:  
$\sigma^i \sigma^j = I \delta_{ij} + i\varepsilon_{ijk} \sigma^k$    

**Altri autostati**

$|\pm  \rangle = \frac 1 {\sqrt 2} (|0 \rangle \pm | 1 \rangle)$  
sono autostati di $\sigma_x$, con autovalore $\pm 1$  

**Altre proprietà**

se  ho due sottosistemi, A e B:  
$[\hat O_A, \hat O_B]= 0$  
cioè, se prend due operatori qualunque, ma uno afisce solo su A e uno solo su B, commutano  
obbio, perchè ognuno non modifica l'altro stato  

#### Hamiltoniana

**H generale**

so che H deve essere autoaggiunta, quindi la forma generale è:  
$\begin{cases}
\varepsilon_0 & w^* \\
w & \varepsilon_1 \end{cases}$  

nota: $|0 \rangle, |1 \rangle$ NON sono autostati di H  

$E_0 =\frac{\varepsilon_0  +\varepsilon_1 }2$   
$\Delta = \frac{\varepsilon_0  -\varepsilon_1 }2$  

w: costante di accoppiaento  
può essere complessa: $w = w' + i w''$ 
dice quale è la scala di energia che accoppia gli stati    
se $\Delta \gg |\omega|$: 0 e 1 sono troppo distanti, w non è abbastanza forte per mischiarli  

**Autostati**


tip: scrivi  _{
$\hat H = E_0I- w'\sigma?x + w'' \sigma^y +\Delta \sigma^z$  
$= U H_{diag} U^*$  
usando proprietà di amtrici di Pauli    

$H_{diag} = E_0 I + E \sigma^z$   
è una matrice diagonale:  
$H_{diag}  |0 \rangle = E_0 + E |0 \rangle $  
$H_{diag}  |1 \rangle = E_0 - E |1 \rangle $    

$U = exp[i \theta \sigma^y /2] exp[i \varphi \sigma^z /2]$    
è una matrice di cambio di base   

so gli autostati di $H_{diag}$ -> per sapere quelli di H, devo cambiargli la base con la stessa matrice U   
autostati di H:  
$|\phi _+ \rangle = U|1 \rangle \quad E_+ = E_0 + \sqrt{\Delta^2+|\omega|^2}$  
$|\phi _- \rangle = U|0 \rangle \quad E_- = E_0 - \sqrt{\Delta^2+|\omega|^2}$  

**Level repulsion**

quando c'è un crossing di livelli energetici  
si generano due nuovi livelli energetici, di energie diverse  

cioè: ???  
ho $\varepsilon_0, \varepsilon_1$  
si uniscono a darmi, invece: $E_+, E_-$  

### Sfera di Bloch

**Cosa**

parametrizzazione degli stati possibili di 1 qubit  

come:   
ogni stato è un vettore $\vec u$ all'interno di una sfera di R=1    
stati misti: $|u|^2 < 1$, cioè punti dentro la sfera  
stati puri: $|u|^2 = 1$, cioè punti sul bordo della sfera  

**Nota**

le coordinate di un vettore lungo le assi x,y,z è il valore atteso di $\sigma_x, \sigma_y,\sigma_z$ su quello stato  

### Sistema con 2 qubits

**Setup**

base:    
$H = span \{|0,0 \rangle, \, |1,1 \rangle, \, |0,1 \rangle, \, |1,0 \rangle, \,\}$  

hamiltoniana di Heisenberg:  
$\hat H = J \vec \sigma^1  \cdot \vec \sigma^2$  

**Risolvi**

scrivo $\hat H$ sulla base sopra  
come: calcolo che vettore ottengo in output quando applico $\hat H$ ad ogni vettore della base, ed ottengo ogni colonna della matrice $\hat H$    

$H =J \begin{bmatrix} 1 & & &  \\  & -1 & 2 &  \\  & 2 & -1 &  \\  & & & 1  \end{bmatrix}$  
il resto sono 0  

autovettori | autovalori 
---|---  
$\|0,0 \rangle$ | J 
$\|1,1 \rangle$ | J
