#### 2. Eq di Schrodinger

**Caso: potenziale costante $V_0$**

eq autovalori di H:   
$H\phi_n(x) = -\frac {\hbar^2}{2m} \frac{\partial^2}{\partial x^2} \phi_n(x) + V_0 = E_n \phi(x)$  
$\phi'' +k^2 \phi =0 \quad k = \frac{\sqrt{2m(E-V_0)}}{\hbar}$ 

sol:  
$\phi(x) = Ae^{ikx}+B^{-ikx}$  

caso | $E>V_0$ | $E<V_0$ 
---|---|--
come è k | $k \in R$ | $k \in C$  
sol | onda piana (oscillante) | onda che decade (o cresce, ma di solito quel coeff lo cancello)
come scrivo sol | $\phi(x) = Ae^{ikx}+B^{-ikx}$ <br> $k = \frac{\sqrt{2m(E-V_0)}}{\hbar}$ | $\phi(x) = Ae^{-k'x}+B^{k'x}$ <br> $ k'=ik =  \frac{\sqrt{2m(V_0-E)}}{\hbar}$ <br> (visto che k è complesso, così $k' \in R$)   
tip per ricordare | E>V -> la particella può essere qui, quidi ci oscilla tranquillamente | E<V -> la particella non può essere qui, quindi ci decade

**Casi speciali**

1- $V_0=0$  
c'è solo il caso $E>V_0$, cioè onda oscillante  

2- $V_0 = \infty$   
c'è solo il caso $E<V_0$, cioè onda che decade
