### Inecrtezze

**Recap**

autostati normalizzati  
l odd: $\pi = 1 \quad \psi(x) = \sqrt {\frac 2 a} \cos (\frac {\pi l} a x)$  
l even: $\pi = -1 \quad \psi(x) = \sqrt {\frac 2 a} \sin (\frac {\pi l} a x)$  

**Formule incertezze**

$\Delta x^2 = \langle \psi | \hat X^2 \psi \rangle - (\psi | \hat X | \psi \rangle )^2$    

$\langle \psi | \hat X | \psi \rangle = \int_{-a/2}^{a/2} dx \; x \; |\psi|^2$  
$\langle \psi | \hat P | \psi \rangle = \int_{-a/2}^{a/2} dx \;\psi(x)  -i \hbar \frac{\partial} {\partial x} \; \psi$  

**Osservazioni**

1- $\langle \psi | \hat X | \psi \rangle = \int_{-a/2}^{a/2} dx \; x \; |\psi|^2$  = 0$  
$|\psi \rangle$ è un autostato di $\Pi$ (pari o dispari, a seconda se l sia pari o dispari, dovrei scrivere $\psi_l(x)$)    
quindi $|\psi|^2$ è pari  
$x$ è dispari    
quindi: integranda dispari -> = 0

2- $\langle \psi | \hat P | \psi \rangle = \int_{-a/2}^{a/2} dx \;\psi(x)  -i \hbar \frac{\partial} {\partial x} \; \psi = 0$  
la derivata cambia il segno della funzione: se $\psi$ è pari, $\frac {d}{dx} \psi$ è dispari  
quindi l'integranda è dispari -> = 0  

3- gli altri addnedi, col quadrato degli operatori, invece sono integrali che devo svolgere  

ATTENZIONE  

$\langle \psi | \hat P | \psi \rangle  = \int dx \;\psi^*(x)  P \; \psi(x)$   
è importante separarle, perchè P potrebbe agire su quella di destra  
se $\psi$ è reale (di solito assumo di sì), allora $\psi^* = \psi$    

**COsa ci faccio**

controllo che $\Delta x^2$ e $\Delta  p^2$ rispettino il principio di indeterminazione
