funzione d'onda: non normalizzabile  
$\phi_k(x)= e^{ikx}$  
k: numero d'onda  

pacchettod'onda: normalizzabile  
sommando onde con diverso k  
$\psi(x,0)= \int A(k) e^{ikx} \, dx$   

esempio:  
pacchetto gaussiano  
$\psi(x,0)= \frac 1 {(2\pi \sigma^2)^{1/4}}exp[-\frac{(x-x_0)^2}{4 \sigma^2} \; ik_0x]$  

Se il pacchetto si muove verso una barriera, inizialmente è localizzato prima della barriera. Dopo l’interazione, può separarsi in una parte riflessa e una trasmessa. È una descrizione più realistica di una particella localizzata rispetto a una singola onda piana.

**Stati scattering**

scattering: sono onde piane (non normalizzabili)  
physical stationary states made of a single plane wave does not exist  
but can use a linear combination of them to create a physical state    
eg. pacchetto gaussiano in free space
