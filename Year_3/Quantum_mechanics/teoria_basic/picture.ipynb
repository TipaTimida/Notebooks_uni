**Rappresentazione/picture**

rappresentazione | di Schrodinger | di Heisenberg
---|---|---
chi cambia nel tempo | gli stati $\| \psi(t) \rangle $|  gli operatori $\hat O(t)$ 
equazione dell'evoluzione temporale | $i \hbar \frac d{dt} \|\psi(t) \rangle = \hat H \| \psi(t) \rangle $ | $\frac{d}{dt}\hat O (t) = \frac i \hbar [\hat O, \hat O(t)]$  
legge dell' evoluzione temporale |  $\| \psi(t) \rangle = U \|\psi(t_0)\rangle $ <br> cioè: applico matrice <br> $U= exp[-\frac i \hbar \hat H(t-t_0)]$  | $ \hat O(t) = \hat U^\dagger \hat O \hat U$ <br> cioè: cambio base  
misurare osservabile: $\langle \hat O \rangle =$   | $\langle \psi(t) \| \hat O \| \psi(t) \rangle$ |  $\langle \psi(t_0) \| \hat O(t) \| \psi(t_0) \rangle$

io uso sempre quella di Schrodinger  

##### Spiegazioni

**Dipendenza di osservabili dal tempo**

$\langle \hat O \rangle = \langle \phi_n | \hat O | \phi_n \rangle = \int |\phi_n|^2 $  
se c'è una fase unica: sparisce quando prendo la norma (oscillazioni hanno norma 1)  
se ho fasi diverse per componente: non sparisce    
perchè $|e^{iA}|=1$, ma $|e^{iA}+ e^{iB}| \neq 1$  


**U**

U: operatore di evoluzione temporale  
la sua formula si trova riarrangiando con l'algebra l'eq di Schrodinger (eq. di evol temporale)  

**A cosa servono i pezzi**

i: garantisce che $\hat U$ sia unitario  
$\hbar$: energia x tempo , fa tornare i conti a liello dimensionale  

**Proprietà di eq. Schrodinger**

1- è lineare e omogenea in $\psi$  
quindi vale il principio di sovrapposizione (per rappresentare sistemi complessi, sommo $\psi$ di ogni componente) 

2- rispetto al tempo, è di primo ordine  
quindi lo stato al tempo $t_0$ determina lo stato in tutti i momenti successivi
