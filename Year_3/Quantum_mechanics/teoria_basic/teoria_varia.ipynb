### Simmetrie

**Come si fa una trasformazione**

trasformazione = cambio base  
quindi devo fare il cambio base sul vettore dello stato e sulle matrici degli osservabili  

$|\psi ' \rangle = \hat U |\psi \rangle$  
$\hat A' = \hat U \hat A \hat U ^\dagger$  

**Cosa vuol dire rispettare la simmetria**

osservabile:  
i possibili outcome, cioè gli autovalori, non devono cambiare dopo la trasformazione  
$\hat A | \phi_n \rangle  = \alpha_n | \phi_n \rangle$  
$\hat A' | \phi_n' \rangle  = \alpha_n | \phi_n' \rangle$  

stati:  
le componenti non devono cambiare (a meno di una fase)  
$|\psi \rangle = \sum c_n |\phi_n \rangle$   
$|\psi ' \rangle = \sum c_n' |\phi_n' \rangle$  
$|c_n|^2 = |c_n'|^2$  

**Cosa è una traformazione**

è descritta da un parametro (continuo) s  
è uguale a quelle classiche:  
traslazione: $\hat U(x)$  
rotazione: $\hat U(\theta)$

### Densità e flusso di probabilità

**Flusso**

mi dice come (verso dove, e quanto veloce) la probabilità si muove nello spazio  

**Formule**

probability density:  
$\rho(\vec r,t)= |\psi(r,t)|^2$  

probability-current density:  
$\vec j(\vec r,t) = \frac{\hbar}{2mi} (\psi^*\nabla \psi-\psi \nabla \psi^*)=\frac\hbar m Im(\psi^* \nabla \psi)$  
in 1D: $j(x,t) = \frac \hbar m Im(\psi^* \frac{\partial \psi}{\partial x})$

flusso:  
$\psi_s = \int_s \vec j \cdot dS$  

equazione di continuità:  
$\frac{\partial \rho}{\partial t} + \nabla \cdot j = 0$  

le equazioni sono molto simili a quelle della corrente normale  

**Esempio**

wave in 1D: $\psi = Ae^{i(kx-\omega t)}$ 
$\rho = |A|^2$   
$j = \frac{\hbar k}{m} |A|^2$  

quindi, la velocità della particella è:  
$v = \frac{ \hbar  k}m$  
perchè???
così: $j = \rho v$
