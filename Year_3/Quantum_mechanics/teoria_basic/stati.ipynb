### Puri e misti

**Incertezza**

correlazione tra stati = incertezza  

correlazione | quantistica (entanglement) | statistica
---|---|---
quando c'è | sempre | solo in stati misti, non puri 
significato fisico | SUPERPOSIZIONE (no mistura) | MISTURA PROBABILISTICA CLASSICA
spiegazione significato fisico | il sistema è davvero in tutti quegli stati contemporaneamente, finchè non lo misuro e decade in uno | il sistema è già in uno solo degli stati, non in una sovrapposizione, ma per MIA IGNORANZA non so in quale
c'è interferenza? | sì <br> perchè sovrapposizione quantistica, ci sono tutti gli stati, interagiscono | no <br> perchè in realtà c'è un solo stato, non può interagire con nessuno

entanglement = correlazione quantistica

**Come si scrivono**

cosa | puro | misto
---|---|---
def | $\rho = \|\psi \rangle \langle \psi \|$  | $\rho = \sum_n \rho_n \|\phi_n \rangle \langle \phi_n\|$ 
spiegazione | caso speciale, n=1 | caso generale <br> $\rho_n$: distribuzione di prob ($\sum \rho_n = 1 \quad \rho_k > 0$) <br> cioè, COMBINAZIONE CONVESSA di stati puri  
autovalori (i $\rho_n$) <br> e rank matrice| ha un eigenvalue 1, il resto 0 <br> rank = 1| ha almeno 2 non-0 eigennvalues  <br> rank>1  
che incertezza | quantistica | quantistica + STATISTICA  
come si scrive | scrivi $\|\psi\rangle $, usando $\|\phi\rangle$ | scrivi $\rho $, usando $\|\phi\rangle \langle \phi \|$

**Esempio**

cosa | puro | misto
---|---|---
come si scrive in base 0,1  | $\rho=\|\psi \rangle \langle \| \psi \| $  <br> $\|\psi \rangle = \frac 1 {\sqrt 2} (\|0 \rangle + \| 1 \rangle )$ | $\rho = \frac 12 (\| 0 \rangle \langle 0 \| + \| 1 \rangle \langle 1 \|)$   
matrice in base 0,1|  $\rho = \frac 12 \begin{bmatrix} 1 & 1 \\ 1 & 1 \end{bmatrix}$ | $\rho = \frac 12 \begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$  
significato fisico | il sistema è in sovrapposizione quanstica tra $\|0 \rangle$ e $\|1 \rangle$  | il sistema è o in $\|0 \rangle$ o in $\|1 \rangle$, con prob del 50% per ognuno <br> ma è già in uno stato solo, sono che ignoro quale 
rango | rank = 1  | rank = 2
se misuto in base 0,1 | $P(0)=P(1) = \frac 12$ | ""
se misuro in base +,- | $P(+) = 1 \quad P(-)=0$| $P(+)=P(-) = \frac 12$
come si scrive in base +,- | $\rho = \| + \rangle$ |  $\rho = \frac 12 (\| + \rangle \langle + \| + \| - \rangle \langle - \|)$  

vedo la differenza se misuro in un'altra base

**Proprietà stati puri**

1- teorema degli stati puri    
A pure state cannot be expressed as a nontrivial convex combination (combinazione lineare) of other states  
a non-pure one can (it's its definition)     

2- evol temporale  
SE è sistema isolato: uno stato puro, dopo evol temporale, resta puro  

**Matrice di densità**

1- popolazioni  
elementi della diagonale principale  
prob di avere ogni stato  
hanno somma 1, perchè $Tr[\rho]=1$  

2- coerenze  
elementi NON sulla diagonale principale  
permettono di avere sovrapposizione quantistica  

### Stazionari

**Tipi di stati**

cosa | stazionario | non stazionario
---|---|---
cosa | autospazio di $\hat H$ <br> $\hat H \| \phi_n \rangle = E_n \| \phi_n \rangle$  | stato generico, sulla base degli autostati di $\hat H$ <br> $\| \psi \rangle = \sum c_n \|\phi_n \rangle \quad \sum \|c_n\|^2 = 1$
evoluz temporale | $\|\phi_n (t) \rangle = exp[\frac i \hbar E_n t] \, \|\phi_n (0) \rangle$ <br> ottiene un fattore di fase globale| $\|\psi (t) \rangle = \sum c_n exp[\frac i \hbar E_n t] \, \|\phi_n (0) \rangle$ <br> ogni vettore della base ha una fase diversa
osservabili | non dipende dal tempo <br> per questo si chiama stazionario | dipende dal tempo

**Cosa vuol dire**

se misuro l'energia, cioè applico l'operatore $\hat H$:  

1- stato stazionario  
ottengo sicuramente $E_n$  
$\langle \hat H \rangle = E_n$  


2- stato NON stazionario  
ottenere $E_n$ con prob $|c_n|^2$  
$\langle \hat H \rangle = \sum_n |c_n|^2 E_n$  


**Spettro continuo vs discreto**

discreto: $|\psi \rangle = \sum_n c_n |\phi_n \rangle$  
continuo: $| \psi \rangle = \int dE \, \phi(E) \; \|E \rangle $   
l'evoluz temproale si scrive uguale per entrambi  

### Legati e scattered

cosa | Scattering states | Bound states (stati legati)| 
---|---|---
condizione sulla waveform | NON normizzabile | normalizzabile <br> (cioè, $\psi(x) \in L^2$, la norma è finita) <br> (cioè $\lim_{x =\pm \infty} \psi(x) = 0$)
cosa descrive | Onde incidenti, riflesse e/o trasmesse; la particella può allontanarsi indefinitamente | La particella è localizzata: la probabilità è concentrata in una regione e tende a zero all’infinito
waveform tipica | Onde piane: $e^{ikx}+r e^{-ikx}$ a sinistra e $t e^{ikx}$ <br> (incidente, riflessa, trasmessa) | Dipende dal potenziale: può oscillare in una parte dello spazio e avere code esponenziali che decadono nelle regioni esterne
| **Uso** | Studiare riflessione, trasmissione e diffusione; descrivere particelle che arrivano o si allontanano. | Descrivere livelli energetici di particelle confinate. |
| **Spettro dell’energia** | Continuo, nei casi usuali. | Discreto, nei casi usuali. |


**Nota**

per vedere se uno stato è bounded o scattered, devo avere uno stato completo, definito su tutta l'asse x  
e devo vedere se decade all'infinito (cioè se è integrabile)  
un pezzo di stato, cioè solo su una regione, NON è ne scattered nè bounded, non ha senso la domanda  

**Come distinguerli**

1- disegna V(x)  
2- traccia una linea all'altezza E  
3- ricorda meccanica classica: moto ammesso solo dove V<E  
se la regione di moto è limitata -> bounded  
sennò -> scattering  

nota: devo vedere quali valori di E sono accettabili   
E sono i possibili autovalori di H, li trovo risolvendo l'eq agli autovalori di H  

##### Pacchetto

**Packet vs scattered state**

1- scattering eigenstate: $\phi(x)$  
the stationary solution, which happens to be scattering     

2- packet: $\Psi(x,t) = \int_0^\infty A(E) \psi_E(x) exp[-\frac i \hbar Et] \, dE$   
one possible way to combine many scattering eigenstates of different energies into a localized wave  
if use the weights: $A(E)= N exp[-\frac{(E-E_0)^2}{4 \sigma^2}]$, i obtain a GAUSSIAN packet  

**Pacchetto d'onda**

cosa: combinazione lineare di varie funzioni d'onda (stati)  
sommando stati scattered, posso ottenerne uno bounded    

eg.  
stato scattered: $\phi_k(x)= e^{ikx}$  
pacchetto (può essere bounded): $\psi(x,0)= \int A(k) e^{ikx} \, dx$   

a seconda dei coeff A, il pacchetto può avere diversi nomi  
pacchetto gaussiano: $\psi(x,0)= \frac 1 {(2\pi \sigma^2)^{1/4}}exp[-\frac{(x-x_0)^2}{4 \sigma^2} \; ik_0x]$  

**A cosa serve**

Se il pacchetto si muove verso una barriera:  
prima: è localizzato prima della barriera  
dopo interazione: può separarsi in una parte riflessa e una trasmessa  

è una descrizione più realistica di una particella localizzata rispetto a una singola onda piana

#### Traccia parziale

**Setup**

sistema formato da due sottosistemi, A e B   

$\rho_{AB}$  
contiene la prob di fare qualunque misura sullo stato intero  

$\rho_A=Tr_B[\rho_{AB}]$  
contiene la prob di fare una qualunque misar sullo stato A, ingorando cosa ottengo sullo stato B  
cioè, sullo stato B può uscire qualunque cosa, non mi interessa

**Traccia parziale**

$\rho_{AB} = \begin{cases}
\rho_{11} & \rho_{12} & \rho_{13} & \rho_{14}  \\
\rho_{21} & \rho_{22} & \rho_{23} & \rho_{24}  \\
\rho_{31} & \rho_{32} & \rho_{33} & \rho_{34}  \\
\rho_{41} & \rho_{42} & \rho_{43} & \rho_{44}  \\
\end{cases}$  

$|01 \rangle$: 0 è stato di A, 1 di B  

$\rho_A = Tr_B[\rho_{AB}] =\rho_{AB} = \begin{cases}
\rho_{11} + \rho_{22} & \rho_{13} + \rho_{24} \\
\rho_{31} + \rho_{42} & \rho_{33} + \rho_{44} \\
\end{cases}$  

$\rho_B = Tr_A[\rho_{AB}] =\rho_{AB} = \begin{cases}
\rho_{11} + \rho_{33} & \rho_{1} + \rho_{34} \\
\rho_{21} + \rho_{44} & \rho_{22} + \rho_{44} \\
\end{cases}$  

spiegazione celle di B:  
elementi sulla diag: prob that B=0, don't care what A does  
so must add the prob of the states: |00> and |10> (elements on the diagonal in the original matrix $\rho_{AB}$)  
primo elemento in alto a destra: $\langle 00| \rho_{AB} | 00 \rangle + \langle 10| \rho_{AB} | 10 \rangle $  

elementi NON sulla diag: coerenza tra B=0 e B=1  
compare \(B=0\) with \(B=1\) while keeping \(A\) the same in each term  
eg, elemento in alto a destra: devo sommare tutti gli elementi dell'originale in cui B=0  
primo elemento in alto a destra: $\langle 00| \rho_{AB} | 01 \rangle + \langle 10| \rho_{AB} | 11 \rangle $  


**Utilizzo**

If you have a combined system but can measure only \(A\), then \(\rho_A\) is enough to calculate the probabilities of measurements on \(A\). It lets you ignore the details of \(B\) while still accounting for how \(A\) is connected to it.
