### Th di Ehrenfest

**Cosa dice**

the average values of position and momentum in a quantum state evolve according to equations that resemble Newton’s laws  
quindi, se studio il moto del valore atteso degli osservabili, ottengono le leggi della emccanica classica  

**Esempio**

cosa: particella in potenziale V(x)   
posso ottenere:  
$\frac{d \langle x \rangle}{dt} = \frac{ \langle p \rangle}{m}$  
$\frac{d \langle p \rangle}{dt} = - \langle \nabla V(x) \rangle$  

ricordano: $v= \frac pm \quad \dot p = -\partial _x V = F$ 

### Indeterminazione

**Varianza**


$\hat A$: operatore  
$\overline A$: valore medio, un numero (???)    
$\langle \hat A \rangle$: valore atteso, un numero  

MQ: $\Delta_A ^2 = \langle A^2 \rangle -\langle A\rangle ^2 = \langle \psi | \hat A-\overline A| \psi \rangle$   
classico: $Var(x) = E[x^2]-E[x]^2 = E[(x -\ mu)^2]$  


**Robertson**

ipotesi: $[\hat A, \hat B] = i \hat C$  

tesi:  
$\Delta_A \Delta_B \ge \frac 1 i |\langle \psi | \hat C | \psi \rangle| $  

**Principio di indeterminazione**

$\Delta Q   ; \Delta P \ge \frac 12 \hbar$  

da dove viene: è Robertson, applicato a Q e P    
perchè so che: $[\hat Q, \hat P] = i \hbar \hat I$  
quindi non è un assioma/principio, ma è un teorema, con dimostrazione  

#### Saturazione

**Saturazione**

saturare una disug: farla diventare uguaglianza  

gli stati gaussiani sono quelli che saturano il pricipio di indeterminazione  

nota:  
gaussiana, dopo Fourier transform, rimane gaussiana   
quindi Q e P, legati da una Fourier transform, possono essere entrambe gaussiane  

**Intuizione: saturazione**

più una funzione/distribuzione ha picchi, più la sua FT è sbrodolata (bassa e ondulata)   
se Q ha un picco, cioè è molto precisa, ho poca incertezza sul valore, allora P, ottenuto tramite FT, è molto largo deve avere molta incertezza
