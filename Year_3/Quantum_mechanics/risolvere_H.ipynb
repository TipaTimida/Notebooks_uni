### Quantum well

**1. Setup**

$V(x) = \theta(-x-\frac a2) V_0+\theta(x-\frac a2 )V_0$  

$V(x) = \begin{cases}
0 & [-\frac a2, \frac a2] \\
v_0 & else 
\end{cases}$

**2. Simmetrie**

see that: $V(x)=V(-x)$  
quindi c'è la simmetria di parità: $\hat \Pi$   

**3. Base dell'operatore simmetria**

osserva: $\Pi^2 = I \rightarrow \pi = \pm 1$   
perchè: $\Pi^2 | \phi_\pi \rangle = \pi^2 | \phi_\pi \rangle  =  | \phi_\pi \rangle$

therefore:  
$\hat \Pi | \phi_+ \rangle = | \phi_+ \rangle$ (pari)   
$\hat \Pi | \phi_- \rangle = -| \phi_- \rangle$ (dispari)  

That means you can choose energy eigenstates that also have definite parity

**4. Autospazi di H**  

$H\phi_n(x) = -\frac {\hbar^2}{2m} \frac{\partial^2}{\partial x^2} \phi_n(x) + V(x) = E_n \phi(x)$  

**4.1 Inside the well**

1- eq. autovalori  
$\phi'' +k^2 \phi=0$   
$k= \frac {\sqrt {2mE}}\hbar \quad \in \R$  

2- sol  
$\phi_I(x) = Ce^{ikx}+De^{-ikx}$  

3- applico la parità:  
cioè cerco A,B che rispettano $\phi(x) = \pm \phi(x)$  
$\pi = +1 \quad \phi_I(x) = A \cos (kx)$  
$\pi = -1 \quad \phi_I(x) = A \sin (kx)$  

4- normalizzo:  
$\phi_I = \sqrt{\frac a2} \cos(l \frac \pi a x)$  
$\phi_I = \sqrt{\frac a2} \sin(l \frac \pi a x)$

### Oscillatore

$\hat H = \frac{\hat P^2}{2m} + \frac 12 m \omega^2 \hat X^2$  

**Scale**

energie: $E_\omega = \hbar \omega$  
impulsi: $P_\omega = \sqrt {\hbar \omega m}$   
lunghezze: $l_\omega = \sqrt{\frac \hbar {m \omega}}$  

a cosa servono:  
per scrivere degli operatori normalizzati, che mi semplificano H  
$\hat q = \frac {\hat Q}{l_\omega}$  
$\hat p = \frac{\hat P}{p_\omega}$  
$\hat H = \hbar \frac \omega 2(\hat p^2+\hat q^2)$

**Autostati di H**

1- definisco nuovo operatore   
$\hat a = \frac 12 (\hat q + \hat p)$  
$\hat N = \hat a^\dagger \hat a$
così: $\hat H = \hbar \frac \omega 2(1+2a^\dagger a) = \hbar \frac \omega 2(1+\hat N)$  

2- osservazione $H= f(N)$  
quindi gli autovalori di N lo sono anche di H  

3- cerco autovalori di N    
eq. autovalori: $N | v \rangle = v |v\rangle$  
ottengo: $\hat N \hat a|v\rangle = (v-1) a |v \rangle$  
cioè $\hat a |v\rangle$ è autovettore di $\hat N$, con autovalore v-1    
posso iterare con v-1, v-2...   
quindi lo spettro degli autovalori è N, tutti i numeri naturali
