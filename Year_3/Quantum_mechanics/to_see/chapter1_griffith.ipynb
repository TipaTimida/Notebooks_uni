#### Normalization

**Normalization**

$\Psi$ is a probability, so must be normalized  
it's obtained by Schrodinger's equation, which is linear, so it's easy to normalize, just multiply my solution $\Psi$ by a value A  
A cannot depend on time, or it would not satisy SChrodinger's equation anymore  

note:  
if get a result $\Psi=0$ or $\Psi=\infty$, it cannot be normalized, so it's not a physical solution, must be rejected

### Proofs

**Proof: normalization is maintained in time**

what i want to prove: if $\Psi$ is normalized at t=0, it's normalized at all times, with the same A  

$\frac {d}{dt} \int |\Psi(x,t)|^2 dx $  
$ =  \int \frac {\partial}{\partial t} |\Psi(x,t)|^2 dx$  
$ =  \int [\Psi^* \frac{d\Psi}{dt} + \frac{d\Psi^*}{dt} \Psi] dx$   
$= \frac{ih}{2m}  \int [ \Psi^* \frac{\partial^2 \Psi}{\partial x^2} - \Psi \frac{\partial^2 \Psi^*}{\partial x^2} ] \; dx \quad $    
$=   \frac{ih}{2m} \int  \frac{\partial}{\partial x}[ \Psi^* \frac{\partial \Psi}{\partial x} - \Psi \frac{\partial \Psi^*}{\partial x} ] \; dx \quad $    
$ = \frac{ih}{2m} [ \Psi^* \frac{\partial \Psi}{\partial x} - \Psi \frac{\partial \Psi^*}{\partial x} ]_{-\infty}^\infty $  
$= 0$  

means that the integral don't depend on t  
so if it's =1 at one time, it's =1 at all times  

steps:  
1- bring d/dt inside  
2- d/dt of product    
3- plug schrodinger  
4- common shortcut using product rule ($\partial_x [AB_x-A_xB] = AB_{xx} - A_{xx}B$)  
5- $\int$ e d/dx si cancellano   
6- limit of $\psi$  

**Proof: fundamental commutation relation**

ricorda:  
$\hat x \psi(x) = x \psi(x)$  
$\hat p_x \psi(x) = -i \hbar \frac{\partial}{\partial x} \psi(x)$ 

$[\hat x, \hat p_x] \psi(x) = \hat x \hat p_x \psi(x) - \hat p_x \hat x \psi(x)$  
$ = x (- \hbar -i \hbar \frac{\partial}{\partial x} \psi(x)) - (-i \hbar \frac{\partial}{\partial x})(x \psi(x))$    
$ = -i \hbar x \psi'(x) +i \hbar [\psi(x)+x \psi'(x)]$  
$= i \hbar \psi(x)$  

$[\hat x, \hat p_x] = i \hbar$  

steps: semplicemente sostituisci i valori degli operatori  

## Momentum

**Derivative of expected value**

$\frac{d\langle x \rangle }{dt} = \frac {d}{dt}\int x \; |\Psi|^2 \, dx$    
$ = \int x \, \frac{\partial }{\partial t} (\psi^*\psi) \, dx$  
$ = \frac {ih}{2m} \int x \; \frac{\partial}{\partial x}[\Psi^* \frac{d\Psi}{dt} + \frac{d\Psi^*}{dt} \Psi] \, dx$  
$=  \frac {ih}{2m} \int [\Psi^* \frac{d\Psi}{dt} + \frac{d\Psi^*}{dt} \Psi] \, dx \quad$   
$= -\frac {ih}m \int \Psi^* \frac{d\Psi}{dt} \; dx \quad$

steps:  
1- bring d/dt inside  
2- plug schrodinger  
3- integration by parts: (dx/dx=1, and $\psi \rightarrow 0$)    
4- integrate by parts the second addend, so it's equal to the first  

**Time-dependence of $<x>$, ALTERNATIVE VERSION FROM INSTAGRAM**   

$\langle x \rangle = \int \psi^* \, x \psi \, d\tau$  
$\frac {d\langle x \rangle}{dt} = \int \frac{\partial \psi^* }{\partial t} \, x \, \psi \, d\tau+ \int \psi^* \, x \,  \frac{\partial \psi }{\partial t} \, d\tau$  
use the time-dependent Schrondiger equation  
$i\hbar  \frac{\partial \psi}{\partial t} \hat H \psi$  
$\frac {d\langle x \rangle}{dt}  = - \frac i \hbar \int \psi^* \, \hat H x \, \psi \, d\tau -\frac i \hbar \psi^* x \hat H \psi d\tau$  
$= \frac i \hbar \langle [\hat H, x] \rangle$  

se $\hat H = \frac{\hat p^2}{2m}+V(x)$  
$[\hat H, x] = \frac 1 {2m}[\hat p^2, x] = ...??? = \frac 1{2m} (-2i\hbar \hat p)$ = -\frac{i \hbar}m \hat p$  

quindi:  
$\frac {d\langle x \rangle}{dt}  = \frac{\langle p_x \rangle}{m}$  
Ehrenfest's therorem for position

## Schrodinger equation

$ih \frac{\partial \Psi}{\partial t} = -\frac{h^2}{2m}\frac{\partial^2 \Psi}{\partial x^2} +V \Psi$  

it's linear, so can use superposition th on $\psi(x,t)$

**Notes**

V is the potential: $F=-\frac{\partial V}{\partial x}$  

how to use it: like Newton's law  
knoing initial conditions for $\Psi(x,t)$, can compute how it changes in time  

**What it means**

hamiltonian operator:  
$H = \frac{p^2}{2m} +V(x)$  
$\hat H = \frac{1}{2m}(-h^2 \partial_{xx}) + V(x)$   

therefore: $\hat H \Psi$  is the right side of the Schrodinger equation  
and the left side is $E \Psi$  
so the Schrodinger's equation is actually: $\hat H \Psi = E \Psi$  

**Note: stationary state**

can find a solution to the Schrodinger equation:  
$\Psi(x,t) = \psi(x) \, e^{-iEt/h}$   
where E is the energy of the particle in this state
