## Stationary state

**Separable solution**

1- ansatz: $\Psi(x,t) = \psi(x) \, f(t)$  
2- plug into Schrodinger  
3- divide by $\Psi$  
4- as constant, instead pf $\lambda$, use $E$  

solution:  
$i \hbar \frac 1f \frac{df}{dt} = E$   
$-\frac{\hbar^2}{2m} \frac i \psi \frac{d^2 \psi}{dx^2}+V = E$  

$f(t) = e^{-iEt/\hbar}$  
$\psi(x)$ depends on V(x)

**Note**

the time-independent equation can be written:  
$\hat H \psi = E \psi$  

and has:    
$<\hat H> = E$    
$<H^2> =E^2$  
so: $\sigma_H = 0$  
measurements are not spread, obtain ceratinly E  

**stationary state**

what it means: $|\Psi(x,t)|^2 = \Psi^* \Psi = \psi(x)^2$  
the prob density don't depend on t, just on x   

so can simplify the formula for finding the expectation value, use $\psi$ instead of $\Psi$ at the sides 


**IMPORTANT**

to be a physical solution, $\psi(x)$ must be:  
1- finite everywhere (even at $|x| = \infty$) 
2- $|\psi|^2 < \infty$   
3- $\psi$ and $\psi'$ continuous

### Infinite square well

**Situation**

$V(x) = \begin{cases}
0 & x \in [0,a] \\
\infty & else
\end{cases}$

**Solve**

1- eq diff
$\psi_{xx}=-k^2 \psi$  
where $k = \frac{\sqrt{2mE}}{\hbar}$   
$\psi(x) = A \sin(kx) + B \cos (kx)$     

2- bounday condition: V(0)=V(a)=0   
$\psi(0) =0 \rightarrow B=0$  
$\psi(a)=0 \rightarrow A \sin(ka)=0 \rightarrow kL=n\pi, \; n \in N$  
so i get:   
$\psi(x) = A\sin (kx) $   
with $k_n = \frac{n\pi}a \quad n \in N$  

3- normalize  
$\int_0^a |A|^2 \sin^2(kx) \, dx = |A|^2 \, \frac a2 = 1$  
so $|A|^2 = \frac 2a$  

4- solution  
$\psi_n(x) = \sqrt{ \frac 2a} \sin(\frac{n\pi}a x)$  

5- substitute k  
so, i have only some allowed values:  
$E_n = \frac{\hbar^2 k_n^2}{2m} = \frac{n^2\pi^2 \hbar^2}{2ma^2}$
