### States

**How to write**

2 ways to write them  

what | $\psi(x)$ | $ \|\psi ⟩$ 
---|---|---
how it's written |  as wavefuncition: $\psi(x)$  | with braket notation
maths meaning | complex valued function | abstract vector in hilbert space 
dependance | dependent on same basis, eg. coord system | indep of any basis

**Note**

the braket notation has no coord  
depending on the coord system i pick, i get:  
position-space wavefunction: $\psi(x) = ⟨ x |\psi ⟩$   
momentum-space wavefunction: $\psi(p) = ⟨ p |\psi ⟩$   

essentially i'm doing a scalar product between $\psi$ and a base, so i get the components of $\psi$ in that base  
like always in linear algebra: $v_i = ⟨e_i ∣v⟩ = ⟨e_i , v⟩$ 


**Fock states**

since states are vectors, need a base to write them  
the standard base is called Fock base, and its elements are:  
$|0⟩=e_0$, $|1⟩=e_1$, ...

**Meaning of states**

the fock states, in an EM field, mean: 
$|0⟩$ = zero photons (called 'vacuum state')    
$|1⟩$ = on1 photon  

**Wigner function**

$W_\psi(x,p)$  
is a scalar function of two real variables  
it's the one i can actually plot  

**Born rule (Max Born)**  

for some measurables (like position), can only predict te PROBABILITY OF THE OUTCOME of this measurement, and not the certain result  
$|\psi(x)|^2 \, dx$  
the prob that the measurement returns a value in the interval [x, x+dx]  

the wavefunction encodes all the information about the system   

**Examples**

to know the prob of one of them happening (prob of having a spcific outcome), do the scalar product of the waveform with that oucome  
so, means i'm projecting the waveform on the base, so taking the component along each member of the base  

waveform: $| \psi ⟩$   
prob of outcome Fock state 1: $|⟨1|ψ⟩|²$  

if i don't project on a specific state, i obtain a frmula with the prob of all states  

### Operators

**Operators**

observe a physical property in the quantum realm   
means an operator acting on a state  

common operators:  
$\hat X$ position, $\hat P$ momentum, $\hat H$ energy, $\hat N $ particle number, $\hat S$ spin


**Meaning**

1- measurement in quantum mechanics (observable)  
described by: hermitian operator  

2- time evolution and quantum operations  
described by: unitary operator
