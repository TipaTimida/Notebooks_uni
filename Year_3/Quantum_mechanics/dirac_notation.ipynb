### Scalari vs vettori

**Scalare vs vettore**

scalari: $f(x), \quad \langle x| v \rangle$  
vettori: $|f\rangle \quad |v \rangle$    
f(x) è uno scalare, perchè prende in input un valore e dà in ouput uno scalare  

passare in RAPPRESENTAZIONE DELLE COORDINATE:  
vuol dire passare da vettori a scalari, moltiplciando per il bra $\langle x|$ (cioè considerando la x-esima componente)  
se scelgo base discreta -> diventa vettore colonna $\langle x |v \rangle =v_x$  
se scelgo base continua ->  diventa funzione $\langle x |f \rangle =  f(x)$   
$\langle x |f(t) \rangle =  f(x,t)$   

esempio: stessa eq con scalari o vettori   
vettori: $\hat Q |v \rangle = \lambda |v \rangle$  
scalari: $\langle x | \hat Q |v \rangle = \lambda \langle x |v \rangle \quad  \hat Q f(x) = \lambda f(x)$

NOTA: in rappresentazione delle coordinate non si mette hat sopra gli operatori

### Commutatori

$[A,B]=AB-BA$  

**Calcoli**

1- è lineare su entrambi gli argomenti (anti-lineare sul secondo???)  
2- I commuta con tutto, quindi [A,I]=0

**Per calcolare**

$[A,B]=-[B,A]$   
$[AB,C] = A[B,C]+[A,C]B$  

**Proprietà**

se un set di operatori commuta: hanno base di autovettori comune  

**Complete set of commuting operators (CSCO)**

un insieme di operatori $A_i$ tale che:  
1- commutano tutti tra loro  
2- il set di autovalori determina in modo univoco uno stato  

**Spiegazione punto 2**

sistema: atomo di idrogeno  
qui, uno stato si descrive come: $|\psi \rangle = (n,l,m)$  

operatori: $H, L^2, L_z$  

scrivo le equazioi degli autovalori:  
$\begin{cases}
H | \psi \rangle = E_n |\psi \rangle \\
L^2 | \psi \rangle = \hbar^2l(l+1) |\psi \rangle \\
L_z | \psi \rangle = \hbar m |\psi \rangle \\
\end{cases}$  

se so tutti gli autovalori, allora so (n,l,m), cioè posso idenificare in modo univoco lo stato $\psi$

**Note**

se commutano tutti, ho base di autovettori  
ogni autovettore lo applico ad un operatore diverso   
prendo il set degli autovettori restituiti, deve essere unico???  

**Esempio**

osservabili: $\{\hat H, \hat S, \hat \Pi \}$ (enegiaa, spin, simmetria)    

1- scrivili come matrici (graficamente)  
2- controlla che commutano tutte le coppie (calcola i prodotti)  
3- diagonalizza S (è il più facile) e trova base di autovettori $\{s\}$
4- scrivi $\Pi$ nella base di autovettori di S (usa formula di cambio base)  
5- trova base $\{\Pi\}$. di autovettori sia di S che di $\pi$  
6- continuacon H  
alla fine, devo avere una base di autovettori in cui non ci sono due elementi della base che danno gli stessi $\lambda$ applicati a tutto il set  
cioè $v_1$ e $v_2$ posso dare entrambi autovalore 1 quando applicati a H e 2 quando applicati a S, ma allora quando applicati a $\Pi$ devono dare autovalori diversi -> devo poterli distinguere   


**Recap: algebra lineare**

2- trova base: per ogni sol, plugga quel valore in $\lambda$ nella matrice, trova base per quell'autospazio

### Spazi

**Spazio di Hilbert**

spazio vettoriale + prodotto scalare  

**Limitato vs dimensione finita**

dimensione finita: la base ha un numero finito di vettori  
cioè un vettore ha un numero finito di componenti (ma le componenti possono avere qualunque valore)
cioè spaizio $R^n$, dove $n < \infty$  

limitato: nessun vettore ha norma infinita (quindi i vettori non possono avere $\infty$ come coord)  
nota che $R^n$ non può essere limitato, quindi si usa per descrivere sottospazi  

recap:  
dim finita: vettori hanno numero di componenti finito  
limitato: componenti dei vettori NON possono essere $\infty$, o dare norma infinita
