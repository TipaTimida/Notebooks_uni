### Postulati

**1. Var dinamiche**

ogni var dinamica è associata ad un operatore $\hat O$  
i possibili valori della var sono gli autovalori di $\hat O$  

**2. Stati**

ad ogni stato fisico del sistema corrisponde un operatore di stato $\rho$   
il valore medio di una var dinamica O sullo stato $\rho$ è: $\langle \hat O \rangle = \frac{Tr(\rho \hat O)}{Tr (\rho)}$  

simbolo | nome | cosa | proprietà
---|---|---|---
$O$ | variabile dinamica <br> osservabile | concetto fisico misurabile (posizione, energia...) <br> dinamica: dipende dal tempo 
$\hat O$ | operatore dell'osservabile | la misura da fare (matrice da applicaere allo stato) per ottenere il valore della var O | $\hat O$ può avere spettro continuo o discreto <br> quindi la var dinamica può avere valori continui o discreti <br> eg. $\hat X$ ha spettro continuo -> la particella si può localizzare nel continuo dello spazio   
$ \rho$ | operatore densità <br> matrice densità | a livello fisico è uno stato, ma a livello matematico è un operatore | 1- hermitiano <br>2- non negativa (autovalori non negativi) <br> 3- traccia=1, cioè $\sum \|c_n\|^2 = 1$
$ \| \psi \rangle $ | vettore si stato <br> wavefunction | come $\rho$, ma solo per stati puri  | def: $\rho = \rho^2$ oppure $Tr[\rho^2]=1$ <br> $\langle \hat O \rangle = Tr(\rho \hat O) = \langle \psi \| \hat O \| \psi \rangle$  




### Rappresentazione

#### Schrodinger vs Heisenberg

**Rappresentazione/picture**

rappresentazione | di Schrodinger | di Heisenberg
---|---|---
chi cambia nel tempo | gli stati $\| \psi(t) \rangle $|  gli operatori $\hat O(t)$ 
equazione dell'evoluzione temporale | $i \hbar \frac d{dt} \|\psi(t) \rangle = \hat H \| \psi(t) \rangle $ | $\frac{d}{dt}\hat O (t) = \frac i \hbar [\hat O, \hat O(t)]$  
legge dell' evoluzione temporale |  $\| \psi(t) \rangle = U \|\psi(t_0)\rangle $ <br> cioè: applico matrice <br> $U= exp[-\frac i \hbar \hat H(t-t_0)]$  | $ \hat O(t) = \hat U^\dagger \hat O \hat U$ <br> cioè: cambio base  
misurare osservabile: $\langle \hat O \rangle =$   | $\langle \psi(t) \| \hat O \| \psi(t) \rangle$ |  $\langle \psi(t_0) \| \hat O(t) \| \psi(t_0) \rangle$

io uso sempre quella di Schrodinger  

##### Spiegazioni

**Dipendenza di osservabili dal tempo**

$\langle \hat O \rangle = \langle \phi_n | \hat O | \phi_n \rangle = \int |\phi_n|^2 $  
se c'è una fase unica: sparisce quando prendo la norma (oscillazioni hanno norma 1)  
se ho fasi diverse per componente: non sparisce    
perchè $|e^{iA}|=1$, ma $|e^{iA}+ e^{iB}| \neq 1$  


**U**

U: operatore di evoluzione temporale  
la sua formula si trova riarrangiando con l'algebra l'eq di Schrodinger (eq. di evol temporale)  

**A cosa servono i pezzi**

i: garantisce che $\hat U$ sia unitario  
$\hbar$: energia x tempo , fa tornare i conti a liello dimensionale  

**Proprietà di eq. Schrodinger**

1- è lineare e omogenea in $\psi$  
quindi vale il principio di sovrapposizione (per rappresentare sistemi complessi, sommo $\psi$ di ogni componente) 

2- rispetto al tempo, è di primo ordine  
quindi lo stato al tempo $t_0$ determina lo stato in tutti i momenti successivi  

#### Coordinate vs impulso

per ogni eq. scrivila in:  
1- notazione bra-ket    
quindi usando gli operatori  

2- rappr delle coords (o degli impulsi)  
quindi scrivi ogni operatore che ci appare nella sua rappr alle coords  
come: moltiplica per $\langle x |$ or $\langle p |$ 

**Esempi**

cosa | braket | rappr coords
---|---|---
formula di H |  $H \|\phi_n \rangle =  \frac{\hat P^2}{2m} + V(\hat X)$ |$H\phi_n(x) = -\frac {\hbar^2}{2m} \frac{\partial^2}{\partial x^2} \phi_n(x) + V(x)$  
eq. autovalori H |  $\hat H \| \phi_n \rangle = E_n \| \phi_n \rangle$  | $H\phi_n(x)  = E_n \phi_n(x)$  

**Operatori importanti**

Rappresentazioni | $\hat Q$ | $\hat P$
---|---|---
coordinate | $Q \psi(x) = x \psi(x)$ |$P\psi(x) = -i \hbar \frac{\partial}{\partial x} \psi(x)$   
impulsi | $Q\psi(p) = i \hbar \frac{\partial}{\partial p}  \; \tilde \psi(p)$ | $P \tilde \psi(p) = p \psi(p)$  

cioè: se cambio la base dei vettori (rappresentazione), devo fare base change alla matrice prima di usarla  
quindi si scrive in un altro modo  

**Autostati**

Rappr | autostati di $\hat Q$ | autostati di $\hat P$ 
---|---|---
eq. autostati |  $ \hat Q \|x_0 \rangle = x_0 \| x_0 \rangle$ | $ \hat P \|p_0 \rangle = p_0 \| p_0 \rangle$
posizione | $\langle x \| x_0 \rangle = \delta (x-x_0)$   | $\langle x \| p_0 \rangle = \tilde \psi(x) = \frac 1 {\sqrt{2 \pi \hbar}}e^{ip_0x(\hbar)}$ <br> onda piana  
impulso  | $\langle x \| x_0 \rangle = \frac 1 {\sqrt{2 \pi \hbar}}e^{ipx_0(\hbar)}$ <br> onda piana  | $\langle p \| p_0 \rangle = \delta (p-p_0)$  

sia onda piana che delta di Dirac sono pseudo-autovettori, cioè autostati generalizzati   

stato generico:  
$\langle x |\psi \rangle= \psi(x)$  
$\langle p |\psi \rangle= \tilde \psi(p)$  


**Relazione canoniche**

$[\hat Q, \hat P] = i \hbar \; \delta_{ij} \quad (i,j = x,y,z)$   
se Q e P sono sulla stessa retta, $\delta =1$  
di solito lavoro in 1D, quindi lo sono  

$[\hat Q_i, \hat Q_j] = [\hat P_i, \hat P_j]=0$   

**Rapporto**

$\tilde \psi(p) = \frac 1 {\sqrt \hbar} FT[\psi]$    

solo legate da trasformata di Fourier

#### Stazionari

**Tipi di stati**

cosa | stazionario | non stazionario
---|---|---
cosa | autospazio di $\hat H$ <br> $\hat H \| \phi_n \rangle = E_n \| \phi_n \rangle$  | stato generico, sulla base degli autostati di $\hat H$ <br> $\| \psi \rangle = \sum c_n \|\phi_n \rangle \quad \sum \|c_n\|^2 = 1$
evoluz temporale | $\|\phi_n (t) \rangle = exp[\frac i \hbar E_n t] \, \|\phi_n (0) \rangle$ <br> ottiene un fattore di fase globale| $\|\psi (t) \rangle = \sum c_n exp[\frac i \hbar E_n t] \, \|\phi_n (0) \rangle$ <br> ogni vettore della base ha una fase diversa
osservabili | non dipende dal tempo <br> per questo si chiama stazionario | dipende dal tempo

**Cosa vuol dire**

se misuro l'energia, cioè applico l'operatore $\hat H$:  

1- stato stazionario  
ottengo sicuramente $E_n$  
$\langle \hat H \rangle = E_n$  


2- stato NON stazionario  
ottenere $E_n$ con prob $|c_n|^2$  
$\langle \hat H \rangle = \sum_n |c_n|^2 E_n$  


**Spettro continuo vs discreto**

discreto: $|\psi \rangle = \sum_n c_n |\phi_n \rangle$  
continuo: $| \psi \rangle = \int dE \, \phi(E) \; \|E \rangle $   
l'evoluz temproale si scrive uguale per entrambi  

#### Legati e scattered

cosa | Scattering states | Bound states (stati legati)| 
---|---|---
condizione sulla waveform | NON normizzabile | normalizzabile <br> (cioè, $\psi(x) \in L^2$, la norma è finita) <br> (cioè $\lim_{x =\pm \infty} \psi(x) = 0$)
cosa descrive | Onde incidenti, riflesse e/o trasmesse; la particella può allontanarsi indefinitamente | La particella è localizzata: la probabilità è concentrata in una regione e tende a zero all’infinito

| **Uso** | Studiare riflessione, trasmissione e diffusione; descrivere particelle che arrivano o si allontanano. | Descrivere livelli energetici di particelle confinate. |
| **Spettro dell’energia** | Continuo, nei casi usuali. | Discreto, nei casi usuali. |


**Nota**

per vedere se uno stato è bounded o scattered, devo avere uno stato completo, definito su tutta l'asse x  
e devo vedere se decade all'infinito (cioè se è integrabile)  
un pezzo di stato, cioè solo su una regione, NON è ne scattered nè bounded, non ha senso la domanda  

**Come distinguerli**

1- disegna V(x)  
2- traccia una linea all'altezza E  
3- ricorda meccanica classica: moto ammesso solo dove V<E  
se la regione di moto è limitata -> bounded  
sennò -> scattering  

nota: devo vedere quali valori di E sono accettabili

### Operatori

autovalori: possibili outcome della misurazione    
valore di aspettazione: la media dei poissibili outcome, pesati secondo la loro prob  

**Regola di Borne**

$\langle \hat O \rangle = \sum_n P_n \lambda_n$  

$P_n = |c_n|^2 $, prob dell'n-esimo autovalore   
$\lambda_n$: i possibili outcome dell'osservazione (autovalori di O)    

per stati puri: $P_n = |\psi|^2$    
quindi $\langle \hat Q \rangle =  \langle \psi | \hat Q | \psi \rangle$   

**Derivata temporale**

$\frac{d}{dt}\langle \hat O \rangle = \frac i \hbar \langle [\hat H, \hat O] \rangle + \langle \frac{\partial \hat O}{\partial t} \rangle$  

se $\frac{d}{dt}\langle \hat O \rangle=0$, quell'osservabile è una costante del moto  
cioè il suo valore atteso non cambia dopo l'evoluzione temporale

### Simmetrie

**Come si fa una trasformazione**

trasformazione = cambio base  
quindi devo fare il cambio base sul vettore dello stato e sulle matrici degli osservabili  

$|\psi ' \rangle = \hat U |\psi \rangle$  
$\hat A' = \hat U \hat A \hat U ^\dagger$  

**Cosa vuol dire rispettare la simmetria**

osservabile:  
i possibili outcome, cioè gli autovalori, non devono cambiare dopo la trasformazione  
$\hat A | \phi_n \rangle  = \alpha_n | \phi_n \rangle$  
$\hat A' | \phi_n' \rangle  = \alpha_n | \phi_n' \rangle$  

stati:  
le componenti non devono cambiare (a meno di una fase)  
$|\psi \rangle = \sum c_n |\phi_n \rangle$   
$|\psi ' \rangle = \sum c_n' |\phi_n' \rangle$  
$|c_n|^2 = |c_n'|^2$  

**Cosa è una traformazione**

è descritta da un parametro (continuo) s  
è uguale a quelle classiche:  
traslazione: $\hat U(x)$  
rotazione: $\hat U(\theta)$

### Indeterminazione

**Varianza**


$\hat A$: operatore  
$\overline A$: valore medio, un numero (???)    
$\langle \hat A \rangle$: valore atteso, un numero  

MQ: $\Delta_A ^2 = \langle A^2 \rangle -\langle A\rangle ^2 = \langle \psi | \hat A-\overline A| \psi \rangle$   
classico: $Var(x) = E[x^2]-E[x]^2 = E[(x -\ mu)^2]$  


**Robertson**

ipotesi: $[\hat A, \hat B] = i \hat C$  

tesi:  
$\Delta_A \Delta_B \ge \frac 1 i |\langle \psi | \hat C | \psi \rangle| $  

**Principio di indeterminazione**

$\Delta Q   ; \Delta P \ge \frac 12 \hbar$  

da dove viene: è Robertson, applicato a Q e P    
perchè so che: $[\hat Q, \hat P] = i \hbar \hat I$  
quindi non è un assioma/principio, ma è un teorema, con dimostrazione  

#### Saturazione

**Saturazione**

saturare una disug: farla diventare uguaglianza  

gli stati gaussiani sono quelli che saturano il pricipio di indeterminazione  

nota:  
gaussiana, dopo Fourier transform, rimane gaussiana   
quindi Q e P, legati da una Fourier transform, possono essere entrambe gaussiane  

**Intuizione: saturazione**

più una funzione/distribuzione ha picchi, più la sua FT è sbrodolata (bassa e ondulata)   
se Q ha un picco, cioè è molto precisa, ho poca incertezza sul valore, allora P, ottenuto tramite FT, è molto largo deve avere molta incertezza
