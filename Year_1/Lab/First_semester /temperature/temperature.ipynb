### Plot

temps: temperatures at all holes

```python
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
model = LinearRegression()

x = np.arange(0, holes_gray)
x = x*dist_gray
model.fit(x.reshape(-1, 1), temps_gray)
preds = model.predict(x.reshape(-1, 1))

plt.figure(figsize=(10, 6))
plt.errorbar(x, temps_gray, 0.5, fmt='o')
plt.plot(x, preds, c='red')          # regression line

plt.xlabel('Distanza [cm]')
plt.ylabel('Temperatura [deg C]')
plt.title('Plot barra di alluminio', fontsize= 20)
plt.grid()

plt.show()

#plt.savefig('alluminio.png')
```

```python
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
model2 = LinearRegression()

x2 = np.arange(0, holes_red)
x2 = x2*dist_red

model2.fit(x2.reshape(-1, 1), temps_red)
preds2 = model2.predict(x2.reshape(-1, 1))

plt.figure(figsize=(10, 6))
plt.errorbar(x2, temps_red, 0.5, fmt='o')
plt.plot(x2, preds2, c='red')          # regression line

plt.xlabel('Distanza [cm]')
plt.ylabel('Temperatura [deg C]')
plt.title('Plot barra di rame', fontsize=20)
plt.grid()

plt.show()

#plt.savefig('rame.png')
```
