## Refraction

light that hits a surface is partly reflected, partly refracted  

reflection: like surface is normal mirror, ray bounces back  
refraction: ray crosses surface changing a bit its direction  

refraction can not happen if the incident angles is too big, in which case it's called total reflection (bc only reflection is left)  

note that effect of rifraction is more noticeable when the incident angle is bigger

**Refractive index**

$v = \frac{c}{n}$  

therefore: 
$v_1 n_1 = v_2 n_2$  

refraction index: n  
that of void is 1, the others are bigger  

refractivity: $\mu = n-1$  
difference in refraction power between a material and void  

**Gladstone-Dale law**

refraction power of a medium depends on its density  
$n-1 = kp$  

p: density  
k: the constant used in thermodynamics, Boltzmann constant

**Snell law**

$n_1 \sin \theta _1 = n_2 \sin \theta _2$  

**Total refraction**

$\sin \theta _{lim} = \frac{n_2}{n_1}$  

if incident angle is:  
- exactly the limit one, ray is reflected along the separation of mediums (in the line that separates them)  
bc $\sin \theta _2 = 1$, so $\theta _2 = 90$ degrees wrt the normal  

- bigger than the limit one, total reflection, like specchio piano  
bc each ray is both reflected and refracted  
since there isn't a possible angle for the refraction, it simply don' happen, only reflection is left  

1: material the rays comes from  
2: material it's going into  

**Apparent deepness**

$\frac{d_2}{d_1} = \frac{n_2}{n_1}$  

if watch (from air) an obj in water from vertically above it, see it at deepness $\frac{h_{true}}{h_{apparent}} = n_{water}$

## Diottrio sferico

**What is it**

calotta sferica with refractive index $n_2$  
ray hits the curved part, then gets refracted inside it  
so like lense, an obj produces an image  

$n_1$: refractive index outside  
$n_2$: refractive index inside  
p: distance from obj (which is outside)   
q: distance of image (which is inside)

**Legge punti coniugati per il diottro**


$\frac{n_1}{p} + \frac{n_2}{q} = \frac{n_2 - n_1}{R}$  

**Focal points**

unlike lense, the two focal points have different distances $f_1, f_2$  
they're one inside, one outside  

$\frac{n_1}{f_1} = \frac{n_2}{f_2} = \frac{n_2 - n_1}{R}$  

**Ingrandimento**

$G = \frac{h'}{h} = \frac{q-R}{p+R}$
