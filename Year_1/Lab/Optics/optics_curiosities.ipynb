### How do plane fly

thanks to shape of wing  
upper side is longer than lower side  
air splits on left, follows wing, reunites on right at same time  
so, since path above wing is longer, air above must go faster  
faster air means less pressure (intuitvely, stagnates less, same thing happens in fluidodynamics)   
so pressure pushing down from up is less than pressure pushing up from down, there is a force pushing upward  
essentially, highest pressure down tries to go up the fill the pressure deficit, doing so pushes the plane upward  

**Limit of atmosphere**

density of atmosphere decreases with distance, how to decide where the limit is?  
limit is height where, to be kept up, a plane would need a speed equal to the orbital speed a satellite would have at that height  
so it's limit between aerodynamics (plane, need motor) and astroynamics (satellite, uses gravity)

after some distance, light can't be seen for 2 reasons:  
1. portata luminosa  
atmosphere absorbs light

2. portata geografica  
earth shape obscures the view, since lamp goes under the line of the horizon

**How it works**

light comes from left outside, hits the crystalline, which is a lense  
to see, the image must always be formed on the retinical plane (the retina, at the end of the ocular bulb)  
which means q is constant, can't change (distance between crystalline and retina)  

**Adjusting focal point to see near vs far**

$\frac{1}{p}+ \frac{1}{q} = \frac{1}{f}$  
q can't change, it's the anatomical distance between crystalline and retina  

to see far:  
far object means $p = \infty$  
technically p is different for each object, but wrt q it's so much bigger than it can be considered infinite the whole time  
so essentially can write  $0 + \frac{1}{q} = \frac{1}{f}$, so i can pretty much keep f constant as well  

to see near:  
when p is of the order of centimetres, it's comparable to q  
meaning it cannot be approximated as $\infty$ the whole time, must put its actual value  
so f changes contantly to adapt to the distance at which i'm looking   

how to change f:  
the ciliar muscles, which are above and under the crystalline, constract and squeeze the crystalline  
this changes its curvature radius, which determines the focal point of a lense  

in graph, focal distance needed in crystalline vs distance of obj observed  
for far objects, f it's almost constant

### Rotating liquid

a rotating liquid has for equipotential surface a paraboloid of revolution (essentially a 3d parabola, shape like when planet curves space)    
so used to make parabolic mirrors  

**Proof**

there are 2 forces acting on any molecule of fluid: weight and centrifugal force  
the potential acting on a infinitesimal mass is:  

gravitational potential: $Φ_g=gz$  
centrifugal potential: $Φ_c​=\int _0^r ​ω^2r \: dr=\frac{1}{2}​ω^2r^2$  

so the potential function is:  $Φ = gz + \frac{1}{2}​ω^2r^2$  

an equipotential surface has $Φ$ constant, so put it equal to c and rearrange  

$z=−\frac{ω^2}{2g}​r^2+\frac{C}{g}$  
​
this is the equation of a paraboloid  

**How to find potentials**

integrate the force per mass $\frac{F}{m}$ over the space?

gravitaional:  
$F= mg$  
$Φ_g = \int  \frac{F}{m} \; dz = gz$  

centrifgal:  
$F = m \omega ^2 r$  
$Φ_c = \int  \frac{F}{m} \; dr = \frac{1}{2}​ω^2r^2$  


**Potential vs potential energy**

potential energy refers to a whole system of mass m  
potential refers to an infinitesimal mass  
so potential is found by diving potential energy by m  

formally: Potential is the potential energy per unit mass (in gravity) or per unit charge (in electrostatics).

the formulas are similar:  
$F=−∇Φ$  
$F=−∇U$  
however the first one, with potential, consider the force per mass unit, $\frac{F}{m}$  
the second one consider the total force
