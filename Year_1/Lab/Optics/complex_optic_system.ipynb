**What is complex system**

complex systems are made by a series of:  
- interfaces between mediums  
- paths through a medium  

often, a complex system is made by a series of thin lenses  
thin lenses, by definition, have trascurable thickness, so all the paths travelled are in the air  

**How to model complex system**  

model each part (interface and path) as a matrix  
multiply all matrices, get a single matrix that describe the system  
like a black box, i give input ray and give me output ray  

! remember matrices are READ FROM TIGHT TO LEFT  

called 'metodo delle matrici di tarsferimento'  
uses linear algebra  

### Travel through a medium

**Example**

what happens: a ray goes in a lastra di vetro, travels a bit inside it, then goes outside again  

$\begin{bmatrix} x_2 \\ \theta _2 \end{bmatrix} = \begin{bmatrix} 1 & 0 \\ 0 & \frac{n_2}{n_1} \end{bmatrix} \begin{bmatrix} 1 & d \\ 0 & 1 \end{bmatrix} \begin{bmatrix} 1 & 0 \\ 0 & \frac{n_1}{n_2} \end{bmatrix} \begin{bmatrix} x_1 \\ \theta _1 \end{bmatrix}$  

the matrices, read from right to left, represent:  
- input ray  
- interface air-glass 
- travel through glass 
- interface glass-air 
- output ray 

**How to indicate rays**

ray, both in input and output, is indicated with a vector with 2 vars:  
$x$: the height the ray is at that intersection  
$\theta$: the intersection angle (angle with the horizontal direction)  

technically the angle should be represented by $\tan \theta$, but used Gauss approximation for small angles $\tan \theta = \theta$  
this makes calculations easier, but causes aberrazione sferica  

**Interface between 2 mediums**

$\begin{bmatrix} 1 & 0 \\ 0 & \frac{n_1}{n_2} \end{bmatrix}$  

$n_1, n_2$: respectively, index of the mediums before and after the interface (coming from 1, going into 2)  
this matrix describe only one interface/wall  
the opposite wall will have other indexes (maybe just swapped, if going back to air)

this matrix essentially applies Gauss ($n_1 \theta_1 = n_2 \theta _2$) on the $\theta$ variable  

**Travel path thorugh a pedium**

$\begin{bmatrix} 1 & d \\ 0 & 1 \end{bmatrix}$  

$d$: horizontal length of path (so of the medium)  

note that:  
there isn't the index, bc the traveling part don't care where it's traveling, that is told by the matrix before it describing the interface (which essentially says from now on, you're in this medium)  
in thin lense, assume d=0, so this would be identity, so can ignore path travelled inside lens  

### Lenses

**Interface of lens**

$\begin{bmatrix} 1 & 0 \\ \frac{n_1-n_2}{n_1R} & \frac{n_0}{n} \end{bmatrix}$  

$n_1, n_2$: respectively, indexes of where it comes from and where it goes  

each lense has 2 interfaces (left and right), which can have different R and swapped $n_1, n_2$  
so each lense is modeled with two matrices, one for each interface  

**Full thin lense**

$ M_2M_1 = \begin{bmatrix} 1 & 0 \\ -\frac{1}{f} & 1 \end{bmatrix}$  

obtained putting together bith interfaces (rem read from right to left, so 1 is on right)  

**Two lenses**

$ M_2M_1 = \begin{bmatrix} 1 & 0 \\ -\frac{1}{f_e} & 1 \end{bmatrix}$  

where $f_e$ is the equivalent focal length  
represents two lenses of $f_1, f_2$, separated by distance $d$  

the full version is multiplying the two lenses and the path in air in the middle:  
$ \begin{bmatrix} 1 & 0 \\ -\frac{1}{f_2} & 1 \end{bmatrix} \begin{bmatrix} 1 & d \\ 0 & 1 \end{bmatrix}
\begin{bmatrix} 1 & 0 \\ -\frac{1}{f_1} & 1 \end{bmatrix}$  

#### Notes and formulas

**How to find focal distance**

$\frac{1}{f} = (n-1)(\frac{1}{R_1} - \frac{1}{R_2})$  

f>0 : convergent  
f<0 : divergent  

called 'equazione del costruttore di lenti', bs used by opticians  

**Dioptric power**

$\frac{1}{f} = D$, dioptric power of the lense (dioptries)  
if stack many lenses tutte attaccate, $D_{tot} = \Sigma D_i$, if there is space in between them must compute it by hand, depend on how much space and what happens to the ray in this travel  

eg: d=2m means f=1/2=0.5m  
d, f can have any multiple of the meter for measure unit   

**Equivalnt focal length**

$\frac{1}{f_e} = \frac{f_1+f_2 -d}{f_1f_2-d(f_1+f_2)}$  

or if $s \sim 0$; with Taylor approximation:  
$\frac{1}{f_e} = \frac{1}{f_1} + \frac{1}{f_2} - \frac{d}{f_1f_2}$  

note that swapping $f_1$ and $f_2$ don't change anything, the formulas are symmetric wrt them  
bc a ray coming from right or left will be reflected in same way  

### Telescope (telescopio riflettore kepleriano)

**What it is**

made of two lenses put so that their focal points coincide: $f_1+f_2 = d$    
called an afocal system  
so, rays coming in parallel to the optical axis are outputed again parallel to the optical axis  

**What's for**

used to ptroduce 'ingrandimento angolare' (to make stuff appera bigger)  

$G = -\frac{f_1}{f_2} = -\frac{\tan \beta}{\tan \alpha}$  
G is how many times the image is bigger than the original (G=2 means image is double than the original)   

**Note on formula**

$d = f_1 +f_2$ is the 'ingombro dello strumento', essentially how long the telescope is  
the lenses are generally the first convergent, the second divergent, so $f_1 >0, f_2<0$    
maybe something like $f_1 = 105, f_2 = -5, d=100$  
the focal point can be outside the tool, no prob
