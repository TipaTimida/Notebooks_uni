```python
# probability p successes on n tries in binomial

import numpy as np
import scipy.stats

n = 4  # Number of tries
p = 0.1  # Probability of success per trial
k = np.linspace(0, n, n + 1, dtype=int)  # Generates an array [0, 1, 2, 3, 4]

prob = scipy.stats.binom(n, p).pmf(k)  # Computes prob of getting each success
```
