**Extimate f of convergent lense**

can only do after having concluded that the lense is convergent  

like method 2 from before  
move lense up and down above the paper  
find distance where spot on paper is a fuoco, measure distance between paper and lense  

why it works:  
can assume light from above comes all vertically and parallel, thus perpendicular to lense  
which means $p=\infty$  
so $q=f$, and q is distance between paper (where the image is) and lense 

**Extimate f of divergent lense**

need both convergent and divergent lense  

1. put convergent lense in front of light, measure its $f_{conv}$   
2. put divergent lens between convergent one and paper, at a specific distance d  
3. keeping the lenses at distance d, move them together up and down until light converge again on paper, the height it's $f_{eff}$  

now use formula:  
$\frac{1}{f_{eff}} = \frac{1}{f_{conv}} + \frac{1}{f_{div}} - \frac{d}{f_{conv}f_{div}}$  
approximation when:  
lenses are thin  
optical axes of lenses are aligned  
$d \ll p,q$  

for the non-approximated formula, would need matrix ultiplication of: lens div - space d - lens conv   
rem, first thing met is the rightest  

$M = L(f_2) T(d) L(f_1) = \begin{bmatrix} 1 & 0 \\ -1/f_{div} & 1 \end{bmatrix} \begin{bmatrix} 1 & d \\ 0 & 1 \end{bmatrix} \begin{bmatrix} 1 & 0 \\ -1/f_{conv} & 1 \end{bmatrix} = \begin{bmatrix} A & B \\ D & C \end{bmatrix}$  

$f_{eff} = -\frac{1}{C}$  

note that the approximations are needed to write the matrix too

### Virtual vs real

mirrors reflect light, don't refract it, so the real ray is the one that goes back the way it come from, can't cross the mirror  
if real rays cross, it's real image and can be captured on a paper  
if the imaginary continuation of rays cross, it's virtual  

upright -> virtual (UV)  
inverted -> real (IR)  

on the other hand, lenses refract light, don't reflect it, so the ral ray is that that goes on the side opposite from where it comes from  
here too, UV and IR    

signs for lenses:  

p: positive is the left side  
$p=10$ means 10m BEFORE the lense, on the left  

q: positive is the right side (OPPOSITE FROM p)  
$p=10$ means 10m AFTER the lense, on the right  
$p=-10$ means 10m BEFORE the lense, on the left  

f: positive is right side  
$f=10$ means 10m AFTER the lense (converging lense)  
$f=-10$ means 10m BEFORE the lense (divering lense)  



virtuale significa che non può essere catturata su uno schermo  
quindi per vederla devo guardare dentro il sistema ottico (guardare attraverso la lente), non posso catturarla su uno schermo  
perchè oggettivamente lì non c'è niente, ma i miei occhi sono in grado di ricostruire un'immagine se guardo attraverso la lente, inventandosi i prolungamenti dei raggi

per vedere, lo schermo è la retina e la lente il cristallino  

### Drawing

in mirror: draw F and C, where C = 2F  
in lens: only draw F, not C (because the center is the center of the lense, not of the curvature)  

draw:  

mirror: riflette, quindi il vero raggio è quello dal lato da cui è partito  
1 orizzontale, riflesso in fuoco  
1 che passa nel centro e viene riflesso su sè stesso  

lense: rifrange, quindi il vero raggio è quello dal lato opposto  
1 orizzontale, riflesso nel fuoco (avanti se convergente, indietro se divergente)

se convergente, 1 che passa nel fuoco prima della lente, riflesso orizzontale

### Esperimento ottica

ordine oggetti:  

1. disegna oggetto e convergente  

2. disegna l'immagine (sarebbe lo schermo)
è reale, perchè sono i veri raggi, quindi posso catturarla su uno schermo

3. metti la divergente tra la convergente e lo schermo  
p è distanza divergente-schermo  
se la divergente è messa prima dell'immagine di prima, per la divergente quell'immagine è virtuale  
perchè se ce l'aggiungo, i raggi in realtà non arriveranno mai a formare quell'immagine, ma verranp deviati prima della divergente  
quindi l'immagine è virtuale, perchè non è creata davvero  
se mettessi la divergente dopo l'immagine, l'immagine verrbbe davvero creata prima di essere usata dalla divergente, quindi per la divergente sarebbe reale 

4. disegna l'immagine che la divergente fa dell'immagine di prima  
distanza di questa dalla divergente è q
