```python
def radix_sort(numbers):
    max_digits = len(str(max(numbers)))  # how many digit places we need

    for digit_pos in range(max_digits):
        # Create 10 empty buckets for digits 0-9
        buckets = [[] for _ in range(10)]

        # Place each number in the correct bucket based on the current digit
        for num in numbers:
            digit = (num // (10 ** digit_pos)) % 10
            buckets[digit].append(num)

        # Flatten the list of buckets back into numbers
        numbers = [num for bucket in buckets for num in bucket]

    return numbers
```
