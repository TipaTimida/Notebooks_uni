**Height of a sound**

distinguishes sounds acuto and grave  
depends on frequency  

humans hear from 20 Hx to 20 kHz  
lowest DO of piano is 32 Hz, highest DO is 4000 Hz  
central LA is 440 Hz by definition, the same as the diapason (used to accord) 

a,b are frequencies  ($\nu _a$)

**Intervals of frequences**  

Intervals of frequences where extremes have a specific rapport

octave: extremes have rapport 2:1  
number of octaves between two frequencies: $octave(a,b) = \ln _2 (\frac{b}{a})$  

semitone: a 12th of an octave (number of white and black tasti in an ottava)  
number of semitoni between 2 freqencies:  $semitone(a,b) = 12 \ln _2 (\frac{b}{a})$  

so lowest and highest DO of a piano distano 7 octaves  


**Distance between tasti of a guitar**

tasti: the line on the part in wood, to press to make notes  

Marsenne law: ties fundamental frequency of vibrating chord with its physical properties (length, tension, linear mass)   
$\nu = \frac{1}{2l} \sqrt{\frac{\tau}{\mu}}$  

distance between 2 tasti (so length of chord that vibrates) must be:  
$\frac{l_2}{l_1} = \frac{\nu _2}{\nu _1}$  

note that $\frac{\nu _1}{\nu _2} = \sqrt[12]{2} = 0.94$




**Volume sonoro**

determined by pression that sound wave exercitates on the tympan  

$L = 20 \log (\frac{p}{p_0})$  
$p_0 = 20 \mu Pa$: is a pression of reference, the 'soglia of udibilità'  
the lowest sound we can hear, thus, has $L=0 \, dB$  

note that we only hear variations in pressure  
$p$ is a component variable in time of the pression  
so atmospheric pression, which don't change, don't make any sound  
ear is a like a condensatore, it filtrates the continuous component of pression
