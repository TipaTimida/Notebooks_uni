### Compute eq of regression

steps:  

1. compute $S_{xx}, S_{yy}, S_{xy}$  
2. compute $S_e$  
3. derivate $S_e$ wrt a,b, and equate to 0  
4. raccogli a,b  
5. isolate a 
6. find eq

**Formula**  

what are symbols:  
- y: values measured direcly 
- $\overline{y}$ mean of the measured values  
- $\hat{y}$: values estimated from regression eq  


1. sum of squares of residuals 
$S_{xx} = \Sigma (x-\overline{x})^2$  
$S_{yy} = \Sigma (y-\overline{y})^2$  
$S_{xx} = \Sigma (x-\overline{x})(y-\overline{y})$  

2.  
$S_{e} = \Sigma (y-\hat{y})^2$  


eg: x=29, y=77, want to fit linear model  
residual: $y - (ax-b) = 77- (29a-b) = y - \hat{y}$  

3. derivate
$\frac{dS_e}{da}=0$  
$\frac{dS_e}{db}=0$  

4. simplify  
arrive at $b=\overline{y}-\overline{x}a$  

5. isolate  
arrive at $a=\frac{S_{xy}}{S_{xx}}$  

use points 5 and 4 to compute a and b, now know optimal params of my model 

**Correlation coefficient (of Pearson)**

$R=\frac{S_{y\hat{y}}}{\sqrt{S_{yy}S_{\hat{y}\hat{y}}}}$  

$R^2 = 1 - \frac{S_e}{S_{yy}}$  

$R^2$ is the determination coefficient  
there is correlation if $R^2 > 0.5$  

**Variance analysis (ANOVA)**

$\sigma ^2 = \frac{S_e}{num.samples -2} $
