## Measurements and errors

### Express a measurement

$x = \hat{x} \pm \sigma _{x} [units] [C.L]$

**Types of measurements**

- single sampling  
regime of null dispersion: all measurements return the same  
eg: ruler, scale  

- repeated measures  
regime of non-null dispersion: human error gives statistical fluctuations  
eg: stopwatch  

**Single sampling**

- $\hat{x}$: measure obtained  
- $\sigma _{x} = \frac{resolution}{\sqrt{12}}$: tool resolution, over sqrt bc it's uniform distribution  
- $C.L.= 58\%$: bc it's uniform distribution 

**Reapeated measures**

- $\hat{x}$: arithmetic mean of measurements (called sample mean)  
- $\sigma _x:$ standard deviation of the mean (called statistical uncertainty) (NOT of the sample, but of the mean)  
$\sigma _m^2 = \frac{\sigma _s^2}{n} = \frac{1}{n(n-1)} \Sigma (x_i - \hat{x})^2$
- $C.L. = 68\%$: bc it's gausssian distribution  

#### Common examples

**Using a digital tool/scale**

i get $m=58$  

if 8 is last digit shown in the display (resolution is 1)  
m has an uniform distribution, could be between 57.5 and 58.5  

$\mu=58$  
$\sigma = \frac{58.5-57.5}{\sqrt{12}}=0.289 = 0.3$  

note that:  
$0.3$ is the statistical eror (std dev)  
$0.5$, half of the resolution, is the maximum error  

**Not independent measures**

eg: mass and length of an object (to compute density) are not independent  
have N measure (l,m), want to extimate linear density  

2 methods:  

1. compute linear density for each object (without uncertainty, just m/l)  
then find mean and standard deviation of density (like with periods), treat density like repeated measures        

2. find mean and standard deviation of m and l  
then find density using error propagation for dependent vars  
uses covariance, hard, don't do it     

in density experiment i used normal error propagation, so am underextimating uncertainty

is the probability that the real value 'dista' at most 1 sigma from the mean  
so it's the integral of the generative function (PDF) with bounds $[\hat{x}-\sigma_x, \hat{x}+\sigma_x]$  
depends on what i use as uncertainty (sigma) and mean:  

- repeated measures  

uncertainty is the standard deviation computed on the mean  
so get a GAUSSIAN distribution bc of central limit th, so use rule 68-95-99 for gaussian (use 1 sigma, so need 68)

## Theory of errors

### Maximum vs statistical error

**Types of errors**

$x = \hat{x} + \Delta x \quad [units]$  
$x = \hat{x} + \sigma x \quad [units]$  


$x$: misurando (real value)  
$\hat{x}$: best extimate, central value  
$\Delta x$: maximum error (smallest interval where i have the certitude to find $x$, 100\% prob)     
$\sigma_x$: statistical error/standard deviation (smallest interval where i have a certain probability, the C.L., to find $x$)



**Why is maximum error bad?**

bc the more measures i do, the interval $\Delta x$ can only grow (if get a value outside the current interval), not shrink  
it's counterintuitive, most measurements should give me more accuracy, not worsen it  

**Relative error**

$err = \frac{\sigma _x}{\hat{x}}$  

### Statistical vs systematic error

**Types of errors**

$\sigma_{statistical}$ : scales on $\frac{1}{\sqrt{N}}$  
if i make more measures can make it as small as i want  

$\sigma_{sistematic}$: does not scale, can't erase it, but if i notice it i can compensate for it  
caused by errors in using tools, like:
  - zero error: put zero in wrong tick in tool scale, so scale is translated (there is offset)  
  - scale error: draw ticks not at correct distance  
  - errors in modeling: if using small angle approximation, must define how small the angle must be
  
eg: reflexes when using chronometer

how to check for systematic errors?
1. use 2 tools (one of which i know to be correct) and compare readings
2. measure known quantity first

### How to lower the uncertainty

standard deviation of the mean:  

$\sigma_x = \sqrt{\frac{1}{n(n-1)} \Sigma (x_i-\hat{x})^2}$   

**How statistical error grows?**

$ \lim _{n \rightarrow \infty} \Sigma^n (x_i-\hat{x})^2 \propto n$  
bc it's a sum of n positive terms, so grows linearly with n  

$\lim _{n \rightarrow \infty} \sqrt{\frac{n}{n(n-1)}} = \lim _{n \rightarrow \infty} \frac{1}{\sqrt{n}}$

how to say it:  
statistical error decresce come $\frac{1}{\sqrt{n}}$  
to increase precision by a factor $c$, must increase number of measurements by a factor $c^2$  

**How many measurements to get a specific uncertainty?**

$\sigma _i \sqrt{n_i} = const$  

do one try, with $n_1$ measurements, get $\sigma_1$ error  
decide the error $\sigma_2$ i want, compute number of emasures needed $n_2$    
$n_2 = (\frac{\sigma _1}{\sigma _2})^2 n_1$  

#### Propagation

**Independent measurements**

$y_0 = f(a,b...)$  
$\sigma _y^2 = \sum \left  ( \frac{\partial f}{\partial a} |_{y_0} \; \sigma _{a} \right)^2$  

derivative is a value calculated in a point  
the summation has term for each var taken by f  

**Not-independent measurements**

$\sigma _f ^2 = \sum \left  ( \frac{\partial f}{\partial x_i}|_{y_0} \sigma _{x_i} \right)^2 + 2 \sum_{i < j} \left( \frac{\partial f}{\partial x_i} \frac{\partial f}{\partial x_j} \text{Cov}(x_i, x_j) \right)$  

must add a term with the covariance between all pairs  

**Relative error**  

$ y = f(a, b, \ldots) = a^\alpha b^\beta \ldots$    
$ (\frac{\sigma_y}{y_0})^2 = (\alpha \frac{\sigma_a}{a_0})^2 + \ldots $  

#### Approximation

**Recap**

y is something i directly measured, can i ignore error on x when fitting?  
only if error from measurements of y (std dev on measures of y) is much bigger than the error i would put on the y from propagating the error on the x  

so there are 2 types of errors on the y:  
std dev from multiple measures of y  
propagation of error of x on y  

y is only obtained through a formula, not measured  
its error is necessarily the propagation of error of x on y  
if some values in the formula have really small errors, can ignore it those parts (use formula of propagation of relative error to check) 


**Can ignore stuff (a term, an uncertainty)?**

if can ignore it or not depends on how precise i want the measurement to be, not how big the obj is  

1. compute $\sigma _f$ IGNORING THE THING i want to ignore  
so assume my hypothesis that it's trascurable it's true  

2. if, by eye, the value i ignored gets inglobated by uncertainty, ok, my hp was correct    

**Conditions to ignore**

$f(a,b,c) = ...$  

can ignore small imperfection? (like a hole)  
(the imperfection must be a whole term/addend, bc must have same units as f, and thus as $\sigma _f$):   
condition: $V_{hole}\ll \sigma _f $ (where f is considering the total volume of the obj, compreso il buco, i'm ingoring the imperfection)  

can ignore an uncertainty $\sigma _a$? (is the error on it trascurable?)  
condition: $\frac{\sigma _a}{a} \ll \frac{\sigma _f}{f}$   

can approximate a value? (eg, approx $\pi = 3$)  
condition: $\frac{3.14-3}{3} \ll \frac{\sigma _f}{f}$   


**Is error trascurable?**

DO NOT write $\sigma _z \ll \sigma _y$  
it's wrong bc entity of the error depends on how big the value i'm measuring is  
also, it's dimensionally wrong, bc x and y can have different units

**Small angle approximation (ignore a term)**

$T(\theta) = T_0(1+\frac{\theta^2}{16} ...)$  
using taylor approximation for sin  

condition to ignore the addend $T_0 \frac{\theta^2}{16}$:    
$T_0 \frac{\theta^2}{16} \ll \sigma _T$  
it's the condition to ignore an uncertainty, like an hole  

so: $\theta \ll  4 \sqrt{\frac{\sigma_T}{T}}$  

small angle depends on precision i want, must compute after getting std dev    
this is for simple pendulum, not physical pendulum  
can use it anyway, but write that it's not quite correct  

**Simplify error propagation (ignore some uncertainty)**

same resolution gives more uncertainty if i'm measuring something small  
so in relative error propagation can only keep the term that gives the greatest uncertainty

eg: compute volume $V=l^2h$, where $h \ll l$  
if use same tool to measure l and h, h will have a much bigger relative error  

so eror on h will be dominant, can ignor error on l  
(relative errors are added in quadratura anyway, so it would be really small)  
$\frac{\sigma _V}{V} \sim \frac{\sigma _h}{h}$  

**Approximate a value**

have $a = 1.003$, can i assume $a=1$?  
depends on precision i want, the relative error of this approximation must be much smaller than the error of the thing i want to compute with this value  

$\frac{1.003-1}{1} \ll \frac{\sigma _y}{\hat{y}}$  
if this is true, can approximate $a=1$ 

**Approximation**

$x\ll y$ if $y \geq 10x$ 

### Compatibility

**When are compatible**

$ | \hat{a} - \hat{b} | \le n\sqrt{\sigma^2_a+ \sigma^2_b }$  

n usually integer smaller than 5  

$a$ and $b$ are the values i'm checking  
if b is the theoric value, that i'm looking up on internet, as uncertainty can take the digit where i'm rounding/troncating (like a resolution) and divide by sqrt, like in single sample from uniform distribution (bc like reading in on a screen with only up to that digit)  
eg: for $g=9.81$ $\quad \sigma = \frac{0.01}{\sqrt{12}}$  

**What then**

- if compatible: compute relative error $\frac{\hat{a}-\hat{b}}{\hat{b}}$
- if not compatible: try to understand why, maybe sistematic errors?

**What values of n mean**

When you assume that the uncertainties $\sigma_a, \sigma _b$  are standard deviations of Gaussian distributions (like it is after a fit):  

$n=1$: means you're testing for compatibility at a 68% confidence level  
There's a 68% chance the difference is due to random fluctuations  

$n=2$: 95% confidence level  
Widely used in science. You're saying: 'If a and b came from the same value, 95% of the time their difference would fall in this range.

$n=3$:  99.7% confidence level  
Very strict; almost certain compatibility.  

so write either:   
"The two values are compatible at the 95% confidence level."  
"Measured and theoretical values agree within 2 standard deviations (95% confidence level)."

**Tool properties**

1. resolution: smallest change $dx$ it can measure (so the smaller, the better) (eg. distance of 2 ticks on a meter)
2. precision: consistency among measurements (how sure it is of its answer)
3. accuracy: accord between $\hat{x}$ and real value (how correct its answer is, so how correct the best extimate i get from it is)
4. sensibility: smallest value of $x$ it can perceive (below that, it returns 0)
