### Calibro ventesimale

**How to use**

express result in mm  

1. move the bottom thing to the largness of the obj, then block it (so don't risk to move it)  
to read, split integer and decimal parts  

2. integer part: from the main scale, the millimeter tick BEFORE the 0 tick on the nonio  
so read the 0 tick on the NONIO, not the end ob the obj  
gives a 2 digit number, like 23 mm  

3. decimal part: from the nonio, the tick that coincides with one of the ticks of the main scale   
gives a 2-digit number that ends in 0 or 5, like 45  

therefore:  
$23.45 \pm 0.5 \, [58\% C.L.]$   

resolution is 1/20 of mm, so 0.05  

**Why it works**

the nonio, the secondary scale, has a spacing of $\frac{39}{20}$ mm, and has 20 ticks  
so at each time, only one can correspond to a tick of the main axis (or two, 0 and 20)  

### Calibro palmer

**How to use**

express results in mm  

0. check there isn't a 0 error, check where the 0 tick is  

1. rotate the manovella

2. see last tack visible on the body of the cilinder  
ticks above the scale are normal millimeters, ticks under the scale are half millimeter  
so can a number with an integer and a decimal digit, where the decimal digit is 0 or 5 (0 if last tack is above, 5 if it's below)  
eg: 7.5  

3. see, on the manovella, the last tick under the horizontal line on the body  
get a 2 digit number, eg: 16  

put together:  
$7.5+0.16 = 7.66$  
$7.66 \pm 0.01 \, [58\% C.L. ]$  

resolution is 1/100 mm, so 0.01  

often at the start there is a 0 error (the 0 is not correctly drawn on the scale)  
so check at the start, if it is can compensate for it by knowing how much it is
