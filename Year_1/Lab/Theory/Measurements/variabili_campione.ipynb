### Sum of many vars

**What is the question**

i have many distributions $x_i$, with $\mu _i$ and $\sigma^2 _i$ known  
so have many aleatoric vars: $x_i$, each from a distribution  

if i write a new distribution $f(x)=...$, what is $\mu_f$ and $\sigma^2_f$ of this new one?  

**General case**

$\mu _f = f(\mu_1, \mu_2 ...)$  

$\sigma^2_f = (\frac{\partial f}{\partial x_1}|_{\mu _f})^2 \sigma_1^2 + ...$  

**Example with 2 vars**

$x=x_1+x_2$  
$\mu = \mu_1 + \mu _2$  
$\sigma ^2 = \sigma_1^2 + \sigma_2^2 + 2Cov(x_1,x_2)$

#### Don't get

**Mean**

approximate f with Taylor until first order around the mean of x, which i know   

$\mu _f = E[f(x)] = E[f(\mu) + \frac{df}{dx} (\mu)(x-\mu)] = E[f(\mu)] + \frac{df}{dx} (\mu)E[(x-\mu)] $  

since $E[\mu -x]=0$, $\mu _f \sim f(\mu)$  
it's $\sim$, not $=$, bc i truncated the approx at the first order  
this means that, if in $f(x)$  x has at most exponent 1, then can put the = sign, since my approximation is actually the full function  

**Variance**

$\sigma^2_f = E[(f(x)-\mu_f)^2] \sim (\frac{df}{dx})(\mu))^2 \sigma ^2$  

## Variabili campione

mean and variance can be either:  

- if KNOW generative distribution, compute it using formula (so start from distribution)    
it's a mathematical property of the distribution, given by principles of theory  
so get $\mu, \sigma ^2$, which are the mean and variance  

- if DON'T know generative distribution (bc want to measure some value), extimate it from some data i measured  
it's a quantity extimated experimentally (an extimate of a corresponding mathematical property of a distribution)        
so get $m, \sigma ^2$ which are media campione and statistical error  

$\sigma^2$ is used to indicate both variance and statistical error   

**Formulas mean**

mean: $\mu = E[x]$  
media campione: $m = \frac{1}{n}\Sigma^n x_i$  

**Prove media campione is impartial**

impartial: expected value corresponds to the real mathematical property of the generative distrib  

media campione is impartial means: $E[m] = \mu$  
$E[m] = \frac{1}{n}\Sigma E[x_i] = \frac{1}{n} n \mu = \mu$   
since, by def, $E[x_i] = \mu$  

**Formulas variance of the sample**

variance: $\sigma ^2 = E[x^2]-\mu^2$  

when computing campione cannot use this relatioship, bc would need $\mu$ of the real distribution, not $m$ that i can compute without knowing the distrib  
so can either:  

1. $s^2 _n = \frac{1}{n}\Sigma ^n (x_i-m)^2$  
intuitive choice, when pick $m = \mu$     
problem: it's not impartial (the expected value of the varianza campione is different from the value of the varianza)  
$E[\frac{1}{n} \Sigma ^n (x_i-m)^2] = \frac{1}{n} \Sigma ^n E[x_i^2+m^2-2mx_i] = ... =\frac{n-1}{n} \sigma ^2$  
it's only impartial asymptotically  

2. $s^2_{n-1}= \frac{1}{n-1}\Sigma ^n (x_i-m)^2$  
this is normalized to be impartial  
that -1 means i'm not actually considering the whole distribution function, since i don't know it, but only some points, which are those i sampled  

the first one is not inherently wrong, sometimes it might be more accurate  
but generally will use the second one  

**Variance of sample vs of mean**

want to write $x = m \pm \sigma$  
$\sigma$ is applief to the mean, so need fluctuation of the mean, not of a sample, cannot use $\sigma _s$     
intuitively the avg will fluctuate less, bc should be already kinda correct, is less likely to get all outliers   

difference: 

- $\sigma _s^2 = s^2_{n-1}$, fluctuations of a sample     
sample a point, expect it will be in $m \pm s_{n-1}$  

- $\sigma _m^2 = \frac{\sigma_s^2}{n}$, fluctuations of the mean  
sample some points, compute the mean, expect it will be in $m \pm s_m$  

therefore: $\sigma _m = \frac{\sigma _s}{\sqrt{n}}$  

**Covariance and correlation campione**

same formulas as before, but using vars campione    
here too use the n-1, bc it comes from the variance   

cov campione: $q_{xy} =  \frac{1}{n-1}\Sigma (x_i-m_x)(y_i-m_y)$  

corr campione: $r_{xy} = \frac{q_{xy}}{s_xs_y}$  
where $s_x$ is the root of the n-1 variance of x  
called 'coefficiente di correlazione lineare' or 'indice di correlazione di Pearson'
