### Density functions

**Functions**

- PDF: probability density function  
$p(x)$ or $f_X(x)$   
how dense the probability is around a point x  

- CDF: cumulative density function  
$F(x) = p(x \le x_0) = \int_{-\infty}^{x_0} p(t) \, dt$    
prob, when sampling an $x$, of it being lower than $x_0$, the one in the argument    

**Intuitive**

prob in a point: $p(x)$  
prob in an interval: $\int_a^b p(x) \; dx$  
simply integrate in all the points that are ok  

**Discrete vs continuous**

$p(x)$ is called:  

- PMF: probability mass function, for discrete distributions  
is prob of getting exactly that value x  

- PDF: probability density function, for continuous distributions  
relative likelihood density of values near x, but not a real probability  
tells which values are more likely, but likelyness given as a relative property, so it's not a probability  
prob of getting a specific value is 0 (bc there are infinite possibilities)  

**What is a probability?**

the PDF, $p(x)$, is NOT a probability, since it can be $>1$  
but must be normalized, total area (which is CDF) must be 1  

the PMF IS a probability  

the CDF IS a probability  
it's used to find prob of a sample belonging to an interval:  
$p(x_1<x<x_2) = F(x_2) -F(x_1)= \int _{x_1}^{x_2} p(x) \, dx$


**Relationsips between PDF, CDF**

$f_X​(x)=\frac{d}{dx}​ F_X​(x)$  

$F_X(x)= \int _{-\infty}^{x} f_X(x) \, dx$  

**Quantile / percent point function (PPF)**

$F_X^{-1}​(p)=x_q$  

$x_q$, the quantile, is the x coord for which the fraction of area before is $p$  
so the quantil function $F_X^{-1}$ is the inverse of the CDF  
$F_X(x_q) =p = p(x \le x_q) \quad  \rightarrow  \quad 0 < p < 1$  

can be called quantile or percent point function  

**Compute quantile: example**

have CDF: $p= F_X(x) = x^2$   
want: $x = F^{-1}(p) = ?$  

just invert:  
$x^2 = p$  
$x = \pm \sqrt{p}$  

so must write the quantile: $F_X^{-1}(p)$  
$F_X^{-1}(p) = \sqrt{p}$  

note: CDF must be non-decreasing (bc going right must add more area, not delete area, since all graph in is first quadrant)  
thus, if get $\pm$, pick $+$  
in this case, the CDF must have actually been:  
$F_X(x)= \begin{cases}
0 & x<0 \\
x^2 & 0<x<1 \\
1 & x > 1
\end{cases}$ 

**Notes on CDF bounds**

technically must integrate from $-\infty$  
but usually $X$ is restricted to some interval  
so must write proper PDF of X, a tratti  

eg: $X \sim U(0,1)$  
$f_X(x) = \begin{cases}
0 & x<0 \\ 
1 & 0<x<1 \\
0 & x>1 \end{cases}$  

so when want CDF, must split integral in these tratti  
clearly the first and last one are 0, so get:  
$F_X(x) = \int_{-\infty}^x f_X(x) \, dx = \int_{0}^x f_X(x) \, dx$  



### Types of exercises

**Problem**

have an aleator variable $X$ from a distribution  
another aleator variable is a function of the first one $Y=g(X)$  
do they follow the same distribution?  

with an uniform distribution in [0,1], $U(x; 0,1)$   


**Answer**

generally no, they don't follow the same distrib  
when stuff is random, the most likely occurrence is an extremal point (if it exists)  
extremal point is either a point at the extremities of the interval or a critical point, with derivative = 0  

**Check y=f(x)**

is g(x) invertible?  
if yes, easy  
if no, why?  
- has non-elementary funcs, like max() -> solve graphically  
- is not biunivoc, like $y=x^2$ -> a little harder to solve, but doable    

**Method 1: CDF**

1. compute PDF of x   
generally x is uniform distibution, so use formula: $f_X(x) = \frac{1}{b-a}$  

2. find inverse, $x= g^{-1}(y)$  

3. compute CDF of y  
$F_Y​(y)=p(Y≤y)=p(X^2≤y)= ... = p(y_0 \le X \le y_1) = \int _{y_0}^{y_1} f_X(x) \, dx$  
this is an example if $y=x^2$, $y_0$ and $y_1$ must be really numbers     

4. compute PDF of y  
$f_Y​(y)=\frac{d}{dy}​ F_Y​(y)$  

**Other method: change of var**

$f_Y​(y)=\Sigma _i \; f_X(g_i^{-1}(y)) \; \;  |\frac{d}{dy}(g_i^{-1}(y))|$  

so, for all the possible $g^{-1}$ (there can be more if, eg, have $\pm$) must get:  
- its derivative: $\frac{d}{dy}(g_i^{-1}(y))$  
- the PDF of the corresponding x: $f_X(g_i^{-1}(y)) = f_X(x)$  
then multiply them together  

easier to see, where $x= g^{-1}(y)$:  
$f_Y​(y)=\Sigma _i \; f_X(x) \; \;  |\frac{d}{dy}(x)|$  

use if cannot integrate $f_X(x)$, maybe bc X is distributed as a gaussian  

### Analytical exercise

**Easy problem**

$U(x; 0,1)$  
$y=x^2$  

1. compute PDF of x  
$f_X(x) = 1$  

2. find inverse of g  
$g(x) = y= x^2$  
$x = \sqrt{y}$  
only + sign, no $\pm$, bc x must be positive, since it's in [0,1]  

3. find CDF of y  
$F_Y​(y)=p(Y≤y)=p(X^2≤y) = p(X \le \sqrt{y}) =   \int_0^{\sqrt{y}} f_X(x) = \sqrt{y} $  
(must integrate in the interval where x is contained, here lower bound is 0)  

4. find PDF of y  
$f_Y​(y)=\frac{d}{dy}​ F_Y​(y)=\frac{d}{d​y} \sqrt{y} = \frac{1}{2 \sqrt{y}}$  

**Medium problem**

$U(x; -1,1)$  
$y=x^2$  

1. compute PDF of x   
$f_X(x) = \frac{1}{b-a} = \frac{1}{2}$  

2. find inverse of g  
$y = g(x) = x^2$  
$x = \pm \sqrt{y}$  
here need $\pm$  

3. find CDF of y  
$F_Y(y) = p(Y \le y) = p(X^2 \le y)  = p(-\sqrt{y}\le X \le \sqrt{y})= \int _{-\sqrt{y}}^{\sqrt{y}} f_X(x) \, dx = \sqrt{y}$  

4. find PDF of y  
$f_Y(y) = \frac{d}{dy}F_Y(y) = \frac{1}{2 \sqrt{y}}$  


**Medium problem with change of var**

steps 1,2 are the same  

have already found: $g^{-1}(y) = x = \pm \sqrt{y}$  
so have 2 possible inverse functions  

$f_Y(y) = f_X(\sqrt{y}) \, |\frac{d}{dy} \sqrt{y} | \quad+ f_X(- \sqrt{y}) | \, \frac{d}{dy} (-\sqrt{y}) |$

since $y=x^2$ is even, can write:  
$f_Y(y) = 2 \, f_X(\sqrt{y}) \, |\frac{d}{dy} \sqrt{y} |  $  

answer is: $f_Y(y) = \frac{1}{2 \sqrt{y}}$  

**After solving**

note that $F_Y(y)$ can diverge in some points, it's fine, don't care  

1. check $f_Y(y)$ is normalized    
integrate in the values y can assume, integral must be 1  

2. can compute mean and variance with $E[y] = \int y \, p(y) \,  dy$ and $E[y^2]$  
bc $f_Y(y) = p(y)$  
again, integrate in the values y can assume  

### Graphical exercise

**Text**

$y = max(x_1,x_2)$  
$X \sim U(0,1)$  

steps to study a distribution:  

1. find $p(x)$  
this is the hard part  
make sure it's normalized  

2. find cumulative distribution  

3. compute stuff (mean, variance)  
just use def and formula $p(x)$ just found  


**Find distribution function**

$max(x_1,x_2)$ means pick a point in 2d plane, with $x_1$ and $x_2$ on the axes x and y   
if it's under the diagonal pick $x_1$, else $x_2$  

so $p(x)$ for a value $x$ is proportional to length of the segment until there: $p(x)=cx$  

use condition of normalization: $\int_0^1 p(x) = 1$ to find $c=2$  

so $p(x)=2x$
