**How to write a distribution**

2 methods:  

$N(x; \mu,\sigma)$  
$X \sim N(\mu, \sigma)$  

before ; the aleator/random variable (value sampled from a distribution)  
after ; the params needed by the distribution   

**What they are**

- binomial  
2 outcomes, prob p of success, 1-p of failure    
prob of k successes in n tries  
eg: number of heads when tossing a coin  

- poissonian  
describes count data (of stationary phenomena) (conteggi)  
prob of k outcomes in a limited time interval, when IN THAT SAME INTERVAL the average is $\mu$ (or spatial area) (unlike binomial, has no upper limit)  
eg: number of messages received in a day, people entering a shop in an hour    

- uniform  
data where each value has equal prob  
useful for uncertainty on digital tools  

- exponential  
data with higher prob for small than large values   
time intervals between events in a stationary/poissonian process  
growth of bacteria, spread of disease (even though it's not a poissonian event, bc the more events already happened, the more infected there are currently, so the more prob of another event happening)  
eg: time between earthquakes  

- gaussian  
data that gets less probable the furthest from the mean  
attractor of all other distributions (central limit th)

### Random walk

**Model problem**

at each turn, takes a step long l, with equal prob towards left or right  
start at x=0

campionary space: $\Omega = f(x) = \{-l, +l\}$  
probabilities: $p(x) = \{ \frac{1}{2}, \frac{1}{2} \}$

**Notation**

$s$: single step, aleatoric var      
$x_n = \Sigma^n s$: x coord after n steps, gaussian (bc sum of aleatoric vars, th of central limit)  
$d_n = |x_n|$: absolute distance after n steps (function of an distribution/aleatoric var)


tip: $E[x_n] = E[\Sigma^n  s] = nE[s]$

**Aleatoric var: step**

$\mu = E[s] = l \frac{1}{2} - l \frac{1}{l} = 0$  

$ Var(s) = E[s^2] - \mu ^2 = l^2 \frac{1}{2} + (-l)^2 \frac{1}{2} - 0= l^2$  
to compute $E[s^2]$, note that $s$ is discrete, so use formula $\Sigma$, not $\int$

**Gaussian: position**

$\mu = E[x_n] = n \: E[s] = 0$ 

$Var(x_n) =E[x_n^2]-\mu_n^2 = E[x_n^2] = E[\Sigma ^n s^2] = nE[s^2] = nl^2$

**Distance**

use method of var change  

find inverse:  
$x_n = f^{-1}(d_n) = \pm d_m$  
since f is even, compute $\pm$ together, then add factor 2 in final formula  
find inverse means find values of x that give me d  
d must be positive, bc $d=|x|$, so the values of x that give d are $\pm x$  

compute parts i need:  
$p(x_n)$: is PDF of the gaussian $N(x_n; 0, \sqrt{nl^2})$  
$\frac{df^{-1}}{d \, d_n} (\pm d_n) = \pm 1$  

put together:  
$p(d_n) = 2 \, p(\pm x_n) \, \frac{df^{-1}}{d \, d_n} (\pm d_n)$  

so must solve the integral (easy, do with var change) to find the mean:  
$E[d_n] = \int d_n \; p(d_n) \quad dd_n =  l \sqrt{\frac{2}{\pi}}\sqrt{n}$  


## Swap between distributions

**Binomial, Poissonian, Gaussian**

- binomial: 2 outcomes, happens or not  
- poissonian: limit of binomial when do many tries  (n-> inf, p->0, np = $\mu$)  
- gauss: limit of poisson, when n-> inf (so used for continuous vars, not discrete)    

n-> inf means $\mu$ must be big  

$B(x;n,p)$    
$P(x; \mu = np)$    
$N(x; \mu = \mu, \sigma = \sqrt{\mu})$

**Poisson vs binomial**

poisson is limit of the binomial when do many tries, so:  
- $n \rightarrow \infty$    
- $p \rightarrow 0$  
- $\mu = np$  


$B(k; n, p) \rightarrow P(k; \mu)$  
$\mu = np$   

---

differences:  

- binomial:  
    - need 2 params: n,p
    - number of occurrences/successes bounded: $0 \le k \le n$
    
- poisson:  
    - need 1 param: $\mu$   
    - number of occurrences not bounded, can be any number i want   
    - if cannot define a prob p of success for the event, but can only see the average occurrences $\mu$

\int _a^b p(x) \, dx &  continuous
\end{cases}$

- discrete: yes, always, but might be long, so there are tricks to simplify

3. turn it into gaussian (if n is big)

### Poisson distribution

**Poissonian process**

event must have 3 characteristics: 
 
1. independence  
an occurrence has no influence on prob of the next occurrence  

2. stationariety  
prob of occurrence is the same in each instant    

3. non-simultaneity  
the event is something that happens or not, binary choice, can't happen twice in the same instant     

used for counting stuff in a specific interval  
eg: how many cars pass, how many people enter

**Explanation**  

divide time in infinitesimal interval: $n \rightarrow \infty$  
make them so small that at most one event can happen in an subinterval, not more  
so each subinterval has a binomial distribution, event can happen or not with prob p  

$p=\frac{\mu}{n}$  


**Exponential: distribution of time intervals between events**

need frequency of events in the full interval: $\lambda = \frac{\mu}{\Delta t}$  
time intervals between 2 events are distributed like an exponential: $p(x \ge x_0) = e^{−λx_0}$  
so the average time interval is the mean of the exponential, $\frac{1}{\lambda}$  

intuitively, if $\lambda$ is the mean, $\frac{1}{\lambda} = \frac{\Delta}{\mu}$ is the period, the total time over the number of events  

sum of poisson distributions give another poisson distributions, where the mean $\mu$ is sum of means of the other two

**Memorylessness**

probability of an event happening don't depend on how much i have already waited  

written formally:  
$P(x \ge s+t | x \ge s) = P(x \ge t)$  

bc in this case, writing the cumulative function gives:  
$P(x \ge s + t) = P (x \ge s) P(x \ge t)$  

**Memory-less: examples**

intuitive example: flip a coin (it's memory-less, but NOT exponential distribution)  
prob of getting head does not depends on what happened on previous throws  
even if before got all tails, prob is always 0.5  

harder example: decay (it's memory-less AND has exponential distribution)  
a radioactive nucleus has 30% prob of not decaying in the first 1s  
after 10s, i check and it's still there: again, the nucleus has 30% prob of not decaying in the successive 1s (come se riazzerassi il tempo)  
don't know if it's the starting 1s or not, don't care  

essentially only care about how long is the time interval before something happens, not when it started  
make it start at the last time i checked and was sure it didn't happen yet  
don't matter how much time passed before that, it's ininfluent

### Th of central limit

take n independent vars $x_i$, from ANY DISTRIBUTION  
each has finite mean and variance (so no Cauchy distrib)  

$S = \Sigma x_i$  
$S$ is asymptotically distributed (so if n is big) as a Gaussian with:  
$ \mu = \Sigma \mu _i$  
$ \sigma^2 = \Sigma \sigma^2 _i$   

Teorema 5.1 (centrale del limite).  

Siano date n variabili casuali indipendenti $x_i$ , ciascuna con media $µ_i$ e varianza $σ^2_i$ finite (non necessariamente uguali). 
Indipendentemente dalla forma particolare delle funzioni di distribuzione delle singole $x_i$ , la somma $S = \Sigma ^n x_i$ è asintoticamente normale nel limite $n → ∞$  
nel senso che per n grande la funzione di distribuzione di S tende ad una distribuzione di Gauss, con media e varianza date dalla somma delle medie e delle varianze delle singole variabili

**Intuitively**

## Computing mean and variance

### Binomial and possonian  

note: in summation, binomial has n as upper bound, poisonian has $\infty$  

**Normalization**

must prove $\Sigma _{k=0}^n P(k; \mu) = 0$  

easy, simply use some kind of identity  

binomial:  
$\Sigma (\frac{n}{k})p^k(1-p)^{n-k} = (p+ (1-p))^n$  

poissonian:  
$\Sigma \frac{\mu^k}{k!} = e^\mu$

**Mean**

must prove:  
$\mu = E[k] = \Sigma_{k=0}^n  k B(k;n,p) $  

1. addend with $k=0$ is 0, so can start the sum from $k=1$ 
(just change it under the $\Sigma$ sign, nothing else)

2. bring out of the summatory an $np$ (can, since no $k$ appear in it)  
and the k next to the B, remove it from the k! at the denom of the b, so get (k-1)!  
all k must appear as $k-1$ or as $n-k$    

3. now do var change:  
$h=k-1$  
$m=n-1$   
so $n-k = m-h$  

4. reorganize to have:  
$\mu = np \, \Sigma_{h=0}^m B(h; n,p)$  

stuff inside the summatory is normalized, it's 1, so $\mu = np$

**Variance**

must prove:  
$E[k^2] = \Sigma_{k=0}k^2 B(k;n,p)$  

same steps as mean  

var change is also $h=k-1$, no n, bc there isn't n  

but at the end will get inside the summation $\Sigma (h+1)B(h;n,p) = \Sigma hB(h;n,p) + B(h;n,p)$  
the first is $\mu=np$, bc i computed it before, it's the mean  
the second is 1, bc it's normalized

4. Compute
