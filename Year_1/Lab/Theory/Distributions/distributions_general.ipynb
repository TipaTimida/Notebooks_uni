### Properties distributions  

**Normalization**

it's normalized when  
$\int _{-\infty}^{\infty} p(x) dx = 1$  

if it's not, must divide the funct p(x) by this value to normalize it  

### General

### Expected value

$E[f(x)] = \int _{-\infty}^{\infty} f(x) \: p(x) \: dx$   

$f(x)$: array of possible outcomes, campionary space   
$p(x)$: array of same length, each is probability of occurrence of the outcome in f(x) (depends on the distribution used)


**Intuitively**

$E[x]$ is the weighted average  
mutliply each outcome for its probability, find the average between these values  
eg: in standard dice it's 3.5  

**Properties**

linearity:  
$E[cx +y] = cE[x] + E[y]$  

$E[c]=c$  if c is a constant  

**Note when computing**

p(x) is always prob of x, no matter what i'm computing E of  
eg: $E[x^2] = \int x^2 p(x)$  
no $p(x^2)$  

### Stuff to compute

if discrete, change all integrals to summation  

#### Mean

**Mean**  

$\mu = E[x] = \int _{-\infty}^{\infty} x \: p(x) \: dx$      

it's weighted average  

#### Variance

**Variance**

$\sigma ^2 = E[(x-\mu)^2] = E[x^2] - \mu^2$   
$Var(x)=\sigma ^2$  

properties:  
$Var(cx) = c^2 Var(x)$  
$Var(x_1+x_2)= Var(x_1)+Var(x_2)$ if independent  

**How to prove a property**

$\sigma ^2 = E[(x-\mu)^2] = E[x^2-2x\mu + \mu^2] = E[x^2] -2\mu[x]+\mu^2 = E[x^2] -2\mu^2 + \mu^2 = E[x^2]- \mu^2$   


remember:   
$E[x]=\mu$  
$E[c]=c$, and $\mu$ is a const , so $E[\mu]=\mu$   
$E$ is linear  
$\sigma ^2 =  E[x^2]- \mu^2 = \int x^2 \, p(x) \, dx \; - \mu^2$  

**Why variance has square?**

same problem as absolute error, bc positive and negative fluctuations cancel each other out:  
$E[x- \mu] = E[x]-E[\mu] = \mu- \mu = 0$  
bc $E[x] = \mu$ by def  

use std dev for uncertainty bc has same units as misurando  

#### Other stuff

**HWHM (half width half maximum)**
   
find x coords where p(x) is half the max (one on each side of center)   
take half their distance      
usually it's close to sigma, with a factor nearly 1  
(there is also FWHM, where take full distance between the  coords, not half)  

**Confidence Level** 


$C.L. = p(|x-\mu| \le \sigma) = \int_{\mu -\sigma}^{\mu + \sigma} p(x) \, dx$   

prob of a sampeld value to be distant less than a std deviation from the mean 

**Median**  

m such that:  
$\int_{-\infty}^{m} p(x)\, dx = \int^{\infty}_{m} p(x)\, dx = \frac{1}{2}$

#### Momentum of a distribution

momentum of order n around point $x_0$:  

$M_n(x_0) = E[(x-x_0)^n] =  \int _{- \infty}^{\infty} (x- x_0)^n p(x) dx$  

compute fluctuations around a point, using different exponents  

**Types of momentum**

depends around which point, 0 or the mean:    
- algebric: $M_n(0) = \lambda _n$, 
- central: $M_n(\mu) = \mu _n$

normalization: algebric of order 0  
mean: algebric of order 1   
variance: central of order 2  

if distrib is symmetric around $x_0$, the momentum of odd order around $x_0$ is null  

**Skeweness**

$ \gamma = \frac{\mu _3}{\sigma ^3}$

skewness is an adymensional quantity that is:
- 0: symmetric wrt to mean 
- $ > $ 0: right extreme goes further (bc in formula of moment, positive part is right part)
- <0: left extreme goes further 

**To compute skewness**

compute central moment of order 3:  
$M_3(\mu) = \mu _3= E[x^3] - 3\mu \sigma ^2 - \mu ^3$  

it's the first momentum of of odd order that 'non si annulla' (bc central of order 1 always gives zero)  
useful bc it's odd (dispari): fluctuations on left of the mean are negative, on right are positive, so can find asymmetry, using mean value as axis  
use it to define asymmetry coefficient, or skewness

### Multivariable distribution

**Congiunt probability**

function of congiunt probability: function that associates to each set of vars the prob of all of them happening together

$p(x=x_0, y=y_0) = p(x_0, y_0)$

**Densità di probabilità marginale**

$p(x_0) = \int_{-\infty}^{\infty} p(x_0,y_0) dy_0$  
$p(y_0) = \int_{-\infty}^{\infty} p(x_0,y_0) dx_0$  

prof of finding a specific $x_0$, independently of value of y  
the prob of getting a specific $x_0$ depends only on $x_0$ itself, bc $y_0$ is deleted when doing the definite integral   

**Marginalization: intuively**

technique to extract probability distribution of a specific var in a congiunt distribution of many vars  

for discrete distribution, it's:  
$p(x_0) = \Sigma _y \: p(x_0,y)$  
so to obtain prob distrib of x, need to consider for each value of x all its occurrences, whatever the y it is paired with  
this is called marginalizing y

sometimes marginal probability are written with a subscript, like $p_1(x_1), p_2(x_2)$  

**When 2 vars are independent?**

when value of one don't influence value of the other  
can be written either:  
$p(x_1,x_2) = p(x_1)p(x_2)$  
$p(x_1|x_2) = p_1(x_1)$  

so whatever is the value of $x_2$, the prob of getting a specific value for $x_1$ don't change, it only depends on what value exactly i want  

**Check if two vars are independent**

have $p(x_1,x_2)$  
compute the marginal probabilities $p_i(x_i)$ (is always possible)  
if $p(x_1,x_2)=p_1(x_1)p_2(x_2)$, they are independent  

**Expected value**

similar to single-var, just need to add all possible cases for non-fixed vars

3 types:  
- marginal  
most likely value of a single var, ignoring the fact that there are others  
(normal expected value)  
$E[x] = \Sigma _{x_i} \: x_i \: p(x_i)$  

- conditional  
most likely value of of a var, knowing values of other vars  
(consider only allowed cases)   
$E[x|y] = \Sigma _{x_i} p(x_i|y)$


- congiunt  
the most likely config of all vars  
$E[xy] = \Sigma _{x_i} \Sigma _{y_j} \: x_i \: y_j \: p(x_i,y_j)$  

if they are independent, $E[x_1x_2] = E[x_1]E[x_2]$  
actually this means no LINEAR dependence, so it's not necessary but not sufficient for independency  

##### Covariance, correlation

**What they are**  

- covariance  
measures linear relation between vars x and y (extimates m of fit line)   
formula similar to variance, but use both vars  

- correlation  
covariance, but normalized (divided by the standard dev)  

$cov(x,y) = \frac{1}{n-1} \Sigma (x- \mu_x)(y- \mu _y)$  
$corr(x,y) = \frac{cov(x,y)}{\sigma_x \sigma_y}$


**Proper def covariance**

$Cov(x,y) = E[xy] - E[x]E[y] = E[xy]- \mu_x \mu_y$  

**Independent and null cov/corr**

if vars are independent ( so $E[xy] = E[x]E[y]$), cov and corr are 0  
but if they're 0, not necessarily the vars are independent  

eg: $cov(x, x^2) = E[x^3]-E[x]E[x^2]=0-0=0$  
bc x is symmetric wrt 0, so the momenti of odd order $E[x]$ and $E[x^3]$ are null  
but clearly $x$ and $x^2$ aren't independent   

**How to use**

- COVARIANCE   

<u>how big it is has no impact</u>, depend on how big values are  
<u>only care about sign</u>, used to see if m of fit line is positive or negative (if both increase together, or if one increase while the other decrease)  
if vars independent, their covariance is null (the opposite not always true, bc covariance senses only linear dependence, not other types, so could have some other type of dependance)    


- CORRELATION  

measure both strength and direction of correlation 
scaled between -1 and 1 (-1 and 1 are perfect correlation, 0 is no correlation)  

**Properties of covariance**

bilinear, symmetric  

$cov(a,b) = cov(b,a)$  
$cov(a,a) = var(a) = \sigma ^2_a$  
$cov(a, const) = 0$  
$cov(ax_1 + bx_2, y) = a cov(x_1, y) + b cov(x_2, y)$   

eg:  
$cov(x, mx+q) = m cov(x,x) + cov(x, q) = m \sigma_x^2$  
cov between a var and a const (q) is null, bc even it var changes the const doesn't  

**Matrix**

can be covariance or correlation matrix  
if have N vars, NxN matrix, assign each var to a row and col like chessboard  
in each spot, plot $cov(x_i, x_j)$   
note that $cov(x_i, x_i) = var(x_i)$

### What?

**Many times same experiment, find conclusion**

many people do same experiment, of measuring something  
each get result $x_i \pm \sigma_i$   
to combine this results, the final result is the weighted average of the $x_i$, where the weight of each one is $w_i = \frac{1}{\sigma_i^2}$  
uncertainty of final result is $\frac{1}{\Sigma w_i}$  
this final uncertainty is smaller than all of the uncertainties $\sigma_i$  

to combine result of different experiments that measure same quantity:  
weighted sum  
$x = \Sigma \frac{w_ix_i}{w_i}$  

where the weigh is $w_i = \frac{1}{\sigma_i^2}$  

the new uncertainty is:  
$\sigma^2 _x = \frac{1}{\Sigma w_i} = \frac{1}{\Sigma \frac{1}{\sigma _i^2}}$
