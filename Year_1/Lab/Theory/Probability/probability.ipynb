#### Theory

**Probability vs statistic**

probability: branca della matematica pura  
know the model, can say stuff with certainty using deductive reasoning (eg. prob is x%)  
eg: box has 2 red balls, 4 blue balls, what is prob of getting a red ball?  

statistica: mezzo di indagine dei fenomeni fisici  
know some info about a system, but not the model, want to trarne conclusioni sul modello del sistema using inductive reasoning  
eg: box has 5 balls, some red some blue. extract a blue ball, then extract a red ball, what is probably the number of red balls?

**Terminology**

phenomenon: throw a dice  

elementary realizations: get 1, get 2  

campionary space $\Omega$: set of all elementary realizations  
$\Omega = \{1,2,3,4,5,6\}$  

event space $F$, power set of $\Omega$: set of all possible subsets of $\Omega$ 

**Kolmogrov axioms**

- non-negativity  
prob of an event happening can't be negative  

- normalization  
sum of probs is 1  

- additivity  
if events are disjoint, prob of one happening is the sum  

**Definitions**

1. combinatorics: $p = \frac{favorable}{possible} = \frac{n}{N}$

2. frequentist: $ \lim_{N \to \infty}\frac{n}{N}$  
like combinatorics, but assume doing infinite tries  
problem: only works in repeteable experiments, and not work on hypothesis  
   - prob tomorrow rains, not repeatable
   - prob this model is correct, it's an hypothesis  
   
3. bayesian: how much you'd bet on it happening, based on info you have  
when gaining new info, update bet with Bayes theorem

### Bayes theorem / conditioned probability

**Conditioned probability**

$p(A|B) = \frac{p(A \cap B)}{p(B)}$  
  
prob of A happening, knowing B has happened  

intuitively:  
at numerator i'm considering prob of both A and B happening  
then at denominator i take away the prob of B happening, bc i already know it happened, so its prob is 1  


**Bayes theorem**

$p(A|B) = \frac{p(B|A) \; p(A)}{p(B)}$  

i simply rewrote the formula for conditioned probability, expanding the numerator  

to remember it:  
$p(A \cap B) = \quad p(A|B) \, p(B) = p(B|A) \, p(A) \quad$   
bc the union is symmetric  

**What the terms mean**  

i'm interested in finding $p(A|B)$, so this the probability i want  

$p(A|B)$: posterior probability (A happening knowing B has happened)  
$p(A)$: priori probability  
$p(B|A)$: likelihood (versomiglianza) (prob of them happening together)  
$p(B)$: normalizing constant  

**Find normalizing constant**

$p(B) = \Sigma \, p(B \cap C_i) = \Sigma \, p(B|C_i) \, p(C_i)$  

consider $C_i$, a list of disjoint event that form a partition, so exactly one of them must happen  

**Expanded form: sick test**  

$S$: person is sick  
$H$: person is healthy  
$Y$: test says person is sick (test say yes)  
$N$: test says person is healthy (test say no)  

$p(H|N) = \frac{p(N|H)p(H)}{p(N|S)p(S)+p(N|H)p(H)}$  

**Independent events**

possible definitions:  
$p(A\cap B) = p(A)\, p(B)$  
$p(A|B)=p(A)$

the fact that one happened does not change the prob of the other happening (two-way)  
so conditioned prob is equal to normal prob, for both   

### Chebyshev inequality

$p(|x-\mu| \ge c \sigma) \ge \frac{1}{c^2}$

**What it is**

a gross extimation of prob of how many std deviations a value is distant from expected one (rem, the mean is the expected value)  
is always valid, it's not as stringent as it could be  
if know exactly the distribution can find better prob  

the proper way would be $p(x-\sigma<x<x+\sigma) = \int _{x-\sigma}^{x+\sigma} p(x) \, dx$ using the correct $p(x)$ for that distribution  
the proper way is more precise, but longer  

**How to use it ???**

1. get x, $\sigma$ from my experiment  
x is the mean, $\hat{x}$  

2. need to know theorical value of x, usually indicated with $\mu$  
so $\mu$ is the theorical value of $x$, not the mean i found  

3. find c by measuring distance in units of std_dev  

4. now i know the probability  

### Trick question

**Trabocchetto binomiale**

coin has 99% change of giving tails  
if flip 100 times, what is prob of getting 99 tails?  

it's NOT 99%, it's like 37%  
compute it with $B(k=99; n=100, p = 99/100)$  
formula of binomial distribution, the one with newton coefficient  

eg: $(\frac{99}{100})^{99} = (1-\frac{1}{100})^{99} = (1-\frac{99/100}{99})^{99} = e^{-99/100} \sim e^{-1}$
