### Chi squared test

#### What it is

**Statistical test**

$\chi ^2$ is a statistical test for hypothesis testing  
compute a value from data, which tells me <u>if hypothesis when building model are correct</u>    

hypothesis when building model means:  
the form of the function f  
and values picked for params

**What chi represents**

quantifies the scarto between my model and the data points (so between real and expected value)  
the smaller the value, the better my fitted function follows my data points (the better the model represent the data)

$\chi$ is a gaussian var in normal form (chi, not chi squared)  

**What it does**

used to JUDGE and OPTIMIZE params  
params means vars in the function f(a,b,c...), the stuff i want to optimize  

how to use it:  
1. compute value of $\chi ^2$ plugging in current params  
2. judge: is it close to the expected value?  
3. optimize: use gradient descent on the params  
means on the formula for $\chi ^2$, compute derivative wrt all params (with current param values), then change each param accordingly, so $\chi^2$ gets closer to what i want it to be    

if the function f is easy:  
instead of doing gradient descent and improving the params step by step, can just put the derivative wrt all params equal to zero and have equations to find immediately the best params  

#### How to compute

2 formulas, depends on context:  

 
1. to judge a fit (goodness-of-fit test)  

$\chi ^2 = \Sigma (\frac{y_i-f(x_i)}{\sigma_i})^2$  


2. to check if some data follows a specific distribution  

$\chi ^2 = \Sigma \frac{(o_i-e_i)^2}{e_i}$   


**To remember**

the chi square is made by a sum of squares of gaussian vars in normal form  
gaussian var in normal form is: $\frac{x-\mu}{\sigma}$  
this is the var, not the PDF of the var  

**Check if data follows some specific distribution**


$o_i, e_i$: observed and expected occurrences for each category  


La statistica dei conteggi è Poissoniana, e dato un numero di conteggi medio $e_i$, la deviazione standard corrispondente è $\sqrt{e_i}$, per cui il denominatore della $e_i$ rappresenta esattamente l’equivalente
del quadrato dell’errore di misura.  
essentially: samplings from an occurrence (so any count data???) can be thought of like a poissonian distribution (bc take samples one at a time, and with a constant rate)


**What are the elements**

i measured some y and x, such that $f(x, params) = y$  
then i fitted it to find the params  

thus:  
$x,y$: measured values (in machine learning called target and feature)  
$f(x)$: predicted value of y, using the function with the params i fitted   
$f$: the fit function of which i'm looking for params  (eg: $f=ax^2$)  
in this case, $\chi ^2$ is the loss function (it's a kinf of MSE, average squared error)

so each addend is how many std deviations the real and predicted value are distant, squared  

#### Draw conclusion

**What is a good value for $\chi ^2$?**

not too big, but neither too small, must be close to the mean i expect     
same for p-value, must be netiher 0 nor 1  

**Why is $\chi ^2$ too big?**

2 possible causes:  

- numerator is big: f(x) is a bad function to model it, or the params are wrong, so a lot of difference with y  
- denominator is small: underestimating errors, need to put bigger uncertainty  

**Why is $\chi ^2$ too small?**  

2 possible causes: 
- overfitting (so change model, this has too many params)  
- overestimating errors

#### Error at denominator

**What does it mean error on x is trascurable?**

condition for trascurability:  
$\frac{\sigma _x}{\mu _x} \ll \frac{\sigma _y}{\mu _y}$     

if $x_i$ have trascurable error:  
$\sigma _i =  \sigma _{y_i}$, are simply the errors on $y_i$  

if $x_i$ HAVE ERROR:  
2 methods:  
1. use ODR (orthogonal distance rejection), a different loss function    

2. propagate errors on $x$ and $y$ to get total error on denominator  
$σ^2_i = σ^2_{y_i} + (\frac{df}{dx})^2 \sigma _{x_i}^2$  

both troublesome, just try to assume error on $x$ are trascurable  

**Method 1: ODR**

ODR is a different loss function  
useful if have error on both x and y  
instead of minimizing vertical distance between measued and expected point (y-f(x)), tries to minimize distance of point from the graph of the curve  
so the distance is not a vertical line, can be diagonal  

steps: 

1. do a first fit, ignoring error on x
2. use the current best-fit values of params to extimate derivative $\frac{df}{dx}$  
3. do a new fit, using the derivative computed at step 2 in the complete formula for uncertainty to propagate the error on x   
so now have both error on x and y   

repeat steps 2 and 3 many times, values will get always better  
if have easy model, like linear, one iteration should suffice, won't change much even if continue  

ODR vs OLS  
ODR: orthogonal istance regression, considera errori su x e y  
OLS: ordinary least squares, considera errori solo su y

**Method 2: propagate errors**

often the params in $f$ are not independent, but are correlated  
so would need to propagate thorugh $f$ using the hard formula with the covariance  
so too hard    

### Degrees of freedom

**DOF: degrees of freddom**

what it is:  
number of independent terms in summation $\chi ^2 = \Sigma x_i^2$   


DOF: $\nu = n-m$  
n: how many points $x_i$ i have (normal degree of freedom, experimental data, used when considering just a distribution, not a fit)    
m: how many constraint there are  

**Example of constraints**
  
- each param in the f function  
(bc each param is a constraint $\frac{d\chi^2}{dp}=0$)  

- a predetermined number of total occurrences  
$(\Sigma o_i)$

**DOF in table with count data ($o_i, e_i$)**

constraints:  
- predetermined number of occurrences $\Sigma o_i$  
- extimated parameter: $\mu$ if assume data follow poissonian, or $p$ if assume data follow binomial  

$DOF = num\_ rows -2$  

**DOF in dice**

have $o_i$ for each face  
a restraint is the total number of occurrences  
bc if know total throws and occurrences of each face 1-5, know occurrences of face 6, so it's not independent  

$DOF = 6-1 =5$  

### Hypothesis

**Null hypothesis**

$H_0$: null hypothesis (hypothesis that the phenomen do NOT happen)  
eg:  
the dice is NOT tricked (=occurrences follow a binomial distribution)  
the function is NOT wrong  
the data follow such distribution (is this still $H_0$)??  

intuitively, it's the hypothesis that simplifies the modeling of the problem  
bc must allow me to compute the probability of the outcome later  

**Why assume the null hypothesis?**

can simplify the problem, so can assume $p$ or $e_i$, the probabilities of something  

example: if assume that    
dice is not tricked: prob of each face appearing is $\frac{1}{6}$  
count data follow a specific distribution: can compute $e_i$  
value is sampled from a specific distribution: can compute $\mu$ and $\sigma$  

**How they're usually written**

all those about something, an outcome, being plausible/likely  
if outcome is a single value = is it sampled from such distribution?  
if outcome is an array of occurrences = does it follow such distribution?  

eg:  is dice fair? = are these occurrences likely? = are they sampled from a multinomial distribution?   

### P-value

**What it is**

it's the last stage of the chi-test  
quantifies the strength of the evidence against H0  
probability of observing a test statistic (getting a value of $\chi ^2$) equal or more extreme than the one obtained, assuming $H_0$ is true (so assuming the null hypothesis)   
$p(\chi^2  \ge    \chi^2 _0 \; | H_0)$  
is NOT prob of $H_0$, nor a ton of other things  

**What its value mean**

must pick a soglia, generally: $\alpha = 0.05$  

- $p\,value < α$: strong evidence against $H_0$, reject it  
so everything is wrong  

- $p\,value > α$: weak evidence against $H_0$, fail to reject it   
don't mean it's right, just mean i have nor reason to reject it yet, so probably it's right  


### Random

**Rule of 5 $\sigma$**

in physics, a model is wrong if an observation gives a result distant at least $5 \sigma$ from predicted value  

prob of getting value distance at least $5 \sigma$, even if the model is correct (so just being unlucky) is:  
$P(z\ge 5) = 2.87 \cdot 10^{-7}$, small enough

### Hypothesis testing

**What does it mean**

have an hypothesis, eg:  
the dice is fair  
the model accurately represents the data (the function and uncertainties used are correct)  

how to check if such hypothesis is correct?  

**Steps for hypothesis testing** 

1. assume the null hypothesis $H_0$  

2. compute the outcome (usually $\chi ^2$)  

3. find distribution of the outcome, so find its mean and variance $\mu \pm \sigma$  
if have $\chi ^2$, means find DOF  

3. 3 methods to check:  
    - p-test: check prob of an outcome more extreme happening (on table, or compute)  
    - study chi reduced  
    - check $\chi ^2$ is inside one standard dev from the expected value  

**What is the outcome?**

it's $\chi ^2$ if:  
- goodness-of-fit test   
- does count data follow such distribution? (eg. table with $o_i$ and $o_e$)  

it's a value if:  
- is this value $k$ sampled from such distribution (eg. dice, how many times k a single face appears in n throws)  

**Find distribution of outcome**

what is the distribution:  

- if my outcome is a chi: $\chi ^2$ distribution    
- if my outcome is a single value: probably binomial distribution   

expected outcome is $\mu \pm \sigma$  
$\mu$ and $\sigma$ must be computed using the formulas of the distribution of the outcome  
so, if outcome follows $\chi ^2$ distribution, it becomes  $\nu \pm \sqrt{2 \nu}$   

**p-test**

if $\chi ^2_0$ is the outcome, compute prob of getting a more extreme value (bigger or smaller)    
only one is enough, since it's either bigger or smaller    
$p(\chi ^2 \ge \chi ^2_0)$  

must find this probability on the tables for the $\chi^2$ distribution  

both prob of being bigger or smaller must be far from both 0 and 1   
let's say both between 0.1 and 0.9   
else it's suspicious    

**Study reduced chi**

have outcome $\chi ^2 _0$ and its DOF  
compute reduced chi: $\chi ^2 _\nu=  \frac{\chi_0^2}{\nu}$

if value of $\chi^2_\nu$ is:  
- $\sim 1$, the fit is ideal (residuals match noise level)  
- $\ll$ 1, the model might be overfitting or uncertainties are overestimated 
- $\gg 1$, the fit is bad or uncertainties are underestimated 

**Check distance from real and expected chi**

outcome: $\chi ^2 _0$  
distribution of outcome: $\chi ^2$, so expect $\nu \pm \sqrt{2 \nu}$  

check how many deviations the one i computed is distant from the mean  
if just 1 deviation, probably the hypothesis correct  

#### Example

**P-test**  

if got $\chi ^2 = 8.7$, with 13 DOF  
check in table $p(\chi^2_{13} \le 8.7)$  

**Reduced chi**

method 1:  
if got $\chi ^2 = 8.7$, with 13 DOF  
chi distribution has $\mu = \nu, \sigma = \sqrt{2 \nu}$  
reduced chi square is: $\chi ^2 _\mu = \frac{8.7}{13}$  
it's close enough to 1, ok  

method 2:  
since it's a chi distribution, expect $13 \pm \sqrt{26}$  
the one i got is within 1 deviation, so ok

types of exercises:  

- find prob that something happens

- is this outcome plausible/likely? (eg: is dice fair?)  
need hypothesis testing

#### Dice

**Is dice fair?**

2 types of exercises:  

- if have occurrences $o_i$ for each faces in $n$ throws:  

$H_0$: occurrences of all faces follow a multinomial distribution (gives p, allows to compute $e_i$)  
outcome: $\chi ^2$ (to check count data follows some distribution, computed with $o_i$ and $e_i$)  
distribution of the outcome: $\chi^2$, $\nu =6-1=5$  
-1 è perchè N, il totale di lanci, è un vincolo  

- if have occurrences $k$ for one face in $n$ throws:  

$H_0$: value $k$ is sampled from a binomial distribution (allows to compute $p$)  
outcome: $k$  
distribution of the outcome: binomial, with $\mu \pm \sigma$  

#### Table with count data

**Count data follow some distribution?**

like first version of dice exercise  

$H_0$: data follows some distribution X  
have $o_i$, compute $e_i$ (explained below)  
outcome: $\chi ^2$ to check if data follow a specific distribution  
distribution of the outcome: $\chi ^2$, with $\nu = i -1 -1$  
-1 viene da N, il totale di osservazioni, un vincolo  
-1 viene dal fatto che devo stimare p (nel caso del dado non l'ho stimato)

**What i have**

at first, only have cols $k$ and $o_k$  
$k$: index of number of events in subinterval  
$o_k$: how many times such number of events in subinterval was observed  

eg: $2,5$ means that there are 5 intervals in which happened exactly 2 events  

**Get data**

number of observations done (how many intervals/periods were checked):  
$N = \Sigma o_k$  
add all values in col $o_k$  

number of events verified:  
$n = \Sigma k \times o_k$   
add all values of col $o_k$, weighted by the k   
eg: if a row is $2,5$ must add $2 \times 5$, bc 10 events occurred  



**Extimate occurrences with a distribution**

distrib could be binomial or poissonian  
note that binomial could not always have sense, and possonian can only be used if n is big and p small  

probability for each k:  
  
- binomial:  
$ p= \frac{1}{N}$ (not sure, better poisson)  
makes sense intuitively anyway  
$B(k; n,p)$  

- poissonian:  
$\mu = \frac{n}{N}$  
$P(k; \mu)$  

expected occurences for each k (for each row):    
$e_k = n \times P(k; \mu)$  
$n$ is total number of events, each k gets a number proportional to its probability (can use either binomial or poissonian prob)
