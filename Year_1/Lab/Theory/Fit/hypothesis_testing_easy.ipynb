### Hypothesis testing

**Steps**

number | what | what to find 
---|---|---
1 | define the problem | $H_0$ (null hypothesis)
2 | reality: do the experiment | $O_i$(observed outcomes)   
3 | expectations: see expected resulst if $H_0$ is true | $E_i$ (expected values)  
_ | to see stuff well, could write $E_i$ and $O_i$ in a table  
4 | chi test: discrepance between reality and expectations (E and O) | $\chi^2$
5 | count the DOF | DOF
6 | p-value and conclusion: how likely is this discrepance | p-value   

reality = result of the experiment 

### 1. Hypothesis

**Nomenclature**

hypothesis: statement about the population or the probability distribution that generated the data  

in a statistical test, define 2 competing hypothesis:  

1- null hypothesis $H_0$: generally represents the standard situation  
in practice, it's the version where the probability distribution is known, and i can use it to compute the $E_i$  
means nothing weird happens, everything is as modeled  

2- alternative hypothesis: it says that the null hyporhesis is false  

**How to choose null hypothesis**

$H_0$: the data follows some sort of distribution     
$H_1$: the data DON'T follow that distribution

why:  
choose like this bc at step 3 will need to compute $E_i$ assuming $H_0$ to be true  
so pick as $H_0$ the one that allows me to compute the $E_i$, so the where can model the phenomenon with a distribution   

### 2. Experiment

**Steps**

1- choose what 1 experiment means  
eg. rolling the dice once  
eg. counting the people going in in 1 hour  

2- choose how many types to repeat the experiment  
call it N 

3- what are the possible outcomes $i$ of the experiment?  
so, RANGE of i  
if they are $\infty$, make a bigger bin to group all the ones going towards infinity (cannot have infinite outcomes, or $\chi^2$ will be an infinite sum, cannot compute it)  

4- how many occurrences $O_i$ has each outcome i?  
maybe put in it in a table

**3. Infinite outcomes**

if outcomes could potentially be $\infty$:  
after a certain i, regroup all the outcomes da lì in poi as one  

eg. how many poeple enter in 1 hour  
technically, there is not a max to the number of people who can enter, the outcome $i$ could be any number  
but after measuring, see that the outcomes from 1 onward don't happen many times  
so, regroup  them as outcomes $i = 5+$, where $O_{5+} = \Sigma_5^\infty O_i$   
and after, will have also $E_{5+} = \Sigma _5^\infty E_i$  

### 3. Expected values

**Steps**

assuming $H_0$ to be true:  

1- what distribution follows the data?    
find it in $H_0$  

2- what params this distribution has?    
MAYBE find it in $H_0$  
if don't find them, extimate them using the $O_i$ 

3- compute prob of each outcome $P_i(k_i, args)$    

4- compute the expected values $E_i = N \, P_i$  

**2. Missing params**

distributions need params (those passed in brackets) to compute the prob  
they might or not be specified by $H_0$  
if they aren't, must extimate the missing parameters  
each extimated parameter is a constraint (rem when i'll count the DOF)

to know what params i need, go see how disrtibutions are written, what i need to pass

**3,4. Compute**

have: $P(k; args)$    
the outcomes are k=1,2,3... 

for each outcome i:  

1- compute the prob $P(k_i; args)$ (just plug the value of i in k)  

2- compute expected value: $E_i = P(k_i) \, N$  
means: mutiply the total number of experiments performed $N = \Sigma O_i$  per the prob of that outcome happening $P(k_i)$   

### 4. Chi test

**Formula**
  
$\chi^2 = \Sigma \frac{(O_i-E_i)^2}{E_i}$  

**Meaning**

$\chi^2$ measures the discrepancy between reality ($O_i$) and expectations ($E_i$)  

measures how much different the observed counts are from the expected counts  
bc at numerator have difference between 
expectations $E_i$ and reality $O_i$  
the more different realit is from expectation, the bigger $\chi^2$  

rem, expectations are the things i expect to happen IF $H_0$ is true  

**Testing**

questions is: are $O_i$ and $E_i$ close enough that the differences could be explained by chance?  

3- chi-square goodness-of-fit test  
measures how different the observed counts are from the expected counts  
$\chi^2 = \Sigma \frac{(O_i-E_i)^2}{E_i}$  

4- count degrees of freedom

### 5. Degrees of freedom

**Definition**

DOF: the number of quantities that can vary independently, AFTER taking all constraints into account  

or, mathematically:  
how many independent addends are in the $\chi^2$ summation  
so, how many values of $O_i$ are independent, AFTER considering the constraints  

**Formula**

dof = k-m-1    

k: number of possible outcomes  
m: number of params extimated    

**Explain formula**

dof = number of outcomes - number of constraints  

imagine writing a system with alll the constrains: how many $O_i$ i can set before the others are obligated?    

constraint are equations, like:  
1- total number of tries: $N= \Sigma_i O_i$    
appears always, 1 time  
2- an extimated parameter: bc it's a formula using the $O_i$, so ties them together  
could appear, could have m parameters to extimate    

number of constraints  = 1 + m  

**Example: dice**

there are six faces  
but once i know from $O_1$ to $O_5$, so occurrences of first 5 faces, i know $O_6$, the occurrences of the sixth  
bc their sum must be N  

### 6. p-value

**Steps**

1- compute it  
2- compare it with the threshold  $\alpha$  

**1. Compute**

the p-value is the area in the right tail of the appropriate chi-square distribution  
$p=P(\chi^2_{dof}\ge \chi^2_{obtained})$   
$p = \int_{\chi^2_{obtained}}^\infty \, f_{\chi^2_{dof}}(x) \, dx$  

1- take a drawing of the distribution of $\chi^2_{dof}$  
note that the drawing changes if dof changes  

2- draw a vertical line at $x = \chi^2_{obtained}$  

3- measure the area under the distribution, on the right of such vertical line  
called the 'right tail' og the distribution   

in practice, the computer does it  
find a command for python  

**2. Threshold**  

technically, can pick what i want  
a common decision threshold is $\alpha = 0.05$   

1- $p< \alpha$: reject $H_0$,  
2- $p>\alpha$ : NOT reject $H_0$  
this doesn't prove that $H_0$ is true, just that the observed evidence is not strong enough to conclude otherwise  

means that i reject $H_0$ if:    
$p< \alpha$  
prob of observing a discrepancy at least this big bc of random variation <  5%    
so it's improbable that it was just random variation, probably $H_0$ was wrong  

**Why using it**

we have $\chi^2$, a way to measure discrepancy  
we know taht the bigger $\chi^2$ is, the bigger the discrepancy is  
but we have no reference for $\chi^2$, how big is too big? when it becomes suspicious?  
obv, how bad the discrepancy is depends on the type of problem, that we classigy using the DOF  
then, to answer our questions, we define the p-value: the prob of the experiment having more discrepancy if I where to do it again  


**Meaning**

the p-value tells me:  
assuming $H_0$ is true, what is the probability of obtaining a discrepancy (between reality and expectation) at least as large as the one that happened?  

in other words:  
assuming $H_0$ is true, what is the probability of obtaining a $\chi^2$ at least as large as the one i got?  
bc rem that $\chi^2$ measures the discrepancy  



**Def**

if get p=0.22, means:   

assuming $H_0$ is true  
there is a 22% probability of  
obtaining a $\chi^2$ (so, a discrepancy between result and expectation)  
at least as big as the one obtained here  
purely because of random variation  

$p= P(\chi^2_{dof} \ge \chi^2_{obtained} | H_0)$ 

**Explain threshold**

if get p=0.22, means:   
if repeat the experiment, 22% of the times the discrepancy between result and expectations is bigger than the one i got here     

1- if p is small:   
i almost never get a discrepancy as big or bigger than this  
so this discrepancy is huge, suspicious  
probably $H_0$ is wrong  

2- if p is big  
it's common to get a discrepancy as big or bigger than this  
so it's no big deal, relatively close to what i expected, could just be a small fluctuation    
probably $H_0$ is right  

**Other way to explain threshold**

p-value is the prob that if i did the experiment again, it would go worse (worse = more discrepancy, get $\chi^2$ bigger)  

1- if p is small, like 1%:   
it's unlikely to go worse (a discrepancy bigger than this)    
cioè: difficile fare peggio  
which is obv a critic, means this experiment went pretty shitty already  
it's suspicious, maybe $H_0$ was wrong  

2- if p is big, like 99%:  
it's likely that, if i do it again, i'll do worse  
so, this attempt was really good  
good means the discrepancy was small, tutto torna  
so, no reason to critic $H_0$, is probably right  


### Example: fair dice

**1. Null hypothesis**
  
$H_0$: dice follows multinomial distribution (the dice is fair)  
$H_1$: dice do NOT follow multinomial distribution (the dice is NOT fair) 

**2. Experiment**

1- experiment: roll the dice once  
2- how many experiments: N=60   
3- possible outcomes: from 0 to 6  
finite number of outcomes, so don't need to regroup  

**3. Extimated values**

1- which distr: multinomial    
$M(k_1, ... k_m; n, p_1, ... p_m) = \frac{n!}{k_1!...k_m!}p_1^{k_1}...p_m^{k_m}$  
$k_i$: how many times face i appears (so, $E_i$)   
$p_i$: prob of each occurrences  
$n$: how many times i repeated the experiment  

2- params of the distrib: $p_i = \frac 16$, $n=N$    
$H_0$ tells me all the params (by telling me it's a dice), don't need to extimate anything  

3- compute $E_i$ (so, $k_i$)  
$k_i =  N \, p_i $  
why, where get this formula???

| Face | Observed $O_i$ | Expected $E_i$  |
| ---: | -------------: | -------------: |
|    1 |              5 |             10 |
|    2 |              8 |             10 |
|    3 |              9 |             10 |
|    4 |             10 |             10 |
|    5 |             12 |             10 |
|    6 |             16 |             10 |


**4. Chi square**

$\chi^2 = \Sigma \frac{(O_i-E_i)^2}{E_i} = 7$  

**5. DOF**

k = 6 (outcomes are from 1 to 6)  
m = 0 (no params extimated, $H_0$ told me all of them)  

dof = k-1-m = 5

**6- p-value**

1- compute
$p = P(\chi^2_5 \ge 7) \sim 0.22$    

assuming $H_0$ true, the prob of obtaining a discrepancy at least as large as the one observed is about 22%   

2- compare  
$p > 0.05$  
do not reject $H_0$  

### Example: people in a shop

**1. Choose null hypothesis**

$H_0$: the number of people entering the shop in an hour follows Poisson's distrib  
$H_1$: it doesn't follow Poisson's distrib  

note: Poisson's distrib the number of events occurring in a fixed interval of time, space, area, or volume  
but it's continuous, not discrete    
so the number of people entering at exactly an istant t is 0  
so i need to count the number of people entering in an interval $\Delta t$  like an hour  

**2. Experiment**

2- how manye experiments: N=100   
3- possible outcomes: from 0 to $\infty$, technically

later, $E_{5+} = E_5+E_6+E_7$

**3. Expected values**

1- distrib followed: Poisson  
$P(k; \mu) = \frac{m^k}{k!} e^{-\mu}$  
k: how many events happened in a $\Delta t$  
$\mu$: BY AVERAGE, how many events happened in a $\Delta t$  

2- params  
i don't know $\mu$, must extimate it, using the sample mean   
$\hat \mu = \overline x = \frac{\Sigma i \, O_i}{\Sigma O_i} = 2.27$

3- expected values  
$E_i = P(k_i; \hat \mu) \, \Sigma O_i$

Assuming that the number of customers really follows some Poisson distribution, the probability of obtaining a discrepancy at least as large as the one observed is about 3.15%
