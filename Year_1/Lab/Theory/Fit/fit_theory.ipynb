### Weighted mean

**Weighted mean**

if have many values, each with a different uncertainty (maybe bc many people did the same experiment), to combine them:  
$\mu = \frac{\Sigma w_i x_i}{\Sigma w_i}$  

where the weights are inverse of the variance:  
$w_i = \frac{1}{\sigma _i ^2}$  

its uncertainty is $\sigma _y = \frac{1}{\Sigma \frac{1}{\sigma _i^2}}$, so it's smaller than all the $\sigma _i$  

so result is: $\mu \pm \sigma _y$  

note, if all x have same uncertainty, i get as mean the algebric not-weighted mean, obviously, and for uncertainty $\sigma = \frac{\sigma _i}{\sqrt{n}}$, which is exactly the error on the mean, computed from the error on the sample   

**How to prove these are the weights**

weighted mean it's kinda like a fit, so take the chi squared:  
$\chi ^2 = \Sigma (\frac{x_i-\mu}{\sigma_i})^2$  

must find $\mu$ that minimizes the $\chi ^2$:  
$\frac{d \chi ^2}{d \mu} = -2 \Sigma ( \frac{x_i-\mu}{\sigma_i}) =0$  

solving algebrically, get the formula of the weighted mean  

### Linear fit

if f is a linear function, of any degree  
can write a system putting to 0 the derivatives of the chi wrt all its params $\frac{d \chi ^2}{d \, a}= \frac{d \chi ^2}{d \, b}=... = 0$   
this system can always be solved it analytically  

**Assumptions**

1. all the $x_i$ measures are independent  
2. errors on x is trascurable (else, need harder version)  
3. $\sigma _y$ follow gaussian distribution  

**What if errors don't follow gaussian distribution?**

$\chi ^2$ is a sum of gaussian addends  
so if $\sigma _i$ are not gaussian, the addends aren't either, and so i don't get a $\chi ^2$, so don't know its mean and variance  

to extimate the params and the uncertainties on the params no problem even if errors are not gaussian, they're pretty close anyway  
but to judge goodness of fit is problem, bc mean and variance are not those it would be if it were a $\chi ^2$, and they might be really different  
only if DOF is big, thanks to th of central limit know that $\chi ^2$ is distributed as a gaussian, so confront it with mean and variance expected of a gaussian  

### Numerical methods for fitting (non-linear fit)

#### Newton method

what for: extimate roots of eq that can't solve analytically

what it means geometrically:  
start from a point $x_n$  
draw tangent to f in the coord $x_n$  
its intersection with the x axis will be my new $x_{n+1}$  

**How it works**

1. start from generic point $x_0$  

2. do Taylor approx until first order around $x_0$  
$f(x) \sim f(x_0) + \frac{df}{dx}|_{x_0}\, (x-x_0)$  

3. plug $f(x)=0$ (condition for x being a root) and get:  
$x \sim x_0 - \frac{f(x_0)}{\frac{df}{dx}|_{x_0}}$

4. now can do iteratively, putting $x_0=x_n \quad x = x_{n+1}$  

$x_{n+1} \sim x_n - \frac{f(x_n)}{\frac{df}{dx}|_{x_n}}$  

#### Non-linear fit

**How to find best-fit params**

unlike linear fit, cannot solve analytically the system obtained by putting derivatives to 0  
instead, start by an initial extimate (in curve_fit it's $p=[...]$)  
iteratively update values of best-fit params so to minimize the value of the loss function, here the $\chi ^2$  

**Problem of initial values**

linear fit has a unique minimum, that is therefore global    
which is why i can solve the system  
so don't care about initial params, might speed the problem, but eventually will always converge at global minimum, wherever i start  

non-linear fit has many local minimums  
so, depending on the initial extimates for the params, could converge in a local minimum, not reach the global  
so important to give initial extimates  

how to know if converged to the wrong minimum?  
if it does, the $\chi ^2$ will be way too big, so check that  
bc residuals are big, can see them from the residual plot  

#### Absolute sigma

**Absolute sigma**

it's a boolean argument of the function curve_fit 
by default, it's false   

true: $\sigma_i$ are correct in their absolute value    
false: $\sigma_i$ are not correct in thei absolute value, but only in their relative one (relative to each other)  

so before doing pcov, if gave =false, the $\sigma _i$ are scaled by a constant value (chosen so that the reduced chi square is exactly 1)

if uncertaities $\sigma _y$ are all the same (so given as a np.full array)  

passing $\sigma$ or not is the same (if absolute sigma = false), will get same values and uncertainties for the best fit params  

generally fine to leave absolute sigma = false  
if put = true MUST put errors, if i don't it will assume all errors are 1, which cause trouble if my values x and y are way smaller or bigger than 1  
will correctly find the best-fit params, but give them wrong uncertainty  

actually, to be rigorous must put absolute sigma = true, leaving it as false is for beginners  
but then must have extimated the uncertaities correctly  
maybe can check if with and without absolute sigma the best-fit params have similar uncertainty, they should if the uncertainties i extimated are correct and if the fit is good  


### Errors on x

if i assume it's trascurable but it's not, what happens:  

1. bets-fit params are wrong  
2. uncertainties on params are underextimated  
3. $\chi ^2$ is overextimated (and wrong, don't know what the mean and variance should be)  

**What to do if error on x are not trascurable** 

2 options: 

1. use ODR (orthogonal distance rejection), a different loss function    

2. propagate errors on $x$ and $y$ to get total error on denominator  
$σ^2_i = σ^2_{y_i} + (\frac{df}{dx})^2 \sigma _{x_i}^2$  
essentially it's a normal fit with gradient descent and more epochs  

both troublesome, just try to assume error on $x$ are trascurable  

#### Circular fit

want to extimate: $r, x_c, y_c$, radius and center  

as loss function, instead of the $\chi ^2$, use:  
$\Sigma (d_i-r)^2$  
$d_i = \sqrt{(x_i-x_c)^2+(y_i-y_c)^2}$  

**How to solve**

2 methods:  

1. numerically  
do gradient descent on $\Sigma (d_i-r)^2$  
 
2. analytically: Kasa method  
pick as loss function: $\Sigma (d_i^2-r^2)^2$, this i can solve    

**Uncertainty on params in Kasa method**

ONLY if:  
- points used to fit the circle are more or less equidistanti  
- error on x and y are the same  
(always in image, bc error is 1 pixel for both)   

can say:  
$\sigma _{x_c} = \sigma_{y_c} = \sigma \frac{2}{n}$  
$\sigma _r = \sigma \frac{1}{n}$  

where n is the number of points  

if didn't use Kasa method but solved numerically, can find uncertainties from there, no need to have formula for that
