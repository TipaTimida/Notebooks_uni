## Random generators

**True random vs pseudo-random number generators**

- true (TRNG): use natural phenomena intrinsically casual (atmospheric noise, radioactive decay)    
so monitor them and, when asked a number, give their current value  

- pseudo (PRNG): use algorithm taken from math formulas or number tables (and modular arithmetic) to produce predetermined sequence 

differences: why is pseudo better?  
- sequence is reproducibile if use same seed  
- tecnhically it's periodic, after a while sequence restarts, but it's so long it don't matter  
- more efficient (in a time interval can output more numbers than a true one)  

**Why is pseudo-random periodic?**

formula for pseudo-random is:  
$x_n = f(x_{n-1}) \, \mod N$  

it's a succession by recursion, so if obtain a number $x$ i had already gotten, will start repeating some sequence as before  
uses modular arithmetic (bc i don't want to receive numbers too big), so if use only integers, eventually will run out of numbers and will reach one i already reached  

**Discrete logarithm**

ordinary log: $p^k = g$  
discrete log: $p^k (mod \: m) = g$  

in discrete, if know p,m,g, there is no way to compute k other than bruteforcing (bc k is too irregular)    
used in criptography  

### Pseudo-casual methods

**Middle square method**

created by John Neumann  

start from a seed m, of n digits  
at each step compute $m^2$ and pick the n central digits, these will be the next number  

rating:  
today not used, bc risk having short periodical sequences

## Monte carlo

**Def: Monte Carlo methods**

are a broad class of computational algorithms that rely on repeated random sampling to obtain numerical results  
The underlying concept is to use randomness to solve problems that might be deterministic in principle  

The heart of Monte Carlo lies in the generation of random numbers  
Modern computer algorithms produce pseudo-random numbers, forming the basis for simulation experiments
