#### Numbers and colors

**Integer numbers: types**

sign/unsigned: if can be negative or not   
in 8-bit representation:  
Unsigned: range from 0 to 255  
Signed: range from -128 to 127 (using two’s complement)    

Fixed-Size Integers: Stored in memory with a fixed number of bits (e.g., int32, int64)  

**RGB (red, green, blue)**

addictive color model  
3 components, each from 0 to 255 (no intensity, full intensity (bright))  
combination of red, green, blue  
used in digital screens and web design  

eg:   
RGB(255, 0, 0) → Red

**Hex**

six-digit hex representation of RGB values  
used in web design and CSS

each component, rather than in decimal, is written in hex  

eg:  
#FF0000 → Red  
stands for (FF = 255, 00 = 0, 00 = 0)

Short-hand HEX: If each pair has repeating values, it can be shortened (#FFAA00 → #FA0)  

**HSL (hue, saturation, lightness)**

hue: the color type (0-360° on a color wheel)    
saturation: intensity of the color (0%-100%)  
lightness: brightness level (0%-100%)  

HSL(0, 100%, 50%) → Red  

allow easy adjustements of brightness and saturation  

**HSV (hue, saturation, value)**  

change last component  
lightness balances white and black, making desaturated colors appear more gray  
value represents brightness, maintaining vibrancy even when reducing saturation

**CMYK (Cyan, Magenta, Yellow, Black)**

subtractive color model, used for printing  
subtracts light by absorbing wavelengths

**Addictive vs subtractive color model**

addictive: used for digital screens (light-based)  
a digital screen works by turning on different amounts of red, green, and blue pixels  
primary colors are RGB  
all at max intensity give white light, none give black light  

subtractive: used for printing (ink/pigment-based)  
white paper reflects all light, adding ink absorbs (subtract) certain wavelengths  
primary colors CMY  
ink prevents some colouts from being absorbed, so they're refracted, reach our eyes, and that's the color we see  
all together give black, bc all light is absorbed  
in printers, black is given by a specific ink, not by mizing all others, bc:  
black ink it's less expensive  
black ink give a deeper black

### Represent numbers in mobile comma

**Floating-point arithmetic**

IEEE 754 Standard: defines how floating-point numbers are stored and processed

types:  
32 bit: single precision (about 7 decimal places)  
64 bit: double precision (about 15 decimal places)  

means mantissa is long 7 or 15 decimals, so after that it can't store smaller stuff  

most significant bit: leftmost bit  
like most significant digit in significant digits  

$x = (−1)^s × m × 2^{e−b}$  

- s: sign  
1 bit  
0 positive, 1 negative

- e: exponent  
has 8 or 11 bits (precision single or double)  

- b: bias  
used to represent number <1 without needing to add sign to exponent  
it's 127 or 1023 (precicision s or d)

- m: mantissa  
$1 ⩽ m < 2$, 23 or 52 bits   
sequence of digits AFTER the comma (bc digit before comma must be 1, like in scientific notation, so it's not written)  

to represent a number use  string:  
$0b [s][e-b][m]$  
instead of each brackets put all the bits (used or not) assigned to that letter

### Floating point: weird errors

```python
x = 0.1 + 0.1
print(f'{x:.30f}')
print(x == 0.2)

x = 0.1 + 0.2
print(f'{x:.30f}')
print(x == 0.3)
```

**Explanation**

Computers represent decimal numbers in binary  
using IEEE 754 double-precision floating-point format  

Some decimal numbers cannot be exactly represented in binary, like 0.2 or 0.3  
bc the mantissa has an infinite binary representation 

$0.200000000000000011102230246252$  
$0.200000000000000011102230246252$  


$0.300000000000000044408920985006 $ (Stored x)  
$0.299999999999999988897769753748 $ (Stored 0.3)

```python
# e means 10^

x1 = 1.0e25
x2 = 1.0

print(x1 + x2 - x1)
print(x1 - x1 + x2)
```

**Explanation**

$x_1=10^{25}$  
$x_2 = 1$  
in first case, since $x1 >> x2$, <u>the mantissa (sequence of digits after the comma) is not long enough to hold the $1$</u>      
so it's lost  
essentially $x1 +x2 = 10^{25}$  
so, after, deleting x1, it doesn't remember that 1    

**Note mantissa**

the mantissa has 52-bit precision (it's long 52 bits)  
but this is expressed in binary  
thinking itin decimal, it's 
$2^{52} \sim 10^{15.9}$  

meaning, when thinking about a decimal number, the mantissa would be ling 15 digits  

after that, can't represent smaller values  
which is the reason why $10^{25}$ cannot hold the 1, the mantissa is not long enough
