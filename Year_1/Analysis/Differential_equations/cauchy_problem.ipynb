### Cauchy problem

**What is it**  

i'm given  
- a differential equation of order n  

- n initial conditions (for all the derivatives  
from y to $y^{(n-1)}$, computed on the SAME point $x_0$)  
so like: $y(x_0)=y_0, y'(x_0)=y_0, y''(x_0)=y_0$  

must find a solution  

**Steps**

1. find homo sol to diff eq  
2. if non-homo of second order, find particular sol  
3. plug in all initial conditions, see if find a solution  

finding a solution means finding values for the vars for which all the conditions are true at once  
they're A,B if second order, +c if first order

#### Cauchy-Lipschitz property

rem $f=y'$  

**Cauchy-Lipschitz property**

conditions:  

1. f continuous  
in a rectangular interval containing $(x_0, y_0)$   

2. f locally Lipschitzian wrt y  
a sufficient condition for this is: $\frac{\partial f}{\partial y} |_{y=y_0}$ continuous  
(so must derivate, plug in $y_0$ and check it don't cause a C.E.) 

an easier way to express both of these conditions into just one is:   
f is of class $C^1$   
easy to check since my diff eq is $y'=...$  

**What each condition says**

these properties are SUFFICIENT for:   

1: existance  
2: uniqueness  
1+2: existance + uniqueness  

the conditions in the Cauchy-Lipschitz property are SUFFICIENT, NOT NECESSARY  
even if they're not satisfied there might be a solution and it might be unique  

unicity means that trajectories of two solutions will not meet  
NOT that there is only possible trajectory  
in fact it's exactly the opposite, in cauchy-lipschitz property sols fibrano lo spazio, so in each point passes a trajectory  

**Lipschitzian wrt y**

means:  

$|f(x,y_1) - f(x,y_2)| \le L|y_1-y_2| \quad \forall x,y_1>y_2$, with $L>0$
