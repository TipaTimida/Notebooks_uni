## Biologic growth model

types:  

- Malthus: just 1 population, constraint of resources  
- Volterra-Lotka: 2 populations that interact  

### Malthus model

**Growth rate of population**

characteristic of Maltus model: growth rate is proportional to current population  

if infinite resources:  
$x' = Ax$  

if finite resources, growth rate depends on current population too:  
$x' = (A-Bx)x = Ax - Bx^2$  

**Logistic growth rate**

logistic growth: growth rate depends on current population  
so in rx, r is a func of x too  

$x' = r(1- \frac{x}{K})x$  

r: intrinsic growth rate (growth rate in absence of any limiting factor)  
K: saturation level or carrying capacity  

population, if >0, will grow or diminish until it's equal to K, then will stay equal  

in above example: $x' = A(1- \frac{B}{A}x)x$  
so $K= \frac{A}{B}$ is a stable sol  


**Solutions**

infinite resources:  
$x = x_0 e^{At}$  
x tends to $\infty$ if A>1, to 0 if A<1  

finite resources:  
do var change to solve:  
$\begin{cases}  
y= \frac{B}{A}x \\
y' = A(1-y)y
\end{cases}$  
$y(t)= \frac{1}{1+Ce^{-At}}$     
$x(t) = \frac{A}{B} \frac{e^{Ac}}{e^{A(t+c)}\pm 1}$  


y(t) is called logistic equation, which is a sigmoid    
note that $x=0$ and $x = \frac{A}{B}$ are stabile sols  

### Volterra-Lotka model

![alt text](https://upload.wikimedia.org/wikipedia/commons/thumb/e/e2/Lotka-Volterra.svg/310px-Lotka-Volterra.svg.png)

$x(t)$: prey  
$y(t)$: predator  

$\begin{cases}  
x' = Ax - Bxy = x(A-By)\\
y' = Cxy - Dy = y(Cx-D)
\end{cases}$  


there is correlation of second order: $bxy$, bc this term has order 2  


**What components mean**


- Ax: growth rate of prey, Malthus increase  
- -Bxy: killing of preys by predator (the more predator the more killings, the more preys the more prob a predator finds one)  
- Cxy: growth rate of predators, depends on amount of preys which are their food  
- -Dv: mortality rate of predators  

**Stationary sols**

- P $(\frac{D}{C}, \frac{A}{B})$: non-banal equilibrium between the two species, the point in the graph at then center of the concetrical lines    

- O (0,0): banal equilibrium  

if y=0, preys grow infinitely  
if x=0, predators die bc have no food  

**Linearize system**

linearize means delete higher infinitesimals, so stuff with xy  

- linearize next to O, banal equilibrium point  

$\begin{cases}  
x' = Ax \\
y' = - Dy 
\end{cases}$  

$\begin{bmatrix}A & 0 \\ 0 & -C \end{bmatrix}$  
gives a saddle (intuitively, preys tends to grow, predators to shrink)   

- linearize next to P, non-banal equilibrium point:  

$\begin{cases}  
x' = - Bx (y- \frac{A}{B})\\
y' = Dy(x- \frac{C}{D})
\end{cases}$  

do var change: $X = \frac{C}{D}x, Y= \frac{A}{B}y$ ???  


$\begin{bmatrix} 0 & -\frac{BC}{D} \\ \frac{DA}{B} & 0 \end{bmatrix}$  

sols are $\lambda = \pm \sqrt{AC}i$, pure imaginary, no real part  
so it's a center, can't apply linearization theorem  



**Hamiltonian**

$H(x,y) = By+Cx-A \ln y - D \ln x$  
H is constant in all curves  
thus trajectories are limited, bc if x or y go to infiniry, H would go to infinity too, so would not be constant  

**Other form**  

$\begin{cases}  
x' = Ax - Bxy -Hx^2\\
y' = Cxy - Dy 
\end{cases}$  

stationary sols are:  
O: (0,0)  
P: $(\frac{A}{H}, 0)$    
Q: $(\frac{C}{D}, \frac{A}{B}-\frac{HC}{DB})$  

O: saddle, like before  
Q: saddle  
P: stable spiral  

## Equilibrium points

**What they are**  

points where: $x'=0$  
bc population don't change anymore  
so like a stationary solution  
or like in exercise of finding where recursion converges, points where the population converges  

**Types of stability**

- unstable: points near them (on both sides) get further away   
means origin is an unstable solution point, if little change it gets further  

- asymptotically stable: points near them (on both sides) go towards them  
all trajectories eventually fall into it (spiral)  

- stable: trajectories move around the equilibrium point, but never towards it (center)  

- semi-stable: on one side points go towards, on other get further  


if eigenvalue is negative, its eigenvector points towards the origin: stable    
if it's positive, eigenvector points outwards, away from origin: unstable  

**Types of phase planes depending on eigenvalues**  

$x' = Ax$  
depending on form of A, thus on $\lambda, \mu$ eigenvalues  

A diagonalizable in R (eigenvalues are different):  
- same sign: proper node / star (stable if +, unstable if -)
- different sign: saddle point / sella (unstable)

A not diagonalizable (eigenvalues are coincident, have multiplicity 2):  
- improper node (stable or unstable, depends on sign)  

A diagonalizable in C (eigenvalues are complex conjugates):    
- real part non-null: spiral / fuoco (stable if real part is +, unstable if -)  
- real part null: center / centro (stable)  
compute derivative in random point to check directionl clockwise or anticlockwise  

## Solving linear systems  

**Coupled linear system**

when $x_1, x_2$ appear in both diff eq:  

$\begin{cases}
x_1' = ax_1+bx_2 \\
x_2' = cx_1 + dx_2 
\end{cases}$


can be written in matrix form:  
$\vec{x}' = A\vec{x} + \vec{g}(t)$  

the g part make the system not linear

**Turn 2nd order diff eq in system**

1. define new functions:  
$x_1 (t) = y(t)$  
$x_2 (t) = y'(t)$  
if have more, keep defininf $x_2(t)=y''(t)$, depends on order  

2. differentiate:  
$x_1'(t) = y'(t) = x_2(t)$  
$x_2'(t) = y''(t)$  
the last is the full diff eq  
the others are mapped to the next function  

3. rewrite the initial conditions with the new vars:    
$y'(0)=a \rightarrow x_2(0)=a$  

**Solve system**

$x' = Ax$  

solution is:  
$\vec{x}(t) = \vec{\mu} e^{rt}$  

where $\mu, r$ are an eigenvector and an eigenvalue of A  

if eigenvalues are complex, turn the exponential is sum of sin and cos, using polar form  


**Ways to write sols**

normal sol of differential eq, but with vectors:  
$\vec{x}(t)= 2 \begin{bmatrix} 1 \\ -1 \end{bmatrix}e^{\lambda _1} \quad -3  \begin{bmatrix} 2 \\ 4 \end{bmatrix}e^{\lambda _2}$  

written in matrix form:  
$\vec{x}(t) = \begin{bmatrix}  
2 e^{\lambda 1} -6 e^{\lambda _2} \\
-2 e^{\lambda 1} -12 e^{\lambda _2} \\
\end{bmatrix}$  

written in linear system:  
$\begin{cases}
x_1(t) = 2 e^{\lambda _1} -6 e^{\lambda _2} \\
x_2(t) = 2 e^{\lambda _1} -12 e^{\lambda _2}
\end{cases}$




**If have repeated eigenvalues**

solution is:  

$\vec{x} = c_1 \mu e^{\lambda t} + c_2 [\mu t e^{\lambda t} + \rho e^{\lambda t}]$  

where:  
$\mu$: eigenvector   
$\rho$: vector mapped to eigenvector, $A \rho =  \mu$ (generalized eigenvector)  

so sketch only 1 eigenvector in plane phase  
if eigenvelue is positive, moves away from origin  
gives a node

#### Spring

**Problem text**

block with spring back and forth on plane, -bv is damping  
$ma = -kx -bv $

rewrite with derivative with respect to time:  
$v = \dot{x} =  v \\
a = \dot{v} = -\frac{kx}{m} -\frac{bv}{m}
$

in matrix form:
$
\begin{pmatrix} \dot{x} \\ \dot{v} \end{pmatrix} =
\begin{pmatrix} 0 & 1 \\ -\frac{k}{m} & -\frac{b}{m} \end{pmatrix}
\begin{pmatrix} x \\ v \end{pmatrix}
$

let's examine the case b=0  

compute eigenvalue:  
$\lambda ^2 + \frac{k}{m} = 0 \\
\lambda = \sqrt{-\frac{k}{m}}
$

eigenvalues is complex: there are no vectors only sheared and not rotated

therefore, eigenvectors and eigenvalues show long-term behaviour
