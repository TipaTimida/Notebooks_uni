### Definitions

**Ways to write them**

- implicit form:  $F(x, y, y', y'' ...) = 0$  
put all on one side and equal to zero  

- normal form: $y^{(n)} = f(x,y,...y^{n-1})$  
explicitate the derivative of max order  

note that y(x), x is the independent var  
the incognita is y(x), $ y \in C^n$  

called functional, bc incognita is a func, not a value

**ODE vs ODE**

differential eq: there are derivatives  

ODE: ordinary, no partial derivatives  
PDE: partial, yes partial derivatives  

**Mermoryless vs functional**  

MEMORYLESS (senza ritardo /senza memoria)   
func and derivatives y(x), y'(x) ... are all computed in same point x  
so can write y(x) = y

FUNCTIONAL   
not memoryless (not done here)  
musts specify in which point each y is computed  


**Types**  

must be only option of each kind:  
- memoryless / functional  
- ODE / PDE  

**Characteristics**

- order: highest derivative  
if at most have y' order=1  

- degree: power of highest derivative

**Types**

- autonomous: x by itself don't appear, only from y in poi  

- linear: vars (x, y, y'...) have no power, products, exponents...

- homogeneous: no constant term, g(x)=0  

- linear with constant coefficients: if coeffs of vars don't contain vars themselves

**Autonomous eq**

autonomous eq: in F the independent var x itself never appears, only appear in dx of dy/dx     

means if find solution $y(x), y(x-x_0)$ is a solution as well (it's a temporal translation)  
eg, in eq of pendulum t itself don't appear, bc motion don't depend on the hour where i do the experiment  

### Contraction theorem (Teorema delle contrazioni), Banach-Caccioppoli th

(X,d): complete matric space (d is the distance defined in the space)  


hp:  

- $f: X \rightarrow X$ is a contraction  

th:  

- $\exists !$ fixed poinf of f  
fixed point is one where $f(x)=x$  
means a contraction must have one and only one point of convergence (whatever the starting value of the recurrent succession)  
useful if f is thought of as a recurrent succession

**Contraction**

$|f (x) − f (y)| ≤ L  |x − y|$  
if $0<L<1$, f is a contraction  

written using metric spaces:  
$f: X \rightarrow X$  
means f is an endomorphism on X, which is a metric space (X,d)  
f is a contraction is:  
$d(f(x),f(y)) \le C d(x,y)$  
with $C < 1$  

essentially jsut replace abs value with distance  

### Other

**Associated poly**

consider omogeneous equation: $\Sigma a_k \; y^{(k)} (x) = 0$  

want solution of the form $y(x)=e^{\lambda x} \quad \lambda \in C$   
which means, its derivatives are: $y^{(k)}(x) = \lambda ^k e^{\lambda x}$  

so, when plugging the solution in, get:  
$\Sigma a_k \; \lambda ^k e^{\lambda x} = p(\lambda) e^{\lambda x} = 0$    

the associated poly is:  $p(\lambda) = \Sigma a_k \; \lambda ^k$  

like in linear algebra, associated poly has brackets with roots, each with its own multiplicity
