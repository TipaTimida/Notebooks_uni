## Theory

### What the things are

**What the things are**

$f(x)$: function   
$f: X \rightarrow Y$  
$f: x \mapsto f(x)$  

$a_n$: succession, sequence  
$a:N→X$  
$a: n \mapsto a_n$

$f_n(x)$: sequence of functions  
$f_n​:N→F(X,Y)$  


**In words**

function: 
input: dominium is R, can receive all values  
output: a value  

sequence: 
input: dominium is N. can receive in input only integers (the integer given is the index of the value i want)    
output: a value  
so sequence is like a discrete function  

sequence of functions:  
input: dominium is R, can receive any value (input is n, not x)  
output: a function  
its pointwise limit, written $f(x)$, is the function it becomes when $n \rightarrow \infty$  

### Properties

**Concept of continuity**

function: there is can, be either continuous or uniformly continuous  

succession: there isn't, since it's discrete (dominium is N)  

sequence of functions: if plug in a specific $n_i$ can ask if that $f_{n_i}(x)$ is continuous   
or can check if the pointwise limit $f(x)$ is continuous  
but can't study continuity of the sequence itself $f_n(x)$, has no sense  


**Limit existance vs convergence**

existance of limit means limit is FINITE  

existance of limit and convergence are intuitively the same thing   
generally call it: 
- existance of limit: function 
- convergence: sequence, sequence of functions (in sequence of functions means pointwise convergence)

uniform convergence is a concept that only exist for sequence of functions  

**What to do to check limit existance / convergence / pointwise convergence**

function: $x \rightarrow x_0$  
can pick point where to check it  
follow graph of a func until point i picked and see where i goes  

sequence: $n \rightarrow \infty$  
only makes sense at infinity  
same thing as function, follow graph until infinity and see where it goes  

sequence of functions: $n \rightarrow \infty$, $x \; const$  
means fix the number $x$ (consider it a const), make  $n \rightarrow \infty$, it converges if limit is finite   
increasing n, skip between all different functions, see what they end up resembling  
which $x$ to fix it as must be told by the text   

**Continuity**

types of continuity:  
pointwise continuity: local property  
continuity: global property, means local property is true everywhere  
uniform continuity: global property, other conditions on top of continuity  

in both continuity and uniform continuity, must specify an interval  

**Continuity vs uniform continuity**

continuity of f: has no specific definition (nor with epsilon, nor with sequences)  
just need to use def of continuity in a point, and check it's true in all points  
so generally not done with epsilon proof  

uniform continuity: has definition (either epsilon or sequence)  

## Writing proofs

### Notation

**Stuff written the same**

make the limit $n \sim \infty$ equal to 0 can be written:  
$\forall \epsilon \; \exist N: \; \forall n>N \quad ...<\epsilon $  
$\lim _{n \rightarrow \infty} \quad ... = 0$  
$ ... \rightarrow 0 $    

so a same proof (convergence of sequence) can look like:  
$\forall \epsilon \; \exist N: \; \forall n>N \quad |a_n -l | <\epsilon$  
$\lim _{n \rightarrow \infty} |a_n -l | = 0 $   
$ |a_n -l | \rightarrow 0$  

thus these are the same:  
$\lim f(x) = A$  
$a_n \rightarrow A$  

generally $\lim$ is used with functions, $\rightarrow$ with sequences  
bc in functions can pick limit in any point, so must specify the point under $\lim$  
bc in sequences can only take limit at infinite, so no need to specify, so $\lim$ is useless, the arrow is faster  

### Find upper/lower bounds

**What values to plug in**

can either plug:  

- the same value in all x: more stringent, but first must find maximum of $f_n(x)$ by studying $\frac{df}{dx}$  
only do if a previous exercise made me find such maximum point  

- a different value in every x: more lax, but faster and sometimes it's enough  
generally, at numerator and denomiantor put extremes of interval  

**When to plug in: for integrals**

- if bounds of interval are funcs of n: plug them in all x,      

- if bounds have $\infty$: plug into evertything i can, except those where i would want to plug $\infty$  
leave those and integrate  

- if inside integral have trig func: don't plug anything in there  
plug in all others, then say $\sin(nx) \le 1$ to get upper bound or integrate it  

or in general, plug in those x that, after not being there anymore, allow me to integrate normally  


**Attention doing the sup**

to do it well, should plug in every x the same value, the one for which f is maximal  
if plug in every x a different value, to make it the biggest, i'm doing an upper bound $\le$  
so if the upper bound converge my func converge too, but if it diverges then don't know what mine does, since it's smaller  

**Tips**

the fastest method is to plug something in EVERY x that appears  
in this case, integrating means multying per the lengh of the interval, $(b-a)$, since would get a constant function  

if $f_n(x)$ is always positive, to find lower bound can reduce the bounds until something where it's easier to plug in ($\int_0^\infty \ge \int_0^1$)m so cannot plug in $\infty$  

**Example: plug everywhere**

$\Sigma \int _n^{2n} \frac{\ln (x)}{(1+x^2)^\alpha} dx$  

find bounds (bounds are funcs of n, so plug everywhere):  
upper bound:  $a_n \le n \frac{\ln (2n)}{(1+n^2)^\alpha} \sim \frac{1}{n^{2\alpha}}$  
lower bound:  $a_n \ge  n \frac{\ln (n)}{(1+(2n)^2)^\alpha} \sim \frac{1}{(2n)^{2\alpha}}$  

upper converges for $a >1$, lower diverges for $a<1$, so the original integral does the same  
now plug in $a=1$ and check  

**Example: plug only some and tip 2**

delete just some terms, not all  

$a_n = \int _0^\infty \frac{e^{-x}}{n+x} \; dx$  

find bounds:  
bounds are $0$ and $\infty$, so plug only in some  
if there weren't the denominator i could integrate, so get rid of that  

upper: $a_n \le \int^\infty_0 \frac{e^{-x}}{n} \; dx = \frac{1}{n} \int_0^\infty e^{-x} \; dx = \frac{1}{n}$  
lower: $a_n = \int _0^\infty \frac{e^{-x}}{n+x} \; dx \ge \int _0^1 \frac{e^{-x}}{n+x} \; dx = \frac{1}{n+1}\int_0^1 e^{-x} \;dx = \frac{1-e^{-x}}{n+1} = \frac{C}{n+1}$

### Modulo

**Operations with modulo (to extimate)**

rem, in proof i want to create upper bound  

if know $x>0$, $|x|=x$  
so whenever i can, delete the abs value  
maybe delete it at denominator, only lave it at numerator  

$|xy| = |x||y|$

$|x+y|≤∣x∣+∣y∣$  (triangle inequality)  
(now, if know interval, could plug biggest x and y and get upper bound)

corollary of triangle inequality:  
$||x|-|y|| \le |x-y|$  

## Epsilon proofs 

### Conditions

**Steps**

1. write $\forall \epsilon >0$  

2. write the condition i want to prove

3. extimate the thing that must be smaller than $\epsilon$  

4. pick N or $\delta$ accordingly, they're a function of $\epsilon$  

5. now, if i want, can plug $N$ or $\delta$ in to show it holds 

**Conditions for limit existance / convergence**

what | formula | notes 
---|---|---
sequence,  converge to known limit $l$ |  $\exist N: \quad \forall n\ge N \quad \|a_n -l\| <\epsilon$  | 
sequence, converge to unspecified limit | $\exist N: \quad \forall m,n \ge N \quad \|aₙ − aₘ\| < ε$ | (it's the Cauchy  property, bc in R cauchy = convergent)           
function, has known limit $l$ in $x=c$ |$\exist \delta >0 : \quad \|x-c\| < \delta \rightarrow \|f(x)-l\| < \epsilon$ | (unlike in sequence, must specify $c$)
function, has unspecified limit in $c$ |  $\exist \delta >0: \quad 0 < \| x−c \| <δ, \; \; 0 < \| y−c \| <δ \rightarrow \| f(x)−f(y) \| <ε$ | Cauchy condition
sequence of functions, pointwise convergence | $\exist N: \quad \forall n\ge N \quad \| f_n(x)-f(x) \| < \epsilon$ | 
sequence of functions, uniform convergence |$\exist N: \quad  \forall n\ge N \; \; \forall x \in I, \quad \|f_n(x)-f(x)\|< \epsilon$  

remember $\forall \epsilon >0$ at the start

**Conditions for continuity**


what | formula   
---|---
continuity in point $c$ | $\exist \delta >0: \quad \forall x \quad  \|x − c\| < δ ⇒ \|f(x) − f(c)\| < ε$  
uniform continuity | $\exist \delta >0: \quad \forall x,y \quad \|x − y\| < δ ⇒ \|f(x) − f(y)\| < ε$  

### Steps explained

**Tips to remember formulas**


convergence/limit existance to UNSPECIFIED limit: use Cauchy condition  
substitute the limit with another point of the function/sequence (rem must iterate on that too in $\forall$)  

sequence vs functions :   
to pass from sequence to function, specify the point $c$ and swap $a_n$ with $f(x)$  
function: $\epsilon - \delta$ condition   
sequence: $\epsilon - N$ condition   
bc $N$ or $\delta$ is the error on x axis  
so if dominium is R, error must be ble to be any number (float)  
if dominium is N, error can be an integer 

limit existance vs continuity:  
the difference is that a limit can be anything, $l$, while for continuity it must be the image of the function, $f(c)$  
it's more obv if look at conditions using series  

**Extimate thing smaller than epsilon**

if have formula of the thing $(f, a_n)$, often:    
- find upper bound (maybe with taylor, or by plugging in each var the a different value to make it the biggest possible)   
- use triangular inequality   

if don't know formula (weird exercise with words):  
use known properties of f (e.g. Lipschitz, derivative bounded, etc.) or compactness  

if want $\delta$: must leave $|x-y|$, or whatever must be $< \delta$    
if want N: don't leave anything  

**Pick N or delta**

just finished the extimation, put the final step obtained (the upper bound) $< \epsilon$  

modify to get:  
if want $\delta$: $\quad |x-y|<g(\epsilon)$    
if want $N$: $\quad n > g(\epsilon)$  

now pick:  
$\delta = g(\epsilon)$  
$N = ceil(g(\epsilon)) \quad$  or $\quad N = g(\epsilon) +1$  

### Examples

#### Have formula of function/sequence

**Limit of a sequence**

prove:  $\quad$ $\lim_{n \rightarrow \infty} \frac{1}{n} = 0$  

1. what i want to prove: $\quad$ $\exist N : \quad \forall n\ge N, \quad |\frac{1}{n} -0| <\epsilon$  

2. extimate the thing:  $\quad$ $|\frac{1}{n} -0| = \frac{1}{n} < \epsilon \quad n> \frac{1}{\epsilon}$   

3. pick N:  $\quad$ $N = ceil(\frac{1}{\epsilon})$  

**Limit of sequence to unspecified point**

prove $a_n = \frac{1}{n}$ is convergent  

1. what i want: $\quad$ $\exist N : \quad \forall  m, n ≥ N  \quad |aₙ − aₘ| < ε.  $  

2. extimate: $\quad$ $∣a_n​−a_m​∣= |\frac{1}{n}-\frac{1}{m}|​=\frac{1}{m}-\frac{1}{n} < \frac{1}{m} < \frac{1}{N} < \epsilon$  
assume n>m  

3. pick N: $\quad$ $N = ceil (\frac{1}{\epsilon})$   

**Continuity in a point**

prove:  $\quad$ $f(x)=3x+2$ is continuous at $x=1$  

1. what i want: $\quad$ $\exist δ > 0: \quad |x − 1| < δ ⇒ |f(x) − f(1)| < ε$  

2. extimate it:  $\quad$ $∣f(x)−f(1)∣ = 3|x-1|< \epsilon$  

3. pick $\delta$:  $\quad$ $\delta = \frac{\epsilon}{3}$ 

**Uniform continuity**

prove:  $\quad$ $f(x)=x^2$ is uniformly continuous on [0,1]  
intuitively it is, bc continuous + compact space (closed interval) = uniformly continuous   

1. what i want: $\quad$ $\exist δ > 0: \quad \forall x,y ∈ [0, 1] \quad  |x − y| < δ ⇒ |x² − y²| < ε $  

2. extimate:  $\quad$ $∣x^2−y^2∣=∣x−y∣∣x+y∣≤∣x−y∣(|x|+|y|)\le 2|x-y| < \epsilon$  

3. pick $\delta$: $\quad$ $\delta = \frac{\epsilon}{2}$  


**Uniform convergence**

$f_n(t) = \frac{n \sin \frac{t}{n}}{t+t^2}$ $\quad$ $I=[0,b]$

1. what i want  

$\exist N : \quad \forall n\ge N \; \forall x \in I \quad ∣f_n(x)−f(x)|<\epsilon$  

2. extimate  

$f(x)= \frac{1}{1+x}$  

$|f_n(x)-f(x)| \le |\frac{1}{1+x} (\frac{\sin(x/n)}{x/n}-1)| \le \frac{1}{1+x} |(\frac{\sin(x/n)}{x/n}-1)| \le \frac{1}{1+x}  \frac{x^2}{6n^2} \le \frac{b^2}{6n^2}$  
  

upper bound:  $ 1- \frac{\sin t}{t} = \frac{t^2}{6}  - \frac{t^4}{120} \le \frac{t^2}{6} $  
bc $x \le b$, based on interval i'm using  

3. pick N  

$\frac{b^2}{6n^2} \le \epsilon \rightarrow n > \sqrt{\frac{b^2}{6\epsilon}}$  
$N = ceil(\sqrt{\frac{b^2}{6\epsilon}})$

## Proof with sequences

it's an optimized version of the epsilon proofs  

can be used to prove stuff for functions (limit existance, continuity) and sequences of functions  
no for sequences, has no sense, it's already a sequence itself so wouldn't be changing anything  

### Conditions

**Conditions for functions**

what | formula  
---|---
limit existance at $x=c$ | if $x_n \rightarrow c$ then $f(x_n) \rightarrow l$  
continuity at $x=c$ | if $x_n \rightarrow c$ then $ f(x_n) \rightarrow f(c)$  
uniform continuity |  if $\|a_n - b_n \| \rightarrow 0$ then $\|f (a_n) − f (b_n)\| → 0$  

**Conditions for convergence, sequences of functions**

what | formula  
---|---
pointwise to $f(x)$ | $\forall x \quad f_n(x) \rightarrow f(x)$  
uniform | $sup_{x \in I} \; \|f_n(x)-f(x)\| \rightarrow 0$

\end{cases}$

$\begin{cases}

$\begin{cases}

|f (a_n) − f (b_n)| → l \neq 0  
\end{cases}$  

2. prove: $\quad lim _{x \rightarrow \infty} f(x) = \infty$

### Steps explained

**Pointwise vs uniform convergence**

in pointwise, essentially just check that it never diverges  
it's fine if converge to different points for different x, can just say the $f(x)$ it converges to is definita a tratti  

in uniform, taking the sup means taking the upper bound BEFORE doing the limit  
must still have n when doing the upper bound, and must ignore it, don't know it will go to infinity, just treat it as a number  
this actually don't use sequences  
here should plug the same value in all x, but can also plug a differnet one in all x and get upper bound, and if upper converge this convgerge too  
however if upper diverge, don't know what the real one does  


if $f(x)$ defined a tratti:  
can prove it's NOT uniformly continuous (if it isn't, it still might be), by picking a sequence $x_n$ and plugging that instead of $f(x)$

**Intuitively continuity**

$x_n \rightarrow c$ means: check what happes when getting close to c   
$f(x_n) \rightarrow l$ means: the function gets close to l  
so, to be continuous, each time i get close to a point, function must get close to same point  
so, whatever sequences i take to arrive there, the function must send the image to the same point  
so, two sequences that end at same point x, must be sent at same point y  

**Continuous vs uniform continuous conditions**

the first condition for uniform continuous just ask that the sequences get close to each other, not to a specific point $x_0$    
so they might both tend to infinity, might not tend to a specific point $x_0$      
in such case, even if they have different images, it's not a problem for continuity, since they're actually not the same point  
so the condition for uniform continuity failing DOES NOT mean its not continuous  

so uniform continuity mean:  
Whenever two inputs are close — anywhere in the domain — their outputs must also be close  
This is much stronger: you want the function to behave "tame" everywhere, even at infinity.
in normal continuity don't care what it does at infinity, only what it does in specific points  

**Prove continuity**

from the easiest method:  
1. composition of continuous functions is continuous  
2. left and right limits are equal???  
3. sequence method  
4. epsilon method  

**Tips uniform covergence**

if get $\frac{\infty}{\infty}$ it's unsure, don't mean it's not uniform  
2 methods to solve:  

1. find more stringent upper bound  
example:   
$ sup_{[1, \infty]} \; \frac{x-1}{1+nx^2} = \frac{\infty}{1+n} \rightarrow \frac{\infty}{\infty}$  
so, before, do:  
$ sup_{[1, \infty]} \frac{x-1}{1+nx^2} \le sup \frac{x}{nx^2} \le sup \frac{1}{nx} = \frac{1}{n} \rightarrow 0$  

2. plug same x in all the x when doing the sup (to know which x to plug in, compute derivative of f wrt x, find x which gives biggest absolute value)
