### Definizioni

**Conditions for sequences of functions**

$f_n: A\rightarrow R$ converge to $f: A\rightarrow R$ se: 

puntiforme: $\quad \forall x \quad f_n(x) \rightarrow f(x)$  
uniforme: $\quad sup_{x \in A} \; \|f_n(x)-f(x)\| \rightarrow 0$   

in the first one, essentially just check that it never diverges  
it's fine if converge to different points for different x, can just say the $f(x)$ it converges to is definita a tratti  

in the second one, taking the sup means taking the upper bound before doing the limit  

**Domain**

for both, must be stated on a specific domain  
so must specify what is A, the one from which choose the x in the sup (it's an interval)  
could be convergent or uniformly convergent in an interval, but not in another  

**Intuitive definition**

pointwise: for each point, it eventually converges to where i want  
uniform: converges in a manner where after a point N, all next steps will stays within $\epsilon$ of my final function  

so the speed of convergence, which vary for all points in pointwise, is a unique value for all points in unifom (like the $\delta$ valries in continuity)  

**Definizione a parole**

I valori $f_n(x)$ si avvicinano a $f(x)$ con la stessa velocità per tutti i punti $x∈A$. 
Per ogni margine d'errore $ε>0$, esiste un indice N (che non dipende da x) tale che da lì in poi (cioè per tutti $n≥N$ e per tutti i punti $x∈A$ del dominio) la distanza tra $f_n(x)$ e $f(x)$ è minore di $ε$.  

In pratica: posso decidere in anticipo quanto voglio che le funzioni siano vicine al loro limite, e superato un certo indice N, saranno sempre così vicine in ogni punto.  

quindi la velocià di convergenza è la stessa per ogni punto.

**Delta-epsilon conditions**

$f_n: X \rightarrow Y$ converges to $f: X \rightarrow Y $ if  

**Pointwise**

$\forall x \in X \quad \lim _{n \rightarrow \infty}f_n(x) = f(x)$  

**Uniform**

$\forall \epsilon >0, \, \exists N \in N: \quad \forall n \ge N, \, \forall x \in X \quad |f_n(x) - f(x)| < \epsilon$  

### Proprietà

**Continuità e convergenza uniforme**

Se $f_n→f$ uniformemente e ogni $f_n$​ è continua, allora anche $f$ è continua.  
la convergenza puntuale non è abbastanza  
cioè, la convergenza uniforme preserva la continuità  

esempio:  
$f_n(x)=x^n$ has solo convergenza puntuale, quindi il suo limite $f(x)$ non è continuo  
perchè converge a $f(x)=0$ se $x \in [0,1)$ e s $f(x)=1$ se $x=1$  
quindi $f(x)$ è discontinua  

**Continuity sequence of functions** 

sequence of functions: if plug in a specific $n_i$ can ask if that $f_{n_i}(x)$ is continuous, else no  

$f$ is the pointwise limit of $f_n$  (when $n \rightarrow \infty$)  
$f$ is continuous if all the $f_n$ are continuous AND there is uniform convergence  
if all $f_n$ are continuous but there is only pointwise convregence, not uniform, $f$ might not be continuous   

so, pointwise convergence don't preserve continuity  

**Example**

$f(x)=\begin{cases}
0 & x \in [0,\frac{1}{n}] \\
n(x-1+\frac{1}{n}) & x \in [1-\frac{1}{n}, 1]
\end{cases}$  

all functions, whatever the n, are continuous  
limits are:  
$f(x)=\begin{cases}
0 & x \in [0,1) \\
1 & x=1
\end{cases}$  
bc when $n \rightarrow \infty$, all x end up in the $[1-\frac{1}{n}, 1]$ interval, except when $x=1$  
so for $x \neq 1$ follw first row, for $x=1$ follow second row  

all f are continuous, but NOT uniformly convergence, only pointwise    
so f, the pointwise limit, is not continuous  

### Pointwise convergence

**What i'm asked**  

simply means finding limit of succession or series when $n \rightarrow \infty$, <u>considering x as a param</u>    

so for what x this converge:  
succession: $\lim _{n \rightarrow \infty} f_n(x) = l$

**What converge means**

converge means $l$ is FINITE (not infinite)  
if $l$ is a function of x, might converge for some x and not for others, check for what values of x $l$ is finite  
if $l$ is a not function of x, either converge on all R or never does (if $l=\infty$)

for successions, can then say that $f_n(x)$ converges to $l$  

**What am i doing intuitively: convergence of succession**

$f_n(x)$ is a sequence of functions

increasing n gets it closer to the real function i'm trying to represent, so when $n \rightarrow \infty$ i have the real function  
so can say the function converveges to a function, the one when $n \rightarrow \infty$  

**Types of exercises**

### Uniform convergence

**Preliminaries**

if uniform convergence, then pointwise convergence  
but not necessarily the opposite  

so can have uniform convergence on $I$ only if there is punctual convergence in all $I$  

**Condition for succession**  

2 ways to write it:  

$lim _{n \rightarrow \infty} \quad( sup _I \quad |f_n(x)-l|) = 0$



Using Weierstrass theorem:  
$|f_n(x)-l| \le M_n \quad \forall x \in I$  
if $\quad \lim _{n \rightarrow \infty} M_n = 0 \quad$, f is uniformly convergent on I  

essentially:  
first line means finding $sup_I$  
second line means taking limit of sup  


**Condition for series**

condition is same as before, but instead of:    
$f_n(x)$ put $\Sigma f_n(x)$  
$l = \lim f_n(x)$ put $l=\Sigma ^{\infty} f_n(x)$

**Weierstrass theorem**

if $f_n(x)$ is convergent in a compact space, it's uniformly convergent  

note that if my upper bound is valid for all R (is independent of the interval), then there is convergence in any compact space $[-M, M]$ i may pick  
so means there is convergence in all R, bc for any point of R can find a compact space that contains it and where f is uniformly convergent  

#### Example

$f_n(t) = \frac{n \sin \frac{t}{n}}{t+t^2}$  
$I=[0,b]$

1. compute $l$ (pointwise convergence)  

can use Taylor approx, since $l = \lim _{n \rightarrow \infty} f_n(t)$  
$l= \frac{1}{1+t}$  

2. write condition  

$ sup_I \; | \frac{n \sin \frac{t}{n}}{t+t^2} - \frac{1}{1+t} |$  

3. find upper bound   

$ sup_I \; \frac{1}{1+t} (1- \frac{\sin \frac{t}{n}}{\frac{t}{n}}) \le sup_I \; (1- \frac{\sin \frac{t}{n}}{\frac{t}{n}})$  

do a var change  

$sup_{ \; 0 \le y \le \frac{b}{n} \;} \; (1- \frac{\sin y}{y})$  

since $ \frac{\sin y}{y}$ is decreasing in the interval defined under the sup, the minimum is when $x=0$ and gives 1, so $sup = 1-1 = 0$  

not even need to compute limit for $n \rightarrow \infty$  
condition respected, it's uniformly convergent

**Passagio al limite sotto integrale**

$\lim \int f_n(x) \; dx = \int \lim f_n(x) \; dx$  
the, right side is equal to $\int f(x) \; dx$  

note: lim tends $n \rightarrow \infty$  
bounds are $[a,b]$  

can swap if either:  
1. Dominated convergence th (DCT)  
pointwise convergence and exist $g(x)$ such that: $|f_n(x)|\le  |g(x)| \forall n$    

2. Monotone convergence th (MCT)  
$f_n(x)$ is non-negative and increasing pointwise to $f(x)$  

3. uniform convergence  
if $f_n$ convrge uniformly to $f$ on [a,b]

**Passaggio alla derivata sotto il segno di integrale (differentiation under integral sign)**

$\frac{d}{dx} \int f(x,t) \; dt = \int \frac{\partial}{\partial x}\; f(x,t) \; dt$  

bounds are $[a,b]$

- Leibniz integration rule
