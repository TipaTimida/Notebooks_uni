### Definizione continuità

**Definizioni** 
 
$f: A \subseteq R  \rightarrow R$  

$\forall \epsilon >0 \quad  \exists \delta >0:$  
$|x-y|<\delta  \quad \rightarrow \quad |f(x)-f(y)| <   \epsilon$  

$ \forall x \in A$: continuità puntuale in y  
$ \forall x,y \in A$: continuità uniforme in A  

**A parole**

Nell'uniforme, $\delta$ non dipende dal punto y.  

Posso fissare a priori quanto piccoli voglio che siano gli scarti tra i valori della funzione, e trovare una distanza tra i punti di ingresso (le x) che garantisce che ovunque nel dominio, i valori della funzione saranno così vicini tra loro.  

Cioè se due input sono abbastanza vicini, allora anche gli output saranno abbastanza vicini, indipendentemente da dove ci troviamo nel dominio.

In altre parole, la funzione non accelera improvvisamente da nessuna parte: la velocità con cui varia è controllata in modo uniforme su tutto il dominio. Non ha "impennate locali" come accade per $f(x)=\frac{1}{x}$​ vicino a zero.

**Termini**

proprietà globali (da definire rispetto ad un intervallo):  
continuità semplice: continuità puntuale in ogni punto   
continuità uniforme: più restrittiva  

### Esempio

**Def with sequences**

continuity at $x=c$   
if $x_n \rightarrow c$ then $ f(x_n) \rightarrow f(c)$  


uniform continuity   
if $\|a_n - b_n \| \rightarrow 0$ then $\|f (a_n) − f (b_n)\| → 0$  

**Non continuità uniforme**

$f(x)=\frac{1}{x}$, $\quad A = (0, \infty)$  

è continua, ma non uniformemente continua  

perchè continua?  
per ogni $x_0 \in A$, $\lim _{x \rightarrow x_0} \frac{1}{x}$ esiste ed è uguale a $f(x_0)$   

perchè non uniforme?  
sequenze: $x_n = \frac{1}{n} \quad y_n = \frac{1}{n+1}$  
stesso punto: $|x_n-y_n| \rightarrow 0$  
immagini diverse: $|f(x_n)-f(y_n)| = |n-(n+1)| \rightarrow 1$  
quindi non rispetta la definizione di continuità uniforme 

**Altro esempio**

$f(x)=x^2$ 

in R: non uniforme  
pick sequences: $a_n = n \quad b_n = n + \frac{1}{n}$  
to same point: $|a_n -b_n | = \frac{1}{n} \rightarrow 0$  
different image: $|f(a_n)-f(b_n)| = |n^2 -(n+\frac{1}{n}^2)| = 2+\frac{1}{n^2} \rightarrow 0$  

in $[0,1]$: uniforme    
take 2 sequences such that: $|a_n -b_n | \rightarrow 0$  and $a_n, b_n \in [0,1]$  
condition is:  $\quad |f(a_n)-f(b_n)| = |a_n^2-b_n^2| = |a_n+b_n| |a_n-b_n| \le  2[a_b-b_n] \rightarrow 0$  
so condition is proved  
the $\le$ is the upper bound in [0,1]  

**Note**

$\frac{1}{x}$ non è uniformemente contiua su su (0,1) perchè avvicinandosi a 0 la funzione (e la derivata) comincia a crescere velocissima, non riesco a trovare upper bound finito del tasso di crescita  
però lo sarebbe sicuramente su un compatto su cui è continua, tipo su [1,2]  
in realtà anche su (1,2) perchè questo intervallo on dà problemi, niente va ad infinito (e poi è Lipschitz)  

in $x^2$ ciò non accade mai in (0,1), funzione e derivata sono sempre finite  
accadrebbe su R, e per questo non è uniforme su R  

**Per dimostrare**

NON uniforme: scegli due specifiche sequenze che lo negano  
Sì uniforme: prendi sequenza generale, prosegui come in $\delta - \epsilon$  

### Proprietà

**Rapporti**

Lipschitz $\rightarrow$ continuità uniforme $\rightarrow$ continuità semplice  
non vale l'implicazione inversa  

intuitivamente:  

Lipschitz: derivative has upper bound, so f can't change too fast  

uniform continuity: f don't suddenly start changing faster in different regions, must change derivative gradually, but might still reach infinite derivative    

**Proof rapporti**

continuità uniforme e semplice: ovvio, si vede dalle formule (se c'è un $\delta$ che va bene per tutti, ogni punto ha almeno un $\delta$ che va bene)  

Lipschitz e uniforme:  
se Lipschitz, $|f(x)-f(y)|\le L|x-y|$  
scelgo $\delta = \frac{\epsilon}{L}$, così def di uniforme ridà:  
$|x-y|< \delta \quad  \rightarrow \quad |f(x)-f(y)|\le L|x-y| \le  L\delta = \epsilon$  

**Th Heine–Cantor**  

Ogni funzione continua su un insieme compatto (chiuso e limitato) è uniformemente continua.

proof:  

scrivi ipotesi della condizione di continuità uniforme (ho due sequenze a,b con stesso limite, voglio vedere che anche le immagini hanno stesso limite)  
prendo $|a_n - b_n | \rightarrow 0$   

compatto: Bolzano-Weierstrass, sequenza a ha sottosequenza convergente (quindi b converge a stesso punto per costruzione)  
$a_{n_k} \rightarrow x \quad b_{n_k} \rightarrow x$  

continuità: $|f(a_{n_k}) - f(n_{n_k})| = |f(x)-f(x)| \rightarrow 0$  
che è la condizione di continuità uniforme  

**Restriction**

restriction of uniformly continuous function is uniformly continuous  

**Extension**

if f is uniformly continuous defined on (a,b), an open dominium  
it can be extended with pointwise continuity on the closure of its dominium, so on [a,b]

#### Prove it's NOT uniformly continuous

simply reverse the ways to prove it is  
based on the dominium:   

- closed interval:  
prove it's not pointwise continuous on a point of the interval 

- open interval: 2 methods  
    1. prove it's not continuous on a point of the CLOSED interval  
    (bc uniformly continuous func can be extended with pointwise continuity to the closure of its domain)  

   2. prove $lim _{x \rightarrow \infty} f(x) = \infty$  

**Formal method to prove f is NOT uniformly continuous**

to prove it's not continuous, must find 2 successions $y_k, z_k$ such that:   
$|y_k − z_k| → 0$  
$|f (y_k) − f (z_k)| → ℓ \neq 0$  

$\rightarrow$ means tends to such value when $k \sim \infty$

$z_k = \sqrt[3]{(2k+\frac{1}{2})\pi}$

$|f(y_k)-f(z_k)| \rightarrow |0-1| = 1 \neq 0$

### Many definitions

**Modulo of continuity**

$f: A\subset R \rightarrow R$  
can define its modulo of continuity:  
$M: [0, \infty) \rightarrow [0, \infty)$  
note that $M(r)$ is crescent  

$M(r) = sup \{|f(x)-f(y)|: |x-y|\le r \quad x,y \in A\}$  

**Similar definitions**

uniformly continuous: $lim _{r \rightarrow 0} M(r) = 0$  

Lipschitz: $M(r)\le Lr$
