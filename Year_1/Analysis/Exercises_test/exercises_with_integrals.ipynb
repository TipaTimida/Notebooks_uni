### Compute integral for all values of a variable  

2 methods:  
1. recurrence  
2. Feynman trick

#### Recurrence

**Recurrence: what it is**

have $I_n$, an integral with a param n  

want to write as a system with:  
- a recurrence relation  
- a known initial value     

$\begin{cases}  
I_0 = ... \\
I_n = f(I_{n-1}) \\
\end{cases}$ 

**Recurrence: steps**

1. find recurrence relation  
probably integration by parts on $I_n$, find $I_{n-1}$    
in $I_{n-1}$ the integrand is similar, just instead of n there is n-1  
eg: $I_n = A - I_{n-1}$  

2. find initial value  
just compute $I_n$ when n=0, should be easy  

3. if can, after writing system, find $I_n(n)$, a formula to get value of $I_n$ putting in input n  
sometimes uses the factorial or something similar  
to find it, compute first few numbers and see if there is a pattern  

**Example**

$I_n = \int _0 ^\infty x^n e^{-x}$  

1. integrate by parts to find recurrence: $I_n = n \, I_{n-1}$  

2. find initial easy value:  $I_0 = 1$  

$\begin{cases}  
I_n = n \, I_{n-1}  \\
I_0 = 1  \\ 
\end{cases}$  


3. find formula for it  

$I_n = n!$  


#### Feynman trick (Trasformata of Laplace-Fourier)

**What it is**

also called 'differentiating under the integral sign'  
use when integrals depends on a parameter (eventually added by me, or can be the one given by the exercise)  
derivating wrt that param must simplify the integrand  

can differentiate under the integral sign if:  
the integrand function is continuous   
The partial derivative exists and is integrable (or continuous)  
The bounds of integration don't depend on the parameter.

**Steps Feynman trick**

have an integral in x, $I=\int f(x) \, dx$      

1. Define a function with the param t  
$F(t) = \int f(x,t)$  

2. Differentiate under integral sign wrt t  
$\frac{dF(t)}{dt} = \int \frac{\partial f}{\partial t}$   
then finish solving by integrating wrt x  

3. Integrate it wrt t to find the original function  
$ F(t) = \int F'(t) \, dt +c$    

4. Find c using known conditions  
generally some value of t that makes the integrand function go to 0  
eg: a known value (bc the original func $F(t)$ defined in step 1 is easy to compute for some t)  
or stuff like $lim _{t \rightarrow \infty} = 0$  

5. plug my original t, compute what i want  
$F(t=2)$  

**Notes steps**

1. the param t must be added in the integrand function  
make it so for a specific value of $t$, maybe $t=0$, $F(t)$ becomes my original integral  

2. after derivating should find a function i'm capable of integrating  
must differentiate wrt t, but plug bounds in x, so at the end i still have t  

3. bounds are only needed in differentiating step, to compute the itnegral obtained   
in other steps consider it indefinite integral  

**Examples to use Feynman trick**

the integrand is $f(x)$, the param is $t$  
general tip: maybe using chain rule, find some way to derivative numerator to obtain the thing at the denominator, so cancel each other out  


1. $x$ can be used as $x^t$ to cancel $\ln x$  
$\frac{x}{\ln x} \rightarrow \frac{x^t}{\ln x} \rightarrow \frac{1}{\ln x} x^t \ln x$  

2. $e^x$ can be used as $e^{xt}$ to cancel x  
$\frac{e^x}{x} \rightarrow \frac{e^{xt}}{x} \rightarrow \frac{1}{x} e^{xt} x$   

**Easy example**

$I = \int _0^1  \frac{x^2 -1}{\log x} dx$  

step | do | notes
---|---|---
define | $F(t) =  \int \frac{x^t -1}{\log x} dx$ | 
differentiate | $F'(t) = \int _0^1 x^t dx = \frac{1}{t+1}$ | $\frac{d}{dt} x^t = x^t \ln x$  
integrate  | $F(t) = \int  F'(t) dt = \int \frac{dt}{t+1} = \ln (t+1) +c$   
find c | $F(0) =  \ln (0+1) +c \quad \rightarrow \quad c=0$  | $F(0)=  \int  \frac{x^0 -1}{\log x} = \int 0 = 0$ 
compute | $F(2) = \ln 3$   | $F(x) = \ln (n+1)$  



**Tip for step 4**

if i'm capable of solving $F(t=0)$, instead of using it in step 4 to find c, can merge steps 4 and 5 and use the fundamental th of Calculus to say:  
$F(t^*) = \int _0^{t^*} F'(t) dt$   

so on easy example it's:  
$F(2) = \int _0^2 F'(t) dt = \int _0^2 \frac{dt}{t+1} = \log 3$   

**Laplace-Fuorier transform (Trasformata di Laplace-Fourier)**

Trasformata di Fourier: $I =  \int_{0}^{\infty} \frac{\sin t}{t} dt$   
note that here the integral is in dt, my param is x (which is originally 0)  

step | do | notes
---|---|---
define | $F(x) = \int  \frac{\sin t}{t} \; e^{-xt} \; dt$  |

compute |$F(0) = -\arctan(0) + \frac{\pi}{2} = + \frac{\pi}{2} $    |$F(x) = -\arctan (x) + \frac{\pi}{2}$  
 



##### Advanced ways to use it

**Other method for trasformata: 2 params**

1. here introduce 2 params, a and b  

$I(a) = \int _0^\infty \frac{\sin (at)}{t} e^{-bt} \, dt$  

2. differentiate wrt a, first param  

$\frac{dI}{da} = \int _0^\infty \cos (at) e^{-bt} \, dt = \frac{b}{a^2+b^2}$

3. integrate back

$I(a) = \int _0^a \frac{b}{x^2+b^2} \, dx = \tan ^{-1} (\frac{a}{b}) $  

4. Go back to original integral, must put $a=1$ and $b=0$  

$\int _0^\infty \frac{\sin t}{t} dt = lim _{b \rightarrow 0} I(a=1) = \lim _{b \rightarrow 0} \tan ^{-1} ( \frac{1}{b}) = \frac{\pi}{2}$

**Same ex as recurrence one, but with Feynman trick**

recursive method, good for exponentials  

$I_n = \int _0 ^\infty x^n e^{-x}$  

1. define function with a parameter  

$I_n(t) = \int _0^\infty x^n e^{-tx} \, dx$   

2. differentiate wrt t: first differentiate $I_0$, then the successives, to find some recurrence    
bc cannot solve the fisrt derivative, so continue until i find a way to solve  

$\frac{d}{dt} I_0(t) = \int _0^\infty \frac{d}{dt} e^{-tx} \, dx = \int _0^\infty -x e^{-tx} = -I_1(t)$ 

so: $I_1(t) = -\frac{d}{dt}I_0(t) = -\frac{d}{dt} \frac{1}{t}$  

$I_2(t) = \frac{d^2}{dt^2} (\int_0^\infty e^{-ax}) = \frac{d^2}{dt^2} (\frac{1}{t})$  

$I_n(t)= \int_0^\infty x^n e^{-tx} dx = (-1)^n \frac{d^n}{dt^n} (\int_0^\infty e^{-tx} dx) = (-1)^n \frac{d^n}{dt^n} (\frac{1}{t})$  

3. now solve properly  

$\frac{d^t}{dt^n}(\frac{1}{t}) = (-1)^n \frac{n!}{t^{n+1}}$   
$I_n(t)=  (-1)^n  (-1)^n \frac{n!}{t^{n+1}} = \frac{n!}{t^{n+1}}$  

4. plug t from my original function, t=1

### Prove an integral is equal something for every n  

**Steps: induction**

have $I_n = g(n)$, where n is the param   

1. base step: check it's true for $n=0$ (should be easy to compute)  

2. inductive step: assume it's true for $I_n$, prove it's true for $I_{n+1}$  
    - $I_{n+1} = f(I_n)$  
    integrate by part $I_{n+1}$ so the second addend, that remains an integral, contains $I_n$  
    - $I_{n+1} = f(g(n))$  
    plug in $I_n = g(n)$   
    - $I_{n+1} = g(n+1)$  
    manipulate to obtain this   


**How to derivate by parts**

want to get $I_n$ in the part that stays an integral  

1. check what the term with my param, n, is in the $I_n$ and $I_{n+1}$ integrals  
eg: in $I_n$ it's $x^{2n+1}$, so in $I_{n+1}$ is $x^{2n+3}$  

2. check what i need to derivate to get the term in $I_n$  
i want to get $x^{2n+1}$, the original thing, so i want to derivate $x^{2n+2}$ (bc exponent gets a -1)  

3. modify the term, bringing some stuff over to the term to be integrated, so i obtain the form  i want it in to derivate it  
from the thing in $I_{n+1}$ take away an x, put it in the part to be integrated  
$x^{2n+2} \;x$  
derivate first half (the thing i wanted at tep 2) and integrate second  

**Induction vs recurrence**

in both do:  
- find $I_0$  
- integration by parts to get recursion $I_{n+1} = f(I_n)$  

but then changes:  
- recurrence: finish here  
- induction: since i know what $I_n$ is, plug it in, check that $I_{n+1}$ is that thing too 

**Example**

$I_n = \int x^{2n+1} e^{-\alpha x^2} dx = \frac{n!}{2\alpha ^{n+1}}$    
$I_{n+1} = \int x^{2n+3} e^{-\alpha x^2} dx$  

- modify the function inside the integral to get a term that, integrated, gives me what i want:  

$I_{n+1} = \int \frac{x^{2n+2}}{-2 \alpha} \; (-2 \alpha x)e^{-\alpha x^2} dx$  

- do integration by parts:  

$ = A - \int \frac{2n+2}{-2 \alpha}x^{2n+1} \; e^{-\alpha x^2} dx$  

- substitute the integral at the second addend with the thing i know it's equal to:  

$= A + \frac{2n+2}{2 \alpha} \frac{n!}{2 \alpha^{n+1}}$   

now manipulate to get n+1 for of thing it should be equal to

### Find upper/lower bounds

**What values to plug in**

can either plug:  

- the same value in all x: more stringent, but first must find maximum of $f_n(x)$ by studying $\frac{df}{dx}$  
only do if a previous exercise made me find such maximum point  

- a different value in every x: more lax, but faster and sometimes it's enough  
generally, at numerator and denomiantor put extremes of interval  

**When to plug in**

- if bounds of interval are funcs of n: plug them in all x    

- if bounds have $\infty$: plug into evertything i can, except those where i would want to plug $\infty$  
leave those and integrate  

- if inside integral have trig func: don't plug anything in there  
plug in all others, then say $\sin(nx) \le 1$ to get upper bound or integrate it  

or in general, plug in those x that, after not being there anymore, allow me to integrate normally  


**Tips**

the fastest method is to plug something in EVERY x that appears  
in this case, integrating means multying per the lengh of the interval, $(b-a)$, since would get a constant function  

if $f_n(x)$ is always positive, to find lower bound can reduce the bounds until something where it's easier to plug in ($\int_0^\infty \ge \int_0^1$)m so cannot plug in $\infty$  

**Example: plug everywhere**

$\Sigma \int _n^{2n} \frac{\ln (x)}{(1+x^2)^\alpha} dx$  

find bounds (bounds are funcs of n, so plug everywhere):  
upper bound:  $a_n \le n \frac{\ln (2n)}{(1+n^2)^\alpha} \sim \frac{1}{n^{2\alpha}}$  
lower bound:  $a_n \ge  n \frac{\ln (n)}{(1+(2n)^2)^\alpha} \sim \frac{1}{(2n)^{2\alpha}}$  

upper converges for $a >1$, lower diverges for $a<1$, so the original integral does the same  
now plug in $a=1$ and check  

**Example: plug only some and tip 2**

delete just some terms, not all  

$a_n = \int _0^\infty \frac{e^{-x}}{n+x} \; dx$  

find bounds:  
bounds are $0$ and $\infty$, so plug only in some

upper: $a_n \le \int^\infty_0 \frac{e^{-x}}{n} \; dx = \frac{1}{n} \int_0^\infty e^{-x} \; dx = \frac{1}{n}$

## Limit with integral

2 methods:  

1. de l'Hopital

#### De L'hopital  

Il teorema di de l’Hospital si può applicare, nel caso in cui il
denominatore di un rapporto tenda a $+∞$ anche senza alcuna informazione
sul numeratore.  

**Example**

$\lim _{h \rightarrow 0} \frac{1}{h}     \int_0^h cos(x^x) dx$  

the numerator is the integral func, that i can derivate:    
$F(h)= \int _0^h cos(x^2) dx$  
$F'(h) = cos(h^2)$  
 

so limit becomes:    

$lim_{h \rightarrow 0} \frac{ cos(h^2)}{1}$  

**Compute derivative of an integral**

$F'(x) = f(upper bound) \cdot (upper bound)' - f(lower bound) \cdot (lower bound)'$

where f integrand function    
obv, if one bound has termine noto, its derivative is 0, so that term don't appear  

eg:    
$F(x) = \int _{x^2}^{x^4} \frac{1}{ln(t)} dt$  
$F'(x) = \frac{4x^3}{ln(x^4)} - \frac{2x}{ln(x^2)}$  

#### Taylor approx

- integrand function f

#### Both bounds are $\infty$  

if both bounds tend to infinity, can assume the var inside tends to infinity too  

eg: $\lim _{x \rightarrow \infty} \int_x^{x+\frac{1}{x}} f(t) \; dt$   
clearly when $x \sim \infty$, both bounds are $\infty$, and so all $t$ in $f(t)$ must belong in such interval and be $\infty$ too

**Recap**

this extimates the value in a point of the interval, multiply per length of interval    
$\lim _{x \rightarrow \infty}  \int _x^{x+c}f(t) \; dt \sim f(x) \; c$
