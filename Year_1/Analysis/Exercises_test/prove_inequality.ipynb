### Convexity

**Doing it formally**

convex: graph above tangent for all points  
$e^x \ge 1+x \quad$ for $x \ge 0$  
$e^x \le 1+x \quad$ for $x \le 0$  


concave: graph under tangent for all points  
$\sqrt[n]{1 + x} \le 1 + \frac{1}{n}x \quad$ for $x \ge 0$ 

generally need to expand only 2 terms for each non-rational func  
so, chain the the inequalities:  
$e^x > 1+x > 1 + \frac{1}{n}x > \sqrt[n]{1+x}$  

**Trig**

$\sin x \le x \quad \forall x >0$  
$\sin x \ge x \quad \forall x<0$  
$|\sin x| \le |x|\quad $ always  
these are valid bc $f(x)=x- \sin x$ is crescent and $f(0)=0$  

$\sin x \ge x -\frac{x^3}{6} \quad \forall x >0$  


$\ln x \le x-1$  

**Doing it intuitively**

start doing Taylor approx, what sign should i pick, $\le$ or $\ge$?  

depend on first term i nave not included     
if is't positive, pick $\ge$, if negative pick $\le$  

example:  

$e^x = 1+x + x^2 +O(x^3) > 1+x \quad \rightarrow \quad e^x \ge 1+ x$

$\sqrt{1+x} = 1 +x -\frac{x^2}{2} +O(x^3) <1+x  \quad \rightarrow \quad \sqrt{1+x} \le 1 +x$  
first non-included term is $-\frac{1}{2}x^2$
