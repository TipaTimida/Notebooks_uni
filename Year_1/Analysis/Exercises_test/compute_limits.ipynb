## Asymptotic behaviour

**How to do it**

1. SPLIT THE THING IN BRACKETS   
(factors, stuff separated by muliplication)  
each bracket can contain only sums  

2. in each bracket, KEEP ONLY 1 ADDEND  
which one depends on the singularity i'm examining:  
$x \sim \infty$: keep term HIGHEST in infinitum hierarchy  
$x \sim 0$: keep term LOWEST in infinitum hierarchy  
$x \sim 1$ or other stuff that breaks C.E., do like x=0, keep lowest term, but must hold their own infinity (so here keep x-1, but still better to do var change)  

3. put the exponent of the bracket on the addend i chose to represent it  

4. now have a multiplication of stuff  
here don't matter what singularity, WINS ALWAYS THE TERM HIGHEST IN INFINITUM HIERARCHY  
the limit is the value of such term when plugging in the limit  

**Example**

sum:  
$x \sim \infty: \quad x + \ln x \sim x = \infty$  
$x \sim 0: \quad x + \ln x \sim \ln x = -\infty$  

multiplication:  
$x \, \ln x \sim x \quad$ for all singularities  


**Infinitum hierarchy**

$\ln x≪x^a≪a^x≪x!≪x^x$  

$\frac{1}{x^a} \gg e^{-x} \gg e^{-x^2}$  

if have difficult stuff, try to do var change to bring it to a form i know  
or, if have logs: $x ^{\ln x}=e^{\ln x \ln x}=e^{ln^2 x}$

**Recap infinitum hierarchy**

which term wins (is kept):  
addition: depends on singularity (highest for $x \sim \infty$, lowest for $x \sim 0$)  
multiplication: always the highest

**Notes on infinitum hierarchy**

why work different in addiction and multiplication?    
each time i want to keep only the most 'meaningful term', but the meaning of meaningful changes  

addiction:  
meaningful means BIGGEST (in absolute value)  
to decide biggest addend must check what x i'm studying:  
$x \sim \infty$, the higher in the hierarchy the higher its value, so pick the highest  
$x \sim 0$, the higher in the hierarchy the closest to 0, so the biggest is actually the lowest  

multiplication:  
meaningful means that CHANGES FAST  
stuff higher in infinitum hierarchy CHANGES FASTER, both diverge to $\infty$ or converge to $0$ faster  
so the highest in the infinitum hierarchy always wins, no matter what x i'm studying  

**With parameter, if unsure which term to keep**

$\lim _{n \rightarrow \infty} \frac{x^n + 3^n}{4^n}$  

at numerator, what it scales with depends on value of x  
must split the problem in 3 cases:  
$\begin{cases}  
\frac{x^n}{4^n} & if \quad x>3 \\
\frac{3^n}{4^n} & if \quad x<3 \\
\frac{2 (3^n)}{4^n} & if \quad x=3 \\
\end{cases}$

do each case separately  
for each, after finding conditions on x for it to converge, pick only values of x among those that generate such case  

## Rationalization

do if have non-rational funcs like: $e^x, \log x, cos(x)$  

**How to rationalize**  

use Taylor approximation  
only if the ARGUMENTS OF THE NON-RATIONAL FUNC TENDS TO 0  
or for logarithm and binomial, if the arguments tends to 1  

can be either:  
$e^x$ when $x \sim 0$  
$e^\frac{1}{x}$ when $x \sim \infty$  

**When have $\lim x \rightarrow 1$**

best way: DO VAR CHANGE $y=x-1$, so have $y \sim 0$  

or if not var change:  
only $(x-1)$ holds an infinity, nothing else  
so if have something else can just plug in $x=1$, so it don't confuse me  

examples:  

$\frac{1}{x^2-1} = \frac{1}{(x-1)(x+1)} = \frac{1}{2(x-1)}$  
$\frac{1}{x(x-1)} = \frac{1}{x-1}$  

**Taylor approx when $x \sim 1$**

use the long formula:  
$f(x) = f(1) +f'(1)\, (x-1)$

### Exponentials / binomials (n at exponent)  

**Great technique for binomials**  

$x^n = e^{\ln(x^n)} = e^{n \ln (x)}$


use to:  
- compare exponentials with different base, like $x^n$ and $e^n$  
- if have n IN EXPONENT  
- if have $1^\infty$, to recognize a Nepero  


**What i could do then**

maybe something tends to 0 or 1:  
argument of log to 1: taylor approx on $ln (1+x)$   
exponent of e to 0: taylor approx on $e^x$  

$\sqrt[n]{c} = e^{\frac{ln c}{n}} =  1 + \frac{ln(c)}{n}$  
last = use Taylor  

**If arg tends to $\infty$**

$(n+1)^a= n^a(1+\frac{1}{n})^a = n^a (1 + \frac{a}{n} + o(\frac{1}{n}))$  

**Nepero**

instead of rememebring it, obtain it by using his technique for binomials  
for ease, first get $(1+x)$ in argument, so later can easily simplidy the log  

**Examples**

easy:  
$\lim (1+\frac{-3}{x})^{2x} = e^{2x \ln(1-\frac{3}{x})}= e^{(2x)(\frac{-3}{x})} = e^{-6}$    

rem it's a limit $x \sim \infty$, so can approximate $x+a \sim x$  
$(1+ \frac{n}{x+a})^x = e^{x \ln(1+ \frac{n}{x+a})} = e^{x \frac{n}{x+a}}  \sim e^{x \frac{n}{x}} = e^n$   

need some manipulation to get (1+x):  
$(\frac{2n}{2n+1})^n = (1-\frac{1}{2n+1})=  e^{n \ln (1-\frac{1}{2n+1})} = e^{-n\frac{1}{2n+1}} = e^{-\frac{1}{2}}$ 

**Behaviour with param**

$\lim (1+ \frac{c}{n^a})^n =  
\begin{cases}
\infty & a <1 \\
e^c & a=1 \\
1 & a>1
\end{cases}$

why:  
$(1+ \frac{c}{n^a})^n = e^{n \ln(1+ \frac{c}{n^a})} \sim e^{n \frac{c}{n^a}} = e^{\frac{c}{n^{a-1}}}$

if the exponent is not n, so is less than $\infty$, this converges to 1, bc the I.F. is only $1^\infty$, not $1^N$

### Trigonometric

**Complicated argument, with n in it**  

often <u>can turn in something like $(-1)^n$ if there is n inside the argument</u>  
use formulas for sin ad cos of sum, $sin(a+b)$   
this only true in series, bc n only takes integers values  
 
$cos(\pi n) = (-1)^n$  
$ sin((n+1)π + x) = (−1)^{n+1} sin(x) $  
$ sin(nπ + x) = sin(nπ) cos(x) + cos(nπ) sin(x) = (−1)^n sin(x) $

deal with obscillating 1: can add minus depending on parity of number i'm taking away from exponent  
$(-1)^{n+1} = -(-1)^n$   

an obscillating trig is simply a random number, NOT something that scales with x  
so, eg: $cos(x) +x \sim c+  x \sim x$ when $x = \infty$  

**Param in argument**

$\frac{1}{\arctan(x^\alpha)}$  

when $x \sim \infty$, $x^\alpha$ can have results really different  
so must examine all cases separately:  

$x^\alpha = \begin{cases}
 {+} \infty & if & \alpha >0 & \rightarrow \arctan(\infty) = \frac{\pi}{2} \\
1 & if  & \alpha = 0 & \rightarrow \arctan(1) = \frac{\pi}{4}\\
0 & if  & \alpha <0 & \rightarrow \arctan(0) = use \; Taylor \\
\end{cases}$  

when finding $\alpha$ with p-test, remember to only consider $\alpha$ belonging to the current case  
troublesome also for direct functions, like sin(x), not just for inverse ones  

**Inverse trig**  

- domain   
arcsin has domain [-1, +1]  
so if it's arcsin(1), it can only approach from the left, not the right  
(only one limit exist, can't come from second quadrant, bc arcsin is not defined there, is defined from -90 to +90 deg)   
so it's only 90 deg coming from first quadrant, where cos is positive, $0^+$ 


how to work with it:  

1. intuitive  
see what angles (more than one possibility) can be in argument    
then see if the trig thing asked is the same for all those possible angles  

2. formal  
$ sin(arcsin(x)) = x$  
$ cos(arcsin(x)) = sin(arccos(x)) =\sqrt{1−x^2}$  

---

to find formula:  
$sin^2 \theta + cos^2 \theta = 1$  
plug in $x = sin\theta$  
$x^2 + cos^2 \theta = 1$   
$cos^2 \theta = 1-x^2$     
$cos \theta = cos (arcsin(x)) = \sqrt{1-x^2}$  

### Logarithm

**Weird base**

$\ln _n (n!) = \frac{\ln (n!)}{\ln n} < n$  

by infinitum hierarchy:  
$n! < n^n \quad \rightarrow \quad \ln (n!)< n \ln n \quad \rightarrow \quad \frac{\ln (n!)}{\ln n} <n$  

$\ln (n!) \sim \frac{1}{2} \ln (2 \pi n) + n \ln n -n \sim n \ln n - n$  
using Stirling's formula $n! = \sqrt{2 \pi n} (\frac{n}{e})^n$

## Indetermined

**DNE vs IF**

dne = does not exist   
if = indetermined form

limit <u>DNE</u> when:   
1. left and right limits have different values  
likely with absolute value  

2. something obscillates, likely a trig func     

if one side don't exist, does limit DNE? not clear, better to compute the one that does exist  


<u>getting indetermined form does NOT mean limit DNE</u>  
just need to find another way to compute it  

tricky IF:  
$1^\infty$: it's not 1, proof is in expression that gives e  
$0^0$ (depends on how is exaclty the expression that goes to 1)  
$\frac{n}{0}$: bc don't know sign of 0, if i know it it's not indetermined
