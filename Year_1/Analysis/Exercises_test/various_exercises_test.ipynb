### How many solutions: eq unsolvable algebrically

steps:  

0. preliminary: check for what x the function is defined, so what is the dominium  

1. bring all on one side and put equal to zero  
f(x) = ... = 0  
or, if asked to study number of sols depending on a parm, put param on right: $f(x) = ... = \lambda$

2. find monotonic intervals  
(to be sure to catch all solutions)  

3. find sign of extremes of all intervals  
(rem $\pm \infty$ are also extremes of intervals)  

4. use th of zeros (between pos and neg points must cross zero)

**Example**

$e^x = x^3$  


1. bring all on one side    

get expr of x:  
$ x = ln(x^3)$  
$ x = 3 ln(x)$  

make func with it (that i will want the roots of):  
$f(x) = x- 3ln(x) = 0$   

2. monotonic intervals  

compute derivative:  
$f'(x) = 1 - \frac{3}{x}$   

find intervals of monotony:  
x>3: crescent  
x<3: decrescent  


3. compute at bounds of intervals  

$f(0) = + inf$  
$f(3)<0$  
$f(+ inf) = + inf$  


4. th of zeros  

has one sol in interval $[0,3]$, one in $[3, \infty]$    
to have more precision, bisection method (each time halve the interval, check sign at half point)  

**Tip computation**  

to find interval of monotony:  

$1- \frac{8}{x^3} =  \frac{x^3-8}{x^3}$  

try not to multiply by x, causes trouble bc don't know if it's positive or negative  
instead transform in unique fraction, study numer and denom separately  

rem that: $\sqrt{x^2} = |x|$, need the absolute value  
like: $\sqrt{x^3- x} = |x|\sqrt{x-1}$ 

**If asked number of sols of $f(x)=y_0$ al variare di $y_0$**

probably can draw $y(x)$, pay attention to intervals of monotony, explicitate them, must know boundaries    
then looking at graph, seeing each horizontal line how any times touch the function  

#### Other

**Prove at least 1 solution**

find auxiliary eq $g(x)$ to turn it into prob of finding a 0 (root)  
eg: $e^x =x^3 \rightarrow g(x)=e^x-x^3$  
find two points where $g(a)>0, g(b)<0$, bc of Bolzano-Weierstrass there is a root  

**Prove at most 1 solution**

find all roots of $g(x)$  

if it's easy to study:  
find intervals of monotony, compute value at each exteme of interval, see how many intervals have extremes of opposite sign  

if hard to study: depends on the function  
if one of the sides is periodic, draw both functions check in what intervals they could touch  
study behaviour of $g(x)$ in that interval: if it's monotone, there is at most 1 root

### Th of sign permanency

for exercises like  
$\exists \epsilon : f(x)>0 \in [0, \epsilon]$

**Th of sign permanency**

if $lim _{x \rightarrow x_0} g(x) = l >0$, then exists a neighborhood of $x_0$ where $g(x)>0$  

**To prove exists a neighborhood where something is positive or crecent**

positive: find neighborhood of $x_0$ where f>0  
essentially find a point $x_0$ with $\lim _{x\rightarrow x_0} f(x) >0$    

crescent: find neighborhood of $x_0$ where f'>0  
essentially find a point $x_0$ with $\lim _{x\rightarrow x_0} f'(x) >0$
