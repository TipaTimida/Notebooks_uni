### Continuity and derivability

**Intuitive def**

it's continuous in all points that don't violate the C.E. of f  
it's derivable in all points that don't violate the  C.E. of f'  

if in a point it's not continuous it's certainly not derivable in that point  

so NOT DERIVABLE IF:  
- derivate is NOT FINITE  
- left and right derivative are different  
- f not continuous in that point  

**Closed interval**

f is derivable in [a,b] if:  
]a,b[ has derivative  
a has right derivative  
b has left derivative  

a and b don't need derivative on the other side of the interval 

**!!! Formal justification !!!** 

f is continuous/derivable in $I$ (an interval, or dominium D) if:   
it's the composition (means product, sum or chain) of continuous/derivable elementary functions on $I$  

examples:  
$f(x) = \frac{1}{x}e^x: \quad$ is continuous in $D=R-\{0\}$, bc it's composition of elementary functions continuous in D    
$f(x) = x e^x: \quad$ f is continuous in R bc it's composition of continuous elementary functions  
eg. of non-continuous elementary function: $\frac{1}{x}$, give C.E. that denom $\neq 0$  


**Find points where it's not continuous/derivable**

steps:  
1. find singularities of f (points that violate C.E., and also points where it's not defined)    
2. if a point violates the C.E. bc sends the denom to 0, compute the limit with TAYLOR APPROX  
3. if with Taylor the limit is finite, that point is continuous    
if it's still infinite, it's non derivable    

same thing for derivability, but check  singularities of f' always remember to check with Taylor approx if can  

**Example: using Taylor to tell continuity**

when x=0 it violates the C.E., but with Taylor:   
$\frac{x^2}{\sin (x^2)} = \frac{x^2}{x^2} = 1$  
so in x=0 is continuous    

**Example C.E. for trig**

$\frac{1}{sin(x^3)}  \quad \rightarrow \quad  sin(x^3) \neq 0  \quad  \rightarrow  \quad x^3 \neq k \pi  \quad \rightarrow  \quad x \neq \sqrt[3]{k \pi}$

**Check differentiability and continuity of derivative**

f is differentiable in $x_0$ if: $\quad \lim_{h \rightarrow 0} \frac{f(x_0+h)-f(x_0)}{h} < \infty$  
is the incremental rapport, the one fron the definition of derivative   
$f(x_0)$ is the value i gave as extension

f' is continuous in $x_0$ if: $\quad \lim_{h \rightarrow 0} \frac{f(x_0+h)-f(x_0)}{h} = \lim _{x \rightarrow x_0} f'(x)$    
left side is thing computed before  
on right side, $f'$ is first obtained through normal rules of derivation (chain rule, derivative of product...), then take limit of that  

**Example**

$f(x) = \begin{cases}
x^2 \sin (\frac{1}{x}) \quad x \neq 0 \\
0 \quad x=0 
\end{cases}$

is f differentiable at $x=0$?  
$\lim_{h \rightarrow 0} \frac{f(h)-f(0)}{h} = lim \frac{h^2 \sin \frac{1}{h}}{h} = lim \, h \sin \frac{1}{h} = 0$  
it's finite, so it's differentiable  

is f' continuous at $x=0$?  
left side is the one i computed before  
right side is: $lim _{x \rightarrow 0} f'(x) = lim _{x \rightarrow 0} \, 2x \sin (\frac{1}{x}) - \cos (\frac{1}{x})$  
which is obscillating, so not finite limit   
thus not continuous

### Find min/max, bounds

**Find max and min**

must check:  
- point where $f'=0$  (critical points)  
- extremes of interval (if $I=R$, check $x = \pm \infty$)
- points of non-derivability  

**To know if stationary point is relative max or min**  

if get $f'(x_0)=0$, check subsequent derivatives, until the k derivative $f^{(k)}$ that is non-zero  

- k odd: it's flesso (horizontal tangent)  
f is monotonic in that point, no relative max or min  
    - $>0$: crescent  
    - $<0$: decrescent 

- k even: it's max or min  
$>0$ it's min, else max  



**Proof**

write it as a Taylo expansion centered in $x_0$:  
$f(x) =f(x_0)+f'(x_0)(x-x_0)+ ... + \frac{f^{(k)}(x_0)}{n!}(x-x_0)^n + o(x-x_0)^n$  

assume all the derivatives from $f'(x_0)$ until $f^{n-1}(x_0)$ are null (note that $f(x_0)$ is not null)  
so the sign of $f(x)$ close to $x_0$ is controlled by $f^{(n)}$  
consider some $x_1 < x_0 <x_2$, all really close  

if n is even, $(x-x_0)^n>0$  for both $x_1$ and $x_2$  
so the sign is decided by $f^{(n)}(k)$, and is the same for $x_1$ and $x_2$, thus either they're both bigger or smaller than $x_0$  
so if this derivative is positive, get $f(x) \le f(x_0)$, so $x_0$ it's a max, else it's a min  

if n is odd, $(x-x_0)^n$ has opposite sign for $x_1$ and $x_2$  
since numerator is independent on x, the images must be one bigger and smaller than $f(x_0)$, thus cannot be max or min  

**Max/min vs sup/inf**

in both cases, study f' to see where f increase and decrease  

for max/min: $f(x_0)$  
$x_0$ must be a value IN THE DOMINIUM (so no $\infty$ or other extremity with open bracket, if I=(a,b) cannot do f(a))  
must compute f in that point $f(x_0)$, no take limit    
so might not exist, eg. monotonic crescent function on (a,b) has no max or min, bc a,b don't belong to the dominium  

for sup/inf (upper/lower bound): $lim _{x \rightarrow x_0} f(x)$   
$x_0$ can be extremity of open bracket or infinity, don't need to belong to the dominium    
must take LIMIT of where i assume f to be biggest/smallest  
must exist, eventually it's infinite (unlike max/min, which can not exist)  

**What being bounded mean?**

if f has FINITE upper and lower bound, it's bounded  

if f is bounded: f has no asymptotes  
if f' is bounded: f is Lipschitz  

if f is continuous in compact space (closed interval) has finite sup and inf bc of Weierstrass th  
so it's Lipschitzian  



prove f has upper bound: if f is continuous in closed interval, bc of Weierstrass it has bounds

### Lipschitzian

**Definition**

means the derivative (incremental rapport) has an upper bound   
f is L-lipschitzian means L is the max value the derivative can assume, $L= sup \, f'$  

if has a non-derivable point, it's not lipschitzian  
bc not derivable means derivative is infinite, so derivative cannot have an upper bound  

**Definition more formal**

f is lipschitz on $[a,b]$ if:  

continuous on $[a,b]$  
derivable in $(a,b)$ (open interval is ok)    
derivative limited on $(a,b)$ (open interval is ok)  

**Easier def** 

means incremental rapport has upper bound (so there is a limit to the steepness):  

$|\frac{f(a)-f(b)}{a-b}| \le L$  

for every x in the dominium  

**Prove it's Lipschitzian**

2 methods:  
1. compute derivative f', check what the max is  
that is the value of L  

2. actually plug in $f(x)$ and $f(y)$, the same func but changing the letter inside    
continue until i get $|f (x) − f (y)| ≤ L  |x − y|$, the definition of Lipschitzian    

**Example method 2**

$f(x)= \sqrt{|x|+1}$  

$|\sqrt{|x|+1} - \sqrt{|y|+1}| = |\frac{|x|+1- |y|-1}{\sqrt{|x|+1}+\sqrt{|y|+1}}| \le \frac{|x-y|}{2}$  

so f is $\frac{1}{2}$-lipschitzian  

**Banach-Cacciopoli th**

if $0<L<1$, f is a contraction  

a contraction must have one and only one point of convergence (whatever the starting value of the recurrent succession)  
useful if f is thought of as a recurrent succession



**Find Lipschitzian L: formally**

question is to find L where:  
$|f (x) − f (y)| ≤ L  |x − y|$  

bc of Lagrance, know exists a point C where:  
$\frac{f(x)-f(y)}{x-y} = f'(c) \quad$ so:  
$|f(x) -f(y) | = f'(c) |x-y| \le (sup \; f') \; |x-y|$  

**If f has modulo in it/is not derivable in a point**

modulo is not derivable in $x=0$, so this kind of function is not derivable in $x=0$, but might still be Lipschitz   

method 2 has no problem, use properties of modulo operations  
method 1 is troublesome:  
if there is $|x|$, separate the interval $(-\infty, 0), (0, +\infty)$ and prove f' has upper bound in both of these (here x has always same sign, so drop the absolute value)  
also must know that f is continuous in $x=0$  
thus it's lipschitzian on the closed intervals  $(-\infty, 0], [0, +\infty)$  
since they touch in $x=0$, it's Lipschitz on their union, which is all R

### Local and global properties

**Examples**

local | global | (higher) global  
---|---|---
pointwise continuity |continuity | uniform continuity 
pointwise differentiability |derivability | Lipschitz  
pointwise convergence | convergence |  uniform convergence  

**Local vs global**  

- local: property of the function IN A POINT   
so f can have it in some points and not in others  

- global: property of the WHOLE FUNCTION          
so either f has it or not, cannot have it only in some points  

**What are the cols**

local: the local property  
global: if local is true in all points  
(higher) global: if global and something else, generally some param must be acceptable for all points  

higher global is a random name i gave it, it don't have a name  


**Relationship**

in their definition there is var/counter, like $\delta$ or $N$, that:  
in a local and global property can be different for any point  
in a higher global property must be the same for all points  

so the var in higher global is the upper bound of the vars in local and global    
if local is not true in a point, in that point the var is $\infty$  
so higher global, which takes upper bound, has $\infty$ too, so that's not true either
