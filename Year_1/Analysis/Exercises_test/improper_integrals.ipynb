**Integral vs series convergence**

it's similar, except: 
- in series the only singularity is $x=\infty$  
- series has some particular ones (geomteric, power ...) that are not special in integrals  

essentially:  
series have tricks to evaluate it: can use ratio test, test for geometric, power, telescopic...  
integral can't, i always need to rationalize and do asymptotic behaviour  

only tricks that work in integrals are:  
- p-test (if can rationalize)  
- Dirichlet test (only at $x \sim \infty$, generally at $x \sim 0$ can use Taylor anyway)/ upper bound  
- nth term test (necessary, but not sufficient) (if integrand func don't tend to 0, integral is not finite)

steps:  

1. find singularities (doing the C.E.)  
2. check all singularities  

**If the function can be rationalized**

1. rationalize  

2. asymptotic behaviour  

3. p-test  

if function can't be rationalized probably some argument goes to infinity  
below are ways to solve it  

### Find singularities

2 types

- $\pm \infty$ (if they appear in the integration bounds, they're always singularities)

- points that violate the C.E. (existance conditions)   

they are singularities only IF THEY BELONG IN THE INTERVAL in which i'm integrating  

**Attention: singularities with param**  

if i'm not told a param is $>0$, assume it could be negative  
so it's the exponent to something, that something could be at denominator  
so pick as well singularities that would have at denominator  

eg: $\int x^\alpha \; dx$   
if not told $\alpha >0$, could have x at denominator, so $x=0$ is a singularity  

**if have only $\pm \infty$ as singularities**

if can, try to solve the integral analytically, then plug in bounds  
when plugging, remember that the THE SINGULARITY IS A LIMIT, so can use de L'Hopital and inifinitum hierarchy to solve I.F.   
this gives <u>real value of integral, not just character</u>      

### Conclusion

**Conclusion**  

compute the integral on each interval, total integral can be:  

- convergent: only if all parts are convergent  
- divergent: only if all divergent parts go the same infinity (either + or -)  
- indetermined: if diverges at both infinities, so get $\infty - \infty = I.F.$

### Can be rationalized

#### P-test

to do it need the form $\frac{1}{x^p}$  
there can be numbers multiplied to it, but NOTHING ELSE WHERE IT APPEARS AN X  
if it does, can't use p-test, try another test   

can use even if p is a function too:  
$\frac{1}{x^{-ln(x)}} \quad$ so $ \quad p=- \ln x$  

**P-test**

changes depending on the singularity i'm studying  

when converge: 
- $x \sim \infty: \quad p>1$     
like series (bc infinity is the only singularity of series)  

- $x \sim 0 : \quad p <1$  
per esclusione, the other kind  

- other singularities that break the C.E.  follow same rule as $x=0$  
bc the important thing is if the denominator goes to 0 or $ \infty$  
eg: $\frac{1}{(x-1)^p}$ when  $ x \sim 1$:  follow same rule for $x \sim 0$  

**Unsure case with p-test**

if $p=1$ i'm unsure, with all singularities  
check by hand this case  
depends what it's multiplied with, if it's big enough diverges  
if it's not multiplied by anything else, it's the harmonic series, it diverges  


**Unsure case with logarithm**  

if $x \sim \infty$  

$\frac{1}{x \ln^\alpha (x)}$ converge if $\alpha > 1$    
$\frac{1}{ln^\alpha (x)}$ never converge (bc it's bigger than $\frac{1}{x}$, which diverge)    

**If can't do p-test: root test**

root test can only be done at $x \sim \infty$, bc it's the one for series  
use it if have some func of x that can't rationalize  

$x e^{-x}, \quad x \sim \infty$  
there is $e^x$ which prevents me from getting form appropriate for p-test  

use root test:  
$\sqrt[x]{x e^{-x}} = 1 \: e^{-1} =l$  
since $l<1$, the integral converge (when $x \sim \infty$)

##### Dirichlet test

**What it is**

to study improper integral convergence ONLY AT $x \sim + \infty$  
only works at infinity bc it's the same test for infinite series   

  
$\int _a^{\infty}f \; g \; dx$  

conditions for convergence:  
1. f is monotonic  
2. f is infinitesimal  
3. $G = \int g$ is uniformly bounded  
means $\exists M : \forall x |G(x)| \le M$  

combining 1 and 2, means f is either positive and decreasing, or negative and increasing    

**How to use Dirichlet test**

useful if have sin or cos multiplied by something else, like:  
$\int sin(x) \: x^{\alpha}$   
    
- g: pick sin or cos, since they are bounded, so condition 3 is satisfied  
- f: the other thing, just check it's decreasing and infinitesimal (like Leibniz test)  

**Attention: argument of sin**

the var in the argument of sin must be something linear: $x$, or $\pi x$...  
if not, use var change to make it $x$  

example:  
$\int sin(x^\beta) \: x^{\alpha} \rightarrow \int sin(y) \: y^\frac{\beta - \alpha +1}{\alpha}$ 

##### Absolute convergence

**Remember!**

absolute convergence uses upper bound, so it's an if ... then condition, not an if and only if  
so must prove that, if condition is not satisfied, then it don't converge  
for conditional no need to do it, those are if and only if conditions  

**Absolute vs conditional convergence**

- conditional: $\int f \; dx$ converge  
- absolute: $\int |f| \;  dx$ converge  

like in series:    
if absolute convergence, certainly conditional  
but not necessarily the opposite  

**How to study absolute convergence**

steps:  

1. put modulo in the function inside the integral  

2. find bigger function  
eg. can substitute $|\sin x | \le 1$

### Tricky stuff

#### Logarithm

**Singularities**

if it is at denominator, $\frac{1}{ln(x)}$ ha discontinuities in $x = 0, 1, \infty$  

if it is at numerator, $\ln (x)$ has discontinuities in $x=0, \infty$  
but not at x=1, it would simply give 0, which is fine at numerator  

**Common approx**  


- if $x \sim 1: \quad \ln (x) = \ln(1+(x-1)) = x-1$  

- if $x \sim \infty : \quad \ln (x) < x^a  \; \forall a$  
eg: $\frac{1}{\ln x} > \frac{1}{x}, \quad \ln x < x^{-\frac{1}{2}}$  

- if $x \sim 0 : \quad |ln(x)| \le \frac{1}{x^a} \, \forall a\quad$   

log scales slower than any power  
so when $x \sim 0$ or $x \sim \infty$, which would give $\ln = \pm \infty$, it scales slower than any power  
can even make an exponent to log (positive), inequality is still true  

$\int _0^1 |ln(x)| ^\alpha$  

- if $x \sim 0$  

if $\alpha \le 0$ no prob, f becomes $\frac{1}{\infty} = 0$  
if $\alpha > 0$  

TAYLOR: $|ln(x)|^{\alpha} \le \frac{1}{\sqrt{x}}$  
P-TEST: always converge  

- if $x \sim 1$  

TAYLOR:  $\int (1-x)^\alpha$  
P-TEST: converge if $\alpha > -1$  

technically, this integral is something i can integrate analytically, so do it and then plug in bounds  
when plugging in bounds remember it's a limit, so can use de l'Hopital or keep the thing that scales faster  

### Theoretical notions around convergence

**Nth term test**

convergence of integral when $x \sim \infty$ is similar to convergence of series, so:    
if f(x), the integrand function, is positive and not infinitesimal, the integral can't converge  
essentially nth term test  

**No singularity**

intervals without singularity are convergent, bc:    
no $ x \sim \infty$ -> interval is bounded  
no singularity that breaks C.E -> f is continuous  

f continuous on a bounded interval -> the integral is convergent  


**Common divergent integrals**

$\int \frac{e^x}{x}dx \quad x \sim 0$  
$\int \frac{1}{\ln x}dx \quad x \sim 0$  
rhey're the same, do $y=\ln x$

always converge:  
$\int _0 ^\infty e^{-x^2} dx \quad$  (gaussian)  
$\int _0^\infty x^2 e^{-x^2} dx \quad$  

**Using Gamma function to solve**

formulas:  
$\Gamma (n) = \int _0 ^\infty x^{n-1} e^{-x} dx = (n-1)!$  
so: $\quad \Gamma (n+1) = n!$  

another common form it's found in is:  
$\int _0^\infty x^n \; e^{-\lambda x} \; dx = \frac{n!}{\lambda^{n+1}}$  
get to normal one by substitution $y=\lambda x$

**How to find Gamma function**

$\int \frac{\ln x}{x^p} \, dx$  
use substitution: $x=e^{-y}$, so $\ln x = -y$ and $x^{-p}=e^{py}$

$\int \frac{1}{(1+x)^s} \; dx$  
substitute: $x=\frac{y}{1-y}$
