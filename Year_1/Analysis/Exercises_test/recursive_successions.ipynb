### Theory

they define a discrete dynamic system (bc has a fixed number of states, evolves between them)  

**Types**

first-order: $a_n = f(a_{n-1})$  
each term depends on the previous one  

second-order: $a_n = f(a_{n-1}, a_{n-2})$    
each term depends on the 2 previous ones (eg. Fibonacci)  

autonomous: depends on $a_n$, not n itself  
(eg:  $a_{n+1} = n a_n$ is not automous)

**Fixed point**

$f(x) = x$  
also written $a_{n+1} = a_n = x$

**Invariant set**  

$f(A) \subset A$  
starting from there get mapped again in there  

**Crescent succession**  

$f(x) \ge x$

#### Find invariant intervals 

**What to do**

split the R line in invariant intervals   
or in intervals mapped wholly in an invariant interval  

**Condition for invariant interval**

$I=[a,b]$ invariant if:  $\quad f(a), f(b) \in I$   

works if I is monotonic (since it's monotonic, if extremities are in it, so are the values in the middle)  
so MUST PICK AN INTERVAL WITH THE SAME ARROW IN CONDITION 3

**If interval is not invariant**

assume $I_1=[a,b]$ gets mapped to $I_2 = [f(a), f(b)]$    (might be reversed, $[f(b), f(a)]$, put the smaller first)   

if $I_2$ is wholly contained in another interval, follow what happens in that interval  

if $I_2$ stretches across multiple interval, must check what subintervals of $I_2$ are sent in what interval  
for each point $\gamma$ separaring 2 intervals contained in $I_2$, must find point $x$ in $I_1$ such that $f(x)= \gamma$ (x è unico per il teorema dei valori medi e per la stretta monotonia di f)  
so check where the intervals $[a, \gamma], [\gamma, b]$ are mapped, now each should be wholly in one interval  

#### Analyze data

**Bounded vs convergent**

NOT NECESSARILY CONVERGE, might be bounded (limitata) but not convergent  

limited/bounded: means at some point it lands in an invariant CLOSED interval  
so not an interval with $\infty$ as an extremity  

not convergent: means in the (invariant) interval it landed in there is a limit point, but can't arrive there bc it's repulsive  

**Prove it's convergent**

converge if the sequence is:  
- monotonic (sequence, so f(x)>x or opposite sign, condition 2)  
- bounded (so inside invariant interval)   

so if $I$ is invariant, converge to extremity of $I$ indicated by arrow of CONDITION 2 (if it's a fixed point)  
if at that extemity there is $\pm \infty$, it diverges  

to be sure can check that the limit points is attractive, but not really needed???  

**Analyze data: shortcut**

- if BOTH ARROWS POINT RIGHT (NOT LEFT) (arrows of conditions 2 and 3)  
(and in that direction the extremity is a fixed point):   
this is proof that the interval is invariant, converge to right extremity  

- if in interval with extremity $+\infty$ is crescent, it's convergent, no need to check condition 3 or if the interval is invariant  

but don't need these shortcuts anyway, can just study this interval like normal???    


**Recap**

to check an interval is invariant, must pick intervals where arrows of condition 3 point all in same direction

#### Proofs stuff is invariant

what | definition 
---|---
crescent succession | $f(x)>x \quad \forall x $  
crescent function (monotonic) | $a<x<b \quad \rightarrow \quad f(a)<f(x)<f(b) \quad \forall x$  
invariant interval | $a<f(x)<b \quad \forall x \in I=[a,b]$ 

**General invariant**

pick a monotonic interval (crescent function)  
from the definition of monotonic it's clear that:  
if find $f(a),f(b) \in I$, also $f(x) \in I \quad \forall x \in I$

**Interval where point arrows point right**

the succession being crescent is the condition: $f(x)>x$  
and b is a fixed point, so: $f(b)=b$  

plug it into monotonic condition: $a < f(a) <f(x) <f(b) = b$  
got condition of being invariant  

**Prove interval with infinity is invariant**

assume the succession is crescent with $+\infty$, decrescent with $-\infty$ is the same  

succession crescent: $f(x)>x$  
nothing can be bigger than $\infty: f(x) < \infty$  
so: $a<x<f(x)< \infty \quad \forall x \in I$

### Does series converge?

cannot just plug $a_n =x$, must see how much each term scales  
must extimate asymptotic beahviour of $a_n$  

**Steps**

1. find limit of the recursion (with the initial parameter i'm told to use)    
want a limit points, one of the x when $lim(x)=x$  
to know which solve the recursion, see for that initial parameter to what it converge  
if limit is 0, probably will use Taylor approx  
eg: $a_n \rightarrow 0$  

2. linearize  
must get form like: $a_{n+1}=a_n + ...$  
do it by doing a Taylor approximation until the term after $a_{n}$  
$f(x)=x(c)+f'(c)(x-c)$ where c is the limit of the recursion $a_n \rightarrow c$  

3. swap the the subtraction with the derivative  
$a_{n+1}-a_n = \frac{da}{dn}$  

4. solve the diff eq to find $a_n$, how much it scales with n  
after solving it can use asymptotic behaviour  


**Example 1**
 
$a_{n+1}=sin(a_n)$  

1. find limit: $a_n \rightarrow 0$  
2. linearize: $a_{n+1} \sim a_n -\frac{a_n^3}{6}$   
$a_{n+1}-a_n \sim -  \frac{a_n^3}{6}$  
3. swap: $\frac{da}{dn}  \sim -  \frac{a_n^3}{6}$  
4. solve: $a_n = \frac{C}{\sqrt{n}}$  

**Example 2**

$a_{n+1} = \frac{a_n}{1+a_n}$  
1. find limit: $a_n \rightarrow 0$  
2. linearize: $a_{n+1} \sim a_n -a_n^2$  
$a_{n+1} \sim a_n -a_n^2$  
3. swap: $\frac{da}{dn}= -a_n^2$  
4. solve: $a_n \sim \frac{1}{n}$
