**Recap: spaces**

- closed: Whenever a sequence $x_n \in K$ converges to a point $x_n \rightarrow x$, then $x∈K$  
all sequences in K that converge have limit in K  

- compact: Every sequence in K has a convergent subsequence whose limit is also in K (stronger, force the subsequence to converge)  

closed: IF converge, converge inside  
compact: the subsequence MUST converge and do it inside  


- bounded: exist M such that $|x_n| \le M$  
by Bolzano-Weierstrass, every sequence has convergent subsequence (don't know where it converges though, might converge outside of the space unless it is closed, eg might converge at extremity)

- complete: all Cauchy sequences are convergent (and converge inside it)

(0,1): bounded, not closed  
$[0, \infty)$: closed, not bounded (closed means limits of sequences can't escape the set, so the open parenthesis at infinity is fine, bc if limit is at infinity it's still in the set, cannot escape it)  
[0,1]: compact, bounded+closed  

#### Unconnected space

**Definition**

unconnected: if can pick a partition of it made by two subsets (not the banal one, itself and null space), and are both open or both close  

connected: if this is not possible

**Properties**

if T is connected and f is continuous, f(T) is connected  

#### Compact space

**Definition**

3 equivalent ones:  

1. every open cover of it has a finite subcover  
ogni ricoprimento aperto di K ammette un sotto-ricoprimento finito

2. it's closed and bounded   

3. every sequence in K has a sequece that converges in K  

**Explanation def 1**

it means that, when trying to cover it with intervals, only need a finite number of intervals to cover it  

so must be:  
bounded: bc if go to infinity will need infinite intervals  
closed: bc if it's open like (a,b) will need infinite intervals to get close to a, bc can't just put big interval until a, since a don't exist, but must keep piling intervals at the ends  

**Properties**

1. f continuous defined on compact space is:  

    - bounded: inf and sup are not infinity and appear in the dominium (Weierstrass th)   
    (if set is open it's not sure, max could happen at boundary of the interval, which is not part of the dominium)  

    - uniformly continuous (Heine-Cantor th)

2. if T compact, f continuous, f(T) compact  
maps compact set to compact set, so doesn't break connected parts 

3. from any succession of points can extract a convergent subsuccession
