### Discontinuities

**Discontinuities**  



1. jump: left $\neq$ right    
both must exist, be real numbers (no infinite), and be different


$lim_{x \rightarrow x_0^-} f(x) \neq lim_{x \rightarrow x_0^+} f(x)$   


2. second species: either left or right limit don't exist or blow up to infinity   


3. removable (eliminabile): lim is not the actual value of func there  
actual value there can be a specific one or func can be not defined there, bc of C.E.  
limit gets a value bc, as a limit, can compute it using Taylor or de l'Hopital 

$ \exists lim_{x \rightarrow x_0} f(x) = l \neq f(x_0)$

### Neighborhood

**Neighborhood**

neighborhood (intorno) of:  
- a point $x_0$: interval that contains that point (essentially a ball centered in that point)  
- $\pm \infty$: semiretta towards $\pm \infty$, can start wherever  

formally  
$U \subseteq R$ is called neighborhood of:  
- the point $x_0$ if U contains an interval that contains $x_0$
- $\infty$ if U contains a semirect towards $\infty$

**Point of accumulation**

a point to which i can get as close as i want (no matter how much close, there will always be another point at less than that distance)  
formally: every neighborhood of that point contains at least another point on top of the point itself  

in R, every point is a point of accumulation    
in Z, none are, bc they're all isolated, it's not dense  

derivated set of a metric space: set of all points of that metric space which are accumulation points  


if +inf is accumulation point of an interval, means that interval has no upper bound 


### Writing with neighborhoods

**Continuity**

$f:X \rightarrow Y$  
$f$ is continuous if $ \quad \forall B$ neighborhood of $f(x_0) \quad \exists A $ neighborhood of $x_0 : f(A) \subseteq B$  

**Limit existance**

$lim _{x \rightarrow x_0} f(x) = l \quad$  if:  
$ \forall B$ neighborhood of $l \quad \exists A $ neighborhood of $x_0: \quad f(A∖\{x_0\}) \subseteq B$  

**Existance of limit Vs continuity**

conditions for continuity and existance of limit at a point $x_0$:  

continuity: $\quad f(A) = B$  
limit:  $\quad f(A∖\{x_0\}) = B$  

bc in limit don't care about what happens at the exact point, only around  

**Intuitively**

B: interval in y axis, around $f(x_0)$  
A: interval in x axis, around $x_0$  
for every interval B, can choose an interval A such that all points in A map in points of B  

**Intuitive**

tells what happens if i get close to x  
no matter the actual value f(x), could even not exist (be undefined), only care what happens around it, when approaching it  

**Darboux function**

$f : I \rightarrow R$, is a Darboux function if $f(J)$ is an interval $\forall J \subset I$   


continuous $\rightarrow$ Darboux  
but not necessarily the opposite, example is $sin(\frac{1}{x})$


#### Theorems for continuous functions

**Sign permanency**

if $x_0 >0$, i can take a neighborhood  of $x_0$ where all points are positive (just need to take it small enough)  

**Th of zeroes**

if $f(a)\cdot f(b)<0$, in interval [a,b] there is a point where f is zero  
bc f(a) and f(b) have opposite sign  

**Intermediate values**

$f: I \rightarrow R$  
f monotone, assumes every value in interval (a,b) $\longrightarrow$ f is continuous  
$f$ might be still continuous even if not monotone, but i'm not certain just bc assumes every value in interval (might have some jumps)  

f continuous $\longrightarrow$ f assumes all values between its max and min
