## Limit and sup

$\lim sup \: f(x) = max \lim f(x)$  
$\lim inf \: f(x) = min \lim f(x)$  

highest/smallest (sup/inf) value that f(x) approaches INFINITELY OFTEN (it must be reached an infinite number of times)  
infinitely often bc must take limit for $x \rightarrow \infty$, so cannot just happen once, must happen at infinity  

**Example lim sup**

exists even if function don't converge, and obscillates  
eg:  
$f(x) = sin (x), \quad lim \: inf = -1, \quad lim \: sup = 1$   
$f(x) =\frac{1}{x}, \quad lim \: inf = lim \: sup = 0\quad$ bc 0 is the only value it reaches infinite times, when $x \rightarrow \infty$   

**Example sup vs lim sup**

$f(x)=\begin{cases}
1 & x \, odd  \\
n & x \, even
\end{cases}$  

$sup=\infty$: the highest number it reaches  
$\lim \sup = 1$  the highest number it reaches INFINITELY OFTEN (only 1 is reached infinite times)  

**Th**

$ \lim inf \: f \le \lim sup \: f \quad$  is always true  
if it's $=$, not $<$, $f$ has limit, it's this value  

just substitute the stuff above to prove it

### Succession of arithmetic mean

**Definition**

$a_n$ is a succession  
$b_n =  \frac{a_0 + ... + a_{n-1}}{n}$ succession of the arithmetic means  

**Th**

$max \lim b_n  \le max \lim a_n$  (lim to +inf)   


explanation  

if $a_n$ is convergent, $b_n$ is convergent to same limit  
if $a_n$ is not convergent, $b_n$ might still be convergent (eg. $a_n = +1, -1, +1, ... \quad b_n = 1,0, \frac{1}{3}, 0, ... $ is convergent to 0)

intuitively:  
if $a$ reaches something, $b$ don't actually reach it as well, bc must take the average between that and the values before, that were smaller  

**Proof**

set $l = max \lim a_n$  
obv if $l = + \infty$, the demonstration is over, else:  

taken $ \epsilon >0, \exists N :\quad \forall n \ge N,  \quad a_n \le l + \epsilon$  

$\forall n \ge N: b_n =  \frac{a_0 + ... + a_{n-1}}{n} =  \frac{a_0 + ... + a_{N-1}}{n} +  \frac{a_{N} + ... + a_{n-1}}{n}  \le  \frac{a_0 + ... + a_{N-1}}{n} + \frac{n- N}{n}(l + \epsilon) = c_n$  

since from a point onward $b_n \le c_n$, $\quad max \lim b_n \le max \lim c_n = l +\epsilon$  

therefore $max \lim b_n \le l + \epsilon$  
since can take epsilon whatever i want, can take it null:  $max \lim b_n \le l$, which is the thesis  

**Corollary**

if $a_n$ has limit (finite or not), $b_n$, the succession of the means has limit equal to it    

**Proof**

assume $\quad l = \lim a_n$  
then $\quad max \lim b_n \le l, \quad min \lim b_n \ge l$  
bc of the theorem above  

since must always be:  
 $min \lim f(x) \le \lim l \le  max \lim f(x)$:  
$\lim b_n =l$
