$O(g) = \{f \in R^A: \quad \lim sup _{x \rightarrow x_0} |\frac{f(x)}{g(x)}|< \infty\}$   
so could mean any function smaller OR EQUAL to g  

eg: $f(x)=2x, g(x)=x$  
$\frac{f(x)}{g(x)} = 2$, so $f \in O(g)$
