**Conditioned convergence**

$a_n$ is a sequence  
if $\Sigma a_n$ converge conditionally but not absolutely:  
can change the order of terms in $a_n$ to make it converge to any number i want  

**Proof**

note that $a_n$ has no all positive terms, else if converge conditionally would also converge absolutely  
so split positive and negative terms  
imagine want to make it converg to $c$:  
add positive terms until go above $c$  
then add negative terms until go below $c$  
then repeat adding positive terms ...  

since $a_n$ converge conditionally, means the terms keep getting smaller, so i keep getting closer to $c$
