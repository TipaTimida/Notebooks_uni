$\frac{d^2x}{dt^2} = -\frac{k}{m}x $

$\Omega ^2 = \frac{k}{m} $

3. Write general solution:   
$x(t) = A\: cos(\Omega t) + B\: sin (\Omega t)$  

4. Plug initial conds to get A and B:  

$x(0) = A = x_0 \\
\dot{x}(0) = B\Omega = 0 = B$  

Therefore sol is :  
$x(t) = x_0 \: cos(\sqrt{\frac{k}{m}} t)$

**Method 2: Energy conservation**

Start with energy:  
$E = \frac{1}{2}m\dot{x}^2 + \frac{1}{2}kx^2$

Energy conservation means its derivated wrt time is null  
$ \frac{dE}{dt} = \frac{1}{2}m (2\dot{x}) (\ddot{x})  + \frac{1}{2} k (2x) (\dot{x}) = $  
$ = \dot{x} (m \ddot{x} + kx)  = 0 $

**Method 5: hamiltonian flow**

1. Start with differential equation: 

$\frac{d^2x}{dt^2} = -\frac{k}{m}x $ 

2. Replace all velocities with momentum:  

$p= mv = m \frac{dx}{dt} $  
$\frac{dp}{dt} = m \frac{d^2x}{dt^2} = F $  

$\frac{dp}{dt} = -kx $

3. Now i have 2 first order equations:

$\frac{dp}{dt} = -kx $  
$\frac{dx}{dt} = \frac{p}{m} $  

4. Rewrite in matrix form  









$
\frac{d}{dt} 
\begin{bmatrix} x \\ p \end{bmatrix} = \begin{bmatrix} 0 & \frac{1}{m} \\ -k & 0 \end{bmatrix} \begin{bmatrix} x \\ p \end{bmatrix}
$
