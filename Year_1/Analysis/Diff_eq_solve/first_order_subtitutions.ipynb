### First order

#### First order, linear, non-separable

$ y' + p(x)y = g(x) $


steps:  

1. put it in standard form ($y'$ has coeff 1)

2. computing integrating factor: $ \mu (x) = e^{\int p(x) dx} $  
(this don't have the +c)  

3. multiply both sides by $\mu (x)$, so on left find $[\mu (x) y(x)]'$

4. integrate (remember +c, jus put one on right side)  

5. rearrange  

Solution:  

$ y(x) = \frac{ {\int \mu (x) g(x) dx}}{\mu (x)} $

**Example**

$2x y' - y = x^2$

1. Standard form  

$y' - \frac{1}{2x}y = \frac{x}{2}$

2. Integrating factor  
 
$\int p(x) dx = \int - \frac{1}{2x} dx = -\frac{1}{2} ln(x)$  
$ \mu(x) = e^{-\frac{1}{2} ln(x)} = \frac{1}{\sqrt{x}}$  


3. Multiply  

$\frac{1}{\sqrt{x}} y' - \frac{1}{\sqrt{x}}  \frac{1}{2x}y =  \frac{1}{\sqrt{x}}  \frac{x}{2} $  
$ (yx^{−\frac{1}{2}})' = \frac{1}{\sqrt{x}}  \frac{x}{2} $  

4. integrate  

$ yx^{−\frac{1}{2}} = \int \frac{1}{\sqrt{x}}  \frac{x}{2} \; dx$



**Notes**

in the step 4 integration will get +c  
then, if i manipulate the expression to isolate y, c is part of the expression  
eg: if i divide by, get $\frac{c}{x}$  

then g in initial conditions to find c  

#### First order, non-linear, separable

$ N(y) dy = M(x) dx $


steps: 

1. write  $y' = \frac{dy}{dx}$  

2. split stuff with x and y on different sides, with the proper dx or dy    
dx and dy must be split so both are at numerator  

3. integrate (remember +c)

#### First order, exact

$M(x,y) + N(x,y)y' = 0$


to remember method, rewrite as:  
$M \; dx = -N \; dy$  

pick a side and integrate  
then derivate wrt the other var, should obtain the other (if started with N, obtain M)  

$M_y = \frac{\partial{M}}{\partial{y}}$  
it's notation for partial derivatives  

Check sol exists:
-  $M_y = N_x$   

Solve:  
1. $\psi (x,y) = \int M dx \quad $  or   $ \quad \psi (x,y) = \int N dy$  
add $+c(y)$ in first case, $+c(x)$ in second  

2. $\psi _y = N \quad$ or $ \quad \psi _x = M $   
write this equality to find c, which must be a func of only the var in its brackets  

3. write full: $\psi (x,y) = k $ and solve for y  
is k a constant, like +c in another methods??? 

**Example**

$2xy^2 +4 + 2(x^2y-3)y' = 0$  
$M = 2xy^2 +4  \quad N = 2(x^2y-3)$  

0. $M_y = N_x = 4xy$, there is sol  

1. $\psi = \int M dx = x^2y^2 + 4x + c(y) $ 

2. $\psi _y = 2yx^2 + c'(y) = N$  
$c'(y) = -6 \quad$, so $\quad c(y) = -6y$

3. $k = x^2y^2 +4x -6y$

**Why works** 
  
find:  
$\psi _x = M, \quad \psi _y = N \quad$ such that $\psi _x + \psi _y y' = 0$  

so:  
$\frac{d}{dx}\psi (x, y(x)) = 0$  

therefore, the solution is a constant:  
$\psi(x,y) = k$

#### First order, Newton equation

$y'' = f(y)$

**Explanation**

$(\frac{y'^2}{2})' = y'' y' = f(y)y' = F'(y)$  
where $F(y)$ is the primitive of $f(y)$, so by chain rule the two sides are equal  

$y'^2 = 2(F(y)+c)$  
$F(y)+c = \int _{y_0}^y f(y) dy$  
$y_0$ and $y$ are vars, must stay there in the answer  

therefore:    
$y' = \sqrt{2(F(y)+c)}$  

now solve with separation:  
$\int \frac{dy}{\sqrt{2(F(y)+c)}} = \int dt$  
note that the integration bounds must be chosen the same for both sides  
this is often used to compute period of harmonic motion  

**Solutions**

cases:  
$2F(y)+c = 0: y'=0, \quad$ y(x)$ is a stationary sol, a const       
$2F(y)+c \neq 0: \quad G(y) = \int \frac{dy}{\sqrt{2F(y)+c_2}}$
