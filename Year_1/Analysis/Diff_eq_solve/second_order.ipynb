### Theory

**Space and base of solutions**

solutions of a differential eq. of order n form a base for a vector space of dimension n  
(if have non-homo equation, it's an affine space)  
so need n independent solutions to find a base for such space  
such base is the called the solution base, and is said they are a FUNDAMENTAL SET OF SOLUTIONS    

solution base must have as many solutions as the order of the equation  
so first order has 1 solution, second order has 2 

**How to write solution**

example:  

base for solution space: $\quad e^x, \; \; xe^x$  
general solution: $y(x) = Ae^x + Bxe^x$   

**Principle of superimposition**

if they form a fundamental set can put them together to get general solution  
$y = Ay_1 + B y_2$  
A and B are found with initial conditions  

so: solution of second order diff eq. is sum of two solutions belonging to the solution base

### What is a solution made of  

split in homogeneous and particular  

$y_g = y_H + y_p$  

**Depending on non-homo part**

if non-homo term is a SUM, can split in two differnt diff eq, each with a different addend  
find the particular eq separately and then add them  

eg: $y'' -2y' +y = x + \sin(x)$  
$\begin{cases} 
y'' -2y' +y = x  & \rightarrow  y_{p1} \\
y'' -2y' +y = \sin(x) & \rightarrow y_{p2} \\
\end{cases}$  

so final sol is:  
$y_g = y_H + y_{p1}+y_{p2}$  
the two particular solutions can be found using different methods, they're 2 separate problems  

### Second order, homogeneous

#### Second order, constant coefficient

$ay'' + by' +cy = 0$  


steps:  

1. find characteristic poly (incognita is $\lambda$)  
2. compute roots of characteristic poly  
3. pick appropriate solutions for the solution base and add them all

this method can be used even if it's higher than second order, always think to plug in as a guess $y=e^{\lambda x}$  

**Solutions for second order**

- $\Delta >0: y = Ae^{\lambda _1 x}+ Be^{\lambda _2 x}$
- $\Delta = 0: y = (A+Bx)e^{\lambda x} $
- $\Delta <0: \lambda = \alpha \pm i\beta, \quad y = e^{\alpha x}(c_1e^{\beta t}+c_2 e^{-i \beta t}) = e^{\alpha x}(A \cos(\beta x)+ B \sin(\beta x)) = Ae^{\alpha x}A \cos(\beta x+\phi)$ 



example:  
$\lambda = 1 \pm \sqrt{c-1}$  
if $c<1$, sol is: $\lambda = 1 \pm i\sqrt{1-c}$

tip: write $y(x) = Ae^{r_1x} + Be^{r_2x}$ in all cases  
make it easier to solve stuff like for what param her eis solution  

**Find characteristic poly of differential eq**
 
1. bring everthing on left member  
2.  plug: $y'' =\lambda ^2, \quad y' =\lambda, \quad y = 1$   

examples:  
$y'' = c y \quad \rightarrow \quad p(\lambda) = \lambda ^2- c$  
$ay'' + by' +cy = 0 \quad  \rightarrow \quad p(\lambda) = a\lambda ^2 + b \lambda +c = 0$  

(formally, must plug in $ y = e^{\lambda x}$)

### Second-order, non-homogeneous

**How to solve non-homo**

1. solve homogeneous associate  
find $y_h = A y_1 + B y_2$, putting $g(x)=0$  
DO NOT compute A and B yet  

2. find $y_p$, particular solution  
here, do compute A and B for this one, mustn't have any undetermined coefficients  

3. the general solution, bc of the principle of superimposition???, is:  
$y = y_h + y_p$  

4. NOW plug in initial conditions and COMPUTE A AND B of the homogeneous sol  

**Find particular solution: which method?**

2 methods:  

1. similarity method  
can be used only in specific cases  
bases guess on $g(x)$, but it can only contain exponential, power, sin or cos, no hard stuff like logarithm or fraction   
if can do this it's easier, bc A,B are constants, not func of x  

2. variation of constants method  
can always be used, but longer  
bases guess on homogeneous solution  

steps are the same for both methods:  

1. make a guess (this depends on method), will be $y_p$  
2. derivate to get $y'_p, y_p''$  
3. plug $y_p, y_p', y_p''$ in the diff eq i have  
4. find values for the unknown parts in the guess  

#### Similarity method

**How to make guess**

based on form of $g(x)$  
A,B are constants  

**Similarity method: guesses**

$g(x)$ | $y_p$   
--- | ---
$e^{\lambda x}$ | $ A \; e^{\lambda x}$     
$a \cos(\beta t) + b \sin(\beta t) $ | $ A \cos(\beta t) + B \sin(\beta t)$  
$n^{th}$ degree polynomial | $ A_nx^n + \ldots + A_0$

**Examples**

$g(x)$ | $y_p$   
--- | ---
$x$ | $ A+Bx$     
$x \sin (2x)$ | $(Ax+B)\sin (2x) + (Cx + D) \cos (2x)$   
$4e^{3x}$ | $Ae^{3x}$
$e^x \cos x$ | $e^x (A \cos x + B \sin x )$  


**Notes**

in trig case, need to put both sin and cos in my guess, even if g(x) contains only one of them  

**If my guess is already a solution of the homogeneous**

it means that guess is alrady in the solution base, so it's not independent  
so cannot use it again  
instead use the next one available, obtained by multiplying by x until i get a new one, which is independent  

eg:  
solution base of homogeneous: $e^x, xe^x$  
$g(x)=e^x$  
so want to guess $y_p = A e^x$, but is already in solution abse of homogeneous  
instead guess $y_p = Ax^2 e^x$  

if can't use $A \cos x + B \sin x$ will use $x(A \cos x + B \sin x)$   


**Harder guesses**

if $g(x)$ is a SUM:  
split propblem in two, then add the particular sols  
or take as a guess the sum of the guesses  

if $g(x)$ is a MULTIPLICATION:  
my guess is the product of the guesses of the factors

**Examples: $g(x)$ is a multiplication**  


$g(x) = (9x^2−3x)\cos x \quad \mapsto \quad  y_p = (Ax^2+Bx+C)(D \cos x+E \sin x)$  

$g(x) = 16e^{7x} \sin(10x) \quad \mapsto \quad  y_p = e^{7x} (A \cos(10 x)+B \sin(10x))$  

tip:  
multiplication of 2 coefficients is 1 new coefficient  
at first put all coefficients needed, two when 2 combine turn them into 1 with a new name  


so technically in the second example the exponential should have a coefficient C multiplied to it    
but then would become CA and CB, and both would bceome a new coefficient anyway  
so just call those A and B  


#### Variation of constants

**What i need to do**

Take homo solution:  
$y_1, y_2$: base for the homogeneous solution  
$y_h = Ay_1 + By_2$  

swap A and B with $a(x)$ and $b(x)$, unkown functions of x (just write a and b, rem they're functions)  
$y_p = ay_1 +by_2$

note that when derivating $y_p$, will have derivative of product, since a and b are funcs of x  

**Write system**

$\begin{cases}
a'y_1+b'y_2 = 0 \\
a'y_1' + b'y_2' = g(x)
\end{cases}$  

i want to find a and b  
so first solve for a' and b' (treat it like a norm var)  
then integrate to find a and b  

so my sol will be:  
$y_p = ay_1 +by_2$

##### Wronskian

**Alternative method with Wronskian**

$y_p = -y_1 \int \frac{y_2 \; g(x)}{W} dx + y_2 \int \frac{y_1 \; g(x)}{W} dx$  

first compute the wronskian $W(y_1, y_2)$ separatly, then plug it in    
even if not written in full formula, it takes in input $y_1$ and $y_2$  

when computing the integrals DO NOT add the +c  



**Wronskian**


$W = det \begin{vmatrix}
y_1 & y_2 \\
y_1' & y_2'
\end{vmatrix} \quad$

**Example Wronskian**

$y_1 = \cos (3x), y_2 = \sin (3x)$  

$W = det \begin{bmatrix} \cos (3x) & \sin (3x) \\ -3 \sin (3x) & 3 \cos (3x) \end{bmatrix} = 3 \cos^2 (3x) + 3 \sin ^2 (3x) = 3$  

**When 2 solutions are independent**

2 sols $y_1$ and $y_2$ are indipendent if:  
$W \neq 0 \quad \forall x$  
rem x gets plugged in y(x)  

**How to find Wronskian sol**

write the system from variation of coefficients in a matrix:  

$\begin{bmatrix} y_1 & y_2 \\ y_1' & y_2' \end{bmatrix} \begin{bmatrix} a' \\ b' \end{bmatrix} = \begin{bmatrix} 0 \\ g(x) \end{bmatrix}$  

$\begin{bmatrix} a' \\ b' \end{bmatrix} = \begin{bmatrix} y_1 & y_2 \\ y_1' & y_2' \end{bmatrix} ^{-1} \begin{bmatrix} 0 \\ g(x) \end{bmatrix} = \frac{1}{W} \begin{bmatrix} y_2' & -y_2 \\ -y_1' & y_1 \end{bmatrix}$

$a' = - \frac{_2g(x)}{W} \quad b' = \frac{y_1g(x)}{W}$  

$y_p = ay_1 +by_2$, so just integrate a' and b'

### Higher order, constant coefficients  

steps:  
1. find associated poly  
2. compute roots, pay attention to algebric multiplicity  
3. pick solutions for solution base accordingly  
4. add together all solution (th of superimposition) to get solution for the homogeneous  

**Find solutions from roots of associated poly**  

for each root:  
1. check if real or complex  
2. pick as many sols as the multiplicity of the root, in order  

- $\lambda$ real root  
 
$e^{\lambda x}, x e^{\lambda x}, x^2 e^{\lambda x} ...$  

- $\lambda = \alpha \pm i\beta$ complex root 

$e^{\alpha x} \cos (\beta x), e^{\alpha x} \sin(\beta x), xe^{\alpha x} \cos (\beta x), xe^{\alpha x} \sin(\beta x), x^2 e^{\alpha x} \cos (\beta x) ...$  

**Why do i get a trig?**  

$e^{\lambda x} = e^{(\alpha + i \beta)x} = e^{\alpha x} e^{i \beta x} = e^{\alpha x} \cos \beta x$  

$e^{i\beta x} = \cos \beta x + i \sin \beta x$, but only want the real part ???   



**Solutions from associated poly: notes**

sols continue by increasing the power of x outside  

COMPLEX ROOTS:  
if $\alpha + i \beta$ is a solution, so is $\alpha - i \beta$  bc the braket is $(\lambda ^2 + a)$, so conjugates go together    
so multiplicity of the root $\alpha \pm i\beta$ is sum of multiplicities of + and - roots, which are the same, so multiplicity is always even  
for ease don't split brackets: $(\lambda -1)^2 = (\lambda -i)(\lambda +i)$, leave it as the first form, makes it easier to see multiplicity  

examples:  
$(\lambda ^2 + a)$: multiplicity 2  
$(\lambda ^2 + a)^2$: multiplicity 4  



when have real root of multiplicity 2, i usually write solution as:  
$Ae^{\lambda x} + Bx e^{\lambda x} = e^{\lambda x}(A + Bx)$
