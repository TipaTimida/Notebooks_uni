## Definition of integral

### Partition, suddivision

**suddivision**: $D=\{x_0, x_1... \}$  
set of points that act as extremes for the intervals  

**partition**: $P=[a, x_0], [x_0, x_1], ..., $  
set of intervals created  

for proofs, generally use suddivision, not partition (like in X and D)   

**Properties with upper/lower sum and suddivisions**

$X \subset D \quad$ means D is more fine than X, D has more points than X   
means all points in X are also in D, but D then has some more  


$S^*(D) \le S^*(X)$  
$S_*(D) \ge S_*(X)$  

the more fine the partition, the closer is the sum to the real value of the area  
so the upper bound gets smaller, the lower bound gets bigger

**Proof**
 
prove it when D has 1 point more than X  
in the interval of X where there is a point of D which isn't in X, find inf and sup of both D and X  
then ovb from drawing, see the area  

### Extimate area

**Use rectangoloid**  

in each subinterval of the partition, calculate area of rectangle as $b \cdot h$  
base is length interval, height is height of bar (sup or inf of f in that interval to make rectangle inside or outside f)      


- upper sum $ \quad S^*(f,P) = \Sigma (x_{k-1} - x_k)\; sup \: \;f _{[x_{k-1}, x_k]}$  
- lower sum $\quad S_*(f,P) = \Sigma (x_{k-1} - x_k) \; inf \: \; f _{[x_{k-1}, x_k]}$  

means draw a rectangoloid tangent inside and one tangent outside the func, compute their areas  
essentially i'm approximating the func with a function defined a tratti

**Integrals**

- upper integral $\quad I^*(f) = inf \, S^*(f,P)$  
- lower integral $\quad I_*(f) = sup \, S_*(f,P)$  

is integrable if:  
$I^* = I_* = \int _a^b f \, df$  

keep computing bounds with always finer partitions  
so find the lowest upper bound and the highest lower bound  
if their integrals are the same, f is Riemann-integrable and that is the value of the integral

## Properties of integrals

1. additivity (wrt dominium)  
$\int _a^b f = \int _a^c f + \int _c^b f$   

2. monotony  
$f \le g \longrightarrow \int f < \int g$  

3. linearity (wrt argument of f)  
$\int cf = c\int f$  
$\int f+g = \int f + \int g$   

**Properties**
 
$f: I\rightarrow R$:   

uniform continuous $\rightarrow $ continuous on $I$ $\rightarrow $ Riemann integrable on $I$ $\rightarrow $ bounded on $I$

**Absolute value**  

$|\int f \, dt|  \le \int |f| \, dt$  

bc considering all area as positive gives bigger area    

## Euler Gamma function

$\Gamma (n) = \int _0 ^\infty x^{n-1} e^{-x} dx = (n-1)!$  
so: $\quad \Gamma (n+1) = n!$  

other form to find it in (var change):  
$\Gamma (n) = \int _0 ^\infty x^n e^{-x} dx = n!$  



**What is it**

extends the notion of factorial to non-integer numbers  

**How to find the formula**

pick this function $F(t)$, then compute it and differentiate it wrt the parameter many times to see:  

$F(t) =  \int _0^\infty e^{-tx} dx = \frac{1}{t}$   

$\frac{dF(t)}{dx} =  \int _0^\infty x\; e^{-tx} dx = - \frac{1}{t^2}$   

$\frac{d^nF(t)}{dx^n} =   \int _0^\infty x^n\; e^{-tx} dx = \frac{n!}{t^{n+1}}$  

the last one is the formula i want, just plug t=1    

i'm using Feynman trick: introducing a parameter and derivating wrt it  


**Use it co compute integrals**

can look for it with a var change to prove an integral is convergent, since it's convergent  

$\int _0 ^1 \ln^n (x) \, dx\quad $ let: $\begin{cases} t=- \ln x \\ x=e^{-t} \end{cases}$ 

$\int _0^\infty e^{-\sqrt[n]{x}} dx$ let: $\begin{cases} t= x^n \\ dx=nt^{n-1}dt \end{cases}$ 


## Other theory

### Obscillations

all defined in same interval I:  
$osc_I f = sup_I f- inf_I f$  

**Normal vs puntual obscillation**

both create a ball/interval of radius r and center x  
must pick points x,y to get biggest difference between them (biggest $f(x)-f(y)$)  

normal:  $osc_{x,r} f$  
puntual: $osc_{B(x,r)} f$    

both can pick y however they want in that interval  
but in normal x must be the center of the interval, in puntual x can be chosen at will like y  
so in punctual can pick x to extremize the obscillation and make it bigger  
so:  
$ osc_{x,r} f \le osc_{B(x,r)} f$  

### Abelian integral

Integral in 2 vars:  
$\int f(x, y(x)) dx$  

Algebric curve: a function g such that   
$g(x, y(x)) = 0$  
y(x) is a branch of this algenric curve  

the integral in 2 vars is called abelian integral belonging to the algebric curve

### Sup, inf

**Notation**

must be defined on an interval I  
$sup_{I} f$  
$sup_{x \in I}f(x)$  

**Properties with intervals/sets**

$A \subseteq B$, they're both intervals/sets  

$sup B \ge sup A$  
$inf B \le sup A$  

if has more elements, more prob of finding the extreme

**Properties with funcs**

$sup(f+g) \le sup(f) + sup(g)$  
$inf(f+g) \ge inf(f) + inf(g)$  
they're equal obly if sup(f) and sup(g) are on the same x coords, else the sup(f+g) might be somewhere else, made by adding not-maximum values    


$sup(−f)=−inf(f)$    
$inf(−f)=-sup(f)$
