### How to approach

before starting: 

- check if it's even or odd    
odd: $\int_{-a}^a = 0$    
even: $ \int_{-a}^a = 2 \int _0^a$  

- check C.E.  
if there are singularities inside interval of integration it's indefinite integral must check them, can't just compute    
to compute, split interval and put singularities on extremity  

#### Already rational

1. check for reverse chain rule
2. method of rational functions (after doing it, check again for chain rule)  
3. substitutions  

if denominator of second degree:  
(or also if it's second degree but with a square root on all of it, so technically first degree)  

- numerator of first degree: reverse chain rule (might need to split in two integrals, one of which with a const as numerator) 

- numerator of zero degree: might be arctan (or arcsin or arccos if denominator has square root)        
cerca di completare il quadrato al denominatore  
see coeffs of $x^2$ and $2x$, see what must be the termine noto to make a square  

#### Trigonometric

1. if sin, cos both appear, use half-angle formulas (formula di bisezione)    
2. if only one appears, in a power of 1, and multiplied to a power of x, integration by parts  
3. if sin, cos are multiplied per each other with different arguments, try with addition of angle identities

#### Exponential

1. if only exponential appears, rationalizing substitution
2. if it's multiplied by power of x, integration by parts  
3. if multiplied by something else that repeats itself, like sin or cos, integration by parts, try to reobtain starting one  

#### Logarithm

1. if it's divided by x, try to do reverse chain rule substitution, since 1/x is the derivative of ln x  
2. if it's multiplied by a power of x, integration by parts, since it makes appear an 1/x when derivated   
3. if it's at denominator, and at numerator have a power of something minus 1 (like $x^3 -1$), try Feynman trick  

rem integral of 1/x is ln|x|, with abs value  

#### Roots

1. if roots with different exponents, normal substitution  
eg; if have $\sqrt[4]{x-1}$ put $y = \sqrt[4]{x-1}$ 
 
2. if binomial under square root, trigonometric substitution

### Substitution

let: 
$\begin{cases} 
y = x^2 & \rightarrow \sqrt{y} = x \\ 
dy = 2x \: dx & \rightarrow \frac{dy}{\sqrt{y}} = dx\\ 
\end{cases}$

second line is simply derivative of first one  
can either derivate first line as-is or first can express first line as x=f(y), then derivate that  

**How to deal with bounds**

2 methods:  

1. when change x in y, also change the integration bounds following same let formula  

2. first compute indefinite integral (don't write bounds at all)  
  re-plug the original x var in the final result, don't leave it as function of y 
  so can use the original integration bounds  


eg:  
original bounds: [0,1]  
let function: $y = x+1$  
new bounds: [0+1, 1+1] = [1,2]  
old bounds are the x, new bounds are the y in the let function  

**Rationalizing substitution**

if all incognite appear in same form, like $e^x$ or $sin(x)$ and there are no simple x  
can substitute all of them this way, to create  a rational function  

**Reverse chain rule**

if found $\int f(x) f'(x) dx$  

let:  
$\begin{cases}
y = f(x) \\ 
dy = f'(x) dx
\end{cases}$  

**Substitutions for rational functions**

$ \int \frac{1}{(ax +b)^n} \quad$ $\begin{cases}
y = ax+b  \\
dy = a \, dx  \\
\end{cases}$  

$ \int \frac{1}{(ax^2 +b)^n} = \frac{1}{b^n} \int \frac{1}{\frac{a}{b}x^2 +1} \quad$ $\begin{cases} 
y = x\sqrt{\frac{a}{b}} \\
dy = dx \sqrt{\frac{a}{b}}    
\end{cases}$ 


**Tip for rational functions**

$\frac{x^3}{\sqrt{x^2+1}} = \frac{1}{2} \frac{x^2}{\sqrt{x^2+1}} 2x$  

then let $y=x^2$, then $z= 1+y$  

**Substitutions for exponential functions**

$\begin{cases}  
y = e^{\lambda x} &  \\  
dy = e^{\lambda x} dx  & \rightarrow \frac{dy}{\lambda y} = dx 
\end{cases}$ 

other way to write it:  
$\int R(e^{\lambda x}) dx = \int \frac{R(y)}{\lambda y} dy$  


**Trigonometric substitutions**



$\sqrt{a^2 - b^2x^2} \quad$ $\begin{cases}
x = \frac{a}{b} sin(\theta) \\  
cos^2(\theta) = 1 - sin^2 (\theta) \\ 
dx =  \frac{a}{b} cos(\theta) d\theta
\end{cases}$



$\sqrt{b^2x^2 - a^2} \quad$ $\begin{cases}
x = \frac{a}{b} sec(\theta) \\   
tan^2(\theta) = sec^2 (\theta) -1 \\    
dx = \frac{a}{b} sec(\theta) tan (\theta) d\theta    
\end{cases}$  
 

$\sqrt{a^2 + b^2x^2} \quad$ $\begin{cases} 
x = \frac{a}{b} tan(\theta) \\   
sec^2(\theta) = 1 + tan^2 (\theta) \\    
dx = \frac{a}{b} sec^2(\theta) d\theta \\  
\end{cases}$

these also work for $\frac{1}{(1+x^2)^n}$, who cares if there's no aquare root  

the first one might be the eq of a circumference $\sqrt{1-x^2}$. can compute definite integral by drawing it  


**If different exponents**

pick GCD of the exponents  
works with both exponentials and roots (in roots look like mcm, but it's actually GCD)

eg: if appear $e^{1/2x}, e^x, e^{2x}$  
let $y= e^{1/2x}$, so:  $\quad e^x = e^{\frac{1}{2}x \, 2} = y^2, \quad e^{2x} = y^4$    


eg: there are $\sqrt[4]{x}, \sqrt[3]{x}$, which are $x^{1/4}, x^{1/3}$    
let $y= \sqrt[12]{x} = x^{1/12}$   

easy example: if have $\sqrt{x}$, pick:  $\begin{cases}
\sqrt{x} = t &  \rightarrow x = t^2 \\ 
dx = 2t dt
\end{cases}$

if have all hyperbolic trig, so sinhx and coshx, turn them to their form espressed in $e^x$ and the use this  

### Trigonometric

method depends on what i have  

1. when sin, cos have power 1

2. when sin, cos have power 2  
so no just sin and cos, but $sin^2, cos^2, sin \cdot cos$  

3. when sin, cos are multiplied per each other only at numerator and have high exponent  

methods 1,2 work best if sin, cos appear both at numerator and determinator  

**Best method: half angle formulas**

$sin^2(x)=\frac{1−cos(2x)​}{2}$  
$cos^2(x)=\frac{1+cos(2x)​}{2}$ 

combine with bisection formulas:  
$\cos ^2  (x) \sin ^2 (x) = \frac{\sin ^2 (2x)}{4} = \frac{1- \cos (4x)}{8}$

**Method 1**

$\int R(sin(x), cos(x)) dx = \int R(\frac{2y}{1+y^2}, \frac{1-y^2}{1+y^2}) \frac{2}{1+y^2} dy$  

Let:
$\begin{cases}
y = tan(\frac{x}{2}) & \rightarrow x = 2 arctan(y) \\
dy = \frac{2}{1+y^2} dy \\
sin(x) = 2 sin(\frac{x}{2})cos(\frac{x}{2}) = \frac{2y}{1+y^2} \\ 
cos(x) = 2 cos^2(\frac{x}{2}) -1 = \frac{1-y^2}{1+y^2}
\end{cases}$

**Method 2**

Let :  
$ \begin{cases} 
y=tan(x) \\
cos^2(x) = \frac{1}{1+y^2} \\  
sin^2(x) = \frac{y^2}{1+y^2}  \\
\sin x \cos x = \frac{y}{1+y^2} \\
dx = \frac{1}{1+y^2} dy
\end{cases}
$


**Method 3**

$sin^a cos^b$  

cases:

1. a odd  
leave 1 sin, convert rest to cos using $sin^2(x) = 1 − cos^2(x)$    
then substitute $y = cos(x)$
2. b odd  
leave 1 cos, covert rest to sin  
then substitute $y = sin(x)$

3. a,b odd  
pick either 1 or 2  

4. a,b even  
turn it all to sin  
use half-angle formulas recursively until get just addends, trigs  with x with exponent 1 in the argument, which i can integrate  

5. a=b, both even  
can turn it into sin(2x) with some exponent  

**Other method: addition of angle identities**

$\sin A \sin B = \frac{1}{2}(cos(A-B)-cos(A+B))$   
$\cos A \cos B = \frac{1}{2}(cos(A-B)+cos(A+B))$  
$\sin A \cos B = \frac{1}{2}(cos(A+B)+cos(A-B))$  

##### Hyperbolic substitutions

$\int R(x, \sqrt{a^2-x^2}) dx = \int R(a \sin t, a \cos t) |a| \cos t \: dt$    

by doing:  
$x = a \sin t$  
$dx = a \cos t dt$  

then use the one for trigonometric  

$\int R(x, \sqrt{x^2-a^2}) dx = \int R(a \cosh t, a \sinh t) |a| \sinh t \: dt$    

by doing:  
$x = a \cos t$  
$dx = a \sin t dt$  

$\int R(x, \sqrt{x^2+a^2}) dx = \int R(a \sinh t, a \cosh t) |a| \cosh t \: dt$    

by doing:  
$x = a \sinh t$  
$dx = a \cosh t dt$ 

### By parts

$f = x \quad    \quad dg=e^x dx$  
$df = 1 dx  \quad g = e^x$  

assign funcs in first line  
compute stuff in second line  
then transform in:  
$f\:g-\int g \: df$  
main diagon - last line

**When used**

1. multiplication where one of the terms is a power ($x, x^2$)  
derivate the power, so eventually will disappear  

2. stuff that repeats itself after derivating ($e^x sin(x), sin^2(x), cos^2(x)$)  
use it on repeat til get same thing as the start, then bring it on left (remember the +c)   
$\int e^x sin(x) dx = f(x) + \int e^x sin(x)$  
$\int e^x sin(x) dx = \frac{f(x)}{2} + c$  
for the $sin^2(x)$ one rem to change $cos^2(x) = 1-sin^2(x)$ after using it each time  

3. stuff that make x appear in the denom after derivating  
imagine it multiplied by 1, integrate the 1 so becomes x, derivate the thing i know  
eg. $ln(x), arctan(x), arcsin(x)$  

**Note adomain rctan**  

arctan(x) is defined on the domain $[-\frac{\pi}{2}, \frac{\pi}{2}]$  
so if asked $arctan(-\infty)$ the answer is $-\frac{\pi}{2}$, not $\frac{3 \pi}{2}$  

arcsin(0)=0, not $k \pi$  


### Rational funcs

**Rational funcs**

$ f= \frac{p}{q}$  

0. if p is a multiple of q, do the division p/q and leave the rest as numerator of the fraction

1. factorize denominator q

2. write f as sum of fractions  
each fraction has for denominator one bracket of the factorization and for numerator an unkown value A, B...  
if poly at denom is degree 2 at numerator put $Ax +B$, not just $A$  

3. equalize sum of fraction with original func to find vars A,B  

4. plug in vars and can compute integral easy (probably use log)  

note:  
in C can factorize every poly in product of polys of degree 1  
in R can factorize every poly in product of polys of degree 1 and 2 (not 3, can always break it down) (2 happens when roots are complex, like $x^2+1$)  

**Write as sum of fractions**

Depending on the roots of the denominator:  

- No complex roots:  

$\frac{1}{x^2(x-1)} = \frac{A}{x-1} + \frac{B}{x} + \frac{C}{x^2}$  

- Yes complex roots (denom has roots {$\pm i$}, so second degree poly in scomposition):    

$\frac{2x^3 -1}{x^3-2x^2+x-2} = \frac{A}{x-2} + \frac{Bx+C}{x^2+1}$  

**Basic rules**

- if a root has an exponent, must appear for all exponents until the one it appears with  
- in second degree poly need an x at the numerator  

**Tip to rationalize denom**  

- use difference of squares  
need to pick a binomial that makes the coefficient of $x^2$ and $x$ work  
then see what is the difference in the termine noto   

$x^2 +2x -2 = (x+1)^2 -3 = (x+1 - \sqrt{3})(x+1+ \sqrt{3})$  

- if it's a poly of high degree  
try guessing an easy number, like $\pm 1$ 

to check if denom has real roots just compute delta  

usually:  

$\int \frac{1}{x+a} dx = ln|x+a|$

###  Absolute value  

split interval so thing inside the absolute value has constant sign in each subinterval  
so can drop absolute value  

depending on what is inside the abs value:  

- power of x  
split interval in subintervals, eg: $[-1,1] = [-1,0] + [0,1]$   
in each put sign of x accordingly (x or -x)  

- trig:  
if upper bound is $\infty$: split in sommatoria, put obscillating 1 to handle the sign   
if finite bounds: split in subintervals long like the periodicity, since sin has period $\pi$ do: $[0, \pi] + [\pi, 2 \pi]...$  

**Trig**

split in sommatoria:  
$\int _0^\infty e^{-x} |sin(x)| dx =  \Sigma _{n=0}^\infty (-1)^k \int _{k \pi}^{(k+1)\pi}e^{-x} sin(x) dx$  

it's the normal split used to extimate upper and lower bounds, but removing the absolute value makes appear an obscillating 1  


the bounds depend on the periodicity  
sin, cos have periodicity $\pi$, so each subintegral is long $\pi$    

now solve like normal the indefinite integral, ignoring the boundaries  
maybe do integration by parts  
then plug boundaries in and find if series converge   
if need to compute the value of the original integral, it's probably a geometric series, so can find what it converge to
