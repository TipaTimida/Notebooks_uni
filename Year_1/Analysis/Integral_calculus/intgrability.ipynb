### Terminology

**Character of the integral**

- convergent: limit exists and is finite (the function is called integrable in improper sense) 
- divergent: limit exists but is infinite  
- indetermined: limit don't exist  

limit means value of integral  

**Example indetermined integral**

$\int _0^{\infty} sin(x) = -cos(x) |_0^{\infty}$  
$cos(\infty)$ is indetermined, so the whole integral is  

obv if think about it, area of sinusoid obscillates between positive and negative, doesn't tend to a value, never goes to infinity  

**Proper vs improper**

proper integral:  
- f bounded (sup and min of f can't be infinity)  
- dominium bounded (extremities can't be infinity)  

improper integral: extension of proper one where either f or dominium are not bounded

**Terminology recap**

to be integrable in improper sense: means to be convergent (divergent is NOT integrable)  
integral exist: means to be convergent or divergent  

### Riemann integrability

**Criteria of integrability for proofs**

2 ways to say it:  


$\forall \epsilon \; \exists P: \quad \Sigma osc_I f\;(x_{k+1}-x_k) \; < \epsilon$

$\forall \epsilon \; \exists P: \quad S^* - S_* < \epsilon$  

intuitively: no matter how small i want the difference between upper and lower sum, i can pick a partition where it happens  

**General criteria of integrability**


$f:I \rightarrow R$ is integrable if and only if:  

1. $f$ bounded  (sup, inf are not infinite)  . 

2. $disc(f)$ is negligible (has measure zero)   
$disc(f)$: set of all points of discontinuity  

**Improper integrals**

this criteria is valid for proper integrals, bc $f$ and $I$ must be bounded  
using the limit, can extend the concept of Riemann integrability to improper integrals too, but those need new criteria of integrability  
instead, could use Lebesque integrability  

**Locally Riemann-integrable**

Riemann integrability can only be defined wrt a closed interval (local property)   
if f is Riemann-integrable in every interval i can choose, then f is locally Riemann-integrable (global property)      

if f,g are Riemann integrable, f+g and f $\cdot$ g are too

#### Definitions

**Negligible (trascurabile)**

synonim of <u>Lebesque measure zero</u>  

means can cover the set (here disc(f)) with a collection of intervals, whose total length can be made arbitrarily small  

**Is disc(f) trascurable?**  

disc(f) is the set made of all points of discontinuity  

set is countable:    
sufficient, but not necessary condition for negligible      
eg. Cantor set is uncountable but negligible  
countable mean either a finite amount of discontinuity points or a countable infinity  

examples:  
$Q$: measure 0   
$R$: positive measure (>0)  

**What does f must be bounded?**

if f not bounded, even one discontinuity could be enough to make it divergent (eg: $\frac{1}{x}$)  
so even if $disc(f)$ is negligible, it's actually changing a ton the area  
if f bounded, discontinuities cannot cause big troubles, since f cannot go to infinite and put much area under it  

#### Theorems

### Integrability of continuous funcs

hp:  
- f defined in compact interval  
- f continuous

th: f is integrable in that interval  

note that continuity is <u>sufficient, not necessary condition</u> for integrability  
the necessary condition is that $disc(f)$ is negligigble    

**Proof**

obv satisfies the general conditions  
but try to demonstrate it using the conditions for proofs  

f continuous in compact interval -> f is uniformly continuous (Heine-Cantor th)  

bc of uniform continuity:  
pick a partition P where each interval is long $<\delta$  
in each interval k:  
$osc_k \, f = sup_k \, f- inf_k \, \le  \frac{\epsilon}{b-a}$  

therefore, for the whole intervals k:  
$S^*-S_*=  \Sigma _k (osc_k \: f) (x_{k}-x_{k-1}) \le \frac{\epsilon}{b-a} \;  \Sigma _k (x_{k}-x_{k-1}) = \frac{\epsilon}{b-a} (b-a)  = \epsilon$  


so it's integrable, bc of the criteria of integrability  

### Examples

#### Dirichlet function 

**Dirichlet function**

$f: [0,1] \rightarrow N$  

$\begin{cases} 
1 & \text{if x rational} \\
0 & \text{if x irrational}
\end{cases}$  


rational means belong in Q, irrational belong in R\Q


NO: Riemann integrable  
YES: Lebesque integrable  

**Intuitive explanation**  

Riemann:  
no matter how small a subinterval i pick, will have both rational and irrational, since both are dense  
so any interval will have $S^* = 1, S_* = 0$  
can never be equal, is not integrable wrt Riemann  

Lebesque:  
irrationals are way more than rationals  
so its integral is 0


**Riemann integrability**

$disc(f)=R$ (f is discontinuous at all points in R)  
R has positive measure, so $disc(f)$ is not negligible   
not Riemann integrable  

**Lebesque integrability**  

$Q$: countable, so measure 0  
$R - Q$: uncountable, positive measure (bc R has positive measure, Q measure 0)  
can delete the part with measure zero, only the R\Q part is left  
it's 0 in all points, so the integral is 0  

#### Cantor set

**Cantor set**  

$f: [0,1] \rightarrow N$  

$ \begin{cases} 
1 & \text{if x} \in C \\
0 & \text{if x} \notin C
\end{cases}$  

where C is the Cantor set in [0,1]  

Cantor set: iteratively, remove the middle third of an interval  
the Cantor set is what is left  

integrable on Riemann  
all stuff Riemann integrable is also Lebesque integrable  

**Riemann integrability**

$disc(f)=C$ (Cantor set, not complex numbers)  
$C$: measure 0 (even though is uncountable)  
so disc(f) is negligible  
so it's Riemann integrable  

when trascuring point on cantor set where f=1, remain points where f=0  
so the integral is 0
