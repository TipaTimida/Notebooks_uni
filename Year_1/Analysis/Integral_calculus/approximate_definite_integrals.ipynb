### General process

use **Cauchy sum**:  

i need:  
- f defined in an interval [a,b]  
- partition $P$ of n intervals (created by suddivision $D=(x_0, ... x_n)$)  
- n-upla $Ξ = (ξ_1, ... ξ_n)$  of points, each in an interval  

the <u>cauchy sum associated to the partition and n-upla is</u>:  
$S(f,P,Ξ)= \Sigma ^n_i f(ξ_i) (x_i−x_{i−1})$  

intuitively:  

1. divide in subintervals  
2. choose a method to approximate f(x) in that interval  
means draw an easier f(x)    

the difference between the methods is what i use to approximate the f in the subinterval  

**Different methods**  

what is f in a subinterval in each method?  

1. rectangle method  
f is an horizontal line, at the height of the original f at one extremity of the interval  

2. tangent line  
f is the tangent line in the middle point of the interval  
so i'm actually drawing a trapezio, not a rectangle  
    - or can think of it as an horizontal line, but at the height of the middle point, not the extremities  
    (horizontal line at height of middle point or tangent to middle point give the same area in the subinterval, can see it geometrically, but the first makes understanding the formulas $S_n$ easier)  

3. Simpson-Cavalieri  
f is an arc with coincides with the original f in the extremities and middle point  



the first method gives the worst approximation, the third the best  

note:  

approximation with tangent method is:  
convex func: defect  
concav func: excess  


for Simpson method, rememeber that a parabola is uniquely determined if it must pass thrugh 3 points, bc it's a second degree eq, like how a line is uniquely determoined if must pass though 2 points  

**Notes poly**  

- uniquely determined poly  

i have 2 poly of n-th degree  
if they meet at n+1 points, means they're identical throughout  
means a poly of n-th degree is uniquely determined if must pass through n+1 points  

- can i draw poly?  

i have a n-th degree poly  
pick n points, can make poly pass thorugh them, no matter what points are picked  
eg: have 2nd degree poly, a parabola  
even if the points i picked are on a vertical line, i can make a line trough them, and a line is a special case/degeneration of the parabola when a=0, so the 2nd degree poly pass through them  
rem parabola is only a visualization, what matters is the 2nd degree poly, so a line is fine too  

- how many roots has poly  

a poly of n degree can have at most n roots (points where it's 0)  
if has n+1 points where it's 0, it's the identically null poly p(x)=0  



### Formulas

**Condition**

all methods require to find the upper bound in a certain derivative of f  
so to use them i require <u>f and all its derivatives, until and comprised the one i want the upper bound for, to be continuous in the interval</u> where i'm integrating  

**Notation**

$x_0, ...x_n$: points of the suddivision, so extremities ofthe subintervals  
$z_1, ... z_n$: middle points of each subinterval  

where $z_k = \frac{x_{k-1}+ x_{k}}{2}$

**Rectangle method**

$|f'(x)| \le M_1$  

$S_n = \frac{b-a}{n} (f(x_0) ... + f(x_n))$

**Tangent method**

$|f''(x)| \le M_2$  

$S'_n = \frac{b-a}{n} (f(z_1) ... + f(z_n))$  

**Simpson-Cavalieri method**


$|f^{(4)}(x)| \le M_4$  

$S^*_n = \frac{b-a}{n} \Sigma \frac{f(x_k) + f(x_{k-1} + 4f(z_k))}{6} = $  
$ = \frac{b-a}{6n} (f(x_0) +f(x_n) + 2[f(x_1) + ... + f(x_{n-1})]+ 4[f(z_1) + ... + f(z_n)])$

### Error extimation

the error is  
$\varepsilon = | \int _a^b f(x) dx - S_n | $  


where $S_n$ is the values i computed with the formula of my method  

**Rectangle method**

$\varepsilon \le \frac{M_1}{2} \frac{(b-a)^2}{n}$  

**Tangent method**

$ \varepsilon \le  \frac{M_2}{24}\frac{(b-a)^3}{n^2}$  

**Simpson-Cavalieri method**

$\varepsilon \le \frac{M_4 (b-a)^5}{2880} \frac{1}{n^4}$  

mnemonic tip:  
power of bracket is one more of the number of the derivative used  
power of n in denominator is the number of derivative used  

#### Approximations in specific cases (passing thorugh origin)  

here is not error approximation, but directly computing th integral  

**Rectangle method**

condition: $f(0) = 0$  


$\int _0^b f(x) dx = \int _0^b (b-x) f'(x) dx$  

$ | \int _0^b f(x) dx | \le M \frac{b^2}{2}$

**Tangent method**

condition: $f(0)=0$  

$ \int _{-a}^{+a} f(x) dx = \frac{a}{2} \int _0^a (x-a)^2(f''(x)+f''(-x))$  

$|\int_{-a}^{+a} f(x) dx |\le M_2 \frac{a^3}{3}$  



**Simpson-Cavalieri method**

condition: $f(-a) = f(0) = f(a)$  

$\int _{-a}^{+a} f(x) dx = - \frac{1}{72} \int _0^a (a-x)^3(3x+a)(f^{(4)}(x)+f^{(4)}(-x)) dx$ 
 
$| \int _{-a}^a f(x) dx | \le M_4 \frac{a^5}{90}$
