#### Singularity: split in intervals

**Split integral in intervals**

intuitively:  
instead of thinking about intervals, can just think about the singularity itself  
so write $x \sim x_0$, where $x_0$ is a singularity  

formally:  
must split the integral in more intervals, wher in each of them only one bound is a singularity, the other an ok point  
eg: $\pm \infty, 0$ are the singularities, split in $[-\infty, -1], [-1,0], [0, 1], [1, \infty]$  

don't care what happens in intervals that don't have a singularity, those are definitely convergent, bc:  

in an interval without a singularity:  
no $ x \sim \infty$ -> interval is bounded  
no singularity that breaks C.E -> f is continuous  

f continuous on a bounded interval -> the integral is convergent  


#### Theory: comparison test

**Normal vs asymptotic**  

normal is more formal  
but asymptotic is easier to use  

- asymptotic: $\frac{1}{x +ln(x)} \sim \frac{1}{x}$  
  normal:   $\frac{1}{\ln x} > \frac{1}{x}$   


- asymtptotic:  $\frac{1-\sin x}{x^3 +x} \sim  \frac{1}{x^3}$   
  normal:   $\frac{1-\sin x}{x^3 +x} \le \frac{2}{x^3+x} \le \frac{2}{x^3}$  

**Common error for normal comparison**

MUST ALWAYS SPECIFY WITH SINGULARITY I'M EXAMINING  

usually comparison is used for $x \sim \infty$, but can also be used for $x \sim 0$  
so be careful that the inequality must be true for the singularity i'm studying  

eg: $\frac{1}{x} \ge \frac{1}{x^2} \quad$  ONLY IF $ \quad x \ge 1$  
so can use if $x \sim \infty$, but not if $x \sim 0$  

especially tricky is when have both $\infty$ and $-\infty$, bc inequalities are different:  
$ e^{−x^2} \le e^{-x}$ if $x \sim \infty$  
$ e^{−x^2} \le e^{x}$ if $x \sim -\infty$

$lim _{n \rightarrow \infty} \int _0^n \frac{|\sin(\pi x)| \ln x}{x^p} dx = \infty$

**Steps**

1. split trig in sommatoria of intervals  

$\int _0^n \frac{|\sin(\pi x)| \ln x}{x^p} dx = \Sigma ^{n-1}_{k=1} \int _k^{k+1}  \frac{|\sin(\pi x)| \ln x}{x^p} dx$

by plugging into x values of the interval that make it minimum, except inside trig bc don't know what to do  

$\int _k^{k+1}  \frac{|\sin(\pi x)| \ln x}{x^p} dx \ge \int _k^{k+1}  \frac{|\sin(\pi x)| \ln k}{(k+1)^p} dx = \frac{\ln k}{(k+1)^p} \int _0^1 \sin (\pi x) dx$  

3. study series left outside  

$\Sigma _{k=1}^\infty \frac{\ln k}{(k+1)^p}$ diverge if $0<p<1$, so the original integral does too
