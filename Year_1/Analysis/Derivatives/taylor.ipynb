### What it is

x: point where i want to know the function  
a: point where i know the function  

$f(x) = \Sigma  \frac{f^n(a)}{n!}(x-a)^n = f(a) + f'(a)(x-a) + \frac{1}{2}f''(a)(x-a)^2 \ldots$   

if the point i know is a=0, it's called the McLaurin series and becomes:  
$f(x) = f(0) + f'(0)x + \frac{1}{2!}f''(0)x^2 \ldots$

**What it does**

taylor approximation around point $x_0$ is good approximation only close to that point, that it's not similar asnymore  

eg: $e^{-\frac{1}{x^2}}$, approximated around point $x_0 = 0$ gives a taylor series always null, bc all derivatives are 0  
so it's clearly a bad approximation for any point that is not $x_0$ itself

note: when ask poly until order n, means last term must be $x^n$

**Application**

often used to approximate at first order:  
$f(x+ h) =f(x) + f'(x) h$  
often $h=dx$

### Analytic function  

**Note**

analytical can be thought of both like a local or global property  
if not specified, generally means global (so analytical in all points)  

**Definition** 

local:  
f is analytical in $x_0$ if it has a taylor series (power series) centered in $x_0$ that converges to f in a neighborhood of $x_0$  
means can write Taylor series in that point and represents well a neightborhood of it, as small as i want, don't say anything about other point 

global:  
f that can be approximated in every point as a power series (so with taylor)     

**Which funcs are analytical?**

those whose derivatives grow at most with speed n! (n factorial)   
bc Lagrange remainder grows with $\frac{1}{(n+1)!}$, to converge need remainder to tend to 0, so numerator can't grow faster  

eg: sin, cos, poly, exponential, logarithmic  
sin and cos are analytics bc their derivatives are sin and cos with varying sign, so at any point they're at most 1, which is less than $\frac{1}{(n+1)!}$

**Example**

$e^x$ is analytical  
$e^{-\frac{1}{x^2}}$ is not analytical at point $x_0 = 0$, even though it's smooth (class $C^\infty$)  

why:  
all its derivatives in $x=0$ are $0$, so:  
it's smooth, bc are continuous  
it's not analytical, bc the Taylor series is identically zero, but $e^{-1/x^2}>0$ for $x \neq 0$ since it's an exponential  
so there is no neighborhood of $x_0$ where the series represents it properly  
note: at $x=0$ must explicitly defined $f(0)=0$, else it's a singularity and don't exist  

if it's analytical, it's smooth  
but if it's smooth, might still not be analytical (eg: $e^{-1/x^2}$)

### Remainders

$ f(x) = P_n(x) + R_{n+1}(x)$  

after writing n terms, add a remaninder starting from the n+1 term  

**Types of remainder**  

- Peano  
$o((x-x_0)^n)$  

- Langrage  
$\frac{f^{(n+1)}(t)}{(n+1)!} (x-x_0)^{n+1}$  

#### Extimate error in an approximation

use Lagrange remainder  
compute upper bound for numerator  
then check what n needs to be to have the remainder lower than the error i want  

have f(x), know its taylor series centered in point $x_0$, want to approximate its value at point $a$ (which is prob close to $x_0$) with error less than $\epsilon$  

1. must find upper bound for error  
  - $|f^{(n+1)}(t)| = M$  
  find its maximum, taking t in interval $(x_0, a)$  
  for $f(x)=e^x$, all derivatives are the same, $e^x$, just plug in x the biggest value in the interval $(x_0, a)$  
  for sin, cos, its derivatives are all sin, cos, so the max value is $M=1$    

2. plug upper bound in formula and put it less than the wanted error
$∣R​(x)∣ \: \le \: \frac{M}{(n+1)!}​∣a−x_0∣^{n+1} \le \epsilon$  

3. find smallest n that satisfies it (can just try random values of n)   
tells me how many terms in taylor series i need to write to have error smaller than $\epsilon$ in point a
