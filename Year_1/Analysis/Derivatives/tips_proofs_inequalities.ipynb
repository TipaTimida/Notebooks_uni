**Method: minimize a function**

when: have $A\ge B$  

1. move all on one side, get a function $f = A-B \ge 0$ 

2. compute derivative, find value for which f is minimum  
if 2 values x and y appear: compute partial derivative $f_x$, then find $x(y)$ for which it's minimized  
x,y are the values after the $\forall x,y$  

3. plug such value in f, show that for it f is positive  
(so if the minimum is positive, it's always positive)

**Jensen inequality**

general form:  $\quad f (\Sigma w x)  \le \Sigma w f(x)$  
if f is convex sign is correct, if f is concave sign is reversed  
note: all $x, w$ are ctually $x_k, w_k$, didn't write them to keep clean  

eg:  
log is concave: $\quad \ln (\Sigma w x) \ge \Sigma w \ln x$  
e is convex: $\quad e^{\Sigma w x} \le \Sigma (w \, e^{x})$  

weights $w$ and values $x$ can be anything, even somthing complex  
eg: $x_n$ can be $\ln x_n$

**Example: turn inequality in function, Cauchy-Schwarz**

prove: $(\Sigma a_ib_i)^2 \le (\Sigma a_i^2)(\Sigma b_i^2)$  
$B^2-AC\le 0$    

1. write function with that as $\Delta$: $\quad f(x) =  Ax^2+ \frac{B}{2}x+C = \Sigma a_i^2 \; x^2 +2 \Sigma a_ib_i \; x+\Sigma b_i^2 = \Sigma (a_i^2 \; x^2 +2  a_ib_i \; x+ b_i^2)$  

2. prove has no roots (is always positive or 0): $f(x)= (\Sigma a_i \; x + b_i) \ge 0$  
thus delta is negative or 0 (anyway, non-positive, as asked)


**Example: turn inequality in function, AM-GM**

prove $\frac{a+b}{2} \ge \sqrt{ab}$  

modify so has suitable form: $ \frac{(a+b)^2}{4}-ab \ge 0$

1. write function: $f(x)= ax^2 + 2\frac{(a+b)}{2}x +b$

2. prove has 2 roots: $f(x)=(x+1)(ax+b)$    
thus delta is positive  

**Example: use convexity, weighted AM-GM**

prove: $\Sigma w _k x_k \ge \prod x_k ^{w _k}$  

1. change produttoria in sommatoria: $\Sigma w _k x_k \ge e^{\Sigma w _k \ln x_k}$  

2. take log on both sides: $\ln \Sigma w _k x_k \ge \Sigma w _k \ln x_k$  

3. log is concave, so bc of Jensen inequality, this is true  

notes:

2. take exponential: $\Sigma w _k x_k \ge \prod x_k ^{w _k}$  

**Example: use convexity, Young**

prove: $xy \le \frac{x^p}{p} + \frac{y^q}{q}$  
when $ \quad p,q >1, \quad \frac{1}{p}+\frac{1}{q} = 1$  

1. turn product into sum: $e^{\ln x + \ln y}  \le \frac{x^p}{p} + \frac{y^q}{q}$  

2. bring a weight out: $e^{\frac{1}{p} \ln x^p + \frac{1}{q} \ln y^q} \le \frac{x^p}{p} + \frac{y^q}{q} $  

3. take log on both sides: $\frac{1}{p} \ln x^p + \frac{1}{q} \ln y^q \le \ln ( \frac{x^p}{p} + \frac{y^q}{q})$  

4. it's Jensen, proved  

**Example: use convexity, other order, Young**

prove: $xy \le \frac{x^p}{p} + \frac{y^q}{q}$  
when $ \quad p,q >1, \quad \frac{1}{p}+\frac{1}{q} = 1$  

1. turn product into sum: $xy = e^{\ln (xy)} = e^{\ln x + \ln y}$  

2. bring a weight out: $= e^{\frac{1}{p} \ln x^p + \frac{1}{q} \ln y^q}$  

3. use Jensen on e: $\le \frac{1}{p} e^{\ln(x^p)} + \frac{1}{q}e^{\ln(x^q)}$  

4. got the other side: $ = \frac{x^p}{p} + \frac{y^q}{q}$  

**Example: minimize a function, Young**

prove: $xy \le \frac{x^p}{p} + \frac{y^q}{q}$  
when $ \quad p,q >1, \quad \frac{1}{p}+\frac{1}{q} = 1$  


1. write function: $f(x) = \frac{x^p}{p} + \frac{y^q}{q} -xy \ge 0$  

2. compute partial derivative $f_x$, then find $y(x)$ for which it's minimized: $f_x'(x)=x^{p-1}-y = 0$  
$x(y) = y^{q-1}$  

3. plug in: $f(y^{q-1}) = y^q (\frac{1}{p}+\frac{1}{q}-1)=0$  
the minimum is positive, so it's always positive  


**Example: find another known equality, Holder**

prove: $\Sigma a_i b_i \le AB$  
where $p,q > 1, \quad \frac{1}{p}+\frac{1}{q} = 1$   
where $A = (\Sigma a_i^p)^{\frac{1}{p}}, B = (\Sigma b_i^q)^{\frac{1}{q}}$   

1. normalize: $\frac{\Sigma a_i b_i }{AB} \le 1$  
note that: $\frac{\Sigma a_i b_i }{AB} =\Sigma (\frac{a_i}{A} \frac{b_i}{B})$, bc A and B are constants

2. use Young: $\Sigma (\frac{a_i}{A} \frac{b_i}{B}) \le \Sigma (\frac{1}{p}\frac{a_i^p}{A^p} + \frac{1}{q}\frac{b_i^q}{B^q})$  
the Sigma is not part of Young, it's just there, so must be kept  

3. check the right side: $\Sigma (\frac{1}{p}\frac{a_i^p}{A^p} + \frac{1}{q}\frac{b_i^q}{B^q}) = \frac{1}{p}  \frac{\Sigma a_i^p}{A^p} + \frac{1}{q} \frac{\Sigma b_i^q}{B^q} = \frac{1}{p} + \frac{1}{q} = 1$
