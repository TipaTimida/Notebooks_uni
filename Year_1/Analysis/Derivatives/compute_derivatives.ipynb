### Multivar derivative

**Formula**

$f(x,y) = ...$  
$f' = -\frac{f_x}{f_y}$  

$f_x, f_y$ are partial derivatives  

**Implicit differentiation (multivariable)**

do $\frac{d}{dx}$ on both sides  
since $y=y(x)$ need chain rule: $y^2 \rightarrow 2yy'$  

usef for stuff like $x^2 +y^2 = 9$ bc can't write y = f(x) since would get $\pm$  

#### Particular derivatives:

**Reciprocal**  

$\frac{1}{f'(x)} = \frac{-f'(x)}{f(x)^2}$  
use formula division  

**Absolute value**

$\frac{d}{dx} |x| = \frac{x}{|x|}$

#### Feynman trick

$(\prod f_k ^{a_k})' =  \prod f_k ^{a_k} \sum a_k \frac{f'_k}{f_k}$   

steps:  
1. copy whole thing
2. multiply by a poly  
3. each term in poly is the derivative over the func of each individial funcs


useful to derivate complex thing with multiplication/division

**Logarithmic differentiation**

use it to get a proof of Feynman technique  
take log left and right  
then derivate  

eg: $y = \frac{x^2}{x + e^x}$    
$ln(y) = ln(x^2) - ln(x) - ln(e^x)$  

$\frac{y'}{y} = \frac{2x}{x^2} - \frac{1}{x} - \frac{e^x}{e^x}$  

remember that $\frac{d}{dx} ln (f(x)) = \frac{f'(x)}{f(x)}$

**Compute a high derivative in x=0**

instead of doing all the derivatives until there there, just express the function as Taylor series  
the derivative i want is:  
$f^{(n)} = n! \, a_n$  

where $a_n$ is the coefficient of $x^n$ in the series  
actually only care about $a_n$, not the other coefficients  

can be done also if not x=0, just Taylor series is harder
