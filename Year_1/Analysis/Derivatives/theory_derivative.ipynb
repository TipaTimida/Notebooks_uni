#### Hoelder function

**Hoelder function**

given $\alpha \in (0,1]$, f is $\alpha$-Hoelder func if exists a constant $C>0$ such that:  

$|f(a)-f(b)| \le C|x-y|^{\alpha}$


**Properties Hoelder**

Lipschitz = 1-Hoelder  
every Lipschitz is hoelder (for $\alpha=1$)  
not every hoelder is lipschitz, perchè magari è vera solo per degli $\alpha$ più piccolu, non per $\alpha=1$  
hoelder è una versione più debole e generale di Lipschitz  

a hoelder function is uniformly continuous  

if it's true for $\alpha>1$ means f is a constant function, f(x)=c

**Continuity modulo**

$M(r) = sup\{|f(a)-f(b)|: |a-b|\le r\}$  
M(r) rappresenta quanto può variare al massimo la funzione f tra due punti che distano al massimo r

properties:  

if and only if:  
- $\lim_{r \rightarrow 0} M(r)=0$  
f is uniformly continuous  

- exists L: $\quad M(r) \le Lr$  
f is lipschitzian  

- exists C: $\quad M(r) \le Cr^n$  
f is holderian
