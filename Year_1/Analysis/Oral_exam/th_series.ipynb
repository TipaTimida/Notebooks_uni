### General conditions

**How to write condition for convergence**

a series converge if the SEQUENCE OF PARTIAL SUMS does   
and the sequence of partial sums is a sequence, so use rules for sequence  

eg: sequence converge if MONOTONIC AND BOUNDED   (non-decrescent/crescent and upper bounded, non-crescent/decrescent and lower bounded)  

**Conditions for convergence**

sequence of partial sums  
$S_n = \Sigma _0^n a_i$  

conditions of convergence for sequence of partial sums   
both at specifid or unspecified limit (the second is cauchy th)

$\exist N : \quad \forall n \ge N \quad |S_n - l| < \epsilon$  
$\exist N : \quad \forall n,m \ge N \quad |S_n - S_m| < \epsilon$  

converge: $S_n \rightarrow l < \infty$  
diverge: $S_n \rightarrow \infty$  

#### Necessary condition (nth term test, infinitesimal)

**Th**

$\lim a_n = 0$ is necessary for convergence  

**Proof**

assume series converge:  
$\lim S_n = S$   
$\lim S_{n-1} = S$   


therefore:  
$a_n = S_n - S_{n-1}$   
$\lim a_n = \lim S_n - \lim S_{n-1} = 0$  



#### Geometric sum

notable product: $a^n - b^n = (a-b)(a^{n-1} + a^{n-2}b + \ldots + b^{n-1})$  
$1^n - x^n = (1-x)(1 +x +x^2 + \ldots + x^{n-1})$  

second parenthesis is a geometric progression, so:  
$\Sigma _{k=0}^n x^k = \frac{1 - x^{n+1}}{1-x}$

obs:  
if $x>1$, it's divergent, bc have $x^\infty$    
if $0<x<1$, it's bound between 0 and 1

#### Comparison test

**Th**

$\exist N : \quad \forall n>N , b_n  \le a_n$   
so $a_n$ is maggiorante of $b_n$  

if smaller diverge to $+\infty$ (not to -$\infty$), bigger diverge  
if bigger converge, smaller converge    

actually, is enough for this to be true from an n onwards, not for all n  

**Proof**

have: $0\le b_n \le a_n$  
must be $>0$, this proof is only for positive series  

consider sequence of partial sums: $S_a, S_b$  
since all elements are positive, the sequences $S_a, S_b$ are non-decreasing (crescent)  

if A converge:  
$\lim S_b \le \lim S_a \rightarrow l < \infty$  

$S_b <\infty$ means b don't converge  

alternative:  
$b_n \le a_n \le l$  
so $b_n$ is bounded above and non-decreasing: it converges  

if B diverge:  
$+\infty \le b_n \le a_n$  
so $a_n$ diverge too  

**Example: harmonic series**

$a_n = \frac{1}{n^2}$  

compare with Mengoli: $\frac{1}{n^2} \le \frac{1}{n(n+1)}$  

Mengoli is convergent, so this is too

**Mengoli**

it's telescopic:  
$\Sigma _1 \frac{1}{n(n+1)}= \Sigma _1 (\frac{1}{k}-\frac{1}{k+1}) = \Sigma _1^n \frac{1}{k} - \Sigma _2^{n+1} \frac{1}{k} = 1- \frac{1}{n+1} \rightarrow 1$

#### Root criteria

**Th**

converge if:  
$\exists N, h<1 : \;$  
$\forall n>N, \; \; \sqrt[n]{a_n} \le h$  

the part about $\exist N: \forall n>N$ is a formal way to say to take the $\lim _{n \rightarrow \infty}$  

**Proof**

$\forall n>N \; \; \; \sqrt[n]{a_n} \le h \; \rightarrow  \; a_n \le h^n$  
so series has a geometric series of ragione (h) minore di 1 come maggiorante  
by criteria del confronto, converge  

#### Ratio criteria

**Th**

converge if:  
$\exists N, h<1 : $  
$ \forall n>N  \quad \frac{a_{n+1}}{a_n} \le h  $

**Proof**

rewrite the last line:  
$a_n \le a_{n-1}h \le ... \le  a_0 h^n$   
so has for maggiorante a geometric series of ragione less than 1, which converges

### Serie of any sign

**Tips for series of any sign**

method 1:  
split original series in 2 subseries (or more)  
prove both are convergent (sum of convergent series is convergent) 

when splitting, try to make the 2 series both positive, so can use criteria for positive series  
eg: put all positive terms in a series, all negative in the other, changed of sign so are positive  
anyway, sum/difference of convergent series is convergent   

eg: in alternating series, take as subseries the partial sums of only even or odd indexes (and change the sign of the one with the negative terms)  

method 2:  
split the sequence of partial sums in subsequences  
prove both converge TO THE SAME LIMIT (not required in series, yes in sequence of partial sums)


#### Absolute convergence

**Th**

absolute convergence implies conditional convergence  

**Proof**

given $x \in R$, define its positive and negative parts $x^+, x^-$ (essentially they're modulo of x only if x has that sign)    
$x = x^+ - x^-$    
$|x| = x^+ + x^-$  

split original series in 2:  
one with just positive terms, other with just negative terms (but changed of sign so they're positive, so have 2 positive series, can use criteria)      
$a_0^+ + ... a_n^+, \quad a_0^- + ...  a_n^-$

the original series with absolute values $|a_0| + ... + |a_n| \quad$  is a maggiorante on both  
by hypothesis it's convergent, so both of the partial ones are too   

original series is difference of these two partial series, difference of convergent series is a convergent series, so original one it converges too  

essentially: $S_n = S_+ - S_- \le S_+ +S_- = |S_n| < \infty$  
wuinsi $a_n$ converge  

#### Leibniz (convergence alternated sign)

**Th**

Conditions on $a_n$:  
- positive terms   
- not crescent  
- infinitesimal

then $\Sigma (-1)^n a_n = a_0 - a_1 + a_2 ...$ converges   


not crescent means $\le$ is fine, don't need <  
but from now on i'll say that decrescent means $\le$, and call $<$ strictly decrescent  

also can see:  
$S_n$ are partial sums  

**Proof**  

1--- pick partial sums odd and even 

consider the partial sums until term number k: $S_k$  
assume if k is even, the last term has positive sign, if k is odd the last term has negative sign  
eg: $S_3 = a_0-a_1+a_2-a_3$  
$k=3$ is odd, so last term has negative sign   

now consider the sequence of partial sums:  
$S_{2k} = S_0, S_2, ... = S_{2k-2} - a_{2k-1} + a_{2k}$  
$S_{2k +1} = S_1, S_3, ... = S_{2k-1} + a_{2k} - a_{2k+1}$   


2--- study how they behave 

since by hypothesis $a_n$ is decrescent:  
$S_{2k}$ is decrescent:  $S_{2k} \le S_0 \quad$  (bc $  - a_{2k-1} + a_{2k} \le 0$  )  
$S_{2k+1}$ is crescent  $S_{2k+1} \ge S_1 \quad$  (bc $ + a_{2k} - a_{2k+1} \ge 0$)  
intuively: if k is even, the last term was positive  
to get to next partial sum with even k must add 2 numbers, a negative and a positive  
since $a_n$ is decrescent, the negative one is bigger than the positive one, so the partial sum is decreasing  

prove both are bounded:  
$S_0 \ge S_{2k} \ge S_{2k+1} \ge S_1$, bc odd term is negative, so adding it makes the sequence smaller  


thus $S_{2k}$ is decrescent with lower bound $S_1$, so converges bc of monotony theorem for positive series   
same way can prove the other is crescent and with upper bound, so both converge  


3--- convergence of partial sums and full series

since by hypothesis it's infinitesimal ($a_n \rightarrow \infty$), the limits are the same:  
$\lim S_{2k} - \lim S_{2k+1} = \lim a_{2k+1} = 0$   
thus the subseries converge to same limit, and the the original series must converge to the same limit too  
since the original series is the limit of this subseries (these subseries form a partition of the original series) 

### Dyadic blocks / Cauchy

**Dyadic block in proofs**

dyadic block: $B = \Sigma _{n=2^k}^{2^{k+1}-1} \, a_n$  
sum of terms in an interval between 2 consecuive powers of 2  

properties:  
- $\Sigma B = \Sigma a_n$  
- is sum of $2^k$ terms  
- if $a_n$ decreases,  biggest term is $a_{2^k}$, smallest is $a_{2^{k+1}-1}$  
but generally for lower bound is used simply $a_{2^{k+1}}$    

note that $a_n$ must decrease, or won't be infinitesimal, which is a necessary condition for convergence  

**Recap: what to do with it**

1. write the dydic block $B_k$ for the series i'm studying  
2. to prove convergence: prove the higher bound of the dyadic block converge  
3. to prove convergence: prove the lower bound of the dyadic block diverge    

**What to do with it**

1. split the original series into dyadic blocks (intervals)  
2. find lower or higher bound FOR THAT INTERVAL, not for whole series  
3. use the summatory on both sides, B ad its bound, to return to original series $\Sigma B = \Sigma a_n$  
on other side, use upper bound if want to prove convergence, lower bound if want to prove divergence  
4. prove the other side converge or diverge, then use comparison th  

note: if lower bound of the interval converge to something non-zero, series diverge, bc when doing $\Sigma$ to return to original series it gets multiplied infinite times, so lower bound becomes infinite  
thus, to converge, condition is that lower bound of the interval must tend to 0  

**Dyadic block vs Cauchy condensation**

bound of dyadic blocks and condensed series are the same thing  
just hen using condensation th can put = sign between the 2 things, when using dyadic blocks do an inequality  

so condensation is less formal, but faster to write  
however, if doing proofs from scartch, when can't use theorems, use the dyadic block emthod  

#### Cauchy convergence criteria

**Series convrgence, th tail**

necessary and sufficient condition for series convergence:  
$\forall \epsilon > 0, \; \exists N :$  
$ \forall n >N, \forall p>0, \; |a_n + a_{n+1} ... + a_{n+p} | < \epsilon$  

or can write:  
$ \forall n >N, \forall p>0, \; |\Sigma _n^{n+p} a_i  | < \epsilon$  

means:  
however small the error/interval, from a point onward the sum of all next terms will be smaller than that value  

**Example: harmonic series**

$a_n = \frac{1}{n}$    
don't respect Cauchy convergent criteria, thus don't converge  

proof:  
consider the sum of all the terms in between two consecutive powers of 2, eg: $\frac{1}{2^k}, \frac{1}{2^{k+1}}$  
$B_k = \Sigma _{n=2^k}^{2^{k+1}-1} \, a_n$ 

find lower bound ($2^n$ times the lowest term):    
$B_k \ge  2^k \frac{1}{2^{k+1}} = \frac{1}{2}$ 

$\Sigma a_n =  \Sigma B_k \le  \Sigma \frac{1}{2} = \infty$

what it means:  
at any point i start, you can always find a group of consecutive terms after it whose sum is larger than $\frac{1}{2}$  
so cannot pick $\epsilon$ as small as i want and have a tail smaller than it   
thus violate Cauchy critaria  

what it means easily:  
every dyadic block adds 1/2 to the sum of the series  
there are infinite dyadic blocks, so the sum is infinite  

**Example: p-series converge if p>1**

$a_n = \frac{1}{n^p}$  
clearly $a_n$ is positive and decrescent  

pick dyadic block:  
$B_k = \Sigma _{n=2^k}^{2^{k+1}-1} \, \frac{1}{n^p}$  

find upper bound (highest term in interval is: $\frac{1}{(2^k)^p}$, has $2^n$ terms)  
$B_k \le  2^k \frac{1}{2^{kp}} = 2^{-k(p-1)}$   

$\Sigma a_n = \Sigma B_k \le \Sigma  2^{-k(p-1)}$  
it's a geometric series with ratio 2, clearly $k>0$  
converge if exponent tends to $-\infty$, so if $p-1 >0$, $p>1$  
thus $a_n$, that is smaller, converge too  

**Example: p-series diverge if p>1**

pick dyadic block:  
$B_k = \Sigma _{n=2^k}^{2^{k+1}-1} \, \frac{1}{n^p}$  

find lower bound:   
$B_k \ge  2^k \frac{1}{2^{(k+1)p}} = 2^{-p} \, 2^{-k(p-1)}$   

$\Sigma a_n = \Sigma B_k \ge  \Sigma  c \; 2^{-k(p-1)}$  

the $2^{-p}$ can be ignored, it's a constant  
check the other factor: if $p<1$ it grows to infinity  
so $\Sigma a_n$ is bigger than something divergent, it's divergent too  

note: if $p=1$,  $ B_k \ge \frac{1}{2}$, so $\Sigma B_k \ge \Sigma \frac{1}{2} = \infty $, so diverge too   

#### Cauchy condensation th

**Th**

$a_n$ is positive and decreasing  
$\Sigma a_n$ converge $\longleftrightarrow \Sigma 2^n a_{2^n}$ converges  
so it's if and only if condition   
the second one is called condensed series  

**Proof**

take sum of all terms between two powers of 2, like $2^k$ and $2^{k+1}-1$  
must find upper and lower bounds  
note that:  
- there are $2^k$ terms  
- $a_n$ decreases, so biggest term is $a_{2^k}$, smallest is $a_{2^{k+1}}$  

dyadic block: $B = \Sigma _{n=2^k}^{2^{k+1}-1} \, a_n$  
note that: $\Sigma B = \Sigma a_n$  

know a thing converge from hp, want to use comparison th to prove the other converge too  
since $a_n$ is decreasing, can find bounds for the block:  
$2^k a_{2^{k+1}} \le B \le 2^k a_{2^{k}}$  

$\leftarrow$  

$ \Sigma a_n = \Sigma B \le \Sigma 2^k a_{2^{k}}$   
so since by hp f condensed converge, the original converge too  

$\rightarrow$ 

doing some changes:  
$\Sigma _{0} 2^k a_{2^{k+1}} = \Sigma_1  2^{k-1} a_{2^{k}} = \frac{1}{2} \Sigma 2^k a_{2^{k}}$  

$\Sigma 2^k a_{2^{k+1}} \le \Sigma B = \Sigma a_n$  
$\frac{1}{2} \Sigma 2^k a_{2^{k}} \le \Sigma a_n$  

so if the thing on the right converge, the thing on the left does too
