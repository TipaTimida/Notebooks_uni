### Composta

**Derivata funzione composta**

2 methods to write:  

- $(g \, o \, f)' (x) = g'(f(x)) f'(x)$

- $\frac{dz}{dx} = \frac{dz}{dy} \frac{dy}{dx}$  

where $z=g(y)$, $y=f(x)$   
so $z = g(f(x))$  


**Proof**

want $z' = \lim _{h \rightarrow 0} \frac{z(x+h)-z(x)}{h}$

multiply up and down for the middle thing (essentially dy, $f(x+h)-f(x)$):  
$\lim _{h \rightarrow 0} [\frac{g(f(x+h))-g(f(h))}{f(x+h)-f(x)} \frac{f(x+h)-f(x)}{h}] = g'(f(x)) f'(x)$   

### Inversa

$(f^{-1})'(y) = \frac{1}{f'(x)}$  

$\frac{dx}{dy} = \frac{1}{\frac{dy}{dx}}$

**Proof**

$x = g(y)=f^{-1}(y)$  

$f(g(y))=y$, so:  
$\frac{d}{dy} f(g(y)) = \frac{d}{dy} y = 1$  

but doing it properly with the chain rule:  
$f'(g(y)) g'(y) = 1$  
$g'(y) = \frac{1}{f'(g(y))} = \frac{1}{f'(x)}$
