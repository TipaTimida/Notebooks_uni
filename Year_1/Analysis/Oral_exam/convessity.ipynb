### Convexity

**Jensen inequality**

general form:  $\quad f (\Sigma w x)  \le \Sigma w f(x)$  
if f is convex sign is correct, if f is concave sign is reversed  
note: all $x, w$ are ctually $x_k, w_k$, didn't write them to keep clean  

eg:  
log is concave: $\quad \ln (\Sigma w x) \ge \Sigma w \ln x$  
e is convex: $\quad e^{\Sigma w x} \le \Sigma (w \, e^{x})$  

weights $w$ and values $x$ can be anything, even somthing complex  
eg: $x_n$ can be $\ln x_n$

**Proof Jensen inequality**

have $f(x)$  
pick $g(x)$ such that:  $\quad f(x_0)=g(x_0)\quad \quad  g(x) \le f(x) \quad \forall x \in I$  
($g(x)$ could be the tangent, it's a line completely under the graph)  
define $x_0 =  \Sigma w_i x_i$  

$f(\Sigma w_i x_i)  = f(x_0) =g(x_0) = m(\Sigma w_ix_i) +q \le \Sigma w_i (mx_i +q) = \Sigma w_ig(x_i) \le \Sigma w_if(x_i)$  

**Convex combination**

given:  
$x_1, .. x_n \in R$  
$t_1, ... t_n \in [0,1]$, where $\Sigma t_i = 1$  (essentially they're weights)

the point $\Sigma t_i x_i$ is called convex combination of $x_1, .. x_n$  
essentially it's a weighted average, useful to prove inequalities    

geometrically:  
if have only 2 points, it gives me all the points in the line connecting them:  
$w_1x_1 +w_2x_2$, sometimes written: $(1-t)x + ty$  
this formula gives, for t in [0,1], each point in the segment connecting x and y

**Jensen inequality**

can also be written in integral form:  

$f: [a,b] \rightarrow I$ integrable  
$\varphi: I \rightarrow R$ convex  
$\varphi(f)$ integrable  

$\varphi(⨍ _a^b f) \le ⨍ _a^b \varphi (f)$  

where:  
$⨍_a^b f = \frac{1}{b-a}\int_a^b$  
which is the integral average  

written in discrete:  
f convex, $\lambda$ weights with sum 1, all positive  
$f(\Sigma \lambda _k x_k) \le \Sigma \lambda _k f(x_k)$

### Convex set / space

**Formally**

$ E \subseteq R^n : \quad \forall x,y \in E,$ the segment $[x,y] \subseteq E$  

the segment $[x,y]$ can be written:  
$[x,y] = \{ (1-t)x +ty \; : \; t \in [0,1]\}$  
this formula gives, for t in [0,1], each point in the segment  

**Intuitively**

picking any 2 points of the set, the segment connecting them is contained in the set  

**Observations**    

$E_1, E_2 \; convex \longrightarrow E_1 \cap E_2 \; convex$  
true for intersection, not for union (obv, just imagine uniting two balls)  

**Epigraph**

set of points lying above the graph of a function (region of space above the curve f(x))  

if f is convex, the epigraph is a convex set (just imagine it visually)    

### Convex function

#### Definitions

**Recap definitions**

f is convex if:  

1. $f''(x)>0 \forall x$  

2. $R(x,y)$, the incremental rapport, has positive partial derivative in both x and y (is crescent both in x and y)   
essentially the same as def 1, derivative of derivative is positive    

3. $f((1-t)x +ty) \le (1-t)f(x) + tf(y)$  
a version of Jensen inequality with only two points and weights  
bc geometrically means pick 2 points in the function, line connecting them is above the function  

**Useful def**  

$f$ is convex $\longleftrightarrow \forall x \in I, \, \frac{df}{dx}$ is monotone crescent  

right side means f' is crescent (so if the derivative is crescent, NOT if the function f is crescent)   

for $f'$ crescent it's sufficient $f''>0$, but it's not necessary  
bc a function can be convex even if it's not derivable twice everywhere  
eg: $f(x)=|x|$, has a punto angoloso in x=0, so can't be derived twice    
ok discontinuities, but must be at most numerables  

**Def with chords** 

$f: I \rightarrow R: \quad \forall x,y \in I, \quad  \forall t \in [0,1]$  
$f((1-t)x +ty) \le (1-t)f(x) + tf(y)$  

means each cord uniting two points of the curve (the curve is the graph of f) is above the curve   


**Explanation def with chords**:  

pick two points, x and y, belonging to the curve (both x,y are x coords)    
draw a segment between these points

$(1-t)x +ty$  
this formula gives, for t in [0,1], each point in the segment i just drew  

$f((1−t)x+ty)$: value of the function at each point of the segment   
essentially $f(a)$, where a are all points belonging to the segment  

$(1−t)f(x)+tf(y)$: weighted average of values of x and y  


**Def incremental rapport**


write incremental rapport of f:  
$R(x,y) = \frac{f(y)-f(x)}{y-x} \quad x,y \in I$    
note that y is a point on the x axis, an ascissa, NOT an ordinata  


f convex means R(x,y) is crescent in both vars x and y  

$R(x,y)$ is the angular coefficient (like all incremental rapports) of the chord pssing thorugh points $(x, f(x))$ and $(y, f(y))$  

**Definition concave**

$f$ is concave if $-f$ is convex  
all formulas are similar to before, but swap $\ge$ with $le$ and viceversa  

#### Properties

**General properties**

useful to see if imagine a convex parabola (a parabola with $f''>0$)  

if $\quad f: [a,b] \rightarrow R$ convex   

- $f$ bounded, continuous, integrable on $[a,b]$ 

- $f(x) \le max (f(a), f(b))$  
means its max is at one of the extremities 

- its stationary points are also points of minimum absolute 

**Graph over tangent**

$f$ convex and derivable $\longleftrightarrow f(x) \ge f(x_0) + f'(x_0)(x-x_0)$  

$f(x_0) + f(x_0)'(x-x_0)$ is the tangent in a point $x_0$  
so if i draw the tangent in ANY point $x_0$, such tangent will be completely UNDER the graph  


**Proof: graph over tangent**

part $\rightarrow$  

assume $x_0 <x$, the opposite case is analogous:  
assume the proposition is false, so can rewrite:   
$f'(x_0) > \frac{f(x)-f(x_0)}{x-x_0} = f'(c)$    
per Lagrange th, exists at least a point $c \in [x_0,x]$ such that that's its derivative  

but i know $f'$ must be monotonous, so can't be $f'(c)<f'(x_0)$ since $c >x_0$  
so contraddiction 

intuitively: since it's convex, points after must have derivative bigger  


part $\leftarrow$  

(here write $x_0$ as y)   

can rewite as:  
$f'(x) \ge \frac{f(y)-f(x)}{y-x} = R(x,y)$  

therefore (WHY???):  
$f'(y) \ge R(x,y) \ge f'(x)$  

so $f'(x)$ is monotone crescent

$f:I \rightarrow R$ convex, $x_0$ internal point

### Regularity class

**Definition**

f is of class $C^k$  means:  
- f can be derivated k times
- f and all those k derivatives are continuous  

(number on top is number of continuous derivatives) 
so are not considered the funcs derivable k times but where not all these derivatives are continuous  

**Examples**

$C^0$: f is continuous, f' not continuous or don't exist  
eg: $f(x)=|x|$, derivative is not continuous in x=0  

$C^1$: f and f' are continuous, f'' not continuous or don't exist    

$C^\infty$: derivable infinite times, always continuous  
eg: $f(x)=e^x, f(x)=c$  
called SMOOTH FUNCTION, 'funzione liscia'  

### Inequalities

**Young inequality**

$xy \le \frac{x^p}{p} + \frac{y^q}{q} \quad \forall x,y \ge 0$  

when $ \quad p,q >1, \quad \frac{1}{p}+\frac{1}{q} = 1$  


methods to prove:  
- minimize the function, prove minimum is positive  
- special case of weighted AM-GM,  $w_1 = \frac{1}{p}, w _2=\frac{1}{q}$

**Holder inequality**

$a_i, b_i \ge 0$  
$p,q > 1, \quad \frac{1}{p}+\frac{1}{q} = 1$

$\Sigma a_i b_i \le AB$  
where $A = (\Sigma a_i^p)^{\frac{1}{p}}, B = (\Sigma b_i^q)^{\frac{1}{q}}$   

to prove:  
- use Young    

**Cauchy-Schwarz inequality**

$\Sigma |a_ib_i | \le \sqrt{\Sigma a_i^2} \sqrt{\Sigma b_i^2}$  

to prove:  
- the case $p=q=2$ of Holder  
- turn inequality in function    

**AM-GM inequality**

$\frac{1}{n}  \Sigma x_k \ge (\prod x_k )^{\frac{1}{n}}$  


to solve: 
- special case of Jensen with uniform weights $w_n = \frac{1}{n}$   
- use convexity (where i need to use Jensen anyway)

$e^p = \{(a_i)_{i \in N} : \Sigma |a_i|^p < \infty\}$
