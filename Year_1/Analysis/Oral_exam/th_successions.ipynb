### Definition: sequence

**Succession / sequence**

2 ways to write:  

$a:N→X$  
$a: n \rightarrow a_n$

an application whose dominium is N (natural numbers) 

**Limit of a succession /convergence**

$\lim _{n \rightarrow \infty} a_n = l$  if:  

form with neighborhoods:  
$\forall V$ neighborhood of $l, \quad \exists \overline{n} \in N: \quad \forall n \ge \overline{n} \longrightarrow a_n \in V$   

form without neighborhoods:  
$\forall \epsilon >0,  \quad \exists \overline{n} \in N : \quad \forall n \ge \overline{n} \quad \rightarrow | a_n - l| \le \epsilon$   

**Intuitively**

can choose an interval as small as i want, there is a number that, from then on, all values of succession will be in that interval  
in def of existance of limit pick all values in a neighborhood in x axis, here all numbers after a specific n, bc a succession has a direction, since the index increases  

a succession having a limit means it converges  
so this is also the condition for convergence  

#### Subsequence (successione estratta)

**Subsuccession: formal def**  

given $a: N \rightarrow E$ a succession (to a random codominium)  
a subsuccession is one obtained by composition: $N \rightarrow _{\sigma} N \rightarrow _a E$  
where $\sigma$ is crescent (func from N to N)  


$\sigma$ decides which indices $n$ get used, so which terms get picked  
only pick even indices: $a_{2k} \quad \sigma : n \mapsto 2n$   
only pick square indices: $a_{k^2} \quad \sigma : n \mapsto n^2$


**Subsuccession: intuitive def**  

given a sequence a, pick only some indices ($n_k = 0, 2, 5...$ is the list of indices i want to pick)  
sequence of indices must be crescent (that's why $\sigma$ is crescent)  
it's written $a_{n_k}$

### Cauchy sequence

**Definition**

a sequence where: 
 
$\forall \epsilon >0, \quad \exists N \in N : \forall m,n \ge N, \quad |a_n - a_m| < \epsilon$  

so the terms get arbitrarily close to each other as the sequence progresses.

**Cauchy convergence criteria**

a sequence converge IF AND ONLY IF it's a Cauchy sequence   

**Cauchy and convergence definitions**

Cauchy property:  
$\forall \epsilon >0, \quad \exists N \in N : \forall m,n \ge N, \quad |a_n - a_m| < \epsilon$  

convergence definition:  
$\forall ε>0, \quad \exists N∈N:  \quad \forall  n≥N, \quad ∣a_n−L∣<ε$  

in space R they are equivalent, bc of Cauchy convergence criteria    
intuitively, Cauchy says that in the tail of the sequence the points get arbitrarily close to each other, but don't say what value they'are approaching like this    
convergence def says that that the tail gets closed to a specific value $L$, which is its limit

#### Extract monotonic sequence

**Extract monotonic subsequence**  
 
th: from each sequence can extract a monotone subsequence  

proof:  
peak points: elements of a succession bigger than all elements after it  
if peak points are infinite: can make a monotone decrescent subsequence using them, since each is bigger then the next ones    
if they're a finite number: pick the number after the last one, it's not a peak so there will be a bigger number after it, which isn't a peak either bc i started after the end of peaks, so continue forever to pick a bigger one  

#### Bolzano-Weierstrass

if real sequence in bounded space, can extract a convergent subsequence  

bounded is enough, don't need compact  
sequence in bounded space means bounded sequence  

**Bolzano-Weiestrass th**  

proof:  
$x_n$ is real sequence in compact space I (an interval)    
split in half $I$, at least one half has infinitely many entries $x_n$  
iteratively halve the half having infinite elements  
so have nested sequence of interval, with length $\rightarrow 0$, each has infinitely many elements  
their intersection is a single point $x$, can choose a subsequence converging to $x$  

#### Cauchy sequence (R complete)

Ogni successione convergente è di Cauchy: sempre  
ogni successione di Cauchy è convergente: sono in spazio completo  

**converges -> Cauchy**

condition of convergence $a_n \rightarrow L$:  
$\forall \epsilon > 0, \quad \exists N: \forall n \ge N: \quad |a_n -L| < \frac{\epsilon}{2}$  

now use triangle inequality:  
$|a_n - a_m| \le |a_n -L| + |L - a_m| <  \frac{\epsilon}{2} +  \frac{\epsilon}{2} = \epsilon$  

which is the definition of Cauchy sequence

**Cauchy -> converge**

1.  show Cauchy sequence is bounded
  
pick $\epsilon =1, m = N\quad$ in the condition for Cauchy sequence, so get:    
$|a_n| = |a_n-a_N+a_N| < |a_n - a_N| + |a_N| < 1 +|a_N|  \quad \forall n \ge N$   
since $a_N$ is locked, all successive $a_n$ are bounded  
the previous ones are a finite number, so they're limited too  

2. use boundedness to find subsequence

Bolzano-Weierstrass th: every bounded sequence has convergent subsequence  
so $a_{n_k} \rightarrow L$  

3. Cauchy sequence with convergence subsequence converges (at that same L)  

definition of Cauchy:  
$\exists N_1 : \quad |a_n-a_m| < \frac{\epsilon}{2} \quad \forall n,m \ge N_1$  
condition that subsequence converges:  
$\exists N_2: \quad |a_{n_k} - L| < \frac{\epsilon}{2} \quad \forall n \ge N_2$ 

write condition that sequence converges:  
so $\forall n \ge N_1, N_2$:  
$|a_n - L | \le |a_n - a_{n_k}| + |a_{n_k} - L| < \frac{\epsilon}{2} + \frac{\epsilon}{2} = \epsilon$  
which is the condition for convergence  

note that $a_{n_k}$ è comunque un valore della successione, è come un $a_m$  

#### Heine-Borel

closed + bounded = compact

**compact -> bounded** 

recap:  
since it's compact all sequences must have convergent subequence  
if it weren't bounded, could pick a sequence that diverge to $\infty$  
but if diverge can't have convergent subsequence, so can't be  


pick sequence in K  
assume K is not bounded, so can pick: $||x_n|| > n \quad \forall n$  
means a sequence where the terms are bigger than n  
$x_n$ diverge to infinity, has no convergent subsequence  
however, in a compact set, every sequence has convergent subsequence  
$x_n$ doesn't, contraddiction, so K is bounded  

**compact -> closed**

want to prove that if a sequence in K converge, has limit in K  
pick $x_n \rightarrow x$, a random convergent sequence in K (don't know where it converge yet, just that it does)      
since K is compact, there is subsequence converging in K  
but $x_n$ already converge to $x$, so the subsequence must converge to the same $x \in K$  
so all limits of sequences in K are in K, meaning K is closed  

**closed + bounded -> compact**  

take $x_n \in [a,b]$  

bounded: by Bolzano-Weiestrass, there is convergent subsequence $x_{n_k}\rightarrow x$  
closed: a succession in an interval converge to a point in it, so $x$ must be in [a,b] too

thus, every sequence in K has a convergent subsequence with limit in K: K is compact  

#### Weierstrass

$f: K \rightarrow R$  

hp: 
- f continuous  
- K compact  

th:  
- f(K) is bounded on K  
- f attains its bounds (exist $x_{max}$ and $x_{min}$ on K, where $f(x_{max})=sup \; f$)  

$f:[a,b]\rightarrow R$ continuous  
then exist max and min for f in $[a,b]$

**Intuitively**

f continuous in compact space  
th: f has inf and sup finite ($sup < \infty, inf > - \infty$)

rem compact means closed  
bc if interval were open on one extremity, and f were crescent going towards that extremity, then would not have a sup, bc could always grow, would not know when the last term is  
rem that sup is not upper bound, must actually be reached by func somewhere  

1--- f bounded:  
since K compact and f continuous, $f(K)$ is compact   
also, compact = closed + bounded   
thus $sup \, f$ and $inf \, f$ exist and are finite  

2--- attains its bounds (So its sup is really touched, it's a max belonging to the image of f):  
let $M= sup \, f$  
want to prove $f(x)=M$ with unicity of limit  
pick sequence getting close to it: $x_n \in K$, where $f(x_n)>M - \frac{1}{n}$  
thus $f(x_n) \rightarrow M$  
since K is compact, $x_n$ has a convergent  subsequence: $x_{n_k}\rightarrow x \in K$ (x is a random point, i only know it's in K)   
apply continuity:  $f(x_{n_k}) \rightarrow f(x)$  
for uniqueness of limit, $f(x)= M$, the sup is attained at some point x  

#### Heine-Cantor

f continuous, K compact -> f uniformly continuous  

intuitively:  
X compact means dominium/interval is bounded (has start and end)   
f continuous means don't blow to infinite  
so can find biggest derivative in that interval  
that is the maximum steepness, so it's uniformly continuous with that as maximum steepness    

1--- assume f is not uniformly continuous  
then: $\exist \epsilon : \forall \delta , \exist x,y \in K : |x-y| < \delta, |f(x)-f(y)| > \epsilon$  

2--- so can construct sequences:  
$x_n, y_n$ where $|x_n-y_n| < \frac{1}{n} \quad |f(x_n)-f(y_n)|\ge \epsilon$  
intuitively: sequences tend to same point, images no  

3--- use subsequences  
K is compact, so for Bolzano-Weierstrass there is a convergent subsequence  
$x_{n_k} \rightarrow x \in K$  
per costruzione, anche $y_{n_k} \rightarrow x$   
(formalmente, $||x_{n_k}-y_{n_k}||< \frac{1}{n_k} \rightarrow 0$, also $y_{n_k} \rightarrow x$)  

4--- use continuity  
$f(x_{n_k}) \rightarrow f(x)$ and $f(y_{n_k}) \rightarrow f(x)$  
so $|f(x_{n_k}) - f(y_{n_k})| = |f(x)-f(x)| \rightarrow 0$  
which is in contraddiction with the assumption that f is not uniformly convergent $|f(x_k)-f(y_k)| \ge \epsilon$  

**Th Heine–Cantor**  

Ogni funzione continua su un insieme compatto (chiuso e limitato) è uniformemente continua.

proof:  

1--- scrivi ipotesi  
scrivi ipotesi della condizione di continuità uniforme (ho due sequenze a,b con stesso limite, voglio vedere che anche le immagini hanno stesso limite)  
prendo $|a_n - b_n | \rightarrow 0$   

2--- usa compattezza

$a_{n_k} \rightarrow x \quad b_{n_k} \rightarrow x$  

3--- usa continuità per calcolare ciò che voglio  
continuità: $|f(a_{n_k}) - f(b_{n_k})| = |f(x)-f(x)| \rightarrow 0$  
che è la condizione di continuità uniforme  

#### Convergenza uniforme preserva continuità

Se $f_n→f$ uniformemente e ogni $f_n$​ è continua, allora anche $f$ è continua.  

esiste N tale che, oltre quello:  
$f_n \rightarrow f$ uniforme: $|f_n(x)-f(x)| < \frac{\epsilon}{3}$  
$f_n$ continua in $x_0$: $|x-x_0| <\delta \rightarrow |f_n(x)-f_n(x_0)|<  \frac{\epsilon}{3}$  

voglio dimostrare:   $|x-x_0| <\delta \rightarrow |f(x)-f(x_0)|<  \epsilon$  
stimiamo la seconda parte:  
$|f(x)-f(x_0)| \le |f(x)-f_N(x)  |+| f_N(x) - f_N(x_0)| + |f_N(x_0) -f(x_0)| = 3 \frac{\epsilon}{3} = \epsilon$  
dimostrato  

perchè ogni termine è minore di $\frac{\epsilon}{3}$:   
primo e terzo: convergenza uniforme (tra parentesi deve esservi la stessa cosa)  
secondo: continuità di ogni $f_n$, quindi anche di $f_N$ (tra parentesi devono esserci cose diverse)  
in basso c'è N, non n, perchè sto prendendo tra le $f_n$ quelle abbastanza avanti per cui queste definizioni valgono  
ma in pratica $f_N$ sarebbe $f_n$  

**Comce cancellare roba**

$\forall \epsilon > 0 \quad \exists N: \quad \forall n>N$  
f uniforme continua: se $|x-y|<\delta _1$, $\quad |f(x)-f(y)| <\epsilon$  
f convergenza uniforme: $|f_N(x)-f(x)|< \epsilon$  
f continua:  se $|x-x_0|<\delta _2$, $ \quad |f(x)-f(x_0)| <\epsilon$  
limite $\lim f(x)-g(x)=0$: $|f(x)-g(x)| < \epsilon$  

prendiamo $\delta = min\{\delta_1, \delta_2, \delta _3\}$
