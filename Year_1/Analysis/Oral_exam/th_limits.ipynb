### Unicity of limit

**Th**

$f : (E ∖ \{x_0\}) \rightarrow R \quad \quad x_0 \in E$ is an accumulation point  
if possible to extend, there is an unique way to extend $f$ to all $E$ with continuity  
means $f$ can have a single limit in $x_0$   

**Proof**

prove by absurd:  

1. assume there are two limits  

so two values $l_1 \neq l_2$ make f continuous in $x_0$  
so define two funcs, $f_1, f_2$, the same on all R except in $x_0$  

2. take neighborhoods:  

y axis: take neighborhood of each $l$, small enough that they don't intersect: $V_1 \cap V_2 = \empty$  

x axis: since f is continuous in $x_0$, exist a neighborhood of $x_0$ such that $f(U_i∖\{x_0\}) \subset V_i$  
(bc of definition of continuity)


3. intersection of $U$ mapped into intersection of V   

$x_0$ is accumulation point, so INTERSECTION IS NOT EMPTY  
contraddiction: not empty set cannot be mapped to empty set  

**Recap**

prove by absurd  
assume 2 limits exist  
intersection of neighborhoods of $x_0$, not empty, mapped to intersection of neightborhoods of limits, empty  
contraddiction: not empty mapped to empty

### Limiti: criteri di confronto

f,g,h  hanno dominio R  

- confronto normale  
$f(x) \le g(x) \quad \forall x \in A$  
SE hanno limite, $\lim_{x_0} f \le \lim _{x_0} g(x)$  

- conseguenze sul comportamento asintotico  
se la piccola diverge $f(x) \rightarrow \infty$, anche la più grande diverge  
stessa cosa se $g(x) \rightarrow -\infty$  

- teorema dei carabinieri  
$f(x)<h(x)<g(x)$  
se f e g hanno limite l, anche h ha limite l  
di sicuro, non devo controllare che esiste, me lo garantisce  

**Proofs**

1--- 
se fosse $\lim \; f > \lim \; g$:  
funzione ausiliaria $\lim f(x)-g(x) >0$  
ma non può essere perchè $f(x)-g(x) <0$  

3---
se f e g hanno stesso limite l, in ogni intorno U di l si ha deifnitivamente $f(x) \in U$ e $g(x) \in U$  
visto che U è un intervallo, anche $h(x)$ è definitaivamente in U, quindi $h(x) \rightarrow l$

### Discontinuità funzioni monotone

hp:  
- f monotona  

th:  
- $D=disc(f)$ è numerabile

**Proof discontinuità**


1--- Discontinuità sono solo salti:  
in R, tutti i punti sono di accumulazione: quindi per la definizione di limite di una funzione monotona, i limiti unilaterali esistono finiti in ogni punto (eventualmente diversi destro e sinistro)  
quindi l'unica discontinuità possibile è quella di prima specie  

2--- Ad ogni punto associamo un intervallo razionale e costruiamo una mappa:  
definiamo i limiti destro e sinistro $a_x = \lim _{x \rightarrow x^-} f(x), b_x = \lim _{x \rightarrow x^+} f(x)$  
prendiamo un punto $x$ dove questi sono diversi  
assumiamo senza perdita di generalità che f sia crescente, quindi $a_x < b_x$  
l'intervallo $(a_x, b_x)$ è non vuoto, e contiene almeno un numero razionale $q_x \in Q$ (perchè Q è denso)     
ho la mappa $x \mapsto q_x$  
in pratica: identifico ogni salto (discontinuità) con un numero razionale nella parte saltata  

3--- Proviamo che la mappa $x \mapsto q_x$ è iniettiva  
se $x_1 \neq x_2$, gli intervalli $(a_{x_1},b_{x_1})$ e $(a_{x_2},b_{x_2})$ sono disgiunti  
cioè vale l'iniezione $\varphi : D \rightarrow Q$  
iniettiva se $a\neq b \rightarrow f(a) \neq f(b)$  
quindi D è numerabile, perchè in iniezione con un insieme numerabile Q   

**Recap proof**

1. discontinuità sono solo salti  
2. identifica ogni salto con un razionale al suo interno, $q_x$   
3. costruiamo mappa per i salti $\varphi: D \rightarrow Q \quad x \mapsto q_x$  
4. prova mappa è iniettiva, quindi D è numerabile come Q

**Recap: come avevo definito l'esponenziale**

R,S sono due gruppi densi, ordinati e continui

esiste un'unica funzione $\varphi : R \rightarrow S$ tale che:  
- $\varphi(u)=v$ (linee sovrapposte in un punto specifico)  
- isomorfismo: $\varphi(x +_R y)= \varphi(x)+_S \varphi(y)$  
- positività: se $x \ge e_R$, allora $\varphi(x) \ge e_S$
