### Banach-Cacciopoli

X completo

$f: X \rightarrow X$ L-lipschitziana, con $L < 1$ (f is a contraction)  

esiste solo un punto limite (x dove $f(x)=x$)  

**Proof**

1--- scrivo la ricorsione  

$x_0=p$ (p è un punto a caso)  
$x_{k+1}=f(x_k)$  

2--- dimostra che è una successione di Cauchy    

devo stimare la distanza tra due entries, richiesta dalla condizione di Cauchy:  
$|x_2- x_1| = |f(x_1) - f(x_0)| \le L |x_1 - x_0|$  
$|x_3 - x_2| = |f(x_2) - f(x_1)| \le L |x_2 - x_1| \le L^2 |x_1- x_0|$  
$|x_{m+1} - x_m| \le L^m |x_1 - x_0|$  

la distanza tra una coppia qualunque è la somma delle distanze di tutti gli step intermedi  
quindi tra una coppia qualunque ho una serie geometrica:   
$|x_n-  x_m| \le \Sigma _n^{m-1} |x_k - x_{k+1}| \le \Sigma L^k |x_1 - x_0|$  
se $L <1$ la serie converge  
quindi se prendo il limite è 0, $|x_n-x_m| \rightarrow 0$, $x_n$  è una successione di Cauchy  

3--- dimostra che ha un punto limite  

voglio dimostrare $f(x)=x$ con l'unicità del limite  
poichè X è completo, tutte le sequenze di Cauchy convergono, e anche questa: $x_n \rightarrow x \in X$, quindi anche $x_{k+1} \rightarrow x$   
per continuità: $x_{k+1}=f(x_k) \rightarrow f(x)$ (continuità è aggiungere f alla convergenza $x_{k+1} \rightarrow x$)    
per unicità del limite, $x = f(x)$  
quindi x è un punto fisso  
simile al ragionamento fatto nel th di weierstrass

3-- dimostra non ci sono altri limiti 

non possono essercene altri, perchè se y lo fosse avrei:  
$|x-y|= |f(x) - f(y)| \le L \, |x-y|$  
poichè $L<1$, accade solo se $|x-y|=0$, cioè $x=y$

### Esistenza e unicità locale, ordine n

$f  : \Omega \rightarrow R$  

soddisfa cauchy-lipschitz (continua, lipschitz in y)  

problema di cauchy di ordine n ha una unica soluzione locale  
cioè, definito un intervallo I, c'è una sola y che risolve in quell'intervallo  

passo 1: passaggio a sistema del primo ordine  

creo un vettore lungo n, in cui le componenti sono y e le sue n-1 derivate   
$Y = [y, y', y'' ... y^{(n-1)}]$ 

quindi il problema si può riscrivere:  
$\begin{cases}
Y'(x) = F(x, Y(x)) \\
Y(x_0) = y_0 \end{cases}$  

$y_0$ è un vettore con tutte le condizioni iniziali 

perchè  per scrivere le consizioni iniziali, posso prendere le componenti del vettore:  
$y''(x_0)=y_2$ diventa $Y_2(x_0)=y_2$  
fino a $Y_{n-1}$, così copro tutte le condizioni iniziali  

poi, tutte le entries nel vettore sono legate le une alle altre, poichè sono derivate successive di y:    
$Y_1' = Y_2, Y_2' = Y_3 ...$   

prima avevo un problema scalare:  
$y'(x)=f(x,y(x))$  
ora ho un problema vettoriale (F non è l'integrale, è solo per far capire che è diversa da f):    
$Y'(x) = F(x,Y(x))$  

for clarity:  

$ Y = \begin{bmatrix} y \\ y' \\ ... \\ y^{(n-1)} \end{bmatrix} =  \begin{bmatrix} Y_1 \\ Y_2 \\ ... \\ Y_n \end{bmatrix} \quad $  $F(x,Y) = Y' = \begin{bmatrix} y' \\ y'' \\ ... \\ y^{(n)} \end{bmatrix}  = \begin{bmatrix} Y_2 \\ Y_3 \\ ... \\ f(x, Y_1, Y_2, ... Y_n) \end{bmatrix}$ 

$y^{(n)}=f(x,y,y'..y^{n-1})=f(x,Y_1,Y_2,...Y_n) = f(x,Y)$  

passo 2: uso teorema di unicità per eq del primo ordine  

f è continua, quindi F è continua  

controllo che F è lipschitziana in y, sapendo che y lo è:  
poichè è un vettore, calcola la distanza come con Pitagora, con tutte le componenti    

$|F(x,Y)-F(x,Z)| = |Y'-Z'|= \sqrt{(\Sigma _2^n |Y_k-Z_k|^2)+|f(x,Y)-f(x,Z)|^2} \le \sqrt{|Y-Z|^2+L^2|Y-Z|^2} = \sqrt{1+L^2}|Y-Z|$  

$\Sigma _2^n |Y_k-Z_k|^2 \le |Y-Z|^2$, perchè Y' non ha la prima componente $Y_1$, quindi manca una entry e la somma è minore  

verifica Cauchy-Lipschitz, quindi esiste una unica Y soluzione (dal teorema di unicità per il primo ordine)  
 
prendo $Y_1$, la prima componente, che è di classe $C^n$ (in quanto le altre entries del vettore sono le sue derivate, che quindi esistono)  

### Esistenza globale

se f è continua e Lipschitz in y  
E SUBLINEARE in un intervallo I  
la soluzione può essere estesa in tutto I  

La sublinearità è una condizione sufficiente per garantire l'esistenza globale della soluzione.

sublineare: cresce più lenta di una funzione lineare  
f è sublineare se: $|f(x,y)| \le A(x)+B(x)|y|$  
così la derivata f' non può essere infinita, la soluzione non può esplodere in tempo finito  

**Proof**

f continua e Lipschitz, quindi esiste un'unica soluzione  
l'unico motivo per cui potrebbe non essere estendibile è se va a $\infty$ in tempo finito  

1--- usa condizione di sublinearità  
$|y'| = |f(x,y)| \le A+B|y|$  
define $v=|y|$ per comodità  
note y,v,A,B are all funcs of x  

2--- Gronwall inequality  
se $v' \le A +Bv$, allora:  
$v(x_0) \le v(x_0)e^{\int_{x_0}^x B(s) ds} + \int_{x_0}^x A(t) \; e^{\int _t^x B(s) \, ds} dt$  
quindi $|y|=v$ è finita/limitata, purchè A e B siano intergabili, e lo sono perchè f è continua per ipotesi  
poichè y è limitata, non c'è nulla che impedisca di estendere
