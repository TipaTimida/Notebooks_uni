### Fundamental th of integral calculus

hp:  
- f continuous defined in interval $I$, $c \in I$  
- $F(x) = \int _c^x f(t) dt$  

th:  
- F(x) is derivable in $I$  
- $F'(x) = f(x)$

**Proof**

1. write thing to prove, $F'(x)$  

$F'(x) = \lim _{h \rightarrow 0} \frac{F(x+h)-F(x)}{h}$  

for the numerator, use the definition of F:  
$F(x+h)-F(x) = \int _c^{x+h} f \, dt - \int _c^{x} f \, dt = \int _c^{x+h} f \, dt + \int _x^{c} f \, dt =  \int _x^{x+h} f \, dt$  

2. use the MVT  

f continuous on $[x, x+h]$, so exists $c_h \in [x, x+h] : $   
$\int _x^{x+h} f(t) \, dt = f(c_h) \, h$

recap:  
$F'(x) = \lim _{h \rightarrow 0} \frac{1}{h} f(c_h) h = \lim _{h \rightarrow 0} f(c_h) = f(x)$  

last = is bc when $h \rightarrow 0$, $ c \in [x, x+h] \rightarrow c \in [x,x]$

### Criteri confronto asintotico e puntuale

$f, g \ge 0$ and Riemann integrable

- confronto asintotico $\ll$  
se $\frac{f}{g} \rightarrow 0$, se il grande converge pure il piccolo, se il piccolo diverge pure il grande

- confronto asintotico $\sim$

**Proof**

the integrals exist bc of the th of integrability of positive funcs

### Criteri di integrabilità

**Recap**

Riemann integrabile se: $I^* =I_*$  
$I^* = inf \; S^*(P)$  
$I_* = sup \: S_*(P)$  

$I(f)$: the smallest upper bound / biggest lower bound of the integral (don't tell which suddivision, it picks the optimal one)  

$S(f,P)$ upper/lower bound of the integral with some suddivision P  


**Note**

if pick a suddivision with more points, the area is more correct  
so the upper bound is smaller and lower bound bigger  

**Per usare in seguito**

integrabile se:  

metodo epsilon deta:  
$\forall \epsilon >0 \quad \exist P : \quad S^∗ ( 𝑓 , 𝑃) − 𝑆_∗ ( 𝑓 , 𝑃) < 𝜀$  

metodo successione:  
esiste una successione $P_n$ tale che: $S^*(P_n)-S_*(P_n) \rightarrow 0$  

$I^* \le S^*$, perchè il minimo degli upper bound  

$sup (-f)=-inf (f)$, quindi $I^*(-f)=-I_*(f)$

**Th di integrabilità**

1. f è Riemann integrabile se e solo se:

$\forall \epsilon >0 \quad \exist P : \quad S^∗ ( 𝑓 , 𝑃) − 𝑆_∗ ( 𝑓 , 𝑃) < 𝜀$

2. se e solo se f è Riemann integrabile, esiste una successione $P_n$ tale che, per $n \rightarrow \infty$:  

$\lim S^*(f, P_n) = \lim S_*(f, P_n)= \int _a^b f$  

**Proof 1**

se è vero, è integrabile:  
$I^*(f)-I_*(f) \le S^*(f,P)-S_*(f,P) < \epsilon \quad \forall \epsilon >0$  
so means $I^*(f)-I_*(f)=0$, it's condition of integrability  

se è integrabile, è vero:  
esistono suddivisioni dove:  
$I^*(f)\ge S^*(f, Q) - \frac{\epsilon}{2}$   
$I_*(f)\le S_*(f,R) + \frac{\epsilon}{2}$  
prendo $P=Q \cup R$, suddivisione con tutti i punti di Q ed R   
$S^*(P)-S_*(P) \le S^*(Q)-S_*(R) \le (I^* + \frac{\epsilon}{2}) - (I_* -  \frac{\epsilon}{2}) = \epsilon$

**Proof 2**

se è integrabile, è vero:  

uso la descrizione di proof 1 per l'integrabilità, quindi scelgo una suddivisione $P_n$:  
$S^*(P_n)-S_*(P_n) < \frac{1}{n}$  
$I^* \le S^*(P_n) \le S_*(P_n) + \frac{1}{n} \le I_* + \frac{1}{n}$  

prendo il limite $n \rightarrow \infty$  
poichè è integrabile, $I^* = I_* = \int f$, quindi  
$\lim S^* = \lim S_* = \int f$  


se è vero, è integrabile:  
se $\lim S^*(P_n)-S_*(P_n) = 0$, allora $\forall \epsilon >0$ esiste n tale che: $S^*(P_n)- S_*(P_n) < \epsilon$  
che è la definizione di integrabile

### Integrability continuous functions

$f : [a,b] \rightarrow R$ è continua  

allora f è limitata e integrabile  

Weierstrass th: f continua su un compatto è limitata  

Heine-Cantor: continua su compatto è  uniformemente continua  

uniformemente continua vuol dire che se $|x-y|<\delta \rightarrow |f(x)-f(y)|< \epsilon$, e quel $\delta$ vale per tutti  

quindi prendo una suddivisione P in cui ogni intervallo è largo meno di $\delta$, così:  
$S^*-S_* = \Sigma (x_k -x_{k-1})(sup \; f - inf \; f) \le \epsilon \Sigma (x_k -x_{k-1}) = \epsilon (b-a)$  

per $\epsilon \rightarrow 0$ posso renderlo piccolo a piacere, quindi è la condizione di integrabilità    


### Integrability monotone function

se è monotona e in un compatto, è integrabile  

se è monotona e in un compatto, è limitata (perchè $f(a)<f(x)<f(b)$)  
per semplicità, assumo sia crescente  

condizione per essere integrabile è ch esiste una successione $P_n$ tale che $S^*(P_n)-S_*(P_n)\rightarrow 0$  

prendo una suddivisione $P_n$ equispaziata con n punti, così:  
$S^*(P)-S_*(P)= \Sigma (x_k - x_{k-1})(sup \; f - inf \; f) = \Sigma \frac{b-a}{n}(f(x_k)-f(x_{k-1})) = \frac{b-a}{n}(f(b)-f(a))$   

è telescopica  
per $n\rightarrow 0$ tende a 0, come volevo
