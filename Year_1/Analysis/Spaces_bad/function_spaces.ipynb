### Metric spaces, basics

**Spaces**

- topological
- metric (subset of topological)

they're ways to express a geometrical space  


- metric  
need to define a <u>formula for a distance</u> in it, that <u>returns a scalar value</u> (usually euclidean norm)

- topological  
no value for distance between 2 points, can only tell <u>closeness as a yes/no thing</u>, boolean value  
(use sets to group stuff close together, 2 things are close if belong in same set)

**Distance in metric space**  

$(M, d)$

$d: M \times M \rightarrow R$  is the distance between 2 points in a set/space M  
output must be a scalar value  

properties of distance func:  
- always positive
- commutative  
- triangular inequality  




**Types of distance functions**

1. $d_1$: Manhattan distance, $L_1$ norm  
rhombus centered in origin with diagonal long 2   
    - $d(a,b) = |a_1-b_1| + |a_2- b_2|$  
summing the absolute differences across each dimension


2. $d_2$: Euclidean norm, $L_2$ norm   
circle centered in origin with radius 1
    - the commonly used one  
distance in a straight line  


3. $d_{\inf}$: Chebyshev distance  
square centered in origin with side 2  
    - $d(a,b) = max(|a_1-b_1|, |a_2- b_2|)$   
take the largest difference across any dimension (ignore all others dimensions)

---

**Relationship**

$d_1/2 \le d_{\inf} \le d_2 \le d_1 \le 2d_{\inf}$  

note that highest term, $d_1$, is used in expression for lower bound and viceversa  

to remember, manhattan is on the inside, so need to use it more times to get to a point, so it is bigger

**Norma**

$\varphi: V \rightarrow R$  

properties:  
1. positivity  
2. separation 
3. triangular inequality  
4. omogeneity: $𝜑(𝜆𝑣) = |𝜆| · 𝜑(𝑣)$

can be written $𝜑(𝑣), |𝑣|, ∥𝑣∥$

**Norma p**  

all norms can be defined as:  
$|𝒙|_𝑝 =\sqrt[p]{|x_1|^p + |x_2|^p ...}$  

p=1: Manhattan norm (one horizontal or vertical step counts as 1)  
p=2: euclidean norm (usual one)  
p=$\infty$: Chebyshev distance (same as Manhattan, but rotated 45 degree and scaled by $\sqrt{2}$)

topological:

**Topology**

topology: a way of describing which points are "close" to each other—without needing a distance (used in topological spaces)

used to define:  
Continuous functions (using open sets instead of limits!)  
Convergence of sequences or nets  
Compactness, connectedness, and many other deep ideas

inuitively:  
have a set of points  
a topology is a choice of which points i want to call 'open'    
rules are:  
union of open sets stays open  
finite intersection of open sets stays open  
empty space and full space must be open  

formal definition:  
A topology on a set X is a collection T⊆P(X) (i.e., a set of subsets of X) such that:  
-  ∅∈T and X∈T
- Any union of open sets is in T
Any finite intersection of open sets is in T

Elements of T are called open sets  
(so it's a choice of which sets i call open  )


## Definitions 

### Topological spaces

**Def: Topology**

A topology on a set X is a set $T$ of subsets of X, called the open sets,
which satisfies the following conditions: 

1. The empty set ∅ and the set X are open.
2. Any finite intersection of open sets is open.
3. Any union of open sets is open.

intuitively: it's a choice of which sets to call open  

topologies are a bit like a suddivision (in integral notes)  
if a topology has more open sets than another, it's called finer (a finer suddivisions is one with more points)   

**Def: neighborhood**

A neighborhood of a point x ∈ X is an open set U such that x ∈ U

**Nomenclature**

$T$: set of all open sets defining a topology  
$(X, T)$: the set X with topology T  
X is a topological space  

**Def: closed space**

if X is a topological space  
a subset of X is closed if it scomplement is open  

thus closed set satisfy the same 3 conditions as open sets  

note: the empty and full set, in any topology, are the only ones who are at the same time open and closed

### Compact spaces

**Def: compact**

formal:  
A space X is compact if every open cover contains a finite subcover.  
means: Whenever you try to cover it with open sets, you can always do it using just a finite number of those sets.

intuitive:  
closed + bounded = compact (Heine-Borel)  

eg: [0,1]: compact  
(0,1): not compat
