### Neighborhood

neighborhood of $x_0 \in R$ of radius L:  
$U_L(x_0) = \{ x \in R : |x-x_0|<L\}$  

all points closed that L, essentially draw a bubble of radius L around the point $x_0$  

### Open / closed set

a set $E \in R$ is:  
open if it's made up of only internal points  
closed if its complementar is open  

an internal point of a set E is a point $x \in E$ for which exists a neighborhood $U(x_0) \subset E$  

**Example open/closed sets**

E: circle considering border, $E = \{(x,y) : x^2 +y^2 \le R^2\}$  
points of border are not internal points, no matter how small the neighborhood, won't fit in E  
so closed set  

E': circle exclusing border, $E' = \{(x,y) : x^2 +y^2 < R^2\}$  
for all points can find a neighborhood, bc don't know exactly where border is, so can always go further  
so open set  


essentially, if in definition appear strictly bigger/smaller $(<, >)$ it's open, bc not considering the border  
if appear equal as well $(\le, \ge)$ it's closed, bc consider the border too  

so open set can be written: $\{x \in R : f(x) >0\}$  


### Bounded set (limitato)

a set $E \subset R$ is limited if exists $L>0$ such that $E \subset U_L(0)$  
so if all of it is contained in a neighborhood of 0
