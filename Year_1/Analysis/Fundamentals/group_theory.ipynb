**Algebric sructure**

$(S, +)$  
no particular requirements, just basic shape  


**Group**  

$(G, +)$  
  - neuter element
  - inverse element
  - associativity   

if also commutativity, it's abelian/commutative group


**Ring**  

$(R, +, *)$
  - $(R, +)$ is a group
  - $*$ associative
  - $+$ and $*$ distributive


if $*$ has neuter 1, it's a ring with unity  
if $*$ is commutative, it's abelian ring  

essentially, field without inverse for $*$

**Field**  

$(F, +, *)$
  - $(F, +)$ commutative group with neuter 0
  - $(F $ \ $\{0\}, *)$ commutative group with neuter 1
  - $*$ distributive wrt to $+$  

elements 0 and 1 can be any number, but must be 2 different ones



**Other stuff**

- gruppoid: algebric structure wit 1 op

- semi-group: gruppoid + associative op

- monoid: semi-group + neuter element

- group: monoid + inverse element

#### Numerical sets and group theory

- monoid: $(N +), (N, \cdot)$
- group: $(Z, +)$
- ring $(Z, +, \cdot)$
- field: $(Q, +, \cdot)$ ordered, dense, not continuous  
$(R, +, \cdot)$  ordered, dense, continuous

**Properties**

- total ordering: each pair can be confronted (yes if ordering given by <, no if given by $\subseteq$)  
if not, partial ordering  

- dense: between 2 numbers, always a third (yes Q, no N)  

- continuos: no gaps in number line (yes R, no Q, miss for example $\pi$ or $\sqrt{2}$)  
formally, each subset has an upper and lower bound

### Isomorphism

**Isomorphism**

between 2 groups $(G, +), (H, *), \phi$ is an isomorphism if:
1. $\phi$ is bijective
2. $\phi (a+b) = \phi(a) * \phi (b) $  
and also with ops swapped (if true in only one direction, omomorphism)


note that neuter elements of two ops map into each other  

**Construct isomorphism, unicity of R**


between 2 groups, exists only 1 $\phi : G \rightarrow H $ such that:  

1. $\phi$ is isomorphic
2. $\phi(a) = b$    
(make two specific elements correspond)
3. $a \le b \longleftrightarrow \phi(a) \le \phi(b)$  
(keep ordering)  

take 2 number lines, there's only one way to superimpose them so that:  
a specific number a corresponds to a specific one b (point 2)  
the lines are oriented in a specific direction (point 3)  

this gives proof of unicity of R, since every field similar to it is in an isomoprhism to R, so it's R

#### Elementary functions  

**Exponential**  

$ \varphi  = a^x$  such that:  
- $ \varphi : (R, +) \rightarrow (R^+, \cdot)$  
- $\varphi(1) = a$  

---


since it's an isomorphism, get:  

$\varphi(x+y) = \varphi(x) + \varphi(y)$   
$a^{x+y} = a^x \cdot a^y$   

also $ \varphi(0)  = 1$, bc neuter elements map into ecah other
