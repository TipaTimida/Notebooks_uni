**Useful inferences**

De Morgan laws  
$\neg (p \land q) \longleftrightarrow \neg p \lor \neg q$  
$\neg (p \lor q) \longleftrightarrow \neg p \land \neg q$

implications  
$ p \rightarrow q \longleftrightarrow (\neg q \rightarrow \neg p)$  
$ p \rightarrow q \longleftrightarrow (\neg p \lor q)$    (to prove implication is wrong, find counterexample)

$ p \longleftrightarrow q$  p is condition necessary and sufficient for q
$ p \rightarrow q$ p is condition sufficient for q, q is necessary for p


### Set theory

#### Basic theory

**Axiomatic set theory**

axiomatic set theory: beside logic symbols, in the axioms only define $\in$, all other symbols can be derived

**Russel paradox**  


how to define a set: can only create a set taking values from another set  
yes: $ \{ x : x \in A \land p(x) \}$    
no: $\{ x : p(x) \}$  

second way leads to the russel paradox, of sets that contain sets that don't contain themselves  
$ A = \{x : x \notin x\} $ set of sets that don't contain themselves, is $A \in A$? 

other consequence of this paradox is that universe set does not exist  
universe set: hypothetic set that contains everything  
bc if A is the universe set, what is $A \cup \{A\}$


#### Particular sets

**Family of sets**

it's a set of sets  

given a family F, can define union and intersection between all its sets:  

- union set: $\cup F$  
contains all elements at least in one of the sets contained in F

- intersection set: $\cap F$  
contains all elements that appear in all sets contained in F


**power set**: $P(A)$  
called set of parts  
contains all subsets (proper and improper) of A

**quotient set**: $\frac{A}{R}$  
set of all equivalency classes when relationship R is applied to A  
ex. if R is mod 2, $\frac{A}{R} = \{[0], [1]  \}$

**partition**  
set of disjoint sets that, if united, recreate the original set  
the quotient set is a partition

**Cantor set**: C  

start from an interval  
divide in 3 equal parts, take away middle one, obtain 2 subintervals  
for each subinterval, repeat the divide per 3 and take away middle part  
recursive  

properties:  
- uncountable
- has measure zero (bc total length of interval removes is 1: $\frac{1}{3}+ \frac{2}{9} + \frac{4}{27} ...= 1$)

**Measurable set**  

a set for which i can define consistently a Lebesque measure  
rem that any countable set has Lebesque measure zero  

Q: is countable, so has measure zero, so it's measurable  
R: measurable, has measure $\infty$  

R\Q: is the complement of Q in R (which are both measurable), so it's measurable  
has measure $\infty - 0 = \infty$

$\{(x,y) : (x,y) \in A \times B \land R(x,y)\}$

#### Cardinality

**Cardinality**

Dedekind infinite set, 2 possible definitions:
1. those that can be put in bijection with a proper subset of themselves (Hilbert hotel paradox, $x \mapsto 2x$)  
2. S is infinite if exists $\phi : S \rightarrow S$ injective and not surjective

cardinality and existance of functions:  
$\#A \le \#B \longleftrightarrow \exists f: A \rightarrow B  $ injective  
$\#A \ge \#B \longleftrightarrow \exists f: A \rightarrow B  $ surjective  
$\#A = \#B \longleftrightarrow \exists f: A \rightarrow B  $ bijective  

**Types of infinity**

- countable: can reach any numer, can count on them   
can be in biunivoc correspondance with N  
cardinality aleph-null

- uncontable: can't count, any number is at infinite distance, don't know what next number is  
cardinality aleph-one   
 
 
Continuum hypothesis: there is no cardinality in between countable and uncountable  


**Infinity of numerical sets**

countable: N, Z, Q (bc they're quotient sets of $N \times N$)   

uncountable: R, C (bc they're power set of N, and cardinality of power set is always bigger)

### Numerical sets

##### Define N

2 methods, both use recursion:  
- Von Neumann integers  
- Peano axioms  

**Von neumann integers**

$0 = \emptyset$  
$n+1 = n \cup \{n\}$

therefore:  
$0 = \emptyset$   
$1 =  \{\emptyset \}$  
$2 =  \{\emptyset, \{\emptyset \} \}$

**Peano axioms** 

given $(N, 0, \sigma)$:  
set N  
element $0 \in N$  
function $\sigma : N \rightarrow N, x \mapsto  x+1$   

Peano axioms are:  
1. $\sigma$ is injective
2. $\sigma (N) = N \backslash \{0\}$  
to not get a value i alreasy encountered  

2 sets satisfying axioms of Peano are isomoprh, so N is unique

##### Define other sets

**Z**  

$ Z = \frac{(N \times N)}{\sim} $


formal:  <u>equivalency classes in difference</u>   between 2 number of N  
easy:  each number of z defined as difference of 2 numbers in N  under the equivalency relation (so (1,2)=(2,3) = -1)  

$\sim$ is the equivalency relationship, can also be written R  


**Q**  

$ Q = \frac{(Z \times (Z \backslash \{0\}))}{\sim} $  

formal:  <u>equivalency classes in division</u> between 2 number of Z   
easy:  each number of Q defined as fraction of 2 numbers in Z, and the denominator can't be zero  



**R**

$R = \{A \in P(Q): A \neg \emptyset, A$ has upper bound $\}$ 


each number is approximated as the upper bound of the set of all the elements smaller than itself   
eg:  $\sqrt{2} = sup(\{x \in Q : x^2 < 2\})$

called Dedekind section

**Induction**

1. prove it's true for n=0
2. prove if its true for n, it is for n+1 too

ex. $(1+x)^n \ge 1 + nx$
1. obv true for n=0
2. suppose it's true for n:
  - take left side written with n+1, try to reobtain left side with n  
  $(1+x)^{n+1} = (1+x)^n (1+x) \ge$
  - substitute the part with n with the thing i suppose it's equal to  
  $ \ge (1+nx) (1+x) = $
  - try to obtain something like the right side, but with n+1 instead of n  
  $ = 1 + nx + x + nx^2 \ge $  
  $ \ge 1 + (n+1)x $

hint: try to start from both edges, useful to see how to connect them
