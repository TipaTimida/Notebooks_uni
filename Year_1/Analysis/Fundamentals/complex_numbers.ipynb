### Cyclotonic poly - roots of unity

**cyclotonic polynomial**: poly with roots of unity, $z^n -1 = 0$  
the real parts of its roots are sols of $cos(\frac{2\pi}{n})$  
bc the roots are the lines that make such angles (modulo 1 and forming a reguar polygon, draw to check)

#### Ex: compute $cos(\frac{2\pi}{5})$

1. cyclotonic polynomial: $z^5 -1 = 0$  
i'll be using the version with the real part of the roots, bc i want the cos  

2. find obv sols: $z=1$  
$z^5 -1 = (z-1)(z^4+z^3+z^2+z+1) = 0$  

3. find other sols: solve $(z^4+z^3+z^2+z+1) = 0$

tips:  

- multiply by $\overline{z}^2$, since i'm working with roots of 1 and $z\overline{z} = |z|^2 = 1$  -
- subsitute $z+\overline{z} = 2Re(z)$
- substutute again $Re(z) = x$  
obv, if i'm substituting $x=Re(z)$, i'll find the cos  
to find sin should make different moves, to be able to substitute $x=Im(z)$

now i have equation:  $4x^2 +2x -1 = 0$  
solve it to find the real parts of the other roots of 5, so the cosines  
note that the 4 remaining roots have cosines two by two equal, that's why it has 2 solutions  
since $\frac{2\pi}{5}$ is in first quadrant, choose positive solution




### Algebra fundamental theorem

Every non-constant polynomial P(z) with complex coefficients has at least one root in the complex numbers C  

is fundamental because it garantees that the field C is algebraically closed, meaning every polynomial with complex coefficients has roots in C

**Algebrically closed**

C is a field, but not totally ordered -> so no prob with theorem of unicity of R 

Fundamental theorem of algebra: a field is algebrically closed if in it i can factorize any polynomial with coeffs in that field  
that's why n th root in C has always n solutions 

R isn't, bc can't factorize $x^2 +1 = 0$  
C is, and is called the algebric closure of R, bc it's the smallest algebric field that contains R
