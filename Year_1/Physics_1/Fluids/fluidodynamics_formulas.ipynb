### Water bubble and tension

the pression from the membrane tries to make the membrane smaller, to collapse on itself, so pushes towards inside, like the external pressure  
the internal pressure tries to make the ball expand  

so to have a bubble:  
$p_{internal} = p_{external} + p_{membrane}$  
$p_{internal} = p_{external} + \frac{2 \alpha}{R}$  


**superficial tension / membrane pression**

$p_m = \frac{dW}{dV} = \alpha \frac{dA}{dV} = \alpha \frac{4 \pi (R + dR)^2 - 4 \pi R^2}{4 \pi R^2 dR} = \frac{2 \alpha}{R}$


$\alpha$: coefficient of superficial tension  

### Connected air bubbles
