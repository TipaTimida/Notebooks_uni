## Superimpose harmonic motions

eg: wave interference

in exercises, rather than adding $x(t)$, might be more useful adding $\ddot{x}$  

so get $\ddot{x}_1$ in normal way, by doing $\dot{E}=0$  
the other usually i'm given $x(t)$, simply derivate twice to get $\ddot{x_2}$  

then add: $\ddot{x} = \ddot{x_1}+\ddot{x_2}$  

### Harmonic motions in same direction

$\begin{cases}
x_1(t)=A_1 \cos(\omega_1t+\varphi_1) \\
x_2(t)=A_2 \cos(\omega_1t+\varphi_2) \\
\end{cases}$  

$x(t)=x_1(t)+x_2(t)$

**Properties**

$x(t)$: displacement at each moment  

each motion is described by:  
- A: amplitude  
- $\omega$: pulsation 
- $\varphi$: phase  

#### Case: $\omega _1 = \omega _2$

$x(t)= A \cos (\omega t + \varphi)$

$\begin{cases}
\tan \varphi = \frac{A_1 \sin \varphi _1 + A_2 \sin \varphi _2}{A_1 \cos \varphi _1 + A_2 \cos \varphi _2} \\
A^2 = A_1^2+ A_2^2 + 2A_1A_2 \cos (\varphi_1 - \varphi _2)
\end{cases}$

**Intuitively**

eg: 2 springs with same k  

resulting motion is again an HARMONIC MOTION  
can write it as a motion with:  
same pulsation as starting ones  
$A, \varphi$ function of the starting ones  

so the time interval for the two signals/waves to touch the same point is:  
$\Delta t = \frac{\varphi _1 - \varphi _2}{\omega}$  

bc $\Delta t = \frac{\Delta s}{v}$  


**Types of phase difference**

case | condition |  $A$ | $\tan \varphi$ | what happens
--- | --- |--- | --- | ---
concordanza di fase | $\varphi_1 - \varphi_2 = 0$  |  $A_1+A_2$ | $\tan \varphi _1 = \tan \varphi _2$ | constructive interference
controfase |  $\varphi_1 - \varphi_2 =  \pi$ | $\|A_1-A_2\|$ | $\tan \varphi _1 = \tan \varphi _2$ | destructive interference
quadratura | $\varphi_1 - \varphi_2 =  \pm \frac{\pi}{2}$  |  $\sqrt{A_1^2 +A_2^2}$ | no easy formula  

**Case $A_1=A_0$, unsure phases**

$\begin{cases}
A = 2A_1 \cos(\frac{\varphi_1 - \varphi_2}{2}) \\
\tan \varphi = \frac{\sin \varphi _1 + \sin \varphi _2}{\cos \varphi _1 + \cos \varphi _2}
\end{cases}$

so in:  
- concordanza: $A=2A_1$  
- controfase: $A=0$  
- quadratura: $A = A_1 \sqrt{2}$

#### Case: $\omega _1 \neq \omega _2, \varphi_1 = \varphi _2 = 0$

i get $\ddot{x}(t)$ NOT PROPORTIONAL to $x(t)$  
so the resulting motion is NOT HARMONIC  

##### Case $A_1 = A_2$

using formula di Prostaferesi:  
$x(t) = 2A_1 \cos (\frac{(\omega _1 + \omega _2)t}{2})  \cos (\frac{(\omega _1 - \omega _2)t}{2})$  

**Intuitively**

it's not harmonic motion      
but can interpret it as a kind of harmonic motion with AMPLITUDE MODULATED IN TIME  

- pulsation: $\frac{\omega _1 + \omega _2}{2}$  

- amplitude modulated in time: $A(t) = 2A_1 \cos (\frac{(\omega _1 - \omega _2)t}{2})$  

![alt text](https://scuolaelettrica.it/elettronica/telecomunicazioni/ampiezza.gif)

**Battimenti**

this case produces the phenomenon of battimenti  
bc A is not constant, but modulated in time  
so in some moments sound is louder  

so signal is made of:  

- SEGNALE PORTANTE with $\omega _p = \frac{\omega _1 + \omega _2}{2}$  
the continuous one, which sends the most signal  
obv is thicker in the drawing, so has higher frequency  

- SEGNALE MODULANTE with  $\omega _m = \frac{\omega _1 - \omega _2}{2}$  
the rarer one, used only to adjust the more common one  

**Periods**

since there are 2 signals, there are 2 periods:  

- periodo della portante: $T_p = \frac{2 \pi}{\omega_p}$  

- period della modulante: $T_m = \frac{2 \pi}{\omega_m}$

**Trajectory**

note that:  
$-A_x \le x \le A_x$  
$-A_y \le y \le A_y$  

so trajectory is contained in a rectangle  

$\begin{cases}
x(t) = A_x \cos (\omega t) \\
y(t) = A_y \cos (\omega t + \varphi)
\end{cases}$

**Figura di Lissajous**

curves formed by the parametric equation (system above)  
describes the trajectory of a point undergoing two harmonic oscillations in perpendicular directions   
(with possibly different amplitudes and a phase difference φ)  

comes from plotting a graph with:  
$\frac{\omega _y}{\omega _x}$ on y axis  
$\varphi$ on x axis  

#### Case: same pulsation

**Polarization**

an EM wave has both electric and magnetic fields, E and B  
both oscillate perpendicular to each other and to the direction of propagation  

polarization describes the direction and shape of the oscillation of the electric field in an EM wave as it travels through space

**Polarization in Lissajous figures**

only makes sense when both components of the wave have the same pulsation  
bc this models a single electromagnetic wave moving in one direction, with electric field oscillating in two perpendicular directions  
if the frequencies were different, you would not get a closed figure  

**What is polarization fo?***

📡 Communication and antennas

Polarization helps separate signals — two antennas polarized orthogonally can transmit on the same frequency without interference.

Satellites and 5G systems often use polarization multiplexing to double capacity.

😎 Polarized sunglasses

They block light polarized in certain directions (like glare from water or roads).

Glare is often horizontally polarized, so sunglasses filter that direction.

🛰️ Remote sensing and radar

Polarized radar waves can distinguish surface textures and material types (e.g., soil vs water).

🔬 Optics and microscopy

Polarization microscopes reveal anisotropic materials like crystals, biological tissues.

Polarizers and waveplates control light in optical setups (e.g. laser cavities, interferometers).

💡 LCD screens

Liquid crystal displays use polarization to control which pixels let light through.

🧪 Material analysis

Polarization changes when passing through materials — useful for stress testing (photoelasticity), or detecting optical activity (e.g. sugar concentration in solution).

**Cases: polarization**

- $\varphi = 0$ or $\varphi = \pi$  
RECTILINEAR POLARIZATION    
$y = \pm \frac{A_y}{A_x}x$  

- $\varphi = \pm \frac{\pi}{2}$  
ELLYPTICAL POLARIZATION  
if $A_x=A_y$, called circular polarization    
$(\frac{x}{A_x})^2+(\frac{y}{A_y})^2 = 1$  

so in rectilinear, trajectory is a segment  
in ellyptical, it's an ellypse with axes parallel to carthesian axes  
the two cases of the ellypse change the direction of the path, clockwise (+) or antclockwise (-)  


**General case, $\varphi$ is whatever**

$(\frac{x}{A_x})^2+(\frac{y}{A_y})^2 -2\frac{x}{A_x}\frac{y}{A_y}\cos \varphi = \sin ^2 \varphi$  

it's an ellypse with axes not parallel to the cartyesian axes  


## Damped obscillations (smorzate)

**Equation**

2 ways to write it:  
$m\ddot{x} = -\beta \dot{x} - kx$  
$\ddot{x} + \lambda \dot{x} + \omega^2 x=0$  

$\omega ^2 = \frac{k}{m}$  
$\lambda = \frac{\beta}{m}$     

**How to solve**

solution $x(t)$ is usual one of a differential eq of 2nd degree  
solve it by computing the characteristic poly and finding roots, then depending on sign of $\Delta$ write appropriate solutions  

**Cases depending on Delta**

$\Delta = \beta ^2 - 4mk$  

$\Delta >0$: over damping (oscillatore sovrasmorzato)   
$\Delta = 0$: critical damping   
$\Delta <0$: under damping 

**What happens**

in all cases:  
when $t \rightarrow \infty,  \quad x \rightarrow 0$  
so damping succeeds in stopping the obscillator  

overdamping and critical damping: the obscillator is stopped BEFORE it can complete even 1 obscillation  
bc roots are real, no complex part, so there's no trig func in solution  

critical damping: given a k, this case has the smaller $\beta$ for which a complete obscillation will not occur

**Properties obscillator**

- damping time (tempo di smorzamento):  
$\tau = \frac{1}{\lambda} = \frac{m}{\beta}$  

- q factor (chiu factor) (fattore di qualità/merito of the obscillator):  
$Q= \omega \tau$  
    - if $Q \gg 1$ means $\tau \gg \frac{1}{\omega}, \quad T' \sim T$  
    means damping is so small can barely see it  
    - if $Q = \infty$ it's a normal harmonic obscillator, without damping  


**Period**

period is the time to make a full obscillation, so to invert direction twice  

$T' = \frac{T}{\sqrt{1-(\frac{1}{2 Q})^2}}$  
$T$: period it would have without damping  
$T'$: period it has with damping

$T'>T$, bc damping slows the obscillator  

period $T'$ DON'T DEPEND ON TIME t  
bc each obscillation is smaller than the previous bc of damping, but the obscillator moves slower too, so period is constant  



**Energy**

average energy during a period:  

$<\frac{dE}{dt}>_{T'} = \frac{1}{T'} [E(t+T')-E(t)]$  


$<\frac{dE}{dt}>_{T'} = -\frac{2}{\tau} <K>_{T'}$  
K is kinetic energy  

$|\frac{\Delta E}{<E(t)>_T} | = \frac{2 \pi}{Q}$  

## Forced obscillator

$m\ddot{x} + \beta\dot{x} + kx = F_0 \sin (\omega _0 t)$  

what all terms are:  
- $m\ddot{x} + kx$: general harmonic  
- $\beta\dot{x}$: damping  
- $F_0 \sin (\omega _0 t)$: forces (makes the equation non-homogeneous) (can be any $F(t)$)  

### What happens

**What happens**

2 phases:  

1. fase transiente, transient phase  
before the damping terms in the solution decays (saw in damped part that tend to 0 when t tends to infinite)    
the damping part is called transient bc it's tempotaty, won't be felt anymore after a while  

2. fase a regime, steady state  
damping has decaded, obscillates with frequency of forzante

**Behaviour**

$\omega, \lambda$ are of the obscillator  
$\omega_0$ is of the forzante  

obscillator progressively adapts its frequency to frequency $\omega _0$ of the portante  
so at one point will get $\omega = \omega _0$ and will stay like this forever  

formally: solution of the moto del transient has decaying exponential ($e^{-\alpha t}$), so damping part disappears progressively  

**Solution**

it's non-homo differential eq, so:  
$y = y_{homo}+y_{part}$  

- $y_{homo}$: moto del transiente    
compute like before for damped obscillator   

- $y_{part}$: moto a regime  
$x(t)= A \sin (\omega t + \varphi)$  

$\begin{cases}
A = \frac{F_0}{m\sqrt{(\omega ^2 - \omega _0 ^2 )^2+ (\lambda \omega _0)^2}} \\
\tan \varphi = \frac{\lambda \omega _0}{\omega ^2 - \omega _0^2}
\end{cases}$   

### Forzante

**What it is**

external force that gives energy to the obscillator  
this force can't depend only on position, else in a back and forth obscillation would do same work in both directions, but with opposite sign, so in total no work  
so FORCE MUST DEPEND ON TIME    

usually pick an 'forzante armonica': $F(t)=F_0 \sin (\omega_0 t)$  

**How it works**

takes energy from an external source, so it's non-conservative  
since obscillator will always be pushed by it, don't exist an equilibrium position (a position where, if put there with no speed, remains there)  

note that the power of the forza of the transiente alternates moments were it's dissipative and moments where it's not, it gives energy   
however, overall in a period gives more energy than takes  

except in resonance, the forzante always gives energy, never takes, means its power is never dissipative

### Resonance

**What is it**  

condition:  
specific condition within the steady-state regime  
occurs when the driving frequency equals the natural frequency of the system ($\sqrt{\frac{k}{m}} = \omega _0$)  

properties:  
the amplitude of the steady-state oscillation becomes maximal  
physically, it's the most efficient transfer of energy from the driving force to the system  
used by radio, when resonates it receives the maximum amplitude and sygnal  

**Resonance vs steady state**  

$\omega_0$: of forzante  
$\omega$: of obscillator (current one, so changes in time)  
$\sqrt{\frac{k}{m}}$: natural frequency of the obscillator  

steady state: $\omega = \omega _0$  
always happens when the damping dies out  
obscillator takes frequency of forzante, which is different from its starting one    

resonance: $\sqrt{\frac{k}{m}}= \omega _0$  
FROM THE START, frequencies of obscillator and forzante have been the same  
bc the NATURAL FREQUENCY OF THE OBSCILLATOR happens to be equal to the one of the forzante  

**Resonance**

Cacoma bridge: un ponte sospeso  
fell bc started obscillating after taking energy from the wind in an effect of resonance  

### Properties

**Amplitude**

start (transiente): $A(0)= \frac{F_0}{k}$  
a regime: $A(\omega_0 = \omega) = \frac{F_0Q}{k}$  

if it's a good obscillator, so if $Q \gg 1$:  
$A_{max} = A(\omega _0 = \omega) \gg A(0)$ (this is resonance)    

**Tips for integration**

$\int _t^{t+T} \cos \omega t \; dt = 0$  
bc sin, cos are periodic, so in a period the area is null  

$\int _t^{t+T} \sin (\omega t)\cos(\omega t) dt = 0$  

$\int _t^{t+T} \sin (\omega t)^2 dt = \frac{T}{2}$  
to do this use integration method like $\cos^2 x = \frac{1- \cos(2x)}{2}$  



**Cool formulas**

$\sin \varphi = \frac{\tan \varphi}{\sqrt{1+ \tan^2 \varphi}} = \frac{1}{\sec \varphi}$ 

**Energy**

it is not conserved at each moment  
but it is conserved throught each ourse of a period (at start and end of period it's the same, in the points in the middle no)  

average energy dissipated in a period by the damping:  
$<\frac{dE_{dis}}{dt}>_{T} = - \frac{\beta \omega _0^2 A^2}{2}$  

average energy given in a period by the forzante:  
$<\frac{dE_{abs}}{dt}>_{T} = - \frac{F_0 \omega _0 A}{2} \sin \varphi$  

**Decay time**

it's in transient phase  
find solution of homogeneous diff eq  

compare it to this solution  
$x(t)=2A \sin (\omega t) e^{-\frac{t}{\tau}}$  

see what $\tau$ corresponds to  

#### Solving differential equation

**What happens**

it has non-homo differential eq  
the solution is made adding homogeneous and particular sol  

homo sol: moto del transient
part sol: moto a regime  

**Solving eq for damped and forced**

solving homo:  
solve with method of characteristic equation  
assume $\Delta <0$, so get obscillatory solution  
else, if $\Delta \ge 0$, get overdampened obscillator, probably not what i want  

solving particular:  
make a guess (ansatz)

**Situation, non-homo equation**

my diff eq is:  
$\ddot{x}+ a\dot{x}+bx = c \sin (\omega t)$  
the right side is a func of t  

my guess is:  
$x(t) = A \cos (\omega t + \varphi)$  

where:  
- $\omega$ is frequency of the forzante, since at regime they obscillate together  
- $A, \varphi$ are my incognite  

**Strategy**

create a similar prob, swapping $x(t)$ with $z(t)$, a complex function ($z \in C$)   
find solution $z(t)$, then $x(t)$ will be its real part (since to make sense physically, need to have real numbers, what do complex answers mean?)  

my diff eq is:  
$\ddot{z}+ a\dot{z}+bz = ce^{i\omega t}$  

my guess:  
$z = z_0 e^{i \omega t} = Ae^{i(\omega t + \varphi)}$  
$z_0 = Ae^{i\varphi}$ is called complex amplitude  

why:  
$e^{iωt}=\cos(ωt)+i\sin(ωt)$  
the real part of my guess $z$ is $x(t) = A \cos (\omega t + \varphi)$, like my guess  
but this is easier to solve, bc when derivating and plugging in my guess, all the exponentials go away  

**Solution**

$z(t)$ is complex, so has real and imaginary part  

$z = z_{r} + iz_{i}$  

to find params in my original guess for $x(t)$:  

$\begin{cases}
A = \sqrt{z_r^2+z_i^2} \\
\tan \varphi = \frac{z_i}{z_r}
\end{cases}$

**What to do**

1. derivate my guess for $z(t)$  
2. plug it in the diff eq with z  
all $e^{i\omega t}$ will be cancelled since will be in all addens, only $z_0$ is left  
3. find formula for $z_0$  
4. use $z_0$ to compute $A, \varphi$ for my original guess for $x(t)$  

**Tip real, imaginary part**

if the i appears at denominator, need to rationalize to bring it at numerator to understand which parts are real and imaginary  

eg: $\frac{1}{a+ib} = \frac{a-ib}{a^2-(ib)^2} = \frac{a}{a^2 + b^2} - i \frac{b}{a^2+b^2}$
