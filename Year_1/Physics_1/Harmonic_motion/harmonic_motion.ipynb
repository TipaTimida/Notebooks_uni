## Harmonic motion

### Definition

**Equation**

$\ddot{x} = - \omega ^2 x$  

where x is x(t)  

**Solution**

generally can be written in 2 forms:  

$x(t) = A \sin (\omega t) + B \cos (\omega t) + x_{particular}$  
used for spring  


$\theta (t) = \theta _0 - \cos(\omega t + \varphi_0)$   
used for pendulum

**Degrees of freedom**

these easy examples can only obscillate IN A PLANE, following a precise trajectory  
so just need to know in which point of this trajectory it is at each moment   

so motion can be described by just one motion law, $\theta(t)$ or $x(t)$  
so has DOF=1  

### Small obscillation / angle approximation

if told this, when finding diff eq can consider $ \theta \sim 0$ and approximate with Taylor

**When can be used: simple pendulum example**

approximate with Taylor, for $ \theta \sim 0$:  

$\sin \theta = \theta (1-\frac{\theta ^2}{16}+o(\theta^2))$  

small angle approximation is when:  
$\frac{\theta ^2}{16} \ll 1$  
with $\theta$ expressed in radiants  

so corresponds more or less to: $\theta < 15 °$

**Effect on energy**

$U(x) \longrightarrow U(x_0) + \frac{1}{2}\frac{d^2U}{dx^2}(x-x_0)^2$  

potential an any point can be approximated as sum of:  
- potential at equilibrium point $U(x_0)$ 
- an addend that scales on the square of the displacement from the equilibrium point  

so generally, since need to choose level 0, pick it so $U(x_0)=0$  

if have this form, i can find formula for period independent from amplitude  

examples:  
spring, $U(x) = \frac{1}{2}k(x-x_0)^2$  
simple pendulum: $U=mgh \longrightarrow U(\theta) = mgl(1-\frac{\theta^2}{2}) = mgl- \frac{1}{2}mgl \theta^2 $

### Pulsation vs angular speed  

- pulsation: $\omega$  
it's a constant value, a charcateristic of the motion in its entirety   
don't depend on amplitude of motion  

- angular speed: $\dot{\theta}$  
it's a function of a time $\dot{\theta}(t)$, changes in each point  
depends on amplitude of motion  

they appear the same in uniform circular motion (an obj moves in a circle centrated in the origin)  
but actually:  
$\dot{\theta}$ refers to the actual circular motion  
$\omega$ refers to the motion of its projection on the x axis (which is a rectilinear motion)  

the one used in formula of period is the pulsation  

## Find harmonic motion

### Find differential equation

**Methods**

must use some diff eq, for example:  

$\frac{dE}{dt}=0$  

$\frac{dL}{dt} = \tau$  or $I \ddot{\theta} = \tau$  

$\mu \ddot{x} = - F$  

the first is usually easier, for signs as well  

**When can use form with reduced mass?**

when:  
have 2 bodies  
the only force $F$ present is one that acts lungo la loro congiungente (gravitational attraction, spring or thread between them)  

so $F$ is the relative force  
it's fine if there are other forces like weight or N, but must be orthogonal to the plane of obscillation, so don't impact    

$\mu \ddot{x} = - F$  

#### Use eq with momentum of a force

the - in the momentum formula comes from the right hand rule, since it's a vectorial product  
before starting must decide in which direction pojnts the z axis, if entrante or uscente  
generally uscente, bc pick it so x,y,z respect the right hand rule  

intuitively, there is a - bc the weight makes the pendulum move so to decrease the angle, not increase it (since $\theta$ is the angle with the vertical, and the weight tends to bring the pendulum vertical, closing the angle)  

the two versions are the same  

eg. in simple pendulum:

$\vec{\tau} = -mgl \sin \theta$  
$\vec{L} = ml^2 \omega = ml^2 \dot{\theta}$  

$\frac{dL}{dt} = ml^2 \ddot{\theta} = I \ddot{\theta}$  
so can equalize whichever i want  
use the full one with derivating L only if finding I of the obj is hard  

#### Write energy in a point

**Energy in general point**

use $x, \dot{x}$ (NOT v)  

pendulum:  

$E = \frac{1}{2}I_p \dot{\theta}^2 - mgl\cos \theta$   
if U=0 at top of pendulum  

$E = \frac{1}{2}I_p \dot{\theta}^2  + mgl(1−cosθ)$  
if U=0 at bottom of pendulum  
useful in small obscillations, bc can do $1- \cos \theta = \frac{\theta ^2}{2}$  

intuitively, for small obscillations, put $U=0$ on the line where the small obscillation happen  

**Ex. pendulum**

pendulum made by mass m + rod M   

general energy:  
$U= mgl(1-\cos \theta) + Mg\frac{l}{2} (1-\cos \theta)$  
$T = \frac{1}{2}I \dot{\theta}^2$  
$E = U+T$  

**Tip complex system on inclined plane**

can pick a curvilinear axis  
eg for a in inclined plane shaped like a triangle, pick as x axis a curved axis, parallel to the surface  
decide which extremity is the positive one  

write energy like:  
$E =  \frac{1}{2}kx^2 + \frac{1}{2}m\dot{x}^2 + \frac{1}{2}I(\frac{\dot{x}}{R})^2 + mgx$ 

#### Derivating

**Derivating**

use chain rule  
if have more terms with the var, do derivative of product, like $\dot{\theta} \sin (\theta)$  

examples:  
$\frac{d}{dt} \frac{1}{2}m\dot{x}^2 = m \dot{x} \ddot{x}$  
$\frac{d}{dt} \frac{1}{2}k(l_0-x)^2 = -k(l_0-x)\dot{x}$  

can probably delete $\dot{x}$ in all terms  
try to get form:  
$\ddot{x} = -\omega^2 x + c$  



#### Oscillation around a specific eq position

**Method**

want to express my var as the displacement wrt that eq position, so becomes small obscillation  

1. find a diff eq that uses a generic angle $\theta$  
eg. $I \ddot{\theta} =  \tau$ or $\dot{E}=0$  

2. when get to $\ddot{\theta}=...$, cannot say $\sin \theta = \theta$, bc i don't have small angle yet  
so, do var change to get obscillation around a specific eq position:  
$\begin{cases}
\theta = \theta _{eq} +\epsilon \quad \\
\ddot{\theta} = \ddot{\epsilon}
\end{cases}$  

note that now, if small obscillation, can approximate $\epsilon \sim 0$  
so expand the sin and cos of a sum:  
$\begin{cases}
\cos (\theta _{eq} +\epsilon) \sim \cos \theta _{eq} - \sin (\theta _{eq}) \, \epsilon \\
\sin (\theta _{eq} +\epsilon) \sim \sin \theta _{eq} + \cos (\theta _{eq}) \, \epsilon
\end{cases}$  

**Tips var change**

sequence of useful vars is:  

1. coords of point, $x_1, x_2$  

2. displacement from equilibrium position: $\; \overline{x}_1 = x_1 -l_1, \; \overline{x}_2 = x_2 -l_2$  
for this msut first compute $l$, the equilibtrium position, for each particle  

3. stuff to decouple the differential equation  
eg: add or subtract the 2 diff equations till $\ddot{x}$ and $x$ are the same, maybe $\overline{x}_1 \pm \overline{x}_2$

### Solving the diff equation

**Tips unknown constants**

when have:  
$x_1(t) =  c_a \cos(\omega t + \varphi_1),  \quad x_2 = c_2 \cos(\omega t + \varphi_2)$  

IF CAN PICK ORIGIN OF TIME $t=0$:  
can delete one of the phase angles (put it as 0), and the phase difference is comprised in the other  

can't do if can't pick origin of time, so if tells me something happens at $t=0$, like the speed being null or x it being a specific value   
so can't do it if have starting conditions that tell me something on $t=0$, like $x(0), \dot{x}(0)$  

### Stuff i can find

**Period**  

$T = \frac{2 \pi}{\omega}$  

don't need to solve diff eq, just to write it to see $\omega$  

**Amplitude**

DIFFERENT FROM MAX COMPRESSION  
the second is l at EQUILIBRIUM, NOT AT REST  

amplitude: $A =l_{max}-l_{eq}$  
max compression: $\Delta l_{max} = l_{max} - l_0$  

$l_{eq}$: find particular sol of diff eq ($\ddot{x}=0$, find x)  
$l_{max}$: is max compression of spring, explained below  

**Max compression of spring, or max displacement, or max angle in pendulum**

use ENERGY CONSERVATION between two specific moments  
in the point of max compression, where motion is being inverted, put all kinetic energy (both translation and rotational) to 0  


**Amplitude**

$ampl = \sqrt{A^2 + B^2}$  
A,B are the incognite i found when solving the diff eq after applying the initial conditions  

or, if i'm told the extreme point, can find it by $ampl = 2(x_{extreme} - x_{center\_obsc})$  
the extreme point is that where the masses are instantaneously still  


**Center of obscillation / equilibrium length**

<u>particular solution of the diff eq in the stationary state</u>    

in basic case is $l_0$, if add more forces can change    
ex. if spring hangs vertically and there is gravity, center of obscillation is lower

**Minimum k to have harmonic motion**

find (maybe from previous question) the angular speed $\dot{\theta}$  
condition is that k needs to be big enough so that the pulsation $\omega = \sqrt{\frac{k}{m}}$ is equal to $\dot{\theta}$

## Simple pendulum

**Equation**

$\ddot{\theta} = -\frac{g}{l} \sin \theta$  

small-angle approximation:  
$\ddot{\theta} = -\frac{g}{l} \theta$  

**Solution**

$\theta (t) = \theta _0 - \cos(\omega t + \varphi_0)$  

where $\omega ^2 = \frac{g}{l}$  

### Period

2 different formulas, depends on how much i approximate the cos 

the first one approximate until order 2 and don't depend on amplitude  
the second one approimate until order 4, depends on amplitude  

$T = 2 \pi \sqrt{\frac{l}{g}}$  

$T(\theta _0) = T(0) (1 + \frac{\theta _0^2}{16}) = 2 \pi \sqrt{\frac{l}{g}}(1 + \frac{\theta _0^2}{16})$

**Find formula for period**

first, find $dt$ by using separation of variables on eq for speed, bc $\dot{\theta} = \frac{d\theta}{dt}$  

then integrate, period is double the time it takes to go from max angle on left to max angle on right  

$T= \int _0^T dt \; = \; 2 \int _{- \theta _0}^{\theta _0} \frac{d\theta}{\dot{\theta}} \; = \; 2 \int _{- \theta _0}^{\theta _0} \frac{d\theta}{\omega \sqrt{2(\cos \theta _0^2 - \theta ^2)}}$

for spring is done same way, just change formula for speed and bounds are [-A, A]  

$\dot{\theta}^2 = 2 \omega ^2 (\cos \theta - \cos \theta _0)$  
formula taken from energy conservation  

### Energy

it's conservative, bc the forces are:  
- gravity, conservative
- reaction of perno, don't do work bc null displacement  
- reaction of thread on the mass, don't do work bc displacement orthogonal to force  

use energy conservation to find speed in a point  
final formula is:  
$\dot{\theta}^2 = 2 \omega ^2 (\cos \theta - \cos \theta _0)$  

### Find equation  

use 2nd cardinal equation:  

$\vec{\tau} = -mgl \sin \theta$  
$\vec{L} = ml^2 \omega = ml^2 \dot{\theta}$  

$\tau = \frac{dL}{dt}$  

the - in the momentum formula comes from the right hand rule, since it's a vectorial product  
before starting must decide in which direction pojnts the z axis, if entrante or uscente  
generally uscente, bc pick it so x,y,z respect the right hand rule  

intuitively, there is a - bc the weight makes the pendulum move so to decrease the angle, not increase it (since $\theta$ is the angle with the vertical, and the weight tends to bring the pendulum vertical, closing the angle)

## Physical pendulum

$\ddot{\theta} = -\frac{dmg}{I} \sin \theta$  

**Description**

C: cm  
A: pivot, it's the point that stays still, everything rotates around it (the point where the obj is hung up on the wall)  
d: distance between A and C  

### Pulsation

simple pendulum: $\omega = \sqrt{\frac{g}{l}}$  
physical pendulum: $\omega = \sqrt{\frac{dmg}{I}}$   

momentum of inertia must be computed wrt pivot, vincolo from where it's hanging, NOT wrt cm 

### Find equation

**Find differential equation**

feels two forces:  
mg: weight, applied on cm  
N: normal force by the vincolo from where it's hanging, directed up, keep it hanging   

$\tau = - mg \; d \; \sin \theta$  
$\tau = I \ddot{\theta}$  

pick as pole the pivot (vincolo), so N, which is hard to model, has null arm
