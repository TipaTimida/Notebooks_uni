#### Tolemaic

**Why is circle perfect shape?**

but circles have none, all points are important the same

**Backward motion problem (moto retrogrado)**
  

problem of toleimac system  

Earth's angular speed faster than Mars    
so Mars appear to go back when it gets lapped  
apparent inversion of the direction of its motion  
easily seen with close planets (Mars, Venus)  

to solve, Tolomeus came up with epicicli (combining circular orbits)

### Gravitational force

**Cavendish experiment**  

measured G  in 18 century  
with torsion scale experiment (recently re-done with lasers)  

actually Cavendish was a geologist, wanted to measure earth mass, G was just an intermediate step for him  
but it's so important he's remembered for that  

**Inertial and gravitational mass**

defined as:  
- inertial: $F = ma$  
- gravitational: $F = -G \frac{Mm}{r^2}$  

a priori, no reason they should be equal  

equivalency principle (proved experimentally at end '800)  
inertial and gravitational are not the same, but are PROPORTIONAL  
this proportionality factor is stored in G  

proof:  
plug F from gravitational force in F of inertial formula  
if two masses have same acc a, their ratio inertial/grav mass must be the same  



**Problem of action at a distance**

how do Earth and Sun trasmit force between them?  

Newton thought with an instantaneous signal  

but later come up with concept of field  
communication is through gravitational waves, can't go faster than speed of light  
so signal takes 8 minutes to go from Sun to earth  


- modern physics  
interactions, even those 'by contact' or 'on surface', are actually eletromagnetic interactions  
bc bodies can't touch each other  

**Field**  

just a tool to visualize forces  

$\vec{G}(\vec{R}) = \frac{\vec{F}}{m} = - \frac{GM}{R^2}$  

pull caused by an obj of mass M  


field is created if there is a force  
no point asking of the field exists if m (the obj to be attracted) doesn't, bc force is an interaction and needs two bodies  
just a math method to express effects of a force in space  

field line: inviluppo delle tangenti delle forze  
lines from same field can't cross each other, else not deterministic (at the intersection, which force should be the obj follow? bc trajectory must be tangent to force)    

superimposition principle: can add force from each field
