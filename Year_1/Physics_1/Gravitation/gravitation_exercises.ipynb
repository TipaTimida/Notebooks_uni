## Effective potential

### What it is

**What it is**

it's the sum of all stuff scaling wrt r, like a potential energy  
actually includes the kinetic cause by the tangential velocity  

kinetic energy: scales with v (speed)  
potential energy: scales with r (distance)  
potential: it's NOT an energy, has different units  

$U_{eff} = \frac{L^2}{2\mu r^2} + U(r)$  

**How to find formula**

remember  $L= \mu r^2 \dot{\varphi}$


$E= \frac{\mu}{2}({\dot{r}^2 + r^2 \dot{\varphi}^2}) + U(r)=$  
$= \frac{\mu}{2} \dot{r}^2 + \frac{L^2}{2 \mu r^2} + U(r) $  





**Reduced mass**

if $M >> m$, $\mu = m$  
if there is much difference, the reduced mass is the mass of the smallest obj  
bc the bigger obj don't feel enough attraction to move, so can assume it has no speed, and thus no kinetic energy  

**Components**

- $ \frac{L^2}{2 \mu r^2}$: centrifugal potential  

represents angular kinetic energy, but depends on r, like a potential, not $\dot{r}$ like kinetic   
has nothing to do with centrifugal force  
prevents the bodies from getting too close: if r too small, gets big, break energy conservation  

- $U(r)$  
depends on what force between obj (gravitational, coulomb)

$E=0$: parabolic

**What happen with different energies?**

cases:  
- $E > 0$  
1 nodal point: open orbit  
motion allowed in whole region on right  

- $E = 0  $   
2 nodal points, but one is at $+ \infty$   
orbit closes at infinity (happens in comets with long periods)  

- $E^{*} < E < 0$  
limited orbit, 2 nodal points  

- $E^{*} $   
circular motion, can't change r  


$E^*$ is defined as the energy at circular motion  

**Special energies**

$ E = -G\frac{Mm}{2r}$
