## Gravity not on earth surface

**g in point inside earth**

pick point P  
$g_{in}$: gravity at point P (and in circle of same radius)  
$M_{in}$: mass inside the circle of such radius  

$\frac{g_{in}{g}} = \frac{M_{in}}{M} \frac{R^2}{(r-h)^2} =$  
$ \frac{\frac{4}{3}\pi (R-h)^3 \rho}{\frac{4}{3}\pi R^3 \rho} \frac{R^2}{(R-h)^2} = \frac{R-h}{R}$  

therefore:  
$g_{in} = g(1 - \frac{h}{R})$  

**g in point outside earth**

$ \frac{g_{out}{g}} = \frac{R^2}{(R+h)^2} = \frac{1}{(1 + \frac{h}{R})^2} = (1 + \frac{h}{R})^{-2} =$  
$ g_{out} = g(1-2\frac{h}{R})$

## Asteroid collision

in the picture, b is $b_{min}$

if an asteroid arrives:  
- with speed v  
- toward a planet of radius R  

there is a $b_min$ collision parameter, such that:  
- if $b<b_{min}$ the asteroid hits the planet
- else it follows an hyperbolic orbit without hitting  
gets diffused, meaning the grav field of the planet changes its trajectory  

at infinite distance the potential energy is null  
so by conserving energy can see that the asymptotic speed (way before and way after) are the same in modulo, but the direction changed  

**Collision section**

around b can draw a circle, of are $A=\pi b_{min}^2$  
all asteroids (with same speed v) passing through that circle will hit the planet, all those passing outside won't  

for normal collision between balls, ignoring the gravitational attraction, the collision section is simply the ball section  

how big the collision section is depends on how strong the gravitational field of the planet is: the stronger, the bigger  

can be called efficient section of the planet???

**Computing collision section**

- compute minimum b  

$b_{min} = R \sqrt{1 + \frac{2GM}{Rv_0^2}}$  

$v_0$ is the asymptotic speed, back when the planet moves in a straight line  

- compute collision surface  

$A=\pi b_{min}^2$

### Gravitational slingshot (fionda)

if planet moves towards the asteroid (not necessarily thorugh the congiungente, just generally getting closer) the asymptotic speeds of the asteroid before and after the interaction are not the same  

$v_1 = v_0 + 2v_M$  


$v_0$: symptotic speed before interaction  
$v_1$: symptotic speed after interaction  
$v_M$: speed of planet  

does the direction of speeds not matter???

![alt text](https://www.mathpages.com/home/kmath114/kmath114_files/image001.png)

The net effect is almost as if we "bounced" off the front of the planet  
From the planet's perspective we approached at the speed U+v, and therefore we will also recede at the speed U+v relative to the planet  
but the planet is still moving at (virtually) the speed U, so we will be moving at speed 2U+  
This is just like a very small billiard ball bouncing off a very large one  

![alt text](https://www.mathpages.com/home/kmath114/kmath114_files/image004.png)

**Why the speed is the same in the planet’s frame**

Conservation of mechanical energy in a central force field:  

The spacecraft feels only the planet’s gravity (ignoring atmosphere and thrusters during the close pass).

Gravitational force is conservative: the total energy (kinetic + gravitational potential) of the spacecraft relative to the planet is conserved.

Far away from the planet (both on entry and exit), the gravitational potential energy → 0.
So the kinetic energy must be the same.

→ That means the speed at infinity (approach speed and departure speed, relative to the planet) is identical.

In the planet’s frame, it’s like a small ball hitting a huge stationary ball elastically → comes in with speed v, goes out with speed v

In the Sun’s frame, it’s like the big ball was moving. The bounce adds/subtracts that motion, giving you up to ~2U change in speed.
