## Perfect gas

properties:  

- puntiform molecules  
- molecules move randomly  
- elastic collisions between molecules  
- no intermolecular forces (molecules don't interact)  
(so the potential energy is null, so the energy of the gas is just the kinetic one, which is the internal energy)

### Kinetic energy

$E = \frac{l}{2}k_B T$  

it's technically kinetic energy, but it's total energy in ideal gas bc assume that the particles have no potential energy  

l is the DOF (degrees of freedom)  

puntiform particle: DOF=3  
rigid body: DOF=6

note that at absolute zero, K becomes 0  

what my notes say:  

1 molecule: $E = \frac{1}{2}m v^2_{avg} = \frac{l}{2}k_BT$  
while gas: $E = \frac{l}{2}nRT = \frac{l}{2}Nk_BT$  

relationships:  
$R = k_B N_A$  
$N = n N_A$  (number of molecules)  
only here, m is mass os one molecule, so $M = m N_A$  

**Speeds**

$v \propto \sqrt{T}$  

RMS: $v_{RMS} = \sqrt{\frac{3kT}{m}} = \sqrt{\frac{3RT}{M}}$  
average: $v_{avg} = \sqrt{\frac{8kT}{\pi m}}$  
most probable: $v_p = \sqrt{\frac{2kT}{m}}$    

pressure: $p = \frac{1}{2} \rho v^2_{RMS}$  
$\rho$: density
