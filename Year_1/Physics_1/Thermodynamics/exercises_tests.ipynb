**Example: 2 gases**

2 gases separated by setto conduttore, so have same T   

1--- type of transformation  
both do unspecified transformation  
usually assume at the end there is equilibrium, so if the gases are exactly the same (same n, same T)   
will also have same p and same V (bc there's no reason why one should be different, the setup is symmetric)    
so each V will be half the total one, $V_1'=V_2' = \frac{V}{2}$, so a known value    
 
2--- write entropy  
$\Delta S =  [nc_v \ln(\frac{T_f}{T_0}) + nR \ln(\frac{V_1'}{V_1}) ] + [nc_v \ln(\frac{T_f}{T_0})  + nR \ln(\frac{V_2'}{V_2}) ]= 2 nc_v \ln(\frac{T_f}{T_0}) + nR \ln(\frac{V}{4V_1V_2})$

3--- find final temperature  
here the only unkwown is $T_f$, compute that  

4--- compute work  
so now know final temperature and volume, can compute $\Delta U$ for such a transformation  

**Example: gas + obj**

1--- type of transformation  
gas does isobaric transformation and solid obj inside gives heat (isochoric) 

2--- compute entropy change  
$\Delta S = nc_p \ln (\frac{T'}{T_0}) + mc \ln (\frac{T'}{T_0}) = 0$  

3--- find final temperature  
find $T'$ (equalize the argument of the log to solve):  
$\ln((\frac{T'}{T_0})^{2nc_p}) = - \ln ((\frac{T'}{T_1})^{mc}) = \ln ((\frac{T_1}{T'})^{mc})$  
$T' = (T_0^{2nc_p} T_1^{mc})^\frac{1}{2nc_p+mc}$  

4--- compute work  
compute work (sum of work of all things, according to transformation they do)  
since it's adiabatic, $W = \Delta U$, so compute U instead, it's easier bc can input T which i have, instead of V that i need to compute  
here gas do isobaric transformation, solid obj isocoric:    
$W = \Delta U = nc_p(T'-T_0)-mc(T'-T_1)$  

**Example: put machine beween source (not infinite) and gas**

machine operates as long as source and gas have different temperatures, stop when T becomes the same  
each time, absorbes heat from source, lo cede al gas  

1--- type of transformation  
gas does isobaric transformation, and the source is a solid heating (so isochoric)  

2--- write entropy  
$\Delta S = nc_p \ln(\frac{T'}{T_{gas}}) + C \ln (\frac{T'}{T_{source}})=0$  

3--- find final condition   
$ \ln[(\frac{T'}{T_{gas}})^{nc_p}] + \ln [(\frac{T'}{T_{source}})^C]=0$  
$\ln (\frac{T'^{C+nc_p}}{T_{gas}^{nc_p}T_{source}^C}) = 0$

4--- compute work:  
$W = Q_{ass}-|Q_{ced}| = nc_p (T'-T_{gas})-|C(T'-T_{source})|$  

**Other method**

max work of a carnot engine: when efficiency is that of a reversible machine   
$\eta _{rev} = 1-\frac{T_C}{T_H}$  

write work:  
$dW=\eta \; dQ_{abs}$  
$W_{max} = \int _{T_H}^{T'} (1- \frac{T_c}{T}) \; dQ$  
where $dQ = n c_v dT$    

split the sum, the first integrate to give $T$; the second to $\ln(T)$  

$T_H, T_C$: temperatures of the two sources that make the machine work  
$T'$: final temperatures of the two sources   
sources must be finite, they're not infinite reservoirs, else there is no max work
