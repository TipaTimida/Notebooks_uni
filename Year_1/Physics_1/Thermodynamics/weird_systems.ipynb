### Differentiating

**Notation**

if something (U,W,Q) is a function of a single var, write $d$ before it  
if it's function of two vars, write $\partial$ before it  

eg:  
$U= T \rightarrow dU = dT$  
$U=VT \rightarrow \partial U =V dT$
