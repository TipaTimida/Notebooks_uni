## Temperature

### Absolute zero

essentially:  

temp when a system can only receive energy, not give it  
it's not true that at absolute zero all molecules are still

**Temperature and kinetic energy**

$<E_{cin}> \sim T$  

average kinetic energy of a molecule is proportional to temperature ONLY FOR IDEAL GASES  
so only far from liqufying temperature  
so it's not true that at absolute zero all molecules are still, bc this relationship is not valid  

**Absolute zero: quantum explanation**

assume energy is quantizized: each molecule can be in  
$\epsilon _0$: fundamental state  
$\epsilon _1, \epsilon _2...$: excited states  

system is in fundamental state if all its N molecules are in fundamental state  
thus its energy is $E_0 = N \epsilon _0$  

when $T>0$ (kelvin), there is a probability for each molecule to be in an excited state  
in the limit $T \rightarrow 0$, this probability is null, all molecules must be in fundamental state    

so the absolute zero, $T=0$, is the temp at which each systems is in its fundamental state

### Triple point

**Triple point**

triple point of a substance is a point in PV diagram where coexist solid, liquid and gas state  

eg. water:  
at $T=0$ coexist ice and water  
if outside pressure is really low, water can evaporate, thus have gas  
bc something eaporates when the pressure of its molecules is higher than the atmospheric pressure

**Interpretation on microscopic level**

potential energy depends on distance between molecules  
at fundamental level, with energy $\epsilon _0$, must have distance $r_0$  
when temperature increases, which means energy increases, molecules have a range of distances they can be found in  
average distance is the middle point of the segment the horizontal line is cut in by the function (eg. middle point of AB is average distance when have that energy)  

at first, with low temperature, graph is essentially symmetric, so middle point don't change  
increase the amplitude of obscillation of molecules, but not the average distance, since middle point is still $r_0$, so no thermic dilation  

then graph starts being non-symmetrical, so middle point of segment moves towards the right  
means average distance between molecules increase, so thermic dilation  

this is why thermic dilation can only be absorbed from relatively high temperatures, since before the graph is symmetric  
the graph being symemtric around $r_0$ is the reason why we can use approximation for small angle obscillation  

## Heat

### Specific heat

**Formula**

general definition is by reversing that of heat exchange:  
$c(T) = \frac{1}{n} (\frac{dQ}{dT})$  

**Specific heat vs capacity**

$c$: specific heat  
characteristic of SUBSTANCE  

$C = mc$: thermic capacity  
characteristic of BODY  
it's a grandezza estensiva, means depends on quantity of material (mass)  

**Notes on specific heat**

its scale has an arbitrary definition  

depends on the temperature, but often it's approximated constant even with a change in temperature  
at low temperatures, $c(T)$ is a cubic function  

similar to formula for gas, but there is use capacity

### Heating

$Q = m \int _{T_1}^{T_2} C(T) dT$    
$Q = mc \Delta T$ 

**Equilibrium temperature**

$T_e = \frac{C_1T_1 + C_2 T_2}{C_1 + C_2}$  

condition: $\Sigma Q = 0$  

**Definitions**

heat source: can give infinite heat, without changing its own temperature  

triple point: combination of macroscopic variables (temperature and pression) where can coexist 3 states at the same time

### Phase transition

$ Q = m \lambda$

when reach necessary temperature and keep giving heat:  
- TEMPERATURE DON'T CHANGE until the phase change is over

#### Irraggiamento

**Radiation**

all bodies emit radiation  
human body too, nearly 120W a day

is used by sun to send light and heat in space  

**Black body**

black body: a body completely absorbent, don't reflect any radiation that hits it  

black is color of total absorption of radiation  
however actually, to paint a wall for maximum absorbance, better gray paint, since black paint is lucid and reflects a bit 

**Black body with small hole**

hole is small: all radiation remains trapped inside, but a little can get out  
radiation coming out follows Stefan-Boltzmann law  

$P = \sigma T^4$  

$P$: power emitted  
$\sigma$: Stefan-Boltzmann constant  
$T$: temperature of black body  

there are other bodies, not black, that follow law similar to Stefan-Boltzmann  

this formula is used to determine type of star  
bc $T \sim T^4$ means luminosity $\sim T^4$
