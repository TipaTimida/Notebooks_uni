**Formulas**

perfect gas: $PV = nRT$  
1st principle: $\delta Q = dU + \delta W$  

**When can use formulas?**

1st principle: always  
$dW = p \: dV$: only for reversible (bc p must be defined)  
$pV^\gamma = const $: only for reversible 

formulas for $W,Q,\Delta U$  
isochoric and isobaric: always ok, reversible or not  
isothermal: $\Delta U =0$ and $Q = W$ is always ok, but the specific formulas for Q and W only if reversible  
adiabatic: $Q = 0$ and $ \Delta U + W = 0$ is always ok, but $pV^\gamma$ only if reversible  

isothermal gives problem bc all other formulas come from the conservation of energy, thus 1st principle, which is always true  
but in isothermal $\Delta U = 0$, so must get formulas from $dW = pdV$, which is only true in reversible   

### Work

$dW = p \: dV$    
only for reversible  

the ready-made formulas are true also for non-reversible, bc come from the energgy conservation  
except from the one for adiabatic, bc come form $dE=pdV$  
so thatdon't work if non-reversible, need to find it another way  
eg: $W_{gas}=-W_{ext}=p_{ext}\Delta V$  

type | formula 
---|---
P const | $W_P = P \Delta V = nR\Delta T$  
T const (rev)| $W_T = nRT \ln (\frac{V_2}{V_1}) = P_1 V_1 \ln (\frac{V_2}{V_1}) = P_2 V_2 \ln (\frac{V_2}{V_1})$     
V const | $W_V = 0$ 
adiabatic | $W_A = -\Delta U$, since $Q=0$  


other formula for adiabatic:  
$W_A = \frac{P_1V_1-P_2V_2}{\gamma -1} = \frac{nRT_1 - nRT_2}{\gamma -1} = \frac{nRT_1}{\gamma -1}[1- (\frac{V_1}{V_2})^{\gamma -1}]$

**Sign**

$W>0$: done, $W<0$: subito  
if segment goes from left to right volume is increasing, so it's $W>0$   
W is area under PV graph (in a cycle, it's the area inside the boundary)   

**How to get formula**

$dW = F \: dx = P S \: dx = P \: dV$  

useful for external work  
used when have non-reversible isothermal, so can't use formula 

### Heat

$Q = \int nc \: dT = \int C \, dT$  

**Types**

type | formula 
---|---
P const | $Q_P = nc_P \Delta T$  
T const (rev)| $Q_T  = W_T = nRT \ln (\frac{V_2}{V_1})$   
V const | $Q_V = nc_V \Delta T = U_V$ 
adiabatic | $Q_A = 0$


**Sign**
 
$Q>0$ means it's absorbed  

adiabatic: $Q=0$  
isobaric: if expansion, $Q>0$  
isochoric: if heating (T and p increase), $Q>0$  
isothermal: if expansion, $Q>0$  



### Internal energy

$U = \int n c_v \, dT = \int \frac{l}{2} nR \: dT$  

**Types**

type | formula 
---|---
P const | $U_P = c_v n\Delta T$  
T const | $U_T  = 0$   
V const | $U_V = c_v n \Delta T = Q_V$ 

**Specific heat**

$C = (\frac{\delta Q}{dT})_{V \, const}$  
$c = \frac{1}{n}(\frac{\delta Q}{dT})_{V \, const}$  

$C_V = \frac{\delta Q_V}{dT} = \frac{dU}{dT} = \frac{l}{2}nR$  
$C_P = \frac{\delta Q_P}{dT} = \frac{dU}{dT} + \frac{dW}{dT} = C_V + nR = \frac{l+2}{2}nR$   

$l=3$: monoatomic (noble gas)  
$l=5$: biatomic  

remember than $c$ contains the R as well  

**Meyer relation**

$C_p = C_V + nR$  

### Adiabatic

$PV^\gamma = const$  
$TV^{\gamma -1} = const$  
$P^{1-\gamma}T^\gamma = const $

$\gamma = \frac{c_P}{c_V} = \frac{l+2}{l}$  

in PV plane has steeper slope than isothermal  

### Entropy

$dS = \int \frac{dQ}{T}$  

T is temperature of external source, not of gas

types | formula  
--- | ---
P const | $\Delta S _P = nc_P \ln (\frac{T_2}{T_1})$  
V const | $\Delta S_V = nc_V \ln (\frac{T_2}{T_1})$  
T const | $\Delta S_T = nR \ln (\frac{V_2}{V_1}) = \frac{Q}{T}$  
adiabatic (rev) | $\Delta S_A = 0$  
solid heating| $\Delta S = mc \ln (\frac{T_1}{T_0}) = C \ln (\frac{T_1}{T_0}) $
solid phase change |  $\Delta S = \frac{m \lambda}{T}$ 
general | $\Delta S = \Delta S_V + \Delta S_T$

for solid, $Q$ depends on what it happening: might be heat exchanged with a source, $m \lambda$ if phase change, $mc$ if heating an obj   
for general, can split each transformation in an isochoric + an isothermal  

in bracket there can be any rapport, except the thing that is const (really?, don't it depend on sitaution after the transformation)???   
just know if one thing is const, the others have some relationship  
to find formulas, just plug the appropriate thing in palce of $dQ$, eg: $\delta Q=nc_p dT$, and integrate  

for general transformation, the two are in sequence  
so isochoric is done with initial volume, isothrmal with final temperature  

in adiabatic, only $\Delta S_{uni}=0$ always  
but if not reversible, $\Delta S_{gas} \neq 0$  
usually consider it a general transformation, add entropy of isocoric and isothermal  

**Sign**

$Q _{gas} >0 \longrightarrow S_{gas} > 0$  
bc if molecules move faster, there is more chaos  

**Tips**

if system don't exchange heat with outside (adiabatic): entropy of system is constant  
entropy of single gases forming the system can change, but the net entropy, the sum of all, must be constant
