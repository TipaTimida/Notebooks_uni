## Heat flux

### Heat exchange / flux

**Fourier law (heat flux)**  

$\Phi = -k\frac{dT}{dx}$  

where:  
$\Phi = \frac{dQ}{dS \; dt}$  
usually can put $dS = S$  


$\Phi$: heat flux (heat for unit of time and surface)  
k: coefficient of internal conducibility (big = good conductor)   
S: surface of bodies in contact  
$\frac{dT}{dx}$: termic gradient  

**Find $T(x)$**

must find expression for $\frac{dQ}{dt}$ from something else, then it's a seprable diff eq, with $dx$ and $dT$  

##### Chaining stuff

**Useful form for chaining stuff**

$\frac{dQ}{dt} = -kS \frac{dT}{dx} = -\frac{kS}{l}(T_1-T_1)$  

$R = \frac{l}{kS}$  

**Multiple rods**

series: $R_{tot} = R_1 + R_2$  
in each point $\Phi _{in} = \Phi _{out}$  
use it to find $T_{eq}$ at point where two rods touch by equalizing the fluxes of the two rods  

parallel: $\frac{1}{R_{tot}} = \frac{1}{R_1} + \frac{1}{R_2}$

##### Harder stuff

## Fluids

**Stevino law**

$p(z)= p_0 + \rho _f gz$  

$p(z)$: pressure at depth z  
$\delta _f$: fluid density  
z: height of fluid above it  

**Archimede law**

$F_A = m_f g$  

force is equal to weight of fluid displaced  
$m_f = \delta _f V_f$  
$V_f$: volume of fluid moved, so the volume of part of object immersed  

**Does it float?**

$\Sigma F_z = F_A - mg = m_f g - m_o g = gV(\delta _f - \delta _o)$  

f: fluid  
o: object  

$\delta _f <\delta _o$: $F_z<0$, obj sink  
$\delta _f >\delta _o$: $F_z>0$, obj float (immerge enough V that reach $F=0$ and stays like that)  


## Pression

**Formula**

$p = \frac{\vec{F}\cdot \hat{n}}{|\vec{S}|} = \frac{F_{\perp}}{A}$  

$A$: area, scalar  
$\vec{S} = A \hat{n}$: surface, vectorial  


pression is a scalar product, so p is scalar  
only care about component of force perpendicular to surface
