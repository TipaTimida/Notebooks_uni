### Theory on system

#### Systems

**Chaotic system / chaotic behavior**

changing initial conditions even a little bit changes completely the system's evolution  
fluctuations on initial conditions make it impossible to predict  

**Thermodynamic vars**

they're a small set of vars that characterize a system  
they're tied by a relationship, called EQUAZIONE DI STATO: $f(a,b,c...)=0$  
this equation characterize the system, each system can have a different one  

types of vars:  
- mechanical/geometrical: total mass of system, volume, energy  
- not mechanical: pression. temperature, quantity of heat (Q)  

not-mechanical ones need a definizione operative before being used, so need:  
- a scale  
- a unit of measure  
- a technique/tool to measure it  

**State function**

involves state vars  
once its values is fixed, it stays constant  

$f(V,p,T)=0$  

**Types of state**

$A \rightarrow B$  
thermodynamic transformation: passage between 2 equilibrium states, A and B  

between them there are intermediate states, NOT of equilibrium, bc thermodynamical vars are not defined there  
eg, pressure can be different in every point, so can't define a value for the pressure of the gas  

**Types of system**

- open: exhange both  
- closed: can exchange energy, not matter  
- isolated: can't exchange neither energy nor matter  

**Equilibrium of system**

3 types of equilibrium, all must happen to have thermodynamic equilibrium:  

- mechanic: $\Sigma F_{int}=0 \quad \Sigma F_{ext} =0$  
- thermic: all parts have same temperature  
if system is in contact with extenral ambient, $T_{sys} = T_{amb}$  
- chemical: no chemical reactions or matter movements 

**Characteristic of thermodynamics equilibrium**

- independent on time: if reached eq, stays in eq, unless something external happens  
- independent on previous history of system  

#### How to study them

**Descriptions**

microscopic: describe motion of each atom  

macroscopic: pick a limited number ($\sim 5$) of macroscopic variables (not necessarily mechanical)  
called state variables, bc define the state of the system  

**Methods**

1. statistical mechanics  

2. thermodinamics  

**Statistical mechanics**

based on principles of mechanics  
searches relationships (even axiomatic is fine) between microscopic and macroscopic vars, based on laws of mechanics  
eg: relation temperature and kinetic energy of gas, heat and energy exchange between systems  
temperature and heat are not mechanic concepts, energy is  

this method is harder, need hard math

**Thermodynamics**

based on experimental laws, that relate macroscopic vars  
these experimental laws are called principles of thermodynamics  
since are experimental based, have general validity  
work on any gas, not only ideal ones  

#### States of matter

common: solid, liquid, gas/aeriform  

also: plasm (makes up stellar matter, between liquid and gas)  

solid: indeformable and incompressible

**Fluid**

fluids are liquids and gases  

perfect/ideal fluid: a fluid without internal friction  
so strati of fluid slides one wrt the other, like lamine  
water is nearly ideal, oil not at all

property of perfect fluids:  
at equilibrium, after picking a point, p (pression) is the same towards all directions  
however, p could change from a point to another  

gas: p the same at all points  
liquid: p changes depending on depth  

**Why oil to lubricate, not water?**

oil has friction between strati of itself, so can can pass energy from a strato to another  
if put oil between 2 gears (ingranaggi), create the 'miato', a pellicola between them

### Reversible transformation

**Definition**

definition:  
vars are defined at each moment  
so it's a succession of equilibrium states infinitesimally close  
so can make an inverse transformation that crosses exactly the same states, but in the inverse order  
thus ambient exchanges with gas the same heat and work, but with opposite sign  

if it's non-reversible, in the intermediate states vars like p,T are not defined for the gas, as they can be diffrerent in each point  
so cannot write $W =pdV$, since p is not defined  
need to find work in other ways, like $W=F \, l , \quad W_{gas} =-W_{ext}$  

**When is reversible?**

conditions:  

1. must be infinitely slow  
system needs time to reach equilibrium in all intermediate states  

2. cause of transformation must be infintiely small (like dF making something move)  

examples:  
use a weigh to lower a piston: weight must increase infinitesimally each time, like by adding a grain of sand, can't just put the whole weight from the start  
moving wall: same thing, dF moving the wall must be infintiesimaly small and wall must move slowly  
changing temperature: must use infinitely many sources with infinitesimal change in T, can't just use a single source of the temperature i want to reach  

**Quasistatic**

all intermediate states are nearly equilibrium states  
quasistatic $\rightarrow$ reversible  

**Friction**

if there is friction, it's not reversible  
bc friction is a cause, and cannot be made infintiely small  

more formally:  
if try to move a wall and have friction:  
when want to drag wall back and fort, in these cases must apply different force  
compress gas: must overcome both gas and friction  $F = pS  + f$  
expand gas: gas helps me, must overcome friction $F = ps-f$  
so forces have no same sign, friction breaks symemtry  

spontaneous processes are non-reversible bc:  
there is friction  
causes are not infinitesimal  

**Reversible and cyclic**  

a transformation can be cyclical for the gas, but not for the ambient  
meaning that complexively, the ambient has exchanged a net work or heat, thus it's in another state  

cyclic reversible: bring both system and ambient back to initial state  
called totally reversible  

cyclic non-reversible: bring system back to initial state, but ambient willl be in different state (bc will have exchanged a net work or heat)  

**Cyclical vs reversible**

reversible means can be reversed through EXACTLY the same states, but in opposite states  
so when drawing, must pass exacly on the previous line  

cyclical means i can go back to initial state, but following a different path  
so means with cyclical, have a non-null net work, and probably heat too  
so gas and ambient exchange something, state of ambient is different at the end  
so it's not cyclical for ambient, and cannot reverse it  

can be cyclical for system, but not for ambient

## Principles of thermodynamics  

**Principles**

- 0: transitivity of thermal equilibrium  

- 1: $\partial Q = dU + \partial W$  

- 2: Kelvin law  

- 3: Clausius inequality

### 1

**Proof**

consider 2 transformations:  

1: $A \rightarrow B$  
2: $B \rightarrow A$  

therefore:  
-2: $A \rightarrow B$   

consider transformation 1+2, make a cycle, so $\Delta U = 0$  
$\Delta W = \Delta Q$  
$W_1 + W_2 = Q_1 + Q_2$  
$Q_1 - W_1 = W_2 - Q_2 = -W_{-2} - (-Q_{-2}) = Q_{-2} - W_{-2}$  

so, to grom from A to B, work is $Q_i-W_i$, no matter the path  
can write $Q-W = f(A,B)$    

so, if pick a state that i consider as having value 0, i have a scale with which i can essentially express energy of each individual state  

**Internal energy**

associated to the degrees of freedom of the system  
essentially energy of all particles  

### 2

**2nd pinciple: formulations**

comes from the observation of the irreversibility of spontaneous processes  

formulations:  

- Clausius: can't create a machine that doing a cycle, as only result produces heat transfer from cold to hot body  

- Kelvin-Planck: a monothermic machine (exchange heat with only 1 source)  cannot produce work (so its work is $W \le 0$)  
leads to impossibility of perpetual motiond of 2nd species  
easier: cannot create a machine that turns all heat into work, some must be lost as heat  

to prove they're equivalent, prove that if i can build a machine that violates one form, can build one that violates the other

**Monothermic cycle**

monothermic has $W<0$, bc of Kelvin statement  
so is either $W=0$ or irreversible  

why if $W<0$, cannot be reversible?  
bc if it were, i could take the inverted transformation  
this would still be a monothermic transformation, but with $W>0$  
impossible, violates Kelvin principle    

### 3

**Clausius inequality**

are valid only for CYCLIC process:  


for 2 machines:  
$\frac{Q_1}{T_1} + \frac{Q_2}{T_2} \le 0$  
= 0 only if reversible  
 
$\int \frac{\delta Q}{T} \le 0$  

**What it says**

it traduces in formula the Clausius principle ( alternative formulation of the 2nd principle): can't construct a cyclic process whose only effect is to take heat from a reservoir and convert it entirely into work   

the inequality quantifies the irreversibility of real processes  

implies that entropy cannot decrease in an isolated system — it's related to the entropy increase principle

**Explanation**

$\int \frac{\delta Q}{T} = \Delta S_{rev}$   
this is the formula for entropy change of a reversible process  

the inequality says that, for each transformations, the biggest entropy change would be if the transformation were reversible:  
$\int _A^B \frac{\delta Q}{T} \le \Delta S_{rev, AB}$   

so, in a cycle:  
$\Delta S=0$ bc entropy is a state function  
if the cycle is reversible, the equality is true, so $\int \frac{\delta Q}{T} =0$ too  
if cycle is non-reversible, the entropy change must be less than 0  

however the entropy of the universe increses, bc entropy of the ambient increases  

**Proof**

start from:  
$\eta _{rev} > \eta _{irr}$  

substitute the expressions:  
$1-\frac{T_1}{T_2} > 1- \frac{Q_1}{Q_2}$  
$\frac{Q_1}{T_1} + \frac{Q_2}{T_2} \le 0$  
= 0 only if reversible  



**Properties of entropy**

additivity: can add entropy of each sub-system  
extensivity: depends on mass of system  

**Why entropy of isolated system can't dimish?**

bc isolated means that $\delta Q = 0$, so  $\Delta S \ge \oint \frac{\delta Q}{T} = 0$   
so $\Delta S \ge 0$, entropy can't diminish  

**Notes on computing entropy change**

want to use formula $\Delta S = \oint \frac{\delta Q}{T} $, which is only true for reversible transformations (if not reversible have $\le$, not $=$, which is useless)  
but entropy is a funzione di stato, so can use whatever transformation to connect initial and final state, not necessarily the one that really happened   

so:  
if know initial and final state, even if the transformation between them is irreversible, can imagine a revrersible one between them and compute entropy change of that  

formulas for entropy change are essentially always valid, bc if can connect two states with non-reversible isothermal transf, can also connect them with reversible isothermal transf  
except for adiabatic, if have non-reversible adiabatic can't swap it with reversible adiabatic, must use something else, eg isothermal in free expansion  

### Perpetual motion

**Definition of perpetual motion**  

NOT a machine that moves forever, has no sense or purpose, since does no work   

of 1st species: a machine that produces infinitely work, without exchanging heat from outside  

of 2nd species: a machine that can move in condition of bilancio energetico  

**Perpetual motion of 2nd species**

means a body absorbs some energy E, which gives back entirely  
so a body can travel in condition of bilancio energetico nullo  
eg: ship gets heat from sea, turn into work to move, all this energy is given back to the sea as heat again, so can repeat forever  

**Who prohibits it?**

1st principle prohibits perpetual motion of 1st specie  
bc in a cycle $\Delta U =0$, so $W=Q$, so to do work need to exchange heat  


2nd principle prohibits perpetual motion of 2nd specie

## Thermic machine

**Def thermic machine**

dispositivo in grado di operare conversioni tra calore e lavoro e viceversa  
succhia calore dall'ambiente usando delle sorgenti di calore  
funziona in maniera ciclica

**Produce work**

to produce work, need $\ge 2$ sources  
consider as sources only those that exchange a NET $Q \neq 0$  

else it's a monothermic machine, can't do work bc of 2nd principle  

**Leave frigo open**

if leave frigo open, room gets hotter    
bc subtract from inside the frigo $Q_1$  
but reverses outside $Q_2 = Q_1 +W$   
since $|Q_2| > |Q_1|$, room heats up  

**Carnot cycle**

conditions:  

- 2 sources  
consider as sources only those that exchange a net $Q \neq 0$  

- do all reversible transformations  


**Efficiency of carnot cycle**

this is valid for Carnot cycle  
again, consider as sources only those with net $Q \neq 0$  

always: $\eta = \frac{W}{Q_2} = 1- \frac{Q_1}{Q_2}$  
reversible: $\eta = 1- \frac{T_1}{T_2}$  

Carnot theorem: all carnot machines (reverisble) between 2 sources at some temperature have same efficiency
