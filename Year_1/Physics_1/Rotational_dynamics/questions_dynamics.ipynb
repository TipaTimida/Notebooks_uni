## Rotation

### Use energy conservation

**When can conserve**

- without friction: yes, always    

- with friction: only in pure rolling   
if no pure rolling, to conserve must subtract the work $W=f \, x $  

### Minimum energy/height/speed

**Method** 
  
compute mechanical energy at the critical point  
equalize it to initial energy  

eg: going up an inclined block  
critical point is the TOP of where it can reach, where it has NULL KINETIC energy  


**Giro della morte / death circle**

critical point is the top   
if just want to reach top: at top has N=0 and null kinetic energy  
if want to complete the loop: at top has N=0 and speed equal to the one needed by the centripetal force  

solution case 2:  
$-N-mg = -m\frac{v^2}{R}$  
put N=0  
find v, it's speed up there

when doing death circle, check if ball rolls or slides

### Equilibrium

to keep a complex system in equilibrium, DON'T CARE WHO INTERACTS WITH WHO, just write all forces towsrds x and y, the sum of both must be 0  
useful bc internal forces get deleted (like tension in a thread connecting two obj of the system)  
if want one of these internal forces, must write equilibrium condition of the obj that feels/does such force  

**Conditions of equilibrium**

$\Sigma F = 0$  
$\Sigma \tau = 0$  (use as pole the point around which, intuitively, should be rotating)

situation: object rotating around a pivot  
assume the first one is always satisfied, since reaction N of pivot keeping the body up can be however i want in direction and modulo  
but in second condition pick PIVOT AS POLE, so its reaction has no momentum  

**Point of equilibrium**

point $x_0$ where: $\frac{dV}{dx}|_{x_0}=0$   

means point where, if leave the object there WITH NULL SPEED (so only put potential energy), stays there 

**Stable vs unstable eq**

stable eq: relative min (when if i move it a bit, V increases)  
unstable eq: relative max  

generally, that with the lower cm is stable  

**Point of equilibrium, notes**

intuitively:  
an hanging obj is in equilibrium when the cm is directly under the pole  
so:  
mg has null momentum bc has null arm  
N has null momentum bc i picked the pivot as pole  
thus $\tau = 0$  

graphically:  
draw $V(x)$ or $V(\theta)$, the potential energy    
so has $Mgh, \frac{1}{2}kx^2$ etc.  
equilibrium point: where $\frac{dV}{dx} = 0$ (so when the tangent is horizontal in the graph, either relative min or max)  

### Angle to stop

**Formula**

angle to stop: ($\tau$ is the momento frenante)   
$W = \tau \theta = E_1-E_0$  

distance to stop:  
$W = F x =E_1-E_0$

if F or $\tau$ change, it's $W = \int F dx$  
if at end is still, $E_1=0$

**What is F**

must decide which point i'm considering  
if i'm considering the cm must only put the forces APPLIED on the cm  
so for eg. the friction, which is applied at the contact point, don't appear

### Rotation techniques

#### Find Coriolis force

generally done if an obj moves both circular and radial  

1. write using the formula or kinematics:  
$m\vec{a} = m(\ddot r - r \dot{\theta}^2) \hat{r} + (r \ddot{\theta} + 2 \dot{r} \dot{\theta})\hat{\theta}$  

2. write $ma = \Sigma F$, then project (split) them on radial and tangential axes  
pay attention to direction of forces  

$2 \dot{r} \dot{\theta}$ is the coriolis acc  

#### Graphical techniques

**Understand motion of cm**

check what forces there are, to understand motion of cm  
eg: if no horizontal forces, cm mvoes in a vertical line  

eg: an horizontal plane always gives a reaction pointing up, NOT pointing in the inclination of the axis  
so no matter if the axis is inclinated when it hits the plane, the cm will bounce back up  
reaction of plane is impulsive, so L is not conserved, so need second cardinal eq  
depending on how the collision is, might conserve energy  

**Find instantaneous center of rotation**

if know direction of speed of 2 points, draw the perpendiculars to each of them  
they meet in the instantaneous center of obscillation  

**Write system and derivate**

use geometry to write coords of stuff, eg:  

$\begin{cases} 
x = l \cos \theta \\
y = 0
\end{cases}$  

$\begin{cases}
\dot{x} = -l \sin \theta \, \dot{\theta} \\
\dot{y} = 0
\end{cases}$  

**How to handle**

1. write coords of each thing: $r_1(t), r_2(t)$    
use coords with funcs of time  
use versors $\hat{r}, \hat{\theta}$, will be useful when derivating  
eg: $x(t) = l\hat{r}+ a\hat{\theta}$  

2. derivate to find speed of each thing $\dot{r}_1(t)$  
use rules of derivation learned in kynematics  
eg: $\dot{x} (t) = l \dot{\theta} \hat{\theta} - a \dot{\theta} \hat{r} + \dot{a} \dot{\theta} \hat{\theta}$  
here 'a' is a func of time a(t)  
 
3. must know the speed each object should have, maybe bc it's doing rotationa round a point  
in which case, $\dot{x} = a \land \dot{\theta}$  

4. equalize points 2 and 3, find whatever it is i'm missing in the equation

## Translation

### Two objects move together

#### Find displacement

**Method**

if no external forces, $a_{cm}=0$    
so if at start cm is still, will keep being in same position   
so if know displacement of one object of the system, know displacement of the other too  

steps: 
1. find $x_{cm}$ at start      
maybe put coord $x=0$ in that point  

2. find $x_{cm}$ at moment of which i want the displacement:  
$x_{cm} = \frac{Mx_M + mx_m}{m+M}$  

3. equalize them  



**Tips**


if m is on top of M, its displacement can be something like:  
$x_m = x_r + x_M $  
where $x_r$ is the displacement of m relative to M, $x_m, x_M$ are wrt the ground/lab    
m wrt GND = x wrt M + M wrt GND  
attention to sign, probably $x_r$ is in opposite direction, will have negative sign  

#### Max height (obj inside moving conca)

situation:  
there is a moving conca  
there is a block inside  

obj inside reach max height when <u>its velocity wrt conca is null</u>   
so when <u>obj and conca are moving with same speed</u> wrt to lab    

1. find this speed such that momentum is conserved   
2. energy conservation to check potential energy, so height  

#### Two obj attached by spring, one moves towards another

force of spring in internal, so:  
not change momentum  
can model like 2-body problem  

**Max spring compression**

when max compression, obj at sides move with <u>same speed, which is speed of cm</u>  
find it by conservation of momentum    

**Period of obscillation**

formula for two-body system connected by internal force  
$\mu \ddot{r}_{rel} =   \vec{F}_{int} =  -k(r_{rel}-l)$   

longer way  

write eq and swap $x=x_2-x_1$   
when writing eq, stuff in parenthesis is: this coord - other coord  
$m_1\ddot{x_1} = -k(x_1-x_2 - l_0) =-k(-x -l_0)$  
$m_2\ddot{x_2} = -k(x_2-x_1 - l_0) = -k(x-l_0)$  

write, keeping the same order 2-1 used to define x:  
$\ddot{x_2} - \ddot{x_1}$   
will get diff eq with reduced mass next to k  


### Complex collisions

##### Conca hit by something external

energy conservation starts from moment right after collision, not from before, unless collision is elastic and conserve energy  

do conservations one step at a time:  
before collision -> after collision  
after collision -> max height 

#### Collision complex systems

**Speed right after collision**

<u>consider only obj hit by colliding obj</u>, bc RIGHT AFTER THE COLLISION THE FORCES (spring, friction) HAVE NOT PROPAGATED YET</u>   
so all obj attached to them but not hit by colliding obj haven't moved yet

eg:  

- collide against obj, attached with spring to another obj  

right after collision spring is not contracted yet, has no propagated force  
so extremity not hit is at rest, not moving  

- collide against obj with an on obj on top

obj on top will move bc of friction once bottom obj will start to move  
immediately after collision bottom obj hasn't moved yet, so no force has reached obj on top, so its speed is null  



**Compute energy lost in inelastic collision**

remember energy lost is relative kinetic energy, so $K_{rel} = \frac{1}{2} \mu v_r^2$  

$v_r$ is the relative speed at which the two obj are getting closer, simply add the vectors of each object's speed

### Energy/momentum problem

**Steps**

1. <u>choose coord system</u>    
ground level can be anywhere, but point y upwards, bc it will give the right signs  

2. <u>check if forces are conservative</u>  
  energy conservation  
$E_1 - E_1 = W_{non \: cons} = \Sigma F \cdot l$    


3. <u>check if forces are internal</u> or external  
external if at least one obj of the two interacting to create it is not an obj i'm interested in (part of my system)   
momentum conservation  
(may be in just one direction)  
pay <u>attention to signs of speeds</u>: put + on all addends, so the var itself will assume negative value if it's in other direction (don't put - in eq, will just get confused, store the - in the var)      

**When is stuff conserved**

- energy: always, just need to account for work of non conservative forces  
but easier if <u>non conservative forces make null work</u> (can exist or not)  

- momentum: if <u>net external force is null</u> (check each direction separately)   


**Internal forces**

weight and vincolar reaction with GND (mg, N) are external, so <u>usually on y axis no momentum conservation</u>    
but <u>on x axis usually yes</u>  (if there is no attrition with GND, bc it's external)

**Find some force or acceleration**

write all forces
