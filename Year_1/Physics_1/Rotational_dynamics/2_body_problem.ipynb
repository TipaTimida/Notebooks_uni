**What is it**  

have two bodies, $m_1, m_2$, connected by a force F  

can model it:  
- IN FRAME OF CM  
- like A SINGLE BODY, with reduced mass, MOVING IN AN EFFECTIVE POTENTIAL  

**What changes in frame of cm**

what NOT changes:  
distance between $m_1, m_2$ (relative distance)    
energy (if cm frame is inertial)

what changes:  
angular momentum   

energy not change from lab to cm frame only if cm frame is inertial (cm not accelerates)  

**Masses really different**

have two bodies: $M > m$  
can swap:  
$M \rightarrow$ SOMETHING INFINITELY BIG, so stays still  
$m \rightarrow \mu$  

## Model stuff

### Differential equation

$ \mu \ddot{r} = F(r)$  

where:  
$\ddot{r}$: the relative acceleration between the bodies  
$F(r)$: the force connecting them and making this a 2-body problem  

**When can use form with reduced mass?**

conditions:  
have 2 bodies  
the only force $F$ present is one that acts lungo la loro congiungente, so their relative force (gravitational attraction, spring or thread between them) 
it's fine if there are other forces like weight or N, but must be orthogonal to the plane of obscillation, so don't impact    

### Energy and momentum

can find easy formulas using Koenig th  
just need a different form for the realtive component around the pole  
pick as pole the cm and obtain:  

1. $K = \frac{1}{2}Mv_{cm}^2 + \frac{1}{2}\mu v_r^2$   

2. $L = M\vec{r}_{cm} \land \vec{v}_{cm} + \mu \vec{r} \land \vec{v}$  

M is mass of whole system, $M = m_1 +m_2$

**Example**

if have 2 masses, one still and one moves at $v_0$, and impact parameter b, relative part of L becomes: $L_{rel} = \mu v_0 b$  
bc model it like one mass $\mu$ moving at speed $v_0$, the other is still 

with two masses, relative part (so in cm pole) is:  
$L_{rel} = \mu v_0 b$  
$\mu$: reduced mass  
$b$: distance between masses  
$v_0$: speed orthogonal to the distance between masses (pay attention if speeds of 2 masses follow rotation in same or opposite direction, it changes sign: $v_0 = v_1 \sin \alpha \pm v_2 \sin \beta$)

**Angular momentum wrt the cm**
 
general method: $L=\Sigma mr_{P}v \sin \theta$  
pick a pole  P, then for each particle add its momentum   
$r_{P}$: connects the particle to the pole  
$v$: speed of each particle  
$\theta$: angle between r and v  

shortcut if pole is the cm: $L = \mu r_{rel} v \sin \theta$  
$r_{rel}$: relative (radial) distance between the masses  
$v$: RELATIVE tangential speed between particles (if one is still, it's simply the speed of the other)

could also use first method and pick cm as pole, but need to compute where it is  
essentially, must either compute the reduced mass or the coord of the cm  
the reduced mass is easier, usually pick that  

if there are two obj connected by a force, and an obj 3 hits obj 2 anelastically, imagine there are 2 objects, obj 1 and obj 2+3

### Energy in polar coords

$ K =  \frac{\mu}{2} \dot{r}^2 + \frac{L^2}{2 \mu r^2}$  

split kinetic energy in polar and tangential components  

useful to do conservation of both energy and angular momentum, like in gravitation or 2-body problem  

**Full formula for problems**

remember to add potential energy  
can regroup some stuff in the effective potential too  

$E = \frac{\mu}{2} \dot{r}^2 +V_{eff} $  

$V_{eff} = \frac{L^2}{2 \mu r^2} + V(r)$   

**How to find formula**

remember  $L= \mu r^2 \dot{\varphi}$


$E= \frac{\mu}{2}({\dot{r}^2 + r^2 \dot{\varphi}^2}) + V(r)=$  
$= \frac{\mu}{2} \dot{r}^2 + \frac{L^2}{2 \mu r^2} + V(r) $  





the formula with $\mu$ is ok only in 2-body problems, not in single body rotating around a point   

for those, if know L to be conserved, simply find tangential speed from L, then use that in normal expression for kinetic energy, so can easily put radial speed to 0 if needed  

### Find potential energy

if know F(r) of their interaction, rem that $F(r) = -\frac{dU(r)}{dr}$  

so $U(r)= \int F(r) dr$  

common ones:  
- $F(r) = \frac{k}{r^2}, \quad U(r)=\frac{k}{r} +c $  
- $F(r) = k(r -r_0) \quad U(r) = \frac{1}{2}k(r-r_0)^2 +c$  

**How to find the +c**

to find c, need to have a condition on the energy in a point  
to find the condition depends on situation:  

- if force acts everywhere: $U(\infty) = 0$  

generally i'm told it by the text itself  
if try to plug it in, it obv implicates $c=0$  

- if force starts from a point onwards: $U(r_0)=0$

potential field/energy must be continuous  
so at the border point where it starts acting, still $U(r_0)=0$, for continuity ($r_0$ is where force starts acting)  





**Example: When potential energy starts from a point onwards**

i know force, find energy:  

$\begin{cases}
F = 0 & U = 0 & r > r_0  \\
F = \frac{k}{r^2} & U = \frac{k}{r}-\frac{k}{r_0} & r <r_0 \\
\end{cases}$  

note that U must be continuous, but F not  

**Given custom energy between particles**

$U(r)=...$  

$\frac{dU}{dr}>0$: attractive  
$\frac{dU}{dr}<0$: repulsive  

bc $F=-\frac{dU}{dr}$, and a negative force means point inwards, towards the center of the system, so the obj is getting attracted  

**Draw energy**

how to draw:  
check what happens at 0 and $\infty$  
pay attention if there are - signs in the definiton of U when drawing  

how to use it:  
when drawing: particles tend to go where U is lower  
imagine putting a ball on the graph, see where it rolls down to  

draw conclusions:  
if it's lowest at 0 it's attractive, if it's lowest at infinity it's repulsive  

eg. grav energy is -1/r, so an iperbole in the 4th quadrant, at 0 goes to infinity, at infinity goes to 0, so particles rolls down towards r=0  


### Interaction starts at distance $r_0$  

**Speed of bodies at end of interaction**  

want speed of bodies, when, after interacting for a period of time, they are back at distance $r_0$  
can model interaction like an elastic collision, bc conserve energy and momentum  

**How long the bodies interact**  

so how long the distance stays $r < r_0$  

$T = \frac{2 \pi}{\omega}$  
$\omega$ is taken by solving the diff equation $\ddot{r}= - \omega ^2 r$, which comes from $\mu \ddot{r}=-F$   

## Common questions

#### Region of motion

**Definition**

K can be at most 0, not negative  
so region of motion are all r where: 
$V_{eff}(r) < E$  


$E$ is the total energy of the system computed in an easy point, since will probably be conserved  

**Graphically: region of motion**

graphically:  

1. plot $U_{eff}$ wrt $r$  
2. draw line horizontal of total energy of system, E  
3. <u>motion only in r where U(r)< E </u>  
bc potential energy must be less or equal total energy, since $K\ge 0$  


might get separate regions, can't jump from one to another, stays in starting one (so depend on initial conditions)  
to jump barrier need quantum tunneling  

**Limited vs closed motion**

limited: can only happen in a region of space  
closed: orbit is closed path, is always repeated  

limited does not imply closed  

#### Inversion points

**Definition**

the extremities of the region of motion  
r where:  
$V(r)_{eff} = E$   

bc radial speed $\dot{r}$ and radial energy are null  
for E use the value computed in an easy point, maybe from a previous question  
for the L in $V_{eff}$ too, use the L computed in an easy point  

in $\pm$, probably - gives the minimum distance, + the maximum one  
should both be positive values  

**What are they**

called nodal points or inversion points  
points where direction of RADIAL speed is inverted, so for a moment $K_{radial}=0$   
<u>min or max distance</u>, after reaching it the obj change direction of speed  
graphically, point of intersection between E (horizontal line) and U (function)  

- radial speed is instantly null  
- but tangential speed is not, so obj are NOT still at this points
- and whole CM of system might still be moving, obv

#### Circular motion

$r^*$ is the radius that allows it  

3 methods:  

1. $V(r)_{eff} \le E$  
  $\Delta (r^*) = 0$     
  find r such that $\Delta = 0$, so write eq $\Delta = 0$ and solve for r  

2. $\frac{d V_{eff}}{dr} |_{r = r^*} = 0$  

3. $F_{centripetal} = F_{internal}$    

**Example method 2**


$V_{eff} = \frac{L^2}{2 \mu r^2} + Ar^n$  
$\frac{dV}{dr} = -\frac{L^2}{\mu r^3} + nAr^{n-1} = 0$   
$r^{n+2} =\frac{L^2}{\mu nA}$  

where L is computed in a known point  

**Methods explanation**

method 1:  
the r that allow the inequality are the region of movement  
so, if there is only r that allows it, it's a circular motion (bc radius can't change)  
$\Delta=0$ is the condition for having only one solution  

method 2  
find radius when kinetic energy don't change (its derivative is null)   
means grav potential is conserved too, so r can't change
