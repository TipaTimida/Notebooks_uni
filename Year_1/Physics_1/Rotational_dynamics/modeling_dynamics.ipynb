## Modeling

### With linear system

generally, 2 cases:  

- complex system, made by many objects  
- just one object, but makes complex movements (translate and rotate)  

#### Complex system

**Steps: complex system**

1. identify all objects (blocks, springs, carrucolas)  

2. write a formula for each:  
block: $\Sigma F = m\ddot{x}$ (x in direction of motion)   
carrucola: $\Sigma \tau = I \alpha$  
spring: $\Sigma F = 0$ (bc m=0)

3. find relationships between the accelerations  
eg: in pure rolling, $a = R \alpha$  
eg: $\ddot{x}_1 = - \ddot{x}_2$

**Example: plane, spring, carrucola**


$\begin{cases}  
M\ddot{x} = Mg\sin \theta - T_2 & block \\
I \frac{\ddot{x}}{R} = R(T_2-T_1) & carrucola \\ 
0 = T_1 - F_{el} = T_1 + kx  & spring
\end{cases}$  

situation:   
slanted plane, cannot move   
block: slide on plane  
carrucola: blocked on cliff of plane   
spring: vertical on floor, next to vertical wall of plane 

consider x axis the curvilinear axis parallel to surfaces of plane (vertical surface as well)  


**Example: block on moving plane**

$\begin{cases}
 m a = m g + R  \\ 
 M A = R sin (\theta) \end{cases}$

plane M (acc A), obj m (acc a) on it  
A is totally horizontal, a is diagonal  

the RELATIVE perpendicular acceleration of the obj above must be null (not simply the perpendicular one)  

**Solution example**

find RELATIVE ACCELERATION of m, the obj:      
$m a_r = m(a-A) = m g + R - m A$  
 
its component perpendicular to plane must be null:   
$ma_{r \perp} = mg \cos \theta + R - mA \sin \theta = 0$  
$R = mg \cos \theta - mA \sin \theta$   

plug R in vector eq for the plane to get A:  
$A = \frac{m \sin \theta \cos \theta}{M + m \sin^2\theta}g$   

plug A in eq of realtive acc to get $a_r$:  
$a_r = \frac{(m + M)\sin \theta}{M + m\sin^2\theta}g$  

**2 balls, one on surface of cone, other dangling in the middle, conneced by rope**

want that:  
outside ball make a circular trajectory with constant $\omega$ (trajectory is an horizontal circle, on plane xy)    
ball inside is still   

$\begin{cases}  
T-mg = 0 \\  
N \cos \theta - T \sin \theta = -mR\omega^2 \\
N \sin \theta + T \cos \theta -mg = 0
\end{cases}$ 

1st eq: ball inside  
2nd eq: horizontal forces on ball outside  
3rd eq: vertical forces on ball outside  

**2 balls: find inversion points**  

r: radius of the circular trajectory  
thus, horizontal distance of ball outside from center of cone  
$\theta$: angle on top of cone  
$\frac{r}{\tan \theta}$: vertical distance of ball outside  
$\frac{r}{\sin \theta}$: length of rope of ball outside  


inversion points: where radial speed is null???  

$K = \frac{1}{2}m(r\omega)^2 = \frac{L^2}{2m r^2}$  
$U = - mg \frac{r}{\tan \theta}-mg(l- \frac{r}{\sin \theta})$  

#### One object

**Steps: just one object**

1. Write Newton's law, separating tangential and perpendicular forces  
if doing a circle, they represent tangential and radial, so put as radial acceleration the centripetal one  
these accelerations are those of the cm of the obj  
if sliding on plane, perpendicular acceleration is null  

2. describe rotation around the cm of the obj (bc before put acceleration of cm)  

3. if pure rolling, put condition on acceleration  

v in centripetal acceleration ususally found with energy conservation  
if pure rolling there MUST be static friction, of which i don't need to know the coefficient, so put -f  

**Recap**


$\begin{cases}
\Sigma F_{rad} = m\frac{v^2}{r} & \Sigma F_{perp} = 0 \\  
\Sigma F_{tan} = ma_{t} \\
\Sigma \tau = I \alpha \\ 
\alpha r = a \\
\end{cases}$  

in the third, the pole is the center of rotation  
and r is distance cm-center of rotation 

ball that rolls: pole is cm, r is radius of ball  
rod blocked at extremity and rotating around it: pole is extremity still, r is distance pole-cm (half length)      

**Example: ball rolling on curved surface**

$\begin{cases}
ma = mg\sin \theta - f \\  
0 = mg \cos \theta -N \\  
I \alpha = R f \\
\alpha r = a
\end{cases}$  

the last one is the condition of the thread not slipping

**Example: sphere rolls down semicircular slope**

$\begin{cases}
ma_t = mg \sin \theta -f \\   
m\frac{v^2}{R} = mg \cos \theta - N \\  
I \alpha = R f \\
\alpha R = a_T
\end{cases}$  

**Friction in pure rolling**

friction is necessary to allow pure rolling, so must appear in eq 1 and 2 when modeling the problem   
but to solve system don't need to be able to model it, so could be that i'm not given the coeff, it's fine  
DO NOT NEED TO KNOW $\mu$ to solve  

**How to solve**

plug f from the (3) in f of the (2), find $a_t$  
use the (4) to find $\alpha$  

**Notes**

for rollingball, pick cm as pole  

$\alpha$: acceleration of rotation AROUND THE POLE CHOSEN (generally cm)    
bc splitting motion is cm part and relative part  

$a$: acceleration of pole chosen (generally cm)  
it's tangent to plane on which it's rolling  
so if rolling on flat surface it's horizontal, if rolling on semicircle it's actually the tangent acceleration  

if carrucola is not ideal, tension on both sides are not necessarily the same  
so $I \alpha = R(T_2-T_1)$  

**What friction does**

ball rolling does a slope  
there is non-zero $a_{cm}$, which comes from the weight of the ball  
so $v_{cm}$ increase  
so, to keep pure rolling, $\alpha$ increase too  
so the friction needs to do a torque for this to happen  
but still, friction don't do work  


ball pure rolling on flat plane  
there is null $a_{cm}$  
so friction is null too  
bc friction can only effectively exist to contrast another force causing translation (slipping)  
here there's no other force causing translation, so friction can't come up from nothing  
since friction is null, $\alpha$ is null too  

### With differential equation

as diff eq, can use either:  

- 2nd cardinal equation $\frac{dL}{dt} = \tau$, $\quad I \ddot{\theta}= \tau$
- energy conservation $\dot{E} = 0$  

steps method 1:  
1. compute $L$  
2. compute $\tau$  
3. put $\frac{dL}{dt} = \tau$  

steps method 2:  

1. write $E(z_1,z_2, \theta)$ 
can use as many vars as i want  

2. derivate $\dot{E}=0$  
rem all are functions of t, so need chain rule  

3. generally know relationship between the first derivatives, so plug them in  
stuff like $\dot{z_1} = -\dot{z_2}, \dot{\theta} R = \dot{z_1}$  
pay attention to sign  
these relationship are known regarding first derivative, not original funcs, which is why i substitute after derivating  

**How to pick pole**  

pick pole so the most N and T (tensions and reactions, those i can't model) have null arm  
better to give null arm to T that to mg, bc T can't be modeled  

generally good to pick:  
- if pure rolling, contact point  (better)  
so energy only has rotational component, not translational (bc in pure rolling does instantaneuos rotation around it)   
use Huygens-Steiner th to compute inertia  

- cm (worse)  
easy to find inertia, but both components in energy  

**Tips**

- momentum is only done by EXTERNAL FORCES (since internal are in pairs, will cancel each other out anyway)  
- pay attention to signs  

**Pay attention to sign**

at first must decide if positive is uscente o entrante in the page  
generally, positive is uscente  
so pick 3D coords, even if problem looks in 2D  

when writing a $L = mr \land v$ or $\tau = b \land F$:  
USE RIGHT HAND RULE to see if L is entrante or uscente, and based on what i decided at start if it's positive or negative  
note that $\vec{r}$ and $\vec{b}$ go from the pole towards the other extremity, pay attention to the direction  

WHEN USING VARS:  
but if using vars, like $L = mr \dot{z}$, then always put + before it, bc the var might be negative  
so usually:  
L has vars, so put always +  
M has no vars ($M = Rmg$), so need to check sign with right hand rule  

**Vincoli between vars**

pay attention to sign  
if when one go up, the other go down, might have same value but different sign  
for angles, must choose which direction of rotation i consider positive (to know when speed and acceleration  are positive)   
choose it however i like, just make sure it's the same in all the problem
