**What is it**

used to write energy and angular momentum of a rigid body or a 2-body problem   

can split KINETIC energy (th 1) and angular momentum (th 2) in:  

- cm component (coords in FRAME OF LAB)  

- relative component (coords in FRAME OF CM)   
to go in frame of cm need to model using the reduced mass   

splitting is called decoupling  

this decoupling method works with any point P  
but Koenig th is the name of the version where P is the cm  

P muat be either a still point or the cm  
cannot be another moving point, or the formulas would change  
so Koenig th is realizing that these formulas work for the cm, even if it's moving  

**What it means to pick P**

energy: the energy must be the same whatever pole i pick  
just pick the one where the calculations are easier  

actually, energy is the same in all inertial frames  
so if cm is accelerating, energy in its frame will be different from energy in lab frame  

L: changes depending on the pole  
so pick the P required by the question  

### Write energy

$K = \frac{1}{2}mv_P^2 + \frac{1}{2}I_P \omega ^2$  

**Explanation**

P can be any point (generally cm or contact point with the ground)   
solution don't depend on which P i pick  

first addend: pretend the obj is a puntiform obj concetrated in P  
write energy of its translation  
(rem puntiform obj has no extension, thus cannot rotate around itself)  

second addend: describe the rotation of the obj ARLUND THE SAME POINT P as before   
describe how the mass distribution moves around the point P (in frame of P)   

**Picking P**

eg: ball attached to thread, does a circular trajectory on horizontal plane  

depending on what i pick as P:  

O, center of circular trajectory: $\frac{1}{2}I_{O}\omega ^2$  
- O don't translate, so first addens is null  
- obj rotates around O, so second addens there is  

cm of ball: $\frac{1}{2}mv_{cm}^2$  
- cm translates, so first addend there is  
- obj don't rotate around cm, its own axis, so second addens is null

**Example**

ball pure rolling in a circular trajectory  

pole: cm  
$K = \frac{1}{2}mv^2+\frac{1}{2}(mr^2)(\frac{v}{r})^2$   
pole translate, and rotation around pole  

pole: center of rotation  
$K = \frac{1}{2}(mr^2+mR^2)(\frac{v}{R}) ^2$   
pole don't translate, only rotation around pole, find its I with Huygens th  

note that angular speed around cm $\frac{v}{r}$ is different from angular speed around center of rotation $\frac{v}{R}$  
r: radius obj  
R: radius circular trajectory  

**Tip**

generally for rigid body use the cm as P, bc it's easier to compure its speed with energy conservation and find its momnentum of inertia   

if pick cm as pole the second addens appear only if body rotates around itself, not if does circuclar trajectory  

alternative ways to write  

Translation | Rotation
--- | ---
$\frac{1}{2}mv^2$ | $\frac{1}{2}I\omega^2$ 
$\frac{p^2}{2m}$ | $\frac{L^2}{2I}$  

obv this decoupling has sense only if the obj rotates (either around itself, or does a circolar trajectory)   
if it doesn't, just pick the cm as pole, and get normal formula for kinetic energy of translation  

**Picking P in pure rolling**

pick cm: there is both translation and rotation, but easy to find V and I  
pick contact point: no translation (bc in pure rplling, contact point is instantly still)  
need huygens-steiner to find I in contact point

### Find work

$dW = \vec{F} \cdot d\vec{r} + \vec{\tau} d\vec{\theta}$

note that: 
this is work only of force F, not of all forces  
$r$ and $\tau$ must be referred to SAME POLE  
can substitute: $d\vec{\theta} = \vec{\omega} \: dt$

**Which pole to pick**

rem must be still pole or cm (cm is only mobile le allowed)  
pick as pole:  

- point where F is applied, so $\vec{\tau} =0$  

- point around which the obj is rotating, so $d\vec{r} = 0$  
only if radius of rotation is constant, so if trajectory is a circle, bc this is how much the radius changes

used when doing energy conservation, to find work done by force
