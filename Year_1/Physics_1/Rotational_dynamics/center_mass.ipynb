### Definition

$\vec{R}_{cm} = \frac{\int \vec{r} \; p(r) dV}{M} = \frac{\int \vec{r} dm}{M}$

M: total mass og the obj  
$\vec{r}$: coords of the infinitesimal mass dm  
p: density  

**Components**

since R is a trivector, better to compute each coord by itself  

$\vec{x}_{cm} = \frac{\int \vec{x} \; p(x,y,x) dV}{M} = \frac{\int \vec{x} dm}{M}$  

need to make sure that the density does not depend on that coord though  
put as bounds the extremes where the coord varies (so mus have specified a coord system)  

#### Common cases

**If have many obj**

just add integrals, or can think it as formula with the summation  
$\vec{R}_{cm} = \frac{\Sigma mr}{M}$

**Obj with hole**

A: obj as big as the hole  
B: obj with the hole  
C: full obj, without hole

so A+B = C  
want to know C  

first find center of mass of A and B separately, so can consider them like puntiform objs them   
$m_A r_A + m_Br_B = m_C r_B$   

**Express infinitesimal volume**

in different coords:  
- carth: $dV = dx \; dy \; dz$  
- cylindrical: $dV = r \: d\theta \; dr \; dx$  
- spherical: $dV = r^2 \: \sin \theta \; dr \; d\theta \; d\varphi$

### Shapes

in all of them the x position of the CM is obvious  
so only care of y position  

use:  

$y_{cm} = \frac{1}{M} \int \vec{y} \; dm$

steps:  

1. express density (linear or superficial, depends on shape)  
2. express dm (using density in it) in terms of another infinitesimal, generally $d \theta$  
3. express the coordinate is terms of the infintiesimal chosen before, $d \theta$   
also take note of its bounds

now put together and compute

**Semicircular ring**

only have upper half, diameter is x axis  

$y_{cm} = \frac{2R}{\pi}$  

1. density  

$\lambda = \frac{M}{\pi R}$

2. infinitesimal mass  

$ds=Rd\theta$  (length of arc sottinteso by infinitesimal angle)  
$dm = \lambda ds$  
$dm = \frac{M}{\pi} d\theta$  

can also think of it like:  $\frac{dm}{M} = \frac{d \theta}{\pi}$  
but better to do properly with density  

3. coordinate  

$ y = R \sin \theta$  
goes in [0, $\pi$]
  

4. compute 

$y_{cm} = \frac{1}{M} \int _0^\pi  R \sin \theta \frac{M}{\pi}d \theta$  
can cancel the two M  

**Spherical shell ????**

**Semicircular disk**

like before, but filled  

each dm is a semicircular ring, concentrical  
they start from center, gradually go towards the border  

1. density  

$\sigma = \frac{2M}{\pi R^2}$  

2. infinitesimal mass

$dA = \pi r \;dr$  
$dm = \sigma \;dA = \frac{2Mr}{R^2} dr$

3. coordinate 

$y = \frac{2 \pi}{R}$  
is the center of mass of a semicircular ring, computed previously  

4. compute

$y_{cm} = \frac{1}{M} \int \frac{2 \pi}{R} dm$

### List momenti inertia

**Things with circular section**

principal axes are 1 normal to the section (thorugh the hole), 2 on plane of section, orthogonal between themselves  
obv all of them pass thorugh the cm, it's intuitive    
to picture them, they're axes of symmetry  

Obj | normal to section | on plane of section
--- | --- | ---
ring | $MR^2$ | $\frac{1}{2}MR^2$  
disk | $\frac{1}{2}MR^2$ | $\frac{1}{4}MR^2$  
cylinder | $\frac{1}{2}MR^2$ | $M(\frac{R^2}{4} + \frac{L^2}{12})$  
sphere | $\frac{2}{5}MR^2$ | equal


note: that of cylinder is sum of those of disk and rod  

**Other stuff**

puntiform particle: $mR^2$

rod: must be thin, so thread more than rod, can ignore size of section  
axis parallel to rod: 0

axis in its extremity:$\frac{1}{3}MR^2$  

rectangular plate (lastra)  
a is the side perpendicular to the axis chosen, the one that keeps it from being a thin rod  s

**Compute**

$I = \int x^2 dm$  
x is distance of that dm fron the axis chosen  
sometime is indicated with r  

example how to compute in ring:  

$I = \int _{ring} R^2 dm = R^2 \int _{ring} dm = R^2 M$  

**Ring**  

axis normal to plane (going thorugh, like needle):    
each dm is a puntiform mass

axis on plane:  
$ r = R \sin \theta$  
$\frac{dm}{M} = \frac{d \theta}{2 \pi}$  

**Disk**

axis normal to plane:    
each dm is a concentrical ring  
$\sigma = \frac{M}{\pi l^2}$  
$dm = \sigma dA$  
$dA = 2 \pi r dr$  

axis on plane:  
each dm is a thin vertical slice on the disk  
$dm =  \sigma dA$  
$dA = 2y \;  dx = 2\sqrt{R^2-x^2} \: dx $  (area of a vertical slice tall y, long dx)

**Cylinder**

axis normal to plane:  
cylinder is stack on disk, dm is a disk thick dz    
$\frac{dm}{M}= \frac{dz}{L}$  

axis on plane:  
I is sum of those of disk and rod, but computing it is hard, never done it  

**Sphere**

$dm = \rho dV$  
$dV = r^2 dr \sin \theta \; d\theta \; d\varphi$  
$x = r \sin \theta$  

integration bounds:  
$0 \le r \le R$  
$0 \le \theta \le \pi$  
$0 \le \varphi \le 2 \pi$  

since there are $dr, d\theta, d\varphi$ in the integral, can split in a multiplication of integrals, in each keep the factors with that var (if there are), else no factor in it, just a 1  

ex: $I = \int _0^R r^4 dr \; \int _0^{2\pi} d\varphi \; \int _0^\pi \sin^3 \theta \: d\theta$


**Rod**

assume that rod is a thread, so section is small  
formally, say: $\quad L \gg$ dimension trasverse  
don't care about shape of section of rod, can be a circle or a square or anything, don't care about section at all since assume it to be infinitesimally small

$\lambda = \frac{M}{L}$  
$dm = \lambda dx$  

**Lastra rettangolare**

axis on plane:  
dm is a thin vertical stripe  
$dm = \sigma dA$  
$dA = b \; dx$  b is perpendicular side, dx is infintiesimally small thickness of stripe

axis normal on plane:  
dm is a small rectangle/square on the lastra  
$dA = dx \; dy$  
$r^2 = x^2 + y^2$
