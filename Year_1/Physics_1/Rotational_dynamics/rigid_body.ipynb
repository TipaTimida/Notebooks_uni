### Definition rigid body

system of material points where distances between points are constant  

properties:  
1. incomprimibile  
2. indeformable  

**Frames to study it**

1. fixed frame, inertial  

2. cm frame  
translates so origin is always in the cm, but not rotates  

3. rigid body frame  
like cm frame, but also rotates  

2 and 3 are generally non-inertial (depends on movement of cm, if it has linear acceleration or rotation)  

**Motions of rigid body**

1. translation  
2. rotation  
3. rototranslation  

rotation means around an obj's own axis, has nothing to do with the trajectory  
therefore puntiform object cannot do rotation, bc has no mass distribution that can rotate  

**Center of instantaneous rotation**

changes each instant  
but in that instant, the rigid body is rotating around that point, which is instantly still  
like contact point with the ground in a wheel  

how to find:  
1. find 2 points whose velocity has a knwon direction in that instant  
    - cm might have speed only vertical, like in a scodella oscillante  
    - contact point with the ground has horizontal speed, bc else the obj would lift off the ground or sink, which is absurd  

2. draw the perpendiculars to the velocities  
3. the point where they meet is the center of instantaneous obscillation  

#### Degrees of freedom 

**Degrees of freddom (DOF)**

<u>number of independent vars</u> necessary to determine motion of a system  
also number of motion laws to describe the system  
DOF = number of coords - number of vincoli  

eg:  
particle moving in 3d space: DOF=3  
particle moving in 3d on surface on sphere: DOF=3-1=2  
particle moving on circulating ring: DOF=1  

**DOF of rigid body**

DOF = 9-3=6  

why:  
rigid body can be schematized with 3 points  
since each point has 3 coords, have 9 vars  
distance between a pair of points is constant, and that's a vincolo, so have 3 vincoli  

means i can:  
pick point A: all 3 coords can be picked  
pick point B: can only pick 2 coord, the third is obligated to respect the vincolo in distance AB  
pick point C: can only pick 1 coord, the others are obligated to respect the vincoli in distance AC and BC  
so can pick 6 coords  

note: if i add a point i add 3 coords and 3 vincoli (bc makes 3 new pairs of points), so 4th point is not needed to schematize a rigid body, since adds no DOF  

of the 6 DOF:  
3 are of translation, bc $\vec{R}(t) = (\vec{x}(t), \vec{y}(t), \vec{x}(t))$  
3 are of rotation, bc can rotate around 3 independent axes (x,y,z)  

### Pure rolling

**What is it**

also called rolling without slipping or without sliding  

is a particular case of instaneous rotation where the center of rotation is in contact with the surface  
(a priori there is no reason why it should be, it may even be outside the body)  

**Formal definition**

2 equivalent definitions of pure rolling:  

1. when $v_{cm} = \omega \land r = \omega R$
2. when there is instantaneous rotation around the contact point, which is still (so no relative motion between surface and contact point)  
so contact point has null speed  

note: v is a vectorial product, so it's perpendicular to both w and r  
another eq of pure rolling can be $a = \alpha \land r$

**Second definition explained**

Q: contact point  
P: a random point  

no relative motion between surface and contact point  
$v_Q = v_{cm} + \omega \land R_{cm \rightarrow Q}$  
if $v_{cm} = - \omega \land R_{cm \rightarrow Q} = \omega \land R_{Q \rightarrow cm}$:  

- $v_Q = 0$  
- $v_P = \omega \land R_{Q \rightarrow P}$  

so Q is instantly still  
all other points do a rotation around Q  

**Rolling vs slipping**

- relative motion between contact point and surface:  
  - rolling: no  
  - slipping: yes  

- attrition:  
  - rolling: static  
  - slipping: dynamic  

**Trivia: Consequences of rolling**

static attrition is conservative -> pure rolling is energetically convenient  

adherence: condition of pure rolling, wheel don't slip on surface  
ABS: when braking, changes speed of wheel to maintain condition of adherence  
when braking, if slip and lose adherence start using dynamic attrition, less strong then static attriction  

**Energy and friction**

to have pure rolling, there MUST BE FRICTION   
however, since contact point is instantaneously still, friction does null work, so energy is conserved    
(as a matter of fact it's static friction, which never does work)  
so speed of cm of object don't change, bc n

### Angular momentum of rigid body

$L_z = \omega I_z $ IF the axis is fixed  

**Particle vs rigid body**  

- particle: must specify pole  

- rigid body: must specify axis  

pole can be chosen anywhere on the axis, don't care, since i need the orthogonal distance anyway  

in rigid body stuff is called axial: axial angular momentum, axial momentum of a force  

**Direction of L**

in rigid body, $\vec{L}$ is not necessarily parallel to $\vec{\omega}$  
it is only if the body has some simmetry conditions  

if it does, $L_{rel} =  \vec{\omega} \; \Sigma m_i d_i^2$  

they weren't necessarily parallel in puntiform body either, only if pole belonged to plane of motion  

#### Axes

**Types of axis of inertia**

- PRINCIPAL axis  
the component of $L_{rel}$ perpendicular to $\hat{\omega}$ is null     
so momentum and angular speed are parallel: $\vec{L}_z$ \\ \\ $\vec{\omega}$  
so $L_{rel} = I_z \vec{\omega}$  


- NON-PRINCIPAL axis  
L has components both parallel and perpendicular to $\hat{\omega}$  
the component of L parallel to the axis follows $(L_{rel})_z = I_z \vec{\omega}$, but there is the perpendicular component as well, so this isn't the whole momentum   

**Existance**  

for every rigid body exist $\ge 3$ principal axes, orthogonal between themselves

**Momentum of inertia**

$I_z = \Sigma m_i d_i^2 = \int d_i^2 \; dm$    

$d_i$: distance between point and axis (distance means segment perpendicular to axis)  

the momentum of inertia is a geometric factor  
depends on the dsitribution of mass wrt the axis  

**2nd cardinal eq. for rigid bodies**

now must specify axis, not pole, unlike with particles    

$\vec{\tau} _{ext} = \frac{d}{dt} \vec{L_z} = I_z \vec{\alpha}$  

if z is a principal axis, thie is the only component of the momentum  
if not, this is only the component parallel to the axis, there is the orthogonal one too  

I: momentum on inertia  

**If I is not constant**

case where $\tau _{ext} = 0$  
eg: ice skaters doing pirouettes  

$I_z \alpha \cdot \omega \frac{d I_z}{dt} = 0$  
$\frac{\alpha}{\omega} = - \frac{1}{I_z} \frac{d I_z}{dt}$  

### Momentum of inertia

#### Huygens-Steiner th

conditions:  

1. z, z' are parallel axes  
2. z' passes THROUGH THE CM

$I_z = I_{z'} + Md^2$  

$d$ is distance between the axes z and z'  
pay attention to sign, it's z' that passes through the cm  
if want to pass from A to B, neither of which is the cm, must do from A to CM and from CM to B  

#### Inertia tensor

$\vec{L} = I \vec{\omega}$  

I can be written as a matrix  
in such case, it's called inertia tensor (tensore di inerzia)  

**Inertia tensor**

$I = \begin{bmatrix} I_{xx} & I_{xy} & I_{xz} \\
I_{xy} & I_{yy} & I_{yz} \\  
I_{xz} & I_{yz} & I_{zz} \\  
\end{bmatrix}$  

defines geometric and mechanic properties of a rigid body  

where:  

momentums of inertia: $I_{xx}, I_{yy}, I_{zz}$  
$I_{xx} = \int (y^2 +z^2) \; dm$  
(contains distance squared from the x axis, so somma in quadratura of the other coords)    

products of inertia: $I_{xy}, I_{xz}, I_{yz}$  
$I_{xy} = - \int xy \; dm$   
these are 0 when considering 3 principal axes   




**Observations**

the inertia tensor is a matrix:  
- symmetric  
- real  
- defined positive  

therefore:  
- it's diagonalizable  
- has 3 positive real eigenvalues: $I_{xx}^P, I_{yy}^P, I_{zz}^P$  
which are the momentums of inertia referred to the principal axes  
- their eigenvectors are the principal axes themselves  

this is proof that every rigid body has at least 3 principal axes  

**Coseno direttore**

given an axis u, in 3D space, its versors projected on each of the carthesian axes are:  
$u_x = \cos \alpha, u_y = \cos \beta, u_y = \cos \gamma$  

where $\alpha, \beta, \gamma$ are the angles that the u axis creates respectively with the x,y,z axis  

called coseno diretto bc cosines are used to define the versors  

**Theorem of Poinsot**

$I_u = u_x^2 I_{xx} + u_y^2 I_{yy} + u_z^2 I_{zz} + 2 u_xu_y I_{xy} +2 u_xu_z I_{xz} + 2 u_yu_z I_{yz}$  

if i know the inertia tensor of an obj  
can compute the momentum of inertia referred to an axis u ($I_u$), of which i know the orientation  
u can be any axis, as long as passes through the cm  


#### Inertia ellypsoid

**Inertia ellypsoid**

$I_{xx}X^2  +I_{yy}Y^2 + I_{zz}Z^2 +2(XY \; I_{xy} + XZ \; I_{xz} + YZ \: I_{yz}) = 1$    

this is formula of an ellypse with minor and major axes not parallel to carthesian axes  

how to get:  

1. divide the formula in Poinsot th per $I_u$  
2. substitute: $X = \frac{u_x}{\sqrt{I_u}}, Y = \frac{u_y}{\sqrt{I_u}},, Z = \frac{u_z}{\sqrt{I_u}}$

**Graphically**

ellypsoid is a 3d shape, like an egg  
draw ellypsoid and axis $u$, the distance of the point wher they touch is:  
$d^2  =  \frac{1}{I_u}$

what it means:  
if i know the eq of the inertia ellypsoid, can find the momentum of inertia wrt any axis  
and it's a geometrical problem, of finding a distance  
the shape of the ellypsoid is determined by the obj, expresss its geometry  

**Formula when using principal axes**

since products are null, becomes:  
$I_{xx}X^2  +I_{yy}Y^2 + I_{zz}Z^2 = 1$  

if compare it to normal formula of ellypsoid:  
$\frac{x^2}{a^2}+ \frac{y^2}{b^2} + \frac{z^2}{c^2} = 1$  

can see:  
$I_xx = \frac{1}{a^2}$  


if body is a sphere: all axis are the same, so $a=b=c=R$  
so there are infinit principal axis, all with same $I$  

#### Precession (moto di precessione)

**Definition**

a type of rotational motion in which THE AXIS OF ROTATION ITSELF ROTATES (or "wobbles") around another axis DUE TO A TORQUE acting on the system  

happens when a torque acts on a rotating body  
and such torque is not aligned with the angular momentum  
so, the obj precesses around the axis defined by the torque  

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRnANkfQikdLI9O_3lM4PRTvazqSY6MXwDYPA&s)

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcS_yj9_T95nzR24CxVvvjtevW4H8wRCGGZfVA&s)

**Example: rotating wheel hang on one side**

pick as pole the fixed point (pivot point), where the obj and rope touch  
where the tension is applied  

find $L$:  
wheel is rotating, use right hand rule (curl hand) to find direction of L  

find $\tau$:  
T is applied there, has null arm  
there is only mg, use right hand rule on it ( $\tau = b \land F$), find direction of torque  

what happens:  
know that $\frac{dL}{dt} = \tau$  
$L$ and $\tau$ are perpendicular, so $\tau$ changes the direction of L, not its magnitude  
so the vector L moves to try to end up over the vector $\tau$ (align itself with $\tau$)  
bc like applying a force changes the direction of the linear momentum, applying a torque changes the direction of the angular momentum  

so, to move L, the wheel must rotate  
this rotate both vectors L and $\tau$, so the situation is still the same, the rotation continues  

**Precession rate**

$\Omega = \frac{d \theta}{dt} = \frac{\frac{dL}{L}}{dt}=  \frac{\tau}{L}$    

$\Omega = \frac{\tau}{L} = \frac{b \; mg}{I\omega}$  

rem pole for $\tau$ and $L$ is the pivot point, or fixed point, not the cm  

$r$: distance from pivot to center of mass   
$ω$: spin angular speed (defines original rotation around its axis)  
$Ω$: precession angular speed (defines rotation of the axis itself)

note that $dL = L \; d\theta$ bc $L$ and $\tau$ are perpendicular, so it's just changing the orientation of L, not the modulo  
this means rotating a fixed-length vector  

**Examples of precession**

object | what is precessing  
---|---
trottola | the vertical axis 
gyroscope | axis of rotation 
earth | earth axis   


#### Gyroscope

**What it is**

rigid body that rotates really fast around one axis  
montato sui 'giunti cardanici', which allow it to rotate around 3 axes at once  
(essentially, there is perno in its cm, allows it to rotate around its cm)  

easier:  
a spinning disk  
mounted so that can freely rotate around one or more axes (gimbals)  

**What it does**

made to rotate really fast round an axis (x axis)  
this generates angular momentum L  

if something happens that makes a torque appear, the axis of rotation starts a precession (change direction to align itself with the torque)  

so the axis of rotation itself starts rotating around other axes  
note that it's not made to rotate around them, it's a consequence of the rotation around the x axis  

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRnMZHz_LbBL0iaafo-W2d_EhUWX4AfzEMlew&s)

**Example usage**

z azis is vertical, xy is horixontal plane    

it's made to ROTATE AROUND X AXIS    
its weight causes force in z axis 
this creates TORQUE IN Y AXIS    
so axis of rotation tries to turn to end up over the vector of the torque, thus rotates in the xy plane (ROTATION AROUND Z AXIS)  
this is the 'moto di precessione' 

**Speed**

$\Omega = \frac{\tau}{L} = \frac{\tau}{I\omega}$  

$\omega$: speed with which it's rotating around x axis, the original one  
$\tau$: torque (caused by weight of obj, or whatever the force is)  
 
so, for this speed to exist, there must be both rotation around x and y axis  

**Application in navigation**

giroscope is used to tenere la rotta in ships and planes  

for conservation of L, the aglar momentum, if thre are no external torque, it keeps its direction  
so if ship or plane changes direction, the gyroscope keeps the rotation axis in the original direction, so now appears storto  
formally: when plane change direction, there is a torque in y direction, which causes moto di precessione  

use system to monitor it  
when feel moto di precessione, know that direction has been changed, restore original one  
usare il moto di precessione per rilevare cambi di rotta  

#### Trottola

**Speed**

$\Omega = \frac{\tau}{L} =  \frac{mg R_{cm}}{I\omega}$  

$R_{cm}$: distance of cm from point of trottola that touches the ground (bc pole is the fixed point, so the contact point with the ground)  
$I$: measured wrt symmetry axis of trottola, which is tilted while rotating, bc defines original rotation around its axis  
$\omega$: speed of rotation around its symmetry axis  

![alt text](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRV6Wezm8CRpD-kB0qq4E5WdLscQwqB-EaezQ&s)

**Why it moves**

consider the contact point with ground as pole (so reaction of ground, N, has null arm)   

first rotates around its own axis (its principal axis), this produces L  
its weight causes a vertical force, thus an horizontal torque (in the drawing, entrante)  

moto di precessione: L rotates to align itself to $\tau$  
thus the axis itself rotates  
imagine the top of the axis draws a circle on the xy plane  

to find direction of $\Omega$ use right hand rule, curl hand to get $L$ to align with $\tau$  
start of fingers (where touch palm) is on L, top of fingers is on $\tau$  
thumb points in $\Omega$  

### Equilibrium

**What it means**

in equilibrium DON'T MEAN STILL  
it means that its motion don't change, don't accelerate (essentially stays subjected to principle of inertia)  
for a material point, means has rectilinear uniform motion  
for rigid body, pure rolling it allowed as well  

equilibrium position: position where, if put there WITH NULL SPEED, it stays still  
so the obj is deliberately put there with no speed, it's not the moment where it passes there dutring the motion  

**Conditions for rigid body**

- $\Sigma \vec{F} = 0$  
- $\Sigma \vec{\tau} = 0$  

can be proven wrt any pole (just one is enough)  
generally, pick pole so the normal reactions have null arm (since they're the hardest forces to model)  
for material point the second one is not needed, since has no extension  

note:  
rigid body has DOF=6, so need 6 conditions to keep it still (they seem 2, but each must applied in all 3 axes)  

**A different condition**

a condition can also be $\frac{dU}{dx}=0$    
x is whatever var $U(x)$ is written in  

if U is a local minimum it's stable, if it's a maximum it's unstable equilibrium  
just imagine plotting $U(x)$, the equilibrium points are on top of dunes or at bottom of bowls  
but this is generally used for puntiform objects  

to distinguish stable and unstable equilibrium: stable has the cm lower
