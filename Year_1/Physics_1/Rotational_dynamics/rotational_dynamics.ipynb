use $v = rw = r \dot{\theta}$ bc it's the component perpendicular to $r$

### Central forces

**Central forces**

$\vec{F}(r) = \hat{r} f(r)$

characteristics:  
1. directed towards the center of force (but don't know verse, could be attractive or repulsive)  
2. modulo is function of the distance from the center of force   

eg: gravitation, eletcrostatic, elastic

properties:  

1. conservative  
called theorem of circulation  
means work done on a closed path is null    

2. momentum null wrt center of force (so conserve angular momentum causes and planar motion)   
called law of areas

#### Law of areas

**What it is**

geometric equivalent of conservation of angular momentum  
formula: $\frac{dArea}{dt} = \frac{|\vec{L_O}|}{2m} = \frac{r^2 \dot{\varphi}}{2}$  

only true for central forces  

in words, it's Kepler's second law:  
the raggio vettore sweep same area in same time  
Kepler has anticipated it, proving it experimentally with the data from planets   

**Proof law of areas**  

1--- conservation of L  
L is wrt the origin O  
since F is central has null arm, so $\tau$ is null  
so L wrt the origin is conserved    

2--- infinitesimal area sweeped by vector ray:  

$ dA = \frac{1}{2} r \: r d \theta = \frac{1}{2} r^2 d \theta $  
$r d \theta$ is height of triangle of base r  

3--- areolar speed  

$\frac{dA}{dt} = \frac{1}{2}r^2 \dot{\theta} = \frac{L}{2m}$  

rem $L = mr^2 \dot{\theta}$

**Area spazzata**

$A = \int _0^T \frac{dA}{dt} dt = \int _0^T \frac{|L|}{2m} dt =  \frac{|L|}{2m} T$   

T: period to do orbit  
A: area of orbit (area, not length of trajectory)  


### For modeling

#### Relative coords

position of A relative to B: $r_{A,B} = r_{A} - r_{B}$  

coords on right wrt lab  
derivate to find speed, just change r in v   

when 2 bodies, usually written:  
$\vec{r} = \vec{r_1} - \vec{r_2}$  

#### Reduced mass

in 2 body problem:  

$\mu = \frac{m_a m_b}{m_a + m_b}$  

obtain that force from an object on the other is:   
$F = \mu \ddot{r}$

**Simplify reduced mass**

if $m<<M$:  

$\mu = \frac{m}{1+\frac{m}{M}}  ≈ m(1-\frac{m}{M})$  
so the reduced mass is equal to the smaller mass  
used Taylor approximation $(1+x)^{-1} ≈ 1 -x$  

#### Center of mass

**Center of mass**

- discrete: $\vec{R} = \frac{\Sigma m\vec{r}}{M}$
- continuous $\vec{R} = \int _V \frac{\vec{r} dm}{M}$  

derivate for speed and acceleration:  
$p_{tot} = mv_{CM}$  
$F_{ext} = ma_{CM}$

**System of CM**

it's the ref system where $v_{CM}=0$, so $\Sigma mv = 0$  
sum of momentums of stuff in the system is null  

**Position of stuff wrt cm**  

$r_{1\_cm} = r_1 - r_{cm} = - \frac{m_2}{M}\vec{r}$  
$r_{2\_cm} = r_2 - r_{cm} =  \frac{m_1}{M}\vec{r}$  

just plug in formula for cm on left  
also true for v instead of r (relative speed)

### Internal and external forces

action-reaction forces are a 'coppia di forze a braccio nullo', so their momentums cancel each other out 

therefore: 
- sum of internal forces is null  

- momentum of internal forces (wrt any pole) is null  
 
bc both things cancel out in pairs  


can be written: $\Sigma _i \Sigma _j \vec{F} _{j \rightarrow i} =0$, when $j \neq i$

**Motion**

motion of cm: depends only on external forces  
motion of obj in system: depends on both external and internal forces

**Energy**

must consider all sources of energy in a system, even if they come from internal forces (eg. internal spring)

## Systems: translational and rotational

### Conservations

**Isolated systems**

$\begin{cases}
\Sigma \vec{F}_{ext} = 0 & \rightarrow  \vec{p} = m\vec{v}_{cm} \\
\Sigma \vec{\tau}_{ext} = 0 & \rightarrow \vec{L} \\
W_{non \: cons} = 0 & \rightarrow E

### Cardinal equations of system dynamics

1. translation  

$ \Sigma \vec{F_{ext}} = \frac{d\vec{p}_{tot}}{dt}$   



2. rotation (M and L must be wrt same pole)

$ \Sigma \vec{\tau}_{ext} = \frac{d\vec{L}_{tot}}{dt}$

$P=\frac{dK}{dt} + \frac{1}{2}\frac{dm}{dt}v^2$

$\Sigma \vec{\tau}_{ext,O} = \frac{d\vec{L}_O}{dt} + \vec{v}_{pole} \land \vec{p}$
