### Stacked boxes

![](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcRqCuY2RpUYnzZpNycBkWDFcNmwgJKjr-SksA&s)

m: top one  
M: bottom one  

**1. Coord system**

carthesian coord  
x parallel to table  

**2. Free body diagram**

objects: floor, M, m  

forces on M:  
weight: Mg  
contact forces: $N_1$ (with floor)$, $N_2$ (with m), $F$ (with person pushing it)   
friction: $f_1$ (with floor), $-f_2$ (with m)

forces on m:  
weight: mg  
contact forces: $N_2$ (with M)   
friction: $-f$ (with M)

have both $f_2$ and $-f_2$, bc both m and M are things i'm interested in  
but i don't care about the person and the floor, so i don't have $-F$ anf $-f_1$  

**Notes**

weight:  
when: on every object with a mass (that i care about)    
direction: always towards down  

contact forces:  
when: when there is interface between two objects (even if only care about one)  
direction: perpendicular to interface, pushes an object away from the interface  

friction: when: when there is interface between two objects AND such interface has friction coeff $\mu$  
direction: parallel to interface, opposes motion  
value: $f=\mu N$, where N is the contact force on that interface    

forces always appear in twos, with same modulo and different direction  
one is acting on object A, the other is acting on object B  

the only distant force i care about is weight, the attraction between the object and earth  
so never put its opposite, bc never care about the earth  

**3. System**

$\begin{cases}
M {\ddot x}_M = F -f_1 -f_2 \\
M {\ddot y}_M = -Mg - N_2 + N_1 
\end{cases}$


$\begin{cases}
m {\ddot x}_m = f_2 \\
m {\ddot y}_m = -mg+N_2 
\end{cases}$

**Notes**

can skip writing everything in vectorial form, project directly on x and y axis  
and put the right direction/sign for each force, by looking at the PICTURE (the name +N and -N don't mean the directiont they're in, just which arrow to look at)  

**4. Vincoli**

$\ddot y = 0$ per entrambi  
$\ddot_m = \ddot_M = \ddot x$ (assumo siano uguali, cioè che l'attrito sia abbastanza forte) 

**5. Risolvo**

trovo forze vericali:  
$N_2 = mg$  
$N_1 = Mg + N_2 = (m+M)g$  

sommo le forze orizzontali:  
$(m+M) \ddot x =  F -f_1$  
$(m+M) \ddot x = F - \mu N_1 $  
$\ddot x = \frac{F}{m+M} - \mu g$  

### Block on moving plane

#### Other way

block on plane, left free to move  

**Coord**

X,Y: the usual coords, X points to the right  

**Forces**

R: between box and plane

**System**

$m \vec a = m \vec g + \vec R$  
$\begin{cases}
m \ddot x =...  \\ 
m \ddot y = R - mg \cos \theta
\end{cases}$

non mi interessa, ma lo scrivo:  
$M \vec A =  M\vec g + \vec R + \vec N$    
$\begin{cases} 
M \ddot X =  R sin (\theta) \\
M \ddot Y = -Mg + N - R \cos \theta
\end{cases}$  

**Vincoli**

$\ddot Y =0$, perchè plane on può sollevarsi da terra  
quindi $A = \ddot X$, è orizzontale    
quindi so proiettare: $M\ddot x =  A \sin \theta$  
mi interessa dopo, quando devo sapere come proiettare A  

$\ddot x_{rel} =0$, perchè il blocco non può nè affondare dentro il plane, nè sollevarsi a esso   
è la velocità perpendicolare all'interfaccia  

**Risolvi**

acc relativa:  
$a_r = \vec a_m- \vec A_M$  

componente perpendicolare:  
$\ddot x_{r} =  \frac Rm- g \cos \theta  \quad - A \sin \theta = 0$    
$R = mg \cos \theta - mA \sin \theta$   


**Solution example**

find RELATIVE ACCELERATION of m, the obj:      
$m \vec a_r = m(\vec a- \vec A) = m \vec g + \vec R - m \vec A$  
 
its component perpendicular to plane must be null:   
$ma_{r \perp} = mg \cos \theta + R - mA \sin \theta = 0$  
$R = mg \cos \theta - mA \sin \theta$   

plug R in vector eq for the plane to get A:  
$A = \frac{m \sin \theta \cos \theta}{M + m \sin^2\theta}g$   

plug A in eq of realtive acc to get $a_r$:  
$a_r = \frac{(m + M)\sin \theta}{M + m\sin^2\theta}g$
