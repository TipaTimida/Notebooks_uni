### Elastic force

Hooke's law: 
$\vec{F}_{el} = -k(\vec{x}-l_0) = -k(\Delta l)$  


**Tips**

remember the MINUS SIGN, says force is in direction opposite to compression  
bc x and F must follow same coord system, and F is in opposite direction of x (if stretched to the right, force is towards the left)     
first form is better, final length - rest length  

$l_0$ is REST LENGTH, not equilibrium length  

**Parameters**

- pulsation:  $\ddot{x} = -w^2 x$   
write newton law and find harmonic equation (to check is correct there must be - sign)  
usually, pulsation: $w^2 = \frac{k}{m}$  

- amplitude: $ampl = \sqrt{A^2 + B^2}$  
they're the incognite i found by applying the initial conditions  
or, if i'm told the extreme point, can find it by $ampl = 2(extreme - center\_ obsc)$  
the extreme point is that where the masses are instantaneously still  

- period:  $T = \frac{2 \pi}{w}$  
note $w$ is the pulsation, not the tangential velocity   

- center of obscillation    
<u>particular solution of diff eq in stationary state</u>    
in basic case is $l_0$, if add more forces can change    
ex. if spring hangs vertically and there is gravity, center of obscillation is lower

**Pulsation vs angular velocity**

$w$: pulsation -> it's a constant value    
$\dot{\theta}$: angular velocity -> it's a function of time   

they're different, they only happen to be the same in circular motion

### Normal attrition

call it f, minuscola  
$f_s$ or $f_d$ for static and dynamic

**Types**


- dynamic 
  - <u>relative motion</u> between body and surface  
  - direction opposite to velocity

- static: 
  - <u>NO relative motion</u> - all contacts points are instantly still (eg pure rolling)
  - direction opposite to net force

static coefficient always stronger than dynamic: harder to make body start moving than to keep it moving  

**Notation**

1. attrition and normal force are perpendicular, so need modulo of N, not vector  
YES: $\vec{F} = -\hat{v} \mu _d |\vec{N}|\;$  or $\;| \vec{F} | = \mu _d | \vec{N} |$  
NO: $\vec{F} = \mu _d \vec{N}$

2. static attrition is an inequality (dynamic is not, no prob there)   
YES: $ |\vec{F} | \le  \mu _s |\vec{N}|$  
NO: $|\vec{F} | =  \mu _s |\vec{N}|$ -> this is the force of first detachment  


when writing vectorial forces can use  $ \vec{F}_{max}  =  \mu _s |\vec{N}|$  
must put max, bc represent the upper limit, not the actual force there  

---


in definite, write:  

$| \vec{F} | = \mu _d | \vec{N} |$   
$ | \vec{F} | \le  \mu _s |\vec{N}|$    


**Notation: find min/max $\mu$**

$f \le N \mu$  
condition for an obj to be still:  
f: the friction force it's actually producing, or the one i need for something  
$N \mu$ : the teoretically maximal force that can be produced  
obj is still if the actual/needed force is less than the theoretically possible force, means can be done  

common problems:  
- must put $\le$, not $=$  
- must write scalar, not vectorial, bc f and N are orthogonal  

so, if i know a formula for f, eg: $f=mg$  
$mg = f \le N \mu$  
$mg \le N \mu$  

now solve  for $\mu$

### Viscuous attrition

**What it is**


- need <u>relative motion between</u> solid body and fluid, similar to dynamic attrition  
- scales with speed: $\vec{F}_v = - \vec{v} f(v)$  
- for slow velocities: $\vec{F}_v = - \alpha \vec{v}$

viscous force always has negative sign, since produces an acceleration in the direction opposite of the current velocity, and speed and velocity follow same coord system  

**Homogeneous sphere**

Stokes's law: $\alpha = 6 \pi R \mu$   
$\mu$ is density of fluid  

**How to solve**

$m\ddot{x} = - \alpha v+ F_0$

useful to write $\ddot{x} = \dot{v}$ and <u>solve differential equation for $v(t)$</u>

**Extract information from solution**

characteristic time:  $\tau = \frac{m}{\alpha}$, time to reduce speed of a factor $1/e$, around 3  
just used to write easier

after finding $v(t)$, to know it at specific moments:  
- at end: lim of t to infinity  
- at beginning: assume $t \ll \tau$, use Taylor approximation of the first order on $e^{-t/ \tau}$ since exponent is nearly zero   


to compute position at a certain time:  
$x(t) = x_0 + \int _0^t v(t') dt' = x_0 +  \int _0^t v_0 e^{-t'/\tau} dt'  = x_0 + v_0 \tau (1- e^{-t/\tau}) $

## Work and energy

**Work**

$W = \int \vec{F} \cdot d\vec{l}$  

it is:   
- force per displacement per cosine of angle  
- force per component of displacement parallel to force

**Power**

$\frac{dK}{dt} = P$  

power is derivative of kinetic energy

**Th of live forces / conservation of mechanical energy**

$K_2 - K_1 = W$  
$E_2 - E_1 = W_{non \: cons}$  (bc work of conservative force is vartiation of potential energy) 

call it conservation of MECHANICAL energy  


**Is energy conserved?**
 
only if work of non conservative forces is null  

- yes: $\frac{dE}{dt} = 0$   

- no: $\frac{dE}{dt} = \frac{d}{dt} W_{non \: cons} = F \cdot v$    

remember chain rule: $\frac{d}{dt} x = 1 \dot{x}$



**Potential energy**  

$\vec{F} = -\frac{d}{dx} U(x)$   
it's the gradient in 1d


**Kinetic energy**

$k = \frac{1}{2}mv^2 = \frac{p^2}{2m} $  


**If system of two objects**

using koenig th, 2 ways to write energy:  
- compute $v_{cm}$ properly, with formula  
- swap cm energy with momentum 


$k = \frac{1}{2}mv_{cm}^2 + \frac{1}{2}\mu v_r^2$
