**Steps for problem**

1. choose coord system (set axes)  
2. free body diagram (draw forces)
3. equations in vectorial form (indepenent of coord system)
4. project equations on axes
5. find vincolar equations
6. solve

### 1. Choose coord system

**Tips**

- can be different for each object   
- choose the one that gives the EASIEST VINCOLAR EQUATIONS (eg. no motion on one axis)   
- if plan to use energy conservation point y upwards, help with sign   

eg:  
plane: x parallel, y perpendicular  

### 2. Draw free body diagram


##### How to draw

order to remember:  

1. weights  
2. contact forces (vincolar reactions, tensions)  
3. elastic forces   

**Tips for drawing**

1. DO NOT DRAW CENTRIPETAL FORCE, it's virtual  
2. only draw forces applied ON the object</u>, not done by the object (omit those ON stuff i don't care about, like the floor)  
3. CONTACT AND ELASTIC forces always IN PAIR (each acting from one body on the other), called F and -F  

first draw all on main picture  
then split drawing for each obj   
start identifying angles, will be useful for projecting  

##### Project force in components

to get components, <u>original F must be hypotenuse</u>  
so draw x and y components with rect angle in between    
must multiply for some trig func, not divide by it  

draw well angles to find right trig func to project (use rule of alterni interni etc to identify angles)   

when asked an angle i don't know  
instead of expressing it like arctan of something, wait to see exactly what trig form it wants (sin or cos) and express that with pytagora based on sides    
like: $sin(\theta) = \frac{opposite}{hypothenuse}$  

##### Identify angles

- identify angles  

if 2 planes are perpendicular, they make equal angles with perpendicolar lines (angle of one with the vertical same as angle of the other with the horizontal)  

##### If plane can move

- what if plane can move?

must add plane to obj i care about  
so add  contact forces felt by the plane (those done by plane on other bodies are already there, don't change them)    

### 3. Vectorial equations

$m\vec{a} = \Sigma \vec{F}$  

##### Checks and notation

**Checks**

- is there attrition?  
<u>to have attrition there must be normal reaction</u> between 2 bodies  
if they just dangle one next to the other no force between them, don't care if there is coeff of attrition  


- where is weight?  
stuff like <u>mg only acts on that mass m, can't appear in vectorial equation of other bodies</u>    
in those would appear stuff like the normal force exercited by it  

**Notation**  

- write $ma$ as the first member, since that is generally what i'm looking for   
- arrow over vector  

- always use + sign, except:  
    - contact forces that have - bc they're the opposite of another  
    - elastic forces: $\vec{F}_{el} = -k\Delta \vec{x}$  
    but generally leave it as first form

### 5. Vincolar equations

**Recap**

1. plane: no perpendicula acc  
2. obj pushing horizontally another: same horizontal acc  
3. thread: all tensions equal, obj at end same acc  
4. carrucola: sum of forces null

1. plane  
obj on it has NULL PERPENDICULAR ACCELERATION/SPEED    
(need coord system parallel to plane)  

2. obj pushing another  
both have SAME HORIZONTAL ACCELERATION/SPEED  

3. thread  
    - all TENSIONS IN A THREAD ALL EQUAL (null mass)  
    write modulo of tension vector for ease: $|\vec{T_1}| = |\vec{T_2}| = T$  

    - OBJ AT END HAVE SAME ACCELERATION  (inextensible)   
    att: if at end is carrucola, that is the obj to consider, not the body pulled by the carrucola  
    for that, need <u>model thread length and derivate</u>   
    to check, draw thread moving by dx, see what other dx it creates, how many are on the other object    

4. carrucola  
SUM OF FORCES IS NULL (null mass)   
use to find relationship between tensions on different threads

#### Example: pendulum

**Polar coords**

- vectorial  

$ m\vec{a} = \vec{T} + m\vec{g}$

- project on axes $\hat{r}, \hat{\theta}$  

$ \vec{T} \cdot \hat{r} = -T$  
$ \vec{T} \cdot \hat{\theta} = 0$

$ m\vec{g} \cdot \hat{r} = mg cos\theta$  
$ m\vec{g} \cdot \hat{\theta} = -mg sin \theta$


$m\vec{a} = m[(\ddot r - r \dot{\theta}^2) \hat{r} + (r \ddot{\theta} + 2 \dot{r} \dot{\theta})\hat{\theta}]$

separate the two components, radial and tangential, to get 2 equations, like in x and y  

**Motion equation**

$\ddot{\theta} = -\frac{g}{l} sin(\theta)$  has no analytical solution  

with taylor approximation:   
$sin(x) = x + \frac{x^3}{6} \ldots$   
so can say      $sin(x) = x$ if $\frac{x^3}{6} << x$  
need to define what much smaller means  
generally ok angles up to 20 degrees  
it's called small obscillation regime
