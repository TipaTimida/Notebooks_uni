## Forces

**Definition**

forces are interactions between bodies  
virtual forces (centripetal) are not forces, don't have interaction

**Properties**

- vectorial
- superposition principle
- scale on distance
- indepenedent on reference system



**Types**

- at a distance: fundamental forces of nature (gravity, EM, nuclear)
- by contact: mascroscopic manifestations of fundamental forces (especially EM)
  - between rigid bodies: vincolar (attrition, tension, reaction)  
  - between deformable bodies: elastic, anelastic (pull too much a spring, more than limit of elasticity, loses springness)
  - rigid body and fluid: viscuous attrition, pression forces (Archimede)


**Virtual forces**

centripetal: label for forces that makes an object follow a circumference  
centrifugal: only appear in non-inertial reference systems, the one that pushes me outwards

### Types

**Conservative forces**

where work is independent from trajectory  
depends only on start and end point   
(intuitively, those that have a particular formula for the potential energy, that depends on the final state)  

examples:  
- static attrition, normal reaction (bc obj don't move, so don't do work, so tecnhically work is independent from trajectory)  
- weight
- elastic

circuitation theorem: work of a conservative force on a closed path is null


**Central forces**

create field with spherical symmetry  

examples:  
- gravity (full formula for astronomy)  
- Coulomb force 


**Impulsive forces**

to schematize a collision, assume it happens instantly, $\Delta t = 0$  
since $\int F dt = \Delta p$, to have a change in momentum must have an infinite force  
an impulsive force is one so big that can be thought to be infinite  
actually the interval is not zero, just really small, so force is not really infinite, just really big  



**Types of friction**

- attrition  (unspecified)  
can be static or dynamic  
happens to indeformable objects  

- attrito volvente  (rolling friction or rolling resistance)  
happens to deformable objects rolling  
like a ball whose contact part get a bit squished  

- viscuous attrition  
happens to obj that move in a fluid or a gas  

**Viscuous attrition**

$f(v) = \alpha + \beta v^2 + \gamma v^3 \ldots $ is basically a polynomial, depending on how big v is dominates a different term  
in class use slow velocities, so dominates lowest term: $\vec{F}_v = - \alpha \vec{v}$  
$\alpha$ depends on: shape of body, density/viscosity of fluid


**Spring**

usually only consider force from spring to body  
there is also force from body to spring, but is ignored  

- equilibrium point   
point where, if both velocity and net force are null, movement don't start   
it's the stationary sol of diff eq

## Principles of dynamics

1. Inertia principle  

harder: exists at least one reference system where $\Sigma F=0 \longleftrightarrow \vec{a}=0$ 

easier: body perseveres in its state of stasis or uniform motion until a force acts on it  

part about reference system is bc in centripetal system this principle it's not valid, only valid in inertial systems  
inertial system: those where $1^{st}$ principle is valid  



2. Fundamental law of dynamics  

in all inertial systems $\Sigma \vec{F} = m \vec{a}$  

also only valid in inertial system  


3. Reaction

$\vec{F}_{A \rightarrow B} = -\vec{F}_{B \rightarrow A}$  

when drawing free body diagram, only draw forces *applied on* the body, not done by the body

**Third law, inverse forces**

how to recognize:  
- must be parallel, but opposite direction
- must be on contact surface between 2 bodies  
- must ACT ON DIFFERENT BODIES (so mg and N are not opposite, bc both act on same body)  

essentially they act on the same 2 bodies, but swap who does it and who receives it  

## Vincolar forces

**Vincolar forces**

to each force, associated a vincolar equation  

types of vincolo:
- smooth / without attrition -> if fixed, force perpendicular to motion (eg. smooth plane that just prevents from falling down, gives $N$)
- scaber / with attrition -> force parallel to motion (eg. scaber plane with dynamic attrition, gives $F_d$)
- fixed: eg. plane
- mobile: eg. elevator, forces acceleration to be its

types of vincolo:
- thread: tension only in traction
- hook (gancio): tension both in compression and traction  
- plane: N perpendicular to it  
- zip (cerniera): forces obj to do a certain motion (eg. door hinges)  

**Examples of vincolar equations**

in all equations, instead of x and y can put their derivatives  



ex 1: horizontal plane, just prevents from falling  

- normal coords:  
$y(t) = 0$  
so $\dot{y} = \ddot{y} = 0$  

- coords perpendicular and parallel to plane:  
$\frac{\Delta y}{\Delta x} = - tan (\theta)$  
could also find upward velocity and say it must be null
  

ex 2: two bodies must move together:

- same coord system:  $x_1 = x_2$

- body on plane has coord system parallel to plane:  
$\frac{\Delta y}{\Delta x} = \frac{\dot{y}_{block}}{\dot{x}_{plane}- \dot{x}_{block}} = - tan (\theta)$   
easier to just project velocity horizontally (of block moving obliquous), then put horizontal velocites equal

## Energy

**Theorem of alive forces**

$W = K_2 - K_1$  

when inverting path:  
- if no dissipative force: switch sign of work  
- if dissipative force: work has same sign  


dissipative forces: those where always $W<0$, resistant work    
direction opposite to velocity, so oppose displacement  

**Gradient**

operator called nabla    
has the derivative in all directions  

$\vec{F} = - \vec{\nabla} U = -(\hat{x}\frac{\partial }{\partial x} + \hat{y}\frac{\partial }{\partial y})U(x,y)$  

in 1d:  $\vec{F} = -\frac{d}{dx} U(x)$  


**Potential energy**

need to define arbitrary zero  
for gravity: ground level  
for elastic: rest length  


**Concepts of dynamics**

- equipotential surface: geometric plane where U(x,y,z) is constant  
can be called superficie di livello (linea in 2d)  
eg: isobaric weather forecast, the circles showing pression areas  

- equilibrium:  
since $\vec{F} = - \vec{\nabla} U$, F points in direction where U diminishes, so down slope  
therefore bodies tend to go to points of minimum potential energy  
types:  
  - stabile: local minimum U, small displacement sends it back
  - instabil: local maximum U, small displacement sends it further away
  - indifferent: equipotential surface (eg. plane), after small displacement stays in new position 

- force field: set of all values assumed by a grandezza fisica

- force lines: imaginary lines used to represent the direction and strength of a vector field  
called inviluppo delle direzioni delle forze  
properties:  
  - lines are tangent to vectors  
  - can't cross (else movement is not deterministic)  
  - can't form closed loop (only in conservative forces)  

**Why no closed loops in conservative forces**

bc of the circulation th, conservative force has null work in a loop  

but if lines form a loop, since lines are tangent to the vector of the force, the force is never perpendicular  
so net work is not null  
so loop would have non-null circulation, which can't be  

## Dynamic system

**Dynamics of systems**

mechanical system: set of material points subjected to dynamics laws  
i choose arbitratily what's part of my system  
can be discrete or continuous, depending on mass distribution: $M = \Sigma m_i, \: M = \int _V dm$  
can be deformable or indeformable (rigid body)  

types of forces:  
- internal: between 2 obj of system  

- external: one obj in system, one not (mg external, bc don't consider earth part of my system)  
sum of all internal forces are null (bc all forces are in pairs)  
but attention, internal forces on an obj can be not-null, so can make the obj move, just the CdM receives no force  

**Koenig theorem**

in a system, can express its kinetic energy separating:

1. energy of CdM
2. energy of relative motion of obj in it  
in kinetic formula, spedd is speed relative to cm       
with 2 obs, imagine single obj with reduced mass and speed equal to relative speed between them  
$k_{rel} = \frac{1}{2} \mu \dot{r}$  (rem r is relative coord)



**Collisions**

interaction between bodies in small time interval, like nanoseconds ($ \Delta t \rightarrow 0$)  
so force big, time small  

impulsive force: 
- hard to schematize, just observe change in momentum, a mascroscopic effect  
- to characterize, study impulse  
- intensity so big that during collision can ignore all other forces (like, gravity don't move an object much in nanoseconds)  

if only impulsive forces are those of collision the system is isolated, speed of CdM is conserved  

**Types**

all have conservation of momentum  

1. elastic  
impulsive forces are conservative -> energy conservation  

2. totally anelastic  
impulsive forces non conservative -> no energy cons  
after colliding, bodies stay together -> speed will be the same   
def: totally anelastic collision dissipates all relative kinetic energy of system (obj have no more relative motion, all move together)  

3. partially anelastic  
forces non conservative    
bodies separate after colliding   
so have no vincoli, harder  
dissipates a bit of relative kinetic energy, but not all (totally anelatic) or none (elastic)
