### Galileian vs relativistic mechanics  


sometimes i'm asked to compute something, generally a length, in both galileian and relativistic  

difference is that in galileian mechanics don't use lorentz transformations, lengths and time are equal in all frame, so don't put $\lambda$  
eg, means decay time is the same in both the particle frame and the lab frame  
while in relativity the time changes, must change it with Lorentz trasnforamtion  

### How do ships meet?

in this kind of exercises, always try changing signs $\pm$, often that is the problem  
need speed of the final frame seen by initial frame to change frame  

note: 1D motion  

in ref system of earth:  
ship 1 starts at $x=0, t=0, v_1$  
ship 2 starts at $x=0, t=t_0, v_2$  
both go in same direction 

steps:  

1. compute time to meet in ref system where they both appear moving, so ref system of planet  
it's $T_m$  

2. compute time to meet perceived by each of them  
it's $T_s$    
each ship has a different v to plug in $\gamma$  

#### In frame of planet

**Not changing frame**  

if the speeds of the obj moving are expressed in the same frame i want to compute the time (or the space) in, then don't use relativity  
relativity is just to change frame  

start in this frame bc i'm given speeds of ships wrt this frame    
as long as i don't change frame (so as long as i'm not asked a time or space **perceived** by someone else, different from the one the dati are expressed in) don't need relativity, it's normal mechanics  

**For 1d problem**

time and coord to meet are:  
$x_m = v_1T_m = v_2 (T_m - t_0)$  
first compute $T_m$, then $x_m$  

**For 2d problem**

to compute time for meeting, draw a triangle:  

A___C  
|////  
|///    
|//  
|/
B

AB=d, distance between objects at start  
AC = vt, space travelled in time t by obj moving at speed v  
BC = ct, space travelled in time t by a photon, which moves at speed t  

use Pytagora on the triangle, get an equation, solve for t  

ex: ship move at speed v from A towards the right  
from point B, a still planet, are shot photons in all direction  
at what time (in frame of B, not of ship) the photons reach the ship?  

#### In frame of ships

**Time**

must use Lorentz transformation   

$T_s = \frac{T_m}{\gamma}$  

do it for each ship  
for each put in v its own speed wrt the planet  

**If ship changes speed**

if, in planet frame, the ship appears to be moving at $v_1$ for $T_1$ and $v_2$ for $T_2$  
must split the formula in two, bc in each interval must have constant speed:  
$T_s = \frac{T_1}{\gamma _1} + \frac{T_2}{\gamma _2}$  

if it's still wrt the earth $\gamma =1$, intuitive, earth and ship perceive same time for that period  

**Common exercise**

given speed of A and B in frame T  
have an event involving A (A reaches somwehere)  

1. compute time for event in frame T  
don't use relativity yet, no frame change  

2. compute time for event, in frame A  
event refers to A, so i'm going in its proper frame  
$t_A = \frac{t_T}{\gamma _{TA}}$  
where in $\gamma$ put speed of A seen in frame T  

3. compute time for event, in frame B  
2 ways:  
    - go from frame A to B, so the proper frame A appear, can use the simplified transformation  
$t_B =t_A \gamma _{AB}$  
in $\gamma$ must put speed of B in frame A, obtained with relativistic velocity addition  
    - use full Lorentz transformation between T and B  

**Full Lorentz transformation**

instead of doing velocity addition, can use full formula to go from frame T (where i know stuff) to frame B (that i want to know)  

$ct_B = \gamma _{TB} (ct_T - \beta _{TB}x_T)$

where:  
$t_T$: time for that event to happen in frame T (computed at point 1)  
$\beta _{TB}$: speed of B in frame T (attention to sign, if B moves towards left it's negative, so the - becomes +)  
$x_T$: space travelled (by the ship? or by the photon? look like by the ship) before the event in frame T  
eg: if the obj interessed moves at speed v and need time $t_T$ for the event to happen (speed and time computed in frame T), $x_T = vt_T$  
essentially it's side AC of the triangle above  

$x_T$ is the space, measured in frame of T, between the position the ship is in at the start of the event(point A, when the photon is thrown) and at the end of the event (point C, when it receives the photon)  
frame of T is needed to know the speed and time to travel from A to C, but it don't mean that T is the origin of my coord system, don't compute distance of point C wrt T, must compute distance of point C wrt A, where the event started    

#### Send light pulse

ships A,B start at time $t$ from opposite direction, go one towards the other   
at first, distance between them is d  

when they meet they send a radio signal (so with speed c)  
how long it takes from when ship 1 departed ($t_0$) to when signal reach the earth, in the different frames?  

**In frame of planet**

can solve in frame of earth, without using relativity  
bc event is light reaching the earth, involves the earth  


so in earth frame, time and position of meeting are:  

$ \begin{cases} 
d= t_m(v_A+v_B) \\
x_m = v_At_m \\
\end{cases}$  

time for signal to reach earth ($t_s$):  
$x_m = c t_s$  

so total time is:  
$t = t_m + t_s$  

note: moving ship sends signal, theorically should use velocity addition to compute speed of this signal  
but since it's c i know it will always stay c, so no need in this case  

**In frame of ship**

<u>can't use shortcut formula of time dilation, bc have no frame where the light pulse is still</u>    
so must do proper Lorentz transformation  

in earth frame, event has coords: $(t, 0)$  
$t$: total time in planet frame, computed before  
$x=0$: bc pick reference system with origin in planet   

$\beta$ is speed of ship in frame of planet  
use complete Lorentz transformations
