## How special relativity was born

**Problem**

in Maxwell theory, speed of light is invariant wrt frame (frame = reference system)  
not compatible with Galileo transformations

**Aether hypothesis**

light waves propagate in etere, which is everywhere in the universe  
(same idea as material waves need a mezzo materiale to propagate)  
so c would be the usual one only in the ref system where the etere is in quiete

**Michelson-Morley experiment**

measure speed of light wrt etere and earth  
use inteferometer to split bean of light, parts travel different paths, they reunite to cause interference  
wanted to measure the etere wind, cause by relative motion between etere and earth  
but found none

### Michelson-Morley experiment

](https://cdn.britannica.com/79/279-050-58C1AAC5/mirror-parts-interferometer-Michelson-light-beam-angle.jpg)

![alt text](https://blogmedia.testbook.com/blog/wp-content/uploads/2022/09/michelson-morley-experiment-1122dfdf.png)

$M$: beam-splitting mirror  
$M_1$: mirror on top  
$M_2$: mirror on right  

$l_1$: distance M and $M_1$ (vertical)   
$v$: speed of earth wrt aether (direction towards right)

**Compute times for travel of rays**  

time to travel from M to their mirror ($M_1$ or $M_2$) back to M


- ray 1  

consider only going up, going down is same thing  
$M_1$ moves bc of earth motion wrt aether, so ray must travel diagonal path, the hypothenuse  
use Pytagora on the triangle:  
$l_1^2 + (v \frac{t_1}{2})^2 = (c\frac{t_1}{2})^2$  

solve to get:  
$t_1 =  \frac{2l_1}{c} \frac{1}{\sqrt{1- \frac{v^2}{c^2}}}$ 

- ray 2  

the whole interferometer moves with speed v (relative speed of earth to the aether) to the right  
when going right, mirror $M_2$ moves away from     
when coming back , mirror M moves towards it  

$t_2 =  \frac{l_2}{c-v} + \frac{l_2}{c+v} =   \frac{2l_2}{c} \frac{1}{1- \frac{v^2}{c^2}}$    



**Recap**

time for light to travel from M to $M_1$ and back to M:  
$t_1 = \frac{2l_1}{c} \frac{1}{\sqrt{1- \frac{v^2}{c^2}}}$ 

time for light to travel from M to $M_2$ and back to M:  
$t_2 = \frac{2l_2}{c} \frac{1}{1- \frac{v^2}{c^2}}$  

**Practical problem**

can't know $l$ with precision good enough  
so do experiment two times  
the second time swap $l_1$ and $l_2$   

$\Delta t$: phase difference before swap  
$\Delta t'$: phase difference after swap  

each $\Delta$ causes an interference pattern in the eyepiece (if the values of the $\Delta$ are different, the pattern will be different)  


**Check phase difference**

$\Delta t' - \Delta t= \frac{l_1+l_2}{c}\frac{v^2}{c^2}$  

theoretically, difference of phase differences is non-zero  
so $\Delta t' \neq \Delta t$  
so interference pattern should be different  

**Problem**

however in practice, interference pattern in both cases (before and after swap) is the same  
so should be $\Delta t' - \Delta t =0$  
possible explanations:  

1. trascinamento dell'etere  
means aether is stationary wrt earth  
this way $v=0$  

2. contraction of Lorentz-Fitzgèrald  
this is NOT relativity, where contraction is just apparent for an observer, it's a real contraction  
this way the formula of $\Delta t' - \Delta t$ is wrong, bc didn't acccount for it  

both theories tested and proved wrong  
therefore initial hypothesis, that c follows galileian relativity for transformtion, must be wrong

## Simultaneity

https://sites.pitt.edu/~jdnorton/teaching/HPS_0410/chapters/Special_relativity_rel_sim/index.html

**Situation**

O: person still in the middle of points A and B    
O': person in car, going from A to B with speed v  
d: distance between O and A (or O and B)  

instead of trees might be better to picture light sources in point A and B, understand better why them moving should affect the speed of the wave (Like in does in the sound case)  

question:  
two events A and B happen (in those point) when O' is at the midpoint (and he knows he's at the midpoint)     
light of these events travel towards observers, are they simultaneous?  
does it change if events are sound instead of light?  

**Preliminaries: distance of events**

both observers know the events happened at the same distance ($d_A = d_B$)  
to know it, maybe events leave a mark in the coord where they happen and after the events the observers can go and check that they're at the same distance (since both observers ultimately think themselves as still, it's easy to see that the events were at same distance)  

don't mean events happened together  
just that event A happened when source of event A was at distance x  
at event B happened when source of event B was at distance x  
this is true for both observers, both can say this  

**What observers know**

observers know that:  
$t_0 = t_1 - \Delta t$  

$t_0$: time where the event happened  
$t_1$: time the event reached the observer  
$\Delta t$: travel time of the light  

times are in the frame of the person doing the calculation

**Wrt O (either sound or light)**

obviously events A,B reach him together: $t_{1A} = t_{1B}$  
O knows that the path travelled and speed is the same for both rays, in both light and sound cases   
so travel time is the same for both rays  $\Delta t_A = \Delta t_B$  

therefore $t_{0A} = t_{0B}$, events are simultaneous  

**Wrt O', sound (Galileian relativity)**

since sound is far from c, can use galileian relativity to find frame of O'  
O' moves at speed v towards B, so in its proper frame just need to add vector -v to all objects  

frame of O:   
O' moves at $+v$  
ray A moves at $+v_{sound}$  
ray B moves at $-v_{sound}$  

frame of O':  
O' is still  
ray moves at $+v_{sound} -v $  
ray B at $-v_{sound} -v$  

conclusion:  
in modulo speed of B increased, of A decreased  
so event B, the one O' is moving towards, reaches him faster: $t_{1B}<t_{1A}$  

observations:  
O' knows the rays must have travelled the same space  
since rays travelled same space, but ray A is slower, $\Delta t_A > \Delta t_B$  
therefore the fact that B arrived first ($t_{1B}<t_{1A}$) is explained by the fact that it had a slower travel time ($\Delta t_A > \Delta t_B$)    
so B too can do calculations using datahe perceived and confirm that the events actually happened simultaneously  

**Wrt O', light**

cannot use galileian relativity  
bc of postulate of relativity, B sees light move towards at same speed  
before it was $v_{sound} \pm v$, now both are $c$  

conclusion:  
again he knows the rays travelled the same space  
but now must have done so with same speed too, so $\Delta t_A = \Delta t_B$  

however B, the one O' is moving towards, reaches him faster: $t_{1B}<t_{1A}$  
if ray B arrived before and both had same travel time, means ray B started earlier  

**Conclusion**

observer knows how physics work  
can reflect on what he perceives and knows and do calculations   
wants to see if the events actually started together  

with sound:  
O' receives the event if different moments  
but knows one of the rays is faster  
so yeah, they actually started together  

with light: 
O' receives the event if different moments  
but knows they must have same speed (bc he knows well physics)  
he must conclude that the events didn't happen together  
and it's true, for him they really didn't

### Compton effect

#### What it is

- what 

diffusion of a photon, which collided against a still electron  
used in particle rivelators  


- why  

helped prove dualism wave-particle of light  
it proved that light (photons) carry momentum and can be treated as a particle  

**What happens**  

photon moves towards the right, on the x axis  
then hits an electron, who is still  
both move, each with an angle wrt the original direction of the photon $\theta _{\gamma}, \theta _e$

**What i want**

find relation between:  
- scattering angle of of photon: $\theta _{\gamma}$    
- energy of photon before and after $E_{\gamma 1}, E_{\gamma 2}$  

$\frac{1}{E_{\gamma 2}} - \frac{1}{E_{\gamma 1}} = \frac{1 - \cos \theta _{\gamma}}{m_e c^2}$  
$m_e$: mass electron  

#### How to solve

**Write 4-momentum**

photon (massless, so $E = p$)  

$w_{\gamma} = (E_{\gamma}, E_{\gamma}, 0)$  
$w'_{\gamma} = (E'_{\gamma}, E'_{\gamma} \cos \Phi, E'_{\gamma} \sin \Phi )$  


electron   

$w_e = (m,0, 0)$

**Conservation laws**

4-impulse conservation:  

$\begin{cases} 
E_{\gamma} + m = E'_{\gamma} + E_e \\  
E_{\gamma}  = E'_{\gamma} \cos \theta _{\gamma} + \vec{p}_e \cos \theta _e \\

\end{cases}$
