## Theory

constant means either invariant or conserved:  

- invariant  
the same in every frame (lab, cm)  
formally, the same in every system tied to the original one via a Lorentz transformation 

- conserved  
not change in time   

**What is constant**

- 4-impulse  
each component ($E, p_x, p_y, p_z$) is CONSERVED  
so can write a system of 4 eq (bc has 4 components) when changing time, NOT when changing frames, bc each component is NOT INVARIANT  

- norm of 4-impulse   
is both constant and invariant  


so:  
norm can be used all the time  
linear system only from $w$ to $w'$ (change time), not from $w_{lab}$ to $w_{cm}$ (change frame)

### Write 4-impulse

might be better to write it as a col vector, each component in a row  

**How to write 4-impulse of 1 particle**

for each particle (A):    

- energy  
  - $E_A$ if the particle moves  
  - $m_A$ if the particle is at rest    
  bc energy is only rest energy, no kinetic  

- momentum  
  - $p_A$ if the particle moves    
  - 0 if the particle is at rest  

example, contribution from a particle:  
moving particle: $(E_A, p_A)$  
particle at rest: $(m_A, 0)$  

**Tips momentum**

- in 2d plane  
$w = (E,p_x, p_y) = (E, p \sin \theta, p \cos \theta)$  

if working on 2d plane can split the vector to have 3 components:  

- cm frame    
$w = (E_A +E_B, p_A+p_B=0)$   

TOTAL MOMENTUM NULL, by def of cm frame (so the second entry in the 4-impulse is null)   
however the INDIVIDUAL PARTICLES CAN HAVE NON-NULL MOMENTUM -> so the energy component has E, not m  
so better to write the components, not just put 0, bc coud use them to swap between E and p of some particles  
 
- when looking for energia di soglia   
$w = (m_A +m_B, 0)$     

TOTAL MOENTUM NULL, bc i'm in cm frame by def  
ALL PARTICLES HAVE NULL MOMENTUM (definition of energia di soglia) -> so energy component has m, not E, since particles only have rest energy

**Example: common case**

before collision. lab frame: A moves, B still  
after collision, cm frame: C still  

$w_{lab}=(E_A +m_B, p_A)$  
$w'_{cm}= (m_c, 0)$  

**Notation**

call the 4-impulse w, so don't get confused with anything else  

put pedice to specify frame (cm or lab)   
put apostrophe to specify before (no) or after (yes)  

$w_{lab}, w_{lab}', w_{cm}, w_{cm}'$  


can give letters to particles, like A + B = C + D, better than $B_1, B_2 ... $ 

must give different names to energy and momentum in different frames, even if belong to same particle

### Norm conservation

**Example**

$\begin{cases}
||w_{lab}||^2 = ||w_{cm}'||^2 \\ 
E_A^2 = m_A^2 + p_A^2 
\end{cases}$ 

**Compute norm**  

$w = (E_{tot}, p_{tot})$  
$||w|| = (E_{tot})^2 - (p_{tot})^2$  

if split the momentum:  
$||w|| = (E_{tot})^2 - (p_x^2 + p_y^2 +p_z^2)$  

**Norm conservation: example, energia di soglia** 

1. write 4-impulse  

$w_{lab} = (E_A + m_B, p_A)$  
$w_{cm}' = (2m_C, 0)$  

2. compute norm  

$||w_{lab}||^2 = E_A^2 + m_B^2 + 2E_Am_B - p_A^2$   
$||w_{cm}'||^2 = 2m_C^2$  

3. equalize it  

$2m_C^2 = E_A^2 + m_B^2 + 2E_Am_B - p_A^2$

4. <u>swap momentum to difference of energy and mass</u>    

$2m_C^2 =  m_B^2 + 2E_Am_B - (E^2_A-p_A^2)$   
$2m_C^2 =  m_B^2 + 2E_Am_B -m_A^2$  

this way only a first degree $E_A$ is left, can compute it  

### Lorentz transformation

**Lorentz transformation**

$ \begin{cases} 
E' = \gamma (E-\beta p) \\
p_x' = \gamma(p_x-\beta E) \\
p_y' = p_y \\ 
\end{cases}$   

left side: frame i want to know about  
right side: values p,E in frame i already know about  

**Notes on stuff**


$\beta, \gamma$ are OF THE CM  
not of the particle i want to compute something of, but describe the transformation itself  

x must be the direction with the relative speed between frames  
so the direction the cm moves in (direction that the particles in lab frame before the collision move in)  
if it's in 1d can just write $p=p_x$   

note that the orthogonal momentum don't change  
if CM moves horizontally, to find stuff in LAB:   
$p_y$: is the same in CM and LAB, bc perpendicular to transformation     
$p_x$: must do Lorentz transformation  

**Attention to sign**

$\beta$: speed of the new system, seen by the old one  
$\gamma$: handles the relative speed between the 2 frames i'm considering, contains the same $\beta$ outside

attention to sign of $\beta$ in second addend in parenthesis:  
if in lab frame particle move towards the right, so speed is positive:  
to go from lab to cm, $\beta>0$, stays a -  
to go from cm to lab frame, $\beta <0$, so the - will become a +  

example: A moves in lab frame at speed $\beta _A$  
so proper frame of A sees lab moving with speed $-\beta _A$    

**Matrix describing it**

$\begin{bmatrix}
\gamma & - \gamma \beta\\
{-} \gamma \beta & \gamma 
\end{bmatrix}$

#### Mandertan invariant

**Mandertan invartiant**

$t=||w_A -w_B||^2 = ||w_A||^2  + ||w_B||^2 - 2w_A \cdot w_B$  

since it's a scalar product $\cdot$ on the last addend it's a scalar product between vectors (inner product)  
so it's computed like canonic in algebra:  
$w_A \cdot w_B = E_A E_B + p_Ap_B$ 

#### Decay

**Formula**

$l_{lab} = v_{lab} \; \tau _{lab} = \beta _{lab} \; \gamma _{lab} \tau \quad (c)$   

simplify:  
$l = \beta \gamma \tau = \frac{p}{E} \frac{E}{m} \tau = \frac{p}{m}\tau \quad (c)$   

**Where to get values**

$\tau$: if unspecified, is in cm frame  

$\beta _{lab}$: speed of particle in lab frame  

$\gamma$: transformation between lab and proper frame of particle  
contains $\beta_{lab}$, the same as outside  

to find $\beta$ and $\gamma$, plug values in lab frame of the particle that it's decaying:    
$\beta = \frac{p}{E}, \quad \gamma = \frac{E}{m}$  

**Tips**

for space travelled want answer IN REAL UNITS, not natural ones  
just MULTIPLY PER $C$ the formula for l    
rem $c = 3 \cdot 10^8$  

**Average life**


$\tau$ is the average life (or decay time)   
$\tau$ is, by definition, given as a time IN THE PROPER FRAME OF THE PARTICLE, so where the particle appear at rest  
so to get it in lab frame need $\gamma \tau$ (shortcut Lorentz transformation)   

**Relativity or not**

there are 2 types of this exercise, where i'm asked to compute distance travelled in LAB FRAME in time before decaying:    

1. using non-relativistic mechanics    
time is the same in all frames, $\tau _{lab} = \tau _{cm}$  
so $l = v \tau$  

2. using relativistic mechanics  
time differs in frames, $\tau _{lab} = \gamma _{cm} \tau _{cm}$  
so $l = v \tau_{lab} = v \gamma _{cm} \tau_{cm}$  

if asked x component of travelled space, just do $x = l \cos \theta$ (or sin, depends)   

the trig func is taken as a fraction of 2 momenti, draw the triangle

## Drawing

**Changing frame (swap lab and cm)**

what changes:  
- angles  
- direction of momentum  


**Tip for drawing**

before collision:  
lab frame: particle A moves to the right, particle B is still, they collide (or B there isn't, and A decays, same thing)  
produce C and D  

after collision:  
lab frame: multiple options:  
C and D have momentum of different modulo and opposite direction, bc total momentum must be that of A  
C and D have momentum of same modulo and forms same angle with direction of momentum of A, so perpendicular compoennets of momentum cancel each other out, paralle components add up to the momentum of A   
cm frame: C and D have momentum of same modulo, but opposite direction  
y compoenent of momentum must be conserved from the lab frame, so if those were horizontal this are too, if those were diagonal these are too, but with opposite direction, not forming same angle with the horizontal  


**Drawing angles**

- before collision, in lab frame:  
a particle move towards the right, another is still

- after collision  
two identical particles are created  
since they're identical, there must be symmetric, no reason why one should behave differently  

    - lab frame:   
    both move in DIAGONAL towards the right, with SAME ANGLE WRT X AXIS, but one above and une under the axis (so momentum in y axis stays null, on x axis is conserved)  

    - cm frame:  
    since particles are equal and have equal momentum, must move in OPPOSITE directions (so total momentum is null)

in general, if still particle splits in 2, the 2 have opposite directions (180 degree angle)

**Symmetries**

useful to find symmetries  
means particles that have no difference in initial setup, and will obv have same energy and momentum (bc no reason they should behave differently)  
see if initial situation, graphically, looks symemtric
