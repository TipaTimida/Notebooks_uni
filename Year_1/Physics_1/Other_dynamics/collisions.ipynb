## Conserving stuff

**What is conserved: general (eg. rotation, translation)**

- energy: if all forces that do work are CONSERVATIVE  
(so no friction)

- angular momentum: if no net TORQUE (momentum from external forces)    
rem to specify which pole, bc could give null arm to an impulsive force  

- linear momentum: if no net external FORCE  


**What is conserved: collision**

like the general case, but consider only the IMPULSIVE FORCES, not all forces (eg, ignore weigth)  

- energy: if all IMPULSIVE forces that do work are CONSERVATIVE   
conserved only if elastic  

- angular momentum: if no net momentum from external IMPULSIVE forces  
always conserved if pick as pole the point where they collided, bc impulsive force from collision has null arm     

- linear momentum: if no net external IMPULSIVE FORCE  


**Vincolated to rotate around a point**

vincolo is like a hang, keeping a rod hanging  
vincolo does IMPULSIVE FORCE in the point where the rod it's hung on the wall  

what is conserved:  
- energy: depends on type of collision  
- linear momentum: no, there is impulsive force  
- angular momentum: conserved only if pick as pole the point where the vincolar force is applied, so has null arm  

**Non-vincolated**

if has no vincoli, there are no impulsive forces:  

what is conserved:  
- energy: depends on type of collision  
- linear momentum: yes, bc no impulsive force  
- angular momentum: yes, wrt any point    

**Which forces are external**

force N from the collision is external? so, does it negate conservation of momentum?  
depends if both the colliding objects are part of the system (usually, obj part of the system are those of which i know the mass)  

eg: ball hits step, step is NOT part of the system (bc cannot write $v_{step}$, assume it's always still), so force between them is external  
linear momentum is not conserved  

eg. ball hits rod, both are part of the system, so force between them is internal  
linear momentum is conserved  

### Explanation conserving stuff

**Energy conservation: thread and carrucola**

intuitively: 
energy is conserved bc thread not slips on carrucola   
so there isn't relative motion between them  
so there's no heat from sfregamento  
so no lost energy  

formally:  
energy is conserved bc friction, which is non-conservative, does null work (all other forces that do work are conservative, like weight)     
bc of 3rd dynamics law, there is friction force from carrucola on thread and viceversa, with opposite direction  
carrucola and thread move in same direction, together (bc thread can't slip)  
so work of friction is:  
$dW = f \; ds - f ds = 0 $  (the two forces are one from carrucola on thread, the other the opposite)   



**Momentum conservation**  

momentum is conserved only if there are no IMPULSIVE FORCES 
 
bc $d p = I = F \; dt$   
impulsive forces, by definition, have $F \rightarrow \infty$ when $dt \rightarrow 0$  
so even in an INFINITESIMAL TIME, like in a collision, give a FINITE IMPULSE  
so they change the momentum 

generally impulsive forces are normal reactions from a vincolo

**Conservation of angular momentum in collision vs rotation**

the external forces, in both moments, are:  
$\vec{N}$: impulsive  
weight: non-impulsive  

collision with step:  
it's almost instantaneous, in an infinitesimal time $dt$, so ONLY IMPULSIVE forces are strong enough to make an inpact  
non-impulsive forces can be ignored, since it's just an instant, and these tend to 0  
so think that there is only $\vec{N}$, and pick a pole where its arm is null  
this way can conserve angular momentum  

rotation around step:  
here there is a finite time interval  
so cannot ignore weight  
so there is momentum of weight wrt contact point, and angular momentum is not conserved

## Theory collision

**Types**

all have conservation of momentum (if no impulsive forces, eg when wheel hits step and goes up it don't conserve momentum)  

momentum conservation happens if, during collision, all external forces are non-impulsive (only impulsive is the internal force causing the collision)

1. elastic: energy conservation  

2. totally anelastic: move together, speed is the same   
(and equal to $v_{cm}$ vefore the collision)     

3. partially anelastic: nothing

**Anelastic**

by definition, lose all the relative component of the energy  
bc after collision there is no more relative motion between the bodies    

$\Delta K = - \frac{1}{2} \mu v_{rel} ^2$  

also:  
- energy not conserved  
- if conserving L or p, rem the mass and ienrtia is the sum of both, bc both are translation or rotating with same speed  

eg: $m_1v = (m_1+m_2)v',\quad  I_1\omega = (I_1+I_2)\omega '$

**Restitution coefficient**

$k_2 = e^2 k_1$  

k is kinetic energy  
can only consider relative kinetic energy, since the one wrt the cm is equal before and after, get canceled  

e=1: elastic  
e=0: totally anelastic    
in between, partially anelastic  

collision between 2 obj:  

$e = \sqrt{\frac{k_2}{k_1}} =  \frac{|v_{rel \: 2|}}{|v_{rel \: 1|}} = \frac{|v_{B2} - v_{A2}|}{|v_{B1} - v_{A1}|}$   

**Scattering angle**


angle between momentum before and after the collision  
each body has its own   


only if:  
$m_A = m_B$  
B starts still (this is necessary)  
elastic collision  


- elastic: 90 deg, bodies are perpendicular  
- totally anelatsic: 0 deg, move together  
- partially anelastic: in between 


**Proof scattering angle is 90 degrees**

conservations:  

$\begin{cases}
mv = mv_A+mv_B   &  v_A+v_B \\ 
\frac{1}{2}mv^2 = \frac{1}{2}mv_A^2 + \frac{1}{2} &  v^2 = v_A^2 + v_B^2 \end{cases}$  

take dot product with itself  
(kind of square it, but use scalar multiplication between vectors, not just multiplication):  

$v \cdot v = (v_A+v_B)  \cdot (v_A+v_B )$  
$v^2 = v_A^2 +v_B^2 + 2 v_A \cdot v_B$  

bc of the enery conservation. i know that $v_A \cdot v_B = 0$, so they're perpendicular  

when using vectors, square of a binomio uses the scalar product for the double product, don't just multiply  

tip:  
to write energy conservation, $k=\frac{p^2}{2m}$ can be useful  

**Reference system**

- lab  
- cm  (center of mass)

in system of cm:  
- always, total momentum is null (if no external forces upon it) (bc in ref system of cm, the speed of cm is null) 
- only in elastic collision, bodies invert their speed relative to CM (add minus sign)  
 $V_A - V_B = v_B - v_A$  
 can  write it with normal speed, since speed of cm to do the transition will cancel out

**Solution**

$V_A = \frac{(m_A- em_B)v_A + m_B(1+e)v_B}{m_a + m_B}$  

to compute $V_B$ just swap all indices A and B to all terms  

**Solution elastic**

$V_A = \frac{(m_A- m_B)}{m_a + m_B}v_A + \frac{2m_B}{m_a + m_B}v_B$

#### Reaction fulcro/vincolo

**Find vincolar reaction while object is moving**

if the obj is traveling a circular path, the sum of its radial forces is the CENTRIPETAL FORCE  
so write sum of all radial forces:  
$\Sigma F_r = m_{tot}a_{centr}$  

note that direction of the weight change at all points, so modulo of R also changes at all points

**What are the things**

$m_{tot}$: mass of body is sum of mass of all objects that make it (eg. rod + ball hitting it)  

$a _{centr} = \frac{v^2}{r}$, where $r$ is th distance between the fulcro and the <u>cm of the system</u> (rod+ball)  
so compute distance of cm of system made by rod and ball, since centripetal force must be thought to be applied in cm
