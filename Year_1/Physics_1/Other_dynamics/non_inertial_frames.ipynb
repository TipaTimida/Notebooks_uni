### Kinematics

S, S' are the frames, of origin O and O', P is a point  
  
S: still    
S': non-inertial  

**Non-inertial frame**

non-inertiality can be caused by 2 things:  
1. S' rotates around an axis passing through its own origin (with speed $\omega$)  
2. O' translates wrt O, with acceleration $A$

#### Acceleration

**Real and apparent forces**

$\vec{a} = \vec{a}'+ \vec{a}_d + \vec{a}_c$  

so Newton law becomes:  
$\vec{F}_{real} = ma' + \vec{F}_{app}$  
$\vec{F}_{app} = m (a_c + a_d)$  

**Terms**

$\vec{a}$: acc of P in S (from real forces, bc in S, inertial, are only real forces)  

$\vec{a}'$: acc of P in S'  

from apparent forces:   
$a_d$: (local) drift acceleration (di trascinamento)  
$a_c$: coriolis acceleration   

**Components of acceleration**

Coriolis:   
$ a_c = 2 \vec{\omega} \land \vec{v}'$  
appear when $v' \neq 0$  (P moves wrt S')  

Drift:   
$a_d = \vec{\omega} \land (\vec{\omega} \land \vec{r}') + \vec{A}+ \ddot{\theta} \land \vec{r'}$  

- $\vec{\omega} \land (\vec{\omega} \land \vec{r}')$: centrifugal  
  when $\omega \neq 0$  (S' rotates)  

- $\vec{A}$: linear component  
when O and O' accelerate wrt each other  

- $\ddot{\theta} \land \vec{r'}$: derivative component  
 when S' changes angular speed of its rotation, so when there is angular acceleration

#### Theory 

**Formulas**

$\vec{v} = \dot{r}\hat{r} + r\dot{\theta} \hat{\theta}$  
$\vec{a} = (\ddot r - r \dot{\theta}^2) \hat{r} + (r \ddot{\theta} + 2 \dot{r} \dot{\theta})\hat{\theta}$

**Notation**  

radial/tangential acc must give a whole parenthesis:  
$(r \ddot{\theta} + 2 \dot{r} \dot{\theta})$: tangential  
$(\ddot r - r \dot{\theta}^2)$: radial  

$\ddot{\theta}$: angular   
$\ddot r$: not specific name  

$- r \dot{\theta}^2 = - \frac{v^2}{r} \quad$  : centripetal (in polar has -, in intrinsic has +, bc points inwards)      
$ 2 \dot{r} \dot{\theta} \quad \quad$  : Coriolis acceleration


**Initial conditions**

when putting initial conds, if not given, choose not-specified stuff like $x(0)=x_0, \dot{x}(0) = v_0$  
don't put stuff to zero if not told to  

**Solutions for second-order diff eq**

- $\Delta >0: y = Ae^{\lambda _1 x}+ Be^{\lambda _2 x}$
- $\Delta = 0: y = (A+Bx)e^{\lambda x} $
- $\Delta <0: \lambda = \alpha \pm \beta, \quad y = e^{\alpha x}(A \cos(\beta x)+ B \sin(\beta x))$

## Hyperbolic functions

useful when solving diff eq with $\Delta >0$, so solution has exponential  

$\cosh x = \frac{1}{2}(e^x + e^{-x}{2})$  
$\sinh x = \frac{1}{2}(e^x - e^{-x}{2})$  

$\cosh ^2 - \sinh ^2 = 1$  

**For derivating motion laws**  

$\frac{d}{dx} \cosh x = \sinh x$  
$\frac{d}{dx} \sinh x = \cosh x$   

there is not - sign in neither, unlike with normal isn and cos
