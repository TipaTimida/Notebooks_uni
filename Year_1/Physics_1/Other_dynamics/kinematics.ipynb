##### Length of trajectory

integrate between the x coords of the displacement  

$ L = \int \limits_{x \:start}^{x\: end} ds $

$ds = \sqrt{(\frac{dy}{dx})^2 +1 } \quad dx$

$ L = \int \limits_{\theta \:start}^{\theta \: end} ds $

### Common motions

#### Curvilinear motion

**General formulas**

$\vec{v} = \dot{r}\hat{r} + r\dot{\theta} \hat{\theta}$  
$\vec{a} = (\ddot r - r \dot{\theta}^2) \hat{r} + (r \ddot{\theta} + 2 \dot{r} \dot{\theta})\hat{\theta}$

**Notation**  

radial/tangential acc must give a whole parenthesis:  
$(r \ddot{\theta} + 2 \dot{r} \dot{\theta})$: tangential  
$(\ddot r - r \dot{\theta}^2)$: radial  

$\ddot{\theta}$: angular   
$\ddot r$: not specific name  

$r \dot{\theta}^2 = \frac{v^2}{r} \quad$  : centripetal (in polar has -, in intrinsic has +, bc points inwards)      
$ 2 \dot{r} \dot{\theta} \quad \quad$  : Coriolis acceleration


#### Harmonic motion

**Carthesian coords**

$ x= R \: cos \varphi (t)  $  
$ y= R \:sin \varphi (t)  $

$ \dot{x}= -R \dot{\varphi}\: sin \varphi (t)  $  
$ \dot{y}= R \dot{\varphi}\: cos \varphi (t)  $

$ \ddot{x}=  -\dot{\varphi^2}x $  
$ \ddot{y}=  -\dot{\varphi^2}y  $  



<u>formula of acceleration like this are sign of harmonic motion</u>  (there must be - sign)   
in circular motion, $\varphi(t) = \theta, \: \dot{\varphi}=\omega$

#### Curvilinear (spiral) motion
