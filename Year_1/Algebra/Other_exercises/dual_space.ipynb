**Dual map**

#### Functional

- functional: $f: V \rightarrow K \quad$   
a function that outputs a number  (eg. scalar prod, trace, determinant)  

- endomorphism: $f: V \rightarrow V \quad$  
a particular case of transformation, where dominium is the same as codominium     
changes the space, outputs new space with changes  

- transformation: $f: V \rightarrow W \quad$   

**Theorem of representation of Riesa**

if $\varphi$ is not degenere    
every functional can be written as a scalar/hermitian product with some vector  

intuitively: a functional is something that outputs a value, like a scalar prod  
just need to pick the right vector to get the value i want  

**Functional and scalar product (row vector)**

every functional can be expressed a scalar product per a specific vector w:  
$\varphi(v) = <w,v>$  
  
$\varphi$: given functional  
<,>: scalar product  


since a functional can be represented as a scalar product with a specific vector w, can represent the functional as w itself  
in the scalar product in the formula above, w is a row vector (bc scalar product works $a^t b$)  
so a functional can be represented as a row vector  
so applying a functional to a vector v means multiplying such row vector per v    

**How to express functional as row vector w**

i'm given a functional $\varphi$  

1. simply plug in the formula above the given expression for $\varphi$  
2. in the scalar product, put random entries for w: $w = [w_1, w_2 ...]$  
3. see what entries of w i need for the equality to stand  

example:  
$\varphi(v)=2x+3y+4z$  
$<w, v> = w_1 x + w_2 y +w_3 z$  
clearly $w = [2,3,4]$  
so i've the functional as a row vector    

#### Dual space

**Definition**

dual space of a space V: space of all functionals that can be applied to that space   
written $V^*$  

$V^* = \{\varphi: V \rightarrow K  \quad | \varphi   \; linear\}$

**Properties**   

$dim(V)=dim(V^*)$  
since have same dimension they're isomorph  

$V$ and $V^{**}$ are isomorph, but also the same thing  

**Examples**

$\varphi(x,y,z)=2x+3y+4z$ is a member of $V^*$, the dual space of $V=R^3$  
bc it's a functional defined in that space  

#### Dual base

each base B has an associated dual base $B^*$  
its elements can be written either $v_i^*$ or $\varphi_i$  

$B = \{v_1 , v_2 , v_3 \}$  
$B^∗ = \{v^∗_1 , v^∗_2 , v^∗_3 \} = \{\varphi _1, \varphi _2, \varphi _3\} $    

**Important formulas**

any vector can be written as linear comb of normal base: $v = \Sigma a_i v_i$  
$v$: random vectors, combination of others in base  
$v_i$: a vector in base B   

any functional can be written as linear comb of dual base:  $\varphi = \Sigma \alpha_i \varphi _i$  
$\varphi$: random functional, combination of others in base  
$\varphi _i$: a functional in base $B^*$     

to finds coords of random vector:  
$\varphi _i (v) =a_i$ 

to find coords of random functional:  
$\varphi(v_i)=\alpha _i$  

so to find coords of a random something (funtional or vector), iterate on the coords of the other thing  
the results are the coords i'm looking for  

**What are functionals for**

the useful property is: $\varphi _i (v) =a_i$    
functionals are used to extract coordinates of vectors wrt base B (pick out coeffs of its normal linear combination)   
where B is the base the dual base is built from  
just evaluate one of the function of the base, $\varphi_i$, on a vector $x$, it will tell me the coeff $v_i$  

**Compute dual base**

consider   
$B$: matrix where each col is a vector of the base    
$\Phi$: matrix where each row vector is a functional belonging to the dual basis (it's $B^*$)  

$\Phi \, B = I$  
since: $\varphi _i(v_j) = I_{ij}$  

so just: $\Phi = B^{-1}$  
rem: functionals are the ROWS, not the cols 

**Explanation: compute dual base**

$\varphi _i (v) =a_i$ 

so if plug in vector from base $B$:    
$\varphi _i (v_j) =1 $ or $0$, depends if indexes i,j are the same or different  
bc the linear comb has one 1 and all the rest 0, like $v_1 = 1 \, v_1 + 0 \, v_2 $
 
$\varphi _1 (v_1) = 1 \quad \varphi _1 (v_2) = 0$  
$\varphi _2 (v_1) = 0 \quad \varphi _2 (v_2) = 1$  
thus:  $\quad \varphi _i(v_j) = I_{ij}$  

**Explanation: extracting coords of a functional**

explanation of: $\varphi(v_i)=\varphi _i$

$\varphi _i (v_j) =1 $ or $0$, depends if indexes i,j are the same or different  
thus only the corrsponding one is left:  
$\varphi(v_i) = ( \alpha_1 \varphi_1 + \alpha_2  \varphi_2 + \alpha_3  \varphi_3)(v_i) =  \alpha_1\varphi_1 (v_i) + \alpha_2  \varphi_2(v_i)  + \alpha_3  \varphi_3 (v_i)  = \alpha _i$  

so to find all coeffs, just plug in all vectors from base B

#### Dual map / dual function

$f^* = f^t$

- f: linear function    
map vectors from a space to another  

- $f^*$: dual function  
map FUNCTIONALS from a space to another  
can also be written $F^*(\varphi)$  

it's called functional map bc it's used to get a functional in V, that replicates a functional in W  
so maps functional in another space  

**Intuitively**

$f^*$ is used when want to apply:  
first a linear transformation from V to W  
then a functional in W  
bc $f^*$ summarizes those two steps into a single functional, that gets applied directly in V  

formally it's:  
$f^*(φ)(v)=φ(f(v))$  
left:  the dual function  
right: applying a linear function, then a functional  

example:  
old linear function: $f(x,y) = (2x,2y)$  (double the vector)  
old functional: $\varphi(x,y) = 3x + 4y$  

new functional: $f^*(\varphi) = 6x+8y$  

the new functional is applied to V directly, no linear transformation is done anymore  
for each vector in V, gives same result i would get from applying both old linear function and old functional
