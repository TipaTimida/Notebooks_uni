## Affine matrix

### Definition

**Definition**

space is translated, so origin is not in origin  
$A = \vec{v} + W $  

$W = span\{w_1, w_2 ...\}$  


A: the affine space  
W: the non-affine space PARALLEL to A  
$\vec{v}$: origin of A (where the origin of W gets translated to)   

carth eq is:  
$ax+by+cz = d$  

in non-affine, d=0  

### Exercises

**To get carth eq of A**

- if have carth eq for W and a vector $v_A$, a random belonging to A, not necessarily the origin:      

in each carth eq of W, instead of zero on right side, write a number (different for each carth eq)   
to find this number plug coords $v_A$ in carth eq of W, pick number so that eq is satisfied    
eg. $ax+by+cz = 0 \quad \longrightarrow \quad ax_v + by_v + cz_v = d$  

- if have base of W and $\vec{v}$ (the origin of A):   

in last col (with general vars $x_1, x_2 ...$) instead of just the var, must subtract coord of its origin  
last col becomes $\vec{x}-\vec{v}$, so first entry is $x_1- v_1$  
last col is: $\begin{bmatrix} x_1-v_1 \\ x_2-v_2 \end{bmatrix}$  

or simply, get chart eq of W, then apply method from before  

**If have 3 points in affine plane**  

if have 3 points B,C,D which lie on affine plane:  

find 2 vectors that belong to W, to have base for W:  
$w_1 = B-D \quad w_2 = C-D$  
(always subtract the same one, it will be $\vec{v}$)  
find carth eq of $W = span(w_1, w_2)$  
then transform it into carth eq of $A$  

**Distance between parallel planes (affine planes)**

have carth eq of both planes, $\pi _1$ and $\pi _2$ (the second is affine)   
they're parallel, so carth eq have same coeffs and different termine noto  

1. pick 2 vectors:  
$v_n$: orthogonal to both planes (just pick coeffs of carth eq)  
$v_0$: any point on $\pi _1$ (just pick random x,y,z to satisfy its carth eq, write point in coords like a vector)  

2. write line passing through $v_0$ and perpendicular to the planes  
$r = v_0 + \lambda v_n$  
$v_0$ and $v_n$ are vectors, so will get single vector with $\lambda$ in it  

3. find point where the line touches $\pi _2$  
simply plug in coords of $r$ in the carth eq of $\pi _2$, will get equation in $\lambda$  
solve eq, find $\lambda$  

4. distance is $\lambda ||v_n||$  
$||v_n||$ is the ROOT of the norm of the vector $v$, the scalar prod with itself, find it with Pytagora


if want distance from origin use similar method, but plane $\pi _1$ passes through origin, so makes sense to pick $v_0 = (0,0,0)$  

**Orthogonal projection of point Q on plane $\pi$**  

need coords of point Q and carth eq of plane $\pi$  
call H the projection point  
H and Q are both vectors and points, coords can be interpreted in same way, so $Q = v_Q$  

1. pick vector:  
$v_n$: perpendicular to plane (use coeffs of carth eq)  

2. write line passing through Q and peroendicular to plane:  
$r = v_Q + span(v_n) = v_Q + \lambda v_n$  

3. find coords of the projection, where projection of Q touches the plane:  
plug coords of $r$ in carth eq of plane, find $\lambda$  
after finding $\lambda _H$, can write projection point H as:  
$v_H = v_Q + \lambda _H v_n$  

4. distance between Q and $\pi$, so between Q and its projection on $\pi$, H:   
$||v_H - v_Q|| = ||\lambda _H v_n||$  
the norm, like in prev exercise   

eg: if $Q = (1,2,3)$ and $v_n = (4,5,6)$, retta of step 2 is $r: \; (1 + 4t, 2+t, 3+6t)$  

it's similar to ex to find distance between 2 planes  
just here $v_0$ is $v_Q$, so i'm not given 2 planes, since one plane was just to pick $v_0$  

**Recap exercise: distance between 2 things**

1. pick vectors  
$v_n$: perpendicular to plane (use coeffs of cart eq) (if have 2 planes, they're parallel, so no prob)  
$v_0$: a point on the non-affine plane or the point $v_Q$ given to me (if want distance from origin, pick $v_0=0$  )

2. write line parallel to affine plane, passing through point:  
$r = v_0 + \lambda v_n$  
$v_0$ and $v_n$ are vectors, so will get single vector with $\lambda$ in it  

3. find coord of point where line touches the affine plane, call it H (coords of prejection of $v_0$)  
by plugging coords of r in eq of affine plane, find $\lambda$  

4. compute distance  
geometrical distance: $v_H = v_0 + \lambda v_n$  
modulo of distance: $||v_H - v_0|| = ||\lambda _H v_n||$  
$||v_n||$ is the ROOT of the norm of the vector $v$, the scalar prod with itself, find it with Pytagora

5. coord of orthogonal projection  
compute coords of $v_H$ plugging in $\lambda$

## Orthogonal projection

### Definition

if $V = W \oplus W^{\perp}$  
can split any vector: $v = w + u \quad w \in W, u \in W^\perp$

$pr_W(\vec{v}) = \vec{w}$  
can project a vector of V on W   
will get component on W closest to original vector  

so if want $u$, projection on $W^\perp$:  
can find projection on W and subtract it from original vector, bc $v = w +u$  

### Computation

**Methods**

want to project on W  

if have base of W:  
SU, of projections on all vectors in base of W    

if have base of $W^\perp$:  
(or carth eq of W and dim=2, which allows me to find a base for $W^\perp$)  
from vector, DELETE projection on all vectors in base of $W^\perp$  

**Method 1: have base of W**

1. find base of W  
2. orthogonalize it with G.S.
3. formula  
$pr_W(\vec{v}) = \Sigma \frac{<v, w_i>}{<w_i, w_i>} w_i$   

where the $w_i$ are the vectors in an orthogonal base of W

**Method 2: have carth eq for W**

can only be done if W is a plane, so dim 2, so has ONLY 1 carth eq  
if it's bigger no, need other method  

1. find vector $w$, perpendicular to W (coeffs of carth eq)  
2. formula: $pr_W(v) = v- \frac{<v,w>}{<w,w>}w$  


**Write matrix describing it**

since $V = W + W^\perp$, can make base for $V$ by uniting bases for those    
then vectors are mapped:  

- vectors from base of $W$: mapped to themselves  
eg. (1,0,0), (0,1,0) 

- vectors from base of $W^\perp$: mapped to zero vector   
bc graphically, are orthogonal to plane where i'm doing the projection  




### Write matrix

**Method 1: pick good base**

1. pick good base (1 vector orthogonal, 2 on plane)  
2. write transformation matrix in that base  
the orthogonal one mapped to 0 in projection or the opposite of itself (-1) in reflection  
those on plane mapped on themselves ($\lambda = 1$)     
3. write matrix for base change between good base and the one i want (usually canonical)  
4. do base change   

**Method 2: brute-force**
 
1. pick w, a vector orthogonal to the plane on which i'm projecting (just pick coeffs from eq of plane)   
2. for each vector $v_i$ in base I want to wite the transformation in, compute:  
projection: $pr_W(v) = v- \frac{<v,w>}{<w,w>}w$  
reflection:   $pr_W(v) = v- 2\frac{<v,w>}{<w,w>}w$  (bc need to reverse it the orthogonal part, not delete it)  
so in canonic base, $v_i = e_i$  
3. write the vectors i obtained as cols, that will be the matrix i want  
(attention to the order in which they appear in the base)

**Tips**

can check some stuff by eye  

if a vector in the base i want to write the transformation in:  
-  respect the carth eq of plane, meaning it belongs to the plane, it's mapped into itself  

- can be written as $v = a +b$ where a belongs to the plane and w is orthogonal, its projection is a  

**To prove something is a projection on a plane**

check that:  
- something on plane stays the same  
if v is on plane, $\varphi(v) = v$  

- something perpendicular goes to null vector  
if v orthogonal plane (made of its coeffs), $\varphi(v)=0$
