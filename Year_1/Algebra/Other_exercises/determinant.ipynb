### Definition

**Intuitive**

algebric: it's the **product of the eigenvalues**  

geometric: scaling of the area/volume  
det(A)=0 when A is not full rank, so when space is squashed in a lower dimension  



**Formal**

$\varphi : R^{ n \times n} \rightarrow R$  

a functional such that:  
- multilinear (linear in all components, components are the rows of the matrix)  
- alternants: swapping two components changes sign, not modulo  
- normalized as $\varphi(I) = 1$  

**Geometrical**

have matrix $R^{nxn}$  
every of the n rows of the matrix has the coords for a point in space $R^n$  
plot all points and create a solid connecting them  

the determinant is:  
- its volume/area (depends on dimension of space), if using space matrix  
- how much the area is scaled, if using transformation matrix    

can only compute the determinant on a square TRANSFORMATION matrix  

### Computation

order to try:  

1. check if special form (block, vandermonde, 2x2)  
2. check if row/col with many zeros -> sviluppo di Laplace  
3. if 3x3, Sarrus  
4. row reduction

can be useful to transpose matrix since det don't change  
since i'm not good at col-reduction, this way can do it as if it was row-reduction  

**Unsure cases, big and complex**  

if has many zeroes, can try: 
- permutation of both rows and cols, try to get block matrix  
- Laplace  

if has not zeroes, but looks like some kinda pattern:  
- row and col-reduction  
until get diagonal matrix, or simply easier form to use Lagrange  

**Notice det=0**

just looking at a matrix, det=0 if:  
- one row is a linear comb of other rows  
- after row-reducing there is a zero row  
- one col is a linear comb of other cols  

#### Row-reduction

**Can i do it?**

row-reduction CAN be used to compute determinant  
reduce until diagonal, then multiply all stuff on diagonal, it's correct    
but VALUES ON DIAGONAL obtained like this are NOT THE EIGENVALUES  

**Stuff to pay attention to**

when doing Gaussian transformation  

- adding a multiple of a row to another: no problem  

- row or col swaps: each time multiply the det by -1 (Change sign of det)  
in other words, count row swaps done, if they're odd at the end multiply det by one  

- multiplying/dividing a row by a scalar: bring that scalar out of the determinant, like raccogliere  

**Multiplying/dividing a row by a value**

if a row is all multiple of something, can bring it out, before the determinant, and MULTIPLY IT to the determinant    
essentially posso raccogliere un valore da una riga e metterlo fuori dal determinante  
so multiplying a row by a coeff means multiplying the determinant per that coeff  

eg: dividing a row  
$det \begin{bmatrix} 2 & 2 \\ 1 & 0 \end{bmatrix} = 2 \; det \begin{bmatrix} 1 & 1 \\ 1 & 0 \end{bmatrix}$  

posso raccogliere anche un valore da una colonna, non solo da una riga  

**Tip to do it**

to get 2 or more zeroes in same col and then Laplace:  

method 1:  
pick a row with an easy number in some entry, like 1  
use it to delete all other entries in that col with row-reduction, by subtracting a multiple of it from each col  
eg, to delete a $-\lambda$, must multiply such row for $\lambda$ (temporarily) and add them   

method 2:  
do row-reduction AND col-reduction  
generally start by adding cols, then add rows  
do it to get matrix with two 0 in a col or row  
both rows and col swaps count for the number of swaps  

#### 3x3: Sarrus rule

- copy first two cols  
- diagonals pointing right - (minus) diagonals pointing left  

#### Sviluppo di Laplace

**Picking sign**

index rows starting from 1  

if pick odd row/col, start from +  
if pick even row/col, start from -  

formally, multiply by $(-1)^{i+j}$  


**Easy explanation**

before, often do row-reduction to get many 0, so it's more useful  

1. pick a row/col  
2. for each entry, find det of matrix deleting that row and col, and multiply entry and det      
3. add all the values gotten, alternating the sign (start from + and must go in order)  

when skipping entries bc an entry is zero, rem that the sign is alternating on that entry as well  

**Formal explanation**

1. pick a row/col  
possibly one with many zeroes  

2. for each entry in that row/col get an addend by multiplying:    
    - the entry itself
    - the sign decided by the coords of the entry  
    - the determinant of the submatrix obtained deleting that row and col  

3. add all addends

$det (A) = \Sigma a_{ij} (-1)^{i+j} det(A_{ij})$

$(-1)^{i+j} det(A_{ij})$ is the algebric complement of the entry  

#### Triangular block matrix

**Triangular/diagonal matrix (block or not)**  

det is product of entries on main digonal (since they're already the eigenvalues)  

triangular is enough  
bc if have triangular inferior (all stuff over diagonal is 0), can compute determinant using Lagrange, starting from first row then going down, and will get product of all values on main diag  
just try and see  

$det(A) = \prod A_i$  

bc row-reducing one of the blocks don't intefere with the other blocks   
only need a triangular block matrix, not diagonal  

$ \begin{bmatrix}
A_1 & A_2 \\
0 & A_3 
\end{bmatrix} $   



the blocks can be of different sizes, like:  

$ \begin{bmatrix}
x & 0 & 0 \\
0 & x & x \\
0 & x & x 
\end{bmatrix} $   

#### Vandermonde matrix

**Vandermonde matrix**

for a square matrix nxn, where each row contains progressive powers, from power 0, so entry 1, until power n-1  

$ V= \begin{bmatrix}
1 & 2 & 2^2 \\
1 & 3 & 3^2 \\ 
1 & 5 & 5^2
\end{bmatrix}$    

MIGHT ALSO BE TRANSPOSED, where start with a row with all 1s instead of a col with all 1s (which is obv, bc $det(A)=det(A^t)$)     
to compute det use same method, use the bases with no exponent, which will be the second row  

**Compute det**

$det(V) = \prod _{i<j} (x_i -x_j)$   
for each pair of bases, take the difference with the positive sign  

here it's:  
$det(V) = (5-3)(5-2)(3-2)$  


**Hard rows to notice**

[1,0,0] is vandermonde row with base 0  
[1,1,1] is vandermonde row with base 1  

### Other exercises

#### Operations on matrices

**Multiplication (Binet th)**  

$det(AB) = det(BA) = det(A)det(B)$  

corollaries of binet theorem used in proofs:  

$det(A^k) = (det(A))^k$  
$det(\lambda A^3) = det(\lambda I) \, det(A^3) = \lambda^n \, (det(A))^3$  
where A is size nxn  

**Others**

- transposition  

$det (A^t) = det(A)$  
useful for proofs with symmetric and anti-symmetric matrices

- inverse  

$det(A^{-1}) = \frac{1}{det(A)}$  
consequence of Binet th  

- similitude  

$A = P^{-1}DP$, where D is diagonal  
$det(A) = det(D)$  
bc determinant is invariant for similitude  


**Explanation: take out scalar**

$det(2I) = 2^n$ if $I$ is nxn    
bc 2I means multiplying every row by 2, and multiplying a row multiplies the det as well  
so multiply per 2 once for each row  

useful example:  
$det(-M) = (-1)^n det(M)$  

**Observation with multiplication**  

if A is 6x2 and B is 2x6:  

AB is 6x6: definitely not full rank, so det=0  
bc its rank must be smaller than both rank of A and B (bc of rank inequality, multiply matrices can only lower the rank/dimensions, not add more)    
since A is at most rank 2, AB is at most rank 2  
since AB is 6x6, it's definitely not full rank  

BA is 2x2: could be full rank, must actually compute  

#### Proofs

determinant is a functional (in proofs can write $\varphi$ inside of det)  

can write the determinant as taking in input each row instead of the whole matrix:  
$det(A) = det(A_1, .. A_i, ... A_j,...A_n)$  

useful when doing operations on rows, remember that it's multilinear:  
$det(A_1, ... \, , \lambda A_i,...A_n) = \lambda \, det (A) \quad$  (bc of linearity)    
$det(A_1, A_i,.. \lambda A_i,...A_n) = 0 \quad$  (bc repeat a line)  

can combine them:    
$det(A_1, .. A_i + \lambda A_j,...A_n) = det(A_1, ..., A_i,...A_n)+  det(A_1, ... \lambda A_j,...A_n) = 0 + \lambda det(A)$

#### Permutations

how to compute permutations in a number list:  

for each enrty/index i, count $n_i$, the number of entries AFTER IT smaller than it (don't care about stuff before, or will count stuff twice)  
then add all $n_i$, that is number of permutations
