### Autoaggiunta

#### Symmetric and anti-symmetric

**Definition**

- symmetric: $A = A^t$  
symmmetry to main diagonal

- anti-symmetric: $A = -A^t$  
symmetry to main diagonal but opposite sign  
main diagonal is zero, bc must be equal to itself with opposite sign

**Decomposition**

direct sum of vector spaces of symmetric and anti-symmetric matrices nxn gives whole matrix space nxn  
so every matrix can be decomposed in sum of symmetric and antisymmetric one  

**Properties anti-symmetric**

if M is anti-symmetric and has only real entries:   
$iM$ is autoaggiunta/HERMITIAN  
(with the meaning in the C field, so $(iM)^* = (\overline{iM})^t$)  

by using spectral th on $iM$ i can conclude:  

- iM has all real eigenvalues  
M has either pure imaginary or null eigenvalues, cannot have real part  

- M is always diagonalizable on C  
bc $iM$ is hermitian, so diagonalizable bc of spectral th, and M is simply $iM$ scaled  

- M is never diagonalizable on R  
bc $iM$ has real eigenvalues (since it's autoagguinta), so $M$ has complex eigenvalues  
to be diagonalizable would need all eigenvalues 0, but then would be a null matrix  

**Proof decomposition symmetric antisymmetric**

S: symmetric  
A: antisymmetric  
$M = S + A$


- method 1  

$ S = \frac{M + M^T}{2}$  

$ A = \frac{M - M^T}{2}$


check S is symmetric by seeing that $S^T = S$ and check A is antisymmetric



- method 2: dimension formula    

intuitive, see that all entries are controlled by something

obv only matrix both symmetryc and anti is null matrix, so  
$dim(S \cap A) = 0$  
therefore  
$dim(S + A) = dim(S) + dim(A) = n$    
dimensions of symmetric and anti matrices are well known, just draw them to see dimensions:  
symmetric: $\frac{n(n+1)}{2}$  
anti-symmetric: $\frac{n(n-1)}{2}$

**Proof iM is hermitian**

M has real entries $\overline{M}=M$  
M is anti-symmetric: $M^t = -M$  

$(iM)^* = \overline{(iM)}^t = \overline{i} \; \overline{M}^t = -i \; M^t = iM$

### Adjoint

**Definition**

$T^*$ is the adjoint of $T$ if:   
$<Tv, w> = <v, T^*w>$  

fields:  
in R: $T^*=T^t$  
in C: $T^* = \overline{T}^t$

thus, adjoint is an extension of transposed, that has some properties even in C  

**Different scalar products**

$T^*$ is adjoint wrt $\varphi$ means: $\varphi(Tx,y)= \varphi(x, T^*y)$  
$T^*$ is adjoint wrt canonic product means: $<Tx,y>= <x, T^*y>$  

if want $T^*A$, but have weird scalar prod, maybe with trace, do:  
$<T^*A, B> = <A,TB>=tr(...) = ... = <known, B>$  
so $T^*A = known$

note that $<Tv, w> = <v, T^*w>$  
can also be written $<T^*v, w> = <v, Tw>$  
to prove, swap v and w, then apply the conjugate symmetry property of the scalar product $\varphi(x,y) = \overline{\varphi(y,x)}$


**Properties**

extension of propertise of transposed matrices  

$(A+B)^* = A^* + B^*$  
$(\lambda A)^* = \overline{\lambda} A^*$  
$(A^*)^* = A$  
$(AB)^* = B^* A^*$  
$I^* = I$  
if A invertible, $A^*$ in vertible and $(A^*)^{-1} = (A^{-1})^*$
