### Inverse

**Condition**

condition: full rank  

can be proved by seeing that has a non-0 determinant  

**Properties**

$(AB)^{-1} = B^{-1} A^{-1}$ 

since to be invertible, A must be full rank:  
- A has kernel of dim 0  
- rank(AB) = rank(B)  
(bc full rank don't change dimensionality)  


**How to remember $(AB)^{-1}$**

write the definition of inverse:  
$(AB) (AB)^{-1} = I$  

clearly i can pick as a solution $(AB)^{-1} = B^{-1} A^{-1}$ bc:  

$A(BB^{-1})A^{-1} = A I A^{-1} = I$

### Diagonal / triangular

**Conditions**

- diagonal: all eigenvaues belong in K and $m_g=m_a$  
- triangular: all eigenvalues belong in K  

**Properties**

commutative: $AB = BA$  

adding or multiplying diagonal matrices give another diagonal matrix  
it's done by adding/multiplying the respective entries on the main diagonal  

**Bases**

matrix written in base $B$  

is ... only if $f(W_i) \subseteq W_i$ (subspaces are invariant)  
what are these subspaces:  

upper triangular: $W_i= span\{b_1, ... b_i\}$ for all $i=1, ... n$  
lower triangular: $W_i= span\{b_i, b_{i+1}, ... b_n\} $ for all $i=1, ... n$  
diagonal: each column (bc each col is an eigenvector)  

in words, means that the image of a vector $e_j$, with $j\le i$, is a linear combination only of vectors $v_1, ... v_i$  
and all the other entries are 0  

another ways of something upper triangular matrix is saying that it has a base a bandiera

**Diagonalizable**

M is a transformation matrix  

1. if $M$ is diagonalizable, $M^n$ is too, for any n  
if $M$ is not diagonalizable, $M^2$ still might be  

2. if $M^2$ is diagonalizable, can't tell anything on $M$, might or not    
if $M^2$ is not diagonalizable, $M$ surely isn't, bc that would implicate $M^2$ being diagonalizable  

3. if $M^n = \lambda I$, M is diagonalizable (ONLY IN FIELD C)  
(all its powers too, bc of lemma 1)   
means if repeating many times it becomes a multiple of the identity  

4. if M is invertible, M is diagonalizable if and only if $M^2$ is diagonalizable   
on field C, bc eigenvalues on M might end up being complex  

**Proof**  

if $M^n = \lambda I$, M is diagonalizable (ONLY IN FIELD C) 

assume $\lambda = 1$, only scales the matrix anyway  
$M^n = I$   
all eigenvalues of M are roots of 1  

minimum poly is:  
$p(M) = (M^n-I)=0$  

$x^n-1$ has distinct roots in C, all with multiplicity 1, so M is diagonalizable  


**Eigenvalues**

if M is diagonalizable with eigenvalues $\lambda_i$, $M^n$ has eigenvalues $\lambda _i ^n$  
each with algebric multiplicity equal to that of their corresponding one in M  
obv, think of a diagonal matrix, raising to a power raise the main diag to that power  

eg: if M has eigenvalues 1,3  
$M^2$ has eigenvalues 1, 9 (the squares)  

to do the opposite:  
if know eigenvalues of $M^n$, each eigenvalue has a corresponding one in M (IF M is diagonalizable, of which i'm not sure just from this)     
the correspoding one in M is a root of the one in $M^n$, but can't tell which root  
an eigenvalue with algebric multiplicity more than 1 in $M^n$ might correspond to different roots   
eg. $M^2$ might have $\lambda =4$ with multiplicity 2, M might have $\lambda = +2$ and $\lambda = -2$, both with multiplicity 1

**Eigenvector**

if v is eigenvector of M it's also eigenvector of $M^n$, bc each time i apply M, v stays in the same eigenspace, so keep being an eigenvector  
this is bc eigenvectors are M-invariant

### Nilpotent

**Definition**
 
$f^k = 0$  

there is a k for which it becomes identically null (so a zero matrix)  
obv must be an endomorphism, so a square matrix   

**Properties**

- its only eigenvalues is 0  
thus can be written as a block matrix where each block is a jordan block with $\lambda =0$  

- can find a base for it written:  
$v_1, ... v_i, f(v_1), ... f(v_i), f^2(v_1), ... f^2(v_i)...$, until some power (which power???)  

**General base**

for a general matrix, not nilpotent, can find a base for each block by using the genralized eigenvectors:  
$v, (A-\lambda I)v, (A-\lambda I)^2 v...$  
do it for all aigenspaces, changing $\lambda$ to be its eigenvalue  
in nilponent, only eigenvalue is $\lambda =0$
