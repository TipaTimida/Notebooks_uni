#### Chaining rotation and reflection

**Squaring stuff**  

reflection: $\pi ^2 = I$   
projection: $\pi ^2 = \pi$   
rotation: if $T$ has angle $\theta$, $T^2$ has angle $2 \theta$  

rem: $\cos(2 \theta) = 2\cos(\theta)^2-1$  

**Group**

rotations form a group  
so composition of rotations is a rotation: $R_1R_2 = R$   
inverse of rotation is rotation: $R^{-1} = R$  
bc a group must have the inverse, and it's still part of a group

**Consecutive rotations**

if have 2x2 rotation matrices, which describe rotations of angle $\alpha$ and $\beta$, their multiplication is a matrix describing a rotation of angle $\alpha + \beta$  
bc 2x2 rotation matrices are isomorph to complex nubers of modulo 1, where this is true ($SO(2)\sim S^1$)  

in 3x3 matrices this is true only if rotations are wrt the same axis, else there isn't any quick way to compute the total angle    

**Changing angle**

rotations of $+\theta$ and $-\theta$ have same eigenvalues (thus same characteristic poly and trace)  
but the eigenvalues are in a different order, so the matrices are different

### Common forms

**Types of rotation matrix**  

3d rotations can act on a 2d plane or in 3d space  

if act on a 2d plane:  
one of the vectors is mapped on the itself  
the other 2 swap   
$\begin{bmatrix}
1 & 0 & 0 \\  
0 & 0 & -1 \\  
0 & 1 & 0 \\  
\end{bmatrix}$  

if act on 3d plane:  
all vectors swap, and whole plane is rotated  
$\begin{bmatrix}
0 & 0 & 1 \\  
1 & 0 & 0 \\  
0 & 1 & 0 \\  
\end{bmatrix}$  
this is called cyclic operator  
this preserve only a vector, the axis  
so here, and not in the other case of rotation in 2d space, can find a base for $R^3$ of the form $v, Av, A^2v$  

all other properties apply to both cases

**Common form rotation**

exists a base wrt which a rotation matrix can be expressed as:  
$\begin{bmatrix}  
1 & 0 & 0 \\
0 & \cos \theta & - \sin \theta \\
0 & \sin \theta & \cos \theta \end{bmatrix}$  

if it's 2x2, just pick the bottom right 2x2 part
