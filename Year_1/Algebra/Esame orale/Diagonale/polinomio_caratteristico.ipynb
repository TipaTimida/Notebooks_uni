### Polinomio caratteristico: teoria

polinomio caratteristico di A (una matrice) 

$p_A(x)=det(A-xI)$  

polinomio caratteristico di un prodotto scalare: $p_A$, dove A è una qualsiasi matrice che rappresenta $\varphi$

cosa vuol dire $p_A(A)$  
se lo faccio su una matrice diagonale, cioè $p_D(D)$, significa sostituire ogni elemento della diagonale col valore che ottengo pluggandolo nel polinomio, cioè: $x \mapsto p_D(x)$   

note per scriverlo:  

o si scrive con det  
oppure, assumo di aver già fatto il det, deve esserci un esponente alla x   
perchè qualcosa deve creare il polinomio, il det mi fa fare una moltiplicazione, quindi crea un polinomio  

esempio:  
$p(x)=det(A-Ix)$  
$p(x)=(\lambda-x)^n$ or $p(x)=(\lambda_1-x)^{\alpha_1}...(\lambda_k -x)^{\alpha_k}$

un polinomio p annulla A vuol dire che:  
$p(A)=0$  

metto A al posto di x, devo ottenere matrice nulla  

### Cayley-Hamilton

cose da fare prima di provarlo:  

1. esiste un polinomio non banale che annulla A   
2. come usare polinomio su matrice diagonale a blocchi  
3. se annulla A, annulla tutte le matrici simili ad A  

steps per provarlo:  
1. A è un blocco di Jordan  
2. A è un'intera matrice di Jordan  
3. per ognuno di questi casi, trova il polinomio caratteristico e vedi che succede se pluggo tutti i blocchi    
4. conclusioni  


**Esiste un polinomio non banale che annulla A**

ho uno spazio di dimensione $n^2$ (matrici quadrate di lato n)    
prendo una matrice A in esso  

considero le matrici: $I, A, A^2... A^{n^2}$  
sono $n^2+1$, lo spazio è di dimensione $n^2$  
quindi non possono essere tutti indipendenti, cioè la condizione di indipendenza ha almeno una soluzione non banale  
$\Sigma a_iA^i =0$ ha una soluzione non banale  
quindi posso usare quegli $a_i$ per creare un polinomio che si annulla in A, $p(x)=a_0+a_1x+...$

**Un polinomio annulla tutte le matrici simili**

voglio dimostrare: se $p(A)=0$, allora $p(J)=0$ se A,J sono simili  

A e B sono simili: $A=MBM^{-1}$   
ho p(x)  
dimostra che o p(x) o le annulla entrambe, o nessuna dei due  

ricorda che: $A^k = (MBM^{-1})^k= MB^kM^{-1}$  
evidente se lo faccio: $A^2 = (MBM^{-1})(MBM^{-1})=MBM^{-1}$ perchè quelli centrali si cancellano    

scrivo p di una matrice, la sostituisco con la sua espressione  
$p(A)=a_m(MBM^{-1})^m + ... + a_1(MBM^{-1})+a_0I = M(a_mB)M^{-1} +...+ M(a_1B)M^{-1}+M(a_0I)M^{-1} = Mp(B)M^{-1}$  
quindi se su uno dei due si annulla, si annulla anche sull'altro  

**Char poly è invariante per similitudine**

voglio dimostrare: $p_A(x)=p_J(x)$ se A,J simili  

poly: $p(t)=det(A-tI)$  
A,B sono simili: $ A= P^{-1}BP$     
ricorda che I è simile solo a sè stesso, resta I in ogni base  
quindi posso fare $I = P^{-1}IP$  

scrivo p di una matrice, la sostituisco con la sua espressione  
$det(A-tI)=det(P^{-1}BP-tI) = det(P^{-1}BP-tP^{-1}IP)=det(P^{-1}(B-tI)P) = det(P^{-1})det(B-tI)det(P)$

**Cose utili per proof**

1-- ricorda che I è simile solo a sè stesso, resta I in ogni base  
quindi posso fare $I = P^{-1}IP$  

2-- ricorda che: $A^k = (MBM^{-1})^k= MB^kM^{-1}$  
evidente se lo faccio: $A^2 = (MBM^{-1})(MBM^{-1})=MBM^{-1}$ perchè quelli centrali si cancellano    

**Cayley-Hamilton th**

cioè: il polinomio caratteristico $p_A(\lambda)$ annulla A, $p_A(A)=0$  

1-- A È UN SINGOLO BLOCCO DI JORDAN  
assumo $A=B_{\lambda_0, n}$, cioè è un blocco di jordan di autovalore $\lambda_0$  

char poly: $p_A(\lambda)=(\lambda_0-\lambda)^n$   
perchè è triangolare  

pluggo A:  
$p_A(A)= (\lambda_0I-A)^n = \pm (A-\lambda_0I)^n =  \pm (B_{\lambda_0}-\lambda_0I)^n = \pm B_{0,n}^n = 0$  
perchè $ B_{0,n}$ è nilpotente di indice n  

quindi il teorema di Cayley è vero sui blocchi di Jordan  


2-- È È UN'INTERA MATRICE DI JORDAN   
ora considero che A sia un'intera matrice di Jordan, con blocchi $B_i$  

char poly: $p_A(x)=p_{B_1}(x) ... p_{B_k}(x)$  
poichè è diagonale a blocchi

pluggo tutti i blocchi (voglio che annulli ogni blocco):  
abbiamo appena provato che $p_{B_i}(B_i)=0$, quindi anche $p_A(B_i)=0$ (visto che, come si vede alla riga sopra, ha la cosa che si annulla come fattore)  
quindi anche qui è verificato: se $p_A$ si annulla in tutti i blocchi che compongono A, si annulla su tutta A  

3--CONCLUSIONE 

una matrice A è sempre simile ad una matrice di jordan  
e ho provato prima che, se A e B sono simili, p o annulla entrambe o nessuna  
quindi dimostrato  

ho dimostrato $p_J(J)=0$  
so che $p_J(J)=Mp_A(A)M^{-1}$, quindi anche il secondo è 0  


### Proprietà polinomio su matrici triangolari

**Corollario: polinomio caratteristico su matrice diagonale**

A triangolare: $A=\lambda I$, di dimensione n  

$p_A(x)=det(A-x I) = det((\lambda -x)I) = (\lambda-x)^n det(I)= (\lambda-x)^n$  

ricorda che per portare fuori dal det un coefficiente, devo elevarlo alla dimensione della matrice  
una più generale, per tutte le triangolari, è al paragrafo sui blocchi di Jordan  

**Corollario: polinomio caratteristico su matrice triangolare a blocchi**

$A = \begin{bmatrix} B & C \\ 0 & D \end{bmatrix}$  
dimostra che vale: $p_A(x)=p_B(x)p_D(x)$  

$p_A(x)=det(A-xI)=det(\begin{bmatrix} B-xI & C \\ 0 & D-xI \end{bmatrix}) = det(B-xI)det(D-xI)= p_B(x)p_D(x)$   
perchè è il determinate di una matrice diagonale  

**Usare polinomio su matrice diagonale a blocchi**

A è matrice diagonale a blocchi: $A= \begin{bmatrix} B_1 & 0 \\ 0 & B_2 \end{bmatrix}$  
generico polinomio: $p(x)=a_mx^m + ... + a_1x+a_0$  

plug it nel polinomio p(x), usa algebra, vedi cosa ti da

moltiplicare una matrice per un coefficiente significa moltiplicare ogni entry per quel coefficinte, quindi fallo e poi somma tutto, ottengo:  

$p(A) = a_m \begin{bmatrix} B_1 & 0 \\ 0 & B_2 \end{bmatrix}^m + ... + a_1  \begin{bmatrix} B_1 & 0 \\ 0 & B_2 \end{bmatrix} + a_0  \begin{bmatrix} I & 0 \\ 0 & I \end{bmatrix} = \begin{bmatrix} a_m B_1^m + .. + a_1B_1 + a_0 & 0 \\ 0 & a_mB_2^m + ... +a_1B_2 + a_0 \end{bmatrix} =  \begin{bmatrix} p(B_1) & 0 \\ 0 & p(B_2) \end{bmatrix} $



### Polinomio minimo

il polinomio monico di grado più basso tale che $\varphi_A(A)=0$  

**Dimostra che è unico**

è unico, perchè se fossero due, $p(x)$ e $q(x)$, allora anche $p(x)-q(x)$ si annullerebbe, e avrebbe grado minore visto che sono monici e la più grande potenza di x si annullerebbe  
ma non può essere che abbia grado minore, impossibile  

**Dimostra che divide tutti gli altri polinomi che annullano A**

$m_A(x)$ polinomio minimo (quindi ha la proprietà $m_A(A)=0$)   
$p(x)$ generico polinomio che annulla A ($p(A)=0$)  

faccio la divisione tra polinomi:  
$p(x)=q(x)m_A(x)+r(x)$  

so che $p(A)=0$, quindi:  
$0 = q(A)m(A)+r(A)= 0 +r(A)$  
quindi $r(A)=0$, cioè $m_A(x)$ è divisore di $p(x)$  

come corollario, polinomio minimo p divisore di quello caratteristico  

### Blocchi di Jordan

**Polinomio caratteristico**

$B_{\lambda_0, n}$  
blocco si autovalore $\lambda_0$, dimensione n  

prova che il suo polinomio caratteristico è $p(x)=(x-\lambda_0)^n$  

demo 1:  
poichè è triangolare, il suo polinomio caratteristico è il prodotto delle entrate sulla diagonale: $p_A(\lambda)=(\lambda_0-\lambda)^n$   

demo 2:  
$p_B=det(B_{\lambda_0}-\lambda I)= det(\lambda_0I + N -\lambda I) = det(I(\lambda_0-\lambda)+N)= (\lambda_0-\lambda)^n$  
perchè $I(\lambda_0-\lambda)+N$ è una matrice triangolare, il suo determinante è il prodotto degli elementi sulla diagonale   
dove N è la matrice con tutti 1 nella diagonale sopra la principale, 0 nel resto, quindi 0 anche nella diagonale principale  

**Polinomio minimo di un blocco di Jordan**

il polinomio caratteristico di un blocco di Jordan $B_{\lambda_0, n}$ è $p(x)=(x-\lambda_0)^n$  

ora voglio dimostrare che quello minimo è uguale  
so che il polinomio minimo divide quello caratteristico, quindi è $m(x)= (x-\lambda_0)^k$ per un $k\le n$  

vediamo cosa succede con un k qualsiasi:  
$(B-\lambda_0I)^n = B_{0,n}^k$  
dove $B_{0,n}$ è nilpotente di ordine n  
quindi è nulla solo se k=n  
quindi minimo e caratteristico sono uguali  

**Blocco di Jordan nilpotente**

$B_{0,n}$ è un blocco di Jordan di autovalore 0, con n righe  

è nilpotente con indice n, cioè $B^n = 0$, e non prima  

note: moltiplicare una matrice per $B_0$ sposta le colonne verso sinistra, cancella la prima, e l'ultima diventa uno 0  
o moltiplicare un vettore colonna per $B_0$ sposta le coordinate di 1 verso l'alto, cancella la prima, l'ultima diventa 0  

**Dimensione blocchi di Jordan e polinomio minimo**

1-- scrivo char e minimum poly  

polinomio caratteristico di A in funzione dei suoi autovalori (lo calcolo dopo aver triangolarizzato A):    
$p_A(x)= \pm(x-\lambda _1)^{m_1} ... (x-\lambda_k)^{m_k}=  \pm \prod (x-\lambda_i)^{m_i}$  
dove m è la molteplicità algebrica di quel'autovalore  

polinomio minimo divide quello caratteristico, quindi è:  
$m_A(x) = \prod (x-\lambda_i)^{d_i}$  
dove $d_i \le m_i$  

2-- cosa voglio provare  

considero J, la matrice di Jordan associata ad A  
volgio dimostrare che d è la massima taglia dei blocchi di Jordan di quell'autovalore  

singolo blocco: (se l'autovalore $\lambda_0$ ha solo un blocco di dimensione n)    
allora $B_{\lambda_0, n}$, come già detto è n-nilpotente, quindi serve $d=n=m$  

intera matrice di Jordan (più blocchi per uno stesso autovalore):    
il polinomio minimo deve annullare tutta J, quindi è il minimo comune multiplo dei polinomi minimi di ogni blocco  
sono tutti blocchi dello stesso autovalore, quindi il minimo comune multiplo è $(x-\lambda_0)^d$ con il d più grande utilizzato  
il d più grande utilizzato è quella del blocco più grande, che si annulla solo se d è uguale alla sua taglia  

come volevo dimostrare, d è la taglia del più grande blocco di Jordan con quell'autovalore  
e chiaramente deve essere almeno 1  

nota:   
d è la dimensione del più grande blocco di Jordan con dell'autovalore  
$m_G$ è il numero di blocchi di Jordan con quell'autovalore  
$m=m_A$ è la molteplicità algebrica  

quindi $d \neq m_G$  

per ogni autovalore, valgono le disuguaglianze:  
$1 \le d \le m=m_A$  
$1 \le m_G \le m_A$   

**Diagonalizzabile se e solo se il polinomio minimo non ha radici ripetute (esponenti 1)**

A è diagonalizzabile se la sua forma di Jordan è diagonale  
ciò accade se tutti i blocchi hanno taglia 1  
quindi tutte le d nella formula sopra sono 1
