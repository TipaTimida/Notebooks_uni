### Teorema spettrale

nota:  

B: base ortonormale  
$B'$: base ortogonale, ma NON ortonormale  
S: matrice di prodotto scalare  
A: matrice di endomorfismo  

ortonormale:  
$[S]_B$: è l'identità  
$[A]_B$: è diagonale con gli autovalori sulla diagonale  

ortogonale:  
$[S]_{B'}$: diagonale, ma non l'identità, ha valore diversa in ogni entry  
valori sulla diagonale sono i quadrati delle norme dei vettori della base  
li chiamo i wuadrati, ma sono il prodotto scalare solito con sè stesso, vuol dire che non ho fatto la radice  
sono quelli da calcolare per vedere la segnatura  
$A$ non ha niente di particolare così  

per avere A diagonale, la condizione è avere una base di autovettori, non avere una base ortogonale, che è irrilevante  

**Cosa dice**

solo se il prodotto scalare è definito positivo  

se e solo se A è autoaggiunto:  
ha una base ortonormale di autovettori e tutti gli autovalori sono reali  

A autoaggiunto $\leftrightarrow$ base ortonormale di autovettori, autovalori reali  

$\leftarrow$  
se B è una base di autovettori, $[A]_B$ è diagonale   
poichè gli autovalori sono reali, ha numeri reali sulla diagonale  
quindi la matrice A è hermitiana   
quindi l'endomorfismo che descrive è autoaggiunto  

$\rightarrow$  

dimostra che gli autovalori sono reali se A è autoaggiunta:  
se $\lambda$ è autovalore e v il suo autovettore:  
$<Av,v>= <v,Av>$  
$<\lambda v,v>=<v, \lambda v>$  
$\lambda<v,v> = \overline{\lambda}<v,v>$

cioè:  
$\lambda<v,v>=<\lambda v,v>= <Av,v>= <v,Av> = <v, \lambda v> = \overline{\lambda}<v,v>$  

$<v,v>\neq 0$ perchè il prodotto scalare è definito positivo, quindi $\lambda = \overline{\lambda}$, cioè è reale  

$\rightarrow$  

dimostra che se T è autoaggiunta ha una base ortonormale di autovettori:

2-- caso n, se è vero per n-1

**Criterio di Cartesio**

$p(x)= a_nx^n+ ... +a_mx^m$, con n>m, polinomio a coefficienti reali, con primo e ultimo coeff non nulli  
allora:  
1- m è numero di radici nulle di p  
2- numero di radici positive di p è il numero di cambiamenti di segno nella sequenza di coeff (rimuovendo quelli nulli)  

1- è ovvio, basta raccogliere per $x^m$  

2- 
per induzione su n, il grado di p  

n=1:  
$p(x)=a_1x+a_0$, c'è una radice positiva se i coeff hanno segno opposto  

generale n, se vero per n-1:  
$p(x)=a_nx^n + ...+a_0$  

se $a_0=0$: posso raccogliere per x e ottenfo $p(x)=xq(x)$   
$q(x)$ è di grado n-1, quindi è vero  
p e q hanno stesse radici positive e coefficienti, quindi vero anche per p  

se $a_0 \neq 0$:  
derivo p:  
$p'=na_nx^{n-1}+ ...+a_1$  
i coeffiienti hanno tutti gli stessi segni di prima, ma manca $a_0$  
p' è di grado n-1, quindi per ipotesi induttiva le radici positive solo in numero t, uguale al numero di cambiamenti di segno  
p ha t o t+1 radici positive, e ha t p t+q cambiamento di segno  
ma non possono differire di un numero dispari, quindi sono uguali
