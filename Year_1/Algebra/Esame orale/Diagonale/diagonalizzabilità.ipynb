### Autovettori

avere una base di autovettori significa essere diagonalizzabile  

**Se hanno autovalori distinti, sono indipendenti**

autovettori: $v_1, ... v_k$  

provo per induzione su k:  

1-- caso k=1  
chiaramente, un solo vettore è indipendente  

2-- caso k, se è vero k-1  
sono indipendenti se l'unica soluzione di $v = \Sigma a_iv_i = 0$ è quella banale  

applico:  
$(A-\lambda _k)( \Sigma a_iv_i) = \Sigma (\lambda_i-\lambda_k)a_iv_i = 0$  
nota che v=0, quindi qualunque trasformazione gli applico prima continuerà a dare 0  

è una somma di k-1 vettori, perchè il termine dove $i=k$ è nullo (nota che $\lambda_k$ resta sempre quello, mentre i itera su tutti)  
quindi, per ipotesi induttiva, tutti gli $a_i$ sono nulli, tranne $a_k$  
ma allora $\Sigma a_iv_i= a_kv_k=0$, quindi anche $a_k=0$

**Autovalori e radici del polinomio caratteristico**

autovalore: $Av = \lambda v$  

posso traformare in:  
$Av-\lambda v=0$  
$(A-\lambda I)v=0$   
$v \in ker(A-\lambda I)$  
$det(A-\lambda I)=0$  perchè c'è qualcosa nel kernel, quindi è rank-deficient   (singolare)  
$p_A(\lambda)=0$  

cioò se pluggo un autovalore nel polinomio carattristico, mi dà 0  


**Cose utili per proof**

$Av=0$  
$ v \in ker A$, non importa quanto A sia complessa  
$det(A)=0$, perchè ha qualcosa nel kernel, quindi è singolare  

### Autospazi

**Molteplicità algebrica e geometrica**

prova che $1 \le m_g \le m_a$ per ogni autovalore  

1-- $1 \le m_g$  
se $\lambda$ è un autovalore, esiste almeno un autovettore che $Av = \lambda v$, quindi l'autospazio ha almeno dimensione 1  

2-- $m_g \le m_a$  
prendo una base  di $V_{\lambda}$, formata da $m_g$ autovettori  
la completo a base B di V  
in questa base la matrice è: $T_B = \begin{bmatrix}\lambda I_k & C \\ 0 & D \end{bmatrix}$  
dove la prima colonna si riferisce agli autovettori in quell'autospazio, le altre al resto, quelli che ho usato per completare a base di V  
il polinomio caratteristico di una triangolare a blocchi è:  
$p_A(x)=p_{\lambda I_k}(x)p_D(x) = (\lambda -x)^kp_D(x)$  
quindi la radice ha almeno molteplicità algebrica k, ma potrebbe di più, se $\lambda$ appare come radici anche in $p_D$  


**Matrici simili**

A e B simili: $A=M^{-1}BM$  
autovalori uguali, con stesse $m_a$ e $m_g$ per ognuno  
ma autovettori sono diversi (sarebbero uguali solo se commutano)  

prova che autovalori uguali etc.  

1-- molteplicità algebrica:  
A e B sono simili, quindi hanno stesso polinomio caratteristico  
quindi le molteplicità algebriche sono uguali  

2-- molteplicità geometriche:  
$m_{gA}(\lambda)= dim \, ker(A-\lambda I) = dim \, ker(M^{-1}BM-M^{-1}\lambda IM) = dim \, ker(M^{-1}(B-\lambda  I)M) = dim \, ker(B-\lambda  I) = m_{gB}(\lambda)$  

simile al metodo usato per provare che polinomio caratteristico è lo stesso  
ricorda che I è uguale in ogni base  
ultima uguaglianza è perchè due matrici simili hanno lo stesso rango, quindi stessa dimensione del nucleo , quindi posso togliere $M$ e $M^{-1}$ (o perchè M è full rank, quindi posso ignorarla se voglio sapere la dimensione del ker)



La molteplicità algebrica $m_a$ è la molteplicità di λ come radice del polinomio caratteristico  
La molteplicità geometrica $m_g$ è la dimensione dell’autospazio associato a λ, cioè $m_g (λ) = dim \,V_λ$

### Diagonalizzabilità

**Polinomio caratteristico ha n radici distinte, allora diagonalizzabile**

autovettori rispetto a diversi autovalori sono indipendenti  
quindi posso fare una base di autovettori  

avere una base di autovettori significa essere diagonalizzabile  

**Gli autospazi sono in somma diretta**

$\oplus V_i$ (cioè sono disgiunti): è vero sempre  
$\oplus V_i=V$ (spannano tutto V): non sempre, solo se diagonalizzabile  

prova che sono disgiunti sempre:  
se un vettore fosse in due di essi, tipo $V_1$ and $V_2$, allora $Av$ sarebbe contemporatneamente $v$ e $2v$, che è impossibile  

prova che è diagonalizzabile se e solo se gli autospazi sono in somma diretta  
diagonalizzabile $\leftrightarrow \oplus V_i=V$  
già dimostrato che sono disgiunti, devo solo dimostrare che spannano V  
$\leftarrow$ unisco le basi degli autospazi, così ottengo una base di autovettori per A, che vuol dire che A è diagonalizzabile  
$\rightarrow$ se è diagonalizzabile, ogni vettore è combinazione lineare di autovettori, quindi $v \in V_1 + ... +V_k$  


**Condizione di diagonalizzabilità**

A è diagonalizzabile in K se:  
tutte le radici sono in K  
$m_a=m_g$ per ogni autovalore  

1-- cosa voglio dimostrare  
gli autospazi sono sempre disgiunti (in somma diretta), quindi definiamo la loro somma $ W = V_{\lambda 1}+ ... V_{\lambda k}$  
sappiamo che, se e solo se è diagonalizzabile, $W=V$, quindi $dim(W)=n$  
quindi se dimostro che $dim \, W = n$, vuol dire che è diagonalizzabile  
cioè, voglio dimostrare che $\oplus V_i = V$, il che significa che è diagonalizzabile  

2-- scrivo la cosa da dimostrare  
$dim \, W = dim \, V_{\lambda 1} +...+dim \, V_{\lambda k} = m_g(\lambda_1) + ... + m_g(\lambda_k) \le m_a(\lambda _1)+ ...+m_a(\lambda_k) \le n$  

prima disuguaglianza: disuguglianza tra molteplicità algebrica e geometrica  
seconda disuguaglianza: $m_a$ sono le radici del polinomio caratteristico, che deve avere grado n, quindi ha al massimo n radici  

quindi $dim \, W =n$ solo se le disuguaglianza sono uguaglianze  
la prima diventa uguaglianza se $m_a=m_g$ per ogni autovalore  
la seconda se le n radici sono tutte in K 

### Traccia e determinante

**Traccia e determinante**

se tutte le radici sono in K:  
$tr = \Sigma \lambda$  
$det = \prod \lambda$  

sono ovvie se A è diagonale  
poichè sono invarianti per similitudine, sono ovvie anche se A è diagonalizzabile  

se A non è diagonalizzabile:  
so che sempre:    
$p_A(\lambda)= (-1)^n \lambda ^n + (-1)^{n-1}tr \, A \lambda ^{n-1} + ...+ det \, A$  

se tutte le radici sono in K, vale anche:  
$p_A(\lambda) = (-1)^n (\lambda-\lambda_1)...(\lambda-\lambda_n) = (-1)^n \lambda ^n + (-1)^{n-1}(\lambda_1 +...+\lambda_n) \lambda ^{n-1} + ...+ (\lambda_1...\lambda_n)$  
basta confrontarle  

**Cosa leggo nel polinomio caratteristico**

$p_A(\lambda)=a_n\lambda^n + ...+ a_0$  

$a_n =(-1)^n$  
$a_{n-1}=(-1)^{n-1}tr \, A$  
$a_0 = det \, A$  

dimostra per induzione su n  

se n=1  
la matrice è A=a  
$p_A(\lambda)=det(A-\lambda I)= det(a-\lambda)=a-\lambda= -\lambda +a$  
si vede che c'è tutto ciò che volevo
