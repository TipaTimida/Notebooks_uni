## Definizione

un funzionale è una funzione lineare $f: V \rightarrow K$  

spazio duale di V:  
$V^* = \{f: V \rightarrow K | f \, lineare\}$

cioè un funzionale è un vettore riga  

so che $f,g \in V^*$  
è vero che $f+g \in V^*$?  

basta usare la linearità:  
$(f+g)(\alpha v + \beta w) = \alpha (f+g)v + \beta (f+g) w$  

quindi f+g è lineare, quindi rispetta le condiioni per essere in $V^*$  

prendo $v_1,...v_n$, una base di V  
prendo una base duale (composta da funzionali):  
$v_i^*: V \rightarrow K$  
$\Sigma a_jv_j \mapsto a_i$  
cioè $v_i^*$ è un funzionale che estrae la i-esima componente della combinazione lineare di un vettore  

dimostro che $v_i^*$ è l'unico funzionale tale che $v_i^*(v_i)=1$ e $v_i^*(v_j)=0$  

in coordinate, $v_i^* = [0,0,0,1,0,0,0]$ con l'1 in posizione i  
quindi è un vettore riga  

può essere utile?  

ora mostro che generano $V^*$  

ho $a_i=f(v_i)$, voglio prova che $f(v)=(\Sigma a_iv_i^*)(v)$  
prendo $v= \Sigma b_iv_i$  
e $f(v_i)=a_i$  

$f(\Sigma b_iv_i) = \Sigma b_i f(v_i) = \Sigma b_i a_i = \Sigma a_ib_iv_i^*(v_i) = \Sigma _i \Sigma _j a_i b_j v_I^*(v_j) = (\Sigma a_iv_i^*)(\Sigma b_jv_j) = (\Sigma a_iv_i^*) (v)$


### Isomorfismo

$V$ e $V^*$ hanno la stessa dimensione, sono isomorfi  

quindi $v_i \mapsto v_i^*$ è un isomorfismo  

anche $V^{**}$ è isomorfico  
e coincide esattamente con $V$  
perchè prima apsso ai funzionali, poi torno ai vettori  

## Teorema di rappresentazione di Riesa

ogni funzionale può essere scritto come prodotto scalare con uno specifico vettore  

se $\varphi$ definito positivo (quindi anche non degenere):  
$\forall f \in V^* \quad \exists w \in W: \quad f(v)=<v,w>$  

significa che, fissato un prodotto scalare, la funzione $V \rightarrow V^*$, $v \mapsto <v,\cdot>$ è un'isomorfismo  

se $\varphi$ è definito positivo, fissare $\varphi$ è come fissare una base, perchè esiste una B dove $\varphi$ è rappresentato dalla matrice identità

**Proof**

$\varphi_w(v)=<v,w>$ è un funzionale lineare  
anche se fosse hermitiano, perchè sia scalare che hermitiano sono lineari nella prima compoenente, e la seconda è fissata   

1-- cosa voglio dimostrare  
che, fissata una base di V $B = \{v_1, ..., v_n\}$, l'insieme $\varphi_{v_1}, .. \varphi_{v_n}$ (rappresenta un prodotto scalare con quel vettore) è una base di $V^*$  
infatti, se fosse così, avrei:  
$(f)(v)=(\Sigma a_i \varphi_{v_i})(v) = \Sigma a_i <v,v_i> = \Sigma <v, \overline{a_i}v_i> = <v, \Sigma \overline{a_i} v_i>$  
cioè f descrive un prodotto scalare/hermitiano col vettore $\Sigma \overline{a_i} v_i$, che è ciò che volgio dimostrare  

2-- dimostro che $\varphi_{v_1}, .. \varphi_{v_n}$ è una base di $V^*$    

indipendenti:  
sia $f= \Sigma a_i\varphi_{v_i}=0$  
poichè $f(v)= <v, \Sigma \overline{a_i} v_i>=0$, allora $\Sigma a_i\varphi_{v_i} \in V^\perp$  
ma il prodotto scalare canonico <> è non degenere, quindi il radicale è vuoto, quindi per avere f=0 bisogna che uno due due vettori sia nullo, cioè $\overline{a_i}=0$  
(tutti i coefficienti devono essere nulli, perchè <> è positivo, quindi tutti gli addendi avrebbero un prodotto positivo, se ne sommo anche solo uno non posso tornare a 0)  
quindi anche $a_i=0$

spannano V:  
poichè $dim(V)=dim(V^*)$, sono già il numero giusto  
so che sono indipendenti, quindi ok
