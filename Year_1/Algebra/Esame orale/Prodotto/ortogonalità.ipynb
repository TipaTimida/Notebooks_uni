### Teorema di Sylvester / Lagrange

Esiste sempre una base $\varphi-ortogonale$ per uno spazio, qualunque sia $\varphi$  
lo dimostro per induzione su n, la dimensione di V  

1-- caso n=1  
base di V è fatta di un solo vettore, per forza ortogonale  

2-- caso generale n, se vero per n-1  

2.1 -- se esiste un vettore v non isotropo:  
divido lo spazio in: $V = W \oplus W^\perp$    
dove $W= span\{v\}$  
sono in somma diretta perchè v è non isotropo, quindi non è perpendicolare a sè, quindi non è in $W^\perp$   
$W^\perp$ ha dimensione n-1, per ipotesi induttiva ha base ortogonale  
la espando con v, ottengo una base ortogonale per V (perchè sono in somma diretta, quindi posso sommare le basi)  

2.2 -- se tutti i vettori sono isotropi:  
il prodotto scalare tra due vettori è, usando la multilinearità sul prodotto scalare:  
$\varphi(v,w) = \frac{1}{2}g(v+w, v+w)-g(v,v)-g(w,w)$  
quindi se tutti sono isotropi, il lato destro è 0, quindi anche il alto sinistro, cioè ogni base è ortogonale, quindi l'ho dimostrato  

**Congruenza matrice simmetrica**

**Dimostra che: $dim \, W^\perp \ge n - dim W$**  

prendo base di W, fatta da $w_1, ... w_k$, perchè W ha dimensione k    
$ v \in V$ è ortogonale ad ogni vettore di W se e solo se è ortogonale ad ogni vettore nella base  

considero T, una trasformazione che fa:  
$T(v) = \begin{bmatrix} g(v,w_1) \\ ... \\ g(v, w_k) \end{bmatrix}$  

$W^\perp$ è il nucleo di $T: V=R^n \rightarrow R^k$ (T manda nel kernel/vettore nullo la roba perpendicolare ad ogni vettore nella base di W)  
per il teorema della dimensione:  
$dim W^\perp = dim \, ker \, T = n- dim \, Im \, T \ge n-k$  
perchè T perde n-k tra dominio e codominio, la dimensione dell'immagine è al massimo k, al massimo quella del codominio, non puù aumentare    
perchè T va da un vettore a n componenti a uno a k componenti   

**se $g|_W$ è non degenere, $V = W \oplus W^\perp$**  

1-- sono disgiunti  
$g|_W$ è non degenere: quindi nessun vettore in W è ortogonale a tutti quelli in W  
quindi $W$ e $W^\perp$ sono disgiunti  

2-- spannano V  
dal punto 1: so che $dim \, W + dim \, W^\perp \ge n$  
quindi per grassman:  
$dim (W+W^\perp) = dim W+ dim W^\perp - dim(W \cap W^\perp) = dim W + dim W^\perp \ge n$  
quindi $W + W^\perp =V$, spannano V  

quindi $V = W \oplus W^\perp$
