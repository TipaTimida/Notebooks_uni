**Ogni vettore si scrive in modo unico come combinazione lineare**

ho base di vettori B  
assumo che v si scriva in 2 modi: 
$v = \Sigma a_iv_i = \Sigma b_iv_i$  

sposto tutto a sinistra:  
$\Sigma (a_i-b_i)v_i = 0$  
poichè B è una base, deve avere come unica soluzione quella banale, quindi $a_i=b_i$  

**Vettori independenti e nel numero giusto sono una base**

ho n $w_i$ vettori indipendenti  
ho n $v_i$ vettori che so che sono una base  
dimostra che anche $w_i$ lo sono  

$v_i$ sono una base, li uso per esprimere $w_1$: $w_1 = \Sigma \lambda_iv_i$   
poichè $w_1 \neq 0$ (vettore nullo non è mai indipendente), almeno un coeff è non-0  
divido per quello, ad esempio $\lambda_1$  
ottengo: $\frac{1}{\lambda_1}w_1=v_1 - \frac{\lambda_2}{\lambda_1}v_2...$  
cioè $v_1$ è combinazione linare di $w_1$ e degli altri v  
quindi nello span, posso sostituire $v_1$ con $w1$  
$V= span\{w_1, v_2, v_3...\}$  

iterativamente:  
prendo il prossimo w, $w_2$  
scrivo come comb linear di roba attualmente nello span  
deve esserci un coeff diverso da quelli associati ad uno dei v, non ad uno dei w già lì (perchè se non ci fossero, l'attuale w sarebbe combin dei precedenti w, ma non può essere, sono indipendenti)  
divido per quel non-0 coeff  
sposto la v di quel non-0 coeff di lato da sola, è linear comb dello span che sostituisce sè stesso con la versione w  
quindi nello span sostituisco quel v con w  



**Iniettività**

dimostra che:  
f iniettiva $\leftrightarrow ker \, f = \{0\}$  

$\rightarrow$  
sempre, $f(0)=0$  
f iniettiva vuol dire $v \neq 0 \rightarrow f(v) \neq 0$ (if vectors are different they ust give different outputs)  
so $ker \, f=\{0\}$  
in words: injective means only one vector can be mapped to each output, and i know that 0 vector must be necessarily mapped to kernel, so that is the only vector in there  

$\leftarrow$  
pick 2 different vectors $v, v' \in V$, so $v-v' \neq 0$ (different so the difference is non-zero)  
visto che solo il vettore nullo è nel kernel, e v-v' non lo è, la sua immagine non è nel kernel: $f(v-v')=f(v)-f(v')\neq 0$  
quindi $f(v) \neq f(v')$, la condizione di iniettività

**Definizione immagine**

se $v_i$ sono generatori di V, allora dimostra che $Im \, f = span\{f(v_1), ...f(v_n)\}$  

inizion scrivendo la definizione di immagine:  
$Im \, f = \{f(v): v \in V\} = \{f(\Sigma \lambda v_i): \lambda_i \in K\} = span\{f(v_1), ...f(v_n)\}$

**Teorema della dimensione**

$f: V \rightarrow W$  
se V ha dimensione finita n, allora:  
$dim \, ker \, f + dim \, Im \, f = n$  

1-- cosa voglio dimostrare  
prendo $v_1, ... v_k$, una base di $ker \, f$  
la completo a base di V, fino a $v_n$ (si può sempre fare)  
voglio dimostrare che $f(v_{k+1}), ... f(v_n)$ sono una base di $Im \, f$  
se fosse così, $dim \, ker \, f = k$ e $dim \, Im \, f = n-k$, che prova il teorema  

2-- dimostro che $f(v_{k+1}), ... f(v_n)$ generano $Im \, f$   
(è un sottospazio, non tutto $R^n$, quindi devo provare che gli appartengono, non solo che sono nel numero giusto)  
so che $f(v_1), ...f(v_n)$ generano l'immagine, perchè le immagini della base sono una base per l'immagine  
so anche che $f(v_1), ...f(v_k)=0$ perchè le v sono nel kernel, quindi posso rimuoverlo dalla lista  
quindi ho ottenuto che che $f(v_{k+1}, ... f(v_n))$ generano $Im \, f$  

3-- dimostro che $f(v_{k+1}), ... f(v_n)$ sono indipendenti  
devo mostrare che $\Sigma_{k+1}^n \lambda _i f(v_i) = 0$ ha solo la soluzione banale $\lambda_i=0$  
per linearità di f ottengo:  
$f(\Sigma_{k+1}^n \lambda_iv_i)=0$  
quindi $\Sigma_{k+1}^n \lambda_iv_i \in ker \, f$  
poichè $v_1, ...v_k$ sono una base del kernel, posso scrivere anche:  
$\Sigma_{k+1}^n \lambda_iv_i = \Sigma ^k_1\alpha_iv_i$  
sposto tutto a sinitra, trovo:  
$-\Sigma ^k_1\alpha_iv_i + \Sigma_{k+1}^n \lambda_iv_i = 0$  
ma tutti i vettori da $v_1$ a $v_n$ sono indipendenti, perchè per ipotesi sono una base di V, quindi l'unica soluzione è quella banale, con $\lambda_i=\alpha_i=0$, che è ciò che volevo

**Composizione di trasformazioni**

$f: V \rightarrow W$  
$g: W \rightarrow Z$  

dimostra che $dim\, Im (g(f))\le min\{dim \, Im \, g, dim \, Im \, f\}$  

vale sempre $Im (g(f)) \subseteq Im(g)$ (perchè comunque l'ultima è g, l'immagine finale deve essere qualcosa che può apparire dopo g)  
quindi $dim \, Im (g(f)) \le dim \, Im \, g$   
e $Im \, (g(f))= Im(g|_{Im \, f})$, quindi $dim \, Im (g(f)) \le dim \, Im \, f$  

questo servirà a dimostrare che $rank(AB) \le min\{rank(A), rank(B)\}$ e che se A è invertibile, $rank(AB)= rank(B)$

dimostra che se f è un isomorfismo, dim \, Im(g(f)) = dim \, Im \, g$  

se f è un isomorfismo, $g(f) f^{-1} = g$  
applicando la disuguaglianzaotteniamo anche $dim \, Im g \le dim \, Im (g(f))$

### Dimensioni

**Formula di Grassman**

1-- cosa voglio dimostrare  
ho $v_1, ... v_k$ base di $U \cap W$  
espando a base di U con $u_1, u_s$  
a base di W con $w_1, w_t$  
voglio dimostrare che, tutte insieme, sono base di U+W  
cosi $dim(U+W)=k+s+t=(k+s)+(k+t)-k = dim(U)-dim(W)-dim(U\cap W)$  

2-- generano  
chiaramente generano U+W  
perchè un generico vettore in U+W è linear comb di u,v,w  

3-- indipendenza  
$\Sigma a_iv_i +\Sigma b_iu_i + \Sigma c_iw_i=0$  
$v+u+w=0$ (riscrivo con nomi di generici vettori)  
$w=-v-u$  
lato sinistro è in W, destro è in $U$ (perchè v è in intersezione di W e U)  
sono uguali, quindi entrambi sono in $W \cap U$  
allora scrivo w, che ricordo essere in W, come combinazione della base di $W \cap U$:  
$w = \Sigma c_iw_i = \Sigma d_iv_i$  
se sposto tutto a destra, ho linear comb della base di W, quindi i coeff $c_i$ sono nulli   
quindi w è nullo, quindi $v+u=0$  
scrivo come linear comb, sono una base di U, anche questi sono tutti nulli  
ok, tutti coeff nulli, sono indipendenti

### Sistemi lineari

**Touchè-Capelli**

cosa dice: sistema ha soluzione se $rank(A|b)=rank(A)$  
e lo spazio delle soluzioni ha dimensione $n-rank(A)=ker(A)$ (numero di variabili libere)  

sistema ha soluzione se b è nello span delle colonne di A: $b \in span\{A^1...,A^n\}$  
se accade: $span\{A^1...,A^n, b\}=span\{A^1...,A^n\}$, quindi $rank(A|b)=rank(A)$  
altrimento c'è il segno $\supsetneq$ , quindi $rank(A|b)>rank(A)$  

se ha soluzione, dimensione dello spazio delle soluzioni S è uguale a dimensione di $S_0$, spazio delle soluzioni del sistema lineare omogeneo associato  
$S_0$ è generato da n-pivots vettori, cioè $n-rank(A)$

note:  

$S_0$: spazio soluzioni sistema omogeneo (associato)  
$S$: spazio soluzioni sistema non omogeneo  

$S_0$: è un sottospazio vettoriale  
$S$ no, perchè non contiene l'origine  

$S_0$ deve sempre esistere, eventualmente è solo il vettore vuoto  
cioè sistema homo deve avere soluzione, eventualmente solo banale  

mentre sistema non-homo può non avere soluzione, S non esiste

### Algoritmi

**Algoritmo di completamento**

V spazio vettoriale di dimV=n  
ho k vettori indipendenti appartenenti a V, con k<n   
posso sempre completarli a base di V così:  
poichè $k<n$, non sono base di V, quindi c'è un vettore in V ma non nello span dei miei k vettori (che vuol dire indipendente ai vettori dello span attuale), lo chiamo $v_{k+1}$
lo aggiungo allo span dei miei vettori e ripeto, finchè n=k  

**Algoritmo di estrazione**

ho spazio V con generatori m vettori  
voglio estrarne base  

se gli m vettori sono indipendenti, sono già una base  
sennò, uno si scrive come combinazione degli altri, lo rimuovo dallo span  
ho ancora uno span che genera V, continuo finchè sono indipendenti
