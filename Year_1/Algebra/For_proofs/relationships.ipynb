### Commutano / simultaneously diagonzalizable

**Definition**

A,B COMMUTANO and BOTH ARE DIAGONALIZABLE</u> $\longleftrightarrow$ A,B are diagonalizable simultaneously  

formally:  
lemma che afferma che due applicazioni lineari separatamente diagonalizzabili che commutano fra loro sono simultaneamente diagonalizzabili  
due endomorfismi commutano se e solo se l’uno preserva gli autospazi dell’altro  

**Terminology**

- $AB=BA$ (A,B commutano)  
means PRESERVE each other's EIGENSPACES  

- diagonalizable simultaneosuly  
means: exist same matrix P that diagonalizes both, they share a base of eigenvectors (the cols of P)     
must check that they're diagonalizable, bc if A=B is a matrix that is not diagonalizable, they commute but obv they're not diagonalizable simultaneously    

#### Preserving eigenspaces (drawing matrices)

**Preserving eigenspaces**

A preserving eigenspace of B means:    
$A(V_\lambda) \subset V_\lambda$  
where $V_\lambda$ are eigenspaces of B  

so <u>eigenspaces of B are invariant wrt A</u>  
v is eigenvector of B  
if apply A to it before, it's still an eigenvector of B for the same eigenvalue !!!  

written:  
$B(Av)= \lambda(Av)$  

def:  
$v \in V_\lambda \rightarrow Av \in V_\lambda$

this require A,B to commutate, NOT to be simultaneously diagonalizable

#### If A has distinct eigenvalues (and AB=BA)

**Eigenvectors**

all eigenvectors of A are eigenvectors of B  
(not necessarily the opposite, bc B might have repeated eigenvalues)  
so B has basis of eigenvectors, thus B is diagonalizable (not necessarily with distinct eigenvalues)

**How to draw B**

B not necessarily has distinct eigenvalues too  
bc would have:  

$A= \begin{bmatrix}

\end{bmatrix}$

$B= \begin{bmatrix}
a & 0 & 0 \\

but there's no reason why a and b couldn't be the same  
so cannot tell anything about multiplicity of eigenvalues of B  

**Proof: every eigenvector of A is an eigenvector for B**

B must respect A's eigenspaces  
but since all eigenspaces of A are 1-dimensional, must simply scale the vector (not necessarily by the same amount as the eigenvalue in A)     
thus, if $Bv \in A_{\lambda}$, $Bv = cv$  
thus $v$ is also an eigenvector of B (with another eigenvalues)  

**Check that B is diagonalizable**

#### Proofs

**Preserve eigenspaces**

prove if AB=BA, B preserve A's eigenspaces  

if W is an eigenspace of A of eigenvalue $\lambda$ and eigenvector v, and A and B commute:    
$AB(v) = BA(v) = B(\lambda v) = \lambda B(v)$  
so: $A(Bv) = \lambda (Bv)$, meaning $Bv$ is an eigenvector of A, so $Bv \in W$  
therefore all $Bv \in W$, the eigenspace is preserved    


#### Base that diagonalizes both

**Find base that diagonalizes both**

1. find base for each eigenspace of both  
2. check how many vectors i need for base, pick as needed  

if there are eigenspaces of dim=1, the vector that makes its base MUST be in the base  

**Example base that diagonalizes both**

$V_{A1} = span(v_1, v_2)$

this is easy to check, just check if it's a multiple of the vectors that generates it  

so:

### Direct sum

**Definition**

addition space when A and B are disjoint  
$ C = A \oplus B $  

means: 

1. $ A \cap B = \{ \overline{0} \}$
2. $ C = A + B $  

if A and B are disjoint:  
- intersection of spaces is the null vector (which has dim=0, but is not empty)   
- so intersection of bases is empty

**Prove 2 spaces are disjoint (condition 1)**

depends on what i have:  

1. have matrix describing space: use DIMENSION FORMULA   
$dim(A+B)=dim(A)+dim(B)$  -> easy with generator matrices, put together bases    
$dim(A\cap B)=0$  -> easy with coefficient matrices, put together carth eq      

2. have description of space (conditions)   
show that the <u>conditions for an element to be in A or B are such that only null vector belong to both</u>    

if asked to prove that A and B are in direct sum, and not asked anything about what they must add up to, means to prove A and B are disjoint

### F-invariant

**f-invariant**

$f(W) \subseteq W$  

a subspace is f-invariant if, after applying f, its image is contained in the dominium  
so if a vector was in it, after applying f it must keep being in it  
so applying f can't escape the subspace  

**Lemma: restriction to invariant subspace**
 
the restriction of a diagonalizable transformation to an INVARIANT SUBSPACE (invariant wrt that trasnformation itself) is invariant too  
essentially, if A is diagonalizable, means the blocks that describe each eigenspaces are diagonal too  

**Properties**

- orthogonal space   
W is f-invariant, $W^{\perp}$ is f-invariant  

- addition  
of A,B are f-invariant, $C=A+B$ is f-invariant
