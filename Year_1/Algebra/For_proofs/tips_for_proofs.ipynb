### Prove A=B

**A, B are subspaces**

1. split in $A \subseteq B, B \subseteq A$  
2. find easier one and prove it  
3. prove $dim(A) = dim(B)$  

if a subspace has same dimension that its ambient space, the two are the same  

**A,B are matrices**

prove $A_{ij} = B_{ij}$  
must know formula to define every cell of both matrices  

### Useful stuff for proofs

**Null vector**

can't get out of kernel  
$f(\vec{0}) = \vec{0}$  
 

**Rank inequalities**

$rank(AB) \le min(rank(A), rank(B))$  
bc can cancel dimensions, not add them  

$rank(A+B) \le rank(A) + rank(B)$   
$rank(A+B) \ge max(rank(A),rank(B)) − min(rank(A),rank(B))$

$rank(A)+ ker(A) = n$  
this is grassman rule with Im and ker  



### Are vectors independent?

**For proofs**

take the things i want to be independent: $\{v_1, ... v_n\}$  
write a null linear comb  $\Sigma a_i v_i = 0$  
prove only sol is $a_i =0 \; \forall i$

**Prove something**

$a_0 x_1+ . . . + a_n x_n = 0$  
write linear comb =0, show only solution is if all $a_i$ are zero  


if instead of a vector $x_i$ i have a polynomial $p_i(x)$, the only solution must be the one with null coeffs <u>for all values of x</u> i plug in the polynomial, not just some specific one   
so can plug in some useful value of x that gives me some condition on the coeffs (maybe some coeff to 0), and conditions found by plugging in values must all be true at the same time, bc eq must be true for any value i plug in  

### Trace

$tr(A)$ is sum of elements on main diagonal  
since it's invariant wrt base change, it's also the sum of the eigenvalues, each taken with its own multiplicity (just compute it when in diagonal or triangular form) (this is true even if the matrix is ot diagonalizable)

**Properties**

to remember properties, remeber trace is LINEAR:  

$tr(AB) = tr(BA)$  
$tr(A+B)=tr(A)+tr(B)$  
$tr(A^n) = \lambda _1^n + \lambda _2^n ...$   


cannot: $tr(A^2) \neq tr(A)^2$  


used only in transformation matrices, bc the cool thing about it is that it's invariant wrt similarity, which is the base change for transformation matrices  
but it's not invariant wrt row-reduction, so make no sense in space matrices  

### Studying a function

2 methods, when have a description of its effects:    

1. write matrix describing it        
2. find regularity    

**Method 2: find regularity**

1. find a combination of that function, maybe a power, that gives me a well-known function (like identity, or null-matrix, or a multiple of itself)  
eg: $f^2 = I$  
this would also mean that f is full rank and invertible  

2. get $q(x)$ from it, a multiple of the minimum poly  
also remember if $f^n =I$, it's diagonalizable    

3. from $q(x)$, find the eigenvalues (rem not all roots are eigenvalues)

4. based on the eigenvalues obtained, try to understand what their eigenspaces could be (so what is scaled in that way when applying this function, maybe are symmetric matrices?)  

5. find algebric multiplicity of the eigenvalues by finding the dimensions of such eigenspaces  



if know eigenvalues it's useful to write characteristic poly  
then can use Cayley th to find informations or properties on the matrix
