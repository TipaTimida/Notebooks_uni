### Understand the problem

**Something = 0: turn into scalar product**  

ONLY IF $\varphi > 0$  
a norm is nul only if the thing itself is null  

so can write stuff = 0 using the scalar product  
$(A+B)v = 0 \rightarrow ⟨(A+B)v, \, (A+B)v⟩=0$  

useful for ker: $Av = 0 \rightarrow ⟨Av, Av⟩=0$  

then something in ker means det=0  

**When scalar product gives 0**

if it's a norm and if $\varphi >0$: means that thing is null  
eg. $⟨Av,Av⟩=0 \rightarrow Av=0$  

if it's NOT a norm: means either  
the two things are orthogonal  
one of the two things is 0  

**Prove it's an isomorphism**

means prove it's injective and surjective  

injective: only thing in kernel is $0$, the kernel is trivial  
prove if $f(v)=0$, then $v=0$  

**Starting proof**

$v \in V, w \in W, A \in L(V,W)$  
means can write: $Av$ or $A^*w$, NOT $Aw$, must apply A to a vector in the input space  

**Proof with autoaggiunta/symmetric**

if want to prove:    
$T^*(w)= something \quad $ or $\quad T^t(v) = something$    

means i need to start and end by writing:  
$T^*(w)=⟨v, T^*w⟩ = ⟨Tv, w⟩ = ... = ⟨v, something⟩$  

where v is simply a v, a random vector, bc must be true for all v  
while w is the thing it takes in input, can be a linear comb or a vector with applied some transformations  
'something' is the thing i want it to be equal to  
T can be a complex thing, just make sure that the adjoint is applied to all of it  

numbers, like $\lambda$, can be moved in whatever place i want, since they're linear transformations: $T (\lambda v) = \lambda (Tv)$   

but multiplied matrices cannot be swapped: $AB \neq BA$

### How to proceed

#### Working with matrices

**Doing a base change**

have $Mx$  
if want instead to express it in another base, do $MPy$, where P describes the base change, $x = Py$  

**Apply a matrix to another**  

can apply a matrix to all the cols of another ($A_i$ are the cols):  
$AB = [A_1B | A_2B | ...| A_n B]$  

or can apply it to all the rows of anotehr ($A^i$ are the rows):  
$AB = \begin{bmatrix} A^1B \\ A^2 B \\ ... \\ A^n B\end{bmatrix}$

#### Scalar product

remember that canonical product is positive  
so all norms are strictly positive, zero only if the obj is a zero vector  

**Writing with symmetric**

$<Av,v>$ where A is symmetric measns the scalar product described by A, so $v^t Av$  

different ways to write same thing  

$\lambda ||v^2|| = \lambda <v,v>= <\lambda v,v>$  

often start from $\lambda ||v^2||$, but what does it mean???  

property: $<Av,v> \in R \quad \forall v$

**What to do**

1. take away scalar product  
left term gets transposed  
if product is not canonic, at left extremity of right member get the matrix describing it (so can either bring to canonical, or take it away completely)  

$⟨Ax,y⟩=(Ax)^t \; y$  
$\varphi _M(Ax,y) = ⟨Ax, My⟩ = (Ax)^t \; M \; y$  

2. keep product, use adjoints  
if using canonical product, can use symmetric 

$⟨Ax,y⟩=⟨x,A^ty⟩$  
$\varphi _M(Ax,y) = \varphi_M (x, A^*y)$  


**Transposed vs adjoint**

to bring a matrix from one side to the other:  

if in canonical product: transposed  
$⟨Ax,y⟩ = ⟨x,A^ty⟩$  

if in non-canonical product: adjoint (wrt that product)  
$\varphi(Ax,y) = \varphi(x,A^*y)$  
note that A is NOT the matrix describing this product  

**Handle stuff in terms**

multiplied matrices: if it's at the left extremity of its side, can transpose on other side    
$⟨ABx,y⟩ = ⟨Bx,A^ty⟩ = ⟨x,B^tA^ty⟩$

added matrices: split with bilinearity  

coefficient: since it's a multiplication, wherever it is, can move it wherever i like  
even outside of the scalar product  
$⟨\lambda x,y⟩=\lambda ⟨x,y⟩$  


can also swap terms: $⟨a,b⟩=\overline{⟨b,a⟩}$

**Cool tricks**

if $\varphi >0$, can use Cauchy-Schwarz  
 $⟨v,w⟩^2 \; \le \;  ⟨v,v⟩⟨w,w⟩$  

other useful thing:  
$⟨v,w⟩= \frac{1}{2}(⟨v+w,v+w⟩-⟨v,v⟩-⟨w,w⟩)$  

**Using linearity**  

can take out coeffs and split sums   

$\begin{cases}
\varphi(a+b,c) = \varphi(a,c) + \varphi(b,c) \\
\varphi(au, v) = a\varphi(u,v)
\end{cases}$

pay attention:  
$\varphi(a,c) + \varphi(b,c) = \varphi(a+b,c) \quad$ NOT $\varphi(a+b, 2c)$  

complete formula:  
$\varphi(a + b, c + d) = \varphi(a, c+d) + \varphi(b, c +d) =  \varphi(a, c) + \varphi(a, d) + \varphi(b, c) + \varphi(b,d)$  

if complex field:  
bringing out coefficient, changes depending on first and second term:    
$⟨\lambda v, w⟩ = \lambda ⟨v,w⟩$  
$⟨v, \lambda w⟩ = \overline{\lambda}⟨v,w⟩$  


#### Matrix expression

**If given poly expression between matrices**

if it's true on the full matrices, it's also true on all their their INVARIANT RESTRICTIONS/subspaces  

eg: $B^2+3B+2I=A$  
if can write A and B as block matrices (diagonal, so blocks not on diagonal are 0, which means restrictions are invariants), then this expression is true in all block  
so instead of A and B, can plug in the blocks $A_1$ and $B_1$, or $A_2$ and $B_2$  

**If matrix expression: use algebra**

can use normal algebra, so stuff with exponent -1 can be brought to denominator and then multiplied to other side  
$A^{-1} = I \quad \frac{I}{A} = I \quad I = A$  

**Inverse functions**

$ker (f) = f^{-1}(0) \subset f^{-1}(W^\perp)$  

when have inverse functions, can apply both sides that function, and $ff^{-1}= I$:    
$v \in f^{-1}(W^\perp) \longrightarrow f(v) \in W^\perp$

even with scalar product:  
$\phi(v,w) = \varphi(f(v),f(w)) \longrightarrow \phi(f^{-1}(v),f^{-1}(w)) = \varphi(v,w)$ 

**Apply a function**

can apply it on both sides     

$v \in V \cap f(V)$  
can become:  
$f(v) \in f(V) \cap f^2(V)$  

**Proof: check signature of prod associated with M**

M can be any expression, maybe $M=-A^2$  

just write:  
$v^t \; M \; v$  
then manipulate to obtain a norm  
eg, if A is antisymmetric:  
$v^t \; (-A^2) \; v = v^t \; (-A \, A)  \; v = v^t \; A^t \, A \; v = (Av)^t \, AV = <Av,Av> $   
got canonic product, which is defined positive  
so this product is defined positive too
