### Matrix associated with scalar product

can represent product with a matrix  
in each cell ij put values of product between vectors i and j of base

$[M_B(\varphi)]_{ij} = \varphi(v_i, v_j)$

**Properties of matrix**  

- scalar: symmetric  
$S = S^t$  

- hermitian: conjugate symmetric  
$H = \overline{H}^t = H^*$  
means on diag are only real values, rest in symmetric positions there are conjugate entries  

**Orthogonal base**

if base is orthogonal, associated matrix is diagonal  
if base is orthonormal, associated matrix is diagonal and all entries on diag are 1

#### Base change

scalar product has only one base, no input and output base like trasnformation matrix   

have: prod in base B  
want: prod in base B'  

compute P: matrix for base change from B to B   
P is made of vectors in base B expressed in coords of vectors in base B'  

$P = M^{B'}_B (Id)$

- scalar  
$M_{B'}(\varphi) = P^t \: M_B(\varphi) \: P$  

- hermitian  
$M_{B'} (\varphi) = \overline{P}^t  M_B(\varphi) \: P$  

the two M are congruent (in second case, called congruent under hermitian equivalence)  

**Proofs and base change**

if have P, matrix of base change wrt the canonical product:  
$\varphi (v,w) = \varphi (Pv,Pw)$  

if have T, matrix describing the scalar product $\varphi$:  
$\varphi (v,w) = <v, Tw>$

### Scalar vs hermitian

generally scalar used in R, Hermitian in C, but both can be used in the other field too

$\varphi : V x V \rightarrow K$  

K is R for scalar, C does hermitian  

properties:  

- scalar:  
  1. bilinear  
  2. symmetric

- hermitian: 
  1. linear only in first component  
  2. conjugate simmetric  
consequence of those is anti-linearity in second component

explanation properties:  

- linearity: can take out coeffs and split sums   
$\varphi(a+b,c) = \varphi(a,c) + \varphi(b,c)$  
$\varphi(au, v) = a\varphi(u,v)$  


- conjugate symmetry: symmetry, but put conjugate  
$\varphi (x,y) = \overline{\varphi (y,x)}$

- anti-linearity: to take coeffs out need to make them conjugate    
$\varphi(x, ay + bz) = \overline{a} \varphi(x,y) + \overline{b} \varphi(x,z)$  
so can uso linearity wrt sum normally, but not wrt coefficients, need conjugate

### More scalar products at once

**Using different scalar product**

exercise:  
have a scalar product A  
want a base orthogonal wrt both A and the standard scalar product (or any positive defined scalar product)  

solution: pick a base of eigenvectors of A referring to different eigenvalues (pretend for a moment A is a transformation matrix)  
means pick base B' of proof just under  

why it works:  
A is symmetric (self-adjoint wrt standard scalar product), bc defines a scalar product  
so, bc of spectral th, the eigenvectors in different eigenspaces are orthogonal wrt standard scalar product  
since v is an eigenvector of A, can also write, by taking v and w eigenvectors of A:  
$\varphi_A(v,w) = w^t Av = w^t \lambda v = \lambda <w,v> = 0$  


**Simultaneuos diagonaliaztion of 2 scalar products / Find base orthogonal wrt 2 scalar products**

have:  
scalar product not-specified (defined by matrice M: $\varphi _M$)    
scalar product positive-defined $\varphi$ (can be canonical, but not necessarily)  

proof:  

M defines a scalar product, so it's symmetric, so can use spectral th on it  
by spectral theorem: exists a $\varphi$-orthonormal basis B that diagonalizes M (made by eigenvectors of M)  
where $\varphi$ can be any positive-defined product   

thus in base B can write the two products:  
$\varphi_M(x,y) = [x]_{B'}^t \; M \;  [y]_{B'} = [x]_B^t \; D \;  [y]_B  $ (bc diagonalizes M)  
$\varphi(x,y) =[x]_B^t \; I \;  [y]_B$ (bc B' is $\varphi$-orthogonal)  

thus in base B both scalar products are diagonalized simultaneously   
if scalar product is written by a diagonal matrix in some base, means that base is orthogonal wrt it  
so $B$ is orthogonal to both  

### Examples proofs

**Proof: eigenvectors of different eigenvalues of symmetric matrix are orthog wrt standard scalar product**

write hypothesis: B is symmetric  
$<Bx,y>=<x,By>$  

if x and y are eigenvectors of different eigenvalues, $\lambda _x \neq \lambda _y$:    
$\lambda_x<x,y>=\lambda_y <x,y>$  
$<x,y> (\lambda_x-\lambda_y)=0$  
$<x,y>=0$

**Orthogonality wrt another scalar product**

take two eigenvectors of B, with distinct eigenvalues:  
$<x,By> = \lambda _y <x,y>$  
already proved that, in different eigenvalues, $<x,y>=0$  
so $<x,By>=0$ too, using another product  

**If non-canonical scalar product**  

eg. $\varphi _M(Tv, w) = \varphi _M(v, T^*w)$  

must make appear the matrix M describing the scalar product, but have 2 ways:  

1. take away completely the scalar product (and if can, take away v and w from start and end)    

$(Tv)^t \; M  \; w =v^tM \; T^*w$  
$T^t \; M = M \; T^*$  
$T^* = M^{-1}T^t M$  

2. swap to standard scalar product  

$<Tv, Mw> = <v, MT^*w>$  
$<v, T^tMw> = <v, MT^*w>$  (here swap side with $T^t$, not $T^*$, bc i'm already in canonic scalar product, ad $T^*$ refers to product M)    
$T^t M = M T^*$  

**Proof: if semi-defined, isotropic implies in radical**

assume v is isotropic, so $<v,v>=0$  
note that $-v$ is isotropic too, bc $<-v,-v> = (-1)^2<v,v>=0$  
assume <> is semi-definite positive  
want to prove v is in radical, so: $<v,w>=0 \quad \forall w$  

use Cauchy-Schwarz: $<v,w>^2 \; \le \;  <v,v><w,w>$  
since v is isotropic: $<v,w>^2 \; \le \; 0$  
clearly, since it's a square, must be $=0$, cannot be negative  

other useful thing:  
$<v,w>= \frac{1}{2}(<v+w,v+w>-<v,v>-<w,w>)$  

note: Cauchy-Schwarz can be usd if $\varphi > 0$ (defined positive)
