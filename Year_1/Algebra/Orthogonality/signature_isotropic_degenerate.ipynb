### Properties scalar product

**Degenere**

- degenere: $i_0>0$  
- not-degenere: $i_0=0$  

not-degenere means radical is empty  
geometrically, degenere means exist a vector orthogonal to all others (so scalar prod with all others is 0)  
$i_0$ is the dim of the radical (kernel of the scalar product)  

so to prove not-degenere, just prove $det \neq 0$, so must be $i_0=0$  

**Definite**

- indefinite: both $i_+, i_- >0$, don't care $i_0$     
- semi-definite: $i_0>0$ and ether $i_+$ or $i_-=0$  
- definite: $i_0=0$ and either $i_+$ or $i_-=0$  

obv, definite implies not-degenerate  

**Are there isotropic vectors?**

indefinite: MUST exist (don't care about $i_0$)   
semi-definite: MUST exist (bc degenere)  
definite: CANNOT exist  

so isotropic vectors  MUST exist if either:  
- indefinite (don't care if degenerate or not)  
- degenerate  

**Radical vs isotropic**

$i_0$: dimension of the radical  
NOT the number of isotropic vectors  

all vectors in radical are isotropic  
so if degenerate, which means $i_0>0$, since there's stuff in radical there  MUST be isotropic vectors  

not all isotropic vectors are in radical  
so even if $i_0=0$ there might be isotropic vectors (but only in indefinite form, not in definite)  

if it's indefinite:    
$\varphi$ can have both negative and positive results     
thus, for IVT (intermdiate value th), it must touch also the result 0  
so there MUST be a vector with product 0, which is isotropic  

if it's definite:  
definition of definite is that if $<v,v>=0$, then $v=0$  
so there can't be isotropic vectors  

if the product is semi-definite:  
isotropic $\rightarrow$ in radical

in indefinite, some isotropic can not be in radical  

in definite radical is empty anyway, no point asking  

**Not-degenere**

so, given a matrix representing a scalar prod, it's not-degenere if has no 0 eigenvalue  

can be written as:   
- dim ker = 0 (bc radical is kernel)  
- it's full rank (bc means dim ker = 0)  
- det $\neq 0$  

**Example**

imagine 3x3 matrix, where the whole matrix is rank 3, but the 2x2 matrix on top left is rank 1  
so big one is non-degenere, small one is degenere  

obv canonic scalar product is non degenere, bc uses normal concept of perpendicularity  

### Isotropic vectors

#### Definition

$\varphi(v,v) = 0$  

scalar product with itself is null (so orthogonal to itself)   

depends what i define as scalar product, generally defined using a matrix  
isotropic are those that become orthogonal after the transformation with such matrix  

**Explanation**

recall that: $\varphi(v,v) = v^t \; Mv$  

M, the matrix describing the scalar product, can be thougt of like a transformation matrix applied to v  
then check if v after the transformation is orthogonal to the original v, with the canonic inner product      

#### Find isotropic vectors

**Tell by eye if there are**

when given matrix describing $\varphi$:

if not on diag don't mean anything, bc means $<v_i, v_j>=0$, so it's not a norm  

**If have matrix of product**

2. else:

$\varphi(ae_1 + be_2, ae_1 + be_2) = a^2 \varphi(e_1, e_1) + b^2 \varphi(e_2, e_2) + 2ab \varphi(e_1, e_2)$  
in place of the products put the corresponding values from the matrix (not diagonalized is fine)

### Restrictions

if there is a restriction of dimension a such that the scalar product is positive in it, then the matrix has $n_+\ge a$  
the opposite is true too: if have $n_+=a$, means there must be a $axa$ restriction defined positive ( and not a bigger one)

if have a restriction with signature (a,b,c), the full matrix will have $i_+\ge a, i_-\ge b, i_0\ge c$

### Minors

**What is it**

minor of order k: determinant of a $kxk$ restriction   
rank of matrix is the maximum order of non-0 minor  

principal minors: determinants of minors at top left corner of the matrix  

**Properties**

if matrix is rank n, exists a non-0 minor of order n and all those of higher order are 0 (and viceversa)  

if all minors of order k vanish, then every set of k rows or columns is linearly dependent  
so if the det (the biggest minor) vanish, some cols are dependent  



if find restriction nxn of $det \neq 0$, then rank of the full matrix is at least n  

tip: if full matrix has det=0, so it's not full rank  
try to find a minor of size n-1 with non-null det, if can the matrix is rank n-1  

useful for matrices with many vars: to have rank 1, all 2x2 restrictions must have null minore (null det)

if A is a square matrix nxn  
find biggest square submatrix kxk of non-zero determinant  
rank of A is k  

Criterio degli orlati: when checking all the bigger matrices have det=0 i actually only need to check the matrices that contain the kxk matrix with non-zero det that i already found  

a matrix (symmetrical) is defined positive (so it's a scalar product) id all the principal minors are positive

**How to check $i_0=0$**

call A the matrix  

must compute i minori of A (all the square submatrix starting from top left corner, each time growing size of 1)  
if all of them have $det \neq 0$, it means A is invertible, so its nucleus must be banal (which means having dimension 0)  
since $i_0 = dim(ker A)$, then $i_0 =0$
