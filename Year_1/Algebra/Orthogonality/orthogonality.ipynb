### Radical and orthogonal space

#### Definitions

given space V, with subspace W:  

- radical: $V^{\perp}$  
set of all vectors of V orthogonal to all other vectors of V, even themselves    
(don't need W)  

- orthogonal space: $W^{\perp}$   
set of vectors of V orthogonal to all vectors of W  

so the difference is if the space the vectors must be orthogonal to is:  
the ambient space (radical)  
a subspace (orthogonal space)  

so if asked $V^\perp$, check if V is the ambient space or a subspace  

**Radical and isotropic vectors**

in radical: orthogonal to all others in space (itself too)  
isotropic: orthogonal to itself  

- is in radical $\rightarrow$ isotropic  
ALWAYS  
(bc to be in radical must be orthogonal to itself too)  

- is in radical $\leftarrow$ isotropic  
ONLY IF $\varphi \le 0$ or $\varphi \ge 0$  
(means product must be defined or semidefined, in whichever way, but can't have values both negative and positive)  

the second one means the radical is the set of all isotropic vectors  
else thare can be isotropic vectors not in radical, orthog to itself but not to all else   

**Proof: if not semidefined, implication isotropic->radical is not true**

note: if scalar product is indefinite, there must be isotropic vector  
but can have $i_0=0$, meaning the radical is empty (indefinite has no restriction on value of $i_0$)  
so if it's indefinite, there can be isotropic vectors not in radical  

**Proof: if semi-defined, isotropic implies in radical**

assume v is isotropic, so $<v,v>=0$  
note that $-v$ is isotropic too, bc $<-v,-v> = (-1)^2<v,v>=0$  
assume <> is semi-definite positive  
want to prove v is in radical, so: $<v,w>=0 \quad \forall w$  

use Cauchy-Schwarz: $<v,w>^2 \le <v,v><w,w>$  
since v is isotropic: $<v,w>^2 \le 0$  
clearly, since it's a square, must be $=0$, cannot be negative  

other useful thing:  
$<v,w>= \frac{1}{2}(<v+w,v+w>-<v,v>-<w,w>)$  

**Example isotropic and radical**

$\begin{bmatrix} 1 & 0 \\ 0 & -1  \end{bmatrix}$  
(if represents a space matrix it's called the hyperbolic space, but now used it to represent a scalar product)  
is not semidefined  
means to consider matrix 'flip wrt y axis' as defining the scalar product  

bisectors are isotropic  
but are not orthogonal to all, so are not in radical  
this is possible bc the product is not semidefined  

formally: bisector is [1,1]  
after applying transformation of the scalar product, becomes [1,-1]  
thus the inner product netween before and after is 0, it's isotropic  

#### What they are

- radical: $ker (M)$   
- orthogonal space: $ker(G^t M)$  

where:  
$M = M(\varphi)$, the matrix describing the scalar prod  
$G$: generator matrix of W   

M and G must be expressed in the same base (generally canonic one)  

remember that  
dim(ker) = number of cols- number of pivots = number of cols without pivots,  
NOT number of zero rows  

**How to remember it**

radical: whatever the other vector, the scalar product must give zero  

orthogonal space: scalar product with any vectors OF SPACE W gives zero, (so on left must put a vector from W, not leave it empty to mean any vector)

**Common stuff to search**

if asked dimension or base of the radical, simply compute dimension or base of the kernel of M or of CM, since that's what they are  

#### Properties

**Orthogonal space**

always:  

$dim(W^{\perp}) = n - rank(MG)$    
bc orthog space is a radical, so dim is n-rank  
$W^{\perp}$ is subspace of V, even if W is not a subspace itself 

**Dimension formula**

if $\varphi |_W$ not degenere ON THE RESTRICTION (so for eg if $\varphi$ is defined positive):  

$(W^{\perp})^{\perp} = W \quad$  
$V = W^{\perp} + W$ (direct sum, useful to take dimensions of thoes three things)  
$dim(W^{\perp}) + dim(W) = dim(V)$

if scalar product is degenere:  
$dim(W^{\perp}) + dim(W) = dim(V)  + dim(W \cap V^{\perp})$  

to find base of $W^\perp$, remember a vector in $W^\perp$ must be orthogonal to all vectors in base of W  

so make a linear system with the condition:  
$<v,b_1>=0, <v,b_2>=0 ...$  

**Proof**

if $\varphi _{|W}$ not degenere, $W \oplus W^\perp = V$  

1. $W∩W^⊥={0}$   

since $\varphi _{|W}$ not degenere, no vector in W is orthogonal to all vectors of W  
so $W∩W^⊥={0}$  

2. $W+W^⊥=V$  

take $v \in V$  
$e_1, ... e_m$ are orthonormal base of W  

define:  
$w = <v,e_1>e_1 + ... + <v, e_m>e_m$  
$u = v- <v,e_1>e_1 - ...- <v, e_m>e_m$  

can say:  
$v = w+u$  

i see that:  
$w \in W$ (bc linear comb of base of W)  
$u \in W^\perp$ (bc scalar prod with all vectors in base of W gives 0, since base of W is orthogonal)    
$<u, e_i> = <v, e_i> - <v, e_i> =0$   

thus $V=W+W^\perp$
