### Definition

$(i_+, i_-, i_0)$  

they represent number of positive, negative, or null eigenvalues  
sum of them is the dimension of the space  

**What each value mean**

$i_0 = dim(V^{\perp}) = dim(ker S)$  
so used to tell degeneracy (S is matrix describing scalar product)  

$i_+$ is the dimension of the biggest subspace of V i can pick where the scalar product is defined positive  
same for $i_-$  

**Vectors of specific norms**

values of signature they define norms of vectors that can be found  
if $i_- >0$, means there is at least a vector with norm <0  

the vector with a negative norm is exactly the eigenvector with the negative eigenvalue, bc if $\lambda <0$, its eigenvector:     
$\varphi(\lambda v, \lambda v) = \lambda \overline{\lambda} \varphi(v,v) = ||\lambda ||^2 v^t M v = ||\lambda ||^2 \, v \, \lambda v = \lambda \, ||\lambda ||^2 \, ||v||^2$  

the canonic product gives positive norm $||v||^2$, so the $\lambda <0$ makes the $\varphi$ product neagtive  

**Other def**

number of pos, neg, null norms taken on a base orthogonal wrt that product  
so if have orthogonal base, compute norms of all vectors to have signature  
bc in orthogonal base, scalar product is written by a diagonal matrix, and stuff on the diagonal is the norm of each vector  

### Compute with shortcuts

**Shortcut to find signature**

1. check main diagonal

2. check restrictions  

3. check determinant    


**Tip**

if i find only one useful restriction, so i'm not totally sure about the signature:  
finding a restriction gave me the sign of 2 eigenvalues  
if it's a 3x3 matrix, compute det, which is product of all eigenvalues  
from sign of det can know sign of final eigenvalue  

**Studying main diagonal**

can be done in any form, don't need a diagonal matrix  
if in main diagonal there is at least a:  
- positive entry: $i+\ge 1$  
- negative entry: $i_- \ge 1$  
- null entry: CANNOT conclude anything, might still be $i_0=0$  
bc the 0 means there is an isotropic vector, but that vector is not necessarily in the radical (it is only if product is semidefined)  

eg: if on main diagonal have 2 positive and 1 negative entries, put $i_+ \ge 1, i_- \ge 1$  
only care about having at least one entry, not how many there are, it won't tell me anything more  

**How to check restrictions**  

better to check 2x2 restrictions  
to make a restriction: pick 2 rows and 2 cols, even not close, make a matrix with the 4 cells where they intersect  
note: must pick cols are rows referring to same index (so can pick 2x2 square in top left, but not 2x2 square in top right, that's not a restriction)  

if the 2x2 restriction has:  
- positive determinant and positive entries on main diagonal     
then $\varphi > 0$, which means signature $(2,0,0)$    

- negative determinant   
then signature is $(1,1,0)$ (bc check determinant rules, it's negative and non-zero)  

if get signature of a restriction, full matrix must have in signature each term at least like that  
so if find restriction $(2,0,0), (1,1,0)$, full matrix must have at least $i_+ \ge 2, i_- \ge 1$   

**Tip restriction**

if a restriction has some value of $i_+$ pr $i_-$; the full matrix must have at least that  
but this isn't true for $i_0$: might have a restriction with %i_0=1$, but the full matrix with $i_0=0$  
so the restriction degenerate, but the full matrix non-degenere  

**Check 2x2 restrictions**

$\begin{bmatrix} a & b \\ c & d \end{bmatrix}$  

$det >0$    
$a >0$: positive definite (2,0,0)  
$a <0$: negative definite (0,2,0)    

$det <0$: indefinite (1,1,0)  

$det = 0$  
$a>0$: semidefinite positive (1,0,1)  
$a<0$: semidefinite negative (0,1,1)  

bc get a possible value for the norm from a  
then, according to the sign of the determinant, check number of positive and null eigenvalues  

if $a=0, c\neq 0$ can use c insetad of a   

**Check rank**

if full-rank (so not-degenere), $i_0 =0$, else $i_0 = dim (ker S)$   
if full-rank: can check sign of det to know if $i_-$ is even or odd  (if not full rank would just give me 0, so no informations)  
if not full-rank: compute $i_0 = dim(ker \; S)$, where S is matrix describing the scalar product  

**If matrix describing the scalar product is diagonal**

can count the signs of the stuff in main diagonal, since they're the eigenvalues  

**Signature and determinant**

determinant is the product of the eigenvalues  
therefore compute it and:  

- det =0  
there is at least one 0 eigenvalue, so $i_0 >0$  

- $det \neq 0$  
there are no 0 eigenvalues, so $i_0 =0$    

- det > 0  
there is an even number of negative eigenvalues, so $i_- = even$  

- det < 0  
there is an odd number of negative eigenvalues, so $i_- = odd$   

### Compute properly

**Methods**

can either:  
1. symmetric reduction  
2. find eigenvalues

**Symmetric row-reduction**

what to do:  
at each step must do 2 operations: if do first row - second row, right after must do first col- second col  
to subtract cols, use the new col obtained after deleting rows  
essentially it must keep being symmetric 
this way can row-reduce the matrix until it's diagonal  
also, if want to swap 2 rows, after must swap the 2 cols with that index too  

note:  
note that, to find radical, thus treating it like a kernel, do normal row-reduction, not this symmetric thing

what i get:  
stuff in main diagonal is NOT the eigenvalues  
bc i found a congruent matrix, for which eigenvalues are not invariant, but the signature is  
so the signs are correct to compute the signature, but are not eigenvalues    

**Example symmetric row-reduction**

Delete:  

$\begin{bmatrix} 0 & -1 & -2 \\ -1 & -3 & -5 \\ -2 & -5 & -8 \end{bmatrix}$  

for one step, do in sequence:  
third row - 2 times second row  

$\begin{bmatrix} 0 & -1 & -2 \\ -1 & -3 & -5 \\ 0 & 1 & 2 \end{bmatrix}$ 

third col - 2 times second col  

$\begin{bmatrix} 0 & -1 & 0 \\ -1 & -3 & 1 \\ 0 & 1 & 0 \end{bmatrix}$ 


Swap:  

$\begin{bmatrix} 0 & -1 & 0 \\ -1 & -3 & 0 \\ 0 & 0 & 0 \end{bmatrix}$   

swap rows 1 and 2  
$\begin{bmatrix} -1 & -3 & 0 \\ 0 & -1 & 0 \\ 0 & 0 & 0 \end{bmatrix}$ 

swap cols 1 and 2  
$\begin{bmatrix} -3 & -1 & 0 \\ -1 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix}$ 



**Split in orthogonal spaces**

if can split in orthogonal spaces/restrictions, can compute signatures of both, total signature is the SUM  

eg. $\begin{bmatrix} 0 & 1 & 0 \\ 1 & 0 & 0 \\ 0 & 0 & 1 \end{bmatrix}$  
restrictions $\{e_1,e_2\}$ and $\{e_3\}$ are orthogonal, and have signature (1,1,0) and (1,0,0)  
so full signature is the sum (2,1,0)  

**Cartesio theorem**

if just want signature, no need to actually compute the eigenvalues   

find characteristic poly:  
$p(x) = a_nx^n + ... a_m x^m$  
m is the last term with non-null coefficient  

$i_0$ = m (bc means can raccogliere by $x^m$, so m is algebric multiplicity of 0 as a root)  
$i_+$: number of sign changes in the sequence $a_i$ (ignoring null coeffs)  
$i_- = n - i_0 -i_+$  

### Isotropic space

when asked isotropic space, means totally isotropic space  
def is that the matrix describing the scalar product is a 0 matrix  

**Totally isotropic space**

a space where every vector in it is orthogonal to all other vectors in it (itself too)  
so all vector in it are:  
- isotropic (bc orthogonal to itself) 
- not necessarily in radical, bc need only to be orthog to all vectors in this subspace, not in the ambient space  


is a space where the matrix describing the scalar product in it (so a restriction of the full scalar product matrix) is a 0 matrix  

so when computing norm of any vector with the method of expanding $\varphi(ae_1+be_2, ae_1+be_2)$ will get all addends 0  

**Depending on scalar product**

defined: has no isotropic vectors, so has neither radical not isotropic vectors  
semi-defined: in radical = isotropic, so radical is biggest isotropic space  
indefined: can have isotropic vectors not in radical, so can find isotropic space bigger than radical  

**Find base for isotropic space**

to find base for it, pick as many vectors i need, linearly independent and isotropic  

if want space of dimension N, take:  
- pick base from radical  
since radical is an isotropic space  
- if dim(radical)<N, need to find others  
if only miss 1, pick an isotropic one, since will be orthog to all other bc they're in radical anyway  
(only the bottom-left entry of the amtrix is left to set to 0) 

isotropic space is less demandin than radical, so can even find one bigger than radical  

**Find biggest isotropic space in indefined product**  

have scalar product of signature $(i_+, i_-, i_0)$  

Witt index: $r= min(i_+,i_-)$  
r is the maximum dimension of an isotropic space in that scalar product (it MUST exist an isotropic space with this dim), ONLY considering the degenerate part  
thus, to it, must add the space made by the radical  

eg. (1,1,1)  
radical has dim=1, Witt index = 1  
add them, so biggest isotropic space has dim=2  

Witt index only needed in INDEFINED product, bc in semi-defiend the biggest isotropic space is the radical, bc isotropic implies being in radical  

**How to find isotropic vectors**

take random x vector (with $x_1, x_2...$ as components)  
set condition: $\varphi _A(\overline{x}, \overline{x}) = 0$  

will get equation with x components, that is condition for vector to be isotropic
