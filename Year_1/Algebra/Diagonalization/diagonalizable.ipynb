### Definition diagonalizable

**Intuitive**

A is diagonalizable if exist P that diagonalizes it:  
$D=P^{−1}AP$  

- diagonalizing matrix: P  
has base of eigenvectors as cols  

- diagonal matrix: D  
has eigenvalues as entries on main diag (each as many times as its multiplicity)  

- diagonalizable matrix: A  

to remember the order: A is the matrix that, read with base P, gets diagonalized to be D  
the other verse is obv: $A=PDP^{−1}$  

**Formal**  

$\varphi: V \rightarrow V$ is diagonalizable if  
$\exists B: M_B(\varphi)$ is diagonal  

so, if the matrix can be made diagonal by writing it in some base  
SUCH BASE IS MADE OF THE EIGENVECTORS (obv, since each vector gets scaled by the eigenvalue when applying the transformation)

### Row-reduction

**Diagonal vs diagonalizable**

diagonal matrix is any matrix where entries not on main diagonal are 0    
don't care how i got it  

generally there are 2 types of diagonal matrix starting from matrix A:    
- matrice a scalini di A  
obtained from A with row-reduction  

- diagonal matrix simile to A (matrice diagonalizzata associata ad A)  
D: diagonal matrix of A (or diagonal matrix simile to A?)   
P: diagonalizing matrix of A (which allows me to do the diagonalizing)  

every matrix can be turned in its form a scalini  
but not every matrix is diagonalizable    
even if it is diagonalizable, D will be different from the matrice a scalini of A  

**Row-reduction**  

can only be done on matrices that describe VECTOR SPACES, not transformations or scalar products  
bc change the transformation itself (easy to visualize with scaling), so all the properties (signature, eigenvalues) change too  

so DIAGONALIZING a matrix DOES NOT MEAN DOING ROW REDUCTION, which must never be done  
diagonalizing means finding eigenvalues and writing diagonal matrix with those on the diagonal  

**When can i do row-reduction on trasformation matrices?**  

only ok to compute: rank, determinant (but remember to keep track of swaps and scalars)  

for eigenvalues, can only do it **AFTER SUBTRACTING LAMBDA** FROM THE MAIN DIAG

### Is it diagonalizable?

**Preliminary checks**

1. only square matrices can be diagonalizable, else the concept has no sense  
2. must specify in what field, R or C

**Conditions for diagonalization**

both must be respected:  

1. eigenvalues must all belong to chosen field  
2. for each eigenvalue, algebric multiplicity is equal to geometric multiplicity  

**Shortcut**

what stuff is certainly diagonalizable  

1. symmetric/hermitian matrices (self-adjoint)

2. matrix of order n with n distinct eigenvalues  
eg. $T^n = I$    
eg. when minimum/characteristic poly has all exponents 1  

3. if A is diagonalizable, any polynomial function of it is too  
$B(A)= A^2 + 3A + I \quad$, B is too  
but the opposite is not always true, if B is diagonalizable not necessarily A is too  

**Multiplicity of eigenvalues**

where $\lambda$ is a specific eigenvalue  

- algebric: $m_a(\lambda)$  
the exponent of the bracket that has it as root in the characteristic poly  
so the multiplicity of $\lambda$ as a root of the characteristic poly  

- geometric: $m_g(\lambda)$  
$dim \; V_{\lambda}$ (dimension of its eigenspace)

alternative def of diagonalizable: it is if the sum of geometric multiplicites of all eigenspaces is equal to the size of the matrix  
bc means can make a base of eigenvectors for the matrix, that thus diagonalizes it  

**For which parameter it's diagonalizable?**  

k is the param  

1. compute characteristic poly  
2. find eigenvalues, some may be a function of k  

depends on case:  

- for values of k where all the eigevalues are distinct: it's diagonalizable  
- for values of k where some eigenvalues have algebric multiplicity >1: must check by hand, compute geometric multiplicity only of the eigenvalue that appears more than once, check if it's the same as algebric multiplicity  

**Notes on diagonalization**

rank cannot tell me anything about being diagonalizable, nor viceversa  
bc having a 0 eigenvalue make the matrix not-full rank, but can still diagonalizable  

**Lemma: restriction to invariant subspace**
 
the restriction of a diagonalizable transformation to an INVARIANT SUBSPACE (invariant wrt that trasnformation itself) is diagonalizable too  
essentially, if A is diagonalizable, means the blocks that describe each eigenspaces are diagonal too  

**Condition: diagonalizable on C**

must still respect the 2 conditions  
the eigenvalues are certainly in field C (bc R is part of C anyway)  
but the multiplicities might not be the same (thus, have not enough linearly independent eigenvectors)  
so, not all matrices are diagonalizable in C  
but all are triangularizable in C  

##### Explanations

**Why need to specify field to check diagonalization**

number of eigenvalues, each one counted as many times as its multiplicity, must be equal to rank of matrix  
means that if have complex eigenvalues can't diagonalize on R, bc those don't count    
this condition is certainly satisfied if the required field is C, bc it's not like the roots can belong to a bigger field (formally, bc C is a field algebrically closed, so all poly with coeffs exprimibile in C have roots in C)     

so some matrices are diagonalizable in C, but not in R  
eg the 90 deg rotation is not in R, but it is in C  
still, not everything is diagonalizable in C, there is always the second condition to check  

**Why is it diagonalizable if all eigenvalues are different**

geometric multiplicity of an eigenvalue is:  
- at least 1  
- not bigger than its algebric one (so equal or smaller)  

$m_a(\lambda) \ge m_g(\lambda) \ge 1$  

so if an eigenvalue has algebric mult one, geometric mult is also one  
that's why, if all eigenvalues have algebric mult 1 (so if for nxn matrix get n different eigenvalues), def all of them have geometric mult 1, so matrix is diagonalizable  

### Triangularizable

**Lemma: strong triangularization**

for all A, exists P unitary such that:  
$P^*\; A \; P = T$  
where T is triangular  

**Definition**

a matrix is triangularizable in K if respects the first condition of diagonalization:  
all eigenvalues must belong in K  

in such case can be written as a triangular matrix with eigenvalues on the diagonal (when the base is made of eigenvectors)  
if it's diagonalizable, triangular and diagonal form are the same (the base to triangularize it actually diagonalizes it)  

to find base that triangularize it:  
find base for each of the eigenspaces i found, won't be enough vectors  
then pick some other vectors to complete it  
to make sure it's triangular, check where these vectors are mapped, secondo the base i'm picking  
eg: pick as last vector $v_n$, such that $f(v_n)=v_1$, so it's (1,0,0)


### Spectral theorem

**Spectral theorem**

if T is self-adjoint wrt a POSITIVE DEFINED scalar product $\varphi$ (so not necessarily the canonic one), then:  
- T is diagonalizable  
- all eigenvalues are real   
- exists an orthonormal (wrt $\varphi$) base of eigenvectors 
- eigenvectors of different eigenspaces are orthogonal wrt $\varphi$  
- T gets diagonalized by a matrix unitary wrt $\varphi$

as $\varphi$ must use the scalar product used in that field, cannot just use the canonical one bc i feel like it  
so to check if it's self-adjoint, if i'm using a weird scalar prodct, must use full formula of self-adjoint where the scalar product appear  

useful formulation for proofs:  

if:  
M autoaggiunto  
$\varphi$ positive defined (eg. canonical, but not necessarily)  

by spectral theorem: exists a $\varphi$-orthonormal basis B that diagonalizes M (made by eigenvectors of M)  
$\varphi_M(x,y)=[x]_c^t \; M \;  [y]_c = [x]_{B}^t \; D \;  [y]_{B} $  


$\varphi$-orthonormal means, in that base, the matrix describing $\varphi$ is the identity  
$\varphi(x,y) =[x]_B^t \; I \;  [y]_B$
