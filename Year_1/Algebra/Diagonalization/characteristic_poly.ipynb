### Characteristic poly

#### Definition

$p_A(\lambda) = det(A- \lambda I)$ 

sometimes it's written $p(x)$ instead of $p(\lambda)$  
polinomio monico: when coefficient of term of highest degree is 1  

its roots are the eigenvalues of A  
means, if $\lambda_i$ is an eigenvalue, $p_A(\lambda _i)=0$  
if it's not an eigenvalue, $\neq 0$ 

**Example**

if 2x2 with eigenvalues 2,3, have as char poly:  

$p_A​(λ)=det(A−λI)=(2−λ)(3−λ)=λ^2−5λ+6$  


so the char poly gives 0 if plug either:  

- an eigenvalue of A  
see from multiplication of brackets  

- A itself (called evaluating the poly on the matrix A)  
see from sum of addends, $A^2-5A+6=0$  

**Compute**  

1. put $-\lambda$ in entries in main diag  
2. compute determinant (any method)  

answer can be written either as a sum of monomi (normal polinomial) or multiplication of binomi (to show roots)   

#### Properties

**Hamilton-Cayley theorem**

$p_A(A)=0$  

means the characteristic poly of a matrix A, evaluated in that matrix A (so when plugging A in place of x) gives 0 (a null matrix)  

**What can i plug in it**

plug in a scalar number $\lambda$: give 0 if that number is eigenvalue, non-zero else  
plug in the matrix itself $A$ (evaluate on the matrix): always give 0, bc of Hamilton-Cayley th  

**Examples**

eg: if $\lambda - 1$ is the minimum or characteristic poly, then $A-1I=0$   
thus, if minimum poly is of rank 1, then $A -kI = 0$, so A is a constant matrix  
thus, in exercises can easily exclude it  

if A is diagonalizable in R:  
yes: $(x+1)^2$  
no: $(x^2+1)$, bc would give complex eigenvalue 

**Stuff i can find from it: det, trace**

expand the char poly: $p_A(x) = det(A-xI)= (-1)^n x^n + a_{n-1}x^{n-1} + .. +a_0$  

- determinant  
$det(A) = p_A(0)$  
obv, bc $p_A(x) = det(A-xI)$   
also, bc det is the product of all the roots, $det = \prod \lambda _i$  

- trace  
$tr(A) = (-1)^{n-1}a_{n-1}$   
bc it's the sum of all the roots, $tr = \Sigma \lambda_i$  
studying poly, i know relationship between roots and coeffs (like $x_1 + x_2 = b, x_1x_1 = c$ )  
if n-1 is odd change sig, if even keep it  

**Invariancy**

characteristic and minimum poly are geometri properties of the transformation, so don't change with a base change(invariant per similitude)  
they're multiplication and sum of the eigenvalues (each taken as many times as its algebric multiplicity)  
even if the matrix is not diagonalizable (think of the jordan form instead)

to do a base change on a matrix, swap both rows and cols with those indexes  
to swap vectors 2 and 3: first swap rows 2 and 3, then swap cols 2 and 3  

### Minimum poly

**Definition**

smallest divisor of the characteristic poly, such that $p(A) = 0$  
so if have something like $p(x)=...$, the minimum poly is the one with the smallest exponent for which it's still true $p(A)=0$  

**What it has**

in it must appear ALL THE EIGENVALUES of the poly (don't know with what exponent, but they must appear)  

**Notes**

characteristic poly is a poly of degree equal to size of the matrix  
minimum poly can have smaller degree   

exponents are NOT the geometric multiplicity for that eigenvalue, but the size of the biggest jordan block  

**To compute**

must appear all brackets that appear in char poly, just change the exponents  

if given matrix, first compute geometric mult of all eigenvalues to check if it's diagonalizable  

- if matrix is diagonalizable  
can just put ALL EXPONENTS OF THE BINOMI of the characterstic poly to 1  
(this is an if and only if condition, so if i can do this it's diagonalizable)  

- if matrix is NOT diagonalizable  
exponents are not simply 1, but are the size of the biggest Jordan block for that eigenvalue  
so need to compute it in the long way shown below  
if matrix is not diagonalizable, exponents CANNOT be all 1, there MUST be at least a 2 (so if there is only one exponent that could be 2, put that one to 2)

**Compute minimum poly if not diagonalizable**

1. write the characteristic poly as a multiplication of binomi  
2. for each binomio, must find new exponent: starting from 1, see if when plugging in A gives null matrix  

**Compute minimum poly if not diagonalizable: explanation**

write the characteristic poly as a multiplication of binomi  
eg: $(\lambda -1 )^3 (\lambda -2)$  

for each binomio, the exponent must be at least 1, but could be more  
possibilities: $(\lambda -1 )^3 (\lambda -2), \quad (\lambda -1 )^2 (\lambda -2),  \quad (\lambda -1 )(\lambda -2)$  

how to pick correct exponent:  
by definition, $p(A) = 0$  (note that this is $p(\lambda)$, so means plug A in place of $\lambda$)  
so, starting from the option where the exponent is the smallest, find the matrix  
$(\lambda -1 )(\lambda -2) \rightarrow (A- I)(A - 2 I)$  
...  
$(\lambda -1 )^3(\lambda -2) \rightarrow (A- I)^3(A - 2 I)$  

the first one i find where the matrix is a zero matrix (all zeros) is the minimum poly  


### When given expression

**Turn expression in multiple of minimum poly**

eg: $A^5 = 2A^3 + I$  
$q(x) = x^5-2x^3 -1$  

$q(x)$ is a MULTIPLE OF THE MINIMUM POLY   
(bc, when swapping A with x, gives 0)    
it's a multiple of the minimum poly bc, by definition, the minimum poly is the smalllest where $p(A)=0$

write formally:  $p_A(x) \; | \; q(x)$   
$p_A(x)$: minimum poly of A  
|: is a divisor of  


**What to do with multiple of minimum poly**

if have $q(x)$ which is a multiple of the minimum poly, and if it's the same rank as the characteristic poly, don't mean it's the characteristic poly  
bc characteristic poly is too a multiple of the minimum poly, but might be a different multiple  
assume minium poly is $x(x-1)$. could have char poly $x^2(x-1)$ and another random poly, multiple of the minimum poly, $x(x-1)^2$, so they wouldn't be the same

**Prove A it's diagonalizable**

if $q(x)$ only has factors with exponent 1, so does the minimum poly  
theory says that if the minimum poly has all exponents 1, it's diagonalizable  
however, still can't tell the characteristic poly, just the minimum one  

**Evaluate polynomial on a matrix**

means x plugged in p(x) is a matrix  
so compute poly pluggin   
x=matrix passed  
1=identity matrix of same shape of the one passed

$p(x) = 2+x-3x^3$  
$p(A) = 2 Id + A -3A^3$  


**Find eigenvalues of A**

it's a multiple of the minimum poly, so all eigenvalues must appear here   
but not all its factors must be in the minimum poly as well  

so find roots of q(x)  
some of those (not necessarily all) are roots of A, with algebric mult at most the one thay have in q  

rem:  
eigenvalue $\longrightarrow$ root of q(x)  
but not necessarily the opposite  
bc q(x) is a multiple of the minimum poly, can have added stuff too

**Find determinant of A**

starting from original expression, don't transform to poly:    

1. take det on left and right  
2. simplify with rules for determinant computation  

eg: $A^3 = A \rightarrow det(A^3) = det(A)$  

eg: $A^3 = 2 \rightarrow det(A^3) = det(2I)$  

### Exercises

**Base of eigenvectors**

if make a base of eigenvectors can write random vector as linear comb of eigenvectors:  

$f(\alpha _1 v_1 + ... + \alpha _n v_n) = \lambda _1 \alpha _1 v_1 + \lambda _n  \alpha _n v_n$  

**Common reasonings**  

M is symmetric, so:  
- can be diagonalized WITH A ROTATION (unitary matrix)  
- eigenvalues are all real  

M is defined positive ($<M,M> >0$):  
- eigenvalues are all positive  

**Diagonalize matrix with rotation**

M symmetric, exists N such that $M = N^2$ (essentially exists square root)  

since $M = N^2$: (can write $P^t$ instead of $P^{-1}$ since M is symmetric and is diagonalized by unitary matrix)    
$M = P^t D P$  
$N = P^t \sqrt{D} P$  

just compute and check $N^2 = M$  

**How to examine $M^2$, check if diagonalizable**

2 methods:  
1. just compute $M^2$ and check if it's diagonalizable like always  
2. try to write diagonal M, then M^2  

examine method 2:  

if M is diagonalizable, $M^2$ is as well  
just write M it in diagonal form  
then to get $M^2$ just square all numbers on main diagonal  

if M is NOT diagonalizable, $M^2$ still might  
find eigenvalues of M, if it's not diagonalizable means don't have enough eigenvectors   
so write M in a base of the eigenvectors, completed by random vectors



**Example method 2**

M, a 3x3 matrix of rank 2, not diagonalizable   
found eigenvalues:  
- 0, $m_a=2, m_g=1$  
- 3, $m_a=m_g=1$  

since both have $m_g=1$, from both get an eigenvector  
since M is 3x3 need one more random vector to create a base  
write M in such base:  

$M = \begin{bmatrix} 3 & 0 & 0 \\ 0 & 0 & x \\ 0 & 0 & 0 \end{bmatrix}$  

the first two cols are the eigenvectors (one with eigenvalue 3, one 0)  
the last col is the random vector  
since M is of rank 2 (checked at the start) this M must be too (bc rank in invariant wrt similitude/base change), that's why i added that x there, to make it rank 2  

now write $M^2$:  
$M^2 = \begin{bmatrix} 9 & 0 & 0 \\ 0 & 0 & 0 \\ 0 & 0 & 0 \end{bmatrix}$  
can do intuitively, just see where each vector gets mapped by applying M twice  
now it's diagonal, so $M^2$ is diagonalizable  

**Eigenvectors of different matrices**
