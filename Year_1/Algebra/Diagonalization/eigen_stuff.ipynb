### Eigenvalues, eigenvectors

**Definition**

$\exists \vec{v} \in V : \quad \varphi(\vec{v}) = \lambda \vec{v}$  

$\lambda$: eigenvalue  
$\vec{v}$: eigevector  

eigenvalues are the same no matter what base the transformation matrix is written in (bc invariant wrt similitude)  
eigenvector changes, bc geometrically stays the same vector, but is indicated in different coords  

**Notes on 0**

$\vec{0}$ CANNOT be an eigenvector  
would make no sense, since technically everything scales it  

0 CAN be an eigenvalue  
its eigenvectors are all those in the kernel, so those mapped to the null vector  
so multiplicity of 0 as eigenvalue is dim of kernel  

**Properties of eigenvectors**

- independence:   
    - if in different eigenspaces (so refer to different eigenvalues): are always independent  
    - if in same eigenspace (so refer to same eigenvalue): an eigenvalue has as many independent vectors as its geometric multiplicity  
    it's exactly the def of geometric multiplicity, how many eigenvalues to form the base of that eigenspace  

- orthogonality  
  - if in different eigenspaces: only if the transformation matrix to which they belong is autoaggiunta (SYMMETRIC)  
  same condition as orthogonality of eigenspaces  
  said by spectral th  
  - if in same eigenspace: no garantee, must check  

**Complex eigenvalues**

remember complex eigenvalues appear in conjugate pair ONLY IF MATRIX HAS ONLY REAL ENTRIES  
bc the reason is that the sum of eigenvalues = the trace, and if the trace is not complex, the sum of eigenvalues can't be either  
in this case, if dimension is odd, like $R^3$, there must be a real eigenvalue (which can be null)  

anti-symmetric matrices (with real entries) have all purely complex eigenvalue (no real part), so if of odd dimension one of them is 0  

#### Proof notation

**Eigenvector notation for proofs**

can be written either:  
$f(v)=\lambda v$  
$v \in span(v)$  

**How to write it in proofs**

$Tv = \lambda _1 v$  
means either:  
- v is an eigenvalue 
- v is a null vector  
seems dumb, but better to explicitly exclude this option  

can use normal algebra, so can be written:  
$Tv -\lambda _1v = (T-\lambda _1 I)v = 0$  

also:  
$T^k v = \lambda _1 ^k v$  


### Compute eigenvalues

remember that complex eigenvalues must always appear in pairs  
can't have only one, need both $\pm$  

**Compute eigenvalues**

1. find characteristic poly  
  $p(x) = det(A- \lambda I)$  

2. put it equal to 0, find roots  
roots are the eigenvalues  
pay attention to algebric multiplicity   

**0 as an eigenvalue**  

if there is a zero row or col: 0 is an eigenvalue  
but 0 can be an eigenvalue even if there is no zero row  
just need a vector combination that gives the null vector, like $v= e_1-e_2, \quad f(v) = \vec{0}$    

the eigenspace corresponding to the 0 eigenvalues is the kernel  

**Tips**

if know all but one eigenvalues, can use that trace is sum of eigenvalues to compute the last one  

### Compute eigenvectors

**Method 1 (better)**

1. get matrix describing the eigenspace of the eigenvalue i want the eigenvector of   
(so just, in the matrix used to find eigenvalues, plug in $\lambda$, the eigenvalue i want)  

2. find base of eigenspace  
(rem eigenspace is the kernel, so find base of the kernel)      

**Method 2**

1. find eigenvalues  
2. plug each into the definition of eigenvector: $A \vec{v} = \lambda \vec{v}$  
3. do the equation, will get conditions on x,y,z  
solve to get eigenvector  
note that all multiples of an eigenvector are eigenvectors too, so write it in the simplest form  

equation to solve:  

A, $\lambda$ i already know  
in the vector write coords  
$\vec{v} = \begin{bmatrix} x \\ y \\z \end{bmatrix}$  

**Find eigenvector by eye**

looking at associated matrix  
find a vector v, linear comb of the canonic ones, for which $f(v) = \lambda v$  
eg: $v=e_1+e_2-e_3 \quad \rightarrow \quad f(v) = f(e_1)+f(e_2)-f(e_3)=\lambda v$  

**How many eigenvectors**

for each eigenvalue need to write as many as its geometric multiplicity, NOT algebric, to make base of the eigenspace    
bc geometric multiplicity is the dimension of its eigenspace  

rem eigenvectors in same eigespace are not always independent, so check they really are a base for that eigenspace  

**Complex eigenvalues (if A has real coeffs)**

eigenvalues are always present in conjugate pairs (bc come from a second degree poly in tha charatcreristic poly)  
if $v$ is the eigenvector wrt $\lambda = a +ib$, $\quad \overline{v}$ is the eigenvector wrt $\overline{\lambda} = a -ib$  

proof:  
$A \overline{v} = \overline{Av} = \overline{\lambda v} = \overline{\lambda} \overline{v}$  
can do it the first equality bc A has real coeffs  
so only if A has real coeffs, or always???  

**Find orthogonal base for eigenspace**

found base of eigenspace $v_1, v_2$, but they're not orthogonal   
to find orthogonal base, orthogonalize with G.S.:  
$w_1 = v_1$  
$w_2 = v_2 - \frac{<v_2,w_1>}{<w_1,w_1>}w_1$  

bc $\{v_1, v_2\}$ are a base for the eigenspace  
not just turn them into orthogonal base (of same space)  

if matrix is symmetric, eigenvectors of other eigenspaces are already orthogonal  

### Eigenspaces

#### Definition

$V_{λ_1} =  ker(M − \lambda _1 I)$  

IT'S A KERNEL  
so for stuff like finding base or dimension/base treat it like a kernel  

**Definition**

$V_{\lambda} = \{ \vec{v} \in V : \varphi(\vec{v}) = \lambda \vec{v}\}$  
space of all the vectors corresponding to an eigenvalue (that get scaled by it when doing the transformation)

so if eigenvalues are 1, 3, can define $V_1, V_3$  
def: $V_{\lambda}$ is the span of the eigenvectors with eigenvalue $\lambda$  

**Generalized eigenspace**

$\overline{V}_\lambda$  

an eigenspace is a kernel  
when applying the matrix M several times, the kernel can only grow, not shrink  
if it don't grow once, then it won't grow ever again, even if keep applying M  

so the generalized eigenspaces are the EIGENSPACES AFTER HAVING FINISHED GROWING  

so the diagonalization condition $m_a = m_g$ for every eigenspace can also be written:  
each eigenspace must be equal to its generalized eigenspace, $V_\lambda = \overline{V}_{\lambda}$

**Generalized eigenvectors**

similar to generalized eigenspaces  

are vectors that are not eigenvectors yet, but are useful to draw Jordan blocks  

so it's v such that:   
$(A−λI)^k v=0$  
for some k in poi, but not for smaller k  

note that it's NOT an eigenvector for $A^k$  
to see it just try to expand the power of the bracket, get a hard expression   

#### Compute

**Find matrix describing it**

1. find eigenvalues  
2. for each, plug the eigenvalue in $\lambda$   
this gives the matrix of the eigenspace for that eigenvalue  

$V_{λ_1} =  ker(M − \lambda _1 I)$  

#### Properties

**Properties**

where M is the matrix describing the transformation  
eigenspaces of different eigenvalues are:  

- orthogonal (between themselves, with canonic scalar prod)    
if and only if M is autoaggiunta (symmetric)  

- in direct sum  $V = \oplus V_{\lambda i}$   
if and only if M is diagonalizable (so it's a property only of the generalized eigenvalues)    

- M-invariant: always    
in particular, on generalized eigenspaces, $f(W) = W$, not $\subset$, since it won't grow anymore either  
$f(W) \subset W$ mean that all stuff in W keeps being in W after applying f, but some more stuff could end up in it too after applying f  

note: when a matrix is diagonalizable, its eigenspaces are generalized eigenspaces, so have their properties
