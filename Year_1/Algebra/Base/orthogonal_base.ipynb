### Orthogonal base

all pair of vectors are orthogonal (must be defined a scalar product)  
no need to be orthogonal to themselves  


**Lagrange th**  

every vector space has an orthogonal base  
even if the matrix describing it is not diagonalizable in R  
if it's not diagonaliazable, the vectors in the base won't be eigenvectors, but other stuff

**Properties**

matrix of scalar product defined on orthogonal base is diagonal  
obv, bc all entries not in diagonal are scalr prod between two vectors, and if they're orthogonal gives 0  

**Conditions**

condition orthogonal: $<v_1,v_2> = 0$  
condition independent: $a_1 v_1 + a_2 v_2 = 0$ has only banal solution  
condition positive product: $<v,v> \, > 0$   

### Orthonormal base

orthogonal base where all vectors have norm 1  
GS gives orthonormal  

if the product is positive defined, each orthogonal can be turned to orthonormal by normalizing all vectors    
means fivide each by the SQUARE ROOT of its norm  

the norm is computed using the product that the base must be orthoonal wrt to  
so if want base orthogonormal wrt $\varphi$, norm is $\varphi(v,v)$  
and in G.S. too, the product used is this one wrt which it must be orthogonal  

### Gram Schmidt algorithm

**Steps**

1. check product is positive (always non-zero, except when plugging in null vector)  
2. pick a base (canonic usually)  
3. transform each vector  

v: vectors in canonic base  
w: vectors in new base i'm making  

$w_{i+i} = v_{i+1} - \Sigma \frac{\varphi(v_{i+1}, w_k)}{\varphi(w_k, w_k)} w_k$  

intuitively:   
to create new vector, pick the one in old base $v_{i+1}$  
and delete from it the projections of all the vectors already in new base $w_k$  

**Example**

$w_1 = v_1$

$w_2 = v_2 - \frac{\varphi(v_2, w_1)}{\varphi(w_1, w_1)} w_1$

**Extend orthogonal base**

if have orthogonal base, want to extend it to bigger orthogonal base:  

1. extend to base of V (for now no longer orthogonal)  
2. do G.S. algorithm on new part (first part can be left as is, already orthogonal, even if i tried would not change)

**How to compute scalar prod**

for a difficult vector: express it as comb of canonic base  
$\varphi(v,v)=\varphi(e_1+3e_2, e_1+3e_2)= \varphi(e_1,e_1)+9\varphi(e_2, e_2) + 6\varphi(e_1,e_2)$  
these values can be taken from the matrix  

### Find orthogonal base: method 2

1. if have already a vector in the radical, start with that one: $v_1$    
everything else will be orthogonal to it, so won't need to check it's orthogonal to stuff

2. pick a random vector $v_2$, the easier i can (like $v_2 = 1$)    
until now it's definitely ok, bc previous one is in radical  

3. must pick one that is orthogonal to $v_2$ (it's definitely orth to $v_1$)  
so put condition $\varphi(v_2, v_3) = ... = 0$  
choose entries on $v_3$ so condition is true    
see how to develop the formula in the ex to find an isotropic vector  

**Useful formula**

prove orthogonality implies independence if product is positive  

take $v_1, ... v_k$ orthogonal  
write linear combination: $\Sigma _i \: a_i v_i = 0$  

condition independent: there is only the obvious solution to the linear combination  
condition orthogonal: $\varphi(\Sigma _i \: a_i v_i, \Sigma _i \: a_i v_i) = 0$   

expand the condition for orthogonal:  
$\varphi(\Sigma _i \: a_i v_i, \Sigma _i \: a_i v_i) = \Sigma _i \Sigma _j \: a_i a_j \varphi(v_i, v_j) =$  
(since orthogonal, product between different vectors is null, only keep terms where $i=j$)    
$ = \Sigma a_i^2 \varphi(v_i,v_i)$  

since by hp $ \varphi > 0 $, to get 0 ( condition for orthogonality) all a must be null  
so only sol is the obvious one, they're independent  

**Easier version for 2 vectors**

consider a pair off orthogonal vector, prove they're independent:  
condition orthogonal: $<v_1,v_2> = 0$  
condition independent: $a_1 v_1 + a_2 v_2 = 0$ has only banal solution  
condition positive product: $<v,v> \, > 0$   

expand orthogonal condition
$<v_1, v_2  > = <v_1, -v_1 \frac{a_1}{a_2}>= -\frac{a_1}{a_2}<v_1,v_1> =0$  
to be 0, since the pduct is stricly positive, must have $a_1=0$  
thus clearly, to get 0, also $a_2=0$  
thus have banal solution, they're independent  


### Find orthogonal base: other methods  

#### Method 1: Lagrange theorem

**Steps** 

1. pick base for V
2. write matrix describing $\varphi$ in that base  
3. pick non-isotropic vector $\vec{v}$  
4. find $(span\{\vec{v}\})^{\perp} = V^n$    

repeat until $V^n$ has dim=1  
each time, pick a vector, then leave as possibilities for the next choice only the ones orthogonal to it

1. pick base  

if just starting, canonic base is fine  
else, must pick base from vectors in $V^n$ found in previous step  

3. non-isotropic vector  

if a number on a digonal is non-zero, pick vector correspoing to that col  (bc product with itself is non-zero, so non-isotropic)   

4. find perpendicular span  

solve:  

$\vec{x} \: M (\varphi) \: \vec{v} = 0$   
x is vector with all $x_i$ entries  
find conditions on them  
get carth eq and base for $V^n$  

**How it looks like**

the + is actually direct sum  

$V_0 = R^3$  given by text  

$V_0 = span{(1,1,0)} + (span\{(1,1,0)\})^{\perp} = span{(1,1,0)} + V_1$  


$V_1 = span{(1,-1,0)} + (span\{(1,-1,0)\})^{\perp} =  span{(1,-1,0)} + V_2$  

$V_2 = span{(-1,-1,1)}$ 
  
dim=1, done  


these three vectors form orthogonal base  

#### Method 2: diagonalization

have A, matrix describing $\varphi _A$  

need to diagonalize A  

bc if can find P such that:  
$P^t A P = diagonal$   
means cols of P are orthogonal base  

**Steps**

work in loops  
each time want to set to zero a number not on the diagonal:

1. choose number to set to zero  
2. find $P_0$ to delete it 
3. multiply $P^t A P$  

when A is diagonalized:  
multiply all P used:  $ P = P_0 P_1 ...$  
cols of P are orthogonal base  


**How to choose P**

start from identity, same shape as A  
pick a number to set to zero  
find a way to do it combining col: eg, a number in second col is put to zero subtracting first from second col  
then second col of $P_0$ put $(-1,1,0,...)$   
all other cols stay identity
