#### With images

**Recap**

1. express vectors of new start base as linear comb of vectors of old start base  

2. compute all things needed to write new matrix on vectors of new base  
by substituting vectors in new base with combination of vectors in old base, and then using linearity  
transformation: need image of each vector   
scalar product: need scalar product of each pair of vectors  

**Base change of transformation matrix**

1. express vectors of new start base as linear comb of vectors of old start base  

2. compute image of each vector in new base  
$f(b_i) = Ab_i$  

3. express these images as linear comb of vectors in new end base   
$[f(b_i)]_{B_o}$  

the coeffs of the linear comb are a col of the matrix  
if end base is canonical, the coeffs is the image itself  

**Example**

$B = \{v_1, v_2, v_3\}, \quad B'=\{e_1, e_2, e_3\}$  

step 1:  
$e_1 = 2v_1 -v_3 ...$  

step 2:  
$e_1 = 2v_1 -v_3 \rightarrow f(e_1) = 2f(v_1) - f(v_3)$  

**Base change of scalar product**

want to find <u>scalar product for each cell of new matrix</u>, so $\varphi(e_1,e_1), \varphi(e_1,e_2), ... \varphi(e_3, e_3)$  

1. express vectors of new base as linear comb of vectors of old base  

2. compute scalar product for all pairs of vectors in new base  
plug the linear comb of the old vectors instead of the new vectors 
split the thing with bilinearity  
now have scalar product of old vectors, whose results are the cells of the matrix in the old base

eg: $\varphi(e_2 , e_3) = \varphi(−v_2 − v_3 , v_1 + v_3) = - \varphi(v_2 , v_1) - \varphi(v_3 , v_1) − \varphi(v_2 , v_3) − \varphi(v_3 , v_3 )$  

note scalar product only has one base, no start and end base, so no step 3  

**Shortcuts for transformation**

- change only starting base  
write <u>each vector of NEW start base as linear comb of OLD start base</u>  
so <u>compute image of vectors in new start base</u>, are cols of new matrix  

- change only end base  
write <u>each vector of OLD end base as linear comb of NEW end base</u>  
so compute image of vectors in new end base, are cols of new matrix
