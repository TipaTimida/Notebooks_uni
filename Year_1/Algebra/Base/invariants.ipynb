### Types of invariants

complete: A,B are in that relation if and only if this thing is the same  

non-complete: if A,B are in that relation, this thing is the same  
thus if such thing is different, A and B cannot be in that relation  
but even if it is the same they're not necessarily in the relation  

- complete: $\quad$  thing invariant $\longleftrightarrow $ relation  
- not complete:  $\quad$ thing invariant $\longleftarrow $ relation  

### Base change

**Congruence vs similitude**

base change for:  
- endomorphisms (transformation matrix): similitude  
- scalar product: congruence  

to do base change on an endomorphism:  
$B = P^{-1}AP$  

base change on a scalar product:  
$B = P^t AP$

all P used must be invertible (and therefore square matrices)  
P needs to be invertible bc it's equivalency, so is symmetrical

all equivalence relations form partition on the space of all matrices
