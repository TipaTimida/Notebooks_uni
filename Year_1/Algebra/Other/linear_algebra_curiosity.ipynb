**Alternative way to find eigenvalue 2x2**

A = $\begin{bmatrix} a & b \\ c & d \end{bmatrix}$

Properties of eigenvalues (can also use them to check):
- mean: $ m = a+d$
- product: $ p = det (A)$

so, to compute them:

i can say:  
$ mean -d = \lambda _1 \\ mean + d = \lambda _2 \\
p = (mean -d ) (mean -d ) = mean^2 - d^2$

$\lambda_{1,2} = m \pm \sqrt{m^2 - p}$

\begin{bmatrix} 1 \\ 0 \end{bmatrix}$

$x_1 = \frac{det \begin{vmatrix} 1 & b \\ 0 & d \end{vmatrix}}{det \begin{vmatrix} a & b \\ c & d \end{vmatrix}}$  


$x_2 = \frac{det \begin{vmatrix} c & 1 \\ c & 0 \end{vmatrix}}{det \begin{vmatrix} a & b \\ c & d \end{vmatrix}}$
