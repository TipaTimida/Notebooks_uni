every matrix is similar to only one jordan matrix  
if M is diagonal, its jordan form is its diagonal form  

a jordan matrix is made of jordan blocks  

**Jordan form vs eigenspace**

each eigenspace can be represented by more than one jordan block  
if eigenspace of eigenvalue $\lambda$ has multiplicty $m_a$ and $m_g$, means:  
- there are $m_g$ jordan blocks with this $\lambda$  
- adding sizes of all jordan blocks with this $\lambda$, get $m_a$  

so diagonalizable matrx, which has $m_a = m_g$, has all blocks of size 1  
also works viceversa, if all blocks are size 1 it's diagonalizable  


**Jordan form vs eigenvectors**

each jordan block contains 1 eigenvector (the leftmost col) plus some generalised eigenvectors (the other cols)  
generalized eigenvector aren't eigenvector yet for M, but will be respectively for $M^2, M^3...$, bc each time i apply M i swipe the vector one col to the left (jordan chain)  

eigenvectors belonging to different blocks are independent, bc represent a different multiplicity (rem an eigenspace has as many independent eigenvectors as its $m_g$)  eigenspace
but cols in same block are linearly dependent

### Jordan block

$J_{\lambda n}$  
it's a square matrix, with n rows, with:  

- $\lambda$ on main diag (always the same in all the block)  
- 1 in the diag over the main one  
- 0 all the rest  

$J_{\lambda 3} = \begin{bmatrix}
\lambda & 1 & 0 \\
0 & \lambda & 1 \\
0 & 0 & \lambda
\end{bmatrix}$

### Jordan theorem

given matrix M, if all its eigenvalues are in field K (meaning $p_m(x)$, its char poly, can be factorized in K)  
then M is similar to a block matrix made of jordan blocks, where $\lambda _i$ are the the eigenvalues (not necessarily distinct???) of M:    

$M \sim \begin{bmatrix} 
J_{\lambda _1 n_1} & 0  \\ 
0 & J_{\lambda _2 n_2} \\
\end{bmatrix}$  

### Nilpotent matrices

**Definition**
 
$f^k = 0$  

there is a k for which it becomes identically null (so a zero matrix)  
obv must be an endomorphism, so a square matrix   

**Properties**

- its only eigenvalues is 0  
thus can be written as a block matrix where each block is a jordan block with $\lambda =0$  
- can find a base for it written:  
$v_1, ... v_i, f(v_1), ... f(v_i), f^2(v_1), ... f^2(v_i)...$, until some power (which power???)  

**General base**

for a general matrix, not nilpotent, can find a base for each block by using the genralized eigenvectors:  
$v, (A-\lambda I)v, (A-\lambda I)^2 v...$  
do it for all aigenspaces, changing $\lambda$ to be its eigenvalue  
in nilponent, only eigenvalue is $\lambda =0$  

### Base for Jordan matrix

generalized eigenector of ordine k: $(A-\lambda I)^kv=0$  
generalized eigenspace of ordine k: $V_\lambda^k = \{v \in R^n :(A-\lambda I)^kv=0 \}$  

nota che $V \subset V^2$ etc...

1. compute m_a and m_g for each eigenvalue  
then for each eigenvalue, call m the size of the biggest jordan block referring to that eigenvalue  

2. find base for: $V, V^2, ...V^m$, until the ordine m  
just means to write $A-\lambda I$ like usual, then multiply m times, then find base of that new matrix  

3. pick $v_m$, a vector in $V_m$ but not in $V_{m-1}$  
this will be the last vector of the base  
the otehr are easy, just multiply that per the matrix of the eigenspace:  
$v_{m-1}=(A-\lambda I)v_$  
$v_1 = (A-\lambda I)^{m-1}v_m = (A-\lambda I)v_2$  
see that have a chain, make a base starting with $v_1, ... v_m$, it's the base for the jordan amtrix
