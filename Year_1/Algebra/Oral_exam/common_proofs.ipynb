### Spectral th

**Eigenvalues of symmetric matrix are real**

symmetric: $<Ax,y>=<x,Ay>$  
real eigenvalues: $Av = \lambda v$ where $\lambda = \overline{\lambda}$  

proof:  
$<Av,v>=<v,Av>$  
$<\lambda v,v>=<v, \lambda v>$  
$\lambda ||v|| = \overline{\lambda} ||v||$  
$\lambda = \overline{\lambda}$

so from now on will just extract $\lambda$ from second component, not $\overline{\lambda}$, since i proved they're the same for symmetric matrices  

**Eigenvectors with distinct eigenvalues are orthogonal**

A symmetric  
$Ax = \lambda _x x$  
$Ay = \lambda _y y$  
$<x,y>=0$ if $\lambda _x \neq \lambda _y$  

proof:  
$<Ax,y>=<x, Ay>$  
$<\lambda _x x,y>=<x, \lambda _y y>$  
$\lambda _x <x,y> = \lambda _y <x,y>$  
$(\lambda _x -\lambda _y)<x,y> =0$  
first factor is non-0, so second factor is 0  





**Is diagonalized by a unitary matrix (there is orthonormal basis of eigenvectors)**  


proof by induction  

1x1 case:  
obv it's already diagonal  
has only 1 eigenvalue and 1 eigenvector, so it forms an orthonormal base  

inductive hypothesis:  
if true for (n-1)x(n-1) matrix, true for nxn matrix  

inductive step:  

1--- split matrix in 2  
pick an eigenvalue $\lambda$ and normalize its eigenvector, so $||v||=1$  
define the orthogonal complement (Spazio ortogonale a questo vettore): $V^\perp = \{x \in R^n : <x,v>=0\}$  

2--- prove it's a space  
show that $V^\perp$ is A-invariant (so if $x \in V^\perp, Av \in V^\perp$)  
$<Ax,v>=<x,Av>=<x, \lambda v>=\lambda <x,v>=0$  
$=0$ is because my hypothesis is that $x \in V^\perp$, so $<x,v>=0$  

3--- conclusions  
consider the restriction of A to $V^\perp$, $A|_{V^\perp}$     
it's an $V^\perp$ is an (n-1)x(n-1) space, and A is symmetric, so has orthonormal base of eigenvector by inductive hypothesis  
also they're all orthogonal to $v$, because they're in its orthogonal complement  

qed   

### Binet theorem

$det(AB)=det(A)det(B)$  

**Definition of determinant**

Leibniz formula:  

$det(M)=\Sigma _{\sigma \in S_n} sgn(\sigma) \prod ^n_{i=1} m_{i, \sigma(i)}$  

let A,B be $R^{nxn}$ matrices  
$C=AB$, so $c_{ij}=\Sigma _{k=1}^n a_{ik}b_{kj}$  
each entry ij is inner product of row i of A per col j of B, inner product is sum of all products  



### Binet th - method 2

assume $det(B)\neq 0$, else it's obvious  

$\varphi(A)=\frac{det(AB)}{det(B)}$  

show that $\varphi(A)$ has properties:  
- multilinearity in rows of A  
- alternating   
- normalization  

if so, by uniqueness of the determinant function, $\varphi(A)=det(A)$  

**Proof multilinear**

let $A=R^{nxn}$  

1--- replace the $i$ row of A with a linear combination, $A_i = \lambda v + \mu w$  
$A_v$ is A with row i replaced by v  
$A_w$ is A with row i replaced by w  

2--- since det is multilinear:  
$det(AB)= \lambda \det(A_vB) + \mu \det(A_wB)$  
$\varphi(A)= \lambda \frac{\det(A_vB)}{\det (B)} + \mu \frac{\det(A_wB)}{\det (B)} = \lambda \; \varphi(A_v)+ \mu \;\varphi(A_w)$  

so $\varphi$ is multilinear too  

note that $\lambda$ is extrated as such from the determinant, not as $\lambda ^n$, bc it's multiplied ONLY TO A ROW, the changed one, not to the whole matrix, it's not $\lambda A$  


**What it means to prove alternating**  

to prove alternating can just prove that:  
- $\varphi(A)=0$ if two rows are equal 
- AND $\varphi$ is multilinear  
  
proof: if multilinear and 0 when two equal rows, it's alternating  

$A$: original matrix  
$A'$: swap rows i,j  

define matrix M where rows i,j are:  
$M_i =M_j = A_i + A_j$  

since f is multilinear, can write  
$f(M)=2f(A)+2f(A')$  
bc of first condition, $f(M)=0$  
so $f(A)=-f(A')$  

**Proof alternating**

just prove $\varphi(A)=0$ if two rows are equal (bc already proves it's multilinear) 

A: original matrix  
A': after swapping 2 rows  

if swapped two rows in A, they're also swapped in $C=AB$  
bc $C_{i*}= A_{i*}B$  alternating 
so a row in C is that row in A; multiplied by B  

so $\det(A'B)=\det(C')=-\det(C)=-\det(AB)$  
$\varphi(A') = \frac{\det(A'B)}{\det(B)} = -\frac{\det(AB)}{\det(B)}=-\varphi(A)$  

**Normalization**

if $A=I$:  

$\varphi(I) = \frac{det(B)}{det(B)} = 1$  



### Signature and restrictions  

have scalar product of signature $(i_+, i_-, i_0)$  

must exist a restriction of size $i_+$  where the scalar product is defined positive (means all NORMS are positive)  
$i_0$ is dim of radical  

1--- diagonalization (Sylvester law of inertia)  
scalar product is symmetric, so can be diagonalized bc of Spectral th  
so can find orthonormal base B where the matrix is diagonal, with entries $+1$ $i_+$ times, $-1$ $i_-$ times, 0 $i_0$ times  

2--- exists a subspace of dimension $i+$  on which the prod is positive definite  
pick $U=span\{v_1, ... v_{i_+}\}$, whose base is the eigenvectors of positive eigenvalues  
vectors in it are: $u = \Sigma_i^\alpha \alpha _i v_i$  
check the scalar product:  
$<u,u> = \Sigma \alpha _i^2 >0$  
so scalar prod is positive  

3--- not exist a bigger one  
assume a bigger positive restriction exist  
by Silvester law, when diagonalizing should have $i_+ +1$ times the $+1$ in the main diagonal, which cannot happen  


computation:  

$<u,u> = < \Sigma_i^\alpha \alpha _i v_i, \Sigma_j^\alpha \alpha _j v_j > = \Sigma_i^\alpha \alpha _i \Sigma_j^\alpha \alpha _j <v_i,v_j>$  

base of U is made of orthogonal vectors (bc in symmetric matrix can diagonalize with orthogonal base of eigenvectors)  
so $<v_i,v_j>=0$ if $i \neq 0$, 1 if $i=j$    
$<u,u>=\Sigma \alpha_i^2 $
