## Definizione

Il determinante è l'unico funzionale multilineare, alternante e normalizzato.  

funzionale si può scrivere: $\varphi: KxK \rightarrow K$

multilineare: lineare in ogni componente  
(le componenti sono le righe della matrice)  

alternante: scambiando due componenti, cambia il segno, ma non il modulo  

normalizzato: $\varphi(I)=1$
