- 0 sol (if zero-row with non-zero final col)

### Check it's base

**Check B is base for W**

know that n independent vectors form base of $R^n$

conditions:
1. independent
2. $W = span(B)$   

2 methods to prove 1st condition:  
- row reduction, check all have pivots  
- make matrix with vectors of base, check $det \neq 0$  

2 methods to prove 2nd condition:    
- check that vectors in B <u>belong in W</u>    
check number of vectors in B is <u>equal to dim(W)</u>  
- express general vector of W (the one with min number of vars) as linear comb of vectors in B

**Vector belong in subspace**

does $\bar{b} \in W$ if $W=span(\bar{a_1}\bar{a_2})$  
solve $\bar{a_1}x_1 + \bar{a_2}x_2 = \bar{b}$  
if has sol, $\bar{b} \in W$  
$\bar{b}$ can also be generic $\bar{x}$, in this case must check that sol exists for all values



### Check it's subspace

**Condition**

1. contains null vector: $\bar{0} \in W$  
need to understand what $\bar{0}$ is in this context  
2. closed wrt addiction
3. closed wrt multiplication per scalar

**Question**

W is subset of something i know to be a vector space  ($R^n, Mat_{2x2}$)  
want to check if W is too   

if subspace of vectors or matrices or linear functions, <u>useful to write general matrix/vector that belongs in it with free vars, make it easier to check conditions</u> for being a space

**How to check conditions formally**

1. null vector  

define what null vector is here, might be null matrix  
unless i'm in an affine space, it's full of zeroes  
check if it respects the condition to belong in the space


2. close wrt sum

pick two objs, $x_1, x_2 \in W$  
plug $x_1 + x_2$ in the condition, check if it's respected

3. close wrt mult  

pick $x \in W$  
plug $\lambda x$ in the condition, check if it's respected  
check if it's true if $\lambda \in C$, eg try with $\lambda=i$    



to prove something is false, either do it with algebra, or show specific case where it fails (eg write 2 specific vectors in W whose sum is not in W)

$ 2x + 3y +1 = 0$ is not through the origin, miss $\bar{0}$, not subspace

**Over which field is it a subspace?**

check both scalars:  
 $\lambda \in R$  
 $\lambda \in C$\ $R$

1. field of values allowed as entries

### Check it's linear

**Linear application**

$ f : V \rightarrow W$

conditions for linearity:
1. $f(\vec{0}_v) = \vec{0}_w$  (don't move origin, map null vectors) 
2. $f(\vec{v}+ \vec{u}) = f(\vec{v}) + f (\vec{u})$
3. $f(\alpha\vec{v}) = \alpha f(\vec{v})$  

ops on left defined in V, on right defined in W  

to prove something is linear, know that sum and composition of linear func is linear (AA linear if A is)  

usually given a formula  $<v_1, v_2>$  

prove it respects the properties of the product in question (linearity etc..)  
can do by plugging in the stuff required, see how behaves
