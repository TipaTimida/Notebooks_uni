#### Kernel and image

**Definition**

$f : V \rightarrow W$  

- kernel: space of vectors of V mapped into null vector  
$ker(f) = \{  \vec{v} \in V : f(\vec{v}) = \vec{0} \}$  

- image: space of vectors in W reachable with f from V  
$Im(f) = \{  \vec{w} \in V :\: \exists    \vec{v} \in V : \: f(\vec{v}) = \vec{w} \}$

kernel and image are subpaces respectively of V and W  

since kernel and image belong to V and W, differnet spaces, the same vectors can be in both ker and Im (and thus be in the base of both)  
if v is in base of both ker and im, means Av=0 (bc base of ker), and also $Ae_1 =v$ (bc base of Im)

**Behaviour kernel**

each time i apply f the kernel can only grow, not shrink ( can't create new dimensions)  
bc stuff in null vector is mapped in null vector, $f(0)=0$, is condition of linear transformations (means origin can't be translated)  

at one specific number of applications of f the kernel will stay the same size: it has stabilized, from now it will stay like this forever  
so first time kernel doesn't grow, won't grow anymore
