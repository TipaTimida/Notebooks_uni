### How to find dimension

**How to find dimension of matrix**

$dim(A)$:  
- number of vectors in base
- number of pivot vars  

how to find it:  

- have <u>span</u> (generatotr matrix)  
count <u>number of independent vectors</u>   
dim= number of pivots  

- have <u>carthesian eq</u> (coefficient matrix)   
count <u>how many vars i can choose</u> before all the others become obligated   
the number of ones i can pick is the degrees of freedom, and the dimension  
dim= number of NON-pivots (n-pivots)


**Tips**

write <u>generic matrix that satisfies all conditions</u> of the space   
see <u>how many free params</u> are there   
free params are params i can pick, before all the others become obligated  
number of free params = dimension, bc each param would be put to 1 to get a different matrix for the base  
so if no conditions, 3x3 matrix has dim=9 


**Write general form for matrix of a subspace**

2 methods: 

1. find images of vectors in input base 

2. find base of that subspace (so a base made of matrices, by combining these matrices can get any matrix in the subspace)   

each matrix of the subspace is a linear comb, so write $M = aA + bB + cC$ if have 3 matrices A,B,C in the base  
then add all to a single matrix with the incognite a,b,c

### Dimensions when putting together spaces

**Dimension and bases of space**

- $dim(A + B)$  
  to make matrix, use GENERATORS of both (bc more free)    
  must count PIVOT VARS  


- $dim(A\cap B)$  
  to make matrix, use CARTH. EQ. of both (bc more constrained)   
  must count FREE VARS, not pivot vars (called the nullity) !!!  


why count free vars in intersection?   
in a system with n vars:  
- rank=n: system has solution, intersection is a point
- rank=n-1: system has one degree of freedom, intersection is a line  
- rank=n-2: system has 2 degrees of freedom, intersection is a plane  


to remember:  
$A+B$ is more free, $A \cap B$ is more stringent, so use more carth eq, that are restraints  

**Grassman formula**

$dim(A+B) = dim(A) + dim(B) - dim(A \cap B)$  

so if one is difficult, can find the others   
know generators -> find sum  
know carth equations -> find union   
then formula

**Image and kernel: dimension formula**

$dim(Im f) + dim(ker f) = dim(dominium f)$  

to remember:  

- pivot vars: $dim(Im)$  = $rank(f)$
- free vars: $dim(ker)$ 
- number of cols: $dim(dominium) $

it's derivated from Grassman formula:  
- intersection of image and kernel is null  
- their direct sum is the dominium

### Exercises with functions

$f: R^a \rightarrow R^b$  

transformation can only delete dimensions, not add them, so:  
$dim(Im(f)) \le a$  
note that can be $b>a$, but $b$ is the space where is the image, not the image  
in general, $dim(Im(f)) \neq b$  

doing more functions can only shrink space:  
$dim(Im(f) \cup Im (g)) \le dim(Im(f))$  

injectivity:  $dim(Im f) = a$  
surjectivity: $dim(Im f) = b$  

**Injective, surjective**

$ f: V \rightarrow W$ 

conditions: 
1. injective: $dim(V) = dim(Im \: f)$    
  so, if V=W (endomorphism), if $dim(ker f) = 0$, or $ker(f) = \{\vec{0}\}$  
  

2. surjective: $dim(W) = dim(Im \:f)$  

**Can it be unjective or surjective?**

remember intuitive definition:  
injective: stuff in W is touched AT MOST once  
surjective: stuff in W is touched AT LEAST once

1. $dim(V) > dim(W)  $  
cannot be injective   
dimension of the kernel is at least difference in dimension   
$f: R^5 \rightarrow R^4 \quad dim(ker f) \ge 1$

2. $dim(V) < dim(W)  $   
cannot be surjective  
so can't be invertible  

3. $dim(V) = dim(W) $  
<u>either both injective and surjective or neither</u>   
prove it's one thing, these use Grassman formula to prove the other    

essentially:  
$dim(Im \, f)$ cannot be bigger than a or b (the dimension of the codominium/dominium)  
so if a and b are different, the dim of the image is at most as big as the smaller one  

composition of injective trasformations is injective  
if A injective, $A^2$ is too
